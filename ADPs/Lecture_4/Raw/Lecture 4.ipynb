## 2. ЗАЧЕМ ВИЗУАЛИЗИРОВАТЬ: ОСНОВНЫЕ ФУНКЦИИ ГРАФИКОВ

Выделим четыре ключевые функции визуализации в психологических исследованиях.

**1. Разведочный анализ (Exploratory Data Analysis)**

График помогает **увидеть** данные до того, как вы начнёте считать статистики. Выбросы, пропуски, аномалии, форма распределения, нелинейные зависимости — всё это становится видимым.

*Пример из практики:* вы построили гистограмму и увидели, что данные имеют две вершины (бимодальное распределение). Это сигнал: возможно, в вашей выборке смешаны две разные группы (например, пациенты с лёгкой и тяжёлой депрессией). Если бы вы не построили график, вы могли бы упустить это важное открытие и неправильно интерпретировать среднее.

**2. Проверка допущений статистических тестов**

Многие статистические тесты (например, t-тест, ANOVA) требуют, чтобы данные были распределены нормально. График (гистограмма или Q-Q-plot) помогает оценить это визуально — до применения тестов. Если распределение сильно скошено, вы можете выбрать непараметрические тесты.

*Пример:* вы видите на гистограмме, что данные имеют длинный "хвост" справа. Это значит, что среднее будет смещено вверх, и для описания центра лучше использовать медиану (как мы обсуждали в Лекции 2).

**3. Сравнение групп и условий**

Графики позволяют сравнить несколько групп на одном рисунке. Это особенно полезно в экспериментальных и клинических исследованиях.

*Пример:* ящик с усами для трёх групп терапии (КПТ, медикаменты, контроль) показывает не только средние, но и разброс, и выбросы. Вы сразу видите, какая группа показывает лучшие результаты и насколько однородны данные внутри каждой группы.

**4. Коммуникация результатов**

График — самый эффективный способ донести результаты до коллег, заказчиков, пациентов [1]. Статья с хорошими графиками читается легче и запоминается лучше. В APA-стиле (Американская психологическая ассоциация) графики являются обязательным элементом публикаций [1].

*Пример:* вместо того чтобы писать "средний балл BDI снизился с 32.5 до 18.3", вы показываете линейный график, где видно плавное снижение. Это убедительнее и нагляднее.

### 4.1. Гистограмма (Histogram)

**Что это:** график распределения одной количественной переменной. Данные разбиваются на интервалы («корзины», bins), и для каждого интервала показывается частота — сколько значений попало в этот интервал.

**Когда использовать:** чтобы понять форму распределения, проверить нормальность, увидеть выбросы и пробелы.

**Психологический пример:**

> **Задача:** исследователь хочет понять, как распределены баллы по шкале депрессии Бека (BDI) в выборке из 150 пациентов. Есть ли много людей с очень высокими баллами? Распределение симметричное или скошенное?

**Как построить гистограмму (блок-схема):**

```mermaid
flowchart LR
    subgraph Данные
        D["150 пациентов, баллы BDI от 0 до 50"]
    end
    subgraph График
        G["Ось X: интервалы баллов BDI (0-5, 6-10, …)"]
        G2["Ось Y: количество пациентов"]
    end
    subgraph Вывод
        V1["Распределение смещено вправо"]
        V2["Есть выбросы >45"]
    end
    Данные --> График --> Вывод
```

#### 📊 Код для гистограммы (с комментариями)

```python
# 1. Импортируем нужные библиотеки
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Создаём пример данных (150 пациентов, баллы BDI от 0 до 50)
#    В реальности вы загрузите свои данные из файла: data = pd.read_csv('file.csv')
np.random.seed(42)  # чтобы результат был воспроизводимым
bdi_scores = np.random.normal(loc=25, scale=10, size=150)  # нормальное распределение
bdi_scores = np.clip(bdi_scores, 0, 50)  # обрезаем до шкалы 0–50
data = pd.DataFrame({'BDI': bdi_scores})

# 3. Строим гистограмму
plt.figure(figsize=(10, 6))                 # размер рисунка (ширина, высота)
plt.hist(data['BDI'],                       # данные
         bins=12,                           # количество интервалов
         edgecolor='black',                 # обводка столбцов
         color='skyblue',                   # цвет заливки
         alpha=0.7)                         # прозрачность (0-1)

# 4. Добавляем линии среднего и медианы
mean_val = data['BDI'].mean()
median_val = data['BDI'].median()
plt.axvline(mean_val, color='red', linestyle='--', linewidth=2,
            label=f'Среднее = {mean_val:.1f}')
plt.axvline(median_val, color='green', linestyle='--', linewidth=2,
            label=f'Медиана = {median_val:.1f}')

# 5. Оформляем по правилам APA
plt.title('Рисунок 1. Распределение баллов по шкале депрессии Бека (N=150)',
          fontsize=14, fontweight='bold')
plt.xlabel('Балл по шкале BDI', fontsize=12)
plt.ylabel('Количество пациентов', fontsize=12)
plt.legend()                                # показываем легенду
plt.grid(axis='y', alpha=0.3)               # сетка по Y для удобства
plt.tight_layout()                          # автоматически подгоняет элементы
plt.show()                                  # выводим график на экран
```

**Что можно увидеть на гистограмме:**
- **Нормальное распределение** — колоколообразная форма, симметричная.
- **Скошенность** — «хвост» справа (высокие значения) или слева (низкие значения). Это важно для выбора мер центральной тенденции (как в Лекции 2).
- **Бимодальность** — две вершины. Это может означать, что в выборке смешаны две разные группы.
- **Выбросы** — отдельные столбцы далеко от основной массы.

**Практические рекомендации:**
- Для небольших выборок (до 30) используйте 5–7 интервалов.
- Для больших выборок (100+) — 10–15 интервалов.
- В психологических статьях обычно указывают количество интервалов в примечании к рисунку.

### 4.2. Ящик с усами (Boxplot)

**Что это:** график, показывающий распределение через пять чисел: минимум, первый квартиль (Q1), медиана, третий квартиль (Q3), максимум. «Усы» показывают размах без выбросов, а выбросы рисуются отдельными точками.  
*Примечание:* в большинстве библиотек усы ограничены 1.5×IQR, а не абсолютным минимумом/максимумом — это стандарт.

**Когда использовать:** чтобы сравнить распределения нескольких групп, увидеть выбросы, показать разброс данных.

**Структура ящика с усами:**

```mermaid
flowchart TD
    subgraph Box["Ящик с усами (Boxplot)"]
        Max["Максимум (верхний ус)"]
        Q3["Q3 — 75-й перцентиль"]
        Median["Медиана"]
        Q1["Q1 — 25-й перцентиль"]
        Min["Минимум (нижний ус)"]
        Outlier["Выбросы — точки за усами"]
    end
    style Box fill:#e1f5fe,stroke:#01579b
```

**Психологический пример:**

> **Задача:** психолог сравнивает уровень тревоги (по шкале Спилбергера) в трёх группах: пациенты с депрессией, пациенты с тревожным расстройством, здоровые испытуемые. Нужно показать не только средние, но и разброс.

#### 📊 Код для ящика с усами (сравнение трёх групп)

```python
# 1. Подготавливаем данные: три группы по 50 человек
np.random.seed(42)
group_dep = np.random.normal(45, 10, 50)    # депрессия
group_anx = np.random.normal(50, 12, 50)    # тревога
group_health = np.random.normal(30, 8, 50)  # здоровые
# Собираем в DataFrame
df_box = pd.DataFrame({
    'Тревога': np.concatenate([group_dep, group_anx, group_health]),
    'Группа': ['Депрессия']*50 + ['Тревога']*50 + ['Здоровые']*50
})

# 2. Строим boxplot с помощью seaborn
plt.figure(figsize=(10, 6))
sns.boxplot(data=df_box, x='Группа', y='Тревога',
            palette='Set2',            # цветовая палитра (дружелюбная к дальтоникам)
            width=0.5)                # ширина ящиков

# 3. Оформление по APA
plt.title('Рисунок 2. Сравнение уровня тревоги между группами',
          fontsize=14, fontweight='bold')
plt.xlabel('Тип группы', fontsize=12)
plt.ylabel('Балл по шкале тревоги Спилбергера', fontsize=12)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

**Что можно увидеть на boxplot:**
- **Медиану** (линия внутри ящика) — центральная тенденция (устойчива к выбросам).
- **Разброс** (высота ящика = IQR) — насколько данные разбросаны.
- **Симметричность** — если медиана по центру ящика, распределение симметрично; если смещена — есть скошенность.
- **Выбросы** (точки за усами) — аномальные значения, требующие внимания.

**Важное предостережение:** boxplot показывает **грубую** картину. Он не даёт информации о форме распределения внутри ящика (там может быть бимодальность). Поэтому его часто комбинируют с гистограммой или скрипичным графиком (violin plot).

### 4.3. Точечный график (Scatter plot)

**Что это:** график, показывающий связь между двумя количественными переменными. Каждая точка — один респондент. X и Y — значения двух переменных.

**Когда использовать:** чтобы увидеть корреляцию, кластеры, нелинейные зависимости, выбросы.

**Психологический пример:**

> **Задача:** исследователь хочет проверить, есть ли связь между уровнем тревоги (Y) и качеством сна (X) у пациентов с депрессией. Точки показывают каждого пациента.

#### 📊 Код для точечного графика с линией регрессии

```python
# 1. Создаём данные: 50 пациентов, качество сна (0–10) и тревога (0–80)
np.random.seed(42)
n = 50
sleep_quality = np.random.uniform(2, 9, n)                # X
# тревога = 80 - 6*сон + шум, чтобы была отрицательная корреляция
anxiety = 80 - 6 * sleep_quality + np.random.normal(0, 8, n)
anxiety = np.clip(anxiety, 0, 80)                         # не выходим за шкалу
df_scatter = pd.DataFrame({'Сон': sleep_quality, 'Тревога': anxiety})

# 2. Строим scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(df_scatter['Сон'], df_scatter['Тревога'],
            alpha=0.7, color='blue', s=60)   # s – размер точек

# 3. Добавляем линию регрессии (тренд)
from scipy import stats
slope, intercept, r_value, p_value, std_err = stats.linregress(df_scatter['Сон'],
                                                                df_scatter['Тревога'])
x_vals = np.linspace(df_scatter['Сон'].min(), df_scatter['Сон'].max(), 100)
y_vals = slope * x_vals + intercept
plt.plot(x_vals, y_vals, color='red', linestyle='-', linewidth=2,
         label=f'Линия регрессии (r = {r_value:.2f})')

# 4. Оформление
plt.title('Рисунок 3. Связь между качеством сна и уровнем тревоги у пациентов с депрессией',
          fontsize=14, fontweight='bold')
plt.xlabel('Качество сна (0–10)', fontsize=12)
plt.ylabel('Уровень тревоги (0–80)', fontsize=12)
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

**Что можно увидеть на scatter plot:**
- **Корреляцию** — точки вытянуты вдоль линии (положительной или отрицательной).
- **Кластеры** — точки образуют группы (возможно, разные подгруппы пациентов).
- **Нелинейную связь** — точки образуют дугу или другую сложную форму.
- **Выбросы** — точки, далёкие от основной массы, которые могут искажать корреляцию.

**Дополнительные возможности:**
- **Цвет** точек = категориальная переменная (например, пол) — тогда добавьте параметр `c=df['Пол']`.
- **Размер** точек = третья количественная переменная (например, возраст) — параметр `s=df['Возраст']*10`.
- **Линия регрессии** — мы добавили её в коде, она показывает общую тенденцию.

**Важно:** scatter plot показывает **наличие** связи, но не её силу. Для измерения силы связи нужна корреляция (коэффициент Пирсона или Спирмена) [2].

### 4.4. Тепловая карта (Heatmap)

**Что это:** график, где значения представлены цветом. Каждая ячейка — это значение на пересечении двух переменных. Цветовая шкала позволяет мгновенно увидеть паттерны.

**Когда использовать:** для визуализации матриц (корреляционной матрицы, таблиц сопряжённости), когда нужно показать паттерн, а не отдельные числа.

**Два основных типа в психологии:**

**А) Тепловая карта корреляционной матрицы:**

> **Задача:** психолог измерил 10 шкал (тревога, депрессия, сон, настроение, социальная активность, самооценка и др.) и хочет увидеть, какие шкалы сильно коррелируют, а какие — нет.

#### 📊 Код для тепловой карты корреляций

```python
# 1. Создаём искусственные данные для 5 шкал (100 испытуемых)
np.random.seed(42)
n_subj = 100
# Генерируем коррелированные данные
depression = np.random.normal(20, 5, n_subj)
anxiety = 0.7 * depression + np.random.normal(0, 3, n_subj)
sleep = -0.5 * depression + np.random.normal(7, 1.5, n_subj)
mood = 0.6 * depression + np.random.normal(5, 2, n_subj)
social = -0.3 * depression + np.random.normal(6, 2, n_subj)

df_corr = pd.DataFrame({
    'Депрессия': depression,
    'Тревога': anxiety,
    'Сон': sleep,
    'Настроение': mood,
    'Социальная активность': social
})

# 2. Считаем корреляционную матрицу
corr_matrix = df_corr.corr()

# 3. Строим тепловую карту
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix,
            annot=True,               # показывать числа в ячейках
            fmt='.2f',                # формат чисел (2 знака)
            cmap='coolwarm',          # цветовая схема (красный-белый-синий)
            center=0,                 # центр цветовой шкалы (0 – нет корреляции)
            square=True,              # квадратные ячейки
            linewidths=0.5,           # границы между ячейками
            cbar_kws={'shrink': 0.8}) # размер цветовой шкалы
plt.title('Рисунок 4. Корреляционная матрица психологических шкал (N=100)',
          fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

**Что можно увидеть:**
- Кластеры сильно коррелирующих шкал (например, депрессия и тревога). Это может указывать на то, что они измеряют сходные конструкты.
- Шкалы, которые почти не коррелируют с другими (изолированные) — возможно, они измеряют что-то уникальное.

**Б) Тепловая карта таблицы сопряжённости:**

> **Задача:** психолог изучает связь между диагнозом (депрессия/тревога/ПТСР) и типом предпочитаемой терапии (КПТ/психодинамическая/гуманистическая).

#### 📊 Код для тепловой карты таблицы сопряжённости

```python
# 1. Создаём таблицу частот (например, опрос 200 пациентов)
np.random.seed(42)
diagnoses = np.random.choice(['Депрессия', 'Тревога', 'ПТСР'], size=200, p=[0.4, 0.35, 0.25])
therapy = []
for d in diagnoses:
    if d == 'Депрессия':
        therapy.append(np.random.choice(['КПТ', 'Психодинамическая', 'Гуманистическая'], p=[0.6, 0.2, 0.2]))
    elif d == 'Тревога':
        therapy.append(np.random.choice(['КПТ', 'Психодинамическая', 'Гуманистическая'], p=[0.3, 0.5, 0.2]))
    else:  # ПТСР
        therapy.append(np.random.choice(['КПТ', 'Психодинамическая', 'Гуманистическая'], p=[0.4, 0.3, 0.3]))

df_contingency = pd.DataFrame({'Диагноз': diagnoses, 'Терапия': therapy})
# Строим таблицу сопряжённости
contingency_table = pd.crosstab(df_contingency['Диагноз'], df_contingency['Терапия'])

# 2. Тепловая карта
plt.figure(figsize=(8, 5))
sns.heatmap(contingency_table,
            annot=True,            # показывать числа
            fmt='d',               # целые числа
            cmap='Blues',          # голубая палитра
            cbar_kws={'label': 'Количество пациентов'})
plt.title('Рисунок 5. Связь между диагнозом и предпочитаемой терапией')
plt.xlabel('Тип терапии')
plt.ylabel('Диагноз')
plt.tight_layout()
plt.show()
```

**Что можно увидеть:** паттерны выбора — например, пациенты с депрессией чаще выбирают КПТ, а с тревогой — психодинамическую терапию.

### 4.5. Линейный график (Line plot)

**Что это:** график, показывающий изменение переменной вдоль оси X (обычно время или порядковая переменная). Точки соединяются линией, чтобы подчеркнуть динамику.

**Когда использовать:** для отображения динамики, трендов, изменений во времени.

**Психологический пример:**

> **Задача:** психолог отслеживает уровень депрессии (по шкале Бека) у пациента в ходе 12 сессий КПТ. Нужно показать, как снижаются симптомы.

#### 📊 Код для линейного графика (один пациент)

```python
# 1. Данные: 12 сессий, баллы BDI
sessions = np.arange(1, 13)
bdi_scores = [32, 28, 25, 21, 19, 17, 16, 18, 15, 13, 12, 11]  # на 8-й сессии подъём

# 2. Строим линейный график
plt.figure(figsize=(10, 6))
plt.plot(sessions, bdi_scores, marker='o', linestyle='-', color='blue',
         linewidth=2, markersize=8, label='Уровень депрессии (BDI)')

# 3. Оформление
plt.title('Рисунок 6. Динамика депрессии в ходе 12 сессий КПТ',
          fontsize=14, fontweight='bold')
plt.xlabel('Номер сессии', fontsize=12)
plt.ylabel('Балл по шкале BDI', fontsize=12)
plt.xticks(sessions)                  # показываем все номера сессий
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

**Что можно увидеть на line plot:**
- **Тренд** — возрастание, убывание или стабильность.
- **Скорость изменений** — крутизна наклона (быстрое или медленное изменение).
- **Аномалии** — резкие скачки (например, подъём на 8-й сессии, который требует дополнительного анализа).

**Варианты с несколькими линиями (сравнение групп):**

```python
# Пример: две группы (КПТ и медикаменты) – средние показатели за 12 недель
weeks = np.arange(1, 13)
cbt_means = [30, 27, 24, 21, 19, 17, 15, 14, 13, 12, 11, 10]
med_means = [31, 29, 27, 26, 25, 24, 23, 22, 21, 20, 19, 18]

plt.figure(figsize=(10, 6))
plt.plot(weeks, cbt_means, marker='s', label='КПТ', color='green', linewidth=2)
plt.plot(weeks, med_means, marker='^', label='Медикаменты', color='orange', linewidth=2)
plt.title('Рисунок 7. Сравнение динамики снижения депрессии в двух группах')
plt.xlabel('Неделя терапии')
plt.ylabel('Средний балл BDI')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

**Доверительные интервалы** можно добавить, используя `plt.fill_between()` (это более сложный вариант, но при желании можно изучить дополнительно).

## 6. ПРАКТИЧЕСКОЕ УПРАЖНЕНИЕ НА ПАРЕ (5–7 минут)

> **Задача:** у вас есть данные 40 пациентов: уровень депрессии (BDI) и тип терапии (КПТ, медикаменты, контроль). Нужно показать различие между группами.

### Вопросы для обсуждения

1. **Какой тип графика вы выберете?**  
   – Лучший выбор — **ящик с усами (boxplot)**, так как он показывает медиану, разброс (IQR) и выбросы для каждой группы, что идеально для сравнения.

2. **Что должно быть на осях?**  
   – Ось X: тип терапии (категориальная переменная).  
   – Ось Y: балл BDI (количественная переменная).

3. **Что вы ожидаете увидеть?**  
   – Например, в группе КПТ медиана ниже, чем в контроле; в группе медикаментов — промежуточное значение. Возможны выбросы (пациенты с аномально высокими или низкими баллами).

4. **Как оформить график для научной статьи?**  
   – Добавить номер и название, подписи осей, легенду (если нужно), примечание со статистиками (среднее, стандартное отклонение, количество наблюдений).

### Готовый код для упражнения

```python
# 1. Импорт библиотек
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Создаём пример данных (40 пациентов)
np.random.seed(42)
group_cbt = np.random.normal(20, 5, 14)      # КПТ, средний балл 20
group_med = np.random.normal(25, 6, 13)      # медикаменты, средний 25
group_control = np.random.normal(30, 7, 13)  # контроль, средний 30

df_ex = pd.DataFrame({
    'BDI': np.concatenate([group_cbt, group_med, group_control]),
    'Терапия': ['КПТ']*14 + ['Медикаменты']*13 + ['Контроль']*13
})

# 3. Строим ящик с усами
plt.figure(figsize=(10, 6))
sns.boxplot(data=df_ex, x='Терапия', y='BDI',
            palette='Set2', width=0.5)

# 4. Оформление по APA
plt.title('Рисунок А. Уровень депрессии (BDI) в разных группах терапии (N=40)',
          fontsize=14, fontweight='bold')
plt.xlabel('Тип терапии', fontsize=12)
plt.ylabel('Балл по шкале депрессии Бека', fontsize=12)
plt.grid(axis='y', alpha=0.3)

# Добавляем примечание со статистиками
means = df_ex.groupby('Терапия')['BDI'].mean().round(1)
stds = df_ex.groupby('Терапия')['BDI'].std().round(1)
note_text = f"Средние (M ± SD): КПТ = {means['КПТ']} ± {stds['КПТ']}, " \
            f"Медикаменты = {means['Медикаменты']} ± {stds['Медикаменты']}, " \
            f"Контроль = {means['Контроль']} ± {stds['Контроль']}"
plt.figtext(0.5, 0.01, note_text, ha='center', fontsize=10,
            bbox=dict(facecolor='white', alpha=0.8))

plt.tight_layout(rect=[0, 0.05, 1, 1])
plt.show()
```

**Обсуждение альтернатив:**  
Столбчатая диаграмма со стандартными ошибками тоже возможна, но она скрывает распределение внутри группы и не показывает выбросы. Поэтому **boxplot** предпочтительнее для разведочного анализа.

## 7. ОБЗОРНЫЙ КОД ДЛЯ ДЕМОНСТРАЦИИ

Этот код охватывает все основные графики, рассмотренные в лекции. Вы можете скопировать его целиком и запустить в своей среде (Jupyter Notebook, Google Colab и др.). Все шаги подробно прокомментированы на русском.

```python
# ============================================================
# 0. ПОДГОТОВКА (импорт библиотек и создание данных)
# ============================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

np.random.seed(42)
bdi = np.random.normal(25, 10, 150)
bdi = np.clip(bdi, 0, 50)

group_labels = ['КПТ']*50 + ['Медикаменты']*50 + ['Контроль']*50
bdi_group = bdi.copy()
bdi_group[0:50] += 2
bdi_group[50:100] -= 1
bdi_group[100:150] += 3
bdi_group = np.clip(bdi_group, 0, 50)

sleep = 10 - bdi/10 + 0.5 * np.random.randn(150) * 2
sleep = np.clip(sleep, 0, 10)

data = pd.DataFrame({
    'BDI': bdi_group,
    'Группа': group_labels,
    'Сон': sleep
})

# ============================================================
# ГРАФИК 1. ГИСТОГРАММА (распределение одной переменной)
# ============================================================
plt.figure(figsize=(10, 6))
plt.hist(data['BDI'], bins=12, edgecolor='black', color='skyblue', alpha=0.7)
plt.axvline(data['BDI'].mean(), color='red', linestyle='--', linewidth=2,
            label=f'Среднее = {data["BDI"].mean():.1f}')
plt.axvline(data['BDI'].median(), color='green', linestyle='--', linewidth=2,
            label=f'Медиана = {data["BDI"].median():.1f}')
plt.title('Рисунок 1. Распределение баллов по шкале депрессии Бека (N=150)',
          fontsize=14, fontweight='bold')
plt.xlabel('Балл BDI', fontsize=12)
plt.ylabel('Количество пациентов', fontsize=12)
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# ============================================================
# ГРАФИК 2. ЯЩИК С УСАМИ (сравнение трёх групп)
# ============================================================
plt.figure(figsize=(10, 6))
sns.boxplot(data=data, x='Группа', y='BDI', palette='Set2', width=0.5)
plt.title('Рисунок 2. Сравнение уровня депрессии между группами терапии',
          fontsize=14, fontweight='bold')
plt.xlabel('Тип терапии', fontsize=12)
plt.ylabel('Балл BDI', fontsize=12)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# ============================================================
# ГРАФИК 3. ТОЧЕЧНЫЙ ГРАФИК (связь двух переменных + регрессия)
# ============================================================
plt.figure(figsize=(10, 6))
plt.scatter(data['Сон'], data['BDI'], alpha=0.6, color='blue', s=50)

slope, intercept, r_value, p_value, std_err = stats.linregress(data['Сон'],
                                                                data['BDI'])
x_vals = np.linspace(data['Сон'].min(), data['Сон'].max(), 100)
y_vals = slope * x_vals + intercept
plt.plot(x_vals, y_vals, color='red', linewidth=2,
         label=f'Линия регрессии (r = {r_value:.2f})')

plt.title('Рисунок 3. Связь между качеством сна и уровнем депрессии',
          fontsize=14, fontweight='bold')
plt.xlabel('Качество сна (0–10)', fontsize=12)
plt.ylabel('Балл BDI', fontsize=12)
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# ============================================================
# ГРАФИК 4. ТЕПЛОВАЯ КАРТА (корреляционная матрица)
# ============================================================
np.random.seed(42)
anxiety = 0.6 * data['BDI'] + np.random.normal(0, 5, 150)
mood = 0.5 * data['BDI'] + np.random.normal(0, 4, 150)
social = -0.3 * data['BDI'] + np.random.normal(0, 3, 150)
df_corr = pd.DataFrame({
    'BDI': data['BDI'],
    'Тревога': anxiety,
    'Сон': data['Сон'],
    'Настроение': mood,
    'Социальная активность': social
})
corr_matrix = df_corr.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm',
            center=0, square=True, linewidths=0.5,
            cbar_kws={'shrink': 0.8})
plt.title('Рисунок 4. Корреляционная матрица психологических шкал (N=150)',
          fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# ============================================================
# ГРАФИК 5. ЛИНЕЙНЫЙ ГРАФИК (динамика одного пациента)
# ============================================================
sessions = np.arange(1, 13)
bdi_dynamics = [32, 28, 25, 21, 19, 17, 16, 18, 15, 13, 12, 11]
plt.figure(figsize=(10, 6))
plt.plot(sessions, bdi_dynamics, marker='o', linestyle='-', color='blue',
         linewidth=2, markersize=8, label='Уровень депрессии')
plt.title('Рисунок 5. Динамика депрессии в ходе 12 сессий КПТ',
          fontsize=14, fontweight='bold')
plt.xlabel('Номер сессии', fontsize=12)
plt.ylabel('Балл BDI', fontsize=12)
plt.xticks(sessions)
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

**Что важно понять из кода (пояснение для студентов):**

| Строка кода | Что делает для психолога |
|-------------|---------------------------|
| `plt.hist(bins=12)` | Разбивает данные на интервалы и показывает частоту – видим форму распределения |
| `sns.boxplot(x='Группа', y='BDI')` | Сравнивает группы по медиане, разбросу и выбросам |
| `plt.scatter()` + линия регрессии | Показывает связь между двумя переменными и её силу (коэффициент корреляции) |
| `sns.heatmap(corr_matrix, annot=True)` | Визуализирует все попарные корреляции в виде цветовой матрицы |
| `plt.plot(marker='o')` | Отображает изменение показателя во времени |

**Главный вывод:** код — это универсальный шаблон. Вы можете заменить искусственные данные на свои (например, загрузить CSV-файл командой `pd.read_csv()`) и получить готовые графики для своей курсовой или диплома. Главное — правильно выбрать тип графика под вашу задачу.

## 8. ЗАКЛЮЧЕНИЕ

Резюмируем пройденный путь: от сырых цифр к графикам, которые помогают увидеть закономерности и донести результаты до других.

**Ключевые этапы выбора графика:**
1. Определите тип переменных (количественные/категориальные).
2. Сформулируйте исследовательский вопрос (распределение? сравнение? связь? динамика?).
3. Выберите график по алгоритму (см. раздел 3).
4. Оформите график по правилам APA (см. раздел 5).

**Основные типы графиков и их задачи:**

| График | Задача | Психологический пример |
|--------|--------|----------------------|
| Гистограмма | Показать распределение | Баллы BDI в выборке |
| Ящик с усами | Сравнить группы | Тревога по типам терапии |
| Точечный график | Показать связь | Корреляция тревоги и сна |
| Тепловая карта | Показать паттерны | Корреляционная матрица |
| Линейный график | Показать динамику | Изменения в ходе терапии |

**Этический аспект:** график должен **отражать реальность**, а не создавать иллюзию. Изменение масштаба, обрезание данных, выбор "удобного" типа графика — это манипуляция. Ответственность за честность визуализации лежит на авторе [3].

**Связь с предыдущими лекциями:**
- В Лекции 2 мы научились описывать данные статистиками. Теперь мы умеем показывать их наглядно.
- В Лекции 3 мы научились превращать текст в числа. Теперь эти числа можно визуализировать.
- В будущих лекциях мы будем использовать визуализацию для интерпретации результатов моделей машинного обучения.

**Главная мысль напоследок:**
> График — это не просто иллюстрация. Это инструмент мышления. Хороший график помогает увидеть то, что скрыто в цифрах, и задать правильные вопросы. Плохой график может ввести в заблуждение. Поэтому ответственность психолога — делать визуализацию честной, прозрачной и понятной.
