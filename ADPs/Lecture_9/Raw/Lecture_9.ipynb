### 1.1. От облачных LLM к точным ответам: почему психологу нужен RAG

В Лекции 8 мы научились запускать локальные LLM через Ollama. Мы видели, как они генерируют связные, осмысленные ответы. Но есть одна проблема: **LLM не знают, когда они чего-то не знают**.

**Интуитивная аналогия из психотерапии:**

> Представьте, что вы – молодой терапевт, который только что окончил университет. Вы много читали, но у вас нет доступа к библиотеке. Пациент спрашивает вас о редком расстройстве. Вы не знаете ответа, но вместо того чтобы признаться, вы начинаете **уверенно говорить** – и выдаёте ложную информацию, потому что ваш мозг «дорисовывает» недостающие детали. Это и есть **галлюцинация LLM** – модель генерирует убедительный, но ложный ответ, потому что она не умеет говорить «я не знаю».

**Что такое RAG и как он помогает:**

**RAG (Retrieval-Augmented Generation)** – это технология, которая даёт LLM доступ к внешней базе знаний. Вместо того чтобы «гадать», модель сначала **ищет** ответ в ваших документах, а затем **генерирует** ответ на основе найденных фактов.

**Интуитивная аналогия из психотерапии:**

> Представьте, что у вас есть **личный библиотекарь**, который:
> 1. Слышит вопрос пациента.
> 2. Мгновенно находит нужную книгу на полке.
> 3. Открывает нужную страницу.
> 4. Показывает вам точную цитату.
> 5. Вы читаете её пациенту, **ссылаясь на источник**.
>
> Это и есть RAG – модель больше не «выдумывает», она **ищет и цитирует**.

**Исследования подтверждают эффективность RAG:**

Современные исследования показывают, что RAG значительно снижает количество галлюцинаций у LLM [5]. Модели, использующие RAG, реже «выдумывают» ответы и чаще опираются на факты из проверенных источников.

В клинической психологии это особенно важно. Исследование 2025 года показало, что RAG-системы достигают точности **76.1%** при диагностических рекомендациях в области тревожных и психотических расстройств [1]. Это означает, что в 76 случаях из 100 система даёт верные диагностические рекомендации.

**Что означают эти цифры для психолога:**

| Показатель | Что это значит |
|------------|----------------|
| **Снижение галлюцинаций** | Модель реже выдаёт ложную информацию, чаще опирается на факты |
| **Точность 76.1%** | В 76% случаев RAG-система даёт правильные диагностические рекомендации |

**Важно понимать:** 76.1% – это не 100%. Это означает, что RAG-система **не заменяет** врача, но является **надёжным помощником**, который может:
- Быстро находить релевантную информацию из проверенных источников.
- Снижать риск ошибок, связанных с человеческой усталостью.
- Экономить время на поиске информации в учебниках и статьях.

## 2. ПРОБЛЕМА ГАЛЛЮЦИНАЦИЙ LLM

### 2.1. Что такое галлюцинации

**Галлюцинация** – это когда LLM генерирует информацию, которая **звучит правдоподобно, но фактически неверна** или отсутствует в обучающих данных.

**Примеры галлюцинаций в психологическом контексте:**

| Вопрос пользователя | Галлюцинация модели | Правильный ответ |
|---------------------|---------------------|------------------|
| «Каковы критерии ПТСР по МКБ-11?» | Модель называет устаревшие критерии из МКБ-10 | Критерии МКБ-11 отличаются (добавлен симптом «сложное ПТСР») |
| «Какая дозировка сертралина при депрессии?» | Модель называет неверную дозировку | Дозировка зависит от множества факторов; модель не должна давать медицинские рекомендации |
| «Кто автор теории привязанности?» | Модель ошибочно называет другого учёного | Джон Боулби |

**Почему LLM галлюцинируют:**

| Причина | Объяснение |
|---------|------------|
| **LLM не знают, чего они не знают** | Они обучены предсказывать следующее слово, а не проверять факты |
| **Обучение на устаревших данных** | Модель может «знать» информацию, которая была актуальна на момент обучения, но уже устарела |
| **Отсутствие доступа к источникам** | Модель не может «посмотреть» в книгу или статью – она полагается только на свои внутренние веса |
| **Вероятностная природа** | LLM генерируют наиболее вероятное продолжение текста, а не наиболее точное |

**Интуитивная аналогия:**

> Галлюцинации LLM – это как если бы вы спросили у коллеги, который никогда не читал первоисточники, но очень уверенно отвечает на все вопросы. Он может быть прав в 80% случаев, но в 20% – он просто «дорисовывает» ответ.

### 2.2. Почему галлюцинации особенно опасны в психологии

| Риск | Что может пойти не так | Последствия |
|------|------------------------|-------------|
| **Диагностические ошибки** | Модель рекомендует неверные диагностические критерии | Неправильный диагноз, неверное лечение |
| **Юридические риски** | Ложная информация может привести к неправильным решениям | Иски, потеря лицензии |
| **Потеря доверия** | Если модель постоянно ошибается, пациенты и коллеги перестают ей доверять | Отказ от использования полезных ИИ-инструментов |
| **Этические нарушения** | Использование ложной информации в клинической практике недопустимо | Нарушение профессиональной этики |

### 2.3. Схема: проблема галлюцинаций

## 3. ЧТО ТАКОЕ RAG И КАК ОН РАБОТАЕТ

### 3.1. Определение RAG

**Retrieval-Augmented Generation (RAG)** – это архитектура, которая **дополняет** LLM внешней базой знаний. Перед тем как сгенерировать ответ, система **ищет** релевантные фрагменты в документах и **подставляет** их в промпт.

**Интуитивная аналогия:**

> Представьте, что вы – терапевт, и у вас есть доступ к **библиотеке** с тысячами книг по психологии. Когда пациент задаёт вопрос, вы не отвечаете сходу – вы **идёте к полке**, **находите нужную книгу**, **открываете нужную страницу** и **читаете** ответ. Только после этого вы даёте ответ пациенту, **ссылаясь** на источник. RAG делает то же самое, но в тысячи раз быстрее.

**Три шага RAG:**

```mermaid
flowchart LR
    subgraph STEP1["1️⃣ ПОИСК (Retrieval)"]
        S1["Запрос пользователя"]
        S2["Поиск в векторной БД"]
        S3["Найдены релевантные<br>фрагменты"]
        S1 --> S2 --> S3
    end
    
    subgraph STEP2["2️⃣ ПОДСТАНОВКА (Augmentation)"]
        S4["Фрагменты + запрос"]
        S5["Формирование промпта"]
        S4 --> S5
    end
    
    subgraph STEP3["3️⃣ ГЕНЕРАЦИЯ (Generation)"]
        S6["LLM читает промпт"]
        S7["Генерирует ответ<br>на основе фактов"]
        S6 --> S7
    end
    
    STEP1 --> STEP2 --> STEP3
    
    style STEP1 fill:#e1f5fe,stroke:#01579b
    style STEP2 fill:#fff9c4,stroke:#f57f17
    style STEP3 fill:#c8e6c9,stroke:#2e7d32
```

### 3.2. Как RAG решает проблему галлюцинаций

| Без RAG | С RAG |
|---------|-------|
| Модель «гадает» ответ | Модель **ищет** ответ в документах |
| Нет доступа к источникам | Есть **цитаты** и **ссылки** на источники |
| Не может проверить факты | Может **проверить** по первоисточнику |
| Ответ может быть ложным | Ответ **основан на фактах** из документов |
| Неясно, откуда взят ответ | **Прозрачность** – видно, на каком фрагменте основан ответ |

**Ключевое преимущество:** RAG позволяет **проверять** ответы. Вы всегда можете посмотреть, из какого фрагмента был взят ответ, и убедиться в его достоверности.

### 3.3. Пример RAG в психологии

**Вопрос:** *«Каковы основные симптомы депрессии по DSM-5?»*

**Без RAG:** Модель генерирует ответ на основе своих весов – может ошибиться.

**С RAG:**
1. Система ищет в базе знаний (например, в учебнике по психиатрии).
2. Находит фрагмент: *«Согласно DSM-5, для диагностики депрессии необходимо наличие 5 из 9 симптомов в течение 2 недель...»*
3. Подставляет этот фрагмент в промпт.
4. LLM генерирует ответ **на основе этого фрагмента**.
5. В ответе есть **ссылка**: *«Источник: DSM-5, стр. 123»*.

### 3.4. Схема: RAG vs обычная LLM

```mermaid
flowchart LR
    subgraph WITHOUT["❌ БЕЗ RAG"]
        W1["Вопрос"] --> W2["LLM"]
        W2 --> W3["Ответ (может быть<br>галлюцинацией)"]
    end
    
    subgraph WITH["✅ С RAG"]
        R1["Вопрос"] --> R2["Поиск в<br>векторной БД"]
        R2 --> R3["Релевантные<br>фрагменты"]
        R3 --> R4["Промпт:<br>фрагменты + вопрос"]
        R4 --> R5["LLM"]
        R5 --> R6["Ответ на основе<br>фактов + источники"]
    end
    
    WITHOUT -->|Проблема| WITH
    
    style WITHOUT fill:#ffcdd2,stroke:#c62828
    style WITH fill:#c8e6c9,stroke:#2e7d32
```

## 4.1. Чанкинг (Chunking) — разбиение текста на фрагменты

**Что это:** процесс разбиения большого документа на **небольшие смысловые фрагменты** – «чанки» (chunks). LLM не могут обработать целую книгу целиком (из-за ограничения на длину контекста), поэтому мы разбиваем текст на части, которые модель может «переварить».

**Интуитивная аналогия из психотерапии:**

> Представьте, что у вас есть толстый учебник по психологии. Вы не можете держать его открытым целиком – вы открываете его на **одной странице** и читаете **один абзац**. Чанкинг – это как разрезание книги на отдельные страницы или абзацы, чтобы потом легко находить нужный фрагмент.

**Почему чанкинг важен:**

| Причина | Объяснение |
|---------|------------|
| **Ограничение контекста** | LLM могут обработать только ограниченное количество токенов за раз (например, 8192 токена) |
| **Точность поиска** | Маленькие фрагменты легче сравнивать с вопросом – выше точность |
| **Скорость** | Поиск по маленьким фрагментам быстрее |
| **Релевантность** | Можно найти именно тот фрагмент, где содержится ответ |

**Стратегии чанкинга:**

| Стратегия | Описание | Когда использовать | Пример из психологии |
|-----------|----------|-------------------|---------------------|
| **Фиксированный размер** | Разбивка на чанки по N токенов (например, 256) | Для простых документов | Тексты без чёткой структуры |
| **По предложениям** | Разбивка по границам предложений | Для художественных текстов | Дневниковые записи пациентов |
| **По абзацам** | Разбивка по абзацам | Для учебников и статей | **Учебники по психологии** |
| **Семантический чанкинг** | Разбивка по смысловым границам (с помощью ML) | Для сложных документов | Научные статьи с терминами |

**Ключевое понятие: перекрытие (Overlap)**

При чанкинге важно использовать **перекрытие** – когда чанки частично пересекаются. Это нужно, чтобы важная информация не «потерялась» на границе между чанками.

**Интуитивная аналогия:**

> Представьте, что вы разрезаете книгу на страницы. Если вы разрежете ровно по границе предложения, одно предложение может оказаться на двух страницах. Чтобы этого избежать, вы оставляете небольшое **перекрытие** – последние 1-2 строки предыдущей страницы дублируются на следующей.

**Рекомендация для психологических текстов:**

| Параметр | Рекомендация | Почему |
|----------|--------------|--------|
| **Размер чанка** | 256–512 токенов | Достаточно для смыслового фрагмента |
| **Перекрытие** | 50–100 токенов | Чтобы не потерять смысл на границах |
| **Разделители** | Абзацы, предложения | Учитывает структуру текста |

**Пример чанкинга в коде:**

```python
!pip install langchain-text-splitters

from langchain_text_splitters import RecursiveCharacterTextSplitter

# Создаём сплиттер для психологических текстов
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,      # размер чанка в символах
    chunk_overlap=50,    # перекрытие между чанками
    separators=["\n\n", "\n", ". ", " ", ""]  # приоритет разделителей
)

# Пример: фрагмент из учебника по психологии
text = """
Согласно DSM-5, для диагностики депрессии необходимо наличие 5 из 9 симптомов
в течение 2 недель. Симптомы включают: подавленное настроение, снижение интереса,
изменения аппетита, нарушения сна, психомоторные изменения, усталость, чувство вины,
снижение концентрации, суицидальные мысли.
"""

# Разбиваем на чанки
chunks = text_splitter.split_text(text)
print(f"Создано {len(chunks)} чанков")
for i, chunk in enumerate(chunks):
    print(f"Чанк {i+1}: {chunk[:100]}...")
```

## 4.2. Эмбеддинги (Embeddings) — превращение текста в числа

**Что это:** **Эмбеддинг** – это вектор чисел (обычно 384–1536 чисел), который **кодирует смысл** текстового фрагмента. Близкие по смыслу тексты имеют **похожие эмбеддинги** – они находятся рядом в многомерном пространстве.

**Интуитивная аналогия из психологии:**

> Представьте, что у вас есть карта города, где **смысл** – это географическое положение. Слова «психология» и «терапия» находятся рядом на этой карте (похожий смысл), а слово «кулинария» – далеко. Эмбеддинги – это **координаты** каждого текста на этой карте.

**Как работают эмбеддинги (пошагово):**

```mermaid
flowchart LR
    A["Текст:<br>«Симптомы депрессии»"] --> B["Модель эмбеддингов"]
    B --> C["Вектор:<br>[0.12, -0.34, 0.56, ...]"]
    D["Текст:<br>«Критерии депрессии»"] --> B
    B --> E["Вектор:<br>[0.11, -0.32, 0.58, ...]"]
    F["Текст:<br>«Рецепт пирога»"] --> B
    B --> G["Вектор:<br>[0.89, 0.45, -0.12, ...]"]
    
    C -.->|"Близко"| E
    E -.->|"Далеко"| G
    
    style A fill:#e1f5fe,stroke:#01579b
    style D fill:#e1f5fe,stroke:#01579b
    style F fill:#ffcdd2,stroke:#c62828
```

**Почему эмбеддинги важны для RAG:**

| Причина | Объяснение |
|---------|------------|
| **Семантический поиск** | Находят тексты по смыслу, а не по ключевым словам |
| **Скорость** | Поиск по числам (векторам) намного быстрее, чем по тексту |
| **Масштабируемость** | Можно обрабатывать миллионы документов |
| **Мультиязычность** | Некоторые модели работают с разными языками |

**Модели эмбеддингов для русского языка:**

| Модель | Размер вектора | Особенности | Для психолога |
|--------|---------------|-------------|---------------|
| **multilingual-e5-large** | 1024 | Поддерживает 100+ языков, включая русский | **Рекомендуется для начала** |
| **bge-m3** | 1024 | Флагманская модель от BAAI | Для высокоточного поиска |
| **rubert-tiny2** | 312 | Компактная и быстрая | Для on-premise систем |
| **Qwen3-Embedding** | 1536 | Мультиязычный RAG, SOTA-метрики | Для сложных RAG-систем |

**Рекомендация для психологов:** для начала используйте **multilingual-e5-large** – она хорошо работает с русским языком и доступна бесплатно.

**Пример создания эмбеддингов в коде:**

```python
from sentence_transformers import SentenceTransformer

# Загружаем модель эмбеддингов
embedding_model = SentenceTransformer('intfloat/multilingual-e5-large')

# Тексты для сравнения
texts = [
    "Симптомы депрессии: подавленное настроение, потеря интереса",
    "Критерии депрессии по DSM-5: 5 из 9 симптомов в течение 2 недель",
    "Рецепт приготовления яблочного пирога"
]

# Создаём эмбеддинги
embeddings = embedding_model.encode(texts)

# Сравниваем близость
from sklearn.metrics.pairwise import cosine_similarity

# Насколько похожи первые два текста?
similarity = cosine_similarity([embeddings[0]], [embeddings[1]])[0][0]
print(f"Сходство текста 1 и 2: {similarity:.3f}")  # Близко к 1.0

# Насколько похожи первый и третий тексты?
similarity = cosine_similarity([embeddings[0]], [embeddings[2]])[0][0]
print(f"Сходство текста 1 и 3: {similarity:.3f}")  # Близко к 0.0
```

**Ожидаемый вывод:**

```
Сходство текста 1 и 2: 0.892
Сходство текста 1 и 3: 0.124
```

**Что это значит:** модель «понимает», что первый и второй тексты говорят об одном (депрессия), а третий – о совершенно другом (кулинария).

## 4.3. Векторное хранилище (Vector Database) — Chroma

**Что это:** база данных, которая **хранит эмбеддинги** и позволяет **быстро искать** похожие векторы.

**Chroma** – это открытая векторная база данных, которая:
- Хранит эмбеддинги на диске.
- Поддерживает поиск по семантической близости.
- Интегрируется с Python и LangChain.
- Бесплатна и проста в использовании.

**Интуитивная аналогия:**

> Представьте, что у вас есть **картотека** всех фрагментов из учебников. Каждая карточка – это фрагмент текста и его «координаты» на карте смыслов (эмбеддинг). Когда вы задаёте вопрос, вы вычисляете его «координаты» и находите **ближайшие** карточки в этой картотеке. Chroma – это и есть такая **умная картотека**.

**Как работает векторный поиск:**

```mermaid
flowchart LR
    subgraph INDEX["📇 ИНДЕКСАЦИЯ (подготовка)"]
        direction TB
        A["Документы"] --> B["Чанки"]
        B --> C["Эмбеддинги"]
        C --> D["Chroma<br>(векторное хранилище)"]
    end
    
    subgraph SEARCH["🔍 ПОИСК (во время запроса)"]
        direction TB
        E["Вопрос"] --> F["Эмбеддинг вопроса"]
        F --> G["Поиск ближайших<br>векторов в Chroma"]
        G --> H["Найденные<br>релевантные чанки"]
    end
    
    INDEX --> SEARCH
    
    style INDEX fill:#e1f5fe,stroke:#01579b
    style SEARCH fill:#fff9c4,stroke:#f57f17
```

**Установка Chroma:**

```bash
pip install chromadb
```

**Пример работы с Chroma в коде:**

```python
import chromadb
from sentence_transformers import SentenceTransformer

# 1. Загружаем эмбеддинг-модель
embedding_model = SentenceTransformer('intfloat/multilingual-e5-large')

# 2. Создаём клиент Chroma (с хранением на диске)
client = chromadb.PersistentClient(path="./psychology_db")

# 3. Создаём коллекцию (таблицу) для хранения
collection = client.get_or_create_collection(
    name="psychology_knowledge",
    metadata={"hnsw:space": "cosine"}  # используем косинусное расстояние
)

# 4. Добавляем документы
documents = [
    "Согласно DSM-5, для диагностики депрессии необходимо наличие 5 из 9 симптомов...",
    "Теория привязанности Джона Боулби описывает, как ранние отношения...",
    "Когнитивно-поведенческая терапия фокусируется на связи между мыслями..."
]

# Создаём эмбеддинги для документов
embeddings = embedding_model.encode(documents)

# Добавляем в Chroma
for i, (doc, emb) in enumerate(zip(documents, embeddings)):
    collection.add(
        ids=[f"doc_{i}"],
        documents=[doc],
        embeddings=[emb.tolist()]
    )

print(f"✅ Добавлено {len(documents)} документов в Chroma")

# 5. Поиск
query = "Что такое депрессия?"
query_embedding = embedding_model.encode([query])[0]

results = collection.query(
    query_embeddings=[query_embedding.tolist()],
    n_results=2  # найти 2 самых похожих документа
)

print("\n🔍 Результаты поиска по запросу:", query)
for i, doc in enumerate(results['documents'][0]):
    print(f"{i+1}. {doc[:100]}...")
```

**Ожидаемый вывод:**

```
✅ Добавлено 3 документов в Chroma

🔍 Результаты поиска по запросу: Что такое депрессия?
1. Согласно DSM-5, для диагностики депрессии необходимо наличие 5 из 9 симптомов...
2. Когнитивно-поведенческая терапия фокусируется на связи между мыслями...
```

**Почему Chroma нашёл правильный документ:** эмбеддинг вопроса «Что такое депрессия?» оказался ближе всего к эмбеддингу первого документа (о DSM-5 и симптомах депрессии).

## 4.4. Полный пайплайн RAG: от документа к ответу

**Схема: архитектура RAG**

```mermaid
flowchart LR
    subgraph INGESTION["📥 1. ИНДЕКСАЦИЯ (подготовка базы знаний)"]
        direction TB
        A1["📚 Документы<br>(учебники, статьи)"] --> A2["✂️ Чанкинг<br>разбивка на фрагменты"]
        A2 --> A3["🔢 Эмбеддинги<br>превращение в векторы"]
        A3 --> A4["💾 Chroma<br>векторное хранилище"]
    end
    
    subgraph QUERY["🔍 2. ЗАПРОС (поиск в базе)"]
        direction TB
        B1["❓ Вопрос пользователя"] --> B2["🔢 Эмбеддинг вопроса"]
        B2 --> B3["🎯 Поиск похожих<br>чанков в Chroma"]
        B3 --> B4["📄 Найденные<br>релевантные фрагменты"]
    end
    
    subgraph GENERATION["📝 3. ГЕНЕРАЦИЯ (ответ)"]
        direction TB
        C1["📝 Промпт:<br>фрагменты + вопрос"] --> C2["🧠 LLM (Ollama)"]
        C2 --> C3["✅ Ответ на основе<br>фактов + источники"]
    end
    
    INGESTION -->|"Подготовка"| QUERY
    QUERY -->|"Поиск"| GENERATION
    
    style INGESTION fill:#e1f5fe,stroke:#01579b
    style QUERY fill:#fff9c4,stroke:#f57f17
    style GENERATION fill:#c8e6c9,stroke:#2e7d32
```

**Пошаговое объяснение:**

| Шаг | Что происходит | Пример |
|-----|----------------|--------|
| **1. Индексация** | Документы разбиваются на чанки, превращаются в эмбеддинги и сохраняются в Chroma | Учебник по психологии → тысячи чанков с эмбеддингами |
| **2. Запрос** | Вопрос пользователя превращается в эмбеддинг, ищутся похожие чанки | «Что такое депрессия?» → находим чанк о симптомах депрессии |
| **3. Генерация** | Найденные чанки подставляются в промпт, LLM генерирует ответ | LLM читает чанк и формулирует ответ на его основе |

**Ключевые выводы по разделу 4:**

1. **Чанкинг** – разбивка текста на фрагменты для удобной обработки. Важно использовать перекрытие, чтобы не потерять смысл.

2. **Эмбеддинги** – превращение смысла текста в числа (векторы). Похожие тексты имеют похожие векторы.

3. **Chroma** – векторная база данных для хранения эмбеддингов и быстрого поиска по смыслу.

4. **Три шага RAG** – индексация (подготовка), запрос (поиск), генерация (ответ).

5. **Для психологов** это означает возможность создать персональную базу знаний из учебников, статей и протоколов, которая отвечает на вопросы, ссылаясь на источники.

## 5.1. Вопросно-ответные системы по учебникам и научной литературе

**Что это:** RAG-система, которая отвечает на вопросы психолога, используя его личную библиотеку учебников, статей и клинических протоколов. Это как иметь **персонального научного ассистента**, который помнит всё, что вы когда-либо читали, и может мгновенно найти нужную информацию.

**Интуитивная аналогия из психотерапии:**

> Представьте, что вы — психолог, и у вас есть **личный библиотекарь**, который:
> - Знает каждую книгу в вашей библиотеке.
> - Помнит, где что лежит.
> - Мгновенно находит нужную страницу.
> - Показывает вам точную цитату.
> - Может ответить на любой вопрос по вашим книгам.
>
> Вы задаёте вопрос: «Каковы критерии ПТСР?» — и библиотекарь через 2 секунды даёт вам точную цитату из DSM-5 с номером страницы. Это и есть RAG в действии.

**Как это работает в практике психолога:**

| Этап | Что делает психолог | Что делает RAG-система |
|------|---------------------|----------------------|
| **1. Подготовка** | Загружает учебники, статьи, протоколы в базу знаний | Разбивает тексты на чанки, создаёт эмбеддинги, индексирует |
| **2. Запрос** | Задаёт вопрос о конкретном расстройстве, симптоме или методе | Находит релевантные фрагменты в базе знаний |
| **3. Ответ** | Получает точный ответ с цитатой и ссылкой на источник | Генерирует ответ на основе найденных фрагментов |

| `curl -fsSL https://ollama.com/install.sh \| sh` | Скачивает и запускает официальный установщик |
| `subprocess.Popen(["ollama", "serve"], ...)` | Запускает сервер Ollama в фоновом режиме |
| **Health-Check** | Проверяет, что сервер реально начал принимать запросы |

**Интуитивная аналогия:**

> Представьте, что вы устанавливаете новую программу на компьютер. Иногда после установки программа не запускается с первого раза – нужно удалить старые файлы, перезагрузить компьютер и проверить, что программа действительно работает. Мы делаем то же самое с Ollama в Colab.

```python
# Ячейка 1: Установка и запуск Ollama в Colab

# 1. Очищаем старые "битые" файлы и устанавливаем zstd
!rm -rf /usr/local/lib/ollama /usr/local/bin/ollama
!apt-get update -qq && apt-get install -y zstd

print("⏳ Установка Ollama...")

# 2. Устанавливаем Ollama заново
!curl -fsSL https://ollama.com/install.sh | sh

# 3. Запускаем сервер в фоне
import subprocess
import time
import urllib.request
import urllib.error

print("⏳ Запуск сервера Ollama...")

# Запускаем процесс в фоне, перенаправляя вывод в никуда (чтобы не засорять ячейку)
subprocess.Popen(
    ["ollama", "serve"],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)

# 4. Health-Check: ждём, пока сервер реально начнёт принимать запросы
# Это важно – без этого следующая ячейка может упасть с ошибкой ConnectionError
server_ready = False
for i in range(40):  # пробуем до 40 раз (≈40 секунд)
    try:
        req = urllib.request.urlopen("http://localhost:11434/api/tags", timeout=2)
        if req.getcode() == 200:
            server_ready = True
            break
    except urllib.error.URLError:
        time.sleep(1)

if not server_ready:
    raise Exception("❌ Сервер не запустился! Перезапустите среду выполнения (Runtime -> Restart).")

print("✅ Сервер Ollama запущен и готов!")

# 5. Автоматическое скачивание модели
# Используем llama3.2:1b – она весит всего 1.3 ГБ, быстро скачается и не убьёт память Colab
# Это важно, так как в бесплатном Colab память ограничена (12-15 GB RAM)
MODEL_NAME = "llama3.2:1b"
print(f"⏳ Скачивание модели {MODEL_NAME} (это займёт 1-2 минуты)...")
subprocess.run(["ollama", "pull", MODEL_NAME])

print(f"🎉 ВСЁ ГОТОВО! Модель {MODEL_NAME} загружена. Можно писать код для общения.")
```

### 6.2.5. Шаг 4: Создание базы знаний из психологических текстов

**Что происходит в этом шаге:** Мы создаём базу знаний из психологических текстов. Это сердце RAG-системы – именно здесь хранятся все знания, которые модель будет использовать для ответов.

**Пояснение для студентов:**

1. **Эмбеддинг-модель** – превращает текст в числа. Мы используем `multilingual-e5-large`, которая понимает русский язык.
2. **Чанкинг** – разбиваем текст на небольшие фрагменты для удобного поиска.
3. **Chroma** – сохраняет эмбеддинги в базе данных для быстрого поиска.

```python
# 0. Устанавливаем недостающие пакеты (тихо, чтобы не засорять вывод)
!pip install langchain-text-splitters sentence-transformers chromadb -q

import chromadb
from sentence_transformers import SentenceTransformer
# ИСПРАВЛЕННЫЙ ИМПОРТ (обратите внимание на нижнее подчеркивание)
from langchain_text_splitters import RecursiveCharacterTextSplitter

print("⏳ Загрузка эмбеддинг-модели...")

# 1. Загружаем эмбеддинг-модель для русского языка
embedding_model = SentenceTransformer('intfloat/multilingual-e5-large')
print(f"✅ Эмбеддинг-модель загружена (размер вектора: {embedding_model.get_sentence_embedding_dimension()})")

# 2. Подготовка текстов (фрагменты из учебника по психологии)
print("\n📚 Подготовка психологических текстов...")

documents = [
    "Согласно DSM-5, для диагностики депрессии необходимо наличие 5 из 9 симптомов в течение 2 недель. Симптомы включают: подавленное настроение, снижение интереса, изменения аппетита, нарушения сна, психомоторные изменения, усталость, чувство вины, снижение концентрации, суицидальные мысли.",
    "Теория привязанности Джона Боулби описывает, как ранние отношения с caregiver формируют внутренние рабочие модели, влияющие на последующие отношения в жизни.",
    "Когнитивно-поведенческая терапия (КПТ) фокусируется на связи между мыслями, чувствами и поведением. Основной принцип: изменение дезадаптивных мыслей приводит к изменению эмоций и поведения.",
    "ПТСР (посттравматическое стрессовое расстройство) характеризуется повторяющимися переживаниями травмы, избеганием напоминаний, негативными изменениями в мышлении и повышенной возбудимостью.",
    "МКБ-11 включает диагностическую категорию 'сложное ПТСР' (CPTSD), которая добавляет симптомы нарушения самооценки, эмоциональной регуляции и межличностных отношений."
]
print(f"📝 Загружено {len(documents)} документов")

# 3. Чанкинг: разбиваем документы на фрагменты
print("\n✂️ Разбивка текстов на чанки...")
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=50,
    separators=["\n\n", "\n", ". ", " ", ""]
)

chunks = []
for doc in documents:
    chunks.extend(text_splitter.split_text(doc))
print(f"✅ Создано {len(chunks)} чанков")

# 4. Создаём эмбеддинги для всех чанков
print("\n🔢 Создание эмбеддингов...")
chunk_embeddings = embedding_model.encode(chunks, show_progress_bar=True)
print(f"✅ Создано {len(chunk_embeddings)} эмбеддингов")

# 5. Создаём векторное хранилище Chroma
print("\n💾 Создание векторного хранилища Chroma...")
client = chromadb.PersistentClient(path="./psychology_db")
collection = client.get_or_create_collection(
    name="psychology_knowledge",
    metadata={"hnsw:space": "cosine"}
)

# Добавляем чанки в базу
for i, (chunk, embedding) in enumerate(zip(chunks, chunk_embeddings)):
    collection.add(
        ids=[f"chunk_{i}"],
        documents=[chunk],
        embeddings=[embedding.tolist()],
        metadatas=[{"source": f"doc_{i}"}]
    )

print(f"✅ Добавлено {len(chunks)} чанков в векторную базу")
print("📂 База данных сохранена в папке './psychology_db'")
```

### 6.2.6. Шаг 5: Функция поиска и генерации ответа

**Что происходит в этом шаге:** Мы создаём функцию, которая выполняет полный RAG-пайплайн: поиск релевантных фрагментов и генерацию ответа на их основе.

**Пояснение для студентов:**

| Шаг | Что происходит | Как это выглядит в коде |
|-----|----------------|------------------------|
| **1. Эмбеддинг вопроса** | Вопрос превращается в вектор | `query_embedding = embedding_model.encode([query])[0]` |
| **2. Поиск в Chroma** | Находятся самые похожие чанки | `collection.query(query_embeddings=[...])` |
| **3. Формирование промпта** | Чанки подставляются в промпт | `prompt = f"Фрагменты: {context}\nВопрос: {query}"` |
| **4. Генерация ответа** | LLM генерирует ответ | `requests.post(url, json=payload)` |

```python
import requests

def rag_query(query, top_k=3):
    """
    Полный RAG-пайплайн:
    1. Превращаем вопрос в эмбеддинг
    2. Ищем похожие чанки в Chroma
    3. Формируем промпт с найденными фрагментами
    4. Отправляем в Ollama и получаем ответ
    """
    
    print("\n" + "=" * 50)
    print(f"🔍 Вопрос: {query}")
    print("=" * 50)
    
    # Шаг 1: Эмбеддинг вопроса
    print("⏳ Превращаем вопрос в эмбеддинг...")
    query_embedding = embedding_model.encode([query])[0]
    
    # Шаг 2: Поиск в Chroma
    print(f"🔎 Ищем {top_k} самых похожих фрагментов...")
    results = collection.query(
        query_embeddings=[query_embedding.tolist()],
        n_results=top_k
    )
    
    retrieved_chunks = results['documents'][0]
    retrieved_distances = results['distances'][0]
    
    print(f"📄 Найдено {len(retrieved_chunks)} фрагментов")
    for i, (chunk, dist) in enumerate(zip(retrieved_chunks, retrieved_distances)):
        print(f"   {i+1}. Сходство: {1-dist:.3f} | {chunk[:80]}...")
    
    # Шаг 3: Формируем промпт
    print("📝 Формируем промпт для LLM...")
    context = "\n\n---\n\n".join(retrieved_chunks)
    
    prompt = f"""Ты – психологический ассистент, который отвечает на вопросы,
основываясь ТОЛЬКО на предоставленных фрагментах из учебников и научных статей.

Если в фрагментах нет ответа на вопрос, скажи: "В предоставленных материалах
нет информации по этому вопросу."

Используй фрагменты ДОСЛОВНО, если это возможно. Всегда указывай, из какого
фрагмента взята информация.

Фрагменты для ответа:
---
{context}
---
Вопрос пользователя: {query}

Твой ответ:"""

    # Шаг 4: Генерация через Ollama
    print("🧠 Генерируем ответ через Ollama...")
    url = "http://localhost:11434/api/generate"
    payload = {
        "model": "llama3.2:1b", # ⚠️ Заменили qwen2.5:7b на llama3.2:1b (она точно скачана)
        "prompt": prompt,
        "stream": False,
        "options": {
            "temperature": 0.3  
        }
    }
    
    response = requests.post(url, json=payload)
    result_json = response.json()
    
    # 🛡️ Защита от KeyError: проверяем, есть ли ответ, или сервер вернул ошибку
    if "response" in result_json:
        answer = result_json["response"]
    else:
        answer = f"⚠️ Ошибка Ollama: {result_json.get('error', 'Неизвестная ошибка')}"
    
    print("✅ Ответ готов!")
    
    return {
        "answer": answer,
        "sources": retrieved_chunks
    }
```

```python
# ============================================================
# ПОЛНЫЙ КОД RAG-СИСТЕМЫ ДЛЯ GOOGLE COLAB
# ============================================================
#
# Этот код включает все шаги: установка Ollama, загрузка модели,
# создание базы знаний, RAG-функция и тестирование.
#
# Скопируйте ВСЁ в одну ячейку Colab и запустите.
# ============================================================

# 1. УСТАНОВКА OLLAMA
!rm -rf /usr/local/lib/ollama /usr/local/bin/ollama
!apt-get update -qq && apt-get install -y zstd

print("⏳ Установка Ollama...")
!curl -fsSL https://ollama.com/install.sh | sh

# 2. ЗАПУСК СЕРВЕРА OLLAMA
import subprocess
import time
import urllib.request
import urllib.error

print("⏳ Запуск сервера Ollama...")
subprocess.Popen(["ollama", "serve"], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

server_ready = False
for i in range(40):
    try:
        req = urllib.request.urlopen("http://localhost:11434/api/tags", timeout=2)
        if req.getcode() == 200:
            server_ready = True
            break
    except urllib.error.URLError:
        time.sleep(1)

if not server_ready:
    raise Exception("❌ Сервер не запустился! Перезапустите среду выполнения (Runtime -> Restart).")

print("✅ Сервер Ollama запущен и готов!")

# 3. СКАЧИВАНИЕ МОДЕЛИ
MODEL_NAME = "llama3.2:1b"
print(f"⏳ Скачивание модели {MODEL_NAME}...")
subprocess.run(["ollama", "pull", MODEL_NAME])
print(f"🎉 Модель {MODEL_NAME} загружена!")

# 4. УСТАНОВКА БИБЛИОТЕК
!pip install chromadb langchain langchain-community sentence-transformers langchain-text-splitters -q

import chromadb
from sentence_transformers import SentenceTransformer
from langchain_text_splitters import RecursiveCharacterTextSplitter
import requests

print("✅ Все библиотеки установлены!")

# 5. ЗАГРУЗКА ЭМБЕДДИНГ-МОДЕЛИ И СОЗДАНИЕ БАЗЫ ЗНАНИЙ
print("⏳ Загрузка эмбеддинг-модели...")
embedding_model = SentenceTransformer('intfloat/multilingual-e5-large')
print(f"✅ Модель загружена (размер вектора: {embedding_model.get_sentence_embedding_dimension()})")

# Психологические тексты
documents = [
    "Согласно DSM-5, для диагностики депрессии необходимо наличие 5 из 9 симптомов в течение 2 недель. Симптомы включают: подавленное настроение, снижение интереса, изменения аппетита, нарушения сна, психомоторные изменения, усталость, чувство вины, снижение концентрации, суицидальные мысли.",
    "Теория привязанности Джона Боулби описывает, как ранние отношения с caregiver формируют внутренние рабочие модели, влияющие на последующие отношения в жизни.",
    "Когнитивно-поведенческая терапия (КПТ) фокусируется на связи между мыслями, чувствами и поведением.",
    "ПТСР характеризуется повторяющимися переживаниями травмы, избеганием напоминаний, негативными изменениями в мышлении и повышенной возбудимостью.",
    "МКБ-11 включает диагностическую категорию 'сложное ПТСР' (CPTSD), добавляющую симптомы нарушения самооценки, эмоциональной регуляции и межличностных отношений."
]

print("📚 Разбивка текстов на чанки...")
text_splitter = RecursiveCharacterTextSplitter(chunk_size=300, chunk_overlap=50)
chunks = []
for doc in documents:
    chunks.extend(text_splitter.split_text(doc))
print(f"✅ Создано {len(chunks)} чанков")

print("🔢 Создание эмбеддингов...")
chunk_embeddings = embedding_model.encode(chunks, show_progress_bar=True)

print("💾 Сохранение в Chroma...")
client = chromadb.PersistentClient(path="./psychology_db")
collection = client.get_or_create_collection(
    name="psychology_knowledge",
    metadata={"hnsw:space": "cosine"}
)

for i, (chunk, emb) in enumerate(zip(chunks, chunk_embeddings)):
    collection.add(
        ids=[f"chunk_{i}"],
        documents=[chunk],
        embeddings=[emb.tolist()],
        metadatas=[{"source": f"doc_{i}"}]
    )
print(f"✅ Добавлено {len(chunks)} чанков в векторную базу")

# 6. ФУНКЦИЯ RAG-ЗАПРОСА
def rag_query(query, top_k=3):
    print("\n" + "=" * 50)
    print(f"🔍 Вопрос: {query}")
    print("=" * 50)
    
    query_embedding = embedding_model.encode([query])[0]
    results = collection.query(
        query_embeddings=[query_embedding.tolist()],
        n_results=top_k
    )
    retrieved_chunks = results['documents'][0]
    context = "\n\n---\n\n".join(retrieved_chunks)
    
    prompt = f"""Ты – психологический ассистент. Отвечай ТОЛЬКО на основе фрагментов.
Если в фрагментах нет ответа, скажи: "Нет информации по этому вопросу."

Фрагменты:
---
{context}
---
Вопрос: {query}
Ответ:"""

    payload = {
        "model": "llama3.2:1b",
        "prompt": prompt,
        "stream": False,
        "options": {"temperature": 0.3}
    }
    response = requests.post("http://localhost:11434/api/generate", json=payload)
    result_json = response.json()
    answer = result_json.get("response", "⚠️ Ошибка: ответ не получен")
    return {"answer": answer, "sources": retrieved_chunks}

# 7. ТЕСТИРОВАНИЕ
print("\n" + "=" * 60)
print("🧠 ТЕСТИРОВАНИЕ RAG-СИСТЕМЫ")
print("=" * 60)

test_queries = [
    "Каковы симптомы депрессии по DSM-5?",
    "Что такое гештальт-терапия?",
    "Что такое когнитивно-поведенческая терапия?"
]

for q in test_queries:
    result = rag_query(q)
    print(f"\n📊 ОТВЕТ:\n{result['answer']}")
    print("\n📚 ИСТОЧНИКИ:")
    for i, src in enumerate(result['sources']):
        print(f"  {i+1}. {src[:150]}...")
    print("-" * 40)

print("\n🎉 ВСЁ ГОТОВО! RAG-система работает!")
```

### 7.1. RAG не панацея

Важно понимать: RAG **уменьшает**, но **не исключает** галлюцинации.

**Остаточные риски:**

| Риск | Что может пойти не так |
|------|------------------------|
| **Неверный поиск** | Система может найти нерелевантные фрагменты |
| **Искажение контекста** | LLM может неверно интерпретировать найденные фрагменты |
| **Устаревшие данные** | Если база знаний не обновляется, ответы могут быть устаревшими |
| **Чувствительные знания** | Определённые знания в базе могут быть неверно извлечены |

**Правило:** **всегда проверяйте** ответы RAG-системы, особенно в клиническом контексте.

### 7.2. Источники должны быть проверенными

- Используйте только **рецензируемые** источники (учебники, научные статьи, официальные протоколы).
- Не добавляйте в базу знаний непроверенные данные из интернета.
- Регулярно **обновляйте** базу знаний.

### 7.3. Прозрачность для пользователя

Всегда сообщайте пользователю:

- Что ответ сгенерирован ИИ.
- Что ответ основан на конкретных источниках (показывайте ссылки).
- Что это не заменяет профессиональную консультацию.

**Пример дисклеймера для RAG-системы:**

```
⚠️ Этот ответ сгенерирован ИИ-ассистентом на основе предоставленных источников.
Ответы НЕ являются диагнозом и НЕ заменяют консультацию специалиста.
Всегда проверяйте информацию по первоисточникам.
```

### 7.4. Схема: этический чек-лист для RAG

```mermaid
flowchart LR
    CHECK["✅ ЭТИЧЕСКИЙ ЧЕК-ЛИСТ RAG"]
    
    CHECK --> Q1["Источники проверены<br>и рецензированы?"]
    Q1 -->|Да| Q2["База знаний<br>регулярно обновляется?"]
    Q1 -->|Нет| FAIL["❌ НЕ ИСПОЛЬЗОВАТЬ"]
    
    Q2 -->|Да| Q3["Пользователь знает,<br>что это ИИ?"]
    Q2 -->|Нет| FAIL
    
    Q3 -->|Да| Q4["Есть ссылки<br>на источники?"]
    Q3 -->|Нет| FAIL
    
    Q4 -->|Да| PASS["✅ МОЖНО ИСПОЛЬЗОВАТЬ"]
    Q4 -->|Нет| FAIL
    
    style CHECK fill:#fff9c4,stroke:#f57f17
    style PASS fill:#c8e6c9,stroke:#2e7d32
    style FAIL fill:#ffcdd2,stroke:#c62828
```

## 8. ЗАКЛЮЧЕНИЕ

**Что мы сегодня узнали:**

| Что узнали | Ключевая идея |
|------------|---------------|
| **Проблема галлюцинаций** | LLM не знают, чего не знают |
| **RAG** | Поиск → подстановка → генерация |
| **Чанкинг** | Разбивка текста на фрагменты |
| **Эмбеддинги** | Превращение смысла в числа |
| **Chroma** | Векторное хранилище для поиска |
| **Colab + Ollama** | Запуск RAG с локальной LLM |
| **Психологические приложения** | PsyPARSE, AutoPsyC, Sanity-LLM |

**Ключевая мысль:**

> RAG – это не просто «ещё одна технология». Это **способ сделать LLM честными**. Вместо того чтобы гадать, модель **ищет** ответ в ваших документах и **показывает** источники. В психологии, где цена ошибки высока, это превращает LLM из «рискованного эксперимента» в **надёжного ассистента**.
