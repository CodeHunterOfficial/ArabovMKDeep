# Примеры кода для работы с файлами

## Текстовые файлы (.txt)

### Пример 1. Чтение файла построчно

Дан файл `names.txt` с ФИО испытуемых. Требуется вывести все ФИО на экран.

```python
with open("names.txt", "r", encoding="utf-8") as f:
    for line in f:
        print(line.strip())
```

---

### Пример 2. Запись списка в файл

Дано число `N` и `N` баллов. Требуется записать их в файл `scores.txt`.

```python
N = int(input("Сколько баллов? "))
with open("scores.txt", "w", encoding="utf-8") as f:
    for i in range(N):
        score = int(input("Балл: "))
        f.write(str(score) + "\n")
print("Данные записаны.")
```

---

## CSV-файлы (.csv)

### Пример 1. Чтение CSV с заголовком

Дан файл `scores.csv` с колонками `name, age, stai`. Требуется вывести имя и балл каждого испытуемого.

```python
import csv

with open("scores.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(f"{row['name']}: STAI = {row['stai']}")
```

---

### Пример 2. Запись CSV с заголовком

Требуется создать файл `report.csv` с колонками `name, stai, level` и записать три записи.

```python
import csv

data = [
    ["Иванов И.И.", 45, "умеренный"],
    ["Петров П.П.", 72, "очень высокий"],
    ["Сидорова А.А.", 38, "умеренный"]
]

with open("report.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["name", "stai", "level"])
    writer.writerows(data)

print("Файл report.csv создан.")
```

---

## JSONL-файлы (.jsonl)

### Пример 1. Чтение JSONL

Дан файл `responses.jsonl`, в каждой строке — JSON-объект с полями `name`, `stai`.  
Требуется вывести имя и балл каждого испытуемого.

```python
import json

with open("responses.jsonl", "r", encoding="utf-8") as f:
    for line in f:
        record = json.loads(line.strip())
        print(f"{record['name']}: STAI = {record['stai']}")
```

---

### Пример 2. Запись JSONL

Требуется создать файл `responses.jsonl` и записать в него две записи.

```python
import json

records = [
    {"id": 1, "name": "Иванов И.И.", "age": 25, "stai": 45},
    {"id": 2, "name": "Петров П.П.", "age": 32, "stai": 72}
]

with open("responses.jsonl", "w", encoding="utf-8") as f:
    for r in records:
        f.write(json.dumps(r, ensure_ascii=False) + "\n")

print("Файл responses.jsonl создан.")
```
