## Критерии оценивания задачи «Объясни выбор»

Студент должен указать хотя бы одно из следующих положений:

- **f-строка** — краткая запись, высокая читаемость, современный синтаксис Python;
- **`.format()`** — удобство при повторном использовании шаблона, работа со словарями и позиционными аргументами;
- **`%`** — устаревший стиль, встречается в существующем коде, полезен для чтения чужих программ.





## Примеры кода

### Fmt1. Карточка испытуемого — три способа
```python
fio = "Иванов И.И."
age = 25
gender = "мужской"

# f-строка
print("=== Карточка испытуемого ===")
print(f"ФИО: {fio}")
print(f"Возраст: {age} лет")
print(f"Пол: {gender}")

# .format()
print("=== Карточка испытуемого ===")
print("ФИО: {}".format(fio))
print("Возраст: {} лет".format(age))
print("Пол: {}".format(gender))

# %
print("=== Карточка испытуемого ===")
print("ФИО: %s" % fio)
print("Возраст: %d лет" % age)
print("Пол: %s" % gender)
```

### Fmt4. Округление балла — три способа
```python
score = 45.6789

# f-строка
print(f"Балл: {score:.2f}")

# .format()
print("Балл: {:.2f}".format(score))

# %
print("Балл: %.2f" % score)
```

### Fmt5. Процент испытуемых — три способа
```python
high = 10
total = 28
percent = high / total * 100

# f-строка
print(f"Высокая тревожность: {percent:.1f}%")

# .format()
print("Высокая тревожность: {:.1f}%".format(percent))

# %
print("Высокая тревожность: %.1f%%" % percent)
```

### Fmt7. Таблица с выравниванием — три способа
```python
data = [("Иван", 45), ("Пётр", 60), ("Александра", 38)]

# f-строка
print(f"{'Имя':<12}{'Балл':>6}")
for name, score in data:
    print(f"{name:<12}{score:>6}")

# .format()
print("{:<12}{:>6}".format("Имя", "Балл"))
for name, score in data:
    print("{:<12}{:>6}".format(name, score))

# %
print("%-12s%6s" % ("Имя", "Балл"))
for name, score in data:
    print("%-12s%6d" % (name, score))
```
