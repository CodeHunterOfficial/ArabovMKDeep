## Примеры кода

### Array1. Шкала тревожности Спилбергера
```python
N = int(input("Сколько испытуемых? "))
anxiety = []
for i in range(N):
    x = int(input("Балл STAI (20–80): "))
    anxiety.append(x)

print("Баллы:", anxiety)
high = 0
for x in anxiety:
    if x > 60:
        high += 1
print("Высокая тревожность у", high, "испытуемых")
```

### Array2. Возрастные периоды по Выготскому
```python
N = int(input("Сколько испытуемых? "))
ages = []
for i in range(N):
    a = int(input("Возраст: "))
    ages.append(a)

for a in ages:
    if 0 <= a <= 1:
        print(a, "— младенчество")
    elif 1 < a <= 3:
        print(a, "— раннее детство")
    elif 3 < a <= 7:
        print(a, "— дошкольный")
    elif 7 < a <= 11:
        print(a, "— младший школьный")
    elif 11 < a <= 15:
        print(a, "— подростковый")
    elif 15 < a <= 18:
        print(a, "— юношеский")
    else:
        print(a, "— взрослость")
```

### Array17. Тревога и депрессия (HADS)
```python
N = int(input("Сколько испытуемых? "))
clinical = 0

for i in range(N):
    A = int(input("Тревога (HADS): "))
    B = int(input("Депрессия (HADS): "))
    if A > 11 and B > 11:
        print("клинический случай")
        clinical += 1
    else:
        print("норма")

print("Клинических случаев:", clinical)
```

### Array21. Возрастные группы
```python
N = int(input("Сколько испытуемых? "))
ages = []
for i in range(N):
    a = int(input("Возраст: "))
    ages.append(a)

children = 0
teens = 0
youth = 0
adults = 0

for a in ages:
    if 0 <= a <= 11:
        children += 1
    elif 12 <= a <= 15:
        teens += 1
    elif 16 <= a <= 18:
        youth += 1
    else:
        adults += 1

print("Дети:", children)
print("Подростки:", teens)
print("Юноши:", youth)
print("Взрослые:", adults)
```

### Array31. Частота баллов (шкала настроения)
```python
N = int(input("Сколько испытуемых? "))
scores = []
for i in range(N):
    x = int(input("Балл от 0 до 10: "))
    scores.append(x)

freq = [0] * 11
for x in scores:
    if 0 <= x <= 10:
        freq[x] += 1

for k in range(11):
    print(k, ":", freq[k])
```

### Array37. Текстовая гистограмма
```python
N = int(input("Сколько испытуемых? "))
scores = []
for i in range(N):
    x = int(input("Балл от 0 до 10: "))
    scores.append(x)

freq = [0] * 11
for x in scores:
    if 0 <= x <= 10:
        freq[x] += 1

for k in range(11):
    print(k, ":", "*" * freq[k])
```

### Array40. Гендерные различия по тревожности
```python
N = int(input("Сколько испытуемых? "))
genders = []
anxieties = []
for i in range(N):
    g = int(input("Пол (1 — муж, 0 — жен): "))
    a = float(input("Балл тревожности: "))
    genders.append(g)
    anxieties.append(a)

sum_m = 0
count_m = 0
sum_f = 0
count_f = 0

for i in range(N):
    if genders[i] == 1:
        sum_m += anxieties[i]
        count_m += 1
    else:
        sum_f += anxieties[i]
        count_f += 1

if count_m > 0:
    avg_m = sum_m / count_m
    print("Средний балл мужчин:", avg_m)
if count_f > 0:
    avg_f = sum_f / count_f
    print("Средний балл женщин:", avg_f)

if count_m > 0 and count_f > 0:
    if avg_m > avg_f:
        print("У мужчин тревожность выше")
    elif avg_f > avg_m:
        print("У женщин тревожность выше")
    else:
        print("Различий нет")
```
