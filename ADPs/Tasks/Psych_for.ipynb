### For7. Сумма N оценок
```python
N = int(input("Сколько оценок? "))
summa = 0
for i in range(N):
    x = int(input("Оценка: "))
    summa = summa + x
print("Сумма:", summa)
```

### For8. Средний возраст
```python
N = int(input("Сколько испытуемых? "))
summa = 0
for i in range(N):
    age = int(input("Возраст: "))
    summa = summa + age
sredniy = summa / N
print("Средний возраст:", sredniy)
```

### For9. Максимальный балл мотивации
```python
N = int(input("Сколько испытуемых? "))
maximum = -1
for i in range(N):
    x = int(input("Балл мотивации: "))
    if x > maximum:
        maximum = x
print("Максимум:", maximum)
```

### For11. Время реакции больше 500

### For13. Валидность анкет
```python
N = int(input("Сколько анкет? "))
for i in range(N):
    propusk = int(input("Пропусков: "))
    if propusk <= 5:
        print("анкета валидна")
    else:
        print("анкета невалидна")
```

### For15. Средний балл тревожности и количество выше среднего
```python
N = int(input("Сколько испытуемых? "))
scores = []
summa = 0
for i in range(N):
    x = int(input("Балл тревожности: "))
    scores.append(x)
    summa = summa + x

sredniy = summa / N
count = 0
for x in scores:
    if x > sredniy:
        count = count + 1

print("Средний балл:", sredniy)
print("Выше среднего:", count)
```

### For17. Риск по тревоге и депрессии
```python
N = int(input("Сколько испытуемых? "))
high = 0
for i in range(N):
    A = int(input("Тревога: "))
    B = int(input("Депрессия: "))
    if A > 60 and B > 60:
        print("высокий риск")
        high = high + 1
    else:
        print("низкий риск")
print("Высокий риск у", high, "человек")
```

### For20. Средняя оценка и количество ниже среднего
```python
N = int(input("Сколько оценок? "))
scores = []
summa = 0
for i in range(N):
    x = int(input("Оценка: "))
    scores.append(x)
    summa = summa + x

sredniy = summa / N
below = 0
for x in scores:
    if x < sredniy:
        below = below + 1

print("Средний балл:", sredniy)
print("Ниже среднего:", below)
```

### For24. Процент правильных ответов
```python
N = int(input("Сколько ответов? "))
correct = 0
for i in range(N):
    x = int(input("1 — верно, 0 — неверно: "))
    if x == 1:
        correct = correct + 1
percent = correct / N * 100
print("Процент правильных:", percent)
```

### For31. Частота баллов от 0 до 10
```python
N = int(input("Сколько испытуемых? "))
freq = [0] * 11
for i in range(N):
    x = int(input("Балл от 0 до 10: "))
    freq[x] = freq[x] + 1

for k in range(11):
    print(k, ":", freq[k])
```

### For40. Средний балл тревожности по полу
```python
N = int(input("Сколько испытуемых? "))
sum_m = 0
count_m = 0
sum_f = 0
count_f = 0

for i in range(N):
    gender = int(input("Пол (1 — муж, 0 — жен): "))
    score = float(input("Балл тревожности: "))
    if gender == 1:
        sum_m = sum_m + score
        count_m = count_m + 1
    else:
        sum_f = sum_f + score
        count_f = count_f + 1

if count_m > 0:
    print("Средний балл мужчин:", sum_m / count_m)
else:
    print("Мужчин нет")

if count_f > 0:
    print("Средний балл женщин:", sum_f / count_f)
else:
    print("Женщин нет")
```
