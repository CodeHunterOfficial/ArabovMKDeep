## Примеры кода для нескольких задач

### While1. Вывести все введённые баллы
```python
x = float(input("Балл (или -1 для выхода): "))
while x != -1:
    print(x)
    x = float(input("Балл (или -1 для выхода): "))
```

### While4. Средний балл тревожности
```python
x = float(input("Балл тревожности (или -1): "))
summa = 0
count = 0
while x != -1:
    summa = summa + x
    count = count + 1
    x = float(input("Балл тревожности (или -1): "))

if count > 0:
    print("Средний балл:", summa / count)
else:
    print("Нет данных")
```

### While9. Валидность анкет
```python
propusk = int(input("Пропусков (или -1): "))
while propusk != -1:
    if propusk <= 5:
        print("анкета валидна")
    else:
        print("анкета невалидна")
    propusk = int(input("Пропусков (или -1): "))
```

### While15. Средний балл и количество выше среднего
```python
scores = []
x = float(input("Балл тревожности (или -1): "))
while x != -1:
    scores.append(x)
    x = float(input("Балл тревожности (или -1): "))

if len(scores) > 0:
    sredniy = sum(scores) / len(scores)
    count = 0
    for s in scores:   # здесь можно использовать for для подсчёта
        if s > sredniy:
            count = count + 1
    print("Средний балл:", sredniy)
    print("Выше среднего:", count)
else:
    print("Нет данных")
```

### While17. Риск по тревоге и депрессии
```python
A = int(input("Тревога (или -1): "))
high = 0
while A != -1:
    B = int(input("Депрессия: "))
    if A > 60 and B > 60:
        print("высокий риск")
        high = high + 1
    else:
        print("низкий риск")
    A = int(input("Тревога (или -1): "))
print("Высокий риск у", high, "человек")
```

### While31. Частота баллов от 0 до 10
```python
freq = [0] * 11
x = int(input("Балл от 0 до 10 (или -1): "))
while x != -1:
    if 0 <= x <= 10:
        freq[x] = freq[x] + 1
    x = int(input("Балл от 0 до 10 (или -1): "))

for k in range(11):
    print(k, ":", freq[k])
```

### While40. Средний балл тревожности по полу
```python
gender = int(input("Пол (1 — муж, 0 — жен, -1 — выход): "))
sum_m = 0
count_m = 0
sum_f = 0
count_f = 0

while gender != -1:
    score = float(input("Балл тревожности: "))
    if gender == 1:
        sum_m = sum_m + score
        count_m = count_m + 1
    else:
        sum_f = sum_f + score
        count_f = count_f + 1
    gender = int(input("Пол (1 — муж, 0 — жен, -1 — выход): "))

if count_m > 0:
    print("Средний балл мужчин:", sum_m / count_m)
else:
    print("Мужчин нет")

if count_f > 0:
    print("Средний балл женщин:", sum_f / count_f)
else:
    print("Женщин нет")
```
