# Блок 1. Целые числа (PsychoInteger1–PsychoInteger20)

**PsychoInteger1.** Дано время реакции $t$ в миллисекундах. Используя деление нацело, найти количество полных секунд.

**PsychoInteger2.** Дано время выполнения теста $t$ в секундах. Используя деление нацело, найти количество полных минут.

**PsychoInteger3.** Дано общее число ответов $N$ в тесте. Используя деление нацело, найти количество полных страниц, если на одной странице $K$ ответов.

$$p = \frac{N}{K}$$

**PsychoInteger4.** Даны общее число испытуемых $N$ и размер группы $G$. Найти количество полных групп.

$$k = \frac{N}{G}$$

**PsychoInteger5.** Даны общее число испытуемых $N$ и размер группы $G$. Найти число испытуемых в неполной группе.

$$r = N - \frac{N}{G} \cdot G$$

**PsychoInteger6.** Дано двузначное число — номер пункта теста. Вывести отдельно цифру десятков и цифру единиц.

**PsychoInteger7.** Дано двузначное число. Найти сумму и произведение его цифр.

**PsychoInteger10.** Дано трёхзначное число. Вывести последнюю цифру (единицы), затем среднюю (десятки).

**PsychoInteger11.** Дано трёхзначное число. Найти сумму и произведение его цифр.

**PsychoInteger12.** Дано трёхзначное число. Вывести число, прочитанное справа налево.

$$Y = e \cdot 100 + d \cdot 10 + h$$

**PsychoInteger13.** Дано трёхзначное число. Первую цифру зачеркнули и приписали справа. Вывести результат.

**PsychoInteger14.** Дано трёхзначное число. Последнюю цифру зачеркнули и приписали слева. Вывести результат.

**PsychoInteger15.** Дано трёхзначное число. Поменять местами сотни и десятки.

**PsychoInteger16.** Дано трёхзначное число. Поменять местами десятки и единицы.

**PsychoInteger17.** Дано целое число, большее 999. Найти цифру разряда сотен.

**PsychoInteger18.** Дано целое число, большее 999. Найти цифру разряда тысяч.

# Блок 2. Посложнее (PsychoInteger21–PsychoInteger40)

**PsychoInteger21.** С начала сессии прошло $N$ секунд. Найти количество секунд с начала последней минуты.

$$s = N - \frac{N}{60} \cdot 60$$

**PsychoInteger22.** С начала сессии прошло $N$ секунд. Найти количество секунд с начала последнего часа.

$$s = N - \frac{N}{3600} \cdot 3600$$

**PsychoInteger23.** С начала сессии прошло $N$ секунд. Найти количество полных минут с начала последнего часа.

$$m = \frac{N - \frac{N}{3600} \cdot 3600}{60}$$

**PsychoInteger24.** Дни недели: 0 — воскресенье, 1 — понедельник, …, 6 — суббота. Дан номер дня года $K$ (1–365). Первое января было понедельником. Найти номер дня недели.

$$D = (K + 0) - \frac{K + 0}{7} \cdot 7$$

**PsychoInteger25.** Дни недели: 0 — воскресенье, …, 6 — суббота. Дан номер дня года $K$. Первое января было четвергом. Найти номер дня недели.

$$D = (K + 3) - \frac{K + 3}{7} \cdot 7$$

**PsychoInteger26.** Дни недели: 1 — понедельник, …, 7 — воскресенье. Дан номер дня года $K$. Первое января было вторником. Найти номер дня недели.

$$D = (K + 0) - \frac{K + 0}{7} \cdot 7 + 1$$

**PsychoInteger27.** Дни недели: 1 — понедельник, …, 7 — воскресенье. Дан номер дня года $K$. Первое января было субботой. Найти номер дня недели.

$$D = (K + 4) - \frac{K + 4}{7} \cdot 7 + 1$$

**PsychoInteger28.** Дни недели: 1 — понедельник, …, 7 — воскресенье. Даны номер дня года $K$ и номер дня недели 1 января $N$. Найти номер дня недели для дня $K$.

$$D = (K + N - 2) - \frac{K + N - 2}{7} \cdot 7 + 1$$

**PsychoInteger29.** Даны размеры листа $A \times B$ и сторона клетки $C$. Найти количество полных клеток и площадь незанятой части.

$$k = \frac{A}{C} \cdot \frac{B}{C}, \quad S = A \cdot B - k \cdot C \cdot C$$

**PsychoInteger30.** Дан год рождения испытуемого (целое положительное число). Определить номер десятилетия, учитывая, что началом десятилетия был год, кратный 10 плюс 1.

$$D = \frac{Y - 1}{10} + 1$$

**PsychoInteger31.** Дано число правильных ответов $R$ и число ошибок $W$. Найти общее число заданий и остаток от деления общего числа на 10.

$$N = R + W, \quad r = N - \frac{N}{10} \cdot 10$$

**PsychoInteger32.** Дано общее время теста $T$ (с) и время одного задания $t$ (с). Найти число полных заданий и остаток времени.

$$N = \frac{T}{t}, \quad r = T - N \cdot t$$
