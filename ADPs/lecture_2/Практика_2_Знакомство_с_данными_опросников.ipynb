## 3. Если нет интернета или датасет не скачивается — создаём синтетические данные

Если по каким-то причинам вы не можете скачать реальный датасет, вы можете создать правдоподобные синтетические данные с помощью numpy. Это не хуже для учебных целей — вы сможете отработать все те же команды.

**Пример генерации данных для 100 «пациентов»:**




````python
import numpy as np
import pandas as pd

np.random.seed(42)  # для воспроизводимости

# Генерируем 100 записей
n = 100

data = {
    'ID': range(1, n+1),
    'Пол': np.random.choice(['М', 'Ж'], n, p=[0.4, 0.6]),
    'Возраст': np.random.randint(18, 65, n),
    'BDI': np.random.randint(0, 50, n),  # баллы по шкале Бека
    'STAI_ситуативная': np.random.randint(20, 70, n),
    'STAI_личностная': np.random.randint(20, 70, n),
    'Группа_терапии': np.random.choice(['КПТ', 'ДБТ', 'Гештальт'], n)
}

df = pd.DataFrame(data)
df.to_csv('synthetic_psych_data.csv', index=False)
print("Данные сохранены в synthetic_psych_data.csv")
print(df.head())
````

## 6. Объединение таблиц (JOIN, merge)

В реальной психологической практике данные часто хранятся в разных таблицах. Например, одна таблица содержит результаты опросников, другая — демографические данные пациентов, третья — заметки терапевта. Чтобы провести анализ, их нужно объединить.

**Основные способы объединения в pandas:**

| Метод | Описание | Когда использовать |
|-------|----------|-------------------|
| `pd.merge(df1, df2, on='ключ')` | Объединение по общему столбцу | Основной способ, как SQL JOIN |
| `pd.merge(df1, df2, on='ключ', how='left')` | LEFT JOIN — все записи из левой таблицы | Когда нужно сохранить все данные из основной таблицы |
| `pd.merge(df1, df2, on='ключ', how='inner')` | INNER JOIN — только совпадающие записи | Когда важны только те, у кого есть данные в обеих таблицах |
| `pd.merge(df1, df2, on='ключ', how='outer')` | FULL OUTER JOIN — все записи из обеих таблиц | Когда нужно сохранить все данные, даже с пропусками |
| `pd.concat([df1, df2])` | Конкатенация — присоединение строк | Когда у таблиц одинаковые столбцы |

### 6.1. Пример: объединение данных опросников и демографии

Представим, что у нас есть две таблицы:

**Таблица 1: результаты опросников (опросники.csv)**

| ID | BDI | STAI_ситуативная | Группа_терапии |
|----|-----|------------------|----------------|
| 1  | 14  | 38               | КПТ            |
| 2  | 22  | 51               | ДБТ            |
| 3  | 18  | 41               | КПТ            |

**Таблица 2: демографические данные (демография.csv)**

| ID | Пол | Возраст | Образование |
|----|-----|---------|-------------|
| 1  | М   | 34      | Высшее      |
| 2  | Ж   | 28      | Среднее     |
| 3  | М   | 45      | Высшее      |

**Код для объединения:**






**Код для объединения:**

```python
import pandas as pd

# Загружаем обе таблицы
df_опросники = pd.read_csv('опросники.csv')
df_демография = pd.read_csv('демография.csv')

# Объединяем по столбцу ID (как SQL INNER JOIN)
df_combined = pd.merge(df_опросники, df_демография, on='ID', how='inner')

print("Объединённая таблица:")
print(df_combined.head())
```

**Результат:**

| ID | BDI | STAI_ситуативная | Группа_терапии | Пол | Возраст | Образование |
|----|-----|------------------|----------------|-----|---------|-------------|
| 1  | 14  | 38               | КПТ            | М   | 34      | Высшее      |
| 2  | 22  | 51               | ДБТ            | Ж   | 28      | Среднее     |
| 3  | 18  | 41               | КПТ            | М   | 45      | Высшее      |

### 6.2. LEFT JOIN: сохраняем все данные из основной таблицы

Если в таблице демографии есть не все пациенты, мы всё равно хотим сохранить всех пациентов из таблицы опросников:

```python
df_left = pd.merge(df_опросники, df_демография, on='ID', how='left')
print(df_left)
```

Если у какого-то пациента нет данных в таблице демографии, в соответствующих столбцах будет `NaN` (пропуск).

### 6.3. Пример из клинической практики: объединение данных до и после терапии

Часто у нас есть данные «до» и «после» терапии. Их нужно объединить, чтобы сравнить изменения.

**Таблица: до терапии (pre_therapy.csv)**

| ID | BDI_pre |
|----|---------|
| 1  | 35      |
| 2  | 28      |
| 3  | 42      |

**Таблица: после терапии (post_therapy.csv)**

| ID | BDI_post |
|----|----------|
| 1  | 14       |
| 2  | 22       |
| 3  | 18       |

**Код:**

```python
df_pre = pd.read_csv('pre_therapy.csv')
df_post = pd.read_csv('post_therapy.csv')

# Объединяем по ID
df_therapy = pd.merge(df_pre, df_post, on='ID', how='inner')

# Добавляем столбец с изменением
df_therapy['Change'] = df_therapy['BDI_pre'] - df_therapy['BDI_post']

print(df_therapy)
```

**Результат:**

| ID | BDI_pre | BDI_post | Change |
|----|---------|----------|--------|
| 1  | 35      | 14       | 21     |
| 2  | 28      | 22       | 6      |
| 3  | 42      | 18       | 24     |

**Интерпретация:** все пациенты показали улучшение, но у кого-то снижение больше (ID=3), у кого-то меньше (ID=2).

## 8. Простая гистограмма

Гистограмма показывает распределение одной количественной переменной. Построим гистограмму для баллов BDI.

```python
import matplotlib.pyplot as plt

```

## 9. Полный пример кода для практики

Ниже приведён полный скрипт, который можно скопировать и запустить. Он включает загрузку данных, объединение таблиц, группировки и гистограмму.




````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Генерация синтетических данных (или загрузка реальных)
np.random.seed(42)
n = 100

df_опросники = pd.DataFrame({
    'ID': range(1, n+1),
    'BDI': np.random.randint(0, 50, n),
    'STAI_ситуативная': np.random.randint(20, 70, n),
    'STAI_личностная': np.random.randint(20, 70, n),
    'Группа_терапии': np.random.choice(['КПТ', 'ДБТ', 'Гештальт'], n)
})

df_демография = pd.DataFrame({
    'ID': range(1, n+1),
    'Пол': np.random.choice(['М', 'Ж'], n, p=[0.4, 0.6]),
    'Возраст': np.random.randint(18, 65, n),
    'Образование': np.random.choice(['Среднее', 'Высшее', 'Ученая степень'], n)
})

# 2. Объединение таблиц
df = pd.merge(df_опросники, df_демография, on='ID', how='inner')
print("=== Объединённая таблица (первые 5 строк) ===")
print(df.head())

# 3. Первое знакомство с данными
print("\n=== Информация о данных ===")
print(df.info())

print("\n=== Описательные статистики ===")
print(df.describe())

# 4. Группировки
print("\n=== Средний BDI по полу ===")
print(df.groupby('Пол')['BDI'].mean())

print("\n=== Средний BDI по полу с статистиками ===")
print(df.groupby('Пол')['BDI'].agg(['mean', 'std', 'count']))

# Создаём возрастные группы
bins = [18, 35, 55, 65]
labels = ['Молодые', 'Средние', 'Пожилые']
df['Возрастная_группа'] = pd.cut(df['Возраст'], bins=bins, labels=labels, right=False)

print("\n=== Средний BDI по возрастным группам ===")
print(df.groupby('Возрастная_группа')['BDI'].mean())

print("\n=== Средний BDI по полу и возрасту ===")
print(df.groupby(['Пол', 'Возрастная_группа'])['BDI'].mean())

# 5. Гистограмма
plt.figure(figsize=(8,5))
plt.hist(df['BDI'], bins=10, edgecolor='black', color='skyblue', alpha=0.7)
plt.title('Распределение баллов BDI в выборке')
plt.xlabel('Балл BDI')
plt.ylabel('Количество пациентов')
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.show()
````

## 11. Частые ошибки и как их избежать

| Ошибка | Решение |
|--------|---------|
| `FileNotFoundError` при загрузке CSV | Проверьте путь к файлу. Используйте `os.listdir()` чтобы увидеть файлы в папке. |
| В `describe()` нет текстовых столбцов | `describe()` работает только с числами. Для текстовых используйте `df['столбец'].value_counts()`. |
| Ошибка при группировке с `pd.cut()` | Проверьте, что границы `bins` покрывают все значения возраста. |
| Гистограмма не показывается | Добавьте `plt.show()` в конце. В Jupyter/Colab графики отображаются автоматически. |
| Ошибка в `merge()` — дублирование столбцов | Используйте параметры `suffixes` для переименования: `pd.merge(df1, df2, on='ID', suffixes=('_1', '_2'))` |
