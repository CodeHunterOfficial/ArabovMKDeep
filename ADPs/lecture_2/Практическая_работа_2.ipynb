**Описание.** В этом задании вы проверите гипотезу о связи между депрессией (BDI) и ситуативной тревогой (STAI).

**Требуется:**

1. Постройте **диаграмму рассеяния** (scatter plot), где:
   - По оси X — балл BDI.
   - По оси Y — балл STAI (ситуативная тревога).
   - Каждая точка — один пациент.

2. Добавьте линию тренда (линейную регрессию).

3. Рассчитайте коэффициент корреляции Пирсона между BDI и STAI (вручную или с помощью инструмента).

4. Напишите интерпретацию:
   - Есть ли видимая связь между переменными?
   - Какова сила и направление корреляции?
   - Что это может означать для клинической практики?

**Что сдать:** график с линией тренда, значение коэффициента корреляции, текстовую интерпретацию.




```python
# Если используете Python:
import numpy as np
from scipy import stats

stai = [38, 51, 41, 48, 55, 35, 42, 50, 39, 46,
        58, 44, 62, 32, 43, 49, 40, 56, 37, 47,
        45, 34, 52, 44, 59, 31, 53, 63, 33, 57]

# Scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(bdi, stai, alpha=0.7, color='darkblue')

# Линия тренда
z = np.polyfit(bdi, stai, 1)
p = np.poly1d(z)
plt.plot(bdi, p(bdi), color='red', linestyle='dashed')

# Коэффициент корреляции
corr, p_value = stats.pearsonr(bdi, stai)
print(f"Корреляция Пирсона: r = {corr:.3f}, p = {p_value:.4f}")

plt.xlabel('Балл BDI')
plt.ylabel('Балл STAI (ситуативная)')
plt.title('Связь между депрессией и ситуативной тревогой')
plt.show()
```
