**BioBegin1. Молярная концентрация**  
Даны масса вещества $m$ (г), молярная масса $M$ (г/моль) и объём раствора $V$ (л).  
Вычислить молярную концентрацию:  
$C = \frac{m}{M \cdot V}$.

**BioBegin2. Разведение раствора**  
Даны исходная концентрация $C_1$ (моль/л), исходный объём $V_1$ (мл) и конечный объём $V_2$ (мл).  
Найти конечную концентрацию:  
$C_2 = \frac{C_1 \cdot V_1}{V_2}$.

**BioBegin3. Фактор разведения**  
Даны объём аликвоты $V_{al}$ (мкл) и общий объём пробы $V_{tot}$ (мкл).  
Вычислить фактор разведения:  
$DF = \frac{V_{tot}}{V_{al}}$.

**BioBegin4. Закон Бугера–Ламберта–Бера**  
Даны коэффициент экстинкции $\varepsilon$ (л/(моль·см)), концентрация $c$ (моль/л) и длина оптического пути $l$ (см).  
Найти оптическую плотность:  
$A = \varepsilon \cdot c \cdot l$.

**BioBegin5. Концентрация по оптической плотности**  
Даны $A$, $\varepsilon$ и $l$.  
Найти концентрацию:  
$c = \frac{A}{\varepsilon \cdot l}$.

**BioBegin6. Водородный показатель**  
Дана концентрация ионов водорода $[H^+]$ (моль/л).  
Вычислить:  
$pH = -\log_{10}([H^+])$.

**BioBegin7. Уравнение Гендерсона–Хассельбаха**  
Даны $pKa$, концентрация основания $[A^-]$ и концентрация кислоты $[HA]$.  
Вычислить:  
$pH = pKa + \log_{10}\left(\frac{[A^-]}{[HA]}\right)$.

**BioBegin8. Кинетика Михаэлиса–Ментен**  
Даны $V_{max}$ (мкмоль/мин), $K_m$ (мМ) и концентрация субстрата $[S]$ (мМ).  
Вычислить скорость реакции:  
$v = \frac{V_{max} \cdot [S]}{K_m + [S]}$.

**BioBegin9. Удельная активность фермента**  
Даны общая активность $A_{total}$ (Ед) и общее количество белка $m_{prot}$ (мг).  
Вычислить удельную активность:  
$A_{spec} = \frac{A_{total}}{m_{prot}}$.

**BioBegin10. Концентрация двуцепочечной ДНК**  
Даны $A260$ и фактор разведения $DF$.  
Концентрация ДНК (нг/мкл) вычисляется как:  
$C = A260 \cdot DF \cdot 50$.

**BioBegin11. Концентрация РНК**  
Даны $A260$ и фактор разведения $DF$.  
Концентрация РНК (нг/мкл):  
$C = A260 \cdot DF \cdot 40$.

**BioBegin12. Белок по Брэдфорду**  
Даны $A595$, калибровочный коэффициент $k$ и свободный член $b$.  
Вычислить концентрацию белка:  
$C = \frac{A595 - b}{k}$.

**BioBegin13. Энергия Гиббса**  
Даны $\Delta H$ (кДж/моль), температура $T$ (K) и $\Delta S$ (Дж/(моль·K)).  
Вычислить $\Delta G$ в кДж/моль:  
$\Delta G = \Delta H - \frac{T \cdot \Delta S}{1000}$.

**BioBegin14. Константа равновесия**  
Даны $\Delta G^\circ$ (кДж/моль) и $T$ (K).  
Универсальная газовая постоянная $R = 8.314$ Дж/(моль·K).  
Вычислить:  
$K_{eq} = \exp\left(\frac{-\Delta G^\circ \cdot 1000}{R \cdot T}\right)$.

**BioBegin15. Радиоактивный распад**  
Даны начальная активность $A_0$, период полураспада $t_{1/2}$ (ч) и время $t$ (ч).  
Вычислить активность:  
$A = A_0 \cdot \exp\left(\frac{-\ln(2) \cdot t}{t_{1/2}}\right)$.

**BioBegin16. Приготовление буфера**  
Даны требуемая концентрация $C$ (М), объём $V$ (л) и молярная масса $M$ (г/моль).  
Найти массу навески:  
$m = C \cdot V \cdot M$.

**BioBegin17. Процентный раствор**  
Даны масса растворённого вещества $m$ (г) и объём раствора $V$ (мл).  
Вычислить процентную концентрацию:  
$percent = \frac{m}{V} \cdot 100$.

**BioBegin18. Плотность раствора**  
Даны масса $m$ (г) и объём $V$ (мл).  
Вычислить плотность:  
$\rho = \frac{m}{V}$.

**BioBegin19. Число молекул**  
Дано количество вещества $n$ (моль).  
Вычислить число молекул:  
$N = n \cdot 6.022 \times 10^{23}$.

**BioBegin20. Молярная масса по осмотическому давлению**  
Даны масса белка $m$ (г), объём $V$ (л), температура $T$ (K) и осмотическое давление $\pi$ (кПа).  
$R = 8.314$ кПа·л/(моль·K).  
Вычислить молярную массу:  
$M = \frac{m \cdot R \cdot T}{\pi \cdot V}$.

**BioBegin21. Относительная центробежная сила**  
Даны радиус ротора $r$ (см) и скорость вращения $rpm$ (об/мин).  
Вычислить:  
$RCF = 1.118 \times 10^{-5} \cdot r \cdot rpm^2$.

**BioBegin22. Степень диссоциации**  
Даны концентрация диссоциированных молекул $C_d$ и общая концентрация $C$.  
Вычислить степень диссоциации:  
$\alpha = \frac{C_d}{C}$.

**BioBegin23. Скорость реакции**  
Даны изменение концентрации $\Delta C$ (мМ) и время $\Delta t$ (с).  
Вычислить скорость:  
$v = \frac{\Delta C}{\Delta t}$.

**BioBegin24. Каталитическая эффективность**  
Даны $k_{cat}$ (с⁻¹) и $K_m$ (М).  
Вычислить:  
$\frac{k_{cat}}{K_m}$.

**BioBegin25. Молярный коэффициент экстинкции**
