# Блок 1. Простые задачи (BioBegin1–BioBegin20)

**BioBegin1.** Даны масса растворённого вещества $m$ (г), молярная масса вещества $M$ (г/моль) и объём раствора $V$ (л). Найти молярную концентрацию.

$$C = \frac{m}{M \cdot V}$$

**BioBegin2.** Даны исходная концентрация $C_1$ (моль/л), исходный объём $V_1$ (мл) и конечный объём $V_2$ (мл). Найти конечную концентрацию.

$$C_2 = \frac{C_1 \cdot V_1}{V_2}$$

**BioBegin3.** Даны объём аликвоты $V_{al}$ (мкл) и общий объём пробы $V_{tot}$ (мкл). Найти фактор разведения.

$$DF = \frac{V_{tot}}{V_{al}}$$

**BioBegin4.** Даны коэффициент экстинкции $\varepsilon$ (л/(моль·см)), концентрация $c$ (моль/л) и длина оптического пути $l$ (см). Найти оптическую плотность.

$$A = \varepsilon \cdot c \cdot l$$

**BioBegin5.** Дана концентрация ионов водорода $[H^+]$ (моль/л). Найти водородный показатель.

$$pH = -\log_{10}([H^+])$$

**BioBegin6.** Даны $pKa$, концентрация основания $[A^-]$ (моль/л) и концентрация кислоты $[HA]$ (моль/л). Найти $pH$.

$$pH = pKa + \log_{10}\left(\frac{[A^-]}{[HA]}\right)$$

**BioBegin7.** Даны максимальная скорость $V_{max}$ (мкмоль/мин), константа Михаэлиса $K_m$ (мМ) и концентрация субстрата $[S]$ (мМ). Найти скорость реакции.

$$v = \frac{V_{max} \cdot [S]}{K_m + [S]}$$

**BioBegin8.** Даны общая активность $A_{total}$ (Ед) и общее количество белка $m_{prot}$ (мг). Найти удельную активность.

$$A_{spec} = \frac{A_{total}}{m_{prot}}$$

**BioBegin9.** Даны оптическая плотность $A260$ и фактор разведения $DF$. Найти концентрацию двуцепочечной ДНК.

$$C = A260 \cdot DF \cdot 50$$

**BioBegin10.** Даны оптическая плотность $A260$ и фактор разведения $DF$. Найти концентрацию РНК.

$$C = A260 \cdot DF \cdot 40$$

**BioBegin11.** Даны оптическая плотность $A595$, калибровочный коэффициент $k$ и свободный член $b$. Найти концентрацию белка.

$$C = \frac{A595 - b}{k}$$

**BioBegin12.** Даны изменение энтальпии $\Delta H$ (кДж/моль), температура $T$ (K) и изменение энтропии $\Delta S$ (Дж/(моль·K)). Найти энергию Гиббса.

$$\Delta G = \Delta H - \frac{T \cdot \Delta S}{1000}$$

**BioBegin13.** Даны требуемая концентрация $C$ (М), объём $V$ (л) и молярная масса $M$ (г/моль). Найти массу навески.

$$m = C \cdot V \cdot M$$

**BioBegin14.** Даны масса растворённого вещества $m$ (г) и объём раствора $V$ (мл). Найти процентную концентрацию.

$$percent = \frac{m}{V} \cdot 100$$

**BioBegin15.** Даны масса $m$ (г) и объём $V$ (мл). Найти плотность.

$$\rho = \frac{m}{V}$$

**BioBegin16.** Дано количество вещества $n$ (моль). Найти число молекул.

$$N = n \cdot 6.022 \times 10^{23}$$

**BioBegin17.** Даны объём раствора $V$ (л) и молярная концентрация $C$ (моль/л). Найти количество вещества.

$$n = C \cdot V$$

**BioBegin18.** Даны количество вещества $n$ (моль) и молярная масса $M$ (г/моль). Найти массу вещества.

$$m = n \cdot M$$

**BioBegin19.** Даны масса вещества $m$ (г) и его молярная масса $M$ (г/моль). Найти количество вещества.

$$n = \frac{m}{M}$$

**BioBegin20.** Даны масса раствора $m_{sol}$ (г) и масса растворённого вещества $m$ (г). Найти массовую долю.

$$w = \frac{m}{m_{sol}}$$

# Блок 2. По сложнее (BioBegin21–BioBegin40)

**BioBegin21.** Даны масса белка $m$ (мг) и объём раствора $V$ (мкл). Найти концентрацию белка в мг/мл.

$$C = \frac{m}{V / 1000}$$

**BioBegin22.** Даны оптическая плотность $A$, коэффициент экстинкции $\varepsilon$ (л/(моль·см)) и длина оптического пути $l$ (см). Найти концентрацию в мкмоль/л.

$$c = \frac{A}{\varepsilon \cdot l} \cdot 10^6$$

**BioBegin23.** Даны коэффициент пропускания $T$ (в процентах) и длина оптического пути $l$ (см). Найти оптическую плотность.

$$A = -\log_{10}\left(\frac{T}{100}\right)$$

**BioBegin24.** Даны оптическая плотность $A$, длина оптического пути $l$ (см) и молярный коэффициент экстинкции $\varepsilon$ (л/(моль·см)). Найти концентрацию, а затем массу вещества в пробе объёмом $V$ (мл) при молярной массе $M$ (г/моль).

$$c = \frac{A}{\varepsilon \cdot l}, \quad m = c \cdot \frac{V}{1000} \cdot M$$

**BioBegin25.** Даны начальная концентрация $C_0$ (моль/л), константа скорости $k$ (с⁻¹) и время $t$ (с). Найти текущую концентрацию для реакции первого порядка.

$$C = C_0 \cdot \exp(-k \cdot t)$$

**BioBegin26.** Даны текущая концентрация $C$ (моль/л), начальная концентрация $C_0$ (моль/л) и время $t$ (с). Найти константу скорости реакции первого порядка.

$$k = \frac{\ln(C_0 / C)}{t}$$

**BioBegin27.** Даны начальная активность $A_0$, период полураспада $t_{1/2}$ (ч) и время $t$ (ч). Найти активность.

$$A = A_0 \cdot \exp\left(\frac{-\ln(2) \cdot t}{t_{1/2}}\right)$$

**BioBegin28.** Даны стандартная энергия Гиббса $\Delta G^\circ$ (кДж/моль) и температура $T$ (K). Универсальная газовая постоянная $R = 8.314$ Дж/(моль·K). Найти константу равновесия.

$$K_{eq} = \exp\left(\frac{-\Delta G^\circ \cdot 1000}{R \cdot T}\right)$$

**BioBegin29.** Даны равновесные концентрации продуктов $[P]$ (моль/л) и реагентов $[S]$ (моль/л). Найти константу равновесия, а затем стандартную энергию Гиббса при температуре $T$ (K).

$$K_{eq} = \frac{[P]}{[S]}, \quad \Delta G^\circ = -\frac{R \cdot T \cdot \ln(K_{eq})}{1000}$$

**BioBegin30.** Даны максимальная скорость $V_{max}$ (мкмоль/мин), константа Михаэлиса $K_m$ (мМ) и скорость реакции $v$ (мкмоль/мин). Найти концентрацию субстрата.

$$[S] = \frac{K_m \cdot v}{V_{max} - v}$$

**BioBegin31.** Даны скорость реакции $v$ (мкМ/с) и концентрация фермента $[E]$ (мкМ). Найти число оборотов фермента, а затем каталитическую эффективность при известной $K_m$ (М).

$$k_{cat} = \frac{v}{[E]}, \quad E = \frac{k_{cat}}{K_m}$$

**BioBegin32.** Даны масса белка $m$ (г), объём $V$ (л), температура $T$ (K) и осмотическое давление $\pi$ (кПа). Универсальная газовая постоянная $R = 8.314$ кПа·л/(моль·K). Найти молярную массу.

$$M = \frac{m \cdot R \cdot T}{\pi \cdot V}$$

**BioBegin33.** Даны давление $P$ (кПа), объём $V$ (л), температура $T$ (K) и универсальная газовая постоянная $R = 8.314$ кПа·л/(моль·K). Найти количество вещества, а затем массу при молярной массе $M$ (г/моль).

$$n = \frac{P \cdot V}{R \cdot T}, \quad m = n \cdot M$$

**BioBegin34.** Даны объём газа $V$ (л) при нормальных условиях и молярная масса $M$ (г/моль). Молярный объём $V_m = 22.4$ л/моль. Найти массу газа.

$$m = \frac{V}{V_m} \cdot M$$

**BioBegin35.** Даны массовая доля $w$ и масса раствора $m_{sol}$ (г). Найти массу растворённого вещества, а затем его количество вещества при молярной массе $M$ (г/моль).

$$m = w \cdot m_{sol}, \quad n = \frac{m}{M}$$

**BioBegin36.** Даны начальная концентрация субстрата $[S]_0$ (мМ) и концентрация продукта $[P]$ (мМ). Найти долю превращённого субстрата, а затем остаточную концентрацию субстрата.

$$\alpha = \frac{[P]}{[S]_0}, \quad [S] = [S]_0 - [P]$$

**BioBegin37.** Даны радиус ротора $r$ (см) и скорость вращения $rpm$ (об/мин). Найти относительную центробежную силу.

$$RCF = 1.118 \times 10^{-5} \cdot r \cdot rpm^2$$

**BioBegin38.** Даны активность фермента $A$ (Ед) и объём пробы $V$ (мл). Найти объёмную активность, а затем удельную активность при концентрации белка $C$ (мг/мл).

$$A_V = \frac{A}{V}, \quad A_{spec} = \frac{A_V}{C}$$

**BioBegin39.** Даны изменение концентрации $\Delta C$ (мМ) и время $\Delta t$ (с). Найти скорость реакции в мМ/с и в мкМ/мин.

$$v = \frac{\Delta C}{\Delta t}, \quad v_2 = \frac{\Delta C \cdot 1000}{\Delta t / 60}$$

**BioBegin40.** Даны масса ДНК $m$ (нг) и объём раствора $V$ (мкл). Найти концентрацию ДНК в нг/мкл, а затем оптическую плотность $A260$ (без учёта разведения).

$$C = \frac{m}{V}, \quad A260 = \frac{C}{50}$$
