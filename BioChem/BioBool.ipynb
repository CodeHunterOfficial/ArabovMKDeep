```
Вход:  x1 = 1, y1 = 5, x2 = 7, y2 = 5
Выход: TRUE
```

**BioBoolean36.** Даны координаты двух лунок. Проверить: «Лунки находятся в одной строке или одном столбце» (пипетирование мультиканальной пипеткой).

```
Вход:  x1 = 3, y1 = 2, x2 = 3, y2 = 10
Выход: TRUE
```

$$|x_1 - x_2| \le 1 \quad \text{и} \quad |y_1 - y_2| \le 1$$

```
Вход:  x1 = 1, y1 = 1, x2 = 5, y2 = 5
Выход: TRUE
```

$$(|x_1 - x_2| = 2 \text{ и } |y_1 - y_2| = 1) \quad \text{или} \quad (|x_1 - x_2| = 1 \text{ и } |y_1 - y_2| = 2)$$
