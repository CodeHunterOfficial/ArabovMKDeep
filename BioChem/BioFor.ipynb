## Часть 3. Суммы и произведения (BioFor7–BioFor11)

**BioFor7.** Даны два целых числа $A$ и $B$ ($A < B$) — номера первой и последней фракции. Найти **суммарный объём** всех фракций от $A$ до $B$ включительно (объём каждой фракции равен её номеру в мкл).

$$V = \sum_{k=A}^{B} k$$

```
Вход:  A = 3, B = 7
Выход: 25
```

**BioFor8.** Даны два целых числа $A$ и $B$ ($A < B$). Найти **произведение** всех целых чисел от $A$ до $B$ включительно — модель накопления кооперативного эффекта.

$$P = \prod_{k=A}^{B} k$$

```
Вход:  A = 2, B = 5
Выход: 120
```

**BioFor9.** Даны два целых числа $A$ и $B$ ($A < B$). Найти **сумму квадратов** всех целых чисел от $A$ до $B$ включительно — модель суммарной кинетической энергии молекул.

$$S = \sum_{k=A}^{B} k^2$$

```
Вход:  A = 1, B = 4
Выход: 30
```

**BioFor10.** Дано целое число $N > 0$ — количество измерений. Найти **сумму** $1 + 1/2 + 1/3 + \ldots + 1/N$ — модель накопления относительной ошибки при последовательных разведениях.

$$S = \sum_{k=1}^{N} \frac{1}{k}$$

```
Вход:  N = 5
Выход: 2.2833
```

**BioFor11.** Дано целое число $N > 0$ — число клеток в начальной популяции. Найти сумму $N^2 + (N+1)^2 + \ldots + (2N)^2$ — модель роста популяции с ускоренным делением.

$$S = \sum_{k=N}^{2N} k^2$$

```
Вход:  N = 2
Выход: 4 + 9 + 16 = 29
```

## Часть 4. Знакопеременные процессы (BioFor12–BioFor14)

**BioFor12.** Дано целое число $N > 0$ — число шагов разведения. Найти **произведение** $1.1 \cdot 1.2 \cdot 1.3 \cdot \ldots$ ($N$ сомножителей) — модель накопления фактора разведения при ступенчатом добавлении.

$$P = \prod_{k=1}^{N} \left(1 + \frac{k}{10}\right)$$

```
Вход:  N = 3
Выход: 1.1 * 1.2 * 1.3 = 1.716
```

**BioFor13.** Дано целое число $N > 0$ — число циклов измерения. Каждое следующее измерение имеет систематическую ошибку, знак которой чередуется (день/ночь, температура/охлаждение). Найти **суммарную ошибку** $1.1 - 1.2 + 1.3 - \ldots$ ($N$ слагаемых).

$$S = \sum_{k=1}^{N} (-1)^{k-1} \left(1 + \frac{k}{10}\right)$$

```
Вход:  N = 4
Выход: 1.1 - 1.2 + 1.3 - 1.4 = -0.2
```

**BioFor14.** Дано целое число $N > 0$ — число слоёв в колонии клеток. Каждый новый слой $k$ содержит $2k-1$ клеток (нечётное число). Найти **общее число клеток** $N^2$, используя формулу $N^2 = 1 + 3 + 5 + \ldots + (2N-1)$. После добавления каждого слоя выводить текущее суммарное число клеток.

## Часть 5. Степени и экспоненты (BioFor15–BioFor18)

**BioFor15.** Дано вещественное число $A$ — начальная концентрация и целое $N > 0$ — число циклов роста. Найти $A^N$ — модель экспоненциального роста без ограничений.

$$A^N = \underbrace{A \cdot A \cdot \ldots \cdot A}_{N}$$

```
Вход:  A = 2, N = 5
Выход: 32
```

**BioFor16.** Дано вещественное число $A$ — начальная концентрация и целое $N > 0$ — число поколений. Вывести все концентрации $A^k$ от 1 до $N$ — модель роста с шагом.

$$S = \sum_{k=0}^{N} A^k$$

```
Вход:  A = 2, N = 3
Выход: 15
```

**BioFor18.** Дано вещественное число $A$ и целое $N > 0$. Найти $1 - A + A^2 - A^3 + \ldots + (-1)^N A^N$ — модель колебательной кинетики (чередование синтеза и распада).

$$S = \sum_{k=0}^{N} (-1)^k A^k$$

**BioFor22.** Дано $X$ — начальная концентрация бактерий (условные единицы) и $N > 0$ — число поколений. **Рост бактериальной популяции** описывается рядом:

$$S = \sum_{k=0}^{N} \frac{X^k}{k!}$$

Вывести приближённое значение $S$ (при $N \to \infty$ $S \to \sin(X)$ — модель циркадного ритма).

```
Вход:  X = 1, N = 3
Выход: 0.8415
```

**BioFor24.** Дано $X$ — амплитуда **колебаний концентрации метаболита** и $N > 0$ — число гармоник. Колебания описываются рядом:

$$S = \sum_{k=0}^{N} (-1)^k \frac{X^{2k}}{(2k)!}$$

Вывести приближённое значение $S$ (при $N \to \infty$ $S \to \cos(X)$ — модель затухающих колебаний).

## Часть 11. Дополнительные биохимические задачи (BioFor36–BioFor40)

**BioFor36.** Даны целые положительные числа $N$ и $K$. Найти сумму $1^K + 2^K + \ldots + N^K$ — модель **дозозависимого ответа** (суммарный эффект $N$ доз, каждая из которых даёт вклад $i^K$). Использовать вещественную переменную.

$$S = \sum_{i=1}^{N} i^K$$

```
Вход:  N = 4, K = 2
Выход: 1 + 4 + 9 + 16 = 30.0
```

**BioFor37.** Дано целое число $N > 0$ — число поколений. Найти сумму $1^1 + 2^2 + 3^3 + \ldots + N^N$ — модель **автокатализа** (каждое поколение усиливает само себя).

$$S = \sum_{i=1}^{N} i^i$$

```
Вход:  N = 3
Выход: 1 + 4 + 27 = 32.0
```

**BioFor38.** Дано целое число $N > 0$ — число молекул. Найти сумму $1^N + 2^{N-1} + \ldots + N^1$ — модель **обратной зависимости** (эффект тем сильнее, чем меньше концентрация).

$$S = \sum_{i=1}^{N} i^{N-i+1}$$

```
Вход:  N = 3
Выход: 1 + 4 + 3 = 8.0
```

**BioFor40.** Даны целые числа $A$ и $B$ ($A < B$) — диапазон номеров фракций. Вывести все номера от $A$ до $B$ включительно; при этом номер $A$ выводится 1 раз, номер $A+1$ — 2 раза и т.д. — модель **нарастающих реплик** при увеличении номера фракции.
