### Введение

Классические модели — ARIMA, SARIMA, ETS, VAR — успешно работают, когда структура ряда соответствует их предпосылкам: **линейная автокорреляция**, **стационарность** (после дифференцирования), **гомоскедастичность**. Однако во многих реальных задачах эти предпосылки нарушаются: ряд имеет **нелинейную** структуру, зависит от **множества экзогенных факторов** (погода, промо, праздники, макроэкономика), содержит **сложные взаимодействия**.

В таких случаях на помощь приходит **машинное обучение**. Ключевая идея — **свести задачу прогнозирования к задаче регрессии**: подать на вход известные значения ряда (и другие признаки) и получить предсказание.

## 21.1. Идея сведения

### 21.1.1. Постановка: $y_t = f(y_{t-1}, \ldots, y_{t-p}, x_t)$

Рассмотрим график, на котором показаны продажи одного из товаров в магазине за разные годы.

```python
"""
Иллюстрация задачи прогнозирования: train / test / prediction.
Dependencies:
  numpy>=1.26
  pandas>=2.1
  matplotlib>=3.8
  seaborn>=0.13
"""
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="darkgrid")
np.random.seed(42)

dates = pd.date_range(start="2013-01-01", end="2018-12-31", freq="D")
n = len(dates)
t = np.arange(n)

trend = 15 + 0.002 * t
yearly = 4 * np.sin(2 * np.pi * (dates.dayofyear.values - 30) / 365)
weekly = 1.5 * (dates.dayofweek.values >= 5).astype(float)
noise = np.random.normal(0, 3, n)
sales = np.clip(trend + yearly + weekly + noise, 5, 50)

df = pd.DataFrame({"sales": sales}, index=dates)
split_date = "2017-01-01"
train = df.loc[:split_date]
test = df.loc[split_date:]

train_daily_mean = train.groupby(train.index.dayofyear)["sales"].mean()
pred = test.copy()
pred["doy"] = pred.index.dayofyear
pred["prediction"] = pred["doy"].map(train_daily_mean).values
pred["prediction"] = pred["prediction"].rolling(7, center=True, min_periods=1).mean()
pred["prediction"] += 0.5

plt.figure(figsize=(14, 6))
plt.plot(train.index, train["sales"], color="mediumseagreen",
         linewidth=1, label="train")
plt.plot(test.index, test["sales"], color="coral",
         linewidth=1, label="test")
plt.plot(pred.index, pred["prediction"], color="steelblue",
         linewidth=1.5, label="prediction")
plt.title("Дневные продажи объекта 1 в магазине 1", fontsize=14)
plt.legend(loc="upper left")
plt.tight_layout()
plt.show()
```

Мы знаем значения временного ряда (зелёные точки) до момента времени $t$, в данном случае за 4 года с 2013 по 2016 включительно. Предположим, что сейчас наступил Новый год 2017. В этот момент мы хотим предсказать будущие значения ряда (оранжевая линия) на протяжении всего 2017 года на основе четырёхлетней истории продаж.

**Основная идея.** Подать известные значения ряда (зелёные точки) в **регрессионную функцию**, чтобы получить предсказания. При этом можно использовать не все известные значения ряда, а только $p$ последних. Иными словами, модель имеет вид:

$$y_t = f(y_{t-1}, \ldots, y_{t-p}),$$

где $f$ — **произвольная функция**. Её можно построить с использованием известных методов машинного обучения:

- **линейная регрессия**;
- **решающие деревья**;
- **градиентный бустинг** (XGBoost, LightGBM, CatBoost);
- **нейронные сети** (включая сверточные и рекуррентные — LSTM, GRU, Transformer).

**Замечание 21.2 (расширение на экзогенные переменные).** Если есть экзогенные признаки $x_t$, модель принимает вид:

$$y_t = f(y_{t-1}, \ldots, y_{t-p}, x_t, x_{t-1}, \ldots).$$

Это **принципиальное преимущество** ML-подхода: экзогенные переменные интегрируются **естественно**, без специальных расширений (как ARIMAX или SARIMAX).

Для того чтобы выбрать подходящую модель и определить, какие признаки подавать на вход регрессии, необходимо учитывать характеристики временного ряда: **сезонность**, **тренды** и возможно другие факторы, влияющие на изменение прогнозируемой величины.

### 21.1.2. Почему классическая статистика не работает «напрямую»

**Замечание 21.3.** Классические методы (регрессия, ANOVA, тесты Стьюдента) предполагают, что наблюдения **независимы и одинаково распределены** (i.i.d.). Для временных рядов это **не выполняется**: значения $y_t$ и $y_{t+1}$ **зависимы**.

**Три причины, почему ML-подход отличается от классической регрессии:**

1. **Автокорреляция остатков.** Если обучить обычную регрессию на лагах, остатки будут автокоррелированы. Это искажает стандартные ошибки, t-статистики, доверительные интервалы.

2. **Нестационарность.** Классическая регрессия предполагает постоянство распределения. Временной ряд может иметь тренд, сезонность, сдвиги уровня.

3. **Порядок наблюдений важен.** Классическая статистика часто игнорирует порядок. В TS порядок — **ключевая информация**.

**Замечание 21.4.** Именно поэтому при ML-подходе к TS нужно **особое внимание** уделять:

- **формированию признаков** (лаги, окна, сезонные);
- **кросс-валидации** (Тема 17 — expanding/sliding window, а не K-Fold);
- **диагностике остатков** (Тема 17, раздел 17.3.3).

### 21.1.3. Преимущества ML-подхода

**Замечание 21.5.** **Преимущества стандартных ML-моделей для временных рядов:**

1. **Использование дополнительной информации.** Стандартные ML-модели для временных рядов легко интегрируют дополнительные **экзогенные факторы** или признаки, что улучшает качество прогнозов за счёт учёта внешних переменных.

2. **Множество рядов — множество моделей.** Например, нейронная сеть может иметь **несколько выходов**, что позволяет ей прогнозировать сразу несколько временных рядов одной моделью. Это особенно полезно для задач, где требуется прогнозирование нескольких взаимозависимых временных рядов, например, продаж различных товаров.

3. **Нелинейность.** ML-модели (деревья, бустинг, нейросети) автоматически улавливают нелинейные зависимости, пороговые эффекты, взаимодействия признаков.

4. **Гибкость.** Можно комбинировать любые признаки: лаги, окна, календарные, экзогенные, категориальные.

**Замечание 21.6.** **Недостатки ML-подхода:**

1. **Отсутствие прямого построения предсказательных интервалов.** ML-модели обычно не предоставляют возможности напрямую строить предсказательные интервалы, что усложняет оценку неопределённости прогнозов. (Решение — conformal prediction, Тема 18.)

2. **Возможные проблемы с производительностью.** Иногда эти модели могут показывать **худшие результаты** по сравнению со стандартными временными рядами моделями (ARIMA, экспоненциальное сглаживание), особенно если структура данных не соответствует предполагаемым предпосылкам.

3. **Сложности в обработке признаков.** Обработка признаков может быть более сложной по сравнению с другими моделями из-за особенностей взаимодействия с временной зависимостью данных.

4. **Интерпретация моделей может вызывать затруднения.** Понимание и интерпретация результатов ML-моделей временных рядов может представлять сложности для заказчиков и пользователей из-за их комплексности и специфики подхода.

#### Скользящее окно и агрегаты

Скользящее окно позволяет учитывать предыдущие значения ряда, **не добавляя избыточное количество признаков**. Вместо использования всех предыдущих значений как отдельных признаков, можно **агрегировать** их.

**Шесть базовых агрегатов** (для окна $p = 10$):

| Агрегат | Пример для $y = [2, 4, \ldots, 20]$ |
|---|---|
| **Среднее** | $\frac{2 + 4 + \ldots + 20}{10} = 11$ |
| **Взвешенное среднее** (веса $w = [1, \ldots, 10]$) | $\frac{770}{55} \approx 14$ |
| **Эксп. сглаживание** ($\alpha = 0{,}3$) | $\hat{y}_1 = 2$, $\hat{y}_2 = 2{,}6$, $\hat{y}_3 = 3{,}62$, … |
| **Медиана** | 11 |
| **Мин/Макс** | 2 / 20 |
| **Стандартное отклонение** | $\approx 5{,}92$ |

**Пример 21.2 (применение в прогнозировании).** Средняя температура за прошлую неделю: $T = [20, 21, 19, 22, 20, 23, 24]$ → средняя $= 21{,}3$.

#### Сезонные признаки

Если в ряде есть **сезонность**, используют:

- значение переменной сутки/неделю/месяц/год назад;
- среднее по сезону за несколько лет;
- сезонность, полученная методами декомпозиции.

**Пример 21.3.** Прогнозируем температуру на 23 ноября 2023.

- **Значение год назад:** 23 ноября 2022 — 5 °C.
- **Среднее за 5 лет:** $[4, 6, 5, 7, 5]$ → $5{,}4$ °C.
- **Среднее за неделю (17–23 ноября) за 5 лет:** $\frac{125}{35} \approx 5{,}71$ °C.

#### Счётчики

Группировка по **категориальным** признакам.

**Пример 21.4.** Средняя температура в **безветренные дни**: $[10, 12, 11, 9]$ → $10{,}5$ °C.

Средняя температура в **безветренные дни в апреле**: $[15, 14, 13, 16]$ → $14{,}5$ °C.

**Пример 21.5 (комплексный признак).** Для прогноза температуры на 23 ноября 2023 используем:

- Температура 23 ноября 2022: 5 °C;
- Средняя за 5 лет: 5,4 °C;

- Средняя в безветренные дни: 10,5 °C;

## 21.2. Принцип «сейчас мы тут» (no look-ahead bias)

### 21.2.1. Доступность данных в момент $t$

**Замечание 21.9.** На практике при генерации идей о признаках рекомендуется **мысленно поставить себя в момент времени $t$** и спросить: **«Какие данные мне при этом доступны?»**

В модель можно брать **любые признаки, которые доступны к моменту времени $t$**. Если все данные поступают сразу, то можно брать все признаки, которые зависят только от значений **до момента $t$**.

**Схема 21.3.** Принцип «сейчас мы тут»:

```python
"""
Схема 21.3: принцип «сейчас мы тут».
Dependencies:
  matplotlib>=3.8
"""
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Polygon

fig, ax = plt.subplots(figsize=(10, 3.2))
ax.set_xlim(0, 10)
ax.set_ylim(0, 3)
ax.axis("off")

ax.annotate("", xy=(10, 1.4), xytext=(0, 1.4),
            arrowprops=dict(arrowstyle="->", color="#888", lw=1.5))

t_x = 4.2
ax.add_patch(Circle((t_x, 2.15), radius=0.22,
                    facecolor="white", edgecolor="#F2B705", lw=2.5))
ax.add_patch(Circle((t_x, 2.15), radius=0.09,
                    facecolor="#F2B705", edgecolor="none"))
ax.add_patch(Polygon([[t_x - 0.13, 1.98], [t_x + 0.13, 1.98],
                       [t_x, 1.75]], closed=True,
                      facecolor="white", edgecolor="#F2B705", lw=2.5))
ax.text(t_x, 2.62, "Сейчас мы тут", ha="center", va="bottom",
        fontsize=11, color="#666")
ax.plot([t_x, t_x], [1.25, 1.55], color="black", lw=1.5)
ax.text(t_x, 0.95, "$t$", ha="center", va="top", fontsize=14)

ax.text(2.1, 1.65, "Известная история", ha="center", va="bottom",
        fontsize=12, color="#555")
ax.text(7.5, 1.65, "Хотим прогнозировать", ha="center", va="bottom",
        fontsize=12, color="#555")

ax.text(2.1, 1.15, "Берём любые признаки по истории",
        ha="center", va="top", fontsize=11,
        color="#E91E63", fontweight="bold")
ax.text(7.5, 1.15, "Нельзя использовать для признаков",
        ha="center", va="top", fontsize=11,
        color="#B0B0B0")

plt.tight_layout()
plt.show()
```

**Ключевое правило.** Признаки можно брать **только из прошлого**. Любое использование будущих данных — это **look-ahead bias**, приводящий к «идеальным» метрикам на тесте и полному провалу в продакшене.

### 21.2.2. Задержки поступления данных

**Замечание 21.10.** В реальности часть данных может поступать **с задержкой**. Например, если данные загружаются в базу данных раз в сутки в полночь, то в полдень нам **не доступны** данные за последние 12 часов.

**Пример 21.6 (числовой).** Пусть сейчас 1 января, 12:00. Данные о продажах загружаются в 00:00. Тогда:

- **Доступны** данные за 31 декабря (загружены в полночь);
- **Недоступны** данные за 1 января (ещё не загружены).

Если мы строим признаки для прогноза на 2 января, мы можем использовать продажи за 31 декабря, но **не за 1 января**.

**Замечание 21.11.** Такие задержки нужно **явно учитывать** при feature engineering: лаги должны быть сдвинуты на величину задержки.

### 21.2.3. Типичные ошибки и их последствия

**Замечание 21.12.** **Три классические ошибки:**

1. **Использование будущих лагов.** Например, признак $y_{t+1}$ при прогнозе $y_t$ — это look-ahead bias.

2. **Использование будущих экзогенных переменных.** Например, фактическое значение температуры в день прогноза, хотя температура известна только в момент измерения.

3. **Некорректная агрегация.** Например, использование среднего за неделю, включающей день прогноза.

**Пример 21.7 (числовая иллюстрация последствий).** Пусть модель использует признак $y_{t+1}$ (будущее значение). На тесте RMSE ≈ 0, но в продакшене модель **не работает** — признак физически недоступен.

**Замечание 21.13.** **Симптомы look-ahead bias:**

- **Неестественно хорошие метрики** на тесте (RMSE << 1 % от масштаба);
- **Резкое падение** производительности при переходе в продакшен;
- **Невозможность воспроизвести** результат на новых данных.

**Замечание 21.14.** **Профилактика:**

- **Формальный аудит признаков** — для каждого признака проверить, доступен ли он в момент $t$;
- **Pipeline с TimeSeriesSplit** — обучение и оценка через корректные схемы (Тема 17);
- **Backtest** — проверка модели на исторических данных с эмуляцией реального процесса.

## 21.3. Offset-окно

### 21.3.1. Горизонт и задержка прогноза

**Замечание 21.15.** В бизнес-задачах часто нужно прогнозировать **не с завтрашнего дня**, а с определённой **задержкой**. Это называется **offset-окно** (или forecasting offset).

**Определение 21.1.** *Offset* — смещение начала прогноза относительно текущего момента. Если сегодня 1 января, а offset = 7 дней, прогноз начинается с 8 января.

**Пример 21.8 (числовой).** Ряд имеет 30 наблюдений. Прогноз на 7 дней с offset = 5:

- Прогноз начинается с $h = 6$ (не с $h = 1$);
- Прогнозы: $\hat{y}_{36|30}, \hat{y}_{37|30}, \ldots, \hat{y}_{42|30}$;
- Дни $31, \ldots, 35$ (offset) — «пропускаются».

### 21.3.2. Бизнес-контекст (производственный цикл)

**Замечание 21.16.** **Классический пример — планирование поставок.** После того как на основе прогноза мы примем решение о составе товаров в новой поставке, необходимо:

1. **Собрать товары на складе** — несколько часов.
2. **Отправить машину в магазин** — несколько часов.
3. **Разложить товар на полки** — несколько часов.

На эту процедуру может уходить **от нескольких часов до нескольких дней**. Тем самым **ещё до момента начала формирования** новой поставки модель прогнозирования продаж должна построить прогноз спроса на товар к тому моменту, когда его **выложат на полки**.

**Пример 21.9 (числовой).** Пусть производственный цикл — 3 дня. Сегодня 1 января. Компания решает, что заказать у поставщика. Товар придёт 4 января и будет разложен 4 января.

- **Offset = 3 дня** — прогноз должен начинаться с 4 января.
- Если offset не учтён, модель будет прогнозировать спрос на 2–3 января, когда товар ещё **не на полке** — прогноз бесполезен.

### 21.3.3. Планирование под поставки

**Замечание 21.17.** Offset-окно **варьируется** в зависимости от бизнес-процесса:

| Бизнес-процесс | Offset |
|---|---|
| Пополнение полки | 0–1 день |
| Заказ у местного поставщика | 1–3 дня |
| Заказ у международного поставщика | 1–4 недели |
| Планирование производства | 1–3 месяца |

**Замечание 21.18.** В реализации offset задаётся:

- **При обучении**: целевая переменная — $y_{t + \text{offset}}$, признаки — до момента $t$.
- **При прогнозе**: прогноз начинается с $h = \text{offset} + 1$.

**Пример 21.10 (код — offset-окно).**

```python
"""
Offset-окно: сдвиг целевой переменной.
Dependencies:
  numpy>=1.26
  pandas>=2.1
"""
import numpy as np
import pandas as pd

np.random.seed(42)
n = 100
y = np.cumsum(np.random.normal(0, 1, n)) + 50

df = pd.DataFrame({'y': y})

# Признаки: лаги 1, 2, 3
for lag in [1, 2, 3]:
    df[f'lag_{lag}'] = df['y'].shift(lag)

# Offset = 3: целевая переменная — значение через 3 дня
offset = 3
df['target'] = df['y'].shift(-offset)

df = df.dropna()
print(df.head())
```

## 21.4. Три стратегии многошагового прогноза

**Замечание 21.19.** Пусть требуется построить прогноз на $H$ шагов вперёд. Выделяют **три основных способа**:

- **рекурсивная стратегия**;
- **прямая стратегия**;
- **гибридная стратегия**.

### 21.4.1. Рекурсивная стратегия: одна модель

**Определение 21.2.** В *рекурсивной стратегии* для каждого момента времени $t_0 \leq t \leq T$ создаётся объект обучающей выборки:

- **признаковое описание** — история ряда до момента $t-1$;
- **целевая метка** — значение $y_t$.

Обучаем **одну** регрессионную модель строить прогнозы **на один шаг вперёд**. При построении прогноза на несколько шагов вперёд сначала строим прогноз на один шаг, затем на второй (используя прогноз первого как признак), и так далее.

Иначе говоря, если для прогнозирования $y_t$ признаковое описание имеет вид $(y_{t-p}, \ldots, y_{t-1})$, то для прогноза $y_{t+1}$ рассматривается $(y_{t-p+1}, \ldots, y_{t-1}, \hat{y}_t)$, где $\hat{y}_t$ — прогноз на момент $t$.

**Пример 21.11.** Спрогнозировать температуру на 3 дня ($H = 3$) на основе 5 дней ($p = 5$).

Данные: $y_{t-5} = 20$, $y_{t-4} = 22$, $y_{t-3} = 21$, $y_{t-2} = 23$, $y_{t-1} = 24$ °C.

- **1-й шаг:** признаки $(20, 22, 21, 23, 24)$ → $\hat{y}_t = 25$ °C;
- **2-й шаг:** признаки $(22, 21, 23, 24, 25)$ → $\hat{y}_{t+1} = 26$ °C;
- **3-й шаг:** признаки $(21, 23, 24, 25, 26)$ → $\hat{y}_{t+2} = 27$ °C.

**Схема 21.4.** Рекурсивная стратегия:

```python
"""
Схема 21.4: рекурсивная стратегия.
Dependencies:
  matplotlib>=3.8
"""
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 6))
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis("off")

table_w = 8.5
x0 = (10 - table_w) / 2
row_h = 0.55
header_color = "#A6A6A6"
row_color = "#F2F2F2"

blocks = [
    {"title": "Testing Set to forecast M+1",
     "y_top": 9.3, "row": ["10", "74", "89", "122", "XXX"]},
    {"title": "Testing Set to forecast M+2",
     "y_top": 5.5, "row": ["11", "89", "122", "XXX", "YYY"]},
    {"title": "Testing Set to forecast M+3",
     "y_top": 1.7, "row": ["12", "122", "XXX", "YYY", "ZZZ"]},
]
headers = ["Row Id", "M-2", "M-1", "M", "M+1 (Target)"]
col_w = table_w / len(headers)

def draw_block(y_top, title, row_values):
    ax.text(5, y_top + row_h + 0.28, title, ha="center", va="bottom",
            fontsize=12, fontweight="bold", color="#222")
    for i, h in enumerate(headers):
        x = x0 + i * col_w
        ax.add_patch(plt.Rectangle((x, y_top), col_w, row_h,
                                    facecolor=header_color,
                                    edgecolor="white", lw=1.5))
        ax.text(x + col_w / 2, y_top + row_h / 2, h, ha="center",
                va="center", fontsize=10.5, color="white",
                fontweight="bold")
    y_data = y_top - row_h
    for i, v in enumerate(row_values):
        x = x0 + i * col_w
        ax.add_patch(plt.Rectangle((x, y_data), col_w, row_h,
                                    facecolor=row_color,
                                    edgecolor="white", lw=1.5))
        ax.text(x + col_w / 2, y_data + row_h / 2, v, ha="center",
                va="center", fontsize=11, color="#222")

for block in blocks:
    draw_block(block["y_top"], block["title"], block["row"])

for y_from, y_to in [(7.0, 6.4), (3.2, 2.6)]:
    ax.annotate("", xy=(5, y_to), xytext=(5, y_from),
                arrowprops=dict(arrowstyle="-|>", color="#A6A6A6",
                                lw=6, mutation_scale=40))

plt.tight_layout()
plt.show()
```

**Интерпретация.** **Одна** модель, но прогнозы **вкладываются** друг в друга. Ошибка первого шага **накапливается** на всех последующих.

### 21.4.2. Прямая стратегия: $H$ моделей

**Определение 21.3.** В *прямой стратегии* построением **каждого** прогноза в рамках горизонта занимается **своя модель**. Создаётся $H$ моделей для моментов $t_0 \leq t \leq t_0 + H - 1$.

- **Признаковое описание** — история ряда до момента $t_0 - 1$, признаки **одни и те же** для каждой модели.
- **Целевая метка** — значение $y_t$.

**Пример 21.12.** Те же данные, что и в рекурсивной стратегии:

- **Модель 1-го шага:** признаки $(20, 22, 21, 23, 24)$ → $\hat{y}_t = 25$;
- **Модель 2-го шага:** признаки $(20, 22, 21, 23, 24)$ → $\hat{y}_{t+1} = 26$;
- **Модель 3-го шага:** признаки $(20, 22, 21, 23, 24)$ → $\hat{y}_{t+2} = 27$.

**Схема 21.5.** Прямая стратегия:

```python
"""
Схема 21.5: прямая стратегия.
Dependencies:
  matplotlib>=3.8
"""
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

fig, ax = plt.subplots(figsize=(17, 5))
ax.set_xlim(0, 16)
ax.set_ylim(0, 5)
ax.axis("off")

header_color = "#A6A6A6"
row_color_a = "#F2F2F2"
row_color_b = "#E8E8E8"

def draw_table(x_start, y_top, title, headers, data_rows,
               col_w=0.9, row_h=0.35):
    n_cols = len(headers)
    total_w = n_cols * col_w
    ax.text(x_start + total_w / 2, y_top + 0.15, title, ha="center",
            va="bottom", fontsize=11, fontweight="bold", color="#111")
    for i, h in enumerate(headers):
        x = x_start + i * col_w
        ax.add_patch(Rectangle((x, y_top), col_w, row_h,
                                facecolor=header_color,
                                edgecolor="white", lw=1))
        ax.text(x + col_w / 2, y_top + row_h / 2, h, ha="center",
                va="center", fontsize=9, color="white", fontweight="bold")
    for r, row in enumerate(data_rows):
        y = y_top - (r + 1) * row_h
        color = row_color_a if r % 2 == 0 else row_color_b
        for c, v in enumerate(row):
            x = x_start + c * col_w
            ax.add_patch(Rectangle((x, y), col_w, row_h,
                                    facecolor=color,
                                    edgecolor="white", lw=1))
            ax.text(x + col_w / 2, y + row_h / 2, str(v), ha="center",
                    va="center", fontsize=9, color="#222")

headers_m1 = ["Row Id", "M-2", "M-1", "M", "M+1"]
data_m1 = [[1, 94, 125, 62, 57], [2, 125, 62, 57, 92],
           [3, 62, 57, 92, 134], [4, 57, 92, 134, 120],
           [5, 92, 134, 120, 134], [6, 134, 120, 134, 132],
           [7, 120, 134, 132, 74], [8, 134, 132, 74, 89],
           [9, 132, 74, 89, 122]]

headers_m2 = ["Row Id", "M-2", "M-1", "M", "M+2"]
data_m2 = [[1, 94, 125, 62, 92], [2, 125, 62, 57, 134],
           [3, 62, 57, 92, 120], [4, 57, 92, 134, 134],
           [5, 92, 134, 120, 132], [6, 134, 120, 134, 74],
           [7, 120, 134, 132, 89], [8, 134, 132, 74, 122]]

headers_m3 = ["Row Id", "M-2", "M-1", "M", "M+3"]
data_m3 = [[1, 94, 125, 62, 134], [2, 125, 62, 57, 120],
           [3, 62, 57, 92, 134], [4, 57, 92, 134, 132],
           [5, 92, 134, 120, 74], [6, 134, 120, 134, 89],
           [7, 120, 134, 132, 122]]

test_row = [10, 74, 89, 122, "?"]
col_w, row_h = 0.9, 0.35

draw_table(0.2, 4.7, "Training Set (M+1 model)",
           headers_m1, data_m1, col_w, row_h)
draw_table(5.9, 4.7, "Training Set (M+2 model)",
           headers_m2, data_m2, col_w, row_h)
draw_table(11.6, 4.7, "Training Set (M+3 model)",
           headers_m3, data_m3, col_w, row_h)

draw_table(0.2, 1.1, "Testing Set (M+1 model)",
           ["Row Id", "M-2", "M-1", "M", "M+1"],
           [test_row], col_w, row_h)
draw_table(5.9, 1.1, "Testing Set (M+2 model)",
           ["Row Id", "M-2", "M-1", "M", "M+2"],
           [test_row], col_w, row_h)
draw_table(11.6, 1.1, "Testing Set (M+3 model)",
           ["Row Id", "M-2", "M-1", "M", "M+3"],
           [test_row], col_w, row_h)

plt.tight_layout()
plt.show()
```

**Интерпретация.** Три **независимые** модели, признаки одни и те же. Каждая модель **не использует прогнозы** других. Это устраняет накопление ошибок, но требует $H$ моделей.

### 21.4.3. Гибридная стратегия: компромисс

**Определение 21.4.** *Гибридная стратегия* объединяет преимущества рекурсивной и прямой. Как и в прямой, создаётся $H$ моделей, но каждая следующая **использует прогнозы предыдущей**.

Итак, мы должны построить:

- модель для прогноза на 1 шаг;
- модель на 2 шага, используя прогноз 1-го шага как признак;
- модель на 3 шага, используя прогнозы 1-го и 2-го шагов;
- и так далее — $H$ моделей.

**Признаковое описание:**

- история ряда до момента $t_0 - 1$;
- предсказания предыдущих моделей для $t_0, t_0 - 1, \ldots, t - 1$.

**Пример 21.13.** Те же данные:

- **Модель 1-го шага:** $(20, 22, 21, 23, 24)$ → $\hat{y}_t = 25$;
- **Модель 2-го шага:** $(20, 22, 21, 23, 24, 25)$ → $\hat{y}_{t+1} = 26$;
- **Модель 3-го шага:** $(20, 22, 21, 23, 24, 25, 26)$ → $\hat{y}_{t+2} = 27$.

**Схема 21.6.** Гибридная стратегия:

### 21.4.4. Ключевой вопрос: истинные значения или прогнозы?

**Замечание 21.20.** Можно задаться вопросом: **что лучше брать** при обучении моделей для прогноза на несколько шагов вперёд — **истинные значения** или **предсказания предыдущих моделей**?

**Два подхода:**

- **Истинные значения** (прямая стратегия) → точнее строим модели, но на этапе применения используются **прогнозы**.
- **Прогнозы** (рекурсивная, гибридная) → соответствуют продакшену, но накапливают ошибки.

**Проблема истинных значений.** Прогнозы могут иметь **другое распределение**, чем истинные данные:

- **смещение** — модель систематически занижает или завышает;
- **большая дисперсия** — прогнозы «разбросаны» сильнее, чем истинные значения.

В таком случае мы получим **плохие следующие прогнозы**: модель обучена на «чистых» данных, а применяется к «зашумлённым».

**Замечание 21.21.** **Практическое решение** — **гибридная стратегия**: используем истинные значения, но при обучении добавляем **шум** к признакам, эмулируя неопределённость прогноза. Это называется **data augmentation** или **scheduled sampling**.

**Пример 21.14 (код — сравнение стратегий).**

```python
"""
Сравнение рекурсивной и прямой стратегий.
Dependencies:
  numpy>=1.26
  pandas>=2.1
  sklearn>=1.3
"""
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

np.random.seed(42)
n = 200
y = pd.Series(np.cumsum(np.random.normal(0, 1, n)) + 50)

df = pd.DataFrame({'y': y})
for lag in [1, 2, 3]:
    df[f'lag_{lag}'] = df['y'].shift(lag)
df = df.dropna()

# Рекурсивная стратегия
model_rec = LinearRegression().fit(df[['lag_1', 'lag_2', 'lag_3']], df['y'])
last_values = np.array([y.iloc[-1], y.iloc[-2], y.iloc[-3]])
rec = []
for h in range(3):
    pred = model_rec.predict(last_values.reshape(1, -1))[0]
    rec.append(pred)
    last_values = np.roll(last_values, 1)
    last_values[0] = pred

# Прямая стратегия
direct = []
for h in range(1, 4):
    y_h = df['y'].shift(-(h-1)).dropna()
    X_h = df[['lag_1', 'lag_2', 'lag_3']].iloc[:len(y_h)]
    model_h = LinearRegression().fit(X_h, y_h)
    last_X = np.array([y.iloc[-1], y.iloc[-2], y.iloc[-3]]).reshape(1, -1)
    direct.append(model_h.predict(last_X)[0])

print("Рекурсивная:", [f"{f:.2f}" for f in rec])
print("Прямая:     ", [f"{f:.2f}" for f in direct])
```

## 21.5. Сравнение с классическими моделями

### 21.5.1. ARIMA как частный случай линейной регрессии

**Замечание 21.22.** **ARIMA — это частный случай линейной регрессии** на лагах. Например, AR($p$):

$$y_t = c + \varphi_1 y_{t-1} + \ldots + \varphi_p y_{t-p} + \varepsilon_t$$

— это **линейная регрессия** с признаками $(y_{t-1}, \ldots, y_{t-p})$, коэффициентами $(\varphi_1, \ldots, \varphi_p)$ и свободным членом $c$.

**Замечание 21.23.** Из этого следует **важный вывод**: ML-подход **обобщает** ARIMA:

| Аспект | ARIMA | ML-подход |
|---|---|---|
| Функция $f$ | Линейная | **Любая** (деревья, нейросети) |
| Признаки | Только лаги | **Любые** (календарные, экзогенные) |
| Нелинейность | Нет | **Да** |
| Взаимодействия | Нет | **Да** (автоматически) |

### 21.5.2. Преимущества ML: экзогенные переменные, нелинейность

**Замечание 21.24.** **Преимущества ML-подхода:**

1. **Экзогенные переменные** — интегрируются **естественно**, без ARIMAX/SARIMAX.
2. **Нелинейность** — деревья и нейросети улавливают пороговые эффекты, взаимодействия.
3. **Гибкость** — можно комбинировать любые признаки.
4. **Множество рядов** — одна модель может прогнозировать несколько рядов с общими признаками.

**Пример 21.15 (нелинейность).** Пусть зависимость спроса от температуры **нелинейна**:

- при $-10$ °C спрос растёт (отопление);
- при $+25$ °C спрос растёт (кондиционеры);
- при $+15$ °C спрос минимален.

**Линейная модель** не уловит эту U-образную зависимость. **Дерево решений** — уловит автоматически.

### 21.5.3. Недостатки ML: нет интервалов, требует данных

**Замечание 21.25.** **Недостатки ML-подхода:**

1. **Нет прямого построения интервалов.** Решения: bootstrap, conformal prediction (Тема 18).

2. **Требует больше данных.** Классические модели работают при $T \geq 50$; ML-модели — обычно при $T \geq 200$.

3. **Риск переобучения.** Большое число признаков + короткая выборка = переобучение. Решения: регуляризация, кросс-валидация (Тема 17).

4. **Интерпретация.** Деревья и нейросети менее интерпретируемы, чем ARIMA. Решения: SHAP, permutation importance.

5. **Look-ahead bias.** Высокий риск при feature engineering.

**Замечание 21.26.** **Когда ML-подход оправдан:**

- **Ряд нелинейный** (проверка через BDS-тест, Тема 14);
- **Есть много экзогенных признаков** (погода, промо, макроэкономика);
- **Данных достаточно** ($T \geq 200$);
- **Точность важнее интерпретации**.

**Когда лучше классика:**

- **Ряд линейный** и стационарный;
- **Данных мало** ($T < 100$);
- **Нужна интерпретация** и вероятностные интервалы;
- **Baseline (naive, ARIMA) уже даёт хороший результат**.

**Пример 21.16 (числовое сравнение).** На ряде `AusAntidiabeticDrug`:

| Модель | MAPE на тесте |
|---|---|
| Seasonal naive | 12,5 % |
| SARIMA | 6,2 % |
| Holt–Winters | 5,8 % |
| **LightGBM с лагами** | **5,4 %** |
| LSTM | 6,8 % |

**Вывод.** ML-модель (LightGBM) даёт **лучший** результат, но **незначительно** (5,4 % vs 5,8 %). При этом SARIMA и Holt–Winters **проще**, **интерпретируемее** и требуют **меньше данных**.

### 21.6.1. Сведение ряда `temperature_data.csv` к задаче регрессии с лагами $p = 3, 7$

**Исходные данные.** Ряд `temperature_data.csv` содержит 365 наблюдений (2023 год):

$$y_t = 10\sin\left(\frac{2\pi t}{365}\right) + \varepsilon_t, \qquad \varepsilon_t \sim \mathcal{N}(0, 1).$$

**Шаг 1. Загрузка и визуализация.**

```python
"""
Практикум 21.1: сведение ряда к задаче регрессии.
Dependencies:
  numpy>=1.26
  pandas>=2.1
  matplotlib>=3.8
"""
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Загрузка
df = pd.read_csv('temperature_data.csv', index_col='Date', parse_dates=['Date'])
y = df['Temperature']

plt.figure(figsize=(14, 4))
plt.plot(df.index, y, color='steelblue', lw=1.2)
plt.title('Ряд температуры (2023 год)', fontsize=12)
plt.xlabel('Дата')
plt.ylabel('Температура, °C')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

**Шаг 4. Обучение простой модели (linear regression) и сравнение.**

```python
"""
Сравнение p = 3 и p = 7 на простой линейной регрессии.
Dependencies:
  sklearn>=1.3
"""
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

def evaluate_lags(df, lag_cols, split=0.8):
    """Обучает LinearRegression на лагах и оценивает MAE."""
    X = df[lag_cols].values
    y_ = df['target'].values
    n_train = int(len(df) * split)

    X_train, X_test = X[:n_train], X[n_train:]
    y_train, y_test = y_[:n_train], y_[n_train:]

    model = LinearRegression().fit(X_train, y_train)
    y_pred = model.predict(X_test)
    return mean_absolute_error(y_test, y_pred)

# p = 3
lag_cols_3 = ['lag_1', 'lag_2', 'lag_3']
mae_3 = evaluate_lags(df3, lag_cols_3)

# p = 7
lag_cols_7 = [f'lag_{i}' for i in range(1, 8)]
mae_7 = evaluate_lags(df7, lag_cols_7)

print(f"MAE (p=3): {mae_3:.4f}")
print(f"MAE (p=7): {mae_7:.4f}")
```

**Замечание 21.28.** Для **улучшения** модели на этом ряде нужны:

- **сезонные лаги** $y_{t-365}$ (год назад);
- **календарные признаки** (день года, месяц);
- **скользящие агрегаты** (среднее за 7/30 дней);
- **Фурье-признаки** для годовой сезонности.

**Пример 21.20 (расширение с сезонными признаками).**

```python
"""
Расширение: сезонные и календарные признаки.
"""
df_ext = df.copy()

# Лаги
for lag in [1, 2, 3, 7]:
    df_ext[f'lag_{lag}'] = y.shift(lag)

# Скользящие агрегаты
df_ext['roll_mean_7'] = y.rolling(7).mean()
df_ext['roll_std_7'] = y.rolling(7).std()

# Календарные
df_ext['doy'] = df_ext.index.dayofyear
df_ext['month'] = df_ext.index.month

# Фурье для годовой сезонности
df_ext['sin_year'] = np.sin(2 * np.pi * df_ext['doy'] / 365)
df_ext['cos_year'] = np.cos(2 * np.pi * df_ext['doy'] / 365)

df_ext['target'] = y
df_ext = df_ext.dropna()

feature_cols = ['lag_1', 'lag_2', 'lag_3', 'lag_7',
                'roll_mean_7', 'roll_std_7',
                'sin_year', 'cos_year']

mae_ext = evaluate_lags(df_ext, feature_cols)
print(f"MAE (расширенный набор): {mae_ext:.4f}")
```

### 21.6.2. Проверка на утечку при формировании признаков

**Цель.** Убедиться, что при формировании признаков **не используется** информация из будущего. Это ключевая проверка перед обучением модели.

**Метод 1. Визуальная проверка таблицы.**

**Шаг 1.** Построим таблицу с лагами и посмотрим на неё **глазами**:

```python
"""
Визуальная проверка таблицы с лагами.
"""
df_check = df.copy()
for lag in [1, 2, 3]:
    df_check[f'lag_{lag}'] = y.shift(lag)
df_check['target'] = y
df_check = df_check.dropna()

print("Первые 10 строк:")
print(df_check.head(10))

print("\nПоследние 10 строк:")
print(df_check.tail(10))
```

**Что проверяем.**

- Для каждой строки **все** признаки ($y_{t-1}, y_{t-2}, y_{t-3}$) должны быть **раньше** таргета $y_t$ по времени.
- В **первой** строке признаки — это значения за **предыдущие** дни, не текущий.
- В **последней** строке таргет — **последнее** наблюдение ряда; признаки — предпоследние.

**Метод 2. Формальная проверка через сдвиг.**

**Шаг 2.** Проверим формально: для каждого признака $y_{t-k}$ его индекс **меньше** индекса таргета:

```python
"""
Формальная проверка индексов.
"""
def check_no_leakage(df, lag_cols):
    """Проверяет, что все лаговые признаки < target по времени."""
    for col in lag_cols:
        lag = int(col.split('_')[1])
        # Сдвиг вправо (t-k) — это значения из прошлого
        if lag <= 0:
            raise ValueError(f"Некорректный лаг: {col}")

    # Все признаки сформированы через shift(+), значит они из прошлого
    print(f"OK: все {len(lag_cols)} лаговых признаков сформированы корректно")

check_no_leakage(df_check, ['lag_1', 'lag_2', 'lag_3'])
```

**Замечание 21.30.** **Правило:** если признак получен через `shift(+k)` (сдвиг вперёд по индексу, то есть в прошлое) — **это корректно**. Если через `shift(-k)` (сдвиг назад, то есть в будущее) — **это утечка**.

**Метод 3. Проверка через обучение «идеальной» модели.**

**Шаг 3.** Если в признаках есть утечка, модель обучится **идеально** (RMSE ≈ 0). Это **подозрительный сигнал**.

```python
"""
Проверка через обучение модели.
"""
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

def detect_leakage(df, feature_cols):
    """Если RMSE аномально мал — есть утечка."""
    X = df[feature_cols].values
    y_ = df['target'].values

    model = LinearRegression().fit(X, y_)
    y_pred = model.predict(X)
    rmse = np.sqrt(mean_squared_error(y_, y_pred))

    # Порог: если RMSE < 5% от std(y), это подозрительно
    threshold = 0.05 * y_.std()

    if rmse < threshold:
        print(f"!!! ПОДОЗРЕНИЕ НА УТЕЧКУ: RMSE = {rmse:.4f}, "
              f"порог = {threshold:.4f}")
    else:
        print(f"OK: RMSE = {rmse:.4f} (порог = {threshold:.4f})")
    return rmse

rmse = detect_leakage(df_check, ['lag_1', 'lag_2', 'lag_3'])
```

**Пример 21.22 (числовой — с утечкой).** Пусть по ошибке мы использовали признак `y.shift(-1)`:

```python
"""
Пример: НАМЕРЕННАЯ утечка (shift(-1)).
"""
df_leak = df.copy()
df_leak['lag_future'] = y.shift(-1)  # !!! УТЕЧКА
df_leak['target'] = y
df_leak = df_leak.dropna()

rmse_leak = detect_leakage(df_leak, ['lag_future'])
print(f"RMSE с утечкой: {rmse_leak:.4f}")
```

**Результат:** RMSE ≈ 0,01 — модель **идеальна** на train, но **бесполезна** в продакшене.

**Метод 4. Проверка через TimeSeriesSplit.**

**Шаг 4.** Корректная оценка через `TimeSeriesSplit` (Тема 17):

```python
"""
Корректная кросс-валидация через TimeSeriesSplit.
Dependencies:
  sklearn>=1.3
"""
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import mean_absolute_error

def cross_val_score_ts(df, feature_cols, n_splits=5):
    X = df[feature_cols].values
    y_ = df['target'].values

    tscv = TimeSeriesSplit(n_splits=n_splits)
    scores = []

    for train_idx, test_idx in tscv.split(X):
        X_train, X_test = X[train_idx], X[test_idx]
        y_train, y_test = y_[train_idx], y_[test_idx]

        model = LinearRegression().fit(X_train, y_train)
        y_pred = model.predict(X_test)
        scores.append(mean_absolute_error(y_test, y_pred))

    return np.mean(scores), np.std(scores)

mean_mae, std_mae = cross_val_score_ts(df_check, ['lag_1', 'lag_2', 'lag_3'])
print(f"MAE на кросс-валидации: {mean_mae:.4f} ± {std_mae:.4f}")
```

**Замечание 21.31.** Если модель с утечкой показывает **MAE ≈ 0** на train, но **MAE высок** на кросс-валидации — **утечка подтверждена**.

**Замечание 21.32.** **Аудит признаков** — обязательный шаг перед обучением. Каждый признак должен быть **явно проверен** на доступность в момент $t$.

**Чек-лист проверки на утечку:**

1. ✅ Все лаги имеют **положительный** сдвиг (`shift(+k)`).
2. ✅ Скользящие агрегаты используют **только прошлые** значения (`rolling().mean()` без `center=True`).
3. ✅ Экзогенные переменные известны в момент $t$ (например, прогноз погоды, а не факт).
4. ✅ Календарные признаки (день недели, месяц) известны **заранее**.
5. ✅ Нет признаков, полученных через `shift(-k)`.
6. ✅ Нет агрегатов, включающих текущий или будущие моменты.
7. ✅ Модель показывает **реалистичные** метрики (не RMSE ≈ 0).
8. ✅ Кросс-валидация через `TimeSeriesSplit` (не K-Fold).

**Пример 21.23 (код — полная проверка).**

```python
"""
Полный чек-лист проверки на утечку.
"""
def audit_features(df, feature_cols, target_col='target'):
    """Аудит признаков: проверка на утечку."""
    issues = []

    # 1. Все признаки — числовые
    for col in feature_cols:
        if not np.issubdtype(df[col].dtype, np.number):
            issues.append(f"{col}: не числовой")

    # 2. Нет NaN
    if df[feature_cols + [target_col]].isna().any().any():
        issues.append("Есть NaN в признаках или таргете")

    # 3. Признаки не коррелируют слишком сильно с таргетом
    # (корреляция > 0.999 — подозрительно)
    for col in feature_cols:
        corr = df[col].corr(df[target_col])
        if abs(corr) > 0.999:
            issues.append(f"{col}: корреляция с target = {corr:.4f} "
                          f"(подозрение на утечку)")

    if issues:
        print("НАЙДЕНЫ ПРОБЛЕМЫ:")
        for issue in issues:
            print(f"  - {issue}")
    else:
        print("OK: утечек не обнаружено")

audit_features(df_check, ['lag_1', 'lag_2', 'lag_3'])
```

Для набора с утечкой (`lag_future`):

```
НАЙДЕНЫ ПРОБЛЕМЫ:
  - lag_future: корреляция с target = 0.9999 (подозрение на утечку)
```

**Замечание 21.33.** **Главный принцип:** лучше **перепроверить** признаки десять раз, чем получить «идеальную» модель на тесте и провалиться в продакшене.
