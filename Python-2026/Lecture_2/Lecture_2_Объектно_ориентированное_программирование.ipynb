# Глава 1. Классы и объекты

## 1.1. Зачем нужны классы

Представь, что мы пишем программу для управления домом. Нужно хранить данные о комнатах: название, площадь, этаж. Первое, что приходит в голову — завести отдельные переменные для каждой комнаты:

```python
name_1 = "Кухня"
area_1 = 12
floor_1 = 2

name_2 = "Спальня"
area_2 = 18
floor_2 = 2

name_3 = "Ванная"
area_3 = 6
floor_3 = 2
```

Пока комнат три — терпимо. Но уже на пятой комнате появляются проблемы:

- **Имена путаются.** Легко перепутать `area_2` и `area_3`, особенно если код большой.
- **Нет связи между данными.** Ничто не говорит, что `name_1`, `area_1`, `floor_1` относятся к одной комнате.
- **Сложно передавать в функции.** Чтобы передать «одну комнату» в функцию, нужно передавать три переменные:

```python
def show_room(name, area, floor):
    print(f"{name} — {area} м², {floor} этаж")

show_room(name_1, area_1, floor_1)
show_room(name_2, area_2, floor_2)
```

- **Сложно расширять.** Если завтра добавится ещё одно поле (например, «тип отделки»), придётся править все функции и все вызовы.

Класс решает все эти проблемы: он **объединяет связанные данные и поведение в единое целое**. Вместо трёх переменных — один объект, в котором лежит всё нужное.

---

## 1.2. Класс и объект: главное различие

**Класс** — это шаблон, чертёж, описание того, какими бывают объекты данного типа. Класс не содержит конкретных данных — он лишь описывает, **какие поля и методы вообще бывают** у объектов этого типа.

**Объект** — это конкретное воплощение класса, экземпляр с конкретными значениями.

Лучшая аналогия — чертёж дома и сам дом:

| Понятие | Аналогия | Пример |
|---------|----------|--------|
| Класс | Чертёж дома | «Комната» — у любой есть название, площадь, этаж |
| Объект | Построенный дом | Кухня, 12 м², 2 этаж |

По одному чертежу можно построить **сколько угодно зданий**, и каждое будет **отдельным объектом** этого класса. Каждое здание будет иметь те же поля, но с разными значениями.

Ключевая мысль: **класс — это не сама комната**, а описание, что у комнаты вообще бывает. Объект — конкретная комната с конкретными значениями.

---

## 1.3. Как определить класс

Класс определяется ключевым словом `class`. После него — имя класса. По договорённости (не требованию языка) имена классов пишут **с большой буквы**, а имена переменных — с маленькой. Так проще отличать классы от переменных при чтении кода.

Простейший класс выглядит так:

```python
class Room:
    pass
```

`pass` — это заглушка. Она говорит Python: «синтаксически здесь должно быть тело класса, но мы пока ничего не определяем». Без `pass` Python выдаст ошибку, потому что тело класса не может быть пустым.

Этот класс `Room` описывает понятие «комната», но пока ничего о ней не говорит: ни полей, ни методов.

---

## 1.4. Создание объектов

Чтобы создать объект, нужно **вызвать класс как функцию**:

```python
kitchen = Room()
bedroom = Room()
```

Мы получили **два разных объекта** класса `Room`. Это не один объект, а два независимых:

```python
print(kitchen is bedroom)   # False
print(id(kitchen))          # например, 140234567890
print(id(bedroom))          # например, 140234567891
```

`is` сравнивает, один ли это объект. `id()` возвращает адрес объекта в памяти. У `kitchen` и `bedroom` адреса разные — значит, это два разных объекта.

Пока эти объекты пусты: в них нет ни названия, ни площади.

---

## 1.5. Конструктор `__init__`

Чтобы объект сразу получал нужные данные, используют **конструктор** — специальный метод с именем `__init__` (по два подчёркивания с каждой стороны).

```python
class Room:
    def __init__(self, name, area, floor):
        self.name = name
        self.area = area
        self.floor = floor
```

Что происходит при создании объекта:

```python
kitchen = Room("Кухня", 12, 2)
```

1. Python создаёт новый пустой объект.
2. Python **автоматически** вызывает метод `__init__`, передавая туда новый объект и наши аргументы: `"Кухня"`, `12`, `2`.
3. Внутри `__init__` мы сохраняем значения в атрибуты через `self`.
4. Готовый объект возвращается и присваивается переменной `kitchen`.

### Нюанс: `__init__` не возвращает ничего

`__init__` — это **инициализатор**, а не создатель. Он не должен возвращать значение. Если попробовать:

```python
def __init__(self, name):
    self.name = name
    return self.name   # ← ошибка!
```

Python выдаст ошибку: `TypeError: __init__() should return None`. Создание объекта делает другой метод — `__new__` (мы разберём его в главе 37). `__init__` только заполняет уже созданный объект.

### Нюанс: имя `__init__` строго фиксировано

Именно `__init__`, а не `_init_`, не `init`, не `__init`. Если ошибиться в подчёркиваниях, Python не вызовет метод автоматически:

```python
class Room:
    def _init_(self, name):    # ← НЕ сработает!
        self.name = name

kitchen = Room("Кухня")        # TypeError: __init__() takes 1 positional argument
```

---

## 1.6. `self` — что это и зачем

`self` — это **ссылка на текущий объект**. Когда метод вызывается у объекта, Python автоматически подставляет этот объект в первый параметр метода.

### Зачем нужен `self`

Когда мы пишем:

```python
self.name = name
```

мы говорим: «возьми значение параметра `name` и положи его в атрибут `name` **именно этого объекта**».

Без `self` Python не понял бы, к какому объекту относится атрибут. Ведь класс один, а объектов может быть много:

```python
kitchen = Room("Кухня", 12, 2)
bedroom = Room("Спальня", 18, 2)
```

`self` в момент создания `kitchen` указывает на `kitchen`, а в момент создания `bedroom` — на `bedroom`. Поэтому каждый объект хранит **свои** значения.

### Нюанс: `self` — это просто имя

`self` — это **не ключевое слово**, а просто соглашение. Первый параметр метода можно назвать как угодно, и код будет работать:

```python
class Room:
    def __init__(this, name):    # работает!
        this.name = name
```

Но так писать **не принято** — весь Python-мир использует `self`. Если назвать иначе, коллеги не поймут код.

### Нюанс: при вызове через объект `self` не передаётся

Это очень важный момент. Когда мы вызываем метод через объект:

```python
kitchen.info()
```

Python **сам** подставляет `kitchen` в `self`. Мы не пишем `kitchen.info(kitchen)`.

Но если вызвать метод через класс:

```python
Room.info(kitchen)
```

то объект нужно передать вручную. Эти два вызова **делают одно и то же**:

```python
kitchen.info()          # Python подставит kitchen в self
Room.info(kitchen)      # то же самое, но вручную
```

Понимание этого помогает понять, что `self` — это не магия, а обычный параметр, который Python подставляет автоматически.

---

## 1.7. Атрибуты объекта

**Атрибуты** — это данные, которые хранятся в объекте. Они определяются через `self` в конструкторе или в методах.

```python
class Room:
    def __init__(self, name, area, floor):
        self.name = name      # атрибут name
        self.area = area      # атрибут area
        self.floor = floor    # атрибут floor
```

Каждый объект имеет **свои значения** этих атрибутов:

```python
kitchen = Room("Кухня", 12, 2)
bedroom = Room("Спальня", 18, 2)

print(kitchen.name)    # Кухня
print(bedroom.name)    # Спальня
print(kitchen.area)    # 12
print(bedroom.area)    # 18
```

### Доступ к атрибутам

Снаружи класса к атрибутам обращаются через точку:

```python
print(kitchen.name)    # чтение
kitchen.area = 15      # изменение
```

Внутри класса — через `self`:

```python
def info(self):
    return self.name       # обращение к атрибуту через self
```

### Нюанс: атрибуты создаются в момент присваивания

Атрибут появляется у объекта **только тогда**, когда ему что-то присвоили. Если атрибут не установлен, обращение к нему вызовет `AttributeError`:

```python
class Room:
    def __init__(self, name):
        self.name = name

kitchen = Room("Кухня")
print(kitchen.area)      # AttributeError: 'Room' object has no attribute 'area'
```

Здесь `area` не был установлен в конструкторе — и его нет.

---

## 1.8. Методы класса

**Метод** — это функция, определённая внутри класса. Методы описывают **поведение** объекта: что он умеет делать.

```python
class Room:
    def __init__(self, name, area, floor):
        self.name = name
        self.area = area
        self.floor = floor

    def info(self):
        return f"{self.name} — {self.area} м², {self.floor} этаж"
```

Разберём подробно:

- `def info(self):` — объявление метода. Первый параметр — всегда `self`.
- `self.name` — обращение к атрибуту текущего объекта.
- `return` — возвращает строку.

### Вызов метода

```python
kitchen = Room("Кухня", 12, 2)
print(kitchen.info())    # Кухня — 12 м², 2 этаж
```

Python автоматически подставил `kitchen` в `self`.

### Нюанс: метод — это тоже атрибут

С точки зрения Python метод — это **атрибут класса**, значением которого является функция:

```python
print(Room.info)              # <function Room.info at 0x...>
print(kitchen.info)           # <bound method Room.info of <Room object at 0x...>>
```

Когда мы обращаемся к `kitchen.info`, Python создаёт «связанный метод» — функцию, в которую уже «вшит» объект `kitchen`. При вызове `kitchen.info()` этот объект автоматически передаётся в `self`.

### Метод vs функция

| | Функция | Метод |
|---|---------|-------|
| Где определена | Вне класса | Внутри класса |
| Первый параметр | Любой | Всегда `self` |
| Вызов | `info(kitchen)` | `kitchen.info()` |
| Доступ к данным | Через аргументы | Через `self` |

---

## 1.9. Как Python ищет атрибуты и методы

Когда мы пишем `kitchen.info()`, Python выполняет следующие шаги:

1. **Ищет `info` в самом объекте** `kitchen`. Если бы у `kitchen` был свой атрибут `info`, был бы использован он.
2. **Если не нашёл — ищет в классе** `Room`. Там метод `info` есть.
3. **Если не нашёл — ищет у родителей класса** (если есть наследование).
4. **Если не нашёл нигде — `AttributeError`**.

Такая последовательность называется **поиск по MRO** (Method Resolution Order) — мы разберём её подробно в главе 19.

### Пример

```python
class Room:
    type = "Комната"    # атрибут класса

    def __init__(self, name):
        self.name = name    # атрибут экземпляра

kitchen = Room("Кухня")
print(kitchen.name)   # Кухня — найден в самом объекте
print(kitchen.type)   # Комната — найден в классе
```

Это отличие между **атрибутами экземпляра** и **атрибутами класса** — мы разберём его подробно в главе 16.

---

## 1.10. Что происходит при вызове `Room("Кухня", 12, 2)`

Давай проследим **пошагово**, что делает Python:

1. **Создаётся новый пустой объект** — с помощью метода `__new__` (мы его разберём в главе 37).
2. **Вызывается `__init__`** — Python автоматически передаёт туда новый объект и наши аргументы:
   ```python
   Room.__init__(new_object, "Кухня", 12, 2)
   ```
3. **Внутри `__init__`** создаются атрибуты:
   - `new_object.name = "Кухня"`
   - `new_object.area = 12`
   - `new_object.floor = 2`
4. **`__init__` завершается**, возвращая `None` (это обязательно).
5. **Готовый объект возвращается** из `Room(...)` и присваивается переменной `kitchen`.

Понимание этого процесса помогает понять, почему `self` — это ссылка на объект, и почему `__init__` не возвращает значение.

---

## 1.11. Динамическое добавление атрибутов

В Python (в отличие от многих языков) атрибуты можно добавлять объекту **в любой момент**, а не только в конструкторе:

```python
kitchen = Room("Кухня", 12, 2)
kitchen.window_count = 2       # добавили новый атрибут
print(kitchen.window_count)    # 2
```

Это работает, но считается **плохой практикой**, потому что:

- Класс перестаёт быть «шаблоном» — объекты становятся разными по структуре.
- IDE не подскажет, какие атрибуты есть у объекта.
- Опечатки превращаются в баги: `kitchen.windows_count = 2` создаст **новый** атрибут, а не исправит старый.

**Правило:** все атрибуты объекта лучше определять в `__init__`. Тогда у всех объектов класса будет одинаковый набор полей.

---

## 1.12. `__dict__` — где хранятся атрибуты

Каждый объект хранит свои атрибуты в специальном словаре `__dict__`:

```python
kitchen = Room("Кухня", 12, 2)
print(kitchen.__dict__)
# {'name': 'Кухня', 'area': 12, 'floor': 2}
```

Это **не магия** — Python действительно хранит атрибуты как пары «ключ-значение». Обращение `kitchen.name` — это фактически поиск в этом словаре.

У класса тоже есть свой `__dict__` — там хранятся методы и атрибуты класса:

```python
print(Room.__dict__)
# {'__init__': ..., 'info': ..., '__dict__': ..., ...}
```

Понимание `__dict__` поможет в главе 21 (`__slots__`), где мы будем экономить память, отказываясь от этого словаря.

---

## 1.13. Типичные ошибки новичка

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Забыли `self` в параметрах метода | `def info():` | `def info(self):` |
| Забыли `self.` при обращении к атрибуту | `return name` | `return self.name` |
| Путают класс и объект | `Room.info()` | `kitchen.info()` |
| Неправильно пишут `__init__` | `_init_`, `init` | `__init__` |
| Возвращают значение из `__init__` | `return self.name` | убрать `return` |
| Передают `self` вручную при вызове | `kitchen.info(kitchen)` | `kitchen.info()` |
| Добавляют атрибуты вне `__init__` | `kitchen.x = 5` | добавить в `__init__` |
| Обращаются к атрибуту до его создания | `kitchen.area` до `__init__` | сначала создать, потом обращаться |
| Думают, что `self` — ключевое слово | называют как угодно | использовать `self` |

---

## 1.14. Что важно запомнить

- **Класс** — шаблон, **объект** — конкретное воплощение.
- **`__init__`** — конструктор, вызывается автоматически при создании объекта.
- **`__init__` не возвращает значение** — только инициализирует.
- **`self`** — ссылка на текущий объект, всегда первый параметр метода.
- **`self` — не ключевое слово**, а соглашение.
- **Из одного класса можно создать сколько угодно объектов** с разными данными.
- **Атрибуты объекта создаются в момент присваивания**, а не заранее.
- **Метод — это атрибут класса**, но с автоматической подстановкой `self`.
- **`__dict__`** — где реально хранятся атрибуты объекта.
- **Все атрибуты лучше определять в `__init__`**, а не добавлять динамически.

## Задача 1 (создание класса, конструктор, метод)

В доме ведётся учёт комнат. Создайте класс `Room` с полями `name` (название), `area` (площадь в м²) и `floor` (этаж). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о комнате в формате:

```
Название — площадь ... м², ... этаж
```

Создайте **два экземпляра** класса с разными данными, вызовите у каждого метод `info()` и выведите результат на экран. Ожидаемый вывод, например:

```
Кухня — площадь 12 м², 2 этаж
Спальня — площадь 18 м², 3 этаж
```

В комментариях поясните:

1. Чем класс отличается от объекта.
2. Зачем нужен параметр `self` в методах.
3. Что произойдёт, если забыть `self` при обращении к атрибуту.

### Решение

```python
class Room:
    def __init__(self, name, area, floor):
        # self — ссылка на конкретный объект.
        # Благодаря self каждая комната хранит свои значения.
        self.name = name
        self.area = area
        self.floor = floor

    def info(self):
        # Возвращаем строку. self.name — название именно этой комнаты.
        # Без self Python не понял бы, из какого объекта брать name.
        return f"{self.name} — площадь {self.area} м², {self.floor} этаж"


kitchen = Room("Кухня", 12, 2)
bedroom = Room("Спальня", 18, 3)

print(kitchen.info())
print(bedroom.info())
```

### Разбор

**1. Чем класс отличается от объекта.**
Класс `Room` — это описание, шаблон: «у любой комнаты есть название, площадь и этаж». Объекты `kitchen` и `bedroom` — конкретные комнаты с конкретными значениями. Из одного класса можно создать сколько угодно объектов с разными данными.

**2. Зачем нужен `self`.**
`self` — это ссылка на текущий объект. Когда мы вызываем `kitchen.info()`, Python автоматически передаёт в `self` объект `kitchen`. Без `self` метод не понял бы, к какой именно комнате относится `name`, `area`, `floor`.

**3. Что произойдёт, если забыть `self`.**
Если написать:

```python
def info(self):
    return f"{name} — площадь {area} м², {floor} этаж"
```

Python будет искать переменные `name`, `area`, `floor` **внутри метода**, не найдёт их и выдаст `NameError`. Атрибуты объекта доступны **только** через `self`: `self.name`, `self.area`, `self.floor`.

# Глава 2. Инкапсуляция и свойства

## 2.1. Проблема: данные доступны всем

В главе 1 мы создали класс `Room` с полями `name`, `area`, `floor`. Всё работает — но есть скрытая проблема: **данные объекта доступны из любого места программы**. Любой может случайно или намеренно изменить их на что угодно:

```python
kitchen = Room("Кухня", 12, 2)
kitchen.area = -12          # площадь стала отрицательной
kitchen.name = 12345        # название стало числом
kitchen.floor = "второй"    # этаж стал строкой
```

Программа не сломается сразу — но данные теперь бессмысленные. Отрицательная площадь, название-число, этаж-строка — с этим невозможно работать. А поймать такую ошибку сложно: она проявится гораздо позже, когда что-то пойдёт не так.

Такая ситуация называется **нарушением целостности данных**. Решение — **инкапсуляция**.

---

## 2.2. Что такое инкапсуляция

**Инкапсуляция** — это принцип объектно-ориентированного программирования, по которому:

1. **Данные объекта скрываются** от прямого доступа извне.
2. **Доступ к данным предоставляется** только через специальные механизмы.
3. **При изменении данных выполняется проверка** — можно ли записать такое значение.

Простыми словами: инкапсуляция — это **защита внутреннего состояния объекта**. Снаружи не видно, что у объекта внутри; снаружи можно только попросить: «прочитай мне это» или «запиши вот это» — и объект сам решит, можно ли это сделать.

### Аналогия

Представь банкомат. Ты не можешь залезть внутрь и взять деньги — там всё защищено. Но можешь **попросить** банкомат выдать сумму. Банкомат **сам проверит**, есть ли у тебя столько денег, и только потом выдаст.

Инкапсуляция работает так же: объект сам решает, что можно с ним сделать, а что — нет.

---

## 2.3. Три уровня доступа в Python

В Python есть **три уровня доступа** к атрибутам. Обозначаются они количеством подчёркиваний перед именем:

| Уровень | Синтаксис | Доступ | Когда использовать |
|---------|-----------|--------|-------------------|
| **Публичный** | `name` | Везде | Обычные данные |
| **Защищённый** | `_name` | Внутри класса и у наследников | Внутренние данные, но не строго скрытые |
| **Приватный** | `__name` | Только внутри класса | Чувствительные данные |

Разберём каждый подробно.

### Публичные атрибуты

Пишутся **без подчёркиваний**:

```python
self.name = name
```

Доступны откуда угодно:

```python
kitchen = Room("Кухня", 12, 2)
print(kitchen.name)      # Кухня — работает
kitchen.name = "Спальня" # работает
```

Такие атрибуты — часть «открытого интерфейса» класса. Использовать их можно свободно.

### Защищённые атрибуты

Начинаются с **одного подчёркивания**:

```python
self._floor = floor
```

Это **не запрет**, а **соглашение**: «не трогай без нужды, это внутреннее». Доступ снаружи **возможен**, но не рекомендуется:

```python
print(kitchen._floor)    # работает, но так делать не принято
```

Защищённые атрибуты используют, когда поле нужно наследникам класса, но не должно попадать в «публичный интерфейс».

### Приватные атрибуты

Начинаются с **двух подчёркиваний**:

```python
self.__area = area
```

Python **переименовывает** их (см. раздел 2.4), и снаружи обратиться напрямую **нельзя**:

```python
print(kitchen.__area)
# AttributeError: 'Room' object has no attribute '__area'
```

Приватные атрибуты — самые защищённые. Доступ к ним возможен **только внутри класса**.

### Важный нюанс

Строго говоря, **в Python нет настоящих приватных полей**. Даже `__area` можно достать через name mangling (см. 2.4). Но программисты договорились: если видишь `__name` — не лезь. Это осознанное решение, и обычно его уважают.

---

## 2.4. Name mangling — что это и зачем

**Name mangling** (искажение имени) — это механизм Python, при котором приватные атрибуты **автоматически переименовываются**. Если в классе `Room` есть поле `__area`, Python превращает его в `_Room__area`.

```python
class Room:
    def __init__(self, area):
        self.__area = area

kitchen = Room(12)

# print(kitchen.__area)              # AttributeError
print(kitchen._Room__area)           # 12 — работает!
```

### Как это работает

Когда Python видит `self.__area` внутри класса `Room`, он **автоматически** заменяет это на `self._Room__area`. То же самое происходит при чтении `self.__area`:

```python
class Room:
    def __init__(self, area):
        self.__area = area       # превратится в self._Room__area

    def show(self):
        return self.__area       # тоже self._Room__area
```

Снаружи класса `kitchen.__area` **не сработает**, потому что реальное имя — `_Room__area`, а не `__area`.

### Зачем это нужно

Name mangling нужен не для **безопасности**, а для **двух других целей**:

1. **Избежать конфликтов имён между родителем и наследником.**
   Если у родителя есть `self.__x`, а у наследника — тоже `self.__x`, без name mangling они бы перезаписывали друг друга. С ним — это `_Parent__x` и `_Child__x`, разные поля.

2. **Сигнализировать другим программистам:** «это внутреннее, не лезь».

### Ключевая мысль

Приватность в Python — **скорее договорённость, чем жёсткая защита**. Программисты договорились: если видишь два подчёркивания — не трогай. Но если очень нужно, обойти можно (через `_Class__name`), и это осознанное решение.

---

## 2.5. Как тогда работать с приватными данными?

Приватное поле `__area` снаружи недоступно. Но нам всё равно нужно **читать** площадь и **менять** её (с проверкой). Как быть?

Ответ: сделать **специальные методы**, через которые и будет идти доступ. Такие методы называют:

- **Геттерами** (от англ. *get* — «получать») — для **чтения**.
- **Сеттерами** (от англ. *set* — «устанавливать») — для **записи**.

Разберём **два способа** их организации:

1. **Классический** — обычные методы `get_` / `set_`.
2. **Современный** — свойства через `@property`.

Начнём с классического — он проще для понимания.

---

## 2.6. Способ 1. Классические геттеры и сеттеры

Идея простая: для каждого приватного поля пишем **два обычных метода**.

- **Сеттер:** `set_имя_поля(value)` — записывает значение (с проверкой).
- **Геттер:** `get_имя_поля()` — возвращает значение.

### Пример

```python
class Room:
    def __init__(self):
        self.__area = None

    def set_area(self, value):
        # Сеттер: проверяем и записываем.
        if 5 <= value <= 100:
            self.__area = value
        else:
            print("Недопустимая площадь")

    def get_area(self):
        # Геттер: возвращаем значение.
        return self.__area


kitchen = Room()
kitchen.set_area(12)       # записываем через сеттер
print(kitchen.get_area())  # 12 — читаем через геттер

kitchen.set_area(-5)       # Недопустимая площадь
print(kitchen.get_area())  # 12 — значение не изменилось
```

### Как это работает

1. Поле `__area` **приватное** — снаружи напрямую не доступно.
2. Чтобы записать — вызываем `set_area(12)`.
3. Чтобы прочитать — вызываем `get_area()`.
4. Внутри сеттера можно **проверить** значение и не записывать, если оно плохое.

### Плюсы классического подхода

- **Явно:** сразу видно, что это метод: `get_area()`, `set_area(12)`.
- **Понятно новичку:** это просто методы, ничего особенного.
- **Гибко:** можно делать любую логику внутри.

### Минусы классического подхода

- **Длинно:** `room.get_area()` вместо `room.area`.
- **Не похоже на обычное поле:** нельзя написать `room.area = 12`.
- **Много кода** для простых случаев.

### Типичная ошибка

Забыть, что поле приватное, и написать `room.__area = 12` — не сработает, потому что Python переименовал поле в `_Room__area`. Создастся **новый** атрибут `__area`, а приватное поле останется прежним:

```python
kitchen.__area = 999       # создастся НОВЫЙ атрибут __area, а не изменится приватный
print(kitchen.get_area())  # 12 — приватное поле не изменилось!
```

---

## 2.7. А можно ли сделать красивее?

Классический подход работает, но выглядит громоздко. Хочется, чтобы можно было писать так:

```python
kitchen.area = 12          # как обычное поле
print(kitchen.area)        # и читать как обычное поле
```

И при этом чтобы **проверка работала** и поле оставалось защищённым.

Именно для этого в Python придумали **свойства через `@property`**.

---

## 2.8. Способ 2. Свойства через `@property`

`@property` — это **декоратор**, который позволяет методу **притвориться обычным полем**. Снаружи кажется, что `area` — это просто атрибут, но за кулисами вызывается метод.

### Как объявить

```python
class Room:
    def __init__(self):
        self.__area = None

    @property
    def area(self):
        # Геттер: вызывается при чтении room.area.
        return self.__area

    @area.setter
    def area(self, value):
        # Сеттер: вызывается при room.area = value.
        if 5 <= value <= 100:
            self.__area = value
        else:
            print("Недопустимая площадь")
```

### Как использовать

```python
kitchen = Room()
kitchen.area = 12          # работает как обычное поле!
print(kitchen.area)        # 12 — читается как обычное поле

kitchen.area = -5          # Недопустимая площадь
print(kitchen.area)        # 12 — значение не изменилось
```

### Разбор по частям

**`@property` над методом `area`** — говорит Python: «при чтении `room.area` вызывай этот метод **без скобок**».

```python
@property
def area(self):
    return self.__area
```

**`@area.setter` над одноимённым методом** — говорит Python: «при записи `room.area = value` вызывай этот метод».

```python
@area.setter
def area(self, value):
    ...
```

Имя `area` **обязательно совпадает** в обоих случаях. Иначе свойство не свяжется с сеттером.

### Схема работы

```
room.area          →  вызывается геттер (@property)
room.area = 12     →  вызывается сеттер (@area.setter)
```

### Плюсы `@property`

- **Красиво:** `room.area` вместо `room.get_area()`.
- **Работает как обычное поле** — естественно и удобно.
- **Проверка работает** — снаружи не видно, но она есть.
- **Легко превратить в «только для чтения»** — достаточно не писать `.setter`.

### Минусы `@property`

- **Менее явно:** новичок не сразу поймёт, что за `room.area` скрывается метод.
- **Если свойств много** — класс становится длинным.

---

## 2.9. Сравнение двух подходов

Одна и та же задача — защита поля — решается двумя способами.

| Критерий | Классический (`get_`/`set_`) | `@property` |
|----------|------------------------------|-------------|
| Синтаксис чтения | `room.get_area()` | `room.area` |
| Синтаксис записи | `room.set_area(12)` | `room.area = 12` |
| Похоже на поле | ❌ Нет | ✅ Да |
| Явность | ✅ Очень явно | ⚠️ Менее явно |
| Современность | Старый стиль | Современный стиль |
| Количество кода | Больше | Меньше |
| Только для чтения | Через отсутствие сеттера | Через отсутствие `.setter` |

**В современном Python предпочитают `@property`.** Классические `get_`/`set_` встречаются в старом коде и в языках вроде Java.

---

## 2.10. Как сделать поле «только для чтения»

Иногда нужно, чтобы поле можно было **только читать**, но нельзя изменить. Например, номер технического паспорта комнаты после регистрации менять нельзя.

### Классический способ

Не писать сеттер:

```python
class RoomPassport:
    def __init__(self, passport_id):
        self.__passport_id = passport_id

    def get_passport_id(self):
        return self.__passport_id

    # set_passport_id не определяем — изменить нельзя


passport = RoomPassport("77:01:0001001:1234")
print(passport.get_passport_id())   # 77:01:0001001:1234
# passport.set_passport_id("...")   # AttributeError
```

### Через `@property`

Определить только геттер, без `.setter`:

```python
class RoomPassport:
    def __init__(self, passport_id):
        self.__passport_id = passport_id

    @property
    def passport_id(self):
        return self.__passport_id

    # @passport_id.setter не определяем


passport = RoomPassport("77:01:0001001:1234")
print(passport.passport_id)     # 77:01:0001001:1234

passport.passport_id = "00:00:0000000:0000"
# AttributeError: can't set attribute
```

Попытка присвоить значение вызовет `AttributeError` — это и есть защита «только для чтения».

---

## 2.11. Сеттер с полной валидацией

Сеттер — идеальное место для проверки данных. Можно не только ограничить диапазон, но и проверить **тип**.

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area       # вызовет сеттер при создании

    @property
    def area(self):
        return self.__area

    @area.setter
    def area(self, value):
        if not isinstance(value, (int, float)):
            raise TypeError("Площадь должна быть числом")
        if not 5 <= value <= 100:
            raise ValueError("Площадь должна быть от 5 до 100 м²")
        self.__area = value
```

Теперь **любое** присваивание — хоть в конструкторе, хоть снаружи — проходит через проверку:

```python
Room("Кухня", 12)      # ок
Room("Кухня", -5)      # ValueError: Площадь должна быть от 5 до 100 м²
Room("Кухня", "12")    # TypeError: Площадь должна быть числом
```

Это одна из главных причин использовать `@property`.

### Важный нюанс

Обрати внимание: в конструкторе мы пишем `self.area = area`, а **не** `self.__area = area`. Это специально — чтобы сработал сеттер и проверил значение.

```python
def __init__(self, name, area):
    self.name = name
    self.area = area       # ← сработает сеттер, проверит
    # self.__area = area   # ← так сеттер НЕ сработает, проверки не будет
```

Если написать `self.__area = area`, значение запишется напрямую, минуя проверку. Тогда можно случайно создать комнату с отрицательной площадью, если передать плохое значение в конструктор.

---

## 2.12. Когда что выбирать

| Ситуация | Рекомендация |
|----------|--------------|
| Простое публичное поле без проверок | Обычный атрибут (`self.name = name`) |
| Нужна проверка при записи | `@property` + `.setter` |
| Поле только для чтения | Только `@property`, без сеттера |
| Старый проект / обучение | Классические `get_`/`set_` |
| Новый проект | `@property` |
| Много простых полей без проверок | Обычные атрибуты |

---

## 2.13. Типичные ошибки

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Пишут одно подчёркивание вместо двух | `_area` вместо `__area` | Использовать `__area` для приватного |
| Думают, что приватность — реальная защита | В Python её можно обойти | Помнить про name mangling |
| Путают имя свойства и имя поля | В `@property def area` имя `area` — публичное, поле `__area` — приватное | Различать |
| Забывают `@area.setter` | Свойство работает только на чтение | Добавить `.setter` |
| Пишут `self.__area = value` в конструкторе при `@property` | Сеттер не сработает — проверка пропустится | Писать `self.area = value` |
| Используют name mangling снаружи класса | Плохая практика | Не трогать приватные поля снаружи |
| Делают `set_area` и одновременно `@property def area` | Дублирование — выберите один способ | Выбрать один способ |
| Забывают, что `@property` вызывается без скобок | `room.area()` — ошибка | `room.area` без скобок |

---

## 2.14. Что важно запомнить

- **Инкапсуляция** — сокрытие данных и доступ через свойства.
- **Три уровня доступа:** `name` (публичный), `_name` (защищённый), `__name` (приватный).
- **Name mangling:** `__name` превращается в `_Class__name`.
- **Приватность в Python** — договорённость, а не жёсткая защита.
- **Два способа доступа к приватным данным:**
  1. **Классический** — методы `get_` / `set_`.
  2. **Современный** — `@property` + `.setter`.
- **`@property` делает код похожим** на работу с обычным полем: `room.area` вместо `room.get_area()`.
- **Без `.setter`** свойство становится **только для чтения**.
- **Сеттер — идеальное место для валидации** данных.
- **В современном Python предпочитают `@property`**.

## Задача 2 (инкапсуляция и свойства)

У каждой комнаты есть технический паспорт с номером, который нельзя менять или читать напрямую.

**Часть 1.** Реализуйте **классический** подход: метод `set_passport_id(value)` — только для установки номера при регистрации, и метод `get_passport_id()` — для чтения. Повторная попытка изменить номер после регистрации должна игнорироваться, а не вызывать ошибку.

**Часть 2.** Реализуйте **второй вариант** того же класса — через `@property`. Сделайте так, чтобы поле `passport_id` было доступно только для чтения (геттер есть, сеттера нет).

**Часть 3.** Продемонстрируйте оба подхода: создайте запись, установите номер, попробуйте изменить его ещё раз. Дополнительно покажите, что снаружи нельзя напрямую прочитать `passport.__passport_id` — будет ошибка `AttributeError`, но через name mangling (`passport._RoomPassport__passport_id`) доступ всё же существует.

В комментариях объясните разницу между классическим подходом и `@property`: в чём плюсы и минусы каждого.

### Решение. Часть 1 — классический подход

```python
class RoomPassport:
    def __init__(self):
        # Приватное поле: снаружи не читается напрямую.
        self.__passport_id = None

    def set_passport_id(self, value):
        # Устанавливаем номер только если он ещё не установлен.
        # Повторная попытка игнорируется без ошибки.
        if self.__passport_id is None:
            self.__passport_id = value

    def get_passport_id(self):
        return self.__passport_id


passport = RoomPassport()
passport.set_passport_id("77:01:0001001:1234")
print("Номер после регистрации:", passport.get_passport_id())

passport.set_passport_id("00:00:0000000:0000")
print("Номер после повторной попытки:", passport.get_passport_id())

# Проверяем, что прямой доступ к приватному полю невозможен.
try:
    print(passport.__passport_id)
except AttributeError as e:
    print("Прямой доступ невозможен:", e)

# Но через name mangling — доступ есть.
print("Через name mangling:", passport._RoomPassport__passport_id)
```

**Ожидаемый вывод:**

```
Номер после регистрации: 77:01:0001001:1234
Номер после повторной попытки: 77:01:0001001:1234
Прямой доступ невозможен: 'RoomPassport' object has no attribute '__passport_id'
Через name mangling: 77:01:0001001:1234
```

### Решение. Часть 2 — через `@property`

```python
class RoomPassport:
    def __init__(self, passport_id):
        # Приватное поле — устанавливается один раз при создании.
        self.__passport_id = passport_id

    @property
    def passport_id(self):
        # Только геттер. Сеттера нет — поле только для чтения.
        return self.__passport_id


passport = RoomPassport("77:01:0001001:1234")
print("Номер паспорта:", passport.passport_id)

# Попытка изменить — вызовет AttributeError.
try:
    passport.passport_id = "00:00:0000000:0000"
except AttributeError as e:
    print("Изменение невозможно:", e)

# Проверяем name mangling.
print("Через name mangling:", passport._RoomPassport__passport_id)
```

**Ожидаемый вывод:**

```
Номер паспорта: 77:01:0001001:1234
Изменение невозможно: can't set attribute 'passport_id'
Через name mangling: 77:01:0001001:1234
```

### Разбор

**Классический подход** (`get_` / `set_`) — **явный**, но **длинный**. Чтение выглядит как вызов метода: `passport.get_passport_id()`. Легко понять новичку: это просто методы. Минус — громоздко, если полей много.

**Подход через `@property`** — **современный** и **красивый**. Чтение выглядит как обращение к полю: `passport.passport_id`. Снаружи кажется, что это обычное поле, но за кулисами работает метод. Если сеттер не определён — поле становится **только для чтения**, и попытка присвоить значение вызовет `AttributeError`.

**Почему повторная попытка в первом варианте игнорируется, а во втором — выбрасывает ошибку?**
В первом варианте мы **сами написали** такую логику: «если номер уже установлен — ничего не делать». Это осознанное решение. Во втором варианте сеттера вообще нет, поэтому Python **сам** не даёт изменить поле — это уже другой механизм.

**Оба подхода решают одну задачу** — защищают приватное поле. В современном Python предпочитают `@property`, потому что код получается чище и естественнее.

# Глава 3. Наследование

## 3.1. Зачем нужно наследование

Вернёмся к нашему дому. У нас есть класс `Room` с полями `name`, `area`, `floor`. Теперь мы хотим описать разные типы комнат: кухню, спальню, ванную. У каждой — свои особенности:

- **Кухня:** есть плита, холодильник.
- **Спальня:** есть кровать, шкаф.
- **Ванная:** есть ванна, раковина.

Первое, что приходит в голову — создать три отдельных класса:

```python
class Kitchen:
    def __init__(self, name, area, floor):
        self.name = name
        self.area = area
        self.floor = floor

    def info(self):
        return f"{self.name} — {self.area} м², {self.floor} этаж"

class Bedroom:
    def __init__(self, name, area, floor):
        self.name = name
        self.area = area
        self.floor = floor

    def info(self):
        return f"{self.name} — {self.area} м², {self.floor} этаж"

class Bathroom:
    def __init__(self, name, area, floor):
        self.name = name
        self.area = area
        self.floor = floor

    def info(self):
        return f"{self.name} — {self.area} м², {self.floor} этаж"
```

Уже здесь видно проблему: **код дублируется**. Конструктор и метод `info` — одинаковые во всех трёх классах. Если завтра нужно будет добавить поле «тип отделки», придётся править в трёх местах.

А теперь представь, что комнат не три, а двадцать. Или что у каждой комнаты есть десять общих методов. Дублирование превращается в катастрофу: любое изменение нужно вносить везде, и легко где-то забыть.

**Наследование решает эту проблему.** Мы описываем общее один раз — в родительском классе, а наследники получают это общее и добавляют своё.

---

## 3.2. Что такое наследование

**Наследование** — это механизм, позволяющий создать новый класс **на основе уже существующего**. Новый класс автоматически получает **все публичные атрибуты и методы** родителя, а также может:

- **добавлять** свои методы и атрибуты,
- **изменять** (переопределять) унаследованные методы.

Простыми словами: наследование — это способ **переиспользовать код** и описывать отношения «**является**».

### Аналогия

Кухня **является** комнатой. У неё есть всё, что есть у комнаты (название, площадь, этаж), плюс свои особенности (плита, холодильник). Поэтому логично, что класс `Kitchen` наследует от класса `Room`.

---

## 3.3. Терминология

В наследовании используются разные названия для одних и тех же понятий:

| Понятие | Синонимы |
|---------|----------|
| **Родительский класс** | Суперкласс, базовый класс (base class), parent class |
| **Дочерний класс** | Наследник, подкласс (subclass), производный класс (derived class), child class |

В этой книге мы будем использовать **родительский** и **дочерний** — так проще.

### Схема

```
        Room  (родительский класс)
        ├── Kitchen   (дочерний класс)
        ├── Bedroom   (дочерний класс)
        └── Bathroom  (дочерний класс)
```

---

## 3.4. Синтаксис наследования

После имени дочернего класса в **круглых скобках** указывается родитель:

```python
class Room:
    def get_type(self):
        return "Комната"

class Kitchen(Room):
    pass
```

Здесь `Kitchen` наследует от `Room`. Запись `class Kitchen(Room)` читается так: «класс `Kitchen` — наследник класса `Room`».

### Что теперь умеет `Kitchen`

Даже если внутри `Kitchen` ничего не написано (`pass`), он **уже умеет** всё, что умеет `Room`:

```python
kitchen = Kitchen()
print(kitchen.get_type())   # Комната — метод унаследован от Room
```

Это ключевая идея: **наследник получает функционал родителя бесплатно**.

---

## 3.5. Что наследуется, а что нет

Не всё наследуется одинаково. Разберём подробно.

| Что | Наследуется? | Пояснение |
|-----|--------------|-----------|
| **Публичные методы** | ✅ Да | `get_type`, `info` и т. д. |
| **Защищённые методы (`_name`)** | ✅ Да | По договорённости доступны наследникам |
| **Приватные методы (`__name`)** | ⚠️ Формально нет | Но доступны через name mangling |
| **Публичные атрибуты класса** | ✅ Да | Атрибуты, определённые на уровне класса |
| **Атрибуты экземпляра** | ❌ Нет | Они создаются в конструкторе родителя; если не вызвать `super().__init__()`, их не будет |
| **Конструктор `__init__`** | ❌ Не наследуется автоматически | Нужно вызывать через `super()` |
| **Магические методы** | ✅ Да | `__str__`, `__eq__` и т. д. |

### Нюанс: приватные атрибуты родителя

Приватные атрибуты родителя (`__name`) **не видны** в наследнике под тем же именем. Это связано с name mangling: `__name` в классе `Room` превращается в `_Room__name`. В классе `Kitchen` тот же `__name` превратится в `_Kitchen__name` — а это **другое поле**.

```python
class Room:
    def __init__(self, name):
        self.__name = name       # превратится в _Room__name

class Kitchen(Room):
    def show(self):
        return self.__name       # превратится в _Kitchen__name — ошибка!
```

Здесь `Kitchen().show()` вызовет `AttributeError`, потому что `_Kitchen__name` не существует — существует `_Room__name`.

**Как обойти:** использовать **защищённые** атрибуты (`_name`) вместо приватных, если поле должно быть доступно наследникам.

```python
class Room:
    def __init__(self, name):
        self._name = name        # защищённое — доступно наследникам

class Kitchen(Room):
    def show(self):
        return self._name        # работает
```

---

## 3.6. Переопределение методов

Часто наследник хочет вести себя **не так, как родитель**. Для этого метод **переопределяют**: в дочернем классе объявляют метод **с тем же именем** — и он **замещает** родительский.

```python
class Room:
    def get_type(self):
        return "Комната"

class Kitchen(Room):
    def get_type(self):
        return "Кухня"

class Bedroom(Room):
    def get_type(self):
        return "Спальня"
```

Теперь:

```python
print(Room().get_type())       # Комната
print(Kitchen().get_type())    # Кухня
print(Bedroom().get_type())    # Спальня
```

Python ищет метод **сначала в самом классе**. Если находит — использует его, а не родительский.

### Как Python ищет метод

Когда вызывается `kitchen.get_type()`, Python:

1. Ищет `get_type` в классе `Kitchen`. **Находит** — использует его.
2. Если бы не нашёл — пошёл бы к родителю `Room`.
3. Если бы не нашёл и там — пошёл бы к `object` (базовый класс всех классов).
4. Если не нашёл нигде — `AttributeError`.

Этот порядок называется **MRO** (Method Resolution Order). Мы разберём его подробно в главе 19.

### Нюанс: переопределение vs перегрузка

| Понятие | Что это | Есть в Python? |
|---------|---------|----------------|
| **Переопределение (override)** | Наследник заменяет метод родителя | ✅ Да |
| **Перегрузка (overload)** | Несколько методов с одним именем, но разными параметрами | ❌ Нет |

Это разные вещи! **Переопределение** — про наследование: два класса, один метод, разные реализации. **Перегрузка** — про параметры: один класс, одно имя, разные сигнатуры. В Python перегрузки нет — она имитируется через значения по умолчанию (см. главу 10).

---

## 3.7. `super()` — вызов родительской версии

Иногда нужно **не заменить** родительский метод, а **дополнить** его. Например, у комнаты есть метод `info()`, а у кухни хочется **добавить** информацию о плите.

Для этого используют функцию **`super()`**. Она возвращает ссылку на родительский класс, и через неё можно вызвать родительскую версию метода.

```python
class Room:
    def info(self):
        return "Комната"

class Kitchen(Room):
    def info(self):
        parent_info = super().info()       # вызываем версию Room
        return f"{parent_info} (с плитой)"
```

Теперь:

```python
print(Kitchen().info())    # Комната (с плитой)
```

Мы **не переписали** логику родителя, а **взяли её и дополнили**.

### Почему не написать `Room.info(self)`

Можно и так:

```python
class Kitchen(Room):
    def info(self):
        parent_info = Room.info(self)      # работает
        return f"{parent_info} (с плитой)"
```

Но это **плохая практика**. Если завтра класс `Kitchen` станет наследником другого класса (не `Room`), придётся менять код. `super()` **автоматически** находит родителя — код становится гибче.

**Правило:** всегда используйте `super()`, а не прямое обращение к родителю по имени.

---

## 3.8. `super()` в конструкторе

Это **самый важный** случай использования `super()`. Разберём его подробно.

### Проблема

Представь, у нас есть `Room` с конструктором и `Kitchen`, который тоже имеет конструктор:

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area

class Kitchen(Room):
    def __init__(self, name, area, has_fridge):
        self.has_fridge = has_fridge
```

Что произойдёт при создании?

```python
kitchen = Kitchen("Кухня", 12, True)
print(kitchen.name)     # AttributeError!
```

`kitchen.name` не существует, потому что конструктор `Kitchen` **не вызвал** конструктор `Room`. Атрибуты `name` и `area` так и не были установлены.

### Решение — `super().__init__()`

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area

class Kitchen(Room):
    def __init__(self, name, area, has_fridge):
        super().__init__(name, area)      # ← вызываем родительский конструктор
        self.has_fridge = has_fridge
```

Теперь всё работает:

```python
kitchen = Kitchen("Кухня", 12, True)
print(kitchen.name)         # Кухня
print(kitchen.area)         # 12
print(kitchen.has_fridge)   # True
```

### Схема работы

```
Kitchen("Кухня", 12, True)
    ↓
Kitchen.__init__(self, "Кухня", 12, True)
    ↓
super().__init__("Кухня", 12)   ← вызов Room.__init__
    ↓
Room.__init__(self, "Кухня", 12)
    ↓
self.name = "Кухня"
self.area = 12
    ↓
возврат в Kitchen.__init__
    ↓
self.has_fridge = True
```

### Типичная ошибка

**Забыть `super().__init__()`** — самая частая ошибка при работе с наследованием. Признак: обращение к атрибутам родителя вызывает `AttributeError`.

```python
class Kitchen(Room):
    def __init__(self, name, area, has_fridge):
        # super().__init__(name, area)   ← забыли!
        self.has_fridge = has_fridge

kitchen = Kitchen("Кухня", 12, True)
print(kitchen.name)    # AttributeError: 'Kitchen' object has no attribute 'name'
```

### Нюанс: где вызывать `super().__init__()`

Порядок в конструкторе наследника важен:

1. **Сначала** `super().__init__(...)` — инициализируем родительскую часть.
2. **Потом** — свои атрибуты.

```python
def __init__(self, name, area, has_fridge):
    super().__init__(name, area)   # сначала родитель
    self.has_fridge = has_fridge   # потом свои
```

Если сделать наоборот — работать будет, но логика ломается: сначала «свои» поля, потом «родительские». Это плохо для читаемости.

---

## 3.9. Множественное наследование

Python поддерживает **множественное наследование** — класс может наследоваться от **нескольких** родителей одновременно.

```python
class Employee:
    def work(self):
        return "Работает"

class Student:
    def study(self):
        return "Учится"

class WorkingStudent(Employee, Student):
    pass
```

Теперь `WorkingStudent` умеет и `work()`, и `study()`:

```python
ws = WorkingStudent()
print(ws.work())     # Работает
print(ws.study())    # Учится
```

### Схема

```
    Employee   Student
         \      /
          \    /
      WorkingStudent
```

### Проблема: конфликты имён

Если **оба** родителя имеют метод с одинаковым именем, возникает вопрос: какой использовать?

```python
class Employee:
    def do(self):
        return "Работает"

class Student:
    def do(self):
        return "Учится"

class WorkingStudent(Employee, Student):
    pass

ws = WorkingStudent()
print(ws.do())    # Работает — метод взят из Employee
```

Python использует **первый** родитель в списке. Это связано с **MRO** — порядком разрешения методов.

### Как посмотреть порядок

У каждого класса есть атрибут `__mro__` (или метод `mro()`):

```python
print(WorkingStudent.__mro__)
# (<class 'WorkingStudent'>, <class 'Employee'>, <class 'Student'>, <class 'object'>)
```

Это означает: сначала ищем в `WorkingStudent`, потом в `Employee`, потом в `Student`, потом в `object`.

### Множественное наследование — осторожно!

Множественное наследование — **мощный, но опасный** инструмент. Оно часто приводит к путанице:

- Какой метод вызван, если он есть у нескольких родителей?
- Как работает `super()` при нескольких родителях?
- Легко получить «ромбовидное» наследование и запутаться.

Поэтому множественное наследование используют **редко** и **осторожно**. Чаще всего — для **миксинов** (см. главу 20). В большинстве случаев **композиция** предпочтительнее (см. главу 8).

---

## 3.10. Наследование vs композиция

Наследование — это отношение «**является**»:
- Кухня **является** комнатой → `class Kitchen(Room)`.
- Спальня **является** комнатой → `class Bedroom(Room)`.

Композиция — это отношение «**содержит**»:
- Дом **содержит** комнаты → `House` имеет поле `rooms`.

### Как выбрать?

| Ситуация | Что использовать |
|----------|------------------|
| «A является B» | Наследование |
| «A содержит B» | Композиция |
| Нужно переиспользовать код | Обычно композиция |
| Есть чёткая иерархия «является» | Наследование |

**Правило:** «**предпочитайте композицию наследованию**». Это один из принципов хорошего дизайна (мы разберём его в главе 28).

### Почему композиция предпочтительнее

- Наследование создаёт **жёсткую связь** между классами: изменение родителя ломает наследника.
- Наследование легко использовать **неправильно**: например, сделать `Car` наследником `Engine`, хотя «машина не является двигателем».
- Композиция гибче: можно заменить внутренний объект без изменения класса.

---

## 3.11. `isinstance()` и `issubclass()`

Python предоставляет две полезные функции для работы с типами.

### `isinstance(obj, Class)`

Проверяет, является ли объект экземпляром класса (или его наследником):

```python
kitchen = Kitchen("Кухня", 12)

print(isinstance(kitchen, Kitchen))   # True
print(isinstance(kitchen, Room))      # True — Kitchen наследует Room
print(isinstance(kitchen, str))       # False
```

Обрати внимание: `isinstance` возвращает `True` **и для наследников**. `kitchen` — это и `Kitchen`, и `Room`.

### `issubclass(Child, Parent)`

Проверяет, является ли класс наследником другого класса:

```python
print(issubclass(Kitchen, Room))      # True
print(issubclass(Room, Kitchen))      # False — Room не наследник Kitchen
print(issubclass(Kitchen, Kitchen))   # True — класс является наследником самого себя
```

### Зачем это нужно

Эти функции полезны, когда нужно **по-разному обрабатывать объекты разных типов**:

```python
def describe(item):
    if isinstance(item, Kitchen):
        print("Это кухня")
    elif isinstance(item, Room):
        print("Это комната")
```

Мы разберём это подробнее в главе 4 (Полиморфизм).

---

## 3.12. Типичные ошибки

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Забыли `super().__init__()` | Родительские атрибуты не создаются | Добавить вызов в начало конструктора |
| Вызывают `super().__init__()` **после** своих полей | Работает, но нелогично | Вызывать **первым** |
| Обращаются к `self.__name` в наследнике | Приватное поле родителя недоступно | Использовать `_name` (защищённое) |
| Наследуют ради переиспользования кода | Нарушает логику «является» | Использовать композицию |
| Путают override и overload | Это разные вещи | Override — переопределение; overload — перегрузка |
| Переопределяют метод с другим именем | Это новый метод, а не переопределение | Имя должно совпадать |
| Используют `Room.info(self)` вместо `super()` | Жёсткая связь | Использовать `super()` |
| Злоупотребляют множественным наследованием | Сложно понять, какой метод вызван | Использовать композицию или миксины |

---

## 3.13. Что важно запомнить

- **Наследование** — создание нового класса на основе существующего.
- **Синтаксис:** `class Child(Parent):`.
- **Наследник получает** все публичные атрибуты и методы родителя.
- **Отношение «является»** — основание для наследования.
- **Переопределение** — своя версия метода родителя (имя совпадает).
- **`super()`** — вызов родительской версии метода.
- **`super().__init__()` обязателен**, если у родителя есть конструктор с параметрами.
- **Приватные поля родителя** недоступны наследнику напрямую.
- **Множественное наследование** — мощно, но опасно.
- **Композиция** часто предпочтительнее наследования.
- **`isinstance()`** — проверка объекта; **`issubclass()`** — проверка класса.

---

## Задача 3 (наследование)

В доме есть разные типы комнат. Создайте базовый класс `Room` с полями `name` (название) и `area` (площадь), а также методом `get_type()`, возвращающим строку «Комната», и методом `info()`, возвращающим строку вида `Название — площадь ... м²`.

**Часть 1.** Унаследуйте от `Room` два класса:

- `Kitchen` — с дополнительным полем `has_fridge` (есть ли холодильник) и переопределённым методом `get_type()`, возвращающим «Кухня».
- `Bedroom` — с дополнительным полем `bed_size` (размер кровати) и переопределённым методом `get_type()`, возвращающим «Спальня».

**Часть 2.** В каждом наследнике реализуйте конструктор с вызовом `super().__init__(...)`, чтобы родительские поля `name` и `area` правильно инициализировались.

**Часть 3.** Переопределите в `Kitchen` метод `info()` так, чтобы он **дополнял** родительский: возвращал строку вида `Кухня — площадь 12 м² (холодильник есть)`. Используйте `super().info()` для получения родительской строки.

**Часть 4.** Создайте по одному объекту `Kitchen` и `Bedroom`. Выведите их тип через `get_type()` и информацию через `info()`. Проверьте через `isinstance()`, что объект `Kitchen` является и `Kitchen`, и `Room`.

**Часть 5.** В комментариях объясните:

1. Что произойдёт, если в `Kitchen.__init__` не вызвать `super().__init__()`.
2. Чем переопределение отличается от перегрузки.
3. Почему `super().info()` лучше, чем `Room.info(self)`.

### Решение

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area

    def get_type(self):
        return "Комната"

    def info(self):
        return f"{self.name} — площадь {self.area} м²"


class Kitchen(Room):
    def __init__(self, name, area, has_fridge):
        # Вызываем родительский конструктор, чтобы установить name и area.
        # Без этой строки self.name и self.area не будут созданы.
        super().__init__(name, area)
        self.has_fridge = has_fridge

    def get_type(self):
        return "Кухня"

    def info(self):
        # Дополняем родительскую строку, а не переписываем её заново.
        parent_info = super().info()
        fridge = "холодильник есть" if self.has_fridge else "холодильника нет"
        return f"{parent_info} ({fridge})"


class Bedroom(Room):
    def __init__(self, name, area, bed_size):
        super().__init__(name, area)
        self.bed_size = bed_size

    def get_type(self):
        return "Спальня"


# Создаём объекты.
kitchen = Kitchen("Кухня", 12, True)
bedroom = Bedroom("Спальня", 18, "King Size")

# Проверяем типы.
print(kitchen.get_type())    # Кухня
print(bedroom.get_type())    # Спальня

# Проверяем info.
print(kitchen.info())        # Кухня — площадь 12 м² (холодильник есть)
print(bedroom.info())        # Спальня — площадь 18 м²

# Проверяем isinstance.
print(isinstance(kitchen, Kitchen))   # True
print(isinstance(kitchen, Room))      # True — Kitchen наследует Room
```

**Ожидаемый вывод:**

```
Кухня
Спальня
Кухня — площадь 12 м² (холодильник есть)
Спальня — площадь 18 м²
True
True
```

### Разбор

**1. Что произойдёт, если в `Kitchen.__init__` не вызвать `super().__init__()`.**

Атрибуты `name` и `area` не будут созданы. При попытке обратиться к `kitchen.name` Python выдаст `AttributeError: 'Kitchen' object has no attribute 'name'`. Это самая частая ошибка при работе с наследованием.

**2. Чем переопределение отличается от перегрузки.**

**Переопределение (override)** — это когда **наследник** даёт свою версию метода **родителя**. Имя метода совпадает, но реализация другая. Работает благодаря наследованию: два класса, один метод, разные версии.

**Перегрузка (overload)** — это когда **в одном классе** есть несколько методов **с одним именем**, но **разными параметрами**. В Python такого нет: имя метода — это один объект-функция, объявить две функции с одним именем нельзя. Перегрузку имитируют через значения по умолчанию (см. главу 10).

**3. Почему `super().info()` лучше, чем `Room.info(self)`.**

`super()` **автоматически** находит родительский класс. Если завтра иерархия изменится — например, `Kitchen` станет наследником не `Room`, а `LivingRoom` — код с `super()` продолжит работать, а код с `Room.info(self)` придётся переписывать. `super()` делает код **гибче** и **устойчивее к изменениям**.

Кроме того, `super()` корректно работает при **множественном наследовании** — он идёт по MRO. Прямое обращение `Room.info(self)` этого не умеет.

# Глава 4. Полиморфизм

## 4.1. Что такое полиморфизм

**Полиморфизм** — это способность разных объектов **реагировать на один и тот же вызов по-разному**. Слово составное: «поли» — много, «морфе» — форма. Буквально: «много форм».

Простыми словами: если у объектов разных классов есть метод **с одинаковым именем**, то вызов этого метода будет приводить к **разным результатам** — в зависимости от того, объект какого класса вызван.

### Аналогия

Представь, что ты даёшь команду «убраться» разным людям:

- **Повар** поймёт это как «помыть плиту и стол».
- **Горничная** — как «пропылесосить и заправить кровать».
- **Сантехник** — как «почистить плитку и раковину».

Команда одна — «убраться». Но каждый делает **своё**. Это и есть полиморфизм: **один вызов — много форм поведения**.

---

## 4.2. Зачем нужен полиморфизм

Без полиморфизма нам пришлось бы писать **разный код** для каждого типа:

```python
def clean_kitchen(kitchen):
    print("Моем плиту и стол")

def clean_bedroom(bedroom):
    print("Пылесосим и заправляем кровать")

def clean_bathroom(bathroom):
    print("Чистим плитку и сантехнику")
```

А теперь представь, что комнат **десять типов**. Придётся писать десять функций. А если появится одиннадцатый тип — снова новая функция и снова правки во всех местах, где вызывается уборка.

**С полиморфизмом** пишем **одну** функцию:

```python
def clean(room):
    room.clean()
```

И она работает со **всеми** типами комнат. Функции **всё равно**, какая именно комната — кухня, спальня или ванная. Ей важно только, что у комнаты есть метод `clean()`.

Это ключевая идея: **полиморфизм позволяет писать общий код, который работает с разными типами объектов**.

---

## 4.3. Пример из жизни

Вернёмся к нашему дому. У нас есть классы `Kitchen`, `Bedroom`, `Bathroom`. У каждой комнаты — свой способ уборки:

```python
class Kitchen:
    def clean(self):
        return "Моем плиту и стол"

class Bedroom:
    def clean(self):
        return "Пылесосим и заправляем кровать"

class Bathroom:
    def clean(self):
        return "Чистим плитку и сантехнику"
```

Теперь напишем **одну** функцию, которая убирает **любую** комнату:

```python
def clean_room(room):
    print(f"Уборка: {room.clean()}")
```

Вызовем её с объектами разных классов:

```python
clean_room(Kitchen())     # Уборка: Моем плиту и стол
clean_room(Bedroom())     # Уборка: Пылесосим и заправляем кровать
clean_room(Bathroom())    # Уборка: Чистим плитку и сантехнику
```

**Один и тот же вызов** `room.clean()` — **три разных результата**. Это полиморфизм.

---

## 4.4. Как Python определяет, какой метод вызвать

Это **ключевой момент**, который часто упускают. Давай разберём пошагово.

Когда Python видит выражение:

```python
room.clean()
```

Он выполняет следующие действия:

1. **Смотрит на фактический класс объекта** `room`. Если `room` — объект класса `Kitchen`, Python запомнит это.
2. **Ищет метод `clean`** в классе `Kitchen`.
3. **Если находит** — вызывает его.
4. **Если не находит** — идёт к родителям класса (по MRO, см. главу 19).
5. **Если не находит нигде** — `AttributeError`.

Всё решается **во время выполнения**, по фактическому типу объекта. Это называется **динамическое связывание** (dynamic binding) или **позднее связывание** (late binding).

### Схема работы

```
room.clean()
    ↓
Определяем фактический класс room → Kitchen
    ↓
Ищем clean в Kitchen → нашли
    ↓
Вызываем Kitchen.clean()
```

### Почему это важно

Функция `clean_room` **не знает заранее**, какой объект ей передадут. Она просто вызывает `room.clean()`. Python сам разберётся, какой метод использовать. Это делает код **универсальным**.

### Сравнение с другими языками

В **статически типизированных** языках (Java, C++) метод выбирается **во время компиляции** или через **виртуальные таблицы**. В Python — **во время выполнения**. Это даёт гибкость, но цена — ошибки видны только в момент вызова.

---

## 4.5. Полиморфизм через наследование

Первый способ полиморфизма — через **наследование**. Все классы наследуют общий родитель с методом:

```python
class Room:
    def clean(self):
        return "Убираем комнату"

class Kitchen(Room):
    def clean(self):
        return "Моем плиту и стол"

class Bedroom(Room):
    def clean(self):
        return "Пылесосим и заправляем кровать"
```

Здесь `Kitchen` и `Bedroom` — наследники `Room`. Они **переопределяют** метод `clean`. Функция `clean_room(room)` работает с любым из них.

### Схема

```
        Room
        ├── Kitchen   (переопределяет clean)
        └── Bedroom   (переопределяет clean)
```

### Плюсы

- **Явная иерархия** — видно, что все комнаты «являются» комнатами.
- **Общий интерфейс** — родитель гарантирует, что метод есть у всех.
- **Работает с `isinstance`** — можно проверить тип.

### Минусы

- **Жёсткая связь** — все классы должны наследовать общего родителя.
- **Ограничение** — нельзя добавить «чужой» класс, который не наследует `Room`.

---

## 4.6. Полиморфизм через утиную типизацию

Python поддерживает **второй способ** полиморфизма — **без наследования**. Достаточно, чтобы у объекта был **нужный метод**.

```python
class Kitchen:
    def clean(self):
        return "Моем плиту и стол"

class Bedroom:
    def clean(self):
        return "Пылесосим и заправляем кровать"

class Robot:
    def clean(self):
        return "Робот-пылесос моет пол"
```

Все три класса — **не связаны** друг с другом. Никакого общего родителя. Но функция работает со всеми:

```python
def clean_room(room):
    print(room.clean())

clean_room(Kitchen())    # Моем плиту и стол
clean_room(Bedroom())    # Пылесосим и заправляем кровать
clean_room(Robot())      # Робот-пылесос моет пол
```

### Утиная типизация — что это

Принцип звучит так:

> **«Если это выглядит как утка, плавает как утка и крякает как утка — это утка».**

В Python это означает: **неважно, какого класса объект**. Важно, что у него есть **нужные методы**. Если у объекта есть `clean()` — его можно убирать, независимо от того, кто он.

### Название

Название пришло из английского: **duck typing**. Принцип приписывают Джеймсу Уиткомбу Райли, но популяризировал его Python-мир.

### Плюсы утиной типизации

- **Гибкость:** не нужно наследовать общий родитель.
- **Естественность:** не нужно создавать искусственные иерархии.
- **Лёгкость:** любой класс с нужным методом подходит.

### Минусы утиной типизации

- **Ошибка видна только в рантайме.** Если у объекта нет метода, программа упадёт **в момент вызова**.
- **IDE не подскажет.** Автодополнение не знает, какие методы есть у объекта.
- **Нет гарантии.** Нельзя проверить заранее, что объект «правильный».

### Утиная типизация vs наследование

| Критерий | Через наследование | Через утиную типизацию |
|----------|-------------------|------------------------|
| Нужен общий родитель | ✅ Да | ❌ Нет |
| Явная иерархия | ✅ Есть | ❌ Нет |
| Гибкость | ⚠️ Ограничена | ✅ Высокая |
| Ошибка видна | Раньше | Только в рантайме |
| IDE подсказывает | ✅ Да | ❌ Нет |

**В Python предпочитают утиную типизацию.** Это одна из ключевых философских идей языка.

---

## 4.7. Полиморфизм через абстрактные классы

Смешанный подход — **абстрактный класс** (см. главу 7). Он задаёт **общий интерфейс** и требует от наследников его реализации.

```python
from abc import ABC, abstractmethod

class HomeItem(ABC):
    @abstractmethod
    def clean(self):
        pass

class Kitchen(HomeItem):
    def clean(self):
        return "Моем плиту и стол"

class Bedroom(HomeItem):
    def clean(self):
        return "Пылесосим и заправляем кровать"
```

**Преимущество:** гарантия, что у всех наследников **точно есть** метод `clean`. Python не даст создать класс-наследник без реализации.

**Недостаток:** работает только с наследниками — «чужие» классы не подойдут.

### Когда использовать

- Если нужна **гарантия** интерфейса — используйте абстрактный класс.
- Если нужна **гибкость** — используйте утиную типизацию.
- В Python чаще комбинируют: абстрактный класс задаёт «официальный» интерфейс, а утиная типизация позволяет добавлять «неофициальные» классы.

---

## 4.8. Полиморфизм в действии: список объектов

Одна из самых частых ситуаций — **список объектов разных типов**, с которыми нужно работать одинаково:

```python
rooms = [
    Kitchen("Кухня"),
    Bedroom("Спальня"),
    Bathroom("Ванная"),
    Kitchen("Вторая кухня"),
]

for room in rooms:
    print(f"{room.name}: {room.clean()}")
```

**Ожидаемый вывод:**

```
Кухня: Моем плиту и стол
Спальня: Пылесосим и заправляем кровать
Ванная: Чистим плитку и сантехнику
Вторая кухня: Моем плиту и стол
```

Цикл **один**, а поведение — **разное** для каждого объекта. Это полиморфизм в действии.

### Почему это удобно

Без полиморфизма пришлось бы писать:

```python
for room in rooms:
    if isinstance(room, Kitchen):
        print(room.clean_kitchen())
    elif isinstance(room, Bedroom):
        print(room.clean_bedroom())
    elif isinstance(room, Bathroom):
        print(room.clean_bathroom())
```

Много кода, много условий. Полиморфизм убирает всё это: **один вызов — работает для всех**.

---

## 4.9. `isinstance()` и полиморфизм

Иногда всё же нужно **проверить тип** объекта. Для этого есть `isinstance()`:

```python
def describe(room):
    if isinstance(room, Kitchen):
        print("Это кухня — здесь готовят")
    elif isinstance(room, Bedroom):
        print("Это спальня — здесь спят")
    else:
        print("Просто комната")
```

### Но осторожно!

**`isinstance` — это антиполиморфизм.** Если вы используете `isinstance` в большом количестве мест, скорее всего, вы **делаете что-то не так**. Полиморфизм существует именно для того, чтобы **избегать** таких проверок.

### Плохой пример

```python
def clean(room):
    if isinstance(room, Kitchen):
        print("Моем плиту и стол")
    elif isinstance(room, Bedroom):
        print("Пылесосим и заправляем кровать")
    elif isinstance(room, Bathroom):
        print("Чистим плитку и сантехнику")
```

Здесь **нарушен полиморфизм**. Если добавится новый тип комнаты, придётся править эту функцию.

### Хороший пример

```python
def clean(room):
    room.clean()      # каждый объект знает, как убирать себя
```

Функция **не проверяет типы** — она просто вызывает метод. Новый тип комнаты — ничего менять не нужно.

**Правило:** `isinstance` — используйте **редко**, когда действительно нужно разное поведение **вне класса**. Обычно полиморфизм решает эту задачу лучше.

---

## 4.10. Полиморфизм и `str()` / `len()` / `print()`

Полиморфизм — это **не только про ваши классы**. Встроенные функции Python тоже полиморфны:

```python
print(len("hello"))      # 5 — длина строки
print(len([1, 2, 3]))    # 3 — длина списка
print(len({"a": 1}))     # 1 — длина словаря
```

`len()` работает с **разными типами** — потому что у каждого из них есть метод `__len__`. Это тоже полиморфизм.

То же самое с `str()`:

```python
print(str(42))           # "42"
print(str(3.14))         # "3.14"
print(str([1, 2, 3]))    # "[1, 2, 3]"
```

Каждый тип реализует `__str__` по-своему — `str()` просто вызывает его. Это **полиморфизм через магические методы**.

### Пример из нашего дома

Если у `Room` определить метод `__str__`:

```python
class Room:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Комната: {self.name}"

print(Room("Кухня"))    # Комната: Кухня
```

`print()` работает с любым объектом, у которого есть `__str__`. Это тоже полиморфизм.

---

## 4.11. Полиморфизм и `+`

Оператор `+` тоже полиморфен:

```python
print(2 + 3)              # 5 — сложение чисел
print("a" + "b")          # "ab" — конкатенация строк
print([1, 2] + [3, 4])    # [1, 2, 3, 4] — объединение списков
```

Один оператор — **разное поведение**. Это реализовано через магические методы:

- `int.__add__` — арифметическое сложение.
- `str.__add__` — конкатенация.
- `list.__add__` — объединение списков.

### Пример из нашего дома

Если у `Room` определить `__add__`, можно будет складывать комнаты:

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area

    def __add__(self, other):
        return Room(f"{self.name}+{other.name}", self.area + other.area)

combined = Room("Кухня", 12) + Room("Спальня", 18)
print(combined.name)    # Кухня+Спальня
print(combined.area)    # 30
```

Мы разберём магические методы подробно в главах 11–13.

---

## 4.12. Полиморфизм vs наследование vs инкапсуляция

Эти три понятия часто путают. Давай разберёмся.

| Понятие | Что это | Отвечает на вопрос |
|---------|---------|-------------------|
| **Инкапсуляция** | Сокрытие данных и доступ через методы | «Как защитить данные?» |
| **Наследование** | Создание класса на основе другого | «Как переиспользовать код?» |
| **Полиморфизм** | Один вызов — разное поведение | «Как работать с разными типами одинаково?» |

### Как они связаны

- **Инкапсуляция** позволяет объекту **самому** управлять своим состоянием.
- **Наследование** позволяет **переиспользовать** код и создавать иерархии.
- **Полиморфизм** позволяет **писать общий код** для разных типов.

Вместе они образуют **три кита ООП**.

### Пример: три кита в одной задаче

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.__area = area      # инкапсуляция

    @property
    def area(self):
        return self.__area

class Kitchen(Room):            # наследование
    def clean(self):
        return "Моем плиту и стол"

class Bedroom(Room):            # наследование
    def clean(self):
        return "Пылесосим и заправляем кровать"

def clean_room(room):           # полиморфизм
    print(room.clean())
```

## 4.13. Полиморфизм в реальных проектах

Где полиморфизм используется в реальной жизни?

### 1. GUI-приложения

Все виджеты (`Button`, `Label`, `Input`) имеют метод `render()`. Отрисовка идёт одним циклом:

```python
for widget in widgets:
    widget.render()
```

### 2. Обработка платежей

Все платёжные системы (`Card`, `PayPal`, `Crypto`) имеют метод `pay()`:

```python
for payment in payments:
    payment.pay(amount)
```

### 3. Логирование

Все логгеры (`FileLogger`, `ConsoleLogger`, `EmailLogger`) имеют метод `log()`:

```python
logger.log("Сообщение")
```

### 4. Игровые объекты

Все игровые объекты (`Player`, `Enemy`, `Item`) имеют метод `update()`:

```python
for obj in game_objects:
    obj.update()
```

### 5. Наш дом

Все комнаты (`Kitchen`, `Bedroom`, `Bathroom`) имеют метод `clean()`:

```python
for room in rooms:
    room.clean()
```

**Общий принцип:** полиморфизм позволяет **писать циклы и функции**, которые работают с **любыми объектами**, имеющими нужный интерфейс.

## 4.14. Типичные ошибки

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Думают, что полиморфизм требует наследования | В Python — нет, достаточно метода | Использовать утиную типизацию |
| Забывают, что ошибка проявляется в рантайме | Программа падает при вызове | Использовать тесты, аннотации типов |
| Путают полиморфизм и перегрузку | Это разные вещи | Полиморфизм — про классы, перегрузка — про параметры |
| Пишут `isinstance` вместо полиморфизма | Нарушает идею ООП | Использовать методы объектов |
| Вызывают метод, которого нет у объекта | `AttributeError` в рантайме | Проверять или использовать `hasattr` |
| Думают, что все методы должны быть в родителе | Родитель задаёт общий интерфейс | Оставлять только общее |
| Называют метод по-разному в разных классах | Не работает полиморфизм | Имя метода должно совпадать |

---

## 4.15. Что важно запомнить

- **Полиморфизм** — один вызов, разное поведение.
- **Python определяет метод по фактическому классу** объекта (динамическое связывание).
- **Два способа полиморфизма:**
  1. **Через наследование** — все классы наследуют общий родитель.
  2. **Через утиную типизацию** — достаточно, чтобы был нужный метод.
- **Утиная типизация:** «Если это выглядит как утка и крякает как утка — это утка».
- **В Python предпочитают утиную типизацию** — она гибче.
- **Ошибка видна только в рантайме** — используйте тесты.
- **Полиморфизм работает и во встроенных функциях** — `len()`, `str()`, `print()`, `+`.
- **`isinstance` — антиполиморфизм.** Используйте редко.
- **Полиморфизм позволяет писать общие циклы** для разных типов.

## Задача 4 (полиморфизм)

Дом убирают по-разному в зависимости от типа комнаты.

**Часть 1.** Создайте три класса — `Kitchen`, `Bedroom` и `Bathroom`, у каждого из которых есть метод `clean()`, возвращающий строку с описанием уборки:

- `Kitchen` → «Моем плиту и стол»
- `Bedroom` → «Пылесосим и заправляем кровать»
- `Bathroom` → «Чистим плитку и сантехнику»

Классы **не должны наследовать** общего родителя — это пример утиной типизации.

**Часть 2.** Напишите функцию `show_cleaning(room)`, которая принимает любую комнату и выводит описание уборки в формате: `Уборка: <описание>`.

**Часть 3.** Создайте список из **четырёх** объектов (например, две кухни, спальня, ванная) и переберите его в цикле `for`, вызывая `show_cleaning` для каждого.

**Часть 4.** Дополнительно создайте класс `Robot` с методом `clean()`, возвращающим «Робот-пылесос моет пол». Убедитесь, что функция `show_cleaning` работает и с ним — **без изменений в коде функции**.

**Часть 5.** В комментариях объясните:

1. Как Python определяет, какой именно метод вызвать.
2. Что такое утиная типизация.
3. Почему ошибка в утиной типизации видна только во время выполнения.
4. Почему `isinstance` — антиполиморфизм.

### Решение

```python
class Kitchen:
    def clean(self):
        return "Моем плиту и стол"


class Bedroom:
    def clean(self):
        return "Пылесосим и заправляем кровать"


class Bathroom:
    def clean(self):
        return "Чистим плитку и сантехнику"


class Robot:
    def clean(self):
        return "Робот-пылесос моет пол"


def show_cleaning(room):
    # Функции всё равно, какого класса объект.
    # Ей важно только, что у объекта есть метод clean().
    print(f"Уборка: {room.clean()}")


# Создаём список объектов разных типов.
rooms = [
    Kitchen(),
    Bedroom(),
    Bathroom(),
    Kitchen(),   # вторая кухня
    Robot(),     # робот — без изменений в функции
]

for room in rooms:
    show_cleaning(room)
```

**Ожидаемый вывод:**

```
Уборка: Моем плиту и стол
Уборка: Пылесосим и заправляем кровать
Уборка: Чистим плитку и сантехнику
Уборка: Моем плиту и стол
Уборка: Робот-пылесос моет пол
```

### Разбор

**1. Как Python определяет, какой именно метод вызвать.**

Когда Python видит `room.clean()`, он смотрит на **фактический класс** объекта `room`. Если это `Kitchen` — ищет метод `clean` в классе `Kitchen`. Если находит — вызывает. Если нет — идёт к родителям (по MRO). Всё решается **во время выполнения**, по фактическому типу объекта. Это называется **динамическое связывание**.

**2. Что такое утиная типизация.**

Утиная типизация — принцип, по которому **важен не класс объекта, а его методы**. «Если это выглядит как утка, плавает как утка и крякает как утка — это утка». В нашем примере классы `Kitchen`, `Bedroom`, `Bathroom` и `Robot` **не связаны** друг с другом. Но у всех есть метод `clean()` — и этого достаточно.

**3. Почему ошибка в утиной типизации видна только во время выполнения.**

Python — **динамически типизированный** язык. Он не проверяет заранее, есть ли у объекта метод `clean()`. Проверка происходит **в момент вызова**. Если у объекта нет метода — Python выдаст `AttributeError` **только тогда**, когда дойдёт до этой строки. Это и плюс (гибкость), и минус (ошибки всплывают позже). Поэтому используют **тесты** и **аннотации типов**.

**4. Почему `isinstance` — антиполиморфизм.**

`isinstance` проверяет **тип** объекта. Если писать код так:

```python
def show_cleaning(room):
    if isinstance(room, Kitchen):
        print("Моем плиту и стол")
    elif isinstance(room, Bedroom):
        print("Пылесосим и заправляем кровать")
    elif isinstance(room, Robot):
        print("Робот-пылесос моет пол")
```

то при добавлении нового типа комнаты придётся **править функцию**. Это **нарушает полиморфизм** — вместо того чтобы объект сам знал, как убирать себя, мы заставляем функцию это решать.

**Правильно:** функция вызывает `room.clean()` и **не знает**, какой это класс. Каждый объект сам знает, как убирать **себя**. Это и есть полиморфизм.

# Глава 5. Модификаторы доступа

## 5.1. Что такое модификаторы доступа

**Модификатор доступа** — это способ указать, **кто и откуда** может обращаться к атрибуту или методу класса. В разных языках это ключевые слова: `public`, `private`, `protected`. В Python таких ключевых слов **нет** — вместо них используется **соглашение об именах**: количество подчёркиваний перед именем.

### Зачем нужны модификаторы доступа

Когда класс большой, важно **разделить** то, что является **публичным интерфейсом**, и то, что является **внутренней кухней**:

- **Публичное** — то, чем могут пользоваться все: `room.info()`, `room.name`.
- **Внутреннее** — то, что нужно только самому классу: кэш, счётчики, служебные поля.

Если всё сделать публичным, программисты начнут использовать **внутренние** методы и поля снаружи. А потом, при рефакторинге, изменить их будет невозможно — весь чужой код сломается.

**Модификаторы доступа помогают** чётко обозначить: «это публичное, используй», «это внутреннее, не трогай».

### Аналогия

Представь автомобиль:

- **Публичное:** руль, педали, кнопки — то, чем пользуется водитель.
- **Защищённое:** капот — можно открыть, но обычно этим занимается механик.
- **Приватное:** электроника двигателя — снаружи не достать.

Ты не открываешь капот каждый раз, чтобы поехать. Но если нужно — механик может. И только механик знает, что под капотом.

---

## 5.2. Три уровня доступа в Python

В Python есть **три уровня доступа**, которые различаются **количеством подчёркиваний**:

| Уровень | Синтаксис | Доступ | Идея |
|---------|-----------|--------|------|
| **Публичный** | `name` | Везде | «Это часть интерфейса» |
| **Защищённый** | `_name` | Внутри класса и у наследников | «Это внутреннее, но наследникам пригодится» |
| **Приватный** | `__name` | Только внутри класса | «Это только моё» |

### Важно понять сразу

**В Python нет настоящих модификаторов доступа.** Всё это — **соглашения**, а не запреты. Приватные поля можно обойти через name mangling (см. 5.5). Защищённые — и так доступны. Публичные — тем более.

Программисты договорились: **если видишь подчёркивания — не трогай**. Это работает, потому что все уважают соглашение. Если кто-то его нарушает — он либо знает, что делает, либо плохой программист.

---

## 5.3. Публичные атрибуты и методы

**Публичные** — без подчёркиваний. Доступны откуда угодно.

### Атрибуты

```python
class Room:
    def __init__(self, name, area):
        self.name = name    # публичный атрибут
        self.area = area    # публичный атрибут
```

```python
kitchen = Room("Кухня", 12)
print(kitchen.name)         # Кухня — работает
kitchen.name = "Спальня"    # работает
```

### Методы

```python
class Room:
    def info(self):         # публичный метод
        return "Комната"
```

```python
kitchen.info()              # работает
```

### Когда использовать

- **Имя, площадь, этаж** — публичные: эти данные видны всем.
- **Метод `info()`** — публичный: это часть интерфейса класса.
- Всё, что вы **готовы поддерживать** в будущем — публичное.

### Правило

**По умолчанию всё делайте публичным.** Переводите в защищённые или приватные только то, что действительно **внутреннее**.

Многие новички делают **всё приватным** — «на всякий случай». Это ошибка: код становится неудобным, тестирование усложняется, а защиты никакой (в Python её нет).

---

## 5.4. Защищённые атрибуты и методы

**Защищённые** — с **одним** подчёркиванием. Это **сигнал**: «это внутреннее, не трогай без нужды».

### Атрибуты

```python
class Room:
    def __init__(self, name, area):
        self.name = name       # публичный
        self._area = area      # защищённый
```

```python
kitchen = Room("Кухня", 12)
print(kitchen._area)           # 12 — работает, но не рекомендуется
```

### Методы

```python
class Room:
    def _validate_area(self, value):   # защищённый метод
        return 5 <= value <= 100
```

### Когда использовать

- **Поле, нужное наследникам.** Например, `_area` — потому что класс `Kitchen` может захотеть работать с площадью.
- **Внутренний метод, который может пригодиться в наследнике.** Например, `_validate_area`.
- **Служебные поля**, которые не должны попадать в публичный интерфейс, но и не являются строго приватными.

### Как понять, что это защищённое

Если поле нужно **только вам и вашим наследникам** — делайте его защищённым.

### Ключевое отличие от приватного

| Защищённое (`_name`) | Приватное (`__name`) |
|----------------------|----------------------|
| Доступно наследникам | Недоступно наследникам (напрямую) |
| Можно обратиться снаружи | Нельзя обратиться снаружи |
| Не искажается | Искажается (name mangling) |

### Пример: зачем нужно защищённое

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self._area = area       # защищённое — нужно наследникам

class Kitchen(Room):
    def describe(self):
        return f"Кухня площадью {self._area} м²"    # работает!
```

Если бы `_area` было `__area`, в наследнике `Kitchen` оно было бы **недоступно** (см. 5.5).

---

## 5.5. Приватные атрибуты и методы

**Приватные** — с **двумя** подчёркиваниями. Python **переименовывает** их, и снаружи обратиться напрямую нельзя.

### Атрибуты

```python
class Room:
    def __init__(self, name, area):
        self.name = name        # публичный
        self.__area = area      # приватный
```

```python
kitchen = Room("Кухня", 12)
print(kitchen.__area)
# AttributeError: 'Room' object has no attribute '__area'
```

### Методы

```python
class Room:
    def __secret_method(self):    # приватный метод
        return "Только для внутреннего использования"
```

Снаружи вызвать нельзя:

```python
kitchen.__secret_method()    # AttributeError
```

### Когда использовать

- **Поле, которое не должно меняться извне и не нужно наследникам.** Например, внутренний кэш.
- **Внутренний служебный метод.** Например, `__validate` — не должен использоваться никем, кроме самого класса.
- **Данные, которые чувствительны** — например, пароль, доход, служебная информация.

### Ошибка: делать всё приватным

Новички часто делают **все** поля приватными. Это плохо:

- Снаружи приходится писать геттеры/сеттеры для всего.
- Код усложняется без пользы.
- Наследники не могут получить доступ к нужным полям.

**Правило:** делайте приватным **только то, что действительно внутреннее**.

---

## 5.6. Name mangling — подробно

Это **самая важная часть** темы. Разберём детально.

### Что происходит

Когда Python видит атрибут с двумя подчёркиваниями в начале — например, `__area` **внутри класса `Room`** — он автоматически **переименовывает** его в `_Room__area`.

Это называется **name mangling** («искажение имени»).

### Пример

```python
class Room:
    def __init__(self, area):
        self.__area = area      # станет self._Room__area
```

На самом деле в объекте лежит поле `_Room__area`:

```python
kitchen = Room(12)
print(kitchen.__dict__)
# {'_Room__area': 12}
```

### Почему нельзя обратиться снаружи

Снаружи вы пишете `kitchen.__area`, но Python **не ищет** поле с таким именем — он ищет `__area` **как есть** (без mangling). А реальное имя — `_Room__area`. Поэтому:

```python
kitchen.__area          # AttributeError
kitchen._Room__area     # работает — это и есть настоящее имя
```

### Внутри класса работает

Внутри класса Python **автоматически** заменяет `self.__area` на `self._Room__area`:

```python
class Room:
    def __init__(self, area):
        self.__area = area         # → self._Room__area

    def get_area(self):
        return self.__area         # → self._Room__area
```

Оба имени внутри класса превращаются в одно. Поэтому внутри класса всё работает.

### Зачем нужен name mangling

**Не для безопасности.** Python не защищает данные — он просто переименовывает. Настоящая цель — **две задачи**:

**1. Избежать конфликтов имён при наследовании.**

```python
class Room:
    def __init__(self):
        self.__x = "Room"

class Kitchen(Room):
    def __init__(self):
        super().__init__()
        self.__x = "Kitchen"
```

Что получится? Два **разных** поля: `_Room__x` и `_Kitchen__x`. Без name mangling они бы перезаписывали друг друга. С ним — существуют независимо.

**2. Сигнализировать другим программистам: «это внутреннее».**

Когда видишь `__area` — понимаешь, что лезть туда не надо.

### Ключевая мысль

**Приватность в Python — договорённость.** Технически доступ есть — через `_Class__name`. Но программисты договорились: **не трогаем приватные поля снаружи**. Это работает, потому что все согласны.

---

## 5.7. Приватные и защищённые методы в наследниках

Это **тонкий момент**, который часто путают.

### Приватные атрибуты и наследники

**Приватные атрибуты родителя недоступны наследнику напрямую** — из-за name mangling:

```python
class Room:
    def __init__(self, area):
        self.__area = area       # → _Room__area

class Kitchen(Room):
    def describe(self):
        return self.__area       # → _Kitchen__area — AttributeError!
```

Внутри `Kitchen` Python ищет `_Kitchen__area`, а не `_Room__area`. Такого поля нет — `AttributeError`.

**Решение:** если поле должно быть доступно наследникам — делайте его **защищённым** (`_area`), а не приватным.

### Защищённые атрибуты и наследники

Защищённые атрибуты **доступны** наследникам напрямую:

```python
class Room:
    def __init__(self, area):
        self._area = area        # защищённое

class Kitchen(Room):
    def describe(self):
        return f"Кухня {self._area} м²"     # работает
```

Это ещё одна причина использовать `_name`, если поле нужно наследникам.

### Приватные методы и наследники

Аналогично: приватные методы родителя недоступны наследнику.

```python
class Room:
    def __validate(self):
        return True

class Kitchen(Room):
    def check(self):
        return self.__validate()     # → _Kitchen__validate — AttributeError
```

### Схема

| Что | Наследник видит? |
|-----|------------------|
| `name` | ✅ Да |
| `_name` | ✅ Да |
| `__name` | ❌ Нет (искажается под имя наследника) |

---

## 5.8. Обращение к приватному полю через name mangling

Иногда **действительно нужно** получить доступ к приватному полю — например, при отладке или тестировании. Python **позволяет** это сделать через `_Class__name`.

```python
class Room:
    def __init__(self, area):
        self.__area = area

kitchen = Room(12)

# Снаружи обычный доступ — не работает.
try:
    print(kitchen.__area)
except AttributeError as e:
    print("Ошибка:", e)

# Через name mangling — работает.
print(kitchen._Room__area)    # 12
```

### Важно

Это **технически возможно**, но **плохая практика**. Если вы обращаетесь к `_Class__name` снаружи — вы **нарушаете соглашение**. Делать это можно только в исключительных случаях:

- **отладка** — когда нужно посмотреть, что внутри;
- **тесты** — когда нужно проверить внутреннее состояние;
- **крайние меры** — когда другого способа нет.

В обычном коде **так писать не надо**.

---

## 5.9. Атрибуты класса тоже имеют уровни доступа

Все три уровня работают и для атрибутов **класса** (не только экземпляра).

```python
class Room:
    type = "Комната"          # публичный
    _default_area = 20        # защищённый
    __version = "1.0"         # приватный
```

```python
print(Room.type)              # Комната — работает
print(Room._default_area)     # 20 — работает, но не рекомендуется
print(Room.__version)         # AttributeError
print(Room._Room__version)    # 1.0 — через name mangling
```

### Когда использовать

- **Публичный** — общая константа для всех объектов: `type = "Комната"`.
- **Защищённый** — значение по умолчанию, нужное наследникам: `_default_area = 20`.
- **Приватный** — служебная версия класса, не нужная никому снаружи.

---

## 5.10. Уровни доступа в реальной жизни

Когда какой уровень использовать — вопрос **дизайна класса**. Приведём несколько примеров.

### Пример 1: класс `Room`

```python
class Room:
    def __init__(self, name, area):
        self.name = name           # публичный — все видят название
        self._area = area          # защищённый — нужно наследникам
        self.__passport = None     # приватный — только сам класс

    def info(self):                # публичный метод
        return f"{self.name} ({self._area} м²)"

    def _validate(self):           # защищённый метод
        return self._area > 0

    def __register(self):          # приватный метод
        self.__passport = "OK"
```

**Почему так:**

- `name` — публичное: название комнаты видно всем.
- `_area` — защищённое: наследник `Kitchen` может захотеть работать с площадью.
- `__passport` — приватное: служебное поле, никому снаружи не нужное.
- `info` — публичный метод: часть интерфейса.
- `_validate` — защищённый: наследник может переопределить проверку.
- `__register` — приватный: внутренний метод, не должен вызываться никем другим.

### Пример 2: класс `Owner`

```python
class Owner:
    def __init__(self, name, income):
        self.name = name           # публичный
        self.__income = income     # приватный — доход не должен быть виден

    @property
    def income(self):              # публичное свойство для чтения
        return self.__income
```

Здесь:

- `name` — публичное: имя владельца может видеть любой.
- `__income` — приватное: доход скрыт.
- `income` — свойство для чтения: можно узнать доход, но не изменить.

---

## 5.11. Разница между `_name`, `__name` и `__name__`

Тут есть **важный нюанс**, который путает новичков. Три разных случая:

| Синтаксис | Название | Что это |
|-----------|----------|---------|
| `name` | Публичный | Обычный атрибут |
| `_name` | Защищённый | Соглашение «не трогай» |
| `__name` | Приватный | Name mangling — искажается в `_Class__name` |
| `__name__` | **Магический** | **НЕ приватный!** Специальные методы Python |

### Магические методы

Два подчёркивания **с двух сторон** (`__init__`, `__str__`, `__eq__`) — это **магические методы**, а не приватные. Они **не** искажаются, и их **можно** вызывать снаружи:

```python
class Room:
    def __init__(self, name):      # магический метод
        self.name = name

    def __str__(self):             # магический метод
        return f"Room({self.name})"

kitchen = Room("Кухня")
print(str(kitchen))                # работает
print(kitchen.__str__())           # тоже работает — магические методы доступны
```

**Не путайте!** `__name` (без завершающих подчёркиваний) — приватный. `__name__` (с завершающими) — магический, публичный.

### Правило именования

- `name` — публичный.
- `_name` — защищённый.
- `__name` — приватный (Python искажает).
- `__name__` — магический метод Python (не трогайте без причины).
- `name_` — если имя конфликтует с ключевым словом: `class_`.

---

## 5.12. Модификаторы доступа в других языках (для сравнения)

Чтобы понять, **чем Python отличается**, сравним с другими языками.

| Язык | Публичный | Защищённый | Приватный |
|------|-----------|-----------|-----------|
| **Python** | `name` | `_name` (соглашение) | `__name` (mangling) |
| **Java** | `public` | `protected` | `private` |
| **C++** | `public:` | `protected:` | `private:` |
| **C#** | `public` | `protected` | `private` |

### Ключевые отличия Python

- **Нет ключевых слов** — только соглашение об именах.
- **Нет настоящей защиты** — приватность можно обойти через name mangling.
- **Всё основано на договорённости** — программисты уважают соглашение.

### Почему так

Python — **динамический язык**. Он верит, что программист **взрослый человек** и сам знает, что можно, а что нет. Не нужно жёстких ограничений: если ты видишь `_name` — не трогай, и всё будет хорошо.

**Философия Python:** «Мы все здесь взрослые люди».

---

## 5.13. Типичные ошибки

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Делают всё приватным | Код становится неудобным, защиты нет | Публичное по умолчанию, приватное — только когда нужно |
| Думают, что `__name` — настоящая защита | Можно обойти через mangling | Помнить: это соглашение |
| Используют `_name` как «приватный» | Это защищённый, доступ открыт | Различать уровни |
| Обращаются к `_Class__name` снаружи | Нарушает соглашение | Не делать этого в обычном коде |
| Обращаются к `__name` в наследнике | Приватное поле родителя недоступно | Использовать `_name` |
| Путают `__name` и `__name__` | Разные вещи | Первое — приватное, второе — магическое |
| Делают публичными внутренние поля | Их начнут использовать снаружи | Скрывать внутреннее |
| Называют метод `_info` вместо `info` | Лишнее подчёркивание | Публичный метод — без подчёркиваний |

---

## 5.14. Что важно запомнить

- **В Python нет настоящих модификаторов доступа** — только соглашения.
- **Публичный** (`name`) — доступен везде.
- **Защищённый** (`_name`) — доступен внутри класса и наследникам; снаружи можно, но не рекомендуется.
- **Приватный** (`__name`) — доступен только внутри класса; снаружи нельзя напрямую.
- **Name mangling:** `__name` превращается в `_Class__name`.
- **Приватность в Python — договорённость**, а не жёсткая защита.
- **Наследник не видит приватные поля родителя** — используйте `_name`, если поле нужно наследникам.
- **`__name__` (магические методы) — не приватные!** Это специальные методы Python.
- **По умолчанию всё делайте публичным.** Скрывайте только внутреннее.
- **Обращение к `_Class__name` снаружи** — плохая практика.
- **Философия Python:** «Мы все здесь взрослые люди».

## Задача 5 (модификаторы доступа)

Создайте класс `Owner` (владелец дома), в котором есть три поля, демонстрирующие три уровня доступа:

- `name` — **публичное** (имя владельца).
- `_house` — **защищённое** (название дома, доступное наследникам).
- `__income` — **приватное** (ежемесячный доход, только для самого класса).

**Часть 1.** Реализуйте конструктор, который принимает `name`, `house`, `income` и сохраняет их в соответствующие поля.

**Часть 2.** Добавьте публичный метод `info()`, возвращающий строку с именем владельца и названием дома. Не включайте доход — он приватный.

**Часть 3.** Создайте объект `owner` и продемонстрируйте доступ к каждому полю:

- к `name` — свободно;
- к `_house` — можно, но с комментарием «не рекомендуется»;
- к `__income` — попробуйте обратиться напрямую, поймайте `AttributeError` и выведите сообщение;
- затем получите доступ к доходу через name mangling: `owner._Owner__income` — с комментарием «так делать не нужно, но технически возможно».

**Часть 4.** Создайте наследника `Tenant` (арендатор), который наследует от `Owner`. В нём определите метод `describe()`, возвращающий название дома через `_house`. Убедитесь, что защищённое поле **доступно** наследнику.

**Часть 5.** В комментариях объясните:

1. Зачем нужны три уровня доступа.
2. Почему `__name` не защищает по-настоящему.
3. Почему наследник **не видит** `__income`, но **видит** `_house`.
4. Почему в Python предпочли соглашение, а не ключевые слова вроде `private`.

### Решение

```python
class Owner:
    def __init__(self, name, house, income):
        self.name = name            # публичное — доступно везде
        self._house = house         # защищённое — доступно наследникам
        self.__income = income      # приватное — только внутри класса

    def info(self):
        # Публичный метод. Не показывает доход — он приватный.
        return f"Владелец: {self.name}, дом: {self._house}"


class Tenant(Owner):
    def describe(self):
        # Защищённое поле _house доступно наследнику.
        return f"Арендатор в доме: {self._house}"


# Создаём объект владельца.
owner = Owner("Иванов И.И.", "ул. Ленина, д. 10", 120_000)

# 1. Публичное поле — доступно свободно.
print("Публичное name:", owner.name)

# 2. Защищённое поле — доступно, но не рекомендуется.
print("Защищённое _house:", owner._house)   # работает, но не рекомендуется

# 3. Приватное поле — напрямую недоступно.
try:
    print(owner.__income)
except AttributeError as e:
    print("Приватное __income недоступно:", e)

# 4. Через name mangling — доступ возможен (но не рекомендуется).
print("Через name mangling:", owner._Owner__income)

# 5. Проверяем, что защищённое поле доступно наследнику.
tenant = Tenant("Петров П.П.", "ул. Садовая, д. 3", 80_000)
print(tenant.describe())
```

**Ожидаемый вывод:**

```
Публичное name: Иванов И.И.
Защищённое _house: ул. Ленина, д. 10
Приватное __income недоступно: 'Owner' object has no attribute '__income'
Через name mangling: 120000
Арендатор в доме: ул. Садовая, д. 3
```

### Разбор

**1. Зачем нужны три уровня доступа.**

Три уровня позволяют **разделить** интерфейс класса и его внутреннее устройство:

- **Публичное** — то, чем могут пользоваться все. Это «официальный интерфейс» класса.
- **Защищённое** — то, что нужно наследникам. Это «интерфейс для расширения».
- **Приватное** — то, что касается только самого класса. Это «внутренняя кухня».

Благодаря этому программист, использующий класс, **сразу видит**, что можно трогать, а что нет.

**2. Почему `__name` не защищает по-настоящему.**

`__name` — это **не запрет**, а **переименование**. Python превращает `__income` в `_Owner__income`. Снаружи к этому полю можно обратиться, если знать настоящее имя. Это **защита от случайного доступа**, а не от злонамеренного.

**Зачем тогда `__name`?**

- Чтобы **избежать конфликтов имён** при наследовании.
- Чтобы **сигнализировать** другим программистам: «не трогай, это внутреннее».

**3. Почему наследник не видит `__income`, но видит `_house`.**

Когда Python видит `self.__income` **внутри класса** `Owner`, он превращает это в `self._Owner__income`. Когда тот же `self.__income` встречается **внутри класса** `Tenant`, Python превращает это в `self._Tenant__income`. Это **разные поля** — и второго не существует.

А `_house` — защищённое поле. Оно **не искажается**, поэтому доступно и в `Owner`, и в `Tenant`, и вообще везде. Именно поэтому если поле нужно наследникам — используют **защищённый** уровень, а не приватный.

**4. Почему в Python предпочли соглашение, а не ключевые слова.**

Python — **динамически типизированный язык** с философией «мы все здесь взрослые люди». Вместо жёстких запретов (как `private` в Java) используется **договорённость**:

- Программист **видит** подчёркивания и понимает: не трогай.
- Если очень нужно — может обойти, но это его ответственность.
- Код становится **проще**, **гибче**, без лишних ограничений.

Это работает, потому что сообщество Python **уважает соглашение**. Нарушать его — плохой тон, и это заметно сразу.

# Глава 6. Виды методов: обычные, статические и классовые

## 6.1. Проблема: где разместить утилиту

Представь, что у нас есть класс `Room`, и мы часто проверяем корректность площади комнаты. Логично, что такая проверка **связана с классом** `Room` — она работает с понятием «площадь комнаты». Но **где её разместить**?

Первое, что приходит в голову — сделать её **обычным методом**:

```python
class Room:
    def is_valid_area(self, area):
        return 5 <= area <= 100
```

Но для вызова нужен **объект**:

```python
room = Room()
print(room.is_valid_area(12))    # работает, но зачем создавать объект?
```

Это **неудобно**. Мы создаём объект `Room` только чтобы проверить число. Это бессмысленно: проверка **не зависит** от конкретного объекта.

Ещё хуже — сделать её **отдельной функцией** вне класса:

```python
def is_valid_area(area):
    return 5 <= area <= 100
```

Функция работает, но **теряется связь** с классом. Непонятно, что это «проверка для комнаты».

Для таких случаев в Python есть **статические методы**. А ещё есть **классовые методы** — для работы с классом в целом. И те, и другие — **разновидности методов**, наряду с обычными.

В этой главе мы разберём **все три вида** и научимся их правильно сочетать.

---

## 6.2. Три вида методов в Python

В Python есть **три вида методов**, которые различаются тем, **что они получают первым аргументом**:

| Вид | Первый параметр | Декоратор | Что получает |
|-----|----------------|-----------|--------------|
| **Обычный (instance)** | `self` | нет | Конкретный объект |
| **Классовый (class)** | `cls` | `@classmethod` | Сам класс |
| **Статический (static)** | нет | `@staticmethod` | Ничего |

### Схема

```
        Обычный метод       Классовый метод       Статический метод
        ─────────────       ───────────────       ──────────────────
        def m(self)         @classmethod          @staticmethod
                            def m(cls)            def m()
                ↓                   ↓                     ↓
        получает объект      получает класс         ничего не получает
                ↓                   ↓                     ↓
        работает с данными   работает с данными      просто функция
        объекта              класса
```

Все три вида могут **сосуществовать в одном классе**. Никаких ограничений нет.

Разберём каждый вид подробно.

---

## 6.3. Обычные методы (instance methods)

**Обычный метод** — самый распространённый. Он **получает `self`** — ссылку на конкретный объект — и работает с его данными.

### Синтаксис

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area

    def info(self):                    # обычный метод
        return f"{self.name}: {self.area} м²"
```

### Как работает

- **Первый параметр — `self`** (ссылка на объект).
- Вызывается через **объект**: `room.info()`.
- Может читать и менять **данные объекта**: `self.name`, `self.area`.
- Python автоматически подставляет объект в `self`.

### Пример

```python
kitchen = Room("Кухня", 12)
print(kitchen.info())          # Кухня: 12 м²
```

### Когда использовать

Обычные методы — **по умолчанию**. Если метод работает с данными объекта — он обычный.

---

## 6.4. Статические методы (`@staticmethod`)

**Статический метод** — метод, который **не получает** ни `self`, ни `cls`. По сути — **обычная функция внутри класса**.

### Синтаксис

```python
class Room:
    @staticmethod
    def is_valid_area(area):          # статический метод
        return 5 <= area <= 100
```

### Как работает

- **Первый параметр — нет** (только те, что передали).
- Вызывается через **класс**: `Room.is_valid_area(12)`.
- **Не работает** ни с объектом, ни с классом.
- Логически связан с классом, но **не зависит** от его состояния.

### Пример

```python
print(Room.is_valid_area(12))      # True
print(Room.is_valid_area(3))       # False
```

### Когда использовать

- **Валидация** — проверка данных.
- **Вычисления** — математические операции.
- **Преобразование** — конвертация форматов.
- **Утилиты** — форматирование, парсинг.

### Признак

Если метод **не использует** ни `self`, ни `cls` — он **статический**.

---

## 6.5. Классовые методы (`@classmethod`)

**Классовый метод** — метод, который **получает `cls`** — ссылку на **класс**, а не на объект.

### Синтаксис

```python
class Room:
    count = 0                          # атрибут класса

    @classmethod
    def get_count(cls):                # классовый метод
        return cls.count
```

### Как работает

- **Первый параметр — `cls`** (ссылка на класс).
- Вызывается через **класс**: `Room.get_count()`.
- Через `cls` может **читать и менять атрибуты класса**: `cls.count`.
- Может **создавать объекты** через `cls(...)`.
- Python автоматически подставляет класс в `cls`.

### Пример

```python
print(Room.get_count())               # 0
```

### Когда использовать

Классовые методы используются в **трёх основных случаях**:

**1. Фабричные методы** — создание объектов из другого представления:

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area

    @classmethod
    def from_string(cls, data):
        name, area = data.split(";")
        return cls(name, int(area))
```

```python
room = Room.from_string("Кухня;12")
print(room.name, room.area)          # Кухня 12
```

**2. Работа с атрибутами класса** — счётчики, реестры, настройки:

```python
class Room:
    count = 0

    def __init__(self, name):
        self.name = name
        Room.count += 1

    @classmethod
    def get_count(cls):
        return cls.count
```

**3. Альтернативные конструкторы** — создание объектов с другим набором параметров:

```python
class Room:
    @classmethod
    def default_kitchen(cls):
        return cls("Кухня", 12)
```

### Ключевое преимущество `@classmethod`

`@classmethod` **автоматически подстраивается** под класс, через который вызван. Если создать наследника, `cls` будет **наследником**, а не родителем:

```python
class Kitchen(Room):
    pass

kitchen = Kitchen.from_string("Кухня;12")
print(type(kitchen))                 # <class 'Kitchen'> — не Room!
```

Это работает благодаря `cls` — он указывает на **фактический класс**, через который вызван метод.

---

## 6.6. Сравнение всех трёх видов

| Критерий | Обычный (`self`) | Классовый (`@classmethod`) | Статический (`@staticmethod`) |
|----------|------------------|---------------------------|-------------------------------|
| Первый параметр | `self` | `cls` | нет |
| Что получает | Объект | Класс | Ничего |
| Вызывается через | Объект | Класс (или объект) | Класс (или объект) |
| Доступ к объекту | ✅ Да | ❌ Нет | ❌ Нет |
| Доступ к классу | ✅ Да (через `type(self)`) | ✅ Да | ❌ Нет |
| Может менять объект | ✅ Да | ❌ Нет | ❌ Нет |
| Может менять класс | ⚠️ Через `type(self)` | ✅ Да | ❌ Нет |
| Типичное использование | Работа с объектом | Фабрика, работа с классом | Утилита |

### Пример: все три вида в одном классе

```python
class Room:
    count = 0

    def __init__(self, name, area):
        self.name = name
        self.area = area
        Room.count += 1

    # 1. Обычный метод — работает с объектом.
    def info(self):
        return f"{self.name}: {self.area} м²"

    # 2. Классовый метод — работает с классом.
    @classmethod
    def get_count(cls):
        return cls.count

    # 3. Статический метод — утилита.
    @staticmethod
    def is_valid_area(area):
        return 5 <= area <= 100
```

Использование:

```python
kitchen = Room("Кухня", 12)
print(kitchen.info())                  # Кухня: 12 м² (обычный)
print(Room.get_count())                # 1 (классовый)
print(Room.is_valid_area(12))          # True (статический)
```

---

## 6.7. Сочетание видов методов в одном классе

**Все три вида отлично сочетаются.** Это **стандартная практика** в реальных проектах.

### Пример: класс `Room` со всеми тремя видами

```python
class Room:
    count = 0

    def __init__(self, name, area):
        self.name = name
        self.area = area
        Room.count += 1

    # ─── Обычные методы ───
    def info(self):
        return f"{self.name}: {self.area} м²"

    def is_big(self):
        return self.area > 50

    # ─── Классовые методы ───
    @classmethod
    def from_string(cls, data):
        name, area = data.split(";")
        return cls(name, int(area))

    @classmethod
    def get_count(cls):
        return cls.count

    # ─── Статические методы ───
    @staticmethod
    def is_valid_area(area):
        return 5 <= area <= 100

    @staticmethod
    def format_area(area):
        return f"{area:.1f} м²"
```

Использование:

```python
# Обычные методы — через объект.
kitchen = Room("Кухня", 12)
print(kitchen.info())                    # Кухня: 12 м²
print(kitchen.is_big())                  # False

# Классовые методы — через класс.
bedroom = Room.from_string("Спальня;18")
print(bedroom.info())                    # Спальня: 18 м²
print(Room.get_count())                  # 2

# Статические методы — через класс.
print(Room.is_valid_area(12))            # True
print(Room.format_area(12.5))            # 12.5 м²
```

### Схема

```
        Room
        ├── __init__        (обычный)
        ├── info            (обычный)
        ├── is_big          (обычный)
        ├── from_string     (@classmethod)
        ├── get_count       (@classmethod)
        ├── is_valid_area   (@staticmethod)
        └── format_area     (@staticmethod)
```

**Никаких ограничений нет.** Класс может содержать **любую комбинацию** методов.

---

## 6.8. Наследование методов всех видов

Все три вида методов **наследуются** и **могут быть переопределены**. Но ведут себя по-разному.

### 6.8.1. Обычные методы при наследовании

**Наследуются и переопределяются** как обычно.

```python
class Room:
    def info(self):
        return "Комната"

class Kitchen(Room):
    def info(self):
        return "Кухня"          # переопределение
```

```python
print(Kitchen().info())          # Кухня
```

**Особенность:** обычные методы **не подстраиваются** под класс. Если метод вызывает `type(self)`, он получит фактический класс объекта — но это работает через `self`, а не через `cls`.

### 6.8.2. Статические методы при наследовании

**Наследуются и могут быть переопределены.**

```python
class Room:
    @staticmethod
    def is_valid_area(area):
        return 5 <= area <= 100

class Kitchen(Room):
    @staticmethod
    def is_valid_area(area):
        return 8 <= area <= 100     # своя версия
```

```python
print(Room.is_valid_area(6))        # True
print(Kitchen.is_valid_area(6))     # False — своя версия
```

**Особенность:** статические методы **не подстраиваются** под класс. Если нужно, чтобы метод **зависел от класса** — используйте `@classmethod`.

### 6.8.3. Классовые методы при наследовании

**Наследуются и работают правильно** с наследниками — благодаря `cls`.

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area

    @classmethod
    def create_default(cls, name):
        # cls — это тот класс, через который вызван метод.
        return cls(name, 20)

class Kitchen(Room):
    pass
```

```python
room = Room.create_default("Комната")
print(type(room))                   # <class 'Room'>

kitchen = Kitchen.create_default("Кухня")
print(type(kitchen))                # <class 'Kitchen'> — cls = Kitchen!
```

**Ключевая особенность:** `cls` указывает на **фактический класс**, через который вызван метод. Это делает `@classmethod` **устойчивым к наследованию**.

### 6.8.4. Смешанный пример: все виды методов в иерархии

```python
class Room:
    count = 0

    def __init__(self, name, area):
        self.name = name
        self.area = area
        Room.count += 1

    # Обычный метод — переопределяется.
    def info(self):
        return f"{self.name}: {self.area} м²"

    # Классовый метод — работает с cls.
    @classmethod
    def create(cls, name, area):
        return cls(name, area)

    # Статический метод — утилита.
    @staticmethod
    def is_valid_area(area):
        return 5 <= area <= 100


class Kitchen(Room):
    # Переопределяем обычный метод.
    def info(self):
        return f"Кухня: {self.name} ({self.area} м²)"

    # Переопределяем статический метод.
    @staticmethod
    def is_valid_area(area):
        return 8 <= area <= 100

    # Классовый метод НЕ переопределяем — работает через cls.


# Обычный метод.
print(Kitchen("Кухня", 12).info())          # Кухня: Кухня (12 м²)

# Классовый метод — cls = Kitchen.
kitchen = Kitchen.create("Кухня", 12)
print(type(kitchen))                        # <class 'Kitchen'>

# Статический метод — своя версия.
print(Room.is_valid_area(6))                # True
print(Kitchen.is_valid_area(6))             # False
```

### 6.8.5. Таблица: как ведут себя методы при наследовании

| Вид метода | Наследуется? | Переопределяется? | Подстраивается под класс? |
|------------|--------------|-------------------|---------------------------|
| **Обычный** (`self`) | ✅ Да | ✅ Да | ❌ Нет |
| **Классовый** (`cls`) | ✅ Да | ✅ Да | ✅ Да (через `cls`) |
| **Статический** | ✅ Да | ✅ Да | ❌ Нет |

---

## 6.9. Когда что использовать

### Правило выбора вида метода

1. **Метод использует `self`?** → **Обычный метод.**
2. **Метод использует `cls`?** → **`@classmethod`.**
3. **Метод не использует ни то, ни другое?** → **`@staticmethod`.**

### Схема выбора

```
        Метод работает с...
                ↓
        ┌───────┴───────┐
        │               │
        ↓               ↓
    с self?          с cls?
        │               │
        ↓               ↓
    Обычный         @classmethod
        │               │
        └───────┬───────┘
                │
                ↓
        не с чем? → @staticmethod
```

### Примеры

| Ситуация | Вид метода |
|----------|-----------|
| Прочитать `self.name` | Обычный |
| Изменить `self.area` | Обычный |
| Увеличить `cls.count` | `@classmethod` |
| Создать объект через `cls(...)` | `@classmethod` |
| Проверить число `is_valid_area(area)` | `@staticmethod` |
| Отформатировать строку `format_area(area)` | `@staticmethod` |

### Случаи использования `@staticmethod`

- **Валидация** — `is_valid_email`, `is_valid_phone`.
- **Расчёты** — `circle_area`, `average`.
- **Преобразования** — `sqm_to_sqft`, `from_iso`.
- **Форматирование** — `format_currency`, `format_date`.

### Случаи использования `@classmethod`

- **Фабричные методы** — `from_string`, `create_default`.
- **Работа с атрибутами класса** — `get_count`, `reset_count`.
- **Альтернативные конструкторы** — `from_dict`, `from_json`.

### Случай использования обычных методов

- **Всё остальное** — работа с данными объекта.

---

## 6.10. Можно ли вызывать методы через объект?

**Да, можно** — но не всегда рекомендуется.

### Обычный метод через объект

```python
room = Room("Кухня", 12)
room.info()                  # ✅ Правильно
```

### Статический метод через класс

```python
Room.is_valid_area(12)       # ✅ Правильно
room.is_valid_area(12)       # ⚠️ Работает, но вводит в заблуждение
```

**Почему не рекомендуется:** кажется, что метод зависит от объекта, но это **не так**. Лучше вызывать через класс.

### Классовый метод через класс

```python
Room.get_count()             # ✅ Правильно
room.get_count()             # ⚠️ Работает, но вводит в заблуждение
```

**Почему не рекомендуется:** кажется, что метод работает с объектом, но он работает с **классом**.

### Правило

- **Обычный метод** → через **объект**.
- **Классовый метод** → через **класс**.
- **Статический метод** → через **класс**.

---

## 6.11. Как Python обрабатывает `@staticmethod` и `@classmethod`

Интересно, **что происходит** при объявлении этих методов.

### `@staticmethod`

Когда вы пишете:

```python
class Room:
    @staticmethod
    def is_valid_area(area):
        return 5 <= area <= 100
```

Python:

1. **Создаёт обычную функцию** `is_valid_area(area)`.
2. **Оборачивает её** в объект `staticmethod`.
3. **Кладёт в `__dict__` класса**.

При вызове `Room.is_valid_area(12)` Python **автоматически разворачивает** `staticmethod` в функцию и вызывает её.

```python
print(Room.__dict__["is_valid_area"])       # <staticmethod object at 0x...>
print(Room.is_valid_area)                   # <function Room.is_valid_area at 0x...>
```

### `@classmethod`

Когда вы пишете:

```python
class Room:
    @classmethod
    def get_count(cls):
        return cls.count
```

Python:

1. **Создаёт обычную функцию** `get_count(cls)`.
2. **Оборачивает её** в объект `classmethod`.
3. **Кладёт в `__dict__` класса**.

При вызове `Room.get_count()` Python **подставляет класс** в `cls`:

```python
print(Room.get_count)       # <bound method Room.get_count of <class 'Room'>>
```

То же самое с объектом: `room.get_count()` подставит **класс объекта** (а не сам объект).

---

## 6.12. Типичные ошибки

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Забыли `@staticmethod` | Python передаст `self` | Добавить декоратор |
| Забыли `@classmethod` | Python передаст `self` вместо `cls` | Добавить декоратор |
| Используют `self` в `@staticmethod` | `self` не передан | Убрать `self` или использовать обычный метод |
| Используют `cls` в `@staticmethod` | `cls` не передан | Использовать `@classmethod` |
| Используют `self` в `@classmethod` | `self` не передан, есть только `cls` | Использовать `cls` |
| Делают статическим метод, работающий с объектом | Нет доступа к `self` | Сделать обычным методом |
| Делают статическим метод, создающий объект класса | Не используется `cls` | Сделать `@classmethod` |
| Делают классовым метод, не использующий `cls` | `cls` не нужен | Сделать `@staticmethod` |
| Думают, что `@staticmethod` = «приватный» | Это разные вещи | `@staticmethod` — про привязку |
| Путают `@staticmethod` и `@classmethod` | Разные параметры | `@staticmethod` — без `cls`, `@classmethod` — с `cls` |
| Вызывают `@staticmethod` через объект | Работает, но вводит в заблуждение | Вызывать через класс |
| Вызывают `@classmethod` через объект | Работает, но вводит в заблуждение | Вызывать через класс |

---

## 6.13. Что важно запомнить

- **Три вида методов:** обычный (`self`), классовый (`cls`), статический (без параметров).
- **Все три могут сосуществовать** в одном классе.
- **Обычный метод** — работает с объектом, вызывается через объект.
- **Классовый метод** — работает с классом, вызывается через класс.
- **Статический метод** — утилита, не работает ни с чем, вызывается через класс.
- **Правило выбора:**
  - использует `self` → обычный;
  - использует `cls` → `@classmethod`;
  - не использует ни то, ни другое → `@staticmethod`.
- **Все три вида наследуются.**
- **`@classmethod` подстраивается** под наследника через `cls`.
- **`@staticmethod` не подстраивается** — это просто функция.
- **Вызывайте методы правильно:** обычные — через объект, классовые и статические — через класс.
- **`@staticmethod` ≠ `@classmethod`:** первый без параметров, второй с `cls`.

## Задача 6 (виды методов)

Дому нужны методы **всех трёх видов**: обычные, статические и классовые. Создайте класс `Room`, в котором будут методы всех трёх видов, и продемонстрируйте их работу.

**Часть 1. Обычные методы.** Создайте класс `Room` с полями `name` (название) и `area` (площадь). Реализуйте:

- конструктор `__init__`;
- обычный метод `info()` — возвращает строку `Название: площадь м²`;
- обычный метод `is_big()` — возвращает `True`, если площадь больше 50 м², и `False` — иначе.

**Часть 2. Статические методы.** Добавьте в класс `Room` два статических метода:

- `is_valid_area(area)` — возвращает `True`, если площадь от 5 до 100 м²;
- `format_area(area)` — возвращает строку вида `12.5 м²`.

**Часть 3. Классовые методы.** Добавьте:

- атрибут класса `count`, увеличивающийся при создании каждого объекта;
- классовый метод `from_string(data)` — создаёт объект из строки `"Название;Площадь"`;
- классовый метод `get_count()` — возвращает значение счётчика.

**Часть 4. Наследование.** Создайте наследника `Kitchen(Room)`:

- переопределите обычный метод `info()` — пусть возвращает `Кухня: Название (площадь м²)`;
- переопределите статический метод `is_valid_area(area)` — пусть возвращает `True`, если площадь от 8 до 100 м²;
- классовый метод **не переопределяйте** — проверьте, что он работает через `cls` и создаёт объекты правильного класса.

**Часть 5. Проверка.** Продемонстрируйте:

- работу обычных методов через объект;
- работу статических методов через класс;
- работу классовых методов через класс;
- что `Kitchen.from_string("Кухня;12")` создаёт объект класса `Kitchen`, а не `Room`;
- что `Room.is_valid_area(6)` возвращает `True`, а `Kitchen.is_valid_area(6)` — `False`.

**Часть 6.** В комментариях объясните:

1. Чем обычный метод отличается от статического.
2. Чем статический метод отличается от классового.
3. Почему `@classmethod` работает правильно при наследовании, а `@staticmethod` — нет.
4. Как выбрать вид метода для конкретной задачи.

### Решение

```python
class Room:
    count = 0                          # атрибут класса

    def __init__(self, name, area):
        self.name = name
        self.area = area
        Room.count += 1

    # ─── Обычные методы ───
    def info(self):
        # Работает с данными объекта.
        return f"{self.name}: {self.area} м²"

    def is_big(self):
        return self.area > 50

    # ─── Классовые методы ───
    @classmethod
    def from_string(cls, data):
        # cls — фактический класс, через который вызван метод.
        name, area = data.split(";")
        return cls(name, int(area))

    @classmethod
    def get_count(cls):
        return cls.count

    # ─── Статические методы ───
    @staticmethod
    def is_valid_area(area):
        # Не зависит ни от объекта, ни от класса.
        return 5 <= area <= 100

    @staticmethod
    def format_area(area):
        return f"{area:.1f} м²"


class Kitchen(Room):
    # Переопределяем обычный метод.
    def info(self):
        return f"Кухня: {self.name} ({self.area} м²)"

    # Переопределяем статический метод.
    @staticmethod
    def is_valid_area(area):
        return 8 <= area <= 100

    # Классовый метод НЕ переопределяем — работает через cls.


# ─── Часть 5. Проверка ───

# Обычные методы — через объект.
kitchen = Room("Кухня", 12)
print(kitchen.info())                       # Кухня: 12 м²
print(kitchen.is_big())                     # False

# Статические методы — через класс.
print(Room.is_valid_area(12))               # True
print(Room.is_valid_area(3))                # False
print(Room.format_area(12.5))               # 12.5 м²

# Классовые методы — через класс.
bedroom = Room.from_string("Спальня;18")
print(bedroom.info())                       # Спальня: 18 м²
print(Room.get_count())                     # 2

# Наследование — классовый метод подстраивается под класс.
kitchen2 = Kitchen.from_string("Кухня;12")
print(type(kitchen2))                       # <class 'Kitchen'>
print(kitchen2.info())                      # Кухня: Кухня (12 м²)

# Наследование — статический метод переопределён.
print(Room.is_valid_area(6))                # True (5 <= 6 <= 100)
print(Kitchen.is_valid_area(6))             # False (6 < 8)
```

**Ожидаемый вывод:**

```
Кухня: 12 м²
False
True
False
12.5 м²
Спальня: 18 м²
2
<class 'Kitchen'>
Кухня: Кухня (12 м²)
True
False
```

### Разбор

**1. Чем обычный метод отличается от статического.**

Обычный метод **получает `self`** и работает с конкретным объектом: читает `self.name`, меняет `self.area`. Вызывается через объект: `room.info()`.

Статический метод **не получает `self`**. Он не работает с объектом, ему передают только аргументы. Вызывается через класс: `Room.is_valid_area(12)`. По сути — это обычная функция, живущая внутри класса.

**2. Чем статический метод отличается от классового.**

Статический метод (`@staticmethod`) не получает **ничего** — только аргументы, которые ему передали.

Классовый метод (`@classmethod`) получает **`cls`** — ссылку на класс. Через `cls` он может читать и менять **атрибуты класса**, создавать объекты через `cls(...)`.

**3. Почему `@classmethod` работает правильно при наследовании, а `@staticmethod` — нет.**

`@classmethod` получает **`cls`** — фактический класс, через который вызван. Когда мы вызываем `Kitchen.from_string(...)`, Python подставляет `cls = Kitchen`. Поэтому создаётся объект `Kitchen`, а не `Room`.

`@staticmethod` не получает **ничего**. Он не знает, через какой класс вызван. Если нужно, чтобы метод зависел от класса — используйте `@classmethod`.

**4. Как выбрать вид метода.**

- **Метод использует `self`?** → обычный метод.
- **Метод использует `cls`?** → `@classmethod`.
- **Метод не использует ни то, ни другое?** → `@staticmethod`.

# Глава 7. Абстрактные методы

## 7.1. Проблема: понятия, которых не существует

Вернёмся к нашему дому. У нас есть базовый класс `Room` и наследники `Kitchen`, `Bedroom`, `Bathroom`. У каждой комнаты — свой способ уборки.

Мы уже знаем (глава 4), что можно написать функцию, которая вызывает `room.clean()` у любого объекта. Но есть проблема: **как гарантировать, что у каждой комнаты точно есть метод `clean()`?**

```python
class Room:
    def info(self):
        return "Комната"

class Kitchen(Room):
    def clean(self):                    # есть clean
        return "Моем плиту и стол"

class Bedroom(Room):
    pass                                # забыли clean!
```

Если создать `Bedroom` и попробовать убрать:

```python
bedroom = Bedroom()
print(bedroom.clean())                  # AttributeError!
```

**Проблема:** мы **забыли** реализовать метод. Python не предупредил — класс `Bedroom` создался нормально. Ошибка всплыла **только в момент вызова** — когда программа уже работает.

Это плохо: ошибка обнаруживается слишком поздно. Хотелось бы, чтобы Python **сразу** сказал: «класс `Bedroom` не реализует метод `clean` — его нельзя создать».

Для этого и нужны **абстрактные методы**.

### Аналогия

Представь, что ты архитектор. У тебя есть **чертёж** «Комната». На чертеже написано: «у каждой комнаты должна быть дверь и окно». Ты не можешь построить «просто комнату» — чертёж описывает **общее понятие**, а не конкретную комнату.

Чтобы построить конкретную комнату, нужно **достроить** чертёж: «вот здесь дверь, вот здесь окно». Только после этого можно строить.

Абстрактный класс — это **недостроенный чертёж**. Он описывает, **что должно быть** у наследников, но **не реализует** это. Наследники **обязаны достроить**.

---

## 7.2. Что такое абстрактный класс

**Абстрактный класс** — это класс, который:

- описывает **общее понятие**, которое само по себе не существует;
- **не может быть создан напрямую** (нельзя сделать объект);
- служит **шаблоном** для наследников;
- может требовать от наследников **реализации** определённых методов.

### Примеры из жизни

| Абстрактное понятие | Конкретные воплощения |
|---------------------|----------------------|
| «Комната» | Кухня, спальня, ванная |
| «Животное» | Кошка, собака, попугай |
| «Геометрическая фигура» | Круг, квадрат, треугольник |
| «Транспорт» | Машина, самолёт, поезд |
| «Платёж» | Оплата картой, наличными, переводом |

Можно ли «просто построить комнату»? Нет — комната бывает **кухней**, **спальней** или **ванной**. «Просто комната» — это абстракция.

Можно ли «просто нарисовать фигуру»? Нет — фигура бывает **кругом**, **квадратом** или **треугольником**.

### Абстрактный метод

**Абстрактный метод** — это метод, у которого **нет реализации**. Он объявлен, но вместо тела стоит только **описание сигнатуры**.

Наследники **обязаны** реализовать абстрактный метод. Если не реализуют — их объект тоже нельзя будет создать.

### Схема

```
        Room (абстрактный)
        ├── clean()        ← абстрактный метод (нет реализации)
        └── info()         ← обычный метод (есть реализация)
                │
        ┌───────┼───────┐
        ↓       ↓       ↓
    Kitchen  Bedroom  Bathroom
    (clean)  (clean)  (clean)
    реализует реализует реализует
```

---

## 7.3. Инструменты в Python: модуль `abc`

В Python инструменты для абстрактных классов находятся в модуле **`abc`** (Abstract Base Classes).

### Что нужно импортировать

```python
from abc import ABC, abstractmethod
```

Три ключевых компонента:

| Компонент | Что это |
|-----------|---------|
| **`ABC`** | Базовый класс для абстрактных классов |
| **`@abstractmethod`** | Декоратор для абстрактных методов |
| **`ABCMeta`** | Метакласс (обычно не нужен напрямую) |

### Как объявить абстрактный класс

```python
from abc import ABC, abstractmethod

class Room(ABC):                       # наследуем от ABC
    @abstractmethod
    def clean(self):                   # абстрактный метод
        pass
```

Разберём по частям:

- **`class Room(ABC):`** — наследуем от `ABC`. Это делает класс абстрактным.
- **`@abstractmethod`** — декоратор, помечающий метод как абстрактный.
- **`def clean(self): pass`** — метод **объявлен**, но **не реализован**. Вместо тела — `pass`.

### Что теперь нельзя

Попытка создать объект абстрактного класса вызовет ошибку:

```python
room = Room()
# TypeError: Can't instantiate abstract class Room with abstract method clean
```

Python **не даст** создать объект, пока не реализованы все абстрактные методы.

---

## 7.4. Наследники обязаны реализовать абстрактный метод

Абстрактный класс описывает **что должно быть**, но не **как**. Наследники обязаны реализовать все абстрактные методы.

### Пример

```python
from abc import ABC, abstractmethod

class Room(ABC):
    @abstractmethod
    def clean(self):
        pass

class Kitchen(Room):
    def clean(self):                   # реализация
        return "Моем плиту и стол"

class Bedroom(Room):
    def clean(self):                   # реализация
        return "Пылесосим и заправляем кровать"
```

Теперь:

```python
kitchen = Kitchen()                    # ✅ работает
print(kitchen.clean())                 # Моем плиту и стол

bedroom = Bedroom()                    # ✅ работает
print(bedroom.clean())                 # Пылесосим и заправляем кровать
```

### Что если наследник НЕ реализует

Если наследник **не реализует** абстрактный метод, его объект **тоже нельзя создать**:

```python
class BadRoom(Room):
    pass                                # забыли clean

bad = BadRoom()
# TypeError: Can't instantiate abstract class BadRoom with abstract method clean
```

**Ключевая идея:** Python проверяет **при создании объекта**, все ли абстрактные методы реализованы. Если нет — объект создать нельзя.

### Схема

```
        Room (ABC)
        ├── clean()   ← абстрактный
                │
        ┌───────┼───────┐
        ↓       ↓       ↓
    Kitchen  Bedroom  BadRoom
    (clean)  (clean)  (нет clean)
        ✅      ✅       ❌
     создаётся создаётся  ошибка!
```

---

## 7.5. Абстрактный класс может содержать обычные методы

Абстрактный класс — это **не только абстрактные методы**. Он может содержать:

- **обычные методы** — с реализацией;
- **конструктор** — с параметрами;
- **атрибуты класса**;
- **свойства** — `@property`;
- **статические методы** — `@staticmethod`;
- **классовые методы** — `@classmethod`.

### Пример: абстрактный класс с обычным методом

```python
from abc import ABC, abstractmethod

class Room(ABC):
    def __init__(self, name, area):
        self.name = name
        self.area = area

    @abstractmethod
    def clean(self):
        pass

    # Обычный метод — есть реализация.
    def info(self):
        return f"{self.name}: {self.area} м²"
```

Наследник:

```python
class Kitchen(Room):
    def clean(self):
        return "Моем плиту и стол"


kitchen = Kitchen("Кухня", 12)
print(kitchen.clean())                 # Моем плиту и стол
print(kitchen.info())                  # Кухня: 12 м² — унаследован от Room
```

### Что наследуется

- **Обычные методы** — наследуются и работают.
- **Конструктор** — вызывается через `super().__init__()`.
- **Атрибуты класса** — наследуются.
- **Абстрактные методы** — **обязательно реализовать** в наследнике.

### Схема

```
        Room (ABC)
        ├── __init__()     ← обычный (наследуется)
        ├── clean()        ← абстрактный (обязателен)
        └── info()         ← обычный (наследуется)
```

---

## 7.6. Абстрактные свойства (`@property` + `@abstractmethod`)

Иногда нужно, чтобы наследники реализовали не метод, а **свойство** (см. главу 2). Для этого комбинируют `@property` и `@abstractmethod`.

### Синтаксис

```python
from abc import ABC, abstractmethod

class Room(ABC):
    @property
    @abstractmethod
    def clean_cost(self):
        # абстрактное свойство — только для чтения
        pass
```

**Важен порядок декораторов:** `@property` — **сверху**, `@abstractmethod` — **снизу**. Это связано с тем, как Python применяет декораторы (снизу вверх).

### Наследники обязаны реализовать

```python
class Kitchen(Room):
    @property
    def clean_cost(self):
        return 500


kitchen = Kitchen()
print(kitchen.clean_cost)              # 500
```

### Если наследник не реализует

```python
class BadRoom(Room):
    pass

BadRoom()
# TypeError: Can't instantiate abstract class BadRoom with abstract method clean_cost
```

### Абстрактное свойство с сеттером

Если нужно, чтобы свойство было **изменяемым**, можно определить абстрактный сеттер:

```python
class Room(ABC):
    @property
    @abstractmethod
    def area(self):
        pass

    @area.setter
    @abstractmethod
    def area(self, value):
        pass
```

Теперь наследник **обязан** реализовать и геттер, и сеттер.

---

## 7.7. Абстрактные статические и классовые методы

Абстрактными можно делать **не только обычные методы**, но и статические, и классовые.

### Абстрактный статический метод

```python
from abc import ABC, abstractmethod

class Room(ABC):
    @staticmethod
    @abstractmethod
    def is_valid_area(area):
        # абстрактный статический метод
        pass


class Kitchen(Room):
    @staticmethod
    def is_valid_area(area):
        return 8 <= area <= 100
```

### Абстрактный классовый метод

```python
class Room(ABC):
    @classmethod
    @abstractmethod
    def from_string(cls, data):
        # абстрактный классовый метод
        pass


class Kitchen(Room):
    def __init__(self, name, area):
        self.name = name
        self.area = area

    @classmethod
    def from_string(cls, data):
        name, area = data.split(";")
        return cls(name, int(area))
```

### Нюанс: порядок декораторов

Когда вы комбинируете `@staticmethod` или `@classmethod` с `@abstractmethod`:

- **`@classmethod`** — **сверху**, `@abstractmethod` — **снизу**:

```python
@classmethod
@abstractmethod
def from_string(cls, data):
    pass
```

- **`@staticmethod`** — **сверху**, `@abstractmethod` — **снизу**:

```python
@staticmethod
@abstractmethod
def is_valid_area(area):
    pass
```

Порядок важен: `@abstractmethod` должен **внутри**, чтобы он пометил функцию до того, как её обернут.

### Когда использовать

- **Абстрактный статический метод** — если наследники обязаны реализовать утилиту.
- **Абстрактный классовый метод** — если наследники обязаны реализовать фабрику.

На практике это используется **редко** — обычно абстрактными делают **обычные методы** и **свойства**.

---

## 7.8. Несколько абстрактных методов

Класс может содержать **сколько угодно** абстрактных методов. Наследник обязан реализовать **все**.

### Пример

```python
from abc import ABC, abstractmethod

class Room(ABC):
    @abstractmethod
    def clean(self):
        pass

    @abstractmethod
    def info(self):
        pass

    @abstractmethod
    def is_big(self):
        pass


class Kitchen(Room):
    def __init__(self, name, area):
        self.name = name
        self.area = area

    def clean(self):
        return "Моем плиту и стол"

    def info(self):
        return f"Кухня: {self.name}"

    def is_big(self):
        return self.area > 50
```

Если наследник реализует **не все** абстрактные методы, объект создать нельзя:

```python
class BadRoom(Room):
    def clean(self):
        return "Убираемся"
    # info и is_big не реализованы

BadRoom()
# TypeError: Can't instantiate abstract class BadRoom
# with abstract methods info, is_big
```

Python **перечислит все нереализованные методы** — удобно для отладки.

---

## 7.9. Абстрактные классы и наследование

Абстрактные классы **вписываются в иерархию наследования** как обычные. Можно строить **цепочки**:

### Цепочка абстрактных классов

```python
from abc import ABC, abstractmethod

class HomeItem(ABC):
    @abstractmethod
    def describe(self):
        pass

class Room(HomeItem):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def clean(self):
        pass

class Kitchen(Room):
    def describe(self):
        return f"Кухня: {self.name}"

    def clean(self):
        return "Моем плиту и стол"
```

Разберём:

- `HomeItem` — абстрактный класс с одним абстрактным методом `describe`.
- `Room` — тоже абстрактный (не реализует `describe`), добавляет ещё один абстрактный метод `clean`.
- `Kitchen` — **конкретный**: реализует **оба** абстрактных метода.

### Схема

```
        HomeItem (ABC)
        ├── describe()  ← абстрактный
                │
        Room (ABC)
        ├── describe()  ← абстрактный (унаследован)
        ├── clean()     ← абстрактный (новый)
                │
        ┌───────┼───────┐
        ↓       ↓       ↓
    Kitchen  Bedroom  Bathroom
    (оба)    (оба)    (оба)
    реализует реализует реализует
```

### Правило

**Класс абстрактный, пока не реализованы все абстрактные методы** — унаследованные и свои.

---

## 7.10. Зачем нужны абстрактные классы

Разберём **четыре главные причины**.

### 1. Гарантия интерфейса

Абстрактный класс **гарантирует**, что у всех наследников есть нужные методы.

```python
from abc import ABC, abstractmethod

class Room(ABC):
    @abstractmethod
    def clean(self):
        pass
```

Теперь **любой** наследник `Room` обязан реализовать `clean`. Если забудет — Python не даст создать объект.

### 2. Раннее обнаружение ошибок

Ошибка видна **при создании класса**, а не **в момент вызова метода**.

```python
class BadRoom(Room):
    pass                                # забыли clean

BadRoom()                               # TypeError — сразу видно!
```

Это гораздо лучше, чем утиная типизация, где ошибка всплывает только в рантайме.

### 3. Документация

Абстрактный класс **документирует** интерфейс: «вот что должны уметь все комнаты». Читая код, программист сразу видит, что обязательно реализовать.

### 4. Общий код

Абстрактный класс может содержать **общую логику**, которую наследники получают бесплатно:

```python
class Room(ABC):
    def __init__(self, name, area):
        self.name = name
        self.area = area

    @abstractmethod
    def clean(self):
        pass

    def info(self):                     # общий метод
        return f"{self.name}: {self.area} м²"
```

Наследники не дублируют `info()` — он общий.

### Сравнение с утиной типизацией

| Критерий | Утиная типизация | Абстрактный класс |
|----------|------------------|-------------------|
| Проверка при создании | ❌ Нет | ✅ Есть |
| Ошибка видна | В рантайме | При создании |
| Гарантия интерфейса | ❌ Нет | ✅ Есть |
| Гибкость | ✅ Высокая | ⚠️ Ограничена |
| Документация | Неявная | Явная |

**Вывод:** абстрактные классы **безопаснее**, но **менее гибкие**. В Python предпочитают **комбинировать**: абстрактные классы для «официального» интерфейса, утиную типизацию для «неофициальных» случаев.

---

## 7.11. Абстрактный класс vs обычный класс

Чтобы не путаться, сравним абстрактный и обычный класс.

| Критерий | Обычный класс | Абстрактный класс |
|----------|---------------|-------------------|
| Создать объект | ✅ Можно | ❌ Нельзя |
| Наследовать | ✅ Можно | ✅ Можно |
| Содержит абстрактные методы | ❌ Нет | ✅ Да |
| Содержит обычные методы | ✅ Да | ✅ Да |
| Обязателен для наследников | ⚠️ Реализация методов не обязательна | ✅ Все абстрактные методы обязательны |
| Родитель | `object` | `ABC` |

### Когда использовать

- **Обычный класс** — если понятие конкретное (есть объекты).
- **Абстрактный класс** — если понятие общее (объектов нет, есть только «виды»).

---

## 7.12. Абстрактные методы в реальных проектах

Где абстрактные классы используются на практике?

### 1. Геометрические фигуры

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass
```

Все фигуры обязаны реализовать `area` и `perimeter`. Легко добавить новые фигуры — без изменения остального кода.

### 2. Платёжные системы

```python
class PaymentProcessor(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CardPayment(PaymentProcessor):
    def pay(self, amount):
        return f"Оплата картой: {amount} ₽"

class CryptoPayment(PaymentProcessor):
    def pay(self, amount):
        return f"Оплата криптой: {amount} ₽"
```

Все платёжные системы обязаны реализовать `pay`.

### 3. Логгеры

```python
class Logger(ABC):
    @abstractmethod
    def log(self, message):
        pass

class FileLogger(Logger): ...
class ConsoleLogger(Logger): ...
class EmailLogger(Logger): ...
```

### 4. Игровые объекты

```python
class GameObject(ABC):
    @abstractmethod
    def update(self):
        pass

    @abstractmethod
    def render(self):
        pass
```

### 5. Наш дом

```python
class Room(ABC):
    @abstractmethod
    def clean(self):
        pass
```

Все комнаты обязаны реализовать уборку.

### Общий признак

Используйте абстрактный класс, если:

- есть **общее понятие** без конкретного воплощения;
- у всех «видов» должен быть **одинаковый интерфейс**;
- важно **гарантировать**, что метод реализован.

---

## 7.13. Нюансы и тонкости

### 7.13.1. `ABC` vs `ABCMeta`

Есть два способа создать абстрактный класс:

**Через `ABC` (современный способ):**

```python
from abc import ABC, abstractmethod

class Room(ABC):
    @abstractmethod
    def clean(self):
        pass
```

**Через `ABCMeta` (старый способ):**

```python
from abc import ABCMeta, abstractmethod

class Room(metaclass=ABCMeta):
    @abstractmethod
    def clean(self):
        pass
```

Оба работают одинаково. **Современный способ** — через `ABC` — короче и понятнее. Используйте его.

### 7.13.2. Абстрактный метод может содержать код

Формально абстрактный метод **может содержать код** — просто наследник **обязан** его переопределить.

```python
class Room(ABC):
    @abstractmethod
    def clean(self):
        # какой-то общий код
        return "Убираемся"
```

Однако на практике в абстрактных методах обычно пишут `pass` — потому что наследники всё равно переопределяют. Если нужен общий код — лучше вынести его в **обычный метод** (см. 7.5).

### 7.13.3. `super()` в наследниках

Абстрактный метод можно вызвать через `super()`, если он содержит код:

```python
class Kitchen(Room):
    def clean(self):
        return f"{super().clean()} — моем плиту"
```

Но если в абстрактном методе `pass` — это бессмысленно.

### 7.13.4. `isinstance` и абстрактные классы

`isinstance` работает с абстрактными классами:

```python
kitchen = Kitchen()
print(isinstance(kitchen, Room))        # True
print(isinstance(kitchen, ABC))         # True
```

Это логично: `Kitchen` наследует `Room`, а `Room` наследует `ABC`.

### 7.13.5. Нельзя создать объект абстрактного класса

**Даже если** все методы реализованы, объект самого абстрактного класса создать нельзя:

```python
Room()                                  # TypeError
```

Только **наследники** могут быть созданы — и только если реализовали **все** абстрактные методы.

### 7.13.6. Порядок декораторов

Запомните: `@property`, `@staticmethod`, `@classmethod` — **сверху**, `@abstractmethod` — **снизу**:

```python
@property
@abstractmethod
def area(self): pass

@classmethod
@abstractmethod
def from_string(cls, data): pass

@staticmethod
@abstractmethod
def is_valid_area(area): pass
```

Если перепутать — Python выдаст ошибку или абстрактность не сработает.

---

## 7.14. Типичные ошибки

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Забыли наследовать от `ABC` | Абстрактность не работает | `class Room(ABC):` |
| Забыли `@abstractmethod` | Метод не абстрактный | Добавить декоратор |
| Пытаются создать объект `Room()` | Абстрактный класс нельзя | Создавать объекты наследников |
| Наследник не реализует метод | `TypeError` при создании | Реализовать все абстрактные методы |
| Путают порядок декораторов | `@abstractmethod` должен быть снизу | Сначала `@property`, потом `@abstractmethod` |
| Пишут `raise NotImplementedError` | Это **не** абстрактный метод | Использовать `@abstractmethod` |
| Думают, что абстрактный метод = приватный | Это разные вещи | Абстрактный — обязателен, приватный — скрыт |
| Делают абстрактными всё подряд | Избыточно | Только то, что обязаны реализовать наследники |
| Не реализуют **все** абстрактные методы | `TypeError` | Реализовать все |
| Используют `ABCMeta` вместо `ABC` | Работает, но старый стиль | Использовать `ABC` |

### Нюанс про `NotImplementedError`

Иногда новички пишут так:

```python
class Room:
    def clean(self):
        raise NotImplementedError("Subclass must implement")
```

Это **не абстрактный метод**. Класс `Room` можно создать, объект создать, и ошибка всплывёт **только при вызове `clean()`**.

**Абстрактный метод через `@abstractmethod`** — лучше: ошибка видна **при создании объекта наследника**, если метод не реализован.

---

## 7.15. Что важно запомнить

- **Абстрактный класс** — описывает общее понятие, не может быть создан напрямую.
- **Абстрактный метод** — метод без реализации, обязательный для наследников.
- **Модуль `abc`:** `ABC` и `@abstractmethod`.
- **Синтаксис:** `class Room(ABC):`, `@abstractmethod`.
- **Наследник обязан** реализовать **все** абстрактные методы.
- **Иначе** — `TypeError` при создании объекта.
- **Абстрактный класс может содержать** обычные методы, конструктор, атрибуты.
- **Абстрактные свойства:** `@property` + `@abstractmethod`.
- **Порядок декораторов:** `@property` сверху, `@abstractmethod` снизу.
- **Абстрактные классы** можно наследовать **цепочкой**.
- **Гарантия интерфейса** — главное преимущество.
- **Ошибка видна раньше**, чем при утиной типизации.
- **`raise NotImplementedError`** — не абстрактный метод.

## Задача 7 (абстрактные методы)

Создайте абстрактный класс `HomeItem` (предмет в доме), который описывает **общее понятие** предмета, но **не может быть создан напрямую**.

**Часть 1.** Импортируйте `ABC` и `abstractmethod` из модуля `abc`. Создайте абстрактный класс `HomeItem(ABC)` с абстрактным методом `describe()` — он должен возвращать краткое описание предмета.

**Часть 2.** Реализуйте в `HomeItem` **обычный** конструктор `__init__(name)`, который принимает название предмета и сохраняет его в `self.name`. Добавьте **обычный** метод `show_name()`, возвращающий строку `Название: <name>`.

**Часть 3.** Создайте двух наследников:

- `Furniture(HomeItem)` — с полем `material` (материал) и методом `describe()`, возвращающим строку `Мебель: <name> из <material>`.
- `Appliance(HomeItem)` — с полем `power` (мощность) и методом `describe()`, возвращающим строку `Техника: <name>, мощность <power> Вт`.

**Часть 4.** Проверьте, что:

- объект `HomeItem("...")` **создать нельзя** — поймайте `TypeError`;
- объекты `Furniture` и `Appliance` **создаются** и работают;
- унаследованный метод `show_name()` доступен у наследников.

**Часть 5.** Создайте **третий** класс-наследник `BadItem(HomeItem)`, который **не реализует** метод `describe()`. Убедитесь, что объект `BadItem("...")` **тоже нельзя создать** — Python должен выдать `TypeError` с указанием, какие абстрактные методы не реализованы.

**Часть 6.** Дополнительно создайте класс `Building(ABC)` с **абстрактным свойством** `total_area` (общая площадь). Реализуйте наследника `House(Building)` со свойством `total_area`, возвращающим конкретное число.

**Часть 7.** В комментариях объясните:

1. Зачем нужны абстрактные классы.
2. Что произойдёт, если наследник не реализует абстрактный метод.
3. Чем `@abstractmethod` лучше, чем `raise NotImplementedError`.
4. Почему абстрактный класс может содержать обычные методы.
5. Почему порядок декораторов `@property` + `@abstractmethod` важен.

### Решение

```python
from abc import ABC, abstractmethod


# ─── Часть 1, 2. Абстрактный класс HomeItem ───
class HomeItem(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def describe(self):
        # Абстрактный метод: наследники обязаны реализовать.
        pass

    # Обычный метод — есть реализация, наследуется.
    def show_name(self):
        return f"Название: {self.name}"


# ─── Часть 3. Наследники ───
class Furniture(HomeItem):
    def __init__(self, name, material):
        super().__init__(name)
        self.material = material

    def describe(self):
        return f"Мебель: {self.name} из {self.material}"


class Appliance(HomeItem):
    def __init__(self, name, power):
        super().__init__(name)
        self.power = power

    def describe(self):
        return f"Техника: {self.name}, мощность {self.power} Вт"


# ─── Часть 5. Наследник без реализации ───
class BadItem(HomeItem):
    pass                                # describe не реализован


# ─── Часть 6. Абстрактное свойство ───
class Building(ABC):
    @property
    @abstractmethod
    def total_area(self):
        # Абстрактное свойство: только для чтения.
        pass


class House(Building):
    @property
    def total_area(self):
        return 120


# ─── Часть 4. Проверка ───

# 1. Объект абстрактного класса создать нельзя.
try:
    item = HomeItem("Мебель")
except TypeError as e:
    print("Нельзя создать HomeItem:", e)

# 2. Объекты наследников создаются.
table = Furniture("Стол", "дуб")
fridge = Appliance("Холодильник", 200)

print(table.describe())         # Мебель: Стол из дуб
print(fridge.describe())        # Техника: Холодильник, мощность 200 Вт

# 3. Унаследованный метод доступен.
print(table.show_name())        # Название: Стол
print(fridge.show_name())       # Название: Холодильник

# 4. Наследник без реализации тоже нельзя создать.
try:
    bad = BadItem("Что-то")
except TypeError as e:
    print("Нельзя создать BadItem:", e)

# 5. Абстрактное свойство.
house = House()
print("Общая площадь дома:", house.total_area)   # 120
```

**Ожидаемый вывод:**

```
Нельзя создать HomeItem: Can't instantiate abstract class HomeItem with abstract method describe
Мебель: Стол из дуб
Техника: Холодильник, мощность 200 Вт
Название: Стол
Название: Холодильник
Нельзя создать BadItem: Can't instantiate abstract class BadItem with abstract method describe
Общая площадь дома: 120
```

### Разбор

**1. Зачем нужны абстрактные классы.**

Абстрактные классы решают **четыре задачи**:

- **Гарантия интерфейса** — все наследники обязаны реализовать нужные методы.
- **Раннее обнаружение ошибок** — если наследник забыл метод, Python скажет **при создании объекта**.
- **Документация** — читая код, программист сразу видит, что обязательно.
- **Общий код** — общие методы и конструктор наследуются бесплатно.

**2. Что произойдёт, если наследник не реализует абстрактный метод.**

Python выдаст `TypeError` **при создании объекта**. В сообщении будет указано, какие именно абстрактные методы не реализованы. Например:

```
TypeError: Can't instantiate abstract class BadItem with abstract method describe
```

Это **гораздо лучше**, чем утиная типизация, где ошибка всплыла бы только **в момент вызова**.

**3. Чем `@abstractmethod` лучше, чем `raise NotImplementedError`.**

`raise NotImplementedError` — это **обычный метод**, который бросает исключение. Проблемы:

- Объект **можно создать** — ошибка всплывёт только **при вызове**.
- Нет **гарантии интерфейса** — наследник может забыть переопределить метод, и ничего не произойдёт.
- IDE не подскажет, что метод нужно реализовать.

`@abstractmethod`:

- Объект **нельзя создать**, пока метод не реализован.
- **Гарантия интерфейса** — Python сам проверит.
- IDE **подсказывает**, что метод абстрактный.

**4. Почему абстрактный класс может содержать обычные методы.**

Абстрактный класс описывает **общее понятие**, но у него может быть **общая логика**:

- **Конструктор** — общий для всех наследников.
- **Общий метод** `info()` — работает одинаково для всех.
- **Общие атрибуты** — наследуются.

Наследники получают эту логику **бесплатно** — не нужно дублировать код. Абстрактные методы — это то, что **должны реализовать** наследники. Обычные — то, что **даётся** наследникам.

**5. Почему порядок декораторов `@property` + `@abstractmethod` важен.**

Декораторы в Python применяются **снизу вверх**. То есть:

```python
@property
@abstractmethod
def area(self):
    pass
```

читается так: сначала применяется `@abstractmethod` к функции `area`, потом — `@property` к результату.

Если поменять порядок:

```python
@abstractmethod
@property       # ← неправильно
def area(self):
    pass
```

— Python выдаст ошибку, потому что `@abstractmethod` ожидает **функцию**, а получает **объект property**.

**Правило:** `@property`, `@staticmethod`, `@classmethod` — **сверху**, `@abstractmethod` — **снизу**.

# Глава 8. Композиция

## 8.1. Проблема: как описывать сложные объекты

Вернёмся к нашему дому. У нас есть класс `Room` с полями `name` и `area`. Но дом — это не просто одна комната. Дом **состоит** из множества комнат: кухня, спальня, ванная, коридор.

Как это описать в коде? Первое, что приходит в голову — добавить все комнаты как отдельные поля:

```python
class House:
    def __init__(self):
        self.kitchen = Kitchen()
        self.bedroom = Bedroom()
        self.bathroom = Bathroom()
```

Это работает, но плохо масштабируется. Если комнат десять — десять полей. Если двадцать — двадцать. А если количество комнат заранее неизвестно? Такой подход не подходит.

Второй вариант — использовать **список**:

```python
class House:
    def __init__(self, rooms):
        self.rooms = rooms
```

Но здесь возникает вопрос: **кто создаёт комнаты?** Если дом создаёт их сам — это **композиция**. Если комнаты передаются снаружи и могут существовать отдельно — это **агрегация**.

В этой главе мы разберём **композицию** — отношение, при котором один объект **содержит** другой как **часть себя**.

---

## 8.2. Что такое композиция

**Композиция** — это отношение между объектами, при котором:

1. Один объект **содержит** другой объект как **часть себя**.
2. Внутренний объект **создаётся вместе** с внешним.
3. Внутренний объект **не существует** отдельно от внешнего.
4. Если внешний объект удаляется — внутренний **тоже исчезает**.

Простыми словами: композиция — это отношение **«часть целого»** (part-of).

### Аналогия

- **Дом и комната.** Комната — часть дома. Без дома комнаты не бывает.
- **Машина и двигатель.** Двигатель — часть машины. Без машины двигатель не нужен.
- **Книга и страница.** Страница — часть книги. Без книги страницы нет.
- **Человек и сердце.** Сердце — часть человека. Без человека сердца нет.

### Ключевая идея

Композиция — это **жёсткая связь**: внутренний объект **принадлежит** внешнему. Он создаётся **внутри** и **умирает вместе** с ним.

---

## 8.3. Композиция vs агрегация

Самое важное — понять **разницу** между композицией и агрегацией. Их часто путают.

### Определения

| Понятие | Как создаётся внутренний объект | Может существовать отдельно? |
|---------|--------------------------------|------------------------------|
| **Композиция** | Внутри конструктора внешнего | ❌ Нет |
| **Агрегация** | Передаётся снаружи | ✅ Да |

### Пример композиции

```python
class Kitchen:
    def show(self):
        print("Кухня с плитой")

class House:
    def __init__(self):
        self.kitchen = Kitchen()    # создаётся внутри — композиция
```

Здесь дом **создаёт** кухню сам. Кухня **не существует** без дома.

### Пример агрегации

```python
class Kitchen:
    def show(self):
        print("Кухня с плитой")

class House:
    def __init__(self, kitchen):
        self.kitchen = kitchen      # передаётся снаружи — агрегация
```

Здесь кухня **передаётся** в дом извне. Она может существовать и **без** дома. Её можно переставить в другой дом.

### Схема

```
        Композиция                  Агрегация
        ──────────                  ─────────
        House                       Kitchen (отдельно)
         └── Kitchen                  ↓
            (создаётся внутри)      House
                                     └── Kitchen
                                        (передаётся снаружи)
```

### Как определить

Задайте вопрос: **«Если внешний объект исчезнет, внутренний тоже должен исчезнуть?»**

- **Да** → композиция.
- **Нет** → агрегация.

### Другие примеры

| Отношение | Тип | Почему |
|-----------|-----|--------|
| Дом — комната | Композиция | Комната не существует без дома |
| Дом — мебель | Агрегация | Мебель можно переставить в другой дом |
| Книга — страница | Композиция | Страница не существует без книги |
| Библиотека — книга | Агрегация | Книга может быть в другой библиотеке |
| Машина — двигатель | Композиция | Двигатель создан для этой машины |
| Человек — питомец | Агрегация | Питомец может жить у другого хозяина |

---

## 8.4. Три типа отношений между объектами

Композиция и агрегация — **не единственные** типы отношений. Всего их **три**:

| Тип | Отношение | Пример |
|-----|-----------|--------|
| **Ассоциация** | «использует» | Врач — пациент |
| **Агрегация** | «содержит» (слабая связь) | Дом — мебель |
| **Композиция** | «состоит из» (сильная связь) | Дом — комната |

### Ассоциация

**Ассоциация** — самый слабый тип связи. Объекты **знают** друг о друге, но **не владеют** друг другом.

```python
class Doctor:
    def __init__(self, name):
        self.name = name
        self.patients = []          # список пациентов

    def add_patient(self, patient):
        self.patients.append(patient)

class Patient:
    def __init__(self, name):
        self.name = name
```

Врач и пациент **связаны**, но **не владеют** друг другом. Пациент может быть у другого врача, врач может иметь других пациентов.

### Агрегация

**Агрегация** — более сильная связь. Один объект **содержит** другой, но внутренний объект **может существовать отдельно**.

```python
class Furniture:
    def __init__(self, name):
        self.name = name

class House:
    def __init__(self, furniture_list):
        self.furniture = furniture_list     # агрегация
```

Дом **содержит** мебель, но мебель **не принадлежит** дому. Её можно вынести и поставить в другой дом.

### Композиция

**Композиция** — самая сильная связь. Внутренний объект **принадлежит** внешнему и **не существует** отдельно.

```python
class Room:
    def __init__(self, name):
        self.name = name

class House:
    def __init__(self):
        self.rooms = [Room("Кухня"), Room("Спальня")]   # композиция
```

Дом **создаёт** комнаты сам. Без дома комнаты не нужны.

### Схема

```
        Ассоциация         Агрегация          Композиция
        ──────────         ──────────         ──────────
        A ─── B            A ◇─── B           A ◆─── B
        (знает)            (содержит)         (состоит из)
        слабая             средняя            сильная
```

Символы UML:
- **Ассоциация** — простая линия.
- **Агрегация** — пустой ромб.
- **Композиция** — закрашенный ромб.

---

## 8.5. Как реализовать композицию в Python

В Python композиция реализуется **очень просто**: внутренний объект создаётся **внутри** конструктора внешнего.

### Простой пример

```python
class Kitchen:
    def show(self):
        print("Кухня с плитой и холодильником")


class House:
    def __init__(self, address):
        self.address = address
        self.kitchen = Kitchen()    # ← композиция

    def show_house(self):
        print(f"Адрес: {self.address}")
        self.kitchen.show()
```

Использование:

```python
house = House("ул. Ленина, д. 10")
house.show_house()
```

**Ожидаемый вывод:**

```
Адрес: ул. Ленина, д. 10
Кухня с плитой и холодильником
```

Кухня **создаётся внутри** `House`. Снаружи её не видно.

### Несколько внутренних объектов

Дом может содержать **несколько** комнат:

```python
class Kitchen:
    def show(self):
        print("Кухня: плита, холодильник")

class Bedroom:
    def show(self):
        print("Спальня: кровать, шкаф")

class Bathroom:
    def show(self):
        print("Ванная: ванна, раковина")


class House:
    def __init__(self, address):
        self.address = address
        # Создаём все комнаты внутри — композиция.
        self.kitchen = Kitchen()
        self.bedroom = Bedroom()
        self.bathroom = Bathroom()

    def show_house(self):
        print(f"Дом по адресу: {self.address}")
        self.kitchen.show()
        self.bedroom.show()
        self.bathroom.show()
```

```python
house = House("ул. Ленина, д. 10")
house.show_house()
```

### Список внутренних объектов

Если комнат может быть **много** — используем список:

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area

    def show(self):
        print(f"{self.name}: {self.area} м²")


class House:
    def __init__(self, address):
        self.address = address
        # Создаём список комнат внутри — композиция.
        self.rooms = [
            Room("Кухня", 12),
            Room("Спальня", 18),
            Room("Ванная", 6),
        ]

    def show_house(self):
        print(f"Дом по адресу: {self.address}")
        for room in self.rooms:
            room.show()
```

### Схема

```
        House
         ├── address: "ул. Ленина, д. 10"
         └── rooms: [
                Room("Кухня", 12),
                Room("Спальня", 18),
                Room("Ванная", 6)
             ]
```

Все комнаты **создаются внутри** `House`. Снаружи их не существует.

---

## 8.6. Зачем нужна композиция

Композиция решает **четыре задачи**.

### 1. Скрытие сложности

Снаружи видно только `House`. Внутри он сам управляет комнатами. Пользователю не нужно знать, как устроен дом.

```python
house = House("ул. Ленина, д. 10")
house.show_house()      # снаружи не видно, что внутри три комнаты
```

### 2. Обеспечение целостности

Если дом создан — **все его комнаты тоже созданы**. Нельзя создать дом без кухни. Это гарантирует **целостность** объекта.

### 3. Избежание жёсткого наследования

Отношение «дом — комната» — это **не** «дом **является** комнатой». Это «дом **содержит** комнаты». Наследование здесь не подходит — нужна композиция.

### 4. Переиспользование кода

Класс `Room` можно использовать **где угодно** — в доме, в офисе, в отеле. Он не привязан к конкретному «внешнему» классу.

---

## 8.7. Композиция vs наследование

Это **ключевой выбор** в ООП. Разберём подробно.

### Отношение

| Отношение | Что использовать |
|-----------|------------------|
| «A **является** B» | **Наследование** |
| «A **содержит** B» | **Композиция** |

### Пример: правильно и неправильно

**Правильно (наследование):**

```python
class Room:
    def info(self):
        return "Комната"

class Kitchen(Room):        # Кухня ЯВЛЯЕТСЯ комнатой
    pass
```

**Правильно (композиция):**

```python
class Room:
    def info(self):
        return "Комната"

class House:
    def __init__(self):
        self.rooms = [Room()]   # Дом СОДЕРЖИТ комнаты
```

**Неправильно (наследование там, где нужна композиция):**

```python
class House(Room):          # Дом НЕ ЯВЛЯЕТСЯ комнатой!
    pass
```

Это классическая ошибка новичков: использовать наследование **ради переиспользования кода**, а не по смыслу.

### Почему композиция часто лучше

| Критерий | Наследование | Композиция |
|----------|--------------|------------|
| Связь | Жёсткая | Гибкая |
| Изменение родителя | Ломает наследника | Не влияет |
| Замена внутреннего объекта | Невозможна | Легко |
| Множественное использование | Ограничено | Свободно |
| Понимание | Иногда путает | Яснее |

### Принцип «Предпочитайте композицию наследованию»

Это один из **принципов хорошего дизайна** (мы разберём его подробнее в главе 28). Смысл:

- Используйте наследование **только** когда есть чёткое отношение «является».
- Во всех остальных случаях — композиция.

### Пример из реальной жизни

**Плохо (наследование):**

```python
class Car(Engine):          # Машина НЕ ЯВЛЯЕТСЯ двигателем
    pass
```

**Хорошо (композиция):**

```python
class Car:
    def __init__(self):
        self.engine = Engine()   # Машина СОДЕРЖИТ двигатель
```

---

## 8.8. Композиция и делегирование

Часто при композиции внешний объект **делегирует** работу внутреннему. Это называется **делегирование**.

### Пример

```python
class Engine:
    def start(self):
        return "Двигатель запущен"

class Car:
    def __init__(self):
        self.engine = Engine()

    def start(self):
        # Делегируем работу двигателю.
        return self.engine.start()
```

Здесь `Car.start()` **делегирует** вызов `Engine.start()`. Снаружи кажется, что машина сама запускается, а на самом деле это делает двигатель.

### Зачем это нужно

- **Скрытие деталей.** Снаружи неважно, что под капотом.
- **Гибкость.** Можно заменить двигатель — код `Car` не изменится.
- **Переиспользование.** `Engine` можно использовать в другой машине.

### Схема

```
        car.start()
            ↓
        Car.start() — делегирует
            ↓
        self.engine.start()
            ↓
        Engine.start() — выполняет
```

---

## 8.9. Композиция и жизненный цикл

Композиция тесно связана с **жизненным циклом** объектов.

### Внутренний объект создаётся вместе с внешним

```python
class House:
    def __init__(self):
        self.kitchen = Kitchen()    # создаётся в конструкторе
```

Как только создан `House`, сразу создаётся `Kitchen`. Нельзя создать дом **без** кухни.

### Внутренний объект живёт, пока жив внешний

```python
house = House()          # создан дом и кухня
# ... используем ...
del house                # дом удалён — кухня тоже (сборщик мусора)
```

Когда дом удаляется, ссылка на кухню теряется. Если на кухню **никто не ссылается** — она удаляется сборщиком мусора.

### Нюанс: циклические ссылки

Если внутренний объект **ссылается на внешний**, может возникнуть **циклическая ссылка**:

```python
class Kitchen:
    def __init__(self, house):
        self.house = house      # кухня знает о доме

class House:
    def __init__(self):
        self.kitchen = Kitchen(self)    # дом знает о кухне
```

Это **нормально** для Python — сборщик мусора умеет обрабатывать циклы. Но нужно помнить, что такие ссылки **затрудняют** удаление объектов.

## 8.10. Композиция в реальных проектах

Где композиция используется на практике?

### 1. Графический интерфейс

```python
class Window:
    def __init__(self):
        self.title_bar = TitleBar()       # композиция
        self.menu = Menu()                # композиция
        self.content = ContentArea()      # композиция
```

Окно **состоит** из частей. Без окна части не нужны.

### 2. Игровые объекты

```python
class Player:
    def __init__(self):
        self.health = Health()            # композиция
        self.inventory = Inventory()      # композиция
        self.weapon = Weapon()            # композиция
```

Игрок **состоит** из здоровья, инвентаря, оружия.

### 3. Документы

```python
class Document:
    def __init__(self):
        self.header = Header()            # композиция
        self.body = Body()                # композиция
        self.footer = Footer()            # композиция
```

### 4. Заказы

```python
class Order:
    def __init__(self):
        self.items = [OrderItem()]        # композиция
        self.payment = Payment()          # композиция
        self.shipping = Shipping()        # композиция
```

### 5. Наш дом

```python
class House:
    def __init__(self):
        self.kitchen = Kitchen()          # композиция
        self.bedroom = Bedroom()          # композиция
        self.bathroom = Bathroom()        # композиция
```

### Общий признак

Композиция используется, когда:

- объект **состоит** из частей;
- части **создаются** вместе с объектом;
- части **не существуют** без объекта;
- снаружи видно только **целое**, а части скрыты.

## 8.11. Нюансы и тонкости

### 8.11.1. Композиция — не всегда список

Внутренний объект может быть **одним** или **списком**. Оба варианта — композиция.

**Один объект:**

```python
class House:
    def __init__(self):
        self.kitchen = Kitchen()
```

**Список объектов:**

```python
class House:
    def __init__(self):
        self.rooms = [Kitchen(), Bedroom(), Bathroom()]
```

### 8.11.2. Ленивая инициализация

Иногда внутренний объект создаётся **не сразу**, а при первом обращении:

```python
class House:
    def __init__(self):
        self._kitchen = None

    @property
    def kitchen(self):
        if self._kitchen is None:
            self._kitchen = Kitchen()    # создаётся при первом обращении
        return self._kitchen
```

Это называется **ленивая инициализация**. Полезно, если объект создаётся редко или долго.

### 8.11.3. Композиция и наследование вместе

Их можно **комбинировать**:

```python
class Room:
    def info(self):
        return "Комната"

class Kitchen(Room):                    # наследование
    def info(self):
        return "Кухня"

class House:
    def __init__(self):
        self.rooms = [Kitchen()]        # композиция
```

Здесь `Kitchen` **наследует** от `Room`, а `House` **содержит** `Kitchen`.

### 8.11.4. Композиция и `__init__`

Все внутренние объекты обычно создаются в `__init__`:

```python
class House:
    def __init__(self):
        self.kitchen = Kitchen()        # в конструкторе
        self.bedroom = Bedroom()
```

Если объект создаётся **вне** `__init__` — это уже **не композиция** (агрегация или ассоциация).

### 8.11.5. Композиция и удаление

Если внутренний объект **не должен удаляться** при удалении внешнего — это **агрегация**, а не композиция.

---

## 8.12. Типичные ошибки

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Путают композицию и агрегацию | Разное отношение владения | Спросить: «исчезнет ли внутренний объект?» |
| Передают внутренний объект снаружи | Это агрегация | Создавать внутри конструктора |
| Наследуют там, где нужна композиция | Нарушает логику «является» | Использовать композицию |
| Создают внутренний объект **вне** `__init__` | Это не композиция | Создавать в конструкторе |
| Делают все поля публичными | Скрытие теряется | Использовать `_name` или `__name` |
| Дублируют код внутреннего объекта | Нарушает DRY | Делегировать через композицию |
| Используют композицию для всего | Иногда нужно наследование | Выбирать по смыслу |
| Забывают инициализировать внутренний объект | `AttributeError` | Создавать в `__init__` |

---

## 8.13. Что важно запомнить

- **Композиция** — отношение «часть целого».
- **Внутренний объект** создаётся **внутри** внешнего.
- **Внутренний объект** не существует **отдельно**.
- **Три типа отношений:** ассоциация, агрегация, композиция.
- **Композиция** — самая сильная связь.
- **Агрегация** — слабее: объект может существовать отдельно.
- **Ассоциация** — ещё слабее: объекты просто знают друг о друге.
- **Правило:** «предпочитайте композицию наследованию».
- **Наследование** — «является», **композиция** — «содержит».
- **Композиция** обеспечивает **целостность** и **скрытие сложности**.
- **Внутренние объекты** обычно создаются в `__init__`.
- **Делегирование** — способ использовать композицию.

## Задача 8 (композиция)

У дома есть несколько комнат — дом **не может** существовать без комнат, и комнаты **создаются вместе** с домом. Это классический пример **композиции**.

**Часть 1.** Создайте класс `Room` с полями `name` (название) и `area` (площадь) и методом `show()`, который выводит строку вида `Комната <name>: <area> м²`.

**Часть 2.** Создайте класс `House`, который **содержит**:

- `address` — адрес дома;
- `rooms` — **список** комнат, созданных **внутри** конструктора (композиция).

Пусть по умолчанию дом содержит **три** комнаты: «Кухня» (12 м²), «Спальня» (18 м²), «Ванная» (6 м²).

**Часть 3.** Реализуйте в `House`:

- метод `show_house()` — выводит адрес и вызывает `show()` у каждой комнаты;
- метод `total_area()` — возвращает суммарную площадь всех комнат.

**Часть 4.** Создайте объект `house` и продемонстрируйте работу методов.

**Часть 5.** Создайте наследника `LuxuryHouse(House)`, который **добавляет** дополнительную комнату «Бассейн» (30 м²). Убедитесь, что:

- композиция сохраняется (комнаты создаются внутри);
- наследование и композиция **работают вместе**.

**Часть 6.** В комментариях объясните:

1. Почему это именно композиция, а не агрегация.
2. Чем композиция отличается от наследования.
3. Почему создание комнат в `__init__` важно для композиции.
4. Что произойдёт, если комнаты передавать снаружи.

### Решение

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area

    def show(self):
        print(f"Комната {self.name}: {self.area} м²")


class House:
    def __init__(self, address):
        self.address = address
        # Комнаты создаются ВНУТРИ конструктора — это композиция.
        self.rooms = [
            Room("Кухня", 12),
            Room("Спальня", 18),
            Room("Ванная", 6),
        ]

    def show_house(self):
        print(f"Дом по адресу: {self.address}")
        for room in self.rooms:
            room.show()

    def total_area(self):
        # Суммируем площади всех комнат.
        return sum(room.area for room in self.rooms)


class LuxuryHouse(House):
    def __init__(self, address):
        # Вызываем родительский конструктор — он создаст базовые комнаты.
        super().__init__(address)
        # Добавляем ещё одну комнату — композиция сохраняется.
        self.rooms.append(Room("Бассейн", 30))


# ─── Часть 4. Проверка ───
house = House("ул. Ленина, д. 10")
house.show_house()
print(f"Общая площадь: {house.total_area()} м²")

print()

# ─── Часть 5. Наследник ───
luxury = LuxuryHouse("ул. Садовая, д. 3")
luxury.show_house()
print(f"Общая площадь: {luxury.total_area()} м²")
```

**Ожидаемый вывод:**

```
Дом по адресу: ул. Ленина, д. 10
Комната Кухня: 12 м²
Комната Спальня: 18 м²
Комната Ванная: 6 м²
Общая площадь: 36 м²

Дом по адресу: ул. Садовая, д. 3
Комната Кухня: 12 м²
Комната Спальня: 18 м²
Комната Ванная: 6 м²
Комната Бассейн: 30 м²
Общая площадь: 66 м²
```

### Разбор

**1. Почему это именно композиция, а не агрегация.**

Комнаты создаются **внутри** конструктора `House`:

```python
self.rooms = [
    Room("Кухня", 12),
    ...
]
```

Мы **не передаём** их снаружи. Комнаты **не существуют** отдельно от дома. Если дом исчезнет — комнаты тоже.

Это **признак композиции**. При агрегации комнаты передавались бы в конструктор извне:

```python
def __init__(self, address, rooms):    # ← агрегация
    self.rooms = rooms
```

Тогда комнаты можно было бы создавать отдельно и переставлять между домами.

**2. Чем композиция отличается от наследования.**

| Композиция | Наследование |
|------------|--------------|
| «Дом **содержит** комнаты» | «Кухня **является** комнатой» |
| Внутренний объект — часть внешнего | Наследник — разновидность родителя |
| Гибкая связь | Жёсткая связь |
| Внутренний объект можно заменить | Родителя заменить нельзя |

В нашем примере **и то, и другое** используется правильно:

- `LuxuryHouse(House)` — **наследование** (люксовый дом **является** домом).
- `self.rooms = [Room(...)]` — **композиция** (дом **содержит** комнаты).

Это классический пример **совместного** использования наследования и композиции.

**3. Почему создание комнат в `__init__` важно для композиции.**

Создание в `__init__` — **ключевой признак** композиции. Если комнаты создаются **вне** конструктора, это уже:

- **агрегация** — если передаются снаружи;
- **ассоциация** — если добавляются позже;
- вообще **не композиция**.

Композиция требует, чтобы внутренний объект **создавался вместе** с внешним. Так обеспечивается **целостность**: дом нельзя создать без комнат.

**4. Что произойдёт, если комнаты передавать снаружи.**

Если написать:

```python
class House:
    def __init__(self, address, rooms):
        self.address = address
        self.rooms = rooms
```

— это будет **агрегация**, а не композиция. Разница:

- Комнаты можно создать **отдельно** от дома.
- Одну и ту же комнату можно передать в **два разных** дома.
- Если дом удалить, комнаты **останутся** (пока на них есть ссылки).

Композиция **гарантирует**, что комнаты принадлежат **именно этому** дому. Агрегация — **нет**.

# Глава 9. Переопределение методов

## 9.1. Проблема: наследник ведёт себя как родитель

Вернёмся к нашему дому. У нас есть базовый класс `Room` и наследники `Kitchen`, `Bedroom`. У `Room` есть метод `info()`:

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area

    def info(self):
        return f"{self.name}: {self.area} м²"
```

Унаследуем `Kitchen`:

```python
class Kitchen(Room):
    pass
```

Теперь создадим кухню и вызовем `info()`:

```python
kitchen = Kitchen("Кухня", 12)
print(kitchen.info())    # Кухня: 12 м²
```

Работает — метод **унаследован** от родителя. Но что, если мы хотим, чтобы кухня выводила **дополнительную** информацию? Например, о наличии плиты:

```
Кухня: 12 м² (с плитой)
```

Родительский `info()` этого не умеет. Нужно, чтобы **наследник вёл себя по-своему**. Для этого и существует **переопределение методов**.

---

## 9.2. Что такое переопределение

**Переопределение (override)** — это когда **наследник даёт свою версию метода родителя**. Имя метода **совпадает**, но реализация — **другая**.

Простыми словами: наследник **заменяет** родительский метод своим.

### Синтаксис

```python
class Room:
    def info(self):
        return "Комната"

class Kitchen(Room):
    def info(self):                    # переопределение
        return "Кухня"
```

Разберём:

- `Room.info()` — родительская версия.
- `Kitchen.info()` — **своя** версия с тем же именем.
- При вызове `Kitchen().info()` Python найдёт метод **в `Kitchen`** и вызовет его.

### Схема

```
        Room
        └── info() → "Комната"
                │
        Kitchen
        └── info() → "Кухня"  ← переопределение
```

### Проверим

```python
print(Room().info())        # Комната — родительская версия
print(Kitchen().info())     # Кухня — своя версия
```

Python **сам** выбирает нужную версию по фактическому классу объекта. Это **динамическое связывание** (см. главу 4).

---

## 9.3. Зачем нужно переопределение

Переопределение решает **четыре задачи**.

### 1. Специализация поведения

Наследник может **дополнить** или **изменить** поведение родителя. Кухня не просто комната — у неё есть плита.

### 2. Уточнение интерфейса

Наследник может **вернуть более точные данные**. `Room.info()` возвращает общее описание, `Kitchen.info()` — конкретное.

### 3. Изменение логики

Наследник может **полностью заменить** реализацию родителя. Например, уборка кухни отличается от уборки спальни.

### 4. Совместимость с полиморфизмом

Переопределение — **основа полиморфизма** (см. главу 4). Благодаря ему одна функция работает по-разному с разными объектами.

---

## 9.4. Как Python ищет метод

Это **ключевой момент**. Разберём пошагово.

Когда мы вызываем `kitchen.info()`, Python:

1. **Смотрит на фактический класс** объекта — `Kitchen`.
2. **Ищет `info` в классе `Kitchen`**. Находит — использует **эту** версию.
3. Если бы не нашёл — пошёл бы к **родителю** `Room`.
4. Если бы и там не нашёл — пошёл бы к `object`.
5. Если не нашёл нигде — `AttributeError`.

Этот порядок называется **MRO** (Method Resolution Order). Мы разберём его подробно в главе 19.

### Схема

```
        kitchen.info()
                ↓
        Ищем в Kitchen → нашли → вызываем
                ↓
        (если не нашли)
                ↓
        Ищем в Room → нашли → вызываем
                ↓
        (если не нашли)
                ↓
        Ищем в object → нашли → вызываем
                ↓
        (если не нашли)
                ↓
        AttributeError
```

### Пример

```python
class Room:
    def info(self):
        return "Комната"

class Kitchen(Room):
    def info(self):
        return "Кухня"

class SmallKitchen(Kitchen):
    pass                                # info не переопределён


print(SmallKitchen().info())            # Кухня — версия из Kitchen
```

`SmallKitchen` не переопределяет `info`, поэтому Python идёт вверх по цепочке и находит версию в `Kitchen`.

---

## 9.5. Переопределение vs перегрузка

**Это разные вещи!** Их часто путают. Разберём подробно.

| Понятие | Что это | Есть в Python? |
|---------|---------|----------------|
| **Переопределение (override)** | Наследник заменяет метод родителя | ✅ Да |
| **Перегрузка (overload)** | Несколько методов с одним именем, но разными параметрами | ❌ Нет |

### Переопределение — про наследование

- **Два класса:** родитель и наследник.
- **Один метод:** одно имя.
- **Разные реализации:** у каждого своя.
- Работает благодаря **MRO**.

```python
class Room:
    def info(self):
        return "Комната"

class Kitchen(Room):
    def info(self):                    # переопределение
        return "Кухня"
```

### Перегрузка — про параметры

- **Один класс.**
- **Одно имя метода.**
- **Разные параметры:** разное количество или типы.

В Python перегрузки **нет**. Нельзя объявить два метода `info` в одном классе:

```python
class Room:
    def info(self):
        return "Комната"

    def info(self, x):                 # ← перезапишет первый!
        return f"Комната {x}"
```

Останется **только второй** метод. Первый будет потерян.

### Как имитировать перегрузку

Через **значения по умолчанию** и `*args` (см. главу 10).

### Сравнение

| Критерий | Переопределение | Перегрузка |
|----------|-----------------|------------|
| Классов | Два (родитель и наследник) | Один |
| Имён методов | Одно | Одно |
| Параметров | Совпадают | Разные |
| Есть в Python | ✅ Да | ❌ Нет (имитируется) |

**Запомни:** переопределение — про **замену**, перегрузка — про **разные параметры**.

---

## 9.6. `super()` — вызов родительской версии

Иногда нужно **не заменить** родительский метод, а **дополнить** его. Например, у `Room` есть `info()`, а у `Kitchen` хочется **добавить** информацию о плите.

Для этого используют **`super()`** — ссылку на родительский класс.

### Синтаксис

```python
class Kitchen(Room):
    def info(self):
        parent_info = super().info()       # вызываем версию Room
        return f"{parent_info} (с плитой)"
```

Теперь:

```python
print(Kitchen("Кухня", 12).info())
# Кухня: 12 м² (с плитой)
```

Мы **не переписали** логику родителя, а **взяли её и дополнили**.

### Схема

```
        Kitchen().info()
                ↓
        Kitchen.info()
                ↓
        super().info() → Room.info()
                ↓
        "Кухня: 12 м²"
                ↓
        добавляем "(с плитой)"
                ↓
        "Кухня: 12 м² (с плитой)"
```

### Что если `super()` не использовать

Если написать `info` без `super()`, родительская логика **потеряется**:

```python
class Kitchen(Room):
    def info(self):
        return "Кухня"                    # только своё, без родителя
```

Результат: `"Кухня"`. Информация о площади потеряна.

### Почему не `Room.info(self)`

Можно написать и так:

```python
class Kitchen(Room):
    def info(self):
        parent_info = Room.info(self)      # работает
        return f"{parent_info} (с плитой)"
```

Но это **плохая практика**:

- Жёсткая связь с `Room`.
- Если иерархия изменится — придётся переписывать.
- При множественном наследовании `super()` идёт по MRO, а прямое обращение — нет.

**Правило:** всегда используйте `super()`.

### `super()` без аргументов

В Python 3 `super()` можно вызывать **без аргументов**:

```python
super().info()                          # современный синтаксис
```

Это эквивалентно `super(Kitchen, self).info()`, но короче и безопаснее.

---

## 9.7. Переопределение конструктора

Это **самый частый** случай переопределения. Разберём подробно.

### Проблема

У `Room` есть конструктор. У `Kitchen` — свой, с дополнительным полем:

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area

class Kitchen(Room):
    def __init__(self, name, area, has_fridge):
        self.has_fridge = has_fridge    # ← а name и area?
```

Если не вызвать родительский конструктор, атрибуты `name` и `area` **не создадутся**:

```python
kitchen = Kitchen("Кухня", 12, True)
print(kitchen.name)
# AttributeError: 'Kitchen' object has no attribute 'name'
```

### Решение — `super().__init__()`

```python
class Kitchen(Room):
    def __init__(self, name, area, has_fridge):
        super().__init__(name, area)    # ← вызываем родительский конструктор
        self.has_fridge = has_fridge
```

Теперь всё работает:

```python
kitchen = Kitchen("Кухня", 12, True)
print(kitchen.name)          # Кухня
print(kitchen.area)          # 12
print(kitchen.has_fridge)    # True
```

### Порядок вызова

**Правильно:**

```python
def __init__(self, name, area, has_fridge):
    super().__init__(name, area)        # сначала родитель
    self.has_fridge = has_fridge        # потом свои
```

**Неправильно (по логике):**

```python
def __init__(self, name, area, has_fridge):
    self.has_fridge = has_fridge        # сначала свои
    super().__init__(name, area)        # потом родитель
```

Работать **будет**, но нарушает логику: сначала инициализируем родителя, потом — свои поля. Это стандарт.

### Что если родительский конструктор не нужен

Если `Room` **не имеет** конструктора с параметрами (или вообще без `__init__`), можно не вызывать `super().__init__()`.

Но если у родителя **есть** `__init__` — вызывайте его **всегда**. Это правило.

### Цепочка конструкторов

Если иерархия длинная — `super().__init__()` вызовет **следующий** конструктор в MRO:

```python
class A:
    def __init__(self):
        print("A")

class B(A):
    def __init__(self):
        super().__init__()
        print("B")

class C(B):
    def __init__(self):
        super().__init__()
        print("C")

C()
# A
# B
# C
```

`super()` автоматически идёт по цепочке.

---

## 9.8. Переопределение свойств (`@property`)

Свойства (см. главу 2) тоже можно **переопределять**.

### Простой случай

```python
class Room:
    def __init__(self, area):
        self._area = area

    @property
    def area(self):
        return self._area

class Kitchen(Room):
    @property
    def area(self):
        # Своя версия свойства.
        return self._area + 5           # например, учитываем нишу
```

Теперь:

```python
room = Room(10)
print(room.area)                # 10

kitchen = Kitchen(10)
print(kitchen.area)             # 15 — своя версия
```

### Переопределение сеттера

```python
class Kitchen(Room):
    @property
    def area(self):
        return self._area

    @area.setter
    def area(self, value):
        if value < 8:
            raise ValueError("Кухня должна быть не меньше 8 м²")
        self._area = value
```

### Вызов родительского свойства через `super()`

```python
class Kitchen(Room):
    @property
    def area(self):
        # super().area — обращение к родительскому геттеру.
        return super().area + 5
```

### Нюанс

При переопределении свойства **важно** определять и геттер, и сеттер, если они нужны. Иначе можно потерять возможность записи.

---

## 9.9. Переопределение магических методов

**Магические методы** (`__str__`, `__eq__`, `__add__`) — тоже методы. Их можно **переопределять** (см. главы 11–13).

### Пример с `__str__`

```python
class Room:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Комната: {self.name}"

class Kitchen(Room):
    def __str__(self):
        return f"Кухня: {self.name}"
```

```python
print(Room("Комната"))          # Комната: Комната
print(Kitchen("Кухня"))         # Кухня: Кухня
```

### Пример с `__eq__`

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area

    def __eq__(self, other):
        return self.name == other.name and self.area == other.area
```

Наследник может **изменить** логику сравнения.

---

## 9.10. Переопределение статических и классовых методов

Статические и классовые методы (см. главу 6) тоже можно переопределять.

### Переопределение `@staticmethod`

```python
class Room:
    @staticmethod
    def is_valid_area(area):
        return 5 <= area <= 100

class Kitchen(Room):
    @staticmethod
    def is_valid_area(area):
        return 8 <= area <= 100
```

```python
print(Room.is_valid_area(6))            # True
print(Kitchen.is_valid_area(6))         # False — своя версия
```

### Переопределение `@classmethod`

```python
class Room:
    @classmethod
    def from_string(cls, data):
        name, area = data.split(";")
        return cls(name, int(area))

class Kitchen(Room):
    @classmethod
    def from_string(cls, data):
        # Своя версия: формат "Кухня;Площадь;Плита".
        name, area, has_fridge = data.split(";")
        return cls(name, int(area), has_fridge == "да")
```

### Нюанс

При переопределении `@classmethod` **`cls` остаётся** тем же — фактическим классом, через который вызван метод. Это ключевое преимущество классовых методов.

---

## 9.11. Когда переопределять, а когда нет

Не всегда переопределение — **хорошая идея**. Разберём, когда стоит и когда нет.

### Когда стоит переопределять

| Ситуация | Пример |
|----------|--------|
| Наследник должен **вести себя иначе** | `Kitchen.info()` возвращает другой текст |
| Нужно **дополнить** родителя | Через `super()` |
| Нужно **специализировать** | `Kitchen` добавляет информацию о плите |
| Наследник **уточняет** логику | `Kitchen.is_valid_area()` строже |

### Когда не стоит переопределять

| Ситуация | Почему плохо |
|----------|--------------|
| Переопределяем **всё** | Ломается логика родителя |
| Переопределяем **без вызова** `super()` | Теряется родительское поведение |
| Переопределяем метод, который **не должен меняться** | Нарушает контракт |
| Переопределяем **конструктор** без `super()` | Атрибуты родителя не создаются |
| Переопределяем метод **с другим именем** | Это уже не переопределение |

### Правило

**Переопределяйте только то, что должно отличаться.** Если поведение совпадает с родителем — не переопределяйте.

### Нюанс: контракт родителя

Переопределяя метод, **уважайте контракт** родителя. Если `Room.info()` возвращает строку — `Kitchen.info()` тоже должна возвращать строку. Иначе код, который использует `info()`, может сломаться.

Это связано с принципом **LSP** (Liskov Substitution Principle) из SOLID — мы разберём его в главе 35.

---

## 9.12. Переопределение vs расширение

Есть **два** подхода при работе с методом родителя.

### Расширение (через `super()`)

Наследник **дополняет** родительский метод:

```python
class Kitchen(Room):
    def info(self):
        return f"{super().info()} (с плитой)"
```

Родительская логика **сохранена**.

### Замена (полное переопределение)

Наследник **полностью заменяет** родительский метод:

```python
class Kitchen(Room):
    def info(self):
        return "Это кухня"
```

Родительская логика **потеряна**.

### Когда что использовать

| Подход | Когда использовать |
|--------|-------------------|
| **Расширение** | Родительская логика нужна, но её надо дополнить |
| **Замена** | Родительская логика не подходит, нужна своя |

### Рекомендация

**По умолчанию — расширяйте** через `super()`. Заменяйте только когда родительская логика **точно не нужна**.

---

## 9.13. Переопределение в реальных проектах

Где переопределение используется на практике?

### 1. GUI-приложения

```python
class Widget:
    def render(self):
        return "Отрисовка виджета"

class Button(Widget):
    def render(self):
        return "Отрисовка кнопки"
```

### 2. Игровые объекты

```python
class GameObject:
    def update(self):
        pass

class Enemy(GameObject):
    def update(self):
        self.move_towards_player()
```

### 3. Логгеры

```python
class Logger:
    def log(self, message):
        print(message)

class FileLogger(Logger):
    def log(self, message):
        with open("log.txt", "a") as f:
            f.write(message + "\n")
```

### 4. Обработчики запросов

```python
class Handler:
    def handle(self, request):
        return "OK"

class AuthHandler(Handler):
    def handle(self, request):
        if not self.is_authenticated(request):
            return "401"
        return super().handle(request)
```

### 5. Наш дом

```python
class Room:
    def info(self):
        return f"{self.name}: {self.area} м²"

class Kitchen(Room):
    def info(self):
        return f"{super().info()} (с плитой)"
```

---

## 9.14. Нюансы и тонкости

### 9.14.1. Переопределение и приватные методы

**Приватные методы** родителя **не видны** наследнику напрямую (см. главу 5). Если наследник определяет метод с тем же именем — это **не переопределение**, а **новый метод**.

```python
class Room:
    def __private(self):
        return "Room"

class Kitchen(Room):
    def __private(self):                # это НЕ переопределение!
        return "Kitchen"
```

Из-за name mangling методы превращаются в `_Room__private` и `_Kitchen__private` — это **разные** методы.

### 9.14.2. Переопределение с разными сигнатурами

Если наследник определяет метод с **другим числом параметров**, это **всё ещё переопределение** (имя совпадает), но может сломать совместимость:

```python
class Room:
    def info(self):
        return "Комната"

class Kitchen(Room):
    def info(self, extra):              # другая сигнатура
        return f"Кухня {extra}"
```

Теперь `Kitchen().info()` без аргумента **не работает**. Это нарушает контракт родителя. **Так лучше не делать.**

### 9.14.3. Переопределение `__init__` с разной сигнатурой

Это **нормально**. Конструкторы часто имеют разные сигнатуры:

```python
class Room:
    def __init__(self, name, area):
        ...

class Kitchen(Room):
    def __init__(self, name, area, has_fridge):    # дополнительный параметр
        super().__init__(name, area)
        ...
```

Это **разрешено**, потому что конструктор **не вызывается через полиморфизм** — его вызывают явно.

### 9.14.4. `super()` и множественное наследование

При множественном наследовании (см. главу 19) `super()` идёт по **MRO-цепочке**:

```python
class A:
    def info(self):
        print("A")

class B(A):
    def info(self):
        super().info()
        print("B")

class C(A):
    def info(self):
        super().info()
        print("C")

class D(B, C):
    def info(self):
        super().info()
        print("D")

D().info()
# A → C → B → D (по MRO)
```

Это тонкий момент — разберём его в главе 19.

### 9.14.5. `super()` внутри `__init__` без аргументов

В Python 3 **всегда** можно писать `super().__init__()`. Это работает и для одиночного, и для множественного наследования.

### 9.14.6. Динамическое переопределение

Метод можно переопределить **динамически** — во время выполнения:

```python
class Room:
    def info(self):
        return "Комната"

room = Room()
room.info = lambda: "Динамически"       # переопределили
print(room.info())                       # Динамически
```

Это работает, но **плохая практика** — усложняет отладку.

---

## 9.15. Типичные ошибки

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Забыли `super().__init__()` | Атрибуты родителя не создаются | Добавить вызов |
| Забыли `super()` в обычном методе | Родительская логика теряется | Добавить вызов или убрать `super()` |
| Путают override и overload | Разные вещи | Override — про замену, overload — про параметры |
| Переопределяют с другим именем | Это новый метод | Имя должно совпадать |
| Переопределяют всё подряд | Ломает логику родителя | Переопределять только нужное |
| Меняют сигнатуру | Нарушает контракт | Сохранять сигнатуру |
| Переопределяют приватные методы | Это не переопределение | Использовать `_name` |
| Используют `Room.info(self)` вместо `super()` | Жёсткая связь | Использовать `super()` |
| Забывают `@property` при переопределении свойства | Метод вместо свойства | Добавить декоратор |
| Забывают `@staticmethod` при переопределении | Python передаст `self` | Добавить декоратор |

---

## 9.16. Что важно запомнить

- **Переопределение (override)** — наследник даёт **свою версию** метода родителя.
- **Имя метода** должно **совпадать**.
- **Python ищет метод** сначала в самом классе, потом у родителей (MRO).
- **Переопределение ≠ перегрузка.** Override — про классы, overload — про параметры.
- **`super()`** — вызов родительской версии метода.
- **`super().__init__()`** — обязателен в конструкторе наследника, если у родителя есть свой конструктор.
- **Конструкторы** переопределяются **чаще всего**.
- **Свойства**, **статические** и **классовые** методы тоже можно переопределять.
- **Расширение** (через `super()`) и **замена** (полная) — два подхода.
- **По умолчанию расширяйте** через `super()`.
- **Уважайте контракт** родителя: сигнатура и тип возврата должны совпадать.
- **Переопределяйте только то**, что должно отличаться.

## Задача 9 (переопределение методов)

В доме есть базовый класс `HomeItem` и несколько наследников. Нужно продемонстрировать **переопределение методов** в разных ситуациях.

**Часть 1.** Создайте класс `HomeItem` с полями `name` (название) и `value` (стоимость). Реализуйте:

- конструктор `__init__(name, value)`;
- обычный метод `info()` — возвращает строку `Предмет: <name>, стоимость <value> ₽`.

**Часть 2.** Создайте наследника `Furniture(HomeItem)` с дополнительным полем `material` (материал). Переопределите метод `info()` **через `super()`** — пусть возвращает родительскую строку **плюс** информацию о материале: `Предмет: <name>, стоимость <value> ₽ (материал: <material>)`.

**Часть 3.** Создайте наследника `Appliance(HomeItem)` с дополнительным полем `power` (мощность). Переопределите метод `info()` **полностью** (без `super()`) — пусть возвращает строку `Техника: <name> на <power> Вт, стоимость <value> ₽`. Родительский формат **не используем**.

**Часть 4.** Создайте наследника `Furniture` — `Chair`, который **не переопределяет** `info()`. Убедитесь, что он использует версию из `Furniture`.

**Часть 5.** Создайте класс `HomeItemExtended(HomeItem)` — с дополнительным методом `_info_prefix()`, возвращающим строку `[предмет]`. Переопределите `info()` так, чтобы он использовал `self._info_prefix()` и родительский метод через `super()`.

**Часть 6.** Продемонстрируйте работу всех классов:

- создайте по одному объекту каждого типа;
- выведите `info()` для каждого;
- покажите, что `Chair` использует версию `Furniture`, а не `HomeItem`;
- покажите разницу между `Furniture` (расширение через `super()`) и `Appliance` (полная замена).

**Часть 7.** В комментариях объясните:

1. Чем переопределение отличается от перегрузки.
2. Зачем нужен `super()` при переопределении.
3. В чём разница между **расширением** через `super()` и **полной заменой**.
4. Почему `Chair` не нужно переопределять `info()`.
5. Что произойдёт, если в `Furniture.__init__` не вызвать `super().__init__()`.

### Решение

```python
# ─── Часть 1. Базовый класс ───
class HomeItem:
    def __init__(self, name, value):
        self.name = name
        self.value = value

    def info(self):
        return f"Предмет: {self.name}, стоимость {self.value} ₽"


# ─── Часть 2. Расширение через super() ───
class Furniture(HomeItem):
    def __init__(self, name, value, material):
        super().__init__(name, value)               # вызываем родительский конструктор
        self.material = material

    def info(self):
        # Расширяем родительскую версию, а не заменяем её.
        parent_info = super().info()
        return f"{parent_info} (материал: {self.material})"


# ─── Часть 3. Полная замена ───
class Appliance(HomeItem):
    def __init__(self, name, value, power):
        super().__init__(name, value)
        self.power = power

    def info(self):
        # Своя версия, без super(). Родительский формат не используем.
        return f"Техника: {self.name} на {self.power} Вт, стоимость {self.value} ₽"


# ─── Часть 4. Наследник без переопределения ───
class Chair(Furniture):
    pass                                            # info не переопределён


# ─── Часть 5. Переопределение с дополнительным методом ───
class HomeItemExtended(HomeItem):
    def _info_prefix(self):
        return "[предмет]"

    def info(self):
        # Используем super() и собственный префикс.
        return f"{self._info_prefix()} {super().info()}"


# ─── Часть 6. Проверка ───
base = HomeItem("Стол", 5000)
table = Furniture("Стол", 5000, "дуб")
fridge = Appliance("Холодильник", 30000, 200)
chair = Chair("Стул", 1500, "берёза")
extended = HomeItemExtended("Лампа", 2000)

print(base.info())          # Предмет: Стол, стоимость 5000 ₽
print(table.info())         # Предмет: Стол, стоимость 5000 ₽ (материал: дуб)
print(fridge.info())        # Техника: Холодильник на 200 Вт, стоимость 30000 ₽
print(chair.info())         # Предмет: Стул, стоимость 1500 ₽ (материал: берёза)
print(extended.info())      # [предмет] Предмет: Лампа, стоимость 2000 ₽
```

**Ожидаемый вывод:**

```
Предмет: Стол, стоимость 5000 ₽
Предмет: Стол, стоимость 5000 ₽ (материал: дуб)
Техника: Холодильник на 200 Вт, стоимость 30000 ₽
Предмет: Стул, стоимость 1500 ₽ (материал: берёза)
[предмет] Предмет: Лампа, стоимость 2000 ₽
```

### Разбор

**1. Чем переопределение отличается от перегрузки.**

**Переопределение (override)** — это когда **наследник** даёт **свою версию** метода **родителя**. Имя метода совпадает, параметры те же, но реализация другая. Работает благодаря наследованию: два класса, один метод, разные версии.

**Перегрузка (overload)** — это когда **в одном классе** есть несколько методов **с одним именем**, но **разными параметрами**. В Python такого нет: имя метода — это один объект-функция, объявить две функции с одним именем нельзя.

**2. Зачем нужен `super()` при переопределении.**

`super()` — это **ссылка на родительский класс**. Через него можно вызвать **родительскую версию метода**. Это нужно, когда наследник хочет **дополнить** родительское поведение, а не **заменить** его полностью.

В примере `Furniture.info()` использует `super().info()`, чтобы **взять** родительскую строку и **добавить** информацию о материале.

**3. В чём разница между расширением через `super()` и полной заменой.**

- **Расширение** — наследник **дополняет** родительский метод: `parent_info + " (материал: дуб)"`. Родительская логика **сохранена**.
- **Полная замена** — наследник **полностью переписывает** метод: свой текст, свой формат. Родительская логика **потеряна**.

**Выбор:**

- **Расширяйте**, если родительская логика **нужна** и её достаточно дополнить.
- **Заменяйте**, если родительская логика **не подходит** и нужен свой формат.

**4. Почему `Chair` не нужно переопределять `info()`.**

`Chair` — это частный случай мебели. У него **нет собственной логики** для `info()`. Он может использовать версию `Furniture`, которая уже даёт информацию о материале. Переопределять её — **дублировать код** без пользы.

Если завтра `Furniture.info()` изменится — `Chair` автоматически получит новые изменения. Это и есть **переиспользование кода** через наследование.

**5. Что произойдёт, если в `Furniture.__init__` не вызвать `super().__init__()`.**

Атрибуты `name` и `value` **не создадутся**. При попытке обратиться к `self.name` Python выдаст `AttributeError`:

```python
class Furniture(HomeItem):
    def __init__(self, name, value, material):
        # super().__init__(name, value)  ← забыли!
        self.material = material

table = Furniture("Стол", 5000, "дуб")
print(table.info())
# AttributeError: 'Furniture' object has no attribute 'name'
```

Это **самая частая ошибка** при переопределении конструктора. Всегда вызывайте `super().__init__(...)` в начале конструктора наследника.

# Глава 10. Имитация перегрузки

## 10.1. Проблема: разные ситуации — один метод

Вернёмся к нашему дому. Управляющему нужно **регистрировать заявки на ремонт**. Заявки бывают разные:

- Иногда известна **только комната** — «нужно что-то починить».
- Иногда **комната и мастер** — «Иванов починит кухню».
- Иногда **комната, мастер и дата** — «Иванов починит кухню 12 мая».

Хочется, чтобы метод `register()` работал **во всех трёх ситуациях**:

```python
desk.register(kitchen)
desk.register(kitchen, "Иванов")
desk.register(kitchen, "Иванов", "12.05.2025")
```

В **Java** или **C++** это делается через **перегрузку методов** — три разных метода с одним именем, но разными параметрами:

```java
// Java
void register(Room room) { ... }
void register(Room room, String master) { ... }
void register(Room room, String master, String date) { ... }
```

**В Python так нельзя.** Одно имя — одна функция. Нельзя объявить три метода с именем `register`.

Но задачу **всё равно нужно решить**. Для этого в Python есть **имитация перегрузки** — несколько техник, которые позволяют одному методу работать по-разному.

---

## 10.2. Что такое перегрузка

**Перегрузка (overload)** — это **несколько методов с одним именем**, но **разными параметрами** (разное количество или типы).

### Как это выглядит в других языках

**Java:**

```java
void print(int x) { System.out.println(x); }
void print(String s) { System.out.println(s); }
void print(int x, int y) { System.out.println(x + y); }
```

Три метода `print` — три разных поведения.

**C++:**

```cpp
void print(int x) { cout << x; }
void print(string s) { cout << s; }
void print(int x, int y) { cout << x + y; }
```

**Python — так нельзя.** Одно имя — одна функция:

```python
class Printer:
    def print(self, x):
        ...

    def print(self, x, y):        # ← перезапишет первый!
        ...
```

Останется **только второй** метод. Первый **потерян**.

---

## 10.3. Почему в Python нет перегрузки

Причина **фундаментальная**: в Python **имя метода — это объект**. Когда вы пишете:

```python
class Room:
    def info(self):
        return "Комната"

    def info(self, extra):
        return f"Комната {extra}"
```

Python выполняет **два присваивания**:

1. `Room.info = <функция info>` (первая).
2. `Room.info = <функция info>` (вторая) — **перезаписывает первую**.

В итоге в словаре класса остаётся **только вторая** функция. Первая потеряна навсегда.

### Схема

```
        class Room:
            def info(self):          ← функция 1
            def info(self, extra):   ← функция 2 (перезаписывает)
                                       ↓
        Room.__dict__["info"] = функция 2
                                       ↓
        функция 1 ПОТЕРЯНА
```

### Философия Python

Python использует **динамическую типизацию** и **утиную типизацию** (см. главу 4). Вместо перегрузки по типам — **один метод, который сам разбирается** с аргументами. Это **проще** и **гибче**:

- Не нужно думать о типах на этапе компиляции.
- Один метод работает со многими вариантами.
- Код становится **короче**.

### Сравнение

| Аспект | Java / C++ | Python |
|--------|-----------|--------|
| Перегрузка | ✅ Есть | ❌ Нет |
| Как выбирается метод | По типам аргументов | — |
| Когда выбирается | При компиляции | — |
| Подход Python | — | Один метод + проверки |

**Запомни:** в Python **нет классической перегрузки** — только **имитация**.

---

## 10.4. Четыре техники имитации перегрузки

В Python есть **четыре** основных техники имитации:

1. **Значения по умолчанию** — самый простой способ.
2. **`*args` и `**kwargs`** — гибкий способ.
3. **Проверка типов** — через `isinstance`.
4. **`functools.singledispatch`** — продвинутый способ.

Разберём каждый.

---

## 10.5. Техника 1: значения по умолчанию

Самый **простой** и **распространённый** способ.

### Идея

Метод принимает **все возможные** параметры, но часть из них — **необязательные** (со значениями по умолчанию). Метод сам решает, что делать, в зависимости от того, **что передано**.

### Синтаксис

```python
class RepairDesk:
    def register(self, room, master=None, date=None):
        if master is None:
            print(f"Заявка по комнате: {room.name}")
        elif date is None:
            print(f"Комната {room.name} — мастер {master}")
        else:
            print(f"Комната {room.name} — мастер {master}, {date}")
```

### Как работает

- **`master=None`** — значение по умолчанию: «мастер не указан».
- **`date=None`** — значение по умолчанию: «дата не указана».
- Метод **проверяет** через `if`, что передано.

### Использование

```python
kitchen = Room("Кухня")
desk = RepairDesk()

desk.register(kitchen)                          # Заявка по комнате: Кухня
desk.register(kitchen, "Иванов")                # Комната Кухня — мастер Иванов
desk.register(kitchen, "Иванов", "12.05.2025")  # Комната Кухня — мастер Иванов, 12.05.2025
```

**Один метод — три варианта работы.**

### Плюсы

- **Просто** и **понятно**.
- **Читаемо**: сразу видно, какие параметры обязательные, какие нет.
- **IDE подсказывает** параметры.

### Минусы

- **Ограничено** числом параметров.
- **Проверки** внутри метода могут быть громоздкими.
- **Не работает**, если вариантов **много**.

### Когда использовать

**Почти всегда.** Значения по умолчанию — **предпочтительный** способ имитации.

---

## 10.6. Техника 2: `*args` и `**kwargs`

Более **гибкий** способ, когда количество аргументов **заранее неизвестно**.

### `*args` — произвольное число позиционных аргументов

```python
class RepairDesk:
    def register(self, *args):
        # args — кортеж всех переданных аргументов.
        if len(args) == 1:
            print(f"Заявка по комнате: {args[0].name}")
        elif len(args) == 2:
            room, master = args
            print(f"Комната {room.name} — мастер {master}")
        elif len(args) == 3:
            room, master, date = args
            print(f"Комната {room.name} — мастер {master}, {date}")
        else:
            print("Неверное число аргументов")
```

### Использование

```python
desk.register(kitchen)
desk.register(kitchen, "Иванов")
desk.register(kitchen, "Иванов", "12.05.2025")
```

### `**kwargs` — произвольное число именованных аргументов

```python
class RepairDesk:
    def register(self, room, **kwargs):
        # kwargs — словарь именованных аргументов.
        master = kwargs.get("master")
        date = kwargs.get("date")

        if master is None:
            print(f"Заявка по комнате: {room.name}")
        elif date is None:
            print(f"Комната {room.name} — мастер {master}")
        else:
            print(f"Комната {room.name} — мастер {master}, {date}")
```

### Использование

```python
desk.register(kitchen)
desk.register(kitchen, master="Иванов")
desk.register(kitchen, master="Иванов", date="12.05.2025")
```

### Комбинация `*args` и `**kwargs`

```python
def method(self, *args, **kwargs):
    ...
```

Работает с **любыми** аргументами.

### Плюсы

- **Максимальная гибкость**: любое число аргументов.
- **Можно передавать** именованные аргументы.

### Минусы

- **Менее читаемо**: непонятно, что передаётся.
- **IDE не подсказывает** параметры.
- **Ошибки** обнаруживаются только в рантайме.
- **Сложно** поддерживать.

### Когда использовать

- Когда **число аргументов** действительно **переменное**.
- Когда **названия аргументов** не важны.
- Как **крайний случай**, если другие техники не подходят.

### Сравнение с значениями по умолчанию

| Критерий | Значения по умолчанию | `*args` / `**kwargs` |
|----------|----------------------|----------------------|
| Читаемость | ✅ Высокая | ⚠️ Низкая |
| Гибкость | ⚠️ Ограничена | ✅ Высокая |
| Подсказки IDE | ✅ Есть | ❌ Нет |
| Проверка | В сигнатуре | Внутри метода |
| Предпочтительность | ✅ Да | ⚠️ Крайний случай |

---

## 10.7. Техника 3: проверка типов через `isinstance`

Иногда нужно **разное поведение** в зависимости от **типа** аргумента.

### Пример

```python
class RepairDesk:
    def register(self, arg):
        if isinstance(arg, Room):
            # Передали комнату.
            print(f"Заявка по комнате: {arg.name}")
        elif isinstance(arg, str):
            # Передали имя мастера.
            print(f"Заявка мастеру: {arg}")
        else:
            print("Неизвестный тип аргумента")
```

### Использование

```python
desk.register(kitchen)           # Заявка по комнате: Кухня
desk.register("Иванов")          # Заявка мастеру: Иванов
```

### Плюсы

- **Гибко**: один метод работает с разными типами.
- **Читаемо**: сразу видно, что для каждого типа своё поведение.

### Минусы

- **Несколько `isinstance`** — громоздко.
- **Ошибка видна** только в рантайме.
- **Нарушает полиморфизм** (см. главу 4).

### Когда использовать

- Когда методы **действительно различаются по типу**.
- Когда **нельзя** использовать полиморфизм.

### Нюанс: `isinstance` vs `type`

Иногда используют `type(arg) is Room` вместо `isinstance(arg, Room)`. Разница:

- **`isinstance`** — учитывает **наследников** (см. главу 3).
- **`type`** — строгая проверка **точного класса**.

**Правило:** используйте `isinstance`, а не `type`.

---

## 10.8. Техника 4: `functools.singledispatch`

**Продвинутый** способ, напоминающий перегрузку в других языках.

### Идея

Один метод — **разные реализации** для разных **типов первого аргумента**. Python **сам** выбирает нужную реализацию.

### Синтаксис

```python
from functools import singledispatch

class RepairDesk:
    @singledispatch
    def register(self, arg):
        # Реализация по умолчанию.
        print("Неизвестный аргумент")
```

Однако `singledispatch` работает **только с функциями**, не с методами напрямую. Есть **обходной путь**: использовать **статический метод** или **функцию вне класса**.

### Пример с функцией

```python
from functools import singledispatch

@singledispatch
def register(arg):
    print("Неизвестный аргумент")

@register.register
def _(arg: Room):
    print(f"Заявка по комнате: {arg.name}")

@register.register
def _(arg: str):
    print(f"Заявка мастеру: {arg}")
```

### Использование

```python
register(kitchen)          # Заявка по комнате: Кухня
register("Иванов")         # Заявка мастеру: Иванов
register(42)               # Неизвестный аргумент
```

Python **сам** выбирает нужную реализацию по типу.

### Как работает

- **`@singledispatch`** — регистрирует «главную» функцию.
- **`@register.register`** — регистрирует реализацию для конкретного типа.
- **`arg: Room`** — аннотация типа: Python использует её для выбора.

### Плюсы

- **Очень похоже** на настоящую перегрузку.
- **Читаемо**: для каждого типа — своя функция.
- **Расширяемо**: можно добавлять новые типы без изменения старых.

### Минусы

- **Не работает** с методами класса напрямую.
- **Только по первому аргументу** (не по количеству).
- **Менее распространено**, чем значения по умолчанию.

### Когда использовать

- Когда **типы** аргументов **важны**.
- Когда вариантов **много** и они **не пересекаются**.
- В **библиотеках**, где нужна расширяемость.

---

## 10.9. Сравнение четырёх техник

| Критерий | Значения по умолчанию | `*args` / `**kwargs` | `isinstance` | `singledispatch` |
|----------|----------------------|---------------------|--------------|------------------|
| Простота | ✅ Высокая | ⚠️ Средняя | ✅ Высокая | ⚠️ Средняя |
| Читаемость | ✅ Высокая | ⚠️ Низкая | ⚠️ Средняя | ✅ Высокая |
| Гибкость | ⚠️ Ограничена | ✅ Высокая | ⚠️ Средняя | ✅ Высокая |
| Подсказки IDE | ✅ Есть | ❌ Нет | ⚠️ Частично | ⚠️ Частично |
| По количеству аргументов | ✅ Да | ✅ Да | ❌ Нет | ❌ Нет |
| По типу аргумента | ⚠️ Через проверки | ⚠️ Через проверки | ✅ Да | ✅ Да |
| Использование | ✅ По умолчанию | ⚠️ Крайний случай | ⚠️ Редко | ⚠️ Библиотеки |

### Как выбирать

1. **Начните со значений по умолчанию.** Это **самый простой** и **читаемый** способ.
2. **Если вариантов много** — используйте `*args` / `**kwargs`.
3. **Если различие по типу** — используйте `isinstance` или `singledispatch`.
4. **В библиотеках** — `singledispatch`.

## 10.10. Имитация перегрузки в реальных проектах

Где имитация перегрузки используется на практике?

### 1. Обработчики запросов

```python
class RequestHandler:
    def handle(self, request, response=None, context=None):
        if response is None:
            # Простой запрос.
            ...
        elif context is None:
            # Запрос с ответом.
            ...
        else:
            # Полный запрос.
            ...
```

### 2. Логгеры

```python
class Logger:
    def log(self, message, level="INFO", *args):
        ...
```

### 3. Конструкторы объектов

```python
class Room:
    @classmethod
    def create(cls, name, area=None):
        if area is None:
            # Создать комнату с площадью по умолчанию.
            return cls(name, 20)
        return cls(name, area)
```

### 4. Работа с базами данных

```python
def query(self, sql, params=None, *args):
    if params is None:
        # Запрос без параметров.
        ...
    ...
```

### 5. Наш дом

```python
class RepairDesk:
    def register(self, room, master=None, date=None):
        ...
```

### Общий признак

Имитация перегрузки используется, когда:

- один метод должен работать в **нескольких ситуациях**;
- количество или типы аргументов **различаются**;
- **не хочется** создавать несколько методов.

---

## 10.11. Нюансы и тонкости

### 10.11.1. `None` как значение по умолчанию

**Правило:** используйте `None`, а не другие значения, как маркер «не передано».

**Плохо:**

```python
def register(self, room, master="", date=""):
    if master == "":
        ...
```

**Хорошо:**

```python
def register(self, room, master=None, date=None):
    if master is None:
        ...
```

**Почему:** `None` — это стандартный маркер «нет значения». Пустая строка или `0` могут быть **валидными** значениями.

### 10.11.2. Не используйте изменяемые значения по умолчанию

Это **классическая ошибка** Python:

```python
def register(self, room, masters=[]):    # ← ПЛОХО
    masters.append("Иванов")
    return masters
```

**Проблема:** список `masters` **один** для всех вызовов. Он **накапливает** значения.

**Решение:**

```python
def register(self, room, masters=None):
    if masters is None:
        masters = []
    masters.append("Иванов")
    return masters
```

**Правило:** для изменяемых значений по умолчанию используйте `None` и создавайте объект внутри.

### 10.11.3. Проверка на `None` через `is`

**Правильно:**

```python
if master is None:
    ...
```

**Неправильно:**

```python
if master == None:      # работает, но не рекомендуется
    ...
```

**Почему:** `==` можно **переопределить** (см. главу 12), а `is` — нет. `is None` всегда работает.

### 10.11.4. Порядок параметров

**Обязательные параметры — сначала, необязательные — потом:**

```python
def register(self, room, master=None, date=None):    # ✅
    ...
```

**Нельзя** так:

```python
def register(self, master=None, room):               # ❌ SyntaxError
    ...
```

**Правило:** обязательные параметры — **до** необязательных.

### 10.11.5. Значения по умолчанию и наследование

Если наследник переопределяет метод с **дополнительным** параметром — **не забудьте** родительские значения по умолчанию:

```python
class Room:
    def info(self, short=False):
        ...

class Kitchen(Room):
    def info(self, short=False, with_fridge=False):    # сохраняем short
        ...
```

### 10.11.6. `*args` и `**kwargs` вместе

```python
def register(self, *args, **kwargs):
    ...
```

Работает со **всеми** аргументами. Но будьте осторожны: **порядок** и **типы** нужно проверять **вручную**.

### 10.11.7. `singledispatch` и методы

`singledispatch` **не работает** с методами напрямую. Обходной путь — **статический метод** или **функция вне класса**. Подробнее — в документации Python.

### 10.11.8. Аннотации типов

Аннотации **не влияют** на выполнение. Они — для **документации** и **IDE**:

```python
def register(self, room: Room, master: str | None = None) -> None:
    ...
```

Python **не проверяет** типы в рантайме (если не использовать `mypy` или подобное).

---

## 10.12. Типичные ошибки

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Пытаются объявить два метода с одним именем | Второй перезапишет первый | Использовать значения по умолчанию |
| Используют `""` или `0` как маркер «не передано» | Может быть валидным значением | Использовать `None` |
| Изменяемые значения по умолчанию (`[]`, `{}`) | Один объект на все вызовы | Использовать `None` + создание внутри |
| Проверяют `== None` | Может быть переопределено | Использовать `is None` |
| Ставят необязательные параметры до обязательных | `SyntaxError` | Обязательные — первыми |
| Используют `type` вместо `isinstance` | Не учитывает наследников | Использовать `isinstance` |
| Злоупотребляют `*args` | Нечитаемо | Значения по умолчанию |
| Забывают проверки внутри метода | Может быть неверный результат | Проверять все варианты |
| Не проверяют «неверное число аргументов» | Может быть ошибка | Добавить `else` с сообщением |
| Думают, что аннотации типов проверяются | Нет, это только документация | Использовать `mypy` для проверки |

---

## 10.13. Что важно запомнить

- **Перегрузка** — несколько методов с одним именем и разными параметрами.
- **В Python перегрузки нет** — одно имя — одна функция.
- **Имитация перегрузки** — четыре техники:
  1. **Значения по умолчанию** — самый простой способ.
  2. **`*args` / `**kwargs`** — гибкий способ.
  3. **`isinstance`** — разное поведение по типу.
  4. **`singledispatch`** — продвинутый способ.
- **Предпочтительно** — значения по умолчанию.
- **`None`** — стандартный маркер «не передано».
- **Изменяемые значения по умолчанию** — **опасны**.
- **`is None`** — правильная проверка, `== None` — не рекомендуется.
- **Обязательные параметры** — **до** необязательных.
- **Аннотации типов** — только для документации, не для проверки.

## Задача 10 (имитация перегрузки)

Управляющему домом нужен метод для **регистрации заявок на ремонт**, который мог бы работать в **разных ситуациях**.

**Часть 1. Класс `Room`.** Создайте класс `Room` с полем `name` (название комнаты) и методом `__str__`, возвращающим название.

**Часть 2. Класс `RepairDesk` — значения по умолчанию.** Создайте класс `RepairDesk` с методом:

```python
def register(self, room, master=None, date=None):
```

Метод должен выводить одно из трёх сообщений:

- если **только комната** — `Заявка по комнате: <name>`;
- если **комната и мастер** — `Комната <name> — мастер <master>`;
- если **комната, мастер и дата** — `Комната <name> — мастер <master>, <date>`.

Продемонстрируйте все три варианта вызова.

**Часть 3. Класс `AdvancedRepairDesk` — `*args`.** Создайте класс `AdvancedRepairDesk` с методом:

```python
def register(self, *args):
```

Метод должен работать **так же**, как в части 2, но определять, что передано, **по количеству** аргументов. Если передано **0 или больше 3** аргументов — вывести `Неверное число аргументов`.

**Часть 4. Класс `TypedRepairDesk` — `isinstance`.** Создайте класс `TypedRepairDesk` с методом:

```python
def register(self, arg):
```

Метод должен:

- если `arg` — объект `Room` — вывести `Заявка по комнате: <name>`;
- если `arg` — строка — вывести `Заявка мастеру: <arg>`;
- иначе — вывести `Неизвестный тип аргумента`.

Продемонстрируйте работу со всеми тремя типами аргументов.

**Часть 5. Проверка.**
Продемонстрируйте работу **всех трёх классов**:

- `RepairDesk` — три вызова;
- `AdvancedRepairDesk` — три вызова + один с неверным числом аргументов;
- `TypedRepairDesk` — три вызова.

**Часть 6.** В комментариях объясните:

1. Почему в Python **нет классической перегрузки**.
2. В чём разница между **значениями по умолчанию** и **`*args`**.
3. Почему **`None`** лучше, чем `""` или `0`, как маркер «не передано».
4. Почему **`is None`** лучше, чем `== None`.
5. Какой способ имитации **предпочтителен** и почему.

### Решение

```python
# ─── Часть 1. Класс Room ───
class Room:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name


# ─── Часть 2. Значения по умолчанию ───
class RepairDesk:
    def register(self, room, master=None, date=None):
        if master is None:
            print(f"Заявка по комнате: {room.name}")
        elif date is None:
            print(f"Комната {room.name} — мастер {master}")
        else:
            print(f"Комната {room.name} — мастер {master}, {date}")


# ─── Часть 3. *args ───
class AdvancedRepairDesk:
    def register(self, *args):
        if len(args) == 1:
            print(f"Заявка по комнате: {args[0].name}")
        elif len(args) == 2:
            room, master = args
            print(f"Комната {room.name} — мастер {master}")
        elif len(args) == 3:
            room, master, date = args
            print(f"Комната {room.name} — мастер {master}, {date}")
        else:
            print("Неверное число аргументов")


# ─── Часть 4. isinstance ───
class TypedRepairDesk:
    def register(self, arg):
        if isinstance(arg, Room):
            print(f"Заявка по комнате: {arg.name}")
        elif isinstance(arg, str):
            print(f"Заявка мастеру: {arg}")
        else:
            print("Неизвестный тип аргумента")


# ─── Часть 5. Проверка ───
kitchen = Room("Кухня")

print("=== RepairDesk ===")
desk1 = RepairDesk()
desk1.register(kitchen)
desk1.register(kitchen, "Иванов")
desk1.register(kitchen, "Иванов", "12.05.2025")

print()
print("=== AdvancedRepairDesk ===")
desk2 = AdvancedRepairDesk()
desk2.register(kitchen)
desk2.register(kitchen, "Иванов")
desk2.register(kitchen, "Иванов", "12.05.2025")
desk2.register()                         # неверное число аргументов

print()
print("=== TypedRepairDesk ===")
desk3 = TypedRepairDesk()
desk3.register(kitchen)                  # комната
desk3.register("Иванов")                 # строка
desk3.register(42)                       # неизвестный тип
```

**Ожидаемый вывод:**

```
=== RepairDesk ===
Заявка по комнате: Кухня
Комната Кухня — мастер Иванов
Комната Кухня — мастер Иванов, 12.05.2025

=== AdvancedRepairDesk ===
Заявка по комнате: Кухня
Комната Кухня — мастер Иванов
Комната Кухня — мастер Иванов, 12.05.2025
Неверное число аргументов

=== TypedRepairDesk ===
Заявка по комнате: Кухня
Заявка мастеру: Иванов
Неизвестный тип аргумента
```

### Разбор

**1. Почему в Python нет классической перегрузки.**

В Python **имя метода — это объект**. Когда вы объявляете два метода с одним именем, **второй перезаписывает первый**. Остаётся только одна функция. Это связано с **динамической природой** языка: Python не знает типов аргументов на этапе «компиляции», поэтому не может выбрать нужный метод заранее.

Вместо перегрузки Python использует **один метод + проверки**. Это **проще** и **гибче**: не нужно думать о типах, код короче.

**2. В чём разница между значениями по умолчанию и `*args`.**

**Значения по умолчанию** — параметры **именованы**. Сигнатура метода **явно показывает**, какие аргументы возможны:

```python
def register(self, room, master=None, date=None):    # читаемо
```

**`*args`** — параметры **не именованы**. Сигнатура **не показывает**, что передаётся:

```python
def register(self, *args):                            # менее читаемо
```

**Плюсы значений по умолчанию:** читаемость, подсказки IDE, явные имена.
**Плюсы `*args`:** максимальная гибкость, любое число аргументов.

**Правило:** используйте **значения по умолчанию**, а `*args` — только если число аргументов **действительно переменное**.

**3. Почему `None` лучше, чем `""` или `0`, как маркер «не передано».**

`""` и `0` — **валидные значения**. Пустая строка может быть именем (хоть и странным), `0` — числом. Если использовать их как маркер «не передано», то **нельзя отличить** «не передано» от «передано пустое/нулевое».

**Пример проблемы:**

```python
def register(self, room, master=""):    # плохо
    if master == "":
        print(f"Заявка по комнате: {room.name}")
    else:
        print(f"Мастер: {master}")

register(kitchen, "")    # ← непонятно: не передали или передали пустое?
```

**Решение:** использовать `None` как маркер:

```python
def register(self, room, master=None):    # хорошо
    if master is None:
        print(f"Заявка по комнате: {room.name}")
    else:
        print(f"Мастер: {master}")
```

`None` — единственное значение, которое **не может быть валидным** в контексте «есть/нет». Поэтому оно идеально подходит как маркер.

**4. Почему `is None` лучше, чем `== None`.**

`==` — это **оператор сравнения**. Он может быть **переопределён** в классе через `__eq__` (см. главу 12). Если класс определяет `__eq__`, то `x == None` может вернуть **не то**, что ожидается.

`is` — это **проверка идентичности**. Она сравнивает **адреса в памяти**. `None` — **синглтон** (один объект на всю программу). Поэтому `x is None` **всегда** работает корректно.

**Правило:** для `None` используйте `is`, а не `==`.

**5. Какой способ имитации предпочтителен и почему.**

**Предпочтительно — значения по умолчанию.** Причины:

- **Читаемо:** сигнатура показывает, какие параметры возможны.
- **Подсказки IDE:** IDE знает параметры и подсказывает.
- **Простота:** не нужно проверять длину `*args`.
- **Гибкость:** можно вызывать с любым набором аргументов.
- **Документация:** сразу видно, что обязательно, что нет.

**`*args`** используйте только в **крайних случаях**:
- Когда число аргументов **действительно** переменное.
- Когда **нельзя** перечислить все параметры заранее.

**`isinstance`** — если различие **по типу**, а не по количеству.

**`singledispatch`** — в **библиотеках**, где нужна расширяемость.

# Глава 11. Магические методы `__str__` и `__repr__`

## 11.1. Проблема: нечитаемый вывод объекта

Вернёмся к агентству недвижимости. У нас есть класс `Apartment`:

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price
```

Создадим квартиру и попробуем вывести её на экран:

```python
flat = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)
print(flat)
```

**Что выведется:**

```
<__main__.Apartment object at 0x7f8b8c0d5e10>
```

Это **нечитаемо**. Мы видим:

- имя модуля (`__main__`),
- имя класса (`Apartment`),
- адрес объекта в памяти (`0x7f8b8c0d5e10`).

Никакой полезной информации о **квартире** здесь нет. Что это за квартира? Какая у неё площадь? Цена? Непонятно.

### Где это мешает

Такая проблема возникает **везде**:

- **В каталоге.** Список квартир выводится как набор адресов в памяти.
- **В отладчике.** Вместо данных — мусорные адреса.
- **В логах.** Записи бессмысленны.
- **В консоли.** Разработчик не может быстро посмотреть состояние объекта.

### Решение

Python позволяет **научить** объект **самому решать**, как его выводить. Для этого есть **два магических метода**:

- **`__str__`** — для **человека**.
- **`__repr__`** — для **разработчика**.

---

## 11.2. Что такое магические методы

**Магические методы** (их ещё называют **dunder-методами** — от **d**ouble **under**score) — это специальные методы Python, которые начинаются и заканчиваются **двумя подчёркиваниями**.

### Примеры

| Метод | Когда вызывается |
|-------|------------------|
| `__init__` | При создании объекта |
| `__str__` | При `print()` и `str()` |
| `__repr__` | При `repr()` и в отладчике |
| `__eq__` | При `==` |
| `__hash__` | При `hash()` |
| `__add__` | При `+` |
| `__len__` | При `len()` |

### Ключевые особенности

- **Не вызываются напрямую.** Их вызывает Python **автоматически**.
- **Обычно не пишутся с нуля.** Их **переопределяют** в своих классах.
- **Есть у всех объектов.** Наследуются от `object` — базового класса всех классов.

### Как вызывать

Магические методы **можно** вызывать напрямую, но **не нужно**:

```python
print(flat.__str__())     # работает
print(str(flat))          # лучше — то же самое
```

**Правило:** используйте **обычные функции** (`str()`, `repr()`, `len()`), а не магические методы напрямую.

### Важное отличие от `_name` и `__name`

Не путайте (см. главу 5):

- **`_name`** — защищённый атрибут.
- **`__name`** — приватный атрибут (name mangling).
- **`__name__`** — **магический метод** (не приватный!).

Магические методы **доступны снаружи** и **не искажаются**.

---

## 11.3. Метод `__str__` — для человека

**`__str__`** — это метод, который Python вызывает, когда нужно **строковое представление** объекта **для человека**.

### Когда вызывается

- При `print(obj)`.
- При `str(obj)`.
- При `f"{obj}"` (f-строки).
- При `"{}".format(obj)`.

### Что должен возвращать

**Строку**, которая **понятна человеку**. Например:

```
Квартира "ул. Ленина, д. 10, кв. 5" (65 м², 8 500 000 ₽)
```

### Как объявить

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    def __str__(self):
        return f"Квартира \"{self.address}\" ({self.area} м², {self.price:,} ₽)".replace(",", " ")
```

### Как использовать

```python
flat = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)
print(flat)
```

**Вывод:**

```
Квартира "ул. Ленина, д. 10, кв. 5" (65 м², 8 500 000 ₽)
```

Теперь объект выводится **понятно**.

### Важные правила

1. **`__str__` должен возвращать строку.** Не число, не список — **строку**.
2. **Не должен иметь побочных эффектов.** Не печатать, не писать в файл.
3. **Не должен быть слишком длинным.** Для человека — коротко и понятно.

### Что если `__str__` не определён

Если `__str__` не определён, Python **использует `__repr__`** (см. 11.4). Если и `__repr__` не определён — выводится стандартное представление:

```
<__main__.Apartment object at 0x7f8b8c0d5e10>
```

---

## 11.4. Метод `__repr__` — для разработчика

**`__repr__`** — это метод, который Python вызывает, когда нужно **техническое представление** объекта **для разработчика**.

### Когда вызывается

- При `repr(obj)`.
- В **отладчике** (при просмотре объекта).
- В **списках** и **словарях** — при выводе их через `print` или `repr`.
- В **интерактивной консоли** — когда вы просто пишете `obj` и нажимаете Enter.
- **Если `__str__` не определён** — при `print()`.

### Что должен возвращать

**Строку**, которая **позволяет воссоздать объект**. Например:

```
Apartment('ул. Ленина, д. 10, кв. 5', 65, 8500000)
```

Это **не просто описание** — это **код**, который можно выполнить, чтобы получить такой же объект.

### Как объявить

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    def __repr__(self):
        return f"Apartment({self.address!r}, {self.area!r}, {self.price!r})"
```

### Как использовать

```python
flat = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)
print(repr(flat))
```

**Вывод:**

```
Apartment('ул. Ленина, д. 10, кв. 5', 65, 8500000)
```

### Нюанс: `!r` в f-строках

В `__repr__` часто используют `!r`:

```python
f"{self.address!r}"
```

Это означает: **вызвать `repr()`** для значения. Если `self.address` — строка `"ул. Ленина"`, то `!r` даст `'ул. Ленина'` (с кавычками). Если без `!r` — даст `ул. Ленина` (без кавычек).

**Почему это важно:** для воссоздания объекта нужны **кавычки** вокруг строк. Иначе `Apartment(ул. Ленина, ...)` — синтаксическая ошибка.

### Важные правила

1. **`__repr__` должен возвращать строку.**
2. **Желательно**, чтобы строка **могла воссоздать объект**.
3. **Не должен иметь побочных эффектов.**
4. **Должен работать для всех объектов** класса.
5. **Если возможно** — избегайте адреса памяти в `__repr__`.

### Что если `__repr__` не определён

Python использует **стандартное представление** из класса `object`:

```
<__main__.Apartment object at 0x7f8b8c0d5e10>
```

Это тоже информация, но **менее полезная**, чем `Apartment('ул. Ленина, д. 10, кв. 5', 65, 8500000)`.

---

## 11.5. Разница между `__str__` и `__repr__`

Это **ключевой момент** главы. Разберём подробно.

### Главное отличие

| Метод | Для кого | Когда вызывается | Цель |
|-------|----------|------------------|------|
| **`__str__`** | Для **человека** | `print()`, `str()` | Понятное описание |
| **`__repr__`** | Для **разработчика** | `repr()`, отладчик, список | Точное представление |

### Аналогия

Представь, что у тебя есть **книга**:

- **`__str__`** — это **аннотация** на обложке: «Роман о любви и войне». Понятно **читателю**.
- **`__repr__`** — это **библиографическая карточка**: «Толстой Л.Н. Война и мир. — М.: Художественная литература, 1869. — 1300 с.». Понятно **библиотекарю**.

Оба описывают **одну и ту же книгу**, но **по-разному** и **для разных людей**.

### Пример: обе реализации

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    def __str__(self):
        # Для человека: понятная строка.
        return f"Квартира \"{self.address}\" ({self.area} м², {self.price:,} ₽)".replace(",", " ")

    def __repr__(self):
        # Для разработчика: техническое представление.
        return f"Apartment({self.address!r}, {self.area!r}, {self.price!r})"
```

### Как проявляется разница

**1. `print()` — использует `__str__`:**

```python
flat = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)
print(flat)
# Квартира "ул. Ленина, д. 10, кв. 5" (65 м², 8 500 000 ₽)
```

**2. `repr()` — использует `__repr__`:**

```python
print(repr(flat))
# Apartment('ул. Ленина, д. 10, кв. 5', 65, 8500000)
```

**3. Список — использует `__repr__`:**

```python
flats = [
    Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000),
    Apartment("ул. Садовая, д. 3, кв. 12", 80, 12_000_000),
]
print(flats)
# [Apartment('ул. Ленина, д. 10, кв. 5', 65, 8500000),
#  Apartment('ул. Садовая, д. 3, кв. 12', 80, 12000000)]
```

**Важно:** даже если у объекта определён `__str__`, в **списке** Python использует `__repr__`. Это связано с тем, что список должен показывать **точное** содержимое, а не «красивое».

**4. Интерактивная консоль — использует `__repr__`:**

```python
>>> flat
Apartment('ул. Ленина, д. 10, кв. 5', 65, 8500000)
```

**5. Отладчик — использует `__repr__`:**

В PyCharm, VS Code и других IDE объекты показываются через `__repr__`.

### Схема

```
        print(flat)         →  __str__   →  Квартира "ул. Ленина..." (65 м², 8 500 000 ₽)
        str(flat)           →  __str__
        f"{flat}"           →  __str__

        repr(flat)          →  __repr__  →  Apartment('ул. Ленина...', 65, 8500000)
        [flat, flat2]       →  __repr__
        flat (в консоли)    →  __repr__
        отладчик            →  __repr__
```

### Правило

**`__str__` — для пользователя. `__repr__` — для программиста.**

Если сомневаетесь — **реализуйте `__repr__`**, а `__str__` — по необходимости.

---

## 11.6. Как реализовать `__str__` через `__repr__`

Иногда `__str__` **похож** на `__repr__`, но чуть понятнее. Тогда можно **вызвать `__repr__`** из `__str__`.

### Пример

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    def __repr__(self):
        return f"Apartment({self.address!r}, {self.area!r}, {self.price!r})"

    def __str__(self):
        # Просто вызываем __repr__ — но не всегда это хорошая идея.
        return self.__repr__()
```

### Когда это полезно

- Если `__str__` **совпадает** с `__repr__` — просто не определяйте `__str__`, Python сам его возьмёт из `__repr__`.
- Если `__str__` **дополняет** `__repr__` — вызвать не получится, нужна своя логика.

### Когда это плохо

Если `__str__` должен быть **понятнее** `__repr__` — вызывать `__repr__` **неправильно**. Пользователь увидит `Apartment('ул. Ленина...', 65, 8500000)` вместо `Квартира "ул. Ленина..." (65 м², 8 500 000 ₽)`.

### Правило

- **Не определяйте `__str__`**, если он совпадает с `__repr__`. Python сам возьмёт `__repr__`.
- **Определяйте `__str__`** только если он **должен отличаться**.

---

## 11.7. Что если определить только `__str__`

Если определить **только `__str__`**, а `__repr__` — нет, то:

- `print(flat)` — использует `__str__` ✅
- `repr(flat)` — использует **стандартный `object.__repr__`** ❌
- В списке — использует **стандартный `object.__repr__`** ❌

### Пример

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    def __str__(self):
        return f"Квартира \"{self.address}\" ({self.area} м²)"


flat = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)

print(flat)              # Квартира "ул. Ленина, д. 10, кв. 5" (65 м²)
print([flat])            # [<__main__.Apartment object at 0x...>]  ← плохо!
```

### Вывод

**Всегда реализуйте `__repr__`**, даже если реализовали `__str__`. `__repr__` — это **базовый** метод, `__str__` — **надстройка**.

### Рекомендация

1. **Сначала** реализуйте `__repr__`.
2. **Потом** реализуйте `__str__`, если он должен отличаться.

---

## 11.8. Что если определить только `__repr__`

Если определить **только `__repr__`**, а `__str__` — нет:

- `print(flat)` — использует **`__repr__`** ✅
- `repr(flat)` — использует **`__repr__`** ✅
- В списке — использует **`__repr__`** ✅

### Пример

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    def __repr__(self):
        return f"Apartment({self.address!r}, {self.area!r}, {self.price!r})"


flat = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)

print(flat)              # Apartment('ул. Ленина, д. 10, кв. 5', 65, 8500000)
print([flat])            # [Apartment('ул. Ленина, д. 10, кв. 5', 65, 8500000)]
```

### Вывод

Если реализован **только `__repr__`** — он используется **везде**. Это **безопасный** подход.

### Рекомендация

**Если не хотите писать `__str__`** — пишите только `__repr__`. Это **приемлемо** и часто используется.

---

## 11.9. `__str__` и `__repr__` в наследовании

Магические методы **наследуются**. Если родитель определяет `__repr__`, наследник его получает.

### Пример

```python
class Apartment:
    def __init__(self, address, area):
        self.address = address
        self.area = area

    def __repr__(self):
        return f"{self.__class__.__name__}({self.address!r}, {self.area!r})"


class Studio(Apartment):
    pass


studio = Studio("ул. Ленина, д. 10, кв. 5", 25)
print(studio)
# Studio('ул. Ленина, д. 10, кв. 5', 25)
```

### Нюанс: `self.__class__.__name__`

В `__repr__` часто используют `self.__class__.__name__` вместо жёсткого имени класса:

```python
f"{self.__class__.__name__}(...)"
```

**Почему:** тогда наследник **автоматически** получит своё имя класса в `__repr__`. Если написать жёстко `Apartment(...)`, наследник `Studio` тоже будет показывать `Apartment(...)` — это **неправильно**.

### Переопределение `__str__` в наследнике

Если нужно, наследник может **переопределить** `__str__`:

```python
class Studio(Apartment):
    def __str__(self):
        return f"Студия \"{self.address}\" ({self.area} м²)"
```

---

## 11.10. `__str__` и `__repr__` в реальных проектах

### 1. Django-модели

```python
class Apartment(models.Model):
    address = models.CharField(max_length=200)
    area = models.IntegerField()

    def __str__(self):
        return f"{self.address} ({self.area} м²)"
```

Django использует `__str__` для отображения объектов в админке.

### 2. SQLAlchemy-модели

```python
class Apartment(Base):
    __tablename__ = "apartments"

    def __repr__(self):
        return f"<Apartment(id={self.id}, address={self.address!r})>"
```

### 3. Игровые объекты

```python
class Player:
    def __repr__(self):
        return f"Player({self.name!r}, hp={self.hp}, level={self.level})"
```

### 4. Логирование

```python
logger.info(f"Создан объект: {flat!r}")
```

Здесь `!r` вызывает `__repr__` — в логи попадает **техническое** представление.

### 5. Наш дом

```python
class Room:
    def __str__(self):
        return f"{self.name}: {self.area} м²"

    def __repr__(self):
        return f"Room({self.name!r}, {self.area!r})"
```

---

## 11.11. Нюансы и тонкости

### 11.11.1. `__str__` должен возвращать только строку

**Правильно:**

```python
def __str__(self):
    return f"Квартира: {self.address}"
```

**Неправильно:**

```python
def __str__(self):
    return 42                              # TypeError: __str__ returned non-string
```

Python выдаст ошибку, если `__str__` вернёт **не строку**.

### 11.11.2. `__repr__` тоже должен возвращать строку

То же самое правило:

```python
def __repr__(self):
    return f"Apartment({self.address!r})"
```

### 11.11.3. Не используйте `print` внутри `__str__`

**Правильно:**

```python
def __str__(self):
    return f"Квартира: {self.address}"
```

**Неправильно:**

```python
def __str__(self):
    print(f"Квартира: {self.address}")     # ← побочный эффект!
    return ""
```

`__str__` может вызываться **много раз** (например, в цикле). Печать внутри него засорит консоль.

### 11.11.4. `__repr__` и рекурсия

Если в `__repr__` вызывается `self.__repr__()` — получится **бесконечная рекурсия**:

```python
def __repr__(self):
    return f"Apartment({self.address!r}, area={self.area}, price={self.__repr__})"    # ← плохо
```

**Правило:** в `__repr__` используйте только **простые атрибуты**, не сам объект.

### 11.11.5. `!r` в f-строках

В `__repr__` используйте `!r`, чтобы **строки** выводились с кавычками:

```python
return f"Apartment({self.address!r}, {self.area!r}, {self.price!r})"
```

Без `!r` строки теряют кавычки, и результат **нельзя выполнить**.

### 11.11.6. `__str__` и `__repr__` — не для сравнения

Иногда новички думают, что если `__str__` возвращает строки, то объекты сравниваются по ним. Это **неправильно**. Для сравнения нужен `__eq__` (см. главу 12).

### 11.11.7. Списки используют `__repr__`, не `__str__`

Это **частая ловушка**. Даже если у объекта есть `__str__`, в списке используется `__repr__`:

```python
print([flat])         # [Apartment('ул. Ленина...', 65, 8500000)]
print(flat)           # Квартира "ул. Ленина..." (65 м², 8 500 000 ₽)
```

**Причина:** список должен показывать **точное** содержимое.

### 11.11.8. Строки в f-строках используют `__format__`

`__format__` — ещё один магический метод. Он вызывается для `f"{obj:>10}"`. По умолчанию использует `__str__`. Но можно переопределить (это **редкий** случай).

### 11.11.9. Юникод и кавычки

В `__repr__` Python использует **одинарные** кавычки по умолчанию. Если хотите — можно и двойные, но это **не стандарт**:

```python
return f'Apartment("{self.address}", {self.area})'    # работает, но не принято
```

### 11.11.10. Плейсхолдеры и очень длинные объекты

Если у объекта **много** полей, `__repr__` может быть **очень длинным**. В таких случаях можно использовать **многоточие**:

```python
def __repr__(self):
    return f"<Apartment {self.address!r} ...>"
```

---

## 11.12. Типичные ошибки

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Реализовали только `__str__` | В списках и отладчике мусор | Реализовать `__repr__` |
| Забыли `!r` в `__repr__` | Строки без кавычек | Добавить `!r` |
| `__str__` возвращает не строку | `TypeError` | Вернуть строку |
| `__repr__` возвращает не строку | `TypeError` | Вернуть строку |
| Печатают в `__str__` | Побочные эффекты | Только `return` |
| Используют `self.__repr__` внутри `__repr__` | Бесконечная рекурсия | Не вызывать себя |
| Пишут имя класса жёстко в `__repr__` | Наследники показывают родителя | `self.__class__.__name__` |
| Путают `__str__` и `__repr__` | Вывод не для того | Помнить: `__str__` — для человека |
| Думают, что `print([obj])` использует `__str__` | Использует `__repr__` | Помнить это |
| Слишком длинный `__repr__` | Нечитаемо в списке | Сократить |

---

## 11.13. Что важно запомнить

- **Магические методы** — dunder-методы, вызываются Python автоматически.
- **`__str__`** — для **человека**, вызывается при `print()`, `str()`, f-строках.
- **`__repr__`** — для **разработчика**, вызывается при `repr()`, в списках, в отладчике.
- **`__repr__` — базовый.** Если нет `__str__` — используется `__repr__`.
- **`__str__` — надстройка.** Определяйте, если он должен отличаться.
- **Оба должны возвращать строку.**
- **Не должны иметь побочных эффектов.**
- **`!r` в f-строках** — для технического представления.
- **`self.__class__.__name__`** — для наследуемости.
- **В списках используется `__repr__`**, не `__str__`.
- **Рекомендация:** всегда реализуйте `__repr__`. `__str__` — по необходимости.

---

## Задача 11 (магические методы `__str__` и `__repr__`)

В агентстве недвижимости ведётся каталог квартир. Квартиры часто выводят на экран — в объявлениях, каталогах, отчётах. По умолчанию `print(apartment)` выводит **нечитаемый адрес** объекта в памяти, что неудобно.

**Часть 1. Класс `Apartment`.** Создайте класс `Apartment` с полями:

- `address` — адрес квартиры (строка);
- `area` — площадь в м² (число);
- `price` — цена (число).

Реализуйте конструктор `__init__`.

**Часть 2. Метод `__str__`.** Добавьте в класс магический метод `__str__`, который возвращает строку вида:

```
Квартира "ул. Ленина, д. 10, кв. 5" (65 м², 8 500 000 ₽)
```

Цена должна выводиться с **пробелами** между разрядами (например, `8 500 000`, а не `8500000`). Убедитесь, что при вызове `print(apartment)` выводится именно эта строка.

**Часть 3. Метод `__repr__`.** Реализуйте метод `__repr__`, который возвращает **техническое** представление объекта:

```
Apartment('ул. Ленина, д. 10, кв. 5', 65, 8500000)
```

Это представление используется в **списках**, в **отладчике** и в **интерактивной консоли**.

**Часть 4. Проверка.** Создайте **две** квартиры и продемонстрируйте:

- работу `print(apartment)` — использует `__str__`;
- работу `repr(apartment)` — использует `__repr__`;
- работу со **списком** — `print([flat1, flat2])` использует `__repr__`;
- работу в **f-строке** — `f"{flat1}"` использует `__str__`;
- работу с `!r` в f-строке — `f"{flat1!r}"` использует `__repr__`.

**Часть 5. Наследование.** Создайте наследника `Studio(Apartment)` с тем же набором полей (но без дополнительных). Убедитесь, что:

- `__repr__` **наследуется** и выводит `Studio(...)`, а не `Apartment(...)` — для этого используйте `self.__class__.__name__`;
- `__str__` тоже наследуется — можно **переопределить** его в `Studio`, чтобы выводить слово «Студия» вместо «Квартира».

**Часть 6.** В комментариях объясните:

1. Чем `__str__` отличается от `__repr__`.
2. Почему `print([flat1, flat2])` использует `__repr__`, а не `__str__`.
3. Зачем нужен `!r` в `__repr__`.
4. Почему в `__repr__` используют `self.__class__.__name__`, а не жёсткое имя класса.
5. Что произойдёт, если реализовать **только** `__str__`, а `__repr__` — нет.

### Решение

```python
# ─── Часть 1. Класс Apartment ───
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    # ─── Часть 2. __str__ — для человека ───
    def __str__(self):
        # Форматируем цену с пробелами между разрядами.
        price_str = f"{self.price:,}".replace(",", " ")
        return f"Квартира \"{self.address}\" ({self.area} м², {price_str} ₽)"

    # ─── Часть 3. __repr__ — для разработчика ───
    def __repr__(self):
        # self.__class__.__name__ — имя фактического класса.
        # !r — вызывает repr() для значений (строки — с кавычками).
        return (
            f"{self.__class__.__name__}("
            f"{self.address!r}, {self.area!r}, {self.price!r}"
            f")"
        )


# ─── Часть 5. Наследник Studio ───
class Studio(Apartment):
    def __str__(self):
        # Переопределяем __str__ — для студии другое название.
        price_str = f"{self.price:,}".replace(",", " ")
        return f"Студия \"{self.address}\" ({self.area} м², {price_str} ₽)"


# ─── Часть 4. Проверка ───
flat1 = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)
flat2 = Apartment("ул. Садовая, д. 3, кв. 12", 80, 12_000_000)
studio = Studio("ул. Пушкина, д. 1, кв. 7", 25, 4_000_000)

print("=== print(flat1) — использует __str__ ===")
print(flat1)

print()
print("=== repr(flat1) — использует __repr__ ===")
print(repr(flat1))

print()
print("=== print([flat1, flat2]) — использует __repr__ ===")
print([flat1, flat2])

print()
print("=== f-строка — использует __str__ ===")
print(f"Квартира: {flat1}")

print()
print("=== f-строка с !r — использует __repr__ ===")
print(f"Квартира: {flat1!r}")

print()
print("=== Наследник Studio ===")
print(studio)                # __str__ из Studio
print(repr(studio))          # __repr__ из Apartment, но с именем Studio
print([studio])              # __repr__ из Apartment, но с именем Studio
```

**Ожидаемый вывод:**

```
=== print(flat1) — использует __str__ ===
Квартира "ул. Ленина, д. 10, кв. 5" (65 м², 8 500 000 ₽)

=== repr(flat1) — использует __repr__ ===
Apartment('ул. Ленина, д. 10, кв. 5', 65, 8500000)

=== print([flat1, flat2]) — использует __repr__ ===
[Apartment('ул. Ленина, д. 10, кв. 5', 65, 8500000), Apartment('ул. Садовая, д. 3, кв. 12', 80, 12000000)]

=== f-строка — использует __str__ ===
Квартира: Квартира "ул. Ленина, д. 10, кв. 5" (65 м², 8 500 000 ₽)

=== f-строка с !r — использует __repr__ ===
Квартира: Apartment('ул. Ленина, д. 10, кв. 5', 65, 8500000)

=== Наследник Studio ===
Студия "ул. Пушкина, д. 1, кв. 7" (25 м², 4 000 000 ₽)
Studio('ул. Пушкина, д. 1, кв. 7', 25, 4000000)
[Studio('ул. Пушкина, д. 1, кв. 7', 25, 4000000)]
```

### Разбор

**1. Чем `__str__` отличается от `__repr__`.**

- **`__str__`** — **для человека**. Вызывается при `print()`, `str()`, f-строках. Возвращает **понятное описание**: «Квартира "ул. Ленина..." (65 м², 8 500 000 ₽)».
- **`__repr__`** — **для разработчика**. Вызывается при `repr()`, в списках, в отладчике, в консоли. Возвращает **техническое представление**: `Apartment('ул. Ленина...', 65, 8500000)`.

**Главное отличие:** `__str__` — про **понятность**, `__repr__` — про **точность**.

**2. Почему `print([flat1, flat2])` использует `__repr__`, а не `__str__`.**

Список — это **коллекция**. Когда Python выводит список, он не знает, **как** пользователь хочет видеть элементы. Список должен показать **точное содержимое**, а не «красивое». Поэтому используется `__repr__` — **техническое** представление.

Если бы список использовал `__str__`, то для **отладки** было бы неудобно: непонятно, какие именно объекты внутри.

**3. Зачем нужен `!r` в `__repr__`.**

`!r` в f-строке — это **вызов `repr()`** для значения. Без `!r` строки выводятся **без кавычек**:

```python
f"{self.address!r}"     # 'ул. Ленина, д. 10, кв. 5'  ← с кавычками
f"{self.address}"       # ул. Ленина, д. 10, кв. 5     ← без кавычек
```

**Почему это важно:** `__repr__` должен возвращать **код**, который можно **выполнить**. Без кавычек `Apartment(ул. Ленина, ...)` — **синтаксическая ошибка**. С кавычками `Apartment('ул. Ленина', ...)` — работает.

**4. Почему в `__repr__` используют `self.__class__.__name__`, а не жёсткое имя класса.**

Если написать жёстко:

```python
def __repr__(self):
    return f"Apartment({self.address!r}, {self.area!r}, {self.price!r})"
```

— то **наследник** `Studio` тоже будет выводить `Apartment(...)`. Это **неправильно**: объект класса `Studio`, а представление — как у `Apartment`.

С `self.__class__.__name__`:

```python
def __repr__(self):
    return f"{self.__class__.__name__}({self.address!r}, {self.area!r}, {self.price!r})"
```

— наследник `Studio` **автоматически** выводит `Studio(...)`. Код **устойчив к наследованию**.

**5. Что произойдёт, если реализовать только `__str__`, а `__repr__` — нет.**

Если реализован **только `__str__`**:

- `print(flat)` — **работает** (использует `__str__`) ✅
- `repr(flat)` — **не работает как надо**: используется **стандартный `object.__repr__`**, который выводит `<__main__.Apartment object at 0x...>` ❌
- `print([flat])` — **не работает как надо**: в списке мусорный адрес ❌
- В **отладчике** — мусорный адрес ❌

**Мораль:** `__repr__` — **базовый** метод. Без него объект **нечитаем везде**, кроме `print()`. **Всегда реализуйте `__repr__`**, даже если реализовали `__str__`.

# Глава 12. Магические методы `__eq__` и `__hash__`

## 12.1. Проблема: одинаковые квартиры — разные объекты

Вернёмся к агентству недвижимости. У нас есть класс `Apartment`:

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price
```

Создадим **две квартиры с одинаковыми данными**:

```python
flat1 = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)
flat2 = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)
```

Это **одна и та же квартира** — просто дважды подано объявление. Но Python считает иначе:

```python
print(flat1 == flat2)     # False!
```

**Почему `False`?** Потому что по умолчанию `==` сравнивает **идентичность** — то есть, **один ли это объект в памяти**. `flat1` и `flat2` — **разные объекты**, хоть и с одинаковыми данными.

### Где это мешает

Такая проблема возникает **везде**:

- **Каталог квартир.** Одна и та же квартира подана дважды — в списке она **должна быть одна**.
- **Множество.** Если добавить `flat1` и `flat2` в `set`, будут **две записи**, хотя квартира одна.
- **Словарь.** Нельзя использовать квартиру как ключ, если её надо сравнивать по данным.
- **Поиск.** Нельзя проверить «есть ли такая квартира уже в базе».

### Решение

Нужно **научить** объект **сравниваться по данным**, а не по адресу в памяти. Для этого существуют **два магических метода**:

- **`__eq__`** — определяет, **когда два объекта равны**.
- **`__hash__`** — определяет **хэш-значение** объекта (для `set` и `dict`).

---

## 12.2. Что такое `__eq__`

**`__eq__`** — это магический метод, который Python вызывает, когда вы используете **оператор `==`**.

### Как работает по умолчанию

Если `__eq__` **не определён**, Python использует **версию из `object`**. Она сравнивает **идентичность**:

```python
print(flat1 == flat2)          # False — разные объекты
print(flat1 == flat1)          # True — один и тот же объект
print(flat1 is flat1)          # True — то же самое
```

### Как переопределить

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    def __eq__(self, other):
        # Сравниваем по адресу и площади.
        if not isinstance(other, Apartment):
            return NotImplemented
        return self.address == other.address and self.area == other.area
```

Теперь:

```python
print(flat1 == flat2)          # True — совпадают адрес и площадь
```

### Разбор кода

- **`def __eq__(self, other):`** — метод принимает **два** аргумента: `self` (текущий объект) и `other` (с чем сравниваем).
- **`isinstance(other, Apartment)`** — проверяем, что `other` — тоже `Apartment`. Если нет — **возвращаем `NotImplemented`**.
- **`self.address == other.address`** — сравниваем поля.

### Что такое `NotImplemented`

`NotImplemented` — это **специальное значение**, а не исключение. Оно говорит Python: «я не знаю, как сравнить эти объекты — попробуй другую сторону».

**Зачем это нужно:**

```python
flat == "какая-то строка"
```

Если `__eq__` не проверит тип, он попытается сравнить `flat.address` со `"строка"` — и это может дать **странный результат** или **ошибку**. `NotImplemented` говорит Python: «сравни наоборот» — то есть `"строка".__eq__(flat)`. Строка тоже вернёт `NotImplemented` — в итоге Python вернёт `False`.

**Правило:** всегда проверяйте тип через `isinstance` и возвращайте `NotImplemented` для чужих типов.

### Нюанс: `is` vs `==`

- **`is`** — проверка **идентичности** (один ли объект в памяти). **Не переопределяется**.
- **`==`** — проверка **равенства** (через `__eq__`). **Переопределяется**.

```python
print(flat1 is flat2)          # False — разные объекты
print(flat1 == flat2)          # True — одинаковые данные (после __eq__)
```

**Важно:** `is` **всегда** будет `False` для разных объектов. `==` зависит от `__eq__`.

---

## 12.3. Что такое `__hash__`

**`__hash__`** — это магический метод, который Python вызывает, когда вы используете функцию **`hash()`** или добавляете объект в **`set`** или **`dict`**.

### Что такое хэш

**Хэш** — это **целое число**, которое Python вычисляет для объекта. Хэш используется для **быстрого поиска** в `set` и `dict`.

**Аналогия:** представь, что у тебя есть **шкаф с ячейками**. Хэш — это **номер ячейки**, куда положить вещь. Когда нужно найти вещь — считаешь хэш и идёшь в нужную ячейку.

### Как работает по умолчанию

Если `__hash__` **не определён**, Python использует **версию из `object`**. Она возвращает хэш на основе **адреса** в памяти:

```python
print(hash(flat1))             # например, 1234567890
print(hash(flat2))             # например, 987654321 — другое число!
```

### Как переопределить

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    def __eq__(self, other):
        if not isinstance(other, Apartment):
            return NotImplemented
        return self.address == other.address and self.area == other.area

    def __hash__(self):
        # Хэш на основе тех же полей, что и в __eq__.
        return hash((self.address, self.area))
```

Теперь:

```python
print(hash(flat1) == hash(flat2))    # True — одинаковые хэши
```

### Почему хэш зависит от тех же полей

**Ключевое правило:** если два объекта **равны** через `__eq__`, то их **хэши должны совпадать**.

Если `__eq__` сравнивает **адрес и площадь**, то `__hash__` тоже должен использовать **адрес и площадь**. Иначе объекты будут **равны**, но иметь **разные хэши** — и `set` перестанет работать.

### Схема

```
        __eq__          →  сравнивает данные
        __hash__        →  считает число для быстрого поиска

        Если flat1 == flat2, то hash(flat1) == hash(flat2)
```

---

## 12.4. Зачем нужен `__hash__`

Хэш нужен для **двух** структур данных.

### 1. `set` (множество)

**Множество** хранит **уникальные** элементы. Чтобы определить, является ли элемент уникальным, Python использует **хэш** и **`__eq__`**.

```python
flats = {flat1, flat2}
print(len(flats))              # 1 — потому что они равны!
```

Если бы `__hash__` не был переопределён, `flat1` и `flat2` попали бы как **два разных** элемента.

### 2. `dict` (словарь)

**Словарь** использует ключи. Ключи должны быть **хэшируемыми**.

```python
prices = {
    flat1: "8 500 000 ₽",
    flat2: "12 000 000 ₽",     # ← перезапишет flat1, потому что они равны!
}
print(len(prices))              # 1
```

### Что если `__hash__` не определён, а `__eq__` — определён

Тогда Python **автоматически** делает объект **нехэшируемым**. Попытка использовать его в `set` или `dict` вызовет ошибку:

```python
class Apartment:
    def __init__(self, address, area):
        self.address = address
        self.area = area

    def __eq__(self, other):
        return self.address == other.address

flat = Apartment("ул. Ленина, д. 10, кв. 5", 65)
{flat}                          # TypeError: unhashable type: 'Apartment'
```

### Почему Python так делает

Это **защита от ошибки**. Если объекты **сравниваются по данным**, но **хэш — по адресу**, получается **несогласованность**:

- `flat1 == flat2` → `True`
- `hash(flat1) != hash(flat2)` → разные хэши

В `set` такой объект **не найдётся** через другой, потому что `set` сначала смотрит **хэш**. Если хэши разные — он даже не будет сравнивать через `__eq__`. Это **нарушает инвариант** множества.

Поэтому Python **запрещает** такой класс: если определён `__eq__`, но не определён `__hash__`, объект **нехэшируем**. Нужно **явно** определить `__hash__` — или указать `__hash__ = None` (тогда объект нехэшируем).

### Схема

```
        Определён __eq__?
                ↓
        ┌───────┴───────┐
        │               │
        Да              Нет
        │               │
        ↓               ↓
    Определён       Хэш от адреса
    __hash__?       (по умолчанию)
        │
    ┌───┴───┐
    │       │
    Да      Нет
    │       │
    ↓       ↓
  Хэш от   ОБЪЕКТ
  полей    НЕХЭШИРУЕМ
```

---

## 12.5. Контракт `__eq__` и `__hash__`

Между этими двумя методами есть **важный контракт**. Его нарушение приводит к **непредсказуемому поведению**.

### Три правила

**1. Если `a == b`, то `hash(a) == hash(b)`.**

Обязательное правило. Если нарушить — `set` и `dict` сломаются.

**2. `__eq__` должен быть рефлексивным, симметричным и транзитивным.**

- **Рефлексивность:** `a == a` всегда `True`.
- **Симметричность:** если `a == b`, то `b == a`.
- **Транзитивность:** если `a == b` и `b == c`, то `a == c`.

**3. Объекты, которые сравниваются, должны иметь одинаковые хэши — даже если они разных классов.**

Если `Studio` и `Apartment` могут быть равны, то их хэши должны совпадать.

### Пример правильной реализации

```python
class Apartment:
    def __init__(self, address, area):
        self.address = address
        self.area = area

    def __eq__(self, other):
        if not isinstance(other, Apartment):
            return NotImplemented
        return self.address == other.address and self.area == other.area

    def __hash__(self):
        # Используем те же поля, что и в __eq__.
        return hash((self.address, self.area))
```

### Схема

```
        __eq__              __hash__
        ──────              ────────
        address             (address,
        area          ←→     area)
                │            │
                └─────┬──────┘
                      ↓
                ОДНИ И ТЕ ЖЕ ПОЛЯ
```

---

## 12.6. Использование в `set` и `dict`

Разберём подробно, как объекты ведут себя в этих структурах.

### 12.6.1. Работа в `set`

**Множество** — коллекция **уникальных** элементов.

```python
flat1 = Apartment("ул. Ленина, д. 10, кв. 5", 65)
flat2 = Apartment("ул. Ленина, д. 10, кв. 5", 65)
flat3 = Apartment("ул. Садовая, д. 3, кв. 12", 80)

flats = {flat1, flat2, flat3}
print(len(flats))              # 2 — потому что flat1 и flat2 равны
```

**Что происходит под капотом:**

1. Когда вы добавляете `flat1`, `set` вычисляет `hash(flat1)` и кладёт объект в **соответствующую ячейку**.
2. Когда вы добавляете `flat2`, `set` вычисляет `hash(flat2)`. Если хэши **совпадают** — он проверяет через `__eq__`, равен ли `flat2` уже лежащему `flat1`. Если равны — **не добавляет**.
3. `flat3` имеет **другой хэш** — он добавляется.

### 12.6.2. Работа в `dict`

**Словарь** — коллекция пар «ключ-значение». Ключи должны быть **уникальными** и **хэшируемыми**.

```python
prices = {
    flat1: "8 500 000 ₽",
    flat3: "12 000 000 ₽",
}

print(prices[flat2])           # "8 500 000 ₽" — нашли по flat1 через хэш
```

**Что происходит под капотом:**

1. `flat1` используется как ключ. Python считает `hash(flat1)` и кладёт значение в ячейку.
2. Когда вы запрашиваете `prices[flat2]`, Python считает `hash(flat2)`. Хэш **совпадает** с хэшем `flat1`. Затем Python **проверяет через `__eq__`**, действительно ли `flat2` равен `flat1`. Если да — возвращает значение.

**Ключевой момент:** без `__hash__` такой поиск **не сработал бы**, потому что Python **не знал бы**, в какой ячейке искать.

### 12.6.3. Что нельзя использовать как ключ

**Изменяемые объекты нельзя делать хэшируемыми**, если их поля могут меняться.

```python
class Apartment:
    def __init__(self, address):
        self.address = address

    def __eq__(self, other):
        return self.address == other.address

    def __hash__(self):
        return hash(self.address)


flat = Apartment("ул. Ленина")
prices = {flat: "8 500 000 ₽"}

flat.address = "ул. Садовая"    # ← изменили адрес
print(prices[flat])              # KeyError! Хэш изменился, ключ не находится.
```

**Правило:** если объект **хэшируемый**, его **поля, участвующие в хэше**, не должны меняться.

**Решение:** либо делать объект **неизменяемым** (immutable), либо **не переопределять `__hash__`** (тогда объект нехэшируем).

## 12.7. `__eq__` и наследование

Магический метод `__eq__` наследуется, но есть **нюансы**.

### Пример с наследованием

```python
class Apartment:
    def __init__(self, address, area):
        self.address = address
        self.area = area

    def __eq__(self, other):
        if not isinstance(other, Apartment):
            return NotImplemented
        return self.address == other.address and self.area == other.area

    def __hash__(self):
        return hash((self.address, self.area))


class Studio(Apartment):
    pass
```

Что произойдёт:

```python
apartment = Apartment("ул. Ленина, д. 10, кв. 5", 65)
studio = Studio("ул. Ленина, д. 10, кв. 5", 65)

print(apartment == studio)     # True — Studio наследует Apartment и проходит isinstance
print(studio == apartment)     # True — симметрично
print(hash(apartment) == hash(studio))    # True — хэш одинаковый
```

**Почему `True`:**

- `isinstance(studio, Apartment)` — `True`, потому что `Studio` наследует `Apartment`.
- `self.address == other.address` — адреса совпадают.

**Это правильно?** Зависит от **логики**:

- Если «студия — это квартира» — да, они могут быть равны.
- Если «студия — это отдельный тип» — нет.

**Решение:** если хотите, чтобы наследник **не был равен** родителю, используйте **строгую проверку типа**:

```python
def __eq__(self, other):
    if type(self) is not type(other):
        return NotImplemented
    return self.address == other.address
```

`type(self) is type(other)` — проверяет **точное совпадение** классов.

### Схема

```
        isinstance(other, Apartment)          type(self) is type(other)
        ────────────────────────────          ──────────────────────────
        Studio == Apartment → True            Studio == Apartment → False
        (Studio — это Apartment)              (строгое совпадение классов)
```

**Выбор зависит от задачи.**

## 12.8. `__eq__` и другие операторы сравнения

`__eq__` — **только один** из операторов сравнения. Есть ещё:

| Оператор | Метод | Смысл |
|----------|-------|-------|
| `==` | `__eq__` | Равно |
| `!=` | `__ne__` | Не равно |
| `<` | `__lt__` | Меньше |
| `<=` | `__le__` | Меньше или равно |
| `>` | `__gt__` | Больше |
| `>=` | `__ge__` | Больше или равно |

### Нюанс: `__ne__` по умолчанию

В Python 3 `__ne__` **автоматически** выводится из `__eq__`:

```python
flat1 != flat2    #  → not (flat1 == flat2)
```

Поэтому **отдельно определять `__ne__` не нужно**, если правильно определён `__eq__`.

### Если нужны другие сравнения

Если нужны `<`, `>`, `<=`, `>=` — их нужно определять **отдельно**.

```python
class Apartment:
    def __init__(self, price):
        self.price = price

    def __lt__(self, other):
        return self.price < other.price

    def __le__(self, other):
        return self.price <= other.price
```

Теперь можно сортировать квартиры по цене:

```python
flats = [flat1, flat2, flat3]
flats.sort()                   # сортирует по цене
```

### Сравнение и `functools.total_ordering`

Если определять **все шесть** операторов сравнения — это **много кода**. Можно использовать декоратор `functools.total_ordering`:

```python
from functools import total_ordering

@total_ordering
class Apartment:
    def __init__(self, price):
        self.price = price

    def __eq__(self, other):
        return self.price == other.price

    def __lt__(self, other):
        return self.price < other.price
```

Теперь Python **автоматически** сгенерирует `<=`, `>`, `>=` на основе `__eq__` и `__lt__`.

---

## 12.9. `__eq__` и `__hash__` в реальных проектах

### 1. Модели Django

Django-модели **не переопределяют** `__eq__` по умолчанию — они сравниваются по **первичному ключу**.

### 2. Значения-объекты (value objects)

В **DDD** (Domain-Driven Design) есть понятие **value object** — объект, который определяется **только своими значениями**. Для них `__eq__` и `__hash__` обязательны.

```python
class Money:
    def __init__(self, amount, currency):
        self.amount = amount
        self.currency = currency

    def __eq__(self, other):
        if not isinstance(other, Money):
            return NotImplemented
        return self.amount == other.amount and self.currency == other.currency

    def __hash__(self):
        return hash((self.amount, self.currency))
```

### 3. Кэширование

Если объект — ключ в кэше, он **должен быть хэшируемым**:

```python
cache = {}
cache[apartment] = "8 500 000 ₽"
```

### 4. Уникальные коллекции

Если нужно хранить **уникальные** объекты — `set` требует `__hash__` и `__eq__`.

### 5. Сравнение в тестах

В тестах часто проверяют равенство объектов:

```python
assert actual_apartment == expected_apartment
```

Без `__eq__` это **не сработает**.

### 6. Наш дом

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area

    def __eq__(self, other):
        if not isinstance(other, Room):
            return NotImplemented
        return self.name == other.name and self.area == other.area

    def __hash__(self):
        return hash((self.name, self.area))
```

## 12.10. Нюансы и тонкости

### 12.10.1. `__eq__` должен возвращать `bool` или `NotImplemented`

**Правильно:**

```python
def __eq__(self, other):
    return self.address == other.address      # bool
```

**Правильно:**

```python
def __eq__(self, other):
    return NotImplemented
```

**Неправильно:**

```python
def __eq__(self, other):
    return "True"                             # строка — ошибка!
```

### 12.10.2. `NotImplemented` vs `NotImplementedError`

Это **разные вещи**:

- **`NotImplemented`** — специальное значение для операторов. **Не исключение**.
- **`NotImplementedError`** — исключение, которое **выбрасывается**.

В `__eq__` используйте **`NotImplemented`** (без `Error`).

### 12.10.3. `__hash__ = None`

Если вы определяете `__eq__`, но хотите, чтобы объект был **нехэшируемым**, явно укажите:

```python
class Apartment:
    def __eq__(self, other):
        return self.address == other.address

    __hash__ = None    # явно указываем: объект нехэшируем
```

Теперь попытка использовать в `set` даст понятную ошибку.

### 12.10.4. Хэш не должен меняться

Если объект попал в `set` или `dict`, его **хэш не должен меняться**. Поэтому **изменяемые** объекты **опасно** делать хэшируемыми.

### 12.10.5. Хэш не обязан быть уникальным

**Хэш** — это **не уникальный идентификатор**. Два разных объекта могут иметь **одинаковый хэш** (это называется **коллизия**). `set` и `dict` дополнительно проверяют через `__eq__`.

### 12.10.6. `hash()` возвращает `int`

`__hash__` должен возвращать **целое число**. Не строку, не float — **`int`**.

```python
def __hash__(self):
    return hash(self.address)     # hash() возвращает int
```

### 12.10.7. Порядок полей в хэше

Порядок полей в хэше **важен**:

```python
hash((self.address, self.area))    # НЕ равно
hash((self.area, self.address))    # этому
```

Главное — **согласованность** между `__eq__` и `__hash__`.

### 12.10.8. Мутабельность и хэшируемость

**Правило:**

- **Изменяемый объект** — **не должен быть хэшируемым**.
- **Неизменяемый объект** — **может быть хэшируемым**.

Пример неизменяемого объекта: `tuple`, `str`, `frozenset`.

### 12.10.9. `__eq__` и `__ne__` в Python 2 vs Python 3

В **Python 2** `__ne__` нужно было определять **отдельно**. В **Python 3** — **автоматически** выводится из `__eq__`. Мы пишем для Python 3.

### 12.10.10. Сравнение с `None`

Всегда проверяйте на `None` **до** сравнения:

```python
def __eq__(self, other):
    if other is None:
        return False
    if not isinstance(other, Apartment):
        return NotImplemented
    ...
```

Это помогает избежать **неожиданных ошибок**.

---

## 12.11. Типичные ошибки

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Определили `__eq__`, но не `__hash__` | Объект нехэшируем | Добавить `__hash__` |
| Определили `__hash__` без `__eq__` | Нет смысла, лучше не надо | Определять вместе |
| Хэш не согласован с `__eq__` | `set` и `dict` сломаются | Использовать те же поля |
| Не проверяют `isinstance` | Сравнение с чужим типом | Добавить проверку |
| Возвращают `False` вместо `NotImplemented` | Проблемы при сравнении с другими типами | Вернуть `NotImplemented` |
| Возвращают не `bool` | TypeError или странное поведение | Вернуть `True`/`False`/`NotImplemented` |
| Хэш на изменяемом поле | После изменения `set` сломается | Использовать неизменяемые поля |
| Забыли `return` в `__hash__` | Возвращается `None` | Вернуть `int` |
| Используют `NotImplementedError` вместо `NotImplemented` | Неправильное исключение | Использовать `NotImplemented` |
| Сравнивают по `is` вместо `==` | Не работает с данными | Использовать `==` |
| Забыли, что хэш — это `int` | Не число | `hash()` возвращает `int` |
| Не проверяют `None` | Может упасть | Проверять в начале |

---

## 12.12. Что важно запомнить

- **`__eq__`** — определяет, **когда два объекта равны** (через `==`).
- **`__hash__`** — определяет **хэш-значение** (для `set` и `dict`).
- **По умолчанию** `==` сравнивает **идентичность** (адрес в памяти).
- **`is`** — идентичность. **`==`** — равенство. Это **разные вещи**.
- **Контракт:** если `a == b`, то `hash(a) == hash(b)`.
- **Оба метода** должны использовать **одни и те же поля**.
- **Возвращайте `NotImplemented`** для чужих типов.
- **`NotImplemented`** — значение, а не исключение. Не путать с `NotImplementedError`.
- **Если определён `__eq__`, но не `__hash__`** — объект **нехэшируем**.
- **Хэш не должен меняться** — иначе `set` и `dict` сломаются.
- **Хэш не обязан быть уникальным** — коллизии возможны.
- **Изменяемые объекты** лучше **не делать хэшируемыми**.
- **`isinstance`** — учитывает наследников. **`type(self) is type(other)`** — строгая проверка.
- **`__ne__`** в Python 3 выводится из `__eq__` автоматически.

## Задача 12 (магические методы `__eq__` и `__hash__`)

В каталоге агентства недвижимости может оказаться **несколько записей об одной и той же квартире** — например, при повторной подаче объявления. Нужно, чтобы программа понимала, что это **одна и та же квартира**, и не дублировала её в списках.

**Часть 1. Класс `Apartment`.** Создайте класс `Apartment` с полями:

- `address` — адрес квартиры;
- `area` — площадь в м²;
- `price` — цена.

Реализуйте конструктор `__init__` и метод `__repr__` для технического вывода.

**Часть 2. Метод `__eq__`.** Реализуйте магический метод `__eq__`, чтобы две квартиры считались **равными**, если совпадают их **адрес и площадь**. Обязательно:

- проверяйте `isinstance(other, Apartment)`;
- возвращайте `NotImplemented` для чужих типов.

**Часть 3. Метод `__hash__`.** Реализуйте магический метод `__hash__`, чтобы квартиры можно было хранить в `set` и использовать как ключи в `dict`. Хэш должен вычисляться **по тем же полям**, что и `__eq__`: адрес и площадь.

**Часть 4. Проверка в `set`.** Создайте **три** квартиры:

- `flat1` — «ул. Ленина, д. 10, кв. 5», 65 м², 8 500 000 ₽;
- `flat2` — те же данные, что у `flat1` (дубликат);
- `flat3` — «ул. Садовая, д. 3, кв. 12», 80 м², 12 000 000 ₽.

Продемонстрируйте:

- `flat1 == flat2` → `True`;
- `flat1 == flat3` → `False`;
- `hash(flat1) == hash(flat2)` → `True`;
- `len({flat1, flat2, flat3})` → `2` (дубликат не добавился).

**Часть 5. Проверка в `dict`.** Создайте словарь с ценами, где ключами будут квартиры:

```python
prices = {flat1: "8 500 000 ₽", flat3: "12 000 000 ₽"}
```

Убедитесь, что:

- `prices[flat2]` возвращает цену `flat1` — потому что `flat2 == flat1`;
- длина словаря — `2`, а не `3`.

**Часть 6. Наследник `Studio`.** Создайте наследника `Studio(Apartment)` с теми же полями. Убедитесь, что:

- `Studio("ул. Ленина, д. 10, кв. 5", 65, 8_500_000) == flat1` → `True` (потому что `Studio` наследует `Apartment` и проходит `isinstance`);
- `hash` тоже совпадает.

**Часть 7. Строгая проверка типа.** Создайте **второй** класс `StudioStrict(Apartment)`, в котором `__eq__` использует **строгую проверку** `type(self) is type(other)`. Убедитесь, что:

- `StudioStrict("ул. Ленина, д. 10, кв. 5", 65, 8_500_000) == flat1` → `False` (типы строго не совпадают).

**Часть 8.** В комментариях объясните:

1. Чем `==` отличается от `is`.
2. Почему при переопределении `__eq__` нужно переопределять `__hash__`.
3. Что произойдёт, если этого не сделать.
4. Зачем возвращать `NotImplemented` для чужих типов.
5. В чём разница между `isinstance(other, Apartment)` и `type(self) is type(other)`.
6. Почему изменяемые объекты опасно делать хэшируемыми.

### Решение

```python
# ─── Часть 1. Класс Apartment ───
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    # ─── Часть 2. __eq__ — сравнение по данным ───
    def __eq__(self, other):
        # Для чужих типов — NotImplemented.
        if not isinstance(other, Apartment):
            return NotImplemented
        # Сравниваем по адресу и площади.
        return self.address == other.address and self.area == other.area

    # ─── Часть 3. __hash__ — по тем же полям ───
    def __hash__(self):
        # Хэш от кортежа (address, area) — тех же полей, что и в __eq__.
        return hash((self.address, self.area))

    # __repr__ для отладки.
    def __repr__(self):
        return (
            f"{self.__class__.__name__}("
            f"{self.address!r}, {self.area!r}, {self.price!r}"
            f")"
        )


# ─── Часть 6. Наследник Studio ───
class Studio(Apartment):
    pass


# ─── Часть 7. Наследник StudioStrict со строгой проверкой ───
class StudioStrict(Apartment):
    def __eq__(self, other):
        # Строгая проверка: типы должны СОВПАДАТЬ точно.
        if type(self) is not type(other):
            return NotImplemented
        return self.address == other.address and self.area == other.area

    def __hash__(self):
        return hash((self.address, self.area))


# ─── Часть 4. Проверка в set ───
flat1 = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)
flat2 = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)
flat3 = Apartment("ул. Садовая, д. 3, кв. 12", 80, 12_000_000)

print("flat1 == flat2:", flat1 == flat2)                    # True
print("flat1 == flat3:", flat1 == flat3)                    # False
print("hash(flat1) == hash(flat2):", hash(flat1) == hash(flat2))    # True
print("len({flat1, flat2, flat3}):", len({flat1, flat2, flat3}))    # 2

# ─── Часть 5. Проверка в dict ───
prices = {
    flat1: "8 500 000 ₽",
    flat3: "12 000 000 ₽",
}
print("prices[flat2]:", prices[flat2])                      # 8 500 000 ₽
print("len(prices):", len(prices))                          # 2

# ─── Часть 6. Наследник Studio ───
studio = Studio("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)
print("studio == flat1:", studio == flat1)                  # True
print("hash(studio) == hash(flat1):", hash(studio) == hash(flat1))  # True

# ─── Часть 7. Строгая проверка типа ───
studio_strict = StudioStrict("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)
print("studio_strict == flat1:", studio_strict == flat1)    # False
print("studio_strict == studio_strict:", studio_strict == studio_strict)    # True
```

**Ожидаемый вывод:**

```
flat1 == flat2: True
flat1 == flat3: False
hash(flat1) == hash(flat2): True
len({flat1, flat2, flat3}): 2
prices[flat2]: 8 500 000 ₽
len(prices): 2
studio == flat1: True
hash(studio) == hash(flat1): True
studio_strict == flat1: False
studio_strict == studio_strict: True
```

### Разбор

**1. Чем `==` отличается от `is`.**

- **`==`** — оператор **равенства**. Вызывает `__eq__`. **Переопределяется** в классе. Сравнивает **данные**.
- **`is`** — оператор **идентичности**. Сравнивает **адреса в памяти**. **Не переопределяется**. Всегда проверяет, **один ли это объект**.

```python
flat1 == flat2          # True — данные совпадают
flat1 is flat2          # False — разные объекты
```

**Правило:** для сравнения **значений** используйте `==`. Для проверки **идентичности** (например, `is None`) — `is`.

**2. Почему при переопределении `__eq__` нужно переопределять `__hash__`.**

`set` и `dict` **сначала** смотрят **хэш**, и только **потом** — `__eq__`. Если хэши **разные**, они даже не будут сравнивать объекты. А если объекты **равны**, но хэши **разные** — `set` и `dict` **не смогут найти** объект по хэшу.

Это **нарушает контракт**:

> Если `a == b`, то `hash(a) == hash(b)`.

Python **защищает** от такой ошибки: если определён `__eq__`, но не `__hash__`, объект становится **нехэшируемым**. Это заставляет программиста **явно** решить, как считать хэш.

**3. Что произойдёт, если этого не сделать.**

```python
class Apartment:
    def __init__(self, address):
        self.address = address

    def __eq__(self, other):
        return self.address == other.address


flat = Apartment("ул. Ленина")
{flat}              # TypeError: unhashable type: 'Apartment'
```

Python **не даст** использовать объект в `set` или `dict`. Это **сознательное** решение: лучше явная ошибка, чем непредсказуемое поведение.

**4. Зачем возвращать `NotImplemented` для чужих типов.**

Если `__eq__` **не проверяет тип**, он может попытаться сравнить **разные вещи**:

```python
def __eq__(self, other):
    return self.address == other.address      # ← other может не иметь .address
```

Если `other` — строка, будет `AttributeError`. Если `other` — число, может быть **неожиданный результат**.

`NotImplemented` говорит Python: «я не знаю, как сравнить — попробуй **другую сторону**». Python попробует `other.__eq__(self)`. Если и тот вернёт `NotImplemented` — итог будет `False`.

**5. В чём разница между `isinstance(other, Apartment)` и `type(self) is type(other)`.**

- **`isinstance(other, Apartment)`** — проверяет, является ли `other` **экземпляром `Apartment`** **или его наследником**. Это **широкая** проверка. `Studio == Apartment` → `True`.
- **`type(self) is type(other)`** — проверяет **точное совпадение** классов. Это **строгая** проверка. `Studio == Apartment` → `False`.

**Когда что использовать:**

- **`isinstance`** — если наследники **могут быть равны** родителю (например, «студия — это квартира»).
- **`type(...) is type(...)`** — если наследники **должны быть равны только сами себе** (например, «студия — отдельный тип»).

**6. Почему изменяемые объекты опасно делать хэшируемыми.**

Хэш используется для **поиска** в `set` и `dict`. Когда объект добавляется, Python **запоминает его хэш** и кладёт в нужную ячейку. Если объект **изменить**, его хэш **может измениться** — но `set` об этом не узнает. В итоге:

- Объект лежит в **одной** ячейке (по старому хэшу).
- Поиск идёт по **новому** хэшу — **другая ячейка**.
- Объект **не находится** — `KeyError`.

**Решение:** либо делать объект **неизменяемым**, либо **не переопределять `__hash__`** (тогда объект нехэшируем, и ошибка будет **сразу** — при попытке использования в `set`).

## 13.1. Проблема: объекты не умеют складываться

Вернёмся к агентству недвижимости. У нас есть жилые комплексы — `Complex`. У каждого комплекса есть **список квартир**:

```python
class Complex:
    def __init__(self, name, apartments):
        self.name = name
        self.apartments = apartments
```

Создадим два комплекса:

```python
complex1 = Complex("ЖК Ленина", [flat1, flat2])
complex2 = Complex("ЖК Садовая", [flat3, flat4])
```

А теперь представим: два ЖК **объединились**. Хочется написать так:

```python
combined = complex1 + complex2
```

Но Python **не знает**, что значит «сложить два комплекса». Он выдаст ошибку:

```
TypeError: unsupported operand type(s) for +: 'Complex' and 'Complex'
```

### Почему так

Python **умеет** складывать:

- **числа:** `2 + 3 = 5`
- **строки:** `"a" + "b" = "ab"`
- **списки:** `[1, 2] + [3, 4] = [1, 2, 3, 4]`

Это работает потому, что у этих типов есть **магический метод `__add__`**. У `int`, `str`, `list` — свой `__add__`. А у нашего `Complex` — **нет**.

### Аналогия

Представь, что ты пришёл в **новый магазин** и просишь: «Пробейте мне два товара вместе». Кассир не знает, что делать — **у него нет инструкции** для этого. Но если ты дашь ему **инструкцию** («сложи цены и дай скидку 5%») — он справится.

Магические методы — это **инструкции** для Python. Мы говорим: «когда складывают два `Complex`, делай **вот так**».

### Вывод

По умолчанию объекты **не умеют** складываться. Чтобы научить — нужен **магический метод `__add__`**.

---

## 13.2. Что такое перегрузка операторов

**Перегрузка операторов** — это возможность **научить** объект **вести себя как встроенный тип** при работе с операторами.

### Что можно перегружать

Почти **все** операторы Python:

| Оператор | Магический метод | Смысл |
|----------|------------------|-------|
| `+` | `__add__` | Сложение |
| `-` | `__sub__` | Вычитание |
| `*` | `__mul__` | Умножение |
| `/` | `__truediv__` | Деление |
| `==` | `__eq__` | Равенство (глава 12) |
| `<` | `__lt__` | Меньше |
| `len()` | `__len__` | Длина |
| `[]` | `__getitem__` | Индексация |
| `in` | `__contains__` | Проверка наличия |
| `str()` | `__str__` | Строковое представление (глава 11) |

### Зачем это нужно

Перегрузка операторов делает код **естественнее**:

**Без перегрузки:**

```python
combined = complex1.merge(complex2)     # метод merge
total = complex1.get_apartment_count()  # метод get_apartment_count
```

**С перегрузкой:**

```python
combined = complex1 + complex2          # как складываем числа
total = len(complex1)                   # как узнаём длину списка
```

Второй вариант **короче** и **понятнее**. Он **похож** на работу с встроенными типами.

### Вывод

Перегрузка операторов — способ **сделать объекты «похожими» на встроенные типы**. Код становится **естественнее** и **короче**.

---

## 13.3. Магический метод `__add__` — сложение

**`__add__`** — это магический метод, который Python вызывает при **использовании оператора `+`**.

### Как работает

Когда Python видит `a + b`, он:

1. Смотрит на **тип `a`**.
2. Ищет метод `a.__add__(b)`.
3. Если находит — вызывает его и возвращает результат.
4. Если не находит — пробует `b.__radd__(a)` (см. 13.7).
5. Если и там нет — `TypeError`.

### Синтаксис

```python
class Complex:
    def __init__(self, name, apartments):
        self.name = name
        self.apartments = apartments

    def __add__(self, other):
        # Возвращаем НОВЫЙ объект — результат сложения.
        return Complex(
            f"{self.name} + {other.name}",
            self.apartments + other.apartments,
        )
```

### Разбор кода

- **`def __add__(self, other):`** — метод принимает **два** аргумента: `self` (левый объект) и `other` (правый объект).
- **`self.name + " + " + other.name`** — создаём новое имя.
- **`self.apartments + other.apartments`** — объединяем списки.
- **`return Complex(...)`** — возвращаем **новый** объект.

### Использование

```python
complex1 = Complex("ЖК Ленина", [flat1, flat2])
complex2 = Complex("ЖК Садовая", [flat3, flat4])

combined = complex1 + complex2

print(combined.name)             # ЖК Ленина + ЖК Садовая
print(len(combined.apartments))  # 4
```

Работает! Теперь `Complex` **умеет складываться**.

### Почему `__add__` возвращает **новый** объект

Это **важное правило**. Оператор `+` **не должен** изменять существующие объекты. Он **создаёт новый**.

**Правильно:**

```python
def __add__(self, other):
    return Complex(...)     # новый объект
```

**Неправильно:**

```python
def __add__(self, other):
    self.apartments += other.apartments    # ← изменяет self!
    return self
```

**Почему:** в математике `a + b` **не меняет** ни `a`, ни `b`. Это **ожидаемое** поведение. Если нарушить — программисты будут **удивлены**.

### Аналогия

Представь, что ты складываешь два числа: `2 + 3 = 5`. Число `2` **не стало** `5`. Создалось **новое** число `5`. То же самое с объектами.

### Вывод

`__add__` определяет, что значит `+` для объекта. **Всегда возвращает новый объект**, не меняя существующие.

---

## 13.4. Что если `other` — другого типа

Что, если сложить `Complex` и число?

```python
complex1 + 5    # ← что должно быть?
```

Если `__add__` не проверяет тип, может быть **ошибка**:

```python
def __add__(self, other):
    return Complex(self.name + other.name, ...)    # other.name — AttributeError!
```

### Правильное решение — проверка типа

```python
def __add__(self, other):
    if not isinstance(other, Complex):
        return NotImplemented
    return Complex(
        f"{self.name} + {other.name}",
        self.apartments + other.apartments,
    )
```

### Что такое `NotImplemented`

**`NotImplemented`** — это специальное **значение** (не исключение!). Оно говорит Python: «я не знаю, как сложить эти объекты — попробуй **другую сторону**».

Python попробует `5.__radd__(complex1)`. Если и там нет — `TypeError`.

### Нюанс: `NotImplemented` vs `NotImplementedError`

- **`NotImplemented`** — **значение** (без `Error`). Используется в магических методах.
- **`NotImplementedError`** — **исключение**. Используется, когда метод **не реализован** (например, в абстрактных классах).

**Не путайте!**

### Вывод

Всегда проверяйте тип `other` через `isinstance`. Для чужих типов возвращайте **`NotImplemented`** — это позволяет Python корректно обработать ситуацию.

---

## 13.5. Магический метод `__len__` — длина

**`__len__`** — это магический метод, который Python вызывает при **использовании функции `len()`**.

### Как работает

Когда Python видит `len(obj)`, он:

1. Смотрит на **тип `obj`**.
2. Ищет метод `obj.__len__()`.
3. Если находит — вызывает и возвращает результат.
4. Если не находит — `TypeError`.

### Синтаксис

```python
class Complex:
    def __init__(self, name, apartments):
        self.name = name
        self.apartments = apartments

    def __len__(self):
        # Возвращаем количество квартир в комплексе.
        return len(self.apartments)
```

### Разбор кода

- **`def __len__(self):`** — метод принимает **только `self`**.
- **`return len(self.apartments)`** — возвращает **целое число**.

### Использование

```python
complex1 = Complex("ЖК Ленина", [flat1, flat2, flat3])

print(len(complex1))              # 3
```

Работает! Теперь `Complex` **умеет** возвращать длину.

### Что должен возвращать `__len__`

**Только `int`** — целое неотрицательное число. Не float, не строку.

```python
def __len__(self):
    return 3.14     # TypeError: 'float' object cannot be interpreted as an integer
```

### Зачем это нужно

Благодаря `__len__` объект становится **«правдивым»** в **условиях**:

```python
complex1 = Complex("ЖК Ленина", [])
complex2 = Complex("ЖК Садовая", [flat1, flat2])

if complex1:                     # False — потому что len == 0
    print("Есть квартиры")

if complex2:                     # True — потому что len == 2
    print("Есть квартиры")
```

Python **автоматически** использует `__len__` для проверки на «истинность». Пустой объект — `False`, непустой — `True`.

### Аналогия

Представь **коробку**. `len(box)` — сколько в ней предметов. Пустая коробка — `0`, значит `False`. Полная — `True`. Это **естественно** и **удобно**.

### Вывод

`__len__` определяет длину объекта. Возвращает **целое число**. Благодаря ему объект **автоматически** получает «истинность» в условиях.

---

## 13.6. Другие магические методы для операторов

`__add__` и `__len__` — **не единственные**. Есть **много** других.

### Арифметические операторы

| Оператор | Метод | Пример |
|----------|-------|--------|
| `+` | `__add__` | `a + b` |
| `-` | `__sub__` | `a - b` |
| `*` | `__mul__` | `a * b` |
| `/` | `__truediv__` | `a / b` |
| `//` | `__floordiv__` | `a // b` |
| `%` | `__mod__` | `a % b` |
| `**` | `__pow__` | `a ** b` |

### Операторы сравнения

| Оператор | Метод | Пример |
|----------|-------|--------|
| `==` | `__eq__` | `a == b` (глава 12) |
| `!=` | `__ne__` | `a != b` |
| `<` | `__lt__` | `a < b` |
| `<=` | `__le__` | `a <= b` |
| `>` | `__gt__` | `a > b` |
| `>=` | `__ge__` | `a >= b` |

### Операторы для коллекций

| Оператор | Метод | Пример |
|----------|-------|--------|
| `len()` | `__len__` | `len(a)` |
| `[]` | `__getitem__` | `a[0]` |
| `[] =` | `__setitem__` | `a[0] = x` |
| `del []` | `__delitem__` | `del a[0]` |
| `in` | `__contains__` | `x in a` |
| `iter()` | `__iter__` | `for x in a` |

### Пример: `__getitem__`

```python
class Complex:
    def __init__(self, apartments):
        self.apartments = apartments

    def __getitem__(self, index):
        return self.apartments[index]


complex1 = Complex([flat1, flat2, flat3])
print(complex1[0])              # flat1
print(complex1[0:2])            # [flat1, flat2]
```

Теперь `Complex` можно **индексировать** как список.

### Пример: `__contains__`

```python
class Complex:
    def __init__(self, apartments):
        self.apartments = apartments

    def __contains__(self, apartment):
        return apartment in self.apartments


if flat1 in complex1:
    print("Квартира в комплексе")
```

### Пример: `__sub__`

```python
class Complex:
    def __init__(self, apartments):
        self.apartments = apartments

    def __sub__(self, other):
        # Убираем из первого комплекса все квартиры второго.
        return Complex([a for a in self.apartments if a not in other.apartments])
```

### Вывод

Магических методов **много**. Каждый отвечает за **свой оператор**. Изучив `__add__` и `__len__`, вы поймёте **принцип** — остальные работают так же.

---

## 13.7. `__radd__` — обратное сложение

Что, если `other` стоит **слева**, а наш объект — **справа**?

```python
5 + complex1        # ← 5.__add__(complex1) вернёт NotImplemented
                    #   Python попробует complex1.__radd__(5)
```

### Синтаксис

```python
class Complex:
    def __radd__(self, other):
        # other — то, что стоит слева.
        # self — наш объект (стоит справа).
        return Complex(f"Unknown + {self.name}", self.apartments)
```

### Как работает

Когда Python видит `a + b`:

1. Пробует `a.__add__(b)`.
2. Если не работает — пробует `b.__radd__(a)`.
3. Если и это не работает — `TypeError`.

### Когда это нужно

- Когда ваш объект **должен** уметь складываться с **другими типами**.
- Когда ваш объект — **«правый»** в выражении.

### Пример

```python
class Money:
    def __init__(self, amount):
        self.amount = amount

    def __add__(self, other):
        if isinstance(other, Money):
            return Money(self.amount + other.amount)
        if isinstance(other, (int, float)):
            return Money(self.amount + other)
        return NotImplemented

    def __radd__(self, other):
        # 100 + money → money.__radd__(100)
        return self.__add__(other)
```

Теперь:

```python
money = Money(500)
print((money + 100).amount)    # 600
print((100 + money).amount)    # 600 — работает через __radd__
```

### Вывод

`__radd__` нужен, когда объект может быть **справа** от оператора. Часто его реализуют **через `__add__`**, чтобы не дублировать логику.

---

## 13.8. `__iadd__` — сложение с присваиванием

Что, если хочется поддержать `+=`?

```python
complex1 += complex2
```

### Как работает

Python видит `a += b` и:

1. Пробует `a.__iadd__(b)`.
2. Если не находит — пробует `a = a + b` (через `__add__`).

### Синтаксис

```python
class Complex:
    def __iadd__(self, other):
        # Изменяем self и возвращаем его.
        self.apartments += other.apartments
        return self
```

### Разница с `__add__`

- **`__add__`** — возвращает **новый** объект.
- **`__iadd__`** — может **изменить** существующий и вернуть его.

### Когда использовать

- Когда `+=` должен быть **быстрее**, чем `a = a + b`.
- Когда нужно **изменить** объект «на месте».

### Пример

```python
complex1 = Complex("ЖК Ленина", [flat1])
complex2 = Complex("ЖК Садовая", [flat2])

complex1 += complex2
print(complex1.name)              # ЖК Ленина (изменился на месте)
print(len(complex1))              # 2
```

### Вывод

`__iadd__` — для `+=`. Позволяет **изменить** объект «на месте», а не создавать новый.

---

## 13.9. Полный пример: класс `Complex`

Соберём всё вместе.

```python
class Complex:
    def __init__(self, name, apartments):
        self.name = name
        self.apartments = apartments

    # Сложение: complex1 + complex2
    def __add__(self, other):
        if not isinstance(other, Complex):
            return NotImplemented
        return Complex(
            f"{self.name} + {other.name}",
            self.apartments + other.apartments,
        )

    # Длина: len(complex1)
    def __len__(self):
        return len(self.apartments)

    # Индексация: complex1[0]
    def __getitem__(self, index):
        return self.apartments[index]

    # Проверка наличия: flat in complex1
    def __contains__(self, apartment):
        return apartment in self.apartments

    # Строковое представление
    def __str__(self):
        return f"Комплекс \"{self.name}\" ({len(self)} квартир)"

    def __repr__(self):
        return f"Complex({self.name!r}, {len(self)} квартир)"
```

### Использование

```python
complex1 = Complex("ЖК Ленина", [flat1, flat2])
complex2 = Complex("ЖК Садовая", [flat3])

combined = complex1 + complex2
print(combined.name)             # ЖК Ленина + ЖК Садовая
print(len(combined))             # 3
print(combined[0])               # flat1
print(flat3 in combined)         # True
```

### Вывод

Один класс — **много возможностей**. Благодаря магическим методам `Complex` **ведёт себя** как встроенный тип: **складывается**, имеет **длину**, **индексируется**, поддерживает **`in`**.

---

## 13.10. Нюансы и тонкости

### 13.10.1. `__add__` не должен изменять объекты

Это **правило математики**. `a + b` создаёт **новый** объект.

### 13.10.2. Всегда проверяйте тип `other`

Без проверки — **ошибки** при сравнении с чужими типами. Используйте `isinstance` и `NotImplemented`.

### 13.10.3. `__len__` возвращает `int`

**Целое неотрицательное** число. Не float, не строку.

### 13.10.4. `__len__` влияет на истинность

`if obj:` работает через `__len__` (или `__bool__`). Пустой объект — `False`.

### 13.10.5. `__iadd__` может изменить `self`

Это **разрешено** и **ожидаемо** для `+=`. Но не забывайте возвращать `self`.

### 13.10.6. `__radd__` для симметрии

Если объект может быть **справа** от оператора — нужен `__radd__`. Часто реализуется **через `__add__`**.

### 13.10.7. Порядок методов

Python **сначала** пробует `__add__` левого объекта, **потом** `__radd__` правого.

### 13.10.8. `NotImplemented` — значение, не исключение

**Без `Error`**. Если написать `NotImplementedError` — будет **ошибка**.

### 13.10.9. `__add__` для разных типов

Можно поддержать сложение с **числами**, **строками** и т. д. Проверяйте тип через `isinstance` и **выбирайте** логику.

### 13.10.10. Магические методы **не вызывают** друг друга напрямую

**Не** `self.__add__(other)`, а **`self + other`**. Python **сам** вызовет `__add__`.

### Вывод

Магические методы — **мощный** инструмент, но требуют **аккуратности**. Проверяйте типы, возвращайте `NotImplemented`, не изменяйте объекты в `__add__`.

---

## 13.11. Типичные ошибки

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| `__add__` изменяет `self` | `a + b` не должен менять `a` | Возвращать новый объект |
| Нет проверки типа | `AttributeError` при чужом типе | Проверять `isinstance` |
| `NotImplementedError` вместо `NotImplemented` | Неправильное исключение | Использовать `NotImplemented` |
| `__len__` возвращает float | `TypeError` | Вернуть `int` |
| `__len__` возвращает отрицательное | `ValueError` | Возвращать `>= 0` |
| Забыли `return` в `__add__` | Возвращается `None` | Вернуть объект |
| Забыли `return self` в `__iadd__` | `complex += complex2` вернёт `None` | Вернуть `self` |
| Не реализовали `__radd__` | `5 + complex` не работает | Реализовать `__radd__` |
| Используют `self.__add__(other)` напрямую | Неправильно | Использовать `self + other` |
| `__add__` возвращает не тот тип | Странное поведение | Возвращать свой класс |

---

## 13.12. Что важно запомнить

- **Перегрузка операторов** — способ **научить объекты** работать с операторами.
- **`__add__`** — сложение через `+`.
- **`__len__`** — длина через `len()`.
- **`__add__` возвращает новый объект**, не изменяет существующие.
- **`__len__` возвращает `int`**, влияет на истинность.
- **Проверяйте тип `other`** через `isinstance`.
- **`NotImplemented`** — для чужих типов (без `Error`).
- **`__radd__`** — для объекта **справа** от оператора.
- **`__iadd__`** — для `+=` (может изменять `self`).
- **Магических методов много** — `__sub__`, `__mul__`, `__getitem__`, `__contains__` и т. д.
- **Принцип один:** переопределите метод — объект получит «встроенное» поведение.

## Задача 13 (магические методы `__add__` и `__len__`)

В агентстве недвижимости есть **жилые комплексы**. Иногда их нужно **объединять** (например, при слиянии двух ЖК) или узнавать, **сколько квартир** в комплексе.

**Часть 1. Класс `Apartment`.** Создайте простой класс `Apartment` с полем `address` и методом `__repr__`. Создайте несколько экземпляров для тестов.

**Часть 2. Класс `Complex`.** Создайте класс `Complex` (жилой комплекс) с полями:

- `name` — название комплекса;
- `apartments` — **список** квартир.

Реализуйте конструктор.

**Часть 3. Метод `__add__`.** Реализуйте магический метод `__add__` для **объединения** двух комплексов в один. Требования:

- проверяйте `isinstance(other, Complex)`;
- для чужих типов возвращайте `NotImplemented`;
- возвращайте **новый** объект `Complex`;
- имя нового комплекса — `"<name1> + <name2>"`;
- список квартир — **объединение** списков.

**Часть 4. Метод `__len__`.** Реализуйте магический метод `__len__`, возвращающий **количество квартир** в комплексе.

**Часть 5. Метод `__repr__`.** Реализуйте `__repr__` для удобного вывода:

```
Complex('ЖК Ленина', 2 квартиры)
```

**Часть 6. Проверка.** Продемонстрируйте:

- создание двух комплексов;
- объединение: `combined = complex1 + complex2`;
- вывод имени объединённого комплекса;
- `len(combined)`;
- `len(complex1)`, `len(complex2)`;
- работу `if complex1:` (пустой vs непустой);
- попытку сложить `complex1 + 5` — должно вернуть `TypeError`.

**Часть 7. Дополнительно.** Реализуйте:

- `__getitem__` — чтобы можно было получить квартиру по индексу: `complex1[0]`;
- `__contains__` — чтобы работал оператор `in`: `flat1 in complex1`.

**Часть 8.** В комментариях объясните:

1. Почему `__add__` возвращает **новый** объект, а не изменяет существующий.
2. Почему важно проверять `isinstance(other, Complex)`.
3. Что такое `NotImplemented` и почему это **не** `NotImplementedError`.
4. Почему `__len__` должен возвращать именно `int`.
5. Как `__len__` влияет на работу `if complex1:`.
6. Как магические методы делают объекты **«похожими»** на встроенные типы.

### Решение

```python
# ─── Часть 1. Класс Apartment ───
class Apartment:
    def __init__(self, address):
        self.address = address

    def __repr__(self):
        return f"Apartment({self.address!r})"


# ─── Часть 2. Класс Complex ───
class Complex:
    def __init__(self, name, apartments):
        self.name = name
        self.apartments = apartments

    # ─── Часть 3. __add__ — объединение комплексов ───
    def __add__(self, other):
        # Проверяем тип — только Complex можно складывать.
        if not isinstance(other, Complex):
            return NotImplemented
        # Возвращаем НОВЫЙ объект.
        return Complex(
            f"{self.name} + {other.name}",
            self.apartments + other.apartments,
        )

    # ─── Часть 4. __len__ — количество квартир ───
    def __len__(self):
        return len(self.apartments)

    # ─── Часть 7. __getitem__ — доступ по индексу ───
    def __getitem__(self, index):
        return self.apartments[index]

    # ─── Часть 7. __contains__ — оператор in ───
    def __contains__(self, apartment):
        return apartment in self.apartments

    # ─── Часть 5. __repr__ ───
    def __repr__(self):
        return f"Complex({self.name!r}, {len(self)} квартиры)"


# ─── Часть 6. Проверка ───
flat1 = Apartment("ул. Ленина, д. 10, кв. 5")
flat2 = Apartment("ул. Ленина, д. 10, кв. 6")
flat3 = Apartment("ул. Садовая, д. 3, кв. 12")
flat4 = Apartment("ул. Садовая, д. 3, кв. 13")

complex1 = Complex("ЖК Ленина", [flat1, flat2])
complex2 = Complex("ЖК Садовая", [flat3, flat4])

print("=== Исходные комплексы ===")
print(complex1)
print(complex2)

print()
print("=== Объединение ===")
combined = complex1 + complex2
print(combined)
print("Имя:", combined.name)
print("Количество квартир:", len(combined))

print()
print("=== Длины ===")
print("complex1:", len(complex1))
print("complex2:", len(complex2))

print()
print("=== Проверка на истинность ===")
empty = Complex("Пустой", [])
if empty:
    print("Пустой — True")
else:
    print("Пустой — False")

if complex1:
    print("complex1 — True")

print()
print("=== Индексация и in ===")
print("Первая квартира:", combined[0])
print("flat3 в complex1:", flat3 in complex1)
print("flat3 в combined:", flat3 in combined)

print()
print("=== Сложение с числом ===")
try:
    result = complex1 + 5
except TypeError as e:
    print("TypeError:", e)
```

**Ожидаемый вывод:**

```
=== Исходные комплексы ===
Complex('ЖК Ленина', 2 квартиры)
Complex('ЖК Садовая', 2 квартиры)

=== Объединение ===
Complex('ЖК Ленина + ЖК Садовая', 4 квартиры)
Имя: ЖК Ленина + ЖК Садовая
Количество квартир: 4

=== Длины ===
complex1: 2
complex2: 2

=== Проверка на истинность ===
Пустой — False
complex1 — True

=== Индексация и in ===
Первая квартира: Apartment('ул. Ленина, д. 10, кв. 5')
flat3 в complex1: False
flat3 в combined: True

=== Сложение с числом ===
TypeError: unsupported operand type(s) for +: 'Complex' and 'int'
```

### Разбор

**1. Почему `__add__` возвращает новый объект, а не изменяет существующий.**

Оператор `+` в математике **не изменяет** слагаемые. `2 + 3` **не превращает** `2` в `5` — создаётся **новое** число. То же самое должно быть с объектами.

Если `__add__` **изменяет** `self`, происходит **неожиданное** поведение:

```python
combined = complex1 + complex2
print(len(complex1))    # ожидаем 2, но получим 4 — complex1 изменился!
```

Это **нарушает** принцип «наименьшего удивления». Программист **ожидает**, что `+` создаёт новое, а не меняет старое.

**2. Почему важно проверять `isinstance(other, Complex)`.**

Без проверки `__add__` попытается обратиться к `other.name` или `other.apartments`. Если `other` — **не** `Complex` (например, число), будет **`AttributeError`**:

```python
complex1 + 5    # AttributeError: 'int' object has no attribute 'name'
```

С проверкой мы **вовремя** возвращаем `NotImplemented`. Python **сам** решает, что делать дальше, и в итоге **корректно** выдаёт `TypeError` — стандартное поведение для **неподдерживаемой** операции.

**3. Что такое `NotImplemented` и почему это не `NotImplementedError`.**

- **`NotImplemented`** — **значение** (без `Error`). Специальный **маркер**: «я не знаю, как это сделать».
- **`NotImplementedError`** — **исключение** (с `Error`). Используется, когда **метод не реализован** (например, в абстрактных классах — см. главу 7).

**В магических методах** используется **`NotImplemented`**. Python **сам** разберётся, что делать дальше (попробует `__radd__` другой стороны, потом `TypeError`).

Если написать `NotImplementedError` — Python **выбросит** это исключение, и программа **упадёт** там, где **должна** была бы просто выдать `TypeError`. Это **неправильно**.

**4. Почему `__len__` должен возвращать именно `int`.**

`len()` — это **функция**, которая возвращает **целое неотрицательное** число. Это **стандарт** Python.

Если `__len__` вернёт **float** — Python выдаст `TypeError`:

```python
def __len__(self):
    return 3.14    # TypeError
```

Если вернёт **отрицательное** — тоже ошибка:

```python
def __len__(self):
    return -1      # ValueError
```

**Правило:** возвращайте **`int >= 0`**.

**5. Как `__len__` влияет на работу `if complex1:`.**

Когда Python видит `if obj:`, он **сначала** пробует `obj.__bool__()`. Если `__bool__` **не определён** — использует `obj.__len__()`. Если `len == 0` — `False`, иначе — `True`.

**Пример:**

- `if empty:` → `len(empty) == 0` → `False`.
- `if complex1:` → `len(complex1) == 2` → `True`.

Это **очень удобно**: пустые объекты ведут себя как **пустые коллекции** (`[]`, `""`, `{}`).

**6. Как магические методы делают объекты «похожими» на встроенные типы.**

Благодаря магическим методам `Complex` **ведёт себя** как:

- **число** — при `complex1 + complex2`;
- **коллекция** — при `len(complex1)`, `complex1[0]`, `flat in complex1`;
- **объект** — при `print(complex1)`, `repr(complex1)`.

Программист **не думает** о том, что `Complex` — это **не** встроенный тип. Он работает с ним **так же**, как со списком или числом. Это **главная цель** магических методов — сделать код **естественным** и **понятным**.

# Глава 14. Свойство `@property`

## 14.1. Проблема: атрибуты без контроля

Вернёмся к агентству недвижимости. У нас есть класс `Apartment`:

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price
```

Это **публичные** атрибуты. Любой может изменить их **как угодно**:

```python
flat = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)

flat.area = -10          # отрицательная площадь — бессмыслица
flat.price = "дорого"    # цена — строка, а не число
flat.address = None      # адрес пропал
```

Программа **не сломается сразу**. Но данные стали **бессмысленными**. А поймать ошибку сложно: она проявится **позже**, когда что-то пойдёт не так.

Представь, что площадь стала отрицательной. Через неделю кто-то попробует **посчитать цену за квадратный метр** — и получит **отрицательное число**. Или **отсортирует** квартиры по площади — и отрицательные окажутся **в начале**. Или **сравнит** с другой квартирой — и результат будет **бессмысленным**. Ошибка проявится **не там**, где была допущена.

### Что мы уже знаем

В главе 2 мы разобрали **инкапсуляцию** и **два способа** доступа к приватным данным:

1. **Классический** — методы `get_area()`, `set_area(value)`.
2. **Современный** — `@property` и `.setter`.

В этой главе мы **углубим** знания о `@property` — рассмотрим его **полностью**: вычисляемые свойства, делетер, наследование, нюансы.

### Аналогия

Представь **банкомат**. Ты не можешь залезть внутрь и взять деньги — там всё защищено. Но можешь **попросить** банкомат выдать сумму. Банкомат **сам проверит**, есть ли у тебя столько денег, и только потом выдаст.

`@property` работает так же: атрибут **выглядит публичным**, но за кулисами работает **метод с проверкой**.

### Вывод

Публичные атрибуты **не защищают** данные. `@property` позволяет **скрыть проверку** за «обычным» атрибутом.

---

## 14.2. Что такое `@property`

**`@property`** — это **декоратор**, который позволяет методу **притвориться атрибутом**. Снаружи кажется, что `area` — просто поле, но за кулисами вызывается **метод**.

### Как это выглядит

```python
class Apartment:
    def __init__(self, address, area):
        self.address = address
        self._area = area

    @property
    def area(self):
        # Это геттер — вызывается при чтении flat.area.
        return self._area
```

Разберём этот код **построчно**:

- **`self._area = area`** — сохраняем значение в **защищённое** поле `_area` (с одним подчёркиванием). Оно **не публичное**, но и **не приватное** — просто «внутреннее».
- **`@property`** — декоратор, который превращает метод `area` в **свойство**.
- **`def area(self):`** — имя метода становится **именем свойства**. Обрати внимание: **нет** `get_` — просто `area`.
- **`return self._area`** — возвращаем значение защищённого поля.

### Использование

```python
flat = Apartment("ул. Ленина, д. 10, кв. 5", 65)
print(flat.area)         # 65 — работает как обычный атрибут, БЕЗ скобок!
```

Обрати внимание: **без скобок**. Не `flat.area()`, а `flat.area`. Это **ключевая** особенность `@property`. Если написать со скобками — Python выдаст ошибку, потому что `flat.area` возвращает **число**, а не функцию.

```python
print(flat.area())       # TypeError: 'int' object is not callable
```

### Что происходит

Когда Python видит `flat.area`:

1. Ищет `area` в **самом объекте** `flat` — не находит.
2. Ищет `area` в **классе** `Apartment` — **находит** `property`-объект.
3. Понимает, что это **свойство**, и вызывает его **геттер**.
4. Возвращает результат.

Снаружи это выглядит как **обычное обращение к атрибуту**. Снаружи не видно, что за `flat.area` скрывается **метод**.

### Аналогия

Представь, что у тебя есть **умный холодильник**. Ты просто **открываешь дверцу** — а внутри уже работает **сложная логика** (подсветка, датчики, вентиляция). Снаружи всё выглядит **просто**, но за кулисами — **система**.

`@property` — то же самое. Снаружи — **атрибут**, внутри — **метод**.

### Вывод

`@property` делает метод **«похожим» на атрибут**. Обращение **без скобок**. За кулисами работает **метод**.

---

## 14.3. Геттер — чтение значения

**Геттер** — это метод для **чтения** значения. Создаётся через `@property`.

### Синтаксис

```python
class Apartment:
    def __init__(self, area):
        self._area = area

    @property
    def area(self):
        # Геттер: вызывается при чтении flat.area.
        return self._area
```

### Использование

```python
flat = Apartment(65)
print(flat.area)         # 65
```

### Разбор по частям

- **`@property`** — декоратор, который превращает метод в **свойство**.
- **`def area(self):`** — метод **без** `set_` или `get_`. Имя — просто `area`.
- **`return self._area`** — возвращает значение.

### Что важно

- Имя метода (`area`) становится **именем свойства**.
- Обращение — **без скобок**: `flat.area`.
- Метод **должен** возвращать значение.

### Что можно делать в геттере

Геттер — **полноценный метод**. В нём можно:

- **вычислять** значение на основе других полей;
- **форматировать** данные;
- **логировать** обращения;
- **кешировать** результат (см. 14.7).

Пример **вычисления**:

```python
@property
def area_in_sqft(self):
    # Вычисляем площадь в квадратных футах.
    return self._area * 10.764
```

Здесь `area_in_sqft` — **вычисляемое** свойство. Значение **не хранится** в объекте — оно **считается** каждый раз заново. Об этом подробно в разделе 14.7.

### Что если геттер вернёт не строку

Геттер **может** возвращать **любой тип** — число, строку, список, другой объект. Никаких ограничений нет.

```python
@property
def rooms(self):
    return ["кухня", "спальня", "ванная"]      # список — ок
```

### Вывод

Геттер — метод для **чтения**. Создаётся через `@property`. Обращение — **без скобок**. Может возвращать **любой тип**.

---

## 14.4. Сеттер — запись значения

**Сеттер** — это метод для **записи** значения. Создаётся через `@имя.setter`.

### Синтаксис

```python
class Apartment:
    def __init__(self, area):
        self._area = area

    @property
    def area(self):
        return self._area

    @area.setter
    def area(self, value):
        # Сеттер: вызывается при flat.area = value.
        if not isinstance(value, (int, float)):
            raise TypeError("Площадь должна быть числом")
        if value <= 0:
            raise ValueError("Площадь должна быть больше нуля")
        self._area = value
```

### Использование

```python
flat = Apartment(65)

flat.area = 70           # работает — вызывает сеттер
print(flat.area)         # 70

flat.area = -10          # ValueError: Площадь должна быть больше нуля
flat.area = "abc"        # TypeError: Площадь должна быть числом
```

### Разбор по частям

- **`@area.setter`** — декоратор, который привязывает сеттер к **свойству** `area`. Обрати внимание: **имя** метода (`area`) **совпадает** с именем геттера. Иначе свойство не свяжется.
- **`def area(self, value):`** — имя метода **совпадает** с именем свойства. **Два** параметра: `self` и `value`.
- **`if not isinstance(value, (int, float)):`** — первая проверка: **тип**. Если не число — `TypeError`.
- **`if value <= 0:`** — вторая проверка: **значение**. Если не больше нуля — `ValueError`.
- **`self._area = value`** — записываем значение **после** проверок.

### Что важно

- **Имя** метода должно **совпадать** с именем геттера (`area`).
- **Первый аргумент** — `self`, **второй** — `value`.
- Обращение — **без скобок**: `flat.area = 70`.
- **Проверки** — до записи. Если проверка не прошла — **значение не меняется**.

### Что можно делать в сеттере

Сеттер — **идеальное место** для проверок:

- **тип** значения (`isinstance`);
- **диапазон** (`0 < value < 1000`);
- **зависимости** от других полей;
- **логирование** изменений;
- **каскадное обновление** других полей.

Пример **каскадного обновления**:

```python
@area.setter
def area(self, value):
    if value <= 0:
        raise ValueError("Площадь должна быть больше нуля")
    self._area = value
    # При изменении площади пересчитываем цену.
    self._price = self._base_price_per_sqm * value
```

### Что если сеттер не определён

Если сеттер **не определён**, попытка `flat.area = 70` вызовет:

```
AttributeError: can't set attribute
```

Python **сам** запретит запись. Свойство становится **только для чтения** (см. 14.8).

### Аналогия

Представь, что ты **заполняешь анкету**. Прежде чем данные попадут в базу, система **проверит**:

- правильный ли формат;
- нет ли опечаток;
- все ли обязательные поля.

Если что-то не так — **отказ**. Сеттер делает то же самое: **проверяет** данные перед записью.

### Вывод

Сеттер — метод для **записи**. Создаётся через `@имя.setter`. Имя **совпадает** с геттером. Обращение — **без скобок**. Проверки — до записи.

---

## 14.5. Делетер — удаление значения

**Делетер** — это метод для **удаления** значения. Создаётся через `@имя.deleter`.

### Синтаксис

```python
class Apartment:
    def __init__(self, area):
        self._area = area

    @property
    def area(self):
        return self._area

    @area.setter
    def area(self, value):
        self._area = value

    @area.deleter
    def area(self):
        # Делетер: вызывается при del flat.area.
        print("Площадь удалена")
        del self._area
```

### Использование

```python
flat = Apartment(65)
print(flat.area)         # 65

del flat.area            # Площадь удалена
print(flat.area)         # AttributeError: 'Apartment' object has no attribute '_area'
```

### Разбор по частям

- **`@area.deleter`** — декоратор, который привязывает делетер к **свойству** `area`.
- **`def area(self):`** — имя **совпадает** с геттером. Параметров, кроме `self`, **нет**.
- **`del self._area`** — удаляет значение защищённого поля.

### Когда использовать

Делетер нужен **редко**. Типичные случаи:

- **Логирование** удаления — записать в лог, что кто-то удалил значение.
- **Очистка связанных данных** — например, при удалении площади обнулить цену.
- **Запрет** удаления — например, через `raise`, если поле **нельзя** удалять.

### Запрет удаления

Если не хотите, чтобы поле **можно было удалить**, определите делетер, который **выбрасывает ошибку**:

```python
@area.deleter
def area(self):
    raise AttributeError("Площадь нельзя удалить")
```

Теперь `del flat.area` вызовет **ошибку**:

```
AttributeError: Площадь нельзя удалить
```

Это **часто** используется для **защиты** важных полей от случайного удаления.

### Что если делетер не определён

Если делетер **не определён**, попытка `del flat.area` вызовет:

```
AttributeError: can't delete attribute
```

Python **сам** запретит удаление. Так же, как и с сеттером — свойство становится **защищённым** от удаления.

### Аналогия

Представь **документ** в сейфе. Ты можешь его **достать** (геттер), **положить** (сеттер), а вот **уничтожить** — только если у тебя есть **специальное разрешение**. Делетер — это **проверка** при попытке уничтожить документ.

### Вывод

Делетер — метод для **удаления**. Создаётся через `@имя.deleter`. Используется **редко**. Если не определён — удаление **запрещено** автоматически.

---

## 14.6. Три метода — одно имя

**Ключевая идея** `@property`: **три разных метода** (геттер, сеттер, делетер) используют **одно имя** (`area`).

### Как Python различает их

Каждый декоратор **привязывает** метод к **своему слоту** свойства:

```
        @property              →  геттер
        @area.setter           →  сеттер
        @area.deleter          →  делетер
```

Все три — методы с именем `area`. Но Python знает, **какой** вызвать:

- **Чтение** `flat.area` → **геттер**.
- **Запись** `flat.area = 70` → **сеттер**.
- **Удаление** `del flat.area` → **делетер**.

### Схема

```
        flat.area          →  геттер (@property)
        flat.area = 70     →  сеттер (@area.setter)
        del flat.area      →  делетер (@area.deleter)
```

Снаружи все три действия выглядят **естественно** — как работа с **обычным** атрибутом. Но за кулисами Python **сам** выбирает нужный метод.

### Аналогия

Представь **дверь**. У неё **три режима**:

- **Открыть** (геттер) — читаешь, что внутри.
- **Положить** (сеттер) — кладёшь что-то.
- **Выбросить** (делетер) — удаляешь.

Дверь **одна**, режимы **разные**. `@property` — то же самое: **одно имя**, **три действия**.

### Важно

- **Геттер обязателен**. Без него не будет свойства. Если определить только `@area.setter` — будет ошибка.
- **Сеттер и делетер** — опциональны.
- **Имя** у всех трёх **одинаковое**. Если сеттер назвать `set_area` — свойство не свяжется.
- **Порядок** важен: сначала геттер (`@property`), потом сеттер (`@area.setter`), потом делетер (`@area.deleter`).

### Порядок определения

```python
@property
def area(self):        # 1. Геттер — первым
    return self._area

@area.setter
def area(self, value): # 2. Сеттер — вторым
    self._area = value

@area.deleter
def area(self):        # 3. Делетер — третьим
    del self._area
```

Если поменять порядок — Python выдаст **ошибку**, потому что `@area.setter` не найдёт `area` (ещё не определён геттер).

### Вывод

`@property` использует **одно имя** для **трёх методов**. Python **сам** выбирает нужный по действию. **Геттер обязателен**, остальные — опциональны.

---

## 14.7. Вычисляемые свойства

**Вычисляемое свойство** — это свойство, значение которого **не хранится**, а **вычисляется** на основе других полей.

### Пример

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self._area = area
        self._price = price

    @property
    def area(self):
        return self._area

    @property
    def price(self):
        return self._price

    @property
    def price_per_sqm(self):
        # Вычисляем цену за квадратный метр.
        return self._price / self._area
```

### Использование

```python
flat = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)
print(flat.price_per_sqm)    # 130769.23...
```

### Что происходит

- Значение **не хранится** в объекте.
- Каждый раз **вычисляется** заново.
- Всегда **актуально** — если изменится `_price` или `_area`, значение пересчитается.

### Преимущества

- **Нет дублирования данных.** Не нужно хранить `price_per_sqm` — оно всегда **производное**.
- **Автоматическая актуальность.** Изменились данные — изменилось и производное.
- **Не нужен сеттер.** Свойство **только для чтения** по своей природе.

### Что если попробовать изменить

```python
flat.price_per_sqm = 500
```

Python выдаст **ошибку**:

```
AttributeError: property 'price_per_sqm' of 'Apartment' object has no setter
```

Это **правильно**: `price_per_sqm` — **производное**. Чтобы его изменить, нужно изменить `price` или `area`.

### Аналогия

Представь **калькулятор**. Ты **не хранишь** результат — ты **считаешь** его при каждом нажатии. Если изменились числа — результат **новый**. Калькулятор не «помнит» старое значение — он **всегда** считает заново.

### Нюанс: кеширование

Если вычисление **дорогое** (сложное, медленное), его можно **кешировать** — сохранить результат и пересчитывать только при изменении базовых полей. Об этом — в главе 22 (дескрипторы).

### Вывод

Вычисляемое свойство — значение **не хранится**, а **вычисляется**. Это уменьшает дублирование и гарантирует **актуальность**. Сеттер **не нужен** — свойство **только для чтения**.

---

## 14.8. Свойство только для чтения

Если определить **только геттер** (без сеттера) — свойство становится **только для чтения**.

### Пример

```python
class Apartment:
    def __init__(self, area):
        self._area = area

    @property
    def area(self):
        return self._area

    # Сеттер не определён!
```

### Использование

```python
flat = Apartment(65)
print(flat.area)         # 65 — работает

flat.area = 70           # AttributeError: can't set attribute
```

Попытка **присвоить** значение вызывает `AttributeError`. Это и есть **защита «только для чтения»**.

### Когда использовать

- **Вычисляемые свойства** — они не должны быть изменяемыми.
- **Идентификаторы** — например, `id` или `created_at`.
- **Производные значения** — например, `price_per_sqm`.
- **Константы** — например, `max_area`.

### Аналогия

Представь **номер паспорта**. Ты можешь **прочитать** его, но **не можешь изменить**. Он **присвоен один раз** и **не меняется**.

### Что если очень нужно изменить

Если **всё-таки** нужно изменить — можно использовать **приватное поле** через name mangling:

```python
flat._Apartment__area = 70       # работает, но это взлом
```

Но это **плохая практика**. Если свойство **только для чтения** — значит, оно **должно** быть только для чтения.

### Вывод

Без сеттера свойство становится **только для чтения**. Это **защищает** данные от случайного изменения. Попытка присвоить — `AttributeError`.

---

## 14.9. Свойства и наследование

Свойства **наследуются** как обычные методы. Наследник **может**:

1. **Использовать** родительское свойство.
2. **Переопределить** геттер.
3. **Добавить** сеттер к родительскому геттеру.
4. **Полностью переопределить** свойство.

Разберём каждый случай.

### 14.9.1. Использование родительского свойства

Самый простой случай: наследник **ничего не делает**, свойство **работает** как есть.

```python
class Apartment:
    def __init__(self, area):
        self._area = area

    @property
    def area(self):
        return self._area


class Studio(Apartment):
    pass


studio = Studio(25)
print(studio.area)       # 25 — унаследовано
```

Здесь `Studio` **не переопределяет** `area` — используется родительская версия.

### 14.9.2. Переопределение геттера

Наследник может **дать свою версию** геттера:

```python
class Studio(Apartment):
    @property
    def area(self):
        # Добавляем к площади балкон.
        return super().area + 3
```

Разберём:

- **`@property`** — снова объявляем свойство.
- **`super().area`** — вызываем **родительский геттер**. Он вернёт `self._area`.
- **`+ 3`** — добавляем **условные** 3 м² (например, балкон).

Теперь `Studio(25).area` вернёт `28`.

**Важно:** `super().area` **не работает** без `@property` — это вызов **геттера** родителя через `super()`.

### 14.9.3. Добавление сеттера в наследнике

Если родитель определил **только геттер**, наследник может **добавить сеттер**:

```python
class Apartment:
    def __init__(self, area):
        self._area = area

    @property
    def area(self):
        return self._area


class MutableApartment(Apartment):
    @Apartment.area.setter
    def area(self, value):
        if value <= 0:
            raise ValueError("Площадь должна быть больше нуля")
        self._area = value
```

Разберём:

- **`@Apartment.area.setter`** — привязываем сеттер к **родительскому** свойству `Apartment.area`. Обрати внимание: **не** `@area.setter`, а **`@Apartment.area.setter`**. Это важно: в наследнике свойство `area` пока **не определено**, оно **унаследовано**, поэтому нужно указать **родительский класс**.
- **`def area(self, value):`** — имя **совпадает** с родительским.

Теперь:

```python
flat = MutableApartment(65)
flat.area = 70            # работает — сеттер добавлен в наследнике
print(flat.area)          # 70
```

**Что важно:** до этого `Apartment.area` был **только для чтения**. Наследник **добавил** возможность записи — но **только для себя**. Родительский `Apartment` **остался** только для чтения.

### 14.9.4. Полное переопределение

Если нужно **полностью** переопределить свойство — определяем **все три** метода заново:

```python
class Studio(Apartment):
    def __init__(self, area, balcony):
        super().__init__(area)
        self._balcony = balcony

    @property
    def area(self):
        return self._area + self._balcony

    @area.setter
    def area(self, value):
        self._area = value
```

Здесь `Studio` **полностью** определяет своё свойство `area`. Оно **не связано** с родительским — используется `self._area` и `self._balcony`.

### Что если не использовать `super()`

Если в переопределённом геттере **не вызывать** `super()`, родительская логика **потеряется**. Иногда это нужно, иногда — ошибка.

**Потеря (правильно, если хотим свою логику):**

```python
@property
def area(self):
    return 100            # игнорируем родителя
```

**Потеря (неправильно, если хотели дополнить):**

```python
@property
def area(self):
    return self._area     # забыли super(), но логика та же
```

**Правило:** если хотите **дополнить** родителя — используйте `super()`. Если хотите **заменить** — не используйте.

### Вывод

Свойства **наследуются**. Наследник может **использовать**, **переопределить** или **дополнить** их. Для добавления сеттера в наследнике используется `@Родитель.свойство.setter`. Для вызова родительского геттера — `super().свойство`.

---

## 14.10. Свойства и приватные поля

`@property` **часто используется** вместе с **приватными** полями (`__name`).

### Пример

```python
class Apartment:
    def __init__(self, address, area):
        self.address = address
        self.__area = area       # приватное поле

    @property
    def area(self):
        return self.__area

    @area.setter
    def area(self, value):
        if value <= 0:
            raise ValueError("Площадь должна быть больше нуля")
        self.__area = value
```

### Что это даёт

- Поле `__area` **скрыто** от снаружи — прямой доступ невозможен.
- Через `flat.area` можно **читать** и **писать** — с проверкой.
- Снаружи кажется, что это **обычный атрибут**.

### Схема

```
        flat.__area           →  AttributeError (приватное)
        flat._Apartment__area →  работает через name mangling (не рекомендуется)
        flat.area             →  геттер (читает __area)
        flat.area = 70        →  сеттер (проверяет и пишет в __area)
```

### Почему `__area`, а не `_area`

**`_area`** — защищённое. Доступно снаружи, но **не рекомендуется**.

**`__area`** — приватное. Python **переименовывает** его в `_Apartment__area`. Снаружи **напрямую** не доступно.

**Когда использовать какое:**

- **`_area`** — если поле нужно наследникам.
- **`__area`** — если поле **только** для самого класса.

### Пример с наследованием

```python
class Apartment:
    def __init__(self, area):
        self.__area = area       # приватное

    @property
    def area(self):
        return self.__area


class Studio(Apartment):
    @property
    def area(self):
        # self.__area → _Studio__area — НЕ СУЩЕСТВУЕТ!
        return self.__area + 3    # AttributeError
```

Здесь **ошибка**: в `Studio` `self.__area` превращается в `_Studio__area`, а **не** в `_Apartment__area`. Поле родителя **не доступно** напрямую.

**Решение:** использовать `super().area`:

```python
class Studio(Apartment):
    @property
    def area(self):
        return super().area + 3    # работает
```

### Аналогия

Представь **сейф**. Внутри — **деньги** (приватное поле). Снаружи — **кодовый замок** (`@property`). Ты не можешь **залезть внутрь**, но можешь **попросить** выдать или положить — с проверкой.

### Вывод

`@property` **идеально сочетается** с приватными полями. Поле — **скрыто**, доступ — **через свойство с проверкой**. В наследниках для доступа к родительскому свойству используйте `super()`.

---

## 14.11. Свойства vs классические геттеры и сеттеры

Мы уже сравнивали их в главе 2. Напомним **ключевое**.

| Критерий | Классический (`get_`/`set_`) | `@property` |
|----------|------------------------------|-------------|
| Чтение | `flat.get_area()` | `flat.area` |
| Запись | `flat.set_area(70)` | `flat.area = 70` |
| Похоже на атрибут | ❌ Нет | ✅ Да |
| Явность | ✅ Высокая | ⚠️ Средняя |
| Современность | Старый стиль | Современный |

### Что важно

- **`@property`** — **предпочтительный** способ в современном Python.
- **Классические** методы — в старом коде и языках вроде Java.

### Когда что использовать

- **Новый код** → `@property`.
- **Старый проект** → поддерживать существующий стиль.
- **Обучение** → показать **оба**.

### Что лучше на практике

В **реальном** коде `@property` **выигрывает**:

- **Код короче** — не нужно писать `get_` и `set_`.
- **Код естественнее** — снаружи не видно, что это метод.
- **Легче рефакторить** — если завтра понадобится проверка, **внешний код не изменится**.
- **IDE подсказывает** — `flat.area` автодополняется как атрибут.

### Вывод

`@property` **короче** и **естественнее**. Классические методы — **явнее**, но **громоздче**. В современном Python используйте **`@property`**.

---

## 14.12. Нюансы и тонкости

### 14.12.1. В конструкторе используйте `self.area = value`

Если определено свойство — в конструкторе **лучше** писать `self.area = value`, а не `self._area = value`:

```python
def __init__(self, area):
    self.area = area       # ← сработает сеттер, проверит
    # self._area = area    # ← проверки не будет
```

Это гарантирует, что **даже в конструкторе** значение проходит проверку. Если передать плохое значение — будет **ошибка** сразу, при создании объекта.

### 14.12.2. Не вызывайте свойство **внутри себя**

**Правильно:**

```python
@area.setter
def area(self, value):
    self._area = value      # ← обращаемся к _area, а не к area
```

**Неправильно:**

```python
@area.setter
def area(self, value):
    self.area = value       # ← бесконечная рекурсия!
```

Если в сеттере написать `self.area = value` — Python **снова** вызовет сеттер. Получится **бесконечная рекурсия**. Python в итоге выдаст `RecursionError`.

**Правило:** внутри сеттера обращайтесь к **приватному** полю (`_area` или `__area`), **не** к самому свойству.

### 14.12.3. Имя свойства и имя поля **должны отличаться**

```python
@property
def area(self):
    return self._area       # ← поле с другим именем
```

Если назвать поле так же, как свойство (`self.area = ...`), будет **рекурсия**.

**Правило:** имя поля должно **отличаться** от имени свойства. Обычно добавляют **подчёркивание**: `_area` или `__area`.

### 14.12.4. `@property` — не для всего

**Не** превращайте **каждое** поле в свойство. Это **избыточно**:

```python
# Плохо — избыточно.
class Apartment:
    def __init__(self, address):
        self._address = address

    @property
    def address(self):
        return self._address
```

Если **нет проверок** — используйте **обычное публичное поле**:

```python
class Apartment:
    def __init__(self, address):
        self.address = address       # публичное — просто и понятно
```

**Правило:** `@property` — **только когда есть логика** (проверка, вычисление).

### 14.12.5. Свойства и `__dict__`

Свойства **не попадают** в `__dict__` объекта. Они живут в `__dict__` **класса**:

```python
flat = Apartment(65)
print(flat.__dict__)          # {'_area': 65} — без area
print(Apartment.__dict__)     # {'area': <property object>, ...}
```

Это **важно** понимать: `flat.area` — это **не** атрибут объекта, а **свойство класса**. Поэтому его **нельзя** удалить через `del flat.__dict__["area"]`.

### 14.12.6. Свойства работают на **классе**, а не на объекте

Обращение к `Apartment.area` возвращает **объект property**, а не значение:

```python
print(Apartment.area)         # <property object at 0x...>
print(flat.area)              # 65
```

Это **важно** для декоратора `@Apartment.area.setter` в наследниках. Когда вы пишете `@Apartment.area.setter`, вы **берёте** объект property из класса `Apartment` и добавляете к нему сеттер.

### 14.12.7. `hasattr` и свойства

`hasattr(flat, "area")` вызывает **геттер**. Если геттер **выбросит исключение**, `hasattr` вернёт `False`:

```python
@property
def area(self):
    raise ValueError("Ошибка")

hasattr(flat, "area")         # False — потому что геттер упал
```

Это может **сбивать с толку**. Будьте **осторожны** с геттерами, которые могут **упасть**.

### 14.12.8. `@property` и логирование

Свойства **удобно** использовать для **логирования**:

```python
@property
def area(self):
    print(f"Читаем площадь: {self._area}")
    return self._area
```

Это **полезно** при отладке, но **осторожно** в продакшене — каждый доступ будет писать в лог. При **частом** обращении лог **засорится**.

### 14.12.9. Свойства и `__slots__`

Если в классе используется `__slots__` (см. главу 21), свойства **работают** — но нужно **добавить** поле в `__slots__`:

```python
class Apartment:
    __slots__ = ("_area",)         # ← _area обязательно

    def __init__(self, area):
        self._area = area

    @property
    def area(self):
        return self._area
```

### 14.12.10. Абстрактные свойства

Свойства могут быть **абстрактными** (см. главу 7):

```python
from abc import ABC, abstractmethod

class HomeItem(ABC):
    @property
    @abstractmethod
    def area(self):
        pass
```

Наследники **обязаны** реализовать свойство.

### Вывод

`@property` **мощный**, но требует **аккуратности**. Не делайте всё свойствами. Обращайтесь к **приватным полям** внутри сеттеров. Помните про **рекурсию**. Учитывайте нюансы с `hasattr`, `__slots__` и наследованием.

---

## 14.13. Типичные ошибки

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| `self.area = value` в сеттере | Бесконечная рекурсия | `self._area = value` |
| Имя поля = имя свойства | Рекурсия | Поле с другим именем (`_area`) |
| `self._area = value` в конструкторе | Проверка пропускается | `self.area = value` |
| Делают всё свойствами | Избыточно | Только где есть логика |
| Забыли `@area.setter` | Только чтение | Добавить декоратор |
| Имя сеттера **не** совпадает с геттером | Свойство не свяжется | Имя должно совпадать |
| Обращение со скобками | `flat.area()` — ошибка | `flat.area` — без скобок |
| Делают `@property` для констант | Не нужно | Обычный атрибут класса |
| Геттер выбрасывает исключение | `hasattr` вернёт `False` | Осторожно |
| Забыли про `@Apartment.area.setter` в наследнике | Сеттер не работает | Использовать родительское свойство |
| Не используют `super()` при переопределении | Теряется логика родителя | Использовать `super().свойство` |
| Пытаются изменить свойство только для чтения | `AttributeError` | Не пытаться или добавить сеттер |

---

## 14.14. Что важно запомнить

- **`@property`** — декоратор, делающий метод **похожим на атрибут**.
- **Геттер** — чтение. Создаётся через `@property`.
- **Сеттер** — запись. Создаётся через `@имя.setter`.
- **Делетер** — удаление. Создаётся через `@имя.deleter`.
- **Имя** у всех трёх — **одинаковое**.
- **Обращение** — **без скобок**: `flat.area`, `flat.area = 70`.
- **Вычисляемое свойство** — значение **не хранится**, а **вычисляется**.
- **Без сеттера** — свойство **только для чтения**.
- **Свойства наследуются**. Сеттер добавляется через `@Родитель.свойство.setter`.
- **`@property` + приватное поле** — **защита с проверкой**.
- **В сеттере** — обращайтесь к **приватному полю**, не к свойству.
- **Не делайте всё свойствами** — только где есть **логика**.
- **`@property` предпочтительнее** классических `get_`/`set_`.
- **В конструкторе** — пишите `self.свойство = value`, а не `self._поле = value`.

## Задача 14 (свойство `@property`)

В агентстве важно, чтобы **площадь** квартиры была **корректной** — нельзя допустить отрицательное значение, ноль или строку.

**Часть 1. Класс `Apartment`.** Создайте класс `Apartment` с полями:

- `address` — публичное поле (без свойства);
- `_area` — защищённое поле для площади;
- `_price` — защищённое поле для цены.

Реализуйте конструктор, где площадь и цена устанавливаются через **свойства** (то есть `self.area = area`, `self.price = price`).

**Часть 2. Свойство `area`.** Реализуйте свойство `area`:

- **геттер** — возвращает `self._area`;
- **сеттер** — проверяет: значение должно быть **числом** (`int` или `float`) и **строго больше нуля**. При нарушении — `TypeError` или `ValueError` с понятным сообщением.

**Часть 3. Свойство `price`.** Реализуйте свойство `price` аналогично:

- **геттер** — возвращает `self._price`;
- **сеттер** — проверяет тип (`int` или `float`) и что значение **не отрицательное**.

**Часть 4. Вычисляемое свойство `price_per_sqm`.** Реализуйте **только геттер** (без сеттера) — свойство возвращает цену за квадратный метр:

```
price_per_sqm = price / area
```

Продемонстрируйте, что:

- свойство **только для чтения** — попытка присвоить значение вызывает `AttributeError`;
- при изменении `area` или `price` значение `price_per_sqm` **автоматически** меняется.

**Часть 5. Свойство `info`.** Создайте **вычисляемое свойство** `info`, которое возвращает строку вида:

```
Квартира "ул. Ленина, д. 10, кв. 5" — 65 м², 8 500 000 ₽ (130 769 ₽/м²)
```

**Часть 6. Делетер.** Реализуйте **делетер** для свойства `area`, который **запрещает** удаление и выбрасывает `AttributeError` с сообщением «Площадь нельзя удалить». Продемонстрируйте, что `del flat.area` вызывает ошибку.

**Часть 7. Наследование.** Создайте наследника `Studio(Apartment)`:

- переопределите **геттер** `area` — пусть возвращает `super().area + 3` (условно «балкон»);
- добавьте **сеттер** для `area` через `@Apartment.area.setter`;
- проверьте, что `Studio(25).area` возвращает `28`.

**Часть 8. Проверка.** Продемонстрируйте:

- создание квартиры с корректными данными;
- `flat.area = -10` → `ValueError`;
- `flat.area = 0` → `ValueError`;
- `flat.area = "abc"` → `TypeError`;
- `flat.area = 70` → успех;
- `flat.price = -100` → `ValueError`;
- `flat.price_per_sqm = 500` → `AttributeError`;
- `del flat.area` → `AttributeError`;
- работу `info`.

**Часть 9.** В комментариях объясните:

1. Чем `@property` удобнее классических геттеров и сеттеров.
2. Что такое **вычисляемый атрибут** и зачем он нужен.
3. Почему в сеттере обращаются к `self._area`, а не к `self.area`.
4. Почему в конструкторе пишут `self.area = area`, а не `self._area = area`.
5. Что произойдёт, если имя поля **совпадёт** с именем свойства.
6. Почему вычисляемое свойство **не имеет** сеттера.
7. Почему в наследнике для добавления сеттера используется `@Apartment.area.setter`, а не `@area.setter`.

### Решение

```python
# ─── Части 1–6. Класс Apartment ───
class Apartment:
    def __init__(self, address, area, price):
        self.address = address       # публичное поле — простое
        # Используем свойства — проверки сработают даже здесь.
        self.area = area
        self.price = price

    # ─── Часть 2. Свойство area ───
    @property
    def area(self):
        # Геттер: читаем значение.
        return self._area

    @area.setter
    def area(self, value):
        # Сеттер: проверяем и записываем.
        if not isinstance(value, (int, float)):
            raise TypeError("Площадь должна быть числом")
        if value <= 0:
            raise ValueError("Площадь должна быть больше нуля")
        self._area = value

    @area.deleter
    def area(self):
        # Часть 6. Делетер: запрещаем удаление.
        raise AttributeError("Площадь нельзя удалить")

    # ─── Часть 3. Свойство price ───
    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, value):
        if not isinstance(value, (int, float)):
            raise TypeError("Цена должна быть числом")
        if value < 0:
            raise ValueError("Цена не может быть отрицательной")
        self._price = value

    # ─── Часть 4. Вычисляемое свойство price_per_sqm ───
    @property
    def price_per_sqm(self):
        # Вычисляем цену за квадратный метр.
        # Сеттера НЕТ — свойство только для чтения.
        return self._price / self._area

    # ─── Часть 5. Вычисляемое свойство info ───
    @property
    def info(self):
        return (
            f'Квартира "{self.address}" — '
            f"{self._area} м², {self._price:,} ₽ "
            f"({self.price_per_sqm:,.0f} ₽/м²)"
        ).replace(",", " ")


# ─── Часть 7. Наследник Studio ───
class Studio(Apartment):
    @property
    def area(self):
        # Переопределяем геттер: добавляем «балкон» +3 м².
        return super().area + 3

    # Добавляем сеттер через родительское свойство.
    @Apartment.area.setter
    def area(self, value):
        # Используем тот же сеттер, что и в Apartment.
        Apartment.area.fset(self, value)


# ─── Часть 8. Проверка ───
print("=== Создание квартиры ===")
flat = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)
print(flat.info)

print()
print("=== Проверки area ===")
for value in [-10, 0, "abc", 70]:
    try:
        flat.area = value
        print(f"  area = {value!r} → OK, area = {flat.area}")
    except (ValueError, TypeError) as e:
        print(f"  area = {value!r} → ошибка: {e}")

print()
print("=== Проверки price ===")
for value in [-100, "дорого", 12_000_000]:
    try:
        flat.price = value
        print(f"  price = {value!r} → OK, price = {flat.price}")
    except (ValueError, TypeError) as e:
        print(f"  price = {value!r} → ошибка: {e}")

print()
print("=== Вычисляемое свойство price_per_sqm ===")
print(f"price_per_sqm = {flat.price_per_sqm:,.0f} ₽/м²".replace(",", " "))

print()
print("=== Свойство только для чтения ===")
try:
    flat.price_per_sqm = 500
except AttributeError as e:
    print(f"Ошибка: {e}")

print()
print("=== Делетер ===")
try:
    del flat.area
except AttributeError as e:
    print(f"Ошибка: {e}")

print()
print("=== Автоматическое обновление price_per_sqm ===")
print(f"До:   area={flat.area}, price={flat.price:,} ₽, price_per_sqm={flat.price_per_sqm:,.0f} ₽/м²".replace(",", " "))
flat.area = 100
print(f"После: area={flat.area}, price={flat.price:,} ₽, price_per_sqm={flat.price_per_sqm:,.0f} ₽/м²".replace(",", " "))

print()
print("=== Наследник Studio ===")
studio = Studio("ул. Пушкина, д. 1, кв. 7", 25, 4_000_000)
print(f"studio.area = {studio.area} (25 + 3 = 28)")
```

**Ожидаемый вывод:**

```
=== Создание квартиры ===
Квартира "ул. Ленина, д. 10, кв. 5" — 65 м², 8 500 000 ₽ (130 769 ₽/м²)

=== Проверки area ===
  area = -10 → ошибка: Площадь должна быть больше нуля
  area = 0 → ошибка: Площадь должна быть больше нуля
  area = 'abc' → ошибка: Площадь должна быть числом
  area = 70 → OK, area = 70

=== Проверки price ===
  price = -100 → ошибка: Цена не может быть отрицательной
  price = 'дорого' → ошибка: Цена должна быть числом
  price = 12000000 → OK, price = 12000000

=== Вычисляемое свойство price_per_sqm ===
price_per_sqm = 171 429 ₽/м²

=== Свойство только для чтения ===
Ошибка: property 'price_per_sqm' of 'Apartment' object has no setter

=== Делетер ===
Ошибка: Площадь нельзя удалить

=== Автоматическое обновление price_per_sqm ===
До:   area=70, price=12 000 000 ₽, price_per_sqm=171 429 ₽/м²
После: area=100, price=12 000 000 ₽, price_per_sqm=120 000 ₽/м²

=== Наследник Studio ===
studio.area = 28 (25 + 3 = 28)
```

### Разбор

**1. Чем `@property` удобнее классических геттеров и сеттеров.**

**Классический способ:**

```python
def get_area(self):
    return self._area

def set_area(self, value):
    ...
```

Использование:

```python
flat.set_area(70)
print(flat.get_area())
```

**С `@property`:**

```python
flat.area = 70
print(flat.area)
```

**Преимущества `@property`:**

- **Короче:** нет `get_` и `set_`.
- **Естественнее:** как обычный атрибут.
- **Прозрачнее:** снаружи не видно, что это метод.
- **Один интерфейс:** если завтра добавится проверка — **внешний код не изменится**.

**2. Что такое вычисляемый атрибут и зачем он нужен.**

**Вычисляемый атрибут** — значение которого **не хранится** в объекте, а **вычисляется** при каждом обращении.

**Зачем нужен:**

- **Не нужно хранить** — экономия памяти.
- **Всегда актуален** — при изменении базовых полей пересчитывается сам.
- **Нельзя рассинхронизировать** — невозможно «забыть обновить».
- **Нет сеттера** — защита от изменения.

**Пример:** `price_per_sqm` — всегда вычисляется из `price` и `area`.

**3. Почему в сеттере обращаются к `self._area`, а не к `self.area`.**

Если в сеттере написать `self.area = value` — это **рекурсивный вызов** сеттера:

1. `flat.area = 70` вызывает сеттер.
2. Внутри сеттера `self.area = 70` снова вызывает сеттер.
3. И снова. И снова. → **Бесконечная рекурсия**.

**Правильно:** в сеттере обращаться к **приватному полю** `self._area` — оно **не** вызывает свойство.

**4. Почему в конструкторе пишут `self.area = area`, а не `self._area = area`.**

Если писать `self._area = area` — значение записывается **напрямую**, **минуя** сеттер. Проверки **не сработают**. Можно случайно создать квартиру с отрицательной площадью.

Если писать `self.area = area` — вызывается **сеттер**, который **проверит** значение. Даже в конструкторе **нельзя** передать плохие данные.

**Правило:** в конструкторе используйте `self.свойство = значение` для полей со свойствами.

**5. Что произойдёт, если имя поля совпадёт с именем свойства.**

```python
@property
def area(self):
    return self.area        # ← рекурсия! Обращение к самому себе.
```

Метод `area` вернёт `self.area` — то есть **вызовет сам себя**. Бесконечная рекурсия → `RecursionError`.

**Правило:** имя поля **должно отличаться** от имени свойства. Обычно добавляют подчёркивание: `_area` или `__area`.

**6. Почему вычисляемое свойство не имеет сеттера.**

Вычисляемое свойство — **производное**. Оно **не хранится** в объекте, а **вычисляется** из других полей.

**Если добавить сеттер** — что он должен делать? Устанавливать `price_per_sqm` напрямую? Но тогда нужно **менять** `price` или `area` — а это **неочевидно** и **противоречиво**.

**Логика:**

- Чтобы изменить `price_per_sqm` — измените `price` или `area`.
- Сеттер **не нужен** — иначе получится **двусмысленность**.

**Поэтому** вычисляемое свойство — **только для чтения**. Это **защищает** от ошибок.

**7. Почему в наследнике для добавления сеттера используется `@Apartment.area.setter`, а не `@area.setter`.**

В наследнике `Studio` **свойство `area` ещё не определено**. Оно **унаследовано** от `Apartment`. Поэтому `@area.setter` **не найдёт** свойство `area` в классе `Studio` — Python выдаст ошибку.

**Правильно:** указать **родительский класс**:

```python
@Apartment.area.setter
def area(self, value):
    Apartment.area.fset(self, value)
```

Здесь:

- **`@Apartment.area.setter`** — берём свойство `area` из класса `Apartment` и добавляем сеттер.
- **`Apartment.area.fset(self, value)`** — вызываем **родительский** сеттер напрямую. `fset` — это атрибут property, который хранит **функцию-сеттер**.

Это **стандартный** способ добавить сеттер в наследнике, когда родитель определил **только геттер**.

# Глава 15. Классовый метод `@classmethod`

## 15.1. Проблема: как создать объект из строки

Вернёмся к нашему агентству недвижимости. За несколько глав мы построили класс `Apartment`, научили его красиво выводиться на экран, сравниваться по данным, складываться в комплексы. Класс работает, но постепенно программа растёт, и возникает новая задача — данные о квартирах теперь приходят **из внешнего мира**. Раньше мы создавали объекты вручную, прямо в коде: писали адрес, площадь, цену и получали готовый объект. Но настоящая программа так не работает. Данные приходят из файлов, из базы данных, из API-ответов, из пользовательского ввода. И почти всегда они приходят **в виде строки** или **в виде словаря**, а не в виде готовых объектов Python.

Например, из файла может прийти такая строка:

```
"ул. Ленина, д. 10, кв. 5;65;8500000"
```

Здесь три поля разделены точкой с запятой: адрес, площадь, цена. Нам нужно **превратить** эту строку в объект `Apartment`. Первое, что приходит в голову начинающему программисту, — написать **обычный метод** в классе:

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    def from_string(self, data):
        address, area, price = data.split(";")
        return Apartment(address, int(area), int(price))
```

Логика вроде бы простая: метод `from_string` принимает строку, разбивает её по точке с запятой, преобразует площадь и цену в целые числа и создаёт новый объект `Apartment`. Всё выглядит хорошо, но есть **одна маленькая проблема** — это **обычный метод**, а значит, для его вызова нужен **объект**. То есть чтобы создать квартиру из строки, нам сначала нужно создать какую-то **другую** квартиру, у которой уже есть метод `from_string`. Получается странная ситуация:

```python
flat = Apartment("какой-то адрес", 0, 0)         # создаём «пустой» объект
new_flat = flat.from_string("ул. Ленина;65;8500000")
```

Мы создаём **бессмысленный объект** только для того, чтобы у него вызвать метод и получить **настоящий** объект. Это нарушает здравую логику: зачем нам «пустая» квартира, если мы хотим получить нормальную? Такой код неудобен и нелогичен. Более того, он **сбивает с толку** других программистов, которые читают код и не понимают, зачем здесь `Apartment("какой-то адрес", 0, 0)`.

Можно пойти **другим путём** — сделать метод **статическим** (мы разбирали их в главе 6):

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    @staticmethod
    def from_string(data):
        address, area, price = data.split(";")
        return Apartment(address, int(area), int(price))
```

Теперь метод можно вызвать **без объекта**, напрямую через класс:

```python
flat = Apartment.from_string("ул. Ленина;65;8500000")
```

Это **уже лучше** — не нужно создавать «пустой» объект. Код стал чище и логичнее. Но и здесь есть **скрытая проблема**, которая проявится **позже** — при **наследовании**. Об этом мы поговорим в разделе 15.3, а пока просто запомним: **статический метод работает, но не идеально**. Есть способ **ещё лучше**.

### Почему это важно понимать

Проблема «создать объект из строки» — **не игрушечная**. Она возникает **очень часто** в реальной разработке:

- **Парсинг файлов.** Данные в формате CSV, JSON, XML. Каждая строка — это будущий объект.
- **Базы данных.** Результат SQL-запроса — это набор строк. Каждую нужно превратить в объект.
- **API.** Ответ от сервера — это JSON или текст. Данные нужно превратить в объекты.
- **Пользовательский ввод.** Пользователь вводит данные вручную — в виде строки.

Везде нужен **удобный способ** создать объект **напрямую из данных**. `@classmethod` решает эту задачу **элегантно** — и, что особенно важно, **правильно работает с наследованием**.

### Аналогия

Представь **фабрику мебели**. Обычно ты **собираешь** мебель из деталей: берёшь доски, винты, чертёж — и собираешь. Но иногда мебель приходит **уже в коробке**: «собери стул из этих деталей». Для этого на фабрике есть **специальные линии** — они **принимают** упаковку и **выдают** готовый стул.

`@classmethod` — это **такая линия**. Не нужно создавать «пустой стул», чтобы потом его переделать. Есть **специальный метод**, который создаёт объект **напрямую из данных**.

### Вывод

Обычный метод **требует объект**. Статический метод **не работает с классом** при наследовании. `@classmethod` решает **обе** проблемы: не нужен объект, и работает правильно с наследниками.

---

## 15.2. Что такое `@classmethod`

**`@classmethod`** — это **декоратор**, который превращает метод в **классовый метод**. Такой метод **получает первым аргументом сам класс** (а не объект, как обычный метод). По сути, это **третий вид метода** в Python, наряду с обычными и статическими. Обычный метод привязан к **объекту**, статический — **ни к чему**, а классовый — **к классу**.

### Синтаксис

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    @classmethod
    def from_string(cls, data):
        address, area, price = data.split(";")
        return cls(address, int(area), int(price))
```

### Разбор по частям

Давай разберём этот код **построчно**, чтобы понять каждую деталь.

- **`@classmethod`** — это декоратор, который делает метод **классовым**. Он ставится **перед** определением метода, как `@property` или `@staticmethod`. Именно этот декоратор «сообщает» Python: «Это не обычный метод, не статический, а **классовый**».

- **`def from_string(cls, data):`** — первый параметр — **`cls`** (а не `self`, как у обычного метода). `cls` — это **ссылка на класс**, а не на объект. Это **соглашение**: как `self` для объектов, так `cls` для классов. Python **сам** подставит класс в `cls` при вызове. Название `cls` — не требование языка, а **договорённость программистов**. Можно назвать и `klass`, и `type`, но принято использовать именно `cls` — так делают все.

- **`cls(...)`** — вызываем **конструктор класса**. Если `cls` — это `Apartment`, то `cls(...)` — это `Apartment(...)`. Если `cls` — это `Studio`, то `cls(...)` — это `Studio(...)`. Это **автоматически** подстраивается под класс, через который был вызван метод. Это **ключевая** особенность `@classmethod`.

- **`return cls(...)`** — возвращаем **новый объект**. Обрати внимание: мы **не изменяем** существующий объект (его и нет), а **создаём новый** и возвращаем его.

### Использование

```python
flat = Apartment.from_string("ул. Ленина, д. 10, кв. 5;65;8500000")
print(flat.address)      # ул. Ленина, д. 10, кв. 5
print(flat.area)         # 65
print(flat.price)        # 8500000
```

Обрати внимание на **три важных** момента:

1. Метод вызывается **через класс**: `Apartment.from_string(...)`. Мы **не создаём** объект заранее — как в случае с обычным методом.
2. **Объект не создаётся** вручную — его создаёт **метод** через `cls(...)`.
3. Метод **возвращает новый объект** класса `Apartment` (или его наследника).

### Что происходит под капотом

Когда Python видит `Apartment.from_string("...")`, происходит **последовательность** шагов:

1. Python ищет `from_string` в классе `Apartment`.
2. Находит **классовый метод** — специальный объект `classmethod`, который оборачивает функцию.
3. **Автоматически** подставляет `Apartment` в `cls`. То есть вызов `Apartment.from_string("...")` превращается в `from_string(Apartment, "...")`.
4. Вызывает `from_string(Apartment, "...")` — то есть функцию с двумя аргументами: класс и строку.
5. Внутри `cls(...)` — это `Apartment(...)`. Создаётся объект `Apartment`.
6. Возвращается готовый объект.

Снаружи это выглядит **просто**: «вызвали метод у класса, получили объект». Но за кулисами Python **сам** подставил класс — программисту **не нужно** об этом думать.

### Что если вызвать через объект

`@classmethod` можно вызвать и через **объект**:

```python
flat = Apartment("...", 0, 0)
new_flat = flat.from_string("ул. Садовая;80;12000000")
```

Это **сработает** — но **не рекомендуется**. Потому что:

- **Вводит в заблуждение.** Кажется, что метод работает с `flat`, но на самом деле он работает с **классом** `Apartment`. Объект `flat` тут **не используется**.
- **Неясно, откуда взялся объект.** `flat` — просто «случайно» оказался первым. Логика **сбивает**.
- **Ломает стиль.** `flat.from_string(...)` выглядит так, будто `flat` важен. На самом деле — нет.

**Правило:** классовые методы вызывайте **через класс**, а не через объект.

### Что можно делать в `@classmethod`

Классовый метод — **полноценный метод**. В нём можно:

- **создавать объекты** через `cls(...)`;
- **читать атрибуты класса** через `cls.attribute`;
- **менять атрибуты класса** через `cls.attribute = value`;
- **вызывать другие классовые методы** через `cls.другой_метод()`;
- **вызывать статические методы** через `cls.статический_метод()`;
- **использовать `super()`** для вызова родительских методов.

Чего **нельзя**:

- **обращаться к `self`** — его нет (метод привязан к классу, а не к объекту);
- **работать с данными конкретного объекта** — объекта нет;
- **вызывать обычные методы** без объекта — их не у кого вызывать.

### Аналогия

Представь **кассу самообслуживания** в магазине. Ты **не идёшь** к обычному кассиру (обычный метод), ты **не зовёшь** кого-то со стороны (статический метод). Ты **идёшь** к автомату, который **знает**, что продаёт, и **сам** оформляет покупку. `cls` — это «знание о магазине», которое автомат получает **автоматически**. Автомат **не спрашивает** тебя: «А какой магазин?» — он **уже знает**.

Так же и `@classmethod` — Python **сам** подставляет класс в `cls`. Программисту **не нужно** об этом думать.

### Вывод

`@classmethod` — метод, который получает **класс** первым аргументом. Вызывается **через класс**. Часто используется как **альтернативный конструктор**. Python **сам** подставляет класс в `cls`, что делает метод **удобным** и **безопасным**.

---

## 15.3. Главная особенность: работа с наследованием

**Ключевое преимущество** `@classmethod` — правильная работа при **наследовании**. Именно здесь он **выигрывает** у `@staticmethod`. Это **главная причина**, почему `@classmethod` вообще существует в языке. Если бы не наследование, можно было бы обойтись статическими методами. Но **наследование есть**, и оно **важно** — поэтому нужен `@classmethod`.

### Проблема со `@staticmethod`

Сравним два подхода. Сначала — **статический метод**:

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    @staticmethod
    def from_string(data):
        address, area, price = data.split(";")
        return Apartment(address, int(area), int(price))    # ← жёстко Apartment


class Studio(Apartment):
    pass
```

Что здесь происходит? Внутри `from_string` **жёстко** прописано имя класса `Apartment`. Не `cls`, не переменная — именно `Apartment`. Это работает, **пока нет наследников**. Но как только появляется `Studio`, начинаются проблемы.

Создадим студию из строки:

```python
studio = Studio.from_string("ул. Пушкина, д. 1, кв. 7;25;4000000")
print(type(studio))    # <class 'Apartment'> — НЕ Studio!
```

**Что не так?** Мы вызвали `Studio.from_string(...)`, но получили объект **`Apartment`**, а не `Studio`. Потому что внутри `from_string` **жёстко** прописано `Apartment(...)`. Метод **не знает**, через какой класс его вызвали. Он всегда создаёт **`Apartment`** — независимо от того, `Apartment.from_string(...)` это или `Studio.from_string(...)`.

### Последствия этой проблемы

Представь, что мы вызвали `Studio.from_string(...)` и получили `Apartment`. Что дальше?

- **Мы хотели создать студию** — а получили обычную квартиру.
- **Если у `Studio` есть свои поля** — например, `balcony` — их **не будет** в объекте.
- **Если у `Studio` есть свои методы** — их **не будет** в объекте.
- **Программа сломается** в неожиданном месте, когда попытается обратиться к полю `balcony`.

Это **серьёзная** проблема. Она проявляется **не сразу** — только когда появляется **наследник**. Программист, который писал `from_string`, **не думал** о наследовании. А когда наследник появился, оказалось, что метод работает **неправильно**.

### Решение с `@classmethod`

Теперь — **классовый метод**:

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    @classmethod
    def from_string(cls, data):
        address, area, price = data.split(";")
        return cls(address, int(area), int(price))    # ← cls, а не Apartment


class Studio(Apartment):
    pass
```

Обрати внимание на **единственное** отличие: `Apartment(...)` заменено на `cls(...)`. Казалось бы, мелочь. Но это **меняет всё**.

Создадим студию:

```python
studio = Studio.from_string("ул. Пушкина, д. 1, кв. 7;25;4000000")
print(type(studio))    # <class 'Studio'> — правильно!
```

**Что произошло?**

1. `Studio.from_string(...)` — Python подставил `cls = Studio`.
2. Внутри `cls(...)` — вызван `Studio(...)`.
3. Возвращён объект **`Studio`**, а не `Apartment`.

### Схема

```
        @staticmethod                    @classmethod
        ─────────────                    ────────────
        Apartment(...) жёстко            cls(...) — фактический класс
                ↓                                ↓
        Studio.from_string → Apartment   Studio.from_string → Studio
        (НЕПРАВИЛЬНО!)                   (ПРАВИЛЬНО!)
```

### Почему так работает

`cls` — это **ссылка на фактический класс**, через который вызван метод. Когда вы вызываете `Studio.from_string(...)`, `cls` автоматически **становится `Studio`**. Когда `Apartment.from_string(...)` — `cls` становится `Apartment`. Python **сам** определяет, через какой класс вызван метод, и **подставляет** его в `cls`.

Это **магия** `@classmethod`: программисту **не нужно** об этом думать. Он пишет `cls(...)` — и это работает **для всех** наследников. Один код — много классов.

### Что это даёт на практике

`@classmethod` даёт **четыре** важных преимущества:

- **Единый код.** Не нужно **дублировать** фабричные методы в каждом наследнике. Метод пишется **один раз** в родителе — и работает для **всех**.
- **Гибкость.** Наследник **автоматически** получает правильное поведение. Программисту **не нужно** ничего переопределять.
- **Расширяемость.** Можно **добавить** нового наследника — и `from_string` **сразу** будет работать для него.
- **Безопасность.** Нельзя случайно создать **не тот** класс. Даже если программист ошибётся, `cls(...)` создаст **правильный**.

### Аналогия

Представь **шаблон документа**. У тебя есть **общий шаблон** «Заявление». Ты можешь создать на его основе **три разных документа**: «Заявление на отпуск», «Заявление на увольнение», «Заявление на повышение». Все они **основаны** на общем шаблоне, но **каждый** — со своими полями.

`@classmethod` работает так же: **один** метод, но работает **правильно** для **всех** наследников. Ты пишешь **один раз** — а работает **для всех**.

### Когда это особенно важно

`@classmethod` **критичен**, когда:

- У класса есть **наследники**. Без `@classmethod` фабричные методы будут **ломаться**.
- Есть **фабричные методы** (`from_string`, `from_dict`, `from_json`). Они должны **правильно** работать в наследниках.
- Метод **создаёт** объекты. Он должен создавать **правильный тип**.

Если **наследников нет** и **не будет** — `@staticmethod` тоже сработает. Но если **наследники появятся** — придётся **переписывать** всё на `@classmethod`. Лучше **сразу** использовать `@classmethod` — «на будущее».

### Вывод

`@classmethod` **автоматически подстраивается** под наследника через `cls`. Это его **главное преимущество** перед `@staticmethod`. Используйте `@classmethod`, если метод **создаёт** объекты класса — это **защитит** от проблем с наследованием в будущем.

---

## 15.4. Сравнение трёх видов методов

Разберём **все три** вида методов в **одной таблице** — чтобы увидеть разницу и понять, когда какой использовать.

| Вид | Декоратор | Первый параметр | Что получает | Когда вызывается |
|-----|-----------|-----------------|--------------|------------------|
| **Обычный** | нет | `self` | Объект | Через объект |
| **Классовый** | `@classmethod` | `cls` | Класс | Через класс или объект |
| **Статический** | `@staticmethod` | нет | Ничего | Через класс или объект |

### Что важно понимать

**Обычный метод** — самый распространённый. Он **привязан к объекту** и работает с его данными: `self.address`, `self.area`. Может **менять** состояние объекта. Вызывается через **объект**: `flat.info()`.

**Классовый метод** — привязан к **классу**. Работает с данными класса: `cls.count`, `cls(...)`. Может **менять** состояние класса (например, счётчик). Вызывается через **класс**: `Apartment.from_string(...)`.

**Статический метод** — **не привязан** ни к чему. Не имеет доступа **ни к объекту, ни к классу**. Работает только с **аргументами**. Вызывается через **класс** (или объект).

### Пример: все три в одном классе

```python
class Apartment:
    count = 0

    def __init__(self, address, area):
        self.address = address
        self.area = area
        Apartment.count += 1

    # Обычный — работает с объектом.
    def info(self):
        return f"{self.address}: {self.area} м²"

    # Классовый — работает с классом.
    @classmethod
    def from_string(cls, data):
        address, area = data.split(";")
        return cls(address, int(area))

    @classmethod
    def get_count(cls):
        return cls.count

    # Статический — утилита.
    @staticmethod
    def is_valid_area(area):
        return 5 <= area <= 500
```

Использование:

```python
# Обычный метод — через объект.
flat = Apartment("ул. Ленина", 65)
print(flat.info())                    # ул. Ленина: 65 м²

# Классовый метод — через класс.
new_flat = Apartment.from_string("ул. Садовая;80")
print(Apartment.get_count())          # 2

# Статический метод — через класс.
print(Apartment.is_valid_area(65))    # True
```

### Как выбрать вид метода

**Правило** очень простое:

1. Метод использует `self`? → **Обычный метод**.
2. Метод использует `cls`? → **`@classmethod`**.
3. Метод не использует ни то, ни другое? → **`@staticmethod`**.

### Примеры выбора

| Ситуация | Вид метода | Почему |
|----------|-----------|--------|
| Прочитать `self.name` | Обычный | Работает с объектом |
| Изменить `self.area` | Обычный | Работает с объектом |
| Увеличить `cls.count` | `@classmethod` | Работает с классом |
| Создать объект через `cls(...)` | `@classmethod` | Работает с классом |
| Проверить число `is_valid_area(area)` | `@staticmethod` | Не работает ни с чем |
| Отформатировать строку | `@staticmethod` | Не работает ни с чем |

### Аналогия

Представь **ресторан**:

- **Официант** — **обычный метод**. Он знает **свой столик** и **своих клиентов**. Он **привязан** к конкретному столику.
- **Менеджер зала** — **классовый метод**. Он знает **весь зал** и может **создать** новый столик. Он **работает** со всем рестораном.
- **Калькулятор чаевых** — **статический метод**. Ему **не важно**, кто клиент, — он просто считает. Он **не привязан** ни к чему.

Все три **работают в одном ресторане**, но **по-разному**.

### Что если использовать не тот вид

**Пример ошибки №1:** сделать метод `@staticmethod`, который **создаёт объект**:

```python
@staticmethod
def from_string(data):
    return Apartment(...)      # ← жёстко Apartment
```

Проблема: **ломается наследование**. При `Studio.from_string(...)` вернётся `Apartment`.

**Пример ошибки №2:** сделать метод **обычным**, который **не использует объект**:

```python
def is_valid_area(self, area):
    return 5 <= area <= 500
```

Проблема: **нужен объект** для вызова. Приходится создавать «пустой» объект ради проверки.

**Пример ошибки №3:** сделать метод `@classmethod`, который **не использует `cls`**:

```python
@classmethod
def is_valid_area(cls, area):
    return 5 <= area <= 500       # cls не используется!
```

Проблема: лишний параметр `cls`, который **не нужен**. Лучше `@staticmethod`.

### Вывод

Три вида методов — **три разных подхода**. Выбор зависит от того, **с чем** работает метод: **с объектом**, **с классом** или **ни с чем**. Правило **простое**, но соблюдать его **важно** — иначе будут **ошибки** и **путаница**.

---

## 15.5. Типичные случаи использования `@classmethod`

Разберём **пять** типичных случаев использования `@classmethod`. Каждый — с примером и подробным объяснением.

### 15.5.1. Фабричные методы (альтернативные конструкторы)

**Самый частый** случай. Класс может иметь **несколько** способов создания объектов. Обычный конструктор `__init__` — **один**. Но данных может приходить **много разных форматов**:

- из строки;
- из словаря;
- из JSON;
- из файла;
- из базы данных.

Для каждого формата — **свой** фабричный метод.

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    @classmethod
    def from_string(cls, data):
        # Формат: "адрес;площадь;цена"
        address, area, price = data.split(";")
        return cls(address, int(area), int(price))

    @classmethod
    def from_dict(cls, data):
        # Формат: {"address": ..., "area": ..., "price": ...}
        return cls(data["address"], data["area"], data["price"])

    @classmethod
    def from_json(cls, json_str):
        import json
        return cls.from_dict(json.loads(json_str))
```

Теперь можно создать квартиру **четырьмя** способами:

```python
flat1 = Apartment("ул. Ленина", 65, 8_500_000)              # обычный
flat2 = Apartment.from_string("ул. Ленина;65;8500000")      # из строки
flat3 = Apartment.from_dict({"address": "ул. Ленина",
                             "area": 65,
                             "price": 8_500_000})           # из словаря
flat4 = Apartment.from_json('{"address": "ул. Ленина", "area": 65, "price": 8500000}')
```

**Каждый метод** — это **альтернативный конструктор**. Он создаёт объект **своим способом**.

**Почему это удобно:**

- **Один класс — много способов создания.** Нет нужды в разных классах или функциях.
- **Читаемость.** `Apartment.from_string(...)` говорит само за себя — понятно, откуда данные.
- **Гибкость.** Легко добавить новый способ — ещё один `@classmethod`.
- **Наследование.** Работает правильно для наследников — они **автоматически** получают все фабричные методы.

**Что важно:** внутри **всегда** используйте `cls(...)`, а не `Apartment(...)`. Это обеспечивает **правильную работу** с наследниками.

### 15.5.2. Работа с атрибутами класса

Если метод **читает** или **меняет** атрибуты **класса** — нужен `@classmethod`:

```python
class Apartment:
    count = 0

    def __init__(self, address):
        self.address = address
        Apartment.count += 1

    @classmethod
    def get_count(cls):
        return cls.count

    @classmethod
    def reset_count(cls):
        cls.count = 0
```

Здесь `cls.count` — **атрибут класса**. Через `cls` метод **видит** его. Когда вызывается `Apartment.get_count()`, `cls = Apartment` — метод видит `Apartment.count`. Когда `Studio.get_count()` — `cls = Studio` — метод видит `Studio.count` (если он есть) или наследует `Apartment.count`.

**Почему не `@staticmethod`?** Потому что статический метод **не имеет** доступа к `cls`. Ему пришлось бы **жёстко** писать `Apartment.count`:

```python
@staticmethod
def get_count():
    return Apartment.count      # ← жёстко Apartment
```

А это **ломает наследование**. Если `Studio` имеет **свой** счётчик:

```python
class Studio(Apartment):
    count = 0                   # ← свой счётчик
```

То `Studio.get_count()` вернёт `Apartment.count`, а не `Studio.count`. Это **неправильно**.

**С `@classmethod`** всё работает **правильно**:

```python
@classmethod
def get_count(cls):
    return cls.count            # ← правильно, работает через cls
```

Здесь `Studio.get_count()` вернёт `Studio.count`.

### 15.5.3. Создание объектов в наследниках

Когда класс **создаёт объекты через `cls`**, наследники автоматически получают **правильный** тип:

```python
class Apartment:
    @classmethod
    def from_string(cls, data):
        address, area = data.split(";")
        return cls(address, int(area))


class Studio(Apartment):
    pass


studio = Studio.from_string("ул. Пушкина;25")
print(type(studio))    # <class 'Studio'>
```

**Без `@classmethod`** — было бы `Apartment`. С `@classmethod` — `Studio`, потому что `cls` **автоматически** стал `Studio`.

**Это особенно важно**, если наследники **добавляют** свои поля. При `cls(...)` наследник **сам** решает, как создавать объект — и его конструктор **дополнит** нужные поля. Если бы метод был статическим, он бы **всегда** создавал родительский класс — и наследник **терял** бы свои поля.

### 15.5.4. Работа с абстрактными классами

В **абстрактных классах** (см. главу 7) часто определяют **абстрактные классовые методы**:

```python
from abc import ABC, abstractmethod

class HomeItem(ABC):
    @classmethod
    @abstractmethod
    def from_string(cls, data):
        pass
```

Наследники **обязаны** реализовать `from_string`. Это **контракт**: «каждый наследник должен уметь создаваться из строки».

**Порядок декораторов:** `@classmethod` **сверху**, `@abstractmethod` **снизу**. Если поменять — абстрактность **не сработает**. Python **выдаст ошибку** или проигнорирует `@abstractmethod`.

**Зачем это нужно:**

- **Гарантия интерфейса.** Все наследники **обязаны** уметь создаваться из строки.
- **Единый стиль.** Разные классы — один способ создания.
- **Полиморфизм.** Можно создать любой `HomeItem` из строки, не зная заранее его класс.

### 15.5.5. Реестры и синглтоны

`@classmethod` удобен для **реестров** — когда класс **хранит список** всех созданных объектов:

```python
class Apartment:
    all_apartments = []

    def __init__(self, address):
        self.address = address
        Apartment.all_apartments.append(self)

    @classmethod
    def find_by_address(cls, address):
        for flat in cls.all_apartments:
            if flat.address == address:
                return flat
        return None
```

Здесь `cls.all_apartments` — **реестр**. Метод **ищет** в нём через `cls`.

**Почему через `cls`:** если наследник `Studio` **переопределит** `all_apartments`, поиск будет **по своему** реестру. Гибко и правильно. Если бы использовали `Apartment.all_apartments` **жёстко**, наследник **не смог** бы переопределить реестр.

### Вывод

`@classmethod` используется **везде**, где метод работает **с классом**, а не с объектом. Пять случаев выше — **самые частые** в реальной практике. Особенно часто — **фабричные методы** и **работа с атрибутами класса**.

---

## 15.6. `@classmethod` и наследование

Разберём **наследование** подробно — это **ключевая** тема для `@classmethod`. Метод `@classmethod` **создан** именно для того, чтобы **правильно работать** с наследниками. Разберём все сценарии.

### Наследование без переопределения

Если наследник **не переопределяет** классовый метод — он **наследуется** и работает **правильно**:

```python
class Apartment:
    @classmethod
    def from_string(cls, data):
        address, area = data.split(";")
        return cls(address, int(area))


class Studio(Apartment):
    pass


studio = Studio.from_string("ул. Пушкина;25")
print(type(studio))    # <class 'Studio'>
```

`cls` **автоматически** становится `Studio`. Метод **не нужно** переопределять. Наследник **получает** работающий метод **бесплатно**.

### Что происходит пошагово

Разберём **пошагово**, что происходит при вызове `Studio.from_string("ул. Пушкина;25")`:

1. Python ищет `from_string` в классе `Studio`. **Не находит** — там только `pass`.
2. Ищет в родителе `Apartment`. **Находит**.
3. Понимает, что это **классовый метод** (`classmethod`).
4. **Подставляет фактический класс** `Studio` в `cls`. То есть вызов превращается в `from_string(Studio, "ул. Пушкина;25")`.
5. Внутри `cls(...)` — это `Studio(...)`. Создаётся объект `Studio`.
6. Возвращает объект `Studio`.

**Ключевой момент** — шаг 4. Python **сам** определяет, что метод вызван через `Studio`, и подставляет именно `Studio` в `cls`. Программисту **не нужно** об этом думать.

### Переопределение классового метода

Наследник **может** переопределить метод — например, если формат строки **другой**:

```python
class Studio(Apartment):
    @classmethod
    def from_string(cls, data):
        # Свой формат: "адрес;площадь;балкон"
        address, area, balcony = data.split(";")
        # Создаём объект через cls — правильный тип.
        obj = cls(address, int(area))
        obj.balcony = int(balcony)
        return obj
```

Теперь:

```python
studio = Studio.from_string("ул. Пушкина;25;3")
print(type(studio))         # <class 'Studio'>
print(studio.balcony)       # 3
```

Наследник **полностью** контролирует создание своих объектов. Он **переопределил** метод — и теперь формат строки **другой**.

### Что если вызвать `Apartment.from_string` с форматом Studio

```python
Apartment.from_string("ул. Пушкина;25;3")    # ValueError: too many values to unpack
```

Родительский метод **не знает** про формат Studio. Это **нормально**: каждый класс **отвечает** за свой формат. Если нужна **совместимость** — проектируйте формат **заранее**.

### `super()` в классовом методе

Если нужно **вызвать** родительский классовый метод:

```python
class Studio(Apartment):
    @classmethod
    def from_string(cls, data):
        # Используем родительскую логику.
        return super().from_string(data)
```

`super()` работает **без аргументов** — Python **сам** найдёт родителя.

**Когда это нужно:**

- Наследник **дополняет** родительский метод (например, добавляет ещё одно поле после создания).
- Наследник **изменяет** формат, но хочет **сохранить** базовую логику.
- Наследник **вызывает** родительский метод в **другом** контексте.

### Наследование и атрибуты класса

Атрибуты класса **наследуются**:

```python
class Apartment:
    count = 0

    def __init__(self):
        Apartment.count += 1

    @classmethod
    def get_count(cls):
        return cls.count


class Studio(Apartment):
    pass


Apartment()
Apartment()
Studio()

print(Apartment.get_count())    # 3 — считает все объекты
print(Studio.get_count())       # 3 — тоже 3, потому что cls.count ссылается на Apartment.count
```

**Нюанс:** если наследник **не переопределяет** `count`, он **использует** родительский. Если **переопределит** — получит **свой**:

```python
class Studio(Apartment):
    count = 0       # ← свой счётчик


Apartment()
Studio()

print(Apartment.get_count())    # 1
print(Studio.get_count())       # 1 — свой счётчик
```

### Схема

```
        Apartment.count = 0                    Studio.count = 0
        ─────────────────────                  ─────────────────
        Apartment.get_count() → 1              Studio.get_count() → 1
        (свой счётчик)                         (свой счётчик)
```

### Что происходит при наследовании

| Что | Наследуется? | Переопределяется? |
|-----|--------------|-------------------|
| Классовый метод | ✅ Да | ✅ Да |
| Атрибут класса | ✅ Да | ✅ Да |
| `cls` в методе | ✅ Подстраивается | ✅ Автоматически |
| `super()` | ✅ Работает | ✅ Работает |

### Вывод

`@classmethod` **правильно работает** с наследованием. `cls` **автоматически** подстраивается под фактический класс. Атрибуты класса **наследуются**, но могут быть **переопределены**. `super()` работает **без аргументов**. Это делает `@classmethod` **надёжным** инструментом для **фабричных методов** в иерархии классов.

---

## 15.7. `@classmethod` и `@property`

Иногда возникает **соблазн** сочетать `@classmethod` с `@property` — чтобы создать **вычисляемое свойство** уровня **класса**. Например, чтобы `Apartment.total` возвращало количество квартир **без скобок**:

```python
class Apartment:
    count = 0

    @classmethod
    @property
    def total(cls):
        return cls.count
```

**Важно понимать:** такой подход **работает не всегда** в Python 3.10 и ниже. В **Python 3.11+** — работает **стабильно**. То есть **один и тот же код** может работать **по-разному** в зависимости от **версии Python**. Это **опасно**.

### Обходной путь

Если вы **не хотите** зависеть от версии Python — используйте **обычный классовый метод**:

```python
class Apartment:
    count = 0

    @classmethod
    def total(cls):
        return cls.count


print(Apartment.total())    # работает
```

**Разница:** без `@property` — `Apartment.total()`. С `@property` — `Apartment.total`. Второй вариант **удобнее** (без скобок), но требует **Python 3.11+**.

### Почему это сложно

`@property` и `@classmethod` — **разные** механизмы, которые работают **на разных уровнях**:

- **`@property`** — это **дескриптор**, который работает **на объектах**. Он перехватывает доступ к атрибуту **экземпляра**.
- **`@classmethod`** — это **дескриптор**, который работает **на классах**. Он перехватывает вызов метода **через класс**.

Их **сочетание** — **нестандартное**. Python **не гарантирует** стабильность во всех версиях. В некоторых версиях поведение **меняется** — и код, который работал, **ломается**.

### Что использовать вместо

**Вариант 1:** обычный классовый метод.

```python
@classmethod
def total(cls):
    return cls.count
```

Просто и работает **везде**. Единственный минус — **скобки** при вызове: `Apartment.total()`.

**Вариант 2:** метакласс (см. главу 23).

```python
class ApartmentMeta(type):
    @property
    def total(cls):
        return cls.count

class Apartment(metaclass=ApartmentMeta):
    count = 0


print(Apartment.total)    # работает через метакласс
```

Сложнее, но **надёжнее**. Метакласс — **правильный** способ создать **свойство уровня класса**.

**Правило:** если хотите **свойство класса** — используйте **классовый метод** или **метакласс**. Не пытайтесь **сочетать** `@classmethod` и `@property` без понимания.

### Вывод

`@classmethod` и `@property` — **разные** инструменты. Их **сочетание** требует **аккуратности**. В большинстве случаев достаточно **классового метода** с обычными скобками.

---

## 15.8. `@classmethod` в реальных проектах

Где `@classmethod` **активно** используется. Разберём **пять** примеров из **реальных библиотек** и **фреймворков**.

### 15.8.1. Django

Django — один из **самых популярных** веб-фреймворков Python. В нём `@classmethod` используется **активно** — особенно в **моделях** и **менеджерах**:

```python
class Apartment(models.Model):
    @classmethod
    def from_string(cls, data):
        ...
```

А вот пример **кастомного менеджера**:

```python
class ApartmentManager(models.Manager):
    @classmethod
    def available(cls):
        return cls.filter(is_available=True)
```

Здесь `cls` — **фактический** класс. Если наследник — запрос идёт по **наследнику**. Это **стандарт** Django.

### 15.8.2. SQLAlchemy

SQLAlchemy — популярная ORM для Python. Здесь `@classmethod` используется для **поисковых** методов:

```python
class Apartment(Base):
    @classmethod
    def find_by_address(cls, address):
        return session.query(cls).filter_by(address=address).first()
```

Здесь `cls` — **фактический** класс. Если наследник — запрос идёт по **наследнику**. Это позволяет **легко расширять** иерархии.

### 15.8.3. Dataclasses

Стандартный модуль `dataclasses` **отлично сочетается** с `@classmethod`:

```python
from dataclasses import dataclass

@dataclass
class Apartment:
    address: str
    area: int

    @classmethod
    def from_string(cls, data):
        address, area = data.split(";")
        return cls(address, int(area))
```

Здесь `@dataclass` создаёт `__init__` и `__repr__`, а `@classmethod` — **альтернативные** конструкторы. Они **дополняют** друг друга. Мы разберём `@dataclass` подробно в главе 20.

### 15.8.4. Парсеры и сериализация

Классовые методы — **стандарт** для **парсинга** и **сериализации**. Почти **каждая** библиотека использует их:

```python
class Config:
    @classmethod
    def from_file(cls, path):
        with open(path) as f:
            data = json.load(f)
        return cls(**data)

    @classmethod
    def from_env(cls):
        return cls(
            host=os.environ["HOST"],
            port=int(os.environ["PORT"]),
        )
```

Такой подход **удобен** и **гибок**. Легко добавить новые способы загрузки.

### 15.8.5. Наш дом

```python
class Room:
    @classmethod
    def from_string(cls, data):
        name, area = data.split(";")
        return cls(name, int(area))
```

### Что общего у всех примеров

Во **всех** примерах есть **три общих** черты:

- **Альтернативный конструктор** — создаёт объект из **другого** формата (строки, словаря, файла).
- **Возвращает объект** класса (или наследника, если метод вызван через наследника).
- **Работает с наследованием** правильно — благодаря `cls(...)`.

### Вывод

`@classmethod` используется **везде**, где есть **альтернативные конструкторы** или **работа с классом**. Это **стандартный** инструмент. Если вы **не используете** `@classmethod` — скорее всего, вы **упускаете** удобный приём, который сделал бы код **короче** и **гибче**.

## 15.9. Нюансы и тонкости

Разберём **десять** нюансов, которые часто упускают даже **опытные** программисты. Каждый нюанс — **важная** деталь, которая влияет на **поведение** кода.

### 15.9.1. `cls` — не ключевое слово

Как и `self`, `cls` — это **соглашение**, а не требование языка. Первый параметр классового метода можно назвать **как угодно**:

```python
@classmethod
def from_string(klass, data):
    return klass(...)
```

Этот код **работает** — `klass` выполняет ту же роль, что и `cls`. Точно так же и в обычных методах можно назвать первый параметр не `self`, а как угодно. Но **принято** использовать именно `cls` — так делают **все** программисты Python. Если назвать иначе — **коллеги не поймут**, IDE может **подсветить** предупреждением, а читаемость **снизится**. Следуйте соглашению: `self` для объектов, `cls` для классов.

### 15.9.2. `@classmethod` можно вызывать через объект

Классовый метод можно вызвать и **через объект**:

```python
flat = Apartment("ул. Ленина", 65, 8_500_000)
new_flat = flat.from_string("ул. Садовая;80;12000000")
```

Это **сработает**, но **не рекомендуется**. Потому что **вводит в заблуждение**: кажется, что метод работает с `flat`, но на самом деле он работает с **классом** `Apartment`. Объект `flat` тут **не используется** — он просто «случайно» оказался первым. Программист, читающий код, может **подумать**, что `flat` важен — а это **не так**. **Правило:** классовые методы вызывайте **через класс**, а не через объект.

### 15.9.3. Не используйте `Apartment(...)` внутри `@classmethod`

Это **самая частая** ошибка новичков. Внутри `@classmethod` **всегда** используйте `cls(...)`, **никогда** — имя класса **напрямую**:

**Плохо:**

```python
@classmethod
def from_string(cls, data):
    ...
    return Apartment(...)      # ← жёстко, ломает наследование
```

**Хорошо:**

```python
@classmethod
def from_string(cls, data):
    ...
    return cls(...)            # ← правильно, работает с наследниками
```

**Почему это важно:** если написать `Apartment(...)`, метод **всегда** возвращает объект `Apartment`, **независимо** от того, через какой класс вызван. При `Studio.from_string(...)` вернётся `Apartment` — **неправильно**. С `cls(...)` метод возвращает **фактический** класс. **Правило:** в `@classmethod` **всегда** `cls(...)`, никогда — имя класса **напрямую**.

### 15.9.4. Абстрактные классовые методы

Если определяете **абстрактный** классовый метод — **порядок декораторов** критически важен:

```python
@classmethod
@abstractmethod
def from_string(cls, data):
    pass
```

`@classmethod` **сверху**, `@abstractmethod` **снизу**. Если поменять местами — **не сработает**. Python либо **выдаст ошибку**, либо **проигнорирует** `@abstractmethod`, и абстрактность **не будет** работать. **Правило:** декораторы применяются **снизу вверх**. Сначала `@abstractmethod` помечает функцию как абстрактную, потом `@classmethod` оборачивает её в классовый метод.

### 15.9.5. `@classmethod` и статические методы в одном классе

Можно **сочетать** без проблем:

```python
class Apartment:
    @classmethod
    def from_string(cls, data):
        ...

    @staticmethod
    def is_valid_area(area):
        ...
```

Оба вида методов **сосуществуют** и **не конфликтуют**. Более того — они **дополняют** друг друга. Классовые методы — для **создания** объектов, статические — для **утилит** и **проверок**. В одном классе может быть **сколько угодно** методов обоих видов.

### 15.9.6. `@classmethod` и `__init__`

`__init__` — **не** классовый метод. Он **обычный**. Но `@classmethod` **часто** вызывает `__init__` **неявно** через `cls(...)`:

```python
@classmethod
def from_string(cls, data):
    ...
    return cls(address, area, price)    # ← вызовет cls.__init__
```

То есть, когда вы пишете `cls(address, area, price)`, Python **создаёт** объект, **вызывает** `__init__` с этими аргументами, и **возвращает** готовый объект. Это **стандартный** механизм создания объектов — тот же, что и при `Apartment(address, area, price)`.

### 15.9.7. `@classmethod` и `super()`

`super()` в `@classmethod` работает **без аргументов** — Python **сам** найдёт родителя:

```python
class Studio(Apartment):
    @classmethod
    def from_string(cls, data):
        # Сначала вызываем родительскую логику.
        result = super().from_string(data)
        # Потом дополняем.
        result.is_studio = True
        return result
```

Здесь `super()` возвращает **родительский класс** — `Apartment`. Через него вызывается **родительский** `from_string`. Python **автоматически** разбирается, какой родитель нужен — даже при **множественном** наследовании (см. главу 19). Это делает `@classmethod` **мощным** инструментом для **расширения** родительских методов.

### 15.9.8. `@classmethod` и `functools.singledispatch`

`singledispatch` (см. главу 10) — это механизм **диспетчеризации** по типу первого аргумента. Он **плохо сочетается** с `@classmethod`: по умолчанию `singledispatch` работает **только с функциями**, а не с методами. Если нужно — используйте **обходные пути**:

- `singledispatchmethod` (Python 3.8+) — специальный декоратор для методов.
- Ручная проверка типа через `isinstance` — вместо диспетчеризации.

**Правило:** не пытайтесь **сочетать** `@classmethod` и `singledispatch` без понимания. Это **сложно** и **редко** нужно.

### 15.9.9. `@classmethod` и `@property` — порядок

Если всё-таки **сочетаете** (что не рекомендуется):

```python
@classmethod
@property
def total(cls):
    ...
```

`@classmethod` **сверху**, `@property` **снизу**. В Python 3.11+ работает **стабильно**. В более старых версиях — **может не работать**. **Правило:** если хотите свойство класса — используйте **метакласс** (см. главу 23) или **обычный классовый метод**. Сочетание `@classmethod` и `@property` — **хрупкое** решение.

### 15.9.10. `@classmethod` и type hints

Аннотации типов для `cls` — **полезны** для **документации** и **проверки** через `mypy`:

```python
from typing import Type

@classmethod
def from_string(cls: Type["Apartment"], data: str) -> "Apartment":
    ...
```

Здесь `Type["Apartment"]` указывает, что `cls` — это **класс** `Apartment` (или его наследник). `-> "Apartment"` указывает, что метод возвращает **объект** `Apartment`. Аннотации **не влияют** на выполнение, но **помогают** IDE и **статическому анализатору** ловить ошибки.

### Вывод

`@classmethod` требует **аккуратности** в нюансах. Правило **простое**: `cls` вместо `self`, `cls(...)` вместо `Apartment(...)`. Остальное — **стандартные** приёмы Python. Соблюдайте нюансы — и код будет **чистым** и **надёжным**.

---

## 15.10. Типичные ошибки

Разберём **десять** самых частых ошибок. Каждая — с примером и **подробным** объяснением, почему это ошибка и как её исправить.

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Используют `Apartment(...)` внутри `@classmethod` | Ломает наследование | Использовать `cls(...)` |
| Забыли `@classmethod` | Python передаст `self` | Добавить декоратор |
| Пишут `self` вместо `cls` | Не работает | Использовать `cls` |
| Вызывают через объект | Работает, но вводит в заблуждение | Вызывать через класс |
| Используют `@classmethod` там, где `self` | Метод не получает объект | Использовать обычный метод |
| Используют `@classmethod` вместо `@staticmethod` | Лишний параметр | `@staticmethod` без `cls` |
| Путают порядок `@classmethod` и `@abstractmethod` | Не сработает | `@classmethod` сверху |
| Пытаются изменить объект в `@classmethod` | Нет `self` | Передать объект параметром |
| Возвращают не объект `cls` | Странный тип | Возвращать `cls(...)` |
| Забыли `return` | Возвращается `None` | Вернуть объект |

### Подробный разбор каждой ошибки

**Ошибка 1.** `Apartment(...)` вместо `cls(...)` — **самая частая**. Ломает наследование. Всегда используйте `cls(...)`. Это **главное правило** работы с `@classmethod`.

**Ошибка 2.** Забыли `@classmethod` — Python передаст **`self`** первым. Внутри метода `cls` будет **объектом**, а не классом. Ошибка **не сразу** — только когда попытаетесь вызвать `cls(...)`. Python выдаст **странное** сообщение. **Правило:** всегда ставьте декоратор.

**Ошибка 3.** `self` вместо `cls` — не работает, потому что `self` — это **объект**. Если в `@classmethod` написать `self`, Python передаст **класс** в `self` — и это **собьёт с толку**. **Правило:** в `@classmethod` — только `cls`.

**Ошибка 4.** Вызов через объект — работает, но **вводит в заблуждение**. Лучше — через **класс**. **Правило:** классовые методы вызывайте через `Apartment.from_string(...)`, а не `flat.from_string(...)`.

**Ошибка 5.** `@classmethod` там, где нужен `self` — метод **не получает** объект. Если нужно работать с объектом — используйте **обычный** метод. **Правило:** если метод использует `self` — он обычный.

**Ошибка 6.** `@classmethod` вместо `@staticmethod` — лишний параметр `cls`, который **не используется**. Лучше — `@staticmethod`. **Правило:** если `cls` не используется — `@staticmethod`.

**Ошибка 7.** Порядок `@classmethod` и `@abstractmethod` — **важен**. `@classmethod` **сверху**, `@abstractmethod` **снизу**. **Правило:** декораторы применяются **снизу вверх**.

**Ошибка 8.** Пытаются изменить объект в `@classmethod` — **нет `self`**. Передайте объект **параметром**. **Правило:** для работы с объектом — обычный метод.

**Ошибка 9.** Возвращают не `cls(...)` — странный тип. Например, `Apartment(...)` — ломает наследование. **Правило:** всегда возвращайте `cls(...)`.

**Ошибка 10.** Забыли `return` — возвращается `None`. **Правило:** всегда **возвращайте** объект.

### Как избежать ошибок

1. **Пишите `cls`, а не `self`.**
2. **Всегда `cls(...)`, не `Apartment(...)`.**
3. **Проверяйте `@classmethod` перед методом.**
4. **Вызывайте через класс.**
5. **Выбирайте вид метода по работе.**

### Вывод

`@classmethod` **прост**, но требует **внимания**. Основные ошибки — **жёсткое имя класса** и **забытый декоратор**. Следуйте правилам — и код будет **чистым** и **надёжным**.

---

## 15.11. Что важно запомнить

- **`@classmethod`** — метод, который получает **класс** первым аргументом.
- **Первый параметр** — `cls` (ссылка на класс).
- **Вызывается** через **класс**: `Apartment.from_string(...)`.
- **Главное преимущество** — правильная работа с **наследованием** (`cls` подстраивается под наследника).
- **Типичное использование** — **фабричные методы** (альтернативные конструкторы).
- **Также используется** для работы с **атрибутами класса**.
- **Внутри** — **всегда** `cls(...)`, а не `ClassName(...)`.
- **`@classmethod` ≠ `@staticmethod`**: первый получает `cls`, второй — нет.
- **`@classmethod` ≠ обычный метод**: первый получает класс, второй — объект.
- **Три вида методов** сосуществуют в одном классе без проблем.
- **Порядок декораторов** с `@abstractmethod`: `@classmethod` сверху.
- **`super()`** в `@classmethod` работает без аргументов.
- **Правило выбора:** `self` → обычный, `cls` → `@classmethod`, ни то ни другое → `@staticmethod`.

## Задача 15 (классовый метод `@classmethod`)

В агентство недвижимости часто поступают данные о квартирах **в виде строк** — например, из файла или базы данных: `"ул. Ленина, д. 10, кв. 5;65;8500000"`. Нужно **создавать** объекты **напрямую** из таких строк.

**Часть 1. Класс `Apartment`.** Создайте класс `Apartment` с полями:

- `address` — адрес квартиры;
- `area` — площадь в м²;
- `price` — цена.

Реализуйте конструктор и метод `__repr__`.

**Часть 2. Классовый метод `from_string`.** Реализуйте **классовый метод** `from_string(cls, data)`, который:

- принимает строку формата `"адрес;площадь;цена"`;
- **разбивает** её по точке с запятой;
- **преобразует** площадь и цену в целые числа;
- **создаёт** объект через `cls(...)` (не через `Apartment(...)`);
- **возвращает** новый объект.

Продемонстрируйте создание квартиры через этот метод.

**Часть 3. Классовый метод `from_dict`.** Реализуйте **второй** альтернативный конструктор — `from_dict(cls, data)`, который принимает **словарь** вида:

```python
{"address": "ул. Ленина, д. 10, кв. 5", "area": 65, "price": 8_500_000}
```

и создаёт объект.

**Часть 4. Счётчик объектов.** Добавьте в класс **атрибут класса** `count = 0`, который увеличивается в конструкторе. Реализуйте **классовый метод** `get_count(cls)`, возвращающий значение счётчика.

Продемонстрируйте:

- создание нескольких квартир (обычным способом, через `from_string`, через `from_dict`);
- вывод `Apartment.get_count()`.

**Часть 5. Наследование.** Создайте наследника `Studio(Apartment)`. Убедитесь, что:

- `Studio.from_string("ул. Пушкина, д. 1, кв. 7;25;4000000")` возвращает объект **`Studio`**, а не `Apartment`;
- `type(studio)` показывает `<class 'Studio'>`;
- `Studio.get_count()` работает **корректно** (наследует или использует свой счётчик — по вашему выбору).

**Часть 6. Переопределение.** В наследнике `Studio` **переопределите** `from_string`, чтобы формат строки был **другим**: `"адрес;площадь;балкон"`. Убедитесь, что:

- `Studio.from_string("ул. Пушкина;25;3")` создаёт `Studio` с полем `balcony = 3`;
- `Apartment.from_string("ул. Ленина;65;8500000")` продолжает работать по-старому.

**Часть 7. Проверка.** Продемонстрируйте работу **всех** методов и наследников.

**Часть 8.** В комментариях объясните:

1. Чем `@classmethod` отличается от **обычного** метода.
2. Чем `@classmethod` отличается от `@staticmethod`.
3. Почему `@classmethod` **правильно работает** с наследованием, а `@staticmethod` — нет.
4. Почему внутри `@classmethod` нужно использовать `cls(...)`, а не `Apartment(...)`.
5. В каких случаях используется `@classmethod` (приведите примеры).
6. Что произойдёт, если в `@classmethod` использовать `Apartment(...)` вместо `cls(...)`.

### Решение

```python
# ─── Части 1–4. Класс Apartment ───
class Apartment:
    count = 0                          # атрибут класса

    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price
        Apartment.count += 1           # увеличиваем счётчик

    # ─── Часть 2. Классовый метод from_string ───
    @classmethod
    def from_string(cls, data):
        # Разбиваем строку и создаём объект через cls.
        address, area, price = data.split(";")
        return cls(address, int(area), int(price))

    # ─── Часть 3. Классовый метод from_dict ───
    @classmethod
    def from_dict(cls, data):
        return cls(data["address"], data["area"], data["price"])

    # ─── Часть 4. Классовый метод get_count ───
    @classmethod
    def get_count(cls):
        return cls.count

    # __repr__ для отладки.
    def __repr__(self):
        return (
            f"{self.__class__.__name__}("
            f"{self.address!r}, {self.area!r}, {self.price!r}"
            f")"
        )


# ─── Часть 5. Наследник Studio ───
class Studio(Apartment):
    # Часть 6. Переопределяем from_string под свой формат.
    @classmethod
    def from_string(cls, data):
        # Формат: "адрес;площадь;балкон"
        address, area, balcony = data.split(";")
        obj = cls(address, int(area), 0)     # цена 0 по умолчанию
        obj.balcony = int(balcony)           # дополнительное поле
        return obj


# ─── Часть 7. Проверка ───
print("=== Создание квартир разными способами ===")

flat1 = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)
flat2 = Apartment.from_string("ул. Садовая, д. 3, кв. 12;80;12000000")
flat3 = Apartment.from_dict({
    "address": "ул. Пушкина, д. 1, кв. 7",
    "area": 50,
    "price": 6_000_000,
})

print(flat1)
print(flat2)
print(flat3)

print()
print("=== Счётчик объектов ===")
print(f"Apartment.get_count() = {Apartment.get_count()}")    # 3

print()
print("=== Наследник Studio ===")
studio = Studio.from_string("ул. Пушкина, д. 1, кв. 7;25;3")
print(f"type(studio) = {type(studio)}")
print(f"studio = {studio}")
print(f"studio.balcony = {studio.balcony}")
print(f"Studio.get_count() = {Studio.get_count()}")           # 4

print()
print("=== Проверка, что родительский метод работает ===")
flat4 = Apartment.from_string("ул. Ленина, д. 10, кв. 5;65;8500000")
print(f"type(flat4) = {type(flat4)}")
print(f"flat4 = {flat4}")
```

**Ожидаемый вывод:**

```
=== Создание квартир разными способами ===
Apartment('ул. Ленина, д. 10, кв. 5', 65, 8500000)
Apartment('ул. Садовая, д. 3, кв. 12', 80, 12000000)
Apartment('ул. Пушкина, д. 1, кв. 7', 50, 6000000)

=== Счётчик объектов ===
Apartment.get_count() = 3

=== Наследник Studio ===
type(studio) = <class '__main__.Studio'>
studio = Studio('ул. Пушкина, д. 1, кв. 7', 25, 0)
studio.balcony = 3
Studio.get_count() = 4

=== Проверка, что родительский метод работает ===
type(flat4) = <class '__main__.Apartment'>
flat4 = Apartment('ул. Ленина, д. 10, кв. 5', 65, 8500000)
```

### Разбор

**1. Чем `@classmethod` отличается от обычного метода.**

- **Обычный метод** — получает `self` (объект). Работает с **данными объекта**: `self.address`, `self.area`. Вызывается через **объект**: `flat.info()`. Привязан **к конкретному объекту**.
- **`@classmethod`** — получает `cls` (класс). Работает с **данными класса**: `cls.count`, `cls(...)`. Вызывается через **класс**: `Apartment.from_string(...)`. Привязан **к классу**.

**Ключевое отличие:** обычный метод **привязан к объекту**, классовый — **к классу**. Обычный может **менять объект**, классовый — **класс**.

**2. Чем `@classmethod` отличается от `@staticmethod`.**

- **`@classmethod`** — получает **`cls`**. Имеет доступ к **классу**: `cls.count`, `cls(...)`, `super()`. Может **создавать** объекты и **менять** атрибуты класса.
- **`@staticmethod`** — **не получает** ни `self`, ни `cls`. Просто **функция** внутри класса. Не имеет доступа **ни к чему**, кроме аргументов.

**Пример:**

```python
@classmethod
def from_string(cls, data):
    return cls(...)                # доступ к классу

@staticmethod
def is_valid_area(area):
    return 5 <= area <= 500        # нет доступа к классу
```

**3. Почему `@classmethod` правильно работает с наследованием, а `@staticmethod` — нет.**

**`@classmethod`** получает **`cls`** — фактический класс, через который вызван метод. Если вызвать `Studio.from_string(...)`, `cls = Studio`. Значит, `cls(...)` — это `Studio(...)`. **Наследник получает правильный тип.**

**`@staticmethod`** не получает **ничего**. Если внутри **жёстко** прописано `Apartment(...)`, то при вызове через `Studio` всё равно создастся `Apartment`. **Наследование ломается.**

**4. Почему внутри `@classmethod` нужно использовать `cls(...)`, а не `Apartment(...)`.**

Если использовать `Apartment(...)`, метод **всегда** возвращает объект **именно** `Apartment`. Наследник `Studio.from_string(...)` тоже вернёт `Apartment` — **неправильно**.

Если использовать `cls(...)`, метод возвращает объект **фактического класса**. Наследник `Studio.from_string(...)` вернёт `Studio` — **правильно**.

**Правило:** в `@classmethod` **всегда** `cls(...)`, никогда — имя класса напрямую.

**5. В каких случаях используется `@classmethod`.**

Основные случаи:

- **Фабричные методы** (альтернативные конструкторы): `from_string`, `from_dict`, `from_json`.
- **Работа с атрибутами класса**: `get_count`, `reset_count`.
- **Работа с реестрами**: `find_by_address`.
- **Создание объектов в наследниках**: `cls(...)`.
- **Абстрактные классовые методы** (в сочетании с `@abstractmethod`).

**6. Что произойдёт, если в `@classmethod` использовать `Apartment(...)` вместо `cls(...)`.**

Наследование **сломается**. При вызове `Studio.from_string(...)`:

- Python подставит `cls = Studio`.
- Внутри метода будет вызван `Apartment(...)` — **не** `cls(...)`.
- Возвращён объект **`Apartment`**, а не `Studio`.
- `type(studio)` покажет `<class 'Apartment'>`.

**Результат:** наследник **теряет** свою «индивидуальность». Все фабричные методы **всегда** возвращают **родительский** класс.

**Правило:** в `@classmethod` **всегда** `cls(...)`.

# Глава 16. Классовые vs экземплярные атрибуты

## 16.1. Проблема: где хранить данные — в классе или в объекте

За несколько глав мы построили класс `Apartment` и научили его многому: красиво выводиться, сравниваться, складываться в комплексы, создаваться из строк. Но есть **один вопрос**, который мы до сих пор **не разбирали** подробно. Где именно **хранятся** данные: в **классе** или в **каждом объекте**? Казалось бы, какая разница? Но разница **есть**, и она **принципиальная**.

Представь наш дом. У каждой квартиры есть **свой** адрес, **своя** площадь, **своя** цена. Эти данные **уникальны** для каждого объекта. Если мы изменим адрес одной квартиры, у других он **не должен** измениться. Такие данные логично хранить **в объекте** — они **принадлежат** конкретной квартире. Это **экземплярные атрибуты**.

А теперь представь, что у **всех** квартир есть что-то **общее**. Например, **название агентства**, через которое они продаются. Или **ставка налога** — одинаковая для всех квартир в этом городе. Или **минимальная площадь** — ниже которой квартиры не продаются. Эти данные **одинаковы** для всех объектов. Их **не нужно** дублировать в каждой квартире — они **принадлежат** классу. Это **классовые атрибуты**.

Но есть ещё **третий** случай — самый **коварный**. Есть данные, которые **выглядят** общими, но на самом деле должны быть **у каждого свои**. Например, **счётчик всех созданных квартир**. Он выглядит как «одно число для всех», но если поместить его в объект — каждая квартира будет считать **только себя**. А нам нужно, чтобы счётчик был **общим**. Это уже **классовый атрибут**, но с **особым** поведением.

### Аналогия

Представь **школьный класс**. У **каждого** ученика есть **своё** имя, **свой** возраст, **своя** оценка по математике. Это **экземплярные атрибуты** — они уникальны для каждого ученика.

Но есть данные, **общие** для всего класса: **номер кабинета** (все учатся в одном кабинете), **классный руководитель** (один на всех), **расписание уроков** (одинаковое для всех). Это **классовые атрибуты** — они принадлежат **классу**, а не отдельному ученику.

И есть **третий** случай: **количество учеников в классе**. Оно **общее** для класса — но **меняется**, когда кто-то приходит или уходит. Это **классовый атрибут**, который **изменяется** при создании/удалении объектов.

### Почему это важно понимать

Если **перепутать** классовые и экземплярные атрибуты — программа будет работать **неправильно**. Причём ошибка может быть **коварной**: код **запустится**, не выдаст ошибку, но данные будут **не те**, что ожидалось. Например:

- Хотели общий счётчик — получили **отдельный** у каждого.
- Хотели общие настройки — при изменении в одном объекте изменились **у всех** (что **неожиданно**).
- Хотели уникальные данные — при изменении в одном объекте изменились **все** (что **катастрофа**).

Все эти ошибки — **из-за непонимания** разницы между классовыми и экземплярными атрибутами. В этой главе мы разберём **всё** подробно.

### Вывод

Данные могут храниться **в классе** (общие для всех) или **в объекте** (уникальные для каждого). Понимание **разницы** критично для правильного кода.

---

## 16.2. Что такое классовые атрибуты

**Классовый атрибут** — это переменная, определённая **внутри класса**, но **вне методов**. Она **принадлежит классу** и **общая** для всех его объектов.

### Синтаксис

```python
class Apartment:
    agency_name = "Дом-Сервис"       # классовый атрибут
    tax_rate = 0.13                  # классовый атрибут
    min_area = 5                     # классовый атрибут

    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price
```

Обрати внимание: атрибуты `agency_name`, `tax_rate` и `min_area` определены **прямо в теле класса**, **до** метода `__init__`. Они **не** используют `self`. Они **принадлежат** классу.

### Как обратиться

К классовому атрибуту можно обратиться **двумя** способами:

**1. Через класс:**

```python
print(Apartment.agency_name)     # Дом-Сервис
print(Apartment.tax_rate)        # 0.13
print(Apartment.min_area)        # 5
```

**2. Через объект:**

```python
flat = Apartment("ул. Ленина", 65, 8_500_000)
print(flat.agency_name)          # Дом-Сервис — тот же атрибут
print(flat.tax_rate)             # 0.13
```

Оба способа возвращают **одно и то же** значение. Это **важная** особенность: классовый атрибут **доступен** и через класс, и через любой объект.

### Что происходит под капотом

Когда Python видит `flat.agency_name`, он **не находит** `agency_name` **в объекте** `flat`. Тогда он **идёт** в **класс** `Apartment` — и **находит** там. То есть классовый атрибут **как бы** «принадлежит» объекту, но **реально** хранится в классе.

### Общая память

Классовый атрибут — **общий** для всех объектов. Если **изменить** его через класс — он **изменится** для **всех** объектов:

```python
Apartment.agency_name = "Новое агентство"
print(Apartment.agency_name)     # Новое агентство
print(flat.agency_name)          # Новое агентство — тот же атрибут
```

Обрати внимание: мы изменили атрибут **через класс**, а он **изменился** и у объекта `flat`. Потому что `flat.agency_name` — это **ссылка** на **классовый** атрибут, а не на **свою копию**.

### Аналогия

Представь **табличку** на дверях класса: «5-А». Она **одна** на весь класс. Если её **изменить** — изменится для **всех**, кто входит. Классовый атрибут — это **такая табличка**. Она **общая** для всех объектов, **одна** на класс.

### Когда использовать

Классовые атрибуты **идеальны**, когда данные:

- **Одинаковы** для всех объектов.
- **Не меняются** от объекта к объекту.
- Логически принадлежат **классу**, а не отдельному объекту.

Примеры:

- **Константы** — число `π`, ставка налога, гравитационная постоянная.
- **Настройки по умолчанию** — значение по умолчанию для новых объектов.
- **Справочная информация** — название компании, версия продукта.
- **Счётчики** — количество созданных объектов.

### Вывод

**Классовый атрибут** — переменная в теле класса, **общая** для всех объектов. Доступен **через класс** и **через объект**. Изменение **влияет на всех**. Используется для **общих** данных.

---

## 16.3. Что такое экземплярные атрибуты

**Экземплярный атрибут** — это переменная, определённая **внутри метода** (обычно в `__init__`) через `self`. Она **принадлежит** конкретному объекту и **уникальна** для него.

### Синтаксис

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address       # экземплярный атрибут
        self.area = area             # экземплярный атрибут
        self.price = price           # экземплярный атрибут
```

Обрати внимание: атрибуты `address`, `area` и `price` определены **через `self`** внутри метода `__init__`. Они **привязаны** к конкретному объекту.

### Как обратиться

К экземплярному атрибуту можно обратиться **только через объект**:

```python
flat = Apartment("ул. Ленина", 65, 8_500_000)

print(flat.address)              # ул. Ленина
print(flat.area)                 # 65
print(flat.price)                # 8500000
```

**Через класс** обратиться **нельзя**:

```python
print(Apartment.address)         # AttributeError!
```

Потому что `address` **не существует** в классе. Он **создаётся** только при **создании** объекта — в методе `__init__`.

### Уникальность

Экземплярный атрибут **уникален** для каждого объекта. У двух разных квартир — **разные** значения:

```python
flat1 = Apartment("ул. Ленина", 65, 8_500_000)
flat2 = Apartment("ул. Садовая", 80, 12_000_000)

print(flat1.address)             # ул. Ленина
print(flat2.address)             # ул. Садовая — другое значение
```

Изменение в одном объекте **не влияет** на другой:

```python
flat1.area = 70
print(flat1.area)                # 70
print(flat2.area)                # 80 — не изменилось
```

### Как хранятся

Экземплярные атрибуты хранятся **внутри объекта**, в специальном словаре `__dict__`:

```python
flat = Apartment("ул. Ленина", 65, 8_500_000)
print(flat.__dict__)
# {'address': 'ул. Ленина', 'area': 65, 'price': 8500000}
```

У каждого объекта — **свой** `__dict__`. Поэтому у каждого объекта — **свои** экземплярные атрибуты.

### Аналогия

Представь **личные вещи** ученика: ручка, тетрадь, дневник. Они **принадлежат** конкретному ученику. У другого ученика — **свои** ручки и тетради. Если один потеряет ручку — у другого она **останется**. Экземплярный атрибут — это **такая личная вещь**. Она **уникальна** для объекта.

### Когда использовать

Экземплярные атрибуты **идеальны**, когда данные:

- **Уникальны** для каждого объекта.
- **Меняются** от объекта к объекту.
- Логически принадлежат **объекту**, а не классу.

Примеры:

- **Идентификаторы** — `id`, `uuid`.
- **Личные данные** — имя, адрес, возраст.
- **Состояние** — баланс, счётчик, статус.
- **Данные из конструктора** — то, что передаётся при создании.

### Вывод

**Экземплярный атрибут** — переменная через `self` внутри метода, **уникальная** для каждого объекта. Доступна **только через объект**. Изменение **не влияет** на другие объекты. Используется для **уникальных** данных.

---

## 16.4. Главное отличие: общее vs уникальное

Теперь у нас есть **два** вида атрибутов. Разница **принципиальна**, и её нужно **чётко** понимать.

| Характеристика | Классовый атрибут | Экземплярный атрибут |
|----------------|-------------------|----------------------|
| Где определён | **В теле класса** | **В методе** через `self` |
| Кому принадлежит | **Классу** | **Объекту** |
| Общий или уникальный | **Общий** для всех | **Уникальный** для каждого |
| Как обратиться | **Через класс** и **через объект** | **Только через объект** |
| Как изменить | Через класс (для всех) | Через объект (для одного) |
| Где хранится | В `__dict__` класса | В `__dict__` объекта |
| Когда исчезает | Когда удаляется класс | Когда удаляется объект |

### Пример: оба вида в одном классе

```python
class Apartment:
    agency_name = "Дом-Сервис"        # классовый атрибут

    def __init__(self, address, area):
        self.address = address         # экземплярный атрибут
        self.area = area               # экземплярный атрибут
```

Здесь:

- `agency_name` — **классовый**. Один для всех квартир.
- `address`, `area` — **экземплярные**. У каждой квартиры свои.

### Как это выглядит в памяти

```
        Класс Apartment
        ├── agency_name = "Дом-Сервис"   ← классовый атрибут
        ├── __init__ (метод)
        └── ...

        Объект flat1                  Объект flat2
        ├── address = "ул. Ленина"    ├── address = "ул. Садовая"
        └── area = 65                 └── area = 80
        (ссылается на agency_name)    (ссылается на agency_name)
```

Обрати внимание: **оба объекта** ссылаются на **один** `agency_name`. А вот `address` и `area` — **свои** у каждого.

### Что происходит при обращении

Когда Python видит `flat1.address`:

1. Ищет `address` в **объекте** `flat1` — **находит**. Возвращает.

Когда Python видит `flat1.agency_name`:

1. Ищет `agency_name` в **объекте** `flat1` — **не находит**.
2. Идёт в **класс** `Apartment` — **находит**. Возвращает.

Это **порядок поиска** атрибутов — **ключевой** момент. О нём подробно — в разделе 16.5.

### Практическое правило

**Задай себе вопрос:** «Это **общее** для всех объектов или **уникальное** для каждого?»

- Если **общее** — **классовый** атрибут.
- Если **уникальное** — **экземплярный** атрибут.

### Аналогия

Представь **здание**. У здания есть **адрес** — он **общий** для всех квартир. У каждой квартиры есть **номер** — он **уникален** для каждой. Адрес — **классовый** атрибут. Номер — **экземплярный**.

### Вывод

**Классовый** атрибут — **общий** для всех объектов. **Экземплярный** — **уникальный** для каждого. Разница **принципиальна**. Понимание её — **основа** правильного ООП.

---

## 16.5. Как Python ищет атрибуты

Это **ключевой** раздел главы. Здесь мы разберём, **что происходит**, когда Python видит обращение к атрибуту. Понимание этого — **критично** для правильного кода.

### Порядок поиска

Когда Python видит `obj.attribute`, он ищет в **таком порядке**:

1. **В объекте** `obj` — в его `__dict__`.
2. **В классе** `obj` — в `__dict__` класса.
3. **У родителей класса** — по MRO (см. главу 19).
4. **Если не найдено нигде** — `AttributeError`.

### Схема

```
        obj.attribute
                ↓
        1. Ищем в obj.__dict__
                ↓ (не нашли)
        2. Ищем в type(obj).__dict__
                ↓ (не нашли)
        3. Ищем у родителей (по MRO)
                ↓ (не нашли)
        AttributeError
```

### Пример

```python
class Apartment:
    agency_name = "Дом-Сервис"        # классовый атрибут

    def __init__(self, address):
        self.address = address         # экземплярный атрибут


flat = Apartment("ул. Ленина")
```

Проверим:

**Обращение к `flat.address`:**

1. Ищем в `flat.__dict__` — **находим** `'address': 'ул. Ленина'`. Возвращаем.

**Обращение к `flat.agency_name`:**

1. Ищем в `flat.__dict__` — **не находим**.
2. Идём в `Apartment.__dict__` — **находим** `'agency_name': 'Дом-Сервис'`. Возвращаем.

**Обращение к `flat.unknown`:**

1. Ищем в `flat.__dict__` — **не находим**.
2. Идём в `Apartment.__dict__` — **не находим**.
3. Идём к `object` — **не находим**.
4. **AttributeError**.

### Что это даёт

Из этого порядка **следует** очень **важное** правило: **экземплярный атрибут перекрывает классовый**.

Если у объекта **есть** свой атрибут с **тем же именем**, что и классовый — Python **найдёт** его **в объекте** и **не пойдёт** в класс. Это **тень** классового атрибута.

### Пример перекрытия

```python
class Apartment:
    agency_name = "Дом-Сервис"        # классовый атрибут

    def __init__(self, address):
        self.address = address


flat1 = Apartment("ул. Ленина")
flat2 = Apartment("ул. Садовая")

# Смотрим оба.
print(flat1.agency_name)              # Дом-Сервис (из класса)
print(flat2.agency_name)              # Дом-Сервис (из класса)

# Создаём экземплярный атрибут.
flat1.agency_name = "Своё агентство"  # ← создаём экземплярный атрибут!
print(flat1.agency_name)              # Своё агентство (из объекта)
print(flat2.agency_name)              # Дом-Сервис (из класса — не изменилось!)
print(Apartment.agency_name)          # Дом-Сервис (класс не изменился)
```

Обрати внимание: `flat1.agency_name = "Своё агентство"` **не изменило** классовый атрибут. Оно **создало** новый **экземплярный** атрибут у `flat1`. Теперь `flat1` **перекрывает** классовый атрибут **своим**.

### Схема перекрытия

```
        Класс Apartment
        ├── agency_name = "Дом-Сервис"   ← классовый

        flat1                                  flat2
        ├── address = "ул. Ленина"             ├── address = "ул. Садовая"
        └── agency_name = "Своё агентство"     └── (нет agency_name)
            ↑ перекрывает классовый                ↑ берёт из класса
```

### Почему это важно

Многие новички **думают**, что `flat1.agency_name = "..."` изменяет **классовый** атрибут. Но это **не так**! Оно создаёт **экземплярный** атрибут, который **перекрывает** классовый **только для этого объекта**.

**Чтобы изменить** классовый атрибут — нужно писать `Apartment.agency_name = "..."`.

### Аналогия

Представь **табличку** на дверях класса: «5-А». Ты **не можешь** её изменить, повесив **свою** табличку у себя дома. Ты можешь **запомнить**, что «мой класс — 5-А», но **табличка** на дверях остаётся прежней.

Когда ты **вешаешь свою** табличку — это **твоя** табличка. Она **перекрывает** общую **для тебя**, но **не** для других. **Изменить общую** — значит **заменить** табличку **на дверях**.

### Вывод

Python ищет атрибуты **сначала в объекте**, **потом в классе**. Экземплярный атрибут **перекрывает** классовый. Чтобы изменить классовый — обращайтесь **через класс**.

---

## 16.6. Когда использовать классовые атрибуты

Разберём **пять** типичных случаев использования классовых атрибутов. Каждый — с примером и **подробным** объяснением.

### 16.6.1. Константы

**Самый частый** случай — **константы**, общие для всех объектов. Например, **ставка налога**, **коэффициент пересчёта**, **фиксированное значение**.

```python
class Apartment:
    TAX_RATE = 0.13                   # ставка налога — общая для всех
    SQM_TO_SQFT = 10.764              # коэффициент пересчёта

    def __init__(self, address, price):
        self.address = address
        self.price = price

    def price_with_tax(self):
        # Используем классовую константу.
        return self.price * (1 + Apartment.TAX_RATE)
```

Здесь `TAX_RATE` и `SQM_TO_SQFT` — **константы**. Они **одинаковы** для всех квартир. Нет смысла дублировать их в каждой.

**Почему не экземплярные?** Если сделать их экземплярными — придётся **передавать** их в конструктор **каждый раз**. И если ставка **изменится** — придётся менять **у каждого** объекта.

**Почему не глобальные переменные?** Потому что **логически** они **принадлежат** классу. Глобальные переменные **засоряют** пространство имён и **не связаны** с классом.

### 16.6.2. Настройки по умолчанию

Классовый атрибут может быть **значением по умолчанию** для новых объектов:

```python
class Apartment:
    DEFAULT_AREA = 30                 # площадь по умолчанию
    DEFAULT_PRICE = 5_000_000         # цена по умолчанию

    def __init__(self, address, area=None, price=None):
        self.address = address
        # Если не передано — берём значение по умолчанию.
        self.area = area if area is not None else Apartment.DEFAULT_AREA
        self.price = price if price is not None else Apartment.DEFAULT_PRICE
```

Здесь `DEFAULT_AREA` и `DEFAULT_PRICE` — **значения по умолчанию**. Если пользователь **не указал** площадь или цену — берём **общие** значения.

**Почему это удобно:**

- **Одно место** для изменения значения по умолчанию.
- **Понятно**, какие значения используются.
- Легко **переопределить** в наследнике.

### 16.6.3. Справочная информация

Классовые атрибуты удобны для **справочной информации** о классе:

```python
class Apartment:
    class_name = "Квартира"           # название класса
    version = "2.0"                   # версия класса
    description = "Жилое помещение"   # описание

    def info(self):
        return f"{Apartment.class_name} (v{Apartment.version})"
```

Здесь `class_name`, `version`, `description` — **справочная** информация. Она **одинакова** для всех объектов.

### 16.6.4. Счётчики объектов

Классовый атрибут **идеален** для **счётчика** — сколько всего объектов **создано**:

```python
class Apartment:
    count = 0                         # счётчик объектов

    def __init__(self, address):
        self.address = address
        # Увеличиваем счётчик класса.
        Apartment.count += 1
```

Здесь `count` — **общий** счётчик. Он **увеличивается** при создании **каждого** объекта. И **не сбрасывается**, пока класс **существует**.

**Почему не экземплярный?** Если сделать `self.count = 0` в `__init__`, то у **каждого** объекта будет **свой** `count`, и он будет **всегда** равен `1` (или `0`). Нам же нужен **общий** счётчик.

**Важный нюанс:** увеличивать счётчик нужно **через класс** (`Apartment.count += 1`), а не через `self` (`self.count += 1`). Об этом подробно — в разделе 16.8.

### 16.6.5. Реестры объектов

Классовый атрибут может быть **списком** всех созданных объектов:

```python
class Apartment:
    all_apartments = []               # реестр всех объектов

    def __init__(self, address):
        self.address = address
        Apartment.all_apartments.append(self)

    @classmethod
    def find_by_address(cls, address):
        for flat in cls.all_apartments:
            if flat.address == address:
                return flat
        return None
```

Здесь `all_apartments` — **список** всех объектов. Он **общий** для класса. Через него можно **искать** нужные объекты.

**Важный нюанс:** `all_apartments` — **изменяемый** объект (список). Это **коварно** — об этом подробно в разделе 16.8.

### Вывод

Классовые атрибуты **идеальны** для: **констант**, **настроек по умолчанию**, **справочной информации**, **счётчиков** и **реестров**. Везде, где данные **общие** для всех объектов.

---

## 16.7. Когда использовать экземплярные атрибуты

Разберём **четыре** типичных случая использования экземплярных атрибутов. Каждый — с примером и **подробным** объяснением.

### 16.7.1. Идентификаторы

**Уникальные** идентификаторы — всегда **экземплярные**:

```python
class Apartment:
    def __init__(self, address):
        self.address = address
        self.id = id(self)             # уникальный идентификатор
```

Здесь `id` — **уникален** для каждого объекта. Он **не может** быть классовым.

### 16.7.2. Данные из конструктора

Всё, что **передаётся** в конструктор — обычно **экземплярное**:

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address         # из конструктора
        self.area = area               # из конструктора
        self.price = price             # из конструктора
```

Здесь `address`, `area`, `price` — **уникальны** для каждого объекта. Они **передаются** при создании.

### 16.7.3. Состояние объекта

**Состояние**, которое **меняется** в процессе работы — **экземплярное**:

```python
class Apartment:
    def __init__(self, address):
        self.address = address
        self.is_sold = False           # состояние: продана или нет
        self.views = 0                 # состояние: сколько просмотров

    def view(self):
        self.views += 1                # меняем состояние ОБЪЕКТА

    def sell(self):
        self.is_sold = True            # меняем состояние ОБЪЕКТА
```

Здесь `is_sold` и `views` — **состояние** конкретной квартиры. Оно **меняется** независимо для каждого объекта.

### 16.7.4. Результаты вычислений

**Промежуточные** результаты вычислений — обычно **экземплярные**:

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price
        # Вычисляем цену за м² при создании.
        self.price_per_sqm = price / area
```

Здесь `price_per_sqm` — **результат** вычисления. Он **уникален** для каждой квартиры.

**Но есть нюанс:** если данные **меняются**, лучше использовать **вычисляемое свойство** (см. главу 14). Тогда значение **всегда** актуально.

### Аналогия

Представь **личный дневник**. В нём **твои** записи, **твои** оценки, **твои** заметки. Они **уникальны** для тебя. Если ты получишь «5» — это **только твоя** оценка, не **всех** учеников. Экземплярный атрибут — это **личная запись** в **твоём** дневнике.

### Вывод

Экземплярные атрибуты **идеальны** для: **идентификаторов**, **данных из конструктора**, **состояния объекта**, **результатов вычислений**. Везде, где данные **уникальны** для каждого объекта.

## 16.8. Коварная ловушка: изменяемые классовые атрибуты

Это **самый важный** раздел главы. Здесь мы разберём **самую коварную** ошибку, связанную с классовыми атрибутами. Её **допускают** даже **опытные** программисты.

### Проблема

Представь, что мы хотим вести **список** всех квартир. Логично сделать его **классовым** атрибутом:

```python
class Apartment:
    all_apartments = []               # ← изменяемый классовый атрибут

    def __init__(self, address):
        self.address = address
        self.all_apartments.append(self)   # ← добавляем объект в список
```

Казалось бы, всё **правильно**. `all_apartments` — **общий** список. Каждый объект **добавляется** в него при создании.

Проверим:

```python
flat1 = Apartment("ул. Ленина")
flat2 = Apartment("ул. Садовая")

print(len(Apartment.all_apartments))     # 2 — правильно
print(len(flat1.all_apartments))         # 2 — тоже 2
print(len(flat2.all_apartments))         # 2 — тоже 2
```

Всё работает. Список **общий** для всех. Пока **всё хорошо**.

### Где ошибка

А теперь **изменим** список **через объект**:

```python
flat1.all_apartments.append("что-то ещё")
print(len(flat1.all_apartments))         # 3
print(len(flat2.all_apartments))         # 3 — тоже 3!
print(len(Apartment.all_apartments))     # 3 — и класс тоже видит!
```

**Что произошло?** Мы добавили элемент **через `flat1`**, а он **появился** у **всех**. Потому что `flat1.all_apartments` — это **ссылка** на **тот же список**, что и у класса.

### Ещё хуже

А теперь **присвоим** новый список **через объект**:

```python
flat1.all_apartments = []                # ← создаём ЭКЗЕМПЛЯРНЫЙ атрибут!
flat1.all_apartments.append("только моё")

print(len(flat1.all_apartments))         # 1 — только у flat1
print(len(flat2.all_apartments))         # 3 — у flat2 старое
print(len(Apartment.all_apartments))     # 3 — у класса старое
```

**Что произошло?** `flat1.all_apartments = []` **создало** **экземплярный** атрибут у `flat1`. Теперь `flat1` **перекрывает** классовый атрибут **своим**. `flat2` и класс **не изменились**.

Это **очень** запутанное поведение. Одна строка `flat1.all_apartments.append(...)` **не** создаёт экземплярный атрибут (изменяет классовый). А строка `flat1.all_apartments = []` — **создаёт** (перекрывает классовый).

### Почему так происходит

**Ключ** — в **разнице** между **изменением** и **присваиванием**.

- **`flat1.all_apartments.append(x)`** — это **изменение** объекта **по ссылке**. Python ищет `all_apartments` **в объекте** — **не находит**. Идёт **в класс** — **находит**. **Изменяет** его. **Классовый** атрибут **изменён**.
- **`flat1.all_apartments = [...]`** — это **присваивание**. Python **создаёт** новый атрибут **в объекте** `flat1`. **Классовый** атрибут **не тронут**. Но теперь `flat1.all_apartments` **ссылается** на **другой** список.

### Схема

```
        flat1.all_apartments.append(x)
        ─────────────────────────────
        Ищем all_apartments в flat1 → не нашли
        Идём в класс → нашли
        Изменяем список класса
        ↓
        Изменение ВИДНО ВСЕМ

        flat1.all_apartments = [x]
        ─────────────────────────
        Создаём НОВЫЙ атрибут в flat1
        Теперь flat1.all_apartments → новый список
        Классовый атрибут НЕ ТРОНУТ
        ↓
        Изменение видно ТОЛЬКО flat1
```

### Как избежать проблемы

**Правило:** **никогда** не используйте **изменяемые** классовые атрибуты **напрямую** через объект. Есть **два** способа избежать проблемы.

**Способ 1: работать через класс.**

```python
class Apartment:
    all_apartments = []

    def __init__(self, address):
        self.address = address
        # Явно через класс — понятно, что меняем классовый атрибут.
        Apartment.all_apartments.append(self)
```

Теперь **очевидно**, что мы изменяем **классовый** атрибут. И **нельзя** случайно **присвоить** новый список через объект.

**Способ 2: использовать `__init__` с явным `cls`.**

```python
class Apartment:
    def __init__(self, address):
        self.address = address
        type(self).all_apartments.append(self)
```

Здесь `type(self)` — это **класс** объекта. Тоже работает.

**Способ 3: не использовать изменяемые атрибуты.**

Если **можно** обойтись — не используйте **изменяемые** классовые атрибуты. Например, **счётчик** — это **число** (неизменяемый тип). С ним **нет** такой проблемы.

### Почему это ловушка

Эта ошибка **коварна** потому что:

- **Код запускается** без ошибок.
- **Данные выглядят** правильными **сначала**.
- **Проблема** проявляется **позже** — когда кто-то **изменит** атрибут **через объект**.
- **Найти ошибку** сложно — она **не в том месте**, где проявляется.

### Аналогия

Представь **общую доску** в классе. Если ты **пишешь** на ней мелом — это **видят все**. Это как `append`. Если ты **приносишь свою** доску и **вешаешь** её **рядом** — теперь у тебя **своя** доска, а **общая** осталась **прежней**. Это как **присваивание**. Ты **думаешь**, что пишешь на **общей** доске, а на самом деле — на **своей**.

### Вывод

**Изменяемые** классовые атрибуты **коварны**. Изменение через объект **влияет на всех** (если через `append`). Присваивание через объект **создаёт экземплярный** атрибут (перекрывает классовый). **Правило:** работайте с изменяемыми классовыми атрибутами **только через класс** — `Apartment.attr`.

## 16.9. Классовые атрибуты и наследование

Классовые атрибуты **наследуются**. Это **важно** понимать, потому что **поведение** при наследовании имеет **нюансы**.

### Наследование без переопределения

Если наследник **не переопределяет** классовый атрибут — он **использует** родительский:

```python
class Apartment:
    agency_name = "Дом-Сервис"

class Studio(Apartment):
    pass

print(Studio.agency_name)             # Дом-Сервис — унаследовано
```

### Переопределение в наследнике

Наследник **может** переопределить классовый атрибут:

```python
class Studio(Apartment):
    agency_name = "Студия-Сервис"     # ← свой атрибут

print(Apartment.agency_name)          # Дом-Сервис
print(Studio.agency_name)             # Студия-Сервис
```

Теперь у `Apartment` **свой** `agency_name`, у `Studio` — **свой**. Они **независимы**.

### Важный нюанс: изменение родительского атрибута

Если **изменить** атрибут **в родителе** — это **не повлияет** на наследника, **если** он **переопределил** его:

```python
Apartment.agency_name = "Новое имя"
print(Apartment.agency_name)          # Новое имя
print(Studio.agency_name)             # Студия-Сервис — не изменилось!
```

Потому что у `Studio` — **свой** `agency_name`. Он **не связан** с родительским.

### Если наследник НЕ переопределил

Если наследник **не переопределил** — он **ссылается** на **родительский**:

```python
class Studio(Apartment):
    pass                              # не переопределил

Apartment.agency_name = "Новое имя"
print(Studio.agency_name)             # Новое имя — изменилось!
```

Потому что `Studio` **использует** `Apartment.agency_name`. Изменили родителя — изменилось и у наследника.

### Схема

```
        Apartment                          Studio (не переопределил)
        └── agency_name = "Дом-Сервис"     └── (нет своего)
                ↑                                   ↓
                └──── Studio ссылается на Apartment ─┘

        Apartment                          Studio (переопределил)
        └── agency_name = "Дом-Сервис"     └── agency_name = "Студия-Сервис"
        (независимый)                      (независимый)
```

### Наследование и счётчик

Счётчик — **особый** случай. Если наследник **не переопределил** счётчик — он **использует** родительский:

```python
class Apartment:
    count = 0

    def __init__(self):
        Apartment.count += 1

class Studio(Apartment):
    pass

Apartment()
Studio()

print(Apartment.count)                # 2 — оба считаются
print(Studio.count)                   # 2 — тоже 2 (унаследовано)
```

Если наследник **переопределит** — получит **свой**:

```python
class Studio(Apartment):
    count = 0                         # ← свой счётчик

    def __init__(self):
        Studio.count += 1

Apartment()
Studio()

print(Apartment.count)                # 1
print(Studio.count)                   # 1 — свой
```

### Вывод

Классовые атрибуты **наследуются**. Наследник **может** переопределить их. Если **не переопределил** — использует **родительский**. Изменение родителя **влияет** на наследника **только если** он **не переопределил** атрибут.

## 16.10. Классовые атрибуты в реальных проектах

Где классовые атрибуты **активно** используются. Разберём **пять** примеров.

### 16.10.1. Django-модели

```python
class Apartment(models.Model):
    class Meta:
        db_table = "apartments"        # имя таблицы — классовый атрибут
        ordering = ["-price"]          # сортировка по умолчанию
```

Здесь `Meta` — **вложенный класс** с **классовыми** атрибутами. Django **читает** их при настройке модели.

### 16.10.2. Enum (перечисления)

```python
from enum import Enum

class Status(Enum):
    AVAILABLE = "доступна"             # классовые константы
    SOLD = "продана"
    RENTED = "сдана"
```

Все значения — **классовые** атрибуты. Они **общие** для всего перечисления.

### 16.10.3. Счётчики в играх

```python
class Player:
    total_players = 0                  # общий счётчик игроков

    def __init__(self, name):
        self.name = name
        Player.total_players += 1
```

В играх часто нужно знать, сколько **всего** игроков **создано**. Это **классовый** атрибут.

### 16.10.4. Конфигурации

```python
class Config:
    DEBUG = False                      # классовые настройки
    VERSION = "1.0.0"
    MAX_CONNECTIONS = 100
```

Класс `Config` хранит **настройки** приложения. Они **общие** для всей программы.

### 16.10.5. Наш дом

```python
class Room:
    min_area = 5                       # минимальная площадь — общая
    max_area = 500                     # максимальная площадь — общая

    def __init__(self, name, area):
        if not (Room.min_area <= area <= Room.max_area):
            raise ValueError("Недопустимая площадь")
        self.name = name
        self.area = area
```

Здесь `min_area` и `max_area` — **классовые** ограничения. Они **одинаковы** для всех комнат.

### Что общего у всех примеров

Во **всех** примерах данные:

- **Одинаковы** для всех объектов.
- Логически **принадлежат** классу.
- **Не меняются** от объекта к объекту.

### Вывод

Классовые атрибуты используются **везде**, где есть **общие** данные: настройки, константы, счётчики, реестры. Это **стандартный** инструмент.

## 16.11. Нюансы и тонкости

Разберём **десять** нюансов, которые часто упускают. Каждый — **важная** деталь.

### 16.11.1. Порядок поиска атрибутов

Python ищет атрибуты **сначала в объекте**, **потом в классе**. Это **ключевое** правило. Экземплярный атрибут **перекрывает** классовый.

### 16.11.2. `__dict__` объекта и класса

Экземплярные атрибуты — в `obj.__dict__`. Классовые — в `Class.__dict__`. Это **разные** словари.

```python
flat = Apartment("ул. Ленина", 65)
print(flat.__dict__)                   # {'address': ..., 'area': 65}
print(Apartment.__dict__)              # {'agency_name': ..., ...}
```

### 16.11.3. Атрибут класса доступен через `type(obj)`

```python
print(type(flat).agency_name)          # Дом-Сервис
```

Это **эквивалент** `Apartment.agency_name`, но работает **универсально** (для любого класса).

### 16.11.4. Изменение через `type(self)`

Внутри методов **лучше** писать `type(self).attr`, а не `ClassName.attr`. Это работает **правильно** с наследниками:

```python
def __init__(self):
    type(self).count += 1             # ← правильно, работает с наследниками
    # Apartment.count += 1            # ← жёстко, не работает с наследниками
```

### 16.11.5. Свойства vs классовые атрибуты

`@property` (см. главу 14) — **свойства объекта**. Классовые атрибуты — **данные класса**. Это **разные** вещи.

Если нужно **свойство класса** — используйте **метакласс** (см. главу 23).

### 16.11.6. `__slots__` и классовые атрибуты

Если в классе определён `__slots__` (см. главу 21), экземплярные атрибуты **ограничены**. Классовые атрибуты **не затрагиваются** `__slots__`.

### 16.11.7. Наследование и `type(self)`

Если в родителе написать `type(self).attr`, наследник получит **свой** атрибут (если переопределил). Это **правильное** поведение.

### 16.11.8. Классовые атрибуты в `@classmethod`

`@classmethod` (см. главу 15) получает `cls` — ссылку на класс. Через `cls.attr` можно **читать** и **менять** классовые атрибуты.

### 16.11.9. Классовые атрибуты и `__init_subclass__`

При наследовании вызывается `__init_subclass__` — специальный метод. Он позволяет **настроить** наследника. Можно использовать для **автоматической** регистрации наследников в **реестре**.

### 16.11.10. Классовые атрибуты и аннотации типов

```python
class Apartment:
    count: int = 0                     # аннотация + значение
    agency_name: str = "Дом-Сервис"
```

Аннотации **не влияют** на выполнение, но **документируют** типы. Полезно для **статической** проверки (mypy).

### Вывод

Классовые атрибуты имеют **много** нюансов. Основное правило — **порядок поиска**: сначала объект, потом класс. Всё остальное — **следствия** этого правила.

---

## 16.12. Типичные ошибки

Разберём **десять** самых частых ошибок. Каждая — с примером и **подробным** объяснением.

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Изменяют атрибут через объект | Создаётся экземплярный | Изменять через класс |
| Думают, что `obj.attr = x` меняет класс | Создаётся экземплярный | Использовать `Class.attr = x` |
| Изменяемый классовый атрибут | Общий для всех — опасно | Использовать `Class.attr.append()` |
| Забыли `type(self)` в наследниках | Жёстко `ClassName` | `type(self).attr` |
| Путают `self.count` и `Class.count` | Разные атрибуты | Явно через класс |
| Изменяют `self.count += 1` вместо `Class.count += 1` | Создаётся экземплярный счётчик | `Class.count += 1` |
| Наследник не переопределяет счётчик | Использует родительский | Переопределить, если нужен свой |
| Классовый атрибут — изменяемый список | Общий для всех | Работать через класс |
| Думают, что `Class.attr = x` меняет у всех наследников | Только у этого класса | Переопределить у наследников |
| Используют классовый атрибут для уникальных данных | Одинаковый у всех | Использовать экземплярный |

### Подробный разбор

**Ошибка 1.** Изменение через объект — **создаёт** экземплярный атрибут, **не** меняет классовый. **Правило:** изменять через **класс**.

**Ошибка 2.** `obj.attr = x` — **не** меняет классовый атрибут. Создаёт **экземплярный**. **Правило:** для изменения класса — `Class.attr = x`.

**Ошибка 3.** Изменяемый классовый атрибут **опасен**. Если **изменить** через объект — **изменится для всех**. Если **присвоить** — создастся экземплярный. **Правило:** работайте **через класс**.

**Ошибка 4.** В наследниках **лучше** использовать `type(self)`, а не `ClassName`. Это работает **правильно** с наследованием.

**Ошибка 5.** `self.count` и `Class.count` — **разные** вещи. Первый — экземплярный, второй — классовый. Будьте **внимательны**.

**Ошибка 6.** `self.count += 1` — **создаст** экземплярный атрибут, **не** изменит классовый. Правильно — `Class.count += 1`.

**Ошибка 7.** Наследник **без переопределения** счётчика — использует **родительский**. Если нужен **свой** — переопределите.

**Ошибка 8.** Изменяемый список как классовый атрибут — **общий** для всех. Если **изменить** через объект — изменится у **всех**. **Правило:** работайте **через класс**.

**Ошибка 9.** `Class.attr = x` меняет **только у этого класса**. Наследники, **не переопределившие** атрибут, **увидят** изменение. Переопределившие — **нет**.

**Ошибка 10.** Классовый атрибут **не подходит** для уникальных данных. Например, **id** — должен быть **экземплярным**.

### Как избежать

1. **Понимайте разницу** между классовым и экземплярным.
2. **Изменяйте** классовые атрибуты **через класс**.
3. **Используйте** `type(self)` в наследниках.
4. **Проверяйте** изменяемые атрибуты **особенно** тщательно.
5. **Пишите тесты** — они поймают ошибки.

### Вывод

Ошибки с классовыми атрибутами **коварны** — код **запускается**, но данные **неверны**. Понимание **порядка поиска** и **правил изменения** поможет **избежать** проблем.

---

## 16.13. Что важно запомнить

- **Классовый атрибут** — определён в теле класса, **общий** для всех объектов.
- **Экземплярный атрибут** — определён через `self`, **уникален** для каждого.
- **Порядок поиска:** сначала **объект**, потом **класс**.
- **Экземплярный** атрибут **перекрывает** классовый.
- **Изменение через объект** (`obj.attr = x`) создаёт **экземплярный** атрибут.
- **Изменение через класс** (`Class.attr = x`) меняет **классовый**.
- **Изменяемые** классовые атрибуты **коварны** — работайте через класс.
- **Наследование:** атрибуты **наследуются**, могут быть **переопределены**.
- **`type(self)`** — правильный способ в наследниках.
- **`@classmethod`** (глава 15) удобен для работы с классовыми атрибутами.
- **Правило выбора:** общее → классовый, уникальное → экземплярный.
- **Счётчик объектов** — классический пример классового атрибута.

## Задача 16 (классовые vs экземплярные атрибуты)

В агентстве недвижимости нужно вести **учёт** квартир и общих **настроек**. Данные бывают **разными** — одни **общие** для всех, другие **уникальные** для каждой квартиры.

**Часть 1. Класс `Apartment`.** Создайте класс `Apartment` со следующими атрибутами:

- **Классовые:** `agency_name = "Дом-Сервис"`, `tax_rate = 0.13`, `count = 0`.
- **Экземплярные:** `address`, `area`, `price` — устанавливаются в конструкторе.

При создании объекта **увеличивайте** классовый счётчик `count`.

**Часть 2. Проверка.** Продемонстрируйте:

- `Apartment.agency_name` — работает через класс;
- `flat.agency_name` — работает через объект;
- создание трёх квартир;
- `Apartment.count` — счётчик равен 3;
- `flat1.count` — тоже 3 (унаследовано от класса).

**Часть 3. Экземплярный атрибут.** Продемонстрируйте, что **экземплярные** атрибуты **уникальны**:

- создайте две квартиры с разными адресами;
- измените `flat1.address`;
- проверьте, что `flat2.address` **не изменился**.

**Часть 4. Перекрытие.** Продемонстрируйте **перекрытие** классового атрибута:

- `flat1.agency_name = "Своё агентство"` — создайте **экземплярный** атрибут;
- проверьте, что `flat1.agency_name` — «Своё агентство»;
- проверьте, что `flat2.agency_name` — «Дом-Сервис» (**не изменилось**);
- проверьте, что `Apartment.agency_name` — «Дом-Сервис» (**не изменилось**).

**Часть 5. Счётчик через класс.** Продемонстрируйте **правильный** способ увеличения счётчика:

- в конструкторе используйте `type(self).count += 1` или `Apartment.count += 1`;
- проверьте, что **неправильный** способ (`self.count += 1`) создаёт **экземплярный** счётчик.

**Часть 6. Классовый метод.** Реализуйте **классовый метод** `get_count(cls)`, возвращающий значение счётчика. Продемонстрируйте, что `Apartment.get_count()` и `Studio.get_count()` возвращают **правильные** значения.

**Часть 7. Наследование.** Создайте наследника `Studio(Apartment)`:

- **не переопределяйте** `count`;
- создайте несколько квартир и студий;
- проверьте, что `Apartment.get_count()` и `Studio.get_count()` возвращают **одинаковые** значения (общий счётчик);
- **переопределите** `count` в `Studio` — теперь у студий **свой** счётчик;
- проверьте, что `Apartment.get_count()` и `Studio.get_count()` возвращают **разные** значения.

**Часть 8. Коварная ловушка.** Продемонстрируйте **коварство** изменяемых классовых атрибутов:

- добавьте **классовый** атрибут `all_apartments = []`;
- при создании объекта **добавляйте** его в список через `type(self).all_apartments.append(self)`;
- создайте несколько объектов;
- проверьте, что `len(Apartment.all_apartments)` **растёт**;
- **НЕПРАВИЛЬНО:** попробуйте присвоить `flat1.all_apartments = []` — убедитесь, что это **создаёт экземплярный** атрибут, и классовый **не меняется**.

**Часть 9. Проверка `__dict__`.** Продемонстрируйте разницу между `__dict__` объекта и класса:

- создайте объект;
- выведите `flat.__dict__` — там **только экземплярные** атрибуты;
- выведите `Apartment.__dict__` — там **классовые** атрибуты и методы.

**Часть 10.** В комментариях объясните:

1. Чем классовый атрибут отличается от экземплярного.
2. Как Python ищет атрибуты при обращении `obj.attr`.
3. Почему `flat1.agency_name = "X"` **не меняет** классовый атрибут.
4. Почему изменяемые классовые атрибуты **коварны**.
5. Что произойдёт, если в конструкторе написать `self.count += 1` вместо `Apartment.count += 1`.
6. Как классовые атрибуты **наследуются** и **переопределяются**.

### Решение

```python
# ─── Часть 1. Класс Apartment ───
class Apartment:
    # Классовые атрибуты — общие для всех.
    agency_name = "Дом-Сервис"
    tax_rate = 0.13
    count = 0
    all_apartments = []                   # реестр всех объектов

    def __init__(self, address, area, price):
        # Экземплярные атрибуты — уникальные для каждого.
        self.address = address
        self.area = area
        self.price = price

        # Часть 5. Правильное увеличение счётчика — через класс.
        type(self).count += 1

        # Часть 8. Добавляем в реестр через класс.
        type(self).all_apartments.append(self)

    def __repr__(self):
        return f"{self.__class__.__name__}({self.address!r}, {self.area!r}, {self.price!r})"

    # ─── Часть 6. Классовый метод get_count ───
    @classmethod
    def get_count(cls):
        return cls.count


# ─── Часть 7. Наследник Studio ───
class Studio(Apartment):
    pass                                   # сначала без переопределения


class StudioWithOwnCount(Apartment):
    count = 0                              # ← свой счётчик

    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price
        # Увеличиваем СВОЙ счётчик.
        type(self).count += 1
        type(self).all_apartments.append(self)


# ─── Часть 2. Проверка классовых и экземплярных атрибутов ───
print("=== Классовые атрибуты ===")
print(f"Apartment.agency_name = {Apartment.agency_name}")
print(f"Apartment.tax_rate = {Apartment.tax_rate}")

flat1 = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)
flat2 = Apartment("ул. Садовая, д. 3, кв. 12", 80, 12_000_000)
flat3 = Apartment("ул. Пушкина, д. 1, кв. 7", 50, 6_000_000)

print(f"flat1.agency_name = {flat1.agency_name}")    # тот же
print(f"Apartment.count = {Apartment.count}")        # 3
print(f"flat1.count = {flat1.count}")                # 3

# ─── Часть 3. Экземплярные атрибуты уникальны ───
print()
print("=== Экземплярные атрибуты ===")
print(f"flat1.address = {flat1.address}")
print(f"flat2.address = {flat2.address}")
flat1.address = "Новый адрес"
print(f"После изменения flat1.address:")
print(f"  flat1.address = {flat1.address}")
print(f"  flat2.address = {flat2.address}")           # не изменился

# ─── Часть 4. Перекрытие классового атрибута ───
print()
print("=== Перекрытие классового атрибута ===")
flat1.agency_name = "Своё агентство"
print(f"flat1.agency_name = {flat1.agency_name}")    # Своё агентство
print(f"flat2.agency_name = {flat2.agency_name}")    # Дом-Сервис
print(f"Apartment.agency_name = {Apartment.agency_name}")   # Дом-Сервис

# ─── Часть 8. Коварство изменяемых атрибутов ───
print()
print("=== Реестр объектов ===")
print(f"len(Apartment.all_apartments) = {len(Apartment.all_apartments)}")   # 3

# НЕПРАВИЛЬНО: присваивание через объект создаёт экземплярный атрибут.
flat1.all_apartments = []
flat1.all_apartments.append("что-то ещё")

print(f"flat1.all_apartments = {flat1.all_apartments}")
print(f"len(flat1.all_apartments) = {len(flat1.all_apartments)}")           # 1
print(f"len(Apartment.all_apartments) = {len(Apartment.all_apartments)}")   # 3

# ─── Часть 9. __dict__ ───
print()
print("=== __dict__ объекта и класса ===")
print(f"flat2.__dict__ = {flat2.__dict__}")
print(f"Apartment.__dict__.keys() = {list(Apartment.__dict__.keys())}")

# ─── Часть 7. Наследование ───
print()
print("=== Наследование без переопределения ===")
studio1 = Studio("ул. Гагарина, д. 5, кв. 3", 25, 4_000_000)
studio2 = Studio("ул. Гагарина, д. 5, кв. 4", 30, 5_000_000)

print(f"Apartment.get_count() = {Apartment.get_count()}")   # 5 (все)
print(f"Studio.get_count() = {Studio.get_count()}")         # 5 (общий)

print()
print("=== Наследование с переопределением ===")
studio3 = StudioWithOwnCount("ул. Мира, д. 1, кв. 1", 28, 4_500_000)

print(f"Apartment.get_count() = {Apartment.get_count()}")              # 5 (не изменился)
print(f"StudioWithOwnCount.get_count() = {StudioWithOwnCount.get_count()}")  # 1 (свой)
```

**Ожидаемый вывод:**

```
=== Классовые атрибуты ===
Apartment.agency_name = Дом-Сервис
Apartment.tax_rate = 0.13
flat1.agency_name = Дом-Сервис
Apartment.count = 3
flat1.count = 3

=== Экземплярные атрибуты ===
flat1.address = ул. Ленина, д. 10, кв. 5
flat2.address = ул. Садовая, д. 3, кв. 12
После изменения flat1.address:
  flat1.address = Новый адрес
  flat2.address = ул. Садовая, д. 3, кв. 12

=== Перекрытие классового атрибута ===
flat1.agency_name = Своё агентство
flat2.agency_name = Дом-Сервис
Apartment.agency_name = Дом-Сервис

=== Реестр объектов ===
len(Apartment.all_apartments) = 3
flat1.all_apartments = ['что-то ещё']
len(flat1.all_apartments) = 1
len(Apartment.all_apartments) = 3

=== __dict__ объекта и класса ===
flat2.__dict__ = {'address': 'ул. Садовая, д. 3, кв. 12', 'area': 80, 'price': 12000000}
Apartment.__dict__.keys() = ['__module__', 'agency_name', 'tax_rate', 'count', 'all_apartments', '__init__', '__repr__', 'get_count', '__dict__', '__weakref__', '__doc__']

=== Наследование без переопределения ===
Apartment.get_count() = 5
Studio.get_count() = 5

=== Наследование с переопределением ===
Apartment.get_count() = 5
StudioWithOwnCount.get_count() = 1
```

### Разбор

**1. Чем классовый атрибут отличается от экземплярного.**

**Классовый атрибут** — определён **в теле класса**, **вне методов**. Он **общий** для всех объектов. Хранится в `Class.__dict__`. Доступен **через класс** и **через объект**. Изменение **через класс** влияет на **всех**.

**Экземплярный атрибут** — определён **внутри метода** через `self`. Он **уникален** для каждого объекта. Хранится в `obj.__dict__`. Доступен **только через объект**. Изменение **через объект** влияет **только на этот объект**.

**2. Как Python ищет атрибуты при обращении `obj.attr`.**

Python ищет в **таком порядке**:

1. **В объекте** — в `obj.__dict__`.
2. **В классе** — в `type(obj).__dict__`.
3. **У родителей** — по MRO.
4. **`AttributeError`**, если не найдено.

**Ключевое:** экземплярный атрибут **перекрывает** классовый. Если у объекта есть **свой** атрибут с тем же именем — он **найдётся первым**.

**3. Почему `flat1.agency_name = "X"` не меняет классовый атрибут.**

`flat1.agency_name = "X"` — это **присваивание**. Python **не ищет** `agency_name` в классе. Он **создаёт** новый атрибут **в объекте** `flat1`. Классовый атрибут **не тронут**. Теперь `flat1.agency_name` **перекрывает** классовый **только для** `flat1`. У `flat2` и `Apartment` — **прежнее** значение.

**4. Почему изменяемые классовые атрибуты коварны.**

Изменяемые атрибуты (`list`, `dict`, `set`) **опасны** как классовые, потому что:

- **Изменение через объект** (`flat1.attr.append(x)`) **изменит** атрибут **для всех**. Потому что `flat1.attr` — **ссылка** на **тот же** объект.
- **Присваивание через объект** (`flat1.attr = []`) **создаст** экземплярный атрибут. Классовый **не изменится**.

Поведение **неочевидно**: одна строка меняет у **всех**, другая — **только у одного**. **Правило:** работайте **через класс** (`Apartment.attr.append(x)`).

**5. Что произойдёт, если в конструкторе написать `self.count += 1` вместо `Apartment.count += 1`.**

`self.count += 1` — это **сокращение** для `self.count = self.count + 1`. Python **сначала** ищет `self.count` **в объекте** — **не находит**. **Идёт в класс** — **находит** `count = 0`. **Возвращает** `0`. **Складывает** с `1` — получает `1`. **Присваивает** `self.count = 1` — **создаёт** **экземплярный** атрибут.

**Результат:** у каждого объекта — **свой** `count = 1`. **Классовый** `count` **не меняется** (остаётся `0`). Счётчик **не работает**.

**Правило:** для изменения **классового** атрибута — **всегда** `Class.attr += 1` или `type(self).attr += 1`.

**6. Как классовые атрибуты наследуются и переопределяются.**

**Наследуются:** если наследник **не переопределяет** — использует **родительский**. Изменение родителя **видно** у наследника.

**Переопределяются:** если наследник **определяет свой** — использует **свой**. Изменение родителя **не видно** у наследника.

**С счётчиком:** если наследник **не переопределил** — использует **общий** счётчик. Если **переопределил** — получает **свой**.

В примере:

- `Studio` **не переопределил** `count` → использует **общий** (`Apartment.count`).
- `StudioWithOwnCount` **переопределил** `count = 0` → имеет **свой**.

**Правило:** используйте `type(self).count += 1` — тогда счётчик **автоматически** работает правильно: у наследников без переопределения — общий, с переопределением — свой.

# Глава 17. Собственные исключения

## 17.1. Проблема: стандартных исключений не всегда хватает

За предыдущие главы мы построили класс `Apartment`, научили его создаваться из строк, красиво выводиться, сравниваться, складываться в комплексы. Но чем больше программа, тем чаще возникают **ситуации**, которые **не должны** происходить. Пользователь вводит отрицательную площадь. Кто-то пытается найти квартиру, которой нет в каталоге. Агент пытается продать уже проданную квартиру. Все эти ситуации — **ошибки**. И их нужно **обрабатывать**.

Python предоставляет **много** встроенных исключений. `ValueError` — когда значение не подходит. `TypeError` — когда тип неправильный. `KeyError` — когда ключа нет в словаре. `IndexError` — когда индекс вне диапазона. `ZeroDivisionError` — при делении на ноль. И ещё десятки других.

Казалось бы, этого **достаточно**. Зачем создавать **свои** исключения, если есть **готовые**? На первый взгляд — незачем. Но когда программа **растёт**, становится **ясно**: стандартных исключений **не хватает**. И вот почему.

### Проблема 1: непонятно, что именно пошло не так

Представь, что в агентстве есть **несколько** мест, где может возникнуть `ValueError`:

- При **проверке площади** — если она отрицательная.
- При **проверке цены** — если она не число.
- При **парсинге строки** — если формат неправильный.
- При **конвертации валюты** — если курс отрицательный.

Везде — `ValueError`. И когда программа **падает** с этим исключением, **непонятно**, какая именно ошибка произошла. Приходится **читать сообщение** и **угадывать**, где искать проблему.

### Проблема 2: невозможно поймать только свою ошибку

Если программа использует **только** `ValueError`, то `except ValueError` **поймает все** — и **свою** ошибку, и **чужую**. Например, ошибку **внутри библиотеки**, которую вы **не хотите** перехватывать. Приходится **угадывать** по сообщению, что именно произошло. Это **ненадёжно**.

### Проблема 3: невозможно передать дополнительную информацию

`ValueError` может хранить **только** сообщение. А иногда нужно **больше**: какое **поле** не прошло проверку, какое **значение** было передано, какие **допустимые** значения. Стандартные исключения **не позволяют** этого сделать **удобно**.

### Решение

**Собственные исключения** решают **все три** проблемы. Они позволяют:

- Дать **понятное** имя ошибке (`ApartmentNotFoundError` вместо `ValueError`).
- Ловить **только свою** ошибку (`except ApartmentNotFoundError`).
- Передавать **дополнительную** информацию (адрес, площадь, допустимый диапазон).

### Аналогия

Представь **больницу**. У каждой болезни есть **своё** название: грипп, ангина, аллергия. Врач **не говорит** просто «вам плохо» — он **ставит диагноз**. Так и с исключениями: **своё имя** помогает **понять**, что именно случилось. С `ValueError` — как с диагнозом «вам плохо». Со **своим** исключением — как с диагнозом «у вас ангина».

### Вывод

Стандартных исключений **не всегда хватает**. **Собственные** исключения делают код **понятнее**, **надёжнее** и **удобнее**.

---

## 17.2. Что такое исключение: краткое напоминание

Прежде чем создавать **свои** исключения, вспомним, что такое **исключение** вообще. Если вы уже знаете — можете **пропустить** этот раздел, но я рекомендую **пробежаться** по нему, чтобы **освежить** понимание.

**Исключение** — это **объект**, который **создаётся** в момент **ошибки** и **передаётся** вверх по стеку вызовов. Если исключение **не поймано** — программа **аварийно завершается** с **сообщением** об ошибке.

### Как это работает

Когда происходит **ошибка**, Python **создаёт** объект-исключение и **выбрасывает** его. Это называется **«raise»** (поднять, выбросить). Исключение **распространяется** вверх по стеку: от **функции**, где произошла ошибка, к **вызывающей** функции, к **её вызывающей** — и так **до самого верха**.

Если **где-то** по пути есть **обработчик** (`try/except`) — исключение **ловится** и **обрабатывается**. Если **нет** — программа **падает** с **traceback** (трассировкой).

### Простой пример

```python
def divide(a, b):
    return a / b

result = divide(10, 0)
```

Здесь произойдёт `ZeroDivisionError`. Python **создаст** объект этого исключения и **выбросит** его. Так как **нет** обработчика — программа **упадёт** с сообщением:

```
ZeroDivisionError: division by zero
```

### Как ловить исключения

```python
try:
    result = divide(10, 0)
except ZeroDivisionError as e:
    print(f"Ошибка: {e}")
```

Здесь `try` — блок, где **может** произойти ошибка. `except ZeroDivisionError` — **обработчик** для **этого** типа. Если исключение **поймано** — программа **продолжает** работу.

### Иерархия исключений

Все исключения Python образуют **иерархию**:

```
        BaseException
        ├── SystemExit
        ├── KeyboardInterrupt
        ├── GeneratorExit
        └── Exception
            ├── ValueError
            ├── TypeError
            ├── ArithmeticError
            │   └── ZeroDivisionError
            ├── LookupError
            │   ├── KeyError
            │   └── IndexError
            └── ...
```

**BaseException** — **корень** иерархии. **Exception** — **основа** для **обычных** исключений. **ValueError**, **TypeError** и другие — **наследники** `Exception`.

### Почему это важно

Понятие **иерархии** — **ключевое** для **собственных** исключений. Мы будем **наследовать** свои исключения от `Exception` (или от **других** исключений). Это **встроит** их в **общую** иерархию, и они будут **вести себя** как **обычные** исключения.

### Вывод

**Исключение** — объект, который **создаётся** при ошибке и **распространяется** вверх по стеку. **Иерархия** исключений — **основа** для **собственных** исключений.

---

## 17.3. Что такое собственное исключение

**Собственное исключение** — это **класс**, унаследованный от `Exception` (или от **другого** исключения). Он **ведёт себя** как **обычное** исключение, но имеет **своё** имя и **свою** логику.

### Минимальный пример

```python
class ApartmentNotFoundError(Exception):
    pass
```

Всё. Это **полноценное** исключение. Его можно **выбрасывать** через `raise` и **ловить** через `except`.

### Синтаксис

```python
class ИмяИсключения(Exception):
    pass
```

- **`class`** — ключевое слово для создания класса.
- **`ИмяИсключения`** — имя. По **соглашению** заканчивается на **`Error`**.
- **`(Exception)`** — наследуем от `Exception`.
- **`pass`** — тело класса. Может быть **пустым** — тогда исключение **просто** имеет **имя**.

### Почему наследуют от `Exception`

**Потому что** `Exception` — **базовый** класс для **всех** обычных исключений. Наследуя от него, вы **встраиваете** своё исключение в **общую** иерархию. Оно **будет** ловиться `except Exception`, **будет** работать с `raise`, **будет** иметь **стандартные** атрибуты.

Если унаследовать от `BaseException` — исключение **не будет** ловиться `except Exception`. Это **неправильно** для **обычных** ошибок. От `BaseException` наследуют **только** системные исключения — `SystemExit`, `KeyboardInterrupt`.

### Что происходит под капотом

Когда вы пишете `class ApartmentNotFoundError(Exception): pass`, Python:

1. **Создаёт** новый класс `ApartmentNotFoundError`.
2. **Регистрирует** его в **общей** иерархии как **наследника** `Exception`.
3. **Добавляет** стандартные методы (например, `__str__`, `__repr__`).
4. **Позволяет** использовать его в `raise` и `except`.

### Аналогия

Представь **каталог** ошибок. В нём есть **разделы**: «Ошибки ввода», «Ошибки сети», «Ошибки базы данных». **Своё** исключение — это **новая** папка в этом каталоге. Она **встраивается** в **общую** структуру, но имеет **своё** имя.

### Что можно делать с собственным исключением

- **Выбрасывать** — `raise ApartmentNotFoundError("...")`.
- **Ловить** — `except ApartmentNotFoundError:`.
- **Наследовать** — создавать **более специфичные** исключения.
- **Добавлять** атрибуты и методы — для **дополнительной** информации.

### Вывод

**Собственное исключение** — класс, унаследованный от `Exception`. Минимально — это `class NameError(Exception): pass`. Оно **встраивается** в общую иерархию и **ведёт себя** как стандартное.

---

## 17.4. Как создать собственное исключение

Разберём **пошагово**, как создать **своё** исключение. Начнём с **минимального** и **постепенно** усложним.

### Шаг 1. Минимальное исключение

```python
class ApartmentNotFoundError(Exception):
    pass
```

Этого **достаточно**, чтобы использовать исключение. Оно имеет **имя** и **встроено** в иерархию. Стандартное сообщение — **пустое**.

### Шаг 2. С сообщением

```python
raise ApartmentNotFoundError("Квартира не найдена")
```

Здесь `"Квартира не найдена"` — **сообщение**. Оно **передаётся** в **конструктор** `Exception`. Когда исключение **печатается**, выводится **это** сообщение.

**Как передаётся сообщение:**

`Exception` (и его наследники) принимает **любое** число аргументов. Обычно один — строка. Все аргументы **сохраняются** в атрибуте `args`.

```python
e = ApartmentNotFoundError("Квартира не найдена", "ул. Ленина")
print(e.args)    # ("Квартира не найдена", "ул. Ленина")
```

### Шаг 3. С дополнительными данными

Иногда нужно **больше**, чем просто сообщение. Например, **адрес**, который не найден. Можно **добавить** свой конструктор:

```python
class ApartmentNotFoundError(Exception):
    def __init__(self, address):
        self.address = address
        super().__init__(f"Квартира по адресу {address!r} не найдена")
```

Здесь:

- **`def __init__(self, address):`** — свой конструктор.
- **`self.address = address`** — сохраняем **дополнительное** поле.
- **`super().__init__(...)`** — вызываем **родительский** конструктор, чтобы **установить** сообщение.

Теперь **адрес** доступен **изнутри** исключения:

```python
try:
    raise ApartmentNotFoundError("ул. Ленина, д. 10, кв. 5")
except ApartmentNotFoundError as e:
    print(e.address)    # ул. Ленина, д. 10, кв. 5
    print(e)            # Квартира по адресу 'ул. Ленина, д. 10, кв. 5' не найдена
```

### Шаг 4. С валидацией

Можно **проверять** данные **внутри** исключения:

```python
class InvalidAreaError(Exception):
    def __init__(self, area, min_area=5, max_area=500):
        if not isinstance(area, (int, float)):
            raise TypeError("area должно быть числом")
        self.area = area
        self.min_area = min_area
        self.max_area = max_area
        super().__init__(
            f"Площадь {area} м² вне диапазона [{min_area}, {max_area}]"
        )
```

Теперь исключение **само** проверяет, что данные **корректны**.

### Шаг 5. С дополнительными методами

Можно **добавить** методы **в** исключение:

```python
class InvalidAreaError(Exception):
    def __init__(self, area, min_area, max_area):
        self.area = area
        self.min_area = min_area
        self.max_area = max_area
        super().__init__(f"Площадь {area} вне диапазона")

    def is_too_small(self):
        return self.area < self.min_area

    def is_too_large(self):
        return self.area > self.max_area
```

Теперь можно **различать**, какая **именно** ошибка:

```python
try:
    raise InvalidAreaError(2, 5, 500)
except InvalidAreaError as e:
    if e.is_too_small():
        print("Площадь слишком маленькая")
    elif e.is_too_large():
        print("Площадь слишком большая")
```

### Что важно помнить при создании

1. **Наследуйте** от `Exception` (или его наследников).
2. **Давайте** понятное имя, заканчивающееся на `Error`.
3. **Вызывайте** `super().__init__(...)` — иначе **сообщение** не будет установлено.
4. **Добавляйте** только **полезные** атрибуты.
5. **Не** переусложняйте — исключение **должно** быть **простым**.

### Аналогия

Создание **своего** исключения — как создание **нового** диагноза **в медицине**. Диагноз **встраивается** в **общую** классификацию (наследование от `Exception`). Он **имеет** своё имя (**имя** класса). У него могут быть **дополнительные** поля (симптомы, стадия). И его **можно** «поставить» (выбросить) или «распознать» (поймать).

### Вывод

Создать **своё** исключение — **просто**. Минимально — `class NameError(Exception): pass`. Но можно **добавить** конструктор, атрибуты, методы — для **удобства** и **дополнительной** информации.

---

## 17.5. Иерархия собственных исключений

**Иерархия** исключений — **мощный** инструмент. Она позволяет **группировать** исключения по **смыслу** и **ловить** их **по группам**.

### Зачем нужна иерархия

Представь, что у нас есть **несколько** ошибок, связанных с **квартирами**:

- `ApartmentNotFoundError` — квартира не найдена.
- `ApartmentAlreadySoldError` — квартира уже продана.
- `ApartmentInvalidAreaError` — площадь недопустима.

Можно **создать** их как **отдельные** исключения от `Exception`. Тогда **каждое** придётся ловить **отдельно**:

```python
try:
    ...
except ApartmentNotFoundError:
    ...
except ApartmentAlreadySoldError:
    ...
except ApartmentInvalidAreaError:
    ...
```

Это **работает**, но **неудобно**, если нужно **обработать** все ошибки **одинаково**. С **иерархией** можно **сгруппировать** их:

```python
class ApartmentError(Exception):
    """Базовое исключение для всех ошибок с квартирами."""
    pass

class ApartmentNotFoundError(ApartmentError):
    pass

class ApartmentAlreadySoldError(ApartmentError):
    pass

class ApartmentInvalidAreaError(ApartmentError):
    pass
```

Теперь можно **ловить** как **конкретные**:

```python
try:
    ...
except ApartmentNotFoundError:
    ...
except ApartmentAlreadySoldError:
    ...
```

Так и **все сразу**:

```python
try:
    ...
except ApartmentError:
    print("Произошла ошибка с квартирой")
```

### Схема иерархии

```
        Exception
        └── ApartmentError           ← базовое
            ├── ApartmentNotFoundError
            ├── ApartmentAlreadySoldError
            └── ApartmentInvalidAreaError
```

### Как работает ловля

Когда Python **ищет** подходящий `except`, он **сравнивает** типы. Если **тип** исключения **совпадает** или **является** **наследником** — `except` **срабатывает**.

```python
try:
    raise ApartmentNotFoundError("...")
except ApartmentError:
    print("Поймано!")    # ← сработает — ApartmentNotFoundError наследует ApartmentError
```

### Преимущества иерархии

- **Группировка** — ошибки одного **типа** объединены **под** общим родителем.
- **Гибкость** — можно ловить **конкретные** или **все** сразу.
- **Расширяемость** — легко **добавить** новое исключение в группу.
- **Понятность** — иерархия **показывает** связи между ошибками.

### Пример посложнее

Можно строить **многоуровневую** иерархию:

```python
class AgencyError(Exception):
    """Все ошибки агентства."""
    pass

class ApartmentError(AgencyError):
    """Ошибки с квартирами."""
    pass

class ClientError(AgencyError):
    """Ошибки с клиентами."""
    pass

class ApartmentNotFoundError(ApartmentError):
    pass

class ClientNotFoundError(ClientError):
    pass
```

Теперь можно ловить на **разных** уровнях:

```python
except ApartmentNotFoundError:      # конкретная ошибка
except ApartmentError:              # все ошибки квартир
except AgencyError:                 # все ошибки агентства
except Exception:                   # вообще всё
```

### Аналогия

Представь **дерево** категорий в **интернет-магазине**. Сверху — «Электроника». Внутри — «Смартфоны», «Ноутбуки». Внутри «Смартфоны» — «Apple», «Samsung». Ты можешь **фильтровать** по **любому** уровню: показать **всю** электронику, **все** смартфоны, или **только** Apple. **Иерархия** исключений работает **так же**.

### Когда создавать иерархию

**Создавайте**, если:

- Есть **несколько** связанных ошибок.
- Нужно **ловить** их **по группам**.
- Логически они **принадлежат** одной **области**.

**Не создавайте**, если:

- Ошибка **одна** — хватит **одного** класса.
- Ошибки **разнородные** — не имеет смысла **группировать**.

### Вывод

**Иерархия** исключений позволяет **группировать** ошибки и **ловить** их **по группам**. Строится через **наследование**. Даёт **гибкость** и **понятность**.

---

## 17.6. Использование собственных исключений

Разберём **четыре** типичных случая использования **собственных** исключений.

### 17.6.1. Валидация данных

**Самый частый** случай — **проверка** данных **при** создании объекта:

```python
class InvalidAreaError(Exception):
    def __init__(self, area, min_area=5, max_area=500):
        self.area = area
        self.min_area = min_area
        self.max_area = max_area
        super().__init__(
            f"Площадь {area} м² вне диапазона [{min_area}, {max_area}]"
        )


class Apartment:
    def __init__(self, address, area, price):
        if not isinstance(area, (int, float)):
            raise InvalidAreaError(area)
        if not (5 <= area <= 500):
            raise InvalidAreaError(area)

        self.address = address
        self.area = area
        self.price = price
```

Здесь `InvalidAreaError` **выбрасывается** при **некорректной** площади. Это **понятнее**, чем `ValueError`, потому что имя **говорит** само за себя.

### 17.6.2. Поиск объектов

Когда **объект** не найден — **выбрасываем** **специфичное** исключение:

```python
class ApartmentNotFoundError(Exception):
    def __init__(self, address):
        self.address = address
        super().__init__(f"Квартира по адресу {address!r} не найдена")


class Agency:
    def __init__(self):
        self.apartments = []

    def add(self, apartment):
        self.apartments.append(apartment)

    def find_by_address(self, address):
        for flat in self.apartments:
            if flat.address == address:
                return flat
        raise ApartmentNotFoundError(address)
```

Теперь код, который **ищет** квартиру, **чётко** знает: если не найдено — **будет** `ApartmentNotFoundError`, а не **что-то** стандартное.

### 17.6.3. Бизнес-правила

**Бизнес-правила** — **отличный** случай для **своих** исключений:

```python
class ApartmentAlreadySoldError(Exception):
    def __init__(self, apartment):
        self.apartment = apartment
        super().__init__(f"Квартира {apartment.address!r} уже продана")


class Apartment:
    def __init__(self, address, price):
        self.address = address
        self.price = price
        self.is_sold = False

    def sell(self):
        if self.is_sold:
            raise ApartmentAlreadySoldError(self)
        self.is_sold = True
```

Здесь `ApartmentAlreadySoldError` **выбрасывается** при попытке **продать** уже **проданную** квартиру. Это **бизнес-правило**, и **своё** исключение **подчёркивает** это.

### 17.6.4. Ошибки парсинга

Когда **парсите** данные — **свои** исключения **помогают**:

```python
class ParseError(Exception):
    def __init__(self, data, reason):
        self.data = data
        self.reason = reason
        super().__init__(f"Не удалось разобрать {data!r}: {reason}")


class Apartment:
    @classmethod
    def from_string(cls, data):
        try:
            address, area, price = data.split(";")
        except ValueError:
            raise ParseError(data, "неверный формат строки")

        try:
            area = int(area)
            price = int(price)
        except ValueError:
            raise ParseError(data, "площадь или цена не числа")

        return cls(address, area, price)
```

Здесь `ParseError` **оборачивает** стандартные ошибки в **своё** исключение с **дополнительной** информацией.

### Что общего у всех примеров

Во **всех** примерах **своё** исключение:

- Даёт **понятное** имя.
- Передаёт **дополнительные** данные.
- Позволяет **ловить** только **свою** ошибку.
- **Отдел**яет свои ошибки от **чужих**.

### Вывод

**Собственные** исключения используются **везде**, где стандартных **не хватает**. Особенно — в **валидации**, **поиске**, **бизнес-правилах**, **парсинге**.

---

## 17.7. Собственные исключения с дополнительной информацией

**Одно** из **главных** преимуществ **своих** исключений — возможность **передавать** **дополнительную** информацию. Разберём это **подробно**.

### Проблема стандартных исключений

`ValueError` хранит **только** сообщение. Если нужно **больше** — приходится **всё** **запихивать** в строку:

```python
raise ValueError("Площадь -10 вне диапазона [5, 500], объект: ул. Ленина")
```

Это **работает**, но **неудобно**:

- Сообщение **длинное** и **трудночитаемое**.
- **Невозможно** получить **отдельные** поля (площадь, диапазон, адрес) **программно**.
- **Невозможно** проверить, **какая** именно **проблема** (маленькая или большая площадь).

### Решение — своё исключение с полями

```python
class InvalidAreaError(Exception):
    def __init__(self, area, address, min_area=5, max_area=500):
        self.area = area
        self.address = address
        self.min_area = min_area
        self.max_area = max_area
        super().__init__(
            f"Квартира {address!r}: площадь {area} м² вне диапазона [{min_area}, {max_area}]"
        )
```

Теперь **все** данные **доступны** как **отдельные** поля:

```python
try:
    raise InvalidAreaError(-10, "ул. Ленина", 5, 500)
except InvalidAreaError as e:
    print(f"Ошибка: {e}")
    print(f"Площадь: {e.area}")
    print(f"Адрес: {e.address}")
    print(f"Диапазон: [{e.min_area}, {e.max_area}]")
```

### Методы для анализа

Можно добавить **методы** для **анализа** ошибки:

```python
class InvalidAreaError(Exception):
    def __init__(self, area, min_area, max_area):
        self.area = area
        self.min_area = min_area
        self.max_area = max_area
        super().__init__(f"Площадь {area} вне [{min_area}, {max_area}]")

    def is_too_small(self):
        return self.area < self.min_area

    def is_too_large(self):
        return self.area > self.max_area

    def describe(self):
        if self.is_too_small():
            return f"Минимальная площадь: {self.min_area} м²"
        if self.is_too_large():
            return f"Максимальная площадь: {self.max_area} м²"
        return "Площадь в допустимом диапазоне"
```

Теперь **обработчик** может **понять**, **что** именно **не так**, и **дать** **полезное** сообщение:

```python
try:
    raise InvalidAreaError(2, 5, 500)
except InvalidAreaError as e:
    print(e.describe())      # Минимальная площадь: 5 м²
```

### Дополнительные поля в иерархии

Если у вас **иерархия**, можно **добавлять** поля **на разных** уровнях:

```python
class AgencyError(Exception):
    def __init__(self, message, code=None):
        self.code = code
        super().__init__(message)


class ApartmentError(AgencyError):
    def __init__(self, message, address, code=None):
        self.address = address
        super().__init__(message, code)


class ApartmentNotFoundError(ApartmentError):
    def __init__(self, address):
        super().__init__(
            f"Квартира {address!r} не найдена",
            address,
            code="APARTMENT_NOT_FOUND"
        )
```

Здесь **каждый** уровень **добавляет** свои поля. У `ApartmentNotFoundError` — **все** поля: `code`, `address`, `args`.

### Аналогия

Представь **медицинскую** карту. В ней **много** полей: имя, возраст, диагноз, симптомы, анализы. **Своё** исключение — это **такая** карта. Ты **не** пишешь всё **в одну** строку. Ты **хранишь** данные **по полям**. Это **удобно** и **программно** доступно.

### Что важно помнить

- **Вызывайте** `super().__init__(...)` — иначе **сообщение** не установится.
- **Сохраняйте** данные в `self.` — чтобы они **были** доступны.
- **Добавляйте** только **полезные** поля.
- **Не** переусложняйте — исключение **должно** быть **простым**.

### Вывод

**Дополнительные** поля в **своих** исключениях — **мощный** инструмент. Они позволяют **передавать** больше **информации** и **обрабатывать** ошибки **программно**.

## 17.8. Собственные исключения в наследовании

**Собственные** исключения **наследуются** как **обычные** классы. Это **важно** понимать, потому что **поведение** при наследовании имеет **нюансы**.

### Наследование исключений

Можно **наследовать** **одно** исключение от **другого**:

```python
class ApartmentError(Exception):
    pass

class ApartmentNotFoundError(ApartmentError):
    pass

class ApartmentAlreadySoldError(ApartmentError):
    pass
```

Здесь `ApartmentNotFoundError` — **наследник** `ApartmentError`. Это **не** **отдельное** исключение, а **разновидность** `ApartmentError`.

### Что это даёт

**Ловля** по **родителю**:

```python
try:
    raise ApartmentNotFoundError("...")
except ApartmentError:
    print("Поймано!")    # сработает
```

**Проверка** через `isinstance`:

```python
e = ApartmentNotFoundError("...")
print(isinstance(e, ApartmentError))    # True
print(isinstance(e, Exception))         # True
```

### Переопределение поведения

Наследник **может** переопределить `__init__`:

```python
class ApartmentError(Exception):
    def __init__(self, message):
        self.message = message
        super().__init__(f"[ApartmentError] {message}")

class ApartmentNotFoundError(ApartmentError):
    def __init__(self, address):
        self.address = address
        super().__init__(f"Квартира {address!r} не найдена")
```

Здесь `ApartmentNotFoundError` **дополняет** родительский конструктор.

### Цепочки исключений

Можно строить **многоуровневые** иерархии:

```python
class AgencyError(Exception): pass
class ApartmentError(AgencyError): pass
class ApartmentNotFoundError(ApartmentError): pass
```

Теперь можно **ловить** на **любом** уровне:

```python
except ApartmentNotFoundError:      # конкретная ошибка
except ApartmentError:              # все ошибки квартир
except AgencyError:                 # все ошибки агентства
```

### Стандартные исключения в иерархии

**Свои** исключения можно **наследовать** от **стандартных**:

```python
class InvalidAreaError(ValueError):
    def __init__(self, area, min_area, max_area):
        self.area = area
        self.min_area = min_area
        self.max_area = max_area
        super().__init__(f"Площадь {area} вне [{min_area}, {max_area}]")
```

Здесь `InvalidAreaError` — **наследник** `ValueError`. Это даёт **два** преимущества:

- **Ловится** через `except InvalidAreaError` — **конкретно**.
- **Ловится** через `except ValueError` — **вместе** с другими ошибками **значений**.

### Когда наследовать от стандартных

**Наследуйте** от **стандартных**, если:

- Ваша ошибка — **разновидность** стандартной.
- Хотите, чтобы код, **ловящий** стандартную ошибку, **ловил** и вашу.
- Например, `InvalidAreaError` — это **разновидность** `ValueError`.

**Наследуйте** от `Exception`, если:

- Ошибка **специфична** для вашей области.
- **Не** является **разновидностью** стандартных.
- Например, `ApartmentNotFoundError` — **не** `ValueError` и **не** `KeyError`.

### Аналогия

Представь **классификацию** животных. Все **млекопитающие** — **наследники** класса «Животные». Кошки — **наследники** «Млекопитающих». Персидские кошки — **наследники** «Кошек». Ты можешь **говорить** о «всех животных», о «всех млекопитающих», о «всех кошках» или о «персидских». **Иерархия** исключений работает **так же**.

### Вывод

**Собственные** исключения **наследуются** как **обычные** классы. Можно строить **многоуровневые** иерархии. Можно **наследовать** от **стандартных** исключений. Это даёт **гибкость** в **обработке**.

## 17.9. Собственные исключения в реальных проектах

Где **собственные** исключения **активно** используются. Разберём **пять** примеров.

### 17.9.1. Django

Django определяет **свои** исключения для **разных** ситуаций:

```python
from django.core.exceptions import ObjectDoesNotExist, ValidationError

try:
    flat = Apartment.objects.get(address="...")
except Apartment.DoesNotExist:
    print("Квартира не найдена")
```

`Apartment.DoesNotExist` — **своё** исключение, определённое **автоматически** для **каждой** модели Django.

### 17.9.2. SQLAlchemy

SQLAlchemy определяет **иерархию** исключений:

```python
from sqlalchemy.exc import IntegrityError, OperationalError

try:
    session.add(flat)
    session.commit()
except IntegrityError:
    print("Нарушение целостности данных")
```

`IntegrityError` — **наследник** `SQLAlchemyError` — **базового** исключения библиотеки.

### 17.9.3. Requests

Библиотека `requests` определяет **свои** исключения:

```python
import requests

try:
    response = requests.get("https://example.com")
except requests.ConnectionError:
    print("Нет соединения")
except requests.Timeout:
    print("Таймаут")
```

`ConnectionError`, `Timeout` — **наследники** `RequestException` — **базового** исключения библиотеки.

### 17.9.4. FastAPI

FastAPI использует **свои** исключения для **HTTP**:

```python
from fastapi import HTTPException

@app.get("/apartments/{id}")
def get_apartment(id: int):
    flat = find_apartment(id)
    if not flat:
        raise HTTPException(status_code=404, detail="Не найдена")
    return flat
```

`HTTPException` — **своё** исключение с **дополнительной** информацией: `status_code`, `detail`.

### 17.9.5. Наш дом

```python
class RoomError(Exception):
    """Базовая ошибка для комнат."""
    pass

class RoomNotFoundError(RoomError):
    def __init__(self, name):
        self.name = name
        super().__init__(f"Комната {name!r} не найдена")

class RoomInvalidAreaError(RoomError):
    def __init__(self, area, min_area, max_area):
        self.area = area
        self.min_area = min_area
        self.max_area = max_area
        super().__init__(f"Площадь {area} вне [{min_area}, {max_area}]")
```

### Что общего у всех примеров

Во **всех** примерах:

- Есть **базовое** исключение для **всей** области.
- Есть **конкретные** исключения для **отдельных** ситуаций.
- Исключения **передают** **дополнительные** данные.
- Можно **ловить** на **разных** уровнях.

### Вывод

**Собственные** исключения — **стандарт** для **реальных** проектов. Почти **каждая** **большая** библиотека определяет **свою** иерархию исключений.

## 17.10. Нюансы и тонкости

Разберём **десять** нюансов, которые часто упускают.

### 17.10.1. `raise` внутри `except` затирает traceback

Если вы ловите исключение и **выбрасываете** новое — **старый** traceback **теряется**:

```python
try:
    int("abc")
except ValueError:
    raise ParseError("Не удалось разобрать")    # ← traceback потерян
```

Чтобы **сохранить** связь с **оригинальной** ошибкой — используйте `from`:

```python
try:
    int("abc")
except ValueError as e:
    raise ParseError("Не удалось разобрать") from e    # ← связь сохранена
```

Теперь traceback **покажет** обе ошибки: «During handling of the above exception, another exception occurred».

### 17.10.2. `raise` без аргументов перебрасывает текущее исключение

Внутри `except` можно **перебросить** текущее исключение **без** изменений:

```python
try:
    ...
except ApartmentError:
    print("Логируем ошибку...")
    raise    # ← перебрасываем то же самое исключение
```

Это **полезно**, когда нужно **залогировать** ошибку, но **не** обработать её **окончательно**.

### 17.10.3. `raise ... from None` убирает контекст

Если хотите **скрыть** оригинальное исключение — используйте `from None`:

```python
try:
    int("abc")
except ValueError:
    raise ParseError("Не удалось разобрать") from None
```

Traceback **покажет** **только** `ParseError`, **без** контекста.

### 17.10.4. Проверка типа исключения через `isinstance`

Можно **проверить** тип **в** `except`:

```python
except (ApartmentNotFoundError, ApartmentAlreadySoldError) as e:
    if isinstance(e, ApartmentNotFoundError):
        print("Не найдена")
    else:
        print("Уже продана")
```

### 17.10.5. `__str__` для исключений

`Exception.__str__` возвращает **сообщение** (первый аргумент). Можно **переопределить**:

```python
class ApartmentError(Exception):
    def __str__(self):
        return f"Ошибка агентства: {super().__str__()}"
```

### 17.10.6. `__repr__` для исключений

`Exception.__repr__` показывает **класс** и **аргументы**. Можно **переопределить**:

```python
class ApartmentError(Exception):
    def __repr__(self):
        return f"{self.__class__.__name__}({self.args!r})"
```

### 17.10.7. Атрибут `args`

Все аргументы конструктора **сохраняются** в `args`:

```python
e = ApartmentError("Сообщение", "Дополнительно")
print(e.args)    # ("Сообщение", "Дополнительно")
```

Это **стандартный** механизм. Если хотите **свои** атрибуты — сохраняйте их **вручную** в `self.`.

### 17.10.8. `__cause__` и `__context__`

У исключения есть **два** атрибута для **связи**:

- **`__context__`** — **автоматически** устанавливается, если исключение **выброшено** внутри другого `except`.
- **`__cause__`** — устанавливается через `from`.

### 17.10.9. Исключения и `with`

**Контекстные менеджеры** (см. главу 33) могут **перехватывать** исключения через `__exit__`. Это **полезно** для **гарантированной** очистки ресурсов.

### 17.10.10. Логирование исключений

Исключения **удобно** логировать через `logging.exception`:

```python
import logging

try:
    ...
except ApartmentError as e:
    logging.exception("Ошибка с квартирой")    # ← логирует с traceback
```

Это **сохранит** всю **информацию** об ошибке.

### Вывод

**Собственные** исключения имеют **много** нюансов. Основные — `raise ... from`, `raise` без аргументов, атрибуты `args`, `__cause__`, `__context__`. Знание этих нюансов делает код **чище** и **надёжнее**.

---

## 17.11. Типичные ошибки

Разберём **десять** самых частых ошибок.

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Наследуют от `BaseException` | Не ловится `except Exception` | Наследовать от `Exception` |
| Забыли `super().__init__()` | Сообщение не установлено | Вызвать `super().__init__(...)` |
| Не дают имени на `Error` | Непонятно по имени | Заканчивать на `Error` |
| Ловят слишком широко | Ловятся чужие ошибки | Ловить конкретный класс |
| Забыли `from` при перебросе | Traceback теряется | Использовать `raise ... from` |
| Дублируют стандартные исключения | `ApartmentValueError` вместо `ValueError` | Наследовать от `ValueError` |
| Слишком много классов исключений | Избыточно | Группировать по смыслу |
| Слишком мало классов | Неинформативно | Разделить по ситуациям |
| Забыли сохранить данные в `self.` | Данные недоступны | Сохранять в атрибуты |
| Используют `Exception` как базу везде | Слишком общо | Создать своё базовое исключение |

### Подробный разбор

**Ошибка 1.** Наследование от `BaseException` — **неправильно** для **обычных** ошибок. Такие исключения **не ловятся** `except Exception`. **Правило:** наследуйте от `Exception`.

**Ошибка 2.** Забыли `super().__init__()` — **сообщение** не установлено. `str(e)` вернёт **пустоту**. **Правило:** всегда вызывайте `super().__init__(...)`.

**Ошибка 3.** Имя **не** заканчивается на `Error` — **непонятно** по коду, что это **исключение**. **Правило:** используйте суффикс `Error`.

**Ошибка 4.** Ловля **слишком** широко — **поймает** чужие ошибки. Например, `except Exception` **скроет** **все** ошибки. **Правило:** ловите **конкретные** исключения.

**Ошибка 5.** При **перебросе** без `from` — **оригинальный** traceback **теряется**. **Правило:** используйте `raise NewError(...) from original`.

**Ошибка 6.** Дублирование стандартных — `ApartmentValueError` **не нужен**, если можно **наследовать** от `ValueError`. **Правило:** наследуйте от стандартных, если ошибка — **разновидность**.

**Ошибка 7.** **Слишком много** классов — **тяжело** поддерживать. **Правило:** группируйте по **смыслу**.

**Ошибка 8.** **Слишком мало** классов — **нельзя** различить ошибки. **Правило:** разделяйте по **ситуациям**.

**Ошибка 9.** Забыли **сохранить** данные — они **недоступны** в обработчике. **Правило:** сохраняйте в `self.` атрибуты.

**Ошибка 10.** Использование `Exception` **напрямую** — **слишком** общо. **Правило:** создайте **своё** базовое исключение — `AgencyError`.

### Как избежать

1. **Наследуйте** от `Exception` (или **стандартных**).
2. **Вызывайте** `super().__init__(...)`.
3. **Давайте** **понятные** имена.
4. **Ловите** **конкретные** исключения.
5. **Группируйте** через **иерархию**.

### Вывод

Ошибки с **собственными** исключениями **типичны** для **новичков**. Следуя **правилам**, легко их **избежать**.

---

## 17.12. Что важно запомнить

- **Собственное исключение** — класс, унаследованный от `Exception`.
- **Минимально** — `class NameError(Exception): pass`.
- **Наследуйте** от `Exception`, а не от `BaseException`.
- **Имя** — заканчивается на `Error`.
- **Вызывайте** `super().__init__(...)`.
- **Иерархия** — для **группировки** ошибок.
- **Дополнительные** поля — через `self.` в конструкторе.
- **Методы** в исключении — для **анализа**.
- **`raise ... from e`** — для **сохранения** traceback.
- **`raise`** без аргументов — **переброс** текущего.
- **`isinstance`** — для **проверки** типа.
- **Наследование** от **стандартных** — если ошибка **разновидность**.
- **Ловите** **конкретные** исключения, а не **всё** подряд.
- **Логируйте** через `logging.exception`.

## Задача 17 (собственные исключения)

В агентстве недвижимости возможны **разные** ошибки: квартира **не найдена**, площадь **недопустима**, квартира **уже продана**. Стандартные исключения **не подходят** — нужно **своё**, говорящее.

**Часть 1. Базовое исключение.** Создайте **базовое** исключение `AgencyError(Exception)` — общее для **всех** ошибок агентства. Оно **не будет** выбрасываться **напрямую**, но послужит **родителем** для **остальных**.

**Часть 2. Конкретные исключения.** Создайте **три** исключения, **наследника** `AgencyError`:

- `ApartmentNotFoundError` — квартира не найдена. **Поля:** `address`. **Сообщение:** `Квартира по адресу <address> не найдена`.
- `ApartmentAlreadySoldError` — квартира уже продана. **Поля:** `address`. **Сообщение:** `Квартира <address> уже продана`.
- `InvalidAreaError` — площадь недопустима. **Поля:** `area`, `min_area`, `max_area`. **Сообщение:** `Площадь <area> м² вне диапазона [<min_area>, <max_area>]`. **Методы:** `is_too_small()`, `is_too_large()`, `describe()`.

**Часть 3. Класс `Apartment`.** Создайте класс `Apartment` с полями `address`, `area`, `is_sold`. В конструкторе **проверяйте** площадь: если она **не** число или **вне** диапазона `[5, 500]` — **выбрасывайте** `InvalidAreaError`.

**Часть 4. Класс `Agency`.** Создайте класс `Agency` с **методами**:

- `add(apartment)` — добавляет квартиру в **каталог**;
- `find_by_address(address)` — ищет квартиру, **возвращает** её или **выбрасывает** `ApartmentNotFoundError`;
- `sell(address)` — продаёт квартиру: если **уже продана** — `ApartmentAlreadySoldError`, если **не найдена** — `ApartmentNotFoundError`.

**Часть 5. Проверка.** Продемонстрируйте:

- создание квартиры с **корректной** площадью — успех;
- создание квартиры с **отрицательной** площадью — `InvalidAreaError`;
- создание квартиры со **строкой** вместо числа — `InvalidAreaError`;
- поиск **существующей** квартиры — успех;
- поиск **несуществующей** — `ApartmentNotFoundError`;
- продажу **непроданной** — успех;
- продажу **проданной** — `ApartmentAlreadySoldError`.

**Часть 6. Ловля по иерархии.** Продемонстрируйте, что:

- `except ApartmentNotFoundError` — ловит **конкретно**;
- `except AgencyError` — ловит **все** ошибки агентства;
- `except Exception` — ловит **вообще** всё.

**Часть 7. Дополнительные данные.** В обработчике `InvalidAreaError` **используйте** поля `area`, `min_area`, `max_area` для **вывода** подробной информации.

**Часть 8. Traceback с `from`.** Продемонстрируйте **разницу**:

- **без** `from` — traceback **теряет** оригинал;
- **с** `from e` — traceback **сохраняет** обе ошибки.

**Часть 9.** В комментариях объясните:

1. Зачем нужны **собственные** исключения вместо стандартных.
2. Почему **наследуют** от `Exception`, а не от `BaseException`.
3. Зачем **вызывать** `super().__init__(...)`.
4. Как работает **иерархия** исключений при **ловле**.
5. Зачем нужен **`from`** при **перебросе** исключений.
6. Когда наследовать от **стандартных** исключений, а когда — от `Exception`.

### Решение

```python
# ─── Часть 1. Базовое исключение ───
class AgencyError(Exception):
    """Базовое исключение для всех ошибок агентства."""
    pass


# ─── Часть 2. Конкретные исключения ───
class ApartmentNotFoundError(AgencyError):
    def __init__(self, address):
        self.address = address
        super().__init__(f"Квартира по адресу {address!r} не найдена")


class ApartmentAlreadySoldError(AgencyError):
    def __init__(self, address):
        self.address = address
        super().__init__(f"Квартира {address!r} уже продана")


class InvalidAreaError(AgencyError):
    def __init__(self, area, min_area=5, max_area=500):
        self.area = area
        self.min_area = min_area
        self.max_area = max_area
        super().__init__(
            f"Площадь {area} м² вне диапазона [{min_area}, {max_area}]"
        )

    def is_too_small(self):
        return self.area < self.min_area

    def is_too_large(self):
        return self.area > self.max_area

    def describe(self):
        if self.is_too_small():
            return f"Минимальная площадь: {self.min_area} м²"
        if self.is_too_large():
            return f"Максимальная площадь: {self.max_area} м²"
        return "Площадь в допустимом диапазоне"


# ─── Часть 3. Класс Apartment ───
class Apartment:
    def __init__(self, address, area):
        if not isinstance(area, (int, float)):
            raise InvalidAreaError(area)
        if not (5 <= area <= 500):
            raise InvalidAreaError(area)

        self.address = address
        self.area = area
        self.is_sold = False

    def __repr__(self):
        return f"Apartment({self.address!r}, {self.area!r})"


# ─── Часть 4. Класс Agency ───
class Agency:
    def __init__(self):
        self.apartments = []

    def add(self, apartment):
        self.apartments.append(apartment)

    def find_by_address(self, address):
        for flat in self.apartments:
            if flat.address == address:
                return flat
        raise ApartmentNotFoundError(address)

    def sell(self, address):
        flat = self.find_by_address(address)     # может выбросить NotFound
        if flat.is_sold:
            raise ApartmentAlreadySoldError(address)
        flat.is_sold = True
        return flat


# ─── Часть 5. Проверка ───
print("=== Создание квартир ===")

try:
    flat1 = Apartment("ул. Ленина, д. 10, кв. 5", 65)
    print(f"OK: {flat1}")
except InvalidAreaError as e:
    print(f"Ошибка: {e}")

try:
    flat_bad = Apartment("ул. Плохая", -10)
except InvalidAreaError as e:
    print(f"Ошибка: {e}")
    print(f"  area = {e.area}")
    print(f"  describe: {e.describe()}")

try:
    flat_bad2 = Apartment("ул. Плохая", "abc")
except InvalidAreaError as e:
    print(f"Ошибка: {e}")

# ─── Часть 5. Работа с агентством ───
print()
print("=== Работа с агентством ===")

agency = Agency()
agency.add(flat1)

try:
    found = agency.find_by_address("ул. Ленина, д. 10, кв. 5")
    print(f"Найдено: {found}")
except ApartmentNotFoundError as e:
    print(f"Ошибка: {e}")

try:
    agency.find_by_address("ул. Несуществующая")
except ApartmentNotFoundError as e:
    print(f"Ошибка: {e}")
    print(f"  address = {e.address}")

agency.sell("ул. Ленина, д. 10, кв. 5")
print("Продали квартиру")

try:
    agency.sell("ул. Ленина, д. 10, кв. 5")
except ApartmentAlreadySoldError as e:
    print(f"Ошибка: {e}")

# ─── Часть 6. Ловля по иерархии ───
print()
print("=== Ловля по иерархии ===")

try:
    agency.find_by_address("ул. Несуществующая")
except ApartmentNotFoundError:
    print("Поймано конкретно: ApartmentNotFoundError")

try:
    agency.find_by_address("ул. Несуществующая")
except AgencyError:
    print("Поймано по базовому: AgencyError")

try:
    agency.find_by_address("ул. Несуществующая")
except Exception:
    print("Поймано как Exception")

# ─── Часть 8. Traceback с from ───
print()
print("=== Traceback с from ===")

def parse_apartment_bad(data):
    try:
        address, area = data.split(";")
        return Apartment(address, int(area))
    except ValueError:
        # Без from — traceback потерян.
        raise AgencyError("Не удалось разобрать строку")

def parse_apartment_good(data):
    try:
        address, area = data.split(";")
        return Apartment(address, int(area))
    except ValueError as e:
        # С from — traceback сохранён.
        raise AgencyError("Не удалось разобрать строку") from e

try:
    parse_apartment_bad("только адрес")
except AgencyError as e:
    print(f"Без from: {e}")
    print(f"  __cause__ = {e.__cause__}")

try:
    parse_apartment_good("только адрес")
except AgencyError as e:
    print(f"С from: {e}")
    print(f"  __cause__ = {e.__cause__!r}")
```

**Ожидаемый вывод:**

```
=== Создание квартир ===
OK: Apartment('ул. Ленина, д. 10, кв. 5', 65)
Ошибка: Площадь -10 м² вне диапазона [5, 500]
  area = -10
  describe: Минимальная площадь: 5 м²
Ошибка: Площадь abc м² вне диапазона [5, 500]

=== Работа с агентством ===
Найдено: Apartment('ул. Ленина, д. 10, кв. 5', 65)
Ошибка: Квартира по адресу 'ул. Несуществующая' не найдена
  address = ул. Несуществующая
Продали квартиру
Ошибка: Квартира 'ул. Ленина, д. 10, кв. 5' уже продана

=== Ловля по иерархии ===
Поймано конкретно: ApartmentNotFoundError
Поймано по базовому: AgencyError
Поймано как Exception

=== Traceback с from ===
Без from: Не удалось разобрать строку
  __cause__ = None
С from: Не удалось разобрать строку
  __cause__ = ValueError("not enough values to unpack (expected 2, got 1)")
```

### Разбор

**1. Зачем нужны собственные исключения вместо стандартных.**

**Свои** исключения дают **три** преимущества:

- **Понятное** имя — `ApartmentNotFoundError` **говорит** само за себя.
- **Возможность** ловить **только** свою ошибку — `except ApartmentNotFoundError`.
- **Дополнительные** данные — `address`, `area`, `min_area`.

С `ValueError` вы **не знали** бы, **что** именно **пошло не так** — площадь, цена или адрес. С **своим** исключением — **всё понятно**.

**2. Почему наследуют от `Exception`, а не от `BaseException`.**

`BaseException` — **корень** иерархии. От него наследуют **системные** исключения — `SystemExit`, `KeyboardInterrupt`. Они **не должны** ловиться `except Exception` — иначе программа **не завершится** при `Ctrl+C`.

`Exception` — **основа** для **обычных** ошибок. Наследуя от него, **ваше** исключение:

- **Ловится** `except Exception`.
- **Не** конфликтует с **системными**.
- **Ведёт** себя как **стандартное**.

**Правило:** для **обычных** ошибок — `Exception`. Для **системных** — `BaseException` (но это **редко**).

**3. Зачем вызывать `super().__init__(...)`.**

`Exception.__init__` **сохраняет** аргументы в `self.args`. Если **не** вызвать — `str(e)` вернёт **пустоту**, а `e.args` будет **пустым** кортежем.

**Правильно:**

```python
def __init__(self, address):
    self.address = address
    super().__init__(f"Квартира {address!r} не найдена")    # ← вызываем
```

**Неправильно:**

```python
def __init__(self, address):
    self.address = address
    # super().__init__() забыли — сообщение не установлено
```

**Правило:** **всегда** вызывайте `super().__init__(...)` в **своём** конструкторе исключения.

**4. Как работает иерархия исключений при ловле.**

Python **сравнивает** тип исключения с типом в `except`. Если **тип** исключения **совпадает** или **является наследником** — `except` **срабатывает**.

**Пример:**

- `ApartmentNotFoundError` — **наследник** `AgencyError`.
- `except ApartmentNotFoundError` — **ловит** только `ApartmentNotFoundError`.
- `except AgencyError` — **ловит** **всех** наследников, включая `ApartmentNotFoundError`.
- `except Exception` — **ловит** вообще **всё**.

**Правило:** **чем выше** в иерархии — **тем шире** ловля.

**5. Зачем нужен `from` при перебросе исключений.**

Когда вы **ловите** одно исключение и **выбрасываете** другое — **оригинальный** traceback **теряется**. Это **затрудняет** отладку: непонятно, **что** было **причиной**.

`raise ... from e` **сохраняет** связь: `e` **становится** `__cause__` **нового** исключения. Traceback **покажет** **обе** ошибки:

```
Traceback (most recent call last):
  ...
ValueError: not enough values to unpack (expected 2, got 1)

The above exception was the direct cause of the following exception:

Traceback (most recent call last):
  ...
AgencyError: Не удалось разобрать строку
```

**Правило:** при **перебросе** исключения **всегда** используйте `from`.

**6. Когда наследовать от стандартных исключений, а когда — от `Exception`.**

**Наследуйте** от **стандартных**, если:

- Ваша ошибка — **разновидность** стандартной.
- Например, `InvalidAreaError` — **разновидность** `ValueError`.
- Хотите, чтобы код, **ловящий** `ValueError`, **ловил** и вашу ошибку.

**Наследуйте** от `Exception`, если:

- Ошибка **специфична** для вашей области.
- **Не** является **разновидностью** стандартных.
- Например, `ApartmentNotFoundError` — **не** `ValueError` и **не** `KeyError`.

**Пример:**

```python
class InvalidAreaError(ValueError):    # ← разновидность ValueError
    pass

class ApartmentNotFoundError(Exception):    # ← специфичная ошибка
    pass
```

**Правило:** думайте о **смысле** ошибки. Если она **разновидность** стандартной — наследуйте от неё. Если **специфична** — от `Exception`.

# Глава 18. Агрегация и ассоциация

## 18.1. Проблема: как объекты связаны друг с другом

За предыдущие главы мы разобрали **композицию** (глава 8) — когда один объект **содержит** другой как **часть себя**. Дом **состоит** из комнат. Книга **состоит** из страниц. Заказ **состоит** из позиций. Это отношение **«часть целого»**: без целого части **не существует**.

Но в реальных программах **не все** отношения такие **жёсткие**. Возьмём наше агентство недвижимости. Есть **арендатор**, который **снимает** квартиру. Есть **квартира**, которая **принадлежит** собственнику. Есть **собственник**, который **сдаёт** квартиру. Есть **агентство**, которое **представляет** квартиру. Как **описать** все эти связи?

**Первое**, что приходит в голову — **композиция**. Пусть `Tenant` **создаёт** квартиру внутри себя:

```python
class Tenant:
    def __init__(self, name):
        self.name = name
        self.apartment = Apartment("ул. Ленина", 65)
```

Но это **неправильно**. Квартира **не принадлежит** арендатору. Она **существует** отдельно. Если арендатор **уедет** — квартира **останется**. Если арендатор **переедет** в другую квартиру — старая **не исчезнет**. Это **не** отношение «часть целого».

Здесь **другое** отношение — **агрегация**. Арендатор **содержит** ссылку на квартиру, но **не владеет** ею. Квартира **создаётся** отдельно и **передаётся** в арендатора. Она **может** существовать **без** арендатора.

А теперь **другая** ситуация. Есть **арендатор** и **агентство**, которое сдаёт квартиру. Арендатор **знает** об агентстве, но **не владеет** им. И агентство **знает** об арендаторе, но **не владеет** им. Это **третье** отношение — **ассоциация**. Объекты **связаны**, но **не владеют** друг другом.

### Аналогия

Представь **библиотеку**. У неё есть **книги**. Библиотека **содержит** книги, но **не владеет** ими — книги могут **переходить** в другую библиотеку, могут быть **подарены**, могут быть **утеряны**. Это **агрегация**. Библиотека **не создаёт** книги **сама** — она их **получает**.

А ещё есть **читатель** и **книга**. Читатель **берёт** книгу, но **не владеет** ею. И книга **не знает** о читателе. Это **ассоциация** — связь между **независимыми** объектами.

И **третье**: у **книги** есть **обложка**. Обложка **создаётся** вместе с книгой и **не существует** отдельно. Это **композиция**.

### Три типа отношений

В ООП есть **три** основных типа отношений между объектами:

| Тип | Отношение | Связь | Существует отдельно? |
|-----|-----------|-------|----------------------|
| **Композиция** | «состоит из» | Сильная | ❌ Нет |
| **Агрегация** | «содержит» | Средняя | ✅ Да |
| **Ассоциация** | «использует» | Слабая | ✅ Да |

**Композиция** — разобрали в главе 8. **Агрегация** и **ассоциация** — эта глава.

### Почему это важно

Если **перепутать** типы отношений — код будет **работать неправильно**. Например:

- Если **квартиру** сделать **композицией** арендатора — при удалении арендатора квартира **исчезнет**. Хотя должна **остаться**.
- Если **агентство** сделать **композицией** квартиры — квартира **не сможет** быть в **двух** агентствах. Хотя может.
- Если **арендатора** сделать **композицией** агентства — при **реорганизации** агентства все арендаторы **исчезнут**. Хотя так **не должно** быть.

Правильный **выбор** типа отношения — **основа** правильной **архитектуры**. В этой главе разберём **всё** подробно.

### Вывод

Отношения между объектами бывают **трёх** типов: **композиция**, **агрегация**, **ассоциация**. Каждое **подходит** для **своих** ситуаций. Понимание **разницы** — **ключ** к правильному ООП.

---

## 18.2. Что такое агрегация

**Агрегация** — это отношение, при котором один объект **содержит** другой, но **не владеет** им. Внутренний объект **создаётся** отдельно и **может существовать** без внешнего.

### Синтаксис

```python
class Apartment:
    def __init__(self, address, area):
        self.address = address
        self.area = area


class Tenant:
    def __init__(self, name, apartment):
        self.name = name
        self.apartment = apartment      # ← ссылка на внешний объект
```

Разберём по частям:

- **`class Apartment`** — обычный класс квартиры. Он **не знает** о существовании арендаторов.
- **`class Tenant`** — класс арендатора. В конструкторе он **принимает** объект `apartment` **снаружи**.
- **`self.apartment = apartment`** — сохраняем **ссылку** на квартиру.

**Ключевой момент:** квартира **передаётся** в конструктор **извне**. Арендатор её **не создаёт**. Она **уже существует** к моменту создания арендатора.

### Использование

```python
# Сначала создаём квартиру — она существует сама по себе.
flat = Apartment("ул. Ленина, д. 10, кв. 5", 65)

# Потом создаём арендатора, передавая ему квартиру.
tenant = Tenant("Иванов И.И.", flat)

print(tenant.apartment.address)    # ул. Ленина, д. 10, кв. 5
print(flat.address)                # та же квартира
```

### Что происходит

Когда мы создаём арендатора, **квартира уже есть**. Арендатор **не создаёт** её — он **получает** её **снаружи**. То есть квартира **существует** **независимо** от арендатора.

### Разница с композицией

Сравним **две** реализации:

**Композиция:**

```python
class Tenant:
    def __init__(self, name):
        self.name = name
        self.apartment = Apartment("ул. Ленина", 65)    # ← создаётся ВНУТРИ
```

Здесь арендатор **сам создаёт** квартиру. Без арендатора квартиры **не будет**. Это **композиция**.

**Агрегация:**

```python
class Tenant:
    def __init__(self, name, apartment):
        self.name = name
        self.apartment = apartment                     # ← передаётся СНАРУЖИ
```

Здесь квартира **передаётся** в конструктор. Она **уже существует** и **может существовать** без арендатора. Это **агрегация**.

### Что происходит при удалении

**Композиция:**

```python
tenant = Tenant("Иванов")     # квартира создана вместе с арендатором
del tenant                     # квартира тоже удалится (если нет других ссылок)
```

**Агрегация:**

```python
flat = Apartment("ул. Ленина", 65)
tenant = Tenant("Иванов", flat)
del tenant                     # квартира ОСТАЁТСЯ — flat всё ещё ссылается на неё
print(flat.address)            # работает
```

### Ключевое правило

**Задайте вопрос:** «Если удалить **внешний** объект, должен ли **внутренний** исчезнуть?»

- **Да** → композиция.
- **Нет** → агрегация.

### Аналогия

Представь **команду** и **игроков**. Команда **содержит** игроков, но **не владеет** ими. Игрок может **перейти** в другую команду, **завершить** карьеру, **стать** тренером. Команда **не создаёт** игроков — она их **нанимает** (получает снаружи). Это **агрегация**.

### Когда использовать

Агрегация **идеальна**, когда:

- Один объект **использует** другой, но **не владеет** им.
- Внутренний объект **создаётся** отдельно.
- Внутренний объект **может** быть **общим** для нескольких внешних.
- При удалении внешнего внутренний **должен** остаться.

### Вывод

**Агрегация** — отношение «содержит, но не владеет». Внутренний объект **создаётся** отдельно и **может существовать** без внешнего. Отличие от композиции — в **момент создания** и **владении**.

---

## 18.3. Что такое ассоциация

**Ассоциация** — это самое **слабое** отношение. Объекты **знают** друг о друге, но **не содержат** и **не владеют** друг другом. Это просто **ссылка** между **независимыми** объектами.

### Синтаксис

```python
class Agency:
    def __init__(self, name):
        self.name = name
        self.apartments = []             # список — но это уже агрегация

    def add(self, apartment):
        self.apartments.append(apartment)


class Apartment:
    def __init__(self, address, agency=None):
        self.address = address
        self.agency = agency             # ← ассоциация: ссылка на агентство
```

Здесь:

- **`Agency`** — агентство. Оно **содержит** список квартир.
- **`Apartment`** — квартира. Она **знает** о своём агентстве через ссылку `agency`.
- **Связь двусторонняя**: агентство знает о квартирах, квартира знает об агентстве.

### Разница с агрегацией

**Агрегация** — это **владение** (пусть и слабое). Арендатор **содержит** квартиру. Это отношение **«содержит»**.

**Ассоциация** — это **знание** без **владения**. Квартира **знает** об агентстве, но **не содержит** его. Агентство **знает** о квартире, но **не содержит** её **как часть себя**. Это отношение **«использует»** или **«связан с»**.

### Пример ассоциации

```python
class Apartment:
    def __init__(self, address):
        self.address = address
        self.agency = None               # ещё не привязана к агентству


class Agency:
    def __init__(self, name):
        self.name = name
        self.apartments = []

    def register(self, apartment):
        # Устанавливаем ДВУСТОРОННЮЮ связь.
        apartment.agency = self
        self.apartments.append(apartment)
```

Теперь:

```python
agency = Agency("Дом-Сервис")
flat = Apartment("ул. Ленина, д. 10, кв. 5")

agency.register(flat)

print(flat.agency.name)              # Дом-Сервис
print(agency.apartments[0].address)  # ул. Ленина, д. 10, кв. 5
```

### Двусторонняя ассоциация

Ассоциация может быть **односторонней** или **двусторонней**.

**Односторонняя:** только один объект знает о другом.

```python
class Apartment:
    def __init__(self, address):
        self.address = address
        self.agency = None               # квартира знает об агентстве


class Agency:
    def __init__(self, name):
        self.name = name
        # агентство НЕ знает о квартирах
```

**Двусторонняя:** оба объекта знают друг о друге.

```python
class Apartment:
    def __init__(self, address):
        self.address = address
        self.agency = None

class Agency:
    def __init__(self, name):
        self.name = name
        self.apartments = []             # агентство знает о квартирах
```

### Когда использовать

Ассоциация **идеальна**, когда:

- Объекты **используют** друг друга **временно**.
- Один объект **знает** о другом, но **не содержит** его.
- Связь может **меняться** — например, квартира может быть **в одном** агентстве, потом **в другом**.
- Нужна **гибкость** — легко **разорвать** связь.

### Аналогия

Представь **врача** и **пациента**. Врач **знает** своих пациентов. Пациент **знает** своего врача. Но врач **не владеет** пациентом, а пациент **не владеет** врачом. Они **связаны**, но **независимы**. Это **ассоциация**.

### Вывод

**Ассоциация** — самое **слабое** отношение. Объекты **знают** друг о друге, но **не содержат** и **не владеют**. Может быть **односторонней** или **двусторонней**. Гибкая связь, легко **изменяется**.

---

## 18.4. Три типа отношений: подробное сравнение

Разберём **все три** типа отношений **подробно**. Это **ключевая** тема главы.

### Таблица сравнения

| Характеристика | Композиция | Агрегация | Ассоциация |
|----------------|-----------|-----------|------------|
| Отношение | «состоит из» | «содержит» | «использует» |
| Связь | Сильная | Средняя | Слабая |
| Создание внутреннего | Внутри внешнего | Передаётся снаружи | Не связано |
| Владение | Полное | Частичное | Нет |
| Существует отдельно | ❌ Нет | ✅ Да | ✅ Да |
| Может быть общим | ❌ Нет | ✅ Да | ✅ Да |
| При удалении внешнего | Внутренний удаляется | Внутренний остаётся | Оба независимы |

### Примеры из жизни

| Отношение | Пример |
|-----------|--------|
| **Композиция** | Дом и комната. Сердце и человек. Книга и страница. |
| **Агрегация** | Команда и игрок. Библиотека и книга. Университет и студент. |
| **Ассоциация** | Врач и пациент. Учитель и ученик. Водитель и машина. |

### Схема UML

В UML (языке моделирования) три типа отношений обозначаются **разными** символами:

```
        Композиция:  A ◆─── B     (закрашенный ромб)
        Агрегация:   A ◇─── B     (пустой ромб)
        Ассоциация:  A ─── B      (простая линия)
```

- **Закрашенный ромб** — композиция. Сильная связь.
- **Пустой ромб** — агрегация. Средняя связь.
- **Простая линия** — ассоциация. Слабая связь.

### Пример с квартирой, домом, арендатором

**Композиция:** дом и комната. Комната **не существует** без дома.

```python
class House:
    def __init__(self):
        self.rooms = [Room("Кухня"), Room("Спальня")]    # создаются внутри
```

**Агрегация:** арендатор и квартира. Квартира **существует** без арендатора.

```python
class Tenant:
    def __init__(self, name, apartment):
        self.apartment = apartment                       # передаётся снаружи
```

**Ассоциация:** квартира и агентство. Квартира **знает** об агентстве, но **не содержит** его.

```python
class Apartment:
    def __init__(self, address):
        self.agency = None                               # просто ссылка
```

### Как выбрать

**Задайте себе вопросы:**

1. **Может ли внутренний объект существовать отдельно?**
   - **Нет** → композиция.
   - **Да** → переходите к пункту 2.

2. **Владеет ли внешний объект внутренним?**
   - **Да** → агрегация.
   - **Нет** → ассоциация.

### Схема выбора

```
        Внутренний может существовать отдельно?
                ↓
        ┌───────┴───────┐
        │               │
        Нет             Да
        │               │
        ↓               ↓
    Композиция     Владеет им?
                        ↓
                ┌───────┴───────┐
                │               │
                Да              Нет
                │               │
                ↓               ↓
            Агрегация      Ассоциация
```

### Аналогия

Представь **семью**:

- **Сердце** — часть **человека**. Без человека — нет сердца. **Композиция**.
- **Домашний питомец** — принадлежит **хозяину**, но может **перейти** к другому. **Агрегация**.
- **Сосед** — знает хозяина, но **не принадлежит** ему. **Ассоциация**.

### Вывод

**Три типа отношений** — **три разных** способа связи. **Композиция** — сильная, **агрегация** — средняя, **ассоциация** — слабая. Выбор зависит от **смысла** отношения и **жизненного цикла** объектов.

---

## 18.5. Агрегация на практике

Разберём **пять** типичных случаев использования агрегации.

### 18.5.1. Команда и игроки

Классический пример агрегации — **спортивная команда** и **игроки**:

```python
class Player:
    def __init__(self, name, number):
        self.name = name
        self.number = number


class Team:
    def __init__(self, name):
        self.name = name
        self.players = []                # агрегация

    def add_player(self, player):
        self.players.append(player)

    def remove_player(self, player):
        self.players.remove(player)
```

Здесь:

- **Игрок** создаётся **отдельно**.
- **Команда** получает игроков **снаружи**.
- Игрок может **уйти** из команды — команда **не удаляется**.
- Игрок может **перейти** в другую команду.

### 18.5.2. Библиотека и книги

**Библиотека** содержит книги, но **не владеет** ими:

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author


class Library:
    def __init__(self, name):
        self.name = name
        self.books = []

    def add(self, book):
        self.books.append(book)

    def remove(self, book):
        self.books.remove(book)
```

Книга может быть **в двух** библиотеках, **подарена**, **продана** — библиотека **не управляет** этим.

### 18.5.3. Арендатор и квартира

**Арендатор** снимает квартиру, но **не владеет** ею:

```python
class Apartment:
    def __init__(self, address, price):
        self.address = address
        self.price = price


class Tenant:
    def __init__(self, name, apartment):
        self.name = name
        self.apartment = apartment
```

Арендатор может **съехать** — квартира **останется**. Квартира может быть **сдана** другому арендатору.

### 18.5.4. Университет и студенты

**Университет** содержит студентов, но **не владеет** ими:

```python
class Student:
    def __init__(self, name, group):
        self.name = name
        self.group = group


class University:
    def __init__(self, name):
        self.name = name
        self.students = []

    def enroll(self, student):
        self.students.append(student)

    def expel(self, student):
        self.students.remove(student)
```

Студент может **перейти** в другой университет, **закончить** обучение — университет **не удаляется**.

### 18.5.5. Наш дом

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area


class Furniture:
    def __init__(self, name, material):
        self.name = name
        self.material = material


class House:
    def __init__(self, address):
        self.address = address
        self.rooms = []                  # агрегация — комнаты создаются отдельно
        self.furniture = []              # агрегация — мебель создаётся отдельно

    def add_room(self, room):
        self.rooms.append(room)

    def add_furniture(self, item):
        self.furniture.append(item)
```

Здесь **комнаты** и **мебель** создаются **снаружи** и **передаются** в дом. Дом **содержит** их, но **не владеет**.

### Что общего у всех примеров

Во **всех** примерах:

- Внутренний объект **создаётся** отдельно.
- Внутренний объект **передаётся** снаружи.
- Внутренний объект **может существовать** без внешнего.
- Внутренний объект **может** быть **общим** для нескольких внешних.
- Удаление внешнего **не удаляет** внутренний.

### Вывод

Агрегация используется **везде**, где объект **содержит** другие, но **не владеет** ими. Это **самый частый** тип отношений **после** композиции.

---

## 18.6. Ассоциация на практике

Разберём **пять** типичных случаев использования ассоциации.

### 18.6.1. Врач и пациент

**Классический** пример ассоциации:

```python
class Doctor:
    def __init__(self, name):
        self.name = name
        self.patients = []               # знает о пациентах

    def add_patient(self, patient):
        self.patients.append(patient)
        patient.doctor = self            # двусторонняя связь


class Patient:
    def __init__(self, name):
        self.name = name
        self.doctor = None               # знает о враче
```

Врач и пациент **знают** друг о друге, но **не владеют** друг другом. Пациент может **сменить** врача. Врач может **иметь** много пациентов.

### 18.6.2. Учитель и ученик

```python
class Teacher:
    def __init__(self, name):
        self.name = name
        self.students = []


class Student:
    def __init__(self, name):
        self.name = name
        self.teachers = []               # может быть несколько учителей
```

Учитель и ученик — **независимые** объекты. Ученик может учиться у **нескольких** учителей. Учитель может учить **многих** учеников.

### 18.6.3. Водитель и машина

**Водитель** может **управлять** машиной, но **не владеет** ею:

```python
class Driver:
    def __init__(self, name):
        self.name = name
        self.car = None

    def drive(self, car):
        self.car = car
        car.driver = self


class Car:
    def __init__(self, model):
        self.model = model
        self.driver = None
```

Водитель может **сменить** машину. Машина может **перейти** к другому водителю.

### 18.6.4. Компания и сотрудник

**Сотрудник** может работать в **нескольких** компаниях:

```python
class Company:
    def __init__(self, name):
        self.name = name
        self.employees = []

    def hire(self, employee):
        self.employees.append(employee)
        employee.companies.append(self)


class Employee:
    def __init__(self, name):
        self.name = name
        self.companies = []
```

Здесь связь **многие-ко-многим** — сотрудник в **нескольких** компаниях, компания имеет **нескольких** сотрудников.

### 18.6.5. Наш дом

```python
class Agency:
    def __init__(self, name):
        self.name = name
        self.apartments = []

    def register(self, apartment):
        apartment.agency = self          # ассоциация
        self.apartments.append(apartment)


class Apartment:
    def __init__(self, address, price):
        self.address = address
        self.price = price
        self.agency = None               # связь с агентством
```

Квартира **знает** об агентстве, агентство **знает** о квартире. Но квартира **не принадлежит** агентству. Она может быть **в нескольких** агентствах.

### Что общего у всех примеров

Во **всех** примерах:

- Объекты **знают** друг о друге через **ссылки**.
- Связь **двусторонняя** (обычно).
- Объекты **независимы** — один может **существовать** без другого.
- Связь **может** быть **изменена** или **разорвана**.
- Нет **владения** — только **знание**.

### Вывод

Ассоциация используется **везде**, где объекты **используют** друг друга **временно** или **знают** друг о друге. Это **самый гибкий** тип связи.

---

## 18.7. Жизненный цикл объектов

**Жизненный цикл** — это **важная** характеристика, которая **отличает** три типа отношений. Разберём подробно.

### Композиция: одна судьба

При **композиции** внутренний объект **создаётся** вместе с внешним и **удаляется** вместе с ним. У них **одна судьба**.

```python
class House:
    def __init__(self):
        self.rooms = [Room("Кухня"), Room("Спальня")]


house = House()
del house          # комнаты тоже удаляются (если нет других ссылок)
```

### Агрегация: разные судьбы

При **агрегации** внутренний объект **создаётся** отдельно и **может** пережить внешний.

```python
flat = Apartment("ул. Ленина", 65)
tenant = Tenant("Иванов", flat)
del tenant         # квартира ОСТАЁТСЯ
print(flat.address)   # работает
```

### Ассоциация: полная независимость

При **ассоциации** объекты **полностью независимы**. Удаление одного **не влияет** на другой.

```python
agency = Agency("Дом-Сервис")
flat = Apartment("ул. Ленина", 65)
agency.register(flat)

del agency         # квартира ОСТАЁТСЯ
print(flat.address)   # работает
```

### Что происходит с ссылками

Важно понимать, что **удаление** переменной (`del`) **не означает** удаления **объекта**. Объект удаляется **только** когда на него **никто не ссылается**. Это **сборщик мусора** Python.

**Пример:**

```python
flat = Apartment("ул. Ленина", 65)
tenant = Tenant("Иванов", flat)

del flat           # переменная удалена, но объект ЖИВ — на него ссылается tenant
print(tenant.apartment.address)   # работает
```

Здесь `flat` — **переменная**. Она удалена. Но объект `Apartment` **жив**, потому что на него **ссылается** `tenant.apartment`.

### Схема жизненного цикла

```
        Композиция:      House ──── Room
                                (одна судьба)

        Агрегация:       Tenant ──── Apartment
                         (разные судьбы)

        Ассоциация:      Agency ──── Apartment
                         (полная независимость)
```

### Почему это важно

**Жизненный цикл** определяет, **как** правильно **проектировать** классы:

- Если **композиция** — внутренний объект **не должен** создаваться **снаружи**.
- Если **агрегация** — внутренний объект **должен** передаваться **снаружи**.
- Если **ассоциация** — связь **устанавливается** через **методы**, а не через конструктор.

### Аналогия

Представь **самолёт** и **пассажиров**:

- **Самолёт** и **крылья** — **композиция**. Крылья **не существуют** без самолёта.
- **Самолёт** и **пассажиры** — **агрегация**. Пассажиры **самостоятельны**, могут **пересесть** на другой рейс.
- **Пассажир** и **багаж** — **композиция**. Багаж **не существует** без пассажира.

### Вывод

**Жизненный цикл** — **ключевое** отличие трёх типов отношений. При композиции — **одна судьба**. При агрегации — **разные**. При ассоциации — **полная независимость**.

---

## 18.8. Агрегация и ассоциация в наследовании

Отношения **влияют** на **наследование**. Разберём подробно.

### Наследование классов, связанных агрегацией

Если у нас есть иерархия классов, связанных агрегацией, наследники **наследуют** эти связи:

```python
class Vehicle:
    def __init__(self, model):
        self.model = model
        self.engine = None               # агрегация

    def set_engine(self, engine):
        self.engine = engine


class Engine:
    def __init__(self, power):
        self.power = power


class Car(Vehicle):
    pass
```

Здесь `Car` **наследует** связь с двигателем. Каждая машина **содержит** двигатель, но **не владеет** им.

### Наследование классов, связанных ассоциацией

Аналогично с ассоциацией:

```python
class Person:
    def __init__(self, name):
        self.name = name
        self.doctor = None               # ассоциация

    def set_doctor(self, doctor):
        self.doctor = doctor


class Patient(Person):
    pass
```

Наследник `Patient` **наследует** связь с врачом.

### Переопределение связей

Наследник **может** переопределить **тип** связи:

```python
class Vehicle:
    def __init__(self):
        self.engine = Engine(200)        # композиция


class Truck(Vehicle):
    def __init__(self, engine):
        self.engine = engine             # агрегация — двигатель передаётся снаружи
```

Здесь `Vehicle` использует **композицию**, а `Truck` — **агрегацию**. Наследник **изменил** отношение.

**Но это не всегда хорошо.** Изменение типа связи может **нарушить** контракт родителя. Будьте **осторожны**.

### Ассоциация через `type(self)`

В методах **правильно** использовать `type(self)` вместо имени класса:

```python
class Agency:
    def register(self, apartment):
        apartment.agency = type(self)    # ← правильно, работает с наследниками
        # apartment.agency = Agency      # ← жёстко, не работает с наследниками
```

### Иерархия и агрегация

Можно строить **иерархии**, где **агрегация** используется для **составных** объектов:

```python
class Building:
    def __init__(self, address):
        self.address = address
        self.rooms = []                  # агрегация

    def add_room(self, room):
        self.rooms.append(room)


class House(Building):
    pass


class Office(Building):
    pass
```

Здесь `House` и `Office` **наследуют** агрегацию комнат от `Building`.

### Схема

```
        Building
        └── rooms (агрегация)
                │
        ┌───────┼───────┐
        ↓       ↓       ↓
        House   Office  ...
```

### Вывод

**Наследование** и **отношения** — **связанные** темы. Наследники **наследуют** связи родителя. Можно **переопределять** связи, но **осторожно**. Для **универсальности** используйте `type(self)`.

## 18.9. Агрегация и ассоциация в реальных проектах

Разберём **пять** примеров из **реальных** библиотек и **фреймворков**.

### 18.9.1. Django ORM

В Django **модели** связаны через **внешние ключи** (`ForeignKey`, `ManyToManyField`):

```python
class Apartment(models.Model):
    address = models.CharField(max_length=200)


class Tenant(models.Model):
    name = models.CharField(max_length=100)
    apartment = models.ForeignKey(Apartment, on_delete=models.SET_NULL, null=True)
```

Здесь `ForeignKey` с `on_delete=models.SET_NULL` — **агрегация**. Если квартира удалена, арендатор **остаётся**, но `apartment` становится `None`.

### 18.9.2. SQLAlchemy

SQLAlchemy использует **relationship** для связей:

```python
class Apartment(Base):
    __tablename__ = "apartments"
    id = Column(Integer, primary_key=True)


class Tenant(Base):
    __tablename__ = "tenants"
    id = Column(Integer, primary_key=True)
    apartment_id = Column(Integer, ForeignKey("apartments.id"))
    apartment = relationship("Apartment")    # ассоциация
```

`relationship` без `cascade="all, delete-orphan"` — **агрегация**. С `cascade` — **композиция**.

### 18.9.3. Игровые движки

В **играх** часто используется **агрегация** для **инвентаря** и **ассоциация** для **связей** между персонажами:

```python
class Player:
    def __init__(self):
        self.inventory = []              # агрегация — предметы существуют отдельно
        self.team = None                 # ассоциация — команда


class Item:
    def __init__(self, name):
        self.name = name


class Team:
    def __init__(self):
        self.players = []
```

### 18.9.4. Веб-фреймворки

В **веб-фреймворках** **сессии** связаны с **пользователями** через **ассоциацию**:

```python
class Session:
    def __init__(self, session_id):
        self.session_id = session_id
        self.user = None                 # ассоциация


class User:
    def __init__(self, name):
        self.name = name
        self.sessions = []
```

### 18.9.5. Наш дом

```python
class Agency:
    def __init__(self, name):
        self.name = name
        self.apartments = []             # агрегация
        self.agents = []                 # агрегация


class Apartment:
    def __init__(self, address):
        self.address = address
        self.agency = None               # ассоциация


class Agent:
    def __init__(self, name):
        self.name = name
        self.agency = None               # ассоциация
```

### Что общего у всех примеров

- **Агрегация** — для **коллекций** (списков) объектов.
- **Ассоциация** — для **двусторонних** связей.
- **Композиция** — для **неотъемлемых** частей.

### Вывод

**Агрегация** и **ассоциация** используются **везде**, где объекты **связаны**, но **независимы**. Понимание **разницы** — **основа** правильной **архитектуры**.

## 18.10. Нюансы и тонкости

Разберём **десять** нюансов.

### 18.10.1. Композиция и агрегация — не всегда очевидны

Иногда **сложно** определить, **композиция** это или **агрегация**. Например, **заказ** и **позиции заказа**:

- Если заказ **удалён** — позиции **тоже** удаляются. Это **композиция**.
- Если позиции могут **существовать** в **истории** — это **агрегация**.

**Правило:** думайте о **смысле**, а не о **синтаксисе**.

### 18.10.2. Ссылки не означают владение

Наличие **ссылки** на объект **не означает** владение. Квартира **ссылается** на агентство, но **не владеет** им. Это **ассоциация**, а не агрегация.

### 18.10.3. Агрегация через список — частый случай

Самая **частая** форма агрегации — **список** объектов:

```python
class Team:
    def __init__(self):
        self.players = []
```

Это **классическая** агрегация. Игроки создаются **отдельно**.

### 18.10.4. Циклические ссылки

При **ассоциации** часто возникают **циклические** ссылки:

```python
class Apartment:
    def __init__(self):
        self.agency = None

class Agency:
    def __init__(self):
        self.apartments = []
```

Здесь `Apartment.agency` ссылается на `Agency`, а `Agency.apartments` — на `Apartment`. Это **нормально** для Python — сборщик мусора **умеет** обрабатывать циклы.

### 18.10.5. Слабые ссылки

Для **ассоциации** иногда используют **слабые ссылки** (`weakref`), чтобы **не** продлевать жизнь объекта:

```python
import weakref

class Apartment:
    def __init__(self):
        self.agency = None

    def set_agency(self, agency):
        self.agency = weakref.ref(agency)
```

Слабая ссылка **не увеличивает** счётчик ссылок. Если на агентство **больше нет** ссылок — оно **удаляется**, а слабая ссылка становится `None`.

### 18.10.6. Композиция vs агрегация в API

При **проектировании API** важно **правильно** выбрать тип:

- **Композиция** — если объект **не имеет смысла** без другого.
- **Агрегация** — если объект **может существовать** отдельно.

### 18.10.7. Сериализация

При **сериализации** (в JSON, pickle) **агрегация** и **ассоциация** требуют **разных** подходов. При композиции объект **включается** целиком. При агрегации — **только ссылка** (id).

### 18.10.8. Наследование и связи

Наследник **наследует** связи родителя. Но **может** их **изменить**. Это **мощно**, но **опасно**.

### 18.10.9. Ассоциация многие-ко-многим

Связь **многие-ко-многим** — частый случай ассоциации:

```python
class Student:
    def __init__(self):
        self.courses = []

class Course:
    def __init__(self):
        self.students = []
```

Каждый студент **связан** с **несколькими** курсами. Каждый курс — с **несколькими** студентами.

### 18.10.10. Проверка на `None`

При **ассоциации** важно **проверять** на `None`:

```python
if self.agency is not None:
    print(self.agency.name)
```

Связь может быть **не установлена**.

### Вывод

**Агрегация** и **ассоциация** имеют **много** нюансов. Основные — **жизненный цикл**, **слабые ссылки**, **циклические связи**.

---

## 18.11. Типичные ошибки

Разберём **десять** самых частых ошибок.

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Путают композицию и агрегацию | Разное владение | Спросить: «исчезнет ли?» |
| Создают агрегируемый объект внутри | Это уже композиция | Передавать снаружи |
| Используют композицию вместо ассоциации | Жёсткая связь | Использовать ссылку |
| Забывают инициализировать ссылку | `AttributeError` | Установить `None` |
| Не проверяют на `None` | Ошибка при обращении | Проверять перед доступом |
| Дублируют ссылки | Путаница | Хранить в одном месте |
| Используют двустороннюю ассоциацию без нужды | Лишняя сложность | Односторонняя, если хватает |
| Забывают обновлять обратную ссылку | Несогласованность | Обновлять оба конца |
| Удаляют объект, не обновив ссылки | Висячие ссылки | Использовать `weakref` |
| Смешивают типы отношений в одном классе | Непонятно | Явно разделять |

### Подробный разбор

**Ошибка 1.** Путают композицию и агрегацию — самая частая. **Правило:** спросите, **исчезнет** ли внутренний объект при удалении внешнего.

**Ошибка 2.** Создают агрегируемый объект **внутри** — это **превращает** агрегацию в **композицию**. **Правило:** при агрегации — **передавайте** объект **снаружи**.

**Ошибка 3.** Композиция вместо ассоциации — **жёсткая** связь. **Правило:** если связь **временная** — ассоциация.

**Ошибка 4.** Забыли инициализировать ссылку — `AttributeError`. **Правило:** инициализируйте `None` в `__init__`.

**Ошибка 5.** Не проверяют на `None` — ошибка при **обращении**. **Правило:** всегда **проверяйте**.

**Ошибка 6.** Дублируют ссылки — **путаница**. **Правило:** храните **ссылку** в **одном** месте.

**Ошибка 7.** Двусторонняя ассоциация **без нужды** — лишняя сложность. **Правило:** используйте **одностороннюю**, если **достаточно**.

**Ошибка 8.** Забыли обновить **обратную** ссылку — **несогласованность**. **Правило:** обновляйте **оба** конца.

**Ошибка 9.** Удаляют объект, **не обновив** ссылки — **висячие** ссылки. **Правило:** используйте `weakref` или **обновляйте** вручную.

**Ошибка 10.** Смешивают типы отношений — **непонятно**. **Правило:** в **одном** классе — **один** тип для **одного** поля.

### Вывод

Ошибки с **агрегацией** и **ассоциацией** **типичны**. Следуя **правилам**, легко их **избежать**.

---

## 18.12. Что важно запомнить

- **Агрегация** — «содержит, но не владеет».
- **Ассоциация** — «использует», просто **ссылка**.
- **Композиция** — «состоит из», сильная связь.
- При **агрегации** объект **создаётся** отдельно.
- При **ассоциации** — только **ссылка**.
- **Жизненный цикл**: композиция — **одна судьба**, агрегация — **разные**, ассоциация — **независимость**.
- **Задайте вопрос:** «Исчезнет ли внутренний при удалении внешнего?»
- **Правило выбора:** существует отдельно? владеет? использует?
- **Список** — типичная форма агрегации.
- **Двусторонняя связь** — типичная форма ассоциации.
- **`weakref`** — для **слабых** ссылок.
- **`type(self)`** — для **наследуемости**.
- **Проверяйте** на `None`.

## Задача 18 (агрегация и ассоциация)

В агентстве недвижимости **арендатор** снимает квартиру. Квартира **не принадлежит** арендатору — она **существует** отдельно. Это **агрегация**. Дополнительно квартира **знает** о своём **агентстве** — это **ассоциация**.

**Часть 1. Класс `Apartment`.** Создайте класс `Apartment` с полями:

- `address` — адрес квартиры;
- `price` — цена аренды;
- `agency` — **ссылка** на агентство (по умолчанию `None`).

Реализуйте метод `__repr__`.

**Часть 2. Класс `Agency`.** Создайте класс `Agency` с полями:

- `name` — название агентства;
- `apartments` — **список** квартир (агрегация).

Реализуйте методы:

- `add_apartment(apartment)` — добавляет квартиру **и устанавливает обратную ссылку** через `apartment.agency = self`;
- `remove_apartment(apartment)` — удаляет квартиру **и обнуляет обратную ссылку** (`apartment.agency = None`).

**Часть 3. Класс `Tenant`.** Создайте класс `Tenant` с полями:

- `name` — имя арендатора;
- `apartment` — **ссылка** на квартиру (агрегация, **передаётся** через конструктор).

Реализуйте метод `__repr__`.

**Часть 4. Демонстрация агрегации.** Продемонстрируйте:

- создание **двух** квартир;
- создание **двух** арендаторов, каждый со **своей** квартирой;
- изменение `tenant1.apartment` — квартира **не удалилась**;
- удаление `tenant1` — квартира **осталась**;
- доступ к квартире через **другую** переменную.

**Часть 5. Демонстрация ассоциации.** Продемонстрируйте:

- создание агентства;
- добавление двух квартир через `add_apartment`;
- проверку **двусторонней** связи: `agency.apartments` содержит квартиры, `apartment.agency` указывает на агентство;
- удаление квартиры через `remove_apartment` — обратная ссылка **обнулилась**.

**Часть 6. Одна квартира — два арендатора.** Продемонстрируйте, что **одна** квартира может быть **передана** в **несколько** объектов:

- создайте одну квартиру;
- создайте `tenant1` и `tenant2` с **этой же** квартирой;
- убедитесь, что `tenant1.apartment is tenant2.apartment`.

**Часть 7. Одна квартира — два агентства.** Продемонстрируйте **ассоциацию** многие-ко-многим:

- создайте **два** агентства;
- создайте квартиру;
- добавьте в **оба** агентства;
- проверьте, что `apartment.agency` указывает на **последнее** (или используйте список).

**Часть 8. Жизненный цикл.** Продемонстрируйте **разницу**:

- при **композиции** — удаление родителя **удаляет** ребёнка;
- при **агрегации** — удаление арендатора **не удаляет** квартиру;
- при **ассоциации** — удаление агентства **не удаляет** квартиру.

**Часть 9.** В комментариях объясните:

1. Чем агрегация отличается от композиции.
2. Чем ассоциация отличается от агрегации.
3. Почему в задаче арендатор связан с квартирой через агрегацию.
4. Почему квартира связана с агентством через ассоциацию.
5. Почему **одна** квартира может быть связана с **несколькими** арендаторами.
6. Как **жизненный цикл** объектов влияет на выбор типа отношения.

### Решение

```python
# ─── Часть 1. Класс Apartment ───
class Apartment:
    def __init__(self, address, price):
        self.address = address
        self.price = price
        self.agency = None               # ассоциация: ссылка на агентство

    def __repr__(self):
        return f"Apartment({self.address!r}, {self.price!r})"


# ─── Часть 2. Класс Agency ───
class Agency:
    def __init__(self, name):
        self.name = name
        self.apartments = []             # агрегация: список квартир

    def add_apartment(self, apartment):
        # Устанавливаем ДВУСТОРОННЮЮ связь.
        apartment.agency = self
        self.apartments.append(apartment)

    def remove_apartment(self, apartment):
        # Разрываем связь с ОБЕИХ сторон.
        apartment.agency = None
        self.apartments.remove(apartment)

    def __repr__(self):
        return f"Agency({self.name!r}, {len(self.apartments)} квартир)"


# ─── Часть 3. Класс Tenant ───
class Tenant:
    def __init__(self, name, apartment):
        self.name = name
        self.apartment = apartment       # агрегация: квартира передаётся снаружи

    def __repr__(self):
        return f"Tenant({self.name!r}, {self.apartment.address!r})"


# ─── Часть 4. Демонстрация агрегации ───
print("=== Агрегация: арендатор и квартира ===")

flat1 = Apartment("ул. Ленина, д. 10, кв. 5", 50_000)
flat2 = Apartment("ул. Садовая, д. 3, кв. 12", 60_000)

tenant1 = Tenant("Иванов И.И.", flat1)
tenant2 = Tenant("Петров П.П.", flat2)

print(tenant1)
print(tenant2)

# Меняем квартиру у tenant1 — старая квартира НЕ удаляется.
tenant1.apartment = flat2
print(f"После смены: tenant1.apartment = {tenant1.apartment}")
print(f"Старая квартира жива: {flat1}")

# Удаляем tenant1 — квартира ОСТАЁТСЯ.
del tenant1
print(f"После удаления tenant1: flat1 = {flat1}, flat2 = {flat2}")

# ─── Часть 5. Демонстрация ассоциации ───
print()
print("=== Ассоциация: квартира и агентство ===")

agency = Agency("Дом-Сервис")
agency.add_apartment(flat1)
agency.add_apartment(flat2)

print(f"agency.apartments = {agency.apartments}")
print(f"flat1.agency = {flat1.agency}")
print(f"flat2.agency = {flat2.agency}")

# Удаляем квартиру из агентства — обратная ссылка обнуляется.
agency.remove_apartment(flat1)
print(f"После удаления flat1: flat1.agency = {flat1.agency}")

# ─── Часть 6. Одна квартира — два арендатора ───
print()
print("=== Одна квартира — два арендатора ===")

shared = Apartment("ул. Пушкина, д. 1, кв. 7", 40_000)
tenant_a = Tenant("Сидоров А.А.", shared)
tenant_b = Tenant("Кузнецов Б.Б.", shared)

print(f"tenant_a.apartment is tenant_b.apartment = "
      f"{tenant_a.apartment is tenant_b.apartment}")

# ─── Часть 7. Одна квартира — два агентства ───
print()
print("=== Одна квартира — два агентства ===")

agency1 = Agency("Первое")
agency2 = Agency("Второе")

shared_flat = Apartment("ул. Мира, д. 5", 55_000)
agency1.add_apartment(shared_flat)
agency2.add_apartment(shared_flat)

print(f"agency1.apartments содержит: {agency1.apartments}")
print(f"agency2.apartments содержит: {agency2.apartments}")
print(f"shared_flat.agency = {shared_flat.agency}")   # последнее — agency2

# ─── Часть 8. Жизненный цикл ───
print()
print("=== Жизненный цикл объектов ===")

# Композиция: удаление родителя удаляет ребёнка.
class House:
    def __init__(self, address):
        self.address = address
        self.rooms = [Room("Кухня"), Room("Спальня")]

class Room:
    def __init__(self, name):
        self.name = name

house = House("ул. Ленина, д. 10")
rooms_ref = house.rooms

del house
# Комнаты живы — на них ссылается rooms_ref.
print(f"После удаления house: rooms = {rooms_ref}")

# Агрегация: удаление арендатора НЕ удаляет квартиру.
tenant_x = Tenant("Тестов Т.Т.", flat2)
del tenant_x
print(f"После удаления tenant_x: flat2 = {flat2}")

# Ассоциация: удаление агентства НЕ удаляет квартиру.
agency_x = Agency("Тестовое")
agency_x.add_apartment(flat2)
del agency_x
print(f"После удаления agency_x: flat2 = {flat2}")
print(f"flat2.agency = {flat2.agency}")   # ссылка осталась на удалённое? нет — на объект
```

**Ожидаемый вывод:**

```
=== Агрегация: арендатор и квартира ===
Tenant('Иванов И.И.', 'ул. Ленина, д. 10, кв. 5')
Tenant('Петров П.П.', 'ул. Садовая, д. 3, кв. 12')
После смены: tenant1.apartment = Apartment('ул. Садовая, д. 3, кв. 12', 60000)
Старая квартира жива: Apartment('ул. Ленина, д. 10, кв. 5', 50000)
После удаления tenant1: flat1 = Apartment('ул. Ленина, д. 10, кв. 5', 50000), flat2 = Apartment('ул. Садовая, д. 3, кв. 12', 60000)

=== Ассоциация: квартира и агентство ===
agency.apartments = [Apartment('ул. Ленина, д. 10, кв. 5', 50000), Apartment('ул. Садовая, д. 3, кв. 12', 60000)]
flat1.agency = Agency('Дом-Сервис', 2 квартир)
flat2.agency = Agency('Дом-Сервис', 2 квартир)
После удаления flat1: flat1.agency = None

=== Одна квартира — два арендатора ===
tenant_a.apartment is tenant_b.apartment = True

=== Одна квартира — два агентства ===
agency1.apartments содержит: [Apartment('ул. Мира, д. 5', 55000)]
agency2.apartments содержит: [Apartment('ул. Мира, д. 5', 55000)]
shared_flat.agency = Agency('Второе', 1 квартир)

=== Жизненный цикл объектов ===
После удаления house: rooms = [<__main__.Room object at ...>, <__main__.Room object at ...>]
После удаления tenant_x: flat2 = Apartment('ул. Садовая, д. 3, кв. 12', 60000)
После удаления agency_x: flat2 = Apartment('ул. Садовая, д. 3, кв. 12', 60000)
flat2.agency = Agency('Тестовое', 1 квартир)
```

### Разбор

**1. Чем агрегация отличается от композиции.**

**Композиция** — внутренний объект **создаётся** вместе с внешним. Без внешнего **не существует**.

```python
class House:
    def __init__(self):
        self.rooms = [Room("Кухня")]    # создаётся внутри
```

**Агрегация** — внутренний объект **создаётся** отдельно и **передаётся** снаружи. Может существовать **без** внешнего.

```python
class Tenant:
    def __init__(self, apartment):
        self.apartment = apartment       # передаётся снаружи
```

**Ключевое** отличие — в **момент создания** и **жизненном цикле**.

**2. Чем ассоциация отличается от агрегации.**

**Агрегация** — это **владение** (пусть и слабое). Арендатор **содержит** квартиру.

**Ассоциация** — это **знание** без **владения**. Квартира **знает** об агентстве, но **не содержит** его.

**Разница:**

- При агрегации — объект **хранится** как **поле**.
- При ассоциации — объект **хранится** как **ссылка**, но не является **частью**.

**3. Почему арендатор связан с квартирой через агрегацию.**

Потому что:

- Арендатор **содержит** квартиру — она **является** частью его состояния.
- Квартира **создаётся** отдельно и **передаётся** в арендатора.
- Квартира **может существовать** без арендатора.

Если бы арендатор **создавал** квартиру сам — это была бы **композиция**. Но это **неправильно**: квартира **не принадлежит** арендатору.

**4. Почему квартира связана с агентством через ассоциацию.**

Потому что:

- Квартира **знает** об агентстве — она **ссылается** на него.
- Агентство **не является** частью квартиры.
- Квартира **может** быть в **нескольких** агентствах.
- Агентство **не владеет** квартирой.

Ассоциация **идеальна** для **временных** связей.

**5. Почему одна квартира может быть связана с несколькими арендаторами.**

Потому что **агрегация** позволяет **делить** объект между **несколькими** владельцами. Квартира **не создаётся** внутри арендатора — она **передаётся** ему. Другой арендатор может получить **ту же** квартиру.

**6. Как жизненный цикл объектов влияет на выбор типа отношения.**

**Жизненный цикл** — **ключевой** критерий:

- Если объекты **умирают вместе** — **композиция**.
- Если объекты **умирают по-разному** — **агрегация**.
- Если объекты **полностью независимы** — **ассоциация**.

**Правило:** спрашивайте: «Что **произойдёт** с внутренним объектом при удалении внешнего?»

- **Удалится** → композиция.
- **Останется** → агрегация.
- **Всё равно** → ассоциация.

# Глава 19. Множественное наследование и MRO

## 19.1. Проблема: когда одного родителя недостаточно

За предыдущие главы мы разобрали **обычное** наследование — когда класс наследует от **одного** родителя. `Kitchen` наследует от `Room`. `Studio` наследует от `Apartment`. `ApartmentNotFoundError` наследует от `AgencyError`. Всё это **просто** и **понятно**: у каждого класса — **один** родитель, **одна** цепочка наследования, **один** источник функциональности.

Но в реальной жизни так бывает **не всегда**. Иногда класс должен **совмещать** возможности **нескольких** родителей. Представь наше агентство недвижимости. Есть **агент** — человек, который **показывает** объекты клиентам. Есть **оценщик** — человек, который **оценивает** стоимость квартир. Это **два разных** специалиста с **разными** навыками.

А теперь представь, что в агентстве работает **универсальный специалист** — он **и агент, и оценщик**. Он **показывает** объекты **и** **оценивает** их стоимость. Как это описать в коде? Первое, что приходит в голову — **унаследовать** от **обоих** классов:

```python
class Agent:
    def get_role(self):
        return "Агент"

class Appraiser:
    def get_role(self):
        return "Оценщик"

class AgentAppraiser(Agent, Appraiser):
    pass
```

Здесь `AgentAppraiser` наследует **и** от `Agent`, **и** от `Appraiser`. Он **получает** методы **обоих**. Это **множественное наследование** — возможность класса иметь **нескольких** родителей.

**Звучит удобно.** Но здесь **есть проблема**. Оба родителя **имеют** метод `get_role()`. Какой из них **вызовется**? Тот, что в `Agent`? Или тот, что в `Appraiser`? **Как Python решает**, какой метод использовать?

Ответ — **MRO** (Method Resolution Order). Это **порядок**, в котором Python **ищет** методы при множественном наследовании. **Понимание** MRO — **ключ** к правильному использованию множественного наследования.

### Аналогия

Представь **студента**, который учится **на двух** факультетах — **математическом** и **физическом**. У него **два** набора преподавателей, **две** программы, **два** расписания. Когда он **задаёт** вопрос — кому? Математику? Физику? Зависит от **вопроса**.

В **Python** MRO **определяет**, к **кому** идти **сначала**. Это **порядок** «приоритета» родителей. Если метод **есть** у **первого** — используется **его**. Если **нет** — идём ко **второму**. И так **далее**.

### Почему это важно

**Множественное наследование** — **мощный**, но **опасный** инструмент. Если **не понимать** MRO, можно **получить** неожиданное **поведение**: вызовется **не тот** метод, **не та** версия, **не тот** родитель. Программа **запустится**, но **результат** будет **не тот**, что ожидался.

**MRO** — **фундамент** правильного множественного наследования. В этой главе разберём **всё** подробно: как работает **поиск методов**, что такое **проблема ромба**, как использовать **`super()`**, как **читать** `__mro__`.

### Вывод

Множественное наследование позволяет классу иметь **нескольких** родителей. Но **порядок поиска** методов (MRO) — **критичен** для правильной работы. Понимание MRO — **обязательно** для работы с множественным наследованием.

---

## 19.2. Что такое множественное наследование

**Множественное наследование** — это возможность класса наследовать **от нескольких** родителей **одновременно**. В Python это **разрешено** и **активно** используется. В отличие от некоторых других языков (например, Java), где можно наследовать только от **одного** класса, Python даёт **полную свободу**.

### Синтаксис

Родители указываются **через запятую** в скобках:

```python
class Agent:
    def show_apartment(self):
        return "Показываю квартиру"

    def get_role(self):
        return "Агент"


class Appraiser:
    def estimate_price(self):
        return "Оцениваю стоимость"

    def get_role(self):
        return "Оценщик"


class AgentAppraiser(Agent, Appraiser):
    pass
```

Здесь `AgentAppraiser` наследует **и** от `Agent`, **и** от `Appraiser`. Он получает **все** методы **обоих**:

```python
specialist = AgentAppraiser()

print(specialist.show_apartment())    # Показываю квартиру (из Agent)
print(specialist.estimate_price())    # Оцениваю стоимость (из Appraiser)
print(specialist.get_role())          # ??? — метод есть у ОБОИХ
```

### Что происходит с `get_role()`

Метод `get_role()` есть **и** у `Agent`, **и** у `Appraiser`. Какой вызовется? Python **не выбирает случайно** — он **идёт** по **MRO**. В данном случае **сначала** `Agent`, **потом** `Appraiser`. Поэтому вызовется **версия** `Agent`:

```python
print(specialist.get_role())          # Агент
```

**Почему:** в списке родителей `Agent` идёт **первым**: `class AgentAppraiser(Agent, Appraiser)`. Поэтому Python **сначала** ищет методы **в** `Agent`.

### Как это выглядит в памяти

```
        Agent                    Appraiser
        ├── show_apartment()     ├── estimate_price()
        └── get_role()           └── get_role()
                │                        │
                └──────────┬─────────────┘
                           ↓
                    AgentAppraiser
                    ├── show_apartment()   ← из Agent
                    ├── estimate_price()   ← из Appraiser
                    └── get_role()         ← из Agent (первый в списке)
```

### Что наследуется

Класс-наследник получает **все публичные** атрибуты и методы **всех** родителей:

- **Методы** — все, что есть у родителей.
- **Атрибуты класса** — все.
- **Свойства** (`@property`) — все.
- **Классовые методы** — все.
- **Статические методы** — все.

Если методы **совпадают** по имени — используется метод **того** родителя, который **идёт раньше** в MRO.

### Аналогия

Представь **швейцарский нож**. В нём есть **нож**, **открывашка**, **штопор**, **пила**. Каждый инструмент — **отдельная** функция. Нож **совмещает** их **все**. Это **множественное наследование**: один объект — **много** возможностей.

### Когда использовать

Множественное наследование **полезно**, когда:

- Класс должен **совмещать** **несколько** независимых ролей.
- Есть **миксины** — «примеси» с **узкой** функциональностью (см. главу 20).
- Нужна **гибкость** в **композиции** поведения.

### Когда НЕ использовать

Множественное наследование **опасно**, когда:

- Родители **имеют** **общие** методы с **разным** поведением.
- Иерархия **сложная** и **трудно** отследить MRO.
- Можно **заменить** **композицией** — предпочтительнее.

**Правило:** используйте множественное наследование **осторожно** и **только** когда это **оправдано**.

### Вывод

**Множественное наследование** позволяет классу иметь **несколько** родителей. Он получает **все** их методы. Если методы **совпадают** — используется метод **первого** родителя по MRO. Это **мощно**, но требует **понимания** MRO.

## 19.3. Проблема ромба (diamond problem)

**Проблема ромба** (diamond problem) — **классическая** проблема, возникающая при множественном наследовании. Она **называется** так, потому что **диаграмма** наследования **напоминает** ромб.

### Что это

Представь **иерархию**:

```python
class Person:
    def get_role(self):
        return "Человек"


class Agent(Person):
    def get_role(self):
        return "Агент"


class Appraiser(Person):
    def get_role(self):
        return "Оценщик"


class AgentAppraiser(Agent, Appraiser):
    pass
```

**Схема:**

```
            Person
           /      \
          /        \
       Agent      Appraiser
          \        /
           \      /
        AgentAppraiser
```

Это **ромб**: `Person` — **сверху**. `Agent` и `Appraiser` — **посередине**. `AgentAppraiser` — **снизу**.

### В чём проблема

Метод `get_role()` определён **и** в `Person`, **и** в `Agent`, **и** в `Appraiser`. Какой **вызвать** для `AgentAppraiser`?

- Версию из `Agent`?
- Версию из `Appraiser`?
- Версию из `Person`?

Если бы мы использовали **простое** наследование, вопрос **не возник** бы. Но при **множественном** — нужно **правило**.

### Как Python решает проблему

Python использует **алгоритм C3-линеаризации** — специальный **алгоритм**, который строит **MRO**. Он **гарантирует**:

- **Порядок** родителей **сохраняется**.
- Каждый класс **появляется** в MRO **один раз**.
- **Родители** идут **после** наследников.

Для нашего примера MRO будет:

```python
print(AgentAppraiser.__mro__)
# (<class 'AgentAppraiser'>, <class 'Agent'>, <class 'Appraiser'>, <class 'Person'>, <class 'object'>)
```

Порядок поиска:

1. `AgentAppraiser` — сам класс.
2. `Agent` — первый родитель.
3. `Appraiser` — второй родитель.
4. `Person` — общий родитель.
5. `object` — базовый класс.

**Метод `get_role()`** найдётся **в** `Agent` — это **первый** в MRO. Поэтому вызовется **версия** `Agent`:

```python
print(AgentAppraiser().get_role())    # Агент
```

### Почему это важно

Без **C3-линеаризации** могла бы быть **неоднозначность**:

- Идти в `Agent` или `Appraiser`?
- Идти в `Person` **до** или **после** родителей?

Python **решает** это **автоматически** через **алгоритм**. Но **программист** должен **понимать**, как это **работает**, чтобы **предсказывать** поведение.

### Схема MRO

```
        MRO AgentAppraiser:
        1. AgentAppraiser
        2. Agent
        3. Appraiser
        4. Person
        5. object
```

Python ищет метод **по этому порядку**. Первый **найденный** — используется.

### Аналогия

Представь **собрание**, где есть **директор**, **два заместителя** и **сотрудники**. Когда нужно **принять** решение — **спрашивают** сначала **директора**. Если **не может** — **первого заместителя**. Потом **второго**. Потом **сотрудников**. Это **порядок** MRO: **сверху вниз**, **слева направо**.

### Ромб с общим методом

Особая **сложность** — когда у **общего** родителя (`Person`) есть метод, **переопределённый** у **обоих** промежуточных родителей. Например, если `Person.get_role()` возвращает «Человек», но `Agent` и `Appraiser` **переопределяют** его — Python **выберет** версию `Agent`, потому что он **первый** в MRO.

### Вывод

**Проблема ромба** — **классическая** проблема множественного наследования. Python **решает** её через **C3-линеаризацию**, строя MRO. **Первое** совпадение в MRO — используется. Понимание MRO — **ключ** к правильному множественному наследованию.

---

## 19.4. Что такое MRO

**MRO** (Method Resolution Order) — это **порядок**, в котором Python **ищет** атрибуты и методы **при** обращении к объекту. Это **список** классов, **упорядоченный** по **приоритету**.

### Как посмотреть MRO

У **каждого** класса есть **атрибут** `__mro__` — **кортеж** классов в **порядке** поиска:

```python
class Person:
    pass

class Agent(Person):
    pass

class Appraiser(Person):
    pass

class AgentAppraiser(Agent, Appraiser):
    pass

print(AgentAppraiser.__mro__)
```

**Вывод:**

```
(<class 'AgentAppraiser'>, <class 'Agent'>, <class 'Appraiser'>, <class 'Person'>, <class 'object'>)
```

Также есть **метод** `mro()`:

```python
print(AgentAppraiser.mro())    # то же самое, но список
```

### Как читать MRO

MRO **читается** **слева направо**:

- **Первый** класс — **сам** класс.
- **Далее** — его **прямые** родители (в порядке **указания**).
- **Далее** — **родители** родителей.
- **Последний** — `object` — **базовый** класс **всех** классов.

### Пример с более сложной иерархией

```python
class A:
    pass

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass

print(D.__mro__)
# (D, B, C, A, object)
```

**Порядок:**

1. `D` — сам класс.
2. `B` — первый родитель.
3. `C` — второй родитель.
4. `A` — общий родитель `B` и `C`.
5. `object` — базовый.

### Правила MRO

Python **соблюдает** **три** правила при построении MRO:

1. **Наследник идёт перед родителем.** `D` идёт **перед** `B` и `C`.
2. **Порядок родителей сохраняется.** `B` идёт **перед** `C` (как указано в `class D(B, C)`).
3. **Общий родитель идёт после всех.** `A` идёт **после** `B` и `C`.

### C3-линеаризация

Алгоритм, который **строит** MRO, называется **C3-линеаризацией**. Он **сложный**, но **гарантирует**:

- **Консистентность** — MRO **не противоречит** себе.
- **Монотонность** — если класс идёт **раньше** в родителе, он идёт **раньше** в наследнике.
- **Единственность** — каждый класс **один раз** в MRO.

Если **построить** MRO **невозможно** (из-за **противоречий**) — Python **выдаст** ошибку `TypeError`.

### Пример ошибки

```python
class A:
    pass

class B(A):
    pass

class C(A, B):    # ← A раньше B, но B наследует A — противоречие!
    pass
```

**Ошибка:**

```
TypeError: Cannot create a consistent method resolution order (MRO) for bases A, B
```

Здесь `A` указан **раньше** `B`, но `B` **наследует** `A`. Это **невозможно**: `A` **должен** идти **после** `B`. Python **не может** построить MRO и **выдаёт ошибку**.

### Схема C3

```
        C3(A, B, C) =
            A +
            merge(MRO(B), MRO(C), [B, C])
```

**C3** — это **рекурсивный** алгоритм, который **объединяет** MRO родителей.

### Аналогия

Представь **очередь** на **приём** к **врачу**. Сначала **идут** люди **без** записи. Потом **по** записи. Потом **льготники**. **Порядок** — **фиксирован**. Так и MRO: **фиксированный** порядок **поиска**.

### Вывод

**MRO** — **порядок** поиска методов. Строится **C3-линеаризацией**. **Три** правила: наследник **перед** родителем, порядок **сохраняется**, общий родитель **после** всех. Если **построить** невозможно — **ошибка**.

---

## 19.5. `super()` при множественном наследовании

**`super()`** — **ключевой** инструмент при множественном наследовании. Он **позволяет** вызывать методы **следующего** класса в MRO.

### Обычное использование

При **простом** наследовании `super()` вызывает **родителя**:

```python
class Person:
    def greet(self):
        return "Привет, я Person"

class Agent(Person):
    def greet(self):
        parent = super().greet()
        return f"{parent} → Агент"
```

Здесь `super().greet()` **вызывает** `Person.greet()`.

### При множественном наследовании

При **множественном** наследовании `super()` **не** обязательно вызывает **прямого** родителя. Он вызывает **следующий** класс в **MRO**:

```python
class Person:
    def greet(self):
        return "Привет, я Person"

class Agent(Person):
    def greet(self):
        return f"Агент → {super().greet()}"

class Appraiser(Person):
    def greet(self):
        return f"Оценщик → {super().greet()}"

class AgentAppraiser(Agent, Appraiser):
    def greet(self):
        return f"Универсал → {super().greet()}"
```

**Что произойдёт:**

```python
specialist = AgentAppraiser()
print(specialist.greet())
```

**Вывод:**

```
Универсал → Агент → Оценщик → Привет, я Person
```

**Разберём пошагово:**

1. `AgentAppraiser.greet()` — вызывает `super().greet()`.
2. `super()` в контексте `AgentAppraiser` → **следующий** в MRO после `AgentAppraiser` — это **`Agent`**.
3. `Agent.greet()` — вызывает `super().greet()`.
4. `super()` в контексте `Agent` (для объекта `AgentAppraiser`) → **следующий** в MRO после `Agent` — это **`Appraiser`**.
5. `Appraiser.greet()` — вызывает `super().greet()`.
6. `super()` в контексте `Appraiser` → **следующий** после `Appraiser` — это **`Person`**.
7. `Person.greet()` — возвращает «Привет, я Person».
8. Результат **собирается** обратно: «Привет, я Person» → «Оценщик → ...» → «Агент → ...» → «Универсал → ...».

### Схема

```
        MRO AgentAppraiser:
        AgentAppraiser → Agent → Appraiser → Person → object

        super() идёт по ЭТОМУ порядку, а не по "прямым родителям".
```

### Кооперативное наследование

Такой **подход** называется **кооперативным** наследованием. Каждый класс **вызывает** `super()` — и **вызов** идёт **по всей** цепочке MRO. Это **позволяет** каждому классу **дополнить** поведение **родителей**.

**Правило:** если используете `super()` — **все** классы в **иерархии** должны использовать `super()`. Иначе **цепочка** **оборвётся**.

### Проблема, если не использовать `super()`

Если в **одном** из классов **не вызвать** `super()` — цепочка **оборвётся**:

```python
class Agent(Person):
    def greet(self):
        return "Агент"    # ← НЕ вызываем super()
```

Теперь `AgentAppraiser.greet()` вызовет `Agent.greet()`, но **не** `Appraiser.greet()` и **не** `Person.greet()`. **Цепочка** прервана.

### `super()` с аргументами

Иногда нужно **явно** указать, **откуда** начинать поиск:

```python
class AgentAppraiser(Agent, Appraiser):
    def greet(self):
        # Начать поиск после Appraiser.
        return super(Agent, self).greet()
```

Здесь `super(Agent, self)` — начать **после** `Agent` в MRO. Это **продвинутый** приём, используется **редко**.

### Аналогия

Представь **эстафету**. Каждый **бегун** передаёт **палочку** **следующему**. Первый бежит, потом передаёт второму, второй — третьему. **Эстафета** — это **цепочка** вызовов. `super()` — это **передача** палочки **следующему** по MRO.

### Вывод

**`super()`** при множественном наследовании **идёт** по **MRO**, а не по **прямым** родителям. Это **позволяет** каждому классу **дополнить** поведение. **Правило:** все классы **должны** использовать `super()` — иначе **цепочка** прервётся.

---

## 19.6. Миксины — введение

**Миксины** — **особый** случай множественного наследования. Это **«примеси»** — классы с **узкой** функциональностью, которые **не предназначены** для **самостоятельного** использования. Подробно они разбираются в **главе 20**, но **кратко** упомянем здесь, потому что они **тесно связаны** с MRO.

### Что такое миксин

**Миксин** — это **класс**, который:

- **Добавляет** одну **узкую** функциональность.
- **Не** используется **самостоятельно**.
- **Не** имеет **своего** состояния (обычно).
- **Подключается** через **множественное** наследование.

### Пример

```python
class LoggableMixin:
    def log(self, message):
        from datetime import datetime
        print(f"[{datetime.now()}] {self.__class__.__name__}: {message}")


class Apartment(LoggableMixin):
    def __init__(self, address):
        self.address = address


flat = Apartment("ул. Ленина")
flat.log("Создана квартира")
# [2024-01-15 10:30:00] Apartment: Создана квартира
```

Здесь `LoggableMixin` — **миксин**. Он **добавляет** метод `log()`. `Apartment` **использует** его через **наследование**.

### Почему миксин идёт первым

**Соглашение:** миксин **ставится** **первым** в списке родителей:

```python
class Apartment(LoggableMixin, SomeBase):
    pass
```

**Причина:** миксин **должен** **переопределить** методы **базового** класса. Если миксин **второй** — его методы **могут** не вызваться.

### Миксины и MRO

Миксин **встраивается** в MRO **перед** базовыми классами:

```python
class Apartment(LoggableMixin, Base):
    pass

print(Apartment.__mro__)
# (Apartment, LoggableMixin, Base, object)
```

**Методы** миксина **имеют** **приоритет** над методами базового класса.

### Вывод

**Миксины** — **особый** случай множественного наследования. Это **узкие** «примеси» функциональности. Ставятся **первыми** в списке родителей. Подробно — в **главе 20**.

---

## 19.7. Практические примеры множественного наследования

Разберём **пять** типичных примеров.

### 19.7.1. Агент и оценщик

**Классический** пример из нашей задачи:

```python
class Agent:
    def show_apartment(self):
        return "Показываю квартиру"

    def get_role(self):
        return "Агент"


class Appraiser:
    def estimate_price(self):
        return "Оцениваю стоимость"

    def get_role(self):
        return "Оценщик"


class AgentAppraiser(Agent, Appraiser):
    def get_role(self):
        # Явно вызываем обоих родителей.
        return f"{Agent.get_role(self)} + {Appraiser.get_role(self)}"
```

Здесь `AgentAppraiser` **совмещает** **две** роли. Метод `get_role()` **переопределён** — он **объединяет** оба значения.

```python
specialist = AgentAppraiser()
print(specialist.get_role())    # Агент + Оценщик
```

### 19.7.2. Сортировка и логирование

**Миксины** часто используются для **добавления** функциональности:

```python
class ComparableMixin:
    def __lt__(self, other):
        return self._key() < other._key()

    def __le__(self, other):
        return self._key() <= other._key()


class LoggableMixin:
    def log(self, message):
        print(f"[LOG] {message}")


class Apartment(ComparableMixin, LoggableMixin):
    def __init__(self, address, price):
        self.address = address
        self.price = price

    def _key(self):
        return self.price
```

Теперь `Apartment` **можно** сравнивать **и** **логировать**.

### 19.7.3. Веб-приложения

В **веб-фреймворках** (Django, Flask) множественное наследование используется **активно**:

```python
class LoginRequiredMixin:
    def dispatch(self, request, *args, **kwargs):
        if not request.user.is_authenticated:
            return redirect("/login")
        return super().dispatch(request, *args, **kwargs)


class ApartmentListView(LoginRequiredMixin, ListView):
    model = Apartment
    template_name = "apartments.html"
```

Здесь `LoginRequiredMixin` **добавляет** **проверку** авторизации. `ListView` — **базовый** класс из Django.

### 19.7.4. Игровые объекты

В **играх** множественное наследование **удобно** для **смешивания** способностей:

```python
class MovableMixin:
    def move(self, dx, dy):
        self.x += dx
        self.y += dy


class DamageableMixin:
    def take_damage(self, amount):
        self.hp -= amount


class Player(MovableMixin, DamageableMixin):
    def __init__(self, x, y, hp):
        self.x = x
        self.y = y
        self.hp = hp
```

Теперь `Player` **умеет** **и** двигаться, **и** получать **урон**.

### 19.7.5. Наш дом

```python
class OpenableMixin:
    def open(self):
        self.is_open = True

    def close(self):
        self.is_open = False


class LockableMixin:
    def lock(self):
        self.is_locked = True

    def unlock(self):
        self.is_locked = False


class Door(OpenableMixin, LockableMixin):
    def __init__(self):
        self.is_open = False
        self.is_locked = False
```

Здесь `Door` **совмещает** **две** функции: **открытие** и **запирание**.

### Что общего у всех примеров

- **Миксины** добавляют **узкую** функциональность.
- **Базовые** классы дают **основное** поведение.
- **Комбинация** создаёт **универсальный** класс.
- **MRO** определяет **порядок** вызовов.

### Вывод

Множественное наследование **активно** используется **везде**: в веб-фреймворках, играх, библиотеках. **Миксины** — **главный** способ применения.

---

## 19.8. Проблемы множественного наследования

Множественное наследование — **мощный**, но **опасный** инструмент. Разберём **пять** типичных проблем.

### 19.8.1. Неоднозначность

Когда **у двух** родителей **есть** метод с **одним** именем, **непонятно**, какой **использовать**. Python **решает** через MRO, но программист **может** не **предсказать** результат.

### 19.8.2. Проблема ромба

Мы уже разобрали **проблему ромба**. Если у **общего** родителя есть метод, **переопределённый** у обоих **промежуточных** — Python **выберет** **первый** по MRO. Это **может** быть **не то**, что **ожидалось**.

### 19.8.3. Сложность отладки

Чем **сложнее** иерархия — тем **труднее** **отследить**, откуда **взялся** метод. MRO **может** быть **длинным** и **запутанным**.

### 19.8.4. Хрупкость

Изменение в **одном** родителе **может** **сломать** наследника **неожиданным** образом. Это **хрупкость** — свойство, **нежелательное** в коде.

### 19.8.5. Необходимость `super()`

**Кооперативное** наследование **требует** использования `super()` **во всех** классах. Если **забыть** — **цепочка** **оборвётся**.

### Как избежать проблем

- **Использовать** множественное наследование **только** для **миксинов**.
- **Избегать** сложных иерархий.
- **Документировать** MRO.
- **Тестировать** все комбинации.

### Аналогия

Представь **команду** из **двух** тренеров. Один **учит** **защите**, другой — **атаке**. Игрок **слушает** обоих. Но если **тренеры** **противоречат** — **непонятно**, кого **слушать**. Множественное наследование — **та же** проблема.

### Вывод

Множественное наследование **мощно**, но **опасно**. Основные проблемы: **неоднозначность**, **ромб**, **сложность**, **хрупкость**, **необходимость `super()`**. **Правило:** используйте **осторожно**, **документируйте**, **тестируйте**.

---

## 19.9. Множественное наследование в реальных проектах

Разберём **пять** примеров.

### 19.9.1. Django CBV

Django Class-Based Views **активно** используют миксины:

```python
from django.views.generic import ListView
from django.contrib.auth.mixins import LoginRequiredMixin

class ApartmentListView(LoginRequiredMixin, ListView):
    model = Apartment
```

`LoginRequiredMixin` **добавляет** **проверку** авторизации. `ListView` — **базовый** класс.

### 19.9.2. Flask

Flask использует **множественное** наследование для **расширений**:

```python
class CustomView(MethodView, LoggableMixin):
    def get(self):
        ...
```

### 19.9.3. SQLAlchemy

SQLAlchemy использует **миксины** для **повторного** использования **колонок**:

```python
class TimestampMixin:
    created_at = Column(DateTime, default=datetime.utcnow)
    updated_at = Column(DateTime, onupdate=datetime.utcnow)


class Apartment(TimestampMixin, Base):
    __tablename__ = "apartments"
```

### 19.9.4. Игровые движки

Pygame, Panda3D используют **множественное** наследование для **игровых** объектов.

### 19.9.5. Наш дом

```python
class SerializableMixin:
    def to_dict(self):
        return {k: v for k, v in self.__dict__.items()}


class Apartment(SerializableMixin):
    def __init__(self, address, area):
        self.address = address
        self.area = area


flat = Apartment("ул. Ленина", 65)
print(flat.to_dict())    # {'address': 'ул. Ленина', 'area': 65}
```

### Что общего у всех примеров

- **Миксины** добавляют **узкую** функциональность.
- **Базовые** классы дают **основное** поведение.
- **MRO** определяет **порядок**.

### Вывод

Множественное наследование — **стандарт** для **больших** библиотек и **фреймворков**. **Миксины** — **главный** способ применения.

## 19.10. Нюансы и тонкости

Разберём **десять** нюансов.

### 19.10.1. `__mro__` и `mro()`

`__mro__` — **атрибут** (кортеж). `mro()` — **метод** (список). Разница **небольшая**, но **есть**.

### 19.10.2. `super()` без аргументов

В Python 3 `super()` **работает** без **аргументов** — Python **сам** подставляет **нужные**.

### 19.10.3. `super()` в `__init__`

При **множественном** наследовании **важно** вызывать `super().__init__()` — тогда **цепочка** инициализации **пройдёт** по **MRO**.

### 19.10.4. MRO и `__init__`

`__init__` **наследуется** по MRO. Если **не** вызвать `super().__init__()`, **родительские** атрибуты **не** создадутся.

### 19.10.5. Проблема с `object`

Все **классы** **наследуют** от `object`. В MRO `object` **идёт** **последним**.

### 19.10.6. Миксины без `__init__`

**Миксины** **обычно** **не** определяют `__init__`. Иначе **нарушается** **порядок** инициализации.

### 19.10.7. `super()` и `type(self)`

`super()` **использует** `type(self)` для **построения** MRO. Поэтому **важно**, чтобы все **классы** в **иерархии** использовали `super()`.

### 19.10.8. Множественное наследование и `@property`

**Свойства** (`@property`) **наследуются** как **обычные** методы. При **конфликтах** — **используется** первое по MRO.

### 19.10.9. `super()` и классовые методы

`super()` **работает** с **классовыми** методами. Это **полезно** для **фабричных** методов.

### 19.10.10. Сложные MRO и читаемость

**Сложные** MRO **ухудшают** читаемость. **Правило:** если MRO **трудно** читать — **упростите** иерархию.

### Вывод

Множественное наследование **имеет** **много** нюансов. Основные: **MRO**, **`super()`**, **инициализация**. Понимание нюансов — **обязательно**.

---

## 19.11. Типичные ошибки

Разберём **десять** самых частых ошибок.

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Забыли `super().__init__()` | Атрибуты не созданы | Вызвать в начале |
| Не используют `super()` | Цепочка прервана | Использовать везде |
| Миксин **вторым** | Методы не переопределяют | Миксин **первым** |
| Сложная иерархия | Трудно понять | Упростить |
| Миксин с `__init__` | Нарушает порядок | Избегать |
| `super(Agent, self)` без нужды | Излишне | `super()` без аргументов |
| Наследование ради **кода** | Нарушает логику | Использовать **композицию** |
| Дублирование методов | Неоднозначность | Явно переопределить |
| Игнорирование MRO | Непонятно поведение | Читать `__mro__` |
| Ромб без осознания | Неожиданный результат | Понимать MRO |

### Подробный разбор

**Ошибка 1.** Забыли `super().__init__()` — атрибуты родителей **не** создаются. **Правило:** всегда вызывайте `super().__init__(...)`.

**Ошибка 2.** Не используют `super()` — цепочка **обрывается**. **Правило:** если **кооперативное** наследование — **все** классы **должны** использовать `super()`.

**Ошибка 3.** Миксин **вторым** — методы **не** переопределяют. **Правило:** миксин — **первым** в списке.

**Ошибка 4.** Сложная иерархия — **трудно** понять. **Правило:** **упрощайте**.

**Ошибка 5.** Миксин с `__init__` — **нарушает** порядок. **Правило:** **не** определяйте `__init__` в миксинах.

**Ошибка 6.** `super(Agent, self)` без нужды — **излишне**. **Правило:** используйте `super()` **без** аргументов.

**Ошибка 7.** Наследование ради **кода** — **нарушает** логику «является». **Правило:** используйте **композицию**.

**Ошибка 8.** Дублирование методов — **неоднозначность**. **Правило:** **явно** переопределяйте.

**Ошибка 9.** Игнорирование MRO — **непонятно** поведение. **Правило:** читайте `__mro__`.

**Ошибка 10.** Ромб без осознания — **неожиданный** результат. **Правило:** понимайте MRO.

### Вывод

Ошибки с множественным наследованием **типичны**. Следуя **правилам**, легко их **избежать**.

---

## 19.12. Что важно запомнить

- **Множественное наследование** — класс с **несколькими** родителями.
- **MRO** — **порядок** поиска методов.
- **`__mro__`** — кортеж классов в порядке поиска.
- **C3-линеаризация** — алгоритм построения MRO.
- **Проблема ромба** — **классическая** проблема множественного наследования.
- **`super()`** идёт по **MRO**, а не по **прямым** родителям.
- **Кооперативное** наследование — **все** классы используют `super()`.
- **Миксины** — **первый** в списке родителей.
- **Миксины** — **узкие** «примеси» функциональности.
- **Правила MRO:** наследник **перед** родителем, порядок **сохраняется**, общий родитель **после** всех.
- **Сложная** иерархия — **плохо**.
- **Множественное** наследование — **осторожно**.

## Задача 19 (множественное наследование и MRO)

В агентстве работает **универсальный** специалист — он **и агент, и оценщик**. Он показывает объекты **и** оценивает их стоимость. Нужно реализовать это через **множественное** наследование и **разобраться** с MRO.

**Часть 1. Класс `Agent`.** Создайте класс `Agent` с методами:

- `show_apartment(apartment)` — возвращает строку `Агент показывает {apartment.address}`;
- `get_role()` — возвращает `Агент`.

**Часть 2. Класс `Appraiser`.** Создайте класс `Appraiser` с методами:

- `estimate_price(apartment)` — возвращает строку `Оценщик оценивает {apartment.address}`;
- `get_role()` — возвращает `Оценщик`.

**Часть 3. Класс `AgentAppraiser`.** Создайте класс `AgentAppraiser(Agent, Appraiser)` **без** переопределения методов. **Определите**, какой `get_role()` будет вызван при обращении к `AgentAppraiser().get_role()`, и **проверьте** это на практике.

**Часть 4. `__mro__`.** Выведите `AgentAppraiser.__mro__`. **Объясните** порядок: почему `Agent` идёт **раньше** `Appraiser`? Что будет, если **поменять** порядок родителей (`class AgentAppraiser(Appraiser, Agent)`)?

**Часть 5. Общий родитель (ромб).** Создайте **общий** родитель `Person` с методом `greet()`:

- `Person.greet()` — возвращает `Привет, я Person`;
- `Agent(Person)` — переопределяет `greet()` и возвращает `Агент: {super().greet()}`;
- `Appraiser(Person)` — переопределяет `greet()` и возвращает `Оценщик: {super().greet()}`;
- `AgentAppraiser(Agent, Appraiser)` — переопределяет `greet()` и возвращает `Универсал: {super().greet()}`.

**Продемонстрируйте**, что `AgentAppraiser().greet()` вызовет **все** методы по **MRO**. Объясните, **почему** порядок именно такой.

**Часть 6. Явный вызов.** Переопределите `get_role()` в `AgentAppraiser` так, чтобы он возвращал **обе** роли через **явный** вызов: `Агент + Оценщик`. **Сравните** с поведением без переопределения.

**Часть 7. Проблема неоднозначности.** Создайте **третий** класс `Consultant` с методом `get_role()`, возвращающим `Консультант`. Создайте `AgentAppraiserConsultant(Agent, Appraiser, Consultant)` и **проверьте**, какой `get_role()` вызовется. Объясните, **почему** именно этот.

**Часть 8.** В комментариях объясните:

1. Что такое **MRO** и **зачем** он нужен.
2. Как Python **выбирает** метод при **совпадении** имён.
3. Что такое **проблема ромба** и как Python её **решает**.
4. Как `super()` работает при **множественном** наследовании.
5. Почему **миксин** ставится **первым** в списке родителей.
6. В каких случаях **не стоит** использовать множественное наследование.

### Решение

```python
# ─── Часть 1. Класс Agent ───
class Agent:
    def show_apartment(self, apartment):
        return f"Агент показывает {apartment.address}"

    def get_role(self):
        return "Агент"


# ─── Часть 2. Класс Appraiser ───
class Appraiser:
    def estimate_price(self, apartment):
        return f"Оценщик оценивает {apartment.address}"

    def get_role(self):
        return "Оценщик"


# ─── Часть 3. Класс AgentAppraiser ───
class AgentAppraiser(Agent, Appraiser):
    pass


# ─── Вспомогательный класс Apartment ───
class Apartment:
    def __init__(self, address, price):
        self.address = address
        self.price = price

    def __repr__(self):
        return f"Apartment({self.address!r}, {self.price!r})"


# ─── Часть 4. __mro__ ───
print("=== Часть 3-4. MRO AgentAppraiser ===")
print(f"MRO: {AgentAppraiser.__mro__}")
print()

flat = Apartment("ул. Ленина, д. 10, кв. 5", 8_500_000)
specialist = AgentAppraiser()

print(f"show_apartment: {specialist.show_apartment(flat)}")
print(f"estimate_price: {specialist.estimate_price(flat)}")
print(f"get_role: {specialist.get_role()}")    # Агент — первый в MRO

# Меняем порядок родителей.
class AgentAppraiserReversed(Appraiser, Agent):
    pass

print(f"MRO reversed: {AgentAppraiserReversed.__mro__}")
print(f"get_role reversed: {AgentAppraiserReversed().get_role()}")   # Оценщик

# ─── Часть 5. Ромб ───
print()
print("=== Часть 5. Проблема ромба ===")

class Person:
    def greet(self):
        return "Привет, я Person"


class AgentPerson(Person):
    def greet(self):
        return f"Агент: {super().greet()}"


class AppraiserPerson(Person):
    def greet(self):
        return f"Оценщик: {super().greet()}"


class AgentAppraiserPerson(AgentPerson, AppraiserPerson):
    def greet(self):
        return f"Универсал: {super().greet()}"


print(f"MRO: {AgentAppraiserPerson.__mro__}")
print(f"greet: {AgentAppraiserPerson().greet()}")

# ─── Часть 6. Явный вызов ───
print()
print("=== Часть 6. Явный вызов get_role ===")

class AgentAppraiserExplicit(Agent, Appraiser):
    def get_role(self):
        # Явно вызываем обоих родителей.
        return f"{Agent.get_role(self)} + {Appraiser.get_role(self)}"


print(f"get_role: {AgentAppraiserExplicit().get_role()}")    # Агент + Оценщик

# ─── Часть 7. Три родителя ───
print()
print("=== Часть 7. Три родителя ===")

class Consultant:
    def get_role(self):
        return "Консультант"


class AgentAppraiserConsultant(Agent, Appraiser, Consultant):
    pass


print(f"MRO: {AgentAppraiserConsultant.__mro__}")
print(f"get_role: {AgentAppraiserConsultant().get_role()}")   # Агент
```

**Ожидаемый вывод:**

```
=== Часть 3-4. MRO AgentAppraiser ===
MRO: (<class '__main__.AgentAppraiser'>, <class '__main__.Agent'>, <class '__main__.Appraiser'>, <class 'object'>)

show_apartment: Агент показывает ул. Ленина, д. 10, кв. 5
estimate_price: Оценщик оценивает ул. Ленина, д. 10, кв. 5
get_role: Агент
MRO reversed: (<class '__main__.AgentAppraiserReversed'>, <class '__main__.Appraiser'>, <class '__main__.Agent'>, <class 'object'>)
get_role reversed: Оценщик

=== Часть 5. Проблема ромба ===
MRO: (<class '__main__.AgentAppraiserPerson'>, <class '__main__.AgentPerson'>, <class '__main__.AppraiserPerson'>, <class '__main__.Person'>, <class 'object'>)
greet: Универсал: Агент: Оценщик: Привет, я Person

=== Часть 6. Явный вызов get_role ===
get_role: Агент + Оценщик

=== Часть 7. Три родителя ===
MRO: (<class '__main__.AgentAppraiserConsultant'>, <class '__main__.Agent'>, <class '__main__.Appraiser'>, <class '__main__.Consultant'>, <class 'object'>)
get_role: Агент
```

### Разбор

**1. Что такое MRO и зачем он нужен.**

**MRO** (Method Resolution Order) — **порядок**, в котором Python **ищет** методы и атрибуты при **обращении** к объекту. Это **список** классов, **упорядоченный** по **приоритету**.

**Зачем нужен:**

- **Определяет**, какой метод **вызвать**, если имя **совпадает** у **нескольких** родителей.
- **Гарантирует**, что каждый класс **вызывается** **один раз**.
- **Позволяет** использовать `super()` **предсказуемо**.

**2. Как Python выбирает метод при совпадении имён.**

Python идёт **по MRO** **слева направо** и **берёт** **первый** найденный метод. В нашем примере:

```
MRO: AgentAppraiser → Agent → Appraiser → object
```

Метод `get_role()` есть **и** в `Agent`, **и** в `Appraiser`. Python **идёт** по MRO: `AgentAppraiser` — нет, `Agent` — **есть**! Используется **версия** `Agent`.

Если **поменять** порядок родителей — `Appraiser` **станет** **первым** в MRO. Тогда вызовется **версия** `Appraiser` (что мы **проверили** в `AgentAppraiserReversed`).

**3. Что такое проблема ромба и как Python её решает.**

**Проблема ромба** — ситуация, когда **два** класса наследуют от **общего** родителя, а **третий** наследует от **обоих**:

```
            Person
           /      \
       AgentPerson  AppraiserPerson
           \      /
       AgentAppraiserPerson
```

**Проблема:** какой `greet()` **вызвать** — из `AgentPerson` или `AppraiserPerson`?

**Решение** — **C3-линеаризация**. Python строит MRO так, чтобы:

- **Наследник** шёл **перед** родителем.
- **Порядок** родителей **сохранялся**.
- **Общий** родитель шёл **после** всех.

**MRO:**

```
AgentAppraiserPerson → AgentPerson → AppraiserPerson → Person → object
```

**При `super()`** вызов **идёт** по **всей** цепочке:

1. `AgentAppraiserPerson.greet()` → `super().greet()`.
2. `AgentPerson.greet()` → `super().greet()`.
3. `AppraiserPerson.greet()` → `super().greet()`.
4. `Person.greet()` → «Привет, я Person».

**Результат:** `Универсал: Агент: Оценщик: Привет, я Person`.

**4. Как `super()` работает при множественном наследовании.**

`super()` **не** вызывает **прямого** родителя. Он вызывает **следующий** класс в **MRO**. Это **позволяет** **каждому** классу **дополнить** поведение **родителей**.

**Ключевое:** `super()` **работает** только если **все** классы в **иерархии** **используют** его. Если **хотя бы** один **не** вызывает — **цепочка** **оборвётся**.

**5. Почему миксин ставится первым в списке родителей.**

**Миксин** — это **«примесь»** с **узкой** функциональностью. Он **должен** иметь **приоритет** над **базовыми** классами. Если поставить **вторым** — методы миксина **могут** не вызваться.

**Пример:**

```python
class Apartment(LoggableMixin, Base):
    pass
```

Здесь `LoggableMixin` — **первый**. Его методы **имеют** **приоритет**.

**6. В каких случаях не стоит использовать множественное наследование.**

**Не используйте**, если:

- **Можно** заменить **композицией**.
- Иерархия **сложная** и **трудно** **отследить** MRO.
- Родители **имеют** **конфликтующие** методы.
- Нет **чёткой** логики «является» **для** **всех** родителей.
- **Достаточно** **одного** родителя.

**Правило:** множественное наследование — **для** **миксинов** и **явных** **случаев**. Во **всех** **остальных** — **композиция**.

# Глава 20. `@dataclass` и миксины

## 20.1. Проблема: много шаблонного кода

За предыдущие главы мы построили класс `Apartment` и **много раз** его переписывали. Добавляли `__init__`, `__repr__`, `__eq__`, `__hash__`, свойства, классовые методы. И каждый раз писали **одно и то же** — только с **разными** именами полей. Взгляни на типичный класс:

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    def __repr__(self):
        return f"Apartment({self.address!r}, {self.area!r}, {self.price!r})"

    def __eq__(self, other):
        if not isinstance(other, Apartment):
            return NotImplemented
        return (self.address == other.address and
                self.area == other.area and
                self.price == other.price)

    def __hash__(self):
        return hash((self.address, self.area, self.price))
```

Здесь **три** поля. И **четыре** метода. Но в **каждом** методе — **одно и то же**: перечисление **тех же** полей. Если полей будет **десять** — придётся **десять** раз **повторить** их в конструкторе, в `__repr__`, в `__eq__`, в `__hash__`. Это **шаблонный** код — его **много**, он **скучен** и в нём **легко** ошибиться.

**Первое**, что приходит в голову — **написать генератор** кода. Но это **сложно** и **неудобно**. Python **предлагает** другое решение — декоратор `@dataclass`. Он **автоматически** создаёт `__init__`, `__repr__`, `__eq__` и другие методы **на основе** аннотаций типов. Мы **не пишем** шаблонный код — Python **делает** это **за нас**.

А ещё есть **вторая** проблема. Мы часто **добавляем** одну и ту же функциональность **разным** классам. Логирование. Сериализация. Сравнение. Валидация. Для этого **не хочется** создавать **общий** родитель — у классов **разная** природа. Здесь приходят на помощь **миксины** — «примеси» с **узкой** функциональностью.

### Аналогия

Представь **мебельный** цех. Раньше **каждый** шкаф делали **вручную**: пилили доски, сверлили отверстия, прикручивали ручки. Это **долго** и **одинаково**. Но появился **автоматический** станок — задаёшь **размеры**, и он **сам** собирает **стандартные** шкафы. Это **`@dataclass`**.

А ещё есть **отдельные** детали: **колёсики**, **полочки**, **подсветка**. Их можно **приделать** к **любому** шкафу — неважно, кухонный он или книжный. Это **миксины**.

### Вывод

`@dataclass` **избавляет** от **шаблонного** кода. **Миксины** дают **переиспользование** функциональности. Вместе они делают код **короче** и **чище**.

---

## 20.2. Что такое `@dataclass`

**`@dataclass`** — это **декоратор** из стандартного модуля `dataclasses`. Он **автоматически** создаёт **методы** для класса на основе **аннотаций типов** полей.

### Синтаксис

```python
from dataclasses import dataclass

@dataclass
class Apartment:
    address: str
    area: int
    price: int
```

**Всё.** Больше **ничего** писать **не нужно**. Python **сам** создаст:

- **`__init__`** — конструктор с параметрами `address`, `area`, `price`.
- **`__repr__`** — читаемое представление.
- **`__eq__`** — сравнение по всем полям.

### Как это выглядит в использовании

```python
flat = Apartment("ул. Ленина", 65, 8_500_000)
print(flat)                                   # Apartment(address='ул. Ленина', area=65, price=8500000)

flat2 = Apartment("ул. Ленина", 65, 8_500_000)
print(flat == flat2)                          # True — сравнение по полям
```

### Что происходит под капотом

Когда Python видит `@dataclass`, он:

1. **Читает** аннотации типов полей класса.
2. **Генерирует** `__init__` с параметрами **по** этим полям.
3. **Генерирует** `__repr__` с **выводом** всех полей.
4. **Генерирует** `__eq__` со **сравнением** всех полей.
5. **Заменяет** класс **новой** версией — с **добавленными** методами.

При этом **исходные** методы и атрибуты **сохраняются**. `@dataclass` **не** заменяет их, а **дополняет**.

### Почему это удобно

**Три** преимущества:

- **Короче** — не нужно писать шаблонный код.
- **Меньше** ошибок — Python **сам** создаёт методы.
- **Читаемее** — видно **только** поля, а не **повторяющиеся** методы.

### Какие методы создаются по умолчанию

| Метод | Создаётся? |
|-------|-----------|
| `__init__` | ✅ Всегда |
| `__repr__` | ✅ Всегда |
| `__eq__` | ✅ Всегда |
| `__hash__` | ⚠️ Только если `frozen=True` или `eq=False` |
| `__lt__`, `__gt__` и другие | ⚠️ Только если `order=True` |

Про **параметры** декоратора — в разделе 20.5.

### Аналогия

Представь **конструктор** LEGO. Ты **выбираешь** детали — и он **сам** собирает **машинку**. Тебе **не нужно** вручную **скреплять** каждую деталь. Так и `@dataclass`: ты **описываешь** поля — Python **сам** создаёт методы.

### Вывод

**`@dataclass`** — декоратор, создающий `__init__`, `__repr__`, `__eq__` **автоматически**. Избавляет от **шаблонного** кода. Делает класс **короче** и **чище**.

---

## 20.3. Как переписать `Apartment` через `@dataclass`

Сравним **две** версии — **обычную** и **через `@dataclass`**.

### Обычная версия

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    def __repr__(self):
        return f"Apartment({self.address!r}, {self.area!r}, {self.price!r})"

    def __eq__(self, other):
        if not isinstance(other, Apartment):
            return NotImplemented
        return (self.address == other.address and
                self.area == other.area and
                self.price == other.price)
```

**Три** метода. **Много** кода. Поля **повторяются** **трижды**.

### Версия через `@dataclass`

```python
from dataclasses import dataclass

@dataclass
class Apartment:
    address: str
    area: int
    price: int
```

**Три** поля. **Ноль** методов. Поля **указаны** **один раз**. Всё **остальное** — **автоматически**.

### Что одинаково

Обе версии **дают** объекты с **одинаковым** поведением:

```python
flat1 = Apartment("ул. Ленина", 65, 8_500_000)
flat2 = Apartment("ул. Ленина", 65, 8_500_000)

print(flat1)                          # Apartment(address='ул. Ленина', area=65, price=8500000)
print(flat1 == flat2)                 # True
print(flat1.address)                  # ул. Ленина
```

### Что различается

- **Длина** кода — в `@dataclass` **короче** в **разы**.
- **Читаемость** — в `@dataclass` видно **только** поля.
- **Расширяемость** — в `@dataclass` **легче** добавить **новое** поле.

### Как добавить новое поле

**Обычная версия** — нужно **дописать** поле в **четырёх** местах:

```python
def __init__(self, address, area, price, floor):    # 1
    self.address = address
    self.area = area
    self.price = price
    self.floor = floor                              # 2

def __repr__(self):
    return f"Apartment({self.address!r}, {self.area!r}, {self.price!r}, {self.floor!r})"    # 3

def __eq__(self, other):
    return (self.address == other.address and
            self.area == other.area and
            self.price == other.price and
            self.floor == other.floor)              # 4
```

**Версия через `@dataclass`** — нужно **дописать** поле **один** раз:

```python
@dataclass
class Apartment:
    address: str
    area: int
    price: int
    floor: int          # ← всё!
```

**Разница** — **очевидна**. Чем **больше** полей — тем **больше** выигрыш.

### Что если нужно добавить логику

`@dataclass` **не** мешает **добавлять** свои методы:

```python
@dataclass
class Apartment:
    address: str
    area: int
    price: int

    def price_per_sqm(self):
        return self.price / self.area

    def info(self):
        return f"{self.address}: {self.area} м², {self.price} ₽"
```

Всё **работает** как **обычно**.

### Аналогия

Представь **сборку** мебели **по инструкции** и **готовый** шкаф из **магазина**. Первый — **долго** и **муторно**. Второй — **сразу** готов. `@dataclass` — это **готовый** шкаф.

### Вывод

`@dataclass` **сокращает** код **в разы**. Особенно **заметно** при **большом** числе полей. При **добавлении** новых полей — **одна** строка вместо **четырёх**.

---

## 20.4. Поля `@dataclass`: значения по умолчанию

`@dataclass` **поддерживает** значения по умолчанию для полей. Синтаксис **простой**:

```python
from dataclasses import dataclass

@dataclass
class Apartment:
    address: str
    area: int = 30                    # значение по умолчанию
    price: int = 5_000_000            # значение по умолчанию
    is_sold: bool = False             # значение по умолчанию
```

Теперь можно создавать объекты **по-разному**:

```python
flat1 = Apartment("ул. Ленина")                              # area=30, price=5_000_000, is_sold=False
flat2 = Apartment("ул. Ленина", 65)                          # price=5_000_000, is_sold=False
flat3 = Apartment("ул. Ленина", 65, 8_500_000)               # is_sold=False
flat4 = Apartment("ул. Ленина", 65, 8_500_000, True)         # всё указано
```

### Порядок полей

**Важно:** поля со **значениями** по умолчанию **должны** идти **после** полей **без** значений. Это **стандартное** правило Python.

**Правильно:**

```python
@dataclass
class Apartment:
    address: str                  # обязательное
    area: int = 30                # со значением по умолчанию
    price: int = 5_000_000        # со значением по умолчанию
```

**Неправильно:**

```python
@dataclass
class Apartment:
    address: str = "неизвестно"   # со значением
    area: int                     # без значения — ошибка!
```

Python **выдаст** `TypeError: non-default argument 'area' follows default argument`.

### Изменяемые значения по умолчанию

**Проблема:** для **изменяемых** типов (списков, словарей) **нельзя** использовать обычное значение по умолчанию:

```python
@dataclass
class Apartment:
    address: str
    rooms: list = []              # ← ОШИБКА!
```

Python **выдаст** `ValueError: mutable default <class 'list'> for field rooms is not allowed`.

**Причина:** список был бы **один** для **всех** объектов. Изменение у **одного** объекта **изменило** бы **все**.

**Решение — `field(default_factory=...)`:**

```python
from dataclasses import dataclass, field

@dataclass
class Apartment:
    address: str
    rooms: list = field(default_factory=list)      # ← правильно
    tags: dict = field(default_factory=dict)       # ← правильно
```

`default_factory` — **функция**, которая **создаёт** новое значение **для каждого** объекта. Так **каждый** объект получает **свой** список.

### Пример

```python
flat1 = Apartment("ул. Ленина")
flat2 = Apartment("ул. Садовая")

flat1.rooms.append("Кухня")
print(flat1.rooms)    # ['Кухня']
print(flat2.rooms)    # [] — пустой, потому что свой
```

### Аналогия

Представь **бланк** заказа. В нём есть **обязательные** поля (имя, адрес) и **необязательные** (комментарий, скидка). Если не **заполнил** — берётся **значение** по умолчанию. `@dataclass` — **такой** бланк.

### Вывод

`@dataclass` **поддерживает** значения по умолчанию. Для **изменяемых** типов используется `field(default_factory=...)`. Порядок полей **важен**: сначала **обязательные**, потом **со** значениями.

---

## 20.5. Параметры `@dataclass`

Декоратор `@dataclass` **принимает** **несколько** параметров. Разберём **основные**.

### `frozen=True` — неизменяемые объекты

```python
@dataclass(frozen=True)
class Apartment:
    address: str
    area: int
```

Теперь объекты **нельзя** изменять:

```python
flat = Apartment("ул. Ленина", 65)
flat.area = 70           # FrozenInstanceError!
```

**Что даёт `frozen=True`:**

- Объект **неизменяем**.
- Автоматически **создаётся** `__hash__` — можно **использовать** в `set` и `dict`.
- **Безопасно** для **многопоточности**.

### `order=True` — сравнение

```python
@dataclass(order=True)
class Apartment:
    address: str
    area: int
    price: int
```

Теперь объекты **можно** сравнивать **через** `<`, `>`, `<=`, `>=`:

```python
flat1 = Apartment("ул. Ленина", 65, 8_500_000)
flat2 = Apartment("ул. Садовая", 80, 12_000_000)

print(flat1 < flat2)     # True — сначала по address, потом по area, потом по price
```

**Порядок** сравнения — **по полям** в **том порядке**, в котором они **объявлены**.

### `eq=False` — отключить `__eq__`

```python
@dataclass(eq=False)
class Apartment:
    address: str
    area: int
```

Теперь `__eq__` **не** создаётся. Объекты **сравниваются** по **идентичности** (как **обычные** объекты).

**Что даёт `eq=False`:**

- **Отключает** автоматическое сравнение.
- **Создаёт** `__hash__` на основе **идентичности** — можно **использовать** в `set`.

### `repr=False` — отключить `__repr__`

```python
@dataclass(repr=False)
class Apartment:
    address: str
    area: int
```

Теперь `__repr__` **не** создаётся. Используется **стандартный** из `object`.

### `init=False` — отключить `__init__`

```python
@dataclass(init=False)
class Apartment:
    address: str
    area: int
```

Теперь `__init__` **не** создаётся. Можно **написать** свой.

### `slots=True` — включить `__slots__`

Python 3.10+:

```python
@dataclass(slots=True)
class Apartment:
    address: str
    area: int
```

Теперь **автоматически** создаётся `__slots__` (см. главу 21). Это **экономит** память.

### Сводная таблица

| Параметр | Что делает |
|----------|-----------|
| `frozen=True` | Объект **неизменяем**, создаёт `__hash__` |
| `order=True` | Создаёт `__lt__`, `__gt__`, `__le__`, `__ge__` |
| `eq=False` | Отключает `__eq__`, создаёт `__hash__` по идентичности |
| `repr=False` | Отключает `__repr__` |
| `init=False` | Отключает `__init__` |
| `slots=True` | Создаёт `__slots__` |

### Аналогия

Представь **конструктор** LEGO с **настройками**. Хочешь **колёсики** — включаешь. Хочешь **подсветку** — включаешь. Хочешь **без** фар — отключаешь. `@dataclass` — **такой** конструктор.

### Вывод

`@dataclass` **принимает** **шесть** основных параметров. Они **включают** или **отключают** разные методы. `frozen` и `order` — самые **полезные**.

---

## 20.6. Что такое миксин

**Миксин** (mixin) — это **класс** с **узкой** функциональностью, который **не предназначен** для **самостоятельного** использования. Он **подключается** к **другим** классам **через** множественное наследование (см. главу 19), чтобы **добавить** им **одну** возможность.

### Ключевые характеристики

- **Узкая** функциональность — **одна** задача.
- **Не** используется **самостоятельно**.
- **Не** имеет **своего** состояния (обычно).
- **Не** определяет `__init__` (обычно).
- **Подключается** **первым** в списке родителей.

### Пример

```python
class LoggableMixin:
    def log(self, message):
        from datetime import datetime
        print(f"[{datetime.now()}] {self.__class__.__name__}: {message}")


class Apartment(LoggableMixin):
    def __init__(self, address):
        self.address = address


flat = Apartment("ул. Ленина")
flat.log("Создана квартира")
```

**Вывод:**

```
[2024-01-15 10:30:00] Apartment: Создана квартира
```

### Как это работает

`LoggableMixin` **добавляет** метод `log()`. `Apartment` **наследует** от него и **получает** этот метод. При этом `Apartment` **не теряет** своей **природы** — это **по-прежнему** квартира.

### Чем миксин отличается от обычного родителя

| Характеристика | Обычный родитель | Миксин |
|----------------|------------------|--------|
| **Смысл** | «является» | «умеет» |
| **Функциональность** | Основная | Дополнительная |
| **Состояние** | Обычно есть | Обычно нет |
| **`__init__`** | Обычно есть | Обычно нет |
| **Самостоятельное использование** | ✅ Да | ❌ Нет |
| **Позиция в наследовании** | Любая | **Первым** |

**Пример:**

- `Kitchen(Room)` — **обычный** родитель. Кухня **является** комнатой.
- `Apartment(LoggableMixin)` — **миксин**. Квартира **умеет** логировать.

### Зачем миксины

Миксины **позволяют** **добавлять** функциональность **разным** классам **без** создания **общей** иерархии. Например, **логирование** нужно **и** `Apartment`, **и** `Agency`, **и** `Tenant`. Но они **не** связаны **общим** родителем. Миксин **решает** эту проблему.

### Аналогия

Представь **одежду**. Есть **базовая** рубашка. А есть **аксессуары**: галстук, шарф, значок. Их можно **надеть** на **любую** рубашку. Это **миксины**. Они **добавляют** **стиль**, но **не** меняют **суть** рубашки.

### Вывод

**Миксин** — класс с **узкой** функциональностью. **Не** используется **самостоятельно**. **Подключается** **первым**. Добавляет **одну** возможность **без** создания **общей** иерархии.

---

## 20.7. Миксин `LoggableMixin`

Разберём **типичный** миксин — **логирование**.

### Реализация

```python
from datetime import datetime


class LoggableMixin:
    def log(self, message):
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        print(f"[{timestamp}] {self.__class__.__name__}: {message}")

    def log_error(self, error):
        self.log(f"ОШИБКА: {error}")
```

### Использование

```python
class Apartment(LoggableMixin):
    def __init__(self, address, area):
        self.address = address
        self.area = area

    def sell(self):
        self.log(f"Продаём квартиру {self.address}")
        # ... логика продажи ...


flat = Apartment("ул. Ленина", 65)
flat.sell()
```

**Вывод:**

```
[2024-01-15 10:30:00] Apartment: Продаём квартиру ул. Ленина
```

### Что важно

- **`self.__class__.__name__`** — **имя** класса **автоматически**. Работает **для** **всех** наследников.
- **`self.log(...)`** — **вызов** метода из **миксина**. Работает **у** **всех** классов, **подключивших** миксин.
- **Миксин** **не** имеет `__init__` — **не** нарушает **порядок** инициализации.

### Почему `self.__class__.__name__`

Если **написать** жёстко `Apartment`, то **наследники** будут **логировать** неправильно:

```python
class Studio(Apartment):
    pass

studio = Studio("ул. Пушкина", 25)
studio.sell()
# [2024-01-15 10:30:00] Apartment: ...  ← неправильно, должен быть Studio
```

С `self.__class__.__name__` — **правильно**:

```
[2024-01-15 10:30:00] Studio: Продаём квартиру ул. Пушкина
```

### Расширение миксина

Можно **добавить** **больше** методов:

```python
class LoggableMixin:
    def log(self, message, level="INFO"):
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        print(f"[{timestamp}] [{level}] {self.__class__.__name__}: {message}")

    def log_info(self, message):
        self.log(message, "INFO")

    def log_warning(self, message):
        self.log(message, "WARNING")

    def log_error(self, message):
        self.log(message, "ERROR")
```

Теперь можно **логировать** с **уровнями**.

### Аналогия

Представь **плеер** с **наушниками**. Наушники — **миксин**. Плеер **работает** **сам** по себе. Наушники **добавляют** **возможность** слушать **музыку** **приватно**. Но **без** плеера наушники **бесполезны**. Так и `LoggableMixin` — **без** класса-носителя он **не** имеет **смысла**.

### Вывод

**`LoggableMixin`** — **типичный** миксин для **логирования**. Использует `self.__class__.__name__` для **универсальности**. **Не** имеет `__init__`. Легко **расширяется**.

---

## 20.8. Другие полезные миксины

Разберём **пять** типичных миксинов.

### 20.8.1. `ComparableMixin` — сравнение

```python
class ComparableMixin:
    def __lt__(self, other):
        return self._compare_key() < other._compare_key()

    def __le__(self, other):
        return self._compare_key() <= other._compare_key()

    def __gt__(self, other):
        return self._compare_key() > other._compare_key()

    def __ge__(self, other):
        return self._compare_key() >= other._compare_key()
```

Использование:

```python
class Apartment(ComparableMixin):
    def __init__(self, address, price):
        self.address = address
        self.price = price

    def _compare_key(self):
        return self.price


flat1 = Apartment("ул. Ленина", 8_500_000)
flat2 = Apartment("ул. Садовая", 12_000_000)

print(flat1 < flat2)     # True
```

### 20.8.2. `SerializableMixin` — сериализация

```python
class SerializableMixin:
    def to_dict(self):
        return {
            key: value
            for key, value in self.__dict__.items()
            if not key.startswith("_")
        }

    def to_json(self):
        import json
        return json.dumps(self.to_dict())
```

Использование:

```python
class Apartment(SerializableMixin):
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price


flat = Apartment("ул. Ленина", 65, 8_500_000)
print(flat.to_dict())    # {'address': 'ул. Ленина', 'area': 65, 'price': 8500000}
```

### 20.8.3. `ValidatableMixin` — валидация

```python
class ValidatableMixin:
    def validate(self):
        errors = []
        for field, rules in self._validation_rules().items():
            value = getattr(self, field)
            for rule in rules:
                if not rule(value):
                    errors.append(f"{field}: не прошло проверку")
        if errors:
            raise ValueError("; ".join(errors))

    def _validation_rules(self):
        return {}
```

Использование:

```python
class Apartment(ValidatableMixin):
    def __init__(self, address, area):
        self.address = address
        self.area = area

    def _validation_rules(self):
        return {
            "address": [lambda x: isinstance(x, str) and len(x) > 0],
            "area": [lambda x: 5 <= x <= 500],
        }


flat = Apartment("ул. Ленина", 65)
flat.validate()    # ок

flat_bad = Apartment("", -10)
flat_bad.validate()    # ValueError
```

### 20.8.4. `TimestampMixin` — временные метки

```python
from datetime import datetime


class TimestampMixin:
    def set_created(self):
        self.created_at = datetime.now()

    def set_updated(self):
        self.updated_at = datetime.now()
```

Использование:

```python
class Apartment(TimestampMixin):
    def __init__(self, address):
        self.address = address
        self.set_created()


flat = Apartment("ул. Ленина")
print(flat.created_at)    # 2024-01-15 10:30:00
```

### 20.8.5. `ReprMixin` — красивое представление

```python
class ReprMixin:
    def __repr__(self):
        fields = ", ".join(
            f"{k}={v!r}" for k, v in self.__dict__.items()
            if not k.startswith("_")
        )
        return f"{self.__class__.__name__}({fields})"
```

### Что общего у всех миксинов

- **Узкая** функциональность.
- **Один** метод или **небольшая** группа.
- **Нет** `__init__`.
- **Нет** своего состояния.
- Работает **через** `self`.

### Вывод

Миксины **могут** добавлять: **сравнение**, **сериализацию**, **валидацию**, **временные метки**, **представление**. Все они — **узкие** и **переиспользуемые**.

---

## 20.9. `@dataclass` и миксины вместе

`@dataclass` и **миксины** **отлично** сочетаются. Разберём **два** подхода.

### Подход 1: миксин **первым**

```python
from dataclasses import dataclass
from datetime import datetime


class LoggableMixin:
    def log(self, message):
        timestamp = datetime.now().strftime("%H:%M:%S")
        print(f"[{timestamp}] {self.__class__.__name__}: {message}")


@dataclass
class Apartment(LoggableMixin):
    address: str
    area: int
    price: int


flat = Apartment("ул. Ленина", 65, 8_500_000)
flat.log("Создана квартира")
print(flat)    # Apartment(address='ул. Ленина', area=65, price=8500000)
```

**Что происходит:**

- `@dataclass` **создаёт** `__init__`, `__repr__`, `__eq__`.
- `LoggableMixin` **добавляет** `log()`.
- **Миксин** идёт **первым** — методы **имеют** **приоритет**.

### Подход 2: несколько миксинов

```python
class LoggableMixin:
    def log(self, message):
        print(f"[LOG] {self.__class__.__name__}: {message}")


class SerializableMixin:
    def to_dict(self):
        return {k: v for k, v in self.__dict__.items() if not k.startswith("_")}


@dataclass
class Apartment(LoggableMixin, SerializableMixin):
    address: str
    area: int
    price: int


flat = Apartment("ул. Ленина", 65, 8_500_000)
flat.log("Создана квартира")
print(flat.to_dict())    # {'address': 'ул. Ленина', 'area': 65, 'price': 8500000}
```

**Порядок** миксинов **важен** — они **идут** в MRO **слева** **направо**.

### Нюанс с `@dataclass` и `frozen`

Если используете `frozen=True`, **миксины** **должны** **учитывать** это. **Методы** миксина **не** должны **изменять** состояние:

```python
@dataclass(frozen=True)
class Apartment(LoggableMixin):
    address: str
    area: int


flat = Apartment("ул. Ленина", 65)
flat.log("Тест")       # ок — log() не меняет состояние
flat.area = 70         # FrozenInstanceError!
```

### Нюанс с `slots=True`

Если `@dataclass(slots=True)`, **миксин** **не** должен **использовать** **дополнительные** атрибуты:

```python
@dataclass(slots=True)
class Apartment(LoggableMixin):
    address: str
    area: int

    # Нельзя добавить поле вне __slots__!
```

### Аналогия

Представь **готовый** шкаф из **магазина** (`@dataclass`) и **аксессуары** **к** нему (миксины). Можно **приделать** **подсветку** (LoggableMixin), **колесики** (SerializableMixin), **замок** (ValidatableMixin). Всё **работает** **вместе**.

### Вывод

`@dataclass` и **миксины** **работают** **вместе**. Миксины **ставятся** **первыми**. Несколько миксинов — **через** запятую. `frozen` и `slots` **требуют** **осторожности**.

---

## 20.10. `@dataclass` и наследование

`@dataclass` **поддерживает** **наследование**. Разберём **нюансы**.

### Наследование и поля

Наследник **получает** поля **родителя** и **добавляет** свои:

```python
@dataclass
class Apartment:
    address: str
    area: int


@dataclass
class Studio(Apartment):
    balcony: int = 0        # дополнительное поле


studio = Studio("ул. Пушкина", 25, 3)
print(studio)    # Studio(address='ул. Пушкина', area=25, balcony=3)
```

### Порядок полей

Наследник **не** может **добавить** **обязательное** поле **после** **необязательного**:

```python
@dataclass
class Apartment:
    address: str
    area: int = 30


@dataclass
class Studio(Apartment):
    balcony: int             # ← ошибка! Обязательное после необязательного
```

Python **выдаст** `TypeError`.

**Решение:** используйте значения по умолчанию **или** `field(default=...)`:

```python
@dataclass
class Studio(Apartment):
    balcony: int = 0         # значение по умолчанию
```

### Переопределение полей

Наследник **может** **переопределить** значение по умолчанию:

```python
@dataclass
class Apartment:
    address: str
    area: int = 30


@dataclass
class Studio(Apartment):
    area: int = 25           # другое значение по умолчанию


studio = Studio("ул. Пушкина")
print(studio.area)           # 25
```

### `__repr__` и `__eq__` в наследниках

`@dataclass` **создаёт** методы **для** **каждого** класса. При **наследовании** — **используется** **версия** **наследника**.

```python
apartment = Apartment("ул. Ленина", 65)
studio = Studio("ул. Пушкина", 25)

print(apartment)    # Apartment(address='ул. Ленина', area=65)
print(studio)       # Studio(address='ул. Пушкина', area=25)
```

### Нюанс: `@dataclass` в родителе без декоратора

Если родитель — **обычный** класс, а наследник — `@dataclass`, **работает** **не всегда** **идеально**:

```python
class Base:
    def __init__(self, x):
        self.x = x


@dataclass
class Derived(Base):        # Base не @dataclass
    y: int
```

`@dataclass` **не** знает о **полях** `Base`. Поэтому `__init__` **будет** **только** для `y`. Нужно **вручную** **вызвать** `Base.__init__`.

### Схема

```
        @dataclass Base              @dataclass Derived(Base)
        ├── field1: int              ├── field1: int (унаследовано)
        └── field2: int              ├── field2: int (унаследовано)
                                     └── field3: int (своё)
```

### Аналогия

Представь **конструктор** LEGO с **базовым** набором. Можно **докупить** **дополнительный** — и получить **расширенный**. `@dataclass` **поддерживает** такое **расширение**.

### Вывод

`@dataclass` **поддерживает** **наследование**. Наследник **получает** поля **родителя**. **Порядок** полей **важен**. Переопределение **значений** по умолчанию — **разрешено**.

---

## 20.11. Нюансы и тонкости

Разберём **десять** нюансов.

### 20.11.1. `@dataclass` не заменяет методы, а добавляет

Если вы **определили** `__repr__`, `@dataclass` **не** перезапишет его. Он **добавит** методы **только** если их **нет**.

### 20.11.2. Порядок полей в `__eq__`

Сравнение идёт **по всем** полям **в порядке** объявления. Если **порядок** разный — **результат** может **отличаться**.

### 20.11.3. `field(compare=False)`

Можно **исключить** поле **из** сравнения:

```python
from dataclasses import dataclass, field

@dataclass
class Apartment:
    address: str
    area: int
    cached_price: float = field(default=0, compare=False)
```

Теперь `cached_price` **не** участвует в `__eq__`.

### 20.11.4. `field(repr=False)`

Можно **исключить** поле **из** `__repr__`:

```python
@dataclass
class Apartment:
    address: str
    password: str = field(repr=False)    # не показывается в repr
```

Полезно для **чувствительных** данных.

### 20.11.5. `field(init=False)`

Поле **не** передаётся **в** конструктор:

```python
@dataclass
class Apartment:
    address: str
    full_info: str = field(init=False, default="")

    def __post_init__(self):
        self.full_info = f"Квартира: {self.address}"
```

### 20.11.6. `__post_init__`

Метод **вызывается** **после** `__init__`. Удобен для **вычисляемых** полей:

```python
@dataclass
class Apartment:
    address: str
    area: int
    price: int
    price_per_sqm: float = field(init=False)

    def __post_init__(self):
        self.price_per_sqm = self.price / self.area
```

### 20.11.7. Миксины и `__slots__`

Миксины **не** должны **определять** `__slots__`. Иначе **конфликт** с родителем.

### 20.11.8. Миксины и `@dataclass(frozen=True)`

Миксин **не** должен **изменять** состояние. Иначе `FrozenInstanceError`.

### 20.11.9. Миксины и MRO

Порядок миксинов **важен**. Если **два** миксина **имеют** **одинаковый** метод — **используется** **первый** по MRO.

### 20.11.10. `@dataclass` и `__hash__`

По умолчанию `@dataclass` **не** создаёт `__hash__`, если `eq=True` (что **по умолчанию**). Тогда объект **нехэшируем**. Для **хэшируемости** используйте `frozen=True`.

### Вывод

`@dataclass` и миксины **имеют** **много** нюансов. Основные: **порядок полей**, **`field()`**, **`__post_init__`**, **frozen** и **slots**. Понимание нюансов — **ключ** к правильному использованию.

---

## 20.12. Типичные ошибки

Разберём **десять** самых частых ошибок.

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Изменяемое значение по умолчанию | Ошибка `ValueError` | Использовать `field(default_factory=...)` |
| Обязательное поле после необязательного | `TypeError` | Значения по умолчанию в конце |
| Забыли `@dataclass` | Методы не создаются | Добавить декоратор |
| Миксин **не** первым | Не переопределяет | Поставить **первым** |
| Миксин с `__init__` | Нарушает порядок | **Не** определять |
| Миксин с состоянием | Конфликты | **Не** хранить состояние |
| `@dataclass` без аннотаций | Поля не создаются | Добавить `: type` |
| Порядок полей разный в наследнике | `TypeError` | Сохранять порядок |
| Миксин меняет `self` при `frozen` | Ошибка | **Не** менять |
| `slots=True` и миксин с атрибутами | Ошибка | **Не** добавлять атрибуты |

### Подробный разбор

**Ошибка 1.** Изменяемое значение по умолчанию — **общий** объект для **всех**. **Правило:** `field(default_factory=...)`.

**Ошибка 2.** Обязательное после необязательного — `TypeError`. **Правило:** сначала **обязательные**, потом **со** значениями.

**Ошибка 3.** Забыли `@dataclass` — методы **не** создаются. **Правило:** всегда **ставьте** декоратор.

**Ошибка 4.** Миксин **не** первым — методы **не** переопределяют. **Правило:** миксин — **первым**.

**Ошибка 5.** Миксин с `__init__` — **нарушает** порядок инициализации. **Правило:** **не** определяйте `__init__`.

**Ошибка 6.** Миксин с состоянием — **конфликты** при **множественном** наследовании. **Правило:** **не** храните состояние.

**Ошибка 7.** `@dataclass` без аннотаций — **не** создаёт поля. **Правило:** аннотации **обязательны**.

**Ошибка 8.** Порядок полей разный — `TypeError`. **Правило:** **сохраняйте** порядок.

**Ошибка 9.** Миксин меняет `self` при `frozen` — `FrozenInstanceError`. **Правило:** **не** меняйте.

**Ошибка 10.** `slots=True` и **миксин** с атрибутами — **конфликт**. **Правило:** **не** добавляйте атрибуты.

### Вывод

Ошибки с `@dataclass` и **миксинами** **типичны**. Следуя **правилам**, легко их **избежать**.

---

## 20.13. Что важно запомнить

- **`@dataclass`** — декоратор, создающий `__init__`, `__repr__`, `__eq__`.
- **Поля** — через **аннотации** типов.
- **Значения по умолчанию** — **обычные** или `field(default_factory=...)`.
- **Изменяемые** — **только** через `default_factory`.
- **Параметры:** `frozen`, `order`, `eq`, `repr`, `init`, `slots`.
- **`frozen=True`** — неизменяемый объект + `__hash__`.
- **`order=True`** — сравнение через `<`, `>`.
- **`__post_init__`** — для **вычисляемых** полей.
- **Миксин** — **узкая** «примесь» функциональности.
- **Миксин** — **первым** в списке родителей.
- **Миксин** — **без** `__init__` и **без** состояния.
- **`LoggableMixin`**, **`ComparableMixin`**, **`SerializableMixin`** — **типичные** миксины.
- **`@dataclass` + миксины** — **работают** **вместе**.
- **`@dataclass` + наследование** — **поддерживается**.

## Задача 20 (`@dataclass` и миксин)

Класс `Apartment` содержит **только** данные и **простые** методы — его **удобно** переписать через `@dataclass`. Дополнительно нужно **добавить** **логирование** через **миксин**.

**Часть 1. Переписать через `@dataclass`.** Создайте класс `Apartment` через `@dataclass` с полями:

- `address` — строка;
- `area` — целое число;
- `price` — целое число.

Реализуйте конструктор, `__repr__`, `__eq__` **автоматически**. Добавьте метод `price_per_sqm()` — цена за м².

**Часть 2. Миксин `LoggableMixin`.** Создайте миксин `LoggableMixin` с методом `log(message)`, который **выводит** **имя** класса и **текущее** время (используйте `from datetime import datetime`).

**Часть 3. Подключение миксина.** Перепишите класс так, чтобы он **наследовал** и от `LoggableMixin`, **и** был `@dataclass`:

```python
@dataclass
class Apartment(LoggableMixin):
    address: str
    area: int
    price: int
```

**Миксин** должен идти **первым**.

**Часть 4. Проверка.** Продемонстрируйте:

- создание объекта;
- работу `__repr__` (автоматически);
- работу `__eq__` (автоматически);
- работу `log()`;
- работу `price_per_sqm()`.

**Часть 5. Значения по умолчанию.** Добавьте значения по умолчанию:

- `area: int = 30`;
- `price: int = 5_000_000`;
- `is_sold: bool = False`.

Продемонстрируйте создание объекта с **разным** количеством аргументов.

**Часть 6. Изменяемое поле.** Добавьте поле `tags: list = field(default_factory=list)`. Продемонстрируйте, что у **каждого** объекта **свой** список.

**Часть 7. `frozen=True`.** Создайте **второй** класс `FrozenApartment` с `frozen=True`. Продемонстрируйте, что **изменить** поля **нельзя**, а **хэшировать** — **можно**.

**Часть 8. Наследование.** Создайте наследника `Studio(Apartment)` с полем `balcony: int = 0`. Продемонстрируйте, что `@dataclass` **работает** с **наследованием**.

**Часть 9.** В комментариях объясните:

1. Чем `@dataclass` **удобнее** обычного класса.
2. Почему **изменяемые** значения по умолчанию **опасны**.
3. Что такое **миксин** и чем он **отличается** от обычного родителя.
4. Почему **миксин** ставится **первым**.
5. Как `self.__class__.__name__` **работает** в миксине.
6. Зачем нужны `frozen=True` и `order=True`.

### Решение

```python
from dataclasses import dataclass, field
from datetime import datetime


# ─── Часть 2. Миксин LoggableMixin ───
class LoggableMixin:
    def log(self, message):
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        print(f"[{timestamp}] {self.__class__.__name__}: {message}")


# ─── Части 1, 3, 5, 6. Класс Apartment ───
@dataclass
class Apartment(LoggableMixin):
    address: str
    area: int = 30
    price: int = 5_000_000
    is_sold: bool = False
    tags: list = field(default_factory=list)

    def price_per_sqm(self):
        return self.price / self.area


# ─── Часть 7. FrozenApartment ───
@dataclass(frozen=True)
class FrozenApartment:
    address: str
    area: int
    price: int


# ─── Часть 8. Наследник Studio ───
@dataclass
class Studio(Apartment):
    balcony: int = 0


# ─── Часть 4. Проверка ───
print("=== Часть 4. Базовый Apartment ===")
flat1 = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)
flat2 = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)

print(f"repr: {flat1}")
print(f"eq: {flat1 == flat2}")
flat1.log("Создана квартира")
print(f"price_per_sqm: {flat1.price_per_sqm():,.0f} ₽".replace(",", " "))

# ─── Часть 5. Значения по умолчанию ───
print()
print("=== Часть 5. Значения по умолчанию ===")
print(Apartment("ул. Ленина"))
print(Apartment("ул. Ленина", 65))
print(Apartment("ул. Ленина", 65, 8_500_000))
print(Apartment("ул. Ленина", 65, 8_500_000, True))

# ─── Часть 6. Изменяемое поле ───
print()
print("=== Часть 6. Изменяемое поле tags ===")
flat_a = Apartment("ул. Ленина")
flat_b = Apartment("ул. Садовая")

flat_a.tags.append("новинка")
flat_a.tags.append("скидка")

print(f"flat_a.tags = {flat_a.tags}")
print(f"flat_b.tags = {flat_b.tags}")   # пустой — свой список

# ─── Часть 7. Frozen ───
print()
print("=== Часть 7. FrozenApartment ===")
frozen = FrozenApartment("ул. Ленина", 65, 8_500_000)
print(f"frozen: {frozen}")
print(f"hash: {hash(frozen)}")

try:
    frozen.area = 70
except Exception as e:
    print(f"Ошибка при изменении: {type(e).__name__}: {e}")

# Использование в set.
frozen_set = {frozen, FrozenApartment("ул. Ленина", 65, 8_500_000)}
print(f"set size: {len(frozen_set)}")   # 1 — потому что объекты равны

# ─── Часть 8. Наследование ───
print()
print("=== Часть 8. Наследник Studio ===")
studio = Studio("ул. Пушкина", 25, 4_000_000, False, [], 3)
print(f"studio: {studio}")
print(f"studio.balcony: {studio.balcony}")
studio.log("Создана студия")
```

**Ожидаемый вывод:**

```
=== Часть 4. Базовый Apartment ===
repr: Apartment(address='ул. Ленина, д. 10, кв. 5', area=65, price=8500000, is_sold=False, tags=[])
eq: True
[2024-01-15 10:30:00] Apartment: Создана квартира
price_per_sqm: 130 769 ₽

=== Часть 5. Значения по умолчанию ===
Apartment(address='ул. Ленина', area=30, price=5000000, is_sold=False, tags=[])
Apartment(address='ул. Ленина', area=65, price=5000000, is_sold=False, tags=[])
Apartment(address='ул. Ленина', area=65, price=8500000, is_sold=False, tags=[])
Apartment(address='ул. Ленина', area=65, price=8500000, is_sold=True, tags=[])

=== Часть 6. Изменяемое поле tags ===
flat_a.tags = ['новинка', 'скидка']
flat_b.tags = []

=== Часть 7. FrozenApartment ===
frozen: FrozenApartment(address='ул. Ленина', area=65, price=8500000)
hash: 1234567890
Ошибка при изменении: FrozenInstanceError: cannot assign to field 'area'
set size: 1

=== Часть 8. Наследник Studio ===
studio: Studio(address='ул. Пушкина', area=25, price=4000000, is_sold=False, tags=[], balcony=3)
studio.balcony: 3
[2024-01-15 10:30:00] Studio: Создана студия
```

### Разбор

**1. Чем `@dataclass` удобнее обычного класса.**

`@dataclass` **избавляет** от **шаблонного** кода:

- **`__init__`** — создаётся **автоматически**.
- **`__repr__`** — **автоматически**.
- **`__eq__`** — **автоматически**.
- **`__hash__`** — при `frozen=True`.

**Обычный** класс требовал бы **написать** всё **это** **вручную**. При **добавлении** поля — **правки** в **четырёх** местах. С `@dataclass` — **в одном**.

**2. Почему изменяемые значения по умолчанию опасны.**

Если написать `tags: list = []` — Python создаст **один** список **на** **все** объекты. Это **значит**: изменение `tags` у **одного** объекта **изменит** `tags` у **всех**.

**Пример проблемы:**

```python
@dataclass
class Apartment:
    tags: list = []           # ← ОШИБКА


flat1 = Apartment()
flat2 = Apartment()

flat1.tags.append("новинка")
print(flat2.tags)             # ['новинка'] — изменилось у ВСЕХ!
```

**Решение — `field(default_factory=list)`:** **каждый** объект **получает** **свой** список.

**3. Что такое миксин и чем он отличается от обычного родителя.**

**Миксин** — **узкая** «примесь» функциональности. **Не** используется **самостоятельно**. **Не** имеет **состояния**. **Не** определяет `__init__`.

**Обычный родитель** — **основное** понятие. Используется **самостоятельно**. **Имеет** состояние. **Определяет** `__init__`.

**Пример:**

- `Kitchen(Room)` — **обычный**. Кухня **является** комнатой.
- `Apartment(LoggableMixin)` — **миксин**. Квартира **умеет** логировать.

**4. Почему миксин ставится первым.**

**Миксин** **должен** **переопределять** методы **базовых** классов. В **MRO** (см. главу 19) **первый** в списке **родителей** **идёт** **первым** в **поиске**. Если миксин **второй** — его **методы** **могут** **не** **вызваться**.

**Правило:** миксин — **всегда** **первым** в списке.

**5. Как `self.__class__.__name__` работает в миксине.**

`self` — **ссылка** на **фактический** объект. `type(self)` — **фактический** класс. `__name__` — **имя** класса.

В `LoggableMixin.log` мы **пишем** `self.__class__.__name__`. Это **даёт**:

- `Apartment` — **если** объект — `Apartment`.
- `Studio` — **если** объект — `Studio`.

**Без** этого пришлось бы **жёстко** **писать** имя класса — но это **не** работает **для** **наследников**.

**6. Зачем нужны `frozen=True` и `order=True`.**

**`frozen=True`** — делает объект **неизменяемым**:

- **Попытка** изменить поле → `FrozenInstanceError`.
- **Автоматически** создаётся `__hash__`.
- Можно использовать в `set` и `dict`.
- **Полезно** для **value objects** (значимых объектов).

**`order=True`** — создаёт **операторы** сравнения:

- `__lt__`, `__le__`, `__gt__`, `__ge__`.
- Сравнение **по** полям **в** порядке **объявления**.
- **Полезно** для **сортировки**.

**Пример:**

```python
@dataclass(order=True)
class Apartment:
    address: str
    area: int
    price: int

flats = [Apartment("в", 50, 6), Apartment("а", 30, 4)]
flats.sort()    # сортировка по address, потом area, потом price
```

# Глава 21. `__slots__`

## 21.1. Проблема: каждый объект хранит словарь

За предыдущие главы мы построили класс `Apartment` и **много раз** его улучшали. Добавили `__init__`, `__repr__`, `__eq__`, свойства, классовые методы, `@dataclass`. Всё работает **отлично** — пока объектов **мало**. Но представь, что агентство **растёт**. В базе — **десятки тысяч** квартир. И тут возникает **неожиданная** проблема: **память**.

Каждый **объект** в Python — это **не только** значения полей. Это ещё **служебная** информация: тип, счётчик ссылок, а главное — **словарь атрибутов** `__dict__`. Именно в нём Python **хранит** атрибуты объекта. И этот словарь **занимает** **много** места.

### Сколько занимает объект

Проверим **экспериментально**:

```python
import sys

class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price


flat = Apartment("ул. Ленина", 65, 8_500_000)
print(sys.getsizeof(flat))       # например, 48 байт — только «оболочка»
print(sys.getsizeof(flat.__dict__))   # например, 296 байт — словарь
```

**Что мы видим:** сам объект `flat` занимает **48** байт. А его **словарь** `__dict__` — **296** байт. **Итого** — **344** байта на **один** объект. Если объектов **10 000** — это **3,4 МБ**. Если **100 000** — **34 МБ**. Только на **словари**!

### Почему словарь такой большой

Словарь в Python — **мощная** структура. Он **поддерживает**:

- **Динамическое** добавление атрибутов.
- **Быстрый** поиск по имени.
- **Изменение** размера при росте.

Но **плата** за это — **память**. Даже если у объекта всего **три** поля, словарь **резервирует** место **с запасом**. Плюс **хранит** ключи (имена атрибутов), хэши, служебные поля.

### Аналогия

Представь **шкаф** с **подписанными** ящиками. Каждый ящик — **атрибут**. Ты **можешь** добавить **новый** ящик в **любой** момент — просто **приклеиваешь** **бирку** с именем. Это **удобно**, но **шкаф** должен **вмещать** **бирки**, **крепления**, **место** для **новых** ящиков. В итоге **шкаф** занимает **много** места, **даже** если ящиков **мало**.

### Что делать

Если объект **всегда** имеет **фиксированный** набор полей — можно **отказаться** от словаря. Вместо него Python **зарезервирует** **слоты** под **конкретные** поля. Это **экономит** память **в разы**. Для этого существует **`__slots__`**.

### Вывод

Каждый объект **по умолчанию** хранит **словарь** `__dict__`. Это **удобно**, но **дорого** по памяти. При **большом** числе объектов это **проблема**. Решение — **`__slots__`**.

---

## 21.2. Что такое `__slots__`

**`__slots__`** — это **специальный** атрибут класса, который **сообщает** Python: «у объектов этого класса **фиксированный** набор полей — вот **список**». Python **создаёт** для них **слоты** вместо словаря.

### Синтаксис

```python
class Apartment:
    __slots__ = ("address", "area", "price")

    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price
```

Разберём по частям:

- **`__slots__`** — имя **фиксировано**. Именно **два** подчёркивания с **каждой** стороны.
- **`("address", "area", "price")`** — **кортеж** имён **полей**. Можно **список**, но **кортеж** — **предпочтительнее** (неизменяемый).
- **`__init__`** — работает **как обычно**. Присваивание `self.address = ...` **создаёт** **слот**.

### Что происходит под капотом

Когда Python видит `__slots__`, он:

1. **Создаёт** **слоты** для **каждого** имени из кортежа.
2. **Отключает** создание `__dict__` у объекта.
3. **Сохраняет** значения **напрямую** в слотах.
4. **Запрещает** добавление **новых** атрибутов.

В **итоге** объект **не имеет** словаря. Вместо него — **массив** слотов **фиксированного** размера.

### Как это выглядит в памяти

```
        Обычный объект:                  Объект с __slots__:
        ┌──────────────┐                 ┌──────────────┐
        │   заголовок  │                 │   заголовок  │
        │   __dict__ ──┼→ словарь        │   слот 1     │
        └──────────────┘    (много       │   слот 2     │
                             памяти)     │   слот 3     │
                                         └──────────────┘
                                         (меньше памяти)
```

### Использование

```python
flat = Apartment("ул. Ленина", 65, 8_500_000)
print(flat.address)     # ул. Ленина
print(flat.area)        # 65
print(flat.price)       # 8500000
```

Снаружи **всё** работает **как обычно**. Разница — **внутри**.

### Что даёт `__slots__`

- **Экономия** памяти — **в разы**.
- **Скорость** доступа — **чуть** быстрее.
- **Защита** от **опечаток** — нельзя **добавить** **новое** поле **случайно**.

### Что теряется

- **Нельзя** добавлять **новые** атрибуты.
- **Нет** `__dict__` — **некоторые** приёмы **перестают** работать.
- **Ограничения** при **наследовании**.
- **Конфликты** с `@property`.

Об этом — в следующих разделах.

### Аналогия

Представь **шкаф** с **ящиками** **фиксированного** размера. Ты **знаешь** **заранее**, **сколько** ящиков **нужно** — **ровно** **три**. Никаких **бирок**, **креплений**, **запаса**. Просто **три** ящика. **Меньше** места, **быстрее** доступ. Но **добавить** **четвёртый** ящик **нельзя**. Это **`__slots__`**.

### Вывод

**`__slots__`** — специальный атрибут с **фиксированным** списком **полей**. **Заменяет** словарь **слотами**. **Экономит** память. **Ограничивает** гибкость.

---

## 21.3. Как `__slots__` экономит память

Разберём **механизм** экономии **подробно**.

### Сколько экономит

Проведём **эксперимент**:

```python
import sys

class ApartmentWithDict:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price


class ApartmentWithSlots:
    __slots__ = ("address", "area", "price")

    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price


flat_dict = ApartmentWithDict("ул. Ленина", 65, 8_500_000)
flat_slots = ApartmentWithSlots("ул. Ленина", 65, 8_500_000)

print(f"С __dict__: {sys.getsizeof(flat_dict)} байт")
print(f"С __slots__: {sys.getsizeof(flat_slots)} байт")
```

**Примерный** вывод:

```
С __dict__: 48 байт (только объект)
С __slots__: 56 байт (только объект)
```

**Странно!** Объект **со** слотами **больше**? В чём дело?

### Почему `getsizeof` показывает не то

`sys.getsizeof(flat_dict)` показывает размер **только** «оболочки» объекта — **без** словаря. А сам словарь — **отдельный** объект, его **нужно** измерять **отдельно**:

```python
print(f"Объект: {sys.getsizeof(flat_dict)} байт")
print(f"Словарь: {sys.getsizeof(flat_dict.__dict__)} байт")
```

**Примерный** вывод:

```
Объект: 48 байт
Словарь: 296 байт
Итого: 344 байта
```

А **объект** со слотами **не** имеет **словаря**:

```python
print(f"Объект: {sys.getsizeof(flat_slots)} байт")
# Объект: 56 байт
# Итого: 56 байт
```

**Реальная** экономия — **288** байт **на** **объект**! Это **в 6 раз** меньше.

### Как измерить честно

**Способ 1:** `sys.getsizeof` + `sys.getsizeof(obj.__dict__)`.

```python
def total_size(obj):
    size = sys.getsizeof(obj)
    if hasattr(obj, "__dict__"):
        size += sys.getsizeof(obj.__dict__)
    return size
```

**Способ 2:** библиотека `pympler.asizeof` — **точный** размер **всей** структуры.

```python
from pympler import asizeof

print(asizeof.asizeof(flat_dict))     # ~344 байта
print(asizeof.asizeof(flat_slots))    # ~56 байт
```

`pympler` **учитывает** **всё**: объект, словарь, значения. **Честное** сравнение.

### Масштабирование

Если объектов **10 000**:

- **Словари:** 10 000 × 344 = **3,4 МБ**.
- **Слоты:** 10 000 × 56 = **560 КБ**.

**Экономия** — **почти 3 МБ**. Если **100 000** — **30 МБ**. **Существенно**!

### Схема экономии

```
        Обычный объект:                  Объект с __slots__:
        ┌──────────────┐                 ┌──────────────┐
        │   заголовок  │                 │   заголовок  │
        │   __dict__ ──┼→ словарь (296)  │   address    │
        └──────────────┘                 │   area       │
        Итого: ~344 байта                │   price      │
                                         └──────────────┘
                                         Итого: ~56 байт
```

### Почему экономия такая большая

Словарь в Python — **сложная** структура:

- **Хэш-таблица** — массив **корзин**.
- **Ключи** — имена атрибутов **хранятся** как **строки**.
- **Хэши** — для **быстрого** поиска.
- **Резерв** — для **роста** словаря.

Даже при **трёх** полях словарь **резервирует** место **с запасом**. **Слоты** — **просто** **массив** значений **фиксированного** размера. **Ничего** лишнего.

### Аналогия

Представь **чемодан** с **подписанными** **кармашками** и **чемодан** с **тремя** **просторными** отделениями **фиксированного** размера. Первый **гибкий** — можно **переложить**, **добавить** карман. Второй **фиксированный** — **только** три **отделения**. Но **второй** **легче** и **меньше**.

### Вывод

**`__slots__`** экономит память **в разы**. **`getsizeof`** показывает **только** «оболочку» — нужен **`pympler`** или **ручной** подсчёт. **Экономия** **масштабируется** с **числом** объектов.

---

## 21.4. Что теряется при использовании `__slots__`

`__slots__` — **не** **бесплатная** оптимизация. За **экономию** памяти **платят** **гибкостью**. Разберём **пять** потерь.

### 21.4.1. Нельзя добавлять новые атрибуты

**Самое главное** ограничение. После `__slots__` **нельзя** добавить **новое** поле:

```python
class Apartment:
    __slots__ = ("address", "area", "price")

    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price


flat = Apartment("ул. Ленина", 65, 8_500_000)
flat.floor = 2        # AttributeError: 'Apartment' object has no attribute 'floor'
```

Python **выдаст** `AttributeError`. **Раньше** это **было** **возможно** — теперь **нет**.

**Плюс:** защита от **опечаток**:

```python
flat.arrea = 70       # AttributeError — сразу видно ошибку!
```

**Минус:** нельзя **динамически** **расширять** объект.

### 21.4.2. Нет `__dict__`

У объекта **со** слотами **нет** словаря:

```python
print(flat.__dict__)    # AttributeError: 'Apartment' object has no attribute '__dict__'
```

**Что это ломает:**

- Нельзя **получить** словарь **всех** атрибутов.
- Не работает **сериализация** через `__dict__`.
- Некоторые **библиотеки** **требуют** `__dict__` (например, старые версии pickle).

**Обходной путь:** **явно** указывать поля:

```python
def to_dict(self):
    return {field: getattr(self, field) for field in self.__slots__}
```

### 21.4.3. Конфликты с `@property`

Если в классе **есть** `__slots__` и `@property` с **тем же** именем — **конфликт**:

```python
class Apartment:
    __slots__ = ("address", "area")    # ← area в слотах

    @property
    def area(self):                    # ← и свойство area
        return self._area
```

Python **выдаст** `ValueError: 'area' in __slots__ conflicts with class variable`.

**Почему:** свойство **само** является **атрибутом** класса. А `__slots__` **создаёт** **дескриптор** с **тем же** именем. **Конфликт**.

**Решение:** **не** включать **имена** свойств **в** `__slots__`. В **слотах** — только **приватные** поля:

```python
class Apartment:
    __slots__ = ("_area",)             # ← приватное поле

    @property
    def area(self):
        return self._area

    @area.setter
    def area(self, value):
        self._area = value
```

### 21.4.4. Наследование требует осторожности

Наследник **может** **использовать** `__slots__` **родителя** или **определить** свои. Разберём **нюансы**:

**Случай 1: наследник без `__slots__`.**

```python
class Base:
    __slots__ = ("x",)


class Child(Base):
    pass      # нет __slots__


child = Child()
child.y = 5    # работает — появился __dict__
```

**Что произошло:** у **наследника** **появился** `__dict__`. **Экономия** **потеряна**. Плюс **объект** теперь **больше**, чем у **родителя**.

**Случай 2: наследник с `__slots__`.**

```python
class Base:
    __slots__ = ("x",)


class Child(Base):
    __slots__ = ("y",)    # ← только свои поля


child = Child()
child.x = 1
child.y = 2
child.z = 3            # AttributeError
```

**Что произошло:** **каждый** класс **добавляет** **свои** слоты. **Экономия** **сохраняется**.

**Правило:** если родитель **имеет** `__slots__`, наследник **тоже** должен **иметь** `__slots__`. Иначе **экономия** **потеряется**.

### 21.4.5. Множественное наследование

При **множественном** наследовании `__slots__` **работает** **ограниченно**. Если **хотя бы** один родитель **не** имеет `__slots__` — у наследника **появится** `__dict__`.

Кроме того, **два** родителя **не** могут **иметь** **слот** с **одним** именем — **конфликт**.

### Схема потерь

```
        __slots__ даёт:              __slots__ отнимает:
        ─────────────────            ────────────────────
        ✅ Экономия памяти           ❌ Нельзя добавлять атрибуты
        ✅ Скорость                  ❌ Нет __dict__
        ✅ Защита от опечаток        ❌ Конфликт с @property
                                     ❌ Наследование — осторожно
                                     ❌ Множественное — ограничено
```

### Аналогия

Представь **чемодан** с **тремя** **отделениями**. Он **лёгкий** и **компактный**. Но **нельзя** **добавить** **четвёртое** отделение. **Нельзя** **переложить** вещи в **другой** чемодан. **Нельзя** **использовать** **вместе** с **чемоданом** **другого** типа. **Взамен** — **экономия** места.

### Вывод

`__slots__` **отнимает** **гибкость**: **нельзя** добавлять **поля**, **нет** `__dict__`, **конфликты** с `@property`, **осторожность** при **наследовании**. **Взамен** — **экономия** памяти.

---

## 21.5. `__slots__` и `@property`

`__slots__` и `@property` **могут** **работать** **вместе**, но **требуют** **осторожности**. Разберём **подробно**.

### Проблема

Если имя **свойства** **совпадает** с **именем** **слота** — **конфликт**:

```python
class Apartment:
    __slots__ = ("area",)              # ← area в слотах

    @property
    def area(self):                    # ← и свойство area
        return self._area
```

**Ошибка:** `ValueError: 'area' in __slots__ conflicts with class variable`.

### Правильное решение

В **слотах** — **приватное** поле (`_area`), а **свойство** — **публичное** (`area`):

```python
class Apartment:
    __slots__ = ("_area",)             # ← приватное поле в слотах

    def __init__(self, area):
        self.area = area               # ← через свойство

    @property
    def area(self):
        return self._area

    @area.setter
    def area(self, value):
        if not isinstance(value, (int, float)):
            raise TypeError("area должно быть числом")
        if not (5 <= value <= 500):
            raise ValueError("area должно быть от 5 до 500")
        self._area = value
```

Теперь **всё** работает:

```python
flat = Apartment(65)
print(flat.area)          # 65 — через свойство

flat.area = 70            # через сеттер
print(flat.area)          # 70

flat.area = -10           # ValueError
```

### Почему это работает

- **Слот** хранит **значение** `_area`.
- **Свойство** `area` — **отдельный** **дескриптор** **класса**.
- **Обращение** `flat.area` → **идёт** в **свойство**.
- **Свойство** **читает** `_area` **из** **слота**.

### Схема

```
        flat.area          →  свойство area   →  слот _area
        flat.area = 70     →  сеттер area     →  слот _area
        flat._area         →  слот _area (напрямую)
```

### Что делать с несколькими свойствами

```python
class Apartment:
    __slots__ = ("_address", "_area", "_price")

    @property
    def area(self):
        return self._area

    @area.setter
    def area(self, value):
        # проверка
        self._area = value

    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, value):
        # проверка
        self._price = value
```

**Правило:** **в** `__slots__` — **приватные** имена (`_area`), **свойства** — **публичные** (`area`). **Никогда** **не** **совпадают**.

### Аналогия

Представь **сейф** с **ячейками** **фиксированного** размера. **Свойство** — **кодовый** **замок** **на** **ячейке**. **Слот** — **сама** ячейка. **Замок** **открывает** **ячейку**. Но **замок** **не** **может** **быть** **ячейкой**. **Разные** вещи.

### Вывод

`__slots__` и `@property` **работают** **вместе**, если **имена** **разные**: **приватные** поля — **в** слотах, **публичные** свойства — **отдельно**. **Правило:** **никогда** **не** **совмещайте** одно **имя** в **слотах** и **свойстве**.

---

## 21.6. `__slots__` и наследование

Наследование — **самое** **сложное** **место** в `__slots__`. Разберём **все** **случаи**.

### Случай 1: родитель с `__slots__`, наследник без

```python
class Base:
    __slots__ = ("x",)

    def __init__(self, x):
        self.x = x


class Child(Base):
    pass      # нет __slots__


child = Child(1)
child.y = 5    # работает
print(child.__dict__)    # {'y': 5}
```

**Что произошло:** у **наследника** **появился** `__dict__`. **Экономия** **потеряна**.

**Почему:** Python **добавляет** `__dict__` **автоматически**, если наследник **не** имеет **своих** `__slots__`. Это **страховка** — чтобы **можно** было **добавлять** **новые** **поля**.

**Вывод:** **потеря** экономии. **Не** **делайте** так.

### Случай 2: оба с `__slots__`

```python
class Base:
    __slots__ = ("x",)

    def __init__(self, x):
        self.x = x


class Child(Base):
    __slots__ = ("y",)    # ← только свои поля

    def __init__(self, x, y):
        super().__init__(x)
        self.y = y


child = Child(1, 2)
print(child.x)    # 1
print(child.y)    # 2
child.z = 3       # AttributeError — нельзя!
```

**Что произошло:** **оба** класса **имеют** `__slots__`. **Экономия** **сохраняется**. **Наследник** **добавляет** **только** **свои** **слоты**.

**Схема:**

```
        Base.__slots__ = ("x",)    →  слот x
        Child.__slots__ = ("y",)   →  слот x (из Base) + слот y
```

**Правило:** **наследник** **указывает** **только** **свои** **поля**. **Поля** **родителя** **не** **повторяются**.

### Случай 3: пустой `__slots__` в наследнике

Если **наследник** **не** **добавляет** **новых** **полей**, но **хочет** **сохранить** **экономию**:

```python
class Child(Base):
    __slots__ = ()    # ← пустой кортеж
```

Это **правильно**: **наследник** **наследует** **слоты** **родителя**, **но** **не** **добавляет** **своих** и **не** **создаёт** `__dict__`.

### Случай 4: несколько уровней

```python
class A:
    __slots__ = ("x",)


class B(A):
    __slots__ = ("y",)


class C(B):
    __slots__ = ("z",)


c = C()
c.x = 1
c.y = 2
c.z = 3
c.w = 4    # AttributeError
```

**Каждый** уровень **добавляет** **свои** **слоты**. **Экономия** **сохраняется** **на** **всех** **уровнях**.

### Случай 5: множественное наследование

```python
class A:
    __slots__ = ("x",)


class B:
    __slots__ = ("y",)


class C(A, B):
    __slots__ = ("z",)
```

**Работает**, если **слоты** **не** **пересекаются**. Если **имя** **повторяется** — **ошибка**:

```python
class A:
    __slots__ = ("x",)


class B:
    __slots__ = ("x",)    # ← то же имя!


class C(A, B):            # ошибка при создании
    pass
```

### Схема наследования

```
        Base.__slots__ = ("x",)
        ├── Child1.__slots__ = ("y",)   → слоты: x, y
        ├── Child2.__slots__ = ("z",)   → слоты: x, z
        └── Child3 без __slots__        → появится __dict__ (ПЛОХО)
```

### Правило

**Если** используете `__slots__` **в** родителе — **используйте** **в** **наследниках** **тоже**. **Иначе** экономия **потеряется**.

### Аналогия

Представь **полки** **фиксированного** **размера**. **Родитель** **даёт** **одну** полку. **Наследник** **может** **добавить** **свои** — **но** **только** **фиксированные**. Если **наследник** **не** **указал** `__slots__` — **получает** **универсальный** **стеллаж** (`__dict__`), **который** **больше**.

### Вывод

`__slots__` **наследуется** **с** **нюансами**. **Наследник** **без** `__slots__` **получает** `__dict__` — **экономия** **теряется**. **Наследник** **с** `__slots__` **добавляет** **только** **свои** **поля**. **Правило:** **если** родитель — **с** слотами, наследник **тоже**.

---

## 21.7. Когда использовать `__slots__`

`__slots__` — **не** **универсальное** **решение**. Разберём, **когда** **стоит**, **а** **когда** — **нет**.

### Когда стоит использовать

**1. Много объектов.**

Если создаёте **тысячи** или **десятки** **тысяч** объектов — **экономия** памяти **существенная**.

**2. Фиксированный набор полей.**

Если поля **известны** **заранее** и **не** **меняются** — `__slots__` **подходит** **идеально**.

**3. Критична память.**

В **мобильных** **приложениях**, **микросервисах**, **встроенных** **системах** — **каждый** **байт** **важен**.

**4. Массовые структуры данных.**

Точки, **векторы**, **записи** — **миллионы** **объектов** **простого** **типа**.

**5. Защита от опечаток.**

`__slots__` **запрещает** **добавлять** **новые** **поля**. Это **ловит** **ошибки** типа `flat.arrea = 70`.

### Когда НЕ стоит использовать

**1. Мало объектов.**

Если объектов **десятки** — **экономия** **незаметна**.

**2. Динамические поля.**

Если **нужно** **добавлять** **поля** **в** **процессе** — `__slots__` **не** **подходит**.

**3. Использование `__dict__`.**

Если **библиотеки** **требуют** `__dict__` — `__slots__` **не** **сработает**.

**4. Множественное наследование.**

Сложные **иерархии** **с** `__slots__` — **трудны** **в** **поддержке**.

**5. `@dataclass` без `slots=True`.**

Если `@dataclass` **создаёт** `__dict__` — **не** **сочетается**. Используйте `@dataclass(slots=True)`.

### Сравнение

| Ситуация | `__slots__`? |
|----------|--------------|
| 10 000+ объектов | ✅ **Да** |
| Фиксированные поля | ✅ **Да** |
| Критична память | ✅ **Да** |
| Массовые точки/векторы | ✅ **Да** |
| Мало объектов | ❌ Нет |
| Динамические поля | ❌ Нет |
| Нужен `__dict__` | ❌ Нет |
| Сложное наследование | ⚠️ **Осторожно** |
| `@dataclass` | ⚠️ **Через** `slots=True` |

### Практический пример

```python
# Подходит: миллионы точек.
class Point:
    __slots__ = ("x", "y")

    def __init__(self, x, y):
        self.x = x
        self.y = y


# Не подходит: сложная модель.
class Apartment:
    def __init__(self, address, **kwargs):
        self.address = address
        for key, value in kwargs.items():
            setattr(self, key, value)


flat = Apartment("ул. Ленина", area=65, price=8_500_000, floor=2)
```

### Аналогия

Представь **инструменты** **для** **пикника**. **Нож**, **вилка**, **ложка** — **фиксированный** набор. Можно **взять** **компактный** **набор** — **легче**, **меньше** **места**. Это **`__slots__`**. Но если **нужно** **брать** **разные** **инструменты** **каждый** **раз** — **лучше** **большой** **ящик** (`__dict__`).

### Вывод

`__slots__` **подходит** **для**: **многих** объектов, **фиксированных** полей, **критичной** памяти. **Не** **подходит** **для**: **гибких** полей, **малого** числа объектов, **сложного** наследования.

---

## 21.8. `__slots__` и `@dataclass`

`@dataclass` **поддерживает** `__slots__` **начиная** с Python **3.10**. Разберём **подробно**.

### Параметр `slots=True`

```python
from dataclasses import dataclass

@dataclass(slots=True)
class Apartment:
    address: str
    area: int
    price: int
```

Python **автоматически** **создаёт** `__slots__` **из** **полей**. **Всё** работает **как** **обычно**, **но** **с** **экономией** памяти.

### Что эквивалентно

`@dataclass(slots=True)` **эквивалентно**:

```python
@dataclass
class Apartment:
    __slots__ = ("address", "area", "price")

    address: str
    area: int
    price: int
```

### Ограничения

**1. Порядок полей.**

Поля **со** **значениями** **по** **умолчанию** — **после** **обязательных** (как **обычно**).

**2. Наследование.**

```python
@dataclass(slots=True)
class Apartment:
    address: str


@dataclass(slots=True)
class Studio(Apartment):
    balcony: int = 0
```

**Работает**, но **нужно** **явно** **указывать** `slots=True` **в** **каждом** **классе**.

**3. `frozen=True` + `slots=True`.**

**Работает** **вместе**:

```python
@dataclass(slots=True, frozen=True)
class Apartment:
    address: str
    area: int
```

Объект **неизменяем**, **хэшируем**, **экономит** память.

**4. `__post_init__`.**

**Работает** **как** **обычно**:

```python
@dataclass(slots=True)
class Apartment:
    address: str
    area: int
    price_per_sqm: float = field(init=False)

    def __post_init__(self):
        self.price_per_sqm = 100
```

### Пример

```python
from dataclasses import dataclass

@dataclass(slots=True)
class Apartment:
    address: str
    area: int
    price: int


flat = Apartment("ул. Ленина", 65, 8_500_000)
print(flat)
print(flat.__slots__)    # ('address', 'area', 'price')

flat.floor = 2           # AttributeError
```

### Схема

```
        @dataclass(slots=True)
        ├── __init__ автоматически
        ├── __repr__ автоматически
        ├── __eq__ автоматически
        └── __slots__ автоматически
```

### Аналогия

Представь **конструктор** LEGO с **опцией** **«компактный** **набор»**. Включаешь **галочку** — и **конструктор** **сам** **собирает** **лёгкий** **набор** **без** **лишних** **деталей**. Это **`slots=True`**.

### Вывод

`@dataclass(slots=True)` — **современный** **способ** **использовать** `__slots__`. Python **3.10+**. **Указывайте** `slots=True` **на** **каждом** **классе** **иерархии**.

---

## 21.9. `__slots__` в реальных проектах

Разберём **пять** **примеров** **из** **реальных** **проектов**.

### 21.9.1. NumPy

**NumPy** использует **плотные** **массивы** — **не** **объекты**. Но **внутри** **много** **классов** **со** `__slots__`.

### 21.9.2. requests

Библиотека **`requests`** **использует** `__slots__` **в** **некоторых** **классах** — **например**, **`Request`**.

### 21.9.3. Django

**Django** **модели** **не** **используют** `__slots__` — **им** **нужен** `__dict__` **для** **динамических** **полей**.

### 21.9.4. attrs

Библиотека **`attrs`** **поддерживает** **slots**:

```python
import attr

@attr.s(slots=True)
class Apartment:
    address = attr.ib()
    area = attr.ib()
```

### 21.9.5. Наш дом

```python
@dataclass(slots=True)
class Room:
    name: str
    area: int
    floor: int


@dataclass(slots=True)
class Apartment:
    address: str
    rooms: list = field(default_factory=list)
```

### Что общего

- **Много** **объектов**.
- **Фиксированный** **набор** **полей**.
- **Критична** **память**.

### Вывод

`__slots__` **используется** в **библиотеках** **для** **оптимизации** **памяти**. **Особенно** — **в** **массовых** **структурах**.

---

## 21.10. Нюансы и тонкости

Разберём **десять** **нюансов**.

### 21.10.1. `__slots__` — кортеж, не список

**Рекомендуется** **кортеж** — **неизменяемый**. **Список** **работает**, но **менее** **эффективен**.

### 21.10.2. `__slots__` — атрибут класса

`__slots__` **определяется** **на** **уровне** **класса**, **не** **объекта**.

### 21.10.3. `__slots__` и `__weakref__`

Если **нужны** **слабые** **ссылки** — **добавьте** `__weakref__` **в** **слоты**:

```python
class Apartment:
    __slots__ = ("address", "__weakref__")
```

### 21.10.4. `__slots__` и `__dict__`

Если **добавить** `__dict__` **в** **слоты** — объект **получит** **словарь**:

```python
class Apartment:
    __slots__ = ("address", "__dict__")
```

Тогда **можно** **добавлять** **новые** **поля**, **но** **экономия** **потеряется**.

### 21.10.5. `__slots__` и `__init__`

`__init__` **работает** **как** **обычно**. **Присваивание** `self.x = ...` **создаёт** **значение** **в** **слоте**.

### 21.10.6. `__slots__` и `super()`

`super()` **работает** **нормально** — **слоты** **родителя** **доступны** **наследнику**.

### 21.10.7. `__slots__` и `__repr__`

`__repr__` **работает**, **но** **не** **может** **получить** `self.__dict__`. **Нужно** **явно** **перечислять** **поля**.

### 21.10.8. `__slots__` и pickle

`pickle` **работает** **не** **всегда**. **Нужно** **реализовать** `__getstate__` и `__setstate__`.

### 21.10.9. `__slots__` и копирование

`copy.copy` **работает**. `copy.deepcopy` — **тоже**, **но** **медленнее**.

### 21.10.10. `__slots__` и `@dataclass(slots=True)`

**Рекомендуется** **использовать** **через** `@dataclass(slots=True)` — **меньше** **кода**.

### Вывод

`__slots__` **имеет** **много** **нюансов**. **Основные:** **кортеж**, **weakref**, **`__dict__`**, **`@dataclass`**.

---

## 21.11. Типичные ошибки

Разберём **десять** **самых** **частых** **ошибок**.

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Наследник без `__slots__` | Появится `__dict__` | Добавить `__slots__ = ()` |
| Имя в слотах = имя свойства | `ValueError` | Приватное поле в слотах |
| Забыли `__slots__` в наследнике | Потеря экономии | Добавить |
| Пытаются добавить поле | `AttributeError` | Не добавлять |
| Используют `__dict__` | `AttributeError` | Не использовать |
| Слоты совпадают в MRO | `TypeError` | Переименовать |
| `__slots__` — строка, не кортеж | Работает, но **хуже** | Использовать кортеж |
| Забыли `__weakref__` | Нет слабых ссылок | Добавить в слоты |
| `@dataclass` без `slots=True` | Нет экономии | Добавить `slots=True` |
| Используют `__slots__` зря | Замедляет разработку | Не использовать |

### Подробный разбор

**Ошибка 1.** Наследник без `__slots__` — **появится** `__dict__`. **Правило:** всегда **указывайте** `__slots__` в **наследнике**.

**Ошибка 2.** Имя в слотах = имя свойства — `ValueError`. **Правило:** **приватные** поля — **в** слотах, **публичные** — **свойства**.

**Ошибка 3.** Забыли `__slots__` в наследнике — **потеря** экономии. **Правило:** **следите** за **иерархией**.

**Ошибка 4.** Пытаются добавить поле — `AttributeError`. **Правило:** **все** поля — **в** `__slots__`.

**Ошибка 5.** Используют `__dict__` — его **нет**. **Правило:** **явно** **перечисляйте** поля.

**Ошибка 6.** Слоты совпадают в MRO — `TypeError`. **Правило:** **уникальные** **имена** **слотов**.

**Ошибка 7.** `__slots__` — строка, не кортеж — **работает**, но **хуже**. **Правило:** **кортеж**.

**Ошибка 8.** Забыли `__weakref__` — **нет** **слабых** **ссылок**. **Правило:** **добавьте** в **слоты**.

**Ошибка 9.** `@dataclass` без `slots=True` — **нет** экономии. **Правило:** **используйте** `slots=True`.

**Ошибка 10.** `__slots__` **зря** — **замедляет** разработку. **Правило:** **используйте** **только** при **необходимости**.

### Вывод

Ошибки с `__slots__` **типичны**. Основные — **наследование**, **`@property`**, **свойства**. Следуя **правилам**, легко их **избежать**.

---

## 21.12. Что важно запомнить

- **`__slots__`** — атрибут класса со **списком** полей.
- **Заменяет** `__dict__` **слотами**.
- **Экономит** память **в разы**.
- **`getsizeof`** показывает **только** оболочку — используйте **`pympler`**.
- **Нельзя** добавлять **новые** атрибуты.
- **Нет** `__dict__`.
- **Конфликты** с `@property` при **совпадении** имён.
- **Наследник** без `__slots__` **получает** `__dict__`.
- **Правило:** если родитель — **с** слотами, наследник **тоже**.
- **`@dataclass(slots=True)`** — **современный** способ.
- **`__weakref__`** — если **нужны** слабые ссылки.
- **`__dict__`** в слотах — **отключает** экономию.
- **Используйте** **только** для: **многих** объектов, **фиксированных** полей, **критичной** памяти.

---

## Задача 21 (`__slots__`)

В базе агентства **может** храниться **огромное** количество записей о квартирах — **десятки тысяч**. При таком объёме **важна** экономия памяти.

**Часть 1. Обычный класс.** Создайте класс `Apartment` с полями `address`, `area`, `price` **без** `__slots__`. Замерьте размер объекта через `sys.getsizeof` **объекта** и **словаря** отдельно.

**Часть 2. Класс с `__slots__`.** Создайте **второй** класс `ApartmentSlots` **с** `__slots__ = ("address", "area", "price")`. Замерьте размер.

**Часть 3. Сравнение.** Продемонстрируйте:

- размер объекта **без** слотов (включая `__dict__`);
- размер объекта **со** слотами;
- **разницу** в байтах и **процентах**;
- **расчёт** для **10 000** объектов.

**Часть 4. Запрет новых атрибутов.** Продемонстрируйте, что в `ApartmentSlots` **нельзя** добавить `floor`:

```python
flat.floor = 2    # AttributeError
```

**Часть 5. Наследование.** Создайте наследника `Studio(ApartmentSlots)`:

- **без** `__slots__` — убедитесь, что **появился** `__dict__`;
- **с** `__slots__ = ("balcony",)` — убедитесь, что **экономия** **сохранилась**.

**Часть 6. `@property`.** Реализуйте в `ApartmentSlots` **свойство** `area` с **проверкой**, используя **приватное** поле `_area`:

```python
class ApartmentSlots:
    __slots__ = ("address", "_area", "price")
```

**Часть 7. `@dataclass(slots=True)`.** Перепишите класс через `@dataclass(slots=True)`. Сравните **количество** кода.

**Часть 8. Реальная экономия.** Создайте **список** из **10 000** объектов:

- **без** слотов;
- **со** слотами.

Замерьте **общее** потребление через `pympler.asizeof.asizeof` (если установлен) или **приблизительно** оцените.

**Часть 9.** В комментариях объясните:

1. **Как** `__slots__` **экономит** память.
2. **Почему** `getsizeof` показывает **неполный** размер.
3. **Что** **теряется** при **использовании** `__slots__`.
4. **Почему** наследник **без** `__slots__` **теряет** экономию.
5. **Почему** `@property` **конфликтует** с **одноимённым** **слотом**.
6. **Когда** **стоит** использовать `__slots__`, а **когда** — **нет**.

### Решение

```python
import sys
from dataclasses import dataclass, field


# ─── Часть 1. Обычный класс ───
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price


# ─── Часть 2. Класс с __slots__ ───
class ApartmentSlots:
    __slots__ = ("address", "area", "price")

    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price


# ─── Часть 5. Наследник без __slots__ ───
class StudioBad(ApartmentSlots):
    pass      # ← появится __dict__


# ─── Часть 5. Наследник с __slots__ ───
class StudioGood(ApartmentSlots):
    __slots__ = ("balcony",)

    def __init__(self, address, area, price, balcony):
        super().__init__(address, area, price)
        self.balcony = balcony


# ─── Часть 6. __slots__ + @property ───
class ApartmentSlotsProp:
    __slots__ = ("address", "_area", "price")

    def __init__(self, address, area, price):
        self.address = address
        self.area = area                # через свойство
        self.price = price

    @property
    def area(self):
        return self._area

    @area.setter
    def area(self, value):
        if not isinstance(value, (int, float)):
            raise TypeError("area должно быть числом")
        if not (5 <= value <= 500):
            raise ValueError("area должно быть от 5 до 500")
        self._area = value


# ─── Часть 7. @dataclass(slots=True) ───
@dataclass(slots=True)
class ApartmentDataclass:
    address: str
    area: int
    price: int


# ─── Часть 3. Сравнение ───
print("=== Часть 3. Сравнение размеров ===")

flat_dict = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)
flat_slots = ApartmentSlots("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)

size_dict_obj = sys.getsizeof(flat_dict)
size_dict_inner = sys.getsizeof(flat_dict.__dict__)
size_dict_total = size_dict_obj + size_dict_inner

size_slots = sys.getsizeof(flat_slots)

print(f"Без __slots__:")
print(f"  объект: {size_dict_obj} байт")
print(f"  __dict__: {size_dict_inner} байт")
print(f"  ИТОГО: {size_dict_total} байт")

print(f"\nС __slots__:")
print(f"  объект: {size_slots} байт")

diff = size_dict_total - size_slots
percent = diff / size_dict_total * 100
print(f"\nЭкономия: {diff} байт ({percent:.1f}%)")
print(f"Для 10 000 объектов: {diff * 10_000 / 1024 / 1024:.2f} МБ")

# ─── Часть 4. Запрет новых атрибутов ───
print()
print("=== Часть 4. Запрет новых атрибутов ===")

try:
    flat_slots.floor = 2
except AttributeError as e:
    print(f"Ошибка: {e}")

# ─── Часть 5. Наследование ───
print()
print("=== Часть 5. Наследование ===")

studio_bad = StudioBad("ул. Пушкина", 25, 4_000_000)
studio_bad.extra = "дополнительно"      # работает — появился __dict__
print(f"StudioBad: extra = {studio_bad.extra}")

studio_good = StudioGood("ул. Пушкина", 25, 4_000_000, 3)
print(f"StudioGood: balcony = {studio_good.balcony}")
try:
    studio_good.extra = "дополнительно"
except AttributeError as e:
    print(f"Ошибка: {e}")

# ─── Часть 6. __slots__ + @property ───
print()
print("=== Часть 6. __slots__ + @property ===")

flat_prop = ApartmentSlotsProp("ул. Ленина", 65, 8_500_000)
print(f"area = {flat_prop.area}")

flat_prop.area = 70
print(f"area после изменения = {flat_prop.area}")

try:
    flat_prop.area = -10
except ValueError as e:
    print(f"Ошибка: {e}")

# ─── Часть 7. @dataclass(slots=True) ───
print()
print("=== Часть 7. @dataclass(slots=True) ===")

flat_dc = ApartmentDataclass("ул. Ленина", 65, 8_500_000)
print(f"{flat_dc}")
print(f"__slots__: {ApartmentDataclass.__slots__}")

try:
    flat_dc.floor = 2
except AttributeError as e:
    print(f"Ошибка: {e}")

# ─── Часть 8. Реальная экономия ───
print()
print("=== Часть 8. Экономия на 10 000 объектов ===")

try:
    from pympler import asizeof

    flats_dict = [Apartment("addr", 65, 8_500_000) for _ in range(10_000)]
    flats_slots = [ApartmentSlots("addr", 65, 8_500_000) for _ in range(10_000)]

    total_dict = asizeof.asizeof(flats_dict)
    total_slots = asizeof.asizeof(flats_slots)

    print(f"Без __slots__: {total_dict / 1024 / 1024:.2f} МБ")
    print(f"С __slots__: {total_slots / 1024 / 1024:.2f} МБ")
    print(f"Экономия: {(total_dict - total_slots) / 1024 / 1024:.2f} МБ")
except ImportError:
    print("pympler не установлен. Установите: pip install pympler")
    print(f"Приблизительная оценка:")
    print(f"  без слотов: {size_dict_total * 10_000 / 1024 / 1024:.2f} МБ")
    print(f"  со слотами: {size_slots * 10_000 / 1024 / 1024:.2f} МБ")
```

**Ожидаемый вывод (примерный):**

```
=== Часть 3. Сравнение размеров ===
Без __slots__:
  объект: 48 байт
  __dict__: 296 байт
  ИТОГО: 344 байт

С __slots__:
  объект: 56 байт

Экономия: 288 байт (83.7%)
Для 10 000 объектов: 2.75 МБ

=== Часть 4. Запрет новых атрибутов ===
Ошибка: 'ApartmentSlots' object has no attribute 'floor'

=== Часть 5. Наследование ===
StudioBad: extra = дополнительно
StudioGood: balcony = 3
Ошибка: 'StudioGood' object has no attribute 'extra'

=== Часть 6. __slots__ + @property ===
area = 65
area после изменения = 70
Ошибка: area должно быть от 5 до 500

=== Часть 7. @dataclass(slots=True) ===
ApartmentDataclass(address='ул. Ленина', area=65, price=8500000)
__slots__: ('address', 'area', 'price')
Ошибка: 'ApartmentDataclass' object has no attribute 'floor'

=== Часть 8. Экономия на 10 000 объектов ===
Без __slots__: 3.36 МБ
С __slots__: 0.55 МБ
Экономия: 2.81 МБ
```

### Разбор

**1. Как `__slots__` экономит память.**

Каждый **объект** **по** **умолчанию** **хранит** **словарь** `__dict__` — **отдельный** **объект** **размером** ~296 байт. **Словарь** **поддерживает** **динамическое** **добавление** **полей**, **но** **платит** **за** **это** **памятью**.

`__slots__` **заменяет** **словарь** **слотами** — **массивом** **фиксированного** **размера**. **Нет** **хэш-таблицы**, **нет** **ключей**, **нет** **резерва**. **Только** **значения**. **Экономия** — **в** **разы**.

**2. Почему `getsizeof` показывает неполный размер.**

`sys.getsizeof(obj)` возвращает **размер** **самого** **объекта** — его **«оболочки»**. Для **объекта** **без** `__slots__` это **48** байт — **без** **словаря**. **Словарь** — **отдельный** **объект** (296 байт), его **нужно** **измерять** **отдельно**.

Для **честного** **сравнения** — **используйте** `pympler.asizeof` или **суммируйте** **размеры** **объекта** **и** **словаря**.

**3. Что теряется при использовании `__slots__`.**

- **Нельзя** **добавлять** **новые** **атрибуты** — `flat.floor = 2` → `AttributeError`.
- **Нет** `__dict__` — не работает **автоматическая** **сериализация**.
- **Конфликты** **с** `@property` — при **одноимённых** **полях**.
- **Наследник** **без** `__slots__` — **теряет** **экономию**.
- **Множественное** **наследование** — **ограничено**.

**4. Почему наследник без `__slots__` теряет экономию.**

Python **не** **знает** **заранее**, **какие** **поля** **добавит** **наследник**. Если **наследник** **не** **указывает** `__slots__`, Python **добавляет** `__dict__` **автоматически** — **чтобы** **можно** было **добавлять** **новые** **поля**.

**Решение:** **наследник** **должен** **указать** `__slots__` — **даже** **пустой** (`__slots__ = ()`).

**5. Почему `@property` конфликтует с одноимённым слотом.**

`@property` — **дескриптор** **класса** **с** **именем** `area`. `__slots__` **создаёт** **дескриптор** **с** **тем** **же** **именем**. **Два** **дескриптора** **с** **одним** **именем** — **конфликт** → `ValueError`.

**Решение:** **приватные** **поля** — **в** **слотах** (`_area`), **публичные** **свойства** — **отдельно** (`area`).

**6. Когда стоит использовать `__slots__`, а когда — нет.**

**Стоит:**

- **Много** объектов (**10 000+**).
- **Фиксированный** **набор** **полей**.
- **Критична** память.
- **Массовые** **структуры** (точки, векторы).

**Не стоит:**

- **Мало** объектов.
- **Динамические** **поля**.
- **Библиотеки** **требуют** `__dict__`.
- **Сложное** наследование.

**Правило:** **используйте** `__slots__` **только** **при** **необходимости**. Если **не** **знаете** — **не** **используйте**.

# Глава 22. Дескрипторы

## 22.1. Проблема: много свойств с одинаковой проверкой

За предыдущие главы мы **много раз** использовали `@property` (глава 14) — **удобный** механизм для **создания** свойств с **проверками**. Помнишь, как это выглядит?

```python
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    @property
    def area(self):
        return self._area

    @area.setter
    def area(self, value):
        if not isinstance(value, (int, float)):
            raise TypeError("Площадь должна быть числом")
        if not (5 <= value <= 500):
            raise ValueError("Площадь должна быть от 5 до 500")
        self._area = value
```

**Работает** отлично. Но теперь представь: у нас **не** одно поле с **проверкой**, а **десять**. Площадь. Цена. Этаж. Количество комнат. Год постройки. Высота потолков. Площадь кухни. Количество балконов. Расстояние до метро. Рейтинг.

Для **каждого** поля — **свой** `@property` с **проверкой**. И **каждая** проверка **похожа** на **другие**: «проверить тип», «проверить диапазон». **Девяносто** строк **шаблонного** кода. **Скучно**, **повторяюще**, **легко** ошибиться.

**Первое**, что приходит в голову — **вынести** проверку в **функцию**:

```python
def check_range(value, min_value, max_value, name):
    if not isinstance(value, (int, float)):
        raise TypeError(f"{name} должно быть числом")
    if not (min_value <= value <= max_value):
        raise ValueError(f"{name} должно быть от {min_value} до {max_value}")
```

Но **вызывать** её **всё равно** приходится **в** **каждом** сеттере. **Девять** сеттеров — **девять** вызовов. Всё ещё **много** кода.

**Второе**, что приходит в голову — **создать** **класс** для **проверок**. И это **правильный** путь. В Python такой **класс** называется **дескриптором**. Он **позволяет** **описать** **правило** **один** раз и **переиспользовать** его **для** **многих** полей.

### Аналогия

Представь **завод** по **производству** **деталей**. У **каждой** детали — **свой** контроль качества. **Раньше** **каждый** контролёр **проверял** детали **вручную** — **по-своему**. **Долго**, **одинаково**, **ошибочно**.

Потом **появился** **автоматический** **станок** — **один** **раз** **настроил**, и он **проверяет** **все** детали **одинаково**. Это **дескриптор**: **одна** **логика** — **много** **применений**.

### Что такое дескриптор

**Дескриптор** — это **класс**, который **определяет** **поведение** **атрибута** при **чтении**, **записи** или **удалении**. Это **«умное»** **поле**, которое **само** **знает**, **как** **с** **ним** **обращаться**.

Свойства `@property` — **частный** **случай** **дескрипторов**. Но **есть** **более** **мощные** **варианты**, которые **позволяют** **переиспользовать** **логику** **для** **многих** **полей**.

### Вывод

Когда **проверок** **много** и они **повторяются** — `@property` **становится** **громоздким**. **Дескрипторы** **позволяют** **описать** **правило** **один** раз и **переиспользовать** его. Это **мощный** инструмент **ООП**.

---

## 22.2. Что такое дескриптор

**Дескриптор** — это **класс**, который **реализует** **один** или **несколько** **методов** из **протокола** **дескрипторов**:

- **`__get__(self, instance, owner)`** — вызывается при **чтении** атрибута.
- **`__set__(self, instance, value)`** — вызывается при **записи**.
- **`__delete__(self, instance)`** — вызывается при **удалении**.

Если класс **имеет** **хотя бы** **один** из **этих** **методов** — он **является** **дескриптором**.

### Синтаксис

```python
class PositiveNumber:
    def __get__(self, instance, owner):
        return instance.__dict__[self.name]

    def __set__(self, instance, value):
        if not isinstance(value, (int, float)):
            raise TypeError("Значение должно быть числом")
        if value <= 0:
            raise ValueError("Значение должно быть положительным")
        instance.__dict__[self.name] = value

    def __set_name__(self, owner, name):
        self.name = name
```

**Что здесь:**

- **`__set_name__`** — вызывается **автоматически**, когда **класс** **создаётся**. Сохраняет **имя** **атрибута**.
- **`__get__`** — возвращает **значение** при **чтении**.
- **`__set__`** — проверяет и **сохраняет** при **записи**.

### Использование

```python
class Apartment:
    area = PositiveNumber()
    price = PositiveNumber()

    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price
```

**Что произошло:**

- **`area`** и **`price`** — **экземпляры** **дескриптора** `PositiveNumber`.
- **При** `flat.area = 65` — **вызывается** `__set__`, проверяет.
- **При** `flat.area` — **вызывается** `__get__`, возвращает.
- **Одна** **логика** — **для** **двух** **полей**.

### Что происходит под капотом

Когда Python видит `flat.area`:

1. **Ищет** `area` **в** **классе** `Apartment`.
2. **Находит** **дескриптор**.
3. **Вызывает** `descriptor.__get__(flat, Apartment)`.
4. **Возвращает** **результат**.

Когда Python видит `flat.area = 65`:

1. **Ищет** `area` **в** **классе**.
2. **Находит** **дескриптор**.
3. **Вызывает** `descriptor.__set__(flat, 65)`.
4. **Метод** **сам** **решает**, **что** **делать**.

### Аналогия

Представь **умную** **дверь** **с** **замком**. Ты **просто** **открываешь** её — но **внутри** **работает** **система** **безопасности**: **проверяет** **ключ**, **записывает** **время**, **включает** **сигнализацию**. **Снаружи** — **обычная** **дверь**. **Внутри** — **целая** **система**.

**Дескриптор** — **такая** **дверь**. Снаружи — **обычный** **атрибут**. Внутри — **логика**.

### Где используются дескрипторы

- **`@property`** — **встроенный** **дескриптор**.
- **`@classmethod`** — **встроенный** **дескриптор**.
- **`@staticmethod`** — **встроенный** **дескриптор**.
- **`__slots__`** — **создаёт** **дескрипторы** **автоматически**.
- **Django ORM** — **поля** **модели** — **дескрипторы**.
- **SQLAlchemy** — **колонки** — **дескрипторы**.

**Дескрипторы** — **основа** **многих** **механизмов** Python. Понимание их — **ключ** к **продвинутому** ООП.

### Вывод

**Дескриптор** — класс с **методами** `__get__`, `__set__`, `__delete__`. **Определяет** **поведение** **атрибута**. **Свойства** `@property` — **частный** случай. **Дескрипторы** **позволяют** **переиспользовать** **логику**.

---

## 22.3. Протокол дескрипторов

**Протокол** дескрипторов — это **набор** **методов**, которые **описывают**, **как** **атрибут** **ведёт** **себя** при **разных** **операциях**. Разберём **каждый** **метод** **подробно**.

### `__get__(self, instance, owner)`

**Вызывается** при **чтении** атрибута. **Возвращает** значение.

```python
def __get__(self, instance, owner):
    # instance — объект, из которого читают.
    # owner — класс объекта.
    # Возвращает значение.
    return instance.__dict__[self.name]
```

**Параметры:**

- **`self`** — **сам** дескриптор.
- **`instance`** — **объект**, у которого **читают** атрибут. Если **обращаются** **через** **класс** — `None`.
- **`owner`** — **класс** объекта. **Всегда** **доступен**.

**Пример:**

```python
flat.area          # instance = flat, owner = Apartment
Apartment.area     # instance = None, owner = Apartment
```

Когда `instance is None` — читают **через** **класс**. **Обычно** **возвращают** **сам** **дескриптор**:

```python
def __get__(self, instance, owner):
    if instance is None:
        return self
    return instance.__dict__[self.name]
```

### `__set__(self, instance, value)`

**Вызывается** при **записи** атрибута. **Ничего** не **возвращает** — **просто** **сохраняет** **значение**.

```python
def __set__(self, instance, value):
    # instance — объект, в который пишут.
    # value — новое значение.
    if not isinstance(value, (int, float)):
        raise TypeError("Значение должно быть числом")
    instance.__dict__[self.name] = value
```

**Параметры:**

- **`self`** — **сам** дескриптор.
- **`instance`** — **объект**, в который **пишут**.
- **`value`** — **новое** значение.

### `__delete__(self, instance)`

**Вызывается** при **удалении** атрибута.

```python
def __delete__(self, instance):
    if self.name not in instance.__dict__:
        raise AttributeError("Атрибут не установлен")
    del instance.__dict__[self.name]
```

**Параметры:**

- **`self`** — **сам** дескриптор.
- **`instance`** — **объект**, у которого **удаляют**.

### `__set_name__(self, owner, name)`

**Вызывается** **автоматически**, когда **класс** **создаётся**. Python **сообщает** дескриптору, **под** **каким** **именем** он **используется**.

```python
def __set_name__(self, owner, name):
    # owner — класс, в котором определено поле.
    # name — имя поля.
    self.name = name
```

**Зачем:** дескриптор **один** — **используется** **для** **многих** **полей**. **Каждый** **экземпляр** **дескриптора** **должен** **знать** **своё** **имя**, **чтобы** **сохранить** **значение** **в** **правильное** **место** `instance.__dict__[self.name]`.

**Пример:**

```python
class Apartment:
    area = PositiveNumber()      # __set_name__ вызовется с name="area"
    price = PositiveNumber()     # __set_name__ вызовется с name="price"
```

Без `__set_name__` дескриптор **не** **знал** бы, **под** **каким** **именем** его **используют**.

### Полный протокол

| Метод | Когда вызывается | Что делает |
|-------|------------------|-----------|
| `__set_name__` | При **создании** класса | Сохраняет имя поля |
| `__get__` | При **чтении** | Возвращает значение |
| `__set__` | При **записи** | Проверяет и сохраняет |
| `__delete__` | При **удалении** | Удаляет значение |

### Схема

```
        class Apartment:
            area = RangeDescriptor(5, 500)

        flat.area         →  RangeDescriptor.__get__(flat, Apartment)
        flat.area = 65    →  RangeDescriptor.__set__(flat, 65)
        del flat.area     →  RangeDescriptor.__delete__(flat)
```

### Аналогия

Представь **пульт** **управления**. У него **четыре** **кнопки**: **«прочитать»**, **«записать»**, **«удалить»**, **«настроить»**. **Каждая** **кнопка** — **метод** **протокола**. **Вместе** — **полный** **контроль**.

### Вывод

**Протокол** дескрипторов — **четыре** метода. `__get__` — **чтение**. `__set__` — **запись**. `__delete__` — **удаление**. `__set_name__` — **настройка** при **создании**.

---

## 22.4. Data и non-data дескрипторы

Дескрипторы **делятся** на **два** типа. Разница **принципиальна** — она **влияет** на **порядок** поиска атрибутов.

### Data-дескрипторы

**Data-дескриптор** — **определяет** `__set__` **или** `__delete__`. **Полностью** **контролирует** атрибут.

```python
class DataDescriptor:
    def __get__(self, instance, owner):
        return instance.__dict__[self.name]

    def __set__(self, instance, value):
        instance.__dict__[self.name] = value

    def __set_name__(self, owner, name):
        self.name = name
```

**Что** **характерно:**

- **Чтение** — через `__get__`.
- **Запись** — через `__set__`.
- **Удаление** — через `__delete__`.
- **Приоритет** — **выше** **экземплярных** **атрибутов**.

### Non-data-дескрипторы

**Non-data-дескриптор** — **определяет** **только** `__get__`. **Не** **контролирует** **запись**.

```python
class NonDataDescriptor:
    def __get__(self, instance, owner):
        return "значение по умолчанию"
```

**Что** **характерно:**

- **Чтение** — через `__get__`.
- **Запись** — **можно** **переопределить** **через** `instance.__dict__`.
- **Приоритет** — **ниже** **экземплярных** **атрибутов**.

### Порядок поиска атрибутов

**Правило** **Python**:

1. **Data-дескрипторы** класса — **высший** **приоритет**.
2. **Атрибуты** экземпляра (`instance.__dict__`).
3. **Non-data-дескрипторы** класса.
4. **Обычные** атрибуты класса.
5. **`__getattr__`** (если определён).

**Схема:**

```
        obj.attr
            ↓
        1. Data-дескриптор в классе?
            ↓ (нет)
        2. Атрибут в obj.__dict__?
            ↓ (нет)
        3. Non-data дескриптор в классе?
            ↓ (нет)
        4. Обычный атрибут класса?
            ↓ (нет)
        5. __getattr__
```

### Почему это важно

**Data-дескрипторы** **перекрывают** **экземплярные** **атрибуты**. Это **значит**:

```python
flat.area = 65      # ← вызывает __set__, не создаёт instance.__dict__["area"]
```

Если бы `area` был **non-data** **дескриптором** — **создался** **бы** `instance.__dict__["area"]`, и **последующее** **чтение** **вернуло** **бы** **его**, а не **результат** `__get__`.

### Пример: `@property` — data-дескриптор

`@property` **определяет** `__set__` (если есть сеттер). Поэтому **свойства** **имеют** **приоритет** **над** **экземплярными** **атрибутами**:

```python
class Apartment:
    @property
    def area(self):
        return self._area

    @area.setter
    def area(self, value):
        self._area = value


flat = Apartment()
flat.area = 65      # ← вызывает сеттер, не создаёт instance.area
```

### Пример: функция — non-data дескриптор

**Обычный** **метод** класса — **non-data** **дескриптор**. Он **определяет** **только** `__get__`. Поэтому **можно** **переопределить** **метод** **через** **экземпляр**:

```python
class Apartment:
    def info(self):
        return "Комната"


flat = Apartment()
flat.info = lambda: "Новый info"    # ← переопределили через экземпляр
print(flat.info())                   # Новый info
```

Это **работает** — **потому** **что** **метод** — **non-data** **дескриптор**.

### Сравнение

| Характеристика | Data-дескриптор | Non-data дескриптор |
|----------------|-----------------|---------------------|
| Методы | `__get__`, `__set__`, `__delete__` | Только `__get__` |
| Приоритет | **Высший** | **Ниже** экземплярных |
| Перекрывает `instance.__dict__` | ✅ Да | ❌ Нет |
| Примеры | `@property`, `__slots__` | Методы класса |

### Аналогия

Представь **сейф** **с** **кодовым** **замком**. **Data-дескриптор** — **замок** **есть**: **нельзя** **положить** **вещь** **без** **кода**. **Non-data** — **замка** **нет**: **можно** **положить** **свою** **вещь** **сверху**. **Результат** **зависит** **от** **наличия** **замка**.

### Вывод

**Data-дескрипторы** **определяют** `__set__` **или** `__delete__`. **Приоритет** — **высший**. **Non-data** — **только** `__get__`. **Приоритет** — **ниже**. **Разница** **влияет** на **порядок** поиска.

---

## 22.5. Как `@property` работает под капотом

`@property` — **встроенный** **дескриптор**. Понимание **его** **устройства** **помогает** **понять** **дескрипторы** **вообще**.

### Что делает `@property`

```python
class Apartment:
    def __init__(self, area):
        self._area = area

    @property
    def area(self):
        return self._area

    @area.setter
    def area(self, value):
        self._area = value
```

**Что** **здесь** **происходит:**

1. **Создаётся** **функция** `area` (геттер).
2. **Оборачивается** **в** **объект** `property`.
3. **Дескриптор** **сохраняется** **в** **класс**.
4. **Обращение** `flat.area` → **вызывает** `property.__get__`.
5. **Запись** `flat.area = 70` → **вызывает** `property.__set__`.

### Как выглядит `property` внутри

**Упрощённая** **реализация**:

```python
class MyProperty:
    def __init__(self, fget=None, fset=None, fdel=None):
        self.fget = fget
        self.fset = fset
        self.fdel = fdel

    def __get__(self, instance, owner=None):
        if instance is None:
            return self
        if self.fget is None:
            raise AttributeError("can't read attribute")
        return self.fget(instance)

    def __set__(self, instance, value):
        if self.fset is None:
            raise AttributeError("can't set attribute")
        self.fset(instance, value)

    def __delete__(self, instance):
        if self.fdel is None:
            raise AttributeError("can't delete attribute")
        self.fdel(instance)

    def getter(self, fget):
        return type(self)(fget, self.fset, self.fdel)

    def setter(self, fset):
        return type(self)(self.fget, fset, self.fdel)

    def deleter(self, fdel):
        return type(self)(self.fget, self.fset, fdel)
```

**Что** **мы** **видим:**

- **`__get__`** — вызывает **геттер**.
- **`__set__`** — вызывает **сеттер**.
- **`__delete__`** — вызывает **делетер**.
- **`setter()`** — создаёт **новый** `MyProperty` с **сеттером**.

**Метод** `setter` — **именно** **то**, **что** **вызывается**, **когда** **мы** **пишем** `@area.setter`. Он **создаёт** **новый** **объект** с **добавленным** **сеттером**.

### Как работает `@area.setter`

```python
@area.setter
def area(self, value):
    self._area = value
```

**Что** **происходит** **пошагово**:

1. **Определена** **функция** `area` (геттер), **обёрнутая** **в** `property`.
2. **Пишем** `@area.setter` — **вызывается** `property.setter(fset)`.
3. **Метод** `setter` **создаёт** **новый** `property` **с** **геттером** и **сеттером**.
4. **Новое** **свойство** **заменяет** **старое** **в** **классе**.

**Важно:** `property` **неизменяем**. **Каждый** **вызов** `.setter()` **создаёт** **новый** **объект**.

### Схема

```
        @property           →  property(fget)
        @area.setter        →  property(fget, fset)
        @area.deleter       →  property(fget, fset, fdel)
```

### Почему это важно понимать

**Понимание** **`property`** **даёт**:

- **Уверенность** в **использовании**.
- **Понимание** **ошибок** (например, «can't set attribute»).
- **Возможность** **писать** **свои** **дескрипторы** **по** **аналогии**.
- **Понимание** **ограничений** `property`.

### Сравнение с дескриптором

`@property` — **встроенный** **дескриптор** **для** **одного** **поля**. **Свой** **дескриптор** — **для** **многих** **полей**. Если **логика** **одинакова** — **свой** **дескриптор** **лучше**.

### Аналогия

Представь **фабрику** **мебели**. `@property` — **ручная** **сборка** **одного** **стула**. **Свой** **дескриптор** — **автоматическая** **линия** для **серии** **стульев**. **Логика** **та** **же**, **но** **для** **многих** **применений**.

### Вывод

`@property` — **дескриптор** с **методами** `__get__`, `__set__`, `__delete__`. **Работает** через **геттеры** и **сеттеры**. **Каждый** `.setter()` **создаёт** **новый** **объект**. **Понимание** **устройства** — **ключ** к **своим** **дескрипторам**.

---

## 22.6. `__set_name__` — зачем нужен

`__set_name__` — **относительно** **новый** **метод** (Python 3.6+). Он **решает** **проблему** **универсальных** **дескрипторов**.

### Проблема без `__set_name__`

Представь **дескриптор** **для** **проверки** **диапазона**:

```python
class RangeDescriptor:
    def __init__(self, min_value, max_value):
        self.min_value = min_value
        self.max_value = max_value

    def __get__(self, instance, owner):
        return instance.__dict__[self.name]    # ← откуда self.name?

    def __set__(self, instance, value):
        if not (self.min_value <= value <= self.max_value):
            raise ValueError(f"{value} вне диапазона")
        instance.__dict__[self.name] = value   # ← откуда self.name?
```

**Проблема:** дескриптор **не** **знает** **своё** **имя**. `self.name` **не** **определён**.

### Решение до Python 3.6

**Приходилось** **передавать** **имя** **вручную**:

```python
class Apartment:
    area = RangeDescriptor("area", 5, 500)      # ← имя вручную
    price = RangeDescriptor("price", 0, 10**9)
```

**Минус:** **можно** **ошибиться** — **передать** **одно** **имя** **в** **двух** **местах**.

### Решение с `__set_name__`

```python
class RangeDescriptor:
    def __init__(self, min_value, max_value):
        self.min_value = min_value
        self.max_value = max_value

    def __set_name__(self, owner, name):
        # Python сам сообщает имя!
        self.name = name

    def __get__(self, instance, owner):
        return instance.__dict__[self.name]

    def __set__(self, instance, value):
        if not (self.min_value <= value <= self.max_value):
            raise ValueError(f"{value} вне диапазона [{self.min_value}, {self.max_value}]")
        instance.__dict__[self.name] = value
```

**Использование:**

```python
class Apartment:
    area = RangeDescriptor(5, 500)      # ← имя не нужно!
    price = RangeDescriptor(0, 10**9)
```

Python **сам** **вызовет** `__set_name__` при **создании** класса:

- Для `area` — с `name="area"`.
- Для `price` — с `name="price"`.

### Когда вызывается `__set_name__`

Python **вызывает** `__set_name__` **автоматически** **после** **создания** **класса**, **но** **до** **его** **использования**. **Для** **каждого** **дескриптора** в **классе** — **один** **раз**.

### Схема

```
        Создание класса Apartment
                ↓
        Для каждого дескриптора в классе:
            descriptor.__set_name__(owner, name)
                ↓
        Класс готов к использованию
```

### Важно: `__set_name__` не наследуется

Если **наследник** **использует** **дескриптор** **родителя** — `__set_name__` **не** **вызывается** **снова**. **Дескриптор** **сохраняет** **имя** **из** **родителя**.

**Пример:**

```python
class Base:
    area = RangeDescriptor(5, 500)


class Child(Base):
    pass    # __set_name__ НЕ вызовется


print(Child.area.name)    # "area" — из родителя
```

### Аналогия

Представь **кабель** **без** **бирки**. Раньше **нужно** было **вручную** **подписывать** — «это для кухни», «это для спальни». **Легко** **ошибиться**. Теперь **система** **сама** **подписывает** **при** **установке**. Это `__set_name__`.

### Вывод

`__set_name__` **автоматически** **сообщает** **дескриптору** **его** **имя**. **Избавляет** от **ручного** **указания**. **Вызывается** **один** **раз** при **создании** **класса**.

---

## 22.7. Создание собственного дескриптора

Разберём **создание** **дескриптора** **пошагово**. В **качестве** **примера** — **дескриптор** для **проверки** **диапазона** `RangeDescriptor`.

### Шаг 1. Определяем класс

```python
class RangeDescriptor:
    def __init__(self, min_value, max_value):
        self.min_value = min_value
        self.max_value = max_value
```

**Параметры** — **границы** **диапазона**. Дескриптор **один** — **используется** **для** **разных** **полей** **с** **разными** **границами**.

### Шаг 2. Добавляем `__set_name__`

```python
    def __set_name__(self, owner, name):
        self.name = name
```

**Сохраняем** **имя** **поля**. Дескриптор **будет** **знать**, **куда** **сохранять** **значение**.

### Шаг 3. Добавляем `__get__`

```python
    def __get__(self, instance, owner):
        if instance is None:
            return self
        return instance.__dict__.get(self.name)
```

**Чтение:** **возвращаем** **значение** **из** `instance.__dict__`. **Если** **обращаются** **через** **класс** — **возвращаем** **сам** **дескриптор**.

**`instance.__dict__.get(self.name)`** — **безопасно**: **если** **значения** **нет** — **вернёт** `None`, **не** **упадёт**.

### Шаг 4. Добавляем `__set__`

```python
    def __set__(self, instance, value):
        if not isinstance(value, (int, float)):
            raise TypeError(f"{self.name} должно быть числом")
        if not (self.min_value <= value <= self.max_value):
            raise ValueError(
                f"{self.name} должно быть от {self.min_value} до {self.max_value}"
            )
        instance.__dict__[self.name] = value
```

**Запись:** **проверяем** тип, **проверяем** диапазон, **сохраняем**.

**Что** **важно:**

- **Используем** `self.name` — **имя** **поля**.
- **Проверки** — **до** **сохранения**.
- **Сохраняем** в `instance.__dict__` — **значение** **у** **объекта**.

### Шаг 5. Используем

```python
class Apartment:
    area = RangeDescriptor(5, 500)
    rooms = RangeDescriptor(1, 10)

    def __init__(self, address, area, rooms):
        self.address = address
        self.area = area
        self.rooms = rooms
```

**Один** **дескриптор** — **для** **двух** **полей**. **Разные** **диапазоны**.

### Проверка

```python
flat = Apartment("ул. Ленина", 65, 3)
print(flat.area)      # 65
print(flat.rooms)     # 3

try:
    flat.area = -10
except ValueError as e:
    print(e)          # area должно быть от 5 до 500

try:
    flat.rooms = 20
except ValueError as e:
    print(e)          # rooms должно быть от 1 до 10
```

### Где сохраняется значение

**Важно** **понять**: **дескриптор** — **один** **объект** **класса**. **Значения** — **у** **каждого** **объекта** **свои**. Где **хранятся**?

**В** `instance.__dict__`:

```python
print(flat.__dict__)
# {'address': 'ул. Ленина', 'area': 65, 'rooms': 3}
```

**Ключи** — `area` и `rooms`. **Значения** — **проверенные** **числа**.

**Проблема:** **имя** **дескриптора** (`area`) **совпадает** **с** **ключом** **в** `__dict__`. **Не** **будет** **ли** **конфликта**?

**Нет.** Потому **что** **дескриптор** — **data-дескриптор**. Он **имеет** **приоритет** **над** `instance.__dict__`. **При** **чтении** `flat.area` Python **идёт** **в** **дескриптор**, **а** **не** **в** `__dict__`.

**Правило:** **сохраняйте** **значение** **под** **тем** **же** **именем**, **что** **и** **дескриптор**. Это **безопасно** для **data-дескрипторов**.

### Альтернатива: сохранять под приватным именем

**Некоторые** **используют** **приватное** **имя**:

```python
def __set__(self, instance, value):
    ...
    instance.__dict__[f"_{self.name}"] = value

def __get__(self, instance, owner):
    return instance.__dict__.get(f"_{self.name}")
```

**Что** **это** **даёт:** **можно** **прочитать** `flat._area` **напрямую**. **Но** **это** **не** **рекомендуется** — **нарушает** **инкапсуляцию**.

### Аналогия

Представь **таможню**. **Каждый** **товар** **проверяется** **по** **правилам**. **Правила** **одинаковые** — **но** **для** **разных** **товаров** **разные** **ограничения**. Таможня — **дескриптор**. Правила — **параметры**. Товары — **значения**.

### Вывод

**Создание** **дескриптора** — **пять** **шагов**. **`__init__`** — **параметры**. **`__set_name__`** — **имя**. **`__get__`** — **чтение**. **`__set__`** — **проверка** и **запись**. **Значения** — **в** `instance.__dict__`.

---

## 22.8. Преимущества дескрипторов перед `@property`

Разберём **ключевые** **преимущества**.

### 22.8.1. Переиспользование

**Главное** **преимущество**. **Один** **дескриптор** — **много** **полей**:

```python
class Apartment:
    area = RangeDescriptor(5, 500)
    price = RangeDescriptor(0, 10**9)
    floor = RangeDescriptor(1, 50)
    rooms = RangeDescriptor(1, 10)
    ceiling_height = RangeDescriptor(2, 5)
```

**Пять** **полей** — **пять** **строк** кода. **С** `@property` — **девяносто** **строк**.

### 22.8.2. Единая логика

**Логика** **проверки** — **в** **одном** **месте**. **Изменить** — **в** **одном** **месте**. **Не** **нужно** **править** **девять** **сеттеров**.

### 22.8.3. Параметризация

**Дескриптор** **принимает** **параметры**. **Один** **класс** — **много** **вариаций**:

```python
area = RangeDescriptor(5, 500)             # для площади
rooms = RangeDescriptor(1, 10)             # для комнат
floor = RangeDescriptor(1, 100)            # для этажа
```

**С** `@property` — **пришлось** **бы** **писать** **разные** **сеттеры** **для** **каждого**.

### 22.8.4. Композиция

**Дескрипторы** **можно** **комбинировать**:

```python
class PositiveNumber:
    def __set__(self, instance, value):
        if value <= 0:
            raise ValueError("Должно быть положительным")
        instance.__dict__[self.name] = value

class MaxValue:
    def __init__(self, max_value):
        self.max_value = max_value

    def __set__(self, instance, value):
        if value > self.max_value:
            raise ValueError(f"Не больше {self.max_value}")
        instance.__dict__[self.name] = value
```

**Можно** **создавать** **дескрипторы** **с** **разными** **проверками** **и** **комбинировать** их.

### 22.8.5. Читаемость

**Код** **становится** **короче** и **понятнее**:

```python
class Apartment:
    area = RangeDescriptor(5, 500)
    price = RangeDescriptor(0, 10**9)
```

**Сразу** **видно**: `area` — **от** 5 **до** 500, `price` — **от** 0 **до** миллиарда. **Не** **нужно** **читать** **сеттеры**.

### Когда `@property` лучше

- **Разная** **логика** для **каждого** **поля**.
- **Мало** **полей** — 1–2.
- **Одноразовая** **проверка**.
- **Вычисляемые** **свойства**.

### Когда дескриптор лучше

- **Много** **полей** с **одинаковой** **логикой**.
- **Параметризация** **проверок**.
- **Переиспользование** **логики**.
- **Библиотеки** и **фреймворки**.

### Аналогия

`@property` — **ручная** **работа**: **каждая** **деталь** — **своими** **руками**. **Дескриптор** — **автоматический** **станок**: **одна** **настройка** — **много** **деталей**. **Для** **двух** **деталей** — **ручная** **быстрее**. **Для** **тысячи** — **станок**.

### Вывод

**Дескрипторы** **выигрывают** при: **многих** **полях**, **одинаковой** **логике**, **параметризации**. **`@property`** — **проще** **для** **малого** **числа** **полей** с **разной** **логикой**.

## 22.9. Дескрипторы в реальных проектах

Разберём **пять** **примеров** **из** **реальных** **проектов**.

### 22.9.1. Django ORM

**Поля** **модели** Django — **дескрипторы**:

```python
from django.db import models

class Apartment(models.Model):
    address = models.CharField(max_length=200)
    area = models.IntegerField()
    price = models.DecimalField(max_digits=12, decimal_places=2)
```

**Каждое** **поле** — **дескриптор**. При **чтении** — **загружает** **из** **БД**. При **записи** — **сохраняет**.

### 22.9.2. SQLAlchemy

**Аналогично**:

```python
from sqlalchemy import Column, Integer, String

class Apartment(Base):
    __tablename__ = "apartments"
    id = Column(Integer, primary_key=True)
    address = Column(String)
    area = Column(Integer)
```

### 22.9.3. attrs

Библиотека **attrs** использует **дескрипторы** для **полей**:

```python
import attr

@attr.s
class Apartment:
    address = attr.ib()
    area = attr.ib(validator=attr.validators.instance_of(int))
```

### 22.9.4. Pydantic

**Pydantic** использует **дескрипторы** для **валидации**:

```python
from pydantic import BaseModel, Field

class Apartment(BaseModel):
    address: str
    area: int = Field(ge=5, le=500)
    price: int = Field(gt=0)
```

### 22.9.5. Наш дом

```python
class RangeDescriptor:
    def __init__(self, min_value, max_value):
        self.min_value = min_value
        self.max_value = max_value

    def __set_name__(self, owner, name):
        self.name = name

    def __get__(self, instance, owner):
        if instance is None:
            return self
        return instance.__dict__.get(self.name)

    def __set__(self, instance, value):
        if not (self.min_value <= value <= self.max_value):
            raise ValueError(
                f"{self.name} должно быть от {self.min_value} до {self.max_value}"
            )
        instance.__dict__[self.name] = value


class Apartment:
    area = RangeDescriptor(5, 500)
    rooms = RangeDescriptor(1, 10)
    floor = RangeDescriptor(1, 50)
```

### Что общего

- **Много** **полей** **с** **одинаковой** **логикой**.
- **Параметризация** **проверок**.
- **Переиспользование** **кода**.

### Вывод

**Дескрипторы** — **основа** **многих** **фреймворков**. Django, SQLAlchemy, Pydantic, attrs — **все** **используют** их.

## 22.10. Нюансы и тонкости

Разберём **десять** **нюансов**.

### 22.10.1. `__set_name__` вызывается только для класса

Если **динамически** **добавлять** **дескриптор** **после** **создания** **класса** — `__set_name__` **не** **вызовется**:

```python
Apartment.new_field = RangeDescriptor(1, 10)    # __set_name__ НЕ вызовется!
```

**Решение:** **вручную** **вызвать** `__set_name__`.

### 22.10.2. Приоритет data-дескрипторов

**Data-дескрипторы** **имеют** **приоритет** **над** `instance.__dict__`. Поэтому **можно** **безопасно** **сохранять** **значения** **под** **тем** **же** **именем**.

### 22.10.3. Дескриптор один, значения разные

**Дескриптор** — **один** **объект** **класса**. **Значения** — **у** **каждого** **объекта**. **Не** **путайте!**

### 22.10.4. Наследование дескрипторов

**Дескрипторы** **наследуются**. `__set_name__` **не** **вызывается** **заново** — **имя** **остаётся** **из** **родителя**.

### 22.10.5. Дескрипторы и `__slots__`

**`__slots__`** **создаёт** **дескрипторы** **автоматически**. **Свои** **дескрипторы** **с** `__slots__` — **сложно** **совмещать**.

### 22.10.6. Дескрипторы в `@dataclass`

**`@dataclass`** **не** **поддерживает** **свои** **дескрипторы** **из** **коробки**. **Нужно** **использовать** `field()` или **ручную** **валидацию**.

### 22.10.7. Только для чтения

**Дескриптор** **без** `__set__` — **только** **для** **чтения**:

```python
class ReadOnly:
    def __get__(self, instance, owner):
        return "значение"

    def __set_name__(self, owner, name):
        self.name = name
```

**При** **попытке** **записи** — **создастся** **атрибут** `instance.__dict__[name]`. **При** **следующем** **чтении** — **вернётся** **это** **значение**, **а** **не** **результат** `__get__`.

### 22.10.8. Дескриптор-класс, а не функция

**Дескриптор** — **класс**. **`property`** — **тоже** **класс**. **Не** **путайте** **дескриптор** **с** **методом**.

### 22.10.9. Кэширование значений

**Дескриптор** **может** **кэшировать** **значения**:

```python
class Cached:
    def __get__(self, instance, owner):
        if self.name not in instance.__dict__:
            instance.__dict__[self.name] = self.compute(instance)
        return instance.__dict__[self.name]
```

**Полезно** для **дорогих** **вычислений**.

### 22.10.10. Дескрипторы и метаклассы

**Метаклассы** (глава 23) **могут** **использовать** **дескрипторы** для **свойств** **класса**.

### Вывод

**Дескрипторы** **имеют** **много** **нюансов**. **Основные:** **`__set_name__`**, **приоритет**, **наследование**, **только** **для** **чтения**.

---

## 22.11. Типичные ошибки

Разберём **десять** **самых** **частых** **ошибок**.

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Забыли `__set_name__` | `self.name` не установлен | Добавить метод |
| Хранят значение в дескрипторе | Общее для всех объектов | Хранить в `instance.__dict__` |
| Используют `self.value` вместо `instance.__dict__` | Общее значение | Использовать `instance.__dict__` |
| Забыли `__get__` | Не читается | Добавить метод |
| Возвращают `self` вместо значения | Неправильный результат | Возвращать значение |
| Дескриптор без `__set__` | Можно переопределить | Добавить `__set__` |
| Имя дескриптора = имя поля | Путаница | Использовать одинаково (безопасно) |
| `__set_name__` для наследника | Не вызывается | Использовать имя родителя |
| Конфликт с `@property` | `ValueError` | Не совмещать имена |
| Забыли `if instance is None` | Ошибка при обращении через класс | Добавить проверку |

### Подробный разбор

**Ошибка 1.** Забыли `__set_name__` — `self.name` **не** **установлен**. **Правило:** всегда **добавляйте** `__set_name__`.

**Ошибка 2.** Хранят значение **в** **дескрипторе**:

```python
class Bad:
    def __set__(self, instance, value):
        self.value = value    # ← ОБЩЕЕ для всех объектов!
```

**Правило:** храните в `instance.__dict__`.

**Ошибка 3.** Используют `self.value` **вместо** `instance.__dict__[self.name]` — **та** **же** **проблема**.

**Ошибка 4.** Забыли `__get__` — **не** **читается** **через** **дескриптор**.

**Ошибка 5.** Возвращают `self` **вместо** **значения** — **неправильный** **результат**.

**Ошибка 6.** Дескриптор **без** `__set__` — **можно** **переопределить** **через** `instance.__dict__`. **Правило:** для **контроля** — **добавьте** `__set__`.

**Ошибка 7.** Имя дескриптора = имя поля — **безопасно** для **data-дескрипторов**. **Правило:** **не** **бойтесь** **совпадения**.

**Ошибка 8.** `__set_name__` **для** **наследника** **не** **вызывается** — **имя** **из** **родителя**. **Правило:** **учитывайте** это.

**Ошибка 9.** Конфликт **с** `@property` — `ValueError`. **Правило:** **не** **совмещайте** **имена**.

**Ошибка 10.** Забыли `if instance is None` — **ошибка** **при** **обращении** **через** **класс**. **Правило:** **всегда** **проверяйте**.

### Вывод

Ошибки **с** **дескрипторами** **типичны**. **Основные:** **`__set_name__`**, **хранение** **значений**, **приоритет**. Следуя **правилам**, легко их **избежать**.

---

## 22.12. Что важно запомнить

- **Дескриптор** — класс с **методами** `__get__`, `__set__`, `__delete__`.
- **`__set_name__`** — **автоматически** **сообщает** **имя**.
- **Data-дескриптор** — **есть** `__set__` или `__delete__`. **Приоритет** **высший**.
- **Non-data** — **только** `__get__`. **Приоритет** **ниже**.
- **`@property`** — **встроенный** **дескриптор**.
- **Значения** — **в** `instance.__dict__`, **не** **в** **дескрипторе**.
- **Один** **дескриптор** — **много** **полей**.
- **Параметризация** — **через** `__init__`.
- **`__set_name__`** — **для** **универсальности**.
- **Приоритет** — data-дескриптор **>** `instance.__dict__` **>** non-data.
- **Дескрипторы** — **основа** **Django**, **SQLAlchemy**, **Pydantic**.
- **Используйте** **для** **многих** **полей** **с** **одинаковой** **логикой**.

## Задача 22 (дескрипторы)

Свойство `@property` **удобно**, но **если** **одинаковых** **проверок** **много** — **лучше** **вынести** **логику** **в** **дескриптор**.

Сравните **количество** **кода**.

- проверки **работают**.

**Часть 8. Кэширование (дополнительно).** Создайте **дескриптор** `CachedProperty`, который **кэширует** **результат** **вычисления**:

```python
class CachedProperty:
    def __init__(self, func):
        self.func = func

    def __set_name__(self, owner, name):
        self.name = name

    def __get__(self, instance, owner):
        if instance is None:
            return self
        if self.name not in instance.__dict__:
            instance.__dict__[self.name] = self.func(instance)
        return instance.__dict__[self.name]
```

Используйте для **вычисления** `price_per_sqm`.

**Часть 9.** В комментариях объясните:

1. Что такое **дескриптор** и **зачем** он **нужен**.
2. Что делает **каждый** **метод** **протокола**.
3. Чем **data-дескриптор** **отличается** от **non-data**.
4. Как **работает** `@property` **под** **капотом**.
5. Зачем **нужен** `__set_name__`.
6. Почему **значения** **хранятся** **в** `instance.__dict__`, **а** **не** **в** **дескрипторе**.
7. Когда **использовать** **дескриптор**, а **когда** — `@property`.

### Решение

```python
# ─── Часть 1. Дескриптор RangeDescriptor ───
class RangeDescriptor:
    def __init__(self, min_value, max_value):
        self.min_value = min_value
        self.max_value = max_value

    def __set_name__(self, owner, name):
        # Часть 6. Автоматически вызывается при создании класса.
        print(f"__set_name__ для {owner.__name__}.{name}")
        self.name = name

    def __get__(self, instance, owner):
        # При обращении через класс — возвращаем сам дескриптор.
        if instance is None:
            return self
        return instance.__dict__.get(self.name)

    def __set__(self, instance, value):
        # Проверяем тип.
        if not isinstance(value, (int, float)):
            raise TypeError(f"{self.name} должно быть числом")
        # Проверяем диапазон.
        if not (self.min_value <= value <= self.max_value):
            raise ValueError(
                f"{self.name} должно быть от {self.min_value} до {self.max_value}"
            )
        # Сохраняем в instance.__dict__.
        instance.__dict__[self.name] = value


# ─── Часть 2. Класс Apartment ───
class Apartment:
    area = RangeDescriptor(5, 500)
    rooms = RangeDescriptor(1, 10)
    floor = RangeDescriptor(1, 50)
    price = RangeDescriptor(0, 10**9)

    def __init__(self, address, area, rooms, floor, price):
        self.address = address
        self.area = area
        self.rooms = rooms
        self.floor = floor
        self.price = price

    def __repr__(self):
        return (
            f"Apartment({self.address!r}, area={self.area}, "
            f"rooms={self.rooms}, floor={self.floor}, price={self.price})"
        )


# ─── Часть 3. Проверка ───
print("=== Часть 3. Проверка ===")

flat = Apartment("ул. Ленина", 65, 3, 5, 8_500_000)
print(flat)

for field, value in [("area", -10), ("rooms", 20), ("floor", 0)]:
    try:
        setattr(flat, field, value)
    except ValueError as e:
        print(f"ValueError: {e}")

try:
    flat.area = "abc"
except TypeError as e:
    print(f"TypeError: {e}")

# Успешное изменение.
flat.area = 70
flat.rooms = 4
flat.floor = 6
flat.price = 9_000_000
print(f"После изменений: {flat}")

# ─── Часть 4. Приоритет data-дескриптора ───
print()
print("=== Часть 4. Приоритет data-дескриптора ===")

flat.__dict__["area"] = -100
print(f"flat.__dict__['area'] = {flat.__dict__['area']}")
print(f"flat.area = {flat.area}")   # 70 — из дескриптора, не -100
print("Дескриптор перекрывает instance.__dict__!")

# ─── Часть 5. @property vs дескриптор ───
print()
print("=== Часть 5. @property vs дескриптор ===")


class ApartmentProperty:
    def __init__(self, area):
        self.area = area

    @property
    def area(self):
        return self._area

    @area.setter
    def area(self, value):
        if not isinstance(value, (int, float)):
            raise TypeError("area должно быть числом")
        if not (5 <= value <= 500):
            raise ValueError("area должно быть от 5 до 500")
        self._area = value


# Дескриптор: одна строка.
class ApartmentDescriptor:
    area = RangeDescriptor(5, 500)

    def __init__(self, area):
        self.area = area


print("Property: ~15 строк для одного поля.")
print("Descriptor: 1 строка для каждого поля.")

# ─── Часть 7. Наследование ───
print()
print("=== Часть 7. Наследование ===")


class Studio(Apartment):
    pass    # __set_name__ НЕ вызывается повторно


studio = Studio("ул. Пушкина", 25, 1, 3, 4_000_000)
print(studio)
print(f"Дескриптор: {Studio.area.name}")    # "area" — из родителя

try:
    studio.area = -5
except ValueError as e:
    print(f"ValueError: {e}")

# ─── Часть 8. CachedProperty ───
print()
print("=== Часть 8. CachedProperty ===")


class CachedProperty:
    def __init__(self, func):
        self.func = func

    def __set_name__(self, owner, name):
        self.name = name

    def __get__(self, instance, owner):
        if instance is None:
            return self
        if self.name not in instance.__dict__:
            print(f"Вычисляем {self.name}...")
            instance.__dict__[self.name] = self.func(instance)
        return instance.__dict__[self.name]


class ApartmentCached:
    def __init__(self, area, price):
        self.area = area
        self.price = price

    @CachedProperty
    def price_per_sqm(self):
        return self.price / self.area


flat_cached = ApartmentCached(65, 8_500_000)
print(f"Первый вызов: {flat_cached.price_per_sqm:.0f}")
print(f"Второй вызов: {flat_cached.price_per_sqm:.0f}")   # Без повторного вычисления
```

**Ожидаемый вывод:**

```
=== Часть 3. Проверка ===
__set_name__ для Apartment.area
__set_name__ для Apartment.rooms
__set_name__ для Apartment.floor
__set_name__ для Apartment.price
Apartment('ул. Ленина', area=65, rooms=3, floor=5, price=8500000)
ValueError: area должно быть от 5 до 500
ValueError: rooms должно быть от 1 до 10
ValueError: floor должно быть от 1 до 50
TypeError: area должно быть числом
После изменений: Apartment('ул. Ленина', area=70, rooms=4, floor=6, price=9000000)

=== Часть 4. Приоритет data-дескриптора ===
flat.__dict__['area'] = -100
flat.area = 70
Дескриптор перекрывает instance.__dict__!

=== Часть 5. @property vs дескриптор ===
Property: ~15 строк для одного поля.
Descriptor: 1 строка для каждого поля.

=== Часть 7. Наследование ===
Apartment('ул. Пушкина', area=25, rooms=1, floor=3, price=4000000)
Дескриптор: area
ValueError: area должно быть от 5 до 500

=== Часть 8. CachedProperty ===
Вычисляем price_per_sqm...
Первый вызов: 130769
Второй вызов: 130769
```

### Разбор

**1. Что такое дескриптор и зачем он нужен.**

**Дескриптор** — **класс** с **методами** `__get__`, `__set__`, `__delete__`. **Определяет** **поведение** **атрибута** при **чтении**, **записи** и **удалении**.

**Зачем:**

- **Переиспользование** **логики** — **один** **дескриптор** **для** **многих** **полей**.
- **Параметризация** — **разные** **границы** **через** `__init__`.
- **Читаемость** — **код** **короче**.
- **Основа** **фреймворков** — Django, SQLAlchemy, Pydantic.

**2. Что делает каждый метод протокола.**

- **`__set_name__`** — **вызывается** **один** **раз** **при** **создании** **класса**. **Сохраняет** **имя** **дескриптора**.
- **`__get__`** — **вызывается** **при** **чтении**. **Возвращает** **значение**.
- **`__set__`** — **вызывается** **при** **записи**. **Проверяет** и **сохраняет**.
- **`__delete__`** — **вызывается** **при** **удалении**. **Удаляет** **значение**.

**3. Чем data-дескриптор отличается от non-data.**

**Data-дескриптор** — **есть** `__set__` **или** `__delete__`. **Приоритет** **высший** — **перекрывает** `instance.__dict__`.

**Non-data** — **только** `__get__`. **Приоритет** **ниже** — **можно** **переопределить** **через** `instance.__dict__`.

**Практически:** `@property` — **data-дескриптор**. **Обычные** **методы** — **non-data**.

**4. Как работает `@property` под капотом.**

`@property` — **встроенный** **дескриптор** с **методами** `__get__`, `__set__`, `__delete__`. **Геттер** — **функция**, **обёрнутая** **в** `property`. **Сеттер** — **через** `.setter()` **создаётся** **новый** `property` **с** **сеттером**.

**5. Зачем нужен `__set_name__`.**

**Без** `__set_name__` дескриптор **не** **знал** бы, **под** **каким** **именем** **сохранять** **значение**. **Приходилось** **бы** **передавать** **имя** **вручную** — **легко** **ошибиться**. `__set_name__` **вызывается** **автоматически** **при** **создании** **класса** и **сохраняет** **имя**.

**6. Почему значения хранятся в `instance.__dict__`, а не в дескрипторе.**

**Дескриптор** — **один** **объект** **класса**. **Все** **объекты** **класса** **делят** **его**. **Если** **хранить** **значение** **в** **дескрипторе** — **все** **объекты** **будут** **иметь** **одно** **значение**. **Значения** **должны** **быть** **у** **каждого** **объекта** **свои** — **поэтому** **в** `instance.__dict__`.

**7. Когда использовать дескриптор, а когда — `@property`.**

**Дескриптор:**

- **Много** **полей** с **одинаковой** **логикой**.
- **Параметризация** **проверок**.
- **Библиотеки** и **фреймворки**.

**`@property`:**

- **Мало** **полей**.
- **Разная** **логика** **для** **каждого**.
- **Вычисляемые** **свойства**.
- **Одноразовая** **проверка**.

# Глава 23. Метаклассы

## 23.1. Проблема: классы — тоже объекты

За предыдущие главы мы **много раз** создавали классы. `Apartment`, `Room`, `Agency`, `Agent`, `Appraiser`. Каждый раз мы **писали** `class Имя:` — и Python **создавал** класс. Это **привычно** и **понятно**.

Но вот **вопрос**, который **редко** **задают** новички: **а что такое класс?** Это **шаблон** для объектов. Это **чертёж**. Это **описание**. Всё **правильно**. Но есть **ещё** **одна** **сторона**: **класс — это тоже объект**. Причём **объект** **особого** **типа**.

### Класс — это объект

Проверим **экспериментально**:

```python
class Apartment:
    pass


print(type(Apartment))    # <class 'type'>
print(type(42))           # <class 'int'>
print(type("строка"))     # <class 'str'>
```

**Что мы видим:** `type(Apartment)` — это `type`. То есть **класс** `Apartment` — **объект** **класса** `type`. **Точно так** же, как **число** `42` — **объект** **класса** `int`, а **строка** — **объект** **класса** `str`.

**Вывод:** **класс** — это **объект**. Его **тип** — `type`. Класс `type` — это **метакласс**. **Метакласс** — это **класс** **классов**.

### Почему это важно

Если **класс** — **объект**, то **можно**:

- **Создавать** **классы** **динамически** — **во** **время** **выполнения**.
- **Изменять** **классы** **после** **создания**.
- **Управлять** **созданием** **классов** через **метаклассы**.

**Метаклассы** **позволяют** **вмешиваться** в **процесс** **создания** **класса**. **Проверять** **атрибуты**. **Добавлять** **методы** **автоматически**. **Регистрировать** **классы** **в** **реестрах**. **Обеспечивать** **синглтоны**.

### Аналогия

Представь **завод** по **производству** **чертежей**. Обычно **чертёж** — **конечный** **продукт** — **с** **ним** **работают**. Но **есть** **другой** **уровень**: **завод**, **который** **производит** **чертежи**. Это **метакласс**. Он **управляет** **тем**, **какими** **будут** **чертежи**.

### Почему это сложно

**Метаклассы** — **одна** из **самых** **сложных** **тем** Python. Их **редко** **используют** **напрямую**. Но **понимать** их **важно**:

- **Фреймворки** (Django, SQLAlchemy) **активно** **используют** метаклассы.
- **Чужой** **код** **может** **использовать** их — **нужно** **понимать**.
- **Собеседования** **часто** **спрашивают** про метаклассы.
- **Понимание** **метаклассов** **углубляет** **понимание** **ООП**.

### Вывод

**Класс** — это **объект** **класса** `type`. **Метакласс** — **класс** **классов**. **Метаклассы** **позволяют** **управлять** **созданием** **классов**. Это **продвинутая**, но **важная** тема.

---

## 23.2. Что такое метакласс

**Метакласс** — это **класс**, **экземплярами** **которого** **являются** **классы**. То есть **метакласс** **определяет**, **как** **создаётся** **класс**.

### Обычная аналогия

- **Объект** — **экземпляр** **класса**.
- **Класс** — **экземпляр** **метакласса**.

**Схема:**

```
        Метакласс (type)
                ↓
        Класс (Apartment)
                ↓
        Объект (flat)
```

**Читается** **так:** `flat` — **экземпляр** `Apartment`. `Apartment` — **экземпляр** `type`.

### Как это проверить

```python
class Apartment:
    pass

flat = Apartment()

print(type(flat))           # <class 'Apartment'> — объект
print(type(Apartment))      # <class 'type'> — класс — экземпляр метакласса
print(isinstance(flat, Apartment))     # True
print(isinstance(Apartment, type))     # True
```

**Три** **уровня:**

1. **Объект** `flat` — **экземпляр** `Apartment`.
2. **Класс** `Apartment` — **экземпляр** `type`.
3. **Метакласс** `type` — **экземпляр** **самого** **себя** (это **отдельная** **магия**).

### Стандартный метакласс — `type`

По **умолчанию** **метакласс** **всех** **классов** — `type`. Именно `type` **создаёт** **классы**.

**Но** `type` **можно** **вызывать** **двумя** **способами**:

**Способ 1: узнать тип объекта.**

```python
print(type(42))    # <class 'int'>
print(type("abc")) # <class 'str'>
```

**Способ 2: создать класс динамически.**

```python
Apartment = type("Apartment", (), {"address": "ул. Ленина"})
print(Apartment)             # <class '__main__.Apartment'>
print(Apartment.address)     # ул. Ленина
```

**Второй** **способ** — **это** **и** **есть** **создание** **класса** **через** **метакласс**! **Обычное** **объявление** `class Apartment:` — **синтаксический** **сахар** **над** **этим** **вызовом**.

### Синтаксис `type`

```python
type(name, bases, dict)
```

- **`name`** — **имя** **класса** (строка).
- **`bases`** — **кортеж** **родителей**.
- **`dict`** — **словарь** **атрибутов** и **методов**.

**Пример:**

```python
def info(self):
    return f"Квартира {self.address}"

Apartment = type(
    "Apartment",                    # имя
    (),                             # родители
    {                               # атрибуты и методы
        "address": "ул. Ленина",
        "info": info,
    }
)

flat = Apartment()
print(flat.info())    # Квартира ул. Ленина
```

### Схема

```
        class Apartment:               Apartment = type(
            address = "..."                "Apartment",
            def info(self): ...            (),
                                           {"address": "...", "info": ...}
        )                              )
                ↓                              ↓
        То же самое!
```

### Аналогия

Представь **документ** **и** **шаблон** **документа**. **Обычно** ты **работаешь** **с** **документами** — **заполняешь** **их**. Но **есть** **более** **высокий** **уровень** — **шаблон**, **по** **которому** **создаются** **документы**. Это **метакласс**.

### Вывод

**Метакласс** — **класс** **классов**. **Стандартный** — `type`. **`type`** **можно** **вызывать** для **динамического** **создания** **классов**. **Обычное** `class Имя:` — **синтаксический** **сахар** **над** `type(...)`.

---

## 23.3. Как создать свой метакласс

Разберём **создание** **своего** **метакласса** **пошагово**.

### Шаг 1. Наследование от `type`

**Метакласс** — **наследник** `type`:

```python
class MyMeta(type):
    pass
```

**Пока** **ничего** **не** **делает** — **просто** **наследник**.

### Шаг 2. Переопределение `__new__`

**Основной** **метод** **метакласса** — `__new__`. Он **вызывается** **при** **создании** **класса**.

```python
class MyMeta(type):
    def __new__(mcs, name, bases, namespace):
        # mcs — сам метакласс.
        # name — имя класса.
        # bases — родители.
        # namespace — словарь атрибутов и методов.
        print(f"Создаётся класс {name}")
        return super().__new__(mcs, name, bases, namespace)
```

**Что** **здесь** **происходит:**

- **`mcs`** — **сам** **метакласс** (аналог `cls` в классовом методе).
- **`name`** — **имя** **создаваемого** **класса**.
- **`bases`** — **кортеж** **родителей**.
- **`namespace`** — **словарь** **всех** **атрибутов** и **методов** класса.
- **`super().__new__(...)`** — **создаёт** **класс** **по-настоящему**.

### Шаг 3. Использование метакласса

```python
class Apartment(metaclass=MyMeta):
    address = "ул. Ленина"


# При создании класса выведется:
# Создаётся класс Apartment
```

**Что** **произошло:** Python **при** **создании** **класса** `Apartment` **вызвал** `MyMeta.__new__`. Тот **вывел** **сообщение** и **создал** **класс**.

### Шаг 4. Изменение класса в метаклассе

Метакласс **может** **изменять** **класс** **до** **его** **создания**:

```python
class AddInfoMeta(type):
    def __new__(mcs, name, bases, namespace):
        # Добавляем метод автоматически.
        namespace["get_class_name"] = lambda self: name
        return super().__new__(mcs, name, bases, namespace)


class Apartment(metaclass=AddInfoMeta):
    pass


flat = Apartment()
print(flat.get_class_name())    # Apartment
```

**Что** **произошло:** метакласс **добавил** **метод** `get_class_name` **всем** **классам**, **которые** **его** **используют**.

### Шаг 5. Проверка при создании

Метакласс **может** **проверять** **класс** **и** **выдавать** **ошибку**:

```python
class RequireDocstringMeta(type):
    def __new__(mcs, name, bases, namespace):
        # Проверяем наличие docstring.
        if "__doc__" not in namespace or namespace["__doc__"] is None:
            raise TypeError(f"Класс {name} должен иметь docstring")
        return super().__new__(mcs, name, bases, namespace)


class Apartment(metaclass=RequireDocstringMeta):
    """Квартира в агентстве."""
    pass


# Если бы docstring не было — была бы ошибка.
```

### Полный пример

```python
class LoggingMeta(type):
    def __new__(mcs, name, bases, namespace):
        print(f"[LOG] Создаётся класс: {name}")
        print(f"[LOG] Методы: {[k for k in namespace if not k.startswith('_')]}")
        return super().__new__(mcs, name, bases, namespace)


class Apartment(metaclass=LoggingMeta):
    def __init__(self, address):
        self.address = address

    def info(self):
        return f"Квартира: {self.address}"


# Вывод:
# [LOG] Создаётся класс: Apartment
# [LOG] Методы: ['__init__', 'info']
```

### Схема

```
        Создание класса Apartment(metaclass=MyMeta)
                ↓
        MyMeta.__new__(mcs, "Apartment", (), namespace)
                ↓
        Изменение namespace
                ↓
        super().__new__(mcs, name, bases, namespace)
                ↓
        Класс Apartment готов
```

### Аналогия

Представь **приёмную** **комиссию** **университета**. **Студент** **подаёт** **документы**. **Комиссия** **проверяет** **их** **по** **правилам**. **Может** **принять**, **может** **отклонить**, **может** **добавить** **что-то**. **Метакласс** — **такая** **комиссия** **для** **классов**.

### Вывод

**Свой** **метакласс** — **наследник** `type`. **Переопределяем** `__new__`. **Изменяем** **класс** **до** **создания**. **Используем** **через** `metaclass=...`.

---

## 23.4. Метакласс `SingletonMeta`

Разберём **практический** **пример** — **метакласс** для **синглтона**. **Синглтон** — **паттерн**, **при** **котором** **у** **класса** **только** **один** **экземпляр**.

### Реализация

```python
class SingletonMeta(type):
    _instances = {}    # словарь всех созданных объектов

    def __call__(cls, *args, **kwargs):
        # __call__ вызывается при Apartment(...).
        if cls not in cls._instances:
            # Первый раз — создаём объект.
            cls._instances[cls] = super().__call__(*args, **kwargs)
        # Возвращаем уже созданный.
        return cls._instances[cls]
```

### Разбор

- **`_instances`** — **словарь** **всех** **объектов**. **Ключ** — **класс**, **значение** — **объект**.
- **`__call__`** — **вызывается** **при** `Apartment(...)`. **Перехватывает** **создание**.
- **`super().__call__(...)`** — **создаёт** **объект** **по-настоящему**.
- **Если** **объект** **уже** **есть** — **возвращаем** **его**.

### Использование

```python
class Agency(metaclass=SingletonMeta):
    def __init__(self, name):
        self.name = name


agency1 = Agency("Дом-Сервис")
agency2 = Agency("Другое имя")

print(agency1 is agency2)    # True — один и тот же объект!
print(agency1.name)          # Дом-Сервис
print(agency2.name)          # Дом-Сервис
```

**Что** **произошло:** **первый** **вызов** `Agency(...)` **создал** **объект**. **Второй** **вызов** — **вернул** **тот** **же** **объект**. `agency2` — **это** **тот** **же** **объект**, **что** **и** `agency1`.

### Проверка

```python
print(agency1 is agency2)    # True
print(id(agency1) == id(agency2))    # True
print(Agency._instances)     # {<class 'Agency'>: <Agency object>}
```

### Почему это работает

- **`__call__`** **перехватывает** **вызов** `Agency(...)`.
- **`_instances`** — **общий** **словарь** **для** **всех** **классов** **с** **этим** **метаклассом**.
- **Если** **класс** **уже** **создан** — **возвращаем** **существующий** **объект**.

### Что важно

- **`_instances`** — **словарь** **на** **уровне** **метакласса**. **Один** **для** **всех** **классов**.
- **Каждый** **класс** — **свой** **ключ**. `Agency` и `Registry` — **разные** **ключи**.
- **`super().__call__(...)`** — **создаёт** **объект** **через** **обычный** **механизм**.

### Схема

```
        Agency("Дом-Сервис")
                ↓
        SingletonMeta.__call__(Agency, "Дом-Сервис")
                ↓
        Agency есть в _instances?
            ↓ (нет)
        super().__call__(...) — создаём
                ↓
        Сохраняем в _instances
                ↓
        Возвращаем объект

        Agency("Другое имя")
                ↓
        SingletonMeta.__call__(Agency, "Другое имя")
                ↓
        Agency есть в _instances?
            ↓ (да)
        Возвращаем существующий
```

### Аналогия

Представь **президента** **страны**. **В** **стране** — **один** **президент**. **Нельзя** **избрать** **второго**, **пока** **первый** **у** **власти**. **Синглтон** — **такая** **должность**. **Метакласс** — **избирательная** **комиссия**, **которая** **гарантирует** **единственность**.

### Вывод

**`SingletonMeta`** — **метакласс** для **синглтона**. **Через** `__call__` **перехватывает** **создание**. **Хранит** **объекты** **в** `_instances`. **Возвращает** **существующий** **объект** **при** **повторном** **вызове**.

---

## 23.5. Метакласс vs `__new__` для синглтона

**Синглтон** **можно** **реализовать** **двумя** **способами**: **через** **метакласс** и **через** `__new__`. Разберём **оба** и **сравним**.

### Способ 1: через метакласс

```python
class SingletonMeta(type):
    _instances = {}

    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]


class Agency(metaclass=SingletonMeta):
    def __init__(self, name):
        self.name = name
```

**Плюсы:**

- **Работает** **автоматически** **для** **всех** **классов** с **этим** **метаклассом**.
- **Наследники** **тоже** **синглтоны**.
- **Не** **нужно** **переопределять** `__new__` **в** **каждом** **классе**.

**Минусы:**

- **Метаклассы** — **сложная** **тема**.
- **Может** **конфликтовать** **с** **другими** **метаклассами**.
- **Трудно** **понять** **новичку**.

### Способ 2: через `__new__`

```python
class Agency:
    _instance = None

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

    def __init__(self, name):
        self.name = name
```

**Плюсы:**

- **Проще** **для** **понимания**.
- **Не** **нужен** **метакласс**.
- **Работает** **без** **дополнительных** **сложностей**.

**Минусы:**

- **Нужно** **переопределять** `__new__` **в** **каждом** **классе**.
- **Наследники** **не** **получают** **автоматически**.
- **Код** **дублируется**.

### Сравнение

| Характеристика | Метакласс | `__new__` |
|----------------|-----------|-----------|
| **Работает** для **всех** классов | ✅ Да | ❌ Нет |
| **Наследники** автоматически | ✅ Да | ❌ Нет |
| **Простота** | ⚠️ Сложно | ✅ Просто |
| **Конфликты** с другими метаклассами | ⚠️ Возможны | ✅ Нет |
| **Количество** кода | **Один** класс | **Каждый** класс |

### Когда что использовать

**Метакласс:**

- **Много** **классов** — **синглтонов**.
- **Библиотека** или **фреймворк**.
- **Автоматическое** **применение** **к** **наследникам**.

**`__new__`:**

- **Один** или **два** **класса**.
- **Простота** **важнее** **гибкости**.
- **Нет** **нужды** **в** **автоматике**.

### Пример: два класса-синглтона

**Метакласс:**

```python
class Agency(metaclass=SingletonMeta):
    pass

class Registry(metaclass=SingletonMeta):
    pass

# Оба — синглтоны, каждая — свой.
```

**`__new__`:**

```python
class Agency:
    _instance = None

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance


class Registry:
    _instance = None

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance
```

**Код** **дублируется**. **Плохо**.

### Аналогия

**Метакласс** — **автоматическая** **линия** на **заводе**. **`__new__`** — **ручная** **сборка** **каждой** **детали**. **Для** **одной** **детали** — **ручная** **быстрее**. **Для** **серии** — **автомат**.

### Вывод

**Метакласс** — **мощнее**, **но** **сложнее**. **`__new__`** — **проще**, **но** **требует** **дублирования**. **Выбор** — **по** **задаче**.

## 23.6. Зачем нужны метаклассы

**Метаклассы** — **редкая** **тема**. **В** **95%** **проектов** их **не** **пишут**. **Но** **они** **важны** **по** **трём** **причинам**.

### 23.6.1. Фреймворки

**Django**, **SQLAlchemy**, **Pydantic** **активно** **используют** **метаклассы**.

**Django ORM:**

```python
class Apartment(models.Model):
    address = models.CharField(max_length=200)
    area = models.IntegerField()
```

Django **через** **метакласс** **автоматически**:

- **Преобразует** **поля** в **колонки** **таблицы**.
- **Создаёт** **менеджер** `objects`.
- **Добавляет** **методы** `save`, `delete`.
- **Устанавливает** **связи** между **моделями**.

**SQLAlchemy:**

```python
class Apartment(Base):
    __tablename__ = "apartments"
    id = Column(Integer, primary_key=True)
    address = Column(String)
```

**Метакласс** **создаёт** **таблицу**, **связи**, **запросы**.

**Pydantic:**

```python
class Apartment(BaseModel):
    address: str
    area: int = Field(ge=5, le=500)
```

**Метакласс** **добавляет** **валидацию**, **сериализацию**, **парсинг**.

### 23.6.2. Понимание чужого кода

Если вы **работаете** **с** **Django** — **вы** **используете** **метаклассы** **косвенно**. **Понимание** **метаклассов** **помогает** **читать** **исходники** **фреймворков** и **отлаживать** **сложные** **проблемы**.

### 23.6.3. Собеседования

**Метаклассы** — **частый** **вопрос** на **собеседованиях**. Даже **если** **вы** **не** **пишете** их **каждый** **день** — **надо** **понимать**.

### 23.6.4. Гибкость фреймворков

**Метаклассы** **позволяют** **фреймворкам**:

- **Автоматически** **регистрировать** **классы**.
- **Проверять** **структуру** **классов**.
- **Добавлять** **методы** **и** **свойства**.
- **Создавать** **DSL** (**предметно-ориентированные** **языки**).

### 23.6.5. Пример: реестр

**Метакласс** **позволяет** **автоматически** **регистрировать** **все** **классы** **в** **реестре**:

```python
class RegistryMeta(type):
    _registry = {}

    def __new__(mcs, name, bases, namespace):
        cls = super().__new__(mcs, name, bases, namespace)
        if name not in ("Base",):
            RegistryMeta._registry[name] = cls
        return cls


class Base(metaclass=RegistryMeta):
    pass


class Apartment(Base):
    pass


class House(Base):
    pass


print(RegistryMeta._registry)
# {'Apartment': <class 'Apartment'>, 'House': <class 'House'>}
```

**Все** **классы** **автоматически** **попали** **в** **реестр**. **Без** **метакласса** — **нужно** **было** **бы** **регистрировать** **вручную** **в** **каждом** **классе**.

### Когда НЕ использовать

- **Простая** **задача** — **достаточно** **функции** или **декоратора**.
- **Можно** **обойтись** `__init_subclass__` (Python 3.6+).
- **Метаклассы** **усложняют** **код**.
- **Легко** **создать** **проблемы** **с** **наследованием**.

### Аналогия

**Метаклассы** — **как** **двигатель** **самолёта**. **В** **машине** — **не** **нужен**. **В** **самолёте** — **необходим**. **Дома** — **избыточен**. **Используйте** **только** **там**, **где** **нужно**.

### Вывод

**Метаклассы** **важны** **для**: **фреймворков**, **понимания** **чужого** **кода**, **собеседований**, **автоматизации**. **Но** **в** **обычном** **коде** — **используются** **редко**.

## 23.7. `__init_subclass__` — альтернатива метаклассам

Начиная с **Python 3.6** есть **более** **простой** **способ** **вмешаться** в **создание** **класса** — `__init_subclass__`.

### Синтаксис

```python
class Base:
    def __init_subclass__(cls, **kwargs):
        super().__init_subclass__(**kwargs)
        print(f"Создан наследник: {cls.__name__}")


class Apartment(Base):
    pass


# Вывод: Создан наследник: Apartment
```

### Что делает

**`__init_subclass__`** **вызывается** **автоматически** **при** **создании** **наследника**. Он **проще** **метакласса** — **не** **нужно** **создавать** **новый** **класс**.

### Преимущества перед метаклассом

- **Проще** **синтаксис**.
- **Не** **конфликтует** **с** **метаклассами**.
- **Понятнее** **новичку**.
- **Достаточно** **для** **большинства** **задач**.

### Когда что использовать

| Задача | Метакласс | `__init_subclass__` |
|--------|-----------|---------------------|
| **Регистрация** наследников | ✅ Да | ✅ Да |
| **Проверка** структуры | ✅ Да | ✅ Да |
| **Изменение** класса | ✅ Да | ⚠️ Ограничено |
| **Управление** созданием экземпляров | ✅ Да | ❌ Нет |
| **Простота** | ⚠️ Сложно | ✅ Просто |

### Пример: регистрация через `__init_subclass__`

```python
class Registry:
    _registry = {}

    def __init_subclass__(cls, **kwargs):
        super().__init_subclass__(**kwargs)
        Registry._registry[cls.__name__] = cls


class Apartment(Registry):
    pass


class House(Registry):
    pass


print(Registry._registry)
# {'Apartment': <class 'Apartment'>, 'House': <class 'House'>}
```

**Работает** **как** **метакласс**, **но** **проще**.

### Правило

**Если** **можно** **обойтись** `__init_subclass__` — **используйте** его. **Метакласс** — **только** **когда** **нужно** **больше** (например, **управлять** **созданием** **экземпляров**).

### Аналогия

`__init_subclass__` — **как** **звонок** **родителю** **после** **рождения** **ребёнка**. **Метакласс** — **как** **генетическая** **инженерия** **до** **рождения**. **Первое** — **проще** и **безопаснее**. **Второе** — **мощнее**, **но** **рискованнее**.

### Вывод

`__init_subclass__` — **простая** **альтернатива** **метаклассам**. **Используйте** **его** **по** **умолчанию**. **Метаклассы** — **только** **при** **необходимости**.

## 23.8. Метаклассы в реальных проектах

Разберём **пять** **примеров**.

### 23.8.1. Django ORM

**Django** **использует** **метаклассы** для **моделей**:

```python
class Apartment(models.Model):
    address = models.CharField(max_length=200)
    area = models.IntegerField()
```

**Метакласс** `ModelBase`:

- **Читает** **поля**.
- **Создаёт** **менеджер** `objects`.
- **Устанавливает** **связи**.
- **Добавляет** **методы**.

### 23.8.2. SQLAlchemy

```python
class Apartment(Base):
    __tablename__ = "apartments"
    id = Column(Integer, primary_key=True)
```

**Метакласс** `DeclarativeMeta` **создаёт** **таблицу** и **связи**.

### 23.8.3. Pydantic

```python
class Apartment(BaseModel):
    address: str
    area: int = Field(ge=5, le=500)
```

**Метакласс** `ModelMetaclass`:

- **Проверяет** **типы**.
- **Добавляет** **валидацию**.
- **Создаёт** **парсинг** и **сериализацию**.

### 23.8.4. ABC (Abstract Base Classes)

**Стандартная** **библиотека** `abc` **использует** **метакласс** `ABCMeta`:

```python
from abc import ABC

class HomeItem(ABC):
    pass
```

`ABCMeta` **проверяет**, что **все** **абстрактные** **методы** **реализованы**.

### 23.8.5. Наш дом

```python
class RegistryMeta(type):
    _registry = {}

    def __new__(mcs, name, bases, namespace):
        cls = super().__new__(mcs, name, bases, namespace)
        if name != "Base":
            RegistryMeta._registry[name] = cls
        return cls


class Base(metaclass=RegistryMeta):
    pass


class Apartment(Base):
    pass


class House(Base):
    pass


print(RegistryMeta._registry)    # {'Apartment': ..., 'House': ...}
```

### Что общего

- **Автоматическая** **регистрация**.
- **Добавление** **методов**.
- **Проверка** **структуры**.
- **Создание** **DSL**.

### Вывод

**Метаклассы** — **основа** **многих** **фреймворков**. **Django**, **SQLAlchemy**, **Pydantic**, **ABC** — **все** **используют** их.

## 23.9. Нюансы и тонкости

Разберём **десять** **нюансов**.

### 23.9.1. `type` — не только метакласс

`type` — **функция** и **метакласс** **одновременно**. Как **функция** — **возвращает** **тип**. Как **класс** — **создаёт** **классы**.

### 23.9.2. `metaclass=...` в объявлении класса

```python
class Apartment(metaclass=MyMeta):
    pass
```

**Синтаксис** **указания** **метакласса**.

### 23.9.3. Конфликт метаклассов

Если **родители** **имеют** **разные** **метаклассы** — **нужен** **общий** **наследник**:

```python
class Meta1(type): pass
class Meta2(type): pass

class CombinedMeta(Meta1, Meta2): pass

class Base1(metaclass=Meta1): pass
class Base2(metaclass=Meta2): pass

class Child(Base1, Base2, metaclass=CombinedMeta):
    pass
```

### 23.9.4. `__new__` vs `__init__` в метаклассе

`__new__` — **создаёт** **класс**. `__init__` — **инициализирует** **после** **создания**. **Обычно** **используют** `__new__`.

### 23.9.5. `mcs` — не `cls` и не `self`

В **метаклассе** **первый** **параметр** **называют** `mcs` (**metaclass**) — **соглашение**. Не `cls` (это **класс**) и не `self` (это **объект**).

### 23.9.6. Метаклассы и наследование

Если **родитель** **имеет** **метакласс** — **наследник** **получает** **его** **автоматически**:

```python
class Base(metaclass=MyMeta): pass
class Child(Base): pass   # metaclass=MyMeta автоматически
```

### 23.9.7. Метакласс и `__slots__`

Метакласс **может** **добавлять** `__slots__` **автоматически** — **полезно** для **оптимизации**.

### 23.9.8. Метакласс и `@dataclass`

`@dataclass` **не** **использует** **метакласс** — **использует** **декоратор**. **Не** **путайте**.

### 23.9.9. Порядок MRO с метаклассами

Если **несколько** **метаклассов** — **они** **идут** **по** **MRO**, **как** **обычные** **классы**.

### 23.9.10. Отладка метаклассов

**Метаклассы** **сложно** **отлаживать**. **Используйте** `print` **в** `__new__` и `__init__`.

### Вывод

**Метаклассы** **имеют** **много** **нюансов**. **Основные:** **конфликты**, **порядок** **вызовов**, **наследование**.

## 23.10. Типичные ошибки

Разберём **десять** **самых** **частых** **ошибок**.

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Забыли `metaclass=` | Метакласс не применяется | Указать `metaclass=MyMeta` |
| Используют `cls` вместо `mcs` | Путаница | `mcs` для метакласса |
| Забыли `super().__new__()` | Класс не создаётся | Вызвать `super()` |
| Возвращают `None` | Класс не создаётся | Вернуть класс |
| `__init__` вместо `__new__` | Не работает как надо | Использовать `__new__` |
| Конфликт метаклассов | Ошибка MRO | Создать общий метакласс |
| Много метаклассов | Сложно | Использовать `__init_subclass__` |
| Не проверяют `name` | Базовая логика не работает | Проверять |
| Метакласс для всего | Избыточно | Использовать декораторы |
| Забыли про наследование | Наследники без метакласса | Учитывать |

### Подробный разбор

**Ошибка 1.** Забыли `metaclass=` — **метакласс** **не** **применяется**. **Правило:** **указывайте** **явно**.

**Ошибка 2.** `cls` вместо `mcs` — **путаница**. **Правило:** `mcs` для **метакласса**.

**Ошибка 3.** Забыли `super().__new__()` — **класс** **не** **создаётся**. **Правило:** **вызывайте** `super()`.

**Ошибка 4.** Возвращают `None` — **класс** **не** **создаётся**. **Правило:** **возвращайте** **класс**.

**Ошибка 5.** `__init__` вместо `__new__` — **не** **работает** **как** **надо**. **Правило:** `__new__`.

**Ошибка 6.** Конфликт **метаклассов** — **ошибка** **MRO**. **Правило:** **создайте** **общий**.

**Ошибка 7.** Много **метаклассов** — **сложно**. **Правило:** `__init_subclass__`.

**Ошибка 8.** Не **проверяют** `name` — **базовая** **логика** **не** **работает**. **Правило:** **проверяйте**.

**Ошибка 9.** Метакласс для **всего** — **избыточно**. **Правило:** **декораторы** **часто** **лучше**.

**Ошибка 10.** Забыли **про** **наследование** — **наследники** **без** **метакласса**. **Правило:** **учитывайте**.

### Вывод

Ошибки **с** **метаклассами** **типичны**. **Основные:** **`super()`**, **`mcs`**, **наследование**.

## 23.11. Что важно запомнить

- **Класс** — **объект** **класса** `type`.
- **Метакласс** — **класс** **классов**.
- **Стандартный** метакласс — `type`.
- **`type(name, bases, dict)`** — **создаёт** **класс** **динамически**.
- **Свой** **метакласс** — **наследник** `type`.

- **`mcs`** — **первый** **параметр** **в** **метаклассе**.
- **`metaclass=MyMeta`** — **синтаксис** **применения**.
- **Метаклассы** **наследуются** **автоматически**.
- **`SingletonMeta`** — **классический** **пример**.
- **`__init_subclass__`** — **простая** **альтернатива**.
- **Фреймворки** — **Django**, **SQLAlchemy**, **Pydantic** — **используют** **метаклассы**.
- **В** **95%** **проектов** — **метаклассы** **не** **нужны**.

## Задача 23 (метаклассы)

В агентстве **должен** **быть** **только** **один** **реестр** **объектов**. **Если** **создать** **второй** — **данные** **могут** **разойтись**. **Метакласс** **гарантирует** **единственность**.

**Часть 1. Метакласс `SingletonMeta`.** Создайте **метакласс** `SingletonMeta`, который **хранит** **единственный** **экземпляр** **каждого** **класса**:

- **словарь** `_instances` — **общий** **для** **всех** **классов**;
- **метод** `__call__` — **проверяет**, **есть** **ли** **уже** **объект**;
- **если** **нет** — **создаёт** через `super().__call__(...)`;
- **если** **есть** — **возвращает** **существующий**.

**Часть 2. Класс `Agency`.** Создайте класс `Agency(metaclass=SingletonMeta)` с полем `name` и методом `add_apartment`.

**Часть 3. Проверка.** Продемонстрируйте:

- **создание** **двух** `Agency` — **убедитесь**, что **это** **один** **объект**;
- `agency1 is agency2` → `True`;
- `id(agency1) == id(agency2)` → `True`;
- **изменения** в `agency1` **видны** в `agency2`.

**Часть 4. Второй синглтон.** Создайте **второй** класс `Registry(metaclass=SingletonMeta)`. Убедитесь, что `Agency` и `Registry` — **разные** **синглтоны**.

**Часть 5. Метакласс с логированием.** Создайте **метакласс** `LoggingMeta`:

- **при** **создании** **класса** **выводит** **имя**;
- **выводит** **список** **методов** класса;
- **применяется** **к** `Apartment`.

**Часть 6. Метакласс с проверкой.** Создайте **метакласс** `RequireDocstringMeta`:

- **требует** **наличие** `docstring` **у** **класса**;
- **если** **нет** — **выбрасывает** `TypeError`.

**Часть 7. `__init_subclass__`.** **Перепишите** **регистрацию** **через** `__init_subclass__`:

- **базовый** класс `Base`;
- **при** **создании** **наследника** — **регистрирует** **его** **в** **словаре**;
- **убедитесь**, что `Apartment` и `House` **попали** **в** **реестр**.

**Часть 8.** В комментариях объясните:

1. Что **такое** **метакласс** и **как** **он** **связан** с **классом**.
2. **Зачем** **нужен** `__new__` **в** **метаклассе**.
3. Как **работает** `SingletonMeta`.
4. Чем **метакласс** **отличается** от `__new__` **для** **синглтона**.
5. Когда **использовать** **метаклассы**, а **когда** — `__init_subclass__`.
6. **Почему** **метаклассы** **редко** **используются** в **обычном** **коде**.

### Решение

```python
# ─── Часть 1. SingletonMeta ───
class SingletonMeta(type):
    _instances = {}    # общий словарь для всех классов

    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            print(f"[SingletonMeta] Создаём {cls.__name__}")
            cls._instances[cls] = super().__call__(*args, **kwargs)
        else:
            print(f"[SingletonMeta] Возвращаем {cls.__name__}")
        return cls._instances[cls]


# ─── Часть 2. Класс Agency ───
class Agency(metaclass=SingletonMeta):
    def __init__(self, name):
        self.name = name
        self.apartments = []

    def add_apartment(self, address):
        self.apartments.append(address)

    def __repr__(self):
        return f"Agency({self.name!r}, {len(self.apartments)} квартир)"


# ─── Часть 4. Второй синглтон ───
class Registry(metaclass=SingletonMeta):
    def __init__(self):
        self.items = []

    def add(self, item):
        self.items.append(item)


# ─── Часть 5. LoggingMeta ───
class LoggingMeta(type):
    def __new__(mcs, name, bases, namespace):
        methods = [k for k in namespace if not k.startswith("_")]
        print(f"[LoggingMeta] Создаётся класс: {name}")
        print(f"[LoggingMeta] Методы: {methods}")
        return super().__new__(mcs, name, bases, namespace)


class Apartment(metaclass=LoggingMeta):
    def __init__(self, address, area):
        self.address = address
        self.area = area

    def info(self):
        return f"{self.address}: {self.area} м²"

    def price_per_sqm(self, price):
        return price / self.area


# ─── Часть 6. RequireDocstringMeta ───
class RequireDocstringMeta(type):
    def __new__(mcs, name, bases, namespace):
        if "__doc__" not in namespace or namespace["__doc__"] is None:
            raise TypeError(f"Класс {name} должен иметь docstring")
        return super().__new__(mcs, name, bases, namespace)


class DocumentedApartment(metaclass=RequireDocstringMeta):
    """Квартира с обязательным docstring."""
    pass


try:
    class BadApartment(metaclass=RequireDocstringMeta):
        pass    # нет docstring
except TypeError as e:
    print(f"Ошибка: {e}")

# ─── Часть 7. __init_subclass__ ───
class Base:
    _registry = {}

    def __init_subclass__(cls, **kwargs):
        super().__init_subclass__(**kwargs)
        Base._registry[cls.__name__] = cls
        print(f"[Base] Зарегистрирован: {cls.__name__}")


class House(Base):
    pass


class Villa(Base):
    pass


# ─── Часть 3. Проверка SingletonMeta ───
print()
print("=== Часть 3. Проверка SingletonMeta ===")

agency1 = Agency("Дом-Сервис")
agency1.add_apartment("ул. Ленина, д. 10")

agency2 = Agency("Другое имя")   # name не изменится!

print(f"agency1 is agency2: {agency1 is agency2}")            # True
print(f"id равны: {id(agency1) == id(agency2)}")              # True
print(f"agency1.name: {agency1.name}")                        # Дом-Сервис
print(f"agency2.name: {agency2.name}")                        # Дом-Сервис (то же!)
print(f"agency1.apartments: {agency1.apartments}")
print(f"agency2.apartments: {agency2.apartments}")            # то же!

# ─── Часть 4. Второй синглтон ───
print()
print("=== Часть 4. Второй синглтон ===")

reg1 = Registry()
reg2 = Registry()
reg1.add("квартира 1")

print(f"reg1 is reg2: {reg1 is reg2}")            # True
print(f"agency1 is reg1: {agency1 is reg1}")      # False
print(f"reg1.items: {reg1.items}")
print(f"reg2.items: {reg2.items}")                # то же

# ─── Часть 7. Проверка __init_subclass__ ───
print()
print("=== Часть 7. Реестр через __init_subclass__ ===")
print(f"Base._registry: {list(Base._registry.keys())}")    # ['House', 'Villa']

# ─── Часть 5. Проверка Apartment (LoggingMeta) ───
print()
print("=== Часть 5. Apartment с LoggingMeta ===")
flat = Apartment("ул. Ленина", 65)
print(flat.info())
print(f"price_per_sqm: {flat.price_per_sqm(8_500_000):.0f}")
```

**Ожидаемый вывод (примерный):**

```
[LoggingMeta] Создаётся класс: Apartment
[LoggingMeta] Методы: ['__init__', 'info', 'price_per_sqm']
Ошибка: Класс BadApartment должен иметь docstring
[Base] Зарегистрирован: House
[Base] Зарегистрирован: Villa

=== Часть 3. Проверка SingletonMeta ===
[SingletonMeta] Создаём Agency
[SingletonMeta] Возвращаем Agency
agency1 is agency2: True
id равны: True
agency1.name: Дом-Сервис
agency2.name: Дом-Сервис
agency1.apartments: ['ул. Ленина, д. 10']
agency2.apartments: ['ул. Ленина, д. 10']

=== Часть 4. Второй синглтон ===
[SingletonMeta] Создаём Registry
[SingletonMeta] Возвращаем Registry
reg1 is reg2: True
agency1 is reg1: False
reg1.items: ['квартира 1']
reg2.items: ['квартира 1']

=== Часть 7. Реестр через __init_subclass__ ===
Base._registry: ['House', 'Villa']

=== Часть 5. Apartment с LoggingMeta ===
ул. Ленина: 65 м²
price_per_sqm: 130769
```

### Разбор

**1. Что такое метакласс и как он связан с классом.**

**Метакласс** — **класс** **классов**. **Класс** — **экземпляр** **метакласса**. По **умолчанию** — `type`. **Свой** **метакласс** — **наследник** `type`.

**Связь:**

- `flat` — **экземпляр** `Apartment`.
- `Apartment` — **экземпляр** `SingletonMeta`.
- `SingletonMeta` — **экземпляр** `type`.

**2. Зачем нужен `__new__` в метаклассе.**

`__new__` **вызывается** **при** **создании** **класса**. В **нём** **можно**:

- **Проверить** **структуру** **класса**.
- **Добавить** **методы**.
- **Изменить** **атрибуты**.
- **Зарегистрировать** **класс**.
- **Выдать** **ошибку**, **если** **что-то** **не** **так**.

**3. Как работает `SingletonMeta`.**

- **`_instances`** — **словарь** **на** **уровне** **метакласса**.
- **`__call__`** **перехватывает** **вызов** `Agency(...)`.
- **Если** **класс** **не** **в** `_instances` — **создаём** **объект**.
- **Если** **в** `_instances` — **возвращаем** **существующий**.

**4. Чем метакласс отличается от `__new__` для синглтона.**

**Метакласс:**

- **Работает** **для** **всех** **классов** **автоматически**.
- **Наследники** — **тоже** **синглтоны**.
- **Один** **раз** **написали** — **используем** **везде**.

**`__new__`:**

- **Нужно** **переопределять** **в** **каждом** **классе**.
- **Наследники** **не** **получают** **автоматически**.
- **Проще** **для** **понимания**.

**5. Когда использовать метаклассы, а когда — `__init_subclass__`.**

**Метаклассы:**

- **Управление** **созданием** **экземпляров** (как в `SingletonMeta`).
- **Изменение** **класса** **до** **создания**.
- **Сложная** **логика** **создания** **классов**.

**`__init_subclass__`:**

- **Регистрация** **наследников**.
- **Проверка** **структуры**.
- **Простая** **логика**.

**Правило:** **если** **можно** `__init_subclass__` — **используйте** его.

**6. Почему метаклассы редко используются в обычном коде.**

- **Сложны** **для** **понимания**.
- **Легко** **создать** **проблемы**.
- **Обычно** **есть** **более** **простое** **решение**.
- **Нужны** **только** **в** **фреймворках** и **библиотеках**.
- **В** **95%** **проектов** — **не** **нужны**.

# Глава 24. Паттерн Singleton

## 24.1. Проблема: когда нужен один объект на всю программу

За предыдущие главы мы построили агентство недвижимости: классы `Apartment`, `Room`, `Agency`, `Agent`, `Appraiser`. Всё работает. Но теперь представь **одну** **важную** **деталь**: у агентства должен быть **один** **реестр** **всех** **квартир**. Не два, не три — **ровно** **один**. Если **кто-то** **создаст** **второй** **реестр** — данные **разойдутся**: в одном будут одни квартиры, в другом — другие. Клиент **придёт** в **агентство**, а **информация** **окажется** **разной** в **разных** **частях** **системы**. **Катастрофа**.

**Первое**, что **приходит** **в** **голову** — **запретить** **создавать** **второй** **реестр**. Но как? Python **не** **даёт** **таких** **запретов** **напрямую**. Можно **написать** **комментарий** «не создавайте второй», но **это** **не** **работает** — **кто-нибудь** **забудет**, **и** **проблема** **вернётся**.

**Второе** — **сделать** **реестр** **глобальной** **переменной**:

```python
registry = {}
```

Но **глобальные** **переменные** — **плохая** **практика**. Они **засоряют** **пространство** **имён**, **их** **трудно** **отслеживать**, **они** **не** **привязаны** **к** **классу**. И **главное** — **их** **можно** **случайно** **перезаписать**.

**Третье** — **создать** **класс**, **который** **сам** **следит** за **своей** **единственностью**. Он **не** **даст** **создать** **второй** **экземпляр**, **а** **вернёт** **уже** **существующий**. Это **и** **есть** **паттерн** **Singleton**.

### Аналогия

Представь **президента** **страны**. В **стране** — **один** **президент**. **Нельзя** **избрать** **второго**, **пока** **первый** **у** **власти**. **Все** **обращения** **идут** **к** **одному** **человеку**. Если **бы** **было** **два** **президента** — **начался** **бы** **хаос**: **два** **набора** **указов**, **две** **внешние** **политики**, **две** **армии**. **Singleton** — **такая** **должность**: **гарантирует**, **что** **руководитель** **один**.

### Что такое Singleton

**Singleton** (Одиночка) — это **порождающий** **паттерн** **проектирования**, **который**:

1. **Гарантирует**, что у **класса** **есть** **только** **один** **экземпляр**.
2. **Предоставляет** **глобальную** **точку** **доступа** к **этому** **экземпляру**.

Простыми **словами**: **Singleton** — это **класс**, **который** **при** **любом** **вызове** **возвращает** **один** **и** **тот** **же** **объект**.

### Где нужен Singleton

- **Реестр** **объектов** — **один** **список** **всех** **квартир**.
- **Конфигурация** **приложения** — **один** **набор** **настроек**.
- **Логгер** — **один** **журнал** **событий**.
- **Кэш** — **один** **набор** **кэшированных** **данных**.
- **Пул** **соединений** — **один** **набор** **подключений** к **БД**.

### Вывод

**Singleton** решает **проблему** **единственности** **объекта**. Он **гарантирует**, что **класс** **имеет** **один** **экземпляр**, и **даёт** **глобальную** **точку** **доступа**. Это **нужно**, **когда** **данные** **должны** **быть** **согласованы** **во** **всей** **программе**.

---

## 24.2. Что такое паттерн Singleton

**Паттерн Singleton** — это **классический** **паттерн** **проектирования** **из** **книги** **«Банды** **четырёх»** (GoF). Он **относится** к **порождающим** **паттернам** — то **есть** **управляет** **созданием** **объектов**.

### Ключевая идея

**Singleton** **работает** **так**:

1. **Конструктор** **класса** **скрыт** (или **перехвачен**).
2. **Создание** **объекта** **происходит** **через** **специальный** **метод**.
3. **Метод** **проверяет**: **есть** **ли** **уже** **объект**.
4. **Если** **есть** — **возвращает** **существующий**.
5. **Если** **нет** — **создаёт** **новый** и **сохраняет** его.

### Схема

```
        Первый вызов Registry():
                ↓
        Объект уже создан?
            ↓ (нет)
        Создаём объект
                ↓
        Сохраняем в _instance
                ↓
        Возвращаем объект

        Второй вызов Registry():
                ↓
        Объект уже создан?
            ↓ (да)
        Возвращаем существующий объект
```

### Что даёт Singleton

- **Единственность** — **ровно** **один** **объект**.
- **Глобальный** **доступ** — **любой** **код** **может** **получить** **объект**.
- **Согласованность** — **все** **видят** **одни** **и** **те** **же** **данные**.
- **Ленивая** **инициализация** — **объект** **создаётся** **при** **первом** **обращении**.

### Что НЕ даёт Singleton

- **Потокобезопасность** — в **многопоточных** **программах** **нужны** **дополнительные** **меры**.
- **Тестируемость** — **сложно** **подменить** **объект** в **тестах**.
- **Гибкость** — **нельзя** **создать** **второй** **объект** **для** **других** **целей**.

### Аналогия

Представь **диспетчера** **такси**. Все **заказы** **идут** **к** **одному** **диспетчеру**. **Он** **знает** **всех** **водителей**, **все** **заказы**, **всё** **расписание**. **Если** **бы** **было** **два** **диспетчера** — **они** **бы** **путались**, **какой** **заказ** **кому** **отдан**. **Singleton** — **такой** **диспетчер**.

### Отличие от глобальной переменной

| Характеристика | Глобальная переменная | Singleton |
|----------------|----------------------|-----------|
| Где живёт | В модуле | В классе |
| Инкапсуляция | ❌ Нет | ✅ Да |
| Ленивая инициализация | ❌ Нет | ✅ Да |
| Наследование | ❌ Нет | ✅ Да |
| Контроль создания | ❌ Нет | ✅ Да |

**Singleton** — **это** **не** **просто** **глобальная** **переменная**. Это **класс**, **который** **сам** **управляет** **своей** **единственностью**.

### Вывод

**Singleton** — **порождающий** **паттерн**. **Гарантирует** **единственность** **объекта** и **даёт** **глобальный** **доступ**. **Лучше** **глобальной** **переменной**, **потому** **что** **инкапсулирует** **логику** и **даёт** **контроль**.

---

## 24.3. Способы реализации Singleton в Python

В Python **есть** **несколько** **способов** **реализовать** **Singleton**. Разберём **четыре** **основных**.

### 24.3.1. Через `__new__`

**Самый** **простой** **способ**. Переопределяем `__new__` — **метод**, **который** **создаёт** **объект**.

```python
class Agency:
    _instance = None

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

    def __init__(self, name):
        self.name = name
```

**Как** **работает:**

- **`_instance`** — **атрибут** **класса** для **хранения** **единственного** **объекта**.
- **`__new__`** **вызывается** **при** `Agency(...)`.
- **Если** `_instance` **пуст** — **создаём** **объект**.
- **Возвращаем** **существующий** **объект**.

**Проверка:**

```python
a1 = Agency("Дом-Сервис")
a2 = Agency("Другое имя")

print(a1 is a2)       # True
print(a1.name)        # Другое имя — __init__ вызвался дважды!
```

**Проблема:** `__init__` **вызывается** **каждый** **раз**! **Даже** **если** **объект** **уже** **создан**. Поэтому `name` **изменился** на **второй** **вызов**. **Нужно** **защитить** `__init__`:

```python
class Agency:
    _instance = None

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

    def __init__(self, name):
        if not hasattr(self, "name"):
            self.name = name
```

Теперь `__init__` **работает** **только** **один** **раз**.

### 24.3.2. Через метакласс

**Способ** **из** **главы 23**. **Работает** **автоматически** **для** **всех** **классов**.

```python
class SingletonMeta(type):
    _instances = {}

    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]


class Agency(metaclass=SingletonMeta):
    def __init__(self, name):
        self.name = name
```

**Плюсы:**

- **Автоматически** **для** **всех** **классов**.
- **`__init__`** **вызывается** **только** **один** **раз** (потому **что** `__call__` **перехватывает** **создание**).
- **Наследники** — **тоже** **синглтоны**.

### 24.3.3. Через декоратор

**Способ** **для** **тех**, **кто** **не** **хочет** **метаклассов**. **Декоратор** **оборачивает** **класс**.

```python
def singleton(cls):
    instances = {}

    def get_instance(*args, **kwargs):
        if cls not in instances:
            instances[cls] = cls(*args, **kwargs)
        return instances[cls]

    return get_instance


@singleton
class Agency:
    def __init__(self, name):
        self.name = name
```

**Плюсы:**

- **Проще** **метакласса**.
- **Работает** **для** **любого** **класса**.

**Минусы:**

- **Возвращает** **функцию**, **не** **класс**. **`isinstance`** **может** **не** **работать**.
- **Не** **работает** **с** **наследованием**.

### 24.3.4. Через модуль

**Самый** **простой** **способ** — **использовать** **модуль** **Python**. **Модуль** **загружается** **один** **раз**, **значит** **объект** в **нём** — **единственный**.

```python
# registry.py
class _Registry:
    def __init__(self):
        self.apartments = []


registry = _Registry()
```

**Использование:**

```python
from registry import registry

registry.apartments.append("ул. Ленина")
```

**Плюсы:**

- **Максимально** **просто**.
- **Работает** **из** **коробки**.
- **Python** **гарантирует** **единственность** **модуля**.

**Минусы:**

- **Не** **класс** — **нельзя** **наследовать**.
- **Нет** **ленивой** **инициализации** — **создаётся** **при** **импорте**.

### Сравнение способов

| Способ | Сложность | `__init__` 1 раз | Наследование | `isinstance` |
|--------|-----------|------------------|--------------|--------------|
| `__new__` | ✅ Просто | ❌ Нужна защита | ✅ Да | ✅ Да |
| Метакласс | ⚠️ Сложно | ✅ Да | ✅ Да | ✅ Да |
| Декоратор | ✅ Просто | ✅ Да | ❌ Нет | ⚠️ Частично |
| Модуль | ✅ Просто | ✅ Да | ❌ Нет | ❌ Нет |

### Какой способ выбрать

- **Метакласс** — **если** **много** **синглтонов**.
- **`__new__`** — **если** **один**-**два** **класса**.
- **Декоратор** — **если** **не** **нужно** **наследование**.
- **Модуль** — **для** **простых** **случаев**.

### Вывод

**Singleton** **реализуется** **четырьмя** **способами**. **Метакласс** — **самый** **мощный**. **`__new__`** — **самый** **простой**. **Декоратор** — **компромисс**. **Модуль** — **для** **простых** **задач**.

---

## 24.4. Singleton через `__new__` — подробно

Разберём **самый** **простой** **способ** **подробно**.

### Реализация

```python
class Agency:
    _instance = None

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

    def __init__(self, name):
        if not hasattr(self, "name"):
            self.name = name
            self.apartments = []
```

### Разбор по частям

- **`_instance = None`** — **атрибут** **класса**. **Хранит** **единственный** **объект**.
- **`__new__(cls, ...)`** — **вызывается** **до** `__init__`. **Создаёт** **объект**.
- **`if cls._instance is None`** — **проверка**: **объект** **уже** **создан**?
- **`super().__new__(cls)`** — **создаёт** **новый** **объект** **через** **родителя**.
- **`return cls._instance`** — **возвращает** **существующий** **объект**.
- **`if not hasattr(self, "name")`** — **защита** **`__init__`** от **повторного** **вызова**.

### Почему нужна защита `__init__`

`__init__` **вызывается** **каждый** **раз** **при** `Agency(...)`, **даже** **если** `__new__` **вернул** **существующий** **объект**. **Без** **защиты**:

```python
a1 = Agency("Дом-Сервис")
a2 = Agency("Другое имя")

print(a1.name)    # Другое имя — ПЕРЕЗАПИСАНО!
```

**С** **защитой**:

```python
a1 = Agency("Дом-Сервис")
a2 = Agency("Другое имя")

print(a1.name)    # Дом-Сервис — сохранено!
```

### Проверка

```python
a1 = Agency("Дом-Сервис")
a1.apartments.append("ул. Ленина")

a2 = Agency("Другое имя")
print(a2.apartments)    # ['ул. Ленина'] — то же!

print(a1 is a2)         # True
print(id(a1) == id(a2)) # True
```

### Схема

```
        Agency("Дом-Сервис")
                ↓
        __new__(Agency, "Дом-Сервис")
                ↓
        _instance is None?
            ↓ (да)
        super().__new__(Agency)
                ↓
        Сохраняем в _instance
                ↓
        __init__("Дом-Сервис")
                ↓
        hasattr(self, "name")?
            ↓ (нет)
        self.name = "Дом-Сервис"
                ↓
        Возвращаем объект

        Agency("Другое имя")
                ↓
        __new__(Agency, "Другое имя")
                ↓
        _instance is None?
            ↓ (нет)
        Возвращаем существующий
                ↓
        __init__("Другое имя")
                ↓
        hasattr(self, "name")?
            ↓ (да)
        Ничего не делаем
```

### Аналогия

Представь **кассу** **в** **магазине**. **Первая** **покупка** — **касса** **открывается**, **выдаёт** **чек**. **Вторая** **покупка** — **та** **же** **касса**, **тот** **же** **чек**. **Касса** **одна**. `__new__` — **открытие** **кассы**. `__init__` — **настройка**. **Защита** — **чтобы** **не** **настраивать** **дважды**.

### Вывод

**`__new__`** — **простой** **способ** **Singleton**. **Нужна** **защита** **`__init__`**. **Работает** **с** **наследованием**. **Подходит** **для** **1–2** **классов**.

---

## 24.5. Singleton через метакласс — подробно

Разберём **самый** **мощный** **способ**.

### Реализация

```python
class SingletonMeta(type):
    _instances = {}

    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]


class Agency(metaclass=SingletonMeta):
    def __init__(self, name):
        self.name = name
        self.apartments = []


class Registry(metaclass=SingletonMeta):
    def __init__(self):
        self.items = []
```

### Разбор по частям

- **`class SingletonMeta(type)`** — **наследник** `type` (стандартного метакласса).
- **`_instances = {}`** — **словарь** для **хранения** **объектов**. **Ключ** — **класс**, **значение** — **объект**.
- **`__call__(cls, ...)`** — **вызывается** **при** `Agency(...)`. **Перехватывает** **создание**.
- **`if cls not in cls._instances`** — **проверяет**, **создан** ли **объект** **для** **этого** **класса**.
- **`super().__call__(...)`** — **создаёт** **объект** через **обычный** **механизм**.
- **`return cls._instances[cls]`** — **возвращает** **существующий** **объект**.

### Почему `_instances` — словарь

**Один** **метакласс** — **много** **классов**. **Каждый** **класс** **имеет** **свой** **синглтон**. **Ключ** — **класс**, **значение** — **объект**.

```python
a1 = Agency("Дом-Сервис")
r1 = Registry()

print(a1 is Agency())    # True — Agency один
print(r1 is Registry())  # True — Registry один
print(a1 is r1)          # False — РАЗНЫЕ синглтоны
```

### Наследование

**Наследник** **автоматически** **становится** **синглтоном**:

```python
class MegaAgency(Agency):
    pass


m1 = MegaAgency("МЕГА")
m2 = MegaAgency("Другое")

print(m1 is m2)              # True
print(m1 is Agency("..."))   # False — разные ключи
```

**Ключ** — **фактический** **класс**. Поэтому `MegaAgency` — **свой** **синглтон**, `Agency` — **свой**.

### Почему `__init__` вызывается один раз

`__call__` **перехватывает** **создание**. Если **объект** **уже** **есть** — **возвращается** **существующий**, **без** **вызова** `__init__`. Поэтому **защита** `__init__` **не** **нужна**.

### Схема

```
        Agency("Дом-Сервис")
                ↓
        SingletonMeta.__call__(Agency, "Дом-Сервис")
                ↓
        Agency в _instances?
            ↓ (нет)
        super().__call__(Agency, "Дом-Сервис")
            ↓
        __new__ + __init__
            ↓
        Сохраняем в _instances[Agency]
            ↓
        Возвращаем объект

        Agency("Другое")
                ↓
        SingletonMeta.__call__(Agency, "Другое")
                ↓
        Agency в _instances?
            ↓ (да)
        Возвращаем существующий
        (__init__ НЕ вызывается)
```

### Аналогия

Представь **мэрию** **города**. **Мэр** — **один**. **Если** **ты** **приходишь** **с** **запросом** — **тебя** **ведут** к **существующему** **мэру**. **Нового** **не** **избирают**. **Метакласс** — **это** **система** **выборов**, **которая** **гарантирует** **единственность**.

### Вывод

**Метакласс** — **самый** **мощный** **способ**. **Работает** **автоматически** для **всех** **классов**. **Наследники** — **тоже** **синглтоны**. **`__init__`** **вызывается** **один** **раз**. **Подходит** **для** **библиотек** и **фреймворков**.

---

## 24.6. Singleton vs `__new__` vs метакласс

Разберём **разницу** **между** **тремя** **подходами** **подробно**.

### 24.6.1. Через `__new__`

```python
class Agency:
    _instance = None

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

    def __init__(self, name):
        if not hasattr(self, "name"):
            self.name = name
```

**Плюсы:**

- **Просто** **для** **понимания**.
- **Работает** **с** **наследованием**.
- **Не** **нужен** **метакласс**.

**Минусы:**

- **Нужно** **переопределять** `__new__` **в** **каждом** **классе**.
- **Нужна** **защита** `__init__`.
- **Дублирование** **кода**.

### 24.6.2. Через метакласс

```python
class SingletonMeta(type):
    _instances = {}

    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]


class Agency(metaclass=SingletonMeta):
    def __init__(self, name):
        self.name = name
```

**Плюсы:**

- **Работает** **автоматически**.
- **Наследники** — **синглтоны**.
- **`__init__`** **вызывается** **один** **раз**.
- **Один** **класс** **метакласса** — **для** **всех** **синглтонов**.

**Минусы:**

- **Сложно** **для** **понимания**.
- **Может** **конфликтовать** **с** **другими** **метаклассами**.

### 24.6.3. Через декоратор

```python
def singleton(cls):
    instances = {}

    def get_instance(*args, **kwargs):
        if cls not in instances:
            instances[cls] = cls(*args, **kwargs)
        return instances[cls]

    return get_instance


@singleton
class Agency:
    def __init__(self, name):
        self.name = name
```

**Плюсы:**

- **Просто**.
- **Работает** **для** **любого** **класса**.

**Минусы:**

- **Возвращает** **функцию** — **`isinstance`** **не** **работает**.
- **Не** **работает** **с** **наследованием**.

### Сравнительная таблица

| Критерий | `__new__` | Метакласс | Декоратор |
|----------|-----------|-----------|-----------|
| **Простота** | ✅ Высокая | ⚠️ Средняя | ✅ Высокая |
| **`__init__` 1 раз** | ❌ Защита | ✅ Да | ✅ Да |
| **Наследование** | ✅ Да | ✅ Да | ❌ Нет |
| **`isinstance`** | ✅ Да | ✅ Да | ⚠️ Частично |
| **Много классов** | ❌ Дублирование | ✅ Один метакласс | ✅ Один декоратор |
| **Конфликты** | ✅ Нет | ⚠️ Возможны | ✅ Нет |

### Как выбрать

- **Один** **класс** — `__new__`.
- **Много** **классов** — **метакласс**.
- **Нет** **наследования** — **декоратор**.

### Аналогия

- **`__new__`** — **ручная** **сборка** **каждого** **изделия**.
- **Метакласс** — **автоматическая** **линия**.
- **Декоратор** — **упаковка** **готового** **изделия**.

### Вывод

**Каждый** **способ** **имеет** **свои** **плюсы** и **минусы**. **Метакласс** — **самый** **мощный**. **`__new__`** — **самый** **понятный**. **Декоратор** — **самый** **быстрый**.

---

## 24.7. Когда использовать Singleton

Разберём **пять** **типичных** **случаев** **использования**.

### 24.7.1. Реестр объектов

**Самый** **частый** **случай** — **реестр** всех **объектов**:

```python
class ApartmentRegistry(metaclass=SingletonMeta):
    def __init__(self):
        self.apartments = []

    def add(self, apartment):
        self.apartments.append(apartment)

    def find(self, address):
        for flat in self.apartments:
            if flat.address == address:
                return flat
        return None
```

**Почему** **Singleton:** **все** **части** **программы** **должны** **видеть** **один** **и** **тот** **же** **реестр**.

### 24.7.2. Конфигурация

**Настройки** **приложения** — **один** **набор**:

```python
class Config(metaclass=SingletonMeta):
    def __init__(self):
        self.debug = False
        self.db_url = "sqlite:///app.db"
        self.max_connections = 10
```

**Почему** **Singleton:** **все** **модули** **читают** **одни** **настройки**.

### 24.7.3. Логгер

**Журнал** **событий** — **один** **файл**:

```python
class Logger(metaclass=SingletonMeta):
    def __init__(self):
        self.log_file = "app.log"

    def log(self, message):
        with open(self.log_file, "a") as f:
            f.write(f"{message}\n")
```

**Почему** **Singleton:** **все** **сообщения** **должны** **идти** в **один** **журнал**.

### 24.7.4. Кэш

**Кэш** **данных** — **один** **на** **всю** **программу**:

```python
class Cache(metaclass=SingletonMeta):
    def __init__(self):
        self.data = {}

    def get(self, key):
        return self.data.get(key)

    def set(self, key, value):
        self.data[key] = value
```

**Почему** **Singleton:** **иначе** **кэш** **будет** **разным** в **разных** **частях**.

### 24.7.5. Пул соединений

**Пул** **подключений** к **БД** — **один**:

```python
class ConnectionPool(metaclass=SingletonMeta):
    def __init__(self, size=5):
        self.connections = [self._create() for _ in range(size)]

    def _create(self):
        return {"connected": True}
```

**Почему** **Singleton:** **все** **части** **программы** **используют** **один** **пул**.

### Что общего

- **Данные** **должны** **быть** **согласованы**.
- **Глобальный** **доступ** **нужен**.
- **Создание** **дорогое** — **лучше** **один** **раз**.
- **Дублирование** **вредно**.

### Вывод

**Singleton** **используется** **везде**, **где** **нужна** **согласованность**: **реестры**, **конфигурация**, **логгеры**, **кэши**, **пулы**.

---

## 24.8. Когда НЕ использовать Singleton

**Singleton** — **не** **всегда** **хорошо**. Разберём **когда** **его** **стоит** **избегать**.

### 24.8.1. Сложность тестирования

**Singleton** **мешает** **тестам**. **Нельзя** **создать** **второй** **объект** для **изоляции**:

```python
def test_registry():
    # Нельзя создать чистый реестр — он один на всю программу.
    registry = ApartmentRegistry()
    registry.add(...)
    # Следующий тест увидит данные предыдущего!
```

**Решение:** **dependency injection** — **передавать** **зависимости** **явно**.

### 24.8.2. Скрытые зависимости

**Singleton** **скрывает** **зависимости**. **Класс** **использует** **Singleton**, **но** **это** **не** **видно** в **сигнатуре**:

```python
class ApartmentService:
    def sell(self, address):
        # Скрытая зависимость от Singleton!
        registry = ApartmentRegistry()
        flat = registry.find(address)
        ...
```

**Проблема:** **непонятно**, **какие** **зависимости** **у** **класса**.

### 24.8.3. Многопоточность

**Singleton** **требует** **особой** **осторожности** в **многопоточных** **программах**. **Два** **потока** **могут** **создать** **два** **объекта** **одновременно**.

**Решение:** **блокировки** (`threading.Lock`).

### 24.8.4. Глобальное состояние

**Singleton** — **это** **глобальное** **состояние**. **Глобальное** **состояние** **сложно** **отслеживать**, **отлаживать**, **тестировать**.

### 24.8.5. Альтернативы

- **Dependency Injection** — **передача** **зависимостей** **явно**.
- **Модуль** — **простой** **Singleton** **из** **коробки**.
- **Фабрика** — **создание** **объектов** **по** **запросу**.

### Сравнение

| Подход | Плюсы | Минусы |
|--------|-------|--------|
| Singleton | Глобальный доступ | Глобальное состояние |
| DI | Явные зависимости | Больше кода |
| Модуль | Просто | Не класс |
| Фабрика | Гибко | Сложнее |

### Когда НЕ использовать

- **Много** **тестов** — **Singleton** **мешает**.
- **Многопоточность** — **нужны** **блокировки**.
- **Простая** **задача** — **модуль** **проще**.
- **Явные** **зависимости** — **DI** **лучше**.

### Аналогия

**Singleton** — **как** **общий** **холодильник** в **офисе**. **Удобно** — **все** **знают**, **где** **еда**. Но **если** **кто-то** **съел** **твоё** — **не** **докажешь**. **И** **для** **тестов** **не** **подходит** — **нельзя** **поставить** **свой**.

### Вывод

**Singleton** — **мощный**, **но** **опасный**. **Используйте** **осторожно**. **Альтернативы:** **DI**, **модули**, **фабрики**.

---

## 24.10. Нюансы и тонкости

Разберём **десять** **нюансов**.

### 24.10.1. Потокобезопасность

**Без** **блокировки** **два** **потока** **могут** **создать** **два** **объекта**:

```python
import threading

class SingletonMeta(type):
    _instances = {}
    _lock = threading.Lock()

    def __call__(cls, *args, **kwargs):
        with cls._lock:
            if cls not in cls._instances:
                cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]
```

### 24.10.2. Ленивая инициализация

**Singleton** **может** **создаваться** **лениво** — **при** **первом** **обращении**:

```python
class SingletonMeta(type):
    _instances = {}

    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]
```

**Объект** **не** **создаётся** **до** **первого** `Agency()`.

### 24.10.3. Сброс Singleton

**Для** **тестов** **нужно** **уметь** **сбрасывать** **Singleton**:

```python
@classmethod
def reset(cls):
    cls._instance = None
```

### 24.10.4. Наследование и `_instance`

**Важно:** `_instance` **должен** **быть** **на** **уровне** **класса**, **не** **экземпляра**. **Иначе** **не** **работает**.

### 24.10.5. Множественное наследование

**Singleton** **может** **конфликтовать** **с** **множественным** **наследованием**. **Будьте** **осторожны**.

### 24.10.6. Синглтон и сериализация

**При** `pickle` **Singleton** **может** **создать** **второй** **объект**. **Нужно** **реализовать** `__reduce__`.

### 24.10.7. Синглтон и `__slots__`

**Совместимы**, **но** **нужно** **добавить** `_instance` **в** `__slots__` **или** **учитывать** это.

### 24.10.8. Синглтон и `@dataclass`

`@dataclass` **не** **поддерживает** **Singleton** **из** **коробки**. **Нужен** **метакласс** или **ручная** **реализация**.

### 24.10.9. Синглтон и `__init_subclass__`

**Можно** **использовать** `__init_subclass__` **вместо** **метакласса** для **простых** **случаев**.

### 24.10.10. Синглтон в многопоточности

**Всегда** **используйте** **блокировки** в **многопоточных** **программах**.

### Вывод

**Singleton** **имеет** **много** **нюансов**. **Основные:** **потокобезопасность**, **ленивость**, **сброс**, **сериализация**.

## 24.11. Типичные ошибки

Разберём **десять** **самых** **частых** **ошибок**.

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Забыли защиту `__init__` | `__init__` вызывается дважды | `if not hasattr(...)` |
| `_instance` в `__init__` | Атрибут экземпляра, не класса | Сделать атрибутом класса |
| Нет блокировки | Проблемы в многопоточности | `threading.Lock` |
| Нельзя сбросить | Мешает тестам | Добавить `reset()` |
| Синглтон для всего | Избыточно | Использовать только при необходимости |
| Забыли про pickle | Второй объект при pickle | `__reduce__` |
| `_instance` — общий для всех классов | Путаница | Словарь с ключом-классом |
| Синглтон + `@dataclass` | Не работает | Метакласс |
| Скрытые зависимости | Сложно тестировать | Dependency Injection |
| Наследование без учёта | Два синглтона | Ключ — фактический класс |

### Подробный разбор

**Ошибка 1.** Забыли **защиту** `__init__` — **`__init__`** **вызывается** **дважды**. **Правило:** **добавьте** **проверку** `hasattr`.

**Ошибка 2.** `_instance` **в** `__init__` — **атрибут** **экземпляра**, **не** **класса**. **Правило:** **сделайте** **атрибутом** **класса**.

**Ошибка 3.** Нет **блокировки** — **проблемы** **в** **многопоточности**. **Правило:** `threading.Lock`.

**Ошибка 4.** Нельзя **сбросить** — **мешает** **тестам**. **Правило:** **добавьте** `reset()`.

**Ошибка 5.** Синглтон **для** **всего** — **избыточно**. **Правило:** **используйте** **только** **где** **нужно**.

**Ошибка 6.** Забыли **про** `pickle` — **второй** **объект** **при** **сериализации**. **Правило:** `__reduce__`.

**Ошибка 7.** `_instance` — **общий** **для** **всех** **классов** — **путаница**. **Правило:** **словарь** **с** **ключом**-**классом**.

**Ошибка 8.** Синглтон + `@dataclass` — **не** **работает**. **Правило:** **метакласс**.

**Ошибка 9.** Скрытые **зависимости** — **сложно** **тестировать**. **Правило:** **DI**.

**Ошибка 10.** Наследование **без** **учёта** — **два** **синглтона**. **Правило:** **ключ** — **фактический** **класс**.

### Вывод

Ошибки **с** **Singleton** **типичны**. **Основные:** **защита** `__init__`, **блокировки**, **сброс**.

---

## 24.12. Что важно запомнить

- **Singleton** — **порождающий** **паттерн**.
- **Гарантирует** **один** **экземпляр** **класса**.
- **Даёт** **глобальный** **доступ**.
- **Реализуется** **четырьмя** **способами:** `__new__`, **метакласс**, **декоратор**, **модуль**.
- **Метакласс** — **самый** **мощный**.
- **`__new__`** — **самый** **простой**.
- **Декоратор** — **компромисс**.
- **Модуль** — **для** **простых** **задач**.
- **Нужна** **защита** `__init__` **для** `__new__`.
- **Нужна** **блокировка** **для** **многопоточности**.
- **Используйте** **для:** **реестров**, **конфигурации**, **логгеров**, **кэшей**.
- **Избегайте** **для:** **тестируемости**, **явных** **зависимостей**.
- **Альтернативы:** **DI**, **модули**, **фабрики**.

## Задача 24 (паттерн Singleton)

В агентстве **должен** **быть** **один** **реестр** **объектов**. **Если** **создать** **второй** — **данные** **разойдутся**. **Реализуйте** **Singleton** **двумя** **способами**.

**Часть 1. Singleton через `__new__`.** Создайте класс `AgencyRegistry`:

- **атрибут** **класса** `_instance = None`;
- **метод** `__new__` — **возвращает** **единственный** **объект**;
- **защита** `__init__` — **чтобы** **не** **перезаписывать** **данные**;
- **поля:** `agencies` — **список** **агентств**;
- **методы:** `register(agency)`, `list_all()`.

**Часть 2. Проверка.** Продемонстрируйте:

- **создание** **двух** `AgencyRegistry` — **один** **объект**;
- `r1 is r2` → `True`;
- `r1.agencies` и `r2.agencies` — **одни** **данные**;
- `id(r1) == id(r2)` → `True`.

**Часть 3. Singleton через метакласс.** Создайте **метакласс** `SingletonMeta` и **второй** **класс** `Config(metaclass=SingletonMeta)` с **настройками** `debug`, `db_url`, `max_connections`.

**Часть 4. Проверка.** Продемонстрируйте, что:

- `Config()` **всегда** **возвращает** **один** **объект**;
- **изменение** `c1.debug = True` **видно** в `c2.debug`;
- `AgencyRegistry` и `Config` — **разные** **синглтоны**.

**Часть 5. Singleton через декоратор.** Реализуйте **третий** **способ** — **декоратор** `singleton`. Создайте **класс** `Logger`, **обёрнутый** **декоратором**.

**Часть 6. Проверка.** Продемонстрируйте, что `Logger()` **возвращает** **один** **объект**.

**Часть 7. Сброс для тестов.** Добавьте **метод** `reset()` **в** `AgencyRegistry` и `Config` — **для** **сброса** **Singleton** в **тестах**.

**Часть 8. Потокобезопасность (дополнительно).** Добавьте **блокировку** в **метакласс** `SingletonMeta`:

```python
import threading

class SingletonMeta(type):
    _instances = {}
    _lock = threading.Lock()

    def __call__(cls, *args, **kwargs):
        with cls._lock:
            if cls not in cls._instances:
                cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]
```

**Часть 9.** В комментариях объясните:

1. Что **такое** **Singleton** и **зачем** он **нужен**.
2. Чем **Singleton** **отличается** от **глобальной** **переменной**.
3. Как **работает** **Singleton** **через** `__new__`.
4. Как **работает** **Singleton** **через** **метакласс**.
5. **Почему** **метакласс** **предпочтительнее** при **многих** **синглтонах**.
6. **Когда** **НЕ** **стоит** **использовать** **Singleton**.

### Решение

```python
import threading


# ─── Часть 1. Singleton через __new__ ───
class AgencyRegistry:
    _instance = None

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

    def __init__(self):
        # Защита: инициализируем только один раз.
        if not hasattr(self, "agencies"):
            self.agencies = []

    def register(self, agency):
        self.agencies.append(agency)

    def list_all(self):
        return list(self.agencies)

    @classmethod
    def reset(cls):
        cls._instance = None


# ─── Часть 3. Singleton через метакласс ───
class SingletonMeta(type):
    _instances = {}
    _lock = threading.Lock()    # Часть 8

    def __call__(cls, *args, **kwargs):
        with cls._lock:
            if cls not in cls._instances:
                cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]


class Config(metaclass=SingletonMeta):
    def __init__(self):
        self.debug = False
        self.db_url = "sqlite:///app.db"
        self.max_connections = 10

    @classmethod
    def reset(cls):
        SingletonMeta._instances.pop(cls, None)


# ─── Часть 5. Singleton через декоратор ───
def singleton(cls):
    instances = {}

    def get_instance(*args, **kwargs):
        if cls not in instances:
            instances[cls] = cls(*args, **kwargs)
        return instances[cls]

    return get_instance


@singleton
class Logger:
    def __init__(self):
        self.log_file = "app.log"
        self.messages = []

    def log(self, message):
        self.messages.append(message)


# ─── Часть 2. Проверка AgencyRegistry ───
print("=== Часть 2. Singleton через __new__ ===")

r1 = AgencyRegistry()
r2 = AgencyRegistry()

r1.register("Дом-Сервис")
r2.register("Новое агентство")

print(f"r1 is r2: {r1 is r2}")
print(f"id равны: {id(r1) == id(r2)}")
print(f"r1.agencies: {r1.agencies}")
print(f"r2.agencies: {r2.agencies}")
print(f"Это один и тот же список: {r1.agencies is r2.agencies}")

# ─── Часть 4. Проверка Config ───
print()
print("=== Часть 4. Singleton через метакласс ===")

c1 = Config()
c2 = Config()

print(f"c1 is c2: {c1 is c2}")

c1.debug = True
print(f"c1.debug = {c1.debug}")
print(f"c2.debug = {c2.debug}")    # True — то же самое

print(f"AgencyRegistry и Config — разные: "
      f"{AgencyRegistry() is not Config()}")

# ─── Часть 6. Проверка Logger ───
print()
print("=== Часть 6. Singleton через декоратор ===")

log1 = Logger()
log2 = Logger()

print(f"log1 is log2: {log1 is log2}")

log1.log("Сообщение 1")
log2.log("Сообщение 2")
print(f"log1.messages: {log1.messages}")

# ─── Часть 7. Сброс ───
print()
print("=== Часть 7. Сброс для тестов ===")

AgencyRegistry.reset()
r3 = AgencyRegistry()
print(f"После reset: r3 is r1? {r3 is r1}")    # False
print(f"r3.agencies: {r3.agencies}")            # []

Config.reset()
c3 = Config()
print(f"После reset: c3 is c1? {c3 is c1}")    # False
print(f"c3.debug: {c3.debug}")                  # False
```

**Ожидаемый вывод:**

```
=== Часть 2. Singleton через __new__ ===
r1 is r2: True
id равны: True
r1.agencies: ['Дом-Сервис', 'Новое агентство']
r2.agencies: ['Дом-Сервис', 'Новое агентство']
Это один и тот же список: True

=== Часть 4. Singleton через метакласс ===
c1 is c2: True
c1.debug = True
c2.debug = True
AgencyRegistry и Config — разные: True

=== Часть 6. Singleton через декоратор ===
log1 is log2: True
log1.messages: ['Сообщение 1', 'Сообщение 2']

=== Часть 7. Сброс для тестов ===
После reset: r3 is r1? False
r3.agencies: []
После reset: c3 is c1? False
c3.debug: False
```

### Разбор

**1. Что такое Singleton и зачем он нужен.**

**Singleton** — **порождающий** **паттерн**. **Гарантирует**, что **у** **класса** **один** **экземпляр**. **Даёт** **глобальный** **доступ**.

**Зачем:**

- **Согласованность** **данных** — **все** **видят** **одно**.
- **Экономия** **ресурсов** — **дорогой** **объект** **создаётся** **один** **раз**.
- **Глобальная** **точка** **доступа** — **без** **глобальных** **переменных**.

**2. Чем Singleton отличается от глобальной переменной.**

**Глобальная** **переменная:**

- Живёт в **модуле**.
- **Нет** **инкапсуляции**.
- **Нет** **ленивой** **инициализации**.
- **Можно** **перезаписать**.

**Singleton:**

- **Класс** — **инкапсулирует** **логику**.
- **Ленивая** **инициализация** — **создаётся** **при** **первом** **обращении**.
- **Нельзя** **создать** **второй** **объект**.
- **Можно** **наследовать**.

**3. Как работает Singleton через `__new__`.**

- `_instance` — **атрибут** **класса**.
- `__new__` **проверяет**: **создан** ли **объект**.
- **Если** **нет** — **создаёт** и **сохраняет**.
- **Возвращает** **существующий**.
- `__init__` **защищён** **проверкой** `hasattr`.

**4. Как работает Singleton через метакласс.**

- `SingletonMeta` — **наследник** `type`.
- `_instances` — **словарь** **всех** **объектов**.
- `__call__` **перехватывает** **создание**.
- **Если** **класс** **уже** **в** `_instances` — **возвращает** **объект**.
- `__init__` **вызывается** **один** **раз**.

**5. Почему метакласс предпочтительнее при многих синглтонах.**

**С** `__new__`:

- **Каждый** **класс** — **свой** `__new__`.
- **Каждый** **класс** — **своя** **защита** `__init__`.
- **Дублирование** **кода**.

**С** **метаклассом:**

- **Один** **метакласс** — **для** **всех** **классов**.
- **Наследники** — **автоматически** **синглтоны**.
- **`__init__`** **вызывается** **один** **раз** **без** **защиты**.

**6. Когда НЕ стоит использовать Singleton.**

- **Много** **тестов** — **Singleton** **мешает** **изоляции**.
- **Многопоточность** — **нужны** **блокировки**.
- **Явные** **зависимости** — **DI** **лучше**.
- **Простая** **задача** — **модуль** **проще**.
- **Глобальное** **состояние** — **сложно** **отлаживать**.

**Правило:** **используйте** **Singleton** **только** **когда** **единственность** **действительно** **нужна**.

# Глава 25. Паттерн Factory

## 25.1. Проблема: как создавать разные типы объектов

За предыдущие главы мы построили агентство недвижимости и научили его **многому**. У нас есть `Apartment`, `House`, `Studio`, `Penthouse`. Каждый тип объекта — **свой** класс с **своими** полями. Всё **работает** — **пока** объектов **мало** и **типы** **известны** **заранее**.

Но теперь представь **реальную** **ситуацию**. Агентство **регулярно** **пополняется** **новыми** **объектами** — **студиями**, **двухкомнатными**, **домами**, **пентхаусами**. **Данные** **приходят** **из** **внешней** **системы** — **например**, **из** **CSV-файла** или **API**. Формат **простой**: **строка** **типа** (`"studio"`, `"two_room"`, `"house"`) и **набор** **параметров** **через** `**kwargs`.

**Первое**, что **приходит** в **голову** — **написать** **код** **с** `if/elif`:

```python
def create_estate(kind, **kwargs):
    if kind == "studio":
        return Studio(**kwargs)
    elif kind == "two_room":
        return TwoRoom(**kwargs)
    elif kind == "house":
        return House(**kwargs)
    else:
        raise ValueError(f"Неизвестный тип: {kind}")
```

**Работает**. Но **есть** **проблема**. Представь, что **через** **полгода** **добавили** **пентхаус**. **Придётся**:

1. **Найти** **все** **места** в **коде**, **где** **есть** **такой** `if/elif`.
2. **Добавить** **новую** **ветку** **в** **каждое** **из** **них**.
3. **Не** **забыть** **ни** **одного** — **иначе** **баг**.

А **если** **таких** **мест** **десять**? **Двадцать**? **Каждый** **раз** **править** **все** — **утомительно** и **опасно**.

**Второе**, что **приходит** в **голову** — **вынести** **создание** в **одно** **место** — **фабрику**. **Тогда** **добавление** **нового** **типа** **затронет** **только** **один** **класс**. **Код** **в** **остальных** **местах** **не** **изменится**.

### Аналогия

Представь **ресторан**. Ты **приходишь** и **говоришь** **официанту**: «**борщ**». **Официант** **идёт** на **кухню** и **передаёт** **заказ** **повару**. **Тебе** **не** **нужно** **знать**, **как** **повар** **готовит** **борщ** — **какие** **ингредиенты**, **какая** **плита**, **сколько** **минут**. Ты **просто** **делаешь** **заказ** — **и** **получаешь** **блюдо**.

**Официант** — **фабрика**. Он **принимает** **заказ** (строку «борщ») и **возвращает** **готовое** **блюдо** (объект). **Тебе** **не** **нужно** **знать** **деталей** **готовки** (создания).

### Что такое Factory

**Factory** (Фабрика) — это **порождающий** **паттерн** **проектирования**, **который**:

1. **Скрывает** **логику** **создания** **объектов**.
2. **Предоставляет** **единый** **интерфейс** для **создания** **разных** **типов**.
3. **Позволяет** **добавлять** **новые** **типы** **без** **изменения** **существующего** **кода**.

Простыми **словами**: **Factory** — **это** **класс** или **функция**, **которая** **создаёт** **объекты** **по** **запросу**.

### Вывод

Создание **объектов** **через** `if/elif` **работает**, **но** **плохо** **масштабируется**. **Фабрика** **выносит** **логику** **в** **одно** **место**, **скрывает** **детали** и **упрощает** **добавление** **новых** **типов**.

---

## 25.2. Что такое паттерн Factory

**Factory** — **один** из **самых** **распространённых** **паттернов** **проектирования**. Он **относится** к **порождающим** — **то** **есть** **управляет** **созданием** **объектов**.

### Ключевая идея

**Factory** **работает** **так**:

1. **Клиент** **говорит** фабрике, **какой** **объект** **нужен** (по **имени**, **ключу**, **типу**).
2. **Фабрика** **решает**, **как** его **создать**.
3. **Фабрика** **возвращает** **готовый** **объект**.
4. **Клиент** **не** **знает** **деталей** **создания**.

### Схема

```
        Клиент                    Фабрика                    Объект
        ──────                    ───────                    ──────
        create("studio", ...) →   Разбор параметров   →      Studio(...)
                                  ↓
                                  Выбор класса
                                  ↓
                                  Создание объекта
                                  ↓
                                  ← возврат объекта
```

### Три варианта Factory

В **ООП** **есть** **три** **разновидности** **фабрик**:

| Вариант | Описание | Пример |
|---------|----------|--------|
| **Simple Factory** | **Функция** или **метод**, **создающий** **объекты** | `create_estate(kind, ...)` |
| **Factory Method** | **Метод** **в** **классе**, **переопределяемый** **наследниками** | `EstateFactory.create(kind)` |
| **Abstract Factory** | **Фабрика** для **создания** **семейств** **объектов** | `FurnitureFactory.create_chair()` |

В **этой** **главе** мы **разберём** **Simple Factory** и **Factory Method**. **Abstract Factory** — **продвинутая** **тема**, **реже** **используется** в **Python**.

### Что даёт Factory

- **Инкапсуляция** **логики** **создания**.
- **Единая** **точка** **изменения** — **при** **добавлении** **нового** **типа**.
- **Гибкость** — **можно** **менять** **тип** **создаваемого** **объекта** **без** **изменения** **клиента**.
- **Читаемость** — **клиент** **не** **думает** **о** **деталях**.

### Что НЕ даёт Factory

- **Не** **заменяет** **конструкторы** — **только** **добавляет** **уровень** **абстракции**.
- **Не** **убирает** **сложность** — **переносит** её в **одно** **место**.
- **Не** **всегда** **нужна** — **для** **одного**-**двух** **классов** **избыточна**.

### Аналогия

Представь **автоматическую** **линию** на **заводе**. **Рабочий** **нажимает** **кнопку** — **линия** **сама** **выбирает** **нужный** **инструмент** и **производит** **деталь**. **Рабочему** **не** **нужно** **знать**, **как** **устроена** **линия** **внутри**. **Factory** — **такая** **линия** **для** **объектов**.

### Вывод

**Factory** — **порождающий** **паттерн**. **Скрывает** **логику** **создания**. **Даёт** **единый** **интерфейс**. **Упрощает** **добавление** **новых** **типов**.

---

## 25.3. Simple Factory — простая фабрика

**Simple Factory** — **самый** **простой** **вариант**. Это **функция** или **метод**, **который** **создаёт** **объект** **по** **запросу**.

### Синтаксис

```python
def create_estate(kind, **kwargs):
    if kind == "studio":
        return Studio(**kwargs)
    elif kind == "two_room":
        return TwoRoom(**kwargs)
    elif kind == "house":
        return House(**kwargs)
    else:
        raise ValueError(f"Неизвестный тип: {kind}")
```

**Что** **здесь** **происходит:**

- **`kind`** — **строка** с **типом** **объекта**.
- **`**kwargs`** — **параметры** **для** **создания**.
- **`if/elif`** — **выбор** **класса**.
- **`Studio(**kwargs)`** — **создание** **объекта**.
- **`raise ValueError`** — **ошибка** **при** **неизвестном** **типе**.

### Использование

```python
studio = create_estate("studio", address="ул. Пушкина", area=25, price=4_000_000)
house = create_estate("house", address="ул. Садовая", area=200, price=20_000_000)
```

**Клиент** **не** **знает** **про** **классы** `Studio` и `House`. Он **просто** **говорит** **тип** — и **получает** **объект**.

### Схема

```
        Клиент                    create_estate()              Результат
        ──────                    ───────────────              ─────────
        create_estate(       →    if kind == "studio":   →     Studio(...)
            "studio", ...)         return Studio(...)
```

### Реализация через класс

**Simple Factory** **можно** **оформить** **как** **класс**:

```python
class EstateFactory:
    @staticmethod
    def create(kind, **kwargs):
        if kind == "studio":
            return Studio(**kwargs)
        elif kind == "two_room":
            return TwoRoom(**kwargs)
        elif kind == "house":
            return House(**kwargs)
        else:
            raise ValueError(f"Неизвестный тип: {kind}")
```

**Использование:**

```python
studio = EstateFactory.create("studio", address="ул. Пушкина", area=25, price=4_000_000)
```

**Плюс** **класса:** **логически** **связанные** **методы** **живут** **вместе**. **Можно** **добавить** **другие** **фабричные** **методы**.

### Реализация через словарь

**Более** **элегантный** **вариант** — **словарь** **классов**:

```python
class EstateFactory:
    _registry = {
        "studio": Studio,
        "two_room": TwoRoom,
        "house": House,
    }

    @classmethod
    def create(cls, kind, **kwargs):
        if kind not in cls._registry:
            raise ValueError(f"Неизвестный тип: {kind}")
        return cls._registry[kind](**kwargs)
```

**Что** **это** **даёт:**

- **Нет** **длинной** **цепочки** `if/elif`.
- **Легко** **добавить** **новый** **тип** — **одна** **строка** в **словаре**.
- **Читаемо** — **сразу** **видно** **все** **типы**.

### Что выбрать

| Вариант | Когда использовать |
|---------|-------------------|
| **Функция** | **Простой** **случай** |
| **Класс** с `if/elif` | **Немного** **типов** |
| **Класс** со **словарём** | **Много** **типов** |

### Аналогия

Представь **меню** в **кафе**. **Официант** **смотрит** в **меню** — **есть** **ли** **такое** **блюдо**. **Если** **есть** — **идёт** **на** **кухню** и **заказывает**. **Если** **нет** — **сообщает** **клиенту**. **Словарь** — **это** **меню**.

### Вывод

**Simple Factory** — **функция** или **класс** для **создания** **объектов**. **Скрывает** `if/elif`. **Словарь** **классов** — **самый** **гибкий** **вариант**.

---

## 25.4. Factory Method — фабричный метод

**Factory Method** — **более** **продвинутый** **вариант**. Здесь **фабрика** — **метод** **класса**, **который** **можно** **переопределить** **в** **наследниках**.

### Ключевая идея

**Базовый** **класс** **определяет** **метод** `create()`. **Наследники** **переопределяют** его — **создавая** **разные** **объекты**.

### Синтаксис

```python
class EstateCreator:
    def create(self, **kwargs):
        raise NotImplementedError("Subclasses must implement create")


class ApartmentCreator(EstateCreator):
    def create(self, **kwargs):
        return Apartment(**kwargs)


class HouseCreator(EstateCreator):
    def create(self, **kwargs):
        return House(**kwargs)
```

**Что** **здесь** **происходит:**

- **`EstateCreator`** — **базовый** **класс** **с** **методом** `create`.
- **`ApartmentCreator`** — **создаёт** `Apartment`.
- **`HouseCreator`** — **создаёт** `House`.
- **Каждый** **класс** — **своя** **фабрика**.

### Использование

```python
creator = ApartmentCreator()
flat = creator.create(address="ул. Ленина", area=65, price=8_500_000)
print(type(flat))    # <class 'Apartment'>
```

### Схема

```
        EstateCreator (базовый)
        ├── ApartmentCreator.create() → Apartment
        ├── HouseCreator.create()     → House
        └── StudioCreator.create()    → Studio
```

### Зачем это нужно

**Factory Method** **полезен**, **когда**:

- **Есть** **несколько** **фабрик** **с** **разной** **логикой**.
- **Логика** **создания** **зависит** от **контекста**.
- **Нужна** **расширяемость** — **добавление** **новых** **фабрик** **без** **изменения** **базового** **класса**.

### Пример: разные способы создания

```python
class EstateCreator:
    def create(self, data):
        # Разные фабрики по-разному интерпретируют data.
        raise NotImplementedError


class FromStringCreator(EstateCreator):
    def create(self, data):
        address, area, price = data.split(";")
        return Apartment(address, int(area), int(price))


class FromDictCreator(EstateCreator):
    def create(self, data):
        return Apartment(data["address"], data["area"], data["price"])
```

**Одна** **задача** — **создать** **квартиру**. **Разные** **форматы** **данных**. **Разные** **фабрики**.

### Сравнение с Simple Factory

| Критерий | Simple Factory | Factory Method |
|----------|---------------|----------------|
| **Сложность** | ✅ Просто | ⚠️ Средне |
| **Расширяемость** | ⚠️ Изменение кода | ✅ Наследование |
| **Использование** | **Простой** **случай** | **Гибкая** **логика** |
| **Количество** **классов** | 1 | **Несколько** |

### Когда использовать

**Simple Factory:**

- **Один** **способ** **создания**.
- **Мало** **типов**.
- **Простой** **выбор** **по** **строке**.

**Factory Method:**

- **Разная** **логика** **создания**.
- **Много** **контекстов**.
- **Нужна** **расширяемость** **через** **наследование**.

### Аналогия

**Simple Factory** — **один** **повар**, **который** **готовит** **всё** **меню**. **Factory Method** — **несколько** **поваров**, **каждый** **специализируется** на **своих** **блюдах**.

### Вывод

**Factory Method** — **фабрика** **через** **наследование**. **Гибче** **Simple Factory**, **но** **сложнее**. **Используется**, **когда** **логика** **создания** **различается**.

---

## 25.5. Фабрика через словарь классов

**Словарь** **классов** — **самый** **элегантный** **способ** **реализации** **Simple Factory**. Разберём **подробно**.

### Реализация

```python
class EstateFactory:
    _registry = {
        "studio": Studio,
        "two_room": TwoRoom,
        "house": House,
    }

    @classmethod
    def create(cls, kind, **kwargs):
        if kind not in cls._registry:
            raise ValueError(f"Неизвестный тип: {kind!r}")
        return cls._registry[kind](**kwargs)
```

### Разбор по частям

- **`_registry`** — **словарь** **типов**. **Ключ** — **строка**, **значение** — **класс**.
- **`@classmethod`** — **классовый** **метод**. **Работает** с **классом**.
- **`kind not in cls._registry`** — **проверка** **типа**.
- **`cls._registry[kind]`** — **получение** **класса**.
- **`(**kwargs)`** — **создание** **объекта** **с** **параметрами**.

### Использование

```python
studio = EstateFactory.create("studio", address="ул. Пушкина", area=25, price=4_000_000)
two_room = EstateFactory.create("two_room", address="ул. Ленина", area=65, price=8_500_000)
house = EstateFactory.create("house", address="ул. Садовая", area=200, price=20_000_000)
```

### Добавление нового типа

```python
class Penthouse(Apartment):
    pass


# Регистрация нового типа.
EstateFactory._registry["penthouse"] = Penthouse
```

**Одна** **строка** — **и** **новый** **тип** **доступен** **во** **всей** **программе**. **Клиентский** **код** **не** **меняется**.

### Динамическая регистрация через декоратор

**Можно** **автоматизировать** **регистрацию**:

```python
class EstateFactory:
    _registry = {}

    @classmethod
    def register(cls, kind):
        def decorator(klass):
            cls._registry[kind] = klass
            return klass
        return decorator

    @classmethod
    def create(cls, kind, **kwargs):
        if kind not in cls._registry:
            raise ValueError(f"Неизвестный тип: {kind!r}")
        return cls._registry[kind](**kwargs)


@EstateFactory.register("studio")
class Studio(Apartment):
    pass


@EstateFactory.register("house")
class House(Apartment):
    pass
```

**Что** **это** **даёт:**

- **Классы** **регистрируются** **сами**.
- **Фабрика** **не** **знает** **заранее** **о** **классах**.
- **Легко** **добавлять** **новые** **типы** **без** **изменения** **фабрики**.

### Схема

```
        @EstateFactory.register("studio")
        class Studio:
            ...
                ↓
        _registry = {"studio": Studio}
                ↓
        EstateFactory.create("studio", ...)
                ↓
        Studio(...)
```

### Плюсы и минусы

**Плюсы:**

- **Гибко** — **классы** **регистрируются** **сами**.
- **Расширяемо** — **новые** **типы** **без** **изменения** **фабрики**.
- **Читаемо** — **декоратор** **явно** **указывает** **тип**.

**Минусы:**

- **Неявная** **связь** — **нужно** **знать** **о** **декораторе**.
- **Порядок** **загрузки** **модулей** **важен** — **класс** **должен** **быть** **импортирован**.

### Аналогия

Представь **справочник** **блюд** в **ресторане**. **Когда** **повар** **изобретает** **новое** **блюдо** — **он** **сам** **добавляет** **его** в **справочник**. **Официант** **просто** **открывает** **справочник** и **находит** **нужное**. **Фабрика** — **такой** **справочник**.

### Вывод

**Словарь** **классов** — **элегантный** **способ**. **Декоратор** **регистрации** — **самый** **гибкий** **вариант**.

---

## 25.6. Преимущества Factory

Разберём **пять** **ключевых** **преимуществ**.

### 25.6.1. Единая точка изменения

**При** **добавлении** **нового** **типа** **меняется** **только** **фабрика**:

```python
class EstateFactory:
    _registry = {
        "studio": Studio,
        "house": House,
        "penthouse": Penthouse,    # ← одна строка!
    }
```

**Клиентский** **код** **не** **трогается**.

### 25.6.2. Скрытие сложности

**Клиент** **не** **знает** **о** **классах**:

```python
# Плохо: клиент знает все классы.
if kind == "studio":
    obj = Studio(...)
elif kind == "house":
    obj = House(...)

# Хорошо: клиент знает только фабрику.
obj = EstateFactory.create(kind, ...)
```

### 25.6.3. Гибкость конфигурации

**Можно** **менять** **логику** **создания** **без** **изменения** **клиента**:

```python
# Сначала фабрика создаёт Apartment.
# Потом — зарегистрировали новый тип.
EstateFactory._registry["studio"] = LuxuryStudio
```

**Все** **дальнейшие** **создания** **studio** — **LuxuryStudio**.

### 25.6.4. Читаемость

**Код** **становится** **понятнее**:

```python
# Непонятно, что здесь создаётся.
obj = Studio(address, area, price, has_balcony, floor, ...)

# Понятно: создаётся объект типа "studio".
obj = EstateFactory.create("studio", address=..., area=..., price=...)
```

### 25.6.5. Связь с OCP

**Factory** **напрямую** **связана** с **принципом** **Open/Closed** (глава 35):

- **Открыт** **для** **расширения** — **можно** **добавлять** **новые** **типы**.
- **Закрыт** **для** **изменения** — **не** **меняем** **существующий** **код**.

**Добавление** **пентхауса** **не** **требует** **правки** **фабрики** (если **через** **регистрацию**) или **требует** **одной** **строки** (если **через** **словарь**).

### Аналогия

Представь **пульт** **от** **телевизора**. **Кнопки** — **типы** **каналов**. **Ты** **нажимаешь** **кнопку** — **телевизор** **показывает** **нужный** **канал**. **Тебе** **не** **нужно** **знать**, **как** **устроен** **телевизор** **внутри**. **Пульт** — **фабрика**. **Кнопки** — **типы**.

### Вывод

**Factory** **даёт** **пять** **преимуществ**: **единая** **точка** **изменения**, **скрытие** **сложности**, **гибкость**, **читаемость**, **связь** с **OCP**.

---

## 25.7. Factory и принцип Open/Closed

**Принцип** **Open/Closed** (OCP) — **один** из **пяти** **принципов** **SOLID**. Он **говорит**:

> **Программные** **сущности** **должны** **быть** **открыты** **для** **расширения**, **но** **закрыты** **для** **изменения**.

### Что это значит

- **Открыты** **для** **расширения** — **можно** **добавлять** **новое** **поведение**.
- **Закрыты** **для** **изменения** — **не** **нужно** **менять** **существующий** **код**.

### Как Factory помогает

**Без** **фабрики** **при** **добавлении** **типа**:

```python
def create_estate(kind, **kwargs):
    if kind == "studio":
        return Studio(**kwargs)
    elif kind == "house":
        return House(**kwargs)
    elif kind == "penthouse":              # ← ИЗМЕНЕНИЕ!
        return Penthouse(**kwargs)
    ...
```

**Каждый** **раз** **меняем** **существующий** **код**. **Нарушает** **OCP**.

**С** **фабрикой** **через** **регистрацию**:

```python
@EstateFactory.register("penthouse")
class Penthouse(Apartment):
    pass
```

**Фабрика** **не** **меняется**. **Только** **добавляем** **новый** **класс**. **Соответствует** **OCP**.

### Схема

```
        Без Factory                    С Factory
        ───────────                    ─────────
        Добавили тип:                  Добавили тип:
            ↓                              ↓
        Изменили if/elif               Зарегистрировали класс
            ↓                              ↓
        НАРУШИЛИ OCP                   СОБЛЮЛИ OCP
```

### Аналогия

Представь **розетку** в **стене**. Ты **можешь** **подключить** **любой** **прибор** — **фен**, **зарядку**, **лампу**. **Розетка** **не** **меняется** — **только** **приборы** **добавляются**. **Розетка** — **фабрика**. **Приборы** — **типы**. **Соответствует** **OCP**.

### Практический пример

**Плохо (нарушение OCP):**

```python
class Report:
    def generate(self, kind):
        if kind == "pdf":
            return self._pdf()
        elif kind == "excel":
            return self._excel()
        elif kind == "csv":                 # изменение!
            return self._csv()
```

**Хорошо (OCP):**

```python
class Report:
    _generators = {}

    @classmethod
    def register(cls, kind):
        def decorator(gen_class):
            cls._generators[kind] = gen_class
            return gen_class
        return decorator

    def generate(self, kind):
        return self._generators[kind]().render()
```

**Добавление** **csv** — **новый** **класс** + **декоратор**. **Существующий** **код** **не** **меняется**.

### Вывод

**Factory** **связана** с **принципом** **OCP**. **Через** **регистрацию** **типов** **мы** **соблюдаем** **OCP**: **расширяем** **поведение**, **не** **меняя** **существующий** **код**.

---

### 25.8.5. Наш дом

```python
class EstateFactory:
    _registry = {}

    @classmethod
    def register(cls, kind):
        def decorator(klass):
            cls._registry[kind] = klass
            return klass
        return decorator

    @classmethod
    def create(cls, kind, **kwargs):
        if kind not in cls._registry:
            raise ValueError(f"Неизвестный тип: {kind!r}")
        return cls._registry[kind](**kwargs)


@EstateFactory.register("studio")
class Studio(Apartment):
    pass


@EstateFactory.register("house")
class House(Apartment):
    pass
```

### Что общего

- **Скрывают** **логику** **создания**.

### Вывод

**Factory** — **стандартный** **паттерн**. **Используется** в **Django**, **SQLAlchemy**, **logging**, **JSON** и **многих** **других** **библиотеках**.

## 25.9. Нюансы и тонкости

Разберём **десять** **нюансов**.

### 25.9.1. Фабрика не обязана быть классом

**Simple Factory** — **часто** **просто** **функция**:

```python
def create_estate(kind, **kwargs):
    ...
```

**Класс** **нужен** **только** **если** **есть** **состояние** или **много** **методов**.

### 25.9.2. `**kwargs` — не всегда хорошо

**`**kwargs`** **скрывает** **параметры**. **IDE** **не** **подскажет**. **Ошибки** — **только** **в** **рантайме**.

**Альтернатива** — **явные** **параметры**:

```python
def create_estate(kind, address, area, price):
    ...
```

### 25.9.3. Валидация типа

**Всегда** **проверяйте** **тип**:

```python
if kind not in cls._registry:
    raise ValueError(f"Неизвестный тип: {kind!r}")
```

**Иначе** — **KeyError** или **странное** **поведение**.

### 25.9.4. Фабрика и наследование

**Наследники** **фабрики** **могут** **добавлять** **свои** **типы**:

```python
class LuxuryEstateFactory(EstateFactory):
    pass


@LuxuryEstateFactory.register("penthouse")
class Penthouse(Apartment):
    pass
```

**Проблема:** `LuxuryEstateFactory._registry` — **тот** **же** **словарь**, **что** и `EstateFactory._registry`. **Нужно** **копировать**:

```python
class LuxuryEstateFactory(EstateFactory):
    _registry = EstateFactory._registry.copy()
```

### 25.9.5. Регистрация типов из разных модулей

**Если** **типы** **в** **разных** **модулях** — **нужно** **их** **импортировать**, **чтобы** **декоратор** **сработал**:

```python
# __init__.py
from . import studio
from . import house
```

**Без** **импорта** — **класс** **не** **зарегистрируется**.

### 25.9.6. Фабрика и `@dataclass`

**Фабрика** **работает** с `@dataclass` **без** **проблем**:

```python
@dataclass
class Studio:
    address: str
    area: int
    price: int


EstateFactory._registry["studio"] = Studio
```

### 25.9.7. Фабрика и абстрактные классы

**Можно** **требовать** **наследования** от **абстракции**:

```python
from abc import ABC

class EstateItem(ABC):
    pass


@EstateFactory.register("studio")
class Studio(EstateItem):
    pass
```

**Проверка** — **через** `isinstance`.

### 25.9.8. Фабрика и `__init__`

**Фабрика** **может** **вызывать** **разные** **конструкторы**:

```python
class EstateFactory:
    _registry = {
        "studio": lambda **kw: Studio(**kw),
        "two_room_from_string": lambda data: TwoRoom.from_string(data),
    }
```

**Гибко**, **но** **менее** **читаемо**.

### 25.9.9. Фабрика и type hints

**Аннотации** **помогают**:

```python
from typing import Type

def create_estate(kind: str, **kwargs) -> "Apartment":
    ...
```

**Не** **влияет** на **выполнение**, **но** **полезно** для **документации**.

### 25.9.10. Фабрика и кэширование

**Фабрика** **может** **кэшировать** **объекты**:

```python
class CachedFactory:
    _cache = {}

    @classmethod
    def create(cls, kind, **kwargs):
        key = (kind, tuple(sorted(kwargs.items())))
        if key not in cls._cache:
            cls._cache[key] = cls._registry[kind](**kwargs)
        return cls._cache[key]
```

**Полезно**, **если** **создание** **дорогое**.

### Вывод

**Factory** **имеет** **много** **нюансов**. **Основные:** `**kwargs`, **регистрация**, **наследование**, **type hints**.

## 25.10. Типичные ошибки

Разберём **десять** **самых** **частых** **ошибок**.

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Забыли валидацию типа | `KeyError` вместо `ValueError` | Проверять |
| `**kwargs` без причины | Ошибки в рантайме | Явные параметры |
| Регистрация не сработала | Класс не импортирован | Импортировать |
| Наследники делят `_registry` | Конфликты | Копировать |
| Фабрика для всего | Избыточно | Только где нужно |
| Нет аннотаций | Непонятно | Добавить type hints |
| Логика в фабрике | Нарушает SRP | Только создание |
| Слишком много `if/elif` | Не масштабируется | Словарь |
| Нет документирования | Непонятно | Docstring |
| Мутация `_registry` | Случайные изменения | Не мутировать |

### Подробный разбор

**Ошибка 1.** Забыли **валидацию** — `KeyError` **вместо** `ValueError`. **Правило:** **всегда** **проверяйте** **тип**.

**Ошибка 2.** `**kwargs` **без** **причины** — **ошибки** **в** **рантайме**. **Правило:** **явные** **параметры**, **если** **возможно**.

**Ошибка 3.** Регистрация **не** **сработала** — **класс** **не** **импортирован**. **Правило:** **импортируйте** **модули** с **типами**.

**Ошибка 4.** Наследники **делят** `_registry` — **конфликты**. **Правило:** **копируйте** **словарь**.

**Ошибка 5.** Фабрика **для** **всего** — **избыточно**. **Правило:** **только** **где** **нужна** **логика** **создания**.

**Ошибка 6.** Нет **аннотаций** — **непонятно**. **Правило:** **type hints**.

**Ошибка 7.** Логика **в** **фабрике** — **нарушает** **SRP**. **Правило:** **только** **создание**.

**Ошибка 8.** Слишком **много** `if/elif` — **не** **масштабируется**. **Правило:** **словарь**.

**Ошибка 9.** Нет **документирования** — **непонятно**. **Правило:** **docstring**.

**Ошибка 10.** Мутация `_registry` — **случайные** **изменения**. **Правило:** **не** **мутируйте** **без** **нужды**.

### Вывод

Ошибки **с** **Factory** **типичны**. **Основные:** **валидация**, **`**kwargs`**, **регистрация**.

---

## 25.11. Что важно запомнить

- **Factory** — **порождающий** **паттерн**.
- **Скрывает** **логику** **создания** **объектов**.
- **Simple Factory** — **функция** или **класс** с `if/elif`.
- **Factory Method** — **через** **наследование**.
- **Словарь** **классов** — **элегантный** **вариант**.
- **Декоратор** **регистрации** — **самый** **гибкий**.
- **Преимущества:** **единая** **точка** **изменения**, **скрытие** **сложности**, **гибкость**, **читаемость**.
- **Связана** с **принципом** **OCP** — **открыт** **для** **расширения**, **закрыт** **для** **изменения**.
- **Валидация** **типа** — **обязательна**.
- **`**kwargs`** — **гибко**, **но** **менее** **читаемо**.
- **Регистрация** **типов** — **через** **декоратор**.
- **Используется** в **Django**, **SQLAlchemy**, **logging**, **JSON**.

## Задача 25 (паттерн Factory)

Агентство **регулярно** **пополняется** **разными** **типами** **объектов** — **студиями**, **двухкомнатными**, **домами**. **Создавать** их **напрямую** **через** **конструкторы** **неудобно**: **код** **разрастается**, **легко** **ошибиться**.

**Часть 1. Базовый класс `EstateItem`.** Создайте класс `EstateItem` с полями:

- `address` — адрес;
- `area` — площадь;
- `price` — цена.

Реализуйте `__init__` и `__repr__`.

**Часть 2. Наследники.** Создайте **три** класса:

- `Studio(EstateItem)` — **студия**;
- `TwoRoom(EstateItem)` — **двухкомнатная**;
- `House(EstateItem)` — **дом**.

**Часть 3. Simple Factory.** Создайте класс `EstateFactory` с методом:

```python
@classmethod
def create(cls, kind, **kwargs):
    ...
```

**Требования:**

- **по** **строке** `"studio"` — **создаёт** `Studio`;
- **по** **строке** `"two_room"` — `TwoRoom`;
- **по** **строке** `"house"` — `House`;
- **неизвестный** **тип** — `ValueError` с **понятным** **сообщением**.

**Часть 4. Реализация через словарь.** **Перепишите** фабрику **через** **словарь** `_registry`, **чтобы** **избежать** **длинной** **цепочки** `if/elif`.

**Часть 5. Проверка.** Продемонстрируйте:

- **создание** **каждого** **типа** **через** **фабрику**;
- `type(obj)` **соответствует** **типу**;
- **неизвестный** **тип** — `ValueError`.

**Часть 6. Динамическая регистрация.** Добавьте **метод** `register(kind)`, **который** **возвращает** **декоратор**:

```python
@EstateFactory.register("penthouse")
class Penthouse(EstateItem):
    pass
```

**Продемонстрируйте**, что **после** **регистрации** фабрика **создаёт** `Penthouse`.

**Часть 7. Обработка `**kwargs`.** Убедитесь, что фабрика **передаёт** все **параметры** в **конструктор**:

```python
studio = EstateFactory.create(
    "studio",
    address="ул. Пушкина",
    area=25,
    price=4_000_000,
)
```

**Часть 8. Валидация.** Продемонстрируйте, что **при** **неизвестном** **типе** **выбрасывается** `ValueError` с **сообщением**, **содержащим** **неизвестный** **тип**.

**Часть 9. Расширение.** Добавьте **новый** **тип** `Penthouse` **через** **регистрацию** — **без** **изменения** **кода** фабрики. **Это** **демонстрирует** **принцип** **OCP**.

**Часть 10.** В комментариях объясните:

1. **Что** **такое** **паттерн** **Factory** и **зачем** он **нужен**.
2. **Чем** **Simple Factory** **отличается** от **Factory Method**.
3. **Почему** **словарь** **классов** **удобнее** **цепочки** `if/elif`.
4. **Как** **Factory** **связана** с **принципом** **Open/Closed**.
5. **Какие** **минусы** у `**kwargs` в **фабрике**.
6. **Когда** **фабрика** **избыточна**.

### Решение

```python
# ─── Часть 1. Базовый класс EstateItem ───
class EstateItem:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price

    def __repr__(self):
        return (
            f"{self.__class__.__name__}("
            f"address={self.address!r}, area={self.area!r}, price={self.price!r}"
            f")"
        )


# ─── Часть 2. Наследники ───
class Studio(EstateItem):
    pass


class TwoRoom(EstateItem):
    pass


class House(EstateItem):
    pass


# ─── Части 3-4. Фабрика через словарь ───
class EstateFactory:
    _registry = {
        "studio": Studio,
        "two_room": TwoRoom,
        "house": House,
    }

    @classmethod
    def create(cls, kind, **kwargs):
        if kind not in cls._registry:
            raise ValueError(
                f"Неизвестный тип: {kind!r}. "
                f"Доступные: {list(cls._registry.keys())}"
            )
        return cls._registry[kind](**kwargs)

    # ─── Часть 6. Динамическая регистрация ───
    @classmethod
    def register(cls, kind):
        def decorator(klass):
            cls._registry[kind] = klass
            return klass
        return decorator

    @classmethod
    def list_kinds(cls):
        return list(cls._registry.keys())


# ─── Часть 5. Проверка ───
print("=== Часть 5. Создание через фабрику ===")

studio = EstateFactory.create(
    "studio",
    address="ул. Пушкина, д. 1, кв. 7",
    area=25,
    price=4_000_000,
)
two_room = EstateFactory.create(
    "two_room",
    address="ул. Ленина, д. 10, кв. 5",
    area=65,
    price=8_500_000,
)
house = EstateFactory.create(
    "house",
    address="ул. Садовая, д. 3",
    area=200,
    price=20_000_000,
)

print(studio)
print(two_room)
print(house)

print(f"type(studio): {type(studio).__name__}")
print(f"type(two_room): {type(two_room).__name__}")
print(f"type(house): {type(house).__name__}")

# ─── Часть 8. Валидация ───
print()
print("=== Часть 8. Валидация типа ===")

try:
    unknown = EstateFactory.create("villa", address="...", area=100, price=10_000_000)
except ValueError as e:
    print(f"ValueError: {e}")

# ─── Часть 9. Расширение через регистрацию ───
print()
print("=== Часть 9. Расширение через регистрацию ===")

print(f"До регистрации: {EstateFactory.list_kinds()}")


@EstateFactory.register("penthouse")
class Penthouse(EstateItem):
    pass


print(f"После регистрации: {EstateFactory.list_kinds()}")

penthouse = EstateFactory.create(
    "penthouse",
    address="ул. Мира, д. 1, кв. 100",
    area=150,
    price=50_000_000,
)
print(penthouse)

# ─── Часть 7. Проверка kwargs ───
print()
print("=== Часть 7. Передача kwargs ===")

custom = EstateFactory.create(
    "studio",
    address="ул. Гагарина",
    area=30,
    price=5_000_000,
)
print(custom)
print(f"Поля переданы: address={custom.address!r}, area={custom.area}, price={custom.price}")
```

**Ожидаемый вывод:**

```
=== Часть 5. Создание через фабрику ===
Studio(address='ул. Пушкина, д. 1, кв. 7', area=25, price=4000000)
TwoRoom(address='ул. Ленина, д. 10, кв. 5', area=65, price=8500000)
House(address='ул. Садовая, д. 3', area=200, price=20000000)
type(studio): Studio
type(two_room): TwoRoom
type(house): House

=== Часть 8. Валидация типа ===
ValueError: Неизвестный тип: 'villa'. Доступные: ['studio', 'two_room', 'house']

=== Часть 9. Расширение через регистрацию ===
До регистрации: ['studio', 'two_room', 'house']
После регистрации: ['studio', 'two_room', 'house', 'penthouse']
Penthouse(address='ул. Мира, д. 1, кв. 100', area=150, price=50000000)

=== Часть 7. Передача kwargs ===
Studio(address='ул. Гагарина', area=30, price=5000000)
Поля переданы: address='ул. Гагарина', area=30, price=5000000
```

### Разбор

**1. Что такое паттерн Factory и зачем он нужен.**

**Factory** — **порождающий** **паттерн**. **Скрывает** **логику** **создания** **объектов**. **Даёт** **единый** **интерфейс** для **создания** **разных** **типов**.

**Зачем:**

- **Единая** **точка** **изменения**.
- **Скрытие** **сложности** от **клиента**.
- **Гибкость** — **можно** **менять** **тип** **создаваемого** **объекта**.
- **Читаемость** — **клиент** **не** **думает** **о** **деталях**.

**2. Чем Simple Factory отличается от Factory Method.**

**Simple Factory** — **одна** **фабрика** с **логикой** **выбора** (через `if/elif` или **словарь**). **Простой** **случай**.

**Factory Method** — **фабрика** **через** **наследование**. **Базовый** **класс** **определяет** **метод** `create`, **наследники** **переопределяют**. **Гибкий** **случай**.

**Правило:** **Simple Factory** — **по** **умолчанию**. **Factory Method** — **если** **логика** **создания** **различается** **в** **разных** **контекстах**.

**3. Почему словарь классов удобнее цепочки `if/elif`.**

**Цепочка** `if/elif`:

- **Длинная** и **громоздкая**.
- **Легко** **забыть** **добавить** **ветку**.
- **Медленно** **при** **многих** **типах**.
- **Не** **позволяет** **динамически** **регистрировать** **типы**.

**Словарь**:

- **Компактный** — **все** **типы** **в** **одном** **месте**.
- **Быстрый** — **поиск** **по** **ключу** O(1).
- **Расширяемый** — **легко** **добавлять** **типы**.
- **Динамический** — **регистрация** **во** **время** **работы**.

**4. Как Factory связана с принципом Open/Closed.**

**OCP** — **классы** **открыты** **для** **расширения** (можно **добавлять** **новое** **поведение**), **закрыты** **для** **изменения** (не **трогаем** **существующий** **код**).

**Factory** с **регистрацией** **через** **декоратор** **соблюдает** **OCP**:

- **Добавляем** **новый** **тип** — **регистрируем** **декоратором**.
- **Существующий** **код** **фабрики** **не** **меняется**.
- **Клиентский** **код** **не** **меняется**.

**Без** **регистрации** (только **словарь**) — **нужно** **добавить** **строку** в `_registry`. Это **минимальное** **изменение**, **близкое** к **OCP**.

**5. Какие минусы у `**kwargs` в фабрике.**

- **IDE** **не** **подсказывает** **параметры**.
- **Ошибки** — **только** **в** **рантайме**.
- **Сложно** **документировать**.
- **Нет** **проверки** **типов** **на** **этапе** **написания**.

**Плюс:** **гибкость** — **можно** **передавать** **разные** **параметры** для **разных** **типов**.

**Правило:** **используйте** `**kwargs`, **только** **если** **параметры** **действительно** **различаются**.

**6. Когда фабрика избыточна.**

- **Один** **тип** **объекта** — **фабрика** **не** **нужна**.
- **Простое** **создание** — `Studio(...)` **проще**.
- **Нет** **логики** **выбора** — **фабрика** **избыточна**.
- **Один**-**два** **класса** — **можно** **обойтись** **без** **фабрики**.

**Правило:** **фабрика** **нужна**, **когда** **есть** **логика** **создания**, **много** **типов** или **нужна** **расширяемость**.

# Глава 26. Паттерн Observer

## 26.1. Проблема: как узнавать об изменениях

За предыдущие главы мы построили агентство недвижимости. У нас есть `Agency`, который **хранит** квартиры, **создаёт** их через фабрику (глава 25), **продаёт**, **сдаёт** в аренду. Всё **работает**. Но теперь представь **новую** **задачу**, которая **возникает** в **любом** **агентстве** — **уведомления** **клиентов**.

Клиенты **хотят** **знать**, **когда** **появляется** **новая** **квартира** с **нужными** **параметрами**. Или **когда** **снижается** **цена** на **интересующий** их **объект**. Или **когда** **освобождается** **квартира**, которую они **давно** **искали**.

**Первое**, что **приходит** в **голову** — **пусть** клиент **сам** **проверяет**. **Например**, **раз** в **час** **звонит** в **агентство** и **спрашивает**: «**Появилось** **что-то** **новое**?». **Это** **работает**, **но** **ужасно** **неудобно**:

- **Клиент** **должен** **помнить** о **звонках**.
- **Агентство** **тратит** **время** на **ответы**.
- **Если** **квартира** **появилась** **5** **минут** **назад** — **клиент** **узнает** **только** **через** **час**.
- **Если** **квартир** **много** — **клиент** **устанет** **звонить** **каждый** **раз**.

**Второе**, что **приходит** в **голову** — **пусть** **агентство** **само** **обзванивает** **клиентов**. **Но** **кого**? **Всех** **подряд**? **У** **каждого** **клиента** — **свои** **интересы**. **Одному** **нужны** **студии**, **другому** — **дома** **за** **городом**. **Обзвон** **всех** — **бессмысленно** и **навязчиво**.

**Идеальное** **решение** — **клиент** **сам** **подписывается** на **уведомления**, **агентство** **при** **появлении** **нового** **объекта** **уведомляет** **всех** **подписчиков**. **Клиент** **не** **звонит** — **агентство** **само** **сообщает**. **Каждый** **клиент** **получает** **то**, **что** **ему** **нужно**. Это **и** **есть** **паттерн** **Observer**.

### Аналогия

Представь **газету**. Ты **подписываешься** — **и** **каждое** **утро** **получаешь** **новый** **выпуск**. **Если** **не** **хочешь** — **отписываешься**, **и** **газета** **больше** **не** **приходит**. **Редакция** **не** **бегает** **за** **тобой** — **просто** **рассылает** **всем** **подписчикам**. **Observer** — **такая** **подписка**.

### Что такое Observer

**Observer** (Наблюдатель) — это **поведенческий** **паттерн** **проектирования**, **который**:

1. **Определяет** **зависимость** «**один**-**ко**-**многим**» между **объектами**.
2. **Позволяет** **одному** **объекту** (издателю) **уведомлять** **многих** **других** (подписчиков) об **изменениях**.
3. **Не** **создаёт** **жёсткой** **связи** между **издателем** и **подписчиками**.

Простыми **словами**: **Observer** — **это** **механизм** **подписки**. **Издатель** **рассылает** **события**, **подписчики** их **получают**.

### Где нужен Observer

- **Уведомления** **клиентов** — **новая** **квартира**, **снижение** **цены**.
- **GUI** — **кнопка** **нажата**, **поле** **изменено**.
- **Событийные** **системы** — **щелчок**, **нажатие** **клавиши**.
- **Мессенджеры** — **новое** **сообщение**.
- **Pub/Sub** — **издатель**-**подписчик**.
- **Реактивное** **программирование** — **изменение** **данных** **вызывает** **пересчёт**.

### Вывод

**Observer** решает **проблему** **уведомлений**. **Издатель** **не** **знает** **своих** **подписчиков** **конкретно** — **только** **что** у **них** **есть** **метод** для **получения** **событий**. Это **слабая** **связь** и **гибкость**.

---

## 26.2. Что такое паттерн Observer

**Observer** — **один** из **самых** **распространённых** **паттернов**. Он **относится** к **поведенческим** — **то** **есть** **управляет** **взаимодействием** **объектов**.

### Ключевая идея

**Observer** **работает** **так**:

1. **Издатель** (Subject) **хранит** **список** **подписчиков**.
2. **Подписчики** (Observers) **регистрируются** у **издателя**.
3. **При** **изменении** **издатель** **уведомляет** **всех** **подписчиков**.
4. **Подписчики** **реагируют** **на** **уведомление** **по-своему**.
5. **Подписчик** **может** **отписаться** — и **больше** **не** **получает** **событий**.

### Схема

```
        Издатель (Agency)
        ├── _subscribers = [Client1, Client2, Client3]
        │
        │   Новое событие
        ↓
        notify()
        ├──→ Client1.update(событие)
        ├──→ Client2.update(событие)
        └──→ Client3.update(событие)
```

### Терминология

Разные **источники** **называют** **участников** **по-разному**:

| Издатель | Подписчик |
|----------|-----------|
| Subject | Observer |
| Publisher | Subscriber |
| Observable | Listener |
| Producer | Consumer |

**Мы** будем **использовать** **Subject** и **Observer** — **классические** **термины**.

### Что даёт Observer

- **Слабую** **связь** — издатель **не** **знает** **конкретных** **подписчиков**.
- **Гибкость** — **легко** **добавлять** и **удалять** **подписчиков**.
- **Расширяемость** — **новые** **подписчики** **не** **требуют** **изменения** **издателя**.
- **Динамичность** — **подписка** **во** **время** **работы**.

### Что НЕ даёт Observer

- **Гарантию** **порядка** **уведомлений** — **порядок** **зависит** от **реализации**.
- **Гарантию** **доставки** — **если** **подписчик** **упал** — **его** **проблемы**.
- **Асинхронности** — **по** **умолчанию** **уведомления** **синхронные**.
- **Защиты** от **утечек** — **если** **подписчик** **не** **отписался** — **остаётся** в **памяти**.

### Аналогия

Представь **рассылку** **новостей**. **Издатель** — **редакция** **газеты**. **Подписчики** — **читатели**. **Редакция** **не** **бегает** **за** **каждым** — **просто** **отправляет** **выпуск** **всем** **подписчикам**. **Каждый** **читатель** **сам** **решает**, **что** **делать** с **новостью**: **прочитать**, **выбросить**, **сохранить**. **Редакции** **всё** **равно** — **главное**, **доставить**.

### Отличие от других паттернов

| Паттерн | Связь | Направление |
|---------|-------|-------------|
| **Observer** | Слабая | Издатель → Подписчики |
| **Mediator** | Средняя | Все → Посредник |
| **Pub/Sub** | Очень слабая | Через брокера |
| **Strategy** | Слабая | Клиент → Стратегия |

**Observer** — **прямая** **связь** **издателя** **с** **подписчиками**. **Pub/Sub** — **через** **посредника** (брокера).

### Вывод

**Observer** — **поведенческий** **паттерн**. **Издатель** **уведомляет** **подписчиков** об **изменениях**. **Слабая** **связь**, **гибкость**, **расширяемость**.

---

## 26.3. Простая реализация Observer

Разберём **простейшую** **реализацию** — **без** **абстракций**, **только** **суть**.

### Издатель

```python
class Agency:
    def __init__(self, name):
        self.name = name
        self.apartments = []
        self._subscribers = []      # список подписчиков

    def subscribe(self, client):
        self._subscribers.append(client)
        print(f"{client.name} подписался на {self.name}")

    def unsubscribe(self, client):
        if client in self._subscribers:
            self._subscribers.remove(client)
            print(f"{client.name} отписался от {self.name}")

    def add_apartment(self, apartment):
        self.apartments.append(apartment)
        # Уведомляем всех подписчиков.
        self.notify(apartment)

    def notify(self, apartment):
        for client in self._subscribers:
            client.update(apartment)
```

### Подписчик

```python
class Client:
    def __init__(self, name):
        self.name = name

    def update(self, apartment):
        print(f"{self.name} получил уведомление: {apartment.address}")
```

### Использование

```python
agency = Agency("Дом-Сервис")

client1 = Client("Иванов")
client2 = Client("Петров")

agency.subscribe(client1)
agency.subscribe(client2)

agency.add_apartment(Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000))
```

**Вывод:**

```
Иванов подписался на Дом-Сервис
Петров подписался на Дом-Сервис
Иванов получил уведомление: ул. Ленина, д. 10, кв. 5
Петров получил уведомление: ул. Ленина, д. 10, кв. 5
```

### Разбор по частям

- **`_subscribers`** — **список** **подписчиков**. **Хранит** **объекты** **Client**.
- **`subscribe(client)`** — **добавляет** **клиента** в **список**.
- **`unsubscribe(client)`** — **удаляет** **клиента**.
- **`add_apartment(apartment)`** — **создаёт** **событие** и **вызывает** `notify`.
- **`notify(apartment)`** — **проходит** по **всем** **подписчикам** и **вызывает** `update`.
- **`Client.update(apartment)`** — **реакция** **клиента** на **событие**.

### Что важно

- **Издатель** **не** **знает** **конкретных** **классов** **подписчиков**. Он **просто** **вызывает** `update`.
- **Подписчики** **регистрируются** **динамически**.
- **Событие** — **это** **объект** (`apartment`), который **передаётся** **всем**.

### Аналогия

Представь **чат** в **мессенджере**. **Участники** **подключаются** к **чату**. **Любое** **сообщение** **видят** **все** **участники**. **Если** **кто-то** **вышел** — **больше** **не** **видит** **сообщений**. **Чат** — **издатель**. **Участники** — **подписчики**. **Сообщение** — **событие**.

### Вывод

**Простейшая** **реализация** — **список** **подписчиков** + **метод** `notify`. **Работает** **для** **простых** **случаев**. **Для** **сложных** — **нужна** **абстракция**.

---

## 26.4. Observer с абстрактным интерфейсом

**Простейшая** **реализация** **работает**, **но** **имеет** **проблемы**:

- **Подписчик** **обязан** **иметь** **метод** `update`. **Если** **нет** — **ошибка**.
- **Нет** **контракта** — **какие** **методы** **должны** **быть** у **подписчика**?
- **Сложно** **расширять** — **разные** **типы** **подписчиков** **должны** **иметь** **одинаковый** **интерфейс**.

**Решение** — **абстрактный** **интерфейс** (глава 7).

### Абстрактный Observer

```python
from abc import ABC, abstractmethod


class Observer(ABC):
    @abstractmethod
    def update(self, event):
        pass
```

**Что** **это** **даёт:**

- **Все** **подписчики** **обязаны** **реализовать** `update`.
- **Проверка** **на** **этапе** **создания** — **нельзя** **создать** **подписчика** **без** `update`.
- **Документирует** **контракт** — **что** **должен** **уметь** **подписчик**.

### Абстрактный Subject

```python
class Subject(ABC):
    @abstractmethod
    def subscribe(self, observer):
        pass

    @abstractmethod
    def unsubscribe(self, observer):
        pass

    @abstractmethod
    def notify(self, event):
        pass
```

**Что** **это** **даёт:**

- **Единый** **интерфейс** для **всех** **издателей**.
- **Можно** **менять** **реализацию** **издателя**.
- **Документирует** **контракт**.

### Конкретные классы

```python
class Agency(Subject):
    def __init__(self, name):
        self.name = name
        self.apartments = []
        self._subscribers = []

    def subscribe(self, observer):
        if observer not in self._subscribers:
            self._subscribers.append(observer)

    def unsubscribe(self, observer):
        if observer in self._subscribers:
            self._subscribers.remove(observer)

    def notify(self, event):
        for observer in self._subscribers:
            observer.update(event)

    def add_apartment(self, apartment):
        self.apartments.append(apartment)
        self.notify(apartment)


class Client(Observer):
    def __init__(self, name):
        self.name = name

    def update(self, event):
        print(f"{self.name}: новое событие — {event.address}")
```

### Проверка контракта

```python
class BadClient(Observer):
    pass    # ← не реализовал update


bad = BadClient()    # TypeError: Can't instantiate abstract class
```

**Что** **произошло:** Python **не** **дал** **создать** **подписчика** **без** **реализации** `update`. **Это** **защита** от **ошибок**.

### Схема

```
        Subject (ABC)
        ├── subscribe()
        ├── unsubscribe()
        └── notify()
                ↑
        Agency (реализация)

        Observer (ABC)
        └── update()
                ↑
        ├── Client
        ├── EmailSubscriber
        └── SMSSubscriber
```

### Что важно

- **ABC** — **гарантирует** **наличие** **методов** у **подписчиков**.
- **Единый** **интерфейс** — **издатель** **работает** с **любым** **подписчиком**.
- **Расширяемость** — **можно** **добавлять** **новые** **типы** **подписчиков** **без** **изменения** **издателя**.

### Аналогия

Представь **подписку** **на** **журнал**. **Все** **подписчики** — **люди** **с** **адресом**. **Издатель** **не** **знает**, **кто** **конкретно** — **только** **что** у **всех** **есть** **адрес**. **ABC** — **это** **правило**: «**у** **каждого** **подписчика** **должен** **быть** **адрес**».

### Вывод

**Абстрактный** **интерфейс** — **лучшая** **практика**. **Гарантирует** **наличие** **методов**. **Документирует** **контракт**. **Позволяет** **расширять** **без** **изменения** **издателя**.

---

## 26.5. События и типы событий

В **простейшей** **реализации** **событие** — **это** **один** **объект** (например, `apartment`). Но в **реальном** **агентстве** **события** **бывают** **разные**:

- **Новая** **квартира**.
- **Снижение** **цены**.
- **Продажа**.
- **Изменение** **статуса**.

**Разные** **подписчики** **хотят** **разные** **события**. **Одному** — **только** **новые** **квартиры**. **Другому** — **только** **снижение** **цены**.

### Введение типов событий

**Решение** — **создать** **иерархию** **событий**:

```python
class Event:
    pass


class NewApartmentEvent(Event):
    def __init__(self, apartment):
        self.apartment = apartment


class PriceDropEvent(Event):
    def __init__(self, apartment, old_price, new_price):
        self.apartment = apartment
        self.old_price = old_price
        self.new_price = new_price


class SoldEvent(Event):
    def __init__(self, apartment, buyer):
        self.apartment = apartment
        self.buyer = buyer
```

### Издатель с типами

```python
class Agency:
    def __init__(self, name):
        self.name = name
        self.apartments = []
        self._subscribers = []

    def subscribe(self, observer):
        self._subscribers.append(observer)

    def unsubscribe(self, observer):
        self._subscribers.remove(observer)

    def notify(self, event):
        for observer in self._subscribers:
            observer.update(event)

    def add_apartment(self, apartment):
        self.apartments.append(apartment)
        self.notify(NewApartmentEvent(apartment))

    def drop_price(self, apartment, new_price):
        old_price = apartment.price
        apartment.price = new_price
        self.notify(PriceDropEvent(apartment, old_price, new_price))

    def sell(self, apartment, buyer):
        apartment.is_sold = True
        self.notify(SoldEvent(apartment, buyer))
```

### Подписчик с фильтрацией

```python
class Client:
    def __init__(self, name):
        self.name = name

    def update(self, event):
        # Обрабатываем только интересующие события.
        if isinstance(event, NewApartmentEvent):
            print(f"{self.name}: новая квартира — {event.apartment.address}")
        elif isinstance(event, PriceDropEvent):
            print(f"{self.name}: цена упала с {event.old_price} до {event.new_price}")
        elif isinstance(event, SoldEvent):
            print(f"{self.name}: квартира продана покупателю {event.buyer}")
```

### Что это даёт

- **Разные** **события** — **разные** **классы**.
- **Подписчики** **сами** **решают**, **какие** **события** **обрабатывать**.
- **Легко** **добавлять** **новые** **типы** **событий**.
- **Расширяемость** **без** **изменения** **издателя**.

### Использование

```python
agency = Agency("Дом-Сервис")
client = Client("Иванов")

agency.subscribe(client)

flat = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)

agency.add_apartment(flat)         # Иванов: новая квартира — ...
agency.drop_price(flat, 8_000_000) # Иванов: цена упала с 8500000 до 8000000
agency.sell(flat, "Петров")        # Иванов: квартира продана покупателю Петров
```

### Схема

```
        Event (базовый)
        ├── NewApartmentEvent
        ├── PriceDropEvent
        └── SoldEvent
                ↓
        notify(event)
                ↓
        observer.update(event)
                ↓
        if isinstance(event, ...):
            обработка
```

### Аналогия

Представь **почтовые** **уведомления**. **Разные** **конверты** — **разные** **типы** **писем**: **счёт**, **реклама**, **поздравление**. **Ты** **открываешь** **только** **нужные**. **Остальные** — **выбрасываешь**. **Типы** **событий** — **такие** **конверты**.

### Вывод

**Типы** **событий** — **мощный** **инструмент**. **Разные** **классы** для **разных** **событий**. **Подписчики** **фильтруют** **по** **типу**. **Расширяемость** **без** **изменения** **издателя**.

---

## 26.6. Реализация `subscribe` и `unsubscribe`

**Методы** **подписки** и **отписки** — **простые**, **но** **есть** **нюансы**. Разберём **подробно**.

### Базовые методы

```python
def subscribe(self, observer):
    if observer not in self._subscribers:
        self._subscribers.append(observer)
    else:
        print(f"{observer} уже подписан")


def unsubscribe(self, observer):
    if observer in self._subscribers:
        self._subscribers.remove(observer)
    else:
        print(f"{observer} не был подписан")
```

### Проверка на дубликаты

**Без** **проверки** — **можно** **подписаться** **дважды**. **Тогда** **уведомления** **приходят** **дважды**:

```python
agency.subscribe(client)
agency.subscribe(client)    # ← ошибка, но не заметно

agency.add_apartment(...)
# client.update вызван ДВАЖДЫ
```

**Правило:** **всегда** **проверяйте** **дубликаты** **в** `subscribe`.

### Проверка на существование

**Без** **проверки** в `unsubscribe` — **можно** **получить** `ValueError`:

```python
agency.unsubscribe(client)  # ValueError: list.remove(x): x not in list
```

**Правило:** **проверяйте** **наличие** **перед** **удалением**.

### Хранение подписчиков

**Список** — **простой**, **но** **медленный** при **большом** **количестве**:

```python
self._subscribers = []       # O(n) поиск
```

**Множество** — **быстрее**, **но** **требует** **хэшируемости**:

```python
self._subscribers = set()    # O(1) поиск
```

**Правило:** **для** **небольшого** **числа** — **список**. **Для** **большого** — **множество**.

### Использование `WeakSet` — защита от утечек

**Проблема:** если **подписчик** **не** **отписался** — он **остаётся** в **списке** и **не** **удаляется** сборщиком **мусора**. Это **утечка** **памяти**.

**Решение** — `weakref.WeakSet`:

```python
import weakref

class Agency:
    def __init__(self):
        self._subscribers = weakref.WeakSet()

    def subscribe(self, observer):
        self._subscribers.add(observer)

    def unsubscribe(self, observer):
        self._subscribers.discard(observer)

    def notify(self, event):
        for observer in list(self._subscribers):
            observer.update(event)
```

**Что** **это** **даёт:**

- **Если** **на** **подписчика** **больше** **нет** **ссылок** — он **удаляется** **автоматически**.
- **Не** **нужно** **явно** **отписываться**.
- **Нет** **утечек** **памяти**.

**Правило:** **используйте** `WeakSet`, **если** **подписчики** **могут** **создаваться** **динамически**.

### Отписка внутри уведомления

**Осторожно:** **если** **подписчик** **отписывается** **во** **время** `notify` — **можно** **сломать** **итерацию**:

```python
def notify(self, event):
    for observer in self._subscribers:      # ← изменение во время итерации!
        observer.update(event)
```

**Решение** — **итерироваться** по **копии**:

```python
def notify(self, event):
    for observer in list(self._subscribers):    # ← копия
        observer.update(event)
```

**Правило:** **всегда** **итерируйтесь** по **копии** `list(...)`.

### Схема

```
        subscribe(observer):
            если observer не в списке:
                добавить

        unsubscribe(observer):
            если observer в списке:
                удалить

        notify(event):
            для каждого observer в КОПИИ списка:
                observer.update(event)
```

### Аналогия

Представь **список** **гостей** на **вечеринке**. **Добавляешь** — **проверяешь**, **не** **в** **списке** **ли** **уже**. **Удаляешь** — **проверяешь**, **есть** **ли** в **списке**. **Обходишь** **всех** — **по** **списку**, **но** **если** **кто-то** **уходит** — **не** **ломай** **обход**.

### Вывод

`subscribe` **и** `unsubscribe` — **простые**, **но** **требуют** **внимания**. **Проверяйте** **дубликаты** и **существование**. **Используйте** `WeakSet` **для** **защиты** **от** **утечек**. **Итерируйтесь** по **копии**.

---

## 26.7. Observer vs другие паттерны

Разберём, **чем** **Observer** **отличается** от **похожих** **паттернов**.

### Observer vs Pub/Sub

**Observer** — **прямая** **связь** **издателя** **с** **подписчиками**. **Издатель** **знает** **о** **подписчиках**.

**Pub/Sub** — **через** **посредника** (**брокер**). **Издатель** **не** **знает** **о** **подписчиках**.

| Критерий | Observer | Pub/Sub |
|----------|----------|---------|
| Связь | Прямая | Через брокера |
| Издатель знает подписчиков | ✅ Да | ❌ Нет |
| Сложность | Простая | Сложная |
| Масштабируемость | Ограниченная | Высокая |
| Примеры | GUI, события | RabbitMQ, Kafka |

### Observer vs Mediator

**Mediator** — **объекты** **общаются** **через** **посредника**. **Никто** **не** **знает** **друг** **о** **друге** **напрямую**.

**Observer** — **издатель** **знает** **подписчиков**.

| Критерий | Observer | Mediator |
|----------|----------|----------|
| Связь | Издатель ↔ подписчики | Все ↔ посредник |
| Направление | Один-ко-многим | Многие-ко-многим |
| Издатель знает подписчиков | ✅ Да | ❌ Нет |

### Observer vs Strategy

**Strategy** — **клиент** **выбирает** **алгоритм** **из** **нескольких**. **Один** **клиент** — **одна** **стратегия**.

**Observer** — **издатель** **уведомляет** **многих**. **Много** **подписчиков**.

| Критерий | Observer | Strategy |
|----------|----------|----------|
| Направление | Один → много | Клиент → алгоритм |
| Динамичность | Подписка/отписка | Выбор алгоритма |
| Цель | Уведомления | Разные алгоритмы |

### Observer vs Callback

**Callback** — **функция**, **которая** **вызывается** **после** **события**. **Обычно** **одна**.

**Observer** — **много** **подписчиков**. **Более** **структурированный**.

| Критерий | Observer | Callback |
|----------|----------|----------|
| Количество | Много | Один или несколько |
| Структура | Класс с методом | Функция |
| Управление | Подписка/отписка | Прямая передача |

### Когда что использовать

- **Observer** — **много** **подписчиков** **на** **события**.
- **Pub/Sub** — **распределённые** **системы**, **микросервисы**.
- **Mediator** — **сложные** **взаимодействия** **многих** **объектов**.
- **Strategy** — **разные** **алгоритмы** **для** **одной** **задачи**.
- **Callback** — **простой** **случай** **одного** **обработчика**.

### Аналогия

- **Observer** — **радио**: **одна** **станция** — **много** **слушателей**.
- **Pub/Sub** — **почта**: **отправитель** — **почтамт** — **получатели**.
- **Mediator** — **диспетчер** **такси**: **все** **через** **него**.
- **Strategy** — **инструменты**: **молоток**, **отвёртка** — **выбираешь** **один**.
- **Callback** — **звонок** **другу**: **один** **человек**.

### Вывод

**Observer** — **прямая** **связь** **издателя** **с** **подписчиками**. **Pub/Sub** — **через** **брокера**. **Mediator** — **через** **посредника**. **Strategy** — **выбор** **алгоритма**. **Callback** — **простая** **функция**.

## 26.8. Observer в реальных проектах

### 26.8.1. GUI-фреймворки

**Сохранение** **модели** — **событие**. **Обработчики** — **подписчики** на **сигнал**.

### 26.8.3. Flask

**Flask** **использует** **Observer** для **сигналов**:

```python
from flask import request_started

@request_started.connect
def log_request(sender, **extra):
    print("Запрос начался")
```

```python
import asyncio

loop = asyncio.get_event_loop()
loop.call_later(1, callback)    # подписка на событие через 1 секунду
```

### 26.8.5. Наш дом

```python
class Agency(Subject):
    def __init__(self, name):
        self.name = name
        self.apartments = []
        self._subscribers = []

    def subscribe(self, observer):
        self._subscribers.append(observer)

    def notify(self, event):
        for observer in list(self._subscribers):
            observer.update(event)

    def add_apartment(self, apartment):
        self.apartments.append(apartment)
        self.notify(NewApartmentEvent(apartment))
```

## 26.9. Нюансы и тонкости

Разберём **десять** **нюансов**.

### 26.9.1. Итерируйтесь по копии

**Всегда** **итерируйтесь** по `list(self._subscribers)`. **Иначе** — **проблемы** при **изменении** **во** **время** **итерации**.

### 26.9.2. Проверка дубликатов

**Не** **давайте** **подписаться** **дважды** — **иначе** **двойные** **уведомления**.

### 26.9.3. Порядок уведомлений

**Порядок** **не** **гарантирован**, **если** **используете** `set`. **Список** — **сохраняет** **порядок**.

### 26.9.4. Слабые ссылки

`WeakSet` **защищает** от **утечек** **памяти**. **Используйте** в **долгоживущих** **издателях**.

### 26.9.5. Исключения в подписчиках

**Если** **один** **подписчик** **упадёт** — **другие** **не** **получат** **уведомление**. **Решение** — **обрабатывать** **исключения**:

```python
def notify(self, event):
    for observer in list(self._subscribers):
        try:
            observer.update(event)
        except Exception as e:
            print(f"Ошибка в {observer}: {e}")
```

### 26.9.6. Асинхронность

**Синхронный** **Observer** — **простой**, **но** **медленный** **подписчик** **блокирует** **всех**. **Асинхронный** — **сложнее**, **но** **быстрее**.

### 26.9.7. Типы событий

**Разные** **события** — **разные** **классы**. **Подписчики** **фильтруют** **по** `isinstance`.

### 26.9.8. Множественные издатели

**Один** **подписчик** **может** **подписаться** на **несколько** **издателей**:

```python
agency1.subscribe(client)
agency2.subscribe(client)
```

### 26.9.9. Отписка внутри подписчика

**Осторожно**: **если** **подписчик** **отписывается** **в** `update`, **итерация** **по** **копии** **спасёт**.

### 26.9.10. Слабые ссылки и обработка

**С** `WeakSet` — **подписчик** **может** **быть** **удалён** **до** `update`. **Проверяйте** **на** `None`.

### Вывод

**Observer** **имеет** **много** **нюансов**. **Основные:** **копия** **при** **итерации**, **проверки**, **исключения**, **слабые** **ссылки**.

## 26.10. Типичные ошибки

Разберём **десять** **самых** **частых** **ошибок**.

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Итерация без копии | Проблемы при изменении | `list(...)` |
| Нет проверки дубликатов | Двойные уведомления | Проверять |
| Подписчик без `update` | `AttributeError` | ABC или проверка |
| Забыли `unsubscribe` | Утечка памяти | WeakSet или явная отписка |
| Изменение списка в цикле | Пропуск элементов | Копия |
| Один издатель | Плохая архитектура | Несколько издателей |
| Синхронный медленный | Блокировка | Асинхронность |
| Нет обработки исключений | Падение всех | `try/except` |
| Тип события — строка | Плохо масштабируется | Классы событий |
| Порядок зависим от `set` | Непредсказуемость | Используйте `list` |

### Подробный разбор

**Ошибка 1.** Итерация **без** **копии** — **проблемы** при **изменении** **списка**. **Правило:** `list(...)`.

**Ошибка 2.** Нет **проверки** **дубликатов** — **двойные** **уведомления**. **Правило:** **проверяйте**.

**Ошибка 3.** Подписчик **без** `update` — `AttributeError`. **Правило:** **ABC** или **проверка** `hasattr`.

**Ошибка 4.** Забыли `unsubscribe` — **утечка** **памяти**. **Правило:** `WeakSet`.

**Ошибка 5.** Изменение **списка** в **цикле** — **пропуск** **элементов**. **Правило:** **копия**.

**Ошибка 6.** Один **издатель** — **плохая** **архитектура**. **Правило:** **несколько** **издателей** **по** **типам** **событий**.

**Ошибка 7.** Синхронный **медленный** — **блокировка** **всех**. **Правило:** **асинхронность** или **очередь**.

**Ошибка 8.** Нет **обработки** **исключений** — **падение** **всех**. **Правило:** `try/except` **в** `notify`.

**Ошибка 9.** Тип **события** — **строка** — **плохо** **масштабируется**. **Правило:** **классы** **событий**.

**Ошибка 10.** Порядок **зависит** от `set` — **непредсказуемость**. **Правило:** **используйте** `list`.

### Вывод

Ошибки **с** **Observer** **типичны**. **Основные:** **копия** **при** **итерации**, **проверки**, **исключения**.

## 26.11. Что важно запомнить

- **Observer** — **поведенческий** **паттерн**.
- **Издатель** **уведомляет** **подписчиков** об **изменениях**.
- **Слабая** **связь** — **издатель** **не** **знает** **подписчиков** **конкретно**.
- **Три** **метода** — `subscribe`, `unsubscribe`, `notify`.
- **ABC** — **гарантирует** **наличие** `update`.
- **Типы** **событий** — **классы** для **разных** **событий**.
- **`WeakSet`** — **защита** от **утечек** **памяти**.
- **Копия** **при** **итерации** — **обязательна**.
- **Обработка** **исключений** — **в** `notify`.
- **Используется** в **GUI**, **Django Signals**, **Flask**, **asyncio**.
- **Отличие** от **Pub/Sub** — **прямая** **связь**.
- **Отличие** от **Mediator** — **издатель** **знает** **подписчиков**.

## Задача 26 (паттерн Observer)

Клиенты **хотят** **получать** **уведомления**, **когда** **появляется** **квартира** с **нужными** **параметрами** или **снижается** **цена**. **Реализуйте** **паттерн** **Observer**.

**Часть 1. Базовый класс `Observer`.** Создайте **абстрактный** **класс** `Observer(ABC)` с **абстрактным** **методом** `update(event)`.

**Часть 2. Базовый класс `Subject`.** Создайте **абстрактный** **класс** `Subject(ABC)` с **абстрактными** **методами**:

- `subscribe(observer)`;
- `unsubscribe(observer)`;
- `notify(event)`.

**Часть 3. Типы событий.** Создайте **иерархию** **событий**:

- `Event` — **базовый**;
- `NewApartmentEvent(Event)` — **с** **полем** `apartment`;
- `PriceDropEvent(Event)` — **с** **полями** `apartment`, `old_price`, `new_price`;
- `SoldEvent(Event)` — **с** **полями** `apartment`, `buyer`.

**Часть 4. Класс `Agency(Subject)`.** Реализуйте:

- **хранение** **подписчиков** — **через** `WeakSet` (или **список**);
- **методы** `subscribe`, `unsubscribe`, `notify`;
- **методы** `add_apartment`, `drop_price`, `sell` — **создают** **события**.

**Часть 5. Класс `Client(Observer)`.** Реализуйте **метод** `update(event)`, **который** **обрабатывает** **разные** **типы** **событий**:

- `NewApartmentEvent` — **выводит** «новая квартира»;
- `PriceDropEvent` — **выводит** «цена упала»;
- `SoldEvent` — **выводит** «квартира продана».

**Часть 6. Дополнительный тип подписчика.** Создайте **класс** `Investor(Observer)` — **обрабатывает** **только** `PriceDropEvent` и `SoldEvent`. **Продемонстрируйте**, что **Investor** **игнорирует** **другие** **события**.

**Часть 7. Демонстрация.** Продемонстрируйте:

- **подписку** **двух** **клиентов**;
- **создание** **новой** **квартиры** — **все** **получают** **уведомление**;
- **снижение** **цены** — **все** **получают** **уведомление**;
- **отписку** **одного** **клиента** — **он** **больше** **не** **получает** **уведомлений**;
- **новое** **событие** — **только** **подписанные** **получают**.

**Часть 8. Дополнительно.** Реализуйте **обработку** **исключений** **в** `notify` — **чтобы** **один** **упавший** **подписчик** **не** **ломал** **всех** **остальных**.

**Часть 9.** В комментариях объясните:

1. **Что** **такое** **паттерн** **Observer** и **зачем** он **нужен**.
2. **Почему** **издатель** **не** **знает** **конкретных** **подписчиков**.
3. **Как** **работает** **подписка** и **отписка**.
4. **Зачем** **нужны** **типы** **событий**.
5. **Какие** **проблемы** **возникают** **при** **итерации** **по** **списку** **подписчиков**.
6. **Где** **используется** **Observer** в **реальных** **проектах**.

### Решение

```python
from abc import ABC, abstractmethod
import weakref


# ─── Часть 1. Observer ───
class Observer(ABC):
    @abstractmethod
    def update(self, event):
        pass


# ─── Часть 2. Subject ───
class Subject(ABC):
    @abstractmethod
    def subscribe(self, observer):
        pass

    @abstractmethod
    def unsubscribe(self, observer):
        pass

    @abstractmethod
    def notify(self, event):
        pass


# ─── Часть 3. Типы событий ───
class Event:
    pass


class NewApartmentEvent(Event):
    def __init__(self, apartment):
        self.apartment = apartment


class PriceDropEvent(Event):
    def __init__(self, apartment, old_price, new_price):
        self.apartment = apartment
        self.old_price = old_price
        self.new_price = new_price


class SoldEvent(Event):
    def __init__(self, apartment, buyer):
        self.apartment = apartment
        self.buyer = buyer


# ─── Вспомогательные классы ───
class Apartment:
    def __init__(self, address, area, price):
        self.address = address
        self.area = area
        self.price = price
        self.is_sold = False

    def __repr__(self):
        return f"Apartment({self.address!r}, {self.price})"


# ─── Часть 4. Agency ───
class Agency(Subject):
    def __init__(self, name):
        self.name = name
        self.apartments = []
        self._subscribers = []

    def subscribe(self, observer):
        if observer not in self._subscribers:
            self._subscribers.append(observer)
            print(f"Подписка оформлена")
        else:
            print(f"Уже подписан")

    def unsubscribe(self, observer):
        if observer in self._subscribers:
            self._subscribers.remove(observer)
            print(f"Подписка отменена")
        else:
            print(f"Не был подписан")

    def notify(self, event):
        # Итерируемся по КОПИИ списка.
        for observer in list(self._subscribers):
            try:
                observer.update(event)
            except Exception as e:
                # Часть 8. Обработка исключений.
                print(f"Ошибка в подписчике: {e}")

    def add_apartment(self, apartment):
        self.apartments.append(apartment)
        self.notify(NewApartmentEvent(apartment))

    def drop_price(self, apartment, new_price):
        old_price = apartment.price
        apartment.price = new_price
        self.notify(PriceDropEvent(apartment, old_price, new_price))

    def sell(self, apartment, buyer):
        apartment.is_sold = True
        self.notify(SoldEvent(apartment, buyer))


# ─── Часть 5. Client ───
class Client(Observer):
    def __init__(self, name):
        self.name = name

    def update(self, event):
        if isinstance(event, NewApartmentEvent):
            print(f"[{self.name}] Новая квартира: {event.apartment.address}")
        elif isinstance(event, PriceDropEvent):
            print(f"[{self.name}] Цена упала: "
                  f"{event.old_price} → {event.new_price} "
                  f"({event.apartment.address})")
        elif isinstance(event, SoldEvent):
            print(f"[{self.name}] Продана: {event.apartment.address} "
                  f"покупателю {event.buyer}")


# ─── Часть 6. Investor ───
class Investor(Observer):
    def __init__(self, name):
        self.name = name

    def update(self, event):
        # Обрабатывает только цену и продажи.
        if isinstance(event, PriceDropEvent):
            print(f"[Инвестор {self.name}] Интересная цена: "
                  f"{event.new_price} за {event.apartment.address}")
        elif isinstance(event, SoldEvent):
            print(f"[Инвестор {self.name}] Продано: {event.apartment.address}")


# ─── Часть 7. Демонстрация ───
print("=== Часть 7. Демонстрация ===")

agency = Agency("Дом-Сервис")

client1 = Client("Иванов")
client2 = Client("Петров")
investor = Investor("Сидоров")

agency.subscribe(client1)
agency.subscribe(client2)
agency.subscribe(investor)

print()
print("--- Новая квартира ---")
flat = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)
agency.add_apartment(flat)

print()
print("--- Снижение цены ---")
agency.drop_price(flat, 8_000_000)

print()
print("--- Отписка Петрова ---")
agency.unsubscribe(client2)

print()
print("--- Ещё снижение цены ---")
agency.drop_price(flat, 7_500_000)

print()
print("--- Продажа ---")
agency.sell(flat, "Смирнов")
```

**Ожидаемый вывод:**

```
=== Часть 7. Демонстрация ===
Подписка оформлена
Подписка оформлена
Подписка оформлена

--- Новая квартира ---
[Иванов] Новая квартира: ул. Ленина, д. 10, кв. 5
[Петров] Новая квартира: ул. Ленина, д. 10, кв. 5
(Инвестор Сидоров игнорирует — не его тип события)

--- Снижение цены ---
[Иванов] Цена упала: 8500000 → 8000000 (ул. Ленина, д. 10, кв. 5)
[Петров] Цена упала: 8500000 → 8000000 (ул. Ленина, д. 10, кв. 5)
[Инвестор Сидоров] Интересная цена: 8000000 за ул. Ленина, д. 10, кв. 5

--- Отписка Петрова ---
Подписка отменена

--- Ещё снижение цены ---
[Иванов] Цена упала: 8000000 → 7500000 (ул. Ленина, д. 10, кв. 5)
[Инвестор Сидоров] Интересная цена: 7500000 за ул. Ленина, д. 10, кв. 5

--- Продажа ---
[Иванов] Продана: ул. Ленина, д. 10, кв. 5 покупателю Смирнов
[Инвестор Сидоров] Продано: ул. Ленина, д. 10, кв. 5
```

### Разбор

**1. Что такое паттерн Observer и зачем он нужен.**

**Observer** — **поведенческий** **паттерн**. **Определяет** **зависимость** «**один**-**ко**-**многим**». **Издатель** **уведомляет** **подписчиков** об **изменениях**.

**Зачем:**

- **Слабая** **связь** — издатель **не** **знает** **подписчиков** **конкретно**.
- **Гибкость** — **легко** **добавлять** и **удалять** **подписчиков**.
- **Расширяемость** — **новые** **подписчики** **без** **изменения** **издателя**.
- **Согласованность** — **все** **подписчики** **получают** **актуальные** **данные**.

**2. Почему издатель не знает конкретных подписчиков.**

**Издатель** **работает** с **абстракцией** `Observer`. Он **знает** **только**, что у **подписчика** **есть** **метод** `update`. **Кто** **именно** — **клиент**, **инвестор**, **email-подписчик** — **издателю** **всё** **равно**.

**Это** **даёт** **гибкость**:

- **Можно** **добавлять** **новые** **типы** **подписчиков** **без** **изменения** **издателя**.
- **Можно** **менять** **подписчиков** **во** **время** **работы**.
- **Издатель** **не** **зависит** от **конкретных** **классов**.

**3. Как работает подписка и отписка.**

**Подписка** (`subscribe`):

- **Проверяем**, **не** **подписан** **ли** **уже**.
- **Если** **нет** — **добавляем** в **список** `_subscribers`.

**Отписка** (`unsubscribe`):

- **Проверяем**, **есть** **ли** в **списке**.
- **Если** **да** — **удаляем** из **списка**.

**Уведомление** (`notify`):

- **Итерируемся** по **копии** **списка**.
- **Вызываем** `update` у **каждого** **подписчика**.

**4. Зачем нужны типы событий.**

**Разные** **события** — **разные** **классы**:

- `NewApartmentEvent` — **новая** **квартира**.
- `PriceDropEvent` — **снижение** **цены**.
- `SoldEvent` — **продажа**.

**Что** **это** **даёт:**

- **Подписчики** **фильтруют** по **типу** (`isinstance`).
- **Разные** **события** — **разные** **данные** в **полях**.
- **Легко** **добавлять** **новые** **типы** **событий**.
- **Тип** **события** **несёт** **смысл** — **не** **просто** **строка**.

**5. Какие проблемы возникают при итерации по списку подписчиков.**

**Проблема:** **если** **подписчик** **отписывается** **во** **время** `notify` — **список** **изменяется** **во** **время** **итерации**. **Это** **приводит** к:

- **Пропуску** **элементов**.
- **Ошибкам** **итерации**.
- **Непредсказуемому** **поведению**.

**Решение:** **итерироваться** по **копии** — `list(self._subscribers)`.

**6. Где используется Observer в реальных проектах.**

- **GUI** — **кнопки**, **поля**, **события** **мыши**.
- **Django Signals** — **сигналы** **моделей**.
- **Flask** — **сигналы** **запроса**.
- **`asyncio`** — **событийный** **цикл**.
- **Pub/Sub** **системы** — **RabbitMQ**, **Kafka**.
- **Reactive** **программирование** — **RxPY**, **ReactiveX**.

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Агентство регулярно пополняется разными типами объектов — студиями, двухкомнатными, домами. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `EstateFactory` с методом `create(kind, **kwargs)`, который по строке `"studio"`, `"two_room"` или `"house"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному объекту каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Клиенты хотят получать уведомления, когда появляется квартира с нужными параметрами или снижается цена. Реализуйте паттерн «Наблюдатель»: класс `Agency` хранит список подписчиков (`Client`) и при появлении нового объекта уведомляет всех через метод `notify()`. Продемонстрируйте, что при появлении объекта все клиенты получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки клиент больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Agency` со временем разросся: он и хранит объекты, и оформляет сделки, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Agency` на несколько классов: `EstateStorage` (хранение объектов и поиск по адресу), `DealService` (оформление сделок, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «клиент купил квартиру». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Пентхаусы (`Penthouse`) похожи на обычные квартиры, но имеют дополнительные поля — терраса, панорамные окна. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`Penthouse(Apartment)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`Penthouse` содержит объект `Apartment` внутри себя как поле `self.apartment`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В агентстве нужно создавать копии карточек объектов — например, для архивного каталога. При этом у квартиры есть вложенный объект `Building`, и важно понимать, копируется ли он вместе с квартирой или остаётся общим. Реализуйте в классе `Apartment` методы `__copy__` и `__deepcopy__`. Создайте квартиру с вложенным объектом `Building` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — дом общий (`shallow.building is apartment.building` → `True`), при глубоком — свой (`deep.building is apartment.building` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В агентстве есть разные объекты, у которых есть метод `describe()` — квартиры, дома, клиенты, агенты, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Apartment`, но и с любым другим классом, у которого есть `describe()` (например, классом `Client`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В жилом комплексе есть список квартир, и хочется перебирать их в цикле `for` — так же удобно, как обычный список. Сейчас `for apartment in complex:` не работает. Реализуйте в классе `Complex` магические методы `__iter__` и `__next__` так, чтобы по комплексу можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте комплекс с несколькими квартирами и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(complex)` должен возвращать список всех квартир.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В комплексе квартиры должны быть доступны по индексу — как элементы списка: `complex[0]` возвращает первую квартиру, `complex[1] = apartment` заменяет вторую. Реализуйте в классе `Complex` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `complex[0]` — получение квартиры по индексу, `complex[1] = new_apartment` — замена квартиры, срезы `complex[0:2]` — получение подсписка, `for apartment in complex` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `apartment in complex`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В агентстве есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных сделок её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `DealDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with DealDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В агентстве есть разные способы расчёта комиссии: фиксированная (50 000 ₽), процентная (3% от цены) и «для постоянных клиентов» (1,5% от цены). Реализуйте паттерн «Стратегия»: базовый класс `CommissionStrategy` с методом `calculate(price)`, и три наследника — `FixedCommission`, `PercentCommission`, `LoyaltyCommission`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `LoyaltyCommission`, чтобы вызвать реализацию `PercentCommission` и уменьшить ставку. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В агентстве появился новый тип объекта — апартаменты (`ServicedApartment`). Если добавлять его через изменение базового класса `EstateItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `EstateItem` с методом `utility_cost()`; наследников `Studio`, `TwoRoom`; функцию `process_item(item: EstateItem)`, которая работает с любым объектом через родительский интерфейс. Добавьте `ServicedApartment` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В агентстве есть служебный объект `MortgageCalculator`, который должен уметь рассчитывать ежемесячный платёж по ипотеке. Удобно, чтобы его можно было вызывать как функцию: `calculator(price, years)`, а не `calculator.calculate(price, years)`. Реализуйте в классе `MortgageCalculator` магический метод `__call__(self, price, years)`, который возвращает примерный платёж (например, `price / (years * 12)`). Продемонстрируйте, что объект можно вызывать как функцию: `calc = MortgageCalculator(); calc(8500000, 20)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В агентстве есть класс `Apartment`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Apartment` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования квартир по адресу): два вызова `Apartment("ул. Ленина, д. 10, кв. 5", 65, 8500000)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В агентстве есть базовый класс `EstateItem` с методом `get_info()`. Нужно добавлять к объекту дополнительную информацию — например, «с ремонтом», «с мебелью», «ипотека возможна». При этом нельзя менять сам класс `EstateItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `EstateItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `RenovatedDecorator` и `FurnishedDecorator`, которые можно комбинировать: `RenovatedDecorator(FurnishedDecorator(apartment))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В агентстве появилась внешняя система учёта, которая возвращает данные о квартирах в другом формате — например, в виде словаря `{"location": "ул. Ленина, д. 10, кв. 5", "size": 65, "cost": 8500000}`. А весь остальной код агентства работает с классом `Apartment`, у которого поля `address`, `area`, `price`. Реализуйте паттерн «Адаптер»: класс `ApartmentAdapter`, который принимает словарь и предоставляет интерфейс `Apartment` — свойства `address`, `area`, `price` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Apartment` — например, передать его в функцию, которая работает с квартирой. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В агентстве есть базовый класс `EstateItem`, и постепенно в нём накопились методы, которые нужны не всем объектам: `sell()` (только для продаваемых), `rent()` (только для сдаваемых), `renovate()` (только для ремонтируемых). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Sellable`, `Rentable`, `Renovatable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `DealService`, который зависит не от конкретных классов (`Apartment`, `House`), а от абстракции `Sellable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип объекта (например, `ServicedApartment`, реализующий только `Sellable`) без изменения `DealService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).
