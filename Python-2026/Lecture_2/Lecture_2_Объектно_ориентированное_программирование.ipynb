# Глава 1. Классы и объекты

## 1.1. Зачем нужны классы

Представь, что мы пишем программу для управления домом. Нужно хранить данные о комнатах: название, площадь, этаж. Первое, что приходит в голову — завести отдельные переменные для каждой комнаты:

```python
name_1 = "Кухня"
area_1 = 12
floor_1 = 2

name_2 = "Спальня"
area_2 = 18
floor_2 = 2

name_3 = "Ванная"
area_3 = 6
floor_3 = 2
```

Пока комнат три — терпимо. Но уже на пятой комнате появляются проблемы:

- **Имена путаются.** Легко перепутать `area_2` и `area_3`, особенно если код большой.
- **Нет связи между данными.** Ничто не говорит, что `name_1`, `area_1`, `floor_1` относятся к одной комнате.
- **Сложно передавать в функции.** Чтобы передать «одну комнату» в функцию, нужно передавать три переменные:

```python
def show_room(name, area, floor):
    print(f"{name} — {area} м², {floor} этаж")

show_room(name_1, area_1, floor_1)
show_room(name_2, area_2, floor_2)
```

- **Сложно расширять.** Если завтра добавится ещё одно поле (например, «тип отделки»), придётся править все функции и все вызовы.

Класс решает все эти проблемы: он **объединяет связанные данные и поведение в единое целое**. Вместо трёх переменных — один объект, в котором лежит всё нужное.

---

## 1.2. Класс и объект: главное различие

**Класс** — это шаблон, чертёж, описание того, какими бывают объекты данного типа. Класс не содержит конкретных данных — он лишь описывает, **какие поля и методы вообще бывают** у объектов этого типа.

**Объект** — это конкретное воплощение класса, экземпляр с конкретными значениями.

Лучшая аналогия — чертёж дома и сам дом:

| Понятие | Аналогия | Пример |
|---------|----------|--------|
| Класс | Чертёж дома | «Комната» — у любой есть название, площадь, этаж |
| Объект | Построенный дом | Кухня, 12 м², 2 этаж |

По одному чертежу можно построить **сколько угодно зданий**, и каждое будет **отдельным объектом** этого класса. Каждое здание будет иметь те же поля, но с разными значениями.

Ключевая мысль: **класс — это не сама комната**, а описание, что у комнаты вообще бывает. Объект — конкретная комната с конкретными значениями.

---

## 1.3. Как определить класс

Класс определяется ключевым словом `class`. После него — имя класса. По договорённости (не требованию языка) имена классов пишут **с большой буквы**, а имена переменных — с маленькой. Так проще отличать классы от переменных при чтении кода.

Простейший класс выглядит так:

```python
class Room:
    pass
```

`pass` — это заглушка. Она говорит Python: «синтаксически здесь должно быть тело класса, но мы пока ничего не определяем». Без `pass` Python выдаст ошибку, потому что тело класса не может быть пустым.

Этот класс `Room` описывает понятие «комната», но пока ничего о ней не говорит: ни полей, ни методов.

---

## 1.4. Создание объектов

Чтобы создать объект, нужно **вызвать класс как функцию**:

```python
kitchen = Room()
bedroom = Room()
```

Мы получили **два разных объекта** класса `Room`. Это не один объект, а два независимых:

```python
print(kitchen is bedroom)   # False
print(id(kitchen))          # например, 140234567890
print(id(bedroom))          # например, 140234567891
```

`is` сравнивает, один ли это объект. `id()` возвращает адрес объекта в памяти. У `kitchen` и `bedroom` адреса разные — значит, это два разных объекта.

Пока эти объекты пусты: в них нет ни названия, ни площади.

## 1.5. Конструктор `__init__`

Чтобы объект сразу получал нужные данные, используют **конструктор** — специальный метод с именем `__init__` (по два подчёркивания с каждой стороны).

```python
class Room:
    def __init__(self, name, area, floor):
        self.name = name
        self.area = area
        self.floor = floor
```

Что происходит при создании объекта:

```python
kitchen = Room("Кухня", 12, 2)
```

1. Python создаёт новый пустой объект.
2. Python **автоматически** вызывает метод `__init__`, передавая туда новый объект и наши аргументы: `"Кухня"`, `12`, `2`.
3. Внутри `__init__` мы сохраняем значения в атрибуты через `self`.
4. Готовый объект возвращается и присваивается переменной `kitchen`.

### Нюанс: `__init__` не возвращает ничего

`__init__` — это **инициализатор**, а не создатель. Он не должен возвращать значение. Если попробовать:

```python
def __init__(self, name):
    self.name = name
    return self.name   # ← ошибка!
```

Python выдаст ошибку: `TypeError: __init__() should return None`. Создание объекта делает другой метод — `__new__` (мы разберём его в главе 37). `__init__` только заполняет уже созданный объект.

### Нюанс: имя `__init__` строго фиксировано

Именно `__init__`, а не `_init_`, не `init`, не `__init`. Если ошибиться в подчёркиваниях, Python не вызовет метод автоматически:

```python
class Room:
    def _init_(self, name):    # ← НЕ сработает!
        self.name = name

kitchen = Room("Кухня")        # TypeError: __init__() takes 1 positional argument
```

## 1.6. `self` — что это и зачем

`self` — это **ссылка на текущий объект**. Когда метод вызывается у объекта, Python автоматически подставляет этот объект в первый параметр метода.

### Зачем нужен `self`

Когда мы пишем:

```python
self.name = name
```

мы говорим: «возьми значение параметра `name` и положи его в атрибут `name` **именно этого объекта**».

Без `self` Python не понял бы, к какому объекту относится атрибут. Ведь класс один, а объектов может быть много:

```python
kitchen = Room("Кухня", 12, 2)
bedroom = Room("Спальня", 18, 2)
```

`self` в момент создания `kitchen` указывает на `kitchen`, а в момент создания `bedroom` — на `bedroom`. Поэтому каждый объект хранит **свои** значения.

### Нюанс: `self` — это просто имя

`self` — это **не ключевое слово**, а просто соглашение. Первый параметр метода можно назвать как угодно, и код будет работать:

```python
class Room:
    def __init__(this, name):    # работает!
        this.name = name
```

Но так писать **не принято** — весь Python-мир использует `self`. Если назвать иначе, коллеги не поймут код.

### Нюанс: при вызове через объект `self` не передаётся

Это очень важный момент. Когда мы вызываем метод через объект:

```python
kitchen.info()
```

Python **сам** подставляет `kitchen` в `self`. Мы не пишем `kitchen.info(kitchen)`.

Но если вызвать метод через класс:

```python
Room.info(kitchen)
```

то объект нужно передать вручную. Эти два вызова **делают одно и то же**:

```python
kitchen.info()          # Python подставит kitchen в self
Room.info(kitchen)      # то же самое, но вручную
```

Понимание этого помогает понять, что `self` — это не магия, а обычный параметр, который Python подставляет автоматически.

## 1.7. Атрибуты объекта

**Атрибуты** — это данные, которые хранятся в объекте. Они определяются через `self` в конструкторе или в методах.

```python
class Room:
    def __init__(self, name, area, floor):
        self.name = name      # атрибут name
        self.area = area      # атрибут area
        self.floor = floor    # атрибут floor
```

Каждый объект имеет **свои значения** этих атрибутов:

```python
kitchen = Room("Кухня", 12, 2)
bedroom = Room("Спальня", 18, 2)

print(kitchen.name)    # Кухня
print(bedroom.name)    # Спальня
print(kitchen.area)    # 12
print(bedroom.area)    # 18
```

### Доступ к атрибутам

Снаружи класса к атрибутам обращаются через точку:

```python
print(kitchen.name)    # чтение
kitchen.area = 15      # изменение
```

Внутри класса — через `self`:

```python
def info(self):
    return self.name       # обращение к атрибуту через self
```

### Нюанс: атрибуты создаются в момент присваивания

Атрибут появляется у объекта **только тогда**, когда ему что-то присвоили. Если атрибут не установлен, обращение к нему вызовет `AttributeError`:

```python
class Room:
    def __init__(self, name):
        self.name = name

kitchen = Room("Кухня")
print(kitchen.area)      # AttributeError: 'Room' object has no attribute 'area'
```

Здесь `area` не был установлен в конструкторе — и его нет.

## 1.8. Методы класса

**Метод** — это функция, определённая внутри класса. Методы описывают **поведение** объекта: что он умеет делать.

```python
class Room:
    def __init__(self, name, area, floor):
        self.name = name
        self.area = area
        self.floor = floor

    def info(self):
        return f"{self.name} — {self.area} м², {self.floor} этаж"
```

Разберём подробно:

- `def info(self):` — объявление метода. Первый параметр — всегда `self`.
- `self.name` — обращение к атрибуту текущего объекта.
- `return` — возвращает строку.

### Вызов метода

```python
kitchen = Room("Кухня", 12, 2)
print(kitchen.info())    # Кухня — 12 м², 2 этаж
```

Python автоматически подставил `kitchen` в `self`.

### Нюанс: метод — это тоже атрибут

С точки зрения Python метод — это **атрибут класса**, значением которого является функция:

```python
print(Room.info)              # <function Room.info at 0x...>
print(kitchen.info)           # <bound method Room.info of <Room object at 0x...>>
```

Когда мы обращаемся к `kitchen.info`, Python создаёт «связанный метод» — функцию, в которую уже «вшит» объект `kitchen`. При вызове `kitchen.info()` этот объект автоматически передаётся в `self`.

### Метод vs функция

| | Функция | Метод |
|---|---------|-------|
| Где определена | Вне класса | Внутри класса |
| Первый параметр | Любой | Всегда `self` |
| Вызов | `info(kitchen)` | `kitchen.info()` |
| Доступ к данным | Через аргументы | Через `self` |

## 1.9. Как Python ищет атрибуты и методы

Когда мы пишем `kitchen.info()`, Python выполняет следующие шаги:

1. **Ищет `info` в самом объекте** `kitchen`. Если бы у `kitchen` был свой атрибут `info`, был бы использован он.
2. **Если не нашёл — ищет в классе** `Room`. Там метод `info` есть.
3. **Если не нашёл — ищет у родителей класса** (если есть наследование).
4. **Если не нашёл нигде — `AttributeError`**.

Такая последовательность называется **поиск по MRO** (Method Resolution Order) — мы разберём её подробно в главе 19.

### Пример

```python
class Room:
    type = "Комната"    # атрибут класса

    def __init__(self, name):
        self.name = name    # атрибут экземпляра

kitchen = Room("Кухня")
print(kitchen.name)   # Кухня — найден в самом объекте
print(kitchen.type)   # Комната — найден в классе
```

Это отличие между **атрибутами экземпляра** и **атрибутами класса** — мы разберём его подробно в главе 16.

## 1.10. Что происходит при вызове `Room("Кухня", 12, 2)`

Давай проследим **пошагово**, что делает Python:

1. **Создаётся новый пустой объект** — с помощью метода `__new__` (мы его разберём в главе 37).
2. **Вызывается `__init__`** — Python автоматически передаёт туда новый объект и наши аргументы:
   ```python
   Room.__init__(new_object, "Кухня", 12, 2)
   ```
3. **Внутри `__init__`** создаются атрибуты:
   - `new_object.name = "Кухня"`
   - `new_object.area = 12`
   - `new_object.floor = 2`
4. **`__init__` завершается**, возвращая `None` (это обязательно).
5. **Готовый объект возвращается** из `Room(...)` и присваивается переменной `kitchen`.

Понимание этого процесса помогает понять, почему `self` — это ссылка на объект, и почему `__init__` не возвращает значение.

## 1.11. Динамическое добавление атрибутов

В Python (в отличие от многих языков) атрибуты можно добавлять объекту **в любой момент**, а не только в конструкторе:

```python
kitchen = Room("Кухня", 12, 2)
kitchen.window_count = 2       # добавили новый атрибут
print(kitchen.window_count)    # 2
```

Это работает, но считается **плохой практикой**, потому что:

- Класс перестаёт быть «шаблоном» — объекты становятся разными по структуре.
- IDE не подскажет, какие атрибуты есть у объекта.
- Опечатки превращаются в баги: `kitchen.windows_count = 2` создаст **новый** атрибут, а не исправит старый.

**Правило:** все атрибуты объекта лучше определять в `__init__`. Тогда у всех объектов класса будет одинаковый набор полей.

## 1.12. `__dict__` — где хранятся атрибуты

Каждый объект хранит свои атрибуты в специальном словаре `__dict__`:

```python
kitchen = Room("Кухня", 12, 2)
print(kitchen.__dict__)
# {'name': 'Кухня', 'area': 12, 'floor': 2}
```

Это **не магия** — Python действительно хранит атрибуты как пары «ключ-значение». Обращение `kitchen.name` — это фактически поиск в этом словаре.

У класса тоже есть свой `__dict__` — там хранятся методы и атрибуты класса:

```python
print(Room.__dict__)
# {'__init__': ..., 'info': ..., '__dict__': ..., ...}
```

Понимание `__dict__` поможет в главе 21 (`__slots__`), где мы будем экономить память, отказываясь от этого словаря.

## 1.13. Типичные ошибки новичка

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Забыли `self` в параметрах метода | `def info():` | `def info(self):` |
| Забыли `self.` при обращении к атрибуту | `return name` | `return self.name` |
| Путают класс и объект | `Room.info()` | `kitchen.info()` |
| Неправильно пишут `__init__` | `_init_`, `init` | `__init__` |
| Возвращают значение из `__init__` | `return self.name` | убрать `return` |
| Передают `self` вручную при вызове | `kitchen.info(kitchen)` | `kitchen.info()` |
| Добавляют атрибуты вне `__init__` | `kitchen.x = 5` | добавить в `__init__` |
| Обращаются к атрибуту до его создания | `kitchen.area` до `__init__` | сначала создать, потом обращаться |
| Думают, что `self` — ключевое слово | называют как угодно | использовать `self` |

## 1.14. Что важно запомнить

- **Класс** — шаблон, **объект** — конкретное воплощение.
- **`__init__`** — конструктор, вызывается автоматически при создании объекта.
- **`__init__` не возвращает значение** — только инициализирует.
- **`self`** — ссылка на текущий объект, всегда первый параметр метода.
- **`self` — не ключевое слово**, а соглашение.
- **Из одного класса можно создать сколько угодно объектов** с разными данными.
- **Атрибуты объекта создаются в момент присваивания**, а не заранее.
- **Метод — это атрибут класса**, но с автоматической подстановкой `self`.
- **`__dict__`** — где реально хранятся атрибуты объекта.
- **Все атрибуты лучше определять в `__init__`**, а не добавлять динамически.

## Задача 1 (создание класса, конструктор, метод)

В доме ведётся учёт комнат. Создайте класс `Room` с полями `name` (название), `area` (площадь в м²) и `floor` (этаж). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о комнате в формате:

```
Название — площадь ... м², ... этаж
```

Создайте **два экземпляра** класса с разными данными, вызовите у каждого метод `info()` и выведите результат на экран. Ожидаемый вывод, например:

```
Кухня — площадь 12 м², 2 этаж
Спальня — площадь 18 м², 3 этаж
```

В комментариях поясните:

1. Чем класс отличается от объекта.
2. Зачем нужен параметр `self` в методах.
3. Что произойдёт, если забыть `self` при обращении к атрибуту.

### Решение

```python
class Room:
    def __init__(self, name, area, floor):
        # self — ссылка на конкретный объект.
        # Благодаря self каждая комната хранит свои значения.
        self.name = name
        self.area = area
        self.floor = floor

    def info(self):
        # Возвращаем строку. self.name — название именно этой комнаты.
        # Без self Python не понял бы, из какого объекта брать name.
        return f"{self.name} — площадь {self.area} м², {self.floor} этаж"


kitchen = Room("Кухня", 12, 2)
bedroom = Room("Спальня", 18, 3)

print(kitchen.info())
print(bedroom.info())
```

### Разбор

**1. Чем класс отличается от объекта.**
Класс `Room` — это описание, шаблон: «у любой комнаты есть название, площадь и этаж». Объекты `kitchen` и `bedroom` — конкретные комнаты с конкретными значениями. Из одного класса можно создать сколько угодно объектов с разными данными.

**2. Зачем нужен `self`.**
`self` — это ссылка на текущий объект. Когда мы вызываем `kitchen.info()`, Python автоматически передаёт в `self` объект `kitchen`. Без `self` метод не понял бы, к какой именно комнате относится `name`, `area`, `floor`.

**3. Что произойдёт, если забыть `self`.**
Если написать:

```python
def info(self):
    return f"{name} — площадь {area} м², {floor} этаж"
```

Python будет искать переменные `name`, `area`, `floor` **внутри метода**, не найдёт их и выдаст `NameError`. Атрибуты объекта доступны **только** через `self`: `self.name`, `self.area`, `self.floor`.

# Глава 2. Инкапсуляция и свойства

## 2.1. Проблема: данные доступны всем

В главе 1 мы создали класс `Room` с полями `name`, `area`, `floor`. Всё работает — но есть скрытая проблема: **данные объекта доступны из любого места программы**. Любой может случайно или намеренно изменить их на что угодно:

```python
kitchen = Room("Кухня", 12, 2)
kitchen.area = -12          # площадь стала отрицательной
kitchen.name = 12345        # название стало числом
kitchen.floor = "второй"    # этаж стал строкой
```

Программа не сломается сразу — но данные теперь бессмысленные. Отрицательная площадь, название-число, этаж-строка — с этим невозможно работать. А поймать такую ошибку сложно: она проявится гораздо позже, когда что-то пойдёт не так.

Такая ситуация называется **нарушением целостности данных**. Решение — **инкапсуляция**.

---

## 2.2. Что такое инкапсуляция

**Инкапсуляция** — это принцип объектно-ориентированного программирования, по которому:

1. **Данные объекта скрываются** от прямого доступа извне.
2. **Доступ к данным предоставляется** только через специальные механизмы.
3. **При изменении данных выполняется проверка** — можно ли записать такое значение.

Простыми словами: инкапсуляция — это **защита внутреннего состояния объекта**. Снаружи не видно, что у объекта внутри; снаружи можно только попросить: «прочитай мне это» или «запиши вот это» — и объект сам решит, можно ли это сделать.

### Аналогия

Представь банкомат. Ты не можешь залезть внутрь и взять деньги — там всё защищено. Но можешь **попросить** банкомат выдать сумму. Банкомат **сам проверит**, есть ли у тебя столько денег, и только потом выдаст.

Инкапсуляция работает так же: объект сам решает, что можно с ним сделать, а что — нет.

---

## 2.3. Три уровня доступа в Python

В Python есть **три уровня доступа** к атрибутам. Обозначаются они количеством подчёркиваний перед именем:

| Уровень | Синтаксис | Доступ | Когда использовать |
|---------|-----------|--------|-------------------|
| **Публичный** | `name` | Везде | Обычные данные |
| **Защищённый** | `_name` | Внутри класса и у наследников | Внутренние данные, но не строго скрытые |
| **Приватный** | `__name` | Только внутри класса | Чувствительные данные |

Разберём каждый подробно.

### Публичные атрибуты

Пишутся **без подчёркиваний**:

```python
self.name = name
```

Доступны откуда угодно:

```python
kitchen = Room("Кухня", 12, 2)
print(kitchen.name)      # Кухня — работает
kitchen.name = "Спальня" # работает
```

Такие атрибуты — часть «открытого интерфейса» класса. Использовать их можно свободно.

### Защищённые атрибуты

Начинаются с **одного подчёркивания**:

```python
self._floor = floor
```

Это **не запрет**, а **соглашение**: «не трогай без нужды, это внутреннее». Доступ снаружи **возможен**, но не рекомендуется:

```python
print(kitchen._floor)    # работает, но так делать не принято
```

Защищённые атрибуты используют, когда поле нужно наследникам класса, но не должно попадать в «публичный интерфейс».

### Приватные атрибуты

Начинаются с **двух подчёркиваний**:

```python
self.__area = area
```

Python **переименовывает** их (см. раздел 2.4), и снаружи обратиться напрямую **нельзя**:

```python
print(kitchen.__area)
# AttributeError: 'Room' object has no attribute '__area'
```

Приватные атрибуты — самые защищённые. Доступ к ним возможен **только внутри класса**.

### Важный нюанс

Строго говоря, **в Python нет настоящих приватных полей**. Даже `__area` можно достать через name mangling (см. 2.4). Но программисты договорились: если видишь `__name` — не лезь. Это осознанное решение, и обычно его уважают.

---

## 2.4. Name mangling — что это и зачем

**Name mangling** (искажение имени) — это механизм Python, при котором приватные атрибуты **автоматически переименовываются**. Если в классе `Room` есть поле `__area`, Python превращает его в `_Room__area`.

```python
class Room:
    def __init__(self, area):
        self.__area = area

kitchen = Room(12)

# print(kitchen.__area)              # AttributeError
print(kitchen._Room__area)           # 12 — работает!
```

### Как это работает

Когда Python видит `self.__area` внутри класса `Room`, он **автоматически** заменяет это на `self._Room__area`. То же самое происходит при чтении `self.__area`:

```python
class Room:
    def __init__(self, area):
        self.__area = area       # превратится в self._Room__area

    def show(self):
        return self.__area       # тоже self._Room__area
```

Снаружи класса `kitchen.__area` **не сработает**, потому что реальное имя — `_Room__area`, а не `__area`.

### Зачем это нужно

Name mangling нужен не для **безопасности**, а для **двух других целей**:

1. **Избежать конфликтов имён между родителем и наследником.**
   Если у родителя есть `self.__x`, а у наследника — тоже `self.__x`, без name mangling они бы перезаписывали друг друга. С ним — это `_Parent__x` и `_Child__x`, разные поля.

2. **Сигнализировать другим программистам:** «это внутреннее, не лезь».

### Ключевая мысль

Приватность в Python — **скорее договорённость, чем жёсткая защита**. Программисты договорились: если видишь два подчёркивания — не трогай. Но если очень нужно, обойти можно (через `_Class__name`), и это осознанное решение.

---

## 2.5. Как тогда работать с приватными данными?

Приватное поле `__area` снаружи недоступно. Но нам всё равно нужно **читать** площадь и **менять** её (с проверкой). Как быть?

Ответ: сделать **специальные методы**, через которые и будет идти доступ. Такие методы называют:

- **Геттерами** (от англ. *get* — «получать») — для **чтения**.
- **Сеттерами** (от англ. *set* — «устанавливать») — для **записи**.

Разберём **два способа** их организации:

1. **Классический** — обычные методы `get_` / `set_`.
2. **Современный** — свойства через `@property`.

Начнём с классического — он проще для понимания.

---

## 2.6. Способ 1. Классические геттеры и сеттеры

Идея простая: для каждого приватного поля пишем **два обычных метода**.

- **Сеттер:** `set_имя_поля(value)` — записывает значение (с проверкой).
- **Геттер:** `get_имя_поля()` — возвращает значение.

### Пример

```python
class Room:
    def __init__(self):
        self.__area = None

    def set_area(self, value):
        # Сеттер: проверяем и записываем.
        if 5 <= value <= 100:
            self.__area = value
        else:
            print("Недопустимая площадь")

    def get_area(self):
        # Геттер: возвращаем значение.
        return self.__area


kitchen = Room()
kitchen.set_area(12)       # записываем через сеттер
print(kitchen.get_area())  # 12 — читаем через геттер

kitchen.set_area(-5)       # Недопустимая площадь
print(kitchen.get_area())  # 12 — значение не изменилось
```

### Как это работает

1. Поле `__area` **приватное** — снаружи напрямую не доступно.
2. Чтобы записать — вызываем `set_area(12)`.
3. Чтобы прочитать — вызываем `get_area()`.
4. Внутри сеттера можно **проверить** значение и не записывать, если оно плохое.

### Плюсы классического подхода

- **Явно:** сразу видно, что это метод: `get_area()`, `set_area(12)`.
- **Понятно новичку:** это просто методы, ничего особенного.
- **Гибко:** можно делать любую логику внутри.

### Минусы классического подхода

- **Длинно:** `room.get_area()` вместо `room.area`.
- **Не похоже на обычное поле:** нельзя написать `room.area = 12`.
- **Много кода** для простых случаев.

### Типичная ошибка

Забыть, что поле приватное, и написать `room.__area = 12` — не сработает, потому что Python переименовал поле в `_Room__area`. Создастся **новый** атрибут `__area`, а приватное поле останется прежним:

```python
kitchen.__area = 999       # создастся НОВЫЙ атрибут __area, а не изменится приватный
print(kitchen.get_area())  # 12 — приватное поле не изменилось!
```

## 2.7. А можно ли сделать красивее?

Классический подход работает, но выглядит громоздко. Хочется, чтобы можно было писать так:

```python
kitchen.area = 12          # как обычное поле
print(kitchen.area)        # и читать как обычное поле
```

И при этом чтобы **проверка работала** и поле оставалось защищённым.

Именно для этого в Python придумали **свойства через `@property`**.

## 2.8. Способ 2. Свойства через `@property`

`@property` — это **декоратор**, который позволяет методу **притвориться обычным полем**. Снаружи кажется, что `area` — это просто атрибут, но за кулисами вызывается метод.

### Как объявить

```python
class Room:
    def __init__(self):
        self.__area = None

    @property
    def area(self):
        # Геттер: вызывается при чтении room.area.
        return self.__area

    @area.setter
    def area(self, value):
        # Сеттер: вызывается при room.area = value.
        if 5 <= value <= 100:
            self.__area = value
        else:
            print("Недопустимая площадь")
```

### Как использовать

```python
kitchen = Room()
kitchen.area = 12          # работает как обычное поле!
print(kitchen.area)        # 12 — читается как обычное поле

kitchen.area = -5          # Недопустимая площадь
print(kitchen.area)        # 12 — значение не изменилось
```

### Разбор по частям

**`@property` над методом `area`** — говорит Python: «при чтении `room.area` вызывай этот метод **без скобок**».

```python
@property
def area(self):
    return self.__area
```

**`@area.setter` над одноимённым методом** — говорит Python: «при записи `room.area = value` вызывай этот метод».

```python
@area.setter
def area(self, value):
    ...
```

Имя `area` **обязательно совпадает** в обоих случаях. Иначе свойство не свяжется с сеттером.

### Схема работы

```
room.area          →  вызывается геттер (@property)
room.area = 12     →  вызывается сеттер (@area.setter)
```

### Плюсы `@property`

- **Красиво:** `room.area` вместо `room.get_area()`.
- **Работает как обычное поле** — естественно и удобно.
- **Проверка работает** — снаружи не видно, но она есть.
- **Легко превратить в «только для чтения»** — достаточно не писать `.setter`.

### Минусы `@property`

- **Менее явно:** новичок не сразу поймёт, что за `room.area` скрывается метод.
- **Если свойств много** — класс становится длинным.

---

## 2.9. Сравнение двух подходов

Одна и та же задача — защита поля — решается двумя способами.

| Критерий | Классический (`get_`/`set_`) | `@property` |
|----------|------------------------------|-------------|
| Синтаксис чтения | `room.get_area()` | `room.area` |
| Синтаксис записи | `room.set_area(12)` | `room.area = 12` |
| Похоже на поле | ❌ Нет | ✅ Да |
| Явность | ✅ Очень явно | ⚠️ Менее явно |
| Современность | Старый стиль | Современный стиль |
| Количество кода | Больше | Меньше |
| Только для чтения | Через отсутствие сеттера | Через отсутствие `.setter` |

**В современном Python предпочитают `@property`.** Классические `get_`/`set_` встречаются в старом коде и в языках вроде Java.

## 2.10. Как сделать поле «только для чтения»

Иногда нужно, чтобы поле можно было **только читать**, но нельзя изменить. Например, номер технического паспорта комнаты после регистрации менять нельзя.

### Классический способ

Не писать сеттер:

```python
class RoomPassport:
    def __init__(self, passport_id):
        self.__passport_id = passport_id

    def get_passport_id(self):
        return self.__passport_id

    # set_passport_id не определяем — изменить нельзя


passport = RoomPassport("77:01:0001001:1234")
print(passport.get_passport_id())   # 77:01:0001001:1234
# passport.set_passport_id("...")   # AttributeError
```

### Через `@property`

Определить только геттер, без `.setter`:

```python
class RoomPassport:
    def __init__(self, passport_id):
        self.__passport_id = passport_id

    @property
    def passport_id(self):
        return self.__passport_id

    # @passport_id.setter не определяем


passport = RoomPassport("77:01:0001001:1234")
print(passport.passport_id)     # 77:01:0001001:1234

passport.passport_id = "00:00:0000000:0000"
# AttributeError: can't set attribute
```

Попытка присвоить значение вызовет `AttributeError` — это и есть защита «только для чтения».

## 2.11. Сеттер с полной валидацией

Сеттер — идеальное место для проверки данных. Можно не только ограничить диапазон, но и проверить **тип**.

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area       # вызовет сеттер при создании

    @property
    def area(self):
        return self.__area

    @area.setter
    def area(self, value):
        if not isinstance(value, (int, float)):
            raise TypeError("Площадь должна быть числом")
        if not 5 <= value <= 100:
            raise ValueError("Площадь должна быть от 5 до 100 м²")
        self.__area = value
```

Теперь **любое** присваивание — хоть в конструкторе, хоть снаружи — проходит через проверку:

```python
Room("Кухня", 12)      # ок
Room("Кухня", -5)      # ValueError: Площадь должна быть от 5 до 100 м²
Room("Кухня", "12")    # TypeError: Площадь должна быть числом
```

Это одна из главных причин использовать `@property`.

### Важный нюанс

Обрати внимание: в конструкторе мы пишем `self.area = area`, а **не** `self.__area = area`. Это специально — чтобы сработал сеттер и проверил значение.

```python
def __init__(self, name, area):
    self.name = name
    self.area = area       # ← сработает сеттер, проверит
    # self.__area = area   # ← так сеттер НЕ сработает, проверки не будет
```

Если написать `self.__area = area`, значение запишется напрямую, минуя проверку. Тогда можно случайно создать комнату с отрицательной площадью, если передать плохое значение в конструктор.

## 2.12. Когда что выбирать

| Ситуация | Рекомендация |
|----------|--------------|
| Простое публичное поле без проверок | Обычный атрибут (`self.name = name`) |
| Нужна проверка при записи | `@property` + `.setter` |
| Поле только для чтения | Только `@property`, без сеттера |
| Старый проект / обучение | Классические `get_`/`set_` |
| Новый проект | `@property` |
| Много простых полей без проверок | Обычные атрибуты |

| Ошибка | Что не так | Как исправить |

## 2.14. Что важно запомнить

- **Инкапсуляция** — сокрытие данных и доступ через свойства.
- **Три уровня доступа:** `name` (публичный), `_name` (защищённый), `__name` (приватный).
- **Name mangling:** `__name` превращается в `_Class__name`.
- **Приватность в Python** — договорённость, а не жёсткая защита.
- **Два способа доступа к приватным данным:**
  1. **Классический** — методы `get_` / `set_`.
  2. **Современный** — `@property` + `.setter`.
- **`@property` делает код похожим** на работу с обычным полем: `room.area` вместо `room.get_area()`.
- **Без `.setter`** свойство становится **только для чтения**.
- **Сеттер — идеальное место для валидации** данных.
- **В современном Python предпочитают `@property`**.

## Задача 2 (инкапсуляция и свойства)

У каждой комнаты есть технический паспорт с номером, который нельзя менять или читать напрямую.

**Часть 1.** Реализуйте **классический** подход: метод `set_passport_id(value)` — только для установки номера при регистрации, и метод `get_passport_id()` — для чтения. Повторная попытка изменить номер после регистрации должна игнорироваться, а не вызывать ошибку.

**Часть 2.** Реализуйте **второй вариант** того же класса — через `@property`. Сделайте так, чтобы поле `passport_id` было доступно только для чтения (геттер есть, сеттера нет).

**Часть 3.** Продемонстрируйте оба подхода: создайте запись, установите номер, попробуйте изменить его ещё раз. Дополнительно покажите, что снаружи нельзя напрямую прочитать `passport.__passport_id` — будет ошибка `AttributeError`, но через name mangling (`passport._RoomPassport__passport_id`) доступ всё же существует.

В комментариях объясните разницу между классическим подходом и `@property`: в чём плюсы и минусы каждого.

### Решение. Часть 1 — классический подход

```python
class RoomPassport:
    def __init__(self):
        # Приватное поле: снаружи не читается напрямую.
        self.__passport_id = None

    def set_passport_id(self, value):
        # Устанавливаем номер только если он ещё не установлен.
        # Повторная попытка игнорируется без ошибки.
        if self.__passport_id is None:
            self.__passport_id = value

    def get_passport_id(self):
        return self.__passport_id


passport = RoomPassport()
passport.set_passport_id("77:01:0001001:1234")
print("Номер после регистрации:", passport.get_passport_id())

passport.set_passport_id("00:00:0000000:0000")
print("Номер после повторной попытки:", passport.get_passport_id())

# Проверяем, что прямой доступ к приватному полю невозможен.
try:
    print(passport.__passport_id)
except AttributeError as e:
    print("Прямой доступ невозможен:", e)

# Но через name mangling — доступ есть.
print("Через name mangling:", passport._RoomPassport__passport_id)
```

**Ожидаемый вывод:**

```
Номер после регистрации: 77:01:0001001:1234
Номер после повторной попытки: 77:01:0001001:1234
Прямой доступ невозможен: 'RoomPassport' object has no attribute '__passport_id'
Через name mangling: 77:01:0001001:1234
```

### Решение. Часть 2 — через `@property`

```python
class RoomPassport:
    def __init__(self, passport_id):
        # Приватное поле — устанавливается один раз при создании.
        self.__passport_id = passport_id

    @property
    def passport_id(self):
        # Только геттер. Сеттера нет — поле только для чтения.
        return self.__passport_id


passport = RoomPassport("77:01:0001001:1234")
print("Номер паспорта:", passport.passport_id)

# Попытка изменить — вызовет AttributeError.
try:
    passport.passport_id = "00:00:0000000:0000"
except AttributeError as e:
    print("Изменение невозможно:", e)

# Проверяем name mangling.
print("Через name mangling:", passport._RoomPassport__passport_id)
```

**Ожидаемый вывод:**

```
Номер паспорта: 77:01:0001001:1234
Изменение невозможно: can't set attribute 'passport_id'
Через name mangling: 77:01:0001001:1234
```

### Разбор

**Классический подход** (`get_` / `set_`) — **явный**, но **длинный**. Чтение выглядит как вызов метода: `passport.get_passport_id()`. Легко понять новичку: это просто методы. Минус — громоздко, если полей много.

**Подход через `@property`** — **современный** и **красивый**. Чтение выглядит как обращение к полю: `passport.passport_id`. Снаружи кажется, что это обычное поле, но за кулисами работает метод. Если сеттер не определён — поле становится **только для чтения**, и попытка присвоить значение вызовет `AttributeError`.

**Почему повторная попытка в первом варианте игнорируется, а во втором — выбрасывает ошибку?**
В первом варианте мы **сами написали** такую логику: «если номер уже установлен — ничего не делать». Это осознанное решение. Во втором варианте сеттера вообще нет, поэтому Python **сам** не даёт изменить поле — это уже другой механизм.

**Оба подхода решают одну задачу** — защищают приватное поле. В современном Python предпочитают `@property`, потому что код получается чище и естественнее.

# Глава 3. Наследование

## 3.1. Зачем нужно наследование

Вернёмся к нашему дому. У нас есть класс `Room` с полями `name`, `area`, `floor`. Теперь мы хотим описать разные типы комнат: кухню, спальню, ванную. У каждой — свои особенности:

- **Кухня:** есть плита, холодильник.
- **Спальня:** есть кровать, шкаф.
- **Ванная:** есть ванна, раковина.

Первое, что приходит в голову — создать три отдельных класса:

```python
class Kitchen:
    def __init__(self, name, area, floor):
        self.name = name
        self.area = area
        self.floor = floor

    def info(self):
        return f"{self.name} — {self.area} м², {self.floor} этаж"

class Bedroom:
    def __init__(self, name, area, floor):
        self.name = name
        self.area = area
        self.floor = floor

    def info(self):
        return f"{self.name} — {self.area} м², {self.floor} этаж"

class Bathroom:
    def __init__(self, name, area, floor):
        self.name = name
        self.area = area
        self.floor = floor

    def info(self):
        return f"{self.name} — {self.area} м², {self.floor} этаж"
```

Уже здесь видно проблему: **код дублируется**. Конструктор и метод `info` — одинаковые во всех трёх классах. Если завтра нужно будет добавить поле «тип отделки», придётся править в трёх местах.

А теперь представь, что комнат не три, а двадцать. Или что у каждой комнаты есть десять общих методов. Дублирование превращается в катастрофу: любое изменение нужно вносить везде, и легко где-то забыть.

**Наследование решает эту проблему.** Мы описываем общее один раз — в родительском классе, а наследники получают это общее и добавляют своё.

---

## 3.2. Что такое наследование

**Наследование** — это механизм, позволяющий создать новый класс **на основе уже существующего**. Новый класс автоматически получает **все публичные атрибуты и методы** родителя, а также может:

- **добавлять** свои методы и атрибуты,
- **изменять** (переопределять) унаследованные методы.

Простыми словами: наследование — это способ **переиспользовать код** и описывать отношения «**является**».

### Аналогия

Кухня **является** комнатой. У неё есть всё, что есть у комнаты (название, площадь, этаж), плюс свои особенности (плита, холодильник). Поэтому логично, что класс `Kitchen` наследует от класса `Room`.

---

## 3.3. Терминология

В наследовании используются разные названия для одних и тех же понятий:

| Понятие | Синонимы |
|---------|----------|
| **Родительский класс** | Суперкласс, базовый класс (base class), parent class |
| **Дочерний класс** | Наследник, подкласс (subclass), производный класс (derived class), child class |

В этой книге мы будем использовать **родительский** и **дочерний** — так проще.

### Схема

```
        Room  (родительский класс)
        ├── Kitchen   (дочерний класс)
        ├── Bedroom   (дочерний класс)
        └── Bathroom  (дочерний класс)
```

---

## 3.4. Синтаксис наследования

После имени дочернего класса в **круглых скобках** указывается родитель:

```python
class Room:
    def get_type(self):
        return "Комната"

class Kitchen(Room):
    pass
```

Здесь `Kitchen` наследует от `Room`. Запись `class Kitchen(Room)` читается так: «класс `Kitchen` — наследник класса `Room`».

### Что теперь умеет `Kitchen`

Даже если внутри `Kitchen` ничего не написано (`pass`), он **уже умеет** всё, что умеет `Room`:

```python
kitchen = Kitchen()
print(kitchen.get_type())   # Комната — метод унаследован от Room
```

Это ключевая идея: **наследник получает функционал родителя бесплатно**.

---

## 3.5. Что наследуется, а что нет

Не всё наследуется одинаково. Разберём подробно.

| Что | Наследуется? | Пояснение |
|-----|--------------|-----------|
| **Публичные методы** | ✅ Да | `get_type`, `info` и т. д. |
| **Защищённые методы (`_name`)** | ✅ Да | По договорённости доступны наследникам |
| **Приватные методы (`__name`)** | ⚠️ Формально нет | Но доступны через name mangling |
| **Публичные атрибуты класса** | ✅ Да | Атрибуты, определённые на уровне класса |
| **Атрибуты экземпляра** | ❌ Нет | Они создаются в конструкторе родителя; если не вызвать `super().__init__()`, их не будет |
| **Конструктор `__init__`** | ❌ Не наследуется автоматически | Нужно вызывать через `super()` |
| **Магические методы** | ✅ Да | `__str__`, `__eq__` и т. д. |

### Нюанс: приватные атрибуты родителя

Приватные атрибуты родителя (`__name`) **не видны** в наследнике под тем же именем. Это связано с name mangling: `__name` в классе `Room` превращается в `_Room__name`. В классе `Kitchen` тот же `__name` превратится в `_Kitchen__name` — а это **другое поле**.

```python
class Room:
    def __init__(self, name):
        self.__name = name       # превратится в _Room__name

class Kitchen(Room):
    def show(self):
        return self.__name       # превратится в _Kitchen__name — ошибка!
```

Здесь `Kitchen().show()` вызовет `AttributeError`, потому что `_Kitchen__name` не существует — существует `_Room__name`.

**Как обойти:** использовать **защищённые** атрибуты (`_name`) вместо приватных, если поле должно быть доступно наследникам.

```python
class Room:
    def __init__(self, name):
        self._name = name        # защищённое — доступно наследникам

class Kitchen(Room):
    def show(self):
        return self._name        # работает
```

---

## 3.6. Переопределение методов

Часто наследник хочет вести себя **не так, как родитель**. Для этого метод **переопределяют**: в дочернем классе объявляют метод **с тем же именем** — и он **замещает** родительский.

```python
class Room:
    def get_type(self):
        return "Комната"

class Kitchen(Room):
    def get_type(self):
        return "Кухня"

class Bedroom(Room):
    def get_type(self):
        return "Спальня"
```

Теперь:

```python
print(Room().get_type())       # Комната
print(Kitchen().get_type())    # Кухня
print(Bedroom().get_type())    # Спальня
```

Python ищет метод **сначала в самом классе**. Если находит — использует его, а не родительский.

### Как Python ищет метод

Когда вызывается `kitchen.get_type()`, Python:

1. Ищет `get_type` в классе `Kitchen`. **Находит** — использует его.
2. Если бы не нашёл — пошёл бы к родителю `Room`.
3. Если бы не нашёл и там — пошёл бы к `object` (базовый класс всех классов).
4. Если не нашёл нигде — `AttributeError`.

Этот порядок называется **MRO** (Method Resolution Order). Мы разберём его подробно в главе 19.

### Нюанс: переопределение vs перегрузка

| Понятие | Что это | Есть в Python? |
|---------|---------|----------------|
| **Переопределение (override)** | Наследник заменяет метод родителя | ✅ Да |
| **Перегрузка (overload)** | Несколько методов с одним именем, но разными параметрами | ❌ Нет |

Это разные вещи! **Переопределение** — про наследование: два класса, один метод, разные реализации. **Перегрузка** — про параметры: один класс, одно имя, разные сигнатуры. В Python перегрузки нет — она имитируется через значения по умолчанию (см. главу 10).

## 3.7. `super()` — вызов родительской версии

Иногда нужно **не заменить** родительский метод, а **дополнить** его. Например, у комнаты есть метод `info()`, а у кухни хочется **добавить** информацию о плите.

Для этого используют функцию **`super()`**. Она возвращает ссылку на родительский класс, и через неё можно вызвать родительскую версию метода.

```python
class Room:
    def info(self):
        return "Комната"

class Kitchen(Room):
    def info(self):
        parent_info = super().info()       # вызываем версию Room
        return f"{parent_info} (с плитой)"
```

Теперь:

```python
print(Kitchen().info())    # Комната (с плитой)
```

Мы **не переписали** логику родителя, а **взяли её и дополнили**.

### Почему не написать `Room.info(self)`

Можно и так:

```python
class Kitchen(Room):
    def info(self):
        parent_info = Room.info(self)      # работает
        return f"{parent_info} (с плитой)"
```

Но это **плохая практика**. Если завтра класс `Kitchen` станет наследником другого класса (не `Room`), придётся менять код. `super()` **автоматически** находит родителя — код становится гибче.

**Правило:** всегда используйте `super()`, а не прямое обращение к родителю по имени.

## 3.8. `super()` в конструкторе

Это **самый важный** случай использования `super()`. Разберём его подробно.

### Проблема

Представь, у нас есть `Room` с конструктором и `Kitchen`, который тоже имеет конструктор:

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area

class Kitchen(Room):
    def __init__(self, name, area, has_fridge):
        self.has_fridge = has_fridge
```

Что произойдёт при создании?

```python
kitchen = Kitchen("Кухня", 12, True)
print(kitchen.name)     # AttributeError!
```

`kitchen.name` не существует, потому что конструктор `Kitchen` **не вызвал** конструктор `Room`. Атрибуты `name` и `area` так и не были установлены.

### Решение — `super().__init__()`

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area

class Kitchen(Room):
    def __init__(self, name, area, has_fridge):
        super().__init__(name, area)      # ← вызываем родительский конструктор
        self.has_fridge = has_fridge
```

Теперь всё работает:

```python
kitchen = Kitchen("Кухня", 12, True)
print(kitchen.name)         # Кухня
print(kitchen.area)         # 12
print(kitchen.has_fridge)   # True
```

### Схема работы

```
Kitchen("Кухня", 12, True)
    ↓
Kitchen.__init__(self, "Кухня", 12, True)
    ↓
super().__init__("Кухня", 12)   ← вызов Room.__init__
    ↓
Room.__init__(self, "Кухня", 12)
    ↓
self.name = "Кухня"
self.area = 12
    ↓
возврат в Kitchen.__init__
    ↓
self.has_fridge = True
```

### Типичная ошибка

**Забыть `super().__init__()`** — самая частая ошибка при работе с наследованием. Признак: обращение к атрибутам родителя вызывает `AttributeError`.

```python
class Kitchen(Room):
    def __init__(self, name, area, has_fridge):
        # super().__init__(name, area)   ← забыли!
        self.has_fridge = has_fridge

kitchen = Kitchen("Кухня", 12, True)
print(kitchen.name)    # AttributeError: 'Kitchen' object has no attribute 'name'
```

### Нюанс: где вызывать `super().__init__()`

Порядок в конструкторе наследника важен:

1. **Сначала** `super().__init__(...)` — инициализируем родительскую часть.
2. **Потом** — свои атрибуты.

```python
def __init__(self, name, area, has_fridge):
    super().__init__(name, area)   # сначала родитель
    self.has_fridge = has_fridge   # потом свои
```

Если сделать наоборот — работать будет, но логика ломается: сначала «свои» поля, потом «родительские». Это плохо для читаемости.

## 3.9. Множественное наследование

Python поддерживает **множественное наследование** — класс может наследоваться от **нескольких** родителей одновременно.

```python
class Employee:
    def work(self):
        return "Работает"

class Student:
    def study(self):
        return "Учится"

class WorkingStudent(Employee, Student):
    pass
```

Теперь `WorkingStudent` умеет и `work()`, и `study()`:

```python
ws = WorkingStudent()
print(ws.work())     # Работает
print(ws.study())    # Учится
```

### Схема

```
    Employee   Student
         \      /
          \    /
      WorkingStudent
```

### Проблема: конфликты имён

Если **оба** родителя имеют метод с одинаковым именем, возникает вопрос: какой использовать?

```python
class Employee:
    def do(self):
        return "Работает"

class Student:
    def do(self):
        return "Учится"

class WorkingStudent(Employee, Student):
    pass

ws = WorkingStudent()
print(ws.do())    # Работает — метод взят из Employee
```

Python использует **первый** родитель в списке. Это связано с **MRO** — порядком разрешения методов.

### Как посмотреть порядок

У каждого класса есть атрибут `__mro__` (или метод `mro()`):

```python
print(WorkingStudent.__mro__)
# (<class 'WorkingStudent'>, <class 'Employee'>, <class 'Student'>, <class 'object'>)
```

Это означает: сначала ищем в `WorkingStudent`, потом в `Employee`, потом в `Student`, потом в `object`.

### Множественное наследование — осторожно!

Множественное наследование — **мощный, но опасный** инструмент. Оно часто приводит к путанице:

- Какой метод вызван, если он есть у нескольких родителей?
- Как работает `super()` при нескольких родителях?
- Легко получить «ромбовидное» наследование и запутаться.

Поэтому множественное наследование используют **редко** и **осторожно**. Чаще всего — для **миксинов** (см. главу 20). В большинстве случаев **композиция** предпочтительнее (см. главу 8).

---

## 3.10. Наследование vs композиция

Наследование — это отношение «**является**»:
- Кухня **является** комнатой → `class Kitchen(Room)`.
- Спальня **является** комнатой → `class Bedroom(Room)`.

Композиция — это отношение «**содержит**»:
- Дом **содержит** комнаты → `House` имеет поле `rooms`.

### Как выбрать?

| Ситуация | Что использовать |
|----------|------------------|
| «A является B» | Наследование |
| «A содержит B» | Композиция |
| Нужно переиспользовать код | Обычно композиция |
| Есть чёткая иерархия «является» | Наследование |

**Правило:** «**предпочитайте композицию наследованию**». Это один из принципов хорошего дизайна (мы разберём его в главе 28).

### Почему композиция предпочтительнее

- Наследование создаёт **жёсткую связь** между классами: изменение родителя ломает наследника.
- Наследование легко использовать **неправильно**: например, сделать `Car` наследником `Engine`, хотя «машина не является двигателем».
- Композиция гибче: можно заменить внутренний объект без изменения класса.

## 3.11. `isinstance()` и `issubclass()`

Python предоставляет две полезные функции для работы с типами.

### `isinstance(obj, Class)`

Проверяет, является ли объект экземпляром класса (или его наследником):

```python
kitchen = Kitchen("Кухня", 12)

print(isinstance(kitchen, Kitchen))   # True
print(isinstance(kitchen, Room))      # True — Kitchen наследует Room
print(isinstance(kitchen, str))       # False
```

Обрати внимание: `isinstance` возвращает `True` **и для наследников**. `kitchen` — это и `Kitchen`, и `Room`.

### `issubclass(Child, Parent)`

Проверяет, является ли класс наследником другого класса:

```python
print(issubclass(Kitchen, Room))      # True
print(issubclass(Room, Kitchen))      # False — Room не наследник Kitchen
print(issubclass(Kitchen, Kitchen))   # True — класс является наследником самого себя
```

### Зачем это нужно

Эти функции полезны, когда нужно **по-разному обрабатывать объекты разных типов**:

```python
def describe(item):
    if isinstance(item, Kitchen):
        print("Это кухня")
    elif isinstance(item, Room):
        print("Это комната")
```

Мы разберём это подробнее в главе 4 (Полиморфизм).

## 3.12. Типичные ошибки

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Забыли `super().__init__()` | Родительские атрибуты не создаются | Добавить вызов в начало конструктора |
| Вызывают `super().__init__()` **после** своих полей | Работает, но нелогично | Вызывать **первым** |
| Обращаются к `self.__name` в наследнике | Приватное поле родителя недоступно | Использовать `_name` (защищённое) |
| Наследуют ради переиспользования кода | Нарушает логику «является» | Использовать композицию |
| Путают override и overload | Это разные вещи | Override — переопределение; overload — перегрузка |
| Переопределяют метод с другим именем | Это новый метод, а не переопределение | Имя должно совпадать |
| Используют `Room.info(self)` вместо `super()` | Жёсткая связь | Использовать `super()` |
| Злоупотребляют множественным наследованием | Сложно понять, какой метод вызван | Использовать композицию или миксины |

## 3.13. Что важно запомнить

- **Наследование** — создание нового класса на основе существующего.
- **Синтаксис:** `class Child(Parent):`.
- **Наследник получает** все публичные атрибуты и методы родителя.
- **Отношение «является»** — основание для наследования.
- **Переопределение** — своя версия метода родителя (имя совпадает).
- **`super()`** — вызов родительской версии метода.
- **`super().__init__()` обязателен**, если у родителя есть конструктор с параметрами.
- **Приватные поля родителя** недоступны наследнику напрямую.
- **Множественное наследование** — мощно, но опасно.
- **Композиция** часто предпочтительнее наследования.
- **`isinstance()`** — проверка объекта; **`issubclass()`** — проверка класса.

## Задача 3 (наследование)

В доме есть разные типы комнат. Создайте базовый класс `Room` с полями `name` (название) и `area` (площадь), а также методом `get_type()`, возвращающим строку «Комната», и методом `info()`, возвращающим строку вида `Название — площадь ... м²`.

**Часть 1.** Унаследуйте от `Room` два класса:

- `Kitchen` — с дополнительным полем `has_fridge` (есть ли холодильник) и переопределённым методом `get_type()`, возвращающим «Кухня».
- `Bedroom` — с дополнительным полем `bed_size` (размер кровати) и переопределённым методом `get_type()`, возвращающим «Спальня».

**Часть 2.** В каждом наследнике реализуйте конструктор с вызовом `super().__init__(...)`, чтобы родительские поля `name` и `area` правильно инициализировались.

**Часть 3.** Переопределите в `Kitchen` метод `info()` так, чтобы он **дополнял** родительский: возвращал строку вида `Кухня — площадь 12 м² (холодильник есть)`. Используйте `super().info()` для получения родительской строки.

**Часть 4.** Создайте по одному объекту `Kitchen` и `Bedroom`. Выведите их тип через `get_type()` и информацию через `info()`. Проверьте через `isinstance()`, что объект `Kitchen` является и `Kitchen`, и `Room`.

**Часть 5.** В комментариях объясните:

1. Что произойдёт, если в `Kitchen.__init__` не вызвать `super().__init__()`.
2. Чем переопределение отличается от перегрузки.
3. Почему `super().info()` лучше, чем `Room.info(self)`.

### Решение

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area

    def get_type(self):
        return "Комната"

    def info(self):
        return f"{self.name} — площадь {self.area} м²"


class Kitchen(Room):
    def __init__(self, name, area, has_fridge):
        # Вызываем родительский конструктор, чтобы установить name и area.
        # Без этой строки self.name и self.area не будут созданы.
        super().__init__(name, area)
        self.has_fridge = has_fridge

    def get_type(self):
        return "Кухня"

    def info(self):
        # Дополняем родительскую строку, а не переписываем её заново.
        parent_info = super().info()
        fridge = "холодильник есть" if self.has_fridge else "холодильника нет"
        return f"{parent_info} ({fridge})"


class Bedroom(Room):
    def __init__(self, name, area, bed_size):
        super().__init__(name, area)
        self.bed_size = bed_size

    def get_type(self):
        return "Спальня"


# Создаём объекты.
kitchen = Kitchen("Кухня", 12, True)
bedroom = Bedroom("Спальня", 18, "King Size")

# Проверяем типы.
print(kitchen.get_type())    # Кухня
print(bedroom.get_type())    # Спальня

# Проверяем info.
print(kitchen.info())        # Кухня — площадь 12 м² (холодильник есть)
print(bedroom.info())        # Спальня — площадь 18 м²

# Проверяем isinstance.
print(isinstance(kitchen, Kitchen))   # True
print(isinstance(kitchen, Room))      # True — Kitchen наследует Room
```

**Ожидаемый вывод:**

```
Кухня
Спальня
Кухня — площадь 12 м² (холодильник есть)
Спальня — площадь 18 м²
True
True
```

### Разбор

**1. Что произойдёт, если в `Kitchen.__init__` не вызвать `super().__init__()`.**

Атрибуты `name` и `area` не будут созданы. При попытке обратиться к `kitchen.name` Python выдаст `AttributeError: 'Kitchen' object has no attribute 'name'`. Это самая частая ошибка при работе с наследованием.

**2. Чем переопределение отличается от перегрузки.**

**Переопределение (override)** — это когда **наследник** даёт свою версию метода **родителя**. Имя метода совпадает, но реализация другая. Работает благодаря наследованию: два класса, один метод, разные версии.

**Перегрузка (overload)** — это когда **в одном классе** есть несколько методов **с одним именем**, но **разными параметрами**. В Python такого нет: имя метода — это один объект-функция, объявить две функции с одним именем нельзя. Перегрузку имитируют через значения по умолчанию (см. главу 10).

**3. Почему `super().info()` лучше, чем `Room.info(self)`.**

`super()` **автоматически** находит родительский класс. Если завтра иерархия изменится — например, `Kitchen` станет наследником не `Room`, а `LivingRoom` — код с `super()` продолжит работать, а код с `Room.info(self)` придётся переписывать. `super()` делает код **гибче** и **устойчивее к изменениям**.

Кроме того, `super()` корректно работает при **множественном наследовании** — он идёт по MRO. Прямое обращение `Room.info(self)` этого не умеет.

# Глава 4. Полиморфизм

## 4.1. Что такое полиморфизм

**Полиморфизм** — это способность разных объектов **реагировать на один и тот же вызов по-разному**. Слово составное: «поли» — много, «морфе» — форма. Буквально: «много форм».

Простыми словами: если у объектов разных классов есть метод **с одинаковым именем**, то вызов этого метода будет приводить к **разным результатам** — в зависимости от того, объект какого класса вызван.

### Аналогия

Представь, что ты даёшь команду «убраться» разным людям:

- **Повар** поймёт это как «помыть плиту и стол».
- **Горничная** — как «пропылесосить и заправить кровать».
- **Сантехник** — как «почистить плитку и раковину».

Команда одна — «убраться». Но каждый делает **своё**. Это и есть полиморфизм: **один вызов — много форм поведения**.

---

## 4.2. Зачем нужен полиморфизм

Без полиморфизма нам пришлось бы писать **разный код** для каждого типа:

```python
def clean_kitchen(kitchen):
    print("Моем плиту и стол")

def clean_bedroom(bedroom):
    print("Пылесосим и заправляем кровать")

def clean_bathroom(bathroom):
    print("Чистим плитку и сантехнику")
```

А теперь представь, что комнат **десять типов**. Придётся писать десять функций. А если появится одиннадцатый тип — снова новая функция и снова правки во всех местах, где вызывается уборка.

**С полиморфизмом** пишем **одну** функцию:

```python
def clean(room):
    room.clean()
```

И она работает со **всеми** типами комнат. Функции **всё равно**, какая именно комната — кухня, спальня или ванная. Ей важно только, что у комнаты есть метод `clean()`.

Это ключевая идея: **полиморфизм позволяет писать общий код, который работает с разными типами объектов**.

---

## 4.3. Пример из жизни

Вернёмся к нашему дому. У нас есть классы `Kitchen`, `Bedroom`, `Bathroom`. У каждой комнаты — свой способ уборки:

```python
class Kitchen:
    def clean(self):
        return "Моем плиту и стол"

class Bedroom:
    def clean(self):
        return "Пылесосим и заправляем кровать"

class Bathroom:
    def clean(self):
        return "Чистим плитку и сантехнику"
```

Теперь напишем **одну** функцию, которая убирает **любую** комнату:

```python
def clean_room(room):
    print(f"Уборка: {room.clean()}")
```

Вызовем её с объектами разных классов:

```python
clean_room(Kitchen())     # Уборка: Моем плиту и стол
clean_room(Bedroom())     # Уборка: Пылесосим и заправляем кровать
clean_room(Bathroom())    # Уборка: Чистим плитку и сантехнику
```

**Один и тот же вызов** `room.clean()` — **три разных результата**. Это полиморфизм.

---

## 4.4. Как Python определяет, какой метод вызвать

Это **ключевой момент**, который часто упускают. Давай разберём пошагово.

Когда Python видит выражение:

```python
room.clean()
```

Он выполняет следующие действия:

1. **Смотрит на фактический класс объекта** `room`. Если `room` — объект класса `Kitchen`, Python запомнит это.
2. **Ищет метод `clean`** в классе `Kitchen`.
3. **Если находит** — вызывает его.
4. **Если не находит** — идёт к родителям класса (по MRO, см. главу 19).
5. **Если не находит нигде** — `AttributeError`.

Всё решается **во время выполнения**, по фактическому типу объекта. Это называется **динамическое связывание** (dynamic binding) или **позднее связывание** (late binding).

### Схема работы

```
room.clean()
    ↓
Определяем фактический класс room → Kitchen
    ↓
Ищем clean в Kitchen → нашли
    ↓
Вызываем Kitchen.clean()
```

### Почему это важно

Функция `clean_room` **не знает заранее**, какой объект ей передадут. Она просто вызывает `room.clean()`. Python сам разберётся, какой метод использовать. Это делает код **универсальным**.

### Сравнение с другими языками

В **статически типизированных** языках (Java, C++) метод выбирается **во время компиляции** или через **виртуальные таблицы**. В Python — **во время выполнения**. Это даёт гибкость, но цена — ошибки видны только в момент вызова.

## 4.5. Полиморфизм через наследование

Первый способ полиморфизма — через **наследование**. Все классы наследуют общий родитель с методом:

```python
class Room:
    def clean(self):
        return "Убираем комнату"

class Kitchen(Room):
    def clean(self):
        return "Моем плиту и стол"

class Bedroom(Room):
    def clean(self):
        return "Пылесосим и заправляем кровать"
```

Здесь `Kitchen` и `Bedroom` — наследники `Room`. Они **переопределяют** метод `clean`. Функция `clean_room(room)` работает с любым из них.

### Схема

```
        Room
        ├── Kitchen   (переопределяет clean)
        └── Bedroom   (переопределяет clean)
```

### Плюсы

- **Явная иерархия** — видно, что все комнаты «являются» комнатами.
- **Общий интерфейс** — родитель гарантирует, что метод есть у всех.
- **Работает с `isinstance`** — можно проверить тип.

### Минусы

- **Жёсткая связь** — все классы должны наследовать общего родителя.
- **Ограничение** — нельзя добавить «чужой» класс, который не наследует `Room`.

---

## 4.6. Полиморфизм через утиную типизацию

Python поддерживает **второй способ** полиморфизма — **без наследования**. Достаточно, чтобы у объекта был **нужный метод**.

```python
class Kitchen:
    def clean(self):
        return "Моем плиту и стол"

class Bedroom:
    def clean(self):
        return "Пылесосим и заправляем кровать"

class Robot:
    def clean(self):
        return "Робот-пылесос моет пол"
```

Все три класса — **не связаны** друг с другом. Никакого общего родителя. Но функция работает со всеми:

```python
def clean_room(room):
    print(room.clean())

clean_room(Kitchen())    # Моем плиту и стол
clean_room(Bedroom())    # Пылесосим и заправляем кровать
clean_room(Robot())      # Робот-пылесос моет пол
```

### Утиная типизация — что это

Принцип звучит так:

> **«Если это выглядит как утка, плавает как утка и крякает как утка — это утка».**

В Python это означает: **неважно, какого класса объект**. Важно, что у него есть **нужные методы**. Если у объекта есть `clean()` — его можно убирать, независимо от того, кто он.

### Название

Название пришло из английского: **duck typing**. Принцип приписывают Джеймсу Уиткомбу Райли, но популяризировал его Python-мир.

### Плюсы утиной типизации

- **Гибкость:** не нужно наследовать общий родитель.
- **Естественность:** не нужно создавать искусственные иерархии.
- **Лёгкость:** любой класс с нужным методом подходит.

### Минусы утиной типизации

- **Ошибка видна только в рантайме.** Если у объекта нет метода, программа упадёт **в момент вызова**.
- **IDE не подскажет.** Автодополнение не знает, какие методы есть у объекта.
- **Нет гарантии.** Нельзя проверить заранее, что объект «правильный».

### Утиная типизация vs наследование

| Критерий | Через наследование | Через утиную типизацию |
|----------|-------------------|------------------------|
| Нужен общий родитель | ✅ Да | ❌ Нет |
| Явная иерархия | ✅ Есть | ❌ Нет |
| Гибкость | ⚠️ Ограничена | ✅ Высокая |
| Ошибка видна | Раньше | Только в рантайме |
| IDE подсказывает | ✅ Да | ❌ Нет |

**В Python предпочитают утиную типизацию.** Это одна из ключевых философских идей языка.

## 4.7. Полиморфизм через абстрактные классы

Смешанный подход — **абстрактный класс** (см. главу 7). Он задаёт **общий интерфейс** и требует от наследников его реализации.

```python
from abc import ABC, abstractmethod

class HomeItem(ABC):
    @abstractmethod
    def clean(self):
        pass

class Kitchen(HomeItem):
    def clean(self):
        return "Моем плиту и стол"

class Bedroom(HomeItem):
    def clean(self):
        return "Пылесосим и заправляем кровать"
```

**Преимущество:** гарантия, что у всех наследников **точно есть** метод `clean`. Python не даст создать класс-наследник без реализации.

**Недостаток:** работает только с наследниками — «чужие» классы не подойдут.

### Когда использовать

- Если нужна **гарантия** интерфейса — используйте абстрактный класс.
- Если нужна **гибкость** — используйте утиную типизацию.
- В Python чаще комбинируют: абстрактный класс задаёт «официальный» интерфейс, а утиная типизация позволяет добавлять «неофициальные» классы.

## 4.8. Полиморфизм в действии: список объектов

Одна из самых частых ситуаций — **список объектов разных типов**, с которыми нужно работать одинаково:

```python
rooms = [
    Kitchen("Кухня"),
    Bedroom("Спальня"),
    Bathroom("Ванная"),
    Kitchen("Вторая кухня"),
]

for room in rooms:
    print(f"{room.name}: {room.clean()}")
```

**Ожидаемый вывод:**

```
Кухня: Моем плиту и стол
Спальня: Пылесосим и заправляем кровать
Ванная: Чистим плитку и сантехнику
Вторая кухня: Моем плиту и стол
```

Цикл **один**, а поведение — **разное** для каждого объекта. Это полиморфизм в действии.

### Почему это удобно

Без полиморфизма пришлось бы писать:

```python
for room in rooms:
    if isinstance(room, Kitchen):
        print(room.clean_kitchen())
    elif isinstance(room, Bedroom):
        print(room.clean_bedroom())
    elif isinstance(room, Bathroom):
        print(room.clean_bathroom())
```

Много кода, много условий. Полиморфизм убирает всё это: **один вызов — работает для всех**.

## 4.9. `isinstance()` и полиморфизм

Иногда всё же нужно **проверить тип** объекта. Для этого есть `isinstance()`:

```python
def describe(room):
    if isinstance(room, Kitchen):
        print("Это кухня — здесь готовят")
    elif isinstance(room, Bedroom):
        print("Это спальня — здесь спят")
    else:
        print("Просто комната")
```

### Но осторожно!

**`isinstance` — это антиполиморфизм.** Если вы используете `isinstance` в большом количестве мест, скорее всего, вы **делаете что-то не так**. Полиморфизм существует именно для того, чтобы **избегать** таких проверок.

### Плохой пример

```python
def clean(room):
    if isinstance(room, Kitchen):
        print("Моем плиту и стол")
    elif isinstance(room, Bedroom):
        print("Пылесосим и заправляем кровать")
    elif isinstance(room, Bathroom):
        print("Чистим плитку и сантехнику")
```

Здесь **нарушен полиморфизм**. Если добавится новый тип комнаты, придётся править эту функцию.

### Хороший пример

```python
def clean(room):
    room.clean()      # каждый объект знает, как убирать себя
```

Функция **не проверяет типы** — она просто вызывает метод. Новый тип комнаты — ничего менять не нужно.

**Правило:** `isinstance` — используйте **редко**, когда действительно нужно разное поведение **вне класса**. Обычно полиморфизм решает эту задачу лучше.

## 4.10. Полиморфизм и `str()` / `len()` / `print()`

Полиморфизм — это **не только про ваши классы**. Встроенные функции Python тоже полиморфны:

```python
print(len("hello"))      # 5 — длина строки
print(len([1, 2, 3]))    # 3 — длина списка
print(len({"a": 1}))     # 1 — длина словаря
```

`len()` работает с **разными типами** — потому что у каждого из них есть метод `__len__`. Это тоже полиморфизм.

То же самое с `str()`:

```python
print(str(42))           # "42"
print(str(3.14))         # "3.14"
print(str([1, 2, 3]))    # "[1, 2, 3]"
```

Каждый тип реализует `__str__` по-своему — `str()` просто вызывает его. Это **полиморфизм через магические методы**.

### Пример из нашего дома

Если у `Room` определить метод `__str__`:

```python
class Room:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Комната: {self.name}"

print(Room("Кухня"))    # Комната: Кухня
```

`print()` работает с любым объектом, у которого есть `__str__`. Это тоже полиморфизм.

## 4.11. Полиморфизм и `+`

Оператор `+` тоже полиморфен:

```python
print(2 + 3)              # 5 — сложение чисел
print("a" + "b")          # "ab" — конкатенация строк
print([1, 2] + [3, 4])    # [1, 2, 3, 4] — объединение списков
```

Один оператор — **разное поведение**. Это реализовано через магические методы:

- `int.__add__` — арифметическое сложение.
- `str.__add__` — конкатенация.
- `list.__add__` — объединение списков.

### Пример из нашего дома

Если у `Room` определить `__add__`, можно будет складывать комнаты:

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.area = area

    def __add__(self, other):
        return Room(f"{self.name}+{other.name}", self.area + other.area)

combined = Room("Кухня", 12) + Room("Спальня", 18)
print(combined.name)    # Кухня+Спальня
print(combined.area)    # 30
```

Мы разберём магические методы подробно в главах 11–13.

## 4.12. Полиморфизм vs наследование vs инкапсуляция

Эти три понятия часто путают. Давай разберёмся.

| Понятие | Что это | Отвечает на вопрос |
|---------|---------|-------------------|
| **Инкапсуляция** | Сокрытие данных и доступ через методы | «Как защитить данные?» |
| **Наследование** | Создание класса на основе другого | «Как переиспользовать код?» |
| **Полиморфизм** | Один вызов — разное поведение | «Как работать с разными типами одинаково?» |

### Как они связаны

- **Инкапсуляция** позволяет объекту **самому** управлять своим состоянием.
- **Наследование** позволяет **переиспользовать** код и создавать иерархии.
- **Полиморфизм** позволяет **писать общий код** для разных типов.

Вместе они образуют **три кита ООП**.

### Пример: три кита в одной задаче

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self.__area = area      # инкапсуляция

    @property
    def area(self):
        return self.__area

class Kitchen(Room):            # наследование
    def clean(self):
        return "Моем плиту и стол"

class Bedroom(Room):            # наследование
    def clean(self):
        return "Пылесосим и заправляем кровать"

def clean_room(room):           # полиморфизм
    print(room.clean())
```

## 4.14. Типичные ошибки

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Думают, что полиморфизм требует наследования | В Python — нет, достаточно метода | Использовать утиную типизацию |
| Забывают, что ошибка проявляется в рантайме | Программа падает при вызове | Использовать тесты, аннотации типов |
| Путают полиморфизм и перегрузку | Это разные вещи | Полиморфизм — про классы, перегрузка — про параметры |
| Пишут `isinstance` вместо полиморфизма | Нарушает идею ООП | Использовать методы объектов |
| Вызывают метод, которого нет у объекта | `AttributeError` в рантайме | Проверять или использовать `hasattr` |
| Думают, что все методы должны быть в родителе | Родитель задаёт общий интерфейс | Оставлять только общее |
| Называют метод по-разному в разных классах | Не работает полиморфизм | Имя метода должно совпадать |

## 4.15. Что важно запомнить

- **Полиморфизм** — один вызов, разное поведение.
- **Python определяет метод по фактическому классу** объекта (динамическое связывание).
- **Два способа полиморфизма:**
  1. **Через наследование** — все классы наследуют общий родитель.
  2. **Через утиную типизацию** — достаточно, чтобы был нужный метод.
- **Утиная типизация:** «Если это выглядит как утка и крякает как утка — это утка».
- **В Python предпочитают утиную типизацию** — она гибче.
- **Ошибка видна только в рантайме** — используйте тесты.
- **Полиморфизм работает и во встроенных функциях** — `len()`, `str()`, `print()`, `+`.
- **`isinstance` — антиполиморфизм.** Используйте редко.
- **Полиморфизм позволяет писать общие циклы** для разных типов.

## Задача 4 (полиморфизм)

Дом убирают по-разному в зависимости от типа комнаты.

**Часть 1.** Создайте три класса — `Kitchen`, `Bedroom` и `Bathroom`, у каждого из которых есть метод `clean()`, возвращающий строку с описанием уборки:

- `Kitchen` → «Моем плиту и стол»
- `Bedroom` → «Пылесосим и заправляем кровать»
- `Bathroom` → «Чистим плитку и сантехнику»

Классы **не должны наследовать** общего родителя — это пример утиной типизации.

**Часть 2.** Напишите функцию `show_cleaning(room)`, которая принимает любую комнату и выводит описание уборки в формате: `Уборка: <описание>`.

**Часть 3.** Создайте список из **четырёх** объектов (например, две кухни, спальня, ванная) и переберите его в цикле `for`, вызывая `show_cleaning` для каждого.

**Часть 4.** Дополнительно создайте класс `Robot` с методом `clean()`, возвращающим «Робот-пылесос моет пол». Убедитесь, что функция `show_cleaning` работает и с ним — **без изменений в коде функции**.

**Часть 5.** В комментариях объясните:

1. Как Python определяет, какой именно метод вызвать.
2. Что такое утиная типизация.
3. Почему ошибка в утиной типизации видна только во время выполнения.
4. Почему `isinstance` — антиполиморфизм.

### Решение

```python
class Kitchen:
    def clean(self):
        return "Моем плиту и стол"


class Bedroom:
    def clean(self):
        return "Пылесосим и заправляем кровать"


class Bathroom:
    def clean(self):
        return "Чистим плитку и сантехнику"


class Robot:
    def clean(self):
        return "Робот-пылесос моет пол"


def show_cleaning(room):
    # Функции всё равно, какого класса объект.
    # Ей важно только, что у объекта есть метод clean().
    print(f"Уборка: {room.clean()}")


# Создаём список объектов разных типов.
rooms = [
    Kitchen(),
    Bedroom(),
    Bathroom(),
    Kitchen(),   # вторая кухня
    Robot(),     # робот — без изменений в функции
]

for room in rooms:
    show_cleaning(room)
```

**Ожидаемый вывод:**

```
Уборка: Моем плиту и стол
Уборка: Пылесосим и заправляем кровать
Уборка: Чистим плитку и сантехнику
Уборка: Моем плиту и стол
Уборка: Робот-пылесос моет пол
```

### Разбор

**1. Как Python определяет, какой именно метод вызвать.**

Когда Python видит `room.clean()`, он смотрит на **фактический класс** объекта `room`. Если это `Kitchen` — ищет метод `clean` в классе `Kitchen`. Если находит — вызывает. Если нет — идёт к родителям (по MRO). Всё решается **во время выполнения**, по фактическому типу объекта. Это называется **динамическое связывание**.

**2. Что такое утиная типизация.**

Утиная типизация — принцип, по которому **важен не класс объекта, а его методы**. «Если это выглядит как утка, плавает как утка и крякает как утка — это утка». В нашем примере классы `Kitchen`, `Bedroom`, `Bathroom` и `Robot` **не связаны** друг с другом. Но у всех есть метод `clean()` — и этого достаточно.

**3. Почему ошибка в утиной типизации видна только во время выполнения.**

Python — **динамически типизированный** язык. Он не проверяет заранее, есть ли у объекта метод `clean()`. Проверка происходит **в момент вызова**. Если у объекта нет метода — Python выдаст `AttributeError` **только тогда**, когда дойдёт до этой строки. Это и плюс (гибкость), и минус (ошибки всплывают позже). Поэтому используют **тесты** и **аннотации типов**.

**4. Почему `isinstance` — антиполиморфизм.**

`isinstance` проверяет **тип** объекта. Если писать код так:

```python
def show_cleaning(room):
    if isinstance(room, Kitchen):
        print("Моем плиту и стол")
    elif isinstance(room, Bedroom):
        print("Пылесосим и заправляем кровать")
    elif isinstance(room, Robot):
        print("Робот-пылесос моет пол")
```

то при добавлении нового типа комнаты придётся **править функцию**. Это **нарушает полиморфизм** — вместо того чтобы объект сам знал, как убирать себя, мы заставляем функцию это решать.

**Правильно:** функция вызывает `room.clean()` и **не знает**, какой это класс. Каждый объект сам знает, как убирать **себя**. Это и есть полиморфизм.

# Глава 5. Модификаторы доступа

## 5.1. Что такое модификаторы доступа

**Модификатор доступа** — это способ указать, **кто и откуда** может обращаться к атрибуту или методу класса. В разных языках это ключевые слова: `public`, `private`, `protected`. В Python таких ключевых слов **нет** — вместо них используется **соглашение об именах**: количество подчёркиваний перед именем.

### Зачем нужны модификаторы доступа

Когда класс большой, важно **разделить** то, что является **публичным интерфейсом**, и то, что является **внутренней кухней**:

- **Публичное** — то, чем могут пользоваться все: `room.info()`, `room.name`.
- **Внутреннее** — то, что нужно только самому классу: кэш, счётчики, служебные поля.

Если всё сделать публичным, программисты начнут использовать **внутренние** методы и поля снаружи. А потом, при рефакторинге, изменить их будет невозможно — весь чужой код сломается.

**Модификаторы доступа помогают** чётко обозначить: «это публичное, используй», «это внутреннее, не трогай».

### Аналогия

Представь автомобиль:

- **Публичное:** руль, педали, кнопки — то, чем пользуется водитель.
- **Защищённое:** капот — можно открыть, но обычно этим занимается механик.
- **Приватное:** электроника двигателя — снаружи не достать.

Ты не открываешь капот каждый раз, чтобы поехать. Но если нужно — механик может. И только механик знает, что под капотом.

---

## 5.2. Три уровня доступа в Python

В Python есть **три уровня доступа**, которые различаются **количеством подчёркиваний**:

| Уровень | Синтаксис | Доступ | Идея |
|---------|-----------|--------|------|
| **Публичный** | `name` | Везде | «Это часть интерфейса» |
| **Защищённый** | `_name` | Внутри класса и у наследников | «Это внутреннее, но наследникам пригодится» |
| **Приватный** | `__name` | Только внутри класса | «Это только моё» |

### Важно понять сразу

**В Python нет настоящих модификаторов доступа.** Всё это — **соглашения**, а не запреты. Приватные поля можно обойти через name mangling (см. 5.5). Защищённые — и так доступны. Публичные — тем более.

Программисты договорились: **если видишь подчёркивания — не трогай**. Это работает, потому что все уважают соглашение. Если кто-то его нарушает — он либо знает, что делает, либо плохой программист.

## 5.3. Публичные атрибуты и методы

**Публичные** — без подчёркиваний. Доступны откуда угодно.

### Атрибуты

```python
class Room:
    def __init__(self, name, area):
        self.name = name    # публичный атрибут
        self.area = area    # публичный атрибут
```

```python
kitchen = Room("Кухня", 12)
print(kitchen.name)         # Кухня — работает
kitchen.name = "Спальня"    # работает
```

### Методы

```python
class Room:
    def info(self):         # публичный метод
        return "Комната"
```

```python
kitchen.info()              # работает
```

### Когда использовать

- **Имя, площадь, этаж** — публичные: эти данные видны всем.
- **Метод `info()`** — публичный: это часть интерфейса класса.
- Всё, что вы **готовы поддерживать** в будущем — публичное.

### Правило

**По умолчанию всё делайте публичным.** Переводите в защищённые или приватные только то, что действительно **внутреннее**.

Многие новички делают **всё приватным** — «на всякий случай». Это ошибка: код становится неудобным, тестирование усложняется, а защиты никакой (в Python её нет).

## 5.4. Защищённые атрибуты и методы

**Защищённые** — с **одним** подчёркиванием. Это **сигнал**: «это внутреннее, не трогай без нужды».

### Атрибуты

```python
class Room:
    def __init__(self, name, area):
        self.name = name       # публичный
        self._area = area      # защищённый
```

```python
kitchen = Room("Кухня", 12)
print(kitchen._area)           # 12 — работает, но не рекомендуется
```

### Методы

```python
class Room:
    def _validate_area(self, value):   # защищённый метод
        return 5 <= value <= 100
```

### Когда использовать

- **Поле, нужное наследникам.** Например, `_area` — потому что класс `Kitchen` может захотеть работать с площадью.
- **Внутренний метод, который может пригодиться в наследнике.** Например, `_validate_area`.
- **Служебные поля**, которые не должны попадать в публичный интерфейс, но и не являются строго приватными.

### Как понять, что это защищённое

Если поле нужно **только вам и вашим наследникам** — делайте его защищённым.

### Ключевое отличие от приватного

| Защищённое (`_name`) | Приватное (`__name`) |
|----------------------|----------------------|
| Доступно наследникам | Недоступно наследникам (напрямую) |
| Можно обратиться снаружи | Нельзя обратиться снаружи |
| Не искажается | Искажается (name mangling) |

### Пример: зачем нужно защищённое

```python
class Room:
    def __init__(self, name, area):
        self.name = name
        self._area = area       # защищённое — нужно наследникам

class Kitchen(Room):
    def describe(self):
        return f"Кухня площадью {self._area} м²"    # работает!
```

Если бы `_area` было `__area`, в наследнике `Kitchen` оно было бы **недоступно** (см. 5.5).

## 5.5. Приватные атрибуты и методы

**Приватные** — с **двумя** подчёркиваниями. Python **переименовывает** их, и снаружи обратиться напрямую нельзя.

### Атрибуты

```python
class Room:
    def __init__(self, name, area):
        self.name = name        # публичный
        self.__area = area      # приватный
```

```python
kitchen = Room("Кухня", 12)
print(kitchen.__area)
# AttributeError: 'Room' object has no attribute '__area'
```

### Методы

```python
class Room:
    def __secret_method(self):    # приватный метод
        return "Только для внутреннего использования"
```

Снаружи вызвать нельзя:

```python
kitchen.__secret_method()    # AttributeError
```

### Когда использовать

- **Поле, которое не должно меняться извне и не нужно наследникам.** Например, внутренний кэш.
- **Внутренний служебный метод.** Например, `__validate` — не должен использоваться никем, кроме самого класса.
- **Данные, которые чувствительны** — например, пароль, доход, служебная информация.

### Ошибка: делать всё приватным

Новички часто делают **все** поля приватными. Это плохо:

- Снаружи приходится писать геттеры/сеттеры для всего.
- Код усложняется без пользы.
- Наследники не могут получить доступ к нужным полям.

**Правило:** делайте приватным **только то, что действительно внутреннее**.

## 5.6. Name mangling — подробно

Это **самая важная часть** темы. Разберём детально.

### Что происходит

Когда Python видит атрибут с двумя подчёркиваниями в начале — например, `__area` **внутри класса `Room`** — он автоматически **переименовывает** его в `_Room__area`.

Это называется **name mangling** («искажение имени»).

### Пример

```python
class Room:
    def __init__(self, area):
        self.__area = area      # станет self._Room__area
```

На самом деле в объекте лежит поле `_Room__area`:

```python
kitchen = Room(12)
print(kitchen.__dict__)
# {'_Room__area': 12}
```

### Почему нельзя обратиться снаружи

Снаружи вы пишете `kitchen.__area`, но Python **не ищет** поле с таким именем — он ищет `__area` **как есть** (без mangling). А реальное имя — `_Room__area`. Поэтому:

```python
kitchen.__area          # AttributeError
kitchen._Room__area     # работает — это и есть настоящее имя
```

### Внутри класса работает

Внутри класса Python **автоматически** заменяет `self.__area` на `self._Room__area`:

```python
class Room:
    def __init__(self, area):
        self.__area = area         # → self._Room__area

    def get_area(self):
        return self.__area         # → self._Room__area
```

Оба имени внутри класса превращаются в одно. Поэтому внутри класса всё работает.

### Зачем нужен name mangling

**Не для безопасности.** Python не защищает данные — он просто переименовывает. Настоящая цель — **две задачи**:

**1. Избежать конфликтов имён при наследовании.**

```python
class Room:
    def __init__(self):
        self.__x = "Room"

class Kitchen(Room):
    def __init__(self):
        super().__init__()
        self.__x = "Kitchen"
```

Что получится? Два **разных** поля: `_Room__x` и `_Kitchen__x`. Без name mangling они бы перезаписывали друг друга. С ним — существуют независимо.

**2. Сигнализировать другим программистам: «это внутреннее».**

Когда видишь `__area` — понимаешь, что лезть туда не надо.

### Ключевая мысль

**Приватность в Python — договорённость.** Технически доступ есть — через `_Class__name`. Но программисты договорились: **не трогаем приватные поля снаружи**. Это работает, потому что все согласны.

## 5.7. Приватные и защищённые методы в наследниках

Это **тонкий момент**, который часто путают.

### Приватные атрибуты и наследники

**Приватные атрибуты родителя недоступны наследнику напрямую** — из-за name mangling:

```python
class Room:
    def __init__(self, area):
        self.__area = area       # → _Room__area

class Kitchen(Room):
    def describe(self):
        return self.__area       # → _Kitchen__area — AttributeError!
```

Внутри `Kitchen` Python ищет `_Kitchen__area`, а не `_Room__area`. Такого поля нет — `AttributeError`.

**Решение:** если поле должно быть доступно наследникам — делайте его **защищённым** (`_area`), а не приватным.

### Защищённые атрибуты и наследники

Защищённые атрибуты **доступны** наследникам напрямую:

```python
class Room:
    def __init__(self, area):
        self._area = area        # защищённое

class Kitchen(Room):
    def describe(self):
        return f"Кухня {self._area} м²"     # работает
```

Это ещё одна причина использовать `_name`, если поле нужно наследникам.

### Приватные методы и наследники

Аналогично: приватные методы родителя недоступны наследнику.

```python
class Room:
    def __validate(self):
        return True

class Kitchen(Room):
    def check(self):
        return self.__validate()     # → _Kitchen__validate — AttributeError
```

### Схема

| Что | Наследник видит? |
|-----|------------------|
| `name` | ✅ Да |
| `_name` | ✅ Да |
| `__name` | ❌ Нет (искажается под имя наследника) |

## 5.8. Обращение к приватному полю через name mangling

Иногда **действительно нужно** получить доступ к приватному полю — например, при отладке или тестировании. Python **позволяет** это сделать через `_Class__name`.

```python
class Room:
    def __init__(self, area):
        self.__area = area

kitchen = Room(12)

# Снаружи обычный доступ — не работает.
try:
    print(kitchen.__area)
except AttributeError as e:
    print("Ошибка:", e)

# Через name mangling — работает.
print(kitchen._Room__area)    # 12
```

### Важно

Это **технически возможно**, но **плохая практика**. Если вы обращаетесь к `_Class__name` снаружи — вы **нарушаете соглашение**. Делать это можно только в исключительных случаях:

- **отладка** — когда нужно посмотреть, что внутри;
- **тесты** — когда нужно проверить внутреннее состояние;
- **крайние меры** — когда другого способа нет.

В обычном коде **так писать не надо**.

## 5.9. Атрибуты класса тоже имеют уровни доступа

Все три уровня работают и для атрибутов **класса** (не только экземпляра).

```python
class Room:
    type = "Комната"          # публичный
    _default_area = 20        # защищённый
    __version = "1.0"         # приватный
```

```python
print(Room.type)              # Комната — работает
print(Room._default_area)     # 20 — работает, но не рекомендуется
print(Room.__version)         # AttributeError
print(Room._Room__version)    # 1.0 — через name mangling
```

### Когда использовать

- **Публичный** — общая константа для всех объектов: `type = "Комната"`.
- **Защищённый** — значение по умолчанию, нужное наследникам: `_default_area = 20`.
- **Приватный** — служебная версия класса, не нужная никому снаружи.

## 5.10. Уровни доступа в реальной жизни

Когда какой уровень использовать — вопрос **дизайна класса**. Приведём несколько примеров.

### Пример 1: класс `Room`

```python
class Room:
    def __init__(self, name, area):
        self.name = name           # публичный — все видят название
        self._area = area          # защищённый — нужно наследникам
        self.__passport = None     # приватный — только сам класс

    def info(self):                # публичный метод
        return f"{self.name} ({self._area} м²)"

    def _validate(self):           # защищённый метод
        return self._area > 0

    def __register(self):          # приватный метод
        self.__passport = "OK"
```

**Почему так:**

- `name` — публичное: название комнаты видно всем.
- `_area` — защищённое: наследник `Kitchen` может захотеть работать с площадью.
- `__passport` — приватное: служебное поле, никому снаружи не нужное.
- `info` — публичный метод: часть интерфейса.
- `_validate` — защищённый: наследник может переопределить проверку.
- `__register` — приватный: внутренний метод, не должен вызываться никем другим.

### Пример 2: класс `Owner`

```python
class Owner:
    def __init__(self, name, income):
        self.name = name           # публичный
        self.__income = income     # приватный — доход не должен быть виден

    @property
    def income(self):              # публичное свойство для чтения
        return self.__income
```

Здесь:

- `name` — публичное: имя владельца может видеть любой.
- `__income` — приватное: доход скрыт.
- `income` — свойство для чтения: можно узнать доход, но не изменить.

## 5.11. Разница между `_name`, `__name` и `__name__`

Тут есть **важный нюанс**, который путает новичков. Три разных случая:

| Синтаксис | Название | Что это |
|-----------|----------|---------|
| `name` | Публичный | Обычный атрибут |
| `_name` | Защищённый | Соглашение «не трогай» |
| `__name` | Приватный | Name mangling — искажается в `_Class__name` |
| `__name__` | **Магический** | **НЕ приватный!** Специальные методы Python |

### Магические методы

Два подчёркивания **с двух сторон** (`__init__`, `__str__`, `__eq__`) — это **магические методы**, а не приватные. Они **не** искажаются, и их **можно** вызывать снаружи:

```python
class Room:
    def __init__(self, name):      # магический метод
        self.name = name

    def __str__(self):             # магический метод
        return f"Room({self.name})"

kitchen = Room("Кухня")
print(str(kitchen))                # работает
print(kitchen.__str__())           # тоже работает — магические методы доступны
```

**Не путайте!** `__name` (без завершающих подчёркиваний) — приватный. `__name__` (с завершающими) — магический, публичный.

### Правило именования

- `name` — публичный.
- `_name` — защищённый.
- `__name` — приватный (Python искажает).
- `__name__` — магический метод Python (не трогайте без причины).
- `name_` — если имя конфликтует с ключевым словом: `class_`.

## 5.12. Модификаторы доступа в других языках (для сравнения)

Чтобы понять, **чем Python отличается**, сравним с другими языками.

| Язык | Публичный | Защищённый | Приватный |
|------|-----------|-----------|-----------|
| **Python** | `name` | `_name` (соглашение) | `__name` (mangling) |
| **Java** | `public` | `protected` | `private` |
| **C++** | `public:` | `protected:` | `private:` |
| **C#** | `public` | `protected` | `private` |

### Ключевые отличия Python

- **Нет ключевых слов** — только соглашение об именах.
- **Нет настоящей защиты** — приватность можно обойти через name mangling.
- **Всё основано на договорённости** — программисты уважают соглашение.

### Почему так

Python — **динамический язык**. Он верит, что программист **взрослый человек** и сам знает, что можно, а что нет. Не нужно жёстких ограничений: если ты видишь `_name` — не трогай, и всё будет хорошо.

**Философия Python:** «Мы все здесь взрослые люди».

## 5.13. Типичные ошибки

| Ошибка | Что не так | Как исправить |
|--------|-----------|---------------|
| Делают всё приватным | Код становится неудобным, защиты нет | Публичное по умолчанию, приватное — только когда нужно |
| Думают, что `__name` — настоящая защита | Можно обойти через mangling | Помнить: это соглашение |
| Используют `_name` как «приватный» | Это защищённый, доступ открыт | Различать уровни |
| Обращаются к `_Class__name` снаружи | Нарушает соглашение | Не делать этого в обычном коде |
| Обращаются к `__name` в наследнике | Приватное поле родителя недоступно | Использовать `_name` |
| Путают `__name` и `__name__` | Разные вещи | Первое — приватное, второе — магическое |
| Делают публичными внутренние поля | Их начнут использовать снаружи | Скрывать внутреннее |
| Называют метод `_info` вместо `info` | Лишнее подчёркивание | Публичный метод — без подчёркиваний |

## 5.14. Что важно запомнить

- **В Python нет настоящих модификаторов доступа** — только соглашения.
- **Публичный** (`name`) — доступен везде.
- **Защищённый** (`_name`) — доступен внутри класса и наследникам; снаружи можно, но не рекомендуется.
- **Приватный** (`__name`) — доступен только внутри класса; снаружи нельзя напрямую.
- **Name mangling:** `__name` превращается в `_Class__name`.
- **Приватность в Python — договорённость**, а не жёсткая защита.
- **Наследник не видит приватные поля родителя** — используйте `_name`, если поле нужно наследникам.
- **`__name__` (магические методы) — не приватные!** Это специальные методы Python.
- **По умолчанию всё делайте публичным.** Скрывайте только внутреннее.
- **Обращение к `_Class__name` снаружи** — плохая практика.
- **Философия Python:** «Мы все здесь взрослые люди».

## Задача 5 (модификаторы доступа)

Создайте класс `Owner` (владелец дома), в котором есть три поля, демонстрирующие три уровня доступа:

- `name` — **публичное** (имя владельца).
- `_house` — **защищённое** (название дома, доступное наследникам).
- `__income` — **приватное** (ежемесячный доход, только для самого класса).

**Часть 1.** Реализуйте конструктор, который принимает `name`, `house`, `income` и сохраняет их в соответствующие поля.

**Часть 2.** Добавьте публичный метод `info()`, возвращающий строку с именем владельца и названием дома. Не включайте доход — он приватный.

**Часть 3.** Создайте объект `owner` и продемонстрируйте доступ к каждому полю:

- к `name` — свободно;
- к `_house` — можно, но с комментарием «не рекомендуется»;
- к `__income` — попробуйте обратиться напрямую, поймайте `AttributeError` и выведите сообщение;
- затем получите доступ к доходу через name mangling: `owner._Owner__income` — с комментарием «так делать не нужно, но технически возможно».

**Часть 4.** Создайте наследника `Tenant` (арендатор), который наследует от `Owner`. В нём определите метод `describe()`, возвращающий название дома через `_house`. Убедитесь, что защищённое поле **доступно** наследнику.

**Часть 5.** В комментариях объясните:

1. Зачем нужны три уровня доступа.
2. Почему `__name` не защищает по-настоящему.
3. Почему наследник **не видит** `__income`, но **видит** `_house`.
4. Почему в Python предпочли соглашение, а не ключевые слова вроде `private`.

### Решение

```python
class Owner:
    def __init__(self, name, house, income):
        self.name = name            # публичное — доступно везде
        self._house = house         # защищённое — доступно наследникам
        self.__income = income      # приватное — только внутри класса

    def info(self):
        # Публичный метод. Не показывает доход — он приватный.
        return f"Владелец: {self.name}, дом: {self._house}"


class Tenant(Owner):
    def describe(self):
        # Защищённое поле _house доступно наследнику.
        return f"Арендатор в доме: {self._house}"


# Создаём объект владельца.
owner = Owner("Иванов И.И.", "ул. Ленина, д. 10", 120_000)

# 1. Публичное поле — доступно свободно.
print("Публичное name:", owner.name)

# 2. Защищённое поле — доступно, но не рекомендуется.
print("Защищённое _house:", owner._house)   # работает, но не рекомендуется

# 3. Приватное поле — напрямую недоступно.
try:
    print(owner.__income)
except AttributeError as e:
    print("Приватное __income недоступно:", e)

# 4. Через name mangling — доступ возможен (но не рекомендуется).
print("Через name mangling:", owner._Owner__income)

# 5. Проверяем, что защищённое поле доступно наследнику.
tenant = Tenant("Петров П.П.", "ул. Садовая, д. 3", 80_000)
print(tenant.describe())
```

**Ожидаемый вывод:**

```
Публичное name: Иванов И.И.
Защищённое _house: ул. Ленина, д. 10
Приватное __income недоступно: 'Owner' object has no attribute '__income'
Через name mangling: 120000
Арендатор в доме: ул. Садовая, д. 3
```

### Разбор

**1. Зачем нужны три уровня доступа.**

Три уровня позволяют **разделить** интерфейс класса и его внутреннее устройство:

- **Публичное** — то, чем могут пользоваться все. Это «официальный интерфейс» класса.
- **Защищённое** — то, что нужно наследникам. Это «интерфейс для расширения».
- **Приватное** — то, что касается только самого класса. Это «внутренняя кухня».

Благодаря этому программист, использующий класс, **сразу видит**, что можно трогать, а что нет.

**2. Почему `__name` не защищает по-настоящему.**

`__name` — это **не запрет**, а **переименование**. Python превращает `__income` в `_Owner__income`. Снаружи к этому полю можно обратиться, если знать настоящее имя. Это **защита от случайного доступа**, а не от злонамеренного.

**Зачем тогда `__name`?**

- Чтобы **избежать конфликтов имён** при наследовании.
- Чтобы **сигнализировать** другим программистам: «не трогай, это внутреннее».

**3. Почему наследник не видит `__income`, но видит `_house`.**

Когда Python видит `self.__income` **внутри класса** `Owner`, он превращает это в `self._Owner__income`. Когда тот же `self.__income` встречается **внутри класса** `Tenant`, Python превращает это в `self._Tenant__income`. Это **разные поля** — и второго не существует.

А `_house` — защищённое поле. Оно **не искажается**, поэтому доступно и в `Owner`, и в `Tenant`, и вообще везде. Именно поэтому если поле нужно наследникам — используют **защищённый** уровень, а не приватный.

**4. Почему в Python предпочли соглашение, а не ключевые слова.**

Python — **динамически типизированный язык** с философией «мы все здесь взрослые люди». Вместо жёстких запретов (как `private` в Java) используется **договорённость**:

- Программист **видит** подчёркивания и понимает: не трогай.
- Если очень нужно — может обойти, но это его ответственность.
- Код становится **проще**, **гибче**, без лишних ограничений.

Это работает, потому что сообщество Python **уважает соглашение**. Нарушать его — плохой тон, и это заметно сразу.
