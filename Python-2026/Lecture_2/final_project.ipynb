# 📘 Проект «Агентство недвижимости»

## Часть 1. Постановка задачи

---

## 1.1. Что мы делаем

Мы разрабатываем **учебную систему учёта агентства недвижимости**. Это полноценное консольное приложение, которое демонстрирует все ключевые темы объектно-ориентированного программирования на языке Python.

Система не претендует на промышленное применение. Её цель — **показать**, как теория ООП превращается в работающий проект. Каждая глава курса находит здесь своё применение: от создания первого класса до принципов SOLID и паттернов проектирования.

---

## 1.2. Зачем это нужно

Курс ООП состоит из сорока глав. Каждая глава разбирает отдельную тему — инкапсуляцию, наследование, магические методы, дескрипторы, метаклассы. Но темы изучаются **по отдельности**. Возникает разрыв: студент знает, что такое `@property`, но не понимает, где это нужно в реальном коде.

## 1.7. Зависимости между слоями

Слои организованы так, что верхние зависят от нижних, но не наоборот.

**CLI** зависит от сервисов. Он вызывает методы сервисов и показывает результаты.

**Сервисы** зависят от моделей, паттернов и интерфейсов. Они используют модели как данные, паттерны как инструменты, интерфейсы как контракты.

**Паттерны** зависят от моделей и интерфейсов. Фабрика создаёт модели, декораторы оборачивают их, адаптеры приводят к нужному виду.

**Модели** зависят от интерфейсов, исключений и утилит. Они реализуют интерфейсы, используют дескрипторы, выбрасывают свои исключения.

**База данных** зависит от моделей. Она сохраняет и загружает их.

**Интерфейсы, исключения, утилиты** не зависят ни от чего, кроме стандартной библиотеки.

Такая структура обеспечивает **слабую связь**. Изменение в одном слое не ломает другие.

## 1. `src/real_estate/interfaces/protocols.py`

### Теория

**Абстрактный класс** — это класс, который нельзя создать напрямую. Он описывает **общий интерфейс** для группы классов. В Python абстрактные классы создаются через модуль `abc`.

**ABC** (`Abstract Base Class`) — базовый класс для абстрактных классов. От него наследуются все абстрактные классы.

**`@abstractmethod`** — декоратор, который помечает метод как обязательный. Наследник **обязан** его реализовать.

**Protocol** — механизм структурной типизации. В отличие от `ABC`, он не требует наследования. Любой объект, у которого есть нужные методы, **соответствует** протоколу.

**Зачем это в проекте.** Интерфейсы описывают, что должен уметь объект недвижимости, уведомитель, хранилище. Без них код был бы связан с конкретными классами.

**Ключевое различие.** `ABC` требует **наследования**. `Protocol` — **структурного соответствия**. Первый подходит для явных контрактов, второй — для гибких случаев.

### Задание

Создайте семь интерфейсов: `Infoable`, `Sellable`, `Rentable`, `Renovatable`, `Describable`, `Notifier`, `Storage`. Первые четыре и последний два — через `ABC`. `Describable` — через `Protocol`.

### Полный код

```python
"""
Абстрактные интерфейсы и протоколы.

Этот модуль описывает контракты, которым должны соответствовать
классы проекта. ABC используется для явных контрактов, Protocol —
для структурной типизации.
"""

from abc import ABC, abstractmethod
from typing import Protocol


class Infoable(ABC):
    """
    Интерфейс объекта, у которого есть текстовое описание.

    Любой объект недвижимости обязан реализовать get_info().
    """

    @abstractmethod
    def get_info(self) -> str:
        """Возвращает текстовое описание объекта."""
        ...


class Sellable(ABC):
    """
    Интерфейс продаваемого объекта.

    Реализуется теми классами, которые могут быть проданы.
    """

    @abstractmethod
    def sell(self) -> str:
        """Продаёт объект. Возвращает результат операции."""
        ...


class Rentable(ABC):
    """
    Интерфейс сдаваемого объекта.

    Реализуется теми классами, которые могут быть сданы в аренду.
    """

    @abstractmethod
    def rent(self) -> str:
        """Сдаёт объект в аренду. Возвращает результат операции."""
        ...


class Renovatable(ABC):
    """
    Интерфейс ремонтируемого объекта.

    Реализуется теми классами, которые можно ремонтировать.
    """

    @abstractmethod
    def renovate(self) -> str:
        """Запускает ремонт. Возвращает результат."""
        ...


class Describable(Protocol):
    """
    Протокол для утиной типизации.

    Не требует наследования. Любой объект с методом describe()
    автоматически соответствует протоколу.
    """

    def describe(self) -> str:
        """Возвращает описание объекта."""
        ...


class Notifier(ABC):
    """
    Интерфейс уведомителя.

    Реализуется теми классами, которые отправляют уведомления.
    """

    @abstractmethod
    def send(self, message: str) -> None:
        """Отправляет сообщение."""
        ...


class Storage(ABC):
    """
    Интерфейс хранилища объектов.

    Реализуется теми классами, которые хранят и ищут объекты.
    """

    @abstractmethod
    def save(self, item) -> None:
        """Сохраняет объект."""
        ...

    @abstractmethod
    def find_by_address(self, address: str):
        """Ищет объект по адресу. Возвращает найденный объект."""
        ...
```

### Разбор

**`Infoable`, `Sellable`, `Rentable`, `Renovatable`** — четыре интерфейса, описывающих возможности объектов недвижимости. Они разделены по принципу ISP (глава 40): у каждого интерфейса — одна ответственность.

**`Describable`** — протокол. Он не требует наследования. Любой объект с методом `describe()` **соответствует** ему. Это позволяет использовать утиную типизацию (глава 30).

**`Notifier`** — интерфейс для уведомителей. Реализуется классами `EmailNotifier` и `SmsNotifier`.

**`Storage`** — интерфейс хранилища. Реализуется классом `EstateStorage`.

**Три точки `...`** — тело метода-заглушки. Можно было бы написать `pass`, но `...` (Ellipsis) — более идиоматично для абстрактных методов.

### Пример использования

```python
from src.real_estate.interfaces import Infoable, Describable


class Apartment(Infoable):
    def get_info(self):
        return "Квартира"


flat = Apartment()
print(flat.get_info())
print(isinstance(flat, Infoable))
```

**Вывод:**

```
Квартира
True
```

## 2. `src/real_estate/interfaces/mixins.py`

### Теория

**Миксин** — это класс, который добавляет одну узкую функциональность. Он не предназначен для самостоятельного использования. Миксин подключается через множественное наследование.

**Отличие от обычного родителя.** Обычный родитель описывает основное понятие. Миксин — дополнительное умение. Например, `Apartment` наследует от `Sellable` (основное) и от `LoggableMixin` (дополнительное).

**Почему миксин идёт первым.** В MRO (глава 19) методы миксина должны иметь приоритет над методами базовых классов. Поэтому миксин ставят первым в списке родителей.

**Зачем это в проекте.** `LoggableMixin` даёт метод `log()` всем объектам, которые должны логировать события. `TimestampMixin` добавляет временные метки.

### Задание

Создайте два миксина. `LoggableMixin` добавляет метод `log(message)`, который печатает сообщение с временной меткой и именем класса. `TimestampMixin` добавляет метод `set_created()`, который устанавливает дату создания.

### Полный код

```python
"""
Миксины — «примеси» с узкой функциональностью.

Миксин не предназначен для самостоятельного использования.
Он подключается к другим классам через множественное наследование
и добавляет им одну возможность.
"""

from datetime import datetime


class LoggableMixin:
    """
    Добавляет метод логирования с временной меткой.

    Используется классами, которые должны логировать события.
    """

    def log(self, message: str) -> None:
        """
        Печатает сообщение с временной меткой и именем класса.

        Имя класса берётся через self.__class__.__name__,
        поэтому метод работает правильно для всех наследников.
        """
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        print(f"[{timestamp}] {self.__class__.__name__}: {message}")


class TimestampMixin:
    """
    Добавляет временные метки создания объекта.

    Используется классами, которым важно знать,
    когда объект был создан.
    """

    def set_created(self) -> None:
        """Устанавливает текущую дату в атрибут created_at."""
        self.created_at = datetime.now()
```

### Разбор

**`self.__class__.__name__`** — динамически определяет имя класса. Для `Apartment` вернёт `Apartment`, для `Studio` — `Studio`. Если написать жёстко `Apartment`, наследники будут логировать неправильно.

**`datetime.now()`** — текущая дата и время. Форматируется через `strftime`.

**Отсутствие `__init__`** — важная особенность миксинов. Они **не должны** определять `__init__`, иначе нарушат порядок инициализации в MRO.

### Пример использования

```python
from src.real_estate.interfaces import LoggableMixin


class Apartment(LoggableMixin):
    def __init__(self, address):
        self.address = address


flat = Apartment("ул. Ленина")
flat.log("Создана квартира")
```

**Вывод:**

```
[2025-01-15 10:30:45] Apartment: Создана квартира
```

## 4. `src/real_estate/exceptions/errors.py`

### Теория

**Собственные исключения** позволяют различать ошибки проекта от ошибок библиотек. Они наследуются от `Exception` (или от базового исключения проекта).

**Зачем это нужно.** Стандартные исключения (`ValueError`, `TypeError`) не говорят, **что именно** пошло не так. Своё исключение `ApartmentNotFoundError` сразу понятно.

**Иерархия исключений** — важный приём. Базовое исключение `AgencyError` объединяет все ошибки проекта. Можно ловить как конкретную ошибку, так и все ошибки агентства.

**Зачем это в проекте.** Сервисы выбрасывают свои исключения. Клиент ловит именно их, а не стандартные.

### Задание

Создайте пять исключений. `AgencyError` — базовое. `ApartmentNotFoundError`, `ApartmentAlreadySoldError`, `InvalidAreaError`, `DatabaseError` — наследники. У каждого, кроме первого, — свои поля.

### Полный код

```python
"""
Собственные исключения проекта.

Все исключения наследуются от AgencyError. Это позволяет
ловить как конкретную ошибку, так и все ошибки агентства сразу.
"""


class AgencyError(Exception):
    """
    Базовое исключение агентства.

    Все остальные ошибки проекта наследуются от него.
    Ловить AgencyError — значит ловить все свои ошибки.
    """
    pass


class ApartmentNotFoundError(AgencyError):
    """Квартира по указанному адресу не найдена."""

    def __init__(self, address):
        self.address = address
        super().__init__(f"Квартира по адресу {address!r} не найдена")


class ApartmentAlreadySoldError(AgencyError):
    """Квартира уже продана."""

    def __init__(self, address):
        self.address = address
        super().__init__(f"Квартира {address!r} уже продана")


class InvalidAreaError(AgencyError):
    """Некорректная площадь — вне допустимого диапазона."""

    def __init__(self, area, min_area=5, max_area=500):
        self.area = area
        self.min_area = min_area
        self.max_area = max_area
        super().__init__(
            f"Площадь {area} м² вне диапазона [{min_area}, {max_area}]"
        )

    def is_too_small(self) -> bool:
        """Проверяет, меньше ли площадь минимальной."""
        return self.area < self.min_area

    def is_too_large(self) -> bool:
        """Проверяет, больше ли площадь максимальной."""
        return self.area > self.max_area


class DatabaseError(AgencyError):
    """Ошибка при работе с базой данных."""

    def __init__(self, message):
        self.message = message
        super().__init__(f"Ошибка БД: {message}")
```

### Разбор

**`super().__init__(...)`** — обязательный вызов. Он передаёт сообщение в родительский класс. Без него `str(exception)` вернёт пустоту.

**`self.address`** — сохраняем дополнительные данные. Клиент может их прочитать в обработчике.

**`!r`** в f-строке — добавляет кавычки вокруг строки. Это делает сообщение понятнее.

**Метод `is_too_small()`** — помогает различить, что именно не так: площадь слишком маленькая или слишком большая. Это пример **богатого исключения**.

### Пример использования

```python

    ApartmentNotFoundError,
    InvalidAreaError,
    AgencyError,
)

try:
    raise ApartmentNotFoundError("ул. Ленина")
except ApartmentNotFoundError as e:
    print(e)
    print(f"Адрес: {e.address}")

try:
    raise InvalidAreaError(2)
except InvalidAreaError as e:
    print(e)
    print(f"Слишком маленькая: {e.is_too_small()}")

try:
    raise ApartmentNotFoundError("ул. Мира")
except AgencyError as e:
    print(f"Поймано как AgencyError: {e}")
```

```

Адрес: ул. Ленина
```

## 6. `src/real_estate/utils/descriptors.py`

### Теория

**Дескриптор** — это класс, который управляет доступом к атрибуту другого класса. Дескриптор определяет методы `__get__`, `__set__`, `__delete__` и `__set_name__`.

**Зачем нужны дескрипторы.** Когда одна и та же проверка повторяется в нескольких полях, `@property` становится громоздким. Дескриптор позволяет вынести логику в отдельный класс и переиспользовать.

**Как работает `RangeDescriptor`.** Дескриптор хранит минимальное и максимальное значение. При записи он проверяет, что значение — число и что оно в диапазоне. Значение хранится в `instance.__dict__`, а не в дескрипторе. Это важно: иначе все объекты делили бы одно значение.

**Метод `__set_name__`.** Python автоматически вызывает его при создании класса. Дескриптор узнаёт своё имя (например, `area` или `rooms`) и использует его для сохранения значения.

**Зачем это в проекте.** Дескриптор `RangeDescriptor` применяется для валидации площади, цены, количества комнат. Одна логика — много полей.

### Задание

Создайте класс `RangeDescriptor`. Он должен принимать минимум и максимум, сохранять имя через `__set_name__`, читать значение из `instance.__dict__` и проверять его при записи.

### Полный код

```python
"""
Дескриптор для проверки диапазона значений.

Используется для полей, которые должны быть числами
в определённом диапазоне. Один дескриптор может применяться
к нескольким полям разных классов.
"""


class RangeDescriptor:
    """
    Дескриптор диапазона.

    При записи значения проверяет, что оно является числом
    и находится в диапазоне [min_value, max_value].
    Значение хранится в instance.__dict__ под именем поля.
    """

    def __init__(self, min_value, max_value):
        self.min_value = min_value
        self.max_value = max_value

    def __set_name__(self, owner, name):
        """
        Вызывается автоматически при создании класса.

        Python сообщает дескриптору, под каким именем он
        используется в классе. Это позволяет сохранять значение
        в instance.__dict__ под правильным ключом.
        """
        self.name = name

    def __get__(self, instance, owner=None):
        """
        Возвращает значение при чтении атрибута.

        Если обращаются через класс (instance is None),
        возвращается сам дескриптор.
        """
        if instance is None:
            return self
        return instance.__dict__.get(self.name)

    def __set__(self, instance, value):
        """
        Проверяет и сохраняет значение при записи.

        Если значение не число — выбрасывается TypeError.
        Если значение вне диапазона — выбрасывается ValueError.
        """
        if not isinstance(value, (int, float)):
            raise TypeError(f"{self.name} должно быть числом")

        if not (self.min_value <= value <= self.max_value):
            raise ValueError(
                f"{self.name} должно быть от {self.min_value} "
                f"до {self.max_value}"
            )

        instance.__dict__[self.name] = value
```

### Разбор

**`self.min_value`** и **`self.max_value`** — параметры дескриптора. Задаются при создании.

**`self.name`** — имя поля. Устанавливается автоматически через `__set_name__`. Без этого дескриптор не знал бы, под каким ключом сохранять значение.

**`instance.__dict__.get(self.name)`** — читаем значение из словаря **конкретного объекта**. Если значения нет, возвращается `None`.

**`instance.__dict__[self.name] = value`** — сохраняем значение **после** проверок. Если проверка не прошла, исключение выброшено, значение не сохранено.

**`if instance is None`** — обработка случая, когда обращаются через класс, а не через объект. Это стандартное поведение дескрипторов.

### Пример использования

```python

class Apartment:
    area = RangeDescriptor(5, 500)
    rooms = RangeDescriptor(1, 10)

    def __init__(self, area, rooms):
        self.area = area
        self.rooms = rooms


flat = Apartment(65, 3)
print(flat.area)
print(flat.rooms)

try:
    flat.area = 1000
except ValueError as e:
    print(f"Ошибка: {e}")

try:
    flat.rooms = 15
except ValueError as e:
    print(f"Ошибка: {e}")
```

## 7. `src/real_estate/utils/validators.py`

### Теория

**Статический метод** — это метод, который не зависит ни от объекта, ни от класса. Он объявляется декоратором `@staticmethod`. Внутри статического метода нет ни `self`, ни `cls`.

**Зачем нужны статические методы.** Когда логика связана с классом по смыслу, но не зависит от его состояния. Например, проверка, является ли число корректной площадью.

**Отличие от обычного метода.** Обычный метод получает `self` и работает с объектом. Статический метод — просто функция внутри класса.

**Отличие от классового метода.** Классовый метод получает `cls` и работает с классом. Статический не получает ничего.

**Зачем это в проекте.** Класс `EstateUtils` содержит утилиты для проверки данных. Их можно вызывать без создания объекта.

### Задание

Создайте класс `EstateUtils` с тремя статическими методами: `is_valid_area`, `is_valid_price`, `is_valid_address`. Каждый возвращает `True` или `False`.

### Полный код

```python
"""
Статические утилиты для проверки данных.

Класс не имеет состояния. Все методы — статические.
Вызываются напрямую через имя класса:

    EstateUtils.is_valid_area(65)
"""


class EstateUtils:
    """
    Утилиты для проверки корректности данных о недвижимости.

    Все методы — статические. Объект класса создавать не нужно.
    """

    @staticmethod
    def is_valid_area(area) -> bool:
        """
        Проверяет, является ли площадь корректной.

        Площадь должна быть числом в диапазоне от 5 до 500 м².
        """
        return isinstance(area, (int, float)) and 5 <= area <= 500

    @staticmethod
    def is_valid_price(price) -> bool:
        """
        Проверяет, является ли цена корректной.

        Цена должна быть положительным числом.
        """
        return isinstance(price, (int, float)) and price > 0

    @staticmethod
    def is_valid_address(address) -> bool:
        """
        Проверяет, является ли адрес корректным.

        Адрес должен быть непустой строкой.
        """
        return isinstance(address, str) and len(address.strip()) > 0
```

### Разбор

**`@staticmethod`** — декоратор, который делает метод статическим. Python не передаёт в него ни `self`, ни `cls`.

**`isinstance(area, (int, float))`** — проверка типа. Кортеж типов означает «любой из перечисленных».

**`5 <= area <= 500`** — цепочка сравнений. В Python это эквивалентно `5 <= area and area <= 500`.

**`address.strip()`** — удаляет пробелы по краям. Это защита от адресов вида `"   "`.

### Пример использования

```python

print(EstateUtils.is_valid_area(-10))
print(EstateUtils.is_valid_area("abc"))

print(EstateUtils.is_valid_price(0))

print(EstateUtils.is_valid_address("ул. Ленина"))
```

## 8. `src/real_estate/utils/logging.py`

### Теория

**Логирование** — это запись событий в журнал. Стандартная библиотека Python содержит модуль `logging` для этого.

**Зачем нужен логгер.** В проекте нужно записывать события: создание объекта, продажу, ошибку. Логгер централизует эту запись.

**Уровни логирования.** Стандартные уровни: `DEBUG`, `INFO`, `WARNING`, `ERROR`, `CRITICAL`. Чем выше уровень, тем серьёзнее событие.

**Зачем это в проекте.** Логгер записывает действия агентства. Это помогает отлаживать и анализировать работу.

### Задание

Создайте функцию `setup_logging(level)` и класс `AgencyLogger`. Функция настраивает логирование и возвращает логгер. Класс — обёртка над логгером для удобства.

### Полный код

```python
"""
Настройка логирования.

Функция setup_logging создаёт и настраивает логгер.
Класс AgencyLogger — удобная обёртка над логгером.
"""

import logging


def setup_logging(level=logging.INFO) -> logging.Logger:
    """
    Настраивает и возвращает логгер для проекта.

    Если у логгера уже есть обработчики, повторная настройка
    не выполняется — это защита от дублирования сообщений.
    """
    logger = logging.getLogger("real_estate")
    logger.setLevel(level)

    if not logger.handlers:
        handler = logging.StreamHandler()
        formatter = logging.Formatter(
            "%(asctime)s [%(levelname)s] %(name)s: %(message)s"
        )
        handler.setFormatter(formatter)
        logger.addHandler(handler)

    return logger


class AgencyLogger:
    """
    Обёртка над стандартным логгером.

    Предоставляет методы info, warning, error
    для удобной записи сообщений.
    """

    def __init__(self, name: str = "agency"):
        self.logger = logging.getLogger(f"real_estate.{name}")

    def info(self, message: str) -> None:
        """Записывает информационное сообщение."""
        self.logger.info(message)

    def warning(self, message: str) -> None:
        """Записывает предупреждение."""
        self.logger.warning(message)

    def error(self, message: str) -> None:
        """Записывает сообщение об ошибке."""
        self.logger.error(message)
```

### Разбор

**`logging.getLogger("real_estate")`** — получает логгер по имени. Если логгер уже существует, возвращается он же.

**`if not logger.handlers`** — защита от повторной настройки. Без неё каждое сообщение дублировалось бы.

**`logging.StreamHandler()`** — обработчик, который выводит сообщения в консоль.

**`logging.Formatter(...)`** — формат сообщений. Включает время, уровень, имя логгера и текст.

**`self.logger.info(...)`** — вызывает соответствующий метод стандартного логгера.

### Пример использования

```python
logger.info("Сделка оформлена")
logger.warning("Клиент не найден")
logger.error("Ошибка базы данных")
```

## 1. `src/real_estate/models/building.py`

### Теория

**Здание** — это часть квартиры. Квартира не может существовать без дома, и дом создаётся вместе с квартирой. Это классический пример **композиции** (глава 8).

**Почему отдельный класс, а не строка.** У здания есть свои характеристики: название, этажность. Если бы адрес дома был строкой, мы не смогли бы расширить его без правки всех мест. Отдельный класс позволяет добавлять поля постепенно.

**Зачем это в проекте.** `Building` используется в композиции с `Apartment`. Также он может использоваться в других классах, где нужна информация о здании.

### Задание

Создайте класс `Building` с полями `name` и `floors`. Добавьте метод `show()`, возвращающий строку описания, и метод `__repr__` для отладки.

### Полный код

```python
"""
Здание.

Используется в композиции с классом Apartment:
квартира содержит объект Building как часть себя.
"""


class Building:
    """Здание — часть квартиры."""

    def __init__(self, name: str, floors: int):
        self.name = name
        self.floors = floors

    def show(self) -> str:
        """Возвращает строковое описание здания."""
        return f"Здание {self.name!r}, {self.floors} этажей"

    def __repr__(self) -> str:
        return f"Building({self.name!r}, {self.floors})"
```

### Разбор

**`name: str`** и **`floors: int`** — аннотации типов. Они не влияют на выполнение, но документируют код и помогают IDE.

**`!r`** в f-строке — вызывает `repr()` для значения. Для строк это добавляет кавычки, что удобно для отладки.

**`__repr__`** — вызывается при `print(building)` в списке или в отладчике. Показывает техническое представление.

### Пример использования

```python
building = Building("Дом-1", 9)
print(building.show())
print(building)
```

**Вывод:**

```
Здание 'Дом-1', 9 этажей
Building('Дом-1', 9)
```

## 2. `src/real_estate/models/estate_item.py`

### Теория

**`EstateItem`** — это **абстрактный базовый класс** (глава 7). Он описывает общее понятие «объект недвижимости», но не может существовать сам по себе. У каждой конкретной недвижимости должно быть своё описание.

**Зачем нужен.** Чтобы все объекты недвижимости имели общий интерфейс — метод `get_info()`. Тогда код может работать с любым объектом недвижимости, не зная, квартира это или дом. Это **полиморфизм** (глава 4).

**Что наследует.** `EstateItem` наследуется от `Infoable` (интерфейс с `get_info()`) и от `ABC` (для абстрактности). Это сочетание даёт и интерфейс, и запрет на создание объекта.

**Метод `describe()`.** Он возвращает то же, что `get_info()`. Это сделано для **утиной типизации** (глава 30): чтобы объект подходил под протокол `Describable`.

### Задание

Создайте абстрактный класс `EstateItem`, наследующий от `Infoable` и `ABC`. Метод `get_info()` должен быть абстрактным. Добавьте метод `describe()` и `__repr__`.

### Полный код

```python
"""
Базовый класс недвижимости.

EstateItem — абстрактный класс. Объект создать нельзя,
но любой наследник обязан реализовать get_info().
"""

from abc import ABC

from ..interfaces import Infoable


class EstateItem(Infoable, ABC):
    """Общий интерфейс для всех объектов недвижимости."""

    def describe(self) -> str:
        """Возвращает описание. Нужен для утиной типизации."""
        return self.get_info()

    def __repr__(self) -> str:
        return f"{self.__class__.__name__}({self.get_info()!r})"
```

### Разбор

**`from abc import ABC`** — импорт базового класса для абстрактных классов.

**`EstateItem(Infoable, ABC)`** — множественное наследование: интерфейс плюс абстрактность. `Infoable` содержит абстрактный метод `get_info()`, поэтому `EstateItem` тоже абстрактный.

**`self.__class__.__name__`** — имя фактического класса. Это делает `__repr__` универсальным для всех наследников.

### Пример использования

```python
# Так нельзя — абстрактный класс:
# item = EstateItem()  # TypeError


class Apartment(EstateItem):
    def get_info(self):
        return "Квартира"


flat = Apartment()
print(flat.get_info())
print(flat.describe())
print(flat)
```

**Вывод:**

```
Квартира
Квартира
Apartment('Квартира')
```

## 3. `src/real_estate/models/apartment.py`

### Теория

**`Apartment`** — главная модель проекта. Она используется почти во всех задачах курса. Здесь собраны семь тем.

| Тема | Глава | Где в классе |
|------|-------|--------------|
| `@dataclass` | 20 | Декоратор класса |
| Миксин | 20 | `LoggableMixin` в наследовании |
| Интерфейсы | 40 | `Sellable`, `Rentable` |
| Классовый счётчик | 16 | Атрибут `count` |
| `__post_init__` | 20 | Метод после `__init__` |
| `__str__` | 11 | Метод |
| Классовый метод | 15 | `from_dict()` |

**Почему `@dataclass`.** Класс содержит только данные и простые методы. Это классический случай для `@dataclass`: конструктор, `__repr__` и `__eq__` создаются автоматически.

**Почему миксин первым.** `LoggableMixin` должен идти первым в списке родителей (глава 20), чтобы его методы имели приоритет в MRO.

**Почему `count` — классовый.** Счётчик должен быть общим для всех объектов. Иначе каждый объект будет считать только себя (глава 16).

**Почему `__post_init__`.** `@dataclass` создаёт `__init__` автоматически, но нам нужно после него увеличить счётчик и установить `is_sold = False`. Для этого есть `__post_init__`.

### Задание

Создайте класс `Apartment` через `@dataclass`, наследуя от `LoggableMixin`, `Sellable`, `Rentable`. Поля: `address`, `area`, `price`. Добавьте классовый счётчик `count`, методы `get_info()`, `sell()`, `rent()`, `to_dict()`, `from_dict()`, `__str__`.

### Полный код

```python
"""
Квартира.

Главная модель проекта. Демонстрирует:
- @dataclass (глава 20)
- миксин LoggableMixin (глава 20)
- интерфейсы Sellable, Rentable (глава 40)
- классовый счётчик count (глава 16)
- __post_init__ (глава 20)
- __str__ (глава 11)
- классовый метод from_dict (глава 15)
"""

from dataclasses import dataclass
from typing import ClassVar

from ..interfaces import LoggableMixin, Sellable, Rentable


@dataclass
class Apartment(LoggableMixin, Sellable, Rentable):
    """Квартира: продаётся и сдаётся в аренду."""

    address: str
    area: int
    price: int

    count: ClassVar[int] = 0

    def __post_init__(self) -> None:
        """Вызывается после автоматического __init__."""
        type(self).count += 1
        self.is_sold = False

    def get_info(self) -> str:
        """Возвращает текстовое описание квартиры."""
        return (
            f"Квартира {self.address}: "
            f"{self.area} м², {self.price:,} ₽"
        ).replace(",", " ")

    def sell(self) -> str:
        """Продаёт квартиру."""
        if self.is_sold:
            return "Уже продана"
        self.is_sold = True
        return "Продано"

    def rent(self) -> str:
        """Сдаёт квартиру в аренду."""
        return "Сдано"

    def to_dict(self) -> dict:
        """Преобразует объект в словарь для сохранения в БД."""
        return {
            "type": self.__class__.__name__,
            "address": self.address,
            "area": self.area,
            "price": self.price,
            "is_sold": self.is_sold,
        }

    @classmethod
    def from_dict(cls, data: dict) -> "Apartment":
        """Создаёт объект из словаря при загрузке из БД."""
        return cls(
            address=data["address"],
            area=data["area"],
            price=data["price"],
        )

    def __str__(self) -> str:
        return self.get_info()
```

### Разбор

**`@dataclass`** автоматически создаёт:

- `__init__(self, address, area, price)` — конструктор;
- `__repr__(self)` — техническое представление;
- `__eq__(self, other)` — сравнение по всем полям.

**`count: ClassVar[int] = 0`** — атрибут класса, не экземпляра. `ClassVar` говорит Python, что это не поле для `@dataclass`, а классовая переменная.

**`type(self).count += 1`** — увеличиваем счётчик через фактический класс. Если создать наследника `Studio`, счётчик `Studio.count` будет свой (если переопределён) или общий с `Apartment`. Это правильный способ: жёсткое `Apartment.count` ломало бы наследование.

**`is_sold = False`** создаётся в `__post_init__`, а не в полях. Если бы мы написали `is_sold: bool = False` в полях, `@dataclass` добавил бы его в конструктор, что не нужно.

**`.replace(",", " ")`** — форматирование цены с пробелами между разрядами: `8 500 000` вместо `8,500,000`.

### Пример использования

```python
flat = Apartment("ул. Ленина, д. 10, кв. 5", 65, 8_500_000)
print(flat.get_info())
print(flat.sell())
print(flat.sell())
print(f"Всего квартир: {Apartment.count}")

flat_dict = flat.to_dict()
print(flat_dict)

restored = Apartment.from_dict(flat_dict)
print(restored)
```

**Вывод:**

```
Квартира ул. Ленина, д. 10, кв. 5: 65 м², 8 500 000 ₽
Продано
Уже продана
Всего квартир: 1
{'type': 'Apartment', 'address': 'ул. Ленина, д. 10, кв. 5', 'area': 65, 'price': 8500000, 'is_sold': True}
Квартира ул. Ленина, д. 10, кв. 5: 65 м², 8 500 000 ₽
```

**Что демонстрирует.** Наследование и переопределение метода `get_info()` (глава 9). Наследник не дублирует конструктор: он наследует его от `Apartment`.

**Почему `@dataclass`.** Студия тоже содержит только данные. `@dataclass` автоматически создаст `__init__` с учётом всех полей: и родительских, и своих.

### Задание

Создайте класс `Studio`, наследник `Apartment`, через `@dataclass`. Добавьте поле `kitchen_area: int = 5`. Переопределите `get_info()` и добавьте метод `utility_cost()`, возвращающий 3000.

### Полный код

```python
"""
Студия.

Наследник Apartment. Демонстрирует:
- наследование (глава 3)
- переопределение метода (глава 9)
- @dataclass с дополнительным полем (глава 20)
"""

from dataclasses import dataclass

from .apartment import Apartment


@dataclass
class Studio(Apartment):
    """Студия — квартира с маленькой кухней."""

    kitchen_area: int = 5
```

**`get_info()`** — переопределён (глава 9). Родительская версия не вызывается. Если бы мы хотели её дополнить, использовали бы `super().get_info()`.

## 7. `src/real_estate/models/penthouse.py`

### Теория

**`Penthouse`** — наследник `Apartment`. Пентхаус — это квартира с террасой. Это разновидность квартиры, поэтому наследование оправдано.

**Что демонстрирует.** Расширение родительского метода через `super()` (главы 9, 28). `Penthouse.get_info()` не переписывает логику родителя, а дополняет её.

**Почему не композиция.** В главе 28 мы обсуждали, что пентхаус можно реализовать через композицию. Но в проекте выбран вариант с наследованием: потому что пентхаус является квартирой в контексте агентства.

### Задание

Создайте класс `Penthouse`, наследник `Apartment`, через `@dataclass`. Добавьте поле `terrace: int = 0`. Переопределите `get_info()` через `super()`.

### Полный код

```python
"""
Пентхаус.

Наследник Apartment. Демонстрирует:
- расширение метода через super() (глава 9)
- добавление поля в @dataclass (глава 20)
"""

from dataclasses import dataclass

from .apartment import Apartment


@dataclass
class Penthouse(Apartment):
    """Пентхаус — квартира с террасой."""

    terrace: int = 0

    def get_info(self) -> str:
        """Дополняет родительское описание информацией о террасе."""

```

### Разбор

**`super().get_info()`** — вызывает родительский метод `Apartment.get_info()`. Это расширение, а не замена (глава 9).

**`terrace: int = 0`** — поле со значением по умолчанию. Если террасы нет, значение 0.

### Пример использования

```python

print(penthouse.get_info())
```


## 8. `src/real_estate/models/client.py`

### Теория

**`Client`** и **`Agent`** — вспомогательные модели. Они не объекты недвижимости, поэтому не наследуют `EstateItem`.

**Что демонстрируют.** Утиную типизацию (глава 30). У обоих есть метод `describe()`, но общего родителя нет. Функция `describe_item(item)` работает с любым объектом, у которого есть `describe()`.

**Client и Observer.** Клиент используется в паттерне Observer (глава 26): у него есть метод `update(event)`, который вызывается при уведомлении.

```python
"""
Клиент и агент.

Демонстрируют:
- утиную типизацию (глава 30)
- Observer (глава 26) — у Client есть update()
"""


class Client:
    """Клиент агентства. Подписывается на уведомления."""

    def __init__(self, name: str, budget: int = 0):
        self.name = name
        self.budget = budget
        self.notifications: list[str] = []

    def describe(self) -> str:
        """Описание клиента."""
        return f"Клиент {self.name}, бюджет {self.budget} ₽"

    def update(self, event) -> None:
        """Реакция на уведомление (Observer, глава 26)."""
        message = f"{self.name}: {event}"

        print(f"  → {message}")

        return f"Client({self.name!r}, {self.budget})"


class Agent:

        self.experience = experience

        return f"Агент {self.name}, стаж {self.experience} лет"

    def __repr__(self) -> str:
        return f"Agent({self.name!r}, {self.experience})"
```

# 📘 Этап 3. Сервисы

## Что изучается в этом этапе

Сервисы — это **бизнес-логика** проекта. Модели описывают данные, сервисы описывают действия. Сервис не хранит данные, он их обрабатывает.

В сервисах применяются: принцип единственной ответственности (глава 27), паттерн Observer (глава 26), паттерн Strategy (глава 34), инверсия зависимостей (глава 40), работа с исключениями (глава 17), магические методы (главы 11–13, 31–32).

| Файл | Классы | Главы курса |
|------|--------|-------------|
| `storage.py` | `EstateStorage` | 27, 31, 32 |
| `commission.py` | 4 класса стратегий | 7, 34 |
| `notification_service.py` | `NotificationService`, `EmailNotifier`, `SmsNotifier` | 26, 40 |
| `deal_service.py` | `DealService` | 27, 40 |
| `rental_service.py` | `RentalService` | 27 |
| `report_service.py` | `ReportService` | 27 |
| `services/__init__.py` | экспорт | — |

**Главная идея этапа.** Каждый сервис отвечает за **одну** задачу. Хранение отдельно, уведомления отдельно, сделки отдельно. Это принцип SRP (глава 27).

**Зачем это нужно.** Если бы вся логика была в одном классе `Agency`, его было бы сложно менять. Разделив на сервисы, мы получаем независимые компоненты. Изменение в одном не ломает другие.

## 1. `src/real_estate/services/storage.py`

### Теория

**Хранилище** — это сервис, который хранит объекты и ищет их. Он не создаёт объекты, не продаёт, не уведомляет. Только хранит и ищет.

**Зачем отдельный класс.** В главе 27 мы обсуждали, что класс `Agency` разросся и делал всё сразу. Хранение — это одна ответственность, вынесенная в отдельный сервис.

**Магические методы.** Хранилище реализует `__len__`, `__iter__`, `__contains__`, `__getitem__`. Это делает его похожим на список: можно узнать длину, перебрать в цикле, проверить наличие, обратиться по индексу.

**Собственное исключение.** При поиске, если объект не найден, выбрасывается `ApartmentNotFoundError`. Это наше исключение, а не стандартное.

### Задание

Создайте класс `EstateStorage`, который хранит объекты и умеет искать их по адресу. Реализуйте методы: `add`, `find_by_address`, `remove`, `list_all`. Добавьте магические методы `__len__`, `__iter__`, `__contains__`, `__getitem__`.

### Полный код

```python
"""
Хранилище объектов недвижимости.

Отвечает только за хранение и поиск (SRP, глава 27).
Реализует магические методы, чтобы работать как список
(главы 31, 32).
"""

from ..exceptions import ApartmentNotFoundError


class EstateStorage:
    """
    Хранилище объектов недвижимости.

    Хранит объекты в списке. Умеет искать по адресу.
    При поиске несуществующего объекта выбрасывает
    ApartmentNotFoundError.
    """

    def __init__(self):
        self._items = []

    def add(self, item) -> None:
        """Добавляет объект в хранилище."""
        self._items.append(item)

    def find_by_address(self, address: str):
        """
        Ищет объект по адресу.

        Если объект найден, возвращает его.
        Если не найден, выбрасывает ApartmentNotFoundError.
        """
        for item in self._items:
            if getattr(item, "address", None) == address:
                return item
        raise ApartmentNotFoundError(address)

    def remove(self, item) -> None:
        """Удаляет объект из хранилища."""
        if item in self._items:
            self._items.remove(item)

    def list_all(self) -> list:
        """Возвращает копию списка всех объектов."""
        return list(self._items)

    def __len__(self) -> int:
        """Количество объектов в хранилище."""
        return len(self._items)

    def __iter__(self):
        """Позволяет перебирать объекты в цикле for."""
        return iter(self._items)

    def __contains__(self, item) -> bool:
        """Позволяет использовать оператор in."""
        return item in self._items

    def __getitem__(self, index):
        """Позволяет обращаться по индексу."""
        return self._items[index]
```

### Разбор

**`self._items`** — внутренний список. Атрибут с одним подчёркиванием означает, что он защищённый: не нужно обращаться к нему снаружи.

**`getattr(item, "address", None)`** — безопасное чтение атрибута. Если у объекта нет `address`, вернётся `None`. Так хранилище может работать с любыми объектами, у которых есть адрес.

**`raise ApartmentNotFoundError(address)`** — выбрасываем своё исключение. Клиент может поймать именно его и понять, что искать нужно иначе.

**`list(self._items)`** — возвращаем копию, а не сам список. Иначе клиент мог бы случайно изменить внутреннее состояние хранилища.

**`__len__`** — позволяет писать `len(storage)`. **`__iter__`** — `for item in storage`. **`__contains__`** — `if item in storage`. **`__getitem__`** — `storage[0]`.

### Пример использования

```python

storage.add(House("ул. Садовая, д. 3", 200, 2))

    print(f"  {item.get_info()}")

print(f"Найдено: {flat.get_info()}")
```

## 2. `src/real_estate/services/commission.py`

### Теория

**Паттерн Strategy** (глава 34) позволяет выбирать алгоритм во время работы. В нашем случае алгоритм — это способ расчёта комиссии.

**Три стратегии.** Фиксированная сумма, процент от цены, льготный процент. Все три реализуют один интерфейс `CommissionStrategy` с методом `calculate(price)`.

**Абстрактный базовый класс.** `CommissionStrategy` наследуется от `ABC` и объявляет `calculate` абстрактным. Каждый наследник обязан реализовать его.

**Использование `super()`.** `LoyaltyCommission` — наследник `PercentCommission`. Он не дублирует логику расчёта процента, а вызывает родительский конструктор с другой ставкой. Это переиспользование кода (глава 9).

**Зачем это в проекте.** Агентство может менять способ расчёта комиссии. Стратегия позволяет делать это без правки `DealService`.

### Задание

Создайте абстрактный класс `CommissionStrategy` с методом `calculate(price)`. Создайте три наследника: `FixedCommission`, `PercentCommission`, `LoyaltyCommission`. Последний наследует от `PercentCommission` и использует `super()` для установки ставки.

### Полный код

```python
"""
Стратегии расчёта комиссии.

Паттерн Strategy (глава 34). Позволяет менять способ
расчёта комиссии без изменения сервиса сделок.
"""

from abc import ABC, abstractmethod


class CommissionStrategy(ABC):
    """
    Абстрактная стратегия расчёта комиссии.

    Наследники обязаны реализовать calculate.
    """

    @abstractmethod
    def calculate(self, price: int) -> float:
        """Рассчитывает комиссию от цены."""
        ...


class FixedCommission(CommissionStrategy):
    """Фиксированная комиссия."""

    def __init__(self, amount: int = 50_000):
        self.amount = amount

    def calculate(self, price: int) -> float:
        """Возвращает фиксированную сумму."""
        return self.amount


class PercentCommission(CommissionStrategy):
    """Процентная комиссия."""

    def __init__(self, rate: float = 0.03):
        self.rate = rate

    def calculate(self, price: int) -> float:
        """Возвращает процент от цены."""
        return price * self.rate


class LoyaltyCommission(PercentCommission):
    """
    Льготная комиссия для постоянных клиентов.

    Наследует логику процентной комиссии,
    но с уменьшенной ставкой.
    """

    def __init__(self):
        super().__init__(rate=0.015)
```

### Разбор

**`CommissionStrategy(ABC)`** — абстрактный класс. Объект создать нельзя, но наследники обязаны реализовать `calculate`.

**`FixedCommission`** — возвращает фиксированную сумму. Ставка не зависит от цены.

**`PercentCommission`** — возвращает процент от цены. Ставка задаётся в конструкторе.

**`LoyaltyCommission(PercentCommission)`** — наследник `PercentCommission`. Не реализует `calculate` заново: он переиспользует родительскую логику. Меняется только ставка через `super().__init__(rate=0.015)`.

**Почему это лучше `if/elif`.** Если бы мы написали `if kind == "fixed": ... elif kind == "percent": ...`, то при добавлении новой стратегии пришлось бы править функцию. Со стратегиями — просто новый класс.

### Пример использования

```python

    FixedCommission, PercentCommission, LoyaltyCommission,
)

strategies = [
    FixedCommission(),
    PercentCommission(),
    LoyaltyCommission(),
]

for strategy in strategies:
    commission = strategy.calculate(price)
    print(f"{strategy.__class__.__name__}: {commission:,.0f} ₽".replace(",", " "))
```

**Вывод:**

```
FixedCommission: 50 000 ₽
PercentCommission: 255 000 ₽
LoyaltyCommission: 127 500 ₽
```

## 3. `src/real_estate/services/notification_service.py`

### Теория

**Observer** (глава 26) — паттерн, при котором издатель уведомляет подписчиков. Издатель не знает, кто именно подписан: он знает только, что у подписчика есть метод `update`.

**Два уровня.** В этом файле два разных понятия. Первое — `NotificationService`, который **управляет подписчиками** и уведомляет их. Второе — `EmailNotifier` и `SmsNotifier`, которые **отправляют сообщения** конкретным способом.

**Разница.** `NotificationService` — это издатель для Observer. `EmailNotifier` и `SmsNotifier` — это реализации интерфейса `Notifier` для отправки сообщений.

**Зачем оба.** Издатель уведомляет подписчиков (клиентов). Но иногда нужно отправить служебное сообщение: например, при продаже квартиры. Для этого есть `Notifier`.

**Инверсия зависимостей.** `NotificationService` не зависит от конкретного класса `EmailNotifier`. Он зависит от абстракции `Notifier`. Это DIP (глава 40).

### Задание

Создайте класс `NotificationService` — управление подписчиками и уведомления. Создайте классы `EmailNotifier` и `SmsNotifier` — реализации интерфейса `Notifier`.

### Полный код

```python
"""
Сервис уведомлений.

Демонстрирует:
- Observer (глава 26) — управление подписчиками
- DIP (глава 40) — зависимость от абстракции Notifier
"""

from ..interfaces import Notifier


class NotificationService:
    """
    Управляет подписчиками и уведомляет их.

    Издатель в паттерне Observer. Не знает конкретных классов
    подписчиков: важно только, что у них есть метод update().
    """

    def __init__(self):
        self._subscribers = []

    def subscribe(self, observer) -> None:
        """Подписывает наблюдателя."""
        if observer not in self._subscribers:
            self._subscribers.append(observer)

    def unsubscribe(self, observer) -> None:
        """Отписывает наблюдателя."""
        if observer in self._subscribers:
            self._subscribers.remove(observer)

    def notify(self, event) -> None:
        """
        Уведомляет всех подписчиков.

        Итерируется по копии списка — это защита от изменения
        списка во время обхода.
        """
        for observer in list(self._subscribers):
            observer.update(event)

    def count(self) -> int:
        """Количество подписчиков."""
        return len(self._subscribers)


class EmailNotifier(Notifier):
    """Уведомитель через электронную почту."""

    def send(self, message: str) -> None:
        """Отправляет сообщение (печатает)."""
        print(f"[Email] {message}")


class SmsNotifier(Notifier):
    """Уведомитель через SMS."""

    def send(self, message: str) -> None:
        """Отправляет сообщение (печатает)."""
        print(f"[SMS] {message}")
```

### Разбор

**`NotificationService`** — издатель. Хранит список подписчиков. Уведомляет их через `notify`.

**`subscribe`** — проверяет, что подписчик ещё не подписан. Иначе один и тот же клиент получал бы уведомления дважды.

**`unsubscribe`** — удаляет подписчика, если он есть.

**`notify`** — итерируется по **копии** списка (`list(self._subscribers)`). Это важно: если подписчик отпишется во время уведомления, оригинальный список изменится, и цикл сломается.

**`EmailNotifier`** и **`SmsNotifier`** — реализации интерфейса `Notifier`. У них есть метод `send`, который принимает сообщение. Способ отправки разный: email и SMS.

**Зачем два интерфейса.** `NotificationService` — издатель для Observer. `Notifier` — интерфейс для отправки сообщений. Это разные задачи, разные сервисы.

```python
)

service.subscribe(client1)
service.subscribe(client2)

print()
print("Отписываем Петрова...")

print("Уведомления Иванова:", client1.notifications)
print("Уведомления Петрова:", client2.notifications)
```

## 4. `src/real_estate/services/deal_service.py`

### Теория

**`DealService`** — сервис, который оформляет сделки. Он отвечает только за сделки: не хранит объекты, не отправляет уведомления, не формирует отчёты. Только продаёт.

**Зависимости.** Сервис получает через конструктор три зависимости: хранилище, уведомитель, стратегию комиссии. Это **инверсия зависимостей** (глава 40).

**Почему через конструктор.** Сервис не создаёт зависимости сам. Он их получает. Это позволяет легко подменить реализации: в тестах использовать mock, в проде — реальные классы.

**Работа с исключениями.** Если квартира не найдена, `storage.find_by_address` выбросит `ApartmentNotFoundError`. Если уже продана, сервис выбросит `ApartmentAlreadySoldError`. Клиент может поймать и обработать эти ошибки.

**Проверка интерфейса.** Перед продажей сервис проверяет, реализует ли объект интерфейс `Sellable`. Это ISP (глава 40): продавать можно только то, что продаётся.

**Возврат результата.** Метод `sell` возвращает словарь с информацией о сделке: объект, покупатель, комиссия. Это позволяет клиенту использовать результат.

### Задание

Создайте класс `DealService`. Он принимает `storage`, `notifier`, `commission`. Реализуйте метод `sell(address, buyer)`, который находит объект, проверяет, что он продаётся, помечает проданным, отправляет уведомление и возвращает информацию о сделке.

### Полный код

```python
"""
Сервис оформления сделок.

Отвечает только за сделки (SRP, глава 27).
Зависит от абстракций: Storage, Notifier, CommissionStrategy (DIP, глава 40).
"""

from ..interfaces import Sellable, Notifier
from ..exceptions import ApartmentAlreadySoldError


class DealService:
    """
    Сервис оформления сделок.

    Зависимости передаются через конструктор. Сервис не создаёт
    их сам: это позволяет подменять реализации.
    """

    def __init__(self, storage, notifier: Notifier, commission=None):
        self.storage = storage
        self.notifier = notifier
        self.commission = commission

    def sell(self, address: str, buyer: str) -> dict:
        """
        Продаёт объект недвижимости по адресу.

        Шаги:
        1. Найти объект в хранилище.
        2. Проверить, что он не продан.
        3. Проверить, что он продаётся (интерфейс Sellable).
        4. Пометить проданным.
        5. Рассчитать комиссию (если стратегия задана).
        6. Отправить уведомление.
        7. Вернуть информацию о сделке.
        """
        item = self.storage.find_by_address(address)

        if getattr(item, "is_sold", False):
            raise ApartmentAlreadySoldError(address)

        if not isinstance(item, Sellable):
            raise TypeError(f"{address} не продаётся")

        item.sell()
        item.is_sold = True

        commission = (
            self.commission.calculate(item.price)
            if self.commission else 0
        )

        self.notifier.send(f"Продано: {address} → {buyer}")

        return {
            "item": item,
            "buyer": buyer,
            "commission": commission,
        }
```

### Разбор

**`__init__`** принимает три зависимости. Все они могут быть заменены на другие реализации. `commission` необязателен: если не передан, комиссия равна нулю.

**`self.storage.find_by_address(address)`** — поиск в хранилище. Если объекта нет, выбрасывается `ApartmentNotFoundError`. Сервис не ловит это исключение — оно уходит клиенту. Клиент сам решает, что делать.

**`getattr(item, "is_sold", False)`** — безопасное чтение. Если у объекта нет `is_sold`, считаем, что он не продан.

**`raise ApartmentAlreadySoldError(address)`** — своё исключение. Клиент поймёт, что объект уже продан, а не потерян.

**`isinstance(item, Sellable)`** — проверка интерфейса. Дом сдаётся? Нет. Значит, `House` может не реализовывать `Sellable`. Но в нашем проекте он реализует. Проверка защищает от попытки продать то, что не продаётся.

**`item.sell()`** — вызываем метод интерфейса. Конкретный класс сам решает, что делать.

**`self.commission.calculate(item.price)`** — если стратегия задана, рассчитываем комиссию. Если нет — 0.

**`self.notifier.send(...)`** — отправляем уведомление. Способ отправки определяет конкретный класс: email, SMS, что угодно.

**`return {...}`** — возвращаем словарь с результатами. Клиент может использовать эти данные дальше.

### Пример использования

```python

    EstateStorage, DealService, NotificationService,
    EmailNotifier, PercentCommission,
)

storage.add(Apartment("ул. Ленина, д. 10", 65, 8_500_000))

notifier = EmailNotifier()
commission = PercentCommission(rate=0.03)

result = service.sell("ул. Ленина, д. 10", "Петров")

print(f"Покупатель: {result['buyer']}")
print(f"Комиссия: {result['commission']:,.0f} ₽".replace(",", " "))
print(f"Продано: {result['item'].is_sold}")
```

```
Покупатель: Петров

```

## 5. `src/real_estate/services/rental_service.py`

### Теория

**`RentalService`** — сервис аренды. По структуре похож на `DealService`, но выполняет другую задачу: сдаёт объект в аренду.

**Почему отдельный сервис.** Аренда и продажа — разные операции. У них разные правила, разные сроки, разная логика. Если бы обе операции были в одном классе, он был бы слишком большим. Это нарушение SRP (глава 27).

**Проверка интерфейса.** Перед сдачей сервис проверяет, реализует ли объект интерфейс `Rentable`. Не всё можно сдать: дом может не сдаваться. Это ISP (глава 40).

**Возврат объекта.** Метод `return_item` возвращает объект в состояние «доступен». Это обратная операция к аренде.

### Задание

Создайте класс `RentalService`. Он принимает `storage` и `notifier`. Реализуйте методы `rent(address, tenant, days)` и `return_item(address)`.

### Полный код

```python
"""
Сервис аренды.

Отвечает только за аренду (SRP, глава 27).
Отделён от DealService, потому что продажа и аренда —
разные бизнес-операции.
"""

from ..interfaces import Rentable, Notifier


class RentalService:
    """
    Сервис аренды недвижимости.

    Зависимости передаются через конструктор (DIP, глава 40).
    """

    def __init__(self, storage, notifier: Notifier):
        self.storage = storage
        self.notifier = notifier

    def rent(self, address: str, tenant: str, days: int) -> dict:
        """
        Сдаёт объект в аренду.

        Шаги:
        1. Найти объект.
        2. Проверить, что он сдаётся (Rentable).
        3. Пометить как арендованный.
        4. Отправить уведомление.
        5. Вернуть информацию об аренде.
        """
        item = self.storage.find_by_address(address)

        if not isinstance(item, Rentable):
            raise TypeError(f"{address} не сдаётся")

        if getattr(item, "is_rented", False):
            raise ValueError(f"{address} уже сдан")

        item.rent()
        item.is_rented = True

        self.notifier.send(f"Сдано в аренду: {address} → {tenant} на {days} дней")

        return {
            "item": item,
            "tenant": tenant,
            "days": days,
        }

    def return_item(self, address: str) -> dict:
        """
        Возвращает объект из аренды.

        Помечает объект как доступный.
        """
        item = self.storage.find_by_address(address)
        item.is_rented = False

        self.notifier.send(f"Возвращено из аренды: {address}")

        return {"item": item}
```

### Разбор

**`RentalService`** и **`DealService`** — два независимых сервиса. Если завтра изменится логика аренды, менять нужно только `RentalService`. `DealService` это не затронет.

**`isinstance(item, Rentable)`** — проверка интерфейса. Квартиры и студии реализуют `Rentable`, потому что их можно сдать. Дом — не реализует (в нашей модели).

**`item.rent()`** — вызывает метод интерфейса. Конкретный класс сам решает, как это делается.

**`getattr(item, "is_rented", False)`** — если у объекта нет флага аренды, считаем, что он не сдан.

**`return_item`** — обратная операция. Просто меняет флаг.

### Пример использования

```python

    EstateStorage, RentalService, EmailNotifier,
)

storage = EstateStorage()
storage.add(Apartment("ул. Ленина, д. 10", 65, 8_500_000))

service = RentalService(storage, EmailNotifier())

result = service.rent("ул. Ленина, д. 10", "Иванов", 30)
print(f"Арендатор: {result['tenant']}, дней: {result['days']}")

service.return_item("ул. Ленина, д. 10")
```

**Вывод:**

```

```

## 6. `src/real_estate/services/report_service.py`

### Теория

**`ReportService`** — сервис отчётов. Он принимает хранилище и список сделок, возвращает текстовый отчёт.

**Почему отдельный сервис.** Формирование отчёта — отдельная задача. Она не связана с хранением или сделками. Это SRP (глава 27).

**Два вида отчёта.** Первый — общий отчёт по всем объектам в хранилище. Второй — отчёт по сделкам: сколько продано, какая суммарная комиссия.

**Формат отчёта.** Отчёт — это строка. Она может быть выведена в консоль или сохранена в файл.

### Задание

Создайте класс `ReportService` с двумя методами: `generate(storage, agency_name)` — общий отчёт, `generate_sales_report(deals)` — отчёт по продажам.

### Полный код

```python
"""
Сервис отчётов.

Отвечает только за формирование отчётов (SRP, глава 27).
Не хранит данные, не оформляет сделки — только генерирует текст.
"""


class ReportService:
    """
    Формирует отчёты о работе агентства.

    Первый метод — отчёт по всем объектам хранилища.
    Второй метод — отчёт по проданным объектам.
    """

    def generate(self, storage, agency_name: str) -> str:
        """
        Формирует общий отчёт по всем объектам.

        Возвращает многострочную строку: название агентства,
        список объектов с их статусом.
        """
        lines = [f"Агентство: {agency_name}"]

        for item in storage.list_all():
            status = "продана" if getattr(item, "is_sold", False) else "в продаже"
            lines.append(f"  - {item.get_info()} [{status}]")

        return "\n".join(lines)

    def generate_sales_report(self, deals: list) -> str:
        """
        Формирует отчёт по сделкам.

        Показывает количество сделок и суммарную комиссию.
        """
        if not deals:
            return "Сделок не было"

        total_commission = sum(deal["commission"] for deal in deals)
        count = len(deals)

        lines = [
            f"Всего сделок: {count}",
            f"Суммарная комиссия: {total_commission:,.0f} ₽".replace(",", " "),
        ]

        for i, deal in enumerate(deals, start=1):
            address = deal["item"].address
            buyer = deal["buyer"]
            commission = deal["commission"]
            lines.append(
                f"  {i}. {address} → {buyer} "
                f"(комиссия: {commission:,.0f} ₽)".replace(",", " ")
            )

        return "\n".join(lines)
```

### Разбор

**`generate`** — общий отчёт. Идёт по всем объектам хранилища и выводит их описание со статусом.

**`getattr(item, "is_sold", False)`** — безопасное чтение статуса. Если у объекта нет флага, считаем, что он не продан.

**`"\\n".join(lines)`** — собираем строки в одну многострочную строку. Это стандартный способ в Python.

**`generate_sales_report`** — отчёт по сделкам. Показывает количество сделок и суммарную комиссию.

**`sum(deal["commission"] for deal in deals)`** — генераторное выражение. Суммирует комиссии всех сделок.

**`enumerate(deals, start=1)`** — нумерует сделки, начиная с 1.

### Пример использования

```python

deals.append(service.sell("ул. Садовая, д. 3", "Сидоров"))

```

**Вывод:**


**Главная идея этапа.** Паттерны — это инструменты. Их не нужно применять везде. Только там, где они решают реальную проблему. В нашем проекте четыре паттерна закрывают четыре конкретные задачи.

**Зачем это нужно.** Без паттернов код становится жёстким: добавление нового типа требует правки существующего. Паттерны делают код гибким: новая функциональность добавляется без изменения старой.

## 1. `src/real_estate/patterns/factory.py`

### Теория

**Factory** (глава 25) — порождающий паттерн. Он скрывает логику создания объектов за единым интерфейсом. Клиент говорит: «создай объект типа X» — и получает готовый объект, не зная, какой именно класс был использован.

**Проблема без фабрики.** Представь, что мы создаём объекты через `if/elif`:

```python
if kind == "apartment":
    return Apartment(...)
elif kind == "house":
    return House(...)
elif kind == "studio":
    return Studio(...)
```

Каждый раз, когда появляется новый тип, приходится править этот код. Это нарушает **OCP** (глава 35): класс должен быть открыт для расширения, но закрыт для изменения.

**Решение через словарь.** Вместо цепочки `if/elif` используем словарь `_registry`. Ключ — строка, значение — класс. Добавление нового типа — одна строка в словаре.

**Декоратор регистрации.** Второй способ — регистрация через декоратор. Класс сам сообщает фабрике о себе. Это ещё гибче: не нужно даже править словарь.

**Зачем это в проекте.** Агентство регулярно пополняется разными типами объектов. Фабрика даёт единый способ их создания.

### Задание

Создайте класс `EstateFactory` со словарём типов. Метод `create(kind, **kwargs)` создаёт объект по строке. Метод `register(kind)` возвращает декоратор для регистрации новых типов. Метод `list_kinds()` возвращает список доступных типов.

### Полный код

```python
"""
Фабрика объектов недвижимости.

Паттерн Factory (глава 25). Скрывает логику создания
объектов за единым интерфейсом. Добавление нового типа
не требует изменения существующего кода (OCP, глава 35).
"""

from ..models import Apartment, Studio, TwoRoom, House, Penthouse


class EstateFactory:
    """
    Фабрика для создания объектов недвижимости.

    Хранит словарь типов. Создаёт объект по строковому
    идентификатору. Позволяет регистрировать новые типы
    через декоратор.
    """

    _registry = {
        "apartment": Apartment,
        "studio": Studio,
        "two_room": TwoRoom,
        "house": House,
        "penthouse": Penthouse,
    }

    @classmethod
    def create(cls, kind: str, **kwargs):
        """
        Создаёт объект указанного типа.

        Если тип неизвестен, выбрасывает ValueError
        с перечислением доступных типов.
        """
        if kind not in cls._registry:
            raise ValueError(
                f"Неизвестный тип: {kind!r}. "
                f"Доступные: {list(cls._registry.keys())}"
            )
        return cls._registry[kind](**kwargs)

    @classmethod
    def register(cls, kind: str):
        """
        Декоратор для регистрации нового типа.

        Использование:
            @EstateFactory.register("villa")
            class Villa(Apartment): ...
        """
        def decorator(klass):
            cls._registry[kind] = klass
            return klass
        return decorator

    @classmethod
    def list_kinds(cls) -> list:
        """Возвращает список зарегистрированных типов."""
        return list(cls._registry.keys())
```

### Разбор

**`_registry`** — словарь с типами. Ключ — строка, значение — класс. Атрибут класса, а не экземпляра: все вызовы используют один и тот же словарь.

**`create`** — классовый метод. Проверяет, что тип известен. Если нет, выбрасывает `ValueError` со списком доступных типов. Так клиент сразу поймёт, что он ошибся.

**`register`** — возвращает декоратор. Декоратор принимает класс и добавляет его в словарь. Возвращает тот же класс — чтобы после регистрации класс можно было использовать как обычно.

**`list_kinds`** — вспомогательный метод. Возвращает список доступных типов. Полезно для CLI: можно показать пользователю, что доступно.

**Почему классовый метод.** Фабрика работает с классом, а не с объектом. Объект фабрики не нужен: словарь общий для всех. Классовый метод даёт доступ к словарю.

### Пример использования

```python
from src.real_estate.patterns import EstateFactory

flat = EstateFactory.create(
    "apartment",
    address="ул. Ленина, д. 10",
    area=65,
    price=8_500_000,
)
print(flat.get_info())

house = EstateFactory.create(
    "house",
    address="ул. Садовая, д. 3",
    area=200,
    floors=2,
)
print(house.get_info())

print(f"Доступные типы: {EstateFactory.list_kinds()}")

try:
    EstateFactory.create("villa")
except ValueError as e:
    print(f"Ошибка: {e}")
```

**Вывод:**

```

Дом ул. Садовая, д. 3: 200 м², 2 этажей
Доступные типы: ['apartment', 'studio', 'two_room', 'house', 'penthouse']
Ошибка: Неизвестный тип: 'villa'. Доступные: ['apartment', 'studio', 'two_room', 'house', 'penthouse']
```


## 2. `src/real_estate/patterns/singleton.py`

### Теория

**Singleton** (глава 24) — порождающий паттерн. Гарантирует, что у класса будет только один экземпляр. Все обращения возвращают один и тот же объект.

**Проблема без Singleton.** В агентстве должен быть один реестр объектов. Если создать два, данные могут разойтись: в одном будут одни квартиры, в другом — другие. Программа начнёт работать неправильно.

**Реализация через метакласс.** Метакласс перехватывает создание объекта. Когда мы пишем `Agency(...)`, Python вызывает не `__init__`, а `__call__` метакласса. Мы перехватываем этот вызов и возвращаем уже существующий объект, если он есть.

**Почему метакласс, а не `__new__`.** Метакласс работает автоматически для всех наследников. Достаточно один раз определить метакласс — и любой класс, использующий его, становится Singleton.

**Зачем это в проекте.** В проекте Singleton применяется к реестру агентства. Реестр — один на всю программу.

**Важно.** `__init__` вызывается только один раз. Метакласс перехватывает создание целиком: если объект уже существует, `__init__` не вызывается повторно.

### Задание

Создайте метакласс `SingletonMeta`, который хранит словарь экземпляров. Метод `__call__` проверяет, есть ли объект для данного класса, и возвращает его или создаёт новый.

### Полный код

```python
"""
Метакласс для паттерна Singleton.

Паттерн Singleton (глава 24). Гарантирует, что у класса
будет только один экземпляр. Метакласс перехватывает
создание объекта через __call__ (глава 23).
"""


class SingletonMeta(type):
    """
    Метакласс, превращающий класс в Singleton.

    Хранит словарь экземпляров по классам. При создании
    объекта проверяет, есть ли уже экземпляр данного класса.
    Если есть — возвращает его. Если нет — создаёт новый.
    """

    _instances = {}

    def __call__(cls, *args, **kwargs):
        """Перехватывает создание объекта."""
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]
```

### Разбор

**`SingletonMeta(type)`** — наследуется от `type`. `type` — стандартный метакласс всех классов. Мы переопределяем его.

**`_instances = {}`** — словарь. Ключ — класс, значение — объект. Это позволяет использовать один метакласс для нескольких классов: каждый класс получит свой единственный экземпляр.

**`__call__(cls, ...)`** — перехватывает `Agency(...)`. Проверяет, есть ли объект для класса. Если нет — создаёт через `super().__call__(...)`. Возвращает объект.

**`super().__call__(...)`** — обычное создание объекта. Вызывает `__new__` и `__init__` класса.

**`_instances` в метаклассе.** Словарь хранится в метаклассе, а не в классе. Это важно: если бы словарь был в классе, наследники делили бы его с родителем. С метаклассом каждый класс получает свой ключ.

### Пример использования

```python
from src.real_estate.patterns import SingletonMeta


class AgencyRegistry(metaclass=SingletonMeta):
    def __init__(self, name="Агентство"):
        self.name = name
        self.items = []


registry1 = AgencyRegistry("Первое")
registry1.items.append("ул. Ленина")

registry2 = AgencyRegistry("Второе")

print(f"registry1 is registry2: {registry1 is registry2}")
print(f"registry2.name: {registry2.name}")
print(f"registry2.items: {registry2.items}")
```

**Вывод:**

```
registry1 is registry2: True
registry2.name: Первое
registry2.items: ['ул. Ленина']
```

### Сравнение с `__new__`

Singleton можно реализовать и через `__new__`. Разница:

| Подход | Плюсы | Минусы |
|--------|-------|--------|
| Метакласс | Автоматически для наследников | Сложнее понять |
| `__new__` | Проще | Нужно переопределять в каждом классе |

В проекте выбран метакласс: он применяется к нескольким классам, и каждый автоматически становится Singleton.

## 3. `src/real_estate/patterns/decorator.py`

### Теория

**Decorator** (глава 38) — структурный паттерн. Он позволяет добавлять поведение объекту, не меняя его класс. Декоратор оборачивает объект и добавляет что-то к его работе.

**Проблема без декоратора.** У квартиры может быть много меток: «с ремонтом», «с мебелью», «ипотека возможна». Комбинаций — 2^n. Если делать подклассы, для 3 меток понадобится 8 подклассов. Для 10 — 1024. Это взрыв классов.

**Решение через декоратор.** Каждая метка — отдельный класс. Они оборачивают друг друга. Один объект можно обернуть в любую комбинацию.

**Базовый декоратор.** `ItemDecorator` содержит внутри себя объект `Infoable` и делегирует ему вызов `get_info()`. Это позволяет оборачивать цепочку.

**Конкретные декораторы.** `RenovatedDecorator`, `FurnishedDecorator`, `MortgageDecorator` дополняют результат родителя. Каждый добавляет свою метку.

**Порядок важен.** `RenovatedDecorator(FurnishedDecorator(flat))` и `FurnishedDecorator(RenovatedDecorator(flat))` дают разный порядок меток. Внутренний декоратор применяется первым.

**Зачем это в проекте.** Агентство добавляет к объектам разные метки. Декораторы позволяют делать это гибко.

### Задание

Создайте базовый класс `ItemDecorator`, наследующий от `Infoable`. Он принимает объект `Infoable` и делегирует ему `get_info()`. Создайте три наследника: `RenovatedDecorator`, `FurnishedDecorator`, `MortgageDecorator`. Каждый добавляет свою метку.

### Полный код

```python
"""
Декораторы для меток объектов недвижимости.

Паттерн Decorator (глава 38). Позволяет добавлять метки
к объекту без изменения его класса. Декораторы
комбинируются в цепочку.
"""

from ..interfaces import Infoable


class ItemDecorator(Infoable):
    """
    Базовый декоратор.

    Содержит обёрнутый объект и делегирует ему вызов
    get_info(). Наследники дополняют результат.
    """

    def __init__(self, item: Infoable):
        self.item = item

    def get_info(self) -> str:
        """Делегирует работу обёрнутому объекту."""
        return self.item.get_info()


class RenovatedDecorator(ItemDecorator):
    """Добавляет метку «с ремонтом»."""

    def get_info(self) -> str:
        return f"{self.item.get_info()} (с ремонтом)"


class FurnishedDecorator(ItemDecorator):
    """Добавляет метку «с мебелью»."""

    def get_info(self) -> str:
        return f"{self.item.get_info()} (с мебелью)"


class MortgageDecorator(ItemDecorator):
    """Добавляет метку «ипотека возможна»."""

    def get_info(self) -> str:
        return f"{self.item.get_info()} (ипотека возможна)"
```

### Разбор

**`ItemDecorator(Infoable)`** — тоже реализует `Infoable`. Это значит, что декоратор можно оборачивать в другой декоратор. Цепочка получается бесконечной.

**`self.item`** — обёрнутый объект. Может быть как обычной квартирой, так и другим декоратором.

**`get_info`** в базовом декораторе — просто делегирует. Наследники вызывают `self.item.get_info()` и **дополняют** результат.

**`self.item.get_info()`** — это ключевой момент. Декоратор не заменяет логику родителя, а **расширяет** её. Каждый уровень цепочки добавляет свою метку.

### Пример использования

```python
from src.real_estate.models import Apartment
from src.real_estate.patterns import (
    RenovatedDecorator, FurnishedDecorator, MortgageDecorator,
)

flat = Apartment("ул. Ленина, д. 10", 65, 8_500_000)

print("Без декораторов:")
print(f"  {flat.get_info()}")

print()
print("Один декоратор:")
print(f"  {RenovatedDecorator(flat).get_info()}")

print()
print("Два декоратора:")
decorated = RenovatedDecorator(FurnishedDecorator(flat))
print(f"  {decorated.get_info()}")

print()
print("Три декоратора:")
full = RenovatedDecorator(FurnishedDecorator(MortgageDecorator(flat)))
print(f"  {full.get_info()}")

print()
print("Смена порядка:")
reversed_order = FurnishedDecorator(RenovatedDecorator(flat))
print(f"  {reversed_order.get_info()}")
```

**Вывод:**

```
Без декораторов:
  Квартира ул. Ленина, д. 10: 65 м², 8 500 000 ₽

Один декоратор:
  Квартира ул. Ленина, д. 10: 65 м², 8 500 000 ₽ (с ремонтом)

Два декоратора:
  Квартира ул. Ленина, д. 10: 65 м², 8 500 000 ₽ (с мебелью) (с ремонтом)

Три декоратора:
  Квартира ул. Ленина, д. 10: 65 м², 8 500 000 ₽ (ипотека возможна) (с мебелью) (с ремонтом)

Смена порядка:
  Квартира ул. Ленина, д. 10: 65 м², 8 500 000 ₽ (с ремонтом) (с мебелью)
```

**Обратите внимание.** Порядок меток в выводе — от внутреннего декоратора к внешнему. Если обернули `Furnished` внутри `Renovated`, сначала идёт метка мебели, потом ремонта.

## 4. `src/real_estate/patterns/adapter.py`

### Теория

**Adapter** (глава 39) — структурный паттерн. Он позволяет работать с объектами, у которых несовместимые интерфейсы. Адаптер оборачивает один объект и предоставляет другой интерфейс.

**Проблема без адаптера.** Внешняя система возвращает данные в виде словаря с ключами `location`, `size`, `cost`. Наш код ожидает объект `Apartment` с полями `address`, `area`, `price`. Интерфейсы не совпадают.

**Решение через адаптер.** `ApartmentAdapter` принимает словарь и предоставляет свойства `address`, `area`, `price`. Снаружи адаптер выглядит как обычная квартира.

**Почему свойства через `@property`.** Ключи словаря читаются при обращении. Если изменить ключ в словаре, свойство вернёт новое значение. Это делает адаптер динамическим.

**Зачем это в проекте.** Агентство может получать данные из внешней системы. Адаптер приводит их к нашему формату.

**Отличие от Decorator.** Decorator добавляет поведение объекту, сохраняя интерфейс. Adapter меняет интерфейс. Это разные задачи.

### Задание

Создайте класс `ApartmentAdapter`, наследующий от `Infoable`. Он принимает словарь с ключами `location`, `size`, `cost` (или `address`, `area`, `price`). Реализуйте свойства `address`, `area`, `price` и метод `get_info()`.

### Полный код

```python
"""
Адаптер для внешней системы учёта.

Паттерн Adapter (глава 39). Приводит словарь внешней
системы к интерфейсу Apartment: свойства address, area, price
и метод get_info().
"""

from ..interfaces import Infoable


class ApartmentAdapter(Infoable):
    """
    Адаптер внешних данных к интерфейсу Apartment.

    Принимает словарь. Предоставляет свойства address, area, price.
    Поддерживает два набора ключей: внешний (location, size, cost)
    и внутренний (address, area, price).
    """

    def __init__(self, data: dict):
        self._data = data

    @property
    def address(self):
        """Адрес: из location или address."""
        return self._data.get("location") or self._data.get("address")

    @property
    def area(self):
        """Площадь: из size или area."""
        return self._data.get("size") or self._data.get("area")

    @property
    def price(self):
        """Цена: из cost или price."""
        return self._data.get("cost") or self._data.get("price")

    def get_info(self) -> str:
        """Описание в том же формате, что у Apartment."""
        return (
            f"Квартира {self.address}: "
            f"{self.area} м², {self.price:,} ₽"
        ).replace(",", " ")
```

### Разбор

**`self._data`** — словарь внешней системы. Атрибут защищённый: клиент не должен обращаться к нему напрямую.

**`@property`** — свойства, которые читают нужный ключ. Снаружи это выглядит как обычное поле.

**`self._data.get("location") or self._data.get("address")`** — адаптер поддерживает два набора ключей. Если первый ключ отсутствует, используется второй. Это делает адаптер гибким: он работает и с внешней системой, и с нашим форматом.

**`get_info()`** — метод, который возвращает описание в том же формате, что `Apartment`. Снаружи адаптер неотличим от настоящей квартиры.

### Пример использования

```python
from src.real_estate.patterns import ApartmentAdapter
from src.real_estate.models import Apartment


def describe_item(item):
    """Функция работает с любым объектом, у которого есть get_info()."""
    print(item.get_info())


flat = Apartment("ул. Ленина, д. 10", 65, 8_500_000)
describe_item(flat)

external_data = {
    "location": "ул. Гагарина, д. 5",
    "size": 70,
    "cost": 7_500_000,
}
adapted = ApartmentAdapter(external_data)
describe_item(adapted)

print()
print(f"address: {adapted.address}")
print(f"area: {adapted.area}")
print(f"price: {adapted.price}")
```

# 📘 Этап 5. База данных

## Что изучается в этом этапе

База данных — это **хранилище между запусками программы**. Без неё объекты исчезают, как только программа завершается. С базой — сохраняются.

В этом этапе применяются: контекстные менеджеры (глава 33), работа с исключениями (глава 17), сериализация объектов (`to_dict`, `from_dict`), паттерн Repository (по смыслу близок к главе 27 — SRP).

| Файл | Классы | Главы курса |
|------|--------|-------------|
| `schema.sql` | — | SQL |
| `manager.py` | `DatabaseManager` | 17, 33 |
| `repositories.py` | `ApartmentRepository`, `ClientRepository`, `DealRepository` | 27 |
| `database/__init__.py` | экспорт | — |

**Главная идея этапа.** База данных — это отдельный слой. Модели не знают о базе. База не знает о бизнес-логике. Связь — через репозитории.

**Что такое SQLite.** Это встроенная база данных. Она хранится в одном файле. Не требует установки сервера. Идеальна для учебных проектов.

**Что такое репозиторий.** Это класс, который умеет сохранять и загружать объекты одного типа. Репозиторий квартир, репозиторий клиентов, репозиторий сделок.

**Зачем разделять.** Если бы вся работа с базой была в одном классе, он был бы огромным. Репозитории делают код чистым: каждый отвечает за своё.

## 1. `src/real_estate/database/schema.sql`

### Теория

**SQL-схема** — это описание таблиц базы данных. Каждая таблица — это список колонок с типами. Также здесь описываются связи между таблицами.

**Четыре таблицы.** В нашем проекте их четыре.

`apartments` — хранит объекты недвижимости. Колонки: `id`, `address`, `area`, `price`, `is_sold`, `type`. Поле `type` хранит имя класса: `Apartment`, `Studio`, `House`. Это позволяет при загрузке понять, какой объект восстанавливать.

`clients` — хранит клиентов. Колонки: `id`, `name`, `budget`.

`agents` — хранит агентов. Колонки: `id`, `name`, `experience`.

`deals` — хранит сделки. Колонки: `id`, `apartment_id`, `client_id`, `agent_id`, `commission`, `date`. Связи через `FOREIGN KEY` указывают на другие таблицы.

**Зачем `type` в apartments.** В одной таблице хранятся разные классы. При загрузке нужно знать, какой именно. Поле `type` решает эту задачу.

**Зачем `FOREIGN KEY`.** Это ограничение целостности: сделка не может ссылаться на несуществующую квартиру. База сама проверяет это.

**`IF NOT EXISTS`** — таблица создаётся только если её нет. Это позволяет запускать скрипт много раз.

### Задание

Создайте SQL-скрипт с четырьмя таблицами: `apartments`, `clients`, `agents`, `deals`. У каждой — первичный ключ. У таблицы сделок — внешние ключи на квартиры, клиентов и агентов.

### Полный код

```sql
-- Схема базы данных агентства недвижимости.

CREATE TABLE IF NOT EXISTS apartments (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    address TEXT UNIQUE NOT NULL,
    area INTEGER NOT NULL,
    price INTEGER NOT NULL,
    is_sold INTEGER DEFAULT 0,
    type TEXT NOT NULL
);

CREATE TABLE IF NOT EXISTS clients (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    budget INTEGER DEFAULT 0
);

CREATE TABLE IF NOT EXISTS agents (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    experience INTEGER DEFAULT 0
);

CREATE TABLE IF NOT EXISTS deals (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    apartment_id INTEGER NOT NULL,
    client_id INTEGER,
    agent_id INTEGER,
    commission REAL DEFAULT 0,
    date TEXT NOT NULL,
    FOREIGN KEY (apartment_id) REFERENCES apartments(id),
    FOREIGN KEY (client_id) REFERENCES clients(id),
    FOREIGN KEY (agent_id) REFERENCES agents(id)
);
```

### Разбор

**`INTEGER PRIMARY KEY AUTOINCREMENT`** — автоматически увеличивающийся идентификатор. Каждая запись получает уникальный номер.

**`TEXT UNIQUE NOT NULL`** — текстовое поле, уникальное, не может быть пустым. Адрес квартиры уникален: две квартиры не могут иметь один адрес.

**`is_sold INTEGER DEFAULT 0`** — целое число: 0 или 1. В SQLite нет отдельного типа для булева значения.

**`FOREIGN KEY`** — внешний ключ. Ссылается на `id` другой таблицы. Обеспечивает целостность.

## 2. `src/real_estate/database/manager.py`

### Теория

**`DatabaseManager`** — это класс, который управляет соединением с базой данных. Он открывает соединение, создаёт таблицы, выполняет запросы, закрывает соединение.

**Контекстный менеджер** (глава 33). `DatabaseManager` реализует `__enter__` и `__exit__`. Это позволяет использовать его с `with`:

```python
with DatabaseManager("agency.db") as db:
    db.save_apartment(flat)
```

При выходе из блока `with` соединение автоматически закроется. Даже если возникнет ошибка.

**Зачем контекстный менеджер.** Работа с базой требует открытия и закрытия. Без `with` пришлось бы писать `try/finally` каждый раз. `with` делает код короче и безопаснее.

**Что делает `__enter__`.** Открывает соединение и возвращает `self`. Так мы можем работать с объектом через `as db`.

**Что делает `__exit__`.** Закрывает соединение. Параметры `exc_type`, `exc_value`, `traceback` содержат информацию об исключении, если оно было. Возврат `False` означает, что исключение не поглощается.

**Собственные исключения.** Ошибки базы данных оборачиваются в `DatabaseError`. Это наше исключение (глава 17).

**Сохранение объектов.** Метод `save_apartment` принимает объект недвижимости, извлекает его данные через `to_dict` и сохраняет в базу. Работает с любым наследником `Apartment`: `Studio`, `TwoRoom`, `Penthouse`.

**Загрузка объектов.** Метод `find_apartment` возвращает `Apartment` по адресу. Он не восстанавливает точный класс (`Studio` или `House`) — это упрощение учебного проекта. Для восстановления класса можно использовать `EstateFactory`, передав ей поле `type`.

### Задание

Создайте класс `DatabaseManager`. Он открывает соединение, создаёт таблицы, сохраняет и ищет квартиры. Реализуйте контекстный менеджер.

### Полный код

```python
"""
Управление базой данных SQLite.

Класс DatabaseManager открывает соединение, создаёт таблицы,
выполняет запросы и закрывает соединение. Реализует
контекстный менеджер (глава 33), поэтому работает с with.
"""

import sqlite3
from pathlib import Path

from ..exceptions import DatabaseError


class DatabaseManager:
    """
    Управление соединением с базой данных.

    Открывает соединение, создаёт таблицы, выполняет запросы.
    Используется как контекстный менеджер:

        with DatabaseManager("agency.db") as db:
            db.save_apartment(flat)
    """

    def __init__(self, db_path: str = "data/agency.db"):
        self.db_path = db_path
        self._connection = None

    def connect(self) -> None:
        """Открывает соединение с базой данных."""
        try:
            Path(self.db_path).parent.mkdir(parents=True, exist_ok=True)
            self._connection = sqlite3.connect(self.db_path)
            self._connection.row_factory = sqlite3.Row
        except sqlite3.Error as e:
            raise DatabaseError(f"Не удалось открыть БД: {e}")

    def close(self) -> None:
        """Закрывает соединение."""
        if self._connection is not None:
            self._connection.close()
            self._connection = None

    def create_tables(self) -> None:
        """Создаёт таблицы из SQL-схемы."""
        schema_path = Path(__file__).parent / "schema.sql"
        with open(schema_path, encoding="utf-8") as f:
            schema = f.read()

        try:
            self._connection.executescript(schema)
            self._connection.commit()
        except sqlite3.Error as e:
            raise DatabaseError(f"Не удалось создать таблицы: {e}")

    def save_apartment(self, apartment) -> int:
        """
        Сохраняет объект недвижимости.

        Возвращает id записи. Если адрес уже есть,
        обновляет существующую запись.
        """
        data = apartment.to_dict()
        try:
            cursor = self._connection.execute(
                """
                INSERT INTO apartments (address, area, price, is_sold, type)
                VALUES (?, ?, ?, ?, ?)
                ON CONFLICT(address) DO UPDATE SET
                    area = excluded.area,
                    price = excluded.price,
                    is_sold = excluded.is_sold,
                    type = excluded.type
                """,
                (
                    data["address"],
                    data["area"],
                    data["price"],
                    int(data["is_sold"]),
                    data["type"],
                ),
            )
            self._connection.commit()
            return cursor.lastrowid
        except sqlite3.Error as e:
            raise DatabaseError(f"Не удалось сохранить квартиру: {e}")

    def find_apartment(self, address: str):
        """Находит квартиру по адресу. Возвращает строку или None."""
        try:
            cursor = self._connection.execute(
                "SELECT * FROM apartments WHERE address = ?",
                (address,),
            )
            return cursor.fetchone()
        except sqlite3.Error as e:
            raise DatabaseError(f"Ошибка поиска: {e}")

    def find_all_apartments(self) -> list:
        """Возвращает все квартиры."""
        try:
            cursor = self._connection.execute("SELECT * FROM apartments")
            return cursor.fetchall()
        except sqlite3.Error as e:
            raise DatabaseError(f"Ошибка чтения: {e}")

    def delete_apartment(self, address: str) -> bool:
        """Удаляет квартиру по адресу. Возвращает True, если удалено."""
        try:
            cursor = self._connection.execute(
                "DELETE FROM apartments WHERE address = ?",
                (address,),
            )
            self._connection.commit()
            return cursor.rowcount > 0
        except sqlite3.Error as e:
            raise DatabaseError(f"Ошибка удаления: {e}")

    def save_client(self, client) -> int:
        """Сохраняет клиента в базу."""
        try:
            cursor = self._connection.execute(
                "INSERT INTO clients (name, budget) VALUES (?, ?)",
                (client.name, client.budget),
            )
            self._connection.commit()
            return cursor.lastrowid
        except sqlite3.Error as e:
            raise DatabaseError(f"Не удалось сохранить клиента: {e}")

    def save_deal(self, deal: dict) -> int:
        """
        Сохраняет сделку.

        deal — словарь с полями apartment_id, client_id,
        agent_id, commission, date.
        """
        try:
            cursor = self._connection.execute(
                """
                INSERT INTO deals
                    (apartment_id, client_id, agent_id, commission, date)
                VALUES (?, ?, ?, ?, ?)
                """,
                (
                    deal["apartment_id"],
                    deal.get("client_id"),
                    deal.get("agent_id"),
                    deal.get("commission", 0),
                    deal["date"],
                ),
            )
            self._connection.commit()
            return cursor.lastrowid
        except sqlite3.Error as e:
            raise DatabaseError(f"Не удалось сохранить сделку: {e}")

    def __enter__(self):
        """Открывает соединение при входе в with."""
        self.connect()
        self.create_tables()
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        """Закрывает соединение при выходе из with."""
        self.close()
        return False
```

### Разбор

**`sqlite3.connect(self.db_path)`** — открывает файл базы. Если файла нет, он будет создан.

**`self._connection.row_factory = sqlite3.Row`** — настраивает строки так, чтобы к колонкам можно было обращаться по имени: `row["address"]`. Это удобнее кортежей.

**`Path(self.db_path).parent.mkdir(parents=True, exist_ok=True)`** — создаёт папку `data`, если её нет. `parents=True` создаёт все родительские папки. `exist_ok=True` не выбрасывает ошибку, если папка уже есть.

**`self._connection.executescript(schema)`** — выполняет SQL-скрипт с несколькими командами.

**`ON CONFLICT(address) DO UPDATE SET ...`** — вставка с обновлением. Если адрес уже есть, обновляем запись. Это простой способ избежать дублирования.

**`cursor.lastrowid`** — id только что вставленной записи.

**`cursor.rowcount`** — количество изменённых строк. Если 0, значит запись не найдена.

**`__enter__`** — открывает соединение и создаёт таблицы. Возвращает `self`, чтобы работало `with DatabaseManager(...) as db`.

**`__exit__`** — закрывает соединение. Возвращает `False`: исключение не поглощается, а пробрасывается дальше.

**`raise DatabaseError(...)`** — оборачиваем стандартные ошибки SQLite в своё исключение. Это позволяет клиенту ловить все ошибки БД одним типом.

## 3. `src/real_estate/database/repositories.py`

### Теория

**Репозиторий** — это класс, который умеет работать с объектами одного типа. Он скрывает от клиента работу с базой: клиент вызывает `save(apartment)`, а не пишет SQL.

**Зачем репозитории.** `DatabaseManager` знает о SQL. Репозиторий знает о **доменных объектах**. Клиент работает с репозиторием, не задумываясь о базе. Это SRP (глава 27): разные ответственности — разные классы.

**Три репозитория.** В проекте три репозитория: для квартир, для клиентов, для сделок.

**ApartmentRepository.** Умеет сохранять, искать, удалять квартиры. Преобразует объект в словарь (`to_dict`) и обратно (`from_dict`).

**ClientRepository.** Умеет сохранять и искать клиентов.

**DealRepository.** Умеет сохранять и читать сделки.

**Зависимость от DatabaseManager.** Каждый репозиторий принимает `DatabaseManager` в конструкторе. Это инверсия зависимостей (глава 40).

### Задание

Создайте три класса-репозитория. Каждый принимает `DatabaseManager` и работает с одним типом объектов.

### Полный код

```python
"""
Репозитории для работы с сущностями.

Репозиторий — это слой между бизнес-логикой и базой данных.
Он скрывает SQL и предоставляет удобный интерфейс для работы
с объектами домена (SRP, глава 27).
"""

from ..models import Apartment


class ApartmentRepository:
    """
    Репозиторий квартир.

    Сохраняет, ищет и удаляет квартиры через DatabaseManager.
    """

    def __init__(self, manager):
        self.manager = manager

    def save(self, apartment) -> int:
        """Сохраняет квартиру в базу."""
        return self.manager.save_apartment(apartment)

    def find_by_address(self, address: str):
        """Возвращает объект Apartment или None."""
        row = self.manager.find_apartment(address)
        if row is None:
            return None
        return Apartment(
            address=row["address"],
            area=row["area"],
            price=row["price"],
        )

    def find_all(self) -> list:
        """Возвращает список всех квартир."""
        rows = self.manager.find_all_apartments()
        return [
            Apartment(
                address=row["address"],
                area=row["area"],
                price=row["price"],
            )
            for row in rows
        ]

    def delete(self, address: str) -> bool:
        """Удаляет квартиру по адресу."""
        return self.manager.delete_apartment(address)


class ClientRepository:
    """
    Репозиторий клиентов.

    Сохраняет и ищет клиентов.
    """

    def __init__(self, manager):
        self.manager = manager

    def save(self, client) -> int:
        """Сохраняет клиента."""
        return self.manager.save_client(client)


class DealRepository:
    """
    Репозиторий сделок.

    Сохраняет сделки.
    """

    def __init__(self, manager):
        self.manager = manager

    def save(self, deal: dict) -> int:
        """Сохраняет сделку."""
        return self.manager.save_deal(deal)
```

### Разбор

**`ApartmentRepository`** принимает `manager` и делегирует ему все операции. Клиент вызывает `repo.save(flat)` — и не знает, что внутри `manager.save_apartment(flat)`.

**`find_by_address`** — читает строку из базы и создаёт объект `Apartment`. Это **реконструкция** объекта из данных БД.

**`find_all`** — использует генератор списка. Для каждой строки создаётся объект.

**Ограничение учебного проекта.** Репозиторий восстанавливает `Apartment`, а не `Studio` или `House`. Чтобы восстановить точный класс, нужно использовать `EstateFactory` и поле `type`. Мы это упростили для наглядности.

**`ClientRepository`** и **`DealRepository`** — короче. Они не восстанавливают объекты: клиенты и сделки не так важны для демонстрации.

## 1. `src/real_estate/cli/parser.py`

### Теория

**Парсер аргументов** — это объект, который знает, какие команды и аргументы принимает программа. Он создаётся через `argparse.ArgumentParser`.

**Подкоманды.** В `argparse` можно создать подкоманды через `add_subparsers`. Каждая подкоманда — это отдельная команда со своими аргументами.

**Структура команд.** В нашем CLI девять команд.

| Команда | Что делает | Аргументы |
|---------|-----------|-----------|
| `add` | Добавить квартиру | `--address`, `--area`, `--price`, `--type` |
| `list` | Показать все объекты | — |
| `find` | Найти по адресу | `--address` |
| `sell` | Продать | `--address`, `--buyer` |
| `rent` | Сдать в аренду | `--address`, `--tenant`, `--days` |
| `subscribe` | Подписать клиента | `--name`, `--budget` |
| `report` | Сгенерировать отчёт | — |
| `save` | Сохранить в БД | — |
| `load` | Загрузить из БД | — |

**Зачем подкоманды.** Пользователь пишет `python cli.py add --address ...`. `add` — это подкоманда. `--address` — её аргумент.

**Обязательные и необязательные аргументы.** `--address` обязателен: без него команда бессмысленна. `--type` необязателен: по умолчанию `apartment`.

### Задание

Создайте функцию `create_parser()`, которая возвращает `argparse.ArgumentParser` с девятью подкомандами. У каждой подкоманды — свои аргументы.

### Полный код

```python
"""
Парсер аргументов командной строки.

Создаёт объект ArgumentParser с подкомандами.
Каждая подкоманда — отдельная операция агентства.
"""

import argparse


def create_parser() -> argparse.ArgumentParser:
    """
    Создаёт парсер аргументов с девятью подкомандами.

    Возвращает объект ArgumentParser, готовый к разбору
    аргументов из sys.argv.
    """
    parser = argparse.ArgumentParser(
        prog="agency",
        description="Система учёта агентства недвижимости",
    )

    subparsers = parser.add_subparsers(
        dest="command",
        help="Доступные команды",
    )

    # ─── add ───
    add_parser = subparsers.add_parser(
        "add",
        help="Добавить объект недвижимости",
    )
    add_parser.add_argument(
        "--address", required=True, help="Адрес объекта",
    )
    add_parser.add_argument(
        "--area", type=int, required=True, help="Площадь в м²",
    )
    add_parser.add_argument(
        "--price", type=int, required=True, help="Цена",
    )
    add_parser.add_argument(
        "--type",
        default="apartment",
        choices=["apartment", "studio", "two_room", "house", "penthouse"],
        help="Тип объекта",
    )

    # ─── list ───
    subparsers.add_parser("list", help="Показать все объекты")

    # ─── find ───
    find_parser = subparsers.add_parser(
        "find",
        help="Найти объект по адресу",
    )
    find_parser.add_argument(
        "--address", required=True, help="Адрес объекта",
    )

    # ─── sell ───
    sell_parser = subparsers.add_parser(
        "sell",
        help="Продать объект",
    )
    sell_parser.add_argument(
        "--address", required=True, help="Адрес объекта",
    )
    sell_parser.add_argument(
        "--buyer", required=True, help="Имя покупателя",
    )

    # ─── rent ───
    rent_parser = subparsers.add_parser(
        "rent",
        help="Сдать объект в аренду",
    )
    rent_parser.add_argument(
        "--address", required=True, help="Адрес объекта",
    )
    rent_parser.add_argument(
        "--tenant", required=True, help="Имя арендатора",
    )
    rent_parser.add_argument(
        "--days", type=int, required=True, help="Срок аренды в днях",
    )

    # ─── subscribe ───
    subscribe_parser = subparsers.add_parser(
        "subscribe",
        help="Подписать клиента на уведомления",
    )
    subscribe_parser.add_argument(
        "--name", required=True, help="Имя клиента",
    )
    subscribe_parser.add_argument(
        "--budget", type=int, default=0, help="Бюджет клиента",
    )

    # ─── report ───
    subparsers.add_parser("report", help="Сгенерировать отчёт")

    # ─── save ───
    subparsers.add_parser("save", help="Сохранить в БД")

    # ─── load ───
    subparsers.add_parser("load", help="Загрузить из БД")

    return parser
```

### Разбор

**`argparse.ArgumentParser(...)`** — создаёт парсер. Параметр `prog` задаёт имя программы для справки. `description` — описание.

**`add_subparsers(dest="command")`** — создаёт контейнер подкоманд. Параметр `dest="command"` означает, что имя выбранной команды будет доступно как `args.command`.

**`subparsers.add_parser("add", help="...")`** — создаёт подкоманду. Возвращает объект, к которому можно добавлять аргументы.

**`add_argument("--address", required=True)`** — обязательный аргумент. Если его не передать, `argparse` сам выдаст ошибку.

**`type=int`** — автоматическое преобразование. Если пользователь введёт строку, `argparse` выдаст ошибку.

**`default="apartment"`** — значение по умолчанию. Если не указать `--type`, будет `apartment`.

**`choices=[...]`** — список допустимых значений. Если пользователь введёт другое, `argparse` выдаст ошибку.

## 2. `src/real_estate/cli/commands.py`

### Теория

**Обработчики команд** — это методы класса, которые вызываются для каждой команды. Их задача: взять аргументы, вызвать нужный сервис, показать результат.

**Что содержит Commands.** Класс `Commands` создаёт у себя все сервисы: хранилище, уведомитель, сервис сделок, сервис аренды, сервис отчётов. При вызове команды он использует эти сервисы.

**Словарь команд.** Каждая команда соответствует методу. Например, `add` → `cmd_add`, `list` → `cmd_list`. Это позволяет вызвать нужный метод по имени команды через `getattr`.

**Работа с БД.** Команды `save` и `load` работают с базой данных. Они используют `DatabaseManager` и репозиторий.

**Обработка ошибок.** Каждая команда может выбросить исключение. CLI ловит его и показывает пользователю понятное сообщение.

**Зачем разделять парсер и команды.** Парсер знает о командах и аргументах. Команды знают о сервисах. Если парсер изменится, команды не пострадают.

### Задание

Создайте класс `Commands` с методами `cmd_add`, `cmd_list`, `cmd_find`, `cmd_sell`, `cmd_rent`, `cmd_subscribe`, `cmd_report`, `cmd_save`, `cmd_load`. Каждый метод принимает `args` и выполняет свою задачу.

### Полный код

```python
"""
Обработчики команд CLI.

Класс Commands связывает команды из парсера с сервисами
проекта. Каждый метод выполняет одну команду.
"""

from ..patterns import EstateFactory
from ..services import (
    EstateStorage,
    DealService,
    RentalService,
    NotificationService,
    ReportService,
    EmailNotifier,
    PercentCommission,
)
from ..database import DatabaseManager, ApartmentRepository
from ..models import Client
from ..exceptions import AgencyError


class Commands:
    """
    Обработчики команд командной строки.

    Создаёт все сервисы проекта и вызывает их
    в соответствующих методах.
    """

    def __init__(self):
        self.storage = EstateStorage()
        self.notifier = NotificationService()
        self.email = EmailNotifier()
        self.reports = ReportService()
        self.deals = DealService(
            self.storage, self.email, PercentCommission()
        )
        self.rentals = RentalService(self.storage, self.email)
        self.clients = []
        self.db_path = "data/agency.db"

    def _get_db(self):
        """Возвращает контекстный менеджер для работы с БД."""
        return DatabaseManager(self.db_path)

    # ─── add ───
    def cmd_add(self, args) -> None:
        """Добавляет новый объект недвижимости."""
        try:
            item = EstateFactory.create(
                args.type,
                address=args.address,
                area=args.area,
                price=args.price,
            )
            self.storage.add(item)
            print(f"✓ Добавлено: {item.get_info()}")
        except (ValueError, TypeError) as e:
            print(f"✗ Ошибка: {e}")

    # ─── list ───
    def cmd_list(self, args) -> None:
        """Показывает все объекты в хранилище."""
        if len(self.storage) == 0:
            print("Хранилище пусто")
            return

        print(f"Всего объектов: {len(self.storage)}")
        for i, item in enumerate(self.storage, start=1):
            print(f"  {i}. {item.get_info()}")

    # ─── find ───
    def cmd_find(self, args) -> None:
        """Ищет объект по адресу."""
        try:
            item = self.storage.find_by_address(args.address)
            print(f"Найдено: {item.get_info()}")
        except AgencyError as e:
            print(f"✗ Ошибка: {e}")

    # ─── sell ───
    def cmd_sell(self, args) -> None:
        """Продаёт объект по адресу."""
        try:
            result = self.deals.sell(args.address, args.buyer)
            commission = result["commission"]
            print(f"✓ Продано: {args.address} → {args.buyer}")
            print(f"  Комиссия: {commission:,.0f} ₽".replace(",", " "))
        except AgencyError as e:
            print(f"✗ Ошибка: {e}")

    # ─── rent ───
    def cmd_rent(self, args) -> None:
        """Сдаёт объект в аренду."""
        try:
            result = self.rentals.rent(
                args.address, args.tenant, args.days
            )
            print(
                f"✓ Сдано: {args.address} → {args.tenant} "
                f"на {result['days']} дней"
            )
        except (AgencyError, TypeError, ValueError) as e:
            print(f"✗ Ошибка: {e}")

    # ─── subscribe ───
    def cmd_subscribe(self, args) -> None:
        """Подписывает клиента на уведомления."""
        client = Client(args.name, args.budget)
        self.notifier.subscribe(client)
        self.clients.append(client)
        print(f"✓ Подписан: {client.describe()}")
        print(f"  Подписчиков: {self.notifier.count()}")

    # ─── report ───
    def cmd_report(self, args) -> None:
        """Генерирует отчёт по хранилищу."""
        report = self.reports.generate(self.storage, "Агентство")
        print(report)

    # ─── save ───
    def cmd_save(self, args) -> None:
        """Сохраняет объекты из хранилища в базу."""
        if len(self.storage) == 0:
            print("Нечего сохранять")
            return

        try:
            with self._get_db() as db:
                repo = ApartmentRepository(db)
                count = 0
                for item in self.storage:
                    repo.save(item)
                    count += 1
            print(f"✓ Сохранено объектов: {count}")
        except AgencyError as e:
            print(f"✗ Ошибка: {e}")

    # ─── load ───
    def cmd_load(self, args) -> None:
        """Загружает объекты из базы в хранилище."""
        try:
            with self._get_db() as db:
                repo = ApartmentRepository(db)
                items = repo.find_all()
                for item in items:
                    self.storage.add(item)
            print(f"✓ Загружено объектов: {len(items)}")
        except AgencyError as e:
            print(f"✗ Ошибка: {e}")
```

**`_get_db`** возвращает контекстный менеджер. Отдельный метод упрощает создание БД в разных командах.

**`cmd_add`** использует `EstateFactory.create`. Если тип неизвестен, `ValueError` ловится и показывается пользователю.

**`cmd_sell`** вызывает сервис сделок. Ловит `AgencyError`.

**`cmd_rent`** вызывает сервис аренды. Ловит `AgencyError`, `TypeError`, `ValueError`: сервис может выбросить разные исключения.

## 4. `cli.py` (в корне проекта)

### Теория

**Точка входа** — это скрипт, который запускает CLI. Пользователь пишет `python cli.py add ...` — этот скрипт вызывается.

**Что делает скрипт.** Он выполняет три шага. Первый: создаёт парсер. Второй: разбирает аргументы из командной строки. Третий: вызывает нужный метод `Commands` по имени команды.

**Динамический вызов.** Используется `getattr(commands, f"cmd_{args.command}")`. Если команда `add`, вызывается `cmd_add`. Это позволяет не писать длинный `if/elif`.

**Обработка отсутствия команды.** Если пользователь запустил скрипт без аргументов, показывается справка.

### Задание

Создайте скрипт `cli.py`, который создаёт парсер, разбирает аргументы и вызывает нужный обработчик.

### Полный код

```python
"""
Точка входа для командной строки.

Запуск:

    python cli.py add --address "ул. Ленина" --area 65 --price 8500000
    python cli.py list
    python cli.py report
"""

import sys
from pathlib import Path

sys.path.insert(0, str(Path(__file__).parent))

from src.real_estate.cli import create_parser, Commands


def main():
    """Точка входа CLI."""
    parser = create_parser()
    args = parser.parse_args()

    if not args.command:
        parser.print_help()
        return

    commands = Commands()
    handler = getattr(commands, f"cmd_{args.command}", None)

    if handler is None:
        print(f"✗ Неизвестная команда: {args.command}")
        parser.print_help()
        return

    handler(args)


if __name__ == "__main__":
    main()
```

### Разбор

**`sys.path.insert(0, ...)`** — добавляет корень проекта в путь поиска модулей. Это нужно, чтобы импорт `from src.real_estate.cli` работал.

**`parser.parse_args()`** — разбирает аргументы из `sys.argv`. Возвращает объект `args` со всеми атрибутами.

**`if not args.command`** — проверка, что пользователь передал команду. Если нет, показываем справку.

**`Commands()`** — создаёт объект команд. При этом создаются все сервисы.

**`getattr(commands, f"cmd_{args.command}", None)`** — динамически получает метод. Например, для `add` получает `cmd_add`. Если метода нет, вернёт `None`.

**`handler(args)`** — вызывает метод с аргументами.

**`if __name__ == "__main__"`** — стандартная проверка. Скрипт запускается только при прямом вызове, не при импорте.

# 📘 Этап 7. Тесты

## Что изучается в этом этапе

Тесты — это **проверка работы кода**. Без тестов мы не знаем, работает ли программа. С тестами — уверены в каждом изменении.

В этом этапе применяются: `pytest`, фикстуры, изоляция тестов, проверка исключений, временные файлы, параметризация.

| Файл | Что содержит | Что проверяет |
|------|--------------|---------------|
| `conftest.py` | Общие фикстуры | Подготовка данных |
| `test_models.py` | Тесты моделей | Создание, методы, счётчик |
| `test_services.py` | Тесты сервисов | Хранилище, сделки, уведомления |
| `test_patterns.py` | Тесты паттернов | Factory, Singleton, Decorator, Adapter |
| `test_database.py` | Тесты базы данных | Сохранение, загрузка, удаление |
| `test_cli.py` | Тесты CLI | Парсер, обработчики |
| `tests/__init__.py` | Пакет | — |

**Главная идея этапа.** Каждый тест проверяет **одну вещь**. Тесты **не зависят** друг от друга. Если один упал — остальные всё равно работают.

**Что такое pytest.** Это стандартный фреймворк для тестирования в Python. Он ищет файлы `test_*.py`, функции `test_*`, классы `Test*`.

**Что такое фикстура.** Это функция, которая подготавливает данные для тестов. Например, создаёт квартиру для проверки. Фикстуры объявляются декоратором `@pytest.fixture`.

**Зачем изоляция.** Если тесты делят данные, один тест может повлиять на другой. Изоляция гарантирует, что каждый тест работает с чистыми данными.

## 1. `tests/conftest.py`

### Теория

**`conftest.py`** — это специальный файл. `pytest` находит его автоматически. Фикстуры, объявленные здесь, доступны во всех тестах без дополнительных импортов.

**Изоляция.** Фикстура выполняется перед каждым тестом. Это гарантирует, что данные чистые.

```python
    print()
    print("=" * 60)
    print(title)
    print("=" * 60)


def main() -> None:
    """Основная демонстрация проекта."""

    section("1. Фабрика создаёт объекты")

    flat = EstateFactory.create(
        "apartment",
        address="ул. Ленина, д. 10, кв. 5",

    )
    house = EstateFactory.create(
        "house",
        address="ул. Садовая, д. 3",
        area=200,
        floors=2,
    )
    penthouse = EstateFactory.create(
        "penthouse",
        address="ул. Мира, д. 1, кв. 100",
        area=150,
        price=50_000_000,
        terrace=30,
    )

    for item in [flat, studio, house, penthouse]:
        print(f"  {item.get_info()}")

    section("2. Хранилище")

    storage = EstateStorage()
    for item in [flat, studio, house, penthouse]:
        storage.add(item)

    print(f"  Добавлено объектов: {len(storage)}")

    found = storage.find_by_address("ул. Ленина, д. 10, кв. 5")
    print(f"  Найдено: {found.get_info()}")

    section("3. Уведомления (Observer)")

    notifier = NotificationService()

    notifier.subscribe(client)
    print(f"  Подписчиков: {notifier.count()}")

    section("4. Продажа квартиры (Strategy)")

    deal_service = DealService(
        storage, EmailNotifier(), PercentCommission()
    )
    result = deal_service.sell("ул. Ленина, д. 10, кв. 5", "Петров")
    print(f"  Покупатель: {result['buyer']}")
    print(f"  Комиссия: {result['commission']:,.0f} ₽".replace(",", " "))

    section("5. Отчёт")

    reports = ReportService()
    print(reports.generate(storage, "Дом-Сервис"))

    section("6. Декораторы")

    decorated = RenovatedDecorator(FurnishedDecorator(flat))
    print(f"  {decorated.get_info()}")

    section("7. Адаптер")

    external = ApartmentAdapter({
        "location": "ул. Гагарина, д. 5",
        "size": 70,
        "cost": 7_500_000,

    print(f"  {external.get_info()}")

    section("8. Собственные исключения")

    try:
        storage.find_by_address("несуществующий адрес")
    except ApartmentNotFoundError as e:

    section("9. Классовый счётчик")

    print(f"  Всего квартир: {Apartment.count}")

    price = 8_500_000
    for strategy in [

        commission = strategy.calculate(price)
        name = strategy.__class__.__name__
        print(f"  {name}: {commission:,.0f} ₽".replace(",", " "))

    section("ГОТОВО")


if __name__ == "__main__":
    main()
```

```python
Демо сервисов и паттернов.

from pathlib import Path

    RenovatedDecorator,
    FurnishedDecorator,

    DealService,
    RentalService,
    NotificationService,
    ReportService,
    EmailNotifier,
    SmsNotifier,
    PercentCommission,
    FixedCommission,
    LoyaltyCommission,

    print()
    print("=== Декораторы ===")

    decorated = RenovatedDecorator(FurnishedDecorator(studio))
    print(f"  {decorated.get_info()}")

    print()
    print("=== Адаптер ===")

    external = ApartmentAdapter({
        "location": "ул. Гагарина, д. 5",
        "size": 70,
        "cost": 7_500_000,
    })
    print(f"  {external.get_info()}")

    print()
    print("=== Стратегии комиссии ===")

    price = 8_500_000
    for strategy in [
        FixedCommission(),
        PercentCommission(),
        LoyaltyCommission(),
    ]:
        commission = strategy.calculate(price)
        name = strategy.__class__.__name__
        print(f"  {name}: {commission:,.0f} ₽".replace(",", " "))


if __name__ == "__main__":
```

## 7. `requirements.txt`

### Теория

**Зависимости** — это внешние пакеты, которые нужны проекту. В нашем случае только `pytest` для тестирования.

**Зачем файл.** Чтобы другой человек мог установить всё одной командой: `pip install -r requirements.txt`.

**Формат.** Каждая строка — имя пакета и версия.

### Задание

Создайте файл с зависимостями.

### Полный код

```

```

### Разбор

**`pytest>=7.4.0`** — версия 7.4.0 или новее. Это позволяет использовать новые возможности, но не ломает совместимость.

**Больше ничего не нужно.** Стандартная библиотека Python покрывает всё остальное: `dataclasses`, `abc`, `typing`, `sqlite3`, `argparse`, `logging`, `copy`, `datetime`.

## 8. `setup.py`

### Теория

**`setup.py`** — файл для установки пакета. Позволяет установить проект командой `pip install -e .`.

**Что содержит.** Метаданные: имя, версия, описание, зависимости, точка входа.

**Зачем это нужно.** В учебном проекте не обязательно. Но это показывает, как правильно оформлять пакет.

### Задание

Создайте `setup.py` с метаданными проекта.

### Полный код

```python
"""
Установка пакета real_estate.

Установка:

    pip install -e .
"""

from setuptools import setup, find_packages


setup(
    name="real-estate-agency",
    version="1.0.0",
    description="Учебный проект: система учёта агентства недвижимости",
    author="Student",
    packages=find_packages(where="src"),
    package_dir={"": "src"},
    python_requires=">=3.10",
    install_requires=[],
    extras_require={
        "test": ["pytest>=7.4.0"],
    },
)
```

### Разбор

**`name`** — имя пакета. Используется при установке.

**`version`** — версия проекта.

**`packages=find_packages(where="src")`** — автоматически находит все пакеты в папке `src`.

**`package_dir={"": "src"}`** — указывает, что исходники лежат в `src`.

**`python_requires=">=3.10"`** — минимальная версия Python.

**`install_requires=[]`** — обязательных зависимостей нет.

**`extras_require={"test": [...]}`** — дополнительные зависимости для тестов. Устанавливаются командой `pip install -e ".[test]"`.

### Использование

```bash
# Установка в режиме разработки
pip install -e .

# Установка с тестами
pip install -e ".[test]"

# После установки можно импортировать из любого места
python -c "from real_estate.models import Apartment; print('OK')"
```
