## 🔵 Уровень 2. Условия, boolean, логика, `is` vs `==`, truthiness (9–16)

**Что изучается:** `if/elif/else`, тернарный оператор, цепочки сравнений, `and/or/not`, приоритет, `bool`, falsy-значения, `is` vs `==`, `is None`, короткое замыкание, `in`.

**Задача 9 (if/elif/else).** Библиотека делит читателей по залам в зависимости от возраста, и это классическая задача на цепочку условий. Запросите возраст. В комментариях поясните: зачем `elif`, а не три отдельных `if`; почему порядок условий важен; что будет, если первое условие написать `age < 18`. Подсказка: `if age < 14: ... elif age < 18: ... else: ...`.

**Задача 10 (truthiness).** В Python почти всё можно использовать в `if`, и важно понимать, какие значения считаются «ложными». **Задача 11 (тернарный, цепочка, `in`).** Три идиомы Python, которые делают код короче и читаемее. (1) Тернарный оператор: `status = "есть" if count > 0 else "нет"`. **Задача 12 (and/or/not + приоритет).** Библиотека выдаёт редкую книгу только при особых условиях, и здесь важно не запутаться в логике. **Задача 14 (try/except в условии).** Прежде чем проверить год, нужно убедиться, что пользователь ввёл число, — иначе `int()` бросит исключение. Запросите год. Если `int()` падает — «не число». **Задача 16 (`is` vs `==`, `None`).** Одна из самых частых ошибок новичков — путать `is` и `==`. Продемонстрируйте: `a = [1, 2]; b = [1, 2]; a == b → True`, `a is b → False`. Покажите, что `None` проверяют через `is None`, а не `== None`. Покажите, что для маленьких `int` и `str` `is` иногда даёт `True` из-за интернирования (`a = 100; b = 100; a is b → True`, но `a = 1000; b = 1000` — уже не факт). ## 🟣 Уровень 3. Циклы, серии, накопление, `any/all` (17–24)

**Что изучается:** `for/while`, `range`, счётчики, накопители, `min/max` с `key`, серии, `break/continue`, `else` у цикла, вложенные циклы, `any/all`, `sum`, `join`, форматированные таблицы.

**Задача 17 (`for` + `range`).** Библиотека нумерует новые книги, и нужно вывести их инвентарные номера. Введите `N` — количество новых книг. Выведите номера от 1 до `N` в столбик. Подсказка: `for i in range(1, n + 1): print(i)`.

**Задача 18 (`for` + накопитель).** Библиотека считает общую сумму штрафов и хочет знать, сколько читателей — «злостные» должники. Введите `N`, затем `N` сумм штрафов. Выведите общую сумму и число штрафов `> 500 ₽`. Ожидаемый вывод при `100, 600, 300`: `Итого: 1000 ₽`, `Больше 500: 1`. В комментариях поясните: зачем нужны накопитель (`total`) и счётчик (`count`); почему нельзя посчитать сумму «в голове» без цикла; что будет, если `N = 0`. Подсказка: `total += fine; if fine > 500: count += 1`.

**Задача 19 (`any/all`).** Иногда не нужно считать — нужно просто узнать «есть ли хоть одна» или «все ли». Для списка книг проверьте: (а) есть ли хоть одна после 2000 года (`any`); (б) все ли старше 1900 (`all`); (в) есть ли книга с названием «X» (`any` с условием). Ожидаемый вывод: `True/False` для каждого. В комментариях поясните: как `any` и `all` используют короткое замыкание; почему `any` останавливается на первом `True`; почему `any(b.year > 2000 for b in books)` — генератор, а не список (экономия памяти). Подсказка: `any(b["year"] > 2000 for b in books)`.

**Задача 20 (`min/max/sorted` с `key`).** В библиотеке нужно быстро находить самую старую, самую новую и самую длинную по названию книгу. Среди `N` книг (у каждой — `title` и `year`) найдите: самую старую (`min` с `key`), самую новую (`max` с `key`), книгу с самым длинным названием (`max` с `key=len`), топ-3 по году (`sorted` + срез). Ожидаемый вывод: `1869`, `2010`, `«Война и мир и мир»`, топ-3. В комментариях поясните: что делает `key=`; почему `min(books)` без `key` упадёт (нет сравнения словарей по смыслу); почему `key=lambda b: b["year"]` читается лучше, чем ручной цикл. Подсказка: `min(books, key=lambda b: b["year"])`.

**Задача 21 (`for` + среднее).** Библиотека анализирует сроки выдачи книг и хочет видеть статистику. Введите `N` — количество записей, затем `N` сроков в днях. Выведите сумму, среднее, минимум, максимум. Проверьте `N > 0` (иначе деление на ноль). Ожидаемый вывод при `30, 14, 7`: `Сумма: 51`, `Среднее: 17.00`, `Мин: 7`, `Макс: 30`. В комментариях поясните: почему `sum / count` даёт `float`; почему нельзя считать среднее как `(min + max) / 2`; что делать при `N = 0`. Подсказка: `total / count` в конце.

**Задача 22 (серия).** Библиотека хочет знать, какая серия одинаковых жанров идёт подряд дольше всего. Введите строку жанров через пробел. Найдите длину самой длинной серии **подряд идущих** одинаковых жанров. Ожидаемый вывод при `роман роман роман детектив детектив`: `Максимальная серия: 3`. В комментариях поясните: зачем два счётчика — `current` (текущая серия) и `best` (лучшая); почему нельзя просто посчитать частоты через `set` (речь о **подряд**); что будет на `роман детектив роман`. Подсказка: сравнивайте текущий элемент с предыдущим.

**Задача 23 (`while` + `break/continue/else`).** Библиотека ищет книгу в каталоге, и нужно не просто найти, а показать разные сценарии выхода из цикла. Ищите книгу в списке. Если нашли — `break` и выведите `найдено`. Если цикл закончился **без** `break` — `else` печатает `не найдено`. Дополнительно: если книга — `None` или пустая строка, `continue` (пропустить). Продемонстрируйте оба сценария — с найденной и ненайденной книгой. Ожидаемый вывод: `найдено` / `не найдено`. В комментариях поясните: зачем `else` у цикла (он срабатывает, только если не было `break`); чем `continue` отличается от `break`; когда `while True` + `break` читается лучше, чем условие в `while`. Подсказка: `for b in books: if b == target: print("найдено"); break\nelse: print("не найдено")`.

**Что изучается:** `dict`, `get/items/keys/values`, `set` (все операции), `tuple`, хешируемость, `Counter`, `defaultdict`, `namedtuple`, `deque`, `itertools` (chain, combinations, permutations, product, groupby), сортировка словаря.

**Задача 49 (dict + get).** Словарь — основная структура для пар «ключ-значение», и важно уметь безопасно читать. Продемонстрируйте: `d["x"]` → `KeyError`; `d.get("x", 0)` → `0`. Объясните, когда безопасный `get` предпочтительнее. Итерируйтесь по `items()`, `keys()`, `values()` и покажите разницу. Ожидаемый вывод: ошибка vs `0` + три варианта итерации. В комментариях поясните: почему `d.get(k, default)` — стандартный приём для счётчиков; чем `items()` удобнее `keys()` + `d[k]`; почему `for k in d:` итерирует по ключам. Подсказка: `for k, v in d.items(): print(k, v)`.

**Задача 50 (счётчик через dict).** Подсчёт частот — классическая задача, и её надо уметь делать через словарь. Посчитайте частоту жанров через `dict.get(g, 0) + 1`. Выведите все жанры с частотами и самый частый. Ожидаемый вывод при `[роман, роман, детектив]`: `роман: 2`, `детектив: 1`, `самый частый: роман`. В комментариях поясните: почему `d[g] = d.get(g, 0) + 1` — идиома; что будет, если написать `d[g] += 1` без `get` (KeyError); как найти максимум по значению (`max(items, key=lambda kv: kv[1])`). Подсказка: `counter = {}; for g in genres: counter[g] = counter.get(g, 0) + 1`.

**Задача 51 (словарь + max по значению).** Словарь `{reader: [книги]}` хранит, кто сколько книг взял, и нужно найти самого активного. Выведите читателя с наибольшим числом книг. Ожидаемый вывод: `Иванов: 3`. В комментариях поясните: почему `max(data, key=...)` без `key` сравнивает ключи, а не значения; как `lambda kv: len(kv[1])` это исправляет; что делать при равенстве (первый попавшийся). Подсказка: `max(data.items(), key=lambda kv: len(kv[1]))`.

**Задача 52 (set: все операции).** Множества — мощный инструмент для работы с уникальными элементами. Продемонстрируйте все операции над `set`: `&` (пересечение), `|` (объединение), `-` (разность), `^` (симметричная разность), `issubset`, `issuperset`. Ожидаемый вывод: `{роман}`, `{роман, детектив, поэма}`, `{детектив}`, `{детектив, поэма}`. В комментариях поясните: чем `set` отличается от `frozenset` (последний хешируем); почему `set` быстрее `list` для проверки `in` (O(1) vs O(n)); почему порядок не гарантирован. Подсказка: `a & b`, `a | b`, `a - b`, `a ^ b`.

**Задача 53 (хешируемость).** Не всё можно положить в `set` или использовать как ключ словаря — и важно понимать почему. Покажите: список нельзя ключом (`TypeError: unhashable type: 'list'`), кортеж — можно. Объясните, как это связано с изменяемостью: хеш должен быть **стабильным**, а изменяемый объект его меняет. Ожидаемый вывод: `hash(())` работает, `hash([])` падает. В комментариях поясните: что такое хеш; почему изменяемые объекты не хешируемы; почему кортеж **из изменяемых** элементов тоже не хешируем (`(1, [2])`). Подсказка: `hash(())` vs `hash([])`.

**Задача 54 (`Counter` + сортировка словаря).** `Counter` — готовая замена ручному счётчику. Перепишите задачу 50 через `collections.Counter`. Отсортируйте словарь по значению через `sorted(counter.items(), key=lambda kv: -kv[1])`. Выведите топ-3 в виде таблицы. Ожидаемый вывод: топ-3 с частотами. В комментариях поясните: чем `Counter` удобнее ручного `dict`; что делает `most_common(1)`; зачем `-kv[1]` для сортировки по убыванию (или `reverse=True`). Подсказка: `from collections import Counter; Counter(genres).most_common(1)`.

**Задача 55 (`defaultdict` + `namedtuple` + `deque`).** Три инструмента из `collections`, которые упрощают типовые задачи. В комментариях поясните: почему `defaultdict` удобнее `dict.get`; чем `namedtuple` лучше обычного кортежа; почему `list.pop(0)` — `O(n)`, а `deque.popleft()` — `O(1)`. Подсказка: `from collections import defaultdict, namedtuple, deque`.

**Задача 56 (`itertools`).** `itertools` — «батарейки» для комбинаторики и итерации. Продемонстрируйте: `chain` (объединить два списка), `combinations` (все пары книг), `permutations` (все порядки 3 книг), `product` (декартово произведение жанров и годов), `groupby` (сгруппировать по жанру — предварительно отсортировав!). В комментариях поясните: почему `groupby` требует предварительной сортировки; чем `combinations` отличается от `permutations`; когда `chain` экономит память (ленивый). Подсказка: `from itertools import chain, combinations, permutations, product, groupby`.

## 🟣 Уровень 8. Матрицы, сложность, сортировки, бинарный поиск (57–64)

**Что изучается:** список списков, `zip(*m)`, сумма/среднее/min/max по матрице, транспонирование, умножение, диагонали, спираль, `O(...)`, линейный vs бинарный поиск, пузырёк/выбор/вставки/quick/merge, `perf_counter`.

**Задача 57 (матрица + format).** Матрица — это список списков, и её надо уметь выводить ровно. Дана матрица `R × C` — число книг на каждой полке. Выведите её построчно ровными колонками (`f"{x:4d}"`), найдите сумму, среднее, min, max. Ожидаемый вывод при `[[3, 2], [1, 4]]`: ровный вывод + `Сумма: 10`, `Среднее: 2.50`, `Мин: 1`, `Макс: 4`. В комментариях поясните: почему матрица в Python — список списков; как «сплющить» матрицу в один список (`[x for row in m for x in row]`); почему нельзя `sum(m)` — получится `TypeError`. Подсказка: `flat = [x for row in m for x in row]`.

**Задача 58 (строка с максимумом).** Иногда нужен не максимум элемента, а максимум суммы по строке. Найдите номер строки с наибольшей суммой элементов. Ожидаемый вывод при `[[1, 1], [5, 5]]`: `Строка 1`. В комментариях поясните: почему нельзя просто `max(m)` (сравнивает списки поэлементно); как `[sum(r) for r in m].index(max(...))` решает задачу; что будет при равенстве сумм (первая). Подсказка: `sums = [sum(row) for row in m]; sums.index(max(sums))`.

**Задача 59 (транспонирование + умножение на k).** Две базовые операции над матрицами. (1) Транспонирование: строки становятся столбцами. (2) Умножение на число `k`. Ожидаемый вывод при `[[1, 2], [3, 4]]`: `1 3 / 2 4` и `2 4 / 6 8` при `k=2`. В комментариях поясните: что делает `zip(*m)` (распаковка строк в аргументы `zip`); почему `[[x * k for x in row] for row in m]` — правильный способ, а `m * k` — нет (повторение списка); где транспонирование используется (линейная алгебра, графика). Подсказка: `[list(row) for row in zip(*m)]`.

**Задача 60 (сложение и умножение матриц).** Стандартные операции линейной алгебры. (1) Сложите две матрицы `R×C`. (2) Умножьте `A (R×K)` на `B (K×C)`. Проверьте, что размеры совпадают. Ожидаемый вывод: `6 8 / 10 12` для сложения; `11` для `[[1, 2]] × [[3], [4]]`. В комментариях поясните: почему при сложении размеры должны совпадать; почему при умножении `K` первой должно совпадать с `K` второй; почему тройной цикл — это `O(R·K·C)`. Подсказка: `for i in range(R): for j in range(C): for t in range(K): C[i][j] += A[i][t] * B[t][j]`.

**Задача 61 (диагонали).** Диагонали квадратной матрицы — частый элемент отчётов. Для квадратной матрицы выведите главную и побочную диагонали и их суммы. Ожидаемый вывод при `[[1, 2], [3, 4]]`: `Главная: 1 4`, `Побочная: 2 3`, суммы `5` и `5`. В комментариях поясните: почему главная — это `m[i][i]`; почему побочная — `m[i][n - 1 - i]`; как это обобщается на не-квадратные (не обобщается). Подсказка: `for i in range(n): print(m[i][i])`.

**Задача 62 (спиральный обход).** Классическая задача на аккуратность с границами. Обойдите матрицу `R×C` по спирали (по часовой стрелке) и выведите элементы. Ожидаемый вывод при `[[1, 2], [3, 4]]`: `1 2 4 3`. В комментариях поясните: зачем четыре границы (`top`, `bottom`, `left`, `right`); как сдвигать границы после каждого прохода; почему легко получить дубликаты, если не проверять `top <= bottom` и `left <= right`. Подсказка: четыре цикла и сдвиг границ.

**Задача 63 (линейный vs бинарный поиск).** Наглядно покажите, почему алгоритмы важны. Реализуйте линейный (`O(n)`) и бинарный (`O(log n)`) поиск. На массиве из 1 000 000 элементов посчитайте число сравнений. Ожидаемый вывод: `~500000` vs `~20`. В комментариях поясните: почему бинарный поиск требует **отсортированного** массива; почему `O(log n)` — это ~20 шагов для миллиона; когда линейный поиск всё равно лучше (маленькие массивы, неотсортированные данные). Подсказка: `lo, hi = 0, len(a) - 1; while lo <= hi: mid = (lo + hi) // 2`.

**Задача 64 (сортировки + замер).** Пять классических сортировок — фундамент алгоритмического мышления. Реализуйте: пузырёк, выбор, вставки, quick, merge (используйте слияние из задачи 32). Замерьте `time.perf_counter()` на `n = 100, 1000, 10000`. Постройте таблицу «сортировка × n × время». Ожидаемый вывод: видно `O(n²)` vs `O(n log n)`. В комментариях поясните: почему пузырёк/выбор/вставки — `O(n²)`; почему quick в среднем `O(n log n)`, но в худшем `O(n²)`; почему merge всегда `O(n log n)`, но требует `O(n)` памяти; почему встроенный `sorted` (Timsort) быстрее всех на реальных данных. Подсказка: `import time; t0 = time.perf_counter(); ...; t1 = time.perf_counter()`.

## 🔴 Уровень 9. Структуры данных, генераторы (65–72)

**Что изучается:** LIFO/FIFO, `deque`, односвязный список, BST + обходы, хеш-таблица с цепочками, `heapq`, `yield`, `yield from`, `send`, генераторные выражения, `itertools.tee` для повторной итерации.

**Задача 65 (стек).** Стек — простая, но важная структура: LIFO (последний пришёл — первый вышел). Реализуйте стек на списке: `push`, `pop`, `peek`, `is_empty`. В комментариях поясните: почему `append`/`pop()` — это `O(1)`; где стек используется (отмена действий, парсинг, обход в глубину); почему `list.pop(0)` — это уже очередь, но `O(n)`. **Задача 66 (очередь на `deque`).** Очередь — FIFO (первый пришёл — первый вышел), и её надо делать на `deque`. Реализуйте очередь на `collections.deque`. Объясните, почему `list.pop(0)` — плохо (`O(n)`), а `deque.popleft()` — `O(1)`. Ожидаемый вывод: `popleft` возвращает первый. В комментариях поясните: чем `deque` отличается от `list` (двусвязный список под капотом); где очередь используется (BFS, планировщики, буферы); почему `deque` ещё и `appendleft` умеет. **Задача 67 (односвязный список).** Связный список — базовая структура, которую надо уметь писать руками. Реализуйте `Node(value, next)` и `LinkedList` с `append` и `print_all`. Ожидаемый вывод: `1 → 2 → 3`. В комментариях поясните: почему в связном списке `append` — `O(1)` при хранении `tail`; почему доступ по индексу — `O(n)`; где связный список лучше массива (вставка/удаление в середине). Подсказка: храните `head` и `tail`.

**Задача 68 (BST + обход).** Двоичное дерево поиска — основа многих структур. Реализуйте `insert`, `search`, in-order обход. Убедитесь, что in-order даёт возрастающий порядок. Ожидаемый вывод: `1 2 3 4 5`. В комментариях поясните: почему в BST `search` — `O(log n)` в среднем и `O(n)` в худшем (вырожденное дерево); почему in-order даёт сортировку; чем BST отличается от кучи (heap). Подсказка: `def insert(node, val): if node is None: return Node(val); if val < node.val: node.left = insert(node.left, val) else: node.right = insert(node.right, val); return node`.

**Задача 69 (хеш-таблица).** Хеш-таблица — одна из самых важных структур, и её надо уметь писать. Реализуйте `put`, `get`, `remove` с разрешением коллизий **цепочками** (список пар в каждом bucket). Ожидаемый вывод: работает. В комментариях поясните: зачем `hash(key) % size`; что такое коллизия и почему цепочки её решают; почему в среднем `put/get` — `O(1)`, а в худшем — `O(n)`; почему важен размер таблицы (обычно простое число). Подсказка: `bucket = hash(key) % size; self.buckets[bucket].append((key, value))`.

**Задача 70 (`heapq`).** Приоритетная очередь — частая задача, и в Python она уже есть. Используйте `heapq` как приоритетную очередь книг (по году). Продемонстрируйте `heappush` и `heappop`. Ожидаемый вывод: книги выдаются в порядке возрастания года. В комментариях поясните: что такое куча (min-heap); почему `heappush`/`heappop` — `O(log n)`; как сделать max-heap (хранить `-year`). Подсказка: `import heapq; heapq.heappush(h, (year, title)); heapq.heappop(h)`.

**Задача 71 (генераторы глубоко).** Генераторы — ленивые последовательности, экономящие память. (1) `fine_days(days)` через `yield` — штраф по дням. (2) `yield from` — делегирование другому генератору. (3) `send` — передача значения в генератор. (4) Генераторное выражение для суммы. Ожидаемый вывод: `[50, 100, 150]`, делегирование, `send`-пример. В комментариях поясните: чем генератор отличается от списка (ленивость, одноразовость); что делает `yield` (приостанавливает функцию, сохраняет состояние); зачем `yield from` (делегирование); когда `send` полезен (корутины). Подсказка: `def fine_days(days): for d in range(1, days + 1): yield d * 50`.

**Задача 72 (итерация дважды).** Генератор можно пройти **только один раз** — и это часто ломает код. Продемонстрируйте проблему: `g = (x for x in [1, 2, 3]); list(g)` → `[1, 2, 3]`; `list(g)` → `[]`. Исправьте через `itertools.tee` или генераторную функцию. Покажите, что `list(...)` теперь можно вызвать два раза. Ожидаемый вывод: оба прохода работают. В комментариях поясните: почему генератор одноразовый; что делает `tee` (создаёт два независимых итератора, буферизует); когда `tee` — плохая идея (память). Подсказка: `it1, it2 = tee(data)`.
