## 🟢 Уровень 1. Git: basics (1–8)

**Что изучается:** `git status`, `git add`, `git commit`, `git log`, `git diff`, `git show`, `git restore`, `git rm`, staging area, HEAD.

Подсказка: `git status` и читайте вывод сверху вниз.

**Задача 2 (`git add`: staging).** Добавьте `catalog.py` в staging (`git add library/catalog.py`) и снова `git status`. Ожидаемый вывод: файл в разделе «Changes to be committed» (зелёный). В комментариях поясните: что делает `git add` (переносит изменения в staging); почему `git add .` (всё) может быть опасно (можно закоммитить мусор); чем `git add -p` лучше (интерактивно, по кускам — `y/n/s`); что значит «staging — это выбор того, что попадёт в коммит». Подсказка: `git add library/catalog.py`.

**Задача 3 (`git commit`: фиксация).** Закоммитьте изменения с сообщением `feat: add debug print to catalog`. Ожидаемый вывод: `[main abc1234] feat: add debug print to catalog`, `1 file changed, 1 insertion(+)`. В комментариях поясните: что делает `git commit` (создаёт снимок staged-изменений + метаданные); почему сообщение важно (история = документация); что такое **Conventional Commits** (`feat:`, `fix:`, `chore:`); почему `git commit -m "fix"` — плохо (бесполезное сообщение); что делать, если забыли `-m` (откроется редактор — `:wq` в vim). Подсказка: `git commit -m "feat: add debug print to catalog"`.

**Задача 4 (`git log`: история).** Выведите историю: `git log`, `git log --oneline`, `git log --graph --oneline --all`. Ожидаемый вывод: список коммитов с хешем, автором, датой, сообщением. В комментариях поясните: что такое **хеш коммита** (SHA-1, 40 символов, уникален); зачем `--oneline` (компактно); что такое `HEAD` (указатель на текущий коммит); как искать коммит по сообщению (`git log --grep="debug"`); как выйти из `git log` (`q`). Подсказка: `git log --oneline --graph --all`.

**Задача 5 (`git diff`: что изменилось).** Измените `loans.py`, затем выполните `git diff` (unstaged), `git diff --staged` (staged), `git diff HEAD` (всё с последнего коммита). Ожидаемый вывод: построчный diff с `+`/`-`. В комментариях поясните: что показывает `git diff` (разница между рабочей копией и staging); чем `--staged` отличается (staging vs HEAD); как читать diff (`+` — добавлено, `-` — удалено, `@@` — контекст); зачем diff перед коммитом (не закоммитить лишнее). Подсказка: `git diff`, `git diff --staged`, `git diff HEAD`.

**Задача 6 (`git show`).** Покажите содержимое последнего коммита: `git show HEAD`. Ожидаемый вывод: метаданные + diff. В комментариях поясните: что показывает `git show` (коммит + изменения); как посмотреть конкретный коммит (`git show abc1234`); как посмотреть файл в конкретном коммите (`git show HEAD:library/catalog.py`); чем `git show` отличается от `git log -p` (один коммит vs все). Подсказка: `git show HEAD`.

**Задача 7 (`git restore`: откат изменений).** Измените `catalog.py`, но **не** добавляйте в staging. Откатите изменения: `git restore library/catalog.py`. Ожидаемый вывод: файл вернулся к состоянию последнего коммита. В комментариях поясните: что делает `git restore` (откатывает рабочую копию к HEAD); **опасность** (изменения теряются безвозвратно); чем `git restore --staged` отличается (убирает из staging, но оставляет в файле); почему `git checkout -- file` — старый синтаксис (сейчас `git restore`); что делать, если уже закоммитили (`git revert`). Подсказка: `git restore library/catalog.py`.

**Задача 8 (`git rm` + `git mv`).** Создайте `library/old.py`, закоммитьте, затем удалите через `git rm library/old.py`. Переименуйте `library/cli.py` в `library/commands.py` через `git mv`. Ожидаемый вывод: `git status` показывает `deleted` и `renamed`. В комментариях поясните: чем `git rm` отличается от `rm` (первый сразу staged); что делает `git mv` (переименование + staging); почему Git **не хранит** переименования явно (определяет по сходству); когда это важно (история файла сохраняется). В комментариях поясните: что такое multi-stage (несколько `FROM`, финальный — маленький); зачем (компиляторы не нужны в runtime); как `COPY --from=builder` работает; когда multi-stage обязателен (C-расширения, Go/Rust). Подсказка: `FROM ... AS builder` → `COPY --from=builder`.
