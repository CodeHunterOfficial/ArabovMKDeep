## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** broadcasting (правила, ошибки, практика), `axis`, `keepdims`, универсальные функции (`ufunc`), методы `reduce`/`accumulate`/`outer`, условная логика (`where`/`select`/`clip`).

**Задача 19 (правила broadcasting).** Дана матрица продаж `sales` формы `(3, 4)` (3 месяца × 4 жанра) и вектор поправок `offset` формы `(4,)`. Сложите их: `sales + offset`. Объясните, как NumPy растянул `offset` на все три строки. Затем создайте:
- `offset_col` формы `(3, 1)` — сложите с `sales`;
- `scalar` — сложите с `sales`;
- `wrong` формы `(3,)` — попробуйте сложить с `sales` и получите ошибку.

Для каждой пары выведите формы и результат. В комментариях объясните три правила broadcasting: выравнивание размерностей; совместимость по осям (равны или одна равна 1); результирующая форма — поэлементный максимум; почему `(3,)` несовместима с `(3, 4)`, а `(4,)` — совместима; как прочитать сообщение об ошибке `operands could not be broadcast together with shapes ...`.

**Задача 22 (`keepdims` и broadcasting).** Дана матрица `sales` формы `(3, 4)`. Вычислите среднее по строкам (по жанрам внутри месяца):

**Задача 23 (универсальные функции ufunc).** Дан массив цен `prices = np.array([450, 320, 780, 1200, 540])`. Примените:

Используйте параметры ufunc:
- `out`: запишите результат в заранее созданный массив;

Для категоризации используйте `np.select`:
```python
conditions = [prices < 300, (prices >= 300) & (prices < 1000), prices >= 1000]
choices = ["бюджет", "стандарт", "премиум"]
categories = np.select(conditions, choices, default="?")
```

**Задача 23 (универсальные функции ufunc).** Дан массив цен `prices = np.array([75000, 3500, 1200, 450, 25000])`. Примените:

- `where`: примените операцию только там, где `prices > 1000`.

**Что продемонстрировать:** вызовы ufunc и параметры `out`/`where`. **Вопросы:** что такое ufunc; почему `np.add(a, b)` и `a + b` эквивалентны; зачем нужны `out` и `where`; как ufunc связаны с broadcasting; в каком случае `np.log` вернёт `-inf` или `nan`; какие ещё ufunc вы знаете (`np.sin`, `np.abs`, `np.maximum`, `np.mod`).

**Задача 24 (условная логика: `where`, `select`, `clip`).** Дан массив цен `prices = np.array([150, 320, 780, 1200, 2100, 450, 90])`. Реализуйте:
- через `np.where(prices < 200, 200, prices)` — поднимите минимальную цену до 200;
- через `np.where(prices > 1500, 1500, prices)` — обрежьте максимальную цену.

Для категоризации используйте `np.select`:
```python
conditions = [prices < 300, (prices >= 300) & (prices < 1000), prices >= 1000]
choices = ["бюджет", "стандарт", "премиум"]
categories = np.select(conditions, choices, default="?")
```
Наконец, `np.clip(prices, 200, 1000)` — обрежьте с обеих сторон. Дополнительно реализуйте `np.add.reduce` и `np.add.accumulate` для массива заказов `orders = np.array([12, 15, 11, 18, 14, 13, 16])`. **Что продемонстрировать:** все три подхода и методы ufunc. **Вопросы:** чем `where` отличается от `select`; почему при перекрывающихся условиях `select` берёт первое; в каком случае `clip` лучше `where`; чем `reduce` отличается от `accumulate`; что такое `identity` у ufunc (`np.add.identity == 0`).

**Задача 24 (условная логика: `where`, `select`, `clip`).** Дан массив баллов `gpas = np.array([2.9, 3.5, 4.2, 4.7, 5.0, 3.8, 2.5])`. Реализуйте:

**Задача 40 (`einsum` и структурированные массивы).** Даны:

Реализуйте через `np.einsum`:

- `v @ v`: `np.einsum('i,i->', v, v)`;

Выведите количество приёмов по врачам через `np.unique(..., return_counts=True)`. **Что продемонстрировать:** `einsum` для разных операций и работу со структурированным массивом. **Вопросы:** когда `einsum` читаемее, чем `@`; почему структурированные массивы не заменяют Pandas; как `einsum` обобщается на тензоры (`'bij,bjk->bik'`); как структурированный массив связан с `@dataclass` (оба хранят разнородные поля, но NumPy — C-структуры в памяти).

**Что изучается:** `np.array`, `dtype`, `shape`/`size`/`ndim`, функции создания (`zeros`, `ones`, `full`, `empty`, `arange`, `linspace`), случайные числа и воспроизводимость, память, единичные и диагональные матрицы.

**Задача 1 (создание массива, dtype, shape/size/ndim).** В банке ведётся учёт балансов семи клиентов: `[150000, 320000, 78000, 1250000, 540000, 89000, 660000]` рублей. Создайте массив NumPy из этих значений. Выведите его `dtype`, `shape`, `size`, `ndim`. Дополнительно создайте второй массив из тех же значений, но с типом `float32`, и сравните `dtype` и `nbytes` с первым. Ожидаемый вывод: для первого массива `dtype=int64`, `shape=(7,)`, `size=7`, `ndim=1`; для второго — `dtype=float32`, `nbytes=28`. В комментариях поясните: почему NumPy выбрал именно `int64` для первого массива; чем `shape` отличается от `size`; почему `shape` у одномерного массива — это кортеж из одного элемента `(7,)`, а не `(7)`; что произойдёт при обращении `balances[7]` и почему; как `nbytes` связан с `itemsize` и `size`.

**Задача 2 (подбор dtype под задачу).** В банке есть три вида данных: балансы клиентов (от 0 до 100 000 000 рублей), возраст клиентов (от 18 до 100 лет) и процентные ставки по вкладам (от 0.0 до 30.0 с одним знаком после запятой). Для каждого вида создайте массив и подберите **минимальный** `dtype`, который вмещает все возможные значения без переполнения. Выведите `dtype`, `itemsize` и `nbytes` для каждого. Дополнительно проверьте, что произойдёт с выражением `np.array([100, 200, 250], dtype=np.uint8) + 10` — какой результат получится и почему. В комментариях объясните: почему для балансов не стоит брать `uint8` (какой максимальный баланс он вмещает и что будет при превышении); почему для возраста клиентов достаточно `uint8`, а для балансов — нет; почему для процентных ставок лучше `float32`, чем `float64`, если точности в 7 знаков хватает; как связаны `dtype`, `itemsize` и `nbytes`; почему NumPy не выбрасывает исключение при переполнении и как это может привести к незаметным ошибкам в финансовых расчётах.

**Задача 3 (арифметические операции).** Даны балансы четырёх клиентов `[150000, 320000, 78000, 1250000]` рублей. Одной строкой без использования циклов:
- начислите всем 5% годовых (умножение на 1.05);
- посчитайте балансы после снятия 10 000 рублей;
- возведите балансы в квадрат (условно, для расчёта бонусов);
- прибавьте к каждому балансу 500 рублей (кэшбэк).

Для каждого результата выведите массив. Ожидаемый вывод для первой операции: `[157500. 336000. 81900. 1312500.]`. В комментариях объясните: почему `balances * 1.05` даёт поэлементное умножение, а не повторение, как у списков Python; что произойдёт с `dtype` после умножения на `1.05` и почему результат становится `float64`, а не остаётся `int64`; что случится, если попробовать `balances + "руб."` и почему операция выбросит ошибку, а не вернёт строковый массив.

**Задача 4 (zeros/ones/full/empty).** Банк выделяет буферы под учёт: массив из 30 нулей — будущие ежедневные операции; массив из 12 единиц — веса месяцев для сезонного анализа; массив из 5 значений `-1` — маркер «счёт закрыт» для отсутствующих данных; пустой массив из 3 элементов — буфер, который сразу заполните значениями `[150000, 320000, 78000]`. Для каждого выведите результат и `dtype`. Ожидаемый вывод для первого: `[0. 0. ... 0.]` с `dtype=float64`. В комментариях объясните: зачем нужен `np.empty`, если его содержимое непредсказуемо (когда это оправдано); почему `np.zeros(5)` по умолчанию даёт `float64`, а не `int64`, и как это связано с безопасностью вычислений; в каком случае `np.empty` быстрее `np.zeros` и насколько это критично для больших массивов; что произойдёт, если создать буфер через `np.empty` и случайно прочитать значения до заполнения.

**Задача 5 (arange, linspace).** Банк строит сетки для отчётов:
- `np.arange(1, 13)` — месяцы 1…12;
- `np.arange(0, 10000001, 1000000)` — пороги баланса от 0 до 10 000 000 с шагом 1 000 000;
- `np.linspace(0, 30, 11)` — 11 равномерных точек процентной ставки от 0 до 30;
- `np.linspace(0, 1, 5)` — 5 точек для нормировки.

Для каждого массива выведите результат и `len()`. Дополнительно проверьте `len(np.arange(0, 1, 0.1))` — получилось ли 10? Ожидаемый вывод для последней проверки: `10`, но может быть и `9` или `11` в зависимости от платформы из-за погрешности float. В комментариях объясните: чем `arange` отличается от `linspace`; почему `arange` с вещественным шагом может дать неожиданное количество точек; в каком случае выбирать `arange` (когда важен шаг), а в каком — `linspace` (когда важно количество точек); как избежать ошибки при построении сеток для финансовых расчётов, где нужна точная длина массива.

**Задача 6 (двумерные массивы, форма).** Банк хранит операции за 3 месяца по 4 типам:
```python
ops = np.array([
    [120, 150, 90, 130],   # январь: переводы, платежи, снятия, пополнения
    [110, 140, 100, 125],  # февраль
    [130, 160, 95, 140],   # март
])
```
Выведите `shape`, `ndim`, `size`, `nbytes`. Получите: операции в феврале (`ops[1]` — одномерный массив); операции третьего типа в марте (`ops[2, 2]` — скаляр); весь второй столбец (`ops[:, 1]` — одномерный массив). Ожидаемый вывод: `shape=(3, 4)`, `ndim=2`, `size=12`, `nbytes=96`. В комментариях объясните: почему `ops[1]` возвращает одномерный массив, а не скаляр; что произойдёт при `ops[5]` (индекс за границей); почему запись `ops[1, 2]` эффективнее, чем `ops[1][2]` (какие промежуточные объекты создаются); как `shape` связан с `ndim` и `size`.

**Задача 7 (память: itemsize и nbytes).** Банк хранит данные о 1 000 000 транзакций. Сравните память, занимаемую: массивом `float64` (суммы транзакций); массивом `float32` (те же суммы); массивом `uint16` (количество операций за день, 0–65535). Для каждого выведите `itemsize` и `nbytes` в мегабайтах. Ожидаемый вывод: `float64` — 7.63 МБ, `float32` — 3.81 МБ, `uint16` — 1.91 МБ. Дополнительно оцените размер Python-списка из 1 000 000 чисел через `sys.getsizeof` (только оболочки) и сравните с массивами NumPy. В комментариях объясните: почему `nbytes` не включает метаданные объекта (что ещё занимает память); почему для больших массивов выбор типа — это не микрооптимизация, а вопрос «поместится ли в RAM»; во сколько раз `uint16` экономнее `float64` и как это влияет на скорость обработки за счёт кэша процессора; в каких случаях экономия памяти важнее точности вычислений.

**Задача 8 (случайные числа, воспроизводимость).** Банк моделирует данные для симуляции кредитного портфеля. Создайте генератор `rng = np.random.default_rng(seed=42)` и сгенерируйте: 10 равномерных значений `rng.random(10)` — вероятности одобрения кредита; 10 нормальных значений `rng.normal(500000, 150000, 10)` — балансы клиентов; 10 целых значений `rng.integers(18, 100, 10)` — возраст клиентов. Создайте второй генератор с тем же `seed=42` и убедитесь, что все три последовательности совпадают. Создайте третий генератор без `seed` и покажите, что значения отличаются. В комментариях объясните: зачем нужен `seed` и как он обеспечивает воспроизводимость экспериментов; чем `default_rng` отличается от старого `np.random.seed` (какой BitGenerator используется по умолчанию); почему в новом коде рекомендуется `Generator`, а не глобальный `np.random`; как `SeedSequence` помогает организовать независимые потоки случайных чисел для параллельных вычислений; в каких ситуациях отсутствие фиксированного `seed` — это хорошо, а в каких — плохо.

**Задача 9 (arange vs linspace на практике).** Банк строит график операций за 30 дней. Реализуйте массив дней двумя способами: `np.arange(1, 31)` — дни 1…30; `np.linspace(1, 30, 30)` — 30 точек от 1 до 30. Проверьте `len()`, первые и последние элементы — они должны совпадать. Дополнительно найдите ситуацию, где `arange` даёт неверное количество точек из-за погрешности float: `np.arange(0, 1, 0.1)` — сколько элементов получилось? Ожидаемый ответ: 10, но может быть и другое число на некоторых платформах. В комментариях объясните: почему `linspace` надёжнее, когда важно именно количество точек; в каком случае `arange` уместнее (когда важен шаг); почему `arange` с вещественным шагом опасен в финансовых расчётах, где неверная длина массива может привести к сдвигу данных; как проверить длину массива перед использованием и что делать, если она не совпадает с ожидаемой.

**Задача 10 (единичная и диагональная матрицы).** Банк хранит матрицу корреляций между 4 показателями клиента (баланс, доход, расход, кредитный рейтинг). Создайте: единичную матрицу `np.eye(4)` — начальное приближение (каждый показатель коррелирует сам с собой); диагональную матрицу `np.diag([1.0, 0.8, 0.6, 0.9])` — дисперсии показателей; матрицу со смещённой диагональю `np.diag([0.5, 0.3], k=1)` — корреляции соседних показателей. Выведите все три. Ожидаемый вывод для первой: диагональ из единиц, остальные нули. В комментариях объясните: чем `np.eye(4)` отличается от `np.identity(4)` (какой параметр позволяет задать вторую размерность); что делает параметр `k` в `np.diag` (положительное значение — диагональ выше главной, отрицательное — ниже); в каком случае диагональная матрица лучше обычной (память и скорость умножения, если большинство элементов нули); почему `np.eye` по умолчанию даёт `float`, а не `int`, и когда это важно.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** индексация, срезы, логическая и fancy-индексация, view vs copy, `reshape`, `ravel`/`flatten`, транспонирование и перестановка осей.

**Задача 11 (индексация).** Дан массив балансов `balances = np.array([150000, 320000, 78000, 1250000, 540000, 89000, 660000])`. Получите: баланс первого клиента (`balances[0]`); баланс последнего (`balances[-1]`); баланс третьего (`balances[2]`). Дана матрица `ops` формы `(3, 4)` из задачи 6. Получите: операции второго типа в январе (`ops[0, 1]`); всю строку за февраль (`ops[1]`); весь второй столбец (`ops[:, 1]`). Ожидаемые результаты: `balances[0] = 150000`, `balances[-1] = 660000`, `ops[0, 1] = 150`, `ops[1] = [110, 140, 100, 125]`, `ops[:, 1] = [150, 140, 160]`. В комментариях объясните: почему `ops[1]` возвращает одномерный массив, а не скаляр; что произойдёт при `ops[5]` (какое исключение и с каким сообщением); почему отрицательные индексы работают и для многомерных массивов; что вернёт `ops[-1, -1]` и почему это последний элемент последней строки.

**Задача 12 (срезы).** Дан массив балансов на 30 клиентов: `balances = np.arange(30) * 50000 + 50000` (условно от 50 000 до 1 500 000). Получите: первые 7 клиентов (`balances[:7]`); последние 7 клиентов (`balances[-7:]`); клиентов с 10-го по 20-й (`balances[10:20]`); каждого второго клиента (`balances[::2]`); массив в обратном порядке (`balances[::-1]`). Дана матрица `ops` формы `(3, 4)`. Получите: первые две строки (`ops[:2, :]`); второй и третий столбцы (`ops[:, 1:3]`); подматрицу `2×2` из левого верхнего угла (`ops[:2, :2]`). В комментариях объясните: включается ли правая граница среза (что вернёт `balances[10:20]` — 10 или 11 элементов); что делает `balances[::-1]` (как получается обратный порядок); почему срез возвращает view, а не копию; как проверить, что `balances[:7].base is balances`, и что это означает; в каком случае срез с шагом `[::2]` может быть опасен при изменении данных.

**Задача 13 (логическая индексация).** Дан массив балансов `balances = np.array([150000, 320000, 78000, 1250000, 540000, 89000, 660000, 45000, 2100000])`. Получите: премиум-клиентов с балансом выше 500 000 (`balances[balances > 500000]`); клиентов с низким балансом менее 100 000 (`balances[balances < 100000]`); клиентов среднего сегмента от 100 000 до 500 000 включительно (`balances[(balances >= 100000) & (balances <= 500000)]`); очень бедных или очень богатых (`balances[(balances < 50000) | (balances > 1000000)]`). Дополнительно замените все балансы ниже 50 000 на 50 000 (`balances[balances < 50000] = 50000`) и выведите массив. В комментариях объясните: почему при комбинировании условий нужны скобки вокруг каждого условия; почему нельзя использовать `and`/`or` вместо `&`/`|` (какая ошибка возникнет); почему логическая индексация возвращает копию, а не view; как `np.where` мог бы заменить последнюю операцию и в каком случае он удобнее.

**Задача 14 (fancy indexing).** Дан массив балансов `balances = np.array([150000, 320000, 78000, 1250000, 540000, 89000, 660000])`. Получите: балансы клиентов с индексами 0, 2, 5 (`balances[[0, 2, 5]]`); те же клиенты в обратном порядке (`balances[[5, 2, 0]]`); клиенты 0, 0, 1, 1, 2 с повторениями (`balances[[0, 0, 1, 1, 2]]`). Дана матрица `ops` формы `(3, 4)`. Получите: строки 0 и 2 (`ops[[0, 2]]`); столбцы 1 и 3 (`ops[:, [1, 3]]`); диагональные элементы `ops[0, 0], ops[1, 1], ops[2, 2]` (`ops[[0, 1, 2], [0, 1, 2]]`); прямоугольную подматрицу через `np.ix_([0, 2], [1, 3])` — сравните с `ops[[0, 2], [1, 3]]` и объясните разницу. В комментариях объясните: чем fancy indexing отличается от среза; почему fancy indexing всегда возвращает копию; зачем нужен `np.ix_` (когда без него выбираются только диагональные элементы); что произойдёт, если индексы выйдут за границу массива.

**Задача 15 (view vs copy).** Дан массив балансов `balances = np.arange(10) * 100000 + 50000`. Создайте: `view = balances[2:5]` — срез; `copy = balances[2:5].copy()` — копия. Измените `view[0] = 9999999` и `copy[0] = 1111111`. Выведите `balances`, `view`, `copy`. Проверьте: `view.base is balances` → `True`; `copy.base is None` → `True`; `np.shares_memory(balances, view)` → `True`; `np.shares_memory(balances, copy)` → `False`. В комментариях объясните: почему NumPy возвращает view для срезов, а не копию (какая экономия достигается); в каком случае это приводит к неожиданным ошибкам (например, при передаче среза в функцию, которая меняет данные); как защититься — вызвать `.copy()`; почему view экономит память, но опасен для больших массивов, если держит ссылку на оригинал (утечка памяти); как проверить, является ли массив view или копией.

**Задача 16 (reshape).** Дан массив из 12 балансов: `balances = np.arange(12) * 100000 + 50000`. Переформируйте его: в матрицу `3×4` (`balances.reshape(3, 4)`); в матрицу `2×6` (`balances.reshape(2, 6)`); в трёхмерный массив `2×2×3` (`balances.reshape(2, 2, 3)`); в матрицу `3×4` с автоматическим определением второй размерности (`balances.reshape(3, -1)`). Попробуйте `balances.reshape(5, 5)` — получите ошибку и объясните, почему. В комментариях объясните: почему `reshape` не меняет данные, а только форму (что происходит с блоком памяти); когда `reshape` возвращает view, а когда копию (проверьте `.base` после транспонирования); зачем нужен `-1` и что произойдёт, если указать две `-1`; как `reshape` связан с порядком хранения в памяти (`order='C'` vs `order='F'` — в каком порядке читаются элементы).

**Задача 17 (ravel и flatten).** Дана матрица `ops` формы `(3, 4)`. Получите одномерный массив тремя способами: `ops.ravel()`; `ops.flatten()`; `ops.reshape(-1)`. Сравните `.base` у каждого результата: у `ravel` и `reshape(-1)` — ссылка на `ops`, у `flatten` — `None`. Измените элемент в `ravel` и убедитесь, что `ops` изменилась; измените элемент в `flatten` и убедитесь, что `ops` не изменилась. В комментариях объясните: в каком случае выбирать `ravel` (когда не нужно изменять), а в каком `flatten` (когда нужна независимая копия); почему `flatten` медленнее (какие данные копируются); что такое «непрерывный массив в C-порядке» и почему `ravel` иногда тоже возвращает копию (например, после транспонирования); как это связано с производительностью при обработке больших массивов.

**Задача 18 (транспонирование и перестановка осей).** Дан трёхмерный массив `data` формы `(2, 3, 4)` — например, 2 отделения банка × 3 месяца × 4 типа операций. Получите: `data.T` — обратный порядок осей, форма `(4, 3, 2)`; `np.transpose(data, (1, 0, 2))` — поменять оси 0 и 1, форма `(3, 2, 4)`; `np.swapaxes(data, 0, 2)` — поменять оси 0 и 2, форма `(4, 3, 2)`; `np.moveaxis(data, 0, -1)` — переместить ось 0 в конец, форма `(3, 4, 2)`. Для двумерной `ops` формы `(3, 4)` сравните `ops.T` и `ops.transpose()` — результат должен совпадать. В комментариях объясните: почему транспонирование возвращает view, а не копию (что меняется в метаданных); как `moveaxis` отличается от `swapaxes` (когда нужен каждый); в каком случае транспонирование реально копирует данные (например, после `ops[::2]`, когда массив не непрерывный); как проверить, является ли результат view или копией.

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Задача 34 (линейная алгебра).** Дана система линейных уравнений `A @ x = b`:

Дополнительно: `np.multiply(balances, 1.05, out=balances)` — ufunc с `out`; `np.clip(balances, 50000, 5000000, out=balances)` — обрезка на месте. **Задача 38 (FFT и обработка сигналов).** Дан сигнал — сумма двух синусоид с шумом (например, суточные и недельные колебания количества операций):
```python
t = np.linspace(0, 1, 1000, endpoint=False)
signal = np.sin(2 * np.pi * 5 * t) + 0.5 * np.sin(2 * np.pi * 12 * t)
noisy = signal + np.random.default_rng(42).normal(0, 0.3, 1000)
```

Отфильтруйте шум: обнулите все частоты выше 20 Гц и выполните `np.fft.ifft`. Сравните ошибку до и после фильтрации. В комментариях объясните: что такое FFT (как раскладывает сигнал на частоты); почему длины, кратные степени двойки, эффективнее (какой алгоритм используется); как амплитуда связана с коэффициентом синусоиды (почему умножаем на 2 и делим на длину); как FFT применяется к анализу сезонности операций в банке (выявление периодических пиков); что такое спектр и как его интерпретировать.

**Задача 39 (полиномы и интерполяция).** Даны дискретные измерения количества операций за 5 дней: `x = np.array([0, 1, 2, 3, 4])`, `y = np.array([12, 15, 11, 18, 14])`. Дополнительно: `np.interp` с параметрами `left=-1`, `right=-1` для точек за пределами диапазона. В комментариях объясните: чем интерполяция отличается от аппроксимации (интерполяция проходит через все точки, аппроксимация — приближается); почему полином высокой степени может «выстреливать» за пределы данных (феномен Рунге) и как это связано с переобучением; в каком случае `np.interp` (кусочно-линейная) лучше полинома (когда данные шумные или их мало); как интерполяция применяется для восстановления пропущенных дней в отчёте банка; какие ещё методы интерполяции существуют (сплайны, кубическая интерполяция).

**Задача 40 (`einsum` и структурированные массивы).** Даны: `A` формы `(3, 4)` — матрица «отделения × типы операций»; `B` формы `(4, 5)` — матрица «типы операций × месяцы»; `v` формы `(4,)` — вектор весов типов операций. Реализуйте через `np.einsum`: матричное умножение `A @ B`: `np.einsum('ik,kj->ij', A, B)`; умножение матрицы на вектор `A @ v`: `np.einsum('ik,k->i', A, v)`; скалярное произведение `v @ v`: `np.einsum('i,i->', v, v)`; внешнее произведение `v` на себя: `np.einsum('i,j->ij', v, v)`; след квадратной матрицы: `np.einsum('ii->', M)`. Дополнительно создайте структурированный массив транзакций:
```python
transactions = np.array([
    ('2025-01-01', 'Иван Петров', 'перевод', 150000),
    ('2025-01-02', 'Мария Иванова', 'платёж', 78000),
    ('2025-01-03', 'Пётр Сидоров', 'снятие', 320000),
], dtype=[('date', 'U10'), ('client', 'U30'), ('type', 'U20'), ('amount', 'i8')])
```
Вычислите среднюю сумму транзакции (`transactions['amount'].mean()`) и отфильтруйте транзакции с суммой выше 100 000. В комментариях объясните: в каком случае `einsum` читаемее, чем `@` (когда операция нестандартная); почему структурированные массивы удобны для табличных данных, но не заменяют Pandas (какие операции Pandas удобнее); как `einsum` обобщается на тензоры (`'bij,bjk->bik'`) и где это применяется (пакетная обработка); как структурированный массив связан с `@dataclass` из Python OOP (оба хранят разнородные поля, но NumPy — это C-структуры в памяти с фиксированным расположением).

**Задача 12 (срезы).** Дан массив цен на 30 рейсов: `prices = np.arange(30) * 5000 + 10000` (условно от 10 000 до 155 000). **Задача 17 (ravel и flatten).** Дана матрица `flights` формы `(3, 4)`. Получите одномерный массив тремя способами: `flights.ravel()`; `flights.flatten()`; `flights.reshape(-1)`. ## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** broadcasting (правила, ошибки, практика), `axis`, `keepdims`, универсальные функции (`ufunc`), методы `reduce`/`accumulate`/`outer`, условная логика (`where`/`select`/`clip`).

**Задача 19 (правила broadcasting).** Дана матрица `flights` формы `(3, 4)` (3 дня × 4 авиакомпании) и вектор поправок `offset` формы `(4,)` (например, изменение числа рейсов по авиакомпаниям). Сложите их: `flights + offset`. Объясните, как NumPy растянул `offset` на все три строки (какие размерности были добавлены виртуально). Затем создайте: `offset_col` формы `(3, 1)` — сложите с `flights` (объясните, как теперь растягивается `offset_col`); `scalar` — сложите с `flights` (объясните, почему скаляр всегда совместим); `wrong` формы `(3,)` — попробуйте сложить с `flights` и получите ошибку (объясните, почему `(3,)` несовместима с `(3, 4)`, а `(4,)` — совместима). Для каждой пары выведите формы и результат. В комментариях объясните три правила broadcasting: выравнивание размерностей (добавление единиц слева); совместимость по осям (равны или одна равна 1); результирующая форма — поэлементный максимум; как прочитать сообщение об ошибке `operands could not be broadcast together with shapes (3,4) (3,)` и что оно означает; в каком случае broadcasting может неявно создать огромный массив и как это предотвратить.

**Задача 20 (broadcasting на практике: нормализация).** Дана матрица характеристик пассажиров формы `(365, 5)` — 365 дней × 5 характеристик (вес багажа, возраст, стоимость билета, номер рейса, класс обслуживания). Проверьте, что после стандартизации среднее по каждому столбцу близко к 0, а стандартное отклонение — к 1. В комментариях объясните: почему `data - data.mean(axis=0)` работает без явного `reshape` (как broadcasting растягивает `(5,)` до `(365, 5)`); что произойдёт, если перепутать `axis=0` и `axis=1` (какая ошибка или неверный результат); как broadcasting экономит память (не создаёт копию среднего на 365 строк); как то же самое сделать через `keepdims=True` и в каком случае это удобнее; зачем вообще нужна стандартизация перед анализом данных пассажиров.

**Задача 21 (ось агрегации `axis`).** Дана матрица `flights` формы `(3, 4)`. Вычислите: `flights.sum()` — одно число; `flights.sum(axis=0)` — форма `(4,)`, сумма по дням (по столбцам); `flights.sum(axis=1)` — форма `(3,)`, сумма по авиакомпаниям (по строкам). **Задача 22 (`keepdims` и broadcasting).** Дана `flights` формы `(3, 4)`. **Задача 23 (универсальные функции ufunc).** Дан массив цен `prices = np.array([10000, 35000, 70000, 15000, 25000])`. В комментариях объясните: что такое ufunc (чем отличается от обычной функции Python); почему `np.add(a, b)` и `a + b` эквивалентны; зачем нужны `out` (экономия памяти) и `where` (условное применение); как ufunc связаны с broadcasting; в каком случае `np.log` вернёт `-inf` или `nan` (при нуле и отрицательных значениях) и как это обработать; какие ещё ufunc вы знаете (`np.sin`, `np.abs`, `np.maximum`, `np.mod`) и где они применяются в расчётах аэропорта.

**Задача 24 (условная логика: `where`, `select`, `clip`).** Дан массив цен `prices = np.array([5000, 15000, 35000, 50000, 70000, 12000, 100000])`. Реализуйте: `np.where(prices < 10000, 10000, prices)` — поднимите минимальную цену до 10 000; `np.where(prices > 70000, 70000, prices)` — обрежьте максимальную цену. Для категоризации используйте `np.select`: условия — цена < 15 000, от 15 000 до 50 000, больше 50 000; значения — «эконом», «бизнес», «первый»; значение по умолчанию — «?». ## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** сортировка (`sort`, `argsort`), частичная сортировка (`partition`, `argpartition`), бинарный поиск (`searchsorted`), многоуровневая сортировка (`lexsort`), `unique` и операции над множествами, `isin`.

**Задача 25 (сортировка).** Дан массив цен `prices = np.array([10000, 35000, 70000, 15000, 25000, 50000, 12000])`. Получите: отсортированный по возрастанию (`np.sort(prices)`); отсортированный по убыванию (`np.sort(-prices) * -1` или `np.sort(prices)[::-1]`); отсортируйте на месте через `prices.sort()` и убедитесь, что исходный массив изменился. Дана матрица `flights` формы `(3, 4)`. Отсортируйте: по умолчанию (`axis=-1`, по строкам); по оси 0 (`axis=0`, по столбцам); с `axis=None` (уплощённый массив). В комментариях объясните: чем `np.sort(arr)` отличается от `arr.sort()` (какой возвращает копию, какой меняет на месте); почему сортировка по умолчанию выполняется по последней оси; как NumPy обрабатывает NaN при сортировке (проверьте, добавив `np.nan` — куда попадут NaN); какой алгоритм сортировки используется по умолчанию и как выбрать стабильный (`kind='stable'`) — в каком случае это важно (например, при сортировке по нескольким ключам).

**Задача 26 (`argsort` и топ-N).** Дан массив цен на 10 рейсов: `prices = np.array([10000, 35000, 70000, 15000, 25000, 50000, 12000, 5000, 100000, 30000])`. Получите: индексы, которые отсортировали бы массив (`np.argsort(prices)`); примените индексы к массиву (`prices[np.argsort(prices)]`) и убедитесь, что результат совпадает с `np.sort(prices)`. Дополнительно даны номера рейсов `flights = np.array(["A","B","C","D","E","F","G","H","I","J"])`. Получите рейсы, отсортированные по цене (`flights[np.argsort(prices)]`). Найдите топ-3 самых дорогих: `flights[np.argsort(-prices)[:3]]`. В комментариях объясните: зачем нужен `argsort`, если есть `sort` (как он позволяет сортировать параллельные массивы); как с его помощью отсортировать номера рейсов по цене, не теряя соответствия; почему `argsort(-prices)` даёт порядок по убыванию; что такое «непрямая сортировка» и в каких задачах она необходима (ранжирование, топ-N, сортировка по нескольким полям).

**Задача 27 (`partition` и `argpartition`).** Дан массив из 1 000 000 случайных цен. Найдите 10 самых дорогих рейсов: через полную сортировку (`np.sort(prices)[-10:]`); через `np.partition(prices, -10)[-10:]`; через `np.argpartition(prices, -10)[-10:]` — индексы. Замерьте время каждого подхода через `time.perf_counter()`. В комментариях объясните: почему `partition` работает за O(n), а `sort` — за O(n log n); что гарантирует `partition` (k-й элемент на своём месте, слева — меньше, справа — больше, но порядок внутри частей не определён); во сколько раз `partition` быстрее `sort` для 1 000 000 элементов (по экспериментальным данным); почему `argpartition` + `argsort` дают корректный топ-N (что нужно сделать после `argpartition`, чтобы упорядочить топ); в каком случае разница в скорости критична (интерактивные приложения, реальное время).

**Задача 28 (`searchsorted` и категоризация).** Даны границы классов обслуживания: `bounds = np.array([0, 15000, 50000, 100000])` (эконом, комфорт, бизнес, первый). Даны цены: `prices = np.array([5000, 12000, 25000, 50000, 70000, 150000])`. Определите класс каждой цены: `np.searchsorted(bounds, prices)` — индексы; `np.searchsorted(bounds, prices, side='right')` — сравните.
```python
t = np.linspace(0, 1, 1000, endpoint=False)
signal = np.sin(2 * np.pi * 5 * t) + 0.5 * np.sin(2 * np.pi * 12 * t)
noisy = signal + np.random.default_rng(42).normal(0, 0.3, 1000)
```
Выполните: `np.fft.fft(noisy)` — прямое преобразование; `np.fft.fftfreq(len(noisy), d=1/1000)` — частоты; `amplitudes = np.abs(fft) / len(noisy) * 2` — амплитуды; найдите топ-2 частоты через `np.argsort(amplitudes)[-2:]`. Отфильтруйте шум: обнулите все частоты выше 20 Гц и выполните `np.fft.ifft`. Сравните ошибку до и после фильтрации. В комментариях объясните: что такое FFT (как раскладывает сигнал на частоты); почему длины, кратные степени двойки, эффективнее (какой алгоритм используется); как амплитуда связана с коэффициентом синусоиды (почему умножаем на 2 и делим на длину); как FFT применяется к анализу сезонности продаж билетов в кинотеатре (выявление периодических пиков); что такое спектр и как его интерпретировать.

**Задача 39 (полиномы и интерполяция).** Даны дискретные измерения количества проданных билетов за 5 дней: `x = np.array([0, 1, 2, 3, 4])`, `y = np.array([120, 150, 110, 180, 140])`. Дополнительно: `np.interp` с параметрами `left=-1`, `right=-1` для точек за пределами диапазона. В комментариях объясните: чем интерполяция отличается от аппроксимации (интерполяция проходит через все точки, аппроксимация — приближается); почему полином высокой степени может «выстреливать» за пределы данных (феномен Рунге) и как это связано с переобучением; в каком случае `np.interp` (кусочно-линейная) лучше полинома (когда данные шумные или их мало); как интерполяция применяется для восстановления пропущенных дней в отчёте кинотеатра; какие ещё методы интерполяции существуют (сплайны, кубическая интерполяция).

**Задача 40 (`einsum` и структурированные массивы).** Даны: `A` формы `(3, 4)` — матрица «залы × фильмы»; `B` формы `(4, 5)` — матрица «фильмы × дни»; `v` формы `(4,)` — вектор весов фильмов. Реализуйте через `np.einsum`: матричное умножение `A @ B`: `np.einsum('ik,kj->ij', A, B)`; умножение матрицы на вектор `A @ v`: `np.einsum('ik,k->i', A, v)`; скалярное произведение `v @ v`: `np.einsum('i,i->', v, v)`; внешнее произведение `v` на себя: `np.einsum('i,j->ij', v, v)`; след квадратной матрицы: `np.einsum('ii->', M)`. Дополнительно создайте структурированный массив сеансов:
```python
sessions_log = np.array([
    ('2025-01-01', 'Интерстеллар', 'зал 1', 12),
    ('2025-01-02', 'Матрица', 'зал 2', 15),
    ('2025-01-03', 'Начало', 'зал 1', 18),
], dtype=[('date', 'U10'), ('movie', 'U30'), ('hall', 'U10'), ('sold', 'i8')])
```
Вычислите среднее число проданных билетов (`sessions_log['sold'].mean()`) и отфильтруйте сеансы с числом проданных билетов выше 14. В комментариях объясните: в каком случае `einsum` читаемее, чем `@` (когда операция нестандартная); почему структурированные массивы удобны для табличных данных, но не заменяют Pandas (какие операции Pandas удобнее); как `einsum` обобщается на тензоры (`'bij,bjk->bik'`) и где это применяется (пакетная обработка); как структурированный массив связан с `@dataclass` из Python OOP (оба хранят разнородные поля, но NumPy — это C-структуры в памяти с фиксированным расположением).

В комментариях объясните: почему `nbytes` не включает метаданные объекта (что ещё занимает память); почему для больших массивов выбор типа — это не микрооптимизация, а вопрос «поместится ли в RAM»; во сколько раз `uint8` экономнее `float64` и как это влияет на скорость обработки за счёт кэша процессора; в каких случаях экономия памяти важнее точности вычислений.

**Задача 8 (случайные числа, воспроизводимость).** Школа моделирует данные для симуляции успеваемости. **Что изучается:** индексация, срезы, логическая и fancy-индексация, view vs copy, `reshape`, `ravel`/`flatten`, транспонирование и перестановка осей.

**Задача 11 (индексация).** Дан массив оценок `marks = np.array([4.5, 3.8, 5.0, 4.2, 4.7, 3.5, 4.1])`. Получите: оценку первого ученика (`marks[0]`); оценку последнего (`marks[-1]`); оценку третьего (`marks[2]`). Дана матрица `lessons` формы `(3, 4)` из задачи 6. Получите: уроки второго предмета в понедельник (`lessons[0, 1]`); все уроки во вторник (`lessons[1]`); все уроки второго предмета (`lessons[:, 1]`). Ожидаемые результаты: `marks[0] = 4.5`, `marks[-1] = 4.1`, `lessons[0, 1] = 4`, `lessons[1] = [5, 4, 6, 3]`, `lessons[:, 1] = [4, 4, 5]`. В комментариях объясните: почему `lessons[1]` возвращает одномерный массив, а не скаляр; что произойдёт при `lessons[5]` (какое исключение и с каким сообщением); почему отрицательные индексы работают и для многомерных массивов; что вернёт `lessons[-1, -1]` и почему это последний элемент последней строки.

**Задача 12 (срезы).** Дан массив оценок на 30 учеников: `marks = np.linspace(2.0, 5.0, 30)`. Получите: первые 7 оценок (`marks[:7]`); последние 7 оценок (`marks[-7:]`); оценки с 10-го по 20-й (`marks[10:20]`); каждую вторую оценку (`marks[::2]`); массив в обратном порядке (`marks[::-1]`). Дана матрица `lessons` формы `(3, 4)`. Получите: первые две строки (`lessons[:2, :]`); второй и третий столбцы (`lessons[:, 1:3]`); подматрицу `2×2` из левого верхнего угла (`lessons[:2, :2]`). В комментариях объясните: включается ли правая граница среза (что вернёт `marks[10:20]` — 10 или 11 элементов); что делает `marks[::-1]` (как получается обратный порядок); почему срез возвращает view, а не копию; как проверить, что `marks[:7].base is marks`, и что это означает; в каком случае срез с шагом `[::2]` может быть опасен при изменении данных.

**Задача 13 (логическая индексация).** Дан массив оценок `marks = np.array([4.5, 3.8, 5.0, 4.2, 4.7, 3.5, 4.1, 2.9, 4.9])`. Получите: отличников (`marks[marks >= 4.5]`); хорошистов (`marks[(marks >= 3.5) & (marks < 4.5)]`); двоечников (`marks[marks < 3.0]`); отличников или двоечников (`marks[(marks >= 4.5) | (marks < 3.0)]`). Дополнительно замените все оценки ниже 3.0 на 3.0 (`marks[marks < 3.0] = 3.0`) и выведите массив. В комментариях объясните: почему при комбинировании условий нужны скобки вокруг каждого условия; почему нельзя использовать `and`/`or` вместо `&`/`|` (какая ошибка возникнет); почему логическая индексация возвращает копию, а не view; как `np.where` мог бы заменить последнюю операцию и в каком случае он удобнее.

**Задача 14 (fancy indexing).** Дан массив оценок `marks = np.array([4.5, 3.8, 5.0, 4.2, 4.7, 3.5, 4.1])`. Получите: оценки учеников с индексами 0, 2, 5 (`marks[[0, 2, 5]]`); те же ученики в обратном порядке (`marks[[5, 2, 0]]`); ученики 0, 0, 1, 1, 2 с повторениями (`marks[[0, 0, 1, 1, 2]]`). **Задача 15 (view vs copy).** Дан массив оценок `marks = np.linspace(2.0, 5.0, 10)`. Создайте: `view = marks[2:5]` — срез; `copy = marks[2:5].copy()` — копия. Измените `view[0] = 5.0` и `copy[0] = 2.0`. Выведите `marks`, `view`, `copy`. Проверьте: `view.base is marks` → `True`; `copy.base is None` → `True`; `np.shares_memory(marks, view)` → `True`; `np.shares_memory(marks, copy)` → `False`. В комментариях объясните: почему NumPy возвращает view для срезов, а не копию (какая экономия достигается); в каком случае это приводит к неожиданным ошибкам (например, при передаче среза в функцию, которая меняет данные); как защититься — вызвать `.copy()`; почему view экономит память, но опасен для больших массивов, если держит ссылку на оригинал (утечка памяти); как проверить, является ли массив view или копией.

**Задача 16 (reshape).** Дан массив из 12 оценок: `marks = np.linspace(2.0, 5.0, 12)`. В комментариях объясните: почему `reshape` не меняет данные, а только форму (что происходит с блоком памяти); когда `reshape` возвращает view, а когда копию (проверьте `.base` после транспонирования); зачем нужен `-1` и что произойдёт, если указать две `-1`; как `reshape` связан с порядком хранения в памяти (`order='C'` vs `order='F'` — в каком порядке читаются элементы).

**Задача 17 (ravel и flatten).** Дана матрица `lessons` формы `(3, 4)`. Получите одномерный массив тремя способами: `lessons.ravel()`; `lessons.flatten()`; `lessons.reshape(-1)`. **Задача 18 (транспонирование и перестановка осей).** Дан трёхмерный массив `data` формы `(2, 3, 4)` — например, 2 параллели × 3 дня × 4 предмета. **Задача 19 (правила broadcasting).** Дана матрица `lessons` формы `(3, 4)` (3 дня × 4 предмета) и вектор поправок `offset` формы `(4,)` (например, изменение числа уроков по предметам). В комментариях объясните три правила broadcasting: выравнивание размерностей (добавление единиц слева); совместимость по осям (равны или одна равна 1); результирующая форма — поэлементный максимум; как прочитать сообщение об ошибке `operands could not be broadcast together with shapes (3,4) (3,)` и что оно означает; в каком случае broadcasting может неявно создать огромный массив и как это предотвратить.

**Задача 20 (broadcasting на практике: нормализация).** Дана матрица характеристик учеников формы `(365, 5)` — 365 дней × 5 характеристик (средняя оценка, возраст, число пропусков, число олимпиад, число замечаний). **Задача 21 (ось агрегации `axis`).** Дана матрица `lessons` формы `(3, 4)`. **Задача 22 (`keepdims` и broadcasting).** Дана `lessons` формы `(3, 4)`. **Задача 23 (универсальные функции ufunc).** Дан массив оценок `marks = np.array([4.5, 3.8, 5.0, 4.2, 4.7])`. В комментариях объясните: что такое ufunc (чем отличается от обычной функции Python); почему `np.add(a, b)` и `a + b` эквивалентны; зачем нужны `out` (экономия памяти) и `where` (условное применение); как ufunc связаны с broadcasting; в каком случае `np.log` вернёт `-inf` или `nan` (при нуле и отрицательных значениях) и как это обработать; какие ещё ufunc вы знаете (`np.sin`, `np.abs`, `np.maximum`, `np.mod`) и где они применяются в расчётах школы.

**Задача 24 (условная логика: `where`, `select`, `clip`).** Дан массив оценок `marks = np.array([2.9, 3.5, 4.2, 4.7, 5.0, 3.8, 2.5])`. Реализуйте: `np.where(marks < 3.0, 3.0, marks)` — поднимите минимальную оценку до 3.0; `np.where(marks > 5.0, 5.0, marks)` — обрежьте максимальную. Для категоризации используйте `np.select`: условия — оценка < 3.5, от 3.5 до 4.5, больше 4.5; значения — «троечник», «хорошист», «отличник»; значение по умолчанию — «?». Наконец, `np.clip(marks, 3.0, 5.0)` — обрежьте с обеих сторон. Дополнительно реализуйте `np.add.reduce` и `np.add.accumulate` для массива ежедневных уроков `lessons_per_day = np.array([6, 5, 6, 5, 6, 4, 5])`. В комментариях объясните: чем `where` отличается от `select` (когда два варианта, когда много); почему при перекрывающихся условиях `select` берёт первое (как это использовать для приоритетов); в каком случае `clip` лучше `where` (краткость, скорость, читаемость); чем `reduce` отличается от `accumulate` (свёртка в одно значение vs промежуточные результаты); что такое `identity` у ufunc (`np.add.identity == 0`) и зачем он нужен.

**Задача 40 (`einsum` и структурированные массивы).** Даны: `A` формы `(3, 4)` — матрица «параллели × предметы»; `B` формы `(4, 5)` — матрица «предметы × дни»; `v` формы `(4,)` — вектор весов предметов. Реализуйте через `np.einsum`: матричное умножение `A @ B`: `np.einsum('ik,kj->ij', A, B)`; умножение матрицы на вектор `A @ v`: `np.einsum('ik,k->i', A, v)`; скалярное произведение `v @ v`: `np.einsum('i,i->', v, v)`; внешнее произведение `v` на себя: `np.einsum('i,j->ij', v, v)`; след квадратной матрицы: `np.einsum('ii->', M)`. Дополнительно создайте структурированный массив оценок:
```python
grades_log = np.array([
    ('2025-01-01', 'Иван Петров', 'математика', 5),
    ('2025-01-02', 'Мария Иванова', 'физика', 4),
    ('2025-01-03', 'Пётр Сидоров', 'литература', 3),
], dtype=[('date', 'U10'), ('pupil', 'U30'), ('subject', 'U20'), ('mark', 'i4')])
```
Вычислите среднюю оценку (`grades_log['mark'].mean()`) и отфильтруйте оценки выше 4. В комментариях объясните: в каком случае `einsum` читаемее, чем `@` (когда операция нестандартная); почему структурированные массивы удобны для табличных данных, но не заменяют Pandas (какие операции Pandas удобнее); как `einsum` обобщается на тензоры (`'bij,bjk->bik'`) и где это применяется (пакетная обработка); как структурированный массив связан с `@dataclass` из Python OOP (оба хранят разнородные поля, но NumPy — это C-структуры в памяти с фиксированным расположением).

**Задача 18 (транспонирование и перестановка осей).** Дан трёхмерный массив `data` формы `(2, 3, 4)` — например, 2 корпуса отеля × 3 дня × 4 категории номеров. **Задача 34 (линейная алгебра).** Дана система линейных уравнений `A @ x = b`:

Дополнительно вычислите: `np.linalg.det(A)` — определитель; `np.linalg.eig(A)` — собственные значения и векторы; `np.linalg.norm(b)` — евклидова норма; `np.linalg.norm(b, ord=1)` и `np.linalg.norm(b, ord=np.inf)` — другие нормы. В комментариях объясните: почему `solve` предпочтительнее `inv(A) @ b` (скорость, устойчивость, накопление ошибок при обращении матрицы); что такое собственные значения (на примере: для диагональной матрицы — сами диагональные элементы); в каком случае матрица не имеет обратной (определитель равен нулю — что это означает геометрически); как нормы используются для оценки сходства гостей (косинусное сходство характеристик, норма разности векторов).

В комментариях объясните: чем интерполяция отличается от аппроксимации (интерполяция проходит через все точки, аппроксимация — приближается); почему полином высокой степени может «выстреливать» за пределы данных (феномен Рунге) и как это связано с переобучением; в каком случае `np.interp` (кусочно-линейная) лучше полинома (когда данные шумные или их мало); как интерполяция применяется для восстановления пропущенных дней в отчёте отеля; какие ещё методы интерполяции существуют (сплайны, кубическая интерполяция).

**Задача 40 (`einsum` и структурированные массивы).** Даны: `A` формы `(3, 4)` — матрица «корпуса × категории номеров»; `B` формы `(4, 5)` — матрица «категории × дни»; `v` формы `(4,)` — вектор весов категорий. Реализуйте через `np.einsum`: матричное умножение `A @ B`: `np.einsum('ik,kj->ij', A, B)`; умножение матрицы на вектор `A @ v`: `np.einsum('ik,k->i', A, v)`; скалярное произведение `v @ v`: `np.einsum('i,i->', v, v)`; внешнее произведение `v` на себя: `np.einsum('i,j->ij', v, v)`; след квадратной матрицы: `np.einsum('ii->', M)`. Дополнительно создайте структурированный массив бронирований:
```python
bookings_log = np.array([
    ('2025-01-01', 'Иван Петров', 'стандарт', 12000),
    ('2025-01-02', 'Мария Иванова', 'люкс', 24000),
    ('2025-01-03', 'Пётр Сидоров', 'эконом', 8000),
], dtype=[('date', 'U10'), ('guest', 'U30'), ('room_type', 'U20'), ('total', 'i8')])
```
Вычислите среднюю сумму (`bookings_log['total'].mean()`) и отфильтруйте бронирования дороже 10000. В комментариях объясните: в каком случае `einsum` читаемее, чем `@` (когда операция нестандартная); почему структурированные массивы удобны для табличных данных, но не заменяют Pandas (какие операции Pandas удобнее); как `einsum` обобщается на тензоры (`'bij,bjk->bik'`) и где это применяется (пакетная обработка); как структурированный массив связан с `@dataclass` из Python OOP (оба хранят разнородные поля, но NumPy — это C-структуры в памяти с фиксированным расположением).

В комментариях объясните: чем `np.eye(4)` отличается от `np.identity(4)` (какой параметр позволяет задать вторую размерность); что делает параметр `k` в `np.diag` (положительное значение — диагональ выше главной, отрицательное — ниже); в каком случае диагональная матрица лучше обычной (память и скорость умножения, если большинство элементов нули); почему `np.eye` по умолчанию даёт `float`, а не `int`, и когда это важно.

**Что изучается:** индексация, срезы, логическая и fancy-индексация, view vs copy, `reshape`, `ravel`/`flatten`, транспонирование и перестановка осей.

**Задача 11 (индексация).** Дан массив возрастов `ages = np.array([7, 3, 12, 5, 9, 2, 15])`. В комментариях объясните: включается ли правая граница среза (что вернёт `ages[10:20]` — 10 или 11 элементов); что делает `ages[::-1]` (как получается обратный порядок); почему срез возвращает view, а не копию; как проверить, что `ages[:7].base is ages`, и что это означает; в каком случае срез с шагом `[::2]` может быть опасен при изменении данных.

**Задача 13 (логическая индексация).** Дан массив возрастов `ages = np.array([7, 3, 12, 5, 9, 2, 15, 1, 20])`. **Задача 14 (fancy indexing).** Дан массив возрастов `ages = np.array([7, 3, 12, 5, 9, 2, 15])`. **Задача 15 (view vs copy).** Дан массив возрастов `ages = np.arange(10) % 20`. Создайте: `view = ages[2:5]` — срез; `copy = ages[2:5].copy()` — копия. Измените `view[0] = 99` и `copy[0] = 11`. **Задача 17 (ravel и flatten).** Дана матрица `feedings` формы `(3, 4)`. Получите одномерный массив тремя способами: `feedings.ravel()`; `feedings.flatten()`; `feedings.reshape(-1)`. **Задача 18 (транспонирование и перестановка осей).** Дан трёхмерный массив `data` формы `(2, 3, 4)` — например, 2 корпуса зоопарка × 3 дня × 4 секции. **Задача 20 (broadcasting на практике: нормализация).** Дана матрица характеристик животных формы `(365, 5)` — 365 дней × 5 характеристик (вес, возраст, аппетит, активность, потребление воды). **Задача 23 (универсальные функции ufunc).** Дан массив весов `weights = np.array([180, 3500, 4.5, 60, 12])`. **Задача 24 (условная логика: `where`, `select`, `clip`).** Дан массив весов `weights = np.array([0.5, 2, 180, 3500, 4.5, 60, 5000])`. ## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Задача 25 (сортировка).** Дан массив весов `weights = np.array([180, 3500, 4.5, 60, 12, 500, 80])`. В комментариях объясните: чем `np.sort(arr)` отличается от `arr.sort()` (какой возвращает копию, какой меняет на месте); почему сортировка по умолчанию выполняется по последней оси; как NumPy обрабатывает NaN при сортировке (проверьте, добавив `np.nan` — куда попадут NaN); какой алгоритм сортировки используется по умолчанию и как выбрать стабильный (`kind='stable'`) — в каком случае это важно (например, при сортировке по нескольким ключам).

**Задача 26 (`argsort` и топ-N).** Дан массив весов на 10 животных: `weights = np.array([180, 3500, 4.5, 60, 12, 500, 80, 2, 5000, 300])`. Получите: индексы, которые отсортировали бы массив (`np.argsort(weights)`); примените индексы к массиву (`weights[np.argsort(weights)]`) и убедитесь, что результат совпадает с `np.sort(weights)`. Дополнительно даны клички животных `names = np.array(["A","B","C","D","E","F","G","H","I","J"])`. Получите животных, отсортированных по весу (`names[np.argsort(weights)]`). Найдите топ-3 самых тяжёлых: `names[np.argsort(-weights)[:3]]`. В комментариях объясните: зачем нужен `argsort`, если есть `sort` (как он позволяет сортировать параллельные массивы); как с его помощью отсортировать клички по весу, не теряя соответствия; почему `argsort(-weights)` даёт порядок по убыванию; что такое «непрямая сортировка» и в каких задачах она необходима (ранжирование, топ-N, сортировка по нескольким полям).

**Задача 27 (`partition` и `argpartition`).** Дан массив из 1 000 000 случайных весов. Найдите 10 самых тяжёлых животных: через полную сортировку (`np.sort(weights)[-10:]`); через `np.partition(weights, -10)[-10:]`; через `np.argpartition(weights, -10)[-10:]` — индексы. **Задача 20 (broadcasting на практике: нормализация).** Дана матрица характеристик клиентов формы `(365, 5)` — 365 дней × 5 характеристик (вес, возраст, число посещений, длительность тренировки, потраченные калории). **Задача 38 (FFT и обработка сигналов).** Дан сигнал — сумма двух синусоид с шумом (например, суточные и недельные колебания посещаемости):
```python
t = np.linspace(0, 1, 1000, endpoint=False)
signal = np.sin(2 * np.pi * 5 * t) + 0.5 * np.sin(2 * np.pi * 12 * t)
noisy = signal + np.random.default_rng(42).normal(0, 0.3, 1000)
```
Выполните: `np.fft.fft(noisy)` — прямое преобразование; `np.fft.fftfreq(len(noisy), d=1/1000)` — частоты; `amplitudes = np.abs(fft) / len(noisy) * 2` — амплитуды; найдите топ-2 частоты через `np.argsort(amplitudes)[-2:]`. Отфильтруйте шум: обнулите все частоты выше 20 Гц и выполните `np.fft.ifft`. Сравните ошибку до и после фильтрации. В комментариях объясните: что такое FFT (как раскладывает сигнал на частоты); почему длины, кратные степени двойки, эффективнее (какой алгоритм используется); как амплитуда связана с коэффициентом синусоиды (почему умножаем на 2 и делим на длину); как FFT применяется к анализу сезонности посещаемости клуба (выявление периодических пиков); что такое спектр и как его интерпретировать.

**Задача 40 (`einsum` и структурированные массивы).** Даны: `A` формы `(3, 4)` — матрица «филиалы × типы тренировок»; `B` формы `(4, 5)` — матрица «типы тренировок × дни»; `v` формы `(4,)` — вектор весов типов тренировок. Реализуйте через `np.einsum`: матричное умножение `A @ B`: `np.einsum('ik,kj->ij', A, B)`; умножение матрицы на вектор `A @ v`: `np.einsum('ik,k->i', A, v)`; скалярное произведение `v @ v`: `np.einsum('i,i->', v, v)`; внешнее произведение `v` на себя: `np.einsum('i,j->ij', v, v)`; след квадратной матрицы: `np.einsum('ii->', M)`. Дополнительно создайте структурированный массив посещений:
```python
visits_log = np.array([
    ('2025-01-01', 'Иван Петров', 'йога', 60),
    ('2025-01-02', 'Мария Иванова', 'бокс', 90),
    ('2025-01-03', 'Пётр Сидоров', 'кроссфит', 120),
], dtype=[('date', 'U10'), ('member', 'U30'), ('workout', 'U20'), ('duration', 'i4')])
```
Вычислите среднюю длительность (`visits_log['duration'].mean()`) и отфильтруйте посещения длительнее 60 минут. В комментариях объясните: в каком случае `einsum` читаемее, чем `@` (когда операция нестандартная); почему структурированные массивы удобны для табличных данных, но не заменяют Pandas (какие операции Pandas удобнее); как `einsum` обобщается на тензоры (`'bij,bjk->bik'`) и где это применяется (пакетная обработка); как структурированный массив связан с `@dataclass` из Python OOP (оба хранят разнородные поля, но NumPy — это C-структуры в памяти с фиксированным расположением).
