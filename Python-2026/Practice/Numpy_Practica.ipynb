**Задача 12 (срезы).** Дан массив цен на 30 книг: `prices = np.arange(30) * 100 + 100`. Получите:

**Задача 13 (логическая индексация).** Дан массив цен `prices = np.array([450, 320, 780, 1200, 540, 890, 660, 150, 2100])`. Получите:

**Задача 14 (fancy indexing).** Дан массив цен `prices = np.array([450, 320, 780, 1200, 540, 890, 660])`. Получите:
- цены книг с индексами 0, 2, 5 (`prices[[0, 2, 5]]`);

- диагональные элементы (`sales[[0, 1, 2], [0, 1, 2]]`);

**Что продемонстрировать:** выборку по массивам индексов и разницу `np.ix_`. **Вопросы:** чем fancy indexing отличается от среза; почему fancy indexing всегда возвращает копию; зачем нужен `np.ix_` (когда без него выбираются только диагональные элементы); что произойдёт, если индексы выйдут за границу; в каком случае fancy indexing удобнее логической индексации.

**Задача 15 (view vs copy).** Дан массив цен `prices = np.arange(10) * 100`. Создайте:
- `view = prices[2:5]` — срез;
- `copy = prices[2:5].copy()` — копия.

- `view.base is prices` → `True`;
- `copy.base is None` → `True`;
- `np.shares_memory(prices, view)` → `True`;

**Что продемонстрировать:** влияние view на исходный массив и независимость copy. **Вопросы:** почему NumPy возвращает view для срезов, а не копию; в каком случае это приводит к неожиданным ошибкам (передача среза в функцию, которая меняет данные); как защититься — вызвать `.copy()`; почему view экономит память, но опасен для больших массивов (держится ссылка на оригинал); как проверить, view это или копия.

**Задача 16 (reshape).** Дан массив из 12 цен: `prices = np.arange(12) * 100`. Переформируйте его:

**Задача 17 (ravel и flatten).** Дана матрица `sales` формы `(3, 4)`. Получите одномерный массив тремя способами:

**Задача 18 (транспонирование и перестановка осей).** Дан трёхмерный массив `data` формы `(2, 3, 4)` — 2 филиала × 3 месяца × 4 жанра. Получите:

Для двумерной `sales` формы `(3, 4)` сравните `sales.T` и `sales.transpose()`. **Что продемонстрировать:** все способы перестановки осей. **Вопросы:** почему транспонирование возвращает view, а не копию; как `moveaxis` отличается от `swapaxes`; в каком случае транспонирование реально копирует данные (например, после `sales[::2]`); как проверить, view это или копия.

**Задача 19 (правила broadcasting).** Дана матрица продаж `sales` формы `(3, 4)` (3 месяца × 4 жанра) и вектор поправок `offset` формы `(4,)`. Сложите их: `sales + offset`. Объясните, как NumPy растянул `offset` на все три строки. Затем создайте:
- `offset_col` формы `(3, 1)` — сложите с `sales`;

Реализуйте через `np.einsum`:
- матричное умножение `A @ B`: `np.einsum('ik,kj->ij', A, B)`;
- умножение матрицы на вектор `A @ v`: `np.einsum('ik,k->i', A, v)`;
- скалярное произведение `v @ v`: `np.einsum('i,i->', v, v)`;
- внешнее произведение `v` на себя: `np.einsum('i,j->ij', v, v)`;
- след: `np.einsum('ii->', M)`.

Дополнительно структурированный массив выдач:
```python
loans = np.array([
    ('2025-01-01', 'Война и мир', 30),
    ('2025-01-02', 'Преступление и наказание', 14),
    ('2025-01-03', 'Мастер и Маргарита', 30),
], dtype=[('date', 'U10'), ('title', 'U30'), ('days', 'i4')])
```
Вычислите средний срок выдачи (`loans['days'].mean()`) и отфильтруйте книги со сроком 30 дней. **Что продемонстрировать:** `einsum` для разных операций и работу со структурированным массивом. **Вопросы:** когда `einsum` читаемее, чем `@`; почему структурированные массивы не заменяют Pandas; как `einsum` обобщается на тензоры (`'bij,bjk->bik'`); как структурированный массив связан с `@dataclass` (оба хранят разнородные поля, но NumPy — C-структуры в памяти).

**Задача 18 (транспонирование и перестановка осей).** Дан трёхмерный массив `data` формы `(2, 3, 4)` — например, 2 филиала магазина × 3 месяца × 4 категории. В комментариях объясните: почему транспонирование возвращает view, а не копию (что меняется в метаданных); как `moveaxis` отличается от `swapaxes` (когда нужен каждый); в каком случае транспонирование реально копирует данные (например, после `sales[::2]`, когда массив не непрерывный); как проверить, является ли результат view или копией.

В комментариях объясните: зачем `argsort` при наличии `sort`; как сортировать параллельные массивы; почему `argsort(-gpas)` даёт убывание; что такое непрямая сортировка; как получить топ-3 самых слабых; как `argsort` связан со стабильностью.

```python
applications = np.array([
    ('2025-01-01', 'Иван Петров', 'матан'),
    ('2025-01-02', 'Мария Иванова', 'физика'),
    ('2025-01-03', 'Пётр Сидоров', 'программирование'),
], dtype=[('date', 'U10'), ('student', 'U30'), ('course', 'U30')])
```

Реализуйте через `np.einsum`:
- `A @ B`: `np.einsum('ik,kj->ij', A, B)`;
- `A @ v`: `np.einsum('ik,k->i', A, v)`;
- `v @ v`: `np.einsum('i,i->', v, v)`;

**Задача 40 (`einsum` и структурированные массивы).** Даны: `A` формы `(3, 4)` — матрица «терминалы × авиакомпании»; `B` формы `(4, 5)` — матрица «авиакомпании × дни»; `v` формы `(4,)` — вектор весов авиакомпаний. Реализуйте через `np.einsum`: матричное умножение `A @ B`: `np.einsum('ik,kj->ij', A, B)`; умножение матрицы на вектор `A @ v`: `np.einsum('ik,k->i', A, v)`; скалярное произведение `v @ v`: `np.einsum('i,i->', v, v)`; внешнее произведение `v` на себя: `np.einsum('i,j->ij', v, v)`; след квадратной матрицы: `np.einsum('ii->', M)`. Дополнительно создайте структурированный массив рейсов:
```python
flights_log = np.array([
    ('2025-01-01', 'SU1234', 'Иван Петров', 10000),
    ('2025-01-02', 'S7001', 'Мария Иванова', 35000),
    ('2025-01-03', 'U6002', 'Пётр Сидоров', 15000),
], dtype=[('date', 'U10'), ('flight', 'U10'), ('passenger', 'U30'), ('price', 'i8')])
```
Вычислите среднюю цену (`flights_log['price'].mean()`) и отфильтруйте рейсы с ценой выше 15 000. В комментариях объясните: в каком случае `einsum` читаемее, чем `@` (когда операция нестандартная); почему структурированные массивы удобны для табличных данных, но не заменяют Pandas (какие операции Pandas удобнее); как `einsum` обобщается на тензоры (`'bij,bjk->bik'`) и где это применяется (пакетная обработка); как структурированный массив связан с `@dataclass` из Python OOP (оба хранят разнородные поля, но NumPy — это C-структуры в памяти с фиксированным расположением).

**Задача 9 (arange vs linspace на практике).** Ресторан строит график заказов за 30 дней. **Задача 14 (fancy indexing).** Дан массив цен `prices = np.array([350, 450, 280, 600, 390, 520, 310])`. **Задача 15 (view vs copy).** Дан массив цен `prices = np.arange(10) * 50 + 100`. Создайте: `view = prices[2:5]` — срез; `copy = prices[2:5].copy()` — копия. **Задача 20 (broadcasting на практике: нормализация).** Дана матрица характеристик блюд формы `(365, 5)` — 365 дней × 5 характеристик (цена, вес, калорийность, время приготовления, число заказов). **Задача 23 (универсальные функции ufunc).** Дан массив цен `prices = np.array([350, 450, 280, 600, 390])`. **Задача 24 (условная логика: `where`, `select`, `clip`).** Дан массив цен `prices = np.array([150, 280, 350, 450, 600, 1200, 310])`. В комментариях объясните: чем `where` отличается от `select` (когда два варианта, когда много); почему при перекрывающихся условиях `select` берёт первое (как это использовать для приоритетов); в каком случае `clip` лучше `where` (краткость, скорость, читаемость); чем `reduce` отличается от `accumulate` (свёртка в одно значение vs промежуточные результаты); что такое `identity` у ufunc (`np.add.identity == 0`) и зачем он нужен.

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** `np.array`, `dtype`, `shape`/`size`/`ndim`, функции создания (`zeros`, `ones`, `full`, `empty`, `arange`, `linspace`), случайные числа и воспроизводимость, память, единичные и диагональные матрицы.

**Задача 1 (создание массива, dtype, shape/size/ndim).** В школе ведётся учёт средних оценок 7 учеников: `[4.5, 3.8, 5.0, 4.2, 4.7, 3.5, 4.1]`. Создайте массив NumPy из этих значений. Выведите его `dtype`, `shape`, `size`, `ndim`. Дополнительно создайте второй массив из тех же значений, но с типом `float32`, и сравните `dtype` и `nbytes` с первым. Ожидаемый вывод: для первого массива `dtype=float64`, `shape=(7,)`, `size=7`, `ndim=1`; для второго — `dtype=float32`, `nbytes=28`. В комментариях поясните: почему NumPy выбрал именно `float64` для первого массива; чем `shape` отличается от `size`; почему `shape` у одномерного массива — это кортеж из одного элемента `(7,)`, а не `(7)`; что произойдёт при обращении `marks[7]` и почему; как `nbytes` связан с `itemsize` и `size`.

**Задача 2 (подбор dtype под задачу).** В школе есть три вида данных: средняя оценка (от 1.0 до 5.0 с одним знаком после запятой); номер класса (от 1 до 11); номер личного дела ученика (от 1 до 5000). Для каждого вида создайте массив и подберите **минимальный** `dtype`, который вмещает все возможные значения без переполнения. Выведите `dtype`, `itemsize` и `nbytes` для каждого. Дополнительно проверьте, что произойдёт с выражением `np.array([100, 200, 250], dtype=np.uint8) + 10` — какой результат получится и почему. В комментариях объясните: почему для средней оценки не годится `int`; почему для номера класса достаточно `uint8`, а для номера личного дела — нет; почему для номера личного дела лучше `uint16`, если максимум 5000; как связаны `dtype`, `itemsize` и `nbytes`; почему NumPy не выбрасывает исключение при переполнении и как это может привести к незаметным ошибкам в учёте учеников.

**Задача 3 (арифметические операции).** Даны средние оценки четырёх учеников `[4.5, 3.8, 5.0, 4.2]`. Одной строкой без использования циклов:
- прибавьте всем по 0.1 (округление до лучшей оценки);
- умножьте на 20 (перевод в 100-балльную шкалу);
- возведите в квадрат (условно, для расчёта бонусов);
- посчитайте отклонение от 4.0 (`marks - 4.0`).

Для каждого результата выведите массив. Ожидаемый вывод для первой операции: `[4.6 3.9 5.1 4.3]`. В комментариях объясните: почему `marks * 2` даёт поэлементное умножение, а не повторение, как у списков Python; что произойдёт с `dtype` после `marks + 0.1` и почему он остаётся `float64`; что случится, если попробовать `marks + "балл"` и почему операция выбросит ошибку, а не вернёт строковый массив.

**Задача 4 (zeros/ones/full/empty).** Школа выделяет буферы под учёт: массив из 30 нулей — будущие ежедневные уроки; массив из 11 единиц — веса классов с 1 по 11; массив из 5 значений `-1` — маркер «оценка не выставлена»; пустой массив из 3 элементов — буфер, который сразу заполните значениями `[4.5, 3.8, 5.0]`. Для каждого выведите результат и `dtype`. Ожидаемый вывод для первого: `[0. 0. ... 0.]` с `dtype=float64`. В комментариях объясните: зачем нужен `np.empty`, если его содержимое непредсказуемо (когда это оправдано); почему `np.zeros(5)` по умолчанию даёт `float64`, а не `int64`, и как это связано с безопасностью вычислений; в каком случае `np.empty` быстрее `np.zeros` и насколько это критично для больших массивов; что произойдёт, если создать буфер через `np.empty` и случайно прочитать значения до заполнения.

**Задача 5 (arange, linspace).** Школа строит сетки для отчётов:
- `np.arange(1, 12)` — классы 1…11;
- `np.arange(1, 5001, 500)` — пороги номеров личных дел от 1 до 5000 с шагом 500;
- `np.linspace(1, 5, 9)` — 9 равномерных точек шкалы оценок от 1 до 5;
- `np.linspace(0, 1, 5)` — 5 точек для нормировки.

Для каждого массива выведите результат и `len()`. Дополнительно проверьте `len(np.arange(0, 1, 0.1))` — получилось ли 10? Ожидаемый вывод для последней проверки: `10`, но может быть и `9` или `11` в зависимости от платформы из-за погрешности float. В комментариях объясните: чем `arange` отличается от `linspace`; почему `arange` с вещественным шагом может дать неожиданное количество точек; в каком случае выбирать `arange` (когда важен шаг), а в каком — `linspace` (когда важно количество точек); как избежать ошибки при построении сеток для расчёта средних оценок, где нужна точная длина массива.

**Задача 6 (двумерные массивы, форма).** Школа хранит количество уроков за 3 дня по 4 предметам:
```python
lessons = np.array([
    [6, 4, 5, 3],   # понедельник: математика, литература, физика, история
    [5, 4, 6, 3],   # вторник
    [6, 5, 5, 4],   # среда
])
```
Выведите `shape`, `ndim`, `size`, `nbytes`. Получите: уроки во вторник (`lessons[1]` — одномерный массив); уроки третьего предмета в среду (`lessons[2, 2]` — скаляр); уроки второго предмета за все дни (`lessons[:, 1]` — одномерный массив). Ожидаемый вывод: `shape=(3, 4)`, `ndim=2`, `size=12`, `nbytes=96`. В комментариях объясните: почему `lessons[1]` возвращает одномерный массив, а не скаляр; что произойдёт при `lessons[5]` (индекс за границей); почему запись `lessons[1, 2]` эффективнее, чем `lessons[1][2]` (какие промежуточные объекты создаются); как `shape` связан с `ndim` и `size`.

**Задача 7 (память: itemsize и nbytes).** Школа хранит данные о 1 000 000 оценок. Сравните память, занимаемую: массивом `float64` (средние оценки); массивом `float32` (те же оценки); массивом `uint8` (номер класса, 1–11). Для каждого выведите `itemsize` и `nbytes` в мегабайтах. Ожидаемый вывод: `float64` — 7.63 МБ, `float32` — 3.81 МБ, `uint8` — 0.95 МБ. Дополнительно оцените размер Python-списка из 1 000 000 чисел через `sys.getsizeof` (только оболочки) и сравните с массивами NumPy. В комментариях объясните: почему `nbytes` не включает метаданные объекта (что ещё занимает память); почему для больших массивов выбор типа — это не микрооптимизация, а вопрос «поместится ли в RAM»; во сколько раз `uint8` экономнее `float64` и как это влияет на скорость обработки за счёт кэша процессора; в каких случаях экономия памяти важнее точности вычислений.

**Задача 8 (случайные числа, воспроизводимость).** Школа моделирует данные для симуляции успеваемости. Создайте генератор `rng = np.random.default_rng(seed=42)` и сгенерируйте: 10 равномерных значений `rng.random(10)` — вероятности сдать экзамен; 10 нормальных значений `rng.normal(4.0, 0.5, 10)` — средние оценки; 10 целых значений `rng.integers(1, 12, 10)` — номера классов. Создайте второй генератор с тем же `seed=42` и убедитесь, что все три последовательности совпадают. Создайте третий генератор без `seed` и покажите, что значения отличаются. В комментариях объясните: зачем нужен `seed` и как он обеспечивает воспроизводимость экспериментов; чем `default_rng` отличается от старого `np.random.seed` (какой BitGenerator используется по умолчанию); почему в новом коде рекомендуется `Generator`, а не глобальный `np.random`; как `SeedSequence` помогает организовать независимые потоки случайных чисел для параллельных вычислений; в каких ситуациях отсутствие фиксированного `seed` — это хорошо, а в каких — плохо.

**Задача 9 (arange vs linspace на практике).** Школа строит график успеваемости за 30 дней. Реализуйте массив дней двумя способами: `np.arange(1, 31)` — дни 1…30; `np.linspace(1, 30, 30)` — 30 точек от 1 до 30. Проверьте `len()`, первые и последние элементы — они должны совпадать. Дополнительно найдите ситуацию, где `arange` даёт неверное количество точек из-за погрешности float: `np.arange(0, 1, 0.1)` — сколько элементов получилось? Ожидаемый ответ: 10, но может быть и другое число на некоторых платформах. В комментариях объясните: почему `linspace` надёжнее, когда важно именно количество точек; в каком случае `arange` уместнее (когда важен шаг); почему `arange` с вещественным шагом опасен в расчётах, где неверная длина массива может привести к сдвигу данных; как проверить длину массива перед использованием и что делать, если она не совпадает с ожидаемой.

**Задача 10 (единичная и диагональная матрицы).** Школа хранит матрицу корреляций между 4 показателями ученика (средняя оценка, возраст, число пропусков, число олимпиад). Создайте: единичную матрицу `np.eye(4)` — начальное приближение (каждый показатель коррелирует сам с собой); диагональную матрицу `np.diag([1.0, 0.8, 0.6, 0.9])` — дисперсии показателей; матрицу со смещённой диагональю `np.diag([0.5, 0.3], k=1)` — корреляции соседних показателей. Выведите все три. Ожидаемый вывод для первой: диагональ из единиц, остальные нули. В комментариях объясните: чем `np.eye(4)` отличается от `np.identity(4)` (какой параметр позволяет задать вторую размерность); что делает параметр `k` в `np.diag` (положительное значение — диагональ выше главной, отрицательное — ниже); в каком случае диагональная матрица лучше обычной (память и скорость умножения, если большинство элементов нули); почему `np.eye` по умолчанию даёт `float`, а не `int`, и когда это важно.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** индексация, срезы, логическая и fancy-индексация, view vs copy, `reshape`, `ravel`/`flatten`, транспонирование и перестановка осей.

**Задача 11 (индексация).** Дан массив оценок `marks = np.array([4.5, 3.8, 5.0, 4.2, 4.7, 3.5, 4.1])`. Получите: оценку первого ученика (`marks[0]`); оценку последнего (`marks[-1]`); оценку третьего (`marks[2]`). Дана матрица `lessons` формы `(3, 4)` из задачи 6. Получите: уроки второго предмета в понедельник (`lessons[0, 1]`); все уроки во вторник (`lessons[1]`); все уроки второго предмета (`lessons[:, 1]`). Ожидаемые результаты: `marks[0] = 4.5`, `marks[-1] = 4.1`, `lessons[0, 1] = 4`, `lessons[1] = [5, 4, 6, 3]`, `lessons[:, 1] = [4, 4, 5]`. В комментариях объясните: почему `lessons[1]` возвращает одномерный массив, а не скаляр; что произойдёт при `lessons[5]` (какое исключение и с каким сообщением); почему отрицательные индексы работают и для многомерных массивов; что вернёт `lessons[-1, -1]` и почему это последний элемент последней строки.

**Задача 12 (срезы).** Дан массив оценок на 30 учеников: `marks = np.linspace(2.0, 5.0, 30)`. Получите: первые 7 оценок (`marks[:7]`); последние 7 оценок (`marks[-7:]`); оценки с 10-го по 20-й (`marks[10:20]`); каждую вторую оценку (`marks[::2]`); массив в обратном порядке (`marks[::-1]`). Дана матрица `lessons` формы `(3, 4)`. Получите: первые две строки (`lessons[:2, :]`); второй и третий столбцы (`lessons[:, 1:3]`); подматрицу `2×2` из левого верхнего угла (`lessons[:2, :2]`). В комментариях объясните: включается ли правая граница среза (что вернёт `marks[10:20]` — 10 или 11 элементов); что делает `marks[::-1]` (как получается обратный порядок); почему срез возвращает view, а не копию; как проверить, что `marks[:7].base is marks`, и что это означает; в каком случае срез с шагом `[::2]` может быть опасен при изменении данных.

**Задача 13 (логическая индексация).** Дан массив оценок `marks = np.array([4.5, 3.8, 5.0, 4.2, 4.7, 3.5, 4.1, 2.9, 4.9])`. Получите: отличников (`marks[marks >= 4.5]`); хорошистов (`marks[(marks >= 3.5) & (marks < 4.5)]`); двоечников (`marks[marks < 3.0]`); отличников или двоечников (`marks[(marks >= 4.5) | (marks < 3.0)]`). Дополнительно замените все оценки ниже 3.0 на 3.0 (`marks[marks < 3.0] = 3.0`) и выведите массив. В комментариях объясните: почему при комбинировании условий нужны скобки вокруг каждого условия; почему нельзя использовать `and`/`or` вместо `&`/`|` (какая ошибка возникнет); почему логическая индексация возвращает копию, а не view; как `np.where` мог бы заменить последнюю операцию и в каком случае он удобнее.

**Задача 14 (fancy indexing).** Дан массив оценок `marks = np.array([4.5, 3.8, 5.0, 4.2, 4.7, 3.5, 4.1])`. Получите: оценки учеников с индексами 0, 2, 5 (`marks[[0, 2, 5]]`); те же ученики в обратном порядке (`marks[[5, 2, 0]]`); ученики 0, 0, 1, 1, 2 с повторениями (`marks[[0, 0, 1, 1, 2]]`). Дана матрица `lessons` формы `(3, 4)`. Получите: строки 0 и 2 (`lessons[[0, 2]]`); столбцы 1 и 3 (`lessons[:, [1, 3]]`); диагональные элементы `lessons[0, 0], lessons[1, 1], lessons[2, 2]` (`lessons[[0, 1, 2], [0, 1, 2]]`); прямоугольную подматрицу через `np.ix_([0, 2], [1, 3])` — сравните с `lessons[[0, 2], [1, 3]]` и объясните разницу. В комментариях объясните: чем fancy indexing отличается от среза; почему fancy indexing всегда возвращает копию; зачем нужен `np.ix_` (когда без него выбираются только диагональные элементы); что произойдёт, если индексы выйдут за границу массива.

**Задача 15 (view vs copy).** Дан массив оценок `marks = np.linspace(2.0, 5.0, 10)`. Создайте: `view = marks[2:5]` — срез; `copy = marks[2:5].copy()` — копия. Измените `view[0] = 5.0` и `copy[0] = 2.0`. Выведите `marks`, `view`, `copy`. Проверьте: `view.base is marks` → `True`; `copy.base is None` → `True`; `np.shares_memory(marks, view)` → `True`; `np.shares_memory(marks, copy)` → `False`. В комментариях объясните: почему NumPy возвращает view для срезов, а не копию (какая экономия достигается); в каком случае это приводит к неожиданным ошибкам (например, при передаче среза в функцию, которая меняет данные); как защититься — вызвать `.copy()`; почему view экономит память, но опасен для больших массивов, если держит ссылку на оригинал (утечка памяти); как проверить, является ли массив view или копией.

**Задача 16 (reshape).** Дан массив из 12 оценок: `marks = np.linspace(2.0, 5.0, 12)`. Переформируйте его: в матрицу `3×4` (`marks.reshape(3, 4)`); в матрицу `2×6` (`marks.reshape(2, 6)`); в трёхмерный массив `2×2×3` (`marks.reshape(2, 2, 3)`); в матрицу `3×4` с автоматическим определением второй размерности (`marks.reshape(3, -1)`). Попробуйте `marks.reshape(5, 5)` — получите ошибку и объясните, почему. В комментариях объясните: почему `reshape` не меняет данные, а только форму (что происходит с блоком памяти); когда `reshape` возвращает view, а когда копию (проверьте `.base` после транспонирования); зачем нужен `-1` и что произойдёт, если указать две `-1`; как `reshape` связан с порядком хранения в памяти (`order='C'` vs `order='F'` — в каком порядке читаются элементы).

**Задача 17 (ravel и flatten).** Дана матрица `lessons` формы `(3, 4)`. Получите одномерный массив тремя способами: `lessons.ravel()`; `lessons.flatten()`; `lessons.reshape(-1)`. Сравните `.base` у каждого результата: у `ravel` и `reshape(-1)` — ссылка на `lessons`, у `flatten` — `None`. Измените элемент в `ravel` и убедитесь, что `lessons` изменилась; измените элемент в `flatten` и убедитесь, что `lessons` не изменилась. В комментариях объясните: в каком случае выбирать `ravel` (когда не нужно изменять), а в каком `flatten` (когда нужна независимая копия); почему `flatten` медленнее (какие данные копируются); что такое «непрерывный массив в C-порядке» и почему `ravel` иногда тоже возвращает копию (например, после транспонирования); как это связано с производительностью при обработке больших массивов.

**Задача 18 (транспонирование и перестановка осей).** Дан трёхмерный массив `data` формы `(2, 3, 4)` — например, 2 параллели × 3 дня × 4 предмета. Получите: `data.T` — обратный порядок осей, форма `(4, 3, 2)`; `np.transpose(data, (1, 0, 2))` — поменять оси 0 и 1, форма `(3, 2, 4)`; `np.swapaxes(data, 0, 2)` — поменять оси 0 и 2, форма `(4, 3, 2)`; `np.moveaxis(data, 0, -1)` — переместить ось 0 в конец, форма `(3, 4, 2)`. Для двумерной `lessons` формы `(3, 4)` сравните `lessons.T` и `lessons.transpose()` — результат должен совпадать. В комментариях объясните: почему транспонирование возвращает view, а не копию (что меняется в метаданных); как `moveaxis` отличается от `swapaxes` (когда нужен каждый); в каком случае транспонирование реально копирует данные (например, после `lessons[::2]`, когда массив не непрерывный); как проверить, является ли результат view или копией.

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** broadcasting (правила, ошибки, практика), `axis`, `keepdims`, универсальные функции (`ufunc`), методы `reduce`/`accumulate`/`outer`, условная логика (`where`/`select`/`clip`).

**Задача 34 (линейная алгебра).** Дана система линейных уравнений `A @ x = b`:
```python
A = np.array([[2.0, 1.0], [1.0, 3.0]])
b = np.array([5.0, 10.0])
```
Решите систему: через `np.linalg.solve(A, b)`; через `np.linalg.inv(A) @ b` — сравните результаты и объясните, какой способ лучше. В комментариях объясните: почему `solve` предпочтительнее `inv(A) @ b` (скорость, устойчивость, накопление ошибок при обращении матрицы); что такое собственные значения (на примере: для диагональной матрицы — сами диагональные элементы); в каком случае матрица не имеет обратной (определитель равен нулю — что это означает геометрически); как нормы используются для оценки сходства учеников (косинусное сходство характеристик, норма разности векторов).

```python
t = np.linspace(0, 1, 1000, endpoint=False)
signal = np.sin(2 * np.pi * 5 * t) + 0.5 * np.sin(2 * np.pi * 12 * t)
noisy = signal + np.random.default_rng(42).normal(0, 0.3, 1000)
```

В комментариях объясните: чем интерполяция отличается от аппроксимации (интерполяция проходит через все точки, аппроксимация — приближается); почему полином высокой степени может «выстреливать» за пределы данных (феномен Рунге) и как это связано с переобучением; в каком случае `np.interp` (кусочно-линейная) лучше полинома (когда данные шумные или их мало); как интерполяция применяется для восстановления пропущенных дней в отчёте школы; какие ещё методы интерполяции существуют (сплайны, кубическая интерполяция).

**Задача 40 (`einsum` и структурированные массивы).** Даны: `A` формы `(3, 4)` — матрица «параллели × предметы»; `B` формы `(4, 5)` — матрица «предметы × дни»; `v` формы `(4,)` — вектор весов предметов. Реализуйте через `np.einsum`: матричное умножение `A @ B`: `np.einsum('ik,kj->ij', A, B)`; умножение матрицы на вектор `A @ v`: `np.einsum('ik,k->i', A, v)`; скалярное произведение `v @ v`: `np.einsum('i,i->', v, v)`; внешнее произведение `v` на себя: `np.einsum('i,j->ij', v, v)`; след квадратной матрицы: `np.einsum('ii->', M)`. Дополнительно создайте структурированный массив оценок:
```python
grades_log = np.array([
    ('2025-01-01', 'Иван Петров', 'математика', 5),
    ('2025-01-02', 'Мария Иванова', 'физика', 4),
    ('2025-01-03', 'Пётр Сидоров', 'литература', 3),
], dtype=[('date', 'U10'), ('pupil', 'U30'), ('subject', 'U20'), ('mark', 'i4')])
```
Вычислите среднюю оценку (`grades_log['mark'].mean()`) и отфильтруйте оценки выше 4. В комментариях объясните: в каком случае `einsum` читаемее, чем `@` (когда операция нестандартная); почему структурированные массивы удобны для табличных данных, но не заменяют Pandas (какие операции Pandas удобнее); как `einsum` обобщается на тензоры (`'bij,bjk->bik'`) и где это применяется (пакетная обработка); как структурированный массив связан с `@dataclass` из Python OOP (оба хранят разнородные поля, но NumPy — это C-структуры в памяти с фиксированным расположением).

В комментариях объясните: почему `checkins[1]` возвращает одномерный массив, а не скаляр; что произойдёт при `checkins[5]` (какое исключение и с каким сообщением); почему отрицательные индексы работают и для многомерных массивов; что вернёт `checkins[-1, -1]` и почему это последний элемент последней строки.

**Задача 12 (срезы).** Дан массив сумм на 30 гостей: `totals = np.arange(30) * 1000 + 5000`. **Задача 13 (логическая индексация).** Дан массив сумм `totals = np.array([12000, 8000, 24000, 6000, 16000, 20000, 10000, 4000, 30000])`. **Задача 15 (view vs copy).** Дан массив сумм `totals = np.arange(10) * 1000 + 5000`. Создайте: `view = totals[2:5]` — срез; `copy = totals[2:5].copy()` — копия. **Задача 17 (ravel и flatten).** Дана матрица `checkins` формы `(3, 4)`. Получите одномерный массив тремя способами: `checkins.ravel()`; `checkins.flatten()`; `checkins.reshape(-1)`. **Задача 18 (транспонирование и перестановка осей).** Дан трёхмерный массив `data` формы `(2, 3, 4)` — например, 2 корпуса отеля × 3 дня × 4 категории номеров. **Задача 19 (правила broadcasting).** Дана матрица `checkins` формы `(3, 4)` (3 дня × 4 категории номеров) и вектор поправок `offset` формы `(4,)` (например, изменение числа заселений по категориям). Для каждой пары выведите формы и результат. В комментариях объясните три правила broadcasting: выравнивание размерностей (добавление единиц слева); совместимость по осям (равны или одна равна 1); результирующая форма — поэлементный максимум; как прочитать сообщение об ошибке `operands could not be broadcast together with shapes (3,4) (3,)` и что оно означает; в каком случае broadcasting может неявно создать огромный массив и как это предотвратить.

**Задача 20 (broadcasting на практике: нормализация).** Дана матрица характеристик номеров формы `(365, 5)` — 365 дней × 5 характеристик (стоимость, площадь, вместимость, рейтинг, число заселений). Проверьте, что после стандартизации среднее по каждому столбцу близко к 0, а стандартное отклонение — к 1. В комментариях объясните: почему `data - data.mean(axis=0)` работает без явного `reshape` (как broadcasting растягивает `(5,)` до `(365, 5)`); что произойдёт, если перепутать `axis=0` и `axis=1` (какая ошибка или неверный результат); как broadcasting экономит память (не создаёт копию среднего на 365 строк); как то же самое сделать через `keepdims=True` и в каком случае это удобнее; зачем вообще нужна стандартизация перед анализом номеров.

**Задача 21 (ось агрегации `axis`).** Дана матрица `checkins` формы `(3, 4)`. **Задача 23 (универсальные функции ufunc).** Дан массив сумм `totals = np.array([12000, 8000, 24000, 6000, 16000])`. **Задача 24 (условная логика: `where`, `select`, `clip`).** Дан массив сумм `totals = np.array([4000, 8000, 12000, 16000, 24000, 30000, 10000])`. ## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Задача 25 (сортировка).** Дан массив сумм `totals = np.array([12000, 8000, 24000, 6000, 16000, 20000, 10000])`. Получите: отсортированный по возрастанию (`np.sort(totals)`); отсортированный по убыванию (`np.sort(-totals) * -1` или `np.sort(totals)[::-1]`); отсортируйте на месте через `totals.sort()` и убедитесь, что исходный массив изменился. Дана матрица `checkins` формы `(3, 4)`. Отсортируйте: по умолчанию (`axis=-1`, по строкам); по оси 0 (`axis=0`, по столбцам); с `axis=None` (уплощённый массив). В комментариях объясните: чем `np.sort(arr)` отличается от `arr.sort()` (какой возвращает копию, какой меняет на месте); почему сортировка по умолчанию выполняется по последней оси; как NumPy обрабатывает NaN при сортировке (проверьте, добавив `np.nan` — куда попадут NaN); какой алгоритм сортировки используется по умолчанию и как выбрать стабильный (`kind='stable'`) — в каком случае это важно (например, при сортировке по нескольким ключам).

**Задача 26 (`argsort` и топ-N).** Дан массив сумм на 10 гостей: `totals = np.array([12000, 8000, 24000, 6000, 16000, 20000, 10000, 4000, 30000, 15000])`. Получите: индексы, которые отсортировали бы массив (`np.argsort(totals)`); примените индексы к массиву (`totals[np.argsort(totals)]`) и убедитесь, что результат совпадает с `np.sort(totals)`. Дополнительно даны имена гостей `names = np.array(["A","B","C","D","E","F","G","H","I","J"])`. Получите гостей, отсортированных по сумме (`names[np.argsort(totals)]`). Найдите топ-3 самых дорогих: `names[np.argsort(-totals)[:3]]`. В комментариях объясните: зачем нужен `argsort`, если есть `sort` (как он позволяет сортировать параллельные массивы); как с его помощью отсортировать имена по сумме, не теряя соответствия; почему `argsort(-totals)` даёт порядок по убыванию; что такое «непрямая сортировка» и в каких задачах она необходима (ранжирование, топ-N, сортировка по нескольким полям).

**Задача 27 (`partition` и `argpartition`).** Дан массив из 1 000 000 случайных сумм. Найдите 10 самых дорогих бронирований: через полную сортировку (`np.sort(totals)[-10:]`); через `np.partition(totals, -10)[-10:]`; через `np.argpartition(totals, -10)[-10:]` — индексы. Замерьте время каждого подхода через `time.perf_counter()`. В комментариях объясните: почему `partition` работает за O(n), а `sort` — за O(n log n); что гарантирует `partition` (k-й элемент на своём месте, слева — меньше, справа — больше, но порядок внутри частей не определён); во сколько раз `partition` быстрее `sort` для 1 000 000 элементов (по экспериментальным данным); почему `argpartition` + `argsort` дают корректный топ-N (что нужно сделать после `argpartition`, чтобы упорядочить топ); в каком случае разница в скорости критична (интерактивные приложения, реальное время).

**Задача 28 (`searchsorted` и категоризация).** Даны границы ценовых категорий: `bounds = np.array([0, 7000, 15000, 25000, 100000])` (эконом, стандарт, комфорт, люкс, президентский). Даны суммы: `totals = np.array([4000, 8000, 12000, 16000, 24000, 30000])`. Определите категорию каждой суммы: `np.searchsorted(bounds, totals)` — индексы; `np.searchsorted(bounds, totals, side='right')` — сравните. Затем подсчитайте частоты: `np.bincount(np.searchsorted(bounds, totals))`. В комментариях объясните: что возвращает `searchsorted` (позицию для вставки, а не индекс найденного элемента); чем `side='left'` отличается от `side='right'` на граничных значениях (например, для суммы ровно 15000); как `searchsorted` + `bincount` заменяют `histogram` (в каком случае это удобнее); как `searchsorted` работает с большими массивами за O(log n) (бинарный поиск) и почему это важно для категоризации миллионов бронирований.

**Задача 24 (условная логика: `where`, `select`, `clip`).** Дан массив весов `weights = np.array([0.5, 2, 180, 3500, 4.5, 60, 5000])`. **Задача 38 (FFT и обработка сигналов).** Дан сигнал — сумма двух синусоид с шумом (например, суточные и недельные колебания аппетита):
```python
t = np.linspace(0, 1, 1000, endpoint=False)
signal = np.sin(2 * np.pi * 5 * t) + 0.5 * np.sin(2 * np.pi * 12 * t)
noisy = signal + np.random.default_rng(42).normal(0, 0.3, 1000)
```

Реализуйте через `np.einsum`: матричное умножение `A @ B`: `np.einsum('ik,kj->ij', A, B)`; умножение матрицы на вектор `A @ v`: `np.einsum('ik,k->i', A, v)`; скалярное произведение `v @ v`: `np.einsum('i,i->', v, v)`; внешнее произведение `v` на себя: `np.einsum('i,j->ij', v, v)`; след квадратной матрицы: `np.einsum('ii->', M)`. Дополнительно создайте структурированный массив посещений:
```python
visits_log = np.array([
    ('2025-01-01', 'Иван Петров', 'йога', 60),
    ('2025-01-02', 'Мария Иванова', 'бокс', 90),
    ('2025-01-03', 'Пётр Сидоров', 'кроссфит', 120),
], dtype=[('date', 'U10'), ('member', 'U30'), ('workout', 'U20'), ('duration', 'i4')])
```

**Задача 9 (arange vs linspace на практике).** Аптека строит график продаж за 30 дней. **Задача 11 (индексация).** Дан массив цен `prices = np.array([150, 320, 780, 90, 540, 1200, 250])`. **Задача 14 (fancy indexing).** Дан массив цен `prices = np.array([150, 320, 780, 90, 540, 1200, 250])`. **Задача 15 (view vs copy).** Дан массив цен `prices = np.arange(10) * 100 + 50`. Создайте: `view = prices[2:5]` — срез; `copy = prices[2:5].copy()` — копия. **Задача 19 (правила broadcasting).** Дана матрица `sales` формы `(3, 4)` (3 дня × 4 категории) и вектор поправок `offset` формы `(4,)` (например, изменение числа продаж по категориям). **Задача 20 (broadcasting на практике: нормализация).** Дана матрица характеристик лекарств формы `(365, 5)` — 365 дней × 5 характеристик (цена, срок годности, спрос, наценка, остаток). **Задача 21 (ось агрегации `axis`).** Дана матрица `sales` формы `(3, 4)`. **Задача 22 (`keepdims` и broadcasting).** Дана `sales` формы `(3, 4)`. **Задача 23 (универсальные функции ufunc).** Дан массив цен `prices = np.array([150, 320, 780, 90, 540])`. **Задача 24 (условная логика: `where`, `select`, `clip`).** Дан массив цен `prices = np.array([60, 90, 150, 320, 540, 1200, 2000])`. Для категоризации используйте `np.select`: условия — цена < 200, от 200 до 500, больше 500; значения — «дешёвое», «среднее», «дорогое»; значение по умолчанию — «?». В комментариях объясните: чем `where` отличается от `select` (когда два варианта, когда много); почему при перекрывающихся условиях `select` берёт первое (как это использовать для приоритетов); в каком случае `clip` лучше `where` (краткость, скорость, читаемость); чем `reduce` отличается от `accumulate` (свёртка в одно значение vs промежуточные результаты); что такое `identity` у ufunc (`np.add.identity == 0`) и зачем он нужен.

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** `np.array`, `dtype`, `shape`/`size`/`ndim`, функции создания (`zeros`, `ones`, `full`, `empty`, `arange`, `linspace`), случайные числа и воспроизводимость, память, единичные и диагональные матрицы.

**Задача 1 (создание массива, dtype, shape/size/ndim).** В IT-компании ведётся учёт зарплат 7 сотрудников: `[160000, 250000, 120000, 350000, 90000, 200000, 180000]` рублей. Создайте массив NumPy из этих значений. Выведите его `dtype`, `shape`, `size`, `ndim`. Дополнительно создайте второй массив из тех же значений, но с типом `float32`, и сравните `dtype` и `nbytes` с первым. Ожидаемый вывод: для первого массива `dtype=int64`, `shape=(7,)`, `size=7`, `ndim=1`; для второго — `dtype=float32`, `nbytes=28`. В комментариях поясните: почему NumPy выбрал именно `int64` для первого массива; чем `shape` отличается от `size`; почему `shape` у одномерного массива — это кортеж из одного элемента `(7,)`, а не `(7)`; что произойдёт при обращении `salaries[7]` и почему; как `nbytes` связан с `itemsize` и `size`.

**Задача 2 (подбор dtype под задачу).** В IT-компании есть три вида данных: зарплата сотрудника (от 30 000 до 1 000 000 рублей); стаж в годах (от 0 до 60); количество задач на доске (от 0 до 5 000). Для каждого вида создайте массив и подберите **минимальный** `dtype`, который вмещает все возможные значения без переполнения. Выведите `dtype`, `itemsize` и `nbytes` для каждого. Дополнительно проверьте, что произойдёт с выражением `np.array([100, 200, 250], dtype=np.uint8) + 10` — какой результат получится и почему. В комментариях объясните: почему для зарплаты не стоит брать `uint8` (какой максимум он вмещает и что будет при превышении); почему для стажа достаточно `uint8`, а для задач — `uint16`; как связаны `dtype`, `itemsize` и `nbytes`; почему NumPy не выбрасывает исключение при переполнении и как это может привести к незаметным ошибкам в расчётах фонда оплаты труда.

**Задача 3 (арифметические операции).** Даны зарплаты четырёх сотрудников `[160000, 250000, 120000, 350000]` рублей. Одной строкой без использования циклов:
- увеличьте все зарплаты на 10% (индексация);
- посчитайте зарплаты со скидкой 13% (НДФЛ);
- возведите зарплаты в квадрат (условно, для расчёта бонусов);
- прибавьте к каждой зарплате 5000 рублей (надбавка за удалёнку).

Для каждого результата выведите массив. Ожидаемый вывод для первой операции: `[176000. 275000. 132000. 385000.]`. В комментариях объясните: почему `salaries * 1.1` даёт поэлементное умножение, а не повторение, как у списков Python; что произойдёт с `dtype` после умножения на `1.1` и почему результат становится `float64`, а не остаётся `int64`; что случится, если попробовать `salaries + "руб."` и почему операция выбросит ошибку, а не вернёт строковый массив.

**Задача 4 (zeros/ones/full/empty).** IT-компания выделяет буферы под учёт: массив из 30 нулей — будущие ежедневные задачи; массив из 12 единиц — веса месяцев для сезонного анализа; массив из 5 значений `-1` — маркер «сотрудник в отпуске»; пустой массив из 3 элементов — буфер, который сразу заполните значениями `[160000, 250000, 120000]`. Для каждого выведите результат и `dtype`. Ожидаемый вывод для первого: `[0. 0. ... 0.]` с `dtype=float64`. В комментариях объясните: зачем нужен `np.empty`, если его содержимое непредсказуемо (когда это оправдано); почему `np.zeros(5)` по умолчанию даёт `float64`, а не `int64`, и как это связано с безопасностью вычислений; в каком случае `np.empty` быстрее `np.zeros` и насколько это критично для больших массивов; что произойдёт, если создать буфер через `np.empty` и случайно прочитать значения до заполнения.

**Задача 5 (arange, linspace).** IT-компания строит сетки для отчётов:
- `np.arange(1, 13)` — месяцы 1…12;
- `np.arange(0, 61, 5)` — стаж 0, 5, 10, …, 60 лет;
- `np.linspace(30000, 1000000, 10)` — 10 равномерных точек зарплаты от 30 000 до 1 000 000;
- `np.linspace(0, 1, 5)` — 5 точек для нормировки.

Для каждого массива выведите результат и `len()`. Дополнительно проверьте `len(np.arange(0, 1, 0.1))` — получилось ли 10? Ожидаемый вывод для последней проверки: `10`, но может быть и `9` или `11` в зависимости от платформы из-за погрешности float. В комментариях объясните: чем `arange` отличается от `linspace`; почему `arange` с вещественным шагом может дать неожиданное количество точек; в каком случае выбирать `arange` (когда важен шаг), а в каком — `linspace` (когда важно количество точек); как избежать ошибки при построении сеток для расчёта ФОТ, где нужна точная длина массива.

**Задача 6 (двумерные массивы, форма).** IT-компания хранит количество закрытых задач за 3 дня по 4 командам:
```python
tasks = np.array([
    [12, 15, 9, 13],   # понедельник: backend, frontend, QA, DevOps
    [11, 14, 10, 12],  # вторник
    [13, 16, 9, 14],   # среда
])
```
Выведите `shape`, `ndim`, `size`, `nbytes`. Получите: задачи во вторник (`tasks[1]` — одномерный массив); задачи третьей команды в среду (`tasks[2, 2]` — скаляр); задачи второй команды за все дни (`tasks[:, 1]` — одномерный массив). Ожидаемый вывод: `shape=(3, 4)`, `ndim=2`, `size=12`, `nbytes=96`. В комментариях объясните: почему `tasks[1]` возвращает одномерный массив, а не скаляр; что произойдёт при `tasks[5]` (индекс за границей); почему запись `tasks[1, 2]` эффективнее, чем `tasks[1][2]` (какие промежуточные объекты создаются); как `shape` связан с `ndim` и `size`.

**Задача 7 (память: itemsize и nbytes).** IT-компания хранит данные о 1 000 000 задач. Сравните память, занимаемую: массивом `float64` (оценка задачи в часах); массивом `float32` (та же оценка); массивом `uint16` (приоритет, 1–1000). Для каждого выведите `itemsize` и `nbytes` в мегабайтах. Ожидаемый вывод: `float64` — 7.63 МБ, `float32` — 3.81 МБ, `uint16` — 1.91 МБ. Дополнительно оцените размер Python-списка из 1 000 000 чисел через `sys.getsizeof` (только оболочки) и сравните с массивами NumPy. В комментариях объясните: почему `nbytes` не включает метаданные объекта (что ещё занимает память); почему для больших массивов выбор типа — это не микрооптимизация, а вопрос «поместится ли в RAM»; во сколько раз `uint16` экономнее `float64` и как это влияет на скорость обработки за счёт кэша процессора; в каких случаях экономия памяти важнее точности вычислений.

**Задача 8 (случайные числа, воспроизводимость).** IT-компания моделирует данные для симуляции загрузки. Создайте генератор `rng = np.random.default_rng(seed=42)` и сгенерируйте: 10 равномерных значений `rng.random(10)` — вероятности выполнения задачи в срок; 10 нормальных значений `rng.normal(160000, 50000, 10)` — зарплаты; 10 целых значений `rng.integers(0, 61, 10)` — стаж. Создайте второй генератор с тем же `seed=42` и убедитесь, что все три последовательности совпадают. Создайте третий генератор без `seed` и покажите, что значения отличаются. В комментариях объясните: зачем нужен `seed` и как он обеспечивает воспроизводимость экспериментов; чем `default_rng` отличается от старого `np.random.seed` (какой BitGenerator используется по умолчанию); почему в новом коде рекомендуется `Generator`, а не глобальный `np.random`; как `SeedSequence` помогает организовать независимые потоки случайных чисел для параллельных вычислений; в каких ситуациях отсутствие фиксированного `seed` — это хорошо, а в каких — плохо.

**Задача 9 (arange vs linspace на практике).** IT-компания строит график закрытых задач за 30 дней. Реализуйте массив дней двумя способами: `np.arange(1, 31)` — дни 1…30; `np.linspace(1, 30, 30)` — 30 точек от 1 до 30. Проверьте `len()`, первые и последние элементы — они должны совпадать. Дополнительно найдите ситуацию, где `arange` даёт неверное количество точек из-за погрешности float: `np.arange(0, 1, 0.1)` — сколько элементов получилось? Ожидаемый ответ: 10, но может быть и другое число на некоторых платформах. В комментариях объясните: почему `linspace` надёжнее, когда важно именно количество точек; в каком случае `arange` уместнее (когда важен шаг); почему `arange` с вещественным шагом опасен в расчётах, где неверная длина массива может привести к сдвигу данных; как проверить длину массива перед использованием и что делать, если она не совпадает с ожидаемой.

**Задача 10 (единичная и диагональная матрицы).** IT-компания хранит матрицу корреляций между 4 метриками сотрудника (зарплата, стаж, число задач, code review). Создайте: единичную матрицу `np.eye(4)` — начальное приближение (каждая метрика коррелирует сама с собой); диагональную матрицу `np.diag([1.0, 0.8, 0.6, 0.9])` — дисперсии метрик; матрицу со смещённой диагональю `np.diag([0.5, 0.3], k=1)` — корреляции соседних метрик. Выведите все три. Ожидаемый вывод для первой: диагональ из единиц, остальные нули. В комментариях объясните: чем `np.eye(4)` отличается от `np.identity(4)` (какой параметр позволяет задать вторую размерность); что делает параметр `k` в `np.diag` (положительное значение — диагональ выше главной, отрицательное — ниже); в каком случае диагональная матрица лучше обычной (память и скорость умножения, если большинство элементов нули); почему `np.eye` по умолчанию даёт `float`, а не `int`, и когда это важно.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** индексация, срезы, логическая и fancy-индексация, view vs copy, `reshape`, `ravel`/`flatten`, транспонирование и перестановка осей.

**Задача 11 (индексация).** Дан массив зарплат `salaries = np.array([160000, 250000, 120000, 350000, 90000, 200000, 180000])`. Получите: зарплату первого сотрудника (`salaries[0]`); зарплату последнего (`salaries[-1]`); зарплату третьего (`salaries[2]`). Дана матрица `tasks` формы `(3, 4)` из задачи 6. Получите: задачи второй команды в понедельник (`tasks[0, 1]`); все задачи во вторник (`tasks[1]`); все задачи второй команды (`tasks[:, 1]`). Ожидаемые результаты: `salaries[0] = 160000`, `salaries[-1] = 180000`, `tasks[0, 1] = 15`, `tasks[1] = [11, 14, 10, 12]`, `tasks[:, 1] = [15, 14, 16]`. В комментариях объясните: почему `tasks[1]` возвращает одномерный массив, а не скаляр; что произойдёт при `tasks[5]` (какое исключение и с каким сообщением); почему отрицательные индексы работают и для многомерных массивов; что вернёт `tasks[-1, -1]` и почему это последний элемент последней строки.

**Задача 12 (срезы).** Дан массив зарплат на 30 сотрудников: `salaries = np.arange(30) * 10000 + 50000`. Получите: первые 7 (`salaries[:7]`); последние 7 (`salaries[-7:]`); с 10-го по 20-й (`salaries[10:20]`); каждого второго (`salaries[::2]`); массив в обратном порядке (`salaries[::-1]`). Дана матрица `tasks` формы `(3, 4)`. Получите: первые две строки (`tasks[:2, :]`); второй и третий столбцы (`tasks[:, 1:3]`); подматрицу `2×2` из левого верхнего угла (`tasks[:2, :2]`). В комментариях объясните: включается ли правая граница среза (что вернёт `salaries[10:20]` — 10 или 11 элементов); что делает `salaries[::-1]` (как получается обратный порядок); почему срез возвращает view, а не копию; как проверить, что `salaries[:7].base is salaries`, и что это означает; в каком случае срез с шагом `[::2]` может быть опасен при изменении данных.

**Задача 13 (логическая индексация).** Дан массив зарплат `salaries = np.array([160000, 250000, 120000, 350000, 90000, 200000, 180000, 60000, 400000])`. Получите: высокооплачиваемых (зарплата выше 200 000) (`salaries[salaries > 200000]`); низкооплачиваемых (менее 100 000) (`salaries[salaries < 100000]`); средних от 100 000 до 200 000 включительно (`salaries[(salaries >= 100000) & (salaries <= 200000)]`); очень низких или очень высоких (`salaries[(salaries < 80000) | (salaries > 300000)]`). Дополнительно замените все зарплаты ниже 80 000 на 80 000 (`salaries[salaries < 80000] = 80000`) и выведите массив. В комментариях объясните: почему при комбинировании условий нужны скобки вокруг каждого условия; почему нельзя использовать `and`/`or` вместо `&`/`|` (какая ошибка возникнет); почему логическая индексация возвращает копию, а не view; как `np.where` мог бы заменить последнюю операцию и в каком случае он удобнее.

**Задача 14 (fancy indexing).** Дан массив зарплат `salaries = np.array([160000, 250000, 120000, 350000, 90000, 200000, 180000])`. Получите: зарплаты сотрудников с индексами 0, 2, 5 (`salaries[[0, 2, 5]]`); те же сотрудники в обратном порядке (`salaries[[5, 2, 0]]`); сотрудники 0, 0, 1, 1, 2 с повторениями (`salaries[[0, 0, 1, 1, 2]]`). Дана матрица `tasks` формы `(3, 4)`. Получите: строки 0 и 2 (`tasks[[0, 2]]`); столбцы 1 и 3 (`tasks[:, [1, 3]]`); диагональные элементы `tasks[0, 0], tasks[1, 1], tasks[2, 2]` (`tasks[[0, 1, 2], [0, 1, 2]]`); прямоугольную подматрицу через `np.ix_([0, 2], [1, 3])` — сравните с `tasks[[0, 2], [1, 3]]` и объясните разницу. В комментариях объясните: чем fancy indexing отличается от среза; почему fancy indexing всегда возвращает копию; зачем нужен `np.ix_` (когда без него выбираются только диагональные элементы); что произойдёт, если индексы выйдут за границу массива.

**Задача 15 (view vs copy).** Дан массив зарплат `salaries = np.arange(10) * 10000 + 50000`. Создайте: `view = salaries[2:5]` — срез; `copy = salaries[2:5].copy()` — копия. Измените `view[0] = 999999` и `copy[0] = 111111`. Выведите `salaries`, `view`, `copy`. Проверьте: `view.base is salaries` → `True`; `copy.base is None` → `True`; `np.shares_memory(salaries, view)` → `True`; `np.shares_memory(salaries, copy)` → `False`. В комментариях объясните: почему NumPy возвращает view для срезов, а не копию (какая экономия достигается); в каком случае это приводит к неожиданным ошибкам (например, при передаче среза в функцию, которая меняет данные); как защититься — вызвать `.copy()`; почему view экономит память, но опасен для больших массивов, если держит ссылку на оригинал (утечка памяти); как проверить, является ли массив view или копией.

**Задача 16 (reshape).** Дан массив из 12 зарплат: `salaries = np.arange(12) * 10000 + 50000`. Переформируйте его: в матрицу `3×4` (`salaries.reshape(3, 4)`); в матрицу `2×6` (`salaries.reshape(2, 6)`); в трёхмерный массив `2×2×3` (`salaries.reshape(2, 2, 3)`); в матрицу `3×4` с автоматическим определением второй размерности (`salaries.reshape(3, -1)`). Попробуйте `salaries.reshape(5, 5)` — получите ошибку и объясните, почему. В комментариях объясните: почему `reshape` не меняет данные, а только форму (что происходит с блоком памяти); когда `reshape` возвращает view, а когда копию (проверьте `.base` после транспонирования); зачем нужен `-1` и что произойдёт, если указать две `-1`; как `reshape` связан с порядком хранения в памяти (`order='C'` vs `order='F'` — в каком порядке читаются элементы).

**Задача 17 (ravel и flatten).** Дана матрица `tasks` формы `(3, 4)`. Получите одномерный массив тремя способами: `tasks.ravel()`; `tasks.flatten()`; `tasks.reshape(-1)`. Сравните `.base` у каждого результата: у `ravel` и `reshape(-1)` — ссылка на `tasks`, у `flatten` — `None`. Измените элемент в `ravel` и убедитесь, что `tasks` изменилась; измените элемент в `flatten` и убедитесь, что `tasks` не изменилась. В комментариях объясните: в каком случае выбирать `ravel` (когда не нужно изменять), а в каком `flatten` (когда нужна независимая копия); почему `flatten` медленнее (какие данные копируются); что такое «непрерывный массив в C-порядке» и почему `ravel` иногда тоже возвращает копию (например, после транспонирования); как это связано с производительностью при обработке больших массивов.

**Задача 18 (транспонирование и перестановка осей).** Дан трёхмерный массив `data` формы `(2, 3, 4)` — например, 2 офиса компании × 3 дня × 4 команды. Получите: `data.T` — обратный порядок осей, форма `(4, 3, 2)`; `np.transpose(data, (1, 0, 2))` — поменять оси 0 и 1, форма `(3, 2, 4)`; `np.swapaxes(data, 0, 2)` — поменять оси 0 и 2, форма `(4, 3, 2)`; `np.moveaxis(data, 0, -1)` — переместить ось 0 в конец, форма `(3, 4, 2)`. Для двумерной `tasks` формы `(3, 4)` сравните `tasks.T` и `tasks.transpose()` — результат должен совпадать. В комментариях объясните: почему транспонирование возвращает view, а не копию (что меняется в метаданных); как `moveaxis` отличается от `swapaxes` (когда нужен каждый); в каком случае транспонирование реально копирует данные (например, после `tasks[::2]`, когда массив не непрерывный); как проверить, является ли результат view или копией.

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** broadcasting (правила, ошибки, практика), `axis`, `keepdims`, универсальные функции (`ufunc`), методы `reduce`/`accumulate`/`outer`, условная логика (`where`/`select`/`clip`).

Для каждой пары выведите формы и результат. В комментариях объясните три правила broadcasting: выравнивание размерностей (добавление единиц слева); совместимость по осям (равны или одна равна 1); результирующая форма — поэлементный максимум; как прочитать сообщение об ошибке `operands could not be broadcast together with shapes (3,4) (3,)` и что оно означает; в каком случае broadcasting может неявно создать огромный массив и как это предотвратить.

**Задача 25 (сортировка).** Дан массив зарплат `salaries = np.array([160000, 250000, 120000, 350000, 90000, 200000, 180000])`. Получите: отсортированный по возрастанию (`np.sort(salaries)`); отсортированный по убыванию (`np.sort(-salaries) * -1` или `np.sort(salaries)[::-1]`); отсортируйте на месте через `salaries.sort()` и убедитесь, что исходный массив изменился. Дана матрица `tasks` формы `(3, 4)`. Отсортируйте: по умолчанию (`axis=-1`, по строкам); по оси 0 (`axis=0`, по столбцам); с `axis=None` (уплощённый массив). В комментариях объясните: чем `np.sort(arr)` отличается от `arr.sort()` (какой возвращает копию, какой меняет на месте); почему сортировка по умолчанию выполняется по последней оси; как NumPy обрабатывает NaN при сортировке (проверьте, добавив `np.nan` — куда попадут NaN); какой алгоритм сортировки используется по умолчанию и как выбрать стабильный (`kind='stable'`) — в каком случае это важно (например, при сортировке по нескольким ключам).

**Задача 26 (`argsort` и топ-N).** Дан массив зарплат на 10 сотрудников: `salaries = np.array([160000, 250000, 120000, 350000, 90000, 200000, 180000, 60000, 400000, 210000])`. Получите: индексы, которые отсортировали бы массив (`np.argsort(salaries)`); примените индексы к массиву (`salaries[np.argsort(salaries)]`) и убедитесь, что результат совпадает с `np.sort(salaries)`. Дополнительно даны имена сотрудников `names = np.array(["A","B","C","D","E","F","G","H","I","J"])`. Получите сотрудников, отсортированных по зарплате (`names[np.argsort(salaries)]`). Найдите топ-3 самых высокооплачиваемых: `names[np.argsort(-salaries)[:3]]`. В комментариях объясните: зачем нужен `argsort`, если есть `sort` (как он позволяет сортировать параллельные массивы); как с его помощью отсортировать имена по зарплате, не теряя соответствия; почему `argsort(-salaries)` даёт порядок по убыванию; что такое «непрямая сортировка» и в каких задачах она необходима (ранжирование, топ-N, сортировка по нескольким полям).

**Задача 27 (`partition` и `argpartition`).** Дан массив из 1 000 000 случайных зарплат. Найдите 10 самых высокооплачиваемых сотрудников: через полную сортировку (`np.sort(salaries)[-10:]`); через `np.partition(salaries, -10)[-10:]`; через `np.argpartition(salaries, -10)[-10:]` — индексы. Замерьте время каждого подхода через `time.perf_counter()`. В комментариях объясните: почему `partition` работает за O(n), а `sort` — за O(n log n); что гарантирует `partition` (k-й элемент на своём месте, слева — меньше, справа — больше, но порядок внутри частей не определён); во сколько раз `partition` быстрее `sort` для 1 000 000 элементов (по экспериментальным данным); почему `argpartition` + `argsort` дают корректный топ-N (что нужно сделать после `argpartition`, чтобы упорядочить топ); в каком случае разница в скорости критична (интерактивные приложения, реальное время).

**Задача 28 (`searchsorted` и категоризация).** Даны границы грейдов: `bounds = np.array([0, 100000, 200000, 350000, 1000000])` (junior, middle, senior, lead, principal). **Задача 34 (линейная алгебра).** Дана система линейных уравнений `A @ x = b`:
```python
A = np.array([[2.0, 1.0], [1.0, 3.0]])
b = np.array([5.0, 10.0])
```
Решите систему: через `np.linalg.solve(A, b)`; через `np.linalg.inv(A) @ b` — сравните результаты и объясните, какой способ лучше. Дополнительно вычислите: `np.linalg.det(A)` — определитель; `np.linalg.eig(A)` — собственные значения и векторы; `np.linalg.norm(b)` — евклидова норма; `np.linalg.norm(b, ord=1)` и `np.linalg.norm(b, ord=np.inf)` — другие нормы. В комментариях объясните: почему `solve` предпочтительнее `inv(A) @ b` (скорость, устойчивость, накопление ошибок при обращении матрицы); что такое собственные значения (на примере: для диагональной матрицы — сами диагональные элементы); в каком случае матрица не имеет обратной (определитель равен нулю — что это означает геометрически); как нормы используются для оценки сходства сотрудников (косинусное сходство компетенций, норма разности векторов).

Дополнительно создайте структурированный массив задач:
```python
tasks_log = np.array([
    ('2025-01-01', 'Иван Петров', 'backend', 8),
    ('2025-01-02', 'Мария Иванова', 'frontend', 6),
    ('2025-01-03', 'Пётр Сидоров', 'devops', 10),
], dtype=[('date', 'U10'), ('employee', 'U30'), ('team', 'U20'), ('hours', 'i4')])
```

В комментариях объясните: чем интерполяция отличается от аппроксимации (интерполяция проходит через все точки, аппроксимация — приближается); почему полином высокой степени может «выстреливать» за пределы данных (феномен Рунге) и как это связано с переобучением; в каком случае `np.interp` (кусочно-линейная) лучше полинома (когда данные шумные или их мало); как интерполяция применяется для восстановления пропущенных дней в отчёте полиции; какие ещё методы интерполяции существуют (сплайны, кубическая интерполяция).

**Задача 40 (`einsum` и структурированные массивы).** Даны: `A` формы `(3, 4)` — матрица «отделы × типы вызовов»; `B` формы `(4, 5)` — матрица «типы × дни»; `v` формы `(4,)` — вектор весов типов вызовов. Реализуйте через `np.einsum`: матричное умножение `A @ B`: `np.einsum('ik,kj->ij', A, B)`; умножение матрицы на вектор `A @ v`: `np.einsum('ik,k->i', A, v)`; скалярное произведение `v @ v`: `np.einsum('i,i->', v, v)`; внешнее произведение `v` на себя: `np.einsum('i,j->ij', v, v)`; след квадратной матрицы: `np.einsum('ii->', M)`. Дополнительно создайте структурированный массив дел:
```python
cases_log = np.array([
    ('2025-01-01', 'Иван Петров', 'кража', 3),
    ('2025-01-02', 'Мария Иванова', 'ДТП', 6),
    ('2025-01-03', 'Пётр Сидоров', 'ЧП', 10),
], dtype=[('date', 'U10'), ('officer', 'U30'), ('type', 'U20'), ('danger', 'i4')])
```
Вычислите средний уровень опасности (`cases_log['danger'].mean()`) и отфильтруйте дела с уровнем опасности выше 5. В комментариях объясните: в каком случае `einsum` читаемее, чем `@` (когда операция нестандартная); почему структурированные массивы удобны для табличных данных, но не заменяют Pandas (какие операции Pandas удобнее); как `einsum` обобщается на тензоры (`'bij,bjk->bik'`) и где это применяется (пакетная обработка); как структурированный массив связан с `@dataclass` из Python OOP (оба хранят разнородные поля, но NumPy — это C-структуры в памяти с фиксированным расположением).

В комментариях объясните: почему `linspace` надёжнее, когда важно именно количество точек; в каком случае `arange` уместнее (когда важен шаг); почему `arange` с вещественным шагом опасен в расчётах, где неверная длина массива может привести к сдвигу данных; как проверить длину массива перед использованием и что делать, если она не совпадает с ожидаемой.

**Задача 10 (единичная и диагональная матрицы).** Салон хранит матрицу корреляций между 4 показателями услуги (цена, длительность, спрос, рейтинг). Создайте: единичную матрицу `np.eye(4)` — начальное приближение (каждый показатель коррелирует сам с собой); диагональную матрицу `np.diag([1.0, 0.8, 0.6, 0.9])` — дисперсии показателей; матрицу со смещённой диагональю `np.diag([0.5, 0.3], k=1)` — корреляции соседних показателей. Выведите все три. Ожидаемый вывод для первой: диагональ из единиц, остальные нули. В комментариях объясните: чем `np.eye(4)` отличается от `np.identity(4)` (какой параметр позволяет задать вторую размерность); что делает параметр `k` в `np.diag` (положительное значение — диагональ выше главной, отрицательное — ниже); в каком случае диагональная матрица лучше обычной (память и скорость умножения, если большинство элементов нули); почему `np.eye` по умолчанию даёт `float`, а не `int`, и когда это важно.

**Что изучается:** индексация, срезы, логическая и fancy-индексация, view vs copy, `reshape`, `ravel`/`flatten`, транспонирование и перестановка осей.

**Задача 11 (индексация).** Дан массив цен `prices = np.array([1500, 800, 2500, 1200, 3000, 600, 1800])`. Ожидаемые результаты: `prices[0] = 1500`, `prices[-1] = 1800`, `bookings[0, 1] = 15`, `bookings[1] = [11, 14, 10, 12]`, `bookings[:, 1] = [15, 14, 16]`. В комментариях объясните: почему `bookings[1]` возвращает одномерный массив, а не скаляр; что произойдёт при `bookings[5]` (какое исключение и с каким сообщением); почему отрицательные индексы работают и для многомерных массивов; что вернёт `bookings[-1, -1]` и почему это последний элемент последней строки.

В комментариях объясните: включается ли правая граница среза (что вернёт `prices[10:20]` — 10 или 11 элементов); что делает `prices[::-1]` (как получается обратный порядок); почему срез возвращает view, а не копию; как проверить, что `prices[:7].base is prices`, и что это означает; в каком случае срез с шагом `[::2]` может быть опасен при изменении данных.

**Задача 13 (логическая индексация).** Дан массив цен `prices = np.array([1500, 800, 2500, 1200, 3000, 600, 1800, 400, 5000])`. **Задача 14 (fancy indexing).** Дан массив цен `prices = np.array([1500, 800, 2500, 1200, 3000, 600, 1800])`. **Задача 15 (view vs copy).** Дан массив цен `prices = np.arange(10) * 100 + 300`. Создайте: `view = prices[2:5]` — срез; `copy = prices[2:5].copy()` — копия. **Задача 16 (reshape).** Дан массив из 12 цен: `prices = np.arange(12) * 100 + 300`. **Задача 17 (ravel и flatten).** Дана матрица `bookings` формы `(3, 4)`. Получите одномерный массив тремя способами: `bookings.ravel()`; `bookings.flatten()`; `bookings.reshape(-1)`. **Задача 19 (правила broadcasting).** Дана матрица `bookings` формы `(3, 4)` (3 дня × 4 услуги) и вектор поправок `offset` формы `(4,)` (например, изменение числа записей по услугам). **Задача 20 (broadcasting на практике: нормализация).** Дана матрица характеристик услуг формы `(365, 5)` — 365 дней × 5 характеристик (цена, длительность, спрос, рейтинг, себестоимость). **Задача 21 (ось агрегации `axis`).** Дана матрица `bookings` формы `(3, 4)`. **Задача 23 (универсальные функции ufunc).** Дан массив цен `prices = np.array([1500, 800, 2500, 1200, 3000])`. **Задача 24 (условная логика: `where`, `select`, `clip`).** Дан массив цен `prices = np.array([400, 600, 800, 1500, 2500, 3000, 5000])`.
```python
t = np.linspace(0, 1, 1000, endpoint=False)
signal = np.sin(2 * np.pi * 5 * t) + 0.5 * np.sin(2 * np.pi * 12 * t)
noisy = signal + np.random.default_rng(42).normal(0, 0.3, 1000)
```

Отфильтруйте шум: обнулите все частоты выше 20 Гц и выполните `np.fft.ifft`. Сравните ошибку до и после фильтрации. В комментариях объясните: что такое FFT (как раскладывает сигнал на частоты); почему длины, кратные степени двойки, эффективнее (какой алгоритм используется); как амплитуда связана с коэффициентом синусоиды (почему умножаем на 2 и делим на длину); как FFT применяется к анализу сезонности спроса в турагентстве (выявление периодических пиков); что такое спектр и как его интерпретировать.
