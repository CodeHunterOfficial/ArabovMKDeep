## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В библиотеке ведётся учёт книг. Создайте класс `Book` с полями `title` (название), `author` (автор) и `year` (год издания). Реализуйте конструктор, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о книге в формате: «Название» — Автор, год. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод: `«Война и мир» — Л. Толстой, 1869`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждой книги в библиотеке есть инвентарный номер, который нельзя менять или читать напрямую — это внутренняя служебная информация. Создайте класс `BookCard`, в котором инвентарный номер хранится в приватном поле `__inventory_id`. Реализуйте метод `set_inventory_id(value)`, предназначенный только для установки номера при регистрации книги, и метод `get_inventory_id()` для чтения значения. Повторная попытка изменить номер после регистрации должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте карточку, установите номер, попробуйте изменить его ещё раз и убедитесь, что номер остался прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `card.__inventory_id` — будет ошибка `AttributeError`, но через name mangling (`card._BookCard__inventory_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В библиотеке хранятся разные типы изданий — книги, журналы, газеты. Создайте базовый класс `Publication` с методом `get_type()`, возвращающим строку «Издание». Унаследуйте от него классы `Book` и `Magazine`, переопределив в каждом метод `get_type()` так, чтобы он возвращал «Книга» и «Журнал» соответственно. Создайте по одному объекту каждого класса и выведите их тип, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Библиотека выдаёт разные типы изданий, и для каждого нужно рассчитывать срок выдачи — у книги он один, у журнала другой, у газеты третий. Создайте классы `Book`, `Magazine` и `Newspaper`, у каждого из которых есть метод `loan_period()`, возвращающий количество дней: для книги — 30, для журнала — 14, для газеты — 3. Напишите функцию `show_loan_period(item)`, которая принимает любое издание и выводит срок его выдачи. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `Librarian` (библиотекарь), в котором есть три поля: `name` — публичное (доступно всем), `_department` — защищённое (отдел, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_department` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_Librarian__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Библиотеке нужна утилита для проверки ISBN-номеров, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `LibraryUtils` со статическим методом `is_valid_isbn(isbn)`, который возвращает `True`, если длина строки равна 13 (дополнительно можно проверить, что все символы — цифры), и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `LibraryUtils.is_valid_isbn("9783161484100")`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `LibraryItem` с абстрактным методом `describe()`. Этот класс описывает общее понятие «издание в библиотеке», но не может существовать сам по себе — у каждого конкретного издания должно быть своё описание. Унаследуйте от него классы `Book` и `Magazine`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание издания. Проверьте, что объект абстрактного класса `LibraryItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждой книги в библиотеке есть обложка — книга не может существовать без неё, и обложка создаётся вместе с книгой. Создайте класс `Cover` с методом `show()`, который выводит строку «Твёрдая обложка». Создайте класс `Book`, который содержит объект класса `Cover` внутри себя как поле (композиция) — объект `Cover` создаётся прямо в конструкторе книги. Метод `show_book()` у книги должен выводить название книги и вызывать метод `show()` у её обложки, демонстрируя, что книга «владеет» своей обложкой. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В библиотеке есть базовый класс `LibraryItem` с методом `get_info()`, который возвращает строку «Издание библиотеки». Унаследуйте от него класс `Book` и переопределите метод `get_info()` так, чтобы он возвращал строку «Книга: <название>». Создайте объект класса `Book`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Библиотекарю нужен метод для регистрации выдачи книг, который мог бы работать в разных ситуациях: иногда известна только книга, иногда книга и читатель, иногда ещё и срок выдачи. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `LoanDesk` с методом `register(book, reader=None, days=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Зарегистрирована выдача книги: <title>», «Книга "<title>" выдана читателю <reader>» или «Книга "<title>" выдана <reader> на <days> дней». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Книги в библиотеке часто выводят на экран — в каталогах, отчётах, списках. По умолчанию `print(book)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Book` магический метод `__str__`, который возвращает строку вида «Книга "Название" (Автор, год)». Убедитесь, что при вызове `print(book)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Book('Название', 'Автор', 1869)` — оно используется при выводе списка книг и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В библиотеке может оказаться несколько экземпляров одной и той же книги — например, при повторной закупке. Нужно, чтобы программа понимала, что это «одна и та же книга», и не дублировала её в каталоге. Реализуйте в классе `Book` магический метод `__eq__`, чтобы две книги считались равными, если совпадают их названия и авторы. Дополнительно реализуйте `__hash__`, чтобы книги можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что две одинаковые книги попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В библиотеке есть книжные полки, и иногда нужно объединять их или узнавать, сколько книг на полке. Создайте класс `Bookshelf` (книжная полка), в котором хранится список книг. Реализуйте магический метод `__add__` для объединения двух полок в одну (возвращается **новая** полка со всеми книгами) и метод `__len__`, возвращающий количество книг на полке. Продемонстрируйте работу: создайте две полки, объедините их через `shelf3 = shelf1 + shelf2` и выведите `len(shelf3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В библиотеке важно, чтобы год издания книги был корректным — нельзя допустить, например, год 3000 или год 1200. Перепишите класс `Book` так, чтобы поле `year` было реализовано через свойство `@property`. Проверки: год должен быть целым числом, и он должен лежать в диапазоне от 1500 до текущего года (используйте `from datetime import date; current_year = date.today().year`). При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `book.year` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `book.year = 3000` → ошибка, `book.year = "abc"` → ошибка, `book.year = 1869` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В библиотеку часто поступают данные о книгах в виде строк — например, из файла или базы данных: `"Война и мир;Л. Толстой;1869"`. Добавьте в класс `Book` классовый метод `from_string(cls, data)`, который создаёт объект книги из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание книги через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Библиотеке нужно вести общий счётчик всех книг — сколько всего книг зарегистрировано в системе. Добавьте в класс `Book` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждой новой книги. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько книг и выведите `Book.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`book.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В библиотеке возможна ситуация, когда читатель просит книгу, которой нет в наличии. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `BookNotFoundError(Exception)` и используйте его в классе `Library` (библиотека). Метод `find_book(title)` должен возвращать книгу, если она есть, и выбрасывать `BookNotFoundError` с понятным сообщением (например, с указанием названия), если книги нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В библиотеке читатель берёт книги — но книги не принадлежат читателю, они существуют отдельно и принадлежат библиотеке. Реализуйте класс `Reader` (читатель), который содержит список книг, взятых в библиотеке. Книги при этом существуют отдельно от читателя — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждой книги есть ссылка на `Library`, в которой она хранится. Продемонстрируйте, что при удалении читателя книги остаются в библиотеке. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В библиотеке работает человек, который одновременно и читатель, и сотрудник — например, библиотекарь, который сам любит читать. Создайте два класса-роли: `Reader` (читатель) и `Employee` (сотрудник). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `LibrarianReader`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `LibrarianReader.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Book` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Book` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Book` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Book(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В библиотеке может храниться огромное количество книг — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Book` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Book` атрибут `__slots__ = ("title", "author", "year")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `book.publisher = "..."` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Book` для полей `year` (например, диапазон 1500–2100) и `pages` (диапазон 1–5000), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В библиотеке должен быть только один каталог — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `Library` будет только один экземпляр. Проверьте, что `Library()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `LibraryCatalog` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Библиотека регулярно пополняется разными типами изданий — книгами, журналами, газетами. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `PublicationFactory` с методом `create(kind, **kwargs)`, который по строке `"book"`, `"magazine"` или `"newspaper"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному изданию каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Читатели библиотеки хотят получать уведомления, когда поступает новая книга по интересующей их теме. Реализуйте паттерн «Наблюдатель»: класс `Library` хранит список подписчиков (`Reader`) и при добавлении новой книги уведомляет всех через метод `notify()`. Продемонстрируйте, что при добавлении книги все читатели получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки читатель больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Library` со временем разросся: он и хранит книги, и выдаёт их, и отправляет уведомления, и ведёт отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Library` на несколько классов: `BookStorage` (хранение книг и поиск по названию), `LoanService` (выдача книг, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «читатель взял книгу». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Электронные книги (`EBook`) похожи на обычные книги, но имеют дополнительные поля — формат файла, размер. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`EBook(LibraryItem)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`EBook` содержит объект `LibraryItem` внутри себя как поле `self.item`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В библиотеке нужно создавать копии книг — например, для резервного каталога. При этом у книги есть вложенный объект `Cover`, и важно понимать, копируется ли он вместе с книгой или остаётся общим. Реализуйте в классе `Book` методы `__copy__` и `__deepcopy__`. Создайте книгу с вложенным объектом `Cover` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — обложка общая (`shallow.cover is book.cover` → `True`), при глубоком — своя (`deep.cover is book.cover` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В библиотеке есть разные объекты, у которых есть метод `describe()` — книги, журналы, газеты, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Book`, но и с любым другим классом, у которого есть `describe()` (например, классом `Robot`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В библиотеке есть книжная полка `Bookshelf`, и хочется перебирать книги на ней в цикле `for` — так же удобно, как список. Сейчас `for book in shelf:` не работает. Реализуйте в классе `Bookshelf` магические методы `__iter__` и `__next__` так, чтобы по полке можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте полку с несколькими книгами и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(shelf)` должен возвращать список всех книг.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В библиотеке книги на полке должны быть доступны по индексу — как элементы списка: `shelf[0]` возвращает первую книгу, `shelf[1] = book` заменяет вторую. Реализуйте в классе `Bookshelf` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `shelf[0]` — получение книги по индексу, `shelf[1] = new_book` — замена книги, срезы `shelf[0:2]` — получение подсписка, `for book in shelf` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `book in shelf`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В библиотеке есть операции, которые нужно выполнять «аккуратно»: например, при работе с каталогом книг его нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `LibraryCatalog` как контекстный менеджер: методы `__enter__` (открывает каталог и возвращает его) и `__exit__` (закрывает каталог, даже если было исключение). Продемонстрируйте использование через `with LibraryCatalog() as catalog:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — каталог всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В библиотеке есть разные способы расчёта штрафа за просрочку книги: фиксированный (100 рублей в день), прогрессивный (100 за первый день, 200 за второй, 300 за третий и т.д.) и «мягкий» (для студентов — половина от фиксированного). Реализуйте паттерн «Стратегия»: базовый класс `FineStrategy` с методом `calculate(days)`, и три наследника — `FixedFine`, `ProgressiveFine`, `StudentFine`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `StudentFine`, чтобы вызвать реализацию родителя и уменьшить результат вдвое. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В библиотеке появился новый тип издания — аудиокнига. Если добавлять её через изменение базового класса `LibraryItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `LibraryItem` с методом `loan_period()`; наследников `Book`, `Magazine`; функцию `process_item(item: LibraryItem)`, которая работает с любым изданием через родительский интерфейс. Добавьте `Audiobook` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В библиотеке есть служебный объект `FineCalculator`, который должен уметь рассчитывать штраф за просрочку. Удобно, чтобы его можно было вызывать как функцию: `calculator(days)`, а не `calculator.calculate(days)`. Реализуйте в классе `FineCalculator` магический метод `__call__(self, days)`, который возвращает сумму штрафа (например, 100 рублей за каждый день). Продемонстрируйте, что объект можно вызывать как функцию: `calc = FineCalculator(); calc(5)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В библиотеке есть класс `Book`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Book` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования книг по названию): два вызова `Book("X", "A", 2000)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В библиотеке есть базовый класс `LibraryItem` с методом `get_info()`. Нужно добавлять к изданию дополнительную информацию — например, «редкая книга», «новинка», «рекомендованная». При этом нельзя менять сам класс `LibraryItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `LibraryItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `RareBookDecorator` и `NewBookDecorator`, которые можно комбинировать: `RareBookDecorator(NewBookDecorator(book))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В библиотеке появилась внешняя система учёта, которая возвращает данные о книгах в другом формате — например, в виде словаря `{"name": "...", "writer": "...", "published": 1869}`. А весь остальной код библиотеки работает с классом `Book`, у которого поля `title`, `author`, `year`. Реализуйте паттерн «Адаптер»: класс `BookAdapter`, который принимает словарь и предоставляет интерфейс `Book` — свойства `title`, `author`, `year` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Book` — например, передать его в функцию, которая работает с книгой. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В библиотеке есть базовый класс `LibraryItem`, и постепенно в нём накопились методы, которые нужны не всем изданиям: `print_pages()` (только для печатных книг), `download()` (только для электронных), `listen()` (только для аудиокниг). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Printable`, `Downloadable`, `Listenable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `LibraryService`, который зависит не от конкретных классов (`Book`, `EBook`), а от абстракции `LibraryItem` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип издания (например, `Audiobook`, реализующий только `Listenable`) без изменения `LibraryService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В интернет-магазине ведётся учёт товаров. Создайте класс `Product` с полями `name` (название), `price` (цена) и `quantity` (количество на складе). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о товаре в формате: `Название — цена, количество на складе`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Ноутбук — 75000, на складе 12 шт.`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого товара в магазине есть закупочная цена, которую нельзя показывать покупателям и менять напрямую. Создайте класс `ProductCard`, в котором закупочная цена хранится в приватном поле `__cost_price`. Реализуйте метод `set_cost_price(value)`, предназначенный только для установки значения при добавлении товара, и метод `get_cost_price()` для внутреннего чтения. Попытка изменить цену после установки должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте карточку товара, установите закупочную цену, попробуйте изменить её ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `card.__cost_price` — будет ошибка `AttributeError`, но через name mangling (`card._ProductCard__cost_price`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В магазине продаются разные категории товаров. Создайте базовый класс `Goods` с методом `get_category()`, возвращающим строку «Товар». Унаследуйте от него классы `Electronics` и `Clothing`, переопределив в каждом метод `get_category()` так, чтобы он возвращал «Электроника» и «Одежда» соответственно. Создайте по одному объекту каждого класса и выведите их категорию, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Магазин рассчитывает стоимость доставки для разных типов товаров. Создайте классы `Electronics`, `Clothing` и `Food`, у каждого из которых есть метод `delivery_cost()`, возвращающий стоимость доставки: для электроники — 500, для одежды — 300, для еды — 150. Напишите функцию `show_delivery(item)`, которая принимает любой товар и выводит стоимость его доставки. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `Manager` (менеджер магазина), в котором есть три поля: `name` — публичное (доступно всем), `_department` — защищённое (отдел, доступно внутри класса и наследникам), `__bonus` — приватное (премия, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_department` — можно, но не рекомендуется, к `__bonus` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_Manager__bonus`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Магазину нужна утилита для проверки артикулов товаров, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `ShopUtils` со статическим методом `is_valid_sku(sku)`, который возвращает `True`, если строка начинается с букв «SKU» и её длина равна 10, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `ShopUtils.is_valid_sku("SKU1234567")`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `ShopItem` с абстрактным методом `describe()`. Этот класс описывает общее понятие «товар в магазине», но не может существовать сам по себе — у каждого конкретного товара должно быть своё описание. Унаследуйте от него классы `Electronics` и `Clothing`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание товара. Проверьте, что объект абстрактного класса `ShopItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждого товара в магазине есть упаковка — товар не может существовать без неё, и упаковка создаётся вместе с товаром. Создайте класс `Packaging` с методом `show()`, который выводит строку «Картонная коробка». Создайте класс `Product`, который содержит объект класса `Packaging` внутри себя как поле (композиция) — объект `Packaging` создаётся прямо в конструкторе товара. Метод `show_product()` у товара должен выводить название товара и вызывать метод `show()` у его упаковки, демонстрируя, что товар «владеет» своей упаковкой. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В магазине есть базовый класс `ShopItem` с методом `get_info()`, который возвращает строку «Товар магазина». Унаследуйте от него класс `Electronics` и переопределите метод `get_info()` так, чтобы он возвращал строку «Электроника: <название>». Создайте объект класса `Electronics`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Менеджеру нужен метод для оформления заказов, который мог бы работать в разных ситуациях: иногда известен только товар, иногда товар и количество, иногда ещё и покупатель. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `OrderDesk` с методом `create(product, count=None, customer=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Оформлен заказ на товар: <name>», «Оформлен заказ: <name> × <count>» или «Оформлен заказ: <name> × <count> для <customer>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Товары в магазине часто выводят на экран — в каталогах, корзине, отчётах. По умолчанию `print(product)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Product` магический метод `__str__`, который возвращает строку вида «Товар "Название" — 75000 ₽ (в наличии 12 шт.)». Убедитесь, что при вызове `print(product)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Product('Ноутбук', 75000, 12)` — оно используется при выводе списка товаров и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В магазине может оказаться несколько экземпляров одного и того же товара — например, при повторной поставке. Нужно, чтобы программа понимала, что это «один и тот же товар», и не дублировала его в каталоге. Реализуйте в классе `Product` магический метод `__eq__`, чтобы два товара считались равными, если совпадают их названия и цена. Дополнительно реализуйте `__hash__`, чтобы товары можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых товара попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В магазине есть корзины покупателей, и иногда нужно объединять их или узнавать, сколько товаров в корзине. Создайте класс `Cart` (корзина), в котором хранится список товаров. Реализуйте магический метод `__add__` для объединения двух корзин в одну (возвращается **новая** корзина со всеми товарами) и метод `__len__`, возвращающий количество товаров в корзине. Продемонстрируйте работу: создайте две корзины, объедините их через `cart3 = cart1 + cart2` и выведите `len(cart3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В магазине важно, чтобы цена товара была корректной — нельзя допустить отрицательную цену или ноль. Перепишите класс `Product` так, чтобы поле `price` было реализовано через свойство `@property`. Проверки: цена должна быть числом (int или float), и она должна быть строго больше нуля. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `product.price` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `product.price = -100` → ошибка, `product.price = 0` → ошибка, `product.price = "abc"` → ошибка, `product.price = 75000` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В магазин часто поступают данные о товарах в виде строк — например, из файла или базы данных: `"Ноутбук;75000;12"`. Добавьте в класс `Product` классовый метод `from_string(cls, data)`, который создаёт объект товара из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание товара через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Магазину нужно вести общий счётчик всех товаров — сколько всего товаров зарегистрировано в системе. Добавьте в класс `Product` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового товара. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько товаров и выведите `Product.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`product.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В магазине возможна ситуация, когда покупатель пытается купить товар, которого нет на складе. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `OutOfStockError(Exception)` и используйте его в классе `Store` (магазин). Метод `buy(product, count)` должен уменьшать количество товара на складе, если его достаточно, и выбрасывать `OutOfStockError` с понятным сообщением (например, с указанием названия и запрошенного количества), если товара не хватает. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В магазине покупатель складывает товары в корзину — но товары не принадлежат покупателю, они существуют отдельно и принадлежат магазину. Реализуйте класс `Customer` (покупатель), который содержит список товаров, добавленных в корзину. Товары при этом существуют отдельно от покупателя — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого товара есть ссылка на `Store`, в котором он продаётся. Продемонстрируйте, что при удалении покупателя товары остаются в магазине. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В магазине работает человек, который одновременно и покупатель, и сотрудник — например, продавец, который сам покупает товары со скидкой. Создайте два класса-роли: `Customer` (покупатель) и `Employee` (сотрудник). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `CustomerEmployee`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `CustomerEmployee.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Product` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Product` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Product` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Product(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В интернет-магазине может храниться огромное количество товаров — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Product` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Product` атрибут `__slots__ = ("name", "price", "quantity")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `product.discount = 10` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Product` для полей `price` (например, диапазон 1–1 000 000) и `quantity` (диапазон 0–10 000), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В магазине должен быть только один каталог товаров — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `Store` будет только один экземпляр. Проверьте, что `Store()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `StoreCatalog` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Магазин регулярно пополняется разными типами товаров — электроникой, одеждой, едой. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `ProductFactory` с методом `create(kind, **kwargs)`, который по строке `"electronics"`, `"clothing"` или `"food"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному товару каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Покупатели магазина хотят получать уведомления, когда поступает новый товар или начинается распродажа. Реализуйте паттерн «Наблюдатель»: класс `Store` хранит список подписчиков (`Customer`) и при добавлении нового товара уведомляет всех через метод `notify()`. Продемонстрируйте, что при добавлении товара все покупатели получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки покупатель больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Store` со временем разросся: он и хранит товары, и оформляет заказы, и отправляет уведомления, и ведёт отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Store` на несколько классов: `ProductStorage` (хранение товаров и поиск по названию), `OrderService` (оформление заказов, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «покупатель оформил заказ». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Цифровые товары (`DigitalProduct`) похожи на обычные товары, но имеют дополнительные поля — ссылка на скачивание, размер файла. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`DigitalProduct(ShopItem)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`DigitalProduct` содержит объект `ShopItem` внутри себя как поле `self.item`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В магазине нужно создавать копии товаров — например, для резервного каталога. При этом у товара есть вложенный объект `Packaging`, и важно понимать, копируется ли он вместе с товаром или остаётся общим. Реализуйте в классе `Product` методы `__copy__` и `__deepcopy__`. Создайте товар с вложенным объектом `Packaging` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — упаковка общая (`shallow.packaging is product.packaging` → `True`), при глубоком — своя (`deep.packaging is product.packaging` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В магазине есть разные объекты, у которых есть метод `describe()` — товары, категории, акции, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Product`, но и с любым другим классом, у которого есть `describe()` (например, классом `Category`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В магазине есть корзина `Cart`, и хочется перебирать товары в ней в цикле `for` — так же удобно, как список. Сейчас `for product in cart:` не работает. Реализуйте в классе `Cart` магические методы `__iter__` и `__next__` так, чтобы по корзине можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте корзину с несколькими товарами и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(cart)` должен возвращать список всех товаров.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В корзине товары должны быть доступны по индексу — как элементы списка: `cart[0]` возвращает первый товар, `cart[1] = product` заменяет второй. Реализуйте в классе `Cart` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `cart[0]` — получение товара по индексу, `cart[1] = new_product` — замена товара, срезы `cart[0:2]` — получение подсписка, `for product in cart` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `product in cart`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В магазине есть операции, которые нужно выполнять «аккуратно»: например, при оформлении заказа нужно открыть сессию, а в конце обязательно её закрыть — даже если произошла ошибка. Реализуйте класс `CheckoutSession` как контекстный менеджер: методы `__enter__` (открывает сессию и возвращает её) и `__exit__` (закрывает сессию, даже если было исключение). Продемонстрируйте использование через `with CheckoutSession() as session:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — сессия всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В магазине есть разные способы расчёта скидки: фиксированная (500 рублей), процентная (10% от суммы) и «VIP» (для постоянных покупателей — 20% от суммы). Реализуйте паттерн «Стратегия»: базовый класс `DiscountStrategy` с методом `calculate(total)`, и три наследника — `FixedDiscount`, `PercentDiscount`, `VipDiscount`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `VipDiscount`, чтобы вызвать реализацию `PercentDiscount` и увеличить процент скидки (или переиспользовать код родителя). Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В магазине появился новый тип товара — подписка (`Subscription`). Если добавлять её через изменение базового класса `ShopItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `ShopItem` с методом `delivery_cost()`; наследников `Electronics`, `Clothing`; функцию `process_item(item: ShopItem)`, которая работает с любым товаром через родительский интерфейс. Добавьте `Subscription` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В магазине есть служебный объект `PriceCalculator`, который должен уметь рассчитывать итоговую цену с учётом налога. Удобно, чтобы его можно было вызывать как функцию: `calculator(1000)`, а не `calculator.calculate(1000)`. Реализуйте в классе `PriceCalculator` магический метод `__call__(self, amount)`, который возвращает сумму с налогом (например, 20%). Продемонстрируйте, что объект можно вызывать как функцию: `calc = PriceCalculator(); calc(1000)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В магазине есть класс `Product`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Product` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования товаров по названию): два вызова `Product("Ноутбук", 75000, 12)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В магазине есть базовый класс `ShopItem` с методом `get_info()`. Нужно добавлять к товару дополнительную информацию — например, «подарочная упаковка», «экспресс-доставка», «страховка». При этом нельзя менять сам класс `ShopItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `ShopItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `GiftWrapDecorator` и `ExpressDeliveryDecorator`, которые можно комбинировать: `GiftWrapDecorator(ExpressDeliveryDecorator(product))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В магазине появилась внешняя система учёта поставщика, которая возвращает данные о товарах в другом формате — например, в виде словаря `{"name": "...", "cost": 75000, "stock": 12}`. А весь остальной код магазина работает с классом `Product`, у которого поля `name`, `price`, `quantity`. Реализуйте паттерн «Адаптер»: класс `ProductAdapter`, который принимает словарь и предоставляет интерфейс `Product` — свойства `name`, `price`, `quantity` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Product` — например, передать его в функцию, которая работает с товаром. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В магазине есть базовый класс `ShopItem`, и постепенно в нём накопились методы, которые нужны не всем товарам: `ship()` (только для физических товаров), `download()` (только для цифровых), `activate()` (только для подписок). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Shippable`, `Downloadable`, `Activatable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `OrderService`, который зависит не от конкретных классов (`PhysicalProduct`, `DigitalProduct`), а от абстракции `ShopItem` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип товара (например, `Subscription`, реализующий только `Activatable`) без изменения `OrderService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В университете ведётся учёт студентов. Создайте класс `Student` с полями `name` (имя), `group` (группа) и `gpa` (средний балл). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о студенте в формате: `Имя — группа, средний балл`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Иван Петров — ИУ7-21Б, средний балл 4.7`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого студента есть номер зачётной книжки, который нельзя менять или читать напрямую. Создайте класс `StudentRecord`, в котором номер зачётной книжки хранится в приватном поле `__record_id`. Реализуйте метод `set_record_id(value)`, предназначенный только для установки номера при зачислении, и метод `get_record_id()` для чтения значения. Повторная попытка изменить номер после зачисления должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте запись, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `record.__record_id` — будет ошибка `AttributeError`, но через name mangling (`record._StudentRecord__record_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В университете обучаются разные категории людей. Создайте базовый класс `Person` с методом `get_role()`, возвращающим строку «Человек». Унаследуйте от него классы `Student` и `Teacher`, переопределив в каждом метод `get_role()` так, чтобы он возвращал «Студент» и «Преподаватель» соответственно. Создайте по одному объекту каждого класса и выведите их роль, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Университет начисляет стипендию разным категориям студентов. Создайте классы `ExcellentStudent`, `GoodStudent` и `AverageStudent`, у каждого из которых есть метод `scholarship()`, возвращающий размер стипендии: для отличника — 5000, для хорошиста — 3000, для троечника — 0. Напишите функцию `show_scholarship(student)`, которая принимает любого студента и выводит размер его стипендии. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `Dean` (декан), в котором есть три поля: `name` — публичное (доступно всем), `_faculty` — защищённое (факультет, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_faculty` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_Dean__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Университету нужна утилита для проверки студенческих билетов, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `UniUtils` со статическим методом `is_valid_ticket(ticket)`, который возвращает `True`, если строка начинается с букв «ST» и её длина равна 8, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `UniUtils.is_valid_ticket("ST123456")`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `UniversityMember` с абстрактным методом `describe()`. Этот класс описывает общее понятие «участник университета», но не может существовать сам по себе — у каждого конкретного участника должно быть своё описание. Унаследуйте от него классы `Student` и `Teacher`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание участника. Проверьте, что объект абстрактного класса `UniversityMember` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждого студента есть студенческий билет — студент не может существовать без него, и билет создаётся вместе со студентом. Создайте класс `StudentID` с методом `show()`, который выводит строку «Студенческий билет выдан». Создайте класс `Student`, который содержит объект класса `StudentID` внутри себя как поле (композиция) — объект `StudentID` создаётся прямо в конструкторе студента. Метод `show_student()` у студента должен выводить имя студента и вызывать метод `show()` у его билета, демонстрируя, что студент «владеет» своим билетом. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В университете есть базовый класс `UniversityMember` с методом `get_info()`, который возвращает строку «Участник университета». Унаследуйте от него класс `Teacher` и переопределите метод `get_info()` так, чтобы он возвращал строку «Преподаватель: <имя>». Создайте объект класса `Teacher`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Секретарю деканата нужен метод для оформления заявок, который мог бы работать в разных ситуациях: иногда известен только студент, иногда студент и тема, иногда ещё и научный руководитель. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `RequestDesk` с методом `submit(student, topic=None, teacher=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Оформлена заявка от студента: <name>», «Оформлена заявка от <name> по теме: <topic>» или «Оформлена заявка от <name> по теме: <topic>, руководитель: <teacher>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Студентов часто выводят на экран — в списках группы, ведомостях, отчётах. По умолчанию `print(student)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Student` магический метод `__str__`, который возвращает строку вида «Студент "Иван Петров" (группа ИУ7-21Б, средний балл 4.7)». Убедитесь, что при вызове `print(student)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Student('Иван Петров', 'ИУ7-21Б', 4.7)` — оно используется при выводе списка студентов и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В университете может оказаться несколько записей об одном и том же студенте — например, при повторной подаче документов. Нужно, чтобы программа понимала, что это «один и тот же студент», и не дублировала его в списках. Реализуйте в классе `Student` магический метод `__eq__`, чтобы два студента считались равными, если совпадают их имя и группа. Дополнительно реализуйте `__hash__`, чтобы студентов можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых студента попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В университете есть учебные группы, и иногда нужно объединять их (например, при слиянии потоков) или узнавать, сколько студентов в группе. Создайте класс `Group`, в котором хранится список студентов. Реализуйте магический метод `__add__` для объединения двух групп в одну (возвращается **новая** группа со всеми студентами) и метод `__len__`, возвращающий количество студентов в группе. Продемонстрируйте работу: создайте две группы, объедините их через `group3 = group1 + group2` и выведите `len(group3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В университете важно, чтобы средний балл студента был корректным — нельзя допустить значение больше 5 или меньше 0. Перепишите класс `Student` так, чтобы поле `gpa` было реализовано через свойство `@property`. Проверки: балл должен быть числом (int или float), и он должен лежать в диапазоне от 0 до 5 включительно. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `student.gpa` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `student.gpa = 6` → ошибка, `student.gpa = -1` → ошибка, `student.gpa = "abc"` → ошибка, `student.gpa = 4.7` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В университет часто поступают данные о студентах в виде строк — например, из файла или базы данных: `"Иван Петров;ИУ7-21Б;4.7"`. Добавьте в класс `Student` классовый метод `from_string(cls, data)`, который создаёт объект студента из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание студента через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Университету нужно вести общий счётчик всех студентов — сколько всего студентов зачислено в систему. Добавьте в класс `Student` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового студента. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько студентов и выведите `Student.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`student.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В университете возможна ситуация, когда студент пытается записаться на курс, которого нет в учебном плане. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `CourseNotFoundError(Exception)` и используйте его в классе `University` (университет). Метод `enroll(student, course_name)` должен записывать студента на курс, если он есть, и выбрасывать `CourseNotFoundError` с понятным сообщением (например, с указанием названия курса), если курса нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В университете студент посещает курсы — но курсы не принадлежат студенту, они существуют отдельно и принадлежат университету. Реализуйте класс `Student` (студент), который содержит список курсов, на которые он записан. Курсы при этом существуют отдельно от студента — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого курса есть ссылка на `University`, в котором он читается. Продемонстрируйте, что при удалении студента курсы остаются в университете. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В университете работает человек, который одновременно и студент, и сотрудник — например, лаборант, который сам учится. Создайте два класса-роли: `Student` (студент) и `Employee` (сотрудник). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `WorkingStudent`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `WorkingStudent.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Student` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Student` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Student` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Student(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В университете может храниться огромное количество записей о студентах — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Student` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Student` атрибут `__slots__ = ("name", "group", "gpa")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `student.email = "..."` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Student` для полей `gpa` (например, диапазон 0–5) и `course_year` (диапазон 1–6), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В университете должен быть только один реестр студентов — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `University` будет только один экземпляр. Проверьте, что `University()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `StudentRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Университет регулярно пополняется разными типами участников — студентами, преподавателями, сотрудниками. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `MemberFactory` с методом `create(kind, **kwargs)`, который по строке `"student"`, `"teacher"` или `"employee"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному участнику каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Студенты хотят получать уведомления, когда появляется новый курс или открывается запись на спецкурс. Реализуйте паттерн «Наблюдатель»: класс `University` хранит список подписчиков (`Student`) и при добавлении нового курса уведомляет всех через метод `notify()`. Продемонстрируйте, что при добавлении курса все студенты получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки студент больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `University` со временем разросся: он и хранит студентов, и записывает на курсы, и отправляет уведомления, и ведёт отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `University` на несколько классов: `StudentStorage` (хранение студентов и поиск по имени), `EnrollmentService` (запись на курсы, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «студент записался на курс». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Аспиранты (`PhDStudent`) похожи на обычных студентов, но имеют дополнительные поля — научный руководитель, тема диссертации. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`PhDStudent(Student)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`PhDStudent` содержит объект `Student` внутри себя как поле `self.student`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В университете нужно создавать копии студенческих записей — например, для архивного каталога. При этом у студента есть вложенный объект `StudentID`, и важно понимать, копируется ли он вместе со студентом или остаётся общим. Реализуйте в классе `Student` методы `__copy__` и `__deepcopy__`. Создайте студента с вложенным объектом `StudentID` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — билет общий (`shallow.student_id is student.student_id` → `True`), при глубоком — свой (`deep.student_id is student.student_id` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В университете есть разные объекты, у которых есть метод `describe()` — студенты, преподаватели, курсы, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_member(member)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Student`, но и с любым другим классом, у которого есть `describe()` (например, классом `Course`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В университете есть учебная группа `Group`, и хочется перебирать студентов в ней в цикле `for` — так же удобно, как список. Сейчас `for student in group:` не работает. Реализуйте в классе `Group` магические методы `__iter__` и `__next__` так, чтобы по группе можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте группу с несколькими студентами и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(group)` должен возвращать список всех студентов.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В группе студенты должны быть доступны по индексу — как элементы списка: `group[0]` возвращает первого студента, `group[1] = student` заменяет второго. Реализуйте в классе `Group` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `group[0]` — получение студента по индексу, `group[1] = new_student` — замена студента, срезы `group[0:2]` — получение подсписка, `for student in group` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `student in group`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В университете есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных студентов её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `StudentDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with StudentDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В университете есть разные способы расчёта стипендии: фиксированная (5000 рублей), повышенная (7000 рублей) и социальная (для льготных категорий — 3000 рублей плюс надбавка). Реализуйте паттерн «Стратегия»: базовый класс `ScholarshipStrategy` с методом `calculate()`, и три наследника — `FixedScholarship`, `IncreasedScholarship`, `SocialScholarship`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `SocialScholarship`, чтобы вызвать реализацию родителя и добавить надбавку. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В университете появился новый тип участника — аспирант (`PhDStudent`). Если добавлять его через изменение базового класса `UniversityMember`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `UniversityMember` с методом `get_status()`; наследников `Student`, `Teacher`; функцию `process_member(member: UniversityMember)`, которая работает с любым участником через родительский интерфейс. Добавьте `PhDStudent` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_member` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В университете есть служебный объект `ScholarshipCalculator`, который должен уметь рассчитывать стипендию по среднему баллу. Удобно, чтобы его можно было вызывать как функцию: `calculator(4.7)`, а не `calculator.calculate(4.7)`. Реализуйте в классе `ScholarshipCalculator` магический метод `__call__(self, gpa)`, который возвращает размер стипендии (например, 5000 при GPA ≥ 4.5, 3000 при GPA ≥ 3.5, иначе 0). Продемонстрируйте, что объект можно вызывать как функцию: `calc = ScholarshipCalculator(); calc(4.7)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В университете есть класс `Student`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Student` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования студентов по имени и группе): два вызова `Student("Иван Петров", "ИУ7-21Б", 4.7)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В университете есть базовый класс `UniversityMember` с методом `get_info()`. Нужно добавлять к участнику дополнительную информацию — например, «отличник», «староста», «член совета». При этом нельзя менять сам класс `UniversityMember`. Реализуйте паттерн «Декоратор»: класс `MemberDecorator`, который содержит объект `UniversityMember` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `ExcellentStudentDecorator` и `GroupLeaderDecorator`, которые можно комбинировать: `ExcellentStudentDecorator(GroupLeaderDecorator(student))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В университете появилась внешняя система учёта абитуриентов, которая возвращает данные в другом формате — например, в виде словаря `{"name": "Иван Петров", "group": "ИУ7-21Б", "score": 4.7}`. А весь остальной код университета работает с классом `Student`, у которого поля `name`, `group`, `gpa`. Реализуйте паттерн «Адаптер»: класс `StudentAdapter`, который принимает словарь и предоставляет интерфейс `Student` — свойства `name`, `group`, `gpa` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Student` — например, передать его в функцию, которая работает со студентом. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В университете есть базовый класс `UniversityMember`, и постепенно в нём накопились методы, которые нужны не всем участникам: `enroll()` (только для студентов), `teach()` (только для преподавателей), `grade()` (только для проверяющих). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Enrollable`, `Teachable`, `Gradable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `EnrollmentService`, который зависит не от конкретных классов (`Student`, `PhDStudent`), а от абстракции `Enrollable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип участника (например, `ExchangeStudent`, реализующий только `Enrollable`) без изменения `EnrollmentService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В больнице ведётся учёт пациентов. Создайте класс `Patient` с полями `name` (имя), `age` (возраст) и `diagnosis` (диагноз). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о пациенте в формате: `Имя — возраст, диагноз`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Иван Петров — 45 лет, гипертония`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого пациента есть номер медицинской карты, который нельзя менять или читать напрямую. Создайте класс `MedicalCard`, в котором номер карты хранится в приватном поле `__card_id`. Реализуйте метод `set_card_id(value)`, предназначенный только для установки номера при поступлении, и метод `get_card_id()` для чтения значения. Повторная попытка изменить номер после регистрации должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте карту, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `card.__card_id` — будет ошибка `AttributeError`, но через name mangling (`card._MedicalCard__card_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В больнице работают разные категории сотрудников. Создайте базовый класс `HospitalWorker` с методом `get_position()`, возвращающим строку «Сотрудник». Унаследуйте от него классы `Doctor` и `Nurse`, переопределив в каждом метод `get_position()` так, чтобы он возвращал «Врач» и «Медсестра» соответственно. Создайте по одному объекту каждого класса и выведите их должность, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Больница рассчитывает стоимость приёма для разных специалистов. Создайте классы `Therapist`, `Surgeon` и `Dentist`, у каждого из которых есть метод `appointment_cost()`, возвращающий стоимость приёма: у терапевта — 1000, у хирурга — 2500, у стоматолога — 1800. Напишите функцию `show_cost(doctor)`, которая принимает любого врача и выводит стоимость его приёма. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `HeadDoctor` (главный врач), в котором есть три поля: `name` — публичное (доступно всем), `_department` — защищённое (отделение, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_department` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_HeadDoctor__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Больнице нужна утилита для проверки номеров страховых полисов, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `HospitalUtils` со статическим методом `is_valid_policy(policy)`, который возвращает `True`, если строка состоит только из цифр и её длина равна 16, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `HospitalUtils.is_valid_policy("1234567890123456")`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `HospitalMember` с абстрактным методом `describe()`. Этот класс описывает общее понятие «участник больницы», но не может существовать сам по себе — у каждого конкретного участника должно быть своё описание. Унаследуйте от него классы `Doctor` и `Patient`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание участника. Проверьте, что объект абстрактного класса `HospitalMember` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждого пациента есть палата — пациент не может существовать без неё в стационаре, и палата создаётся вместе с пациентом. Создайте класс `Room` с методом `show()`, который выводит строку «Палата №» вместе с номером. Создайте класс `Patient`, который содержит объект класса `Room` внутри себя как поле (композиция) — объект `Room` создаётся прямо в конструкторе пациента. Метод `show_patient()` у пациента должен выводить имя пациента и вызывать метод `show()` у его палаты, демонстрируя, что пациент «владеет» своей палатой. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В больнице есть базовый класс `HospitalMember` с методом `get_info()`, который возвращает строку «Участник больницы». Унаследуйте от него класс `Doctor` и переопределите метод `get_info()` так, чтобы он возвращал строку «Врач: <имя>». Создайте объект класса `Doctor`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Регистратору нужен метод для оформления записей на приём, который мог бы работать в разных ситуациях: иногда известен только пациент, иногда пациент и врач, иногда ещё и время приёма. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `ReceptionDesk` с методом `book(patient, doctor=None, time=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Оформлена запись пациента: <name>», «Пациент <name> записан к врачу: <doctor>» или «Пациент <name> записан к <doctor> на <time>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Пациентов часто выводят на экран — в списках отделения, картах, отчётах. По умолчанию `print(patient)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Patient` магический метод `__str__`, который возвращает строку вида «Пациент "Иван Петров" (45 лет, диагноз: гипертония)». Убедитесь, что при вызове `print(patient)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Patient('Иван Петров', 45, 'гипертония')` — оно используется при выводе списка пациентов и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В больнице может оказаться несколько записей об одном и том же пациенте — например, при повторной госпитализации. Нужно, чтобы программа понимала, что это «один и тот же пациент», и не дублировала его в списках. Реализуйте в классе `Patient` магический метод `__eq__`, чтобы два пациента считались равными, если совпадают их имя и возраст. Дополнительно реализуйте `__hash__`, чтобы пациентов можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых пациента попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В больнице есть списки пациентов по отделениям, и иногда нужно объединять их (например, при слиянии отделений) или узнавать, сколько пациентов в списке. Создайте класс `Ward` (отделение), в котором хранится список пациентов. Реализуйте магический метод `__add__` для объединения двух отделений в одно (возвращается **новое** отделение со всеми пациентами) и метод `__len__`, возвращающий количество пациентов в отделении. Продемонстрируйте работу: создайте два отделения, объедините их через `ward3 = ward1 + ward2` и выведите `len(ward3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В больнице важно, чтобы возраст пациента был корректным — нельзя допустить отрицательный возраст или значение больше 150. Перепишите класс `Patient` так, чтобы поле `age` было реализовано через свойство `@property`. Проверки: возраст должен быть целым числом и лежать в диапазоне от 0 до 150 включительно. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `patient.age` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `patient.age = -5` → ошибка, `patient.age = 200` → ошибка, `patient.age = "abc"` → ошибка, `patient.age = 45` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В больницу часто поступают данные о пациентах в виде строк — например, из файла или базы данных: `"Иван Петров;45;гипертония"`. Добавьте в класс `Patient` классовый метод `from_string(cls, data)`, который создаёт объект пациента из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание пациента через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Больнице нужно вести общий счётчик всех пациентов — сколько всего пациентов зарегистрировано в системе. Добавьте в класс `Patient` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового пациента. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько пациентов и выведите `Patient.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`patient.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В больнице возможна ситуация, когда врач пытается выписать пациента, которого нет в отделении. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `PatientNotFoundError(Exception)` и используйте его в классе `Hospital` (больница). Метод `discharge(patient_name)` должен выписывать пациента, если он есть, и выбрасывать `PatientNotFoundError` с понятным сообщением (например, с указанием имени), если пациента нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В больнице врач ведёт приём пациентов — но пациенты не принадлежат врачу, они существуют отдельно и принадлежат больнице. Реализуйте класс `Doctor` (врач), который содержит список пациентов, находящихся на приёме. Пациенты при этом существуют отдельно от врача — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого пациента есть ссылка на `Hospital`, в которой он лечится. Продемонстрируйте, что при удалении врача пациенты остаются в больнице. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В больнице работает человек, который одновременно и врач, и пациент — например, врач, который сам лечится в этой же больнице. Создайте два класса-роли: `Doctor` (врач) и `Patient` (пациент). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `DoctorPatient`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `DoctorPatient.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Patient` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Patient` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Patient` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Patient(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В больнице может храниться огромное количество записей о пациентах — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Patient` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Patient` атрибут `__slots__ = ("name", "age", "diagnosis")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `patient.phone = "..."` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Patient` для полей `age` (например, диапазон 0–150) и `temperature` (диапазон 34.0–42.0), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В больнице должен быть только один реестр пациентов — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `Hospital` будет только один экземпляр. Проверьте, что `Hospital()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `PatientRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Больница регулярно пополняется разными типами сотрудников — врачами, медсёстрами, санитарами. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `WorkerFactory` с методом `create(kind, **kwargs)`, который по строке `"doctor"`, `"nurse"` или `"orderly"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному сотруднику каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Пациенты хотят получать уведомления, когда приходит результат анализа или открывается запись к нужному специалисту. Реализуйте паттерн «Наблюдатель»: класс `Hospital` хранит список подписчиков (`Patient`) и при поступлении нового результата уведомляет всех через метод `notify()`. Продемонстрируйте, что при появлении результата все пациенты получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки пациент больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Hospital` со временем разросся: он и хранит пациентов, и ведёт приёмы, и рассылает уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Hospital` на несколько классов: `PatientStorage` (хранение пациентов и поиск по имени), `AppointmentService` (ведение приёмов, назначения), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «пациент записан на приём». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Пациенты дневного стационара (`DayPatient`) похожи на обычных пациентов, но имеют дополнительные поля — время прихода, время ухода. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`DayPatient(Patient)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`DayPatient` содержит объект `Patient` внутри себя как поле `self.patient`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В больнице нужно создавать копии медицинских карт — например, для архивного хранения. При этом у пациента есть вложенный объект `Room`, и важно понимать, копируется ли он вместе с пациентом или остаётся общим. Реализуйте в классе `Patient` методы `__copy__` и `__deepcopy__`. Создайте пациента с вложенным объектом `Room` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — палата общая (`shallow.room is patient.room` → `True`), при глубоком — своя (`deep.room is patient.room` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В больнице есть разные объекты, у которых есть метод `describe()` — пациенты, врачи, отделения, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_member(member)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Patient`, но и с любым другим классом, у которого есть `describe()` (например, классом `Ward`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В больнице есть отделение `Ward`, и хочется перебирать пациентов в нём в цикле `for` — так же удобно, как список. Сейчас `for patient in ward:` не работает. Реализуйте в классе `Ward` магические методы `__iter__` и `__next__` так, чтобы по отделению можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте отделение с несколькими пациентами и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(ward)` должен возвращать список всех пациентов.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В отделении пациенты должны быть доступны по индексу — как элементы списка: `ward[0]` возвращает первого пациента, `ward[1] = patient` заменяет второго. Реализуйте в классе `Ward` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `ward[0]` — получение пациента по индексу, `ward[1] = new_patient` — замена пациента, срезы `ward[0:2]` — получение подсписка, `for patient in ward` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `patient in ward`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В больнице есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных пациентов её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `PatientDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with PatientDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В больнице есть разные способы расчёта стоимости лечения: фиксированная (5000 рублей), по страховке (страховая покрывает 80% стоимости) и «льготная» (для пенсионеров — 50% от фиксированной). Реализуйте паттерн «Стратегия»: базовый класс `TreatmentCostStrategy` с методом `calculate(base)`, и три наследника — `FixedCost`, `InsuranceCost`, `DiscountCost`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `DiscountCost`, чтобы вызвать реализацию родителя и уменьшить результат вдвое. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В больнице появился новый тип участника — санитар (`Orderly`). Если добавлять его через изменение базового класса `HospitalMember`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `HospitalMember` с методом `get_status()`; наследников `Doctor`, `Patient`; функцию `process_member(member: HospitalMember)`, которая работает с любым участником через родительский интерфейс. Добавьте `Orderly` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_member` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В больнице есть служебный объект `CostCalculator`, который должен уметь рассчитывать стоимость приёма с учётом коэффициента специализации. Удобно, чтобы его можно было вызывать как функцию: `calculator(base, k)`, а не `calculator.calculate(base, k)`. Реализуйте в классе `CostCalculator` магический метод `__call__(self, base, k=1.0)`, который возвращает итоговую стоимость (например, `base * k`). Продемонстрируйте, что объект можно вызывать как функцию: `calc = CostCalculator(); calc(1000, 1.5)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В больнице есть класс `Patient`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Patient` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования пациентов по имени и возрасту): два вызова `Patient("Иван Петров", 45, "гипертония")` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В больнице есть базовый класс `HospitalMember` с методом `get_info()`. Нужно добавлять к участнику дополнительную информацию — например, «льготник», «перенёс операцию», «наблюдается у кардиолога». При этом нельзя менять сам класс `HospitalMember`. Реализуйте паттерн «Декоратор»: класс `MemberDecorator`, который содержит объект `HospitalMember` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `BenefitDecorator` и `CardioDecorator`, которые можно комбинировать: `BenefitDecorator(CardioDecorator(patient))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В больнице появилась внешняя лаборатория, которая возвращает результаты анализов в другом формате — например, в виде словаря `{"name": "Иван Петров", "age": 45, "diagnosis": "гипертония"}`. А весь остальной код больницы работает с классом `Patient`, у которого поля `name`, `age`, `diagnosis`. Реализуйте паттерн «Адаптер»: класс `PatientAdapter`, который принимает словарь и предоставляет интерфейс `Patient` — свойства `name`, `age`, `diagnosis` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Patient` — например, передать его в функцию, которая работает с пациентом. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В больнице есть базовый класс `HospitalMember`, и постепенно в нём накопились методы, которые нужны не всем участникам: `treat()` (только для врачей), `receive_treatment()` (только для пациентов), `assist()` (только для медсестёр). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Treatable`, `Receivable`, `Assistable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `TreatmentService`, который зависит не от конкретных классов (`Doctor`, `Patient`), а от абстракции `Treatable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип участника (например, `Consultant`, реализующий только `Treatable`) без изменения `TreatmentService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В автосалоне ведётся учёт автомобилей. Создайте класс `Car` с полями `brand` (марка), `model` (модель) и `price` (цена). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией об автомобиле в формате: `Марка — модель, цена`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Toyota — Camry, 2 500 000 ₽`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого автомобиля есть VIN-номер, который нельзя менять или читать напрямую. Создайте класс `CarPassport`, в котором VIN-номер хранится в приватном поле `__vin`. Реализуйте метод `set_vin(value)`, предназначенный только для установки номера при постановке на учёт, и метод `get_vin()` для чтения значения. Повторная попытка изменить номер после постановки на учёт должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте паспорт, установите VIN, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `passport.__vin` — будет ошибка `AttributeError`, но через name mangling (`passport._CarPassport__vin`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В автосалоне продаются разные типы транспорта. Создайте базовый класс `Vehicle` с методом `get_kind()`, возвращающим строку «Транспорт». Унаследуйте от него классы `Car` и `Motorcycle`, переопределив в каждом метод `get_kind()` так, чтобы он возвращал «Автомобиль» и «Мотоцикл» соответственно. Создайте по одному объекту каждого класса и выведите их вид, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Автосалон рассчитывает налог для разных типов транспорта. Создайте классы `Sedan`, `SUV` и `SportCar`, у каждого из которых есть метод `tax()`, возвращающий размер налога: для седана — 5000, для внедорожника — 12000, для спорткара — 25000. Напишите функцию `show_tax(vehicle)`, которая принимает любой транспорт и выводит размер его налога. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `SalesManager` (менеджер по продажам), в котором есть три поля: `name` — публичное (доступно всем), `_department` — защищённое (отдел продаж, доступно внутри класса и наследникам), `__commission` — приватное (процент комиссии, доступен только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_department` — можно, но не рекомендуется, к `__commission` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_SalesManager__commission`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Автосалону нужна утилита для проверки государственных номеров, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `AutoUtils` со статическим методом `is_valid_plate(plate)`, который возвращает `True`, если строка начинается с букв «AB» и её длина равна 7, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `AutoUtils.is_valid_plate("AB12345")`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `AutoItem` с абстрактным методом `describe()`. Этот класс описывает общее понятие «транспорт в автосалоне», но не может существовать сам по себе — у каждого конкретного транспорта должно быть своё описание. Унаследуйте от него классы `Car` и `Motorcycle`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание транспорта. Проверьте, что объект абстрактного класса `AutoItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждого автомобиля есть двигатель — автомобиль не может существовать без него, и двигатель создаётся вместе с автомобилем. Создайте класс `Engine` с методом `show()`, который выводит строку «Бензиновый двигатель». Создайте класс `Car`, который содержит объект класса `Engine` внутри себя как поле (композиция) — объект `Engine` создаётся прямо в конструкторе автомобиля. Метод `show_car()` у автомобиля должен выводить модель автомобиля и вызывать метод `show()` у его двигателя, демонстрируя, что автомобиль «владеет» своим двигателем. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В автосалоне есть базовый класс `AutoItem` с методом `get_info()`, который возвращает строку «Транспорт автосалона». Унаследуйте от него класс `Car` и переопределите метод `get_info()` так, чтобы он возвращал строку «Автомобиль: <марка> <модель>». Создайте объект класса `Car`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Менеджеру автосалона нужен метод для оформления заказов, который мог бы работать в разных ситуациях: иногда известен только автомобиль, иногда автомобиль и покупатель, иногда ещё и дополнительные опции. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `OrderDesk` с методом `place(car, buyer=None, extras=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Оформлен заказ на автомобиль: <brand> <model>», «Автомобиль <brand> <model> оформлен для <buyer>» или «Автомобиль <brand> <model> для <buyer> с опциями: <extras>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Автомобили часто выводят на экран — в каталогах, прайс-листах, отчётах. По умолчанию `print(car)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Car` магический метод `__str__`, который возвращает строку вида «Автомобиль "Toyota Camry" — 2 500 000 ₽». Убедитесь, что при вызове `print(car)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Car('Toyota', 'Camry', 2500000)` — оно используется при выводе списка автомобилей и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В автосалоне может оказаться несколько экземпляров одной и той же модели — например, разные комплектации одного авто. Нужно, чтобы программа понимала, что это «один и тот же автомобиль», и не дублировала его в каталоге. Реализуйте в классе `Car` магический метод `__eq__`, чтобы два автомобиля считались равными, если совпадают их марка и модель. Дополнительно реализуйте `__hash__`, чтобы автомобили можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых автомобиля попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В автосалоне есть выставочные площадки, и иногда нужно объединять их (например, при расширении салона) или узнавать, сколько машин на площадке. Создайте класс `Showroom` (выставочная площадка), в котором хранится список автомобилей. Реализуйте магический метод `__add__` для объединения двух площадок в одну (возвращается **новая** площадка со всеми автомобилями) и метод `__len__`, возвращающий количество автомобилей на площадке. Продемонстрируйте работу: создайте две площадки, объедините их через `showroom3 = showroom1 + showroom2` и выведите `len(showroom3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В автосалоне важно, чтобы цена автомобиля была корректной — нельзя допустить отрицательную цену или ноль. Перепишите класс `Car` так, чтобы поле `price` было реализовано через свойство `@property`. Проверки: цена должна быть числом (int или float), и она должна быть строго больше нуля. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `car.price` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `car.price = -100` → ошибка, `car.price = 0` → ошибка, `car.price = "abc"` → ошибка, `car.price = 2500000` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В автосалон часто поступают данные об автомобилях в виде строк — например, из файла или базы данных: `"Toyota;Camry;2500000"`. Добавьте в класс `Car` классовый метод `from_string(cls, data)`, который создаёт объект автомобиля из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание автомобиля через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Автосалону нужно вести общий счётчик всех автомобилей — сколько всего автомобилей зарегистрировано в системе. Добавьте в класс `Car` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового автомобиля. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько автомобилей и выведите `Car.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`car.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В автосалоне возможна ситуация, когда покупатель просит автомобиль, которого нет в наличии. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `CarNotFoundError(Exception)` и используйте его в классе `Dealership` (автосалон). Метод `find_car(brand, model)` должен возвращать автомобиль, если он есть, и выбрасывать `CarNotFoundError` с понятным сообщением (например, с указанием марки и модели), если автомобиля нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В автосалоне покупатель оформляет автомобиль — но автомобиль не принадлежит покупателю до момента продажи, он существует отдельно и принадлежит автосалону. Реализуйте класс `Buyer` (покупатель), который содержит список выбранных автомобилей. Автомобили при этом существуют отдельно от покупателя — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого автомобиля есть ссылка на `Dealership`, в котором он продаётся. Продемонстрируйте, что при удалении покупателя автомобили остаются в салоне. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В автосалоне работает человек, который одновременно и продавец, и покупатель — например, менеджер, который сам покупает машину со скидкой. Создайте два класса-роли: `Buyer` (покупатель) и `Employee` (сотрудник). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `BuyerEmployee`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `BuyerEmployee.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Car` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Car` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Car` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Car(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В автосалоне может храниться огромное количество автомобилей — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Car` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Car` атрибут `__slots__ = ("brand", "model", "price")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `car.color = "красный"` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Car` для полей `price` (например, диапазон 100 000–100 000 000) и `year` (диапазон 1950–2100), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В автосалоне должен быть только один каталог автомобилей — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `Dealership` будет только один экземпляр. Проверьте, что `Dealership()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `CarCatalog` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Автосалон регулярно пополняется разными типами транспорта — автомобилями, мотоциклами, грузовиками. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `VehicleFactory` с методом `create(kind, **kwargs)`, который по строке `"car"`, `"motorcycle"` или `"truck"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному транспортному средству каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Покупатели автосалона хотят получать уведомления, когда поступает новая модель или начинается распродажа. Реализуйте паттерн «Наблюдатель»: класс `Dealership` хранит список подписчиков (`Buyer`) и при поступлении нового автомобиля уведомляет всех через метод `notify()`. Продемонстрируйте, что при добавлении автомобиля все покупатели получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки покупатель больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Dealership` со временем разросся: он и хранит автомобили, и оформляет сделки, и отправляет уведомления, и ведёт отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Dealership` на несколько классов: `CarStorage` (хранение автомобилей и поиск по марке/модели), `SaleService` (оформление сделок, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «покупатель оформил покупку». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Электромобили (`ElectricCar`) похожи на обычные автомобили, но имеют дополнительные поля — ёмкость батареи, запас хода. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`ElectricCar(Car)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`ElectricCar` содержит объект `Car` внутри себя как поле `self.car`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В автосалоне нужно создавать копии карточек автомобилей — например, для резервного каталога. При этом у автомобиля есть вложенный объект `Engine`, и важно понимать, копируется ли он вместе с автомобилем или остаётся общим. Реализуйте в классе `Car` методы `__copy__` и `__deepcopy__`. Создайте автомобиль с вложенным объектом `Engine` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — двигатель общий (`shallow.engine is car.engine` → `True`), при глубоком — свой (`deep.engine is car.engine` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В автосалоне есть разные объекты, у которых есть метод `describe()` — автомобили, мотоциклы, грузовики, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Car`, но и с любым другим классом, у которого есть `describe()` (например, классом `Accessory`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В автосалоне есть выставочная площадка `Showroom`, и хочется перебирать автомобили на ней в цикле `for` — так же удобно, как список. Сейчас `for car in showroom:` не работает. Реализуйте в классе `Showroom` магические методы `__iter__` и `__next__` так, чтобы по площадке можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте площадку с несколькими автомобилями и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(showroom)` должен возвращать список всех автомобилей.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** На выставочной площадке автомобили должны быть доступны по индексу — как элементы списка: `showroom[0]` возвращает первый автомобиль, `showroom[1] = car` заменяет второй. Реализуйте в классе `Showroom` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `showroom[0]` — получение автомобиля по индексу, `showroom[1] = new_car` — замена автомобиля, срезы `showroom[0:2]` — получение подсписка, `for car in showroom` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `car in showroom`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В автосалоне есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных автомобилей её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `CarDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with CarDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В автосалоне есть разные способы расчёта скидки: фиксированная (100 000 ₽), процентная (5% от цены) и «трейд-ин» (для покупателей, сдающих старый автомобиль — 10% от цены). Реализуйте паттерн «Стратегия»: базовый класс `DiscountStrategy` с методом `calculate(price)`, и три наследника — `FixedDiscount`, `PercentDiscount`, `TradeInDiscount`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `TradeInDiscount`, чтобы вызвать реализацию `PercentDiscount` и увеличить процент скидки (или переиспользовать код родителя). Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В автосалоне появился новый тип транспорта — электромобиль (`ElectricCar`). Если добавлять его через изменение базового класса `AutoItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `AutoItem` с методом `tax()`; наследников `Sedan`, `SUV`; функцию `process_item(item: AutoItem)`, которая работает с любым транспортом через родительский интерфейс. Добавьте `ElectricCar` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В автосалоне есть служебный объект `CreditCalculator`, который должен уметь рассчитывать ежемесячный платёж по кредиту. Удобно, чтобы его можно было вызывать как функцию: `calculator(2500000, 5)`, а не `calculator.calculate(2500000, 5)`. Реализуйте в классе `CreditCalculator` магический метод `__call__(self, price, years)`, который возвращает примерный ежемесячный платёж (например, `price / (years * 12)`). Продемонстрируйте, что объект можно вызывать как функцию: `calc = CreditCalculator(); calc(2500000, 5)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В автосалоне есть класс `Car`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Car` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования автомобилей по марке и модели): два вызова `Car("Toyota", "Camry", 2500000)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В автосалоне есть базовый класс `AutoItem` с методом `get_info()`. Нужно добавлять к автомобилю дополнительную информацию — например, «премиум-комплектация», «зимний пакет», «гарантия 5 лет». При этом нельзя менять сам класс `AutoItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `AutoItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `PremiumDecorator` и `WinterPackageDecorator`, которые можно комбинировать: `PremiumDecorator(WinterPackageDecorator(car))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В автосалоне появилась внешняя система аукциона, которая возвращает данные об автомобилях в другом формате — например, в виде словаря `{"make": "Toyota", "model": "Camry", "cost": 2500000}`. А весь остальной код автосалона работает с классом `Car`, у которого поля `brand`, `model`, `price`. Реализуйте паттерн «Адаптер»: класс `CarAdapter`, который принимает словарь и предоставляет интерфейс `Car` — свойства `brand`, `model`, `price` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Car` — например, передать его в функцию, которая работает с автомобилем. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В автосалоне есть базовый класс `AutoItem`, и постепенно в нём накопились методы, которые нужны не всем типам транспорта: `drive()` (только для автомобилей), `ride()` (только для мотоциклов), `load()` (только для грузовиков). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Drivable`, `Ridable`, `Loadable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `SalesService`, который зависит не от конкретных классов (`Car`, `Motorcycle`), а от абстракции `AutoItem` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип транспорта (например, `ElectricCar`, реализующий только `Drivable`) без изменения `SalesService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В банке ведётся учёт клиентов. Создайте класс `Client` с полями `name` (имя), `account_number` (номер счёта) и `balance` (баланс). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о клиенте в формате: `Имя — счёт №..., баланс: ...`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Иван Петров — счёт №40817810099910004312, баланс: 150 000 ₽`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого банковского счёта есть ПИН-код, который нельзя менять или читать напрямую. Создайте класс `BankCard`, в котором ПИН-код хранится в приватном поле `__pin`. Реализуйте метод `set_pin(value)`, предназначенный только для установки ПИН-кода при выпуске карты, и метод `check_pin(value)` для проверки введённого ПИН-кода. Повторная попытка изменить ПИН-код после выпуска должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте карту, установите ПИН-код, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `card.__pin` — будет ошибка `AttributeError`, но через name mangling (`card._BankCard__pin`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В банке обслуживаются разные типы счетов. Создайте базовый класс `Account` с методом `get_account_type()`, возвращающим строку «Счёт». Унаследуйте от него классы `DepositAccount` и `CreditAccount`, переопределив в каждом метод `get_account_type()` так, чтобы он возвращал «Депозитный счёт» и «Кредитный счёт» соответственно. Создайте по одному объекту каждого класса и выведите их тип, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Банк начисляет процент по разным типам вкладов. Создайте классы `ShortDeposit`, `MediumDeposit` и `LongDeposit`, у каждого из которых есть метод `interest_rate()`, возвращающий процентную ставку: для краткосрочного — 5, для среднесрочного — 8, для долгосрочного — 12. Напишите функцию `show_rate(deposit)`, которая принимает любой вклад и выводит его ставку. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `BankManager` (управляющий банка), в котором есть три поля: `name` — публичное (доступно всем), `_branch` — защищённое (филиал, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_branch` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_BankManager__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Банку нужна утилита для проверки номеров карт, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `BankUtils` со статическим методом `is_valid_card(card)`, который возвращает `True`, если строка состоит только из цифр и её длина равна 16, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `BankUtils.is_valid_card("4276123456789012")`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `BankProduct` с абстрактным методом `describe()`. Этот класс описывает общее понятие «банковский продукт», но не может существовать сам по себе — у каждого конкретного продукта должно быть своё описание. Унаследуйте от него классы `DepositAccount` и `CreditAccount`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание продукта. Проверьте, что объект абстрактного класса `BankProduct` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждого клиента банка есть паспорт — клиент не может пользоваться услугами без документа, и паспорт создаётся вместе с клиентом. Создайте класс `Passport` с методом `show()`, который выводит строку «Паспорт предъявлен». Создайте класс `Client`, который содержит объект класса `Passport` внутри себя как поле (композиция) — объект `Passport` создаётся прямо в конструкторе клиента. Метод `show_client()` у клиента должен выводить имя клиента и вызывать метод `show()` у его паспорта, демонстрируя, что клиент «владеет» своим паспортом. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В банке есть базовый класс `BankProduct` с методом `get_info()`, который возвращает строку «Банковский продукт». Унаследуйте от него класс `DepositAccount` и переопределите метод `get_info()` так, чтобы он возвращал строку «Депозит: <имя клиента>». Создайте объект класса `DepositAccount`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Операционисту нужен метод для оформления операций, который мог бы работать в разных ситуациях: иногда известен только клиент, иногда клиент и сумма, иногда ещё и тип операции. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `OperationDesk` с методом `process(client, amount=None, operation=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Принят запрос от клиента: <name>», «Операция для <name> на сумму: <amount>» или «Операция "<operation>" для <name> на сумму <amount>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Клиентов часто выводят на экран — в выписках, отчётах, списках обслуживания. По умолчанию `print(client)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Client` магический метод `__str__`, который возвращает строку вида «Клиент "Иван Петров" (счёт №40817..., баланс: 150 000 ₽)». Убедитесь, что при вызове `print(client)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Client('Иван Петров', '40817810099910004312', 150000)` — оно используется при выводе списка клиентов и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В банке может оказаться несколько записей об одном и том же клиенте — например, при повторной регистрации. Нужно, чтобы программа понимала, что это «один и тот же клиент», и не дублировала его в списках. Реализуйте в классе `Client` магический метод `__eq__`, чтобы два клиента считались равными, если совпадают их имя и номер счёта. Дополнительно реализуйте `__hash__`, чтобы клиентов можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых клиента попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В банке есть клиентские портфели, и иногда нужно объединять их (например, при слиянии отделений) или узнавать, сколько клиентов в портфеле. Создайте класс `Portfolio` (портфель клиентов), в котором хранится список клиентов. Реализуйте магический метод `__add__` для объединения двух портфелей в один (возвращается **новый** портфель со всеми клиентами) и метод `__len__`, возвращающий количество клиентов. Продемонстрируйте работу: создайте два портфеля, объедините их через `p3 = p1 + p2` и выведите `len(p3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В банке важно, чтобы баланс счёта был корректным — нельзя допустить отрицательный баланс (кроме кредитного счёта, но в этой задаче считаем, что баланс не может быть отрицательным). Перепишите класс `Client` так, чтобы поле `balance` было реализовано через свойство `@property`. Проверки: баланс должен быть числом (int или float) и не может быть отрицательным. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `client.balance` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `client.balance = -100` → ошибка, `client.balance = "abc"` → ошибка, `client.balance = 150000` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В банк часто поступают данные о клиентах в виде строк — например, из файла или базы данных: `"Иван Петров;40817810099910004312;150000"`. Добавьте в класс `Client` классовый метод `from_string(cls, data)`, который создаёт объект клиента из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание клиента через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Банку нужно вести общий счётчик всех клиентов — сколько всего клиентов зарегистрировано в системе. Добавьте в класс `Client` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового клиента. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько клиентов и выведите `Client.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`client.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В банке возможна ситуация, когда клиент пытается снять больше денег, чем есть на счёте. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `InsufficientFundsError(Exception)` и используйте его в классе `Bank` (банк). Метод `withdraw(client, amount)` должен уменьшать баланс, если денег достаточно, и выбрасывать `InsufficientFundsError` с понятным сообщением (например, с указанием запрошенной суммы и текущего баланса), если денег не хватает. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В банке клиент открывает счета — но счета не принадлежат клиенту, они существуют отдельно и принадлежат банку. Реализуйте класс `Client` (клиент), который содержит список открытых счетов. Счета при этом существуют отдельно от клиента — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого счёта есть ссылка на `Bank`, в котором он открыт. Продемонстрируйте, что при удалении клиента счета остаются в банке. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В банке работает человек, который одновременно и клиент, и сотрудник — например, сотрудник банка, который сам держит там вклад. Создайте два класса-роли: `Client` (клиент) и `Employee` (сотрудник). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `ClientEmployee`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `ClientEmployee.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Client` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Client` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Client` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Client(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В банке может храниться огромное количество записей о клиентах — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Client` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Client` атрибут `__slots__ = ("name", "account_number", "balance")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `client.phone = "..."` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Client` для полей `balance` (например, диапазон 0–100 000 000) и `credit_limit` (диапазон 0–5 000 000), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В банке должен быть только один реестр клиентов — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `Bank` будет только один экземпляр. Проверьте, что `Bank()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `BankRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Банк регулярно пополняется разными типами счетов — депозитными, кредитными, сберегательными. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `AccountFactory` с методом `create(kind, **kwargs)`, который по строке `"deposit"`, `"credit"` или `"savings"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному счёту каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Клиенты хотят получать уведомления об изменениях курса валют или о новых выгодных предложениях. Реализуйте паттерн «Наблюдатель»: класс `Bank` хранит список подписчиков (`Client`) и при появлении нового предложения уведомляет всех через метод `notify()`. Продемонстрируйте, что при появлении предложения все клиенты получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки клиент больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Bank` со временем разросся: он и хранит клиентов, и проводит операции, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Bank` на несколько классов: `ClientStorage` (хранение клиентов и поиск по имени), `OperationService` (проведение операций, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «клиент снял деньги со счёта». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Премиум-клиенты (`PremiumClient`) похожи на обычных клиентов, но имеют дополнительные поля — персональный менеджер, повышенный кэшбэк. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`PremiumClient(Client)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`PremiumClient` содержит объект `Client` внутри себя как поле `self.client`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В банке нужно создавать копии карточек клиентов — например, для архивного хранения. При этом у клиента есть вложенный объект `Passport`, и важно понимать, копируется ли он вместе с клиентом или остаётся общим. Реализуйте в классе `Client` методы `__copy__` и `__deepcopy__`. Создайте клиента с вложенным объектом `Passport` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — паспорт общий (`shallow.passport is client.passport` → `True`), при глубоком — свой (`deep.passport is client.passport` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В банке есть разные объекты, у которых есть метод `describe()` — клиенты, счета, карты, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Client`, но и с любым другим классом, у которого есть `describe()` (например, классом `Transaction`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В банке есть портфель клиентов `Portfolio`, и хочется перебирать клиентов в нём в цикле `for` — так же удобно, как список. Сейчас `for client in portfolio:` не работает. Реализуйте в классе `Portfolio` магические методы `__iter__` и `__next__` так, чтобы по портфелю можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте портфель с несколькими клиентами и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(portfolio)` должен возвращать список всех клиентов.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В портфеле клиенты должны быть доступны по индексу — как элементы списка: `portfolio[0]` возвращает первого клиента, `portfolio[1] = client` заменяет второго. Реализуйте в классе `Portfolio` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `portfolio[0]` — получение клиента по индексу, `portfolio[1] = new_client` — замена клиента, срезы `portfolio[0:2]` — получение подсписка, `for client in portfolio` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `client in portfolio`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В банке есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных клиентов её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `ClientDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with ClientDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В банке есть разные способы начисления процентов по вкладам: фиксированная ставка (5%), прогрессивная (чем больше срок, тем выше ставка) и «премиум» (для VIP-клиентов — повышенная ставка). Реализуйте паттерн «Стратегия»: базовый класс `InterestStrategy` с методом `calculate(amount, months)`, и три наследника — `FixedInterest`, `ProgressiveInterest`, `PremiumInterest`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `PremiumInterest`, чтобы вызвать реализацию `FixedInterest` и увеличить результат (например, на 3 процентных пункта). Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В банке появился новый тип счёта — накопительный (`SavingsAccount`). Если добавлять его через изменение базового класса `BankProduct`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `BankProduct` с методом `interest_rate()`; наследников `DepositAccount`, `CreditAccount`; функцию `process_product(product: BankProduct)`, которая работает с любым продуктом через родительский интерфейс. Добавьте `SavingsAccount` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_product` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В банке есть служебный объект `InterestCalculator`, который должен уметь рассчитывать доход по вкладу. Удобно, чтобы его можно было вызывать как функцию: `calculator(amount, months)`, а не `calculator.calculate(amount, months)`. Реализуйте в классе `InterestCalculator` магический метод `__call__(self, amount, months)`, который возвращает начисленный доход (например, `amount * 0.05 * months / 12`). Продемонстрируйте, что объект можно вызывать как функцию: `calc = InterestCalculator(); calc(100000, 6)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В банке есть класс `Client`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Client` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования клиентов по номеру счёта): два вызова `Client("Иван Петров", "40817810099910004312", 150000)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В банке есть базовый класс `BankProduct` с методом `get_info()`. Нужно добавлять к продукту дополнительную информацию — например, «страховка», «кэшбэк 5%», «обслуживание бесплатно». При этом нельзя менять сам класс `BankProduct`. Реализуйте паттерн «Декоратор»: класс `ProductDecorator`, который содержит объект `BankProduct` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `InsuranceDecorator` и `CashbackDecorator`, которые можно комбинировать: `InsuranceDecorator(CashbackDecorator(account))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В банке появилась внешняя платёжная система, которая возвращает данные о клиентах в другом формате — например, в виде словаря `{"full_name": "Иван Петров", "account": "40817810099910004312", "amount": 150000}`. А весь остальной код банка работает с классом `Client`, у которого поля `name`, `account_number`, `balance`. Реализуйте паттерн «Адаптер»: класс `ClientAdapter`, который принимает словарь и предоставляет интерфейс `Client` — свойства `name`, `account_number`, `balance` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Client` — например, передать его в функцию, которая работает с клиентом. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В банке есть базовый класс `BankProduct`, и постепенно в нём накопились методы, которые нужны не всем продуктам: `deposit()` (только для депозитов), `withdraw()` (только для текущих счетов), `issue_card()` (только для карточных продуктов). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Depositable`, `Withdrawable`, `CardIssuable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `TransactionService`, который зависит не от конкретных классов (`DepositAccount`, `CreditAccount`), а от абстракции `BankProduct` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип продукта (например, `SavingsAccount`, реализующий только `Depositable`) без изменения `TransactionService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В аэропорту ведётся учёт пассажиров. Создайте класс `Passenger` с полями `name` (имя), `flight_number` (номер рейса) и `seat` (место). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о пассажире в формате: `Имя — рейс №..., место ...`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Иван Петров — рейс №SU1234, место 12A`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого пассажира есть номер паспорта, который нельзя менять или читать напрямую. Создайте класс `Passport`, в котором номер паспорта хранится в приватном поле `__passport_id`. Реализуйте метод `set_passport_id(value)`, предназначенный только для установки номера при регистрации, и метод `get_passport_id()` для чтения значения. Повторная попытка изменить номер после регистрации должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте паспорт, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `passport.__passport_id` — будет ошибка `AttributeError`, но через name mangling (`passport._Passport__passport_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В аэропорту работают разные категории сотрудников. Создайте базовый класс `AirportWorker` с методом `get_position()`, возвращающим строку «Сотрудник». Унаследуйте от него классы `Pilot` и `Stewardess`, переопределив в каждом метод `get_position()` так, чтобы он возвращал «Пилот» и «Стюардесса» соответственно. Создайте по одному объекту каждого класса и выведите их должность, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Аэропорт рассчитывает стоимость билета для разных классов обслуживания. Создайте классы `EconomyClass`, `BusinessClass` и `FirstClass`, у каждого из которых есть метод `ticket_price()`, возвращающий стоимость билета: для эконом-класса — 10000, для бизнес-класса — 35000, для первого класса — 70000. Напишите функцию `show_price(ticket)`, которая принимает любой класс обслуживания и выводит стоимость билета. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `Dispatcher` (диспетчер), в котором есть три поля: `name` — публичное (доступно всем), `_shift` — защищённое (смена, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_shift` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_Dispatcher__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Аэропорту нужна утилита для проверки номеров рейсов, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `AirportUtils` со статическим методом `is_valid_flight(flight)`, который возвращает `True`, если строка начинается с букв «SU» и её длина равна 6, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `AirportUtils.is_valid_flight("SU1234")`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `AirportMember` с абстрактным методом `describe()`. Этот класс описывает общее понятие «участник аэропорта», но не может существовать сам по себе — у каждого конкретного участника должно быть своё описание. Унаследуйте от него классы `Pilot` и `Passenger`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание участника. Проверьте, что объект абстрактного класса `AirportMember` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждого пассажира есть багаж — пассажир не может лететь без багажа (в рамках этой задачи), и багаж создаётся вместе с пассажиром. Создайте класс `Luggage` с методом `show()`, который выводит строку «Багаж зарегистрирован». Создайте класс `Passenger`, который содержит объект класса `Luggage` внутри себя как поле (композиция) — объект `Luggage` создаётся прямо в конструкторе пассажира. Метод `show_passenger()` у пассажира должен выводить имя пассажира и вызывать метод `show()` у его багажа, демонстрируя, что пассажир «владеет» своим багажом. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В аэропорту есть базовый класс `AirportMember` с методом `get_info()`, который возвращает строку «Участник аэропорта». Унаследуйте от него класс `Pilot` и переопределите метод `get_info()` так, чтобы он возвращал строку «Пилот: <имя>». Создайте объект класса `Pilot`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Сотруднику стойки регистрации нужен метод для оформления регистрации, который мог бы работать в разных ситуациях: иногда известен только пассажир, иногда пассажир и рейс, иногда ещё и место. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `CheckInDesk` с методом `check_in(passenger, flight=None, seat=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Зарегистрирован пассажир: <name>», «Пассажир <name> зарегистрирован на рейс: <flight>» или «Пассажир <name> зарегистрирован на рейс <flight>, место <seat>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Пассажиров часто выводят на экран — в списках регистрации, посадочных талонах, отчётах. По умолчанию `print(passenger)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Passenger` магический метод `__str__`, который возвращает строку вида «Пассажир "Иван Петров" (рейс SU1234, место 12A)». Убедитесь, что при вызове `print(passenger)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Passenger('Иван Петров', 'SU1234', '12A')` — оно используется при выводе списка пассажиров и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В аэропорту может оказаться несколько записей об одном и том же пассажире — например, при повторной регистрации. Нужно, чтобы программа понимала, что это «один и тот же пассажир», и не дублировала его в списках. Реализуйте в классе `Passenger` магический метод `__eq__`, чтобы два пассажира считались равными, если совпадают их имя и номер рейса. Дополнительно реализуйте `__hash__`, чтобы пассажиров можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых пассажира попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В аэропорту есть списки пассажиров по рейсам, и иногда нужно объединять их (например, при объединении рейсов) или узнавать, сколько пассажиров в списке. Создайте класс `Flight` (рейс), в котором хранится список пассажиров. Реализуйте магический метод `__add__` для объединения двух рейсов в один (возвращается **новый** рейс со всеми пассажирами) и метод `__len__`, возвращающий количество пассажиров в рейсе. Продемонстрируйте работу: создайте два рейса, объедините их через `flight3 = flight1 + flight2` и выведите `len(flight3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В аэропорту важно, чтобы номер места был корректным — нельзя допустить пустое значение или неправильный формат. Перепишите класс `Passenger` так, чтобы поле `seat` было реализовано через свойство `@property`. Проверки: место должно быть строкой, состоящей из цифр и, возможно, буквы в конце (например, «12A», «7B», «23»). При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `passenger.seat` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `passenger.seat = ""` → ошибка, `passenger.seat = 12` → ошибка, `passenger.seat = "12A"` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В аэропорт часто поступают данные о пассажирах в виде строк — например, из файла или базы данных: `"Иван Петров;SU1234;12A"`. Добавьте в класс `Passenger` классовый метод `from_string(cls, data)`, который создаёт объект пассажира из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание пассажира через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Аэропорту нужно вести общий счётчик всех пассажиров — сколько всего пассажиров зарегистрировано в системе. Добавьте в класс `Passenger` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового пассажира. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько пассажиров и выведите `Passenger.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`passenger.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В аэропорту возможна ситуация, когда пассажир пытается зарегистрироваться на рейс, которого нет в расписании. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `FlightNotFoundError(Exception)` и используйте его в классе `Airport` (аэропорт). Метод `check_in_passenger(passenger, flight_number)` должен регистрировать пассажира, если рейс есть, и выбрасывать `FlightNotFoundError` с понятным сообщением (например, с указанием номера рейса), если рейса нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В аэропорту пассажир летит рейсом — но рейс не принадлежит пассажиру, он существует отдельно и принадлежит авиакомпании. Реализуйте класс `Passenger` (пассажир), который содержит список рейсов, на которые он зарегистрирован. Рейсы при этом существуют отдельно от пассажира — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого рейса есть ссылка на `Airline` (авиакомпанию), которой он принадлежит. Продемонстрируйте, что при удалении пассажира рейсы остаются в расписании. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В аэропорту работает человек, который одновременно и пассажир, и сотрудник — например, пилот, который сам летит куда-то как пассажир. Создайте два класса-роли: `Passenger` (пассажир) и `Employee` (сотрудник). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `PilotPassenger`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `PilotPassenger.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Passenger` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Passenger` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Passenger` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Passenger(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В аэропорту может храниться огромное количество записей о пассажирах — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Passenger` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Passenger` атрибут `__slots__ = ("name", "flight_number", "seat")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `passenger.email = "..."` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Passenger` для полей `age` (например, диапазон 0–120) и `baggage_weight` (диапазон 0–32), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В аэропорту должен быть только один реестр рейсов — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `Airport` будет только один экземпляр. Проверьте, что `Airport()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `FlightRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Аэропорт регулярно пополняется разными типами самолётов — пассажирскими, грузовыми, чартерными. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `AircraftFactory` с методом `create(kind, **kwargs)`, который по строке `"passenger"`, `"cargo"` или `"charter"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному самолёту каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Пассажиры хотят получать уведомления об изменениях времени вылета или открытии онлайн-регистрации на их рейс. Реализуйте паттерн «Наблюдатель»: класс `Flight` хранит список подписчиков (`Passenger`) и при изменении времени вылета уведомляет всех через метод `notify()`. Продемонстрируйте, что при изменении времени все пассажиры получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки пассажир больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Airport` со временем разросся: он и хранит рейсы, и регистрирует пассажиров, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Airport` на несколько классов: `FlightStorage` (хранение рейсов и поиск по номеру), `CheckInService` (регистрация пассажиров, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «пассажир зарегистрировался на рейс». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Бизнес-класс (`BusinessPassenger`) похож на обычного пассажира, но имеет дополнительные поля — доступ в лаунж, приоритетную посадку. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`BusinessPassenger(Passenger)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`BusinessPassenger` содержит объект `Passenger` внутри себя как поле `self.passenger`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В аэропорту нужно создавать копии посадочных талонов — например, для резервного архива. При этом у пассажира есть вложенный объект `Luggage`, и важно понимать, копируется ли он вместе с пассажиром или остаётся общим. Реализуйте в классе `Passenger` методы `__copy__` и `__deepcopy__`. Создайте пассажира с вложенным объектом `Luggage` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — багаж общий (`shallow.luggage is passenger.luggage` → `True`), при глубоком — свой (`deep.luggage is passenger.luggage` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В аэропорту есть разные объекты, у которых есть метод `describe()` — пассажиры, пилоты, рейсы, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_member(member)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Passenger`, но и с любым другим классом, у которого есть `describe()` (например, классом `Flight`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В аэропорту есть рейс `Flight`, и хочется перебирать пассажиров на нём в цикле `for` — так же удобно, как список. Сейчас `for passenger in flight:` не работает. Реализуйте в классе `Flight` магические методы `__iter__` и `__next__` так, чтобы по рейсу можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте рейс с несколькими пассажирами и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(flight)` должен возвращать список всех пассажиров.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** На рейсе пассажиры должны быть доступны по индексу — как элементы списка: `flight[0]` возвращает первого пассажира, `flight[1] = passenger` заменяет второго. Реализуйте в классе `Flight` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `flight[0]` — получение пассажира по индексу, `flight[1] = new_passenger` — замена пассажира, срезы `flight[0:2]` — получение подсписка, `for passenger in flight` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `passenger in flight`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В аэропорту есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных рейсов её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `FlightDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with FlightDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В аэропорту есть разные способы расчёта стоимости багажа: фиксированная (1000 ₽ за место), по весу (500 ₽ за каждый килограмм свыше 20 кг) и «эконом» (для эконом-класса — фиксированная, но с доплатой за перевес). Реализуйте паттерн «Стратегия»: базовый класс `BaggageCostStrategy` с методом `calculate(weight)`, и три наследника — `FixedBaggageCost`, `WeightBaggageCost`, `EconomyBaggageCost`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `EconomyBaggageCost`, чтобы вызвать реализацию `FixedBaggageCost` и добавить надбавку за перевес. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В аэропорту появился новый тип участника — бортпроводник (`FlightAttendant`). Если добавлять его через изменение базового класса `AirportMember`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `AirportMember` с методом `get_status()`; наследников `Pilot`, `Passenger`; функцию `process_member(member: AirportMember)`, которая работает с любым участником через родительский интерфейс. Добавьте `FlightAttendant` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_member` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В аэропорту есть служебный объект `TicketCalculator`, который должен уметь рассчитывать итоговую стоимость билета с учётом сборов. Удобно, чтобы его можно было вызывать как функцию: `calculator(base_price)`, а не `calculator.calculate(base_price)`. Реализуйте в классе `TicketCalculator` магический метод `__call__(self, base_price)`, который возвращает итоговую стоимость (например, `base_price + 2000` — топливный сбор). Продемонстрируйте, что объект можно вызывать как функцию: `calc = TicketCalculator(); calc(10000)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В аэропорту есть класс `Passenger`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Passenger` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования пассажиров по имени и номеру рейса): два вызова `Passenger("Иван Петров", "SU1234", "12A")` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В аэропорту есть базовый класс `AirportMember` с методом `get_info()`. Нужно добавлять к участнику дополнительную информацию — например, «VIP», «с животным», «нужна помощь при посадке». При этом нельзя менять сам класс `AirportMember`. Реализуйте паттерн «Декоратор»: класс `MemberDecorator`, который содержит объект `AirportMember` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `VipDecorator` и `AssistanceDecorator`, которые можно комбинировать: `VipDecorator(AssistanceDecorator(passenger))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В аэропорту появилась внешняя система бронирования, которая возвращает данные о пассажирах в другом формате — например, в виде словаря `{"full_name": "Иван Петров", "flight": "SU1234", "seat_number": "12A"}`. А весь остальной код аэропорта работает с классом `Passenger`, у которого поля `name`, `flight_number`, `seat`. Реализуйте паттерн «Адаптер»: класс `PassengerAdapter`, который принимает словарь и предоставляет интерфейс `Passenger` — свойства `name`, `flight_number`, `seat` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Passenger` — например, передать его в функцию, которая работает с пассажиром. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В аэропорту есть базовый класс `AirportMember`, и постепенно в нём накопились методы, которые нужны не всем участникам: `fly()` (только для пилотов), `serve()` (только для бортпроводников), `board()` (только для пассажиров). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Flyable`, `Servable`, `Boardable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `BoardingService`, который зависит не от конкретных классов (`Pilot`, `Passenger`), а от абстракции `Boardable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип участника (например, `FlightAttendant`, реализующий только `Servable`) без изменения `BoardingService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В кинотеатре ведётся учёт фильмов. Создайте класс `Movie` с полями `title` (название), `genre` (жанр) и `duration` (длительность в минутах). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о фильме в формате: `Название — жанр, длительность ... мин`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Интерстеллар — фантастика, длительность 169 мин`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого билета в кинотеатр есть уникальный код, который нельзя менять или читать напрямую. Создайте класс `Ticket`, в котором код билета хранится в приватном поле `__ticket_code`. Реализуйте метод `set_ticket_code(value)`, предназначенный только для установки кода при продаже билета, и метод `get_ticket_code()` для чтения значения. Повторная попытка изменить код после продажи должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте билет, установите код, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `ticket.__ticket_code` — будет ошибка `AttributeError`, но через name mangling (`ticket._Ticket__ticket_code`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В кинотеатре есть разные типы сеансов. Создайте базовый класс `Session` с методом `get_session_type()`, возвращающим строку «Сеанс». Унаследуйте от него классы `MorningSession` и `EveningSession`, переопределив в каждом метод `get_session_type()` так, чтобы он возвращал «Утренний сеанс» и «Вечерний сеанс» соответственно. Создайте по одному объекту каждого класса и выведите их тип, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Кинотеатр рассчитывает стоимость билета для разных категорий зрителей. Создайте классы `AdultTicket`, `ChildTicket` и `StudentTicket`, у каждого из которых есть метод `price()`, возвращающий стоимость билета: для взрослого — 800, для детского — 400, для студенческого — 600. Напишите функцию `show_price(ticket)`, которая принимает любой билет и выводит его стоимость. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `Administrator` (администратор кинотеатра), в котором есть три поля: `name` — публичное (доступно всем), `_hall` — защищённое (зал, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_hall` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_Administrator__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Кинотеатру нужна утилита для проверки кодов бронирования, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `CinemaUtils` со статическим методом `is_valid_booking(code)`, который возвращает `True`, если строка начинается с букв «BK» и её длина равна 8, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `CinemaUtils.is_valid_booking("BK123456")`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `CinemaItem` с абстрактным методом `describe()`. Этот класс описывает общее понятие «объект кинотеатра», но не может существовать сам по себе — у каждого конкретного объекта должно быть своё описание. Унаследуйте от него классы `Movie` и `Session`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `CinemaItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждого фильма есть возрастное ограничение — фильм не может существовать без него, и ограничение создаётся вместе с фильмом. Создайте класс `AgeRating` с методом `show()`, который выводит строку «18+». Создайте класс `Movie`, который содержит объект класса `AgeRating` внутри себя как поле (композиция) — объект `AgeRating` создаётся прямо в конструкторе фильма. Метод `show_movie()` у фильма должен выводить название фильма и вызывать метод `show()` у его возрастного ограничения, демонстрируя, что фильм «владеет» своим ограничением. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В кинотеатре есть базовый класс `CinemaItem` с методом `get_info()`, который возвращает строку «Объект кинотеатра». Унаследуйте от него класс `Movie` и переопределите метод `get_info()` так, чтобы он возвращал строку «Фильм: <название>». Создайте объект класса `Movie`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Кассиру нужен метод для продажи билетов, который мог бы работать в разных ситуациях: иногда известен только фильм, иногда фильм и количество билетов, иногда ещё и сеанс. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `CashDesk` с методом `sell(movie, count=None, session=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Продан билет на фильм: <title>», «Продано <count> билетов на фильм: <title>» или «Продано <count> билетов на фильм <title> на сеанс <session>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Фильмы часто выводят на экран — в афишах, расписаниях, отчётах. По умолчанию `print(movie)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Movie` магический метод `__str__`, который возвращает строку вида «Фильм "Интерстеллар" (фантастика, 169 мин)». Убедитесь, что при вызове `print(movie)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Movie('Интерстеллар', 'фантастика', 169)` — оно используется при выводе списка фильмов и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В кинотеатре может оказаться несколько записей об одном и том же фильме — например, при повторном добавлении в репертуар. Нужно, чтобы программа понимала, что это «один и тот же фильм», и не дублировала его в афише. Реализуйте в классе `Movie` магический метод `__eq__`, чтобы два фильма считались равными, если совпадают их название и жанр. Дополнительно реализуйте `__hash__`, чтобы фильмы можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых фильма попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В кинотеатре есть залы с расписанием, и иногда нужно объединять расписания (например, при объединении залов) или узнавать, сколько сеансов в расписании. Создайте класс `Schedule` (расписание), в котором хранится список сеансов. Реализуйте магический метод `__add__` для объединения двух расписаний в одно (возвращается **новое** расписание со всеми сеансами) и метод `__len__`, возвращающий количество сеансов. Продемонстрируйте работу: создайте два расписания, объедините их через `s3 = s1 + s2` и выведите `len(s3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В кинотеатре важно, чтобы длительность фильма была корректной — нельзя допустить отрицательное значение или ноль. Перепишите класс `Movie` так, чтобы поле `duration` было реализовано через свойство `@property`. Проверки: длительность должна быть целым числом и строго больше нуля. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `movie.duration` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `movie.duration = -10` → ошибка, `movie.duration = 0` → ошибка, `movie.duration = "abc"` → ошибка, `movie.duration = 169` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В кинотеатр часто поступают данные о фильмах в виде строк — например, из файла или базы данных: `"Интерстеллар;фантастика;169"`. Добавьте в класс `Movie` классовый метод `from_string(cls, data)`, который создаёт объект фильма из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание фильма через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Кинотеатру нужно вести общий счётчик всех фильмов — сколько всего фильмов зарегистрировано в системе. Добавьте в класс `Movie` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового фильма. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько фильмов и выведите `Movie.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`movie.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В кинотеатре возможна ситуация, когда зритель пытается купить билет на сеанс, которого нет в расписании. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `SessionNotFoundError(Exception)` и используйте его в классе `Cinema` (кинотеатр). Метод `buy_ticket(movie, session_time)` должен продавать билет, если сеанс есть, и выбрасывать `SessionNotFoundError` с понятным сообщением (например, с указанием времени сеанса), если сеанса нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В кинотеатре зритель покупает билеты на сеансы — но сеансы не принадлежат зрителю, они существуют отдельно и принадлежат кинотеатру. Реализуйте класс `Viewer` (зритель), который содержит список купленных билетов. Билеты при этом существуют отдельно от зрителя — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого сеанса есть ссылка на `Cinema`, в котором он проходит. Продемонстрируйте, что при удалении зрителя сеансы остаются в кинотеатре. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В кинотеатре работает человек, который одновременно и зритель, и сотрудник — например, кассир, который сам ходит на фильмы. Создайте два класса-роли: `Viewer` (зритель) и `Employee` (сотрудник). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `ViewerEmployee`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `ViewerEmployee.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Movie` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Movie` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Movie` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Movie(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В кинотеатре может храниться огромное количество записей о фильмах — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Movie` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Movie` атрибут `__slots__ = ("title", "genre", "duration")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `movie.rating = 9.5` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Movie` для полей `duration` (например, диапазон 1–300) и `year` (диапазон 1900–2100), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В кинотеатре должен быть только один реестр фильмов — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `Cinema` будет только один экземпляр. Проверьте, что `Cinema()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `MovieRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Кинотеатр регулярно пополняется разными типами сеансов — утренними, дневными, вечерними. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `SessionFactory` с методом `create(kind, **kwargs)`, который по строке `"morning"`, `"day"` или `"evening"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному сеансу каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Зрители хотят получать уведомления, когда появляется новый фильм или открывается предпродажа билетов. Реализуйте паттерн «Наблюдатель»: класс `Cinema` хранит список подписчиков (`Viewer`) и при добавлении нового фильма уведомляет всех через метод `notify()`. Продемонстрируйте, что при добавлении фильма все зрители получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки зритель больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Cinema` со временем разросся: он и хранит фильмы, и продаёт билеты, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Cinema` на несколько классов: `MovieStorage` (хранение фильмов и поиск по названию), `TicketService` (продажа билетов, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «зритель купил билет». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Фильмы в формате 3D (`Movie3D`) похожи на обычные фильмы, но имеют дополнительные поля — стоимость очков, технология. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`Movie3D(Movie)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`Movie3D` содержит объект `Movie` внутри себя как поле `self.movie`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В кинотеатре нужно создавать копии карточек фильмов — например, для резервного каталога. При этом у фильма есть вложенный объект `AgeRating`, и важно понимать, копируется ли он вместе с фильмом или остаётся общим. Реализуйте в классе `Movie` методы `__copy__` и `__deepcopy__`. Создайте фильм с вложенным объектом `AgeRating` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — рейтинг общий (`shallow.rating is movie.rating` → `True`), при глубоком — свой (`deep.rating is movie.rating` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В кинотеатре есть разные объекты, у которых есть метод `describe()` — фильмы, сеансы, билеты, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Movie`, но и с любым другим классом, у которого есть `describe()` (например, классом `Hall`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В кинотеатре есть расписание `Schedule`, и хочется перебирать сеансы в нём в цикле `for` — так же удобно, как список. Сейчас `for session in schedule:` не работает. Реализуйте в классе `Schedule` магические методы `__iter__` и `__next__` так, чтобы по расписанию можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте расписание с несколькими сеансами и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(schedule)` должен возвращать список всех сеансов.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В расписании сеансы должны быть доступны по индексу — как элементы списка: `schedule[0]` возвращает первый сеанс, `schedule[1] = session` заменяет второй. Реализуйте в классе `Schedule` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `schedule[0]` — получение сеанса по индексу, `schedule[1] = new_session` — замена сеанса, срезы `schedule[0:2]` — получение подсписка, `for session in schedule` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `session in schedule`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В кинотеатре есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных фильмов её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `MovieDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with MovieDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В кинотеатре есть разные способы расчёта скидки: фиксированная (100 ₽), процентная (10% от стоимости) и «студенческая» (для студентов — половина от фиксированной). Реализуйте паттерн «Стратегия»: базовый класс `DiscountStrategy` с методом `calculate(price)`, и три наследника — `FixedDiscount`, `PercentDiscount`, `StudentDiscount`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `StudentDiscount`, чтобы вызвать реализацию `FixedDiscount` и уменьшить результат вдвое. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В кинотеатре появился новый тип сеанса — ночной (`NightSession`). Если добавлять его через изменение базового класса `CinemaItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `CinemaItem` с методом `get_price()`; наследников `MorningSession`, `EveningSession`; функцию `process_item(item: CinemaItem)`, которая работает с любым сеансом через родительский интерфейс. Добавьте `NightSession` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В кинотеатре есть служебный объект `PriceCalculator`, который должен уметь рассчитывать стоимость билета с учётом скидки. Удобно, чтобы его можно было вызывать как функцию: `calculator(base_price)`, а не `calculator.calculate(base_price)`. Реализуйте в классе `PriceCalculator` магический метод `__call__(self, base_price)`, который возвращает итоговую стоимость (например, `base_price * 0.9` — скидка 10%). Продемонстрируйте, что объект можно вызывать как функцию: `calc = PriceCalculator(); calc(800)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В кинотеатре есть класс `Movie`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Movie` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования фильмов по названию): два вызова `Movie("Интерстеллар", "фантастика", 169)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В кинотеатре есть базовый класс `CinemaItem` с методом `get_info()`. Нужно добавлять к объекту дополнительную информацию — например, «премьера», «рекомендуем», «для детей». При этом нельзя менять сам класс `CinemaItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `CinemaItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `PremiereDecorator` и `KidsDecorator`, которые можно комбинировать: `PremiereDecorator(KidsDecorator(movie))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В кинотеатре появилась внешняя система онлайн-продаж, которая возвращает данные о фильмах в другом формате — например, в виде словаря `{"name": "Интерстеллар", "category": "фантастика", "length": 169}`. А весь остальной код кинотеатра работает с классом `Movie`, у которого поля `title`, `genre`, `duration`. Реализуйте паттерн «Адаптер»: класс `MovieAdapter`, который принимает словарь и предоставляет интерфейс `Movie` — свойства `title`, `genre`, `duration` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Movie` — например, передать его в функцию, которая работает с фильмом. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В кинотеатре есть базовый класс `CinemaItem`, и постепенно в нём накопились методы, которые нужны не всем объектам: `show()` (только для фильмов), `start()` (только для сеансов), `sell()` (только для билетов). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Showable`, `Startable`, `Sellable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `TicketService`, который зависит не от конкретных классов (`Movie`, `Session`), а от абстракции `Sellable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип объекта (например, `GiftCard`, реализующий только `Sellable`) без изменения `TicketService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В ресторане ведётся учёт блюд. Создайте класс `Dish` с полями `name` (название), `category` (категория: суп, горячее, десерт) и `price` (цена). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о блюде в формате: `Название — категория, цена`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Борщ — суп, 350 ₽`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого заказа в ресторане есть внутренний номер, который нельзя менять или читать напрямую. Создайте класс `Order`, в котором номер заказа хранится в приватном поле `__order_id`. Реализуйте метод `set_order_id(value)`, предназначенный только для установки номера при оформлении заказа, и метод `get_order_id()` для чтения значения. Повторная попытка изменить номер после оформления должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте заказ, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `order.__order_id` — будет ошибка `AttributeError`, но через name mangling (`order._Order__order_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В ресторане работают разные категории сотрудников. Создайте базовый класс `Staff` с методом `get_position()`, возвращающим строку «Сотрудник». Унаследуйте от него классы `Chef` и `Waiter`, переопределив в каждом метод `get_position()` так, чтобы он возвращал «Повар» и «Официант» соответственно. Создайте по одному объекту каждого класса и выведите их должность, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Ресторан рассчитывает стоимость приготовления для разных типов блюд. Создайте классы `Soup`, `MainCourse` и `Dessert`, у каждого из которых есть метод `cook_time()`, возвращающий время приготовления в минутах: для супа — 30, для горячего — 45, для десерта — 15. Напишите функцию `show_cook_time(dish)`, которая принимает любое блюдо и выводит время его приготовления. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `RestaurantManager` (управляющий ресторана), в котором есть три поля: `name` — публичное (доступно всем), `_hall` — защищённое (зал, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_hall` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_RestaurantManager__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Ресторану нужна утилита для проверки номеров столиков, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `RestaurantUtils` со статическим методом `is_valid_table(number)`, который возвращает `True`, если число находится в диапазоне от 1 до 50, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `RestaurantUtils.is_valid_table(12)`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `MenuItem` с абстрактным методом `describe()`. Этот класс описывает общее понятие «позиция меню», но не может существовать сам по себе — у каждой конкретной позиции должно быть своё описание. Унаследуйте от него классы `Dish` и `Drink`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание позиции меню. Проверьте, что объект абстрактного класса `MenuItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждого блюда есть тарелка для подачи — блюдо не может быть подано без тарелки, и тарелка создаётся вместе с блюдом. Создайте класс `Plate` с методом `show()`, который выводит строку «Круглая тарелка». Создайте класс `Dish`, который содержит объект класса `Plate` внутри себя как поле (композиция) — объект `Plate` создаётся прямо в конструкторе блюда. Метод `show_dish()` у блюда должен выводить название блюда и вызывать метод `show()` у его тарелки, демонстрируя, что блюдо «владеет» своей тарелкой. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В ресторане есть базовый класс `MenuItem` с методом `get_info()`, который возвращает строку «Позиция меню». Унаследуйте от него класс `Dish` и переопределите метод `get_info()` так, чтобы он возвращал строку «Блюдо: <название>». Создайте объект класса `Dish`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Официанту нужен метод для оформления заказов, который мог бы работать в разных ситуациях: иногда известно только блюдо, иногда блюдо и количество порций, иногда ещё и столик. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `OrderDesk` с методом `take(dish, quantity=None, table=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Оформлен заказ на блюдо: <name>», «Оформлен заказ: <name> × <quantity>» или «Оформлен заказ: <name> × <quantity> на столик <table>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Блюда часто выводят на экран — в меню, заказах, отчётах. По умолчанию `print(dish)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Dish` магический метод `__str__`, который возвращает строку вида «Блюдо "Борщ" (суп, 350 ₽)». Убедитесь, что при вызове `print(dish)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Dish('Борщ', 'суп', 350)` — оно используется при выводе списка блюд и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В ресторане может оказаться несколько записей об одном и том же блюде — например, при повторном добавлении в меню. Нужно, чтобы программа понимала, что это «одно и то же блюдо», и не дублировала его в меню. Реализуйте в классе `Dish` магический метод `__eq__`, чтобы два блюда считались равными, если совпадают их название и категория. Дополнительно реализуйте `__hash__`, чтобы блюда можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых блюда попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В ресторане есть меню, и иногда нужно объединять их (например, при объединении двух ресторанов) или узнавать, сколько блюд в меню. Создайте класс `Menu` (меню), в котором хранится список блюд. Реализуйте магический метод `__add__` для объединения двух меню в одно (возвращается **новое** меню со всеми блюдами) и метод `__len__`, возвращающий количество блюд. Продемонстрируйте работу: создайте два меню, объедините их через `menu3 = menu1 + menu2` и выведите `len(menu3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В ресторане важно, чтобы цена блюда была корректной — нельзя допустить отрицательную цену или ноль. Перепишите класс `Dish` так, чтобы поле `price` было реализовано через свойство `@property`. Проверки: цена должна быть числом (int или float), и она должна быть строго больше нуля. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `dish.price` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `dish.price = -100` → ошибка, `dish.price = 0` → ошибка, `dish.price = "abc"` → ошибка, `dish.price = 350` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В ресторан часто поступают данные о блюдах в виде строк — например, из файла или базы данных: `"Борщ;суп;350"`. Добавьте в класс `Dish` классовый метод `from_string(cls, data)`, который создаёт объект блюда из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание блюда через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Ресторану нужно вести общий счётчик всех блюд — сколько всего блюд зарегистрировано в системе. Добавьте в класс `Dish` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового блюда. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько блюд и выведите `Dish.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`dish.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В ресторане возможна ситуация, когда официант пытается добавить в заказ блюдо, которого нет в меню. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `DishNotFoundError(Exception)` и используйте его в классе `Restaurant` (ресторан). Метод `add_to_order(order, dish_name)` должен добавлять блюдо в заказ, если оно есть в меню, и выбрасывать `DishNotFoundError` с понятным сообщением (например, с указанием названия), если блюда нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В ресторане официант обслуживает столики — но столики не принадлежат официанту, они существуют отдельно и принадлежат ресторану. Реализуйте класс `Waiter` (официант), который содержит список столиков, за которыми он обслуживает. Столики при этом существуют отдельно от официанта — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого столика есть ссылка на `Restaurant`, в котором он находится. Продемонстрируйте, что при удалении официанта столики остаются в ресторане. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В ресторане работает человек, который одновременно и официант, и повар — например, универсальный сотрудник, который и готовит, и обслуживает. Создайте два класса-роли: `Chef` (повар) и `Waiter` (официант). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `ChefWaiter`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `ChefWaiter.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Dish` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Dish` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Dish` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Dish(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В ресторане может храниться огромное количество записей о блюдах — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Dish` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Dish` атрибут `__slots__ = ("name", "category", "price")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `dish.weight = 250` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Dish` для полей `price` (например, диапазон 1–100000) и `weight` (диапазон 1–5000), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В ресторане должен быть только один реестр заказов — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `Restaurant` будет только один экземпляр. Проверьте, что `Restaurant()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `OrderRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Ресторан регулярно пополняется разными типами блюд — супами, горячим, десертами. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `DishFactory` с методом `create(kind, **kwargs)`, который по строке `"soup"`, `"main"` или `"dessert"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному блюду каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Клиенты ресторана хотят получать уведомления, когда появляется новое блюдо или начинается акция. Реализуйте паттерн «Наблюдатель»: класс `Restaurant` хранит список подписчиков (`Customer`) и при добавлении нового блюда уведомляет всех через метод `notify()`. Продемонстрируйте, что при добавлении блюда все клиенты получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки клиент больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Restaurant` со временем разросся: он и хранит меню, и принимает заказы, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Restaurant` на несколько классов: `MenuStorage` (хранение блюд и поиск по названию), `OrderService` (приём заказов, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «клиент сделал заказ». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Вегетарианские блюда (`VeganDish`) похожи на обычные блюда, но имеют дополнительные поля — список разрешённых ингредиентов, калорийность. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`VeganDish(Dish)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`VeganDish` содержит объект `Dish` внутри себя как поле `self.dish`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В ресторане нужно создавать копии карточек блюд — например, для резервного меню. При этом у блюда есть вложенный объект `Plate`, и важно понимать, копируется ли он вместе с блюдом или остаётся общим. Реализуйте в классе `Dish` методы `__copy__` и `__deepcopy__`. Создайте блюдо с вложенным объектом `Plate` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — тарелка общая (`shallow.plate is dish.plate` → `True`), при глубоком — своя (`deep.plate is dish.plate` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В ресторане есть разные объекты, у которых есть метод `describe()` — блюда, напитки, десерты, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Dish`, но и с любым другим классом, у которого есть `describe()` (например, классом `Ingredient`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В ресторане есть меню `Menu`, и хочется перебирать блюда в нём в цикле `for` — так же удобно, как список. Сейчас `for dish in menu:` не работает. Реализуйте в классе `Menu` магические методы `__iter__` и `__next__` так, чтобы по меню можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте меню с несколькими блюдами и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(menu)` должен возвращать список всех блюд.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В меню блюда должны быть доступны по индексу — как элементы списка: `menu[0]` возвращает первое блюдо, `menu[1] = dish` заменяет второе. Реализуйте в классе `Menu` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `menu[0]` — получение блюда по индексу, `menu[1] = new_dish` — замена блюда, срезы `menu[0:2]` — получение подсписка, `for dish in menu` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `dish in menu`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В ресторане есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных заказов её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `OrderDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with OrderDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В ресторане есть разные способы расчёта скидки: фиксированная (100 ₽), процентная (10% от суммы) и «для постоянных клиентов» (15% от суммы). Реализуйте паттерн «Стратегия»: базовый класс `DiscountStrategy` с методом `calculate(total)`, и три наследника — `FixedDiscount`, `PercentDiscount`, `LoyaltyDiscount`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `LoyaltyDiscount`, чтобы вызвать реализацию `PercentDiscount` и увеличить процент скидки (или переиспользовать код родителя). Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В ресторане появился новый тип позиции меню — напиток (`Drink`). Если добавлять его через изменение базового класса `MenuItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `MenuItem` с методом `price()`; наследников `Dish`, `Dessert`; функцию `process_item(item: MenuItem)`, которая работает с любой позицией через родительский интерфейс. Добавьте `Drink` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В ресторане есть служебный объект `BillCalculator`, который должен уметь рассчитывать итоговый счёт с учётом чаевых. Удобно, чтобы его можно было вызывать как функцию: `calculator(total)`, а не `calculator.calculate(total)`. Реализуйте в классе `BillCalculator` магический метод `__call__(self, total)`, который возвращает итоговую сумму с чаевыми (например, +10%). Продемонстрируйте, что объект можно вызывать как функцию: `calc = BillCalculator(); calc(1500)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В ресторане есть класс `Dish`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Dish` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования блюд по названию): два вызова `Dish("Борщ", "суп", 350)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В ресторане есть базовый класс `MenuItem` с методом `get_info()`. Нужно добавлять к позиции дополнительную информацию — например, «острое», «вегетарианское», «новинка». При этом нельзя менять сам класс `MenuItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `MenuItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `SpicyDecorator` и `VeganDecorator`, которые можно комбинировать: `SpicyDecorator(VeganDecorator(dish))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В ресторане появилась внешняя система доставки, которая возвращает данные о блюдах в другом формате — например, в виде словаря `{"dish_name": "Борщ", "type": "суп", "cost": 350}`. А весь остальной код ресторана работает с классом `Dish`, у которого поля `name`, `category`, `price`. Реализуйте паттерн «Адаптер»: класс `DishAdapter`, который принимает словарь и предоставляет интерфейс `Dish` — свойства `name`, `category`, `price` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Dish` — например, передать его в функцию, которая работает с блюдом. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В ресторане есть базовый класс `MenuItem`, и постепенно в нём накопились методы, которые нужны не всем позициям: `cook()` (только для блюд), `pour()` (только для напитков), `serve()` (только для десертов). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Cookable`, `Pourable`, `Servable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `KitchenService`, который зависит не от конкретных классов (`Dish`, `Drink`), а от абстракции `Cookable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип позиции (например, `Drink`, реализующий только `Pourable`) без изменения `KitchenService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В школе ведётся учёт учеников. Создайте класс `Pupil` с полями `name` (имя), `grade` (класс) и `average_mark` (средняя оценка). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией об ученике в формате: `Имя — класс ..., средняя оценка ...`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Иван Петров — класс 9А, средняя оценка 4.5`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого ученика есть номер личного дела, который нельзя менять или читать напрямую. Создайте класс `StudentFile`, в котором номер личного дела хранится в приватном поле `__file_id`. Реализуйте метод `set_file_id(value)`, предназначенный только для установки номера при зачислении, и метод `get_file_id()` для чтения значения. Повторная попытка изменить номер после зачисления должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте личное дело, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `file.__file_id` — будет ошибка `AttributeError`, но через name mangling (`file._StudentFile__file_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В школе работают и учатся разные категории людей. Создайте базовый класс `SchoolMember` с методом `get_role()`, возвращающим строку «Участник школы». Унаследуйте от него классы `Pupil` и `Teacher`, переопределив в каждом метод `get_role()` так, чтобы он возвращал «Ученик» и «Учитель» соответственно. Создайте по одному объекту каждого класса и выведите их роль, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Школа рассчитывает недельную нагрузку для разных предметов. Создайте классы `MathLesson`, `LiteratureLesson` и `PhysicsLesson`, у каждого из которых есть метод `hours_per_week()`, возвращающий количество часов в неделю: для математики — 6, для литературы — 4, для физики — 5. Напишите функцию `show_hours(lesson)`, которая принимает любой урок и выводит количество часов. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `Principal` (директор школы), в котором есть три поля: `name` — публичное (доступно всем), `_office` — защищённое (кабинет, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_office` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_Principal__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Школе нужна утилита для проверки номеров дневников, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `SchoolUtils` со статическим методом `is_valid_diary(number)`, который возвращает `True`, если число находится в диапазоне от 1 до 1000, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `SchoolUtils.is_valid_diary(42)`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `SchoolItem` с абстрактным методом `describe()`. Этот класс описывает общее понятие «участник школы», но не может существовать сам по себе — у каждого конкретного участника должно быть своё описание. Унаследуйте от него классы `Pupil` и `Teacher`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание участника. Проверьте, что объект абстрактного класса `SchoolItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждого ученика есть портфель — ученик не ходит в школу без портфеля, и портфель создаётся вместе с учеником. Создайте класс `Backpack` с методом `show()`, который выводит строку «Портфель полон учебников». Создайте класс `Pupil`, который содержит объект класса `Backpack` внутри себя как поле (композиция) — объект `Backpack` создаётся прямо в конструкторе ученика. Метод `show_pupil()` у ученика должен выводить имя ученика и вызывать метод `show()` у его портфеля, демонстрируя, что ученик «владеет» своим портфелем. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В школе есть базовый класс `SchoolItem` с методом `get_info()`, который возвращает строку «Объект школы». Унаследуйте от него класс `Teacher` и переопределите метод `get_info()` так, чтобы он возвращал строку «Учитель: <имя>». Создайте объект класса `Teacher`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Секретарю школы нужен метод для оформления справок, который мог бы работать в разных ситуациях: иногда известен только ученик, иногда ученик и причина справки, иногда ещё и учитель, заверяющий её. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `DocumentDesk` с методом `issue(pupil, reason=None, teacher=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Оформлена справка ученику: <name>», «Оформлена справка ученику <name> по причине: <reason>» или «Оформлена справка ученику <name> по причине <reason>, заверил: <teacher>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Учеников часто выводят на экран — в списках класса, журналах, отчётах. По умолчанию `print(pupil)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Pupil` магический метод `__str__`, который возвращает строку вида «Ученик "Иван Петров" (класс 9А, средняя оценка 4.5)». Убедитесь, что при вызове `print(pupil)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Pupil('Иван Петров', '9А', 4.5)` — оно используется при выводе списка учеников и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В школе может оказаться несколько записей об одном и том же ученике — например, при повторной подаче документов. Нужно, чтобы программа понимала, что это «один и тот же ученик», и не дублировала его в списках. Реализуйте в классе `Pupil` магический метод `__eq__`, чтобы два ученика считались равными, если совпадают их имя и класс. Дополнительно реализуйте `__hash__`, чтобы учеников можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых ученика попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В школе есть параллели, и иногда нужно объединять их (например, при слиянии классов) или узнавать, сколько учеников в параллели. Создайте класс `Grade` (параллель), в котором хранится список учеников. Реализуйте магический метод `__add__` для объединения двух параллелей в одну (возвращается **новая** параллель со всеми учениками) и метод `__len__`, возвращающий количество учеников. Продемонстрируйте работу: создайте две параллели, объедините их через `g3 = g1 + g2` и выведите `len(g3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В школе важно, чтобы средняя оценка была корректной — нельзя допустить значение больше 5 или меньше 0. Перепишите класс `Pupil` так, чтобы поле `average_mark` было реализовано через свойство `@property`. Проверки: оценка должна быть числом (int или float) и лежать в диапазоне от 0 до 5 включительно. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `pupil.average_mark` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `pupil.average_mark = 6` → ошибка, `pupil.average_mark = -1` → ошибка, `pupil.average_mark = "abc"` → ошибка, `pupil.average_mark = 4.5` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В школу часто поступают данные об учениках в виде строк — например, из файла или базы данных: `"Иван Петров;9А;4.5"`. Добавьте в класс `Pupil` классовый метод `from_string(cls, data)`, который создаёт объект ученика из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание ученика через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Школе нужно вести общий счётчик всех учеников — сколько всего учеников зарегистрировано в системе. Добавьте в класс `Pupil` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового ученика. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько учеников и выведите `Pupil.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`pupil.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В школе возможна ситуация, когда ученик пытается записаться на урок, которого нет в расписании. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `LessonNotFoundError(Exception)` и используйте его в классе `School` (школа). Метод `enroll(pupil, lesson_name)` должен записывать ученика на урок, если он есть, и выбрасывать `LessonNotFoundError` с понятным сообщением (например, с указанием названия урока), если урока нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В школе ученик посещает уроки — но уроки не принадлежат ученику, они существуют отдельно и принадлежат школе. Реализуйте класс `Pupil` (ученик), который содержит список уроков, на которые он записан. Уроки при этом существуют отдельно от ученика — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого урока есть ссылка на `School`, в которой он проводится. Продемонстрируйте, что при удалении ученика уроки остаются в школе. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В школе работает человек, который одновременно и ученик, и сотрудник — например, лаборант, который сам учится в старших классах. Создайте два класса-роли: `Pupil` (ученик) и `Employee` (сотрудник). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `WorkingPupil`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `WorkingPupil.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Pupil` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Pupil` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Pupil` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Pupil(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В школе может храниться огромное количество записей об учениках — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Pupil` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Pupil` атрибут `__slots__ = ("name", "grade", "average_mark")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `pupil.email = "..."` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Pupil` для полей `average_mark` (например, диапазон 0–5) и `age` (диапазон 6–19), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В школе должен быть только один реестр учеников — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `School` будет только один экземпляр. Проверьте, что `School()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `PupilRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Школа регулярно пополняется разными типами участников — учениками, учителями, администраторами. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `MemberFactory` с методом `create(kind, **kwargs)`, который по строке `"pupil"`, `"teacher"` или `"admin"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному участнику каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Ученики и родители хотят получать уведомления о появлении новых уроков или изменении расписания. Реализуйте паттерн «Наблюдатель»: класс `School` хранит список подписчиков (`Pupil`) и при добавлении нового урока уведомляет всех через метод `notify()`. Продемонстрируйте, что при добавлении урока все ученики получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки ученик больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `School` со временем разросся: он и хранит учеников, и ведёт расписание, и рассылает уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `School` на несколько классов: `PupilStorage` (хранение учеников и поиск по имени), `ScheduleService` (ведение расписания, запись на уроки), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «ученик записался на урок». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Ученики профильных классов (`ProfilePupil`) похожи на обычных учеников, но имеют дополнительные поля — профиль обучения, углублённые предметы. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`ProfilePupil(Pupil)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`ProfilePupil` содержит объект `Pupil` внутри себя как поле `self.pupil`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В школе нужно создавать копии личных дел учеников — например, для архивного хранения. При этом у ученика есть вложенный объект `Backpack`, и важно понимать, копируется ли он вместе с учеником или остаётся общим. Реализуйте в классе `Pupil` методы `__copy__` и `__deepcopy__`. Создайте ученика с вложенным объектом `Backpack` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — портфель общий (`shallow.backpack is pupil.backpack` → `True`), при глубоком — свой (`deep.backpack is pupil.backpack` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В школе есть разные объекты, у которых есть метод `describe()` — ученики, учителя, уроки, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_member(member)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Pupil`, но и с любым другим классом, у которого есть `describe()` (например, классом `Lesson`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В школе есть класс `Grade`, и хочется перебирать учеников в нём в цикле `for` — так же удобно, как список. Сейчас `for pupil in grade:` не работает. Реализуйте в классе `Grade` магические методы `__iter__` и `__next__` так, чтобы по классу можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте класс с несколькими учениками и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(grade)` должен возвращать список всех учеников.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В классе ученики должны быть доступны по индексу — как элементы списка: `grade[0]` возвращает первого ученика, `grade[1] = pupil` заменяет второго. Реализуйте в классе `Grade` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `grade[0]` — получение ученика по индексу, `grade[1] = new_pupil` — замена ученика, срезы `grade[0:2]` — получение подсписка, `for pupil in grade` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `pupil in grade`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В школе есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных учеников её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `PupilDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with PupilDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В школе есть разные способы выставления итоговой оценки: средняя арифметическая, средняя взвешенная (с учётом веса контрольных) и «щадящая» (для отстающих — округление вверх при спорной оценке). Реализуйте паттерн «Стратегия»: базовый класс `GradeStrategy` с методом `calculate(marks)`, и три наследника — `AverageStrategy`, `WeightedStrategy`, `LenientStrategy`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `LenientStrategy`, чтобы вызвать реализацию `AverageStrategy` и применить округление вверх. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В школе появился новый тип участника — психолог (`Psychologist`). Если добавлять его через изменение базового класса `SchoolItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `SchoolItem` с методом `get_status()`; наследников `Pupil`, `Teacher`; функцию `process_item(item: SchoolItem)`, которая работает с любым участником через родительский интерфейс. Добавьте `Psychologist` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В школе есть служебный объект `AverageCalculator`, который должен уметь считать средний балл по списку оценок. Удобно, чтобы его можно было вызывать как функцию: `calculator(marks)`, а не `calculator.calculate(marks)`. Реализуйте в классе `AverageCalculator` магический метод `__call__(self, marks)`, который возвращает средний балл (например, `sum(marks) / len(marks)`). Продемонстрируйте, что объект можно вызывать как функцию: `calc = AverageCalculator(); calc([4, 5, 5, 3])`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В школе есть класс `Pupil`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Pupil` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования учеников по имени и классу): два вызова `Pupil("Иван Петров", "9А", 4.5)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В школе есть базовый класс `SchoolItem` с методом `get_info()`. Нужно добавлять к участнику дополнительную информацию — например, «отличник», «староста», «участник олимпиад». При этом нельзя менять сам класс `SchoolItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `SchoolItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `ExcellentDecorator` и `GroupLeaderDecorator`, которые можно комбинировать: `ExcellentDecorator(GroupLeaderDecorator(pupil))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В школе появилась внешняя система электронного журнала, которая возвращает данные об учениках в другом формате — например, в виде словаря `{"fio": "Иван Петров", "class": "9А", "gpa": 4.5}`. А весь остальной код школы работает с классом `Pupil`, у которого поля `name`, `grade`, `average_mark`. Реализуйте паттерн «Адаптер»: класс `PupilAdapter`, который принимает словарь и предоставляет интерфейс `Pupil` — свойства `name`, `grade`, `average_mark` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Pupil` — например, передать его в функцию, которая работает с учеником. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В школе есть базовый класс `SchoolItem`, и постепенно в нём накопились методы, которые нужны не всем участникам: `teach()` (только для учителей), `learn()` (только для учеников), `manage()` (только для администрации). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Teachable`, `Learnable`, `Manageable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `LessonService`, который зависит не от конкретных классов (`Pupil`, `Teacher`), а от абстракции `Learnable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип участника (например, `Psychologist`, реализующий только `Manageable`) без изменения `LessonService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В отеле ведётся учёт гостей. Создайте класс `Guest` с полями `name` (имя), `room_number` (номер комнаты) и `nights` (количество ночей). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о госте в формате: `Имя — комната №..., ночей: ...`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Иван Петров — комната №305, ночей: 4`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого гостя есть номер брони, который нельзя менять или читать напрямую. Создайте класс `Booking`, в котором номер брони хранится в приватном поле `__booking_id`. Реализуйте метод `set_booking_id(value)`, предназначенный только для установки номера при заселении, и метод `get_booking_id()` для чтения значения. Повторная попытка изменить номер после заселения должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте бронь, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `booking.__booking_id` — будет ошибка `AttributeError`, но через name mangling (`booking._Booking__booking_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В отеле есть разные типы номеров. Создайте базовый класс `Room` с методом `get_room_type()`, возвращающим строку «Номер». Унаследуйте от него классы `StandardRoom` и `LuxRoom`, переопределив в каждом метод `get_room_type()` так, чтобы он возвращал «Стандарт» и «Люкс» соответственно. Создайте по одному объекту каждого класса и выведите их тип, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Отель рассчитывает стоимость проживания для разных типов номеров. Создайте классы `EconomyRoom`, `StandardRoom` и `LuxRoom`, у каждого из которых есть метод `price_per_night()`, возвращающий стоимость за ночь: для эконом-номера — 2000, для стандарта — 4000, для люкса — 9000. Напишите функцию `show_price(room)`, которая принимает любой номер и выводит его стоимость. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `HotelManager` (управляющий отеля), в котором есть три поля: `name` — публичное (доступно всем), `_floor` — защищённое (этаж, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_floor` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_HotelManager__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Отелю нужна утилита для проверки номеров комнат, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `HotelUtils` со статическим методом `is_valid_room(number)`, который возвращает `True`, если число находится в диапазоне от 100 до 999, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `HotelUtils.is_valid_room(305)`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `HotelItem` с абстрактным методом `describe()`. Этот класс описывает общее понятие «объект отеля», но не может существовать сам по себе — у каждого конкретного объекта должно быть своё описание. Унаследуйте от него классы `Room` и `Service`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `HotelItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждого номера в отеле есть ключ-карта — номер не может существовать без неё, и карта создаётся вместе с номером. Создайте класс `KeyCard` с методом `show()`, который выводит строку «Ключ-карта активирована». Создайте класс `Room`, который содержит объект класса `KeyCard` внутри себя как поле (композиция) — объект `KeyCard` создаётся прямо в конструкторе номера. Метод `show_room()` у номера должен выводить его тип и вызывать метод `show()` у ключ-карты, демонстрируя, что номер «владеет» своей картой. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В отеле есть базовый класс `HotelItem` с методом `get_info()`, который возвращает строку «Объект отеля». Унаследуйте от него класс `LuxRoom` и переопределите метод `get_info()` так, чтобы он возвращал строку «Люкс: комната <номер>». Создайте объект класса `LuxRoom`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Портье нужен метод для оформления услуг, который мог бы работать в разных ситуациях: иногда известен только гость, иногда гость и услуга, иногда ещё и время. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `ServiceDesk` с методом `order(guest, service=None, time=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Принят запрос от гостя: <name>», «Гость <name> заказал услугу: <service>» или «Гость <name> заказал <service> на <time>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Гостей часто выводят на экран — в списках заселения, отчётах, счетах. По умолчанию `print(guest)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Guest` магический метод `__str__`, который возвращает строку вида «Гость "Иван Петров" (комната №305, 4 ночи)». Убедитесь, что при вызове `print(guest)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Guest('Иван Петров', 305, 4)` — оно используется при выводе списка гостей и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В отеле может оказаться несколько записей об одном и том же госте — например, при повторном бронировании. Нужно, чтобы программа понимала, что это «один и тот же гость», и не дублировала его в списках. Реализуйте в классе `Guest` магический метод `__eq__`, чтобы два гостя считались равными, если совпадают их имя и номер комнаты. Дополнительно реализуйте `__hash__`, чтобы гостей можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых гостя попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В отеле есть этажи с номерами, и иногда нужно объединять списки номеров (например, при объединении этажей) или узнавать, сколько номеров на этаже. Создайте класс `Floor` (этаж), в котором хранится список номеров. Реализуйте магический метод `__add__` для объединения двух этажей в один (возвращается **новый** этаж со всеми номерами) и метод `__len__`, возвращающий количество номеров. Продемонстрируйте работу: создайте два этажа, объедините их через `f3 = f1 + f2` и выведите `len(f3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В отеле важно, чтобы количество ночей было корректным — нельзя допустить отрицательное значение или ноль. Перепишите класс `Guest` так, чтобы поле `nights` было реализовано через свойство `@property`. Проверки: количество ночей должно быть целым числом и строго больше нуля. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `guest.nights` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `guest.nights = -1` → ошибка, `guest.nights = 0` → ошибка, `guest.nights = "abc"` → ошибка, `guest.nights = 4` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В отель часто поступают данные о гостях в виде строк — например, из файла или базы данных: `"Иван Петров;305;4"`. Добавьте в класс `Guest` классовый метод `from_string(cls, data)`, который создаёт объект гостя из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание гостя через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Отелю нужно вести общий счётчик всех гостей — сколько всего гостей зарегистрировано в системе. Добавьте в класс `Guest` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового гостя. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько гостей и выведите `Guest.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`guest.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В отеле возможна ситуация, когда гость пытается заселиться в номер, который уже занят. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `RoomOccupiedError(Exception)` и используйте его в классе `Hotel` (отель). Метод `check_in(guest, room_number)` должен заселять гостя, если номер свободен, и выбрасывать `RoomOccupiedError` с понятным сообщением (например, с указанием номера), если номер занят. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В отеле гость бронирует номера — но номера не принадлежат гостю, они существуют отдельно и принадлежат отелю. Реализуйте класс `Guest` (гость), который содержит список забронированных номеров. Номера при этом существуют отдельно от гостя — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого номера есть ссылка на `Hotel`, в котором он находится. Продемонстрируйте, что при удалении гостя номера остаются в отеле. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В отеле работает человек, который одновременно и гость, и сотрудник — например, сотрудник отеля, который сам живёт в нём во время командировки. Создайте два класса-роли: `Guest` (гость) и `Employee` (сотрудник). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `GuestEmployee`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `GuestEmployee.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Guest` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Guest` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Guest` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Guest(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В отеле может храниться огромное количество записей о гостях — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Guest` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Guest` атрибут `__slots__ = ("name", "room_number", "nights")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `guest.email = "..."` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Guest` для полей `nights` (например, диапазон 1–365) и `room_number` (диапазон 100–999), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В отеле должен быть только один реестр гостей — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `Hotel` будет только один экземпляр. Проверьте, что `Hotel()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `GuestRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Отель регулярно пополняется разными типами номеров — эконом, стандарт, люкс. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `RoomFactory` с методом `create(kind, **kwargs)`, который по строке `"economy"`, `"standard"` или `"lux"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному номеру каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Гости хотят получать уведомления, когда освобождается номер нужной категории или начинается акция. Реализуйте паттерн «Наблюдатель»: класс `Hotel` хранит список подписчиков (`Guest`) и при освобождении номера уведомляет всех через метод `notify()`. Продемонстрируйте, что при освобождении номера все гости получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки гость больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Hotel` со временем разросся: он и хранит номера, и заселяет гостей, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Hotel` на несколько классов: `RoomStorage` (хранение номеров и поиск по типу), `BookingService` (заселение гостей, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «гость заселился в номер». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Апартаменты (`Apartment`) похожи на обычные номера, но имеют дополнительные поля — кухня, отдельный вход. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`Apartment(Room)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`Apartment` содержит объект `Room` внутри себя как поле `self.room`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В отеле нужно создавать копии карточек гостей — например, для архивного хранения. При этом у гостя есть вложенный объект `KeyCard`, и важно понимать, копируется ли он вместе с гостем или остаётся общим. Реализуйте в классе `Guest` методы `__copy__` и `__deepcopy__`. Создайте гостя с вложенным объектом `KeyCard` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — карта общая (`shallow.card is guest.card` → `True`), при глубоком — своя (`deep.card is guest.card` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В отеле есть разные объекты, у которых есть метод `describe()` — гости, номера, услуги, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Guest`, но и с любым другим классом, у которого есть `describe()` (например, классом `Service`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В отеле есть этаж `Floor`, и хочется перебирать номера на нём в цикле `for` — так же удобно, как список. Сейчас `for room in floor:` не работает. Реализуйте в классе `Floor` магические методы `__iter__` и `__next__` так, чтобы по этажу можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте этаж с несколькими номерами и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(floor)` должен возвращать список всех номеров.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** На этаже номера должны быть доступны по индексу — как элементы списка: `floor[0]` возвращает первый номер, `floor[1] = room` заменяет второй. Реализуйте в классе `Floor` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `floor[0]` — получение номера по индексу, `floor[1] = new_room` — замена номера, срезы `floor[0:2]` — получение подсписка, `for room in floor` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `room in floor`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В отеле есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных бронирований её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `BookingDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with BookingDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В отеле есть разные способы расчёта стоимости проживания: фиксированная (4000 ₽/ночь), сезонная (летом — 6000 ₽/ночь, зимой — 3000 ₽/ночь) и «для постоянных гостей» (скидка 15% от фиксированной). Реализуйте паттерн «Стратегия»: базовый класс `PricingStrategy` с методом `calculate(nights)`, и три наследника — `FixedPricing`, `SeasonPricing`, `LoyaltyPricing`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `LoyaltyPricing`, чтобы вызвать реализацию `FixedPricing` и применить скидку. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В отеле появился новый тип номера — апартаменты (`Apartment`). Если добавлять их через изменение базового класса `HotelItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `HotelItem` с методом `price_per_night()`; наследников `StandardRoom`, `LuxRoom`; функцию `process_item(item: HotelItem)`, которая работает с любым объектом через родительский интерфейс. Добавьте `Apartment` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В отеле есть служебный объект `BillCalculator`, который должен уметь рассчитывать итоговый счёт с учётом налога. Удобно, чтобы его можно было вызывать как функцию: `calculator(total)`, а не `calculator.calculate(total)`. Реализуйте в классе `BillCalculator` магический метод `__call__(self, total)`, который возвращает итоговую сумму с налогом (например, +20%). Продемонстрируйте, что объект можно вызывать как функцию: `calc = BillCalculator(); calc(20000)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В отеле есть класс `Guest`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Guest` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования гостей по имени и номеру комнаты): два вызова `Guest("Иван Петров", 305, 4)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В отеле есть базовый класс `HotelItem` с методом `get_info()`. Нужно добавлять к объекту дополнительную информацию — например, «с завтраком», «с видом на море», «для некурящих». При этом нельзя менять сам класс `HotelItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `HotelItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `BreakfastDecorator` и `SeaViewDecorator`, которые можно комбинировать: `BreakfastDecorator(SeaViewDecorator(room))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В отеле появилась внешняя система бронирования, которая возвращает данные о гостях в другом формате — например, в виде словаря `{"full_name": "Иван Петров", "room": 305, "stay": 4}`. А весь остальной код отеля работает с классом `Guest`, у которого поля `name`, `room_number`, `nights`. Реализуйте паттерн «Адаптер»: класс `GuestAdapter`, который принимает словарь и предоставляет интерфейс `Guest` — свойства `name`, `room_number`, `nights` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Guest` — например, передать его в функцию, которая работает с гостем. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В отеле есть базовый класс `HotelItem`, и постепенно в нём накопились методы, которые нужны не всем объектам: `book()` (только для номеров), `serve()` (только для услуг), `clean()` (только для уборки). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Bookable`, `Servable`, `Cleanable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `BookingService`, который зависит не от конкретных классов (`StandardRoom`, `LuxRoom`), а от абстракции `Bookable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип номера (например, `Apartment`, реализующий только `Bookable`) без изменения `BookingService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В зоопарке ведётся учёт животных. Создайте класс `Animal` с полями `name` (кличка), `species` (вид) и `age` (возраст). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о животном в формате: `Кличка — вид, возраст ... лет`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Симба — лев, возраст 7 лет`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого животного есть номер чипа, который нельзя менять или читать напрямую. Создайте класс `Chip`, в котором номер чипа хранится в приватном поле `__chip_id`. Реализуйте метод `set_chip_id(value)`, предназначенный только для установки номера при регистрации животного, и метод `get_chip_id()` для чтения значения. Повторная попытка изменить номер после регистрации должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте чип, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `chip.__chip_id` — будет ошибка `AttributeError`, но через name mangling (`chip._Chip__chip_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В зоопарке живут разные классы животных. Создайте базовый класс `Creature` с методом `get_class()`, возвращающим строку «Существо». Унаследуйте от него классы `Mammal` и `Bird`, переопределив в каждом метод `get_class()` так, чтобы он возвращал «Млекопитающее» и «Птица» соответственно. Создайте по одному объекту каждого класса и выведите их класс, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Зоопарк рассчитывает суточный рацион для разных животных. Создайте классы `Lion`, `Elephant` и `Monkey`, у каждого из которых есть метод `daily_food()`, возвращающий количество килограммов еды в день: для льва — 8, для слона — 150, для обезьяны — 3. Напишите функцию `show_food(animal)`, которая принимает любое животное и выводит его рацион. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `ZooDirector` (директор зоопарка), в котором есть три поля: `name` — публичное (доступно всем), `_sector` — защищённое (сектор, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_sector` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_ZooDirector__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Зоопарку нужна утилита для проверки номеров вольеров, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `ZooUtils` со статическим методом `is_valid_enclosure(number)`, который возвращает `True`, если число находится в диапазоне от 1 до 200, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `ZooUtils.is_valid_enclosure(42)`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `ZooItem` с абстрактным методом `describe()`. Этот класс описывает общее понятие «объект зоопарка», но не может существовать сам по себе — у каждого конкретного объекта должно быть своё описание. Унаследуйте от него классы `Animal` и `Enclosure`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `ZooItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждого животного есть вольер — животное не может жить в зоопарке без вольера, и вольер создаётся вместе с животным. Создайте класс `Enclosure` с методом `show()`, который выводит строку «Просторный вольер». Создайте класс `Animal`, который содержит объект класса `Enclosure` внутри себя как поле (композиция) — объект `Enclosure` создаётся прямо в конструкторе животного. Метод `show_animal()` у животного должен выводить кличку и вызывать метод `show()` у его вольера, демонстрируя, что животное «владеет» своим вольером. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В зоопарке есть базовый класс `ZooItem` с методом `get_info()`, который возвращает строку «Объект зоопарка». Унаследуйте от него класс `Animal` и переопределите метод `get_info()` так, чтобы он возвращал строку «Животное: <кличка>». Создайте объект класса `Animal`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Смотрителю зоопарка нужен метод для регистрации кормлений, который мог бы работать в разных ситуациях: иногда известно только животное, иногда животное и корм, иногда ещё и время. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `FeedDesk` с методом `feed(animal, food=None, time=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Зарегистрировано кормление животного: <name>», «Животное <name> кормят: <food>» или «Животное <name> кормят <food> в <time>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Животных часто выводят на экран — в каталогах зоопарка, отчётах, табличках у вольеров. По умолчанию `print(animal)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Animal` магический метод `__str__`, который возвращает строку вида «Животное "Симба" (лев, 7 лет)». Убедитесь, что при вызове `print(animal)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Animal('Симба', 'лев', 7)` — оно используется при выводе списка животных и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В зоопарке может оказаться несколько записей об одном и том же животном — например, при повторной регистрации. Нужно, чтобы программа понимала, что это «одно и то же животное», и не дублировала его в списках. Реализуйте в классе `Animal` магический метод `__eq__`, чтобы два животных считались равными, если совпадают их кличка и вид. Дополнительно реализуйте `__hash__`, чтобы животных можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых животных попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В зоопарке есть секции с животными, и иногда нужно объединять их (например, при объединении секторов) или узнавать, сколько животных в секции. Создайте класс `Section` (секция), в котором хранится список животных. Реализуйте магический метод `__add__` для объединения двух секций в одну (возвращается **новая** секция со всеми животными) и метод `__len__`, возвращающий количество животных. Продемонстрируйте работу: создайте две секции, объедините их через `s3 = s1 + s2` и выведите `len(s3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В зоопарке важно, чтобы возраст животного был корректным — нельзя допустить отрицательное значение или значение больше 100. Перепишите класс `Animal` так, чтобы поле `age` было реализовано через свойство `@property`. Проверки: возраст должен быть целым числом и лежать в диапазоне от 0 до 100 включительно. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `animal.age` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `animal.age = -1` → ошибка, `animal.age = 200` → ошибка, `animal.age = "abc"` → ошибка, `animal.age = 7` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В зоопарк часто поступают данные о животных в виде строк — например, из файла или базы данных: `"Симба;лев;7"`. Добавьте в класс `Animal` классовый метод `from_string(cls, data)`, который создаёт объект животного из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание животного через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Зоопарку нужно вести общий счётчик всех животных — сколько всего животных зарегистрировано в системе. Добавьте в класс `Animal` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового животного. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько животных и выведите `Animal.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`animal.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В зоопарке возможна ситуация, когда смотритель пытается покормить животное, которого нет в секции. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `AnimalNotFoundError(Exception)` и используйте его в классе `Zoo` (зоопарк). Метод `feed_animal(name, food)` должен кормить животное, если оно есть, и выбрасывать `AnimalNotFoundError` с понятным сообщением (например, с указанием клички), если животного нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В зоопарке смотритель ухаживает за животными — но животные не принадлежат смотрителю, они существуют отдельно и принадлежат зоопарку. Реализуйте класс `Keeper` (смотритель), который содержит список животных, за которыми он ухаживает. Животные при этом существуют отдельно от смотрителя — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого животного есть ссылка на `Zoo`, в котором оно живёт. Продемонстрируйте, что при удалении смотрителя животные остаются в зоопарке. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В зоопарке работает человек, который одновременно и смотритель, и ветеринар — например, ветеринар, который сам ухаживает за частью животных. Создайте два класса-роли: `Keeper` (смотритель) и `Vet` (ветеринар). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `KeeperVet`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `KeeperVet.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Animal` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Animal` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Animal` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Animal(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В зоопарке может храниться огромное количество записей о животных — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Animal` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Animal` атрибут `__slots__ = ("name", "species", "age")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `animal.weight = 200` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Animal` для полей `age` (например, диапазон 0–100) и `weight` (диапазон 1–10000), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В зоопарке должен быть только один реестр животных — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `Zoo` будет только один экземпляр. Проверьте, что `Zoo()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `AnimalRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Зоопарк регулярно пополняется разными классами животных — млекопитающими, птицами, рептилиями. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `AnimalFactory` с методом `create(kind, **kwargs)`, который по строке `"mammal"`, `"bird"` или `"reptile"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному животному каждого класса через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Посетители зоопарка хотят получать уведомления, когда рождается детёныш или открывается новое шоу. Реализуйте паттерн «Наблюдатель»: класс `Zoo` хранит список подписчиков (`Visitor`) и при появлении важного события уведомляет всех через метод `notify()`. Продемонстрируйте, что при появлении события все посетители получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки посетитель больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Zoo` со временем разросся: он и хранит животных, и регистрирует кормления, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Zoo` на несколько классов: `AnimalStorage` (хранение животных и поиск по кличке), `FeedingService` (регистрация кормлений, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «смотритель покормил животное». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Морские животные (`SeaAnimal`) похожи на обычных животных, но имеют дополнительные поля — глубина обитания, солёность воды. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`SeaAnimal(Animal)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`SeaAnimal` содержит объект `Animal` внутри себя как поле `self.animal`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В зоопарке нужно создавать копии карточек животных — например, для архивного хранения. При этом у животного есть вложенный объект `Enclosure`, и важно понимать, копируется ли он вместе с животным или остаётся общим. Реализуйте в классе `Animal` методы `__copy__` и `__deepcopy__`. Создайте животное с вложенным объектом `Enclosure` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — вольер общий (`shallow.enclosure is animal.enclosure` → `True`), при глубоком — свой (`deep.enclosure is animal.enclosure` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В зоопарке есть разные объекты, у которых есть метод `describe()` — животные, вольеры, секции, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Animal`, но и с любым другим классом, у которого есть `describe()` (например, классом `Section`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В зоопарке есть секция `Section`, и хочется перебирать животных в ней в цикле `for` — так же удобно, как список. Сейчас `for animal in section:` не работает. Реализуйте в классе `Section` магические методы `__iter__` и `__next__` так, чтобы по секции можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте секцию с несколькими животными и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(section)` должен возвращать список всех животных.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В секции животные должны быть доступны по индексу — как элементы списка: `section[0]` возвращает первое животное, `section[1] = animal` заменяет второе. Реализуйте в классе `Section` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `section[0]` — получение животного по индексу, `section[1] = new_animal` — замена животного, срезы `section[0:2]` — получение подсписка, `for animal in section` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `animal in section`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В зоопарке есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных животных её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `AnimalDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with AnimalDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В зоопарке есть разные способы расчёта суточного рациона: фиксированный (по норме), по весу (граммов на килограмм массы) и «диетический» (для животных на диете — половина от фиксированного). Реализуйте паттерн «Стратегия»: базовый класс `FeedingStrategy` с методом `calculate(weight)`, и три наследника — `FixedFeeding`, `WeightFeeding`, `DietFeeding`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `DietFeeding`, чтобы вызвать реализацию `FixedFeeding` и уменьшить результат вдвое. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В зоопарке появился новый класс животных — рептилии (`Reptile`). Если добавлять их через изменение базового класса `Creature`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `Creature` с методом `daily_food()`; наследников `Mammal`, `Bird`; функцию `process_creature(creature: Creature)`, которая работает с любым животным через родительский интерфейс. Добавьте `Reptile` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_creature` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В зоопарке есть служебный объект `FoodCalculator`, который должен уметь рассчитывать порцию для животного по его весу. Удобно, чтобы его можно было вызывать как функцию: `calculator(weight)`, а не `calculator.calculate(weight)`. Реализуйте в классе `FoodCalculator` магический метод `__call__(self, weight)`, который возвращает порцию в граммах (например, `weight * 50`). Продемонстрируйте, что объект можно вызывать как функцию: `calc = FoodCalculator(); calc(200)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В зоопарке есть класс `Animal`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Animal` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования животных по кличке и виду): два вызова `Animal("Симба", "лев", 7)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В зоопарке есть базовый класс `ZooItem` с методом `get_info()`. Нужно добавлять к объекту дополнительную информацию — например, «находится на карантине», «участвует в шоу», «редкий вид». При этом нельзя менять сам класс `ZooItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `ZooItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `QuarantineDecorator` и `RareSpeciesDecorator`, которые можно комбинировать: `QuarantineDecorator(RareSpeciesDecorator(animal))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В зоопарке появилась внешняя система учёта из другого зоопарка, которая возвращает данные о животных в другом формате — например, в виде словаря `{"nickname": "Симба", "type": "лев", "years": 7}`. А весь остальной код зоопарка работает с классом `Animal`, у которого поля `name`, `species`, `age`. Реализуйте паттерн «Адаптер»: класс `AnimalAdapter`, который принимает словарь и предоставляет интерфейс `Animal` — свойства `name`, `species`, `age` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Animal` — например, передать его в функцию, которая работает с животным. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В зоопарке есть базовый класс `ZooItem`, и постепенно в нём накопились методы, которые нужны не всем объектам: `feed()` (только для животных), `clean()` (только для вольеров), `show()` (только для выставочных объектов). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Feedable`, `Cleanable`, `Showable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `FeedingService`, который зависит не от конкретных классов (`Lion`, `Elephant`), а от абстракции `Feedable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип животного (например, `Reptile`, реализующий только `Feedable`) без изменения `FeedingService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В фитнес-клубе ведётся учёт клиентов. Ожидаемый вывод, например: `Иван Петров — абонемент "Премиум", на 12 мес.`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого клиента есть номер клубной карты, который нельзя менять или читать напрямую. Реализуйте метод `set_card_id(value)`, предназначенный только для установки номера при оформлении, и метод `get_card_id()` для чтения значения. Повторная попытка изменить номер после оформления должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте карту, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `card.__card_id` — будет ошибка `AttributeError`, но через name mangling (`card._ClubCard__card_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В фитнес-клубе есть разные типы тренировок. Создайте базовый класс `Workout` с методом `get_workout_type()`, возвращающим строку «Тренировка». Унаследуйте от него классы `Yoga` и `Boxing`, переопределив в каждом метод `get_workout_type()` так, чтобы он возвращал «Йога» и «Бокс» соответственно. Создайте по одному объекту каждого класса и выведите их тип, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Фитнес-клуб рассчитывает стоимость занятия для разных типов тренировок. Создайте классы `Yoga`, `Boxing` и `CrossFit`, у каждого из которых есть метод `session_price()`, возвращающий стоимость занятия: для йоги — 500, для бокса — 800, для кроссфита — 1000. Напишите функцию `show_price(workout)`, которая принимает любую тренировку и выводит её стоимость. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `Trainer` (тренер), в котором есть три поля: `name` — публичное (доступно всем), `_specialization` — защищённое (специализация, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_specialization` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_Trainer__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Фитнес-клубу нужна утилита для проверки номеров шкафчиков, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `FitnessUtils` со статическим методом `is_valid_locker(number)`, который возвращает `True`, если число находится в диапазоне от 1 до 300, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `FitnessUtils.is_valid_locker(42)`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `FitnessItem` с абстрактным методом `describe()`. Проверьте, что объект абстрактного класса `FitnessItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. **Задача 8 (композиция).** У каждого клиента есть абонемент — клиент не может посещать клуб без него, и абонемент создаётся вместе с клиентом. Создайте класс `Member`, который содержит объект класса `Subscription` внутри себя как поле (композиция) — объект `Subscription` создаётся прямо в конструкторе клиента. Метод `show_member()` у клиента должен выводить имя клиента и вызывать метод `show()` у его абонемента, демонстрируя, что клиент «владеет» своим абонементом. **Задача 9 (переопределение методов).** В фитнес-клубе есть базовый класс `FitnessItem` с методом `get_info()`, который возвращает строку «Объект клуба». **Задача 10 (имитация перегрузки).** Администратору фитнес-клуба нужен метод для записи на тренировки, который мог бы работать в разных ситуациях: иногда известен только клиент, иногда клиент и тренировка, иногда ещё и тренер. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Клиентов часто выводят на экран — в списках, отчётах, журналах посещений. По умолчанию `print(member)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Member` магический метод `__str__`, который возвращает строку вида «Клиент "Иван Петров" (абонемент "Премиум", 12 мес.)». Убедитесь, что при вызове `print(member)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Member('Иван Петров', 'Премиум', 12)` — оно используется при выводе списка клиентов и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В фитнес-клубе может оказаться несколько записей об одном и том же клиенте — например, при повторной регистрации. Нужно, чтобы программа понимала, что это «один и тот же клиент», и не дублировала его в списках. Реализуйте в классе `Member` магический метод `__eq__`, чтобы два клиента считались равными, если совпадают их имя и тип абонемента. Дополнительно реализуйте `__hash__`, чтобы клиентов можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых клиента попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В фитнес-клубе есть группы тренировок, и иногда нужно объединять их (например, при объединении залов) или узнавать, сколько клиентов в группе. Создайте класс `Group` (группа), в котором хранится список клиентов. Реализуйте магический метод `__add__` для объединения двух групп в одну (возвращается **новая** группа со всеми клиентами) и метод `__len__`, возвращающий количество клиентов. Продемонстрируйте работу: создайте две группы, объедините их через `g3 = g1 + g2` и выведите `len(g3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В фитнес-клубе важно, чтобы срок абонемента был корректным — нельзя допустить отрицательное значение или ноль. Перепишите класс `Member` так, чтобы поле `months` было реализовано через свойство `@property`. Проверки: срок должен быть целым числом и строго больше нуля. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `member.months` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `member.months = -1` → ошибка, `member.months = 0` → ошибка, `member.months = "abc"` → ошибка, `member.months = 12` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В фитнес-клуб часто поступают данные о клиентах в виде строк — например, из файла или базы данных: `"Иван Петров;Премиум;12"`. Добавьте в класс `Member` классовый метод `from_string(cls, data)`, который создаёт объект клиента из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание клиента через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Фитнес-клубу нужно вести общий счётчик всех клиентов — сколько всего клиентов зарегистрировано в системе. Добавьте в класс `Member` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового клиента. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько клиентов и выведите `Member.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`member.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В фитнес-клубе возможна ситуация, когда клиент пытается записаться на тренировку, которой нет в расписании. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `WorkoutNotFoundError(Exception)` и используйте его в классе `FitnessClub` (фитнес-клуб). Метод `book_workout(member, workout_name)` должен записывать клиента, если тренировка есть, и выбрасывать `WorkoutNotFoundError` с понятным сообщением (например, с указанием названия), если тренировки нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В фитнес-клубе тренер ведёт тренировки — но тренировки не принадлежат тренеру, они существуют отдельно и принадлежат клубу. Реализуйте класс `Trainer` (тренер), который содержит список тренировок, которые он ведёт. Тренировки при этом существуют отдельно от тренера — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждой тренировки есть ссылка на `FitnessClub`, в котором она проводится. Продемонстрируйте, что при удалении тренера тренировки остаются в клубе. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В фитнес-клубе работает человек, который одновременно и тренер, и клиент — например, тренер, который сам занимается в этом же клубе. Создайте два класса-роли: `Member` (клиент) и `Employee` (сотрудник). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `TrainerMember`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `TrainerMember.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Member` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Member` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Member` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Member(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В фитнес-клубе может храниться огромное количество записей о клиентах — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Member` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Member` атрибут `__slots__ = ("name", "membership_type", "months")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `member.phone = "..."` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Member` для полей `months` (например, диапазон 1–60) и `visits` (диапазон 0–1000), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В фитнес-клубе должен быть только один реестр клиентов — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `FitnessClub` будет только один экземпляр. Проверьте, что `FitnessClub()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `MemberRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Фитнес-клуб регулярно пополняется разными типами тренировок — йогой, боксом, кроссфитом. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `WorkoutFactory` с методом `create(kind, **kwargs)`, который по строке `"yoga"`, `"boxing"` или `"crossfit"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одной тренировке каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Клиенты хотят получать уведомления, когда открывается запись на новую тренировку или меняется расписание. Реализуйте паттерн «Наблюдатель»: класс `FitnessClub` хранит список подписчиков (`Member`) и при появлении новой тренировки уведомляет всех через метод `notify()`. Продемонстрируйте, что при появлении тренировки все клиенты получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки клиент больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `FitnessClub` со временем разросся: он и хранит клиентов, и ведёт расписание, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `FitnessClub` на несколько классов: `MemberStorage` (хранение клиентов и поиск по имени), `ScheduleService` (ведение расписания, запись на тренировки), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «клиент записался на тренировку». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Персональные тренировки (`PersonalTraining`) похожи на обычные групповые тренировки, но имеют дополнительные поля — персональный тренер, индивидуальный план. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`PersonalTraining(Workout)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`PersonalTraining` содержит объект `Workout` внутри себя как поле `self.workout`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В фитнес-клубе нужно создавать копии карточек клиентов — например, для архивного хранения. При этом у клиента есть вложенный объект `Subscription`, и важно понимать, копируется ли он вместе с клиентом или остаётся общим. Реализуйте в классе `Member` методы `__copy__` и `__deepcopy__`. Создайте клиента с вложенным объектом `Subscription` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — абонемент общий (`shallow.subscription is member.subscription` → `True`), при глубоком — свой (`deep.subscription is member.subscription` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В фитнес-клубе есть разные объекты, у которых есть метод `describe()` — тренировки, клиенты, тренеры, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Workout`, но и с любым другим классом, у которого есть `describe()` (например, классом `Locker`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В фитнес-клубе есть группа `Group`, и хочется перебирать клиентов в ней в цикле `for` — так же удобно, как список. Сейчас `for member in group:` не работает. Реализуйте в классе `Group` магические методы `__iter__` и `__next__` так, чтобы по группе можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте группу с несколькими клиентами и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(group)` должен возвращать список всех клиентов.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В группе клиенты должны быть доступны по индексу — как элементы списка: `group[0]` возвращает первого клиента, `group[1] = member` заменяет второго. Реализуйте в классе `Group` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `group[0]` — получение клиента по индексу, `group[1] = new_member` — замена клиента, срезы `group[0:2]` — получение подсписка, `for member in group` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `member in group`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В фитнес-клубе есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных клиентов её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `MemberDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with MemberDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В фитнес-клубе есть разные способы расчёта стоимости абонемента: фиксированная (1000 ₽/мес.), сезонная (летом — 800 ₽/мес., зимой — 1200 ₽/мес.) и «для постоянных клиентов» (скидка 20% от фиксированной). Реализуйте паттерн «Стратегия»: базовый класс `PricingStrategy` с методом `calculate(months)`, и три наследника — `FixedPricing`, `SeasonPricing`, `LoyaltyPricing`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `LoyaltyPricing`, чтобы вызвать реализацию `FixedPricing` и применить скидку. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В фитнес-клубе появился новый тип тренировки — пилатес (`Pilates`). Если добавлять его через изменение базового класса `FitnessItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `FitnessItem` с методом `session_price()`; наследников `Yoga`, `Boxing`; функцию `process_item(item: FitnessItem)`, которая работает с любой тренировкой через родительский интерфейс. Добавьте `Pilates` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В фитнес-клубе есть служебный объект `CalorieCalculator`, который должен уметь рассчитывать сожжённые калории по длительности тренировки. Удобно, чтобы его можно было вызывать как функцию: `calculator(minutes)`, а не `calculator.calculate(minutes)`. Реализуйте в классе `CalorieCalculator` магический метод `__call__(self, minutes)`, который возвращает количество калорий (например, `minutes * 10`). Продемонстрируйте, что объект можно вызывать как функцию: `calc = CalorieCalculator(); calc(45)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В фитнес-клубе есть класс `Member`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Member` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования клиентов по имени и типу абонемента): два вызова `Member("Иван Петров", "Премиум", 12)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В фитнес-клубе есть базовый класс `FitnessItem` с методом `get_info()`. Нужно добавлять к объекту дополнительную информацию — например, «с бассейном», «с сауной», «с персональным тренером». При этом нельзя менять сам класс `FitnessItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `FitnessItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `PoolDecorator` и `SaunaDecorator`, которые можно комбинировать: `PoolDecorator(SaunaDecorator(subscription))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В фитнес-клубе появилась внешняя система учёта, которая возвращает данные о клиентах в другом формате — например, в виде словаря `{"full_name": "Иван Петров", "plan": "Премиум", "duration": 12}`. А весь остальной код клуба работает с классом `Member`, у которого поля `name`, `membership_type`, `months`. Реализуйте паттерн «Адаптер»: класс `MemberAdapter`, который принимает словарь и предоставляет интерфейс `Member` — свойства `name`, `membership_type`, `months` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Member` — например, передать его в функцию, которая работает с клиентом. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В фитнес-клубе есть базовый класс `FitnessItem`, и постепенно в нём накопились методы, которые нужны не всем объектам: `train()` (только для тренировок), `coach()` (только для тренеров), `lock()` (только для шкафчиков). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Trainable`, `Coachable`, `Lockable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `BookingService`, который зависит не от конкретных классов (`Yoga`, `Boxing`), а от абстракции `Trainable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип тренировки (например, `Pilates`, реализующий только `Trainable`) без изменения `BookingService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В аптеке ведётся учёт лекарств. Создайте класс `Medicine` с полями `name` (название), `form` (форма выпуска: таблетки, сироп, мазь) и `price` (цена). Ожидаемый вывод, например: `Аспирин — таблетки, 150 ₽`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого лекарства есть номер партии, который нельзя менять или читать напрямую. Создайте класс `Batch`, в котором номер партии хранится в приватном поле `__batch_id`. Реализуйте метод `set_batch_id(value)`, предназначенный только для установки номера при поставке, и метод `get_batch_id()` для чтения значения. Повторная попытка изменить номер после поставки должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте партию, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `batch.__batch_id` — будет ошибка `AttributeError`, но через name mangling (`batch._Batch__batch_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В аптеке есть разные категории товаров. Создайте базовый класс `PharmacyItem` с методом `get_category()`, возвращающим строку «Товар аптеки». Унаследуйте от него классы `Medicine` и `Vitamin`, переопределив в каждом метод `get_category()` так, чтобы он возвращал «Лекарство» и «Витамин» соответственно. Создайте по одному объекту каждого класса и выведите их категорию, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Аптека рассчитывает размер скидки для разных категорий товаров. Создайте классы `Medicine`, `Vitamin` и `Cosmetic`, у каждого из которых есть метод `discount()`, возвращающий процент скидки: для лекарства — 5, для витамина — 10, для косметики — 15. Напишите функцию `show_discount(item)`, которая принимает любой товар и выводит его скидку. **Задача 5 (модификаторы доступа).** Создайте класс `Pharmacist` (фармацевт), в котором есть три поля: `name` — публичное (доступно всем), `_department` — защищённое (отдел, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_department` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_Pharmacist__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Аптеке нужна утилита для проверки штрих-кодов, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `PharmacyUtils` со статическим методом `is_valid_barcode(code)`, который возвращает `True`, если строка состоит только из цифр и её длина равна 13, и `False` — в противном случае. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `PharmacyProduct` с абстрактным методом `describe()`. Этот класс описывает общее понятие «товар аптеки», но не может существовать сам по себе — у каждого конкретного товара должно быть своё описание. Унаследуйте от него классы `Medicine` и `Vitamin`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `PharmacyProduct` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. Метод `show_medicine()` у лекарства должен выводить название и вызывать метод `show()` у его инструкции, демонстрируя, что лекарство «владеет» своей инструкцией. **Задача 9 (переопределение методов).** В аптеке есть базовый класс `PharmacyProduct` с методом `get_info()`, который возвращает строку «Товар аптеки». **Задача 10 (имитация перегрузки).** Провизору нужен метод для оформления заказов, который мог бы работать в разных ситуациях: иногда известно только лекарство, иногда лекарство и количество, иногда ещё и покупатель. Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Лекарства часто выводят на экран — в каталогах, чеках, отчётах. По умолчанию `print(medicine)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Medicine` магический метод `__str__`, который возвращает строку вида «Лекарство "Аспирин" (таблетки, 150 ₽)». Убедитесь, что при вызове `print(medicine)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Medicine('Аспирин', 'таблетки', 150)` — оно используется при выводе списка лекарств и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В аптеке может оказаться несколько записей об одном и том же лекарстве — например, при повторной поставке. Нужно, чтобы программа понимала, что это «одно и то же лекарство», и не дублировала его в каталоге. Реализуйте в классе `Medicine` магический метод `__eq__`, чтобы два лекарства считались равными, если совпадают их название и форма выпуска. Дополнительно реализуйте `__hash__`, чтобы лекарства можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых лекарства попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В аптеке есть складские полки, и иногда нужно объединять их (например, при объединении отделов) или узнавать, сколько упаковок на полке. Создайте класс `Shelf` (полка), в котором хранится список упаковок. Реализуйте магический метод `__add__` для объединения двух полок в одну (возвращается **новая** полка со всеми упаковками) и метод `__len__`, возвращающий количество упаковок. Продемонстрируйте работу: создайте две полки, объедините их через `s3 = s1 + s2` и выведите `len(s3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В аптеке важно, чтобы цена лекарства была корректной — нельзя допустить отрицательную цену или ноль. Перепишите класс `Medicine` так, чтобы поле `price` было реализовано через свойство `@property`. Проверки: цена должна быть числом (int или float), и она должна быть строго больше нуля. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `medicine.price` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `medicine.price = -100` → ошибка, `medicine.price = 0` → ошибка, `medicine.price = "abc"` → ошибка, `medicine.price = 150` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В аптеку часто поступают данные о лекарствах в виде строк — например, из файла или базы данных: `"Аспирин;таблетки;150"`. Добавьте в класс `Medicine` классовый метод `from_string(cls, data)`, который создаёт объект лекарства из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание лекарства через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Аптеке нужно вести общий счётчик всех лекарств — сколько всего лекарств зарегистрировано в системе. Добавьте в класс `Medicine` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового лекарства. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько лекарств и выведите `Medicine.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`medicine.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В аптеке возможна ситуация, когда покупатель просит лекарство, которого нет в наличии. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `MedicineNotFoundError(Exception)` и используйте его в классе `Pharmacy` (аптека). Метод `sell(medicine_name, quantity)` должен продавать лекарство, если оно есть, и выбрасывать `MedicineNotFoundError` с понятным сообщением (например, с указанием названия), если лекарства нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В аптеке покупатель складывает товары в корзину — но товары не принадлежат покупателю, они существуют отдельно и принадлежат аптеке. Реализуйте класс `Customer` (покупатель), который содержит список выбранных товаров. Товары при этом существуют отдельно от покупателя — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого товара есть ссылка на `Pharmacy`, в которой он продаётся. Продемонстрируйте, что при удалении покупателя товары остаются в аптеке. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В аптеке работает человек, который одновременно и фармацевт, и покупатель — например, сотрудник аптеки, который сам покупает лекарства со скидкой. Создайте два класса-роли: `Customer` (покупатель) и `Employee` (сотрудник). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `CustomerEmployee`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `CustomerEmployee.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Medicine` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Medicine` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Medicine` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Medicine(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В аптеке может храниться огромное количество записей о лекарствах — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Medicine` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Medicine` атрибут `__slots__ = ("name", "form", "price")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `medicine.country = "..."` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Medicine` для полей `price` (например, диапазон 1–100000) и `quantity` (диапазон 0–10000), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В аптеке должен быть только один реестр лекарств — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `Pharmacy` будет только один экземпляр. Проверьте, что `Pharmacy()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `MedicineRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Аптека регулярно пополняется разными категориями товаров — лекарствами, витаминами, косметикой. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `ProductFactory` с методом `create(kind, **kwargs)`, который по строке `"medicine"`, `"vitamin"` или `"cosmetic"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному товару каждой категории через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Покупатели хотят получать уведомления, когда поступает нужное лекарство или начинается акция. Реализуйте паттерн «Наблюдатель»: класс `Pharmacy` хранит список подписчиков (`Customer`) и при поступлении нового лекарства уведомляет всех через метод `notify()`. Продемонстрируйте, что при поступлении лекарства все покупатели получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки покупатель больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Pharmacy` со временем разросся: он и хранит лекарства, и продаёт их, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Pharmacy` на несколько классов: `MedicineStorage` (хранение лекарств и поиск по названию), `SaleService` (продажа, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «покупатель купил лекарство». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Рецептурные лекарства (`PrescriptionMedicine`) похожи на обычные, но имеют дополнительные поля — номер рецепта, срок действия рецепта. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`PrescriptionMedicine(Medicine)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`PrescriptionMedicine` содержит объект `Medicine` внутри себя как поле `self.medicine`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В аптеке нужно создавать копии карточек лекарств — например, для резервного каталога. При этом у лекарства есть вложенный объект `Instruction`, и важно понимать, копируется ли он вместе с лекарством или остаётся общим. Реализуйте в классе `Medicine` методы `__copy__` и `__deepcopy__`. Создайте лекарство с вложенным объектом `Instruction` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — инструкция общая (`shallow.instruction is medicine.instruction` → `True`), при глубоком — своя (`deep.instruction is medicine.instruction` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В аптеке есть разные объекты, у которых есть метод `describe()` — лекарства, витамины, косметика, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Medicine`, но и с любым другим классом, у которого есть `describe()` (например, классом `Supplier`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В аптеке есть складская полка `Shelf`, и хочется перебирать упаковки на ней в цикле `for` — так же удобно, как список. Сейчас `for package in shelf:` не работает. Реализуйте в классе `Shelf` магические методы `__iter__` и `__next__` так, чтобы по полке можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте полку с несколькими упаковками и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(shelf)` должен возвращать список всех упаковок.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** На полке упаковки должны быть доступны по индексу — как элементы списка: `shelf[0]` возвращает первую упаковку, `shelf[1] = package` заменяет вторую. Реализуйте в классе `Shelf` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `shelf[0]` — получение упаковки по индексу, `shelf[1] = new_package` — замена упаковки, срезы `shelf[0:2]` — получение подсписка, `for package in shelf` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `package in shelf`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В аптеке есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных лекарств её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `MedicineDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with MedicineDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В аптеке есть разные способы расчёта скидки: фиксированная (50 ₽), процентная (10% от суммы) и «для пенсионеров» (половина от фиксированной). Реализуйте паттерн «Стратегия»: базовый класс `DiscountStrategy` с методом `calculate(total)`, и три наследника — `FixedDiscount`, `PercentDiscount`, `PensionerDiscount`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `PensionerDiscount`, чтобы вызвать реализацию `FixedDiscount` и уменьшить результат вдвое. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В аптеке появилась новая категория товаров — медицинская техника (`MedicalDevice`). Если добавлять её через изменение базового класса `PharmacyProduct`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `PharmacyProduct` с методом `discount()`; наследников `Medicine`, `Vitamin`; функцию `process_product(product: PharmacyProduct)`, которая работает с любым товаром через родительский интерфейс. Добавьте `MedicalDevice` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_product` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В аптеке есть служебный объект `PriceCalculator`, который должен уметь рассчитывать итоговую стоимость с учётом НДС. Удобно, чтобы его можно было вызывать как функцию: `calculator(price)`, а не `calculator.calculate(price)`. Реализуйте в классе `PriceCalculator` магический метод `__call__(self, price)`, который возвращает итоговую стоимость (например, `price * 1.1`). Продемонстрируйте, что объект можно вызывать как функцию: `calc = PriceCalculator(); calc(150)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В аптеке есть класс `Medicine`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Medicine` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования лекарств по названию и форме): два вызова `Medicine("Аспирин", "таблетки", 150)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В аптеке есть базовый класс `PharmacyProduct` с методом `get_info()`. Нужно добавлять к товару дополнительную информацию — например, «по рецепту», «детский», «новинка». При этом нельзя менять сам класс `PharmacyProduct`. Реализуйте паттерн «Декоратор»: класс `ProductDecorator`, который содержит объект `PharmacyProduct` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `PrescriptionDecorator` и `KidsDecorator`, которые можно комбинировать: `PrescriptionDecorator(KidsDecorator(medicine))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В аптеке появилась внешняя система поставщика, которая возвращает данные о лекарствах в другом формате — например, в виде словаря `{"title": "Аспирин", "type": "таблетки", "cost": 150}`. А весь остальной код аптеки работает с классом `Medicine`, у которого поля `name`, `form`, `price`. Реализуйте паттерн «Адаптер»: класс `MedicineAdapter`, который принимает словарь и предоставляет интерфейс `Medicine` — свойства `name`, `form`, `price` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Medicine` — например, передать его в функцию, которая работает с лекарством. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В аптеке есть базовый класс `PharmacyProduct`, и постепенно в нём накопились методы, которые нужны не всем товарам: `prescribe()` (только для рецептурных лекарств), `apply()` (только для мазей), `take()` (только для таблеток и сиропов). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Prescribable`, `Applicable`, `Takable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `SaleService`, который зависит не от конкретных классов (`Medicine`, `Vitamin`), а от абстракции `PharmacyProduct` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип товара (например, `MedicalDevice`, реализующий только `Applicable`) без изменения `SaleService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В IT-компании ведётся учёт сотрудников. Создайте класс `Employee` с полями `name` (имя), `position` (должность) и `experience` (стаж в годах). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о сотруднике в формате: `Имя — должность, стаж ... лет`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Иван Петров — Senior Python Developer, стаж 5 лет`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого сотрудника есть размер зарплаты, который нельзя менять или читать напрямую. Создайте класс `Contract`, в котором зарплата хранится в приватном поле `__salary`. Реализуйте метод `set_salary(value)`, предназначенный только для установки зарплаты при приёме на работу, и метод `get_salary()` для чтения значения. Повторная попытка изменить зарплату после подписания контракта должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте контракт, установите зарплату, попробуйте изменить её ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `contract.__salary` — будет ошибка `AttributeError`, но через name mangling (`contract._Contract__salary`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В IT-компании работают разные категории сотрудников. Создайте базовый класс `Staff` с методом `get_role()`, возвращающим строку «Сотрудник». Унаследуйте от него классы `Developer` и `Manager`, переопределив в каждом метод `get_role()` так, чтобы он возвращал «Разработчик» и «Менеджер» соответственно. Создайте по одному объекту каждого класса и выведите их роль, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** IT-компания рассчитывает месячную стоимость сотрудника для разных ролей. Создайте классы `Junior`, `Middle` и `Senior`, у каждого из которых есть метод `monthly_cost()`, возвращающий стоимость: для джуна — 100000, для мидла — 200000, для сеньора — 350000. Напишите функцию `show_cost(dev)`, которая принимает любого разработчика и выводит его стоимость. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `TeamLead` (тимлид), в котором есть три поля: `name` — публичное (доступно всем), `_project` — защищённое (проект, доступно внутри класса и наследникам), `__bonus` — приватное (бонус, доступен только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_project` — можно, но не рекомендуется, к `__bonus` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_TeamLead__bonus`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** IT-компании нужна утилита для проверки email-адресов сотрудников, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `CompanyUtils` со статическим методом `is_valid_email(email)`, который возвращает `True`, если строка содержит символ `@` и заканчивается на `.com`, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `CompanyUtils.is_valid_email("ivan@company.com")`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `CompanyMember` с абстрактным методом `describe()`. Этот класс описывает общее понятие «участник компании», но не может существовать сам по себе — у каждого конкретного участника должно быть своё описание. Унаследуйте от него классы `Developer` и `Manager`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `CompanyMember` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждого сотрудника есть ноутбук — сотрудник не может работать без него, и ноутбук создаётся вместе с сотрудником. Создайте класс `Laptop` с методом `show()`, который выводит строку «Рабочий ноутбук выдан». Создайте класс `Employee`, который содержит объект класса `Laptop` внутри себя как поле (композиция) — объект `Laptop` создаётся прямо в конструкторе сотрудника. Метод `show_employee()` у сотрудника должен выводить имя и вызывать метод `show()` у его ноутбука, демонстрируя, что сотрудник «владеет» своим ноутбуком. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В IT-компании есть базовый класс `CompanyMember` с методом `get_info()`, который возвращает строку «Участник компании». Унаследуйте от него класс `Developer` и переопределите метод `get_info()` так, чтобы он возвращал строку «Разработчик: <имя>». Создайте объект класса `Developer`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** HR-менеджеру нужен метод для оформления заявок, который мог бы работать в разных ситуациях: иногда известен только сотрудник, иногда сотрудник и тип заявки (отпуск, больничный), иногда ещё и количество дней. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `HRDesk` с методом `submit(employee, type=None, days=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Принята заявка от сотрудника: <name>», «Сотрудник <name> подал заявку на: <type>» или «Сотрудник <name> подал заявку на <type> на <days> дней». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Сотрудников часто выводят на экран — в списках команды, отчётах, профилях. По умолчанию `print(employee)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Employee` магический метод `__str__`, который возвращает строку вида «Сотрудник "Иван Петров" (Senior Python Developer, 5 лет стажа)». Убедитесь, что при вызове `print(employee)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Employee('Иван Петров', 'Senior Python Developer', 5)` — оно используется при выводе списка сотрудников и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В IT-компании может оказаться несколько записей об одном и том же сотруднике — например, при переводе с одной должности на другую. Нужно, чтобы программа понимала, что это «один и тот же сотрудник», и не дублировала его в списках. Реализуйте в классе `Employee` магический метод `__eq__`, чтобы два сотрудника считались равными, если совпадают их имя и должность. Дополнительно реализуйте `__hash__`, чтобы сотрудников можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых сотрудника попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В IT-компании есть команды, и иногда нужно объединять их (например, при реорганизации) или узнавать, сколько человек в команде. Создайте класс `Team` (команда), в котором хранится список сотрудников. Реализуйте магический метод `__add__` для объединения двух команд в одну (возвращается **новая** команда со всеми сотрудниками) и метод `__len__`, возвращающий количество сотрудников. Продемонстрируйте работу: создайте две команды, объедините их через `t3 = t1 + t2` и выведите `len(t3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В IT-компании важно, чтобы стаж сотрудника был корректным — нельзя допустить отрицательное значение или значение больше 60. Перепишите класс `Employee` так, чтобы поле `experience` было реализовано через свойство `@property`. Проверки: стаж должен быть целым числом и лежать в диапазоне от 0 до 60 включительно. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `employee.experience` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `employee.experience = -1` → ошибка, `employee.experience = 100` → ошибка, `employee.experience = "abc"` → ошибка, `employee.experience = 5` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В IT-компанию часто поступают данные о сотрудниках в виде строк — например, из файла или базы данных: `"Иван Петров;Senior Python Developer;5"`. Добавьте в класс `Employee` классовый метод `from_string(cls, data)`, который создаёт объект сотрудника из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание сотрудника через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** IT-компании нужно вести общий счётчик всех сотрудников — сколько всего сотрудников зарегистрировано в системе. Добавьте в класс `Employee` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового сотрудника. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько сотрудников и выведите `Employee.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`employee.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В IT-компании возможна ситуация, когда сотрудник пытается взять задачу, которой нет в трекере. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `TaskNotFoundError(Exception)` и используйте его в классе `Company` (компания). Метод `assign_task(employee, task_id)` должен назначать задачу, если она есть, и выбрасывать `TaskNotFoundError` с понятным сообщением (например, с указанием идентификатора), если задачи нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В IT-компании сотрудник работает над проектами — но проекты не принадлежат сотруднику, они существуют отдельно и принадлежат компании. Реализуйте класс `Employee` (сотрудник), который содержит список проектов, над которыми он работает. Проекты при этом существуют отдельно от сотрудника — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого проекта есть ссылка на `Company`, в которой он ведётся. Продемонстрируйте, что при удалении сотрудника проекты остаются в компании. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В IT-компании работает человек, который одновременно и разработчик, и менеджер — например, тимлид, который и код пишет, и командой управляет. Создайте два класса-роли: `Developer` (разработчик) и `Manager` (менеджер). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `TechLead`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `TechLead.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Employee` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Employee` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Employee` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Employee(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В IT-компании может храниться огромное количество записей о сотрудниках — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Employee` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Employee` атрибут `__slots__ = ("name", "position", "experience")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `employee.phone = "..."` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Employee` для полей `experience` (например, диапазон 0–60) и `grade` (диапазон 1–10), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В IT-компании должен быть только один реестр сотрудников — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `Company` будет только один экземпляр. Проверьте, что `Company()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `EmployeeRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** IT-компания регулярно пополняется разными типами сотрудников — разработчиками, менеджерами, тестировщиками. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `EmployeeFactory` с методом `create(kind, **kwargs)`, который по строке `"developer"`, `"manager"` или `"qa"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному сотруднику каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Сотрудники хотят получать уведомления, когда появляется новая задача, меняется дедлайн или начинается новый проект. Реализуйте паттерн «Наблюдатель»: класс `Project` хранит список подписчиков (`Employee`) и при добавлении новой задачи уведомляет всех через метод `notify()`. Продемонстрируйте, что при добавлении задачи все сотрудники получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки сотрудник больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Company` со временем разросся: он и хранит сотрудников, и распределяет задачи, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Company` на несколько классов: `EmployeeStorage` (хранение сотрудников и поиск по имени), `TaskService` (распределение задач, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «тимлид назначил задачу разработчику». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Тимлиды (`TechLead`) похожи на обычных разработчиков, но имеют дополнительные поля — количество подчинённых, зона ответственности. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`TechLead(Developer)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`TechLead` содержит объект `Developer` внутри себя как поле `self.developer`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В IT-компании нужно создавать копии профилей сотрудников — например, для архивного хранения. При этом у сотрудника есть вложенный объект `Laptop`, и важно понимать, копируется ли он вместе с сотрудником или остаётся общим. Реализуйте в классе `Employee` методы `__copy__` и `__deepcopy__`. Создайте сотрудника с вложенным объектом `Laptop` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — ноутбук общий (`shallow.laptop is employee.laptop` → `True`), при глубоком — свой (`deep.laptop is employee.laptop` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В IT-компании есть разные объекты, у которых есть метод `describe()` — сотрудники, проекты, задачи, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Employee`, но и с любым другим классом, у которого есть `describe()` (например, классом `Project`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В IT-компании есть команда `Team`, и хочется перебирать сотрудников в ней в цикле `for` — так же удобно, как список. Сейчас `for emp in team:` не работает. Реализуйте в классе `Team` магические методы `__iter__` и `__next__` так, чтобы по команде можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте команду с несколькими сотрудниками и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(team)` должен возвращать список всех сотрудников.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В команде сотрудники должны быть доступны по индексу — как элементы списка: `team[0]` возвращает первого сотрудника, `team[1] = employee` заменяет второго. Реализуйте в классе `Team` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `team[0]` — получение сотрудника по индексу, `team[1] = new_employee` — замена сотрудника, срезы `team[0:2]` — получение подсписка, `for emp in team` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `employee in team`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В IT-компании есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных задач её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `TaskDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with TaskDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В IT-компании есть разные способы расчёта бонуса сотрудника: фиксированный (10 000 ₽), процентный (10% от оклада) и «за переработку» (половина фиксированного плюс надбавка). Реализуйте паттерн «Стратегия»: базовый класс `BonusStrategy` с методом `calculate(salary)`, и три наследника — `FixedBonus`, `PercentBonus`, `OvertimeBonus`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `OvertimeBonus`, чтобы вызвать реализацию `FixedBonus` и добавить надбавку. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В IT-компании появился новый тип сотрудника — DevOps-инженер (`DevOps`). Если добавлять его через изменение базового класса `CompanyMember`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `CompanyMember` с методом `monthly_cost()`; наследников `Developer`, `Manager`; функцию `process_member(member: CompanyMember)`, которая работает с любым сотрудником через родительский интерфейс. Добавьте `DevOps` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_member` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В IT-компании есть служебный объект `CostCalculator`, который должен уметь рассчитывать стоимость часа работы сотрудника. Удобно, чтобы его можно было вызывать как функцию: `calculator(salary)`, а не `calculator.calculate(salary)`. Реализуйте в классе `CostCalculator` магический метод `__call__(self, salary)`, который возвращает стоимость часа (например, `salary / 160`). Продемонстрируйте, что объект можно вызывать как функцию: `calc = CostCalculator(); calc(160000)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В IT-компании есть класс `Employee`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Employee` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования сотрудников по имени и должности): два вызова `Employee("Иван Петров", "Senior Python Developer", 5)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В IT-компании есть базовый класс `CompanyMember` с методом `get_info()`. Нужно добавлять к сотруднику дополнительную информацию — например, «на испытательном сроке», «удалёнщик», «ментор». При этом нельзя менять сам класс `CompanyMember`. Реализуйте паттерн «Декоратор»: класс `MemberDecorator`, который содержит объект `CompanyMember` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `ProbationDecorator` и `RemoteDecorator`, которые можно комбинировать: `ProbationDecorator(RemoteDecorator(employee))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В IT-компании появилась внешняя HR-система, которая возвращает данные о сотрудниках в другом формате — например, в виде словаря `{"full_name": "Иван Петров", "role": "Senior Python Developer", "years": 5}`. А весь остальной код компании работает с классом `Employee`, у которого поля `name`, `position`, `experience`. Реализуйте паттерн «Адаптер»: класс `EmployeeAdapter`, который принимает словарь и предоставляет интерфейс `Employee` — свойства `name`, `position`, `experience` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Employee` — например, передать его в функцию, которая работает с сотрудником. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В IT-компании есть базовый класс `CompanyMember`, и постепенно в нём накопились методы, которые нужны не всем сотрудникам: `code()` (только для разработчиков), `manage()` (только для менеджеров), `test()` (только для тестировщиков). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Codable`, `Manageable`, `Testable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `TaskService`, который зависит не от конкретных классов (`Developer`, `Manager`), а от абстракции `Codable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип сотрудника (например, `DevOps`, реализующий только `Codable`) без изменения `TaskService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** На почте ведётся учёт отправлений. Создайте класс `Parcel` с полями `sender` (отправитель), `recipient` (получатель) и `weight` (вес в граммах). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией в формате: `От: ... Кому: ..., вес ... г`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `От: Иван Петров Кому: Мария Сидорова, вес 1500 г`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого отправления есть трек-номер, который нельзя менять или читать напрямую. Создайте класс `Tracking`, в котором трек-номер хранится в приватном поле `__track_id`. Реализуйте метод `set_track_id(value)`, предназначенный только для установки номера при приёме отправления, и метод `get_track_id()` для чтения значения. Повторная попытка изменить номер после приёма должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте трек, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `track.__track_id` — будет ошибка `AttributeError`, но через name mangling (`track._Tracking__track_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** На почте есть разные типы отправлений. Создайте базовый класс `MailItem` с методом `get_mail_type()`, возвращающим строку «Отправление». Унаследуйте от него классы `Letter` и `Parcel`, переопределив в каждом метод `get_mail_type()` так, чтобы он возвращал «Письмо» и «Посылка» соответственно. Создайте по одному объекту каждого класса и выведите их тип, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Почта рассчитывает стоимость доставки для разных типов отправлений. Создайте классы `Letter`, `Parcel` и `ExpressMail`, у каждого из которых есть метод `delivery_cost()`, возвращающий стоимость: для письма — 50, для посылки — 300, для экспресс-почты — 800. Напишите функцию `show_cost(item)`, которая принимает любое отправление и выводит его стоимость. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `Postmaster` (начальник почты), в котором есть три поля: `name` — публичное (доступно всем), `_branch` — защищённое (отделение, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_branch` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_Postmaster__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Почте нужна утилита для проверки почтовых индексов, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `PostUtils` со статическим методом `is_valid_index(index)`, который возвращает `True`, если строка состоит только из цифр и её длина равна 6, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `PostUtils.is_valid_index("123456")`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `PostItem` с абстрактным методом `describe()`. Этот класс описывает общее понятие «почтовое отправление», но не может существовать сам по себе — у каждого конкретного отправления должно быть своё описание. Унаследуйте от него классы `Letter` и `Parcel`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `PostItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждой посылки есть упаковочная коробка — посылка не может существовать без неё, и коробка создаётся вместе с посылкой. Создайте класс `Box` с методом `show()`, который выводит строку «Картонная коробка». Создайте класс `Parcel`, который содержит объект класса `Box` внутри себя как поле (композиция) — объект `Box` создаётся прямо в конструкторе посылки. Метод `show_parcel()` у посылки должен выводить имя получателя и вызывать метод `show()` у её коробки, демонстрируя, что посылка «владеет» своей коробкой. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** На почте есть базовый класс `PostItem` с методом `get_info()`, который возвращает строку «Почтовое отправление». Унаследуйте от него класс `Parcel` и переопределите метод `get_info()` так, чтобы он возвращал строку «Посылка для: <получатель>». Создайте объект класса `Parcel`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Оператору почты нужен метод для оформления отправлений, который мог бы работать в разных ситуациях: иногда известна только посылка, иногда посылка и адрес, иногда ещё и срочность. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `PostDesk` с методом `send(parcel, address=None, express=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Оформлена отправка посылки: <recipient>», «Посылка для <recipient> отправлена по адресу: <address>» или «Посылка для <recipient> отправлена по адресу <address>, срочность: <express>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Отправления часто выводят на экран — в квитанциях, отчётах, списках. По умолчанию `print(parcel)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Parcel` магический метод `__str__`, который возвращает строку вида «Посылка "От: Иван Петров → Кому: Мария Сидорова" (1500 г)». Убедитесь, что при вызове `print(parcel)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Parcel('Иван Петров', 'Мария Сидорова', 1500)` — оно используется при выводе списка отправлений и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** На почте может оказаться несколько записей об одном и том же отправлении — например, при повторной регистрации. Нужно, чтобы программа понимала, что это «одно и то же отправление», и не дублировала его в списках. Реализуйте в классе `Parcel` магический метод `__eq__`, чтобы две посылки считались равными, если совпадают отправитель и получатель. Дополнительно реализуйте `__hash__`, чтобы посылки можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что две одинаковые посылки попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** На почте есть контейнеры для отправлений, и иногда нужно объединять их (например, при отправке сборного груза) или узнавать, сколько отправлений внутри. Создайте класс `MailContainer` (контейнер), в котором хранится список отправлений. Реализуйте магический метод `__add__` для объединения двух контейнеров в один (возвращается **новый** контейнер со всеми отправлениями) и метод `__len__`, возвращающий количество отправлений. Продемонстрируйте работу: создайте два контейнера, объедините их через `c3 = c1 + c2` и выведите `len(c3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** На почте важно, чтобы вес отправления был корректным — нельзя допустить отрицательный вес или ноль. Перепишите класс `Parcel` так, чтобы поле `weight` было реализовано через свойство `@property`. Проверки: вес должен быть числом (int или float) и строго больше нуля. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `parcel.weight` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `parcel.weight = -100` → ошибка, `parcel.weight = 0` → ошибка, `parcel.weight = "abc"` → ошибка, `parcel.weight = 1500` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** На почту часто поступают данные об отправлениях в виде строк — например, из файла или базы данных: `"Иван Петров;Мария Сидорова;1500"`. Добавьте в класс `Parcel` классовый метод `from_string(cls, data)`, который создаёт объект посылки из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание посылки через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Почте нужно вести общий счётчик всех отправлений — сколько всего отправлений зарегистрировано в системе. Добавьте в класс `Parcel` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового отправления. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько посылок и выведите `Parcel.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`parcel.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** На почте возможна ситуация, когда клиент пытается получить отправление, которого нет в базе. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `ParcelNotFoundError(Exception)` и используйте его в классе `PostOffice` (почтовое отделение). Метод `find_parcel(track_id)` должен возвращать посылку, если она есть, и выбрасывать `ParcelNotFoundError` с понятным сообщением (например, с указанием трек-номера), если её нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** На почте курьер развозит посылки — но посылки не принадлежат курьеру, они существуют отдельно и принадлежат почте. Реализуйте класс `Courier` (курьер), который содержит список посылок для доставки. Посылки при этом существуют отдельно от курьера — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждой посылки есть ссылка на `PostOffice`, в котором она зарегистрирована. Продемонстрируйте, что при удалении курьера посылки остаются на почте. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** На почте работает человек, который одновременно и курьер, и клиент — например, сотрудник, отправляющий посылки своей семье. Создайте два класса-роли: `Customer` (клиент) и `Employee` (сотрудник). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `CourierCustomer`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `CourierCustomer.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Parcel` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Parcel` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Parcel` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Parcel(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** На почте может храниться огромное количество записей об отправлениях — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Parcel` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Parcel` атрибут `__slots__ = ("sender", "recipient", "weight")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `parcel.note = "..."` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Parcel` для полей `weight` (диапазон 1–50000) и `declared_value` (диапазон 0–1000000), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** На почте должен быть только один реестр отправлений — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `PostOffice` будет только один экземпляр. Проверьте, что `PostOffice()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `ParcelRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Почта регулярно пополняется разными типами отправлений — письмами, посылками, экспресс-почтой. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `MailFactory` с методом `create(kind, **kwargs)`, который по строке `"letter"`, `"parcel"` или `"express"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному отправлению каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Клиенты хотят получать уведомления, когда их посылка прибыла в отделение или изменился статус доставки. Реализуйте паттерн «Наблюдатель»: класс `PostOffice` хранит список подписчиков (`Customer`) и при изменении статуса посылки уведомляет всех через метод `notify()`. Продемонстрируйте, что при изменении статуса все подписчики получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки клиент больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `PostOffice` со временем разросся: он и хранит отправления, и оформляет доставку, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `PostOffice` на несколько классов: `ParcelStorage` (хранение отправлений и поиск по трек-номеру), `DeliveryService` (оформление и доставка, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «клиент отправил посылку». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Международные посылки (`InternationalParcel`) похожи на обычные, но имеют дополнительные поля — страна назначения, таможенная декларация. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`InternationalParcel(Parcel)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`InternationalParcel` содержит объект `Parcel` внутри себя как поле `self.parcel`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** На почте нужно создавать копии квитанций — например, для архивного хранения. При этом у посылки есть вложенный объект `Box`, и важно понимать, копируется ли он вместе с посылкой или остаётся общим. Реализуйте в классе `Parcel` методы `__copy__` и `__deepcopy__`. Создайте посылку с вложенным объектом `Box` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — коробка общая (`shallow.box is parcel.box` → `True`), при глубоком — своя (`deep.box is parcel.box` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** На почте есть разные объекты, у которых есть метод `describe()` — письма, посылки, курьеры, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Parcel`, но и с любым другим классом, у которого есть `describe()` (например, классом `Courier`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** На почте есть контейнер `MailContainer`, и хочется перебирать отправления в нём в цикле `for` — так же удобно, как список. Сейчас `for parcel in container:` не работает. Реализуйте в классе `MailContainer` магические методы `__iter__` и `__next__` так, чтобы по контейнеру можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте контейнер с несколькими отправлениями и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(container)` должен возвращать список всех отправлений.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В контейнере отправления должны быть доступны по индексу — как элементы списка: `container[0]` возвращает первое отправление, `container[1] = parcel` заменяет второе. Реализуйте в классе `MailContainer` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `container[0]` — получение отправления по индексу, `container[1] = new_parcel` — замена отправления, срезы `container[0:2]` — получение подсписка, `for parcel in container` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `parcel in container`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** На почте есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных отправлений её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `PostDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with PostDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** На почте есть разные способы расчёта стоимости доставки: фиксированная (100 ₽), по весу (10 ₽ за каждые 100 г) и «для постоянных клиентов» (скидка 20% от фиксированной). Реализуйте паттерн «Стратегия»: базовый класс `DeliveryCostStrategy` с методом `calculate(parcel)`, и три наследника — `FixedCost`, `WeightCost`, `LoyaltyCost`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `LoyaltyCost`, чтобы вызвать реализацию `FixedCost` и применить скидку. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** На почте появился новый тип отправления — бандероль (`BookPost`). Если добавлять его через изменение базового класса `PostItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `PostItem` с методом `delivery_cost()`; наследников `Letter`, `Parcel`; функцию `process_item(item: PostItem)`, которая работает с любым отправлением через родительский интерфейс. Добавьте `BookPost` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** На почте есть служебный объект `CostCalculator`, который должен уметь рассчитывать стоимость доставки по весу посылки. Удобно, чтобы его можно было вызывать как функцию: `calculator(weight)`, а не `calculator.calculate(weight)`. Реализуйте в классе `CostCalculator` магический метод `__call__(self, weight)`, который возвращает стоимость доставки (например, 10 ₽ за каждые 100 г). Продемонстрируйте, что объект можно вызывать как функцию: `calc = CostCalculator(); calc(1500)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** На почте есть класс `Parcel`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Parcel` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования посылок по отправителю и получателю): два вызова `Parcel("Иван Петров", "Мария Сидорова", 1500)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** На почте есть базовый класс `PostItem` с методом `get_info()`. Нужно добавлять к отправлению дополнительную информацию — например, «хрупкое», «срочное», «с объявленной ценностью». При этом нельзя менять сам класс `PostItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `PostItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `FragileDecorator` и `UrgentDecorator`, которые можно комбинировать: `FragileDecorator(UrgentDecorator(parcel))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** На почте появилась внешняя система доставки, которая возвращает данные о посылках в другом формате — например, в виде словаря `{"from": "Иван Петров", "to": "Мария Сидорова", "mass": 1500}`. А весь остальной код почты работает с классом `Parcel`, у которого поля `sender`, `recipient`, `weight`. Реализуйте паттерн «Адаптер»: класс `ParcelAdapter`, который принимает словарь и предоставляет интерфейс `Parcel` — свойства `sender`, `recipient`, `weight` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Parcel` — например, передать его в функцию, которая работает с посылкой. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** На почте есть базовый класс `PostItem`, и постепенно в нём накопились методы, которые нужны не всем отправлениям: `track()` (только для отслеживаемых), `insure()` (только для страхуемых), `express()` (только для срочных). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Trackable`, `Insurable`, `Expressable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `DeliveryService`, который зависит не от конкретных классов (`Letter`, `Parcel`), а от абстракции `Trackable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип отправления (например, `BookPost`, реализующий только `Trackable`) без изменения `DeliveryService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В полицейском участке ведётся учёт сотрудников. Создайте класс `Officer` с полями `name` (имя), `rank` (звание) и `years_of_service` (выслуга лет). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о сотруднике в формате: `Имя — звание, выслуга ... лет`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Иван Петров — капитан, выслуга 8 лет`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого сотрудника есть номер служебного удостоверения, который нельзя менять или читать напрямую. Создайте класс `Badge`, в котором номер удостоверения хранится в приватном поле `__badge_id`. Реализуйте метод `set_badge_id(value)`, предназначенный только для установки номера при приёме на службу, и метод `get_badge_id()` для чтения значения. Повторная попытка изменить номер после выдачи должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте удостоверение, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `badge.__badge_id` — будет ошибка `AttributeError`, но через name mangling (`badge._Badge__badge_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В полиции есть разные подразделения. Создайте базовый класс `PoliceUnit` с методом `get_unit()`, возвращающим строку «Подразделение». Унаследуйте от него классы `Patrol` и `Investigation`, переопределив в каждом метод `get_unit()` так, чтобы он возвращал «Патрульная служба» и «Следственный отдел» соответственно. Создайте по одному объекту каждого класса и выведите их подразделение, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Полиция рассчитывает уровень опасности для разных вызовов. Создайте классы `TheftCall`, `AccidentCall` и `EmergencyCall`, у каждого из которых есть метод `danger_level()`, возвращающий уровень опасности по шкале от 1 до 10: для кражи — 3, для ДТП — 6, для ЧП — 10. Напишите функцию `show_danger(call)`, которая принимает любой вызов и выводит его уровень опасности. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `Chief` (начальник отдела), в котором есть три поля: `name` — публичное (доступно всем), `_department` — защищённое (отдел, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_department` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_Chief__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Полиции нужна утилита для проверки номеров протоколов, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `PoliceUtils` со статическим методом `is_valid_protocol(number)`, который возвращает `True`, если строка начинается с букв «PR» и её длина равна 7, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `PoliceUtils.is_valid_protocol("PR12345")`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `PoliceItem` с абстрактным методом `describe()`. Этот класс описывает общее понятие «объект полиции», но не может существовать сам по себе — у каждого конкретного объекта должно быть своё описание. Унаследуйте от него классы `Officer` и `Case`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `PoliceItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждого сотрудника есть пистолет — сотрудник не может выйти на службу без оружия, и пистолет создаётся вместе с сотрудником. Создайте класс `Weapon` с методом `show()`, который выводит строку «Служебный пистолет выдан». Создайте класс `Officer`, который содержит объект класса `Weapon` внутри себя как поле (композиция) — объект `Weapon` создаётся прямо в конструкторе сотрудника. Метод `show_officer()` у сотрудника должен выводить имя и вызывать метод `show()` у его оружия, демонстрируя, что сотрудник «владеет» своим оружием. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В полиции есть базовый класс `PoliceItem` с методом `get_info()`, который возвращает строку «Объект полиции». Унаследуйте от него класс `Officer` и переопределите метод `get_info()` так, чтобы он возвращал строку «Сотрудник: <имя>». Создайте объект класса `Officer`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Дежурному нужен метод для регистрации вызовов, который мог бы работать в разных ситуациях: иногда известен только вызов, иногда вызов и адрес, иногда ещё и назначенный сотрудник. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `DutyDesk` с методом `register(call, address=None, officer=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Зарегистрирован вызов: <call>», «Вызов <call> по адресу: <address>» или «Вызов <call> по адресу <address>, назначен сотрудник: <officer>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Сотрудников часто выводят на экран — в списках личного состава, отчётах, рапортах. По умолчанию `print(officer)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Officer` магический метод `__str__`, который возвращает строку вида «Сотрудник "Иван Петров" (капитан, выслуга 8 лет)». Убедитесь, что при вызове `print(officer)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Officer('Иван Петров', 'капитан', 8)` — оно используется при выводе списка сотрудников и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В полиции может оказаться несколько записей об одном и том же сотруднике — например, при переводе из другого отдела. Нужно, чтобы программа понимала, что это «один и тот же сотрудник», и не дублировала его в списках. Реализуйте в классе `Officer` магический метод `__eq__`, чтобы два сотрудника считались равными, если совпадают их имя и звание. Дополнительно реализуйте `__hash__`, чтобы сотрудников можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых сотрудника попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В полиции есть отряды, и иногда нужно объединять их (например, при совместной операции) или узнавать, сколько сотрудников в отряде. Создайте класс `Squad` (отряд), в котором хранится список сотрудников. Реализуйте магический метод `__add__` для объединения двух отрядов в один (возвращается **новый** отряд со всеми сотрудниками) и метод `__len__`, возвращающий количество сотрудников. Продемонстрируйте работу: создайте два отряда, объедините их через `squad3 = squad1 + squad2` и выведите `len(squad3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В полиции важно, чтобы выслуга лет была корректной — нельзя допустить отрицательное значение или значение больше 60. Перепишите класс `Officer` так, чтобы поле `years_of_service` было реализовано через свойство `@property`. Проверки: выслуга должна быть целым числом и лежать в диапазоне от 0 до 60 включительно. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `officer.years_of_service` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `officer.years_of_service = -1` → ошибка, `officer.years_of_service = 100` → ошибка, `officer.years_of_service = "abc"` → ошибка, `officer.years_of_service = 8` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В полицию часто поступают данные о сотрудниках в виде строк — например, из файла или базы данных: `"Иван Петров;капитан;8"`. Добавьте в класс `Officer` классовый метод `from_string(cls, data)`, который создаёт объект сотрудника из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание сотрудника через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Полиции нужно вести общий счётчик всех сотрудников — сколько всего сотрудников зарегистрировано в системе. Добавьте в класс `Officer` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового сотрудника. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько сотрудников и выведите `Officer.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`officer.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В полиции возможна ситуация, когда сотрудник пытается открыть дело, которое уже закрыто или не существует. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `CaseNotFoundError(Exception)` и используйте его в классе `Police` (полиция). Метод `open_case(case_id)` должен открывать дело, если оно есть, и выбрасывать `CaseNotFoundError` с понятным сообщением (например, с указанием номера дела), если дела нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В полиции сотрудник ведёт дела — но дела не принадлежат сотруднику, они существуют отдельно и принадлежат участку. Реализуйте класс `Officer` (сотрудник), который содержит список дел, которые он ведёт. Дела при этом существуют отдельно от сотрудника — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого дела есть ссылка на `Police`, в котором оно ведётся. Продемонстрируйте, что при удалении сотрудника дела остаются в участке. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В полиции работает человек, который одновременно и следователь, и патрульный — например, сотрудник, который совмещает обе роли. Создайте два класса-роли: `Detective` (следователь) и `Patrolman` (патрульный). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `DetectivePatrolman`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `DetectivePatrolman.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Officer` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Officer` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Officer` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Officer(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В полиции может храниться огромное количество записей о сотрудниках — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Officer` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Officer` атрибут `__slots__ = ("name", "rank", "years_of_service")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `officer.phone = "..."` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Officer` для полей `years_of_service` (например, диапазон 0–60) и `rank_level` (диапазон 1–10), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В полиции должен быть только один реестр сотрудников — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `Police` будет только один экземпляр. Проверьте, что `Police()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `CaseRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Полиция регулярно пополняется разными типами вызовов — кражами, ДТП, чрезвычайными происшествиями. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `CallFactory` с методом `create(kind, **kwargs)`, который по строке `"theft"`, `"accident"` или `"emergency"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному вызову каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Сотрудники хотят получать уведомления о новых вызовах или изменениях в делах, которые они ведут. Реализуйте паттерн «Наблюдатель»: класс `Police` хранит список подписчиков (`Officer`) и при регистрации нового вызова уведомляет всех через метод `notify()`. Продемонстрируйте, что при регистрации вызова все сотрудники получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки сотрудник больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Police` со временем разросся: он и хранит сотрудников, и регистрирует вызовы, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Police` на несколько классов: `OfficerStorage` (хранение сотрудников и поиск по имени), `CaseService` (регистрация вызовов и дел, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «зарегистрирован новый вызов». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Сотрудники спецназа (`SpecialOfficer`) похожи на обычных сотрудников, но имеют дополнительные поля — уровень подготовки, спецсредства. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`SpecialOfficer(Officer)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`SpecialOfficer` содержит объект `Officer` внутри себя как поле `self.officer`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В полиции нужно создавать копии личных дел сотрудников — например, для архивного хранения. При этом у сотрудника есть вложенный объект `Weapon`, и важно понимать, копируется ли он вместе с сотрудником или остаётся общим. Реализуйте в классе `Officer` методы `__copy__` и `__deepcopy__`. Создайте сотрудника с вложенным объектом `Weapon` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — оружие общее (`shallow.weapon is officer.weapon` → `True`), при глубоком — своё (`deep.weapon is officer.weapon` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В полиции есть разные объекты, у которых есть метод `describe()` — сотрудники, дела, вызовы, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Officer`, но и с любым другим классом, у которого есть `describe()` (например, классом `Case`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В полиции есть отряд `Squad`, и хочется перебирать сотрудников в нём в цикле `for` — так же удобно, как список. Сейчас `for officer in squad:` не работает. Реализуйте в классе `Squad` магические методы `__iter__` и `__next__` так, чтобы по отряду можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте отряд с несколькими сотрудниками и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(squad)` должен возвращать список всех сотрудников.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В отряде сотрудники должны быть доступны по индексу — как элементы списка: `squad[0]` возвращает первого сотрудника, `squad[1] = officer` заменяет второго. Реализуйте в классе `Squad` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `squad[0]` — получение сотрудника по индексу, `squad[1] = new_officer` — замена сотрудника, срезы `squad[0:2]` — получение подсписка, `for officer in squad` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `officer in squad`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В полиции есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных дел её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `CaseDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with CaseDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В полиции есть разные способы расчёта штрафа за нарушение: фиксированный (1000 ₽), прогрессивный (за повторное нарушение — двойной штраф) и «мягкий» (для несовершеннолетних — половина от фиксированного). Реализуйте паттерн «Стратегия»: базовый класс `FineStrategy` с методом `calculate(violations)`, и три наследника — `FixedFine`, `ProgressiveFine`, `JuvenileFine`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `JuvenileFine`, чтобы вызвать реализацию `FixedFine` и уменьшить результат вдвое. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В полиции появился новый тип вызова — киберпреступление (`CybercrimeCall`). Если добавлять его через изменение базового класса `PoliceItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `PoliceItem` с методом `danger_level()`; наследников `TheftCall`, `AccidentCall`; функцию `process_item(item: PoliceItem)`, которая работает с любым вызовом через родительский интерфейс. Добавьте `CybercrimeCall` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В полиции есть служебный объект `FineCalculator`, который должен уметь рассчитывать штраф за нарушение. Удобно, чтобы его можно было вызывать как функцию: `calculator(violations)`, а не `calculator.calculate(violations)`. Реализуйте в классе `FineCalculator` магический метод `__call__(self, violations)`, который возвращает сумму штрафа (например, 1000 ₽ за каждое нарушение). Продемонстрируйте, что объект можно вызывать как функцию: `calc = FineCalculator(); calc(3)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В полиции есть класс `Officer`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Officer` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования сотрудников по имени и званию): два вызова `Officer("Иван Петров", "капитан", 8)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В полиции есть базовый класс `PoliceItem` с методом `get_info()`. Нужно добавлять к объекту дополнительную информацию — например, «секретно», «срочно», «в архиве». При этом нельзя менять сам класс `PoliceItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `PoliceItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `ClassifiedDecorator` и `UrgentDecorator`, которые можно комбинировать: `ClassifiedDecorator(UrgentDecorator(case))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В полиции появилась внешняя федеральная база данных, которая возвращает данные о сотрудниках в другом формате — например, в виде словаря `{"full_name": "Иван Петров", "rank_title": "капитан", "service_years": 8}`. А весь остальной код полиции работает с классом `Officer`, у которого поля `name`, `rank`, `years_of_service`. Реализуйте паттерн «Адаптер»: класс `OfficerAdapter`, который принимает словарь и предоставляет интерфейс `Officer` — свойства `name`, `rank`, `years_of_service` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Officer` — например, передать его в функцию, которая работает с сотрудником. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В полиции есть базовый класс `PoliceItem`, и постепенно в нём накопились методы, которые нужны не всем объектам: `arrest()` (только для сотрудников), `patrol()` (только для патрульных), `investigate()` (только для следователей). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Arrestable`, `Patrollable`, `Investigable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `CaseService`, который зависит не от конкретных классов (`Detective`, `Patrolman`), а от абстракции `Investigable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип сотрудника (например, `CyberOfficer`, реализующий только `Investigable`) без изменения `CaseService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В пекарне ведётся учёт выпечки. Создайте класс `BakeryItem` с полями `name` (название), `type` (тип: хлеб, булочка, пирог) и `price` (цена). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о выпечке в формате: `Название — тип, цена`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Багет — хлеб, 80 ₽`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого рецепта есть секретный код, который нельзя менять или читать напрямую. Создайте класс `Recipe`, в котором код рецепта хранится в приватном поле `__recipe_code`. Реализуйте метод `set_recipe_code(value)`, предназначенный только для установки кода при разработке рецепта, и метод `get_recipe_code()` для чтения значения. Повторная попытка изменить код после утверждения должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте рецепт, установите код, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `recipe.__recipe_code` — будет ошибка `AttributeError`, но через name mangling (`recipe._Recipe__recipe_code`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В пекарне есть разные виды изделий. Создайте базовый класс `Baked` с методом `get_kind()`, возвращающим строку «Выпечка». Унаследуйте от него классы `Bread` и `Cake`, переопределив в каждом метод `get_kind()` так, чтобы он возвращал «Хлеб» и «Торт» соответственно. Создайте по одному объекту каждого класса и выведите их вид, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Пекарня рассчитывает время выпекания для разных изделий. Создайте классы `Bread`, `Bun` и `Pie`, у каждого из которых есть метод `bake_time()`, возвращающий время в минутах: для хлеба — 45, для булочки — 20, для пирога — 60. Напишите функцию `show_bake_time(item)`, которая принимает любое изделие и выводит время его выпекания. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `Baker` (пекарь), в котором есть три поля: `name` — публичное (доступно всем), `_shift` — защищённое (смена, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_shift` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_Baker__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Пекарне нужна утилита для проверки количества муки, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `BakeryUtils` со статическим методом `is_enough_flour(kg)`, который возвращает `True`, если количество муки больше 10 кг, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `BakeryUtils.is_enough_flour(25)`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `BakeryProduct` с абстрактным методом `describe()`. Этот класс описывает общее понятие «изделие пекарни», но не может существовать сам по себе — у каждого конкретного изделия должно быть своё описание. Унаследуйте от него классы `Bread` и `Cake`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `BakeryProduct` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждого торта есть коробка — торт не может продаваться без неё, и коробка создаётся вместе с тортом. Создайте класс `CakeBox` с методом `show()`, который выводит строку «Праздничная коробка». Создайте класс `Cake`, который содержит объект класса `CakeBox` внутри себя как поле (композиция) — объект `CakeBox` создаётся прямо в конструкторе торта. Метод `show_cake()` у торта должен выводить название и вызывать метод `show()` у его коробки, демонстрируя, что торт «владеет» своей коробкой. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В пекарне есть базовый класс `BakeryProduct` с методом `get_info()`, который возвращает строку «Изделие пекарни». Унаследуйте от него класс `Cake` и переопределите метод `get_info()` так, чтобы он возвращал строку «Торт: <название>». Создайте объект класса `Cake`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Продавцу пекарни нужен метод для оформления заказов, который мог бы работать в разных ситуациях: иногда известно только изделие, иногда изделие и количество, иногда ещё и покупатель. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `OrderDesk` с методом `take(item, quantity=None, customer=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Оформлен заказ на изделие: <name>», «Оформлен заказ: <name> × <quantity>» или «Оформлен заказ: <name> × <quantity> для <customer>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Изделия часто выводят на экран — в витринах, чеках, отчётах. По умолчанию `print(item)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `BakeryItem` магический метод `__str__`, который возвращает строку вида «Изделие "Багет" (хлеб, 80 ₽)». Убедитесь, что при вызове `print(item)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `BakeryItem('Багет', 'хлеб', 80)` — оно используется при выводе списка изделий и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В пекарне может оказаться несколько записей об одном и том же изделии — например, при повторном добавлении в ассортимент. Нужно, чтобы программа понимала, что это «одно и то же изделие», и не дублировала его в меню. Реализуйте в классе `BakeryItem` магический метод `__eq__`, чтобы два изделия считались равными, если совпадают их название и тип. Дополнительно реализуйте `__hash__`, чтобы изделия можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых изделия попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В пекарне есть противни с изделиями, и иногда нужно объединять их (например, при объединении партий) или узнавать, сколько изделий на противне. Создайте класс `Tray` (противень), в котором хранится список изделий. Реализуйте магический метод `__add__` для объединения двух противней в один (возвращается **новый** противень со всеми изделиями) и метод `__len__`, возвращающий количество изделий. Продемонстрируйте работу: создайте два противня, объедините их через `t3 = t1 + t2` и выведите `len(t3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В пекарне важно, чтобы цена изделия была корректной — нельзя допустить отрицательную цену или ноль. Перепишите класс `BakeryItem` так, чтобы поле `price` было реализовано через свойство `@property`. Проверки: цена должна быть числом (int или float), и она должна быть строго больше нуля. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `item.price` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `item.price = -100` → ошибка, `item.price = 0` → ошибка, `item.price = "abc"` → ошибка, `item.price = 80` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В пекарню часто поступают данные об изделиях в виде строк — например, из файла или базы данных: `"Багет;хлеб;80"`. Добавьте в класс `BakeryItem` классовый метод `from_string(cls, data)`, который создаёт объект изделия из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание изделия через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Пекарне нужно вести общий счётчик всех изделий — сколько всего изделий зарегистрировано в системе. Добавьте в класс `BakeryItem` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового изделия. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько изделий и выведите `BakeryItem.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`item.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В пекарне возможна ситуация, когда пекарь пытается испечь изделие по рецепту, которого нет в базе. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `RecipeNotFoundError(Exception)` и используйте его в классе `Bakery` (пекарня). Метод `find_recipe(name)` должен возвращать рецепт, если он есть, и выбрасывать `RecipeNotFoundError` с понятным сообщением (например, с указанием названия рецепта), если его нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В пекарне пекарь работает с рецептами — но рецепты не принадлежат пекарю, они существуют отдельно и принадлежат пекарне. Реализуйте класс `Baker` (пекарь), который содержит список рецептов, с которыми он работает. Рецепты при этом существуют отдельно от пекаря — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого рецепта есть ссылка на `Bakery`, в которой он используется. Продемонстрируйте, что при удалении пекаря рецепты остаются в пекарне. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В пекарне работает человек, который одновременно и пекарь, и кондитер — например, универсальный специалист, который и хлеб печёт, и торты делает. Создайте два класса-роли: `Baker` (пекарь) и `Confectioner` (кондитер). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `BakerConfectioner`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `BakerConfectioner.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `BakeryItem` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `BakeryItem` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `BakeryItem` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class BakeryItem(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В пекарне может храниться огромное количество записей об изделиях — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `BakeryItem` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `BakeryItem` атрибут `__slots__ = ("name", "type", "price")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `item.weight = 500` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `BakeryItem` для полей `price` (например, диапазон 1–100000) и `weight` (диапазон 1–5000), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В пекарне должен быть только один реестр рецептов — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `Bakery` будет только один экземпляр. Проверьте, что `Bakery()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `RecipeRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Пекарня регулярно пополняется разными типами изделий — хлебом, булочками, пирогами. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `BakeryFactory` с методом `create(kind, **kwargs)`, который по строке `"bread"`, `"bun"` или `"pie"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному изделию каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Пекари хотят получать уведомления, когда появляется новый рецепт или начинается смена. Реализуйте паттерн «Наблюдатель»: класс `Bakery` хранит список подписчиков (`Baker`) и при добавлении нового рецепта уведомляет всех через метод `notify()`. Продемонстрируйте, что при добавлении рецепта все пекари получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки пекарь больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Bakery` со временем разросся: он и хранит изделия, и принимает заказы, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Bakery` на несколько классов: `ItemStorage` (хранение изделий и поиск по названию), `OrderService` (приём заказов, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «покупатель сделал заказ». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Сдобная выпечка (`SweetBun`) похожа на обычную булочку, но имеет дополнительные поля — вид начинки, глазурь. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`SweetBun(Bun)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`SweetBun` содержит объект `Bun` внутри себя как поле `self.bun`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В пекарне нужно создавать копии карточек изделий — например, для резервного меню. При этом у торта есть вложенный объект `CakeBox`, и важно понимать, копируется ли он вместе с тортом или остаётся общим. Реализуйте в классе `Cake` методы `__copy__` и `__deepcopy__`. Создайте торт с вложенным объектом `CakeBox` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — коробка общая (`shallow.box is cake.box` → `True`), при глубоком — своя (`deep.box is cake.box` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В пекарне есть разные объекты, у которых есть метод `describe()` — хлеб, булочки, торты, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `BakeryItem`, но и с любым другим классом, у которого есть `describe()` (например, классом `Baker`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В пекарне есть противень `Tray`, и хочется перебирать изделия на нём в цикле `for` — так же удобно, как список. Сейчас `for item in tray:` не работает. Реализуйте в классе `Tray` магические методы `__iter__` и `__next__` так, чтобы по противню можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте противень с несколькими изделиями и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(tray)` должен возвращать список всех изделий.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** На противне изделия должны быть доступны по индексу — как элементы списка: `tray[0]` возвращает первое изделие, `tray[1] = item` заменяет второе. Реализуйте в классе `Tray` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `tray[0]` — получение изделия по индексу, `tray[1] = new_item` — замена изделия, срезы `tray[0:2]` — получение подсписка, `for item in tray` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `item in tray`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В пекарне есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных рецептов её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `RecipeDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with RecipeDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В пекарне есть разные способы расчёта скидки: фиксированная (20 ₽), процентная (10% от суммы) и «для постоянных покупателей» (скидка 15% от суммы). Реализуйте паттерн «Стратегия»: базовый класс `DiscountStrategy` с методом `calculate(total)`, и три наследника — `FixedDiscount`, `PercentDiscount`, `LoyaltyDiscount`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `LoyaltyDiscount`, чтобы вызвать реализацию `PercentDiscount` и увеличить скидку (или переиспользовать код родителя). Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В пекарне появился новый вид выпечки — круассан (`Croissant`). Если добавлять его через изменение базового класса `BakeryProduct`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `BakeryProduct` с методом `bake_time()`; наследников `Bread`, `Bun`; функцию `process_item(item: BakeryProduct)`, которая работает с любым изделием через родительский интерфейс. Добавьте `Croissant` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В пекарне есть служебный объект `PriceCalculator`, который должен уметь рассчитывать итоговую стоимость с учётом НДС. Удобно, чтобы его можно было вызывать как функцию: `calculator(price)`, а не `calculator.calculate(price)`. Реализуйте в классе `PriceCalculator` магический метод `__call__(self, price)`, который возвращает итоговую стоимость (например, `price * 1.2`). Продемонстрируйте, что объект можно вызывать как функцию: `calc = PriceCalculator(); calc(80)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В пекарне есть класс `BakeryItem`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `BakeryItem` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования изделий по названию и типу): два вызова `BakeryItem("Багет", "хлеб", 80)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В пекарне есть базовый класс `BakeryProduct` с методом `get_info()`. Нужно добавлять к изделию дополнительную информацию — например, «без сахара», «новинка», «рекомендуем». При этом нельзя менять сам класс `BakeryProduct`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `BakeryProduct` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `SugarFreeDecorator` и `NewItemDecorator`, которые можно комбинировать: `SugarFreeDecorator(NewItemDecorator(bun))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В пекарне появилась внешняя система поставок, которая возвращает данные об изделиях в другом формате — например, в виде словаря `{"title": "Багет", "category": "хлеб", "cost": 80}`. А весь остальной код пекарни работает с классом `BakeryItem`, у которого поля `name`, `type`, `price`. Реализуйте паттерн «Адаптер»: класс `BakeryItemAdapter`, который принимает словарь и предоставляет интерфейс `BakeryItem` — свойства `name`, `type`, `price` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `BakeryItem` — например, передать его в функцию, которая работает с изделием. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В пекарне есть базовый класс `BakeryProduct`, и постепенно в нём накопились методы, которые нужны не всем изделиям: `bake()` (только для хлеба и булочек), `decorate()` (только для тортов), `pack()` (только для подарочной выпечки). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Bakeable`, `Decorable`, `Packable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `OrderService`, который зависит не от конкретных классов (`Bread`, `Cake`), а от абстракции `Bakeable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип изделия (например, `Croissant`, реализующий только `Bakeable`) без изменения `OrderService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В салоне красоты ведётся учёт услуг. Создайте класс `Service` с полями `name` (название), `category` (категория: стрижка, маникюр, массаж) и `price` (цена). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией об услуге в формате: `Название — категория, цена`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Стрижка — парикмахерские услуги, 1500 ₽`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого клиента есть номер карты лояльности, который нельзя менять или читать напрямую. Создайте класс `LoyaltyCard`, в котором номер карты хранится в приватном поле `__card_id`. Реализуйте метод `set_card_id(value)`, предназначенный только для установки номера при регистрации клиента, и метод `get_card_id()` для чтения значения. Повторная попытка изменить номер после регистрации должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте карту, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `card.__card_id` — будет ошибка `AttributeError`, но через name mangling (`card._LoyaltyCard__card_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В салоне красоты работают разные специалисты. Создайте базовый класс `BeautyMaster` с методом `get_specialty()`, возвращающим строку «Специалист». Унаследуйте от него классы `Hairdresser` и `Manicurist`, переопределив в каждом метод `get_specialty()` так, чтобы он возвращал «Парикмахер» и «Мастер маникюра» соответственно. Создайте по одному объекту каждого класса и выведите их специальность, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Салон красоты рассчитывает длительность для разных услуг. Создайте классы `Haircut`, `Manicure` и `Massage`, у каждого из которых есть метод `duration()`, возвращающий длительность в минутах: для стрижки — 60, для маникюра — 90, для массажа — 120. Напишите функцию `show_duration(service)`, которая принимает любую услугу и выводит её длительность. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `SalonManager` (управляющий салона), в котором есть три поля: `name` — публичное (доступно всем), `_cabinet` — защищённое (кабинет, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_cabinet` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_SalonManager__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Салону нужна утилита для проверки времени записи, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `SalonUtils` со статическим методом `is_working_hours(hour)`, который возвращает `True`, если час находится в диапазоне от 9 до 21, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `SalonUtils.is_working_hours(14)`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `SalonItem` с абстрактным методом `describe()`. Этот класс описывает общее понятие «объект салона», но не может существовать сам по себе — у каждого конкретного объекта должно быть своё описание. Унаследуйте от него классы `Service` и `Master`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `SalonItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждого мастера есть рабочее место — мастер не может работать без него, и рабочее место создаётся вместе с мастером. Создайте класс `Workplace` с методом `show()`, который выводит строку «Косметическое кресло готово». Создайте класс `Master`, который содержит объект класса `Workplace` внутри себя как поле (композиция) — объект `Workplace` создаётся прямо в конструкторе мастера. Метод `show_master()` у мастера должен выводить имя и вызывать метод `show()` у его рабочего места, демонстрируя, что мастер «владеет» своим рабочим местом. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В салоне есть базовый класс `SalonItem` с методом `get_info()`, который возвращает строку «Объект салона». Унаследуйте от него класс `Master` и переопределите метод `get_info()` так, чтобы он возвращал строку «Мастер: <имя>». Создайте объект класса `Master`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Администратору салона нужен метод для записи клиентов, который мог бы работать в разных ситуациях: иногда известен только клиент, иногда клиент и услуга, иногда ещё и мастер. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `ReceptionDesk` с методом `book(client, service=None, master=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Записан клиент: <name>», «Клиент <name> записан на услугу: <service>» или «Клиент <name> записан на <service> к мастеру <master>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Услуги часто выводят на экран — в прайсе, чеках, отчётах. По умолчанию `print(service)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Service` магический метод `__str__`, который возвращает строку вида «Услуга "Стрижка" (парикмахерские услуги, 1500 ₽)». Убедитесь, что при вызове `print(service)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Service('Стрижка', 'парикмахерские услуги', 1500)` — оно используется при выводе списка услуг и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В салоне может оказаться несколько записей об одной и той же услуге — например, при повторном добавлении в прайс. Нужно, чтобы программа понимала, что это «одна и та же услуга», и не дублировала её в списке. Реализуйте в классе `Service` магический метод `__eq__`, чтобы две услуги считались равными, если совпадают их название и категория. Дополнительно реализуйте `__hash__`, чтобы услуги можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что две одинаковые услуги попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В салоне есть расписания, и иногда нужно объединять их (например, при объединении смен) или узнавать, сколько услуг в расписании. Создайте класс `Schedule` (расписание), в котором хранится список услуг. Реализуйте магический метод `__add__` для объединения двух расписаний в одно (возвращается **новое** расписание со всеми услугами) и метод `__len__`, возвращающий количество услуг. Продемонстрируйте работу: создайте два расписания, объедините их через `s3 = s1 + s2` и выведите `len(s3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В салоне важно, чтобы цена услуги была корректной — нельзя допустить отрицательную цену или ноль. Перепишите класс `Service` так, чтобы поле `price` было реализовано через свойство `@property`. Проверки: цена должна быть числом (int или float) и должна быть строго больше нуля. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `service.price` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `service.price = -100` → ошибка, `service.price = 0` → ошибка, `service.price = "abc"` → ошибка, `service.price = 1500` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В салон часто поступают данные об услугах в виде строк — например, из файла или базы данных: `"Стрижка;парикмахерские услуги;1500"`. Добавьте в класс `Service` классовый метод `from_string(cls, data)`, который создаёт объект услуги из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание услуги через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Салону нужно вести общий счётчик всех услуг — сколько всего услуг зарегистрировано в системе. Добавьте в класс `Service` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждой новой услуги. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько услуг и выведите `Service.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`service.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В салоне возможна ситуация, когда клиент пытается записаться на услугу, которой нет в прайсе. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `ServiceNotFoundError(Exception)` и используйте его в классе `Salon` (салон). Метод `find_service(name)` должен возвращать услугу, если она есть, и выбрасывать `ServiceNotFoundError` с понятным сообщением (например, с указанием названия), если её нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В салоне мастер выполняет услуги — но услуги не принадлежат мастеру, они существуют отдельно и принадлежат салону. Реализуйте класс `Master` (мастер), который содержит список услуг, которые он выполняет. Услуги при этом существуют отдельно от мастера — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждой услуги есть ссылка на `Salon`, в котором она предоставляется. Продемонстрируйте, что при удалении мастера услуги остаются в салоне. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В салоне работает человек, который одновременно и парикмахер, и мастер маникюра — например, универсальный специалист. Создайте два класса-роли: `Hairdresser` (парикмахер) и `Manicurist` (мастер маникюра). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `UniversalMaster`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `UniversalMaster.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Service` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Service` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Service` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Service(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В салоне может храниться огромное количество записей об услугах — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Service` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Service` атрибут `__slots__ = ("name", "category", "price")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `service.duration = 60` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Service` для полей `price` (например, диапазон 1–100000) и `duration` (диапазон 1–300), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В салоне должен быть только один реестр услуг — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `Salon` будет только один экземпляр. Проверьте, что `Salon()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `ServiceRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Салон регулярно пополняется разными типами услуг — стрижками, маникюром, массажем. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `ServiceFactory` с методом `create(kind, **kwargs)`, который по строке `"haircut"`, `"manicure"` или `"massage"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одной услуге каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Клиенты хотят получать уведомления, когда появляется новая услуга или начинается акция. Реализуйте паттерн «Наблюдатель»: класс `Salon` хранит список подписчиков (`Client`) и при добавлении новой услуги уведомляет всех через метод `notify()`. Продемонстрируйте, что при добавлении услуги все клиенты получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки клиент больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Salon` со временем разросся: он и хранит услуги, и записывает клиентов, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Salon` на несколько классов: `ServiceStorage` (хранение услуг и поиск по названию), `BookingService` (запись клиентов, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «клиент записался на услугу». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Премиум-услуги (`PremiumService`) похожи на обычные услуги, но имеют дополнительные поля — уровень мастера, используемые материалы. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`PremiumService(Service)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`PremiumService` содержит объект `Service` внутри себя как поле `self.service`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В салоне нужно создавать копии карточек услуг — например, для архивного хранения. При этом у мастера есть вложенный объект `Workplace`, и важно понимать, копируется ли он вместе с мастером или остаётся общим. Реализуйте в классе `Master` методы `__copy__` и `__deepcopy__`. Создайте мастера с вложенным объектом `Workplace` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — рабочее место общее (`shallow.workplace is master.workplace` → `True`), при глубоком — своё (`deep.workplace is master.workplace` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В салоне есть разные объекты, у которых есть метод `describe()` — услуги, мастера, клиенты, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Service`, но и с любым другим классом, у которого есть `describe()` (например, классом `Client`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В салоне есть расписание `Schedule`, и хочется перебирать услуги в нём в цикле `for` — так же удобно, как список. Сейчас `for service in schedule:` не работает. Реализуйте в классе `Schedule` магические методы `__iter__` и `__next__` так, чтобы по расписанию можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте расписание с несколькими услугами и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(schedule)` должен возвращать список всех услуг.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В расписании услуги должны быть доступны по индексу — как элементы списка: `schedule[0]` возвращает первую услугу, `schedule[1] = service` заменяет вторую. Реализуйте в классе `Schedule` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `schedule[0]` — получение услуги по индексу, `schedule[1] = new_service` — замена услуги, срезы `schedule[0:2]` — получение подсписка, `for service in schedule` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `service in schedule`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В салоне есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных услуг её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `ServiceDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with ServiceDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В салоне есть разные способы расчёта скидки: фиксированная (100 ₽), процентная (10% от суммы) и «VIP» (для постоянных клиентов — 20% от суммы). Реализуйте паттерн «Стратегия»: базовый класс `DiscountStrategy` с методом `calculate(total)`, и три наследника — `FixedDiscount`, `PercentDiscount`, `VipDiscount`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `VipDiscount`, чтобы вызвать реализацию `PercentDiscount` и увеличить скидку (или переиспользовать код родителя). Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В салоне появился новый тип услуги — спа-процедура (`Spa`). Если добавлять её через изменение базового класса `SalonItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `SalonItem` с методом `duration()`; наследников `Haircut`, `Manicure`; функцию `process_item(item: SalonItem)`, которая работает с любой услугой через родительский интерфейс. Добавьте `Spa` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В салоне есть служебный объект `PriceCalculator`, который должен уметь рассчитывать итоговую стоимость с учётом надбавки за срочность. Удобно, чтобы его можно было вызывать как функцию: `calculator(base_price)`, а не `calculator.calculate(base_price)`. Реализуйте в классе `PriceCalculator` магический метод `__call__(self, base_price)`, который возвращает итоговую стоимость (например, `base_price * 1.3`). Продемонстрируйте, что объект можно вызывать как функцию: `calc = PriceCalculator(); calc(1500)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В салоне есть класс `Service`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Service` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования услуг по названию и категории): два вызова `Service("Стрижка", "парикмахерские услуги", 1500)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В салоне есть базовый класс `SalonItem` с методом `get_info()`. Нужно добавлять к услуге дополнительную информацию — например, «премиум», «срочно», «со скидкой». При этом нельзя менять сам класс `SalonItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `SalonItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `PremiumDecorator` и `UrgentDecorator`, которые можно комбинировать: `PremiumDecorator(UrgentDecorator(service))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В салоне появилась внешняя система онлайн-записи, которая возвращает данные об услугах в другом формате — например, в виде словаря `{"title": "Стрижка", "group": "парикмахерские услуги", "cost": 1500}`. А весь остальной код салона работает с классом `Service`, у которого поля `name`, `category`, `price`. Реализуйте паттерн «Адаптер»: класс `ServiceAdapter`, который принимает словарь и предоставляет интерфейс `Service` — свойства `name`, `category`, `price` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Service` — например, передать его в функцию, которая работает с услугой. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В салоне есть базовый класс `SalonItem`, и постепенно в нём накопились методы, которые нужны не всем объектам: `cut()` (только для парикмахерских услуг), `paint()` (только для маникюра и педикюра), `relax()` (только для массажа и спа). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Cuttable`, `Paintable`, `Relaxable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `BookingService`, который зависит не от конкретных классов (`Haircut`, `Manicure`), а от абстракции `SalonItem` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип услуги (например, `Spa`, реализующий только `Relaxable`) без изменения `BookingService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В турагентстве ведётся учёт туров. Создайте класс `Tour` с полями `country` (страна), `days` (количество дней) и `price` (цена). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о туре в формате: `Страна — ... дней, цена`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Турция — 10 дней, 80000 ₽`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого клиента есть номер паспорта, который нельзя менять или читать напрямую. Создайте класс `ClientPassport`, в котором номер паспорта хранится в приватном поле `__passport_id`. Реализуйте метод `set_passport_id(value)`, предназначенный только для установки номера при оформлении тура, и метод `get_passport_id()` для чтения значения. Повторная попытка изменить номер после оформления должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте паспорт, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `passport.__passport_id` — будет ошибка `AttributeError`, но через name mangling (`passport._ClientPassport__passport_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В турагентстве есть разные типы туров. Создайте базовый класс `TravelPackage` с методом `get_package_type()`, возвращающим строку «Тур». Унаследуйте от него классы `BeachTour` и `ExcursionTour`, переопределив в каждом метод `get_package_type()` так, чтобы он возвращал «Пляжный тур» и «Экскурсионный тур» соответственно. Создайте по одному объекту каждого класса и выведите их тип, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Турагентство рассчитывает стоимость тура для разных направлений. Создайте классы `TourToTurkey`, `TourToEgypt` и `TourToItaly`, у каждого из которых есть метод `tour_price()`, возвращающий стоимость: в Турцию — 80000, в Египет — 70000, в Италию — 120000. Напишите функцию `show_price(tour)`, которая принимает любой тур и выводит его стоимость. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `AgencyDirector` (директор агентства), в котором есть три поля: `name` — публичное (доступно всем), `_office` — защищённое (офис, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_office` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_AgencyDirector__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Турагентству нужна утилита для проверки номеров путёвок, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `TravelUtils` со статическим методом `is_valid_voucher(code)`, который возвращает `True`, если строка начинается с букв «TV» и её длина равна 8, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `TravelUtils.is_valid_voucher("TV123456")`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `TravelItem` с абстрактным методом `describe()`. Этот класс описывает общее понятие «объект турагентства», но не может существовать сам по себе — у каждого конкретного объекта должно быть своё описание. Унаследуйте от него классы `Tour` и `Client`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `TravelItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждого тура есть страховка — тур не может продаваться без страховки, и страховка создаётся вместе с туром. Создайте класс `Insurance` с методом `show()`, который выводит строку «Страховка оформлена». Создайте класс `Tour`, который содержит объект класса `Insurance` внутри себя как поле (композиция) — объект `Insurance` создаётся прямо в конструкторе тура. Метод `show_tour()` у тура должен выводить страну и вызывать метод `show()` у его страховки, демонстрируя, что тур «владеет» своей страховкой. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В турагентстве есть базовый класс `TravelItem` с методом `get_info()`, который возвращает строку «Объект турагентства». Унаследуйте от него класс `Tour` и переопределите метод `get_info()` так, чтобы он возвращал строку «Тур в <страна>». Создайте объект класса `Tour`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Менеджеру турагентства нужен метод для оформления заявок, который мог бы работать в разных ситуациях: иногда известен только тур, иногда тур и клиент, иногда ещё и дата выезда. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `BookingDesk` с методом `book(tour, client=None, date=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Оформлена заявка на тур: <country>», «Тур в <country> оформлен для <client>» или «Тур в <country> для <client> с выездом <date>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Туры часто выводят на экран — в каталогах, чеках, отчётах. По умолчанию `print(tour)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Tour` магический метод `__str__`, который возвращает строку вида «Тур "Турция" (10 дней, 80000 ₽)». Убедитесь, что при вызове `print(tour)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Tour('Турция', 10, 80000)` — оно используется при выводе списка туров и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В турагентстве может оказаться несколько записей об одном и том же туре — например, при повторном добавлении в каталог. Нужно, чтобы программа понимала, что это «один и тот же тур», и не дублировала его в списке. Реализуйте в классе `Tour` магический метод `__eq__`, чтобы два тура считались равными, если совпадают их страна и количество дней. Дополнительно реализуйте `__hash__`, чтобы туры можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых тура попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В турагентстве есть пакеты туров, и иногда нужно объединять их (например, при формировании сборного тура) или узнавать, сколько туров в пакете. Создайте класс `Package` (пакет), в котором хранится список туров. Реализуйте магический метод `__add__` для объединения двух пакетов в один (возвращается **новый** пакет со всеми турами) и метод `__len__`, возвращающий количество туров. Продемонстрируйте работу: создайте два пакета, объедините их через `p3 = p1 + p2` и выведите `len(p3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В турагентстве важно, чтобы цена тура была корректной — нельзя допустить отрицательную цену или ноль. Перепишите класс `Tour` так, чтобы поле `price` было реализовано через свойство `@property`. Проверки: цена должна быть числом (int или float) и должна быть строго больше нуля. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `tour.price` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `tour.price = -100` → ошибка, `tour.price = 0` → ошибка, `tour.price = "abc"` → ошибка, `tour.price = 80000` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В турагентство часто поступают данные о турах в виде строк — например, из файла или базы данных: `"Турция;10;80000"`. Добавьте в класс `Tour` классовый метод `from_string(cls, data)`, который создаёт объект тура из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание тура через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Турагентству нужно вести общий счётчик всех туров — сколько всего туров зарегистрировано в системе. Добавьте в класс `Tour` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового тура. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько туров и выведите `Tour.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`tour.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В турагентстве возможна ситуация, когда клиент пытается забронировать тур, которого нет в каталоге. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `TourNotFoundError(Exception)` и используйте его в классе `Agency` (турагентство). Метод `find_tour(country)` должен возвращать тур, если он есть, и выбрасывать `TourNotFoundError` с понятным сообщением (например, с указанием страны), если его нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В турагентстве клиент бронирует туры — но туры не принадлежат клиенту, они существуют отдельно и принадлежат агентству. Реализуйте класс `Client` (клиент), который содержит список забронированных туров. Туры при этом существуют отдельно от клиента — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого тура есть ссылка на `Agency`, в которой он продаётся. Продемонстрируйте, что при удалении клиента туры остаются в агентстве. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В турагентстве работает человек, который одновременно и клиент, и сотрудник — например, менеджер, который сам покупает туры со скидкой. Создайте два класса-роли: `Client` (клиент) и `Employee` (сотрудник). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `ClientManager`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `ClientManager.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Tour` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Tour` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Tour` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Tour(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В турагентстве может храниться огромное количество записей о турах — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Tour` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Tour` атрибут `__slots__ = ("country", "days", "price")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `tour.hotel = "..."` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Tour` для полей `price` (например, диапазон 1000–1000000) и `days` (диапазон 1–90), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В турагентстве должен быть только один реестр туров — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `Agency` будет только один экземпляр. Проверьте, что `Agency()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `TourRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Турагентство регулярно пополняется разными типами туров — пляжными, экскурсионными, круизами. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `TourFactory` с методом `create(kind, **kwargs)`, который по строке `"beach"`, `"excursion"` или `"cruise"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному туру каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Клиенты хотят получать уведомления, когда появляется горящий тур или начинается акция. Реализуйте паттерн «Наблюдатель»: класс `Agency` хранит список подписчиков (`Client`) и при появлении горящего тура уведомляет всех через метод `notify()`. Продемонстрируйте, что при появлении горящего тура все клиенты получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки клиент больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Agency` со временем разросся: он и хранит туры, и оформляет бронирования, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Agency` на несколько классов: `TourStorage` (хранение туров и поиск по стране), `BookingService` (оформление бронирований, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «клиент забронировал тур». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** VIP-туры (`VipTour`) похожи на обычные туры, но имеют дополнительные поля — уровень отеля, персональный гид. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`VipTour(Tour)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`VipTour` содержит объект `Tour` внутри себя как поле `self.tour`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В турагентстве нужно создавать копии карточек туров — например, для архивного каталога. При этом у тура есть вложенный объект `Insurance`, и важно понимать, копируется ли он вместе с туром или остаётся общим. Реализуйте в классе `Tour` методы `__copy__` и `__deepcopy__`. Создайте тур с вложенным объектом `Insurance` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — страховка общая (`shallow.insurance is tour.insurance` → `True`), при глубоком — своя (`deep.insurance is tour.insurance` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В турагентстве есть разные объекты, у которых есть метод `describe()` — туры, клиенты, менеджеры, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Tour`, но и с любым другим классом, у которого есть `describe()` (например, классом `Client`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В турагентстве есть пакет туров `Package`, и хочется перебирать туры в нём в цикле `for` — так же удобно, как список. Сейчас `for tour in package:` не работает. Реализуйте в классе `Package` магические методы `__iter__` и `__next__` так, чтобы по пакету можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте пакет с несколькими турами и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(package)` должен возвращать список всех туров.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В пакете туры должны быть доступны по индексу — как элементы списка: `package[0]` возвращает первый тур, `package[1] = tour` заменяет второй. Реализуйте в классе `Package` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `package[0]` — получение тура по индексу, `package[1] = new_tour` — замена тура, срезы `package[0:2]` — получение подсписка, `for tour in package` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `tour in package`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В турагентстве есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных туров её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `TourDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with TourDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В турагентстве есть разные способы расчёта скидки: фиксированная (5000 ₽), процентная (10% от стоимости) и «раннее бронирование» (15% от стоимости). Реализуйте паттерн «Стратегия»: базовый класс `DiscountStrategy` с методом `calculate(price)`, и три наследника — `FixedDiscount`, `PercentDiscount`, `EarlyBirdDiscount`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `EarlyBirdDiscount`, чтобы вызвать реализацию `PercentDiscount` и увеличить процент скидки (или переиспользовать код родителя). Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В турагентстве появился новый тип тура — горнолыжный (`SkiTour`). Если добавлять его через изменение базового класса `TravelItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `TravelItem` с методом `tour_price()`; наследников `TourToTurkey`, `TourToEgypt`; функцию `process_item(item: TravelItem)`, которая работает с любым туром через родительский интерфейс. Добавьте `SkiTour` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В турагентстве есть служебный объект `PriceCalculator`, который должен уметь рассчитывать итоговую стоимость тура с учётом сборов. Удобно, чтобы его можно было вызывать как функцию: `calculator(price)`, а не `calculator.calculate(price)`. Реализуйте в классе `PriceCalculator` магический метод `__call__(self, price)`, который возвращает итоговую стоимость (например, `price * 1.1` — с сервисным сбором 10%). Продемонстрируйте, что объект можно вызывать как функцию: `calc = PriceCalculator(); calc(80000)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В турагентстве есть класс `Tour`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Tour` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования туров по стране и количеству дней): два вызова `Tour("Турция", 10, 80000)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В турагентстве есть базовый класс `TravelItem` с методом `get_info()`. Нужно добавлять к туру дополнительную информацию — например, «всё включено», «горящий», «рекомендуем». При этом нельзя менять сам класс `TravelItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `TravelItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `AllInclusiveDecorator` и `HotTourDecorator`, которые можно комбинировать: `AllInclusiveDecorator(HotTourDecorator(tour))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В турагентстве появилась внешняя система бронирования отелей, которая возвращает данные о турах в другом формате — например, в виде словаря `{"destination": "Турция", "length": 10, "cost": 80000}`. А весь остальной код турагентства работает с классом `Tour`, у которого поля `country`, `days`, `price`. Реализуйте паттерн «Адаптер»: класс `TourAdapter`, который принимает словарь и предоставляет интерфейс `Tour` — свойства `country`, `days`, `price` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Tour` — например, передать его в функцию, которая работает с туром. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В турагентстве есть базовый класс `TravelItem`, и постепенно в нём накопились методы, которые нужны не всем объектам: `book()` (только для туров), `insure()` (только для страхуемых туров), `cancel()` (только для отменяемых бронирований). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Bookable`, `Insurable`, `Cancelable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `BookingService`, который зависит не от конкретных классов (`TourToTurkey`, `TourToEgypt`), а от абстракции `Bookable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип тура (например, `SkiTour`, реализующий только `Bookable`) без изменения `BookingService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** На автомойке ведётся учёт автомобилей. Создайте класс `Car` с полями `brand` (марка), `plate` (номер) и `type` (тип кузова: седан, внедорожник, микроавтобус). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией об автомобиле в формате: `Марка — номер ..., тип ...`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Toyota — номер А123БВ, тип седан`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждой мойки есть внутренний код заказа, который нельзя менять или читать напрямую. Создайте класс `WashOrder`, в котором код заказа хранится в приватном поле `__order_code`. Реализуйте метод `set_order_code(value)`, предназначенный только для установки кода при оформлении заказа, и метод `get_order_code()` для чтения значения. Повторная попытка изменить код после оформления должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте заказ, установите код, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `order.__order_code` — будет ошибка `AttributeError`, но через name mangling (`order._WashOrder__order_code`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** На автомойке есть разные типы обслуживания. Создайте базовый класс `WashService` с методом `get_service_type()`, возвращающим строку «Услуга». Унаследуйте от него классы `ExpressWash` и `DeepWash`, переопределив в каждом метод `get_service_type()` так, чтобы он возвращал «Экспресс-мойка» и «Глубокая мойка» соответственно. Создайте по одному объекту каждого класса и выведите их тип, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Автомойка рассчитывает стоимость для разных типов кузова. Создайте классы `Sedan`, `SUV` и `Minivan`, у каждого из которых есть метод `wash_price()`, возвращающий стоимость: для седана — 500, для внедорожника — 800, для микроавтобуса — 1000. Напишите функцию `show_price(car)`, которая принимает любой автомобиль и выводит стоимость мойки. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `WashManager` (управляющий автомойки), в котором есть три поля: `name` — публичное (доступно всем), `_shift` — защищённое (смена, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_shift` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_WashManager__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Автомойке нужна утилита для проверки длины очереди, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `WashUtils` со статическим методом `is_queue_free(count)`, который возвращает `True`, если количество машин в очереди меньше 3, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `WashUtils.is_queue_free(2)`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `WashItem` с абстрактным методом `describe()`. Этот класс описывает общее понятие «объект автомойки», но не может существовать сам по себе — у каждого конкретного объекта должно быть своё описание. Унаследуйте от него классы `Car` и `WashService`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `WashItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждой услуги мойки есть набор моющих средств — услуга не может быть оказана без них, и средства создаются вместе с услугой. Создайте класс `Detergent` с методом `show()`, который выводит строку «Автошампунь premium». Создайте класс `WashService`, который содержит объект класса `Detergent` внутри себя как поле (композиция) — объект `Detergent` создаётся прямо в конструкторе услуги. Метод `show_service()` у услуги должен выводить её тип и вызывать метод `show()` у моющего средства, демонстрируя, что услуга «владеет» своим средством. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** На автомойке есть базовый класс `WashItem` с методом `get_info()`, который возвращает строку «Объект автомойки». Унаследуйте от него класс `Car` и переопределите метод `get_info()` так, чтобы он возвращал строку «Автомобиль: <марка> <номер>». Создайте объект класса `Car`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Оператору автомойки нужен метод для оформления заказов, который мог бы работать в разных ситуациях: иногда известен только автомобиль, иногда автомобиль и тип мойки, иногда ещё и время. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `WashDesk` с методом `accept(car, service=None, time=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Принят автомобиль: <brand> <plate>», «Автомобиль <brand> <plate> принят на услугу: <service>» или «Автомобиль <brand> <plate> принят на <service> в <time>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Автомобили часто выводят на экран — в очереди, чеках, отчётах. По умолчанию `print(car)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Car` магический метод `__str__`, который возвращает строку вида «Автомобиль "Toyota А123БВ" (седан)». Убедитесь, что при вызове `print(car)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Car('Toyota', 'А123БВ', 'седан')` — оно используется при выводе списка автомобилей и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** На автомойке может оказаться несколько записей об одном и том же автомобиле — например, при повторном заезде. Нужно, чтобы программа понимала, что это «один и тот же автомобиль», и не дублировала его в очереди. Реализуйте в классе `Car` магический метод `__eq__`, чтобы два автомобиля считались равными, если совпадают их марка и номер. Дополнительно реализуйте `__hash__`, чтобы автомобили можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых автомобиля попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** На автомойке есть очередь, и иногда нужно объединять очереди (например, при слиянии постов) или узнавать, сколько машин в очереди. Создайте класс `Queue` (очередь), в котором хранится список автомобилей. Реализуйте магический метод `__add__` для объединения двух очередей в одну (возвращается **новая** очередь со всеми автомобилями) и метод `__len__`, возвращающий количество автомобилей. Продемонстрируйте работу: создайте две очереди, объедините их через `q3 = q1 + q2` и выведите `len(q3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** На автомойке важно, чтобы количество машин в очереди было корректным — нельзя допустить отрицательное значение. Перепишите класс `Queue` так, чтобы поле `count` было реализовано через свойство `@property`. Проверки: количество машин должно быть целым числом и не может быть отрицательным. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `queue.count` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `queue.count = -1` → ошибка, `queue.count = "abc"` → ошибка, `queue.count = 5` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** На автомойку часто поступают данные об автомобилях в виде строк — например, из файла или базы данных: `"Toyota;А123БВ;седан"`. Добавьте в класс `Car` классовый метод `from_string(cls, data)`, который создаёт объект автомобиля из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание автомобиля через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Автомойке нужно вести общий счётчик всех обслуженных автомобилей — сколько всего машин зарегистрировано в системе. Добавьте в класс `Car` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового автомобиля. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько автомобилей и выведите `Car.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`car.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** На автомойке возможна ситуация, когда оператор пытается найти автомобиль, которого нет в очереди. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `CarNotFoundError(Exception)` и используйте его в классе `CarWash` (автомойка). Метод `find_car(plate)` должен возвращать автомобиль, если он есть, и выбрасывать `CarNotFoundError` с понятным сообщением (например, с указанием номера), если его нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** На автомойке мойщик обслуживает автомобили — но автомобили не принадлежат мойщику, они существуют отдельно и принадлежат клиентам. Реализуйте класс `Washer` (мойщик), который содержит список автомобилей, которые он обслуживает. Автомобили при этом существуют отдельно от мойщика — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого автомобиля есть ссылка на `CarWash`, в котором он обслуживается. Продемонстрируйте, что при удалении мойщика автомобили остаются на мойке. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** На автомойке работает человек, который одновременно и мойщик, и оператор — например, универсальный сотрудник. Создайте два класса-роли: `Washer` (мойщик) и `Operator` (оператор). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `WasherOperator`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `WasherOperator.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Car` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Car` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Car` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Car(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** На автомойке может храниться огромное количество записей об автомобилях — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Car` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Car` атрибут `__slots__ = ("brand", "plate", "type")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `car.color = "красный"` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `WashOrder` для полей `price` (например, диапазон 100–10000) и `duration` (диапазон 5–180), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** На автомойке должен быть только один реестр заказов — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `CarWash` будет только один экземпляр. Проверьте, что `CarWash()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `OrderRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Автомойка регулярно пополняется разными типами услуг — экспресс-мойкой, глубокой мойкой, химчисткой салона. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `ServiceFactory` с методом `create(kind, **kwargs)`, который по строке `"express"`, `"deep"` или `"interior"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одной услуге каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Клиенты хотят получать уведомления, когда их автомобиль готов или когда освобождается пост. Реализуйте паттерн «Наблюдатель»: класс `CarWash` хранит список подписчиков (`Client`) и при изменении статуса заказа уведомляет всех через метод `notify()`. Продемонстрируйте, что при изменении статуса все клиенты получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки клиент больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `CarWash` со временем разросся: он и хранит автомобили, и оформляет заказы, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `CarWash` на несколько классов: `CarStorage` (хранение автомобилей и поиск по номеру), `OrderService` (оформление заказов, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «клиент оформил заказ». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Мойка с полировкой (`PolishWash`) похожа на обычную мойку, но имеет дополнительные поля — тип полироли, время обработки. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`PolishWash(WashService)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`PolishWash` содержит объект `WashService` внутри себя как поле `self.service`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** На автомойке нужно создавать копии карточек заказов — например, для архивного хранения. При этом у услуги есть вложенный объект `Detergent`, и важно понимать, копируется ли он вместе с услугой или остаётся общим. Реализуйте в классе `WashService` методы `__copy__` и `__deepcopy__`. Создайте услугу с вложенным объектом `Detergent` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — средство общее (`shallow.detergent is service.detergent` → `True`), при глубоком — своё (`deep.detergent is service.detergent` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** На автомойке есть разные объекты, у которых есть метод `describe()` — автомобили, услуги, мойщики, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Car`, но и с любым другим классом, у которого есть `describe()` (например, классом `Washer`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** На автомойке есть очередь `Queue`, и хочется перебирать автомобили в ней в цикле `for` — так же удобно, как список. Сейчас `for car in queue:` не работает. Реализуйте в классе `Queue` магические методы `__iter__` и `__next__` так, чтобы по очереди можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте очередь с несколькими автомобилями и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(queue)` должен возвращать список всех автомобилей.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В очереди автомобили должны быть доступны по индексу — как элементы списка: `queue[0]` возвращает первый автомобиль, `queue[1] = car` заменяет второй. Реализуйте в классе `Queue` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `queue[0]` — получение автомобиля по индексу, `queue[1] = new_car` — замена автомобиля, срезы `queue[0:2]` — получение подсписка, `for car in queue` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `car in queue`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** На автомойке есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных заказов её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `OrderDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with OrderDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** На автомойке есть разные способы расчёта скидки: фиксированная (100 ₽), процентная (10% от стоимости) и «для постоянных клиентов» (15% от стоимости). Реализуйте паттерн «Стратегия»: базовый класс `DiscountStrategy` с методом `calculate(price)`, и три наследника — `FixedDiscount`, `PercentDiscount`, `LoyaltyDiscount`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `LoyaltyDiscount`, чтобы вызвать реализацию `PercentDiscount` и увеличить процент скидки (или переиспользовать код родителя). Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** На автомойке появился новый тип услуги — химчистка салона (`InteriorCleaning`). Если добавлять её через изменение базового класса `WashItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `WashItem` с методом `wash_price()`; наследников `ExpressWash`, `DeepWash`; функцию `process_item(item: WashItem)`, которая работает с любой услугой через родительский интерфейс. Добавьте `InteriorCleaning` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** На автомойке есть служебный объект `PriceCalculator`, который должен уметь рассчитывать итоговую стоимость мойки с учётом налога. Удобно, чтобы его можно было вызывать как функцию: `calculator(base_price)`, а не `calculator.calculate(base_price)`. Реализуйте в классе `PriceCalculator` магический метод `__call__(self, base_price)`, который возвращает итоговую стоимость (например, `base_price * 1.1`). Продемонстрируйте, что объект можно вызывать как функцию: `calc = PriceCalculator(); calc(500)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** На автомойке есть класс `Car`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Car` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования автомобилей по номеру): два вызова `Car("Toyota", "А123БВ", "седан")` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** На автомойке есть базовый класс `WashItem` с методом `get_info()`. Нужно добавлять к услуге дополнительную информацию — например, «с сушкой», «с полировкой», «эко-средства». При этом нельзя менять сам класс `WashItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `WashItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `DryingDecorator` и `EcoDecorator`, которые можно комбинировать: `DryingDecorator(EcoDecorator(wash))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** На автомойке появилась внешняя система учёта, которая возвращает данные об автомобилях в другом формате — например, в виде словаря `{"make": "Toyota", "number": "А123БВ", "body": "седан"}`. А весь остальной код автомойки работает с классом `Car`, у которого поля `brand`, `plate`, `type`. Реализуйте паттерн «Адаптер»: класс `CarAdapter`, который принимает словарь и предоставляет интерфейс `Car` — свойства `brand`, `plate`, `type` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Car` — например, передать его в функцию, которая работает с автомобилем. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** На автомойке есть базовый класс `WashItem`, и постепенно в нём накопились методы, которые нужны не всем объектам: `wash()` (только для услуг мойки), `dry()` (только для услуг с сушкой), `polish()` (только для полировки). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Washable`, `Dryable`, `Polishable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `OrderService`, который зависит не от конкретных классов (`ExpressWash`, `DeepWash`), а от абстракции `Washable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип услуги (например, `InteriorCleaning`, реализующий только `Washable`) без изменения `OrderService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В химчистке ведётся учёт вещей. Создайте класс `ClothingItem` с полями `name` (название вещи), `material` (материал) и `price` (стоимость чистки). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о вещи в формате: `Название — материал, стоимость ...`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Пальто — шерсть, стоимость 1200 ₽`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждой вещи есть номер квитанции, который нельзя менять или читать напрямую. Создайте класс `Receipt`, в котором номер квитанции хранится в приватном поле `__receipt_id`. Реализуйте метод `set_receipt_id(value)`, предназначенный только для установки номера при приёме вещи, и метод `get_receipt_id()` для чтения значения. Повторная попытка изменить номер после приёма должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте квитанцию, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `receipt.__receipt_id` — будет ошибка `AttributeError`, но через name mangling (`receipt._Receipt__receipt_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В химчистке обрабатываются разные типы вещей. Создайте базовый класс `LaundryItem` с методом `get_item_type()`, возвращающим строку «Вещь». Унаследуйте от него классы `Coat` и `Dress`, переопределив в каждом метод `get_item_type()` так, чтобы он возвращал «Пальто» и «Платье» соответственно. Создайте по одному объекту каждого класса и выведите их тип, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Химчистка рассчитывает время обработки для разных типов вещей. Создайте классы `Coat`, `Dress` и `Curtains`, у каждого из которых есть метод `processing_time()`, возвращающий время в часах: для пальто — 24, для платья — 12, для штор — 48. Напишите функцию `show_time(item)`, которая принимает любую вещь и выводит время её обработки. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `Cleaner` (сотрудник химчистки), в котором есть три поля: `name` — публичное (доступно всем), `_workshop` — защищённое (цех, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_workshop` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_Cleaner__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Химчистке нужна утилита для проверки допустимого веса вещи, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `LaundryUtils` со статическим методом `is_valid_weight(kg)`, который возвращает `True`, если вес не превышает 10 кг, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `LaundryUtils.is_valid_weight(5)`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `LaundryProduct` с абстрактным методом `describe()`. Этот класс описывает общее понятие «продукт химчистки», но не может существовать сам по себе — у каждого конкретного объекта должно быть своё описание. Унаследуйте от него классы `ClothingItem` и `Service`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `LaundryProduct` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждой вещи есть защитный чехол — вещь не может быть сдана в обработку без него, и чехол создаётся вместе с вещью. Создайте класс `Cover` с методом `show()`, который выводит строку «Защитный чехол надет». Создайте класс `ClothingItem`, который содержит объект класса `Cover` внутри себя как поле (композиция) — объект `Cover` создаётся прямо в конструкторе вещи. Метод `show_item()` у вещи должен выводить её название и вызывать метод `show()` у её чехла, демонстрируя, что вещь «владеет» своим чехлом. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В химчистке есть базовый класс `LaundryProduct` с методом `get_info()`, который возвращает строку «Продукт химчистки». Унаследуйте от него класс `ClothingItem` и переопределите метод `get_info()` так, чтобы он возвращал строку «Вещь: <название>». Создайте объект класса `ClothingItem`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Приёмщику химчистки нужен метод для оформления заказов, который мог бы работать в разных ситуациях: иногда известна только вещь, иногда вещь и количество, иногда ещё и клиент. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `ReceptionDesk` с методом `accept(item, count=None, client=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Принята вещь: <name>», «Принято <count> вещей: <name>» или «Принято <count> вещей <name> от клиента <client>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Вещи часто выводят на экран — в квитанциях, отчётах, списках. По умолчанию `print(item)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `ClothingItem` магический метод `__str__`, который возвращает строку вида «Вещь "Пальто" (шерсть, 1200 ₽)». Убедитесь, что при вызове `print(item)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `ClothingItem('Пальто', 'шерсть', 1200)` — оно используется при выводе списка вещей и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В химчистке может оказаться несколько записей об одной и той же вещи — например, при повторной сдаче. Нужно, чтобы программа понимала, что это «одна и та же вещь», и не дублировала её в списке. Реализуйте в классе `ClothingItem` магический метод `__eq__`, чтобы две вещи считались равными, если совпадают их название и материал. Дополнительно реализуйте `__hash__`, чтобы вещи можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что две одинаковые вещи попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В химчистке есть партии вещей, и иногда нужно объединять их (например, при объединении заказов) или узнавать, сколько вещей в партии. Создайте класс `Batch` (партия), в котором хранится список вещей. Реализуйте магический метод `__add__` для объединения двух партий в одну (возвращается **новая** партия со всеми вещами) и метод `__len__`, возвращающий количество вещей. Продемонстрируйте работу: создайте две партии, объедините их через `b3 = b1 + b2` и выведите `len(b3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В химчистке важно, чтобы стоимость чистки была корректной — нельзя допустить отрицательную стоимость или ноль. Перепишите класс `ClothingItem` так, чтобы поле `price` было реализовано через свойство `@property`. Проверки: цена должна быть числом (int или float) и должна быть строго больше нуля. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `item.price` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `item.price = -100` → ошибка, `item.price = 0` → ошибка, `item.price = "abc"` → ошибка, `item.price = 1200` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В химчистку часто поступают данные о вещах в виде строк — например, из файла или базы данных: `"Пальто;шерсть;1200"`. Добавьте в класс `ClothingItem` классовый метод `from_string(cls, data)`, который создаёт объект вещи из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание вещи через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Химчистке нужно вести общий счётчик всех вещей — сколько всего вещей зарегистрировано в системе. Добавьте в класс `ClothingItem` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждой новой вещи. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько вещей и выведите `ClothingItem.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`item.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В химчистке возможна ситуация, когда клиент пытается получить вещь по квитанции, которой нет в базе. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `ReceiptNotFoundError(Exception)` и используйте его в классе `DryCleaner` (химчистка). Метод `find_by_receipt(receipt_id)` должен возвращать вещь, если она есть, и выбрасывать `ReceiptNotFoundError` с понятным сообщением (например, с указанием номера квитанции), если её нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В химчистке клиент сдаёт вещи — но вещи не принадлежат приёмщику, они существуют отдельно и принадлежат клиенту. Реализуйте класс `Client` (клиент), который содержит список сданных вещей. Вещи при этом существуют отдельно от приёмщика — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждой вещи есть ссылка на `DryCleaner`, в которой она обрабатывается. Продемонстрируйте, что при удалении клиента вещи остаются в химчистке. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В химчистке работает человек, который одновременно и приёмщик, и чистильщик — например, универсальный сотрудник. Создайте два класса-роли: `Receptionist` (приёмщик) и `Cleaner` (чистильщик). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `UniversalWorker`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `UniversalWorker.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `ClothingItem` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `ClothingItem` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `ClothingItem` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class ClothingItem(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В химчистке может храниться огромное количество записей о вещах — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `ClothingItem` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `ClothingItem` атрибут `__slots__ = ("name", "material", "price")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `item.color = "чёрный"` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `ClothingItem` для полей `price` (например, диапазон 100–100000) и `weight` (диапазон 1–10), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В химчистке должен быть только один реестр заказов — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `DryCleaner` будет только один экземпляр. Проверьте, что `DryCleaner()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `OrderRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Химчистка регулярно пополняется разными типами вещей — пальто, платьями, шторами. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `ItemFactory` с методом `create(kind, **kwargs)`, который по строке `"coat"`, `"dress"` или `"curtains"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одной вещи каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Клиенты хотят получать уведомления, когда их вещь готова к выдаче. Реализуйте паттерн «Наблюдатель»: класс `DryCleaner` хранит список подписчиков (`Client`) и при изменении статуса заказа уведомляет всех через метод `notify()`. Продемонстрируйте, что при изменении статуса все клиенты получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки клиент больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `DryCleaner` со временем разросся: он и хранит вещи, и принимает заказы, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `DryCleaner` на несколько классов: `ItemStorage` (хранение вещей и поиск по квитанции), `OrderService` (приём заказов, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «клиент сдал вещь». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Деликатная чистка (`DelicateWash`) похожа на обычную чистку, но имеет дополнительные поля — тип растворителя, температура обработки. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`DelicateWash(Service)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`DelicateWash` содержит объект `Service` внутри себя как поле `self.service`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В химчистке нужно создавать копии карточек вещей — например, для архивного хранения. При этом у вещи есть вложенный объект `Cover`, и важно понимать, копируется ли он вместе с вещью или остаётся общим. Реализуйте в классе `ClothingItem` методы `__copy__` и `__deepcopy__`. Создайте вещь с вложенным объектом `Cover` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — чехол общий (`shallow.cover is item.cover` → `True`), при глубоком — свой (`deep.cover is item.cover` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В химчистке есть разные объекты, у которых есть метод `describe()` — вещи, услуги, сотрудники, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `ClothingItem`, но и с любым другим классом, у которого есть `describe()` (например, классом `Cleaner`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В химчистке есть партия вещей `Batch`, и хочется перебирать вещи в ней в цикле `for` — так же удобно, как список. Сейчас `for item in batch:` не работает. Реализуйте в классе `Batch` магические методы `__iter__` и `__next__` так, чтобы по партии можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте партию с несколькими вещами и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(batch)` должен возвращать список всех вещей.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В партии вещи должны быть доступны по индексу — как элементы списка: `batch[0]` возвращает первую вещь, `batch[1] = item` заменяет вторую. Реализуйте в классе `Batch` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `batch[0]` — получение вещи по индексу, `batch[1] = new_item` — замена вещи, срезы `batch[0:2]` — получение подсписка, `for item in batch` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `item in batch`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В химчистке есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных заказов её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `OrderDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with OrderDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В химчистке есть разные способы расчёта скидки: фиксированная (200 ₽), процентная (10% от стоимости) и «для постоянных клиентов» (15% от стоимости). Реализуйте паттерн «Стратегия»: базовый класс `DiscountStrategy` с методом `calculate(price)`, и три наследника — `FixedDiscount`, `PercentDiscount`, `LoyaltyDiscount`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `LoyaltyDiscount`, чтобы вызвать реализацию `PercentDiscount` и увеличить процент скидки (или переиспользовать код родителя). Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В химчистке появился новый тип вещи — ковёр (`Carpet`). Если добавлять его через изменение базового класса `LaundryProduct`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `LaundryProduct` с методом `processing_time()`; наследников `Coat`, `Dress`; функцию `process_item(item: LaundryProduct)`, которая работает с любой вещью через родительский интерфейс. Добавьте `Carpet` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В химчистке есть служебный объект `PriceCalculator`, который должен уметь рассчитывать итоговую стоимость чистки с учётом наценки за срочность. Удобно, чтобы его можно было вызывать как функцию: `calculator(base_price)`, а не `calculator.calculate(base_price)`. Реализуйте в классе `PriceCalculator` магический метод `__call__(self, base_price)`, который возвращает итоговую стоимость (например, `base_price * 1.5`). Продемонстрируйте, что объект можно вызывать как функцию: `calc = PriceCalculator(); calc(1200)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В химчистке есть класс `ClothingItem`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `ClothingItem` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования вещей по названию и материалу): два вызова `ClothingItem("Пальто", "шерсть", 1200)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В химчистке есть базовый класс `LaundryProduct` с методом `get_info()`. Нужно добавлять к вещи дополнительную информацию — например, «срочно», «деликатно», «с пятновыведением». При этом нельзя менять сам класс `LaundryProduct`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `LaundryProduct` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `UrgentDecorator` и `DelicateDecorator`, которые можно комбинировать: `UrgentDecorator(DelicateDecorator(item))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В химчистке появилась внешняя система учёта, которая возвращает данные о вещах в другом формате — например, в виде словаря `{"title": "Пальто", "fabric": "шерсть", "cost": 1200}`. А весь остальной код химчистки работает с классом `ClothingItem`, у которого поля `name`, `material`, `price`. Реализуйте паттерн «Адаптер»: класс `ClothingItemAdapter`, который принимает словарь и предоставляет интерфейс `ClothingItem` — свойства `name`, `material`, `price` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `ClothingItem` — например, передать его в функцию, которая работает с вещью. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В химчистке есть базовый класс `LaundryProduct`, и постепенно в нём накопились методы, которые нужны не всем объектам: `wash()` (только для стираемых вещей), `dry_clean()` (только для сухой чистки), `iron()` (только для глажки). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Washable`, `DryCleanable`, `Ironable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `OrderService`, который зависит не от конкретных классов (`Coat`, `Dress`), а от абстракции `Washable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип вещи (например, `Carpet`, реализующий только `Washable`) без изменения `OrderService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В прокате автомобилей ведётся учёт машин. Создайте класс `RentalCar` с полями `brand` (марка), `model` (модель) и `price_per_day` (цена за сутки). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией об автомобиле в формате: `Марка — модель, цена за сутки ...`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Toyota — Camry, цена за сутки 4000 ₽`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого клиента проката есть номер водительского удостоверения, который нельзя менять или читать напрямую. Создайте класс `DriverLicense`, в котором номер удостоверения хранится в приватном поле `__license_id`. Реализуйте метод `set_license_id(value)`, предназначенный только для установки номера при оформлении аренды, и метод `get_license_id()` для чтения значения. Повторная попытка изменить номер после оформления должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте удостоверение, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `license.__license_id` — будет ошибка `AttributeError`, но через name mangling (`license._DriverLicense__license_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В прокате есть разные классы автомобилей. Создайте базовый класс `RentalVehicle` с методом `get_class()`, возвращающим строку «Автомобиль». Унаследуйте от него классы `EconomyCar` и `PremiumCar`, переопределив в каждом метод `get_class()` так, чтобы он возвращал «Эконом-класс» и «Премиум-класс» соответственно. Создайте по одному объекту каждого класса и выведите их класс, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Прокат рассчитывает стоимость аренды для разных классов автомобилей. Создайте классы `EconomyCar`, `ComfortCar` и `PremiumCar`, у каждого из которых есть метод `rental_price()`, возвращающий цену за сутки: для эконома — 2000, для комфорта — 4000, для премиума — 9000. Напишите функцию `show_price(car)`, которая принимает любой автомобиль и выводит цену аренды. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `RentalManager` (менеджер проката), в котором есть три поля: `name` — публичное (доступно всем), `_branch` — защищённое (филиал, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_branch` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_RentalManager__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Прокату нужна утилита для проверки возраста клиента, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `RentalUtils` со статическим методом `is_valid_age(age)`, который возвращает `True`, если клиенту не меньше 21 года, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `RentalUtils.is_valid_age(25)`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `RentalItem` с абстрактным методом `describe()`. Этот класс описывает общее понятие «объект проката», но не может существовать сам по себе — у каждого конкретного объекта должно быть своё описание. Унаследуйте от него классы `RentalCar` и `Client`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `RentalItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждого автомобиля в прокате есть GPS-навигатор — автомобиль не может быть сдан в аренду без него, и навигатор создаётся вместе с автомобилем. Создайте класс `GPS` с методом `show()`, который выводит строку «GPS-навигатор включён». Создайте класс `RentalCar`, который содержит объект класса `GPS` внутри себя как поле (композиция) — объект `GPS` создаётся прямо в конструкторе автомобиля. Метод `show_car()` у автомобиля должен выводить модель и вызывать метод `show()` у его навигатора, демонстрируя, что автомобиль «владеет» своим навигатором. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В прокате есть базовый класс `RentalItem` с методом `get_info()`, который возвращает строку «Объект проката». Унаследуйте от него класс `RentalCar` и переопределите метод `get_info()` так, чтобы он возвращал строку «Автомобиль: <марка> <модель>». Создайте объект класса `RentalCar`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Менеджеру проката нужен метод для оформления аренды, который мог бы работать в разных ситуациях: иногда известен только автомобиль, иногда автомобиль и клиент, иногда ещё и срок аренды. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `RentalDesk` с методом `rent(car, client=None, days=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Оформлена аренда автомобиля: <brand> <model>», «Автомобиль <brand> <model> арендован клиентом <client>» или «Автомобиль <brand> <model> арендован <client> на <days> дней». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Автомобили часто выводят на экран — в каталогах, договорах, отчётах. По умолчанию `print(car)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `RentalCar` магический метод `__str__`, который возвращает строку вида «Автомобиль "Toyota Camry" (4000 ₽/сутки)». Убедитесь, что при вызове `print(car)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `RentalCar('Toyota', 'Camry', 4000)` — оно используется при выводе списка автомобилей и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В прокате может оказаться несколько записей об одном и том же автомобиле — например, при повторной регистрации. Нужно, чтобы программа понимала, что это «один и тот же автомобиль», и не дублировала его в списке. Реализуйте в классе `RentalCar` магический метод `__eq__`, чтобы два автомобиля считались равными, если совпадают их марка и модель. Дополнительно реализуйте `__hash__`, чтобы автомобили можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых автомобиля попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В прокате есть автопарки, и иногда нужно объединять их (например, при слиянии филиалов) или узнавать, сколько машин в автопарке. Создайте класс `Fleet` (автопарк), в котором хранится список автомобилей. Реализуйте магический метод `__add__` для объединения двух автопарков в один (возвращается **новый** автопарк со всеми автомобилями) и метод `__len__`, возвращающий количество автомобилей. Продемонстрируйте работу: создайте два автопарка, объедините их через `f3 = f1 + f2` и выведите `len(f3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В прокате важно, чтобы цена за сутки была корректной — нельзя допустить отрицательную цену или ноль. Перепишите класс `RentalCar` так, чтобы поле `price_per_day` было реализовано через свойство `@property`. Проверки: цена должна быть числом (int или float) и должна быть строго больше нуля. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `car.price_per_day` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `car.price_per_day = -100` → ошибка, `car.price_per_day = 0` → ошибка, `car.price_per_day = "abc"` → ошибка, `car.price_per_day = 4000` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В прокат часто поступают данные об автомобилях в виде строк — например, из файла или базы данных: `"Toyota;Camry;4000"`. Добавьте в класс `RentalCar` классовый метод `from_string(cls, data)`, который создаёт объект автомобиля из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание автомобиля через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Прокату нужно вести общий счётчик всех автомобилей — сколько всего машин зарегистрировано в системе. Добавьте в класс `RentalCar` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового автомобиля. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько автомобилей и выведите `RentalCar.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`car.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В прокате возможна ситуация, когда клиент пытается арендовать автомобиль, которого нет в автопарке. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `CarNotAvailableError(Exception)` и используйте его в классе `RentalAgency` (прокат). Метод `rent_car(client, car_model)` должен выдавать автомобиль, если он есть и свободен, и выбрасывать `CarNotAvailableError` с понятным сообщением (например, с указанием модели), если автомобиля нет или он занят. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В прокате клиент арендует автомобили — но автомобили не принадлежат клиенту, они существуют отдельно и принадлежат прокату. Реализуйте класс `Client` (клиент), который содержит список арендованных автомобилей. Автомобили при этом существуют отдельно от клиента — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого автомобиля есть ссылка на `RentalAgency`, в которой он числится. Продемонстрируйте, что при удалении клиента автомобили остаются в прокате. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В прокате работает человек, который одновременно и менеджер, и клиент — например, сотрудник, который сам арендует машину со скидкой. Создайте два класса-роли: `Client` (клиент) и `Employee` (сотрудник). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `ClientEmployee`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `ClientEmployee.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `RentalCar` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `RentalCar` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `RentalCar` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class RentalCar(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В прокате может храниться огромное количество записей об автомобилях — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `RentalCar` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `RentalCar` атрибут `__slots__ = ("brand", "model", "price_per_day")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `car.color = "красный"` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `RentalCar` для полей `price_per_day` (например, диапазон 500–50000) и `year` (диапазон 1990–2100), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В прокате должен быть только один реестр автомобилей — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `RentalAgency` будет только один экземпляр. Проверьте, что `RentalAgency()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `CarRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Прокат регулярно пополняется разными классами автомобилей — эконом, комфорт, премиум. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `CarFactory` с методом `create(kind, **kwargs)`, который по строке `"economy"`, `"comfort"` или `"premium"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному автомобилю каждого класса через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Клиенты хотят получать уведомления, когда освобождается нужный автомобиль или начинается акция. Реализуйте паттерн «Наблюдатель»: класс `RentalAgency` хранит список подписчиков (`Client`) и при возврате автомобиля уведомляет всех через метод `notify()`. Продемонстрируйте, что при возврате автомобиля все подписчики получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки клиент больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `RentalAgency` со временем разросся: он и хранит автомобили, и оформляет аренду, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `RentalAgency` на несколько классов: `CarStorage` (хранение автомобилей и поиск по модели), `RentalService` (оформление аренды, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «клиент арендовал машину». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Электромобили в прокате (`ElectricRentalCar`) похожи на обычные автомобили, но имеют дополнительные поля — ёмкость батареи, запас хода. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`ElectricRentalCar(RentalCar)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`ElectricRentalCar` содержит объект `RentalCar` внутри себя как поле `self.car`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В прокате нужно создавать копии карточек автомобилей — например, для архивного каталога. При этом у автомобиля есть вложенный объект `GPS`, и важно понимать, копируется ли он вместе с автомобилем или остаётся общим. Реализуйте в классе `RentalCar` методы `__copy__` и `__deepcopy__`. Создайте автомобиль с вложенным объектом `GPS` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — навигатор общий (`shallow.gps is car.gps` → `True`), при глубоком — свой (`deep.gps is car.gps` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В прокате есть разные объекты, у которых есть метод `describe()` — автомобили, клиенты, менеджеры, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `RentalCar`, но и с любым другим классом, у которого есть `describe()` (например, классом `Client`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В прокате есть автопарк `Fleet`, и хочется перебирать автомобили в нём в цикле `for` — так же удобно, как список. Сейчас `for car in fleet:` не работает. Реализуйте в классе `Fleet` магические методы `__iter__` и `__next__` так, чтобы по автопарку можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте автопарк с несколькими автомобилями и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(fleet)` должен возвращать список всех автомобилей.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В автопарке автомобили должны быть доступны по индексу — как элементы списка: `fleet[0]` возвращает первый автомобиль, `fleet[1] = car` заменяет второй. Реализуйте в классе `Fleet` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `fleet[0]` — получение автомобиля по индексу, `fleet[1] = new_car` — замена автомобиля, срезы `fleet[0:2]` — получение подсписка, `for car in fleet` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `car in fleet`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В прокате есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных аренды её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `RentalDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with RentalDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В прокате есть разные способы расчёта стоимости аренды: фиксированная (4000 ₽/сутки), суточная по классу (2000/4000/9000 в зависимости от класса), и «для постоянных клиентов» (скидка 15% от фиксированной). Реализуйте паттерн «Стратегия»: базовый класс `RentalPricingStrategy` с методом `calculate(days)`, и три наследника — `FixedPricing`, `ClassPricing`, `LoyaltyPricing`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `LoyaltyPricing`, чтобы вызвать реализацию `FixedPricing` и применить скидку. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В прокате появился новый класс автомобилей — внедорожники (`SUV`). Если добавлять их через изменение базового класса `RentalItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `RentalItem` с методом `rental_price()`; наследников `EconomyCar`, `ComfortCar`; функцию `process_item(item: RentalItem)`, которая работает с любым автомобилем через родительский интерфейс. Добавьте `SUV` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В прокате есть служебный объект `RentalCalculator`, который должен уметь рассчитывать стоимость аренды за нужное количество дней. Удобно, чтобы его можно было вызывать как функцию: `calculator(days)`, а не `calculator.calculate(days)`. Реализуйте в классе `RentalCalculator` магический метод `__call__(self, days)`, который возвращает стоимость аренды (например, 4000 ₽ × дни). Продемонстрируйте, что объект можно вызывать как функцию: `calc = RentalCalculator(); calc(5)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В прокате есть класс `RentalCar`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `RentalCar` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования автомобилей по марке и модели): два вызова `RentalCar("Toyota", "Camry", 4000)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В прокате есть базовый класс `RentalItem` с методом `get_info()`. Нужно добавлять к автомобилю дополнительную информацию — например, «с полным баком», «с детским креслом», «с дополнительной страховкой». При этом нельзя менять сам класс `RentalItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `RentalItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `FullTankDecorator` и `ChildSeatDecorator`, которые можно комбинировать: `FullTankDecorator(ChildSeatDecorator(car))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В прокате появилась внешняя система бронирования, которая возвращает данные об автомобилях в другом формате — например, в виде словаря `{"make": "Toyota", "model_name": "Camry", "daily_rate": 4000}`. А весь остальной код проката работает с классом `RentalCar`, у которого поля `brand`, `model`, `price_per_day`. Реализуйте паттерн «Адаптер»: класс `RentalCarAdapter`, который принимает словарь и предоставляет интерфейс `RentalCar` — свойства `brand`, `model`, `price_per_day` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `RentalCar` — например, передать его в функцию, которая работает с автомобилем. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В прокате есть базовый класс `RentalItem`, и постепенно в нём накопились методы, которые нужны не всем объектам: `drive()` (только для автомобилей), `insure()` (только для страхуемых), `refuel()` (только для машин с ДВС). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Drivable`, `Insurable`, `Refuelable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `RentalService`, который зависит не от конкретных классов (`EconomyCar`, `PremiumCar`), а от абстракции `Drivable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип автомобиля (например, `SUV`, реализующий только `Drivable`) без изменения `RentalService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В музее ведётся учёт экспонатов. Создайте класс `Exhibit` с полями `title` (название), `era` (эпоха) и `value` (оценочная стоимость). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией об экспонате в формате: `Название — эпоха, стоимость ...`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Мона Лиза — Возрождение, стоимость 1000000000 ₽`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого экспоната есть инвентарный номер, который нельзя менять или читать напрямую. Создайте класс `ExhibitCard`, в котором инвентарный номер хранится в приватном поле `__inventory_id`. Реализуйте метод `set_inventory_id(value)`, предназначенный только для установки номера при постановке на учёт, и метод `get_inventory_id()` для чтения значения. Повторная попытка изменить номер после постановки на учёт должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте карточку, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `card.__inventory_id` — будет ошибка `AttributeError`, но через name mangling (`card._ExhibitCard__inventory_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В музее хранятся разные типы экспонатов. Создайте базовый класс `MuseumItem` с методом `get_item_type()`, возвращающим строку «Экспонат». Унаследуйте от него классы `Painting` и `Sculpture`, переопределив в каждом метод `get_item_type()` так, чтобы он возвращал «Картина» и «Скульптура» соответственно. Создайте по одному объекту каждого класса и выведите их тип, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Музей рассчитывает стоимость реставрации для разных типов экспонатов. Создайте классы `Painting`, `Sculpture` и `Manuscript`, у каждого из которых есть метод `restoration_cost()`, возвращающий стоимость: для картины — 50000, для скульптуры — 80000, для рукописи — 30000. Напишите функцию `show_cost(item)`, которая принимает любой экспонат и выводит стоимость реставрации. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `Curator` (куратор музея), в котором есть три поля: `name` — публичное (доступно всем), `_hall` — защищённое (зал, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_hall` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_Curator__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Музею нужна утилита для проверки возраста посетителя, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `MuseumUtils` со статическим методом `is_child_ticket(age)`, который возвращает `True`, если посетителю меньше 14 лет, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `MuseumUtils.is_child_ticket(10)`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `MuseumObject` с абстрактным методом `describe()`. Этот класс описывает общее понятие «объект музея», но не может существовать сам по себе — у каждого конкретного объекта должно быть своё описание. Унаследуйте от него классы `Exhibit` и `Hall`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `MuseumObject` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждого экспоната есть витрина — экспонат не может выставляться без неё, и витрина создаётся вместе с экспонатом. Создайте класс `Showcase` с методом `show()`, который выводит строку «Стеклянная витрина с подсветкой». Создайте класс `Exhibit`, который содержит объект класса `Showcase` внутри себя как поле (композиция) — объект `Showcase` создаётся прямо в конструкторе экспоната. Метод `show_exhibit()` у экспоната должен выводить название и вызывать метод `show()` у его витрины, демонстрируя, что экспонат «владеет» своей витриной. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В музее есть базовый класс `MuseumObject` с методом `get_info()`, который возвращает строку «Объект музея». Унаследуйте от него класс `Painting` и переопределите метод `get_info()` так, чтобы он возвращал строку «Картина: <название>». Создайте объект класса `Painting`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Экскурсоводу нужен метод для оформления экскурсий, который мог бы работать в разных ситуациях: иногда известен только посетитель, иногда посетитель и группа, иногда ещё и гид. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `TourDesk` с методом `book(visitor, group=None, guide=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Оформлено посещение: <visitor>», «Оформлено посещение группы <group> от <visitor>» или «Оформлена экскурсия группы <group> с гидом <guide> от <visitor>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Экспонаты часто выводят на экран — в каталогах, путеводителях, отчётах. По умолчанию `print(exhibit)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Exhibit` магический метод `__str__`, который возвращает строку вида «Экспонат "Мона Лиза" (Возрождение, 1000000000 ₽)». Убедитесь, что при вызове `print(exhibit)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Exhibit('Мона Лиза', 'Возрождение', 1000000000)` — оно используется при выводе списка экспонатов и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В музее может оказаться несколько записей об одном и том же экспонате — например, при повторной постановке на учёт. Нужно, чтобы программа понимала, что это «один и тот же экспонат», и не дублировала его в каталоге. Реализуйте в классе `Exhibit` магический метод `__eq__`, чтобы два экспоната считались равными, если совпадают их название и эпоха. Дополнительно реализуйте `__hash__`, чтобы экспонаты можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых экспоната попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В музее есть коллекции, и иногда нужно объединять их (например, при объединении выставок) или узнавать, сколько экспонатов в коллекции. Создайте класс `Collection` (коллекция), в котором хранится список экспонатов. Реализуйте магический метод `__add__` для объединения двух коллекций в одну (возвращается **новая** коллекция со всеми экспонатами) и метод `__len__`, возвращающий количество экспонатов. Продемонстрируйте работу: создайте две коллекции, объедините их через `c3 = c1 + c2` и выведите `len(c3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В музее важно, чтобы оценочная стоимость экспоната была корректной — нельзя допустить отрицательную стоимость или ноль. Перепишите класс `Exhibit` так, чтобы поле `value` было реализовано через свойство `@property`. Проверки: стоимость должна быть числом (int или float) и должна быть строго больше нуля. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `exhibit.value` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `exhibit.value = -100` → ошибка, `exhibit.value = 0` → ошибка, `exhibit.value = "abc"` → ошибка, `exhibit.value = 50000` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В музей часто поступают данные об экспонатах в виде строк — например, из файла или базы данных: `"Мона Лиза;Возрождение;1000000000"`. Добавьте в класс `Exhibit` классовый метод `from_string(cls, data)`, который создаёт объект экспоната из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание экспоната через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Музею нужно вести общий счётчик всех экспонатов — сколько всего экспонатов зарегистрировано в системе. Добавьте в класс `Exhibit` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового экспоната. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько экспонатов и выведите `Exhibit.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`exhibit.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В музее возможна ситуация, когда посетитель хочет увидеть экспонат, которого нет в зале. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `ExhibitNotFoundError(Exception)` и используйте его в классе `Museum` (музей). Метод `find_exhibit(title)` должен возвращать экспонат, если он есть, и выбрасывать `ExhibitNotFoundError` с понятным сообщением (например, с указанием названия), если его нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В музее куратор ведёт залы — но залы не принадлежат куратору, они существуют отдельно и принадлежат музею. Реализуйте класс `Curator` (куратор), который содержит список залов, за которые он отвечает. Залы при этом существуют отдельно от куратора — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого экспоната есть ссылка на `Museum`, в котором он хранится. Продемонстрируйте, что при удалении куратора залы остаются в музее. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В музее работает человек, который одновременно и куратор, и реставратор — например, универсальный специалист. Создайте два класса-роли: `Curator` (куратор) и `Restorer` (реставратор). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `CuratorRestorer`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `CuratorRestorer.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Exhibit` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Exhibit` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Exhibit` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Exhibit(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В музее может храниться огромное количество записей об экспонатах — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Exhibit` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Exhibit` атрибут `__slots__ = ("title", "era", "value")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `exhibit.location = "зал 5"` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Exhibit` для полей `value` (например, диапазон 1–1_000_000_000) и `year` (диапазон 1000–2100), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В музее должен быть только один реестр экспонатов — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `Museum` будет только один экземпляр. Проверьте, что `Museum()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `ExhibitRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Музей регулярно пополняется разными типами экспонатов — картинами, скульптурами, рукописями. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `ExhibitFactory` с методом `create(kind, **kwargs)`, который по строке `"painting"`, `"sculpture"` или `"manuscript"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному экспонату каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Посетители хотят получать уведомления, когда открывается новая выставка или появляется новый экспонат. Реализуйте паттерн «Наблюдатель»: класс `Museum` хранит список подписчиков (`Visitor`) и при появлении нового экспоната уведомляет всех через метод `notify()`. Продемонстрируйте, что при появлении экспоната все посетители получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки посетитель больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Museum` со временем разросся: он и хранит экспонаты, и проводит экскурсии, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Museum` на несколько классов: `ExhibitStorage` (хранение экспонатов и поиск по названию), `TourService` (проведение экскурсий, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «посетитель записался на экскурсию». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Цифровые экспонаты (`DigitalExhibit`) похожи на обычные, но имеют дополнительные поля — формат файла, разрешение. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`DigitalExhibit(Exhibit)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`DigitalExhibit` содержит объект `Exhibit` внутри себя как поле `self.exhibit`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В музее нужно создавать копии карточек экспонатов — например, для архивного каталога. При этом у экспоната есть вложенный объект `Showcase`, и важно понимать, копируется ли он вместе с экспонатом или остаётся общим. Реализуйте в классе `Exhibit` методы `__copy__` и `__deepcopy__`. Создайте экспонат с вложенным объектом `Showcase` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — витрина общая (`shallow.showcase is exhibit.showcase` → `True`), при глубоком — своя (`deep.showcase is exhibit.showcase` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В музее есть разные объекты, у которых есть метод `describe()` — экспонаты, залы, кураторы, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Exhibit`, но и с любым другим классом, у которого есть `describe()` (например, классом `Hall`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В музее есть коллекция `Collection`, и хочется перебирать экспонаты в ней в цикле `for` — так же удобно, как список. Сейчас `for exhibit in collection:` не работает. Реализуйте в классе `Collection` магические методы `__iter__` и `__next__` так, чтобы по коллекции можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте коллекцию с несколькими экспонатами и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(collection)` должен возвращать список всех экспонатов.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В коллекции экспонаты должны быть доступны по индексу — как элементы списка: `collection[0]` возвращает первый экспонат, `collection[1] = exhibit` заменяет второй. Реализуйте в классе `Collection` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `collection[0]` — получение экспоната по индексу, `collection[1] = new_exhibit` — замена экспоната, срезы `collection[0:2]` — получение подсписка, `for exhibit in collection` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `exhibit in collection`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В музее есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных экспонатов её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `ExhibitDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with ExhibitDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В музее есть разные способы расчёта стоимости билета: фиксированная (500 ₽), по возрасту (дети — 200 ₽, взрослые — 500 ₽, пенсионеры — 300 ₽) и «для групп» (скидка 20% от фиксированной). Реализуйте паттерн «Стратегия»: базовый класс `TicketPricingStrategy` с методом `calculate(visitor)`, и три наследника — `FixedPricing`, `AgePricing`, `GroupPricing`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `GroupPricing`, чтобы вызвать реализацию `FixedPricing` и применить скидку. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В музее появился новый тип экспоната — инсталляция (`Installation`). Если добавлять её через изменение базового класса `MuseumObject`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `MuseumObject` с методом `restoration_cost()`; наследников `Painting`, `Sculpture`; функцию `process_item(item: MuseumObject)`, которая работает с любым экспонатом через родительский интерфейс. Добавьте `Installation` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В музее есть служебный объект `TicketCalculator`, который должен уметь рассчитывать стоимость билета для посетителя. Удобно, чтобы его можно было вызывать как функцию: `calculator(age)`, а не `calculator.calculate(age)`. Реализуйте в классе `TicketCalculator` магический метод `__call__(self, age)`, который возвращает стоимость билета (например, 200 ₽ для детей до 14 лет и 500 ₽ для остальных). Продемонстрируйте, что объект можно вызывать как функцию: `calc = TicketCalculator(); calc(10)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В музее есть класс `Exhibit`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Exhibit` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования экспонатов по названию и эпохе): два вызова `Exhibit("Мона Лиза", "Возрождение", 1000000000)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В музее есть базовый класс `MuseumObject` с методом `get_info()`. Нужно добавлять к экспонату дополнительную информацию — например, «отреставрирован», «на реставрации», «выставлен на обозрение». При этом нельзя менять сам класс `MuseumObject`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `MuseumObject` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `RestoredDecorator` и `OnDisplayDecorator`, которые можно комбинировать: `RestoredDecorator(OnDisplayDecorator(painting))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В музее появилась внешняя система учёта, которая возвращает данные об экспонатах в другом формате — например, в виде словаря `{"name": "Мона Лиза", "period": "Возрождение", "cost": 1000000000}`. А весь остальной код музея работает с классом `Exhibit`, у которого поля `title`, `era`, `value`. Реализуйте паттерн «Адаптер»: класс `ExhibitAdapter`, который принимает словарь и предоставляет интерфейс `Exhibit` — свойства `title`, `era`, `value` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Exhibit` — например, передать его в функцию, которая работает с экспонатом. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В музее есть базовый класс `MuseumObject`, и постепенно в нём накопились методы, которые нужны не всем объектам: `restore()` (только для реставрируемых), `display()` (только для выставляемых), `store()` (только для складируемых). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Restorable`, `Displayable`, `Storable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `RestorationService`, который зависит не от конкретных классов (`Painting`, `Sculpture`), а от абстракции `Restorable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип экспоната (например, `Installation`, реализующий только `Restorable`) без изменения `RestorationService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В картинной галерее ведётся учёт работ. Создайте класс `Artwork` с полями `title` (название), `artist` (художник) и `year` (год создания). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о работе в формате: `Название — художник, год`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Звёздная ночь — Винсент ван Гог, 1889`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждой картины есть каталожный номер, который нельзя менять или читать напрямую. Создайте класс `Catalog`, в котором каталожный номер хранится в приватном поле `__catalog_id`. Реализуйте метод `set_catalog_id(value)`, предназначенный только для установки номера при добавлении картины в каталог, и метод `get_catalog_id()` для чтения значения. Повторная попытка изменить номер после добавления в каталог должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте каталог, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `catalog.__catalog_id` — будет ошибка `AttributeError`, но через name mangling (`catalog._Catalog__catalog_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В галерее представлены разные направления искусства. Создайте базовый класс `ArtStyle` с методом `get_style()`, возвращающим строку «Стиль». Унаследуйте от него классы `Impressionism` и `Realism`, переопределив в каждом метод `get_style()` так, чтобы он возвращал «Импрессионизм» и «Реализм» соответственно. Создайте по одному объекту каждого класса и выведите их стиль, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Галерея рассчитывает страховую стоимость для разных типов работ. Создайте классы `Painting`, `Sculpture` и `Graphic`, у каждого из которых есть метод `insurance_value()`, возвращающий стоимость страховки: для живописи — 100000, для скульптуры — 150000, для графики — 60000. Напишите функцию `show_value(art)`, которая принимает любую работу и выводит стоимость страховки. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `GalleryDirector` (директор галереи), в котором есть три поля: `name` — публичное (доступно всем), `_exposition` — защищённое (экспозиция, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_exposition` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_GalleryDirector__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Галерее нужна утилита для проверки времени работы, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `GalleryUtils` со статическим методом `is_open(day)`, который возвращает `True`, если день недели не понедельник, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `GalleryUtils.is_open("вторник")`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `GalleryItem` с абстрактным методом `describe()`. Этот класс описывает общее понятие «объект галереи», но не может существовать сам по себе — у каждого конкретного объекта должно быть своё описание. Унаследуйте от него классы `Artwork` и `Artist`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `GalleryItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждой картины есть рама — картина не может выставляться без неё, и рама создаётся вместе с картиной. Создайте класс `Frame` с методом `show()`, который выводит строку «Золочёная рама». Создайте класс `Artwork`, который содержит объект класса `Frame` внутри себя как поле (композиция) — объект `Frame` создаётся прямо в конструкторе картины. Метод `show_artwork()` у картины должен выводить название и вызывать метод `show()` у её рамы, демонстрируя, что картина «владеет» своей рамой. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В галерее есть базовый класс `GalleryItem` с методом `get_info()`, который возвращает строку «Объект галереи». Унаследуйте от него класс `Artwork` и переопределите метод `get_info()` так, чтобы он возвращал строку «Работа: <название> — <художник>». Создайте объект класса `Artwork`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Администратору галереи нужен метод для оформления экскурсий, который мог бы работать в разных ситуациях: иногда известен только посетитель, иногда посетитель и группа, иногда ещё и гид. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `GuideDesk` с методом `book(visitor, group=None, guide=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Оформлен визит: <visitor>», «Оформлен визит группы <group> от <visitor>» или «Оформлена экскурсия группы <group> с гидом <guide> от <visitor>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Работы часто выводят на экран — в каталогах, буклетах, отчётах. По умолчанию `print(artwork)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Artwork` магический метод `__str__`, который возвращает строку вида «Работа "Звёздная ночь" (Винсент ван Гог, 1889)». Убедитесь, что при вызове `print(artwork)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Artwork('Звёздная ночь', 'Винсент ван Гог', 1889)` — оно используется при выводе списка работ и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В галерее может оказаться несколько записей об одной и той же работе — например, при повторном добавлении в каталог. Нужно, чтобы программа понимала, что это «одна и та же работа», и не дублировала её в экспозиции. Реализуйте в классе `Artwork` магический метод `__eq__`, чтобы две работы считались равными, если совпадают их название и художник. Дополнительно реализуйте `__hash__`, чтобы работы можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что две одинаковые работы попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В галерее есть экспозиции, и иногда нужно объединять их (например, при объединении выставок) или узнавать, сколько работ в экспозиции. Создайте класс `Exposition` (экспозиция), в котором хранится список работ. Реализуйте магический метод `__add__` для объединения двух экспозиций в одну (возвращается **новая** экспозиция со всеми работами) и метод `__len__`, возвращающий количество работ. Продемонстрируйте работу: создайте две экспозиции, объедините их через `e3 = e1 + e2` и выведите `len(e3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В галерее важно, чтобы год создания работы был корректным — нельзя допустить год 3000 или год 1000. Перепишите класс `Artwork` так, чтобы поле `year` было реализовано через свойство `@property`. Проверки: год должен быть целым числом и лежать в диапазоне от 1200 до текущего года (используйте `from datetime import date; current_year = date.today().year`). При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `artwork.year` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `artwork.year = 1000` → ошибка, `artwork.year = 3000` → ошибка, `artwork.year = "abc"` → ошибка, `artwork.year = 1889` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В галерею часто поступают данные о работах в виде строк — например, из файла или базы данных: `"Звёздная ночь;Винсент ван Гог;1889"`. Добавьте в класс `Artwork` классовый метод `from_string(cls, data)`, который создаёт объект работы из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание работы через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Галерее нужно вести общий счётчик всех работ — сколько всего работ зарегистрировано в системе. Добавьте в класс `Artwork` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждой новой работы. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько работ и выведите `Artwork.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`artwork.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В галерее возможна ситуация, когда посетитель хочет увидеть работу, которой нет в экспозиции. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `ArtworkNotFoundError(Exception)` и используйте его в классе `Gallery` (галерея). Метод `find_artwork(title)` должен возвращать работу, если она есть, и выбрасывать `ArtworkNotFoundError` с понятным сообщением (например, с указанием названия), если её нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В галерее художник выставляет работы — но работы не принадлежат художнику после продажи, они существуют отдельно и принадлежат галерее. Реализуйте класс `Artist` (художник), который содержит список работ, выставленных в галерее. Работы при этом существуют отдельно от художника — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждой работы есть ссылка на `Gallery`, в которой она выставляется. Продемонстрируйте, что при удалении художника работы остаются в галерее. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В галерее работает человек, который одновременно и куратор, и искусствовед — например, специалист, совмещающий обе роли. Создайте два класса-роли: `Curator` (куратор) и `ArtCritic` (искусствовед). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `CuratorCritic`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `CuratorCritic.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Artwork` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Artwork` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Artwork` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Artwork(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В галерее может храниться огромное количество записей о работах — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Artwork` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Artwork` атрибут `__slots__ = ("title", "artist", "year")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `artwork.location = "зал 3"` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Artwork` для полей `year` (например, диапазон 1200–2100) и `price` (диапазон 1–1000000000), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В галерее должен быть только один реестр работ — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `Gallery` будет только один экземпляр. Проверьте, что `Gallery()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `ArtworkRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Галерея регулярно пополняется разными типами работ — живописью, скульптурой, графикой. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `ArtworkFactory` с методом `create(kind, **kwargs)`, который по строке `"painting"`, `"sculpture"` или `"graphic"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одной работе каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Посетители хотят получать уведомления, когда открывается новая выставка или привозят новую работу. Реализуйте паттерн «Наблюдатель»: класс `Gallery` хранит список подписчиков (`Visitor`) и при добавлении новой работы уведомляет всех через метод `notify()`. Продемонстрируйте, что при добавлении работы все посетители получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки посетитель больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Gallery` со временем разросся: он и хранит работы, и проводит экскурсии, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Gallery` на несколько классов: `ArtworkStorage` (хранение работ и поиск по названию), `TourService` (проведение экскурсий, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «посетитель записался на экскурсию». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Цифровые работы (`DigitalArtwork`) похожи на обычные работы, но имеют дополнительные поля — формат файла, разрешение. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`DigitalArtwork(Artwork)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`DigitalArtwork` содержит объект `Artwork` внутри себя как поле `self.artwork`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В галерее нужно создавать копии карточек работ — например, для архивного каталога. При этом у работы есть вложенный объект `Frame`, и важно понимать, копируется ли он вместе с работой или остаётся общим. Реализуйте в классе `Artwork` методы `__copy__` и `__deepcopy__`. Создайте работу с вложенным объектом `Frame` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — рама общая (`shallow.frame is artwork.frame` → `True`), при глубоком — своя (`deep.frame is artwork.frame` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В галерее есть разные объекты, у которых есть метод `describe()` — работы, художники, кураторы, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Artwork`, но и с любым другим классом, у которого есть `describe()` (например, классом `Artist`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В галерее есть экспозиция `Exposition`, и хочется перебирать работы в ней в цикле `for` — так же удобно, как список. Сейчас `for artwork in exposition:` не работает. Реализуйте в классе `Exposition` магические методы `__iter__` и `__next__` так, чтобы по экспозиции можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте экспозицию с несколькими работами и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(exposition)` должен возвращать список всех работ.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В экспозиции работы должны быть доступны по индексу — как элементы списка: `exposition[0]` возвращает первую работу, `exposition[1] = artwork` заменяет вторую. Реализуйте в классе `Exposition` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `exposition[0]` — получение работы по индексу, `exposition[1] = new_artwork` — замена работы, срезы `exposition[0:2]` — получение подсписка, `for artwork in exposition` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `artwork in exposition`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В галерее есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных работ её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `ArtworkDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with ArtworkDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В галерее есть разные способы расчёта стоимости билета: фиксированная (700 ₽), по возрасту (дети — 300 ₽, взрослые — 700 ₽, пенсионеры — 400 ₽) и «для групп» (скидка 20% от фиксированной). Реализуйте паттерн «Стратегия»: базовый класс `TicketPricingStrategy` с методом `calculate(visitor)`, и три наследника — `FixedPricing`, `AgePricing`, `GroupPricing`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `GroupPricing`, чтобы вызвать реализацию `FixedPricing` и применить скидку. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В галерее появился новый тип работы — инсталляция (`Installation`). Если добавлять её через изменение базового класса `GalleryItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `GalleryItem` с методом `insurance_value()`; наследников `Painting`, `Sculpture`; функцию `process_item(item: GalleryItem)`, которая работает с любой работой через родительский интерфейс. Добавьте `Installation` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В галерее есть служебный объект `TicketCalculator`, который должен уметь рассчитывать стоимость билета для посетителя по возрасту. Удобно, чтобы его можно было вызывать как функцию: `calculator(age)`, а не `calculator.calculate(age)`. Реализуйте в классе `TicketCalculator` магический метод `__call__(self, age)`, который возвращает стоимость билета (например, 300 ₽ для детей до 14 лет, 400 ₽ для пенсионеров, 700 ₽ для остальных). Продемонстрируйте, что объект можно вызывать как функцию: `calc = TicketCalculator(); calc(10)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В галерее есть класс `Artwork`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Artwork` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования работ по названию и художнику): два вызова `Artwork("Звёздная ночь", "Винсент ван Гог", 1889)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В галерее есть базовый класс `GalleryItem` с методом `get_info()`. Нужно добавлять к работе дополнительную информацию — например, «отреставрирована», «на реставрации», «выставлена на обозрение». При этом нельзя менять сам класс `GalleryItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `GalleryItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `RestoredDecorator` и `OnDisplayDecorator`, которые можно комбинировать: `RestoredDecorator(OnDisplayDecorator(painting))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В галерее появилась внешняя система учёта, которая возвращает данные о работах в другом формате — например, в виде словаря `{"name": "Звёздная ночь", "author": "Винсент ван Гог", "created": 1889}`. А весь остальной код галереи работает с классом `Artwork`, у которого поля `title`, `artist`, `year`. Реализуйте паттерн «Адаптер»: класс `ArtworkAdapter`, который принимает словарь и предоставляет интерфейс `Artwork` — свойства `title`, `artist`, `year` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Artwork` — например, передать его в функцию, которая работает с работой. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В галерее есть базовый класс `GalleryItem`, и постепенно в нём накопились методы, которые нужны не всем объектам: `restore()` (только для реставрируемых), `display()` (только для выставляемых), `insure()` (только для страхуемых). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Restorable`, `Displayable`, `Insurable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `RestorationService`, который зависит не от конкретных классов (`Painting`, `Sculpture`), а от абстракции `Restorable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип работы (например, `Installation`, реализующий только `Restorable`) без изменения `RestorationService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В аквапарке ведётся учёт посетителей. Создайте класс `Visitor` с полями `name` (имя), `age` (возраст) и `ticket_type` (тип билета: взрослый, детский). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о посетителе в формате: `Имя — возраст ..., билет: ...`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Иван Петров — возраст 32, билет: взрослый`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого посетителя есть номер браслета, который нельзя менять или читать напрямую. Создайте класс `Bracelet`, в котором номер браслета хранится в приватном поле `__bracelet_id`. Реализуйте метод `set_bracelet_id(value)`, предназначенный только для установки номера при входе, и метод `get_bracelet_id()` для чтения значения. Повторная попытка изменить номер после входа должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте браслет, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `bracelet.__bracelet_id` — будет ошибка `AttributeError`, но через name mangling (`bracelet._Bracelet__bracelet_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В аквапарке есть разные типы горок. Создайте базовый класс `WaterSlide` с методом `get_slide_type()`, возвращающим строку «Горка». Унаследуйте от него классы `OpenSlide` и `TubeSlide`, переопределив в каждом метод `get_slide_type()` так, чтобы он возвращал «Открытая горка» и «Закрытая труба» соответственно. Создайте по одному объекту каждого класса и выведите их тип, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Аквапарк рассчитывает пропускную способность для разных бассейнов. Создайте классы `KidsPool`, `WavePool` и `DeepPool`, у каждого из которых есть метод `capacity()`, возвращающий максимальное число людей: для детского — 50, для волнового — 200, для глубокого — 100. Напишите функцию `show_capacity(pool)`, которая принимает любой бассейн и выводит его вместимость. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `ParkManager` (управляющий аквапарка), в котором есть три поля: `name` — публичное (доступно всем), `_zone` — защищённое (зона, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_zone` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_ParkManager__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Аквапарку нужна утилита для проверки температуры воды, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `AquaUtils` со статическим методом `is_comfortable(temp)`, который возвращает `True`, если температура от 24 до 28 градусов, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `AquaUtils.is_comfortable(26)`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `AquaItem` с абстрактным методом `describe()`. Этот класс описывает общее понятие «объект аквапарка», но не может существовать сам по себе — у каждого конкретного объекта должно быть своё описание. Унаследуйте от него классы `WaterSlide` и `Pool`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `AquaItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждой горки есть спасательный круг — горка не может эксплуатироваться без него, и круг создаётся вместе с горкой. Создайте класс `LifeRing` с методом `show()`, который выводит строку «Спасательный круг на месте». Создайте класс `WaterSlide`, который содержит объект класса `LifeRing` внутри себя как поле (композиция) — объект `LifeRing` создаётся прямо в конструкторе горки. Метод `show_slide()` у горки должен выводить её тип и вызывать метод `show()` у спасательного круга, демонстрируя, что горка «владеет» своим кругом. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В аквапарке есть базовый класс `AquaItem` с методом `get_info()`, который возвращает строку «Объект аквапарка». Унаследуйте от него класс `WaterSlide` и переопределите метод `get_info()` так, чтобы он возвращал строку «Горка: <тип>». Создайте объект класса `WaterSlide`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Кассиру аквапарка нужен метод для продажи билетов, который мог бы работать в разных ситуациях: иногда известен только посетитель, иногда посетитель и количество билетов, иногда ещё и зона. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `TicketDesk` с методом `sell(visitor, count=None, zone=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Продан билет посетителю: <name>», «Продано <count> билетов посетителю: <name>» или «Продано <count> билетов в зону <zone> посетителю <name>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Посетителей часто выводят на экран — в списках, отчётах, на стойке регистрации. По умолчанию `print(visitor)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Visitor` магический метод `__str__`, который возвращает строку вида «Посетитель "Иван Петров" (32 года, билет: взрослый)». Убедитесь, что при вызове `print(visitor)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Visitor('Иван Петров', 32, 'взрослый')` — оно используется при выводе списка посетителей и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В аквапарке может оказаться несколько записей об одном и том же посетителе — например, при повторной регистрации. Нужно, чтобы программа понимала, что это «один и тот же посетитель», и не дублировала его в списках. Реализуйте в классе `Visitor` магический метод `__eq__`, чтобы два посетителя считались равными, если совпадают их имя и возраст. Дополнительно реализуйте `__hash__`, чтобы посетителей можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых посетителя попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В аквапарке есть группы посетителей, и иногда нужно объединять их (например, при объединении экскурсий) или узнавать, сколько людей в группе. Создайте класс `Group` (группа), в котором хранится список посетителей. Реализуйте магический метод `__add__` для объединения двух групп в одну (возвращается **новая** группа со всеми посетителями) и метод `__len__`, возвращающий количество посетителей. Продемонстрируйте работу: создайте две группы, объедините их через `g3 = g1 + g2` и выведите `len(g3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В аквапарке важно, чтобы возраст посетителя был корректным — нельзя допустить отрицательное значение или значение больше 120. Перепишите класс `Visitor` так, чтобы поле `age` было реализовано через свойство `@property`. Проверки: возраст должен быть целым числом и лежать в диапазоне от 0 до 120 включительно. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `visitor.age` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `visitor.age = -1` → ошибка, `visitor.age = 200` → ошибка, `visitor.age = "abc"` → ошибка, `visitor.age = 32` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В аквапарк часто поступают данные о посетителях в виде строк — например, из файла или базы данных: `"Иван Петров;32;взрослый"`. Добавьте в класс `Visitor` классовый метод `from_string(cls, data)`, который создаёт объект посетителя из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание посетителя через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Аквапарку нужно вести общий счётчик всех посетителей — сколько всего человек зарегистрировано в системе. Добавьте в класс `Visitor` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового посетителя. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько посетителей и выведите `Visitor.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`visitor.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В аквапарке возможна ситуация, когда посетитель пытается попасть в зону, которая закрыта или не существует. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `ZoneNotFoundError(Exception)` и используйте его в классе `AquaPark` (аквапарк). Метод `enter_zone(visitor, zone_name)` должен пускать посетителя, если зона есть, и выбрасывать `ZoneNotFoundError` с понятным сообщением (например, с указанием названия зоны), если её нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В аквапарке инструктор ведёт группы — но группы не принадлежат инструктору, они существуют отдельно и принадлежат аквапарку. Реализуйте класс `Instructor` (инструктор), который содержит список групп, с которыми он работает. Группы при этом существуют отдельно от инструктора — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждой горки есть ссылка на `AquaPark`, в котором она установлена. Продемонстрируйте, что при удалении инструктора группы остаются в аквапарке. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В аквапарке работает человек, который одновременно и инструктор, и спасатель — например, универсальный сотрудник. Создайте два класса-роли: `Instructor` (инструктор) и `Lifeguard` (спасатель). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `InstructorLifeguard`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `InstructorLifeguard.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Visitor` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Visitor` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Visitor` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Visitor(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В аквапарке может храниться огромное количество записей о посетителях — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Visitor` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Visitor` атрибут `__slots__ = ("name", "age", "ticket_type")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `visitor.phone = "..."` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Visitor` для полей `age` (например, диапазон 0–120) и `height` (диапазон 50–250), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В аквапарке должен быть только один реестр посетителей — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `AquaPark` будет только один экземпляр. Проверьте, что `AquaPark()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `VisitorRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Аквапарк регулярно пополняется разными типами горок — открытыми, закрытыми трубами, спиральными. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `SlideFactory` с методом `create(kind, **kwargs)`, который по строке `"open"`, `"tube"` или `"spiral"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одной горке каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Посетители хотят получать уведомления, когда открывается новая зона или начинается шоу с дельфинами. Реализуйте паттерн «Наблюдатель»: класс `AquaPark` хранит список подписчиков (`Visitor`) и при появлении события уведомляет всех через метод `notify()`. Продемонстрируйте, что при появлении события все посетители получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки посетитель больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `AquaPark` со временем разросся: он и хранит посетителей, и продаёт билеты, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `AquaPark` на несколько классов: `VisitorStorage` (хранение посетителей и поиск по имени), `TicketService` (продажа билетов, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «посетитель купил билет». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Экстремальные горки (`ExtremeSlide`) похожи на обычные, но имеют дополнительные поля — максимальная скорость, ограничение по возрасту. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`ExtremeSlide(WaterSlide)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`ExtremeSlide` содержит объект `WaterSlide` внутри себя как поле `self.slide`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В аквапарке нужно создавать копии карточек посетителей — например, для архивного хранения. При этом у посетителя есть вложенный объект `Bracelet`, и важно понимать, копируется ли он вместе с посетителем или остаётся общим. Реализуйте в классе `Visitor` методы `__copy__` и `__deepcopy__`. Создайте посетителя с вложенным объектом `Bracelet` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — браслет общий (`shallow.bracelet is visitor.bracelet` → `True`), при глубоком — свой (`deep.bracelet is visitor.bracelet` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В аквапарке есть разные объекты, у которых есть метод `describe()` — горки, бассейны, инструкторы, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `WaterSlide`, но и с любым другим классом, у которого есть `describe()` (например, классом `Pool`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В аквапарке есть группа `Group`, и хочется перебирать посетителей в ней в цикле `for` — так же удобно, как список. Сейчас `for visitor in group:` не работает. Реализуйте в классе `Group` магические методы `__iter__` и `__next__` так, чтобы по группе можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте группу с несколькими посетителями и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(group)` должен возвращать список всех посетителей.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В группе посетители должны быть доступны по индексу — как элементы списка: `group[0]` возвращает первого посетителя, `group[1] = visitor` заменяет второго. Реализуйте в классе `Group` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `group[0]` — получение посетителя по индексу, `group[1] = new_visitor` — замена посетителя, срезы `group[0:2]` — получение подсписка, `for visitor in group` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `visitor in group`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В аквапарке есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных посетителей её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `VisitorDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with VisitorDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В аквапарке есть разные способы расчёта стоимости билета: фиксированная (1500 ₽), по возрасту (дети — 800 ₽, взрослые — 1500 ₽, пенсионеры — 1000 ₽) и «семейная» (скидка 25% от фиксированной). Реализуйте паттерн «Стратегия»: базовый класс `TicketPricingStrategy` с методом `calculate(visitor)`, и три наследника — `FixedPricing`, `AgePricing`, `FamilyPricing`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `FamilyPricing`, чтобы вызвать реализацию `FixedPricing` и применить скидку. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В аквапарке появился новый тип объекта — ленивая река (`LazyRiver`). Если добавлять её через изменение базового класса `AquaItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `AquaItem` с методом `capacity()`; наследников `KidsPool`, `WavePool`; функцию `process_item(item: AquaItem)`, которая работает с любым объектом через родительский интерфейс. Добавьте `LazyRiver` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В аквапарке есть служебный объект `TicketCalculator`, который должен уметь рассчитывать стоимость билета для посетителя по возрасту. Удобно, чтобы его можно было вызывать как функцию: `calculator(age)`, а не `calculator.calculate(age)`. Реализуйте в классе `TicketCalculator` магический метод `__call__(self, age)`, который возвращает стоимость билета (например, 800 ₽ для детей до 14 лет, 1000 ₽ для пенсионеров, 1500 ₽ для остальных). Продемонстрируйте, что объект можно вызывать как функцию: `calc = TicketCalculator(); calc(10)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В аквапарке есть класс `Visitor`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Visitor` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования посетителей по имени и возрасту): два вызова `Visitor("Иван Петров", 32, "взрослый")` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В аквапарке есть базовый класс `AquaItem` с методом `get_info()`. Нужно добавлять к объекту дополнительную информацию — например, «новинка», «для взрослых», «только по записи». При этом нельзя менять сам класс `AquaItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `AquaItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `NewDecorator` и `AdultsOnlyDecorator`, которые можно комбинировать: `NewDecorator(AdultsOnlyDecorator(slide))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В аквапарке появилась внешняя система бронирования, которая возвращает данные о посетителях в другом формате — например, в виде словаря `{"full_name": "Иван Петров", "years": 32, "pass": "взрослый"}`. А весь остальной код аквапарка работает с классом `Visitor`, у которого поля `name`, `age`, `ticket_type`. Реализуйте паттерн «Адаптер»: класс `VisitorAdapter`, который принимает словарь и предоставляет интерфейс `Visitor` — свойства `name`, `age`, `ticket_type` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Visitor` — например, передать его в функцию, которая работает с посетителем. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В аквапарке есть базовый класс `AquaItem`, и постепенно в нём накопились методы, которые нужны не всем объектам: `slide()` (только для горок), `swim()` (только для бассейнов), `relax()` (только для спа-зон). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Slidable`, `Swimmable`, `Relaxable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `VisitService`, который зависит не от конкретных классов (`WaterSlide`, `Pool`), а от абстракции `Slidable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип объекта (например, `LazyRiver`, реализующий только `Swimmable`) без изменения `VisitService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В боулинг-клубе ведётся учёт дорожек. Создайте класс `Lane` с полями `number` (номер дорожки), `type` (тип: стандарт, VIP) и `price_per_hour` (цена за час). Добавьте метод `info()`, который возвращает строку с информацией о дорожке в формате: `Дорожка №... — тип ..., цена за час ...`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Дорожка №5 — тип VIP, цена за час 2000 ₽`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждой игры есть номер брони, который нельзя менять или читать напрямую. Создайте класс `GameBooking`, в котором номер брони хранится в приватном поле `__booking_id`. Реализуйте метод `set_booking_id(value)`, предназначенный только для установки номера при бронировании, и метод `get_booking_id()` для чтения значения. Повторная попытка изменить номер после бронирования должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте бронь, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `booking.__booking_id` — будет ошибка `AttributeError`, но через name mangling (`booking._GameBooking__booking_id`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В боулинг-клубе есть разные типы шаров. Создайте базовый класс `BowlingBall` с методом `get_ball_type()`, возвращающим строку «Шар». Унаследуйте от него классы `LightBall` и `HeavyBall`, переопределив в каждом метод `get_ball_type()` так, чтобы он возвращал «Лёгкий шар» и «Тяжёлый шар» соответственно. Создайте по одному объекту каждого класса и выведите их тип, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Боулинг-клуб рассчитывает стоимость часа для разных типов дорожек. Создайте классы `StandardLane`, `VIPLane` и `PremiumLane`, у каждого из которых есть метод `price_per_hour()`, возвращающий стоимость: для стандарта — 1000, для VIP — 2000, для премиум — 3500. Напишите функцию `show_price(lane)`, которая принимает любую дорожку и выводит цену. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `ClubManager` (управляющий клуба), в котором есть три поля: `name` — публичное (доступно всем), `_hall` — защищённое (зал, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_hall` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_ClubManager__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Боулинг-клубу нужна утилита для проверки номера дорожки, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `BowlingUtils` со статическим методом `is_valid_lane(number)`, который возвращает `True`, если число находится в диапазоне от 1 до 20, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `BowlingUtils.is_valid_lane(5)`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `BowlingItem` с абстрактным методом `describe()`. Этот класс описывает общее понятие «объект клуба», но не может существовать сам по себе — у каждого конкретного объекта должно быть своё описание. Унаследуйте от него классы `Lane` и `Ball`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `BowlingItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждой дорожки есть механизм подачи шаров — дорожка не может работать без него, и механизм создаётся вместе с дорожкой. Создайте класс `BallReturn` с методом `show()`, который выводит строку «Механизм возврата готов». Создайте класс `Lane`, который содержит объект класса `BallReturn` внутри себя как поле (композиция) — объект `BallReturn` создаётся прямо в конструкторе дорожки. Метод `show_lane()` у дорожки должен выводить её номер и вызывать метод `show()` у механизма возврата, демонстрируя, что дорожка «владеет» своим механизмом. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В боулинг-клубе есть базовый класс `BowlingItem` с методом `get_info()`, который возвращает строку «Объект клуба». Унаследуйте от него класс `VIPLane` и переопределите метод `get_info()` так, чтобы он возвращал строку «VIP-дорожка №...». Создайте объект класса `VIPLane`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Администратору боулинга нужен метод для оформления игр, который мог бы работать в разных ситуациях: иногда известна только дорожка, иногда дорожка и количество часов, иногда ещё и число игроков. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `GameDesk` с методом `book(lane, hours=None, players=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Забронирована дорожка №<number>», «Дорожка №<number> забронирована на <hours> ч.», «Дорожка №<number> забронирована на <hours> ч. для <players> игроков». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Дорожки часто выводят на экран — в расписании, чеках, отчётах. По умолчанию `print(lane)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Lane` магический метод `__str__`, который возвращает строку вида «Дорожка №5 (VIP, 2000 ₽/час)». Убедитесь, что при вызове `print(lane)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Lane(5, 'VIP', 2000)` — оно используется при выводе списка дорожек и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В клубе может оказаться несколько записей об одной и той же дорожке — например, при повторной настройке. Нужно, чтобы программа понимала, что это «одна и та же дорожка», и не дублировала её в расписании. Реализуйте в классе `Lane` магический метод `__eq__`, чтобы две дорожки считались равными, если совпадают их номер и тип. Дополнительно реализуйте `__hash__`, чтобы дорожки можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что две одинаковые дорожки попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В клубе есть залы с дорожками, и иногда нужно объединять их (например, при объединении залов) или узнавать, сколько дорожек в зале. Создайте класс `Hall` (зал), в котором хранится список дорожек. Реализуйте магический метод `__add__` для объединения двух залов в один (возвращается **новый** зал со всеми дорожками) и метод `__len__`, возвращающий количество дорожек. Продемонстрируйте работу: создайте два зала, объедините их через `h3 = h1 + h2` и выведите `len(h3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В клубе важно, чтобы цена за час была корректной — нельзя допустить отрицательную цену или ноль. Перепишите класс `Lane` так, чтобы поле `price_per_hour` было реализовано через свойство `@property`. Проверки: цена должна быть числом (int или float) и должна быть строго больше нуля. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `lane.price_per_hour` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `lane.price_per_hour = -100` → ошибка, `lane.price_per_hour = 0` → ошибка, `lane.price_per_hour = "abc"` → ошибка, `lane.price_per_hour = 2000` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В клуб часто поступают данные о дорожках в виде строк — например, из файла или базы данных: `"5;VIP;2000"`. Добавьте в класс `Lane` классовый метод `from_string(cls, data)`, который создаёт объект дорожки из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание дорожки через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Клубу нужно вести общий счётчик всех дорожек — сколько всего дорожек зарегистрировано в системе. Добавьте в класс `Lane` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждой новой дорожки. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько дорожек и выведите `Lane.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`lane.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В боулинг-клубе возможна ситуация, когда клиент пытается забронировать дорожку, которой нет в клубе. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `LaneNotFoundError(Exception)` и используйте его в классе `BowlingClub` (боулинг-клуб). Метод `find_lane(number)` должен возвращать дорожку, если она есть, и выбрасывать `LaneNotFoundError` с понятным сообщением (например, с указанием номера), если её нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В боулинг-клубе игрок арендует дорожки — но дорожки не принадлежат игроку, они существуют отдельно и принадлежат клубу. Реализуйте класс `Player` (игрок), который содержит список забронированных дорожек. Дорожки при этом существуют отдельно от игрока — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждой дорожки есть ссылка на `BowlingClub`, в котором она находится. Продемонстрируйте, что при удалении игрока дорожки остаются в клубе. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В клубе работает человек, который одновременно и администратор, и тренер — например, универсальный сотрудник. Создайте два класса-роли: `Admin` (администратор) и `Coach` (тренер). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `AdminCoach`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `AdminCoach.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Lane` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Lane` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Lane` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Lane(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В клубе может храниться огромное количество записей о дорожках — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Lane` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Lane` атрибут `__slots__ = ("number", "type", "price_per_hour")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `lane.surface = "дерево"` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Lane` для полей `price_per_hour` (например, диапазон 100–10000) и `number` (диапазон 1–20), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В клубе должен быть только один реестр дорожек — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `BowlingClub` будет только один экземпляр. Проверьте, что `BowlingClub()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `LaneRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Боулинг-клуб регулярно пополняется разными типами дорожек — стандартными, VIP, премиум. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `LaneFactory` с методом `create(kind, **kwargs)`, который по строке `"standard"`, `"vip"` или `"premium"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одной дорожке каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Игроки хотят получать уведомления, когда освобождается нужная дорожка или начинается турнир. Реализуйте паттерн «Наблюдатель»: класс `BowlingClub` хранит список подписчиков (`Player`) и при освобождении дорожки уведомляет всех через метод `notify()`. Продемонстрируйте, что при освобождении дорожки все подписчики получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки игрок больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `BowlingClub` со временем разросся: он и хранит дорожки, и бронирует их, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `BowlingClub` на несколько классов: `LaneStorage` (хранение дорожек и поиск по номеру), `BookingService` (бронирование, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «игрок забронировал дорожку». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Турнирные дорожки (`TournamentLane`) похожи на обычные дорожки, но имеют дополнительные поля — уровень освещения, наличие датчиков счёта. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`TournamentLane(Lane)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`TournamentLane` содержит объект `Lane` внутри себя как поле `self.lane`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В клубе нужно создавать копии карточек дорожек — например, для архивного расписания. При этом у дорожки есть вложенный объект `BallReturn`, и важно понимать, копируется ли он вместе с дорожкой или остаётся общим. Реализуйте в классе `Lane` методы `__copy__` и `__deepcopy__`. Создайте дорожку с вложенным объектом `BallReturn` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — механизм общий (`shallow.ball_return is lane.ball_return` → `True`), при глубоком — свой (`deep.ball_return is lane.ball_return` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В клубе есть разные объекты, у которых есть метод `describe()` — дорожки, шары, игроки, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Lane`, но и с любым другим классом, у которого есть `describe()` (например, классом `BowlingBall`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В клубе есть зал `Hall`, и хочется перебирать дорожки в нём в цикле `for` — так же удобно, как список. Сейчас `for lane in hall:` не работает. Реализуйте в классе `Hall` магические методы `__iter__` и `__next__` так, чтобы по залу можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте зал с несколькими дорожками и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(hall)` должен возвращать список всех дорожек.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В зале дорожки должны быть доступны по индексу — как элементы списка: `hall[0]` возвращает первую дорожку, `hall[1] = lane` заменяет вторую. Реализуйте в классе `Hall` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `hall[0]` — получение дорожки по индексу, `hall[1] = new_lane` — замена дорожки, срезы `hall[0:2]` — получение подсписка, `for lane in hall` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `lane in hall`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В клубе есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных бронирований её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `BookingDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with BookingDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В клубе есть разные способы расчёта стоимости игры: фиксированная (1000 ₽/час), по типу дорожки (стандарт — 1000, VIP — 2000, премиум — 3500) и «для постоянных клиентов» (скидка 20% от фиксированной). Реализуйте паттерн «Стратегия»: базовый класс `PricingStrategy` с методом `calculate(hours)`, и три наследника — `FixedPricing`, `LaneTypePricing`, `LoyaltyPricing`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `LoyaltyPricing`, чтобы вызвать реализацию `FixedPricing` и применить скидку. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В клубе появился новый тип дорожки — детская (`KidsLane`). Если добавлять её через изменение базового класса `BowlingItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `BowlingItem` с методом `price_per_hour()`; наследников `StandardLane`, `VIPLane`; функцию `process_item(item: BowlingItem)`, которая работает с любой дорожкой через родительский интерфейс. Добавьте `KidsLane` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В клубе есть служебный объект `PriceCalculator`, который должен уметь рассчитывать стоимость игры по количеству часов. Удобно, чтобы его можно было вызывать как функцию: `calculator(hours)`, а не `calculator.calculate(hours)`. Реализуйте в классе `PriceCalculator` магический метод `__call__(self, hours)`, который возвращает стоимость игры (например, 1000 ₽ × часы). Продемонстрируйте, что объект можно вызывать как функцию: `calc = PriceCalculator(); calc(3)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В клубе есть класс `Lane`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Lane` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования дорожек по номеру и типу): два вызова `Lane(5, "VIP", 2000)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В клубе есть базовый класс `BowlingItem` с методом `get_info()`. Нужно добавлять к объекту дополнительную информацию — например, «турнирная», «с подсветкой», «для детей». При этом нельзя менять сам класс `BowlingItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `BowlingItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `TournamentDecorator` и `LightingDecorator`, которые можно комбинировать: `TournamentDecorator(LightingDecorator(lane))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В клубе появилась внешняя система бронирования, которая возвращает данные о дорожках в другом формате — например, в виде словаря `{"lane_number": 5, "lane_kind": "VIP", "hour_price": 2000}`. А весь остальной код клуба работает с классом `Lane`, у которого поля `number`, `type`, `price_per_hour`. Реализуйте паттерн «Адаптер»: класс `LaneAdapter`, который принимает словарь и предоставляет интерфейс `Lane` — свойства `number`, `type`, `price_per_hour` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Lane` — например, передать его в функцию, которая работает с дорожкой. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В клубе есть базовый класс `BowlingItem`, и постепенно в нём накопились методы, которые нужны не всем объектам: `play()` (только для дорожек), `roll()` (только для шаров), `refill()` (только для механизмов возврата). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Playable`, `Rollable`, `Refillable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `BookingService`, который зависит не от конкретных классов (`StandardLane`, `VIPLane`), а от абстракции `Playable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип дорожки (например, `KidsLane`, реализующий только `Playable`) без изменения `BookingService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Задача 2 (инкапсуляция).** У каждого клиента тира есть номер карты посетителя, который нельзя менять или читать напрямую. **Задача 3 (наследование).** В тире есть разные виды оружия. Создайте базовый класс `Firearm` с методом `get_weapon_kind()`, возвращающим строку «Оружие». **Задача 4 (полиморфизм).** Тир рассчитывает стоимость выстрела для разных типов оружия. **Задача 5 (модификаторы доступа).** Создайте класс `RangeManager` (управляющий тира), в котором есть три поля: `name` — публичное (доступно всем), `_hall` — защищённое (зал, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). **Задача 6 (статические методы).** Тиру нужна утилита для проверки возраста клиента, которая не зависит от конкретного объекта и может вызываться сама по себе. **Задача 10 (имитация перегрузки).** Инструктору тира нужен метод для оформления сеансов, который мог бы работать в разных ситуациях: иногда известно только оружие, иногда оружие и количество выстрелов, иногда ещё и стрелок. **Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Оружие часто выводят на экран — в каталоге, чеках, отчётах. По умолчанию `print(weapon)` выводит нечитаемый адрес объекта в памяти, что неудобно. **Задача 12 (магические методы `__eq__` и `__hash__`).** В тире может оказаться несколько записей об одном и том же оружии — например, при повторной инвентаризации. Нужно, чтобы программа понимала, что это «одно и то же оружие», и не дублировала его в списке. Реализуйте в классе `Weapon` магический метод `__eq__`, чтобы два оружия считались равными, если совпадают их модель и калибр. Дополнительно реализуйте `__hash__`, чтобы оружие можно было хранить во множестве `set` и использовать как ключи в словаре. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В тире есть стойки с оружием, и иногда нужно объединять их (например, при объединении залов) или узнавать, сколько единиц оружия на стойке. Продемонстрируйте работу: создайте две стойки, объедините их через `r3 = r1 + r2` и выведите `len(r3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В тире важно, чтобы цена за выстрел была корректной — нельзя допустить отрицательную цену или ноль. Перепишите класс `Weapon` так, чтобы поле `price_per_shot` было реализовано через свойство `@property`. Проверки: цена должна быть числом (int или float) и должна быть строго больше нуля. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `weapon.price_per_shot` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `weapon.price_per_shot = -100` → ошибка, `weapon.price_per_shot = 0` → ошибка, `weapon.price_per_shot = "abc"` → ошибка, `weapon.price_per_shot = 50` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В тир часто поступают данные об оружии в виде строк — например, из файла или базы данных: `"Glock 17;9 мм;50"`. Добавьте в класс `Weapon` классовый метод `from_string(cls, data)`, который создаёт объект оружия из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание оружия через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. **Задача 16 (классовые vs экземплярные атрибуты).** Тиру нужно вести общий счётчик всего оружия — сколько всего единиц зарегистрировано в системе. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько единиц и выведите `Weapon.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`weapon.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В тире возможна ситуация, когда клиент пытается взять оружие, которого нет на стойке. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `WeaponNotFoundError(Exception)` и используйте его в классе `ShootingRange` (тир). Метод `find_weapon(model)` должен возвращать оружие, если оно есть, и выбрасывать `WeaponNotFoundError` с понятным сообщением (например, с указанием модели), если его нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В тире инструктор закрепляет за собой оружие — но оружие не принадлежит инструктору, оно существует отдельно и принадлежит тиру. Дополнительно реализуйте **ассоциацию**: у каждой единицы оружия есть ссылка на `ShootingRange`, в котором она используется. Продемонстрируйте, что при удалении инструктора оружие остаётся в тире. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В тире работает человек, который одновременно и инструктор, и оружейник — например, универсальный специалист. Создайте два класса-роли: `Instructor` (инструктор) и `Armorer` (оружейник). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `InstructorArmorer`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `InstructorArmorer.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Weapon` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Weapon` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Weapon` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Weapon(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В тире может храниться огромное количество записей об оружии — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Weapon` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Weapon` атрибут `__slots__ = ("model", "caliber", "price_per_shot")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `weapon.country = "Австрия"` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Weapon` для полей `price_per_shot` (например, диапазон 10–1000) и `caliber_mm` (диапазон 1–20), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В тире должен быть только один реестр оружия — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `ShootingRange` будет только один экземпляр. Проверьте, что `ShootingRange()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `WeaponRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Тир регулярно пополняется разными типами оружия — пистолетами, винтовками, дробовиками. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `WeaponFactory` с методом `create(kind, **kwargs)`, который по строке `"pistol"`, `"rifle"` или `"shotgun"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одной единице оружия каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Стрелки хотят получать уведомления, когда освобождается дорожка или начинается турнир. Реализуйте паттерн «Наблюдатель»: класс `ShootingRange` хранит список подписчиков (`Shooter`) и при появлении события уведомляет всех через метод `notify()`. Продемонстрируйте, что при появлении события все стрелки получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки стрелок больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `ShootingRange` со временем разросся: он и хранит оружие, и ведёт сеансы, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `ShootingRange` на несколько классов: `WeaponStorage` (хранение оружия и поиск по модели), `SessionService` (проведение сеансов, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «стрелок начал сеанс». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Спортивное оружие (`SportWeapon`) похоже на обычное, но имеет дополнительные поля — оптический прицел, точность. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`SportWeapon(Weapon)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`SportWeapon` содержит объект `Weapon` внутри себя как поле `self.weapon`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В тире нужно создавать копии карточек оружия — например, для архивного каталога. При этом у оружия есть вложенный объект `Magazine`, и важно понимать, копируется ли он вместе с оружием или остаётся общим. Реализуйте в классе `Weapon` методы `__copy__` и `__deepcopy__`. Создайте оружие с вложенным объектом `Magazine` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — магазин общий (`shallow.magazine is weapon.magazine` → `True`), при глубоком — свой (`deep.magazine is weapon.magazine` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В тире есть разные объекты, у которых есть метод `describe()` — оружие, мишени, инструкторы, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Weapon`, но и с любым другим классом, у которого есть `describe()` (например, классом `Target`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В тире есть стойка `Rack`, и хочется перебирать оружие в цикле `for` — так же удобно, как список. Сейчас `for weapon in rack:` не работает. Реализуйте в классе `Rack` магические методы `__iter__` и `__next__` так, чтобы по стойке можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте стойку с несколькими единицами оружия и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(rack)` должен возвращать список всего оружия.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** На стойке оружие должно быть доступно по индексу — как элементы списка: `rack[0]` возвращает первое оружие, `rack[1] = weapon` заменяет второе. Реализуйте в классе `Rack` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `rack[0]` — получение оружия по индексу, `rack[1] = new_weapon` — замена оружия, срезы `rack[0:2]` — получение подсписка, `for weapon in rack` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `weapon in rack`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В тире есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных сеансов её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `SessionDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with SessionDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В тире есть разные способы расчёта стоимости сеанса: фиксированная (500 ₽), по числу выстрелов (50 ₽ за выстрел), и «для постоянных клиентов» (скидка 15% от фиксированной). Реализуйте паттерн «Стратегия»: базовый класс `SessionPricingStrategy` с методом `calculate(shots)`, и три наследника — `FixedPricing`, `PerShotPricing`, `LoyaltyPricing`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `LoyaltyPricing`, чтобы вызвать реализацию `FixedPricing` и применить скидку. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В тире появился новый тип оружия — арбалет (`Crossbow`). Если добавлять его через изменение базового класса `RangeItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `RangeItem` с методом `shot_price()`; наследников `Pistol`, `Rifle`; функцию `process_item(item: RangeItem)`, которая работает с любым оружием через родительский интерфейс. Добавьте `Crossbow` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В тире есть служебный объект `ShotCalculator`, который должен уметь рассчитывать стоимость сеанса по числу выстрелов. Удобно, чтобы его можно было вызывать как функцию: `calculator(shots)`, а не `calculator.calculate(shots)`. Реализуйте в классе `ShotCalculator` магический метод `__call__(self, shots)`, который возвращает стоимость сеанса (например, 50 ₽ × выстрелы). Продемонстрируйте, что объект можно вызывать как функцию: `calc = ShotCalculator(); calc(20)`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В тире есть класс `Weapon`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Weapon` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования оружия по модели и калибру): два вызова `Weapon("Glock 17", "9 мм", 50)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В тире есть базовый класс `RangeItem` с методом `get_info()`. Нужно добавлять к объекту дополнительную информацию — например, «служебное», «для новичков», «с оптикой». При этом нельзя менять сам класс `RangeItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `RangeItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `ServiceDecorator` и `OpticsDecorator`, которые можно комбинировать: `ServiceDecorator(OpticsDecorator(rifle))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В тире появилась внешняя система учёта, которая возвращает данные об оружии в другом формате — например, в виде словаря `{"name": "Glock 17", "cal": "9 мм", "shot_cost": 50}`. А весь остальной код тира работает с классом `Weapon`, у которого поля `model`, `caliber`, `price_per_shot`. Реализуйте паттерн «Адаптер»: класс `WeaponAdapter`, который принимает словарь и предоставляет интерфейс `Weapon` — свойства `model`, `caliber`, `price_per_shot` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Weapon` — например, передать его в функцию, которая работает с оружием. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В тире есть базовый класс `RangeItem`, и постепенно в нём накопились методы, которые нужны не всем объектам: `shoot()` (только для оружия), `reload()` (только для огнестрельного оружия), `aim()` (только для прицеливаемых объектов). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Shootable`, `Reloadable`, `Aimable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `SessionService`, который зависит не от конкретных классов (`Pistol`, `Rifle`), а от абстракции `Shootable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип оружия (например, `Crossbow`, реализующий только `Shootable`) без изменения `SessionService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

# 📚 Вариант 29. Квест-комната (полный курс, 40 задач)

> **Тема:** учёт сценариев, комнат, реквизита, бронирований и всей системы квест-комнаты.
> **Цель:** пройти путь от создания первого класса до архитектуры реального проекта.
> **Формат каждой задачи:** краткая теория → задание → что продемонстрировать → вопросы для размышления.
> **Совет:** после каждой задачи запускай код и сверяй вывод с ожидаемым — так закрепляется материал.

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В квест-комнате ведётся учёт сценариев. Создайте класс `Quest` с полями `title` (название), `difficulty` (сложность: лёгкий, средний, сложный) и `duration` (длительность в минутах). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о квесте в формате: `Название — сложность ..., длительность ... мин`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Тайна старого особняка — сложность средний, длительность 60 мин`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждой игры есть код брони, который нельзя менять или читать напрямую. Создайте класс `QuestBooking`, в котором код брони хранится в приватном поле `__booking_code`. Реализуйте метод `set_booking_code(value)`, предназначенный только для установки кода при бронировании, и метод `get_booking_code()` для чтения значения. Повторная попытка изменить код после бронирования должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте бронь, установите код, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `booking.__booking_code` — будет ошибка `AttributeError`, но через name mangling (`booking._QuestBooking__booking_code`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В квест-комнате есть разные типы квестов. Создайте базовый класс `QuestType` с методом `get_quest_type()`, возвращающим строку «Квест». Унаследуйте от него классы `HorrorQuest` и `DetectiveQuest`, переопределив в каждом метод `get_quest_type()` так, чтобы он возвращал «Хоррор-квест» и «Детективный квест» соответственно. Создайте по одному объекту каждого класса и выведите их тип, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Квест-комната рассчитывает стоимость игры для разных сценариев. Создайте классы `EasyQuest`, `MediumQuest` и `HardQuest`, у каждого из которых есть метод `price()`, возвращающий стоимость: для лёгкого — 3000, для среднего — 5000, для сложного — 8000. Напишите функцию `show_price(quest)`, которая принимает любой квест и выводит стоимость. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `QuestManager` (управляющий квестами), в котором есть три поля: `name` — публичное (доступно всем), `_room` — защищённое (комната, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_room` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_QuestManager__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Квест-комнате нужна утилита для проверки размера команды, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `QuestUtils` со статическим методом `is_valid_team(size)`, который возвращает `True`, если число игроков от 2 до 6, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `QuestUtils.is_valid_team(4)`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `QuestItem` с абстрактным методом `describe()`. Этот класс описывает общее понятие «объект квеста», но не может существовать сам по себе — у каждого конкретного объекта должно быть своё описание. Унаследуйте от него классы `Quest` и `Room`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `QuestItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждой квест-комнаты есть набор реквизита — комната не может работать без него, и реквизит создаётся вместе с комнатой. Создайте класс `Props` с методом `show()`, который выводит строку «Реквизит подготовлен». Создайте класс `Room`, который содержит объект класса `Props` внутри себя как поле (композиция) — объект `Props` создаётся прямо в конструкторе комнаты. Метод `show_room()` у комнаты должен выводить название квеста и вызывать метод `show()` у реквизита, демонстрируя, что комната «владеет» своим реквизитом. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В квест-комнате есть базовый класс `QuestItem` с методом `get_info()`, который возвращает строку «Объект квеста». Унаследуйте от него класс `HorrorQuest` и переопределите метод `get_info()` так, чтобы он возвращал строку «Хоррор-квест: <название>». Создайте объект класса `HorrorQuest`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Администратору квест-комнаты нужен метод для оформления игр, который мог бы работать в разных ситуациях: иногда известен только квест, иногда квест и команда, иногда ещё и время. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `BookingDesk` с методом `book(quest, team=None, time=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Забронирован квест: <title>», «Квест <title> забронирован для команды <team>» или «Квест <title> забронирован для команды <team> на <time>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Квесты часто выводят на экран — в каталоге, чеках, отчётах. По умолчанию `print(quest)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Quest` магический метод `__str__`, который возвращает строку вида «Квест "Тайна старого особняка" (средний, 60 мин)». Убедитесь, что при вызове `print(quest)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Quest('Тайна старого особняка', 'средний', 60)` — оно используется при выводе списка квестов и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В квест-комнате может оказаться несколько записей об одном и том же квесте — например, при повторной регистрации. Нужно, чтобы программа понимала, что это «один и тот же квест», и не дублировала его в каталоге. Реализуйте в классе `Quest` магический метод `__eq__`, чтобы два квеста считались равными, если совпадают их название и сложность. Дополнительно реализуйте `__hash__`, чтобы квесты можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых квеста попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В квест-комнате есть расписания, и иногда нужно объединять их (например, при объединении филиалов) или узнавать, сколько квестов в расписании. Создайте класс `Schedule` (расписание), в котором хранится список квестов. Реализуйте магический метод `__add__` для объединения двух расписаний в одно (возвращается **новое** расписание со всеми квестами) и метод `__len__`, возвращающий количество квестов. Продемонстрируйте работу: создайте два расписания, объедините их через `s3 = s1 + s2` и выведите `len(s3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В квест-комнате важно, чтобы длительность квеста была корректной — нельзя допустить отрицательное значение или ноль. Перепишите класс `Quest` так, чтобы поле `duration` было реализовано через свойство `@property`. Проверки: длительность должна быть целым числом и строго больше нуля. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `quest.duration` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `quest.duration = -10` → ошибка, `quest.duration = 0` → ошибка, `quest.duration = "abc"` → ошибка, `quest.duration = 60` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В квест-комнату часто поступают данные о квестах в виде строк — например, из файла или базы данных: `"Тайна старого особняка;средний;60"`. Добавьте в класс `Quest` классовый метод `from_string(cls, data)`, который создаёт объект квеста из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание квеста через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Квест-комнате нужно вести общий счётчик всех квестов — сколько всего квестов зарегистрировано в системе. Добавьте в класс `Quest` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового квеста. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько квестов и выведите `Quest.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`quest.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В квест-комнате возможна ситуация, когда игрок пытается забронировать квест, которого нет в каталоге. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `QuestNotFoundError(Exception)` и используйте его в классе `QuestRoom` (квест-комната). Метод `find_quest(title)` должен возвращать квест, если он есть, и выбрасывать `QuestNotFoundError` с понятным сообщением (например, с указанием названия), если его нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В квест-комнате игрок проходит квесты — но квесты не принадлежат игроку, они существуют отдельно и принадлежат квест-комнате. Реализуйте класс `Player` (игрок), который содержит список пройденных квестов. Квесты при этом существуют отдельно от игрока — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждой комнаты есть ссылка на `QuestRoom`, в которой она находится. Продемонстрируйте, что при удалении игрока квесты остаются в квест-комнате. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В квест-комнате работает человек, который одновременно и актёр, и администратор — например, сотрудник, который и роли играет, и залы бронирует. Создайте два класса-роли: `Actor` (актёр) и `Admin` (администратор). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `ActorAdmin`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `ActorAdmin.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Quest` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Quest` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Quest` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Quest(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В квест-комнате может храниться огромное количество записей о квестах — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Quest` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Quest` атрибут `__slots__ = ("title", "difficulty", "duration")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `quest.age_limit = 18` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Quest` для полей `duration` (например, диапазон 15–180) и `max_players` (диапазон 2–10), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В квест-комнате должен быть только один реестр квестов — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `QuestRoom` будет только один экземпляр. Проверьте, что `QuestRoom()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `QuestRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Квест-комната регулярно пополняется разными типами квестов — хоррорами, детективами, приключениями. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `QuestFactory` с методом `create(kind, **kwargs)`, который по строке `"horror"`, `"detective"` или `"adventure"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному квесту каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Игроки хотят получать уведомления, когда открывается новый квест или освобождается слот. Реализуйте паттерн «Наблюдатель»: класс `QuestRoom` хранит список подписчиков (`Player`) и при появлении нового квеста уведомляет всех через метод `notify()`. Продемонстрируйте, что при появлении квеста все игроки получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки игрок больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `QuestRoom` со временем разросся: он и хранит квесты, и бронирует их, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `QuestRoom` на несколько классов: `QuestStorage` (хранение квестов и поиск по названию), `BookingService` (бронирование, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «игрок забронировал квест». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Выездные квесты (`OutdoorQuest`) похожи на обычные, но имеют дополнительные поля — локация, мобильный реквизит. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`OutdoorQuest(Quest)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`OutdoorQuest` содержит объект `Quest` внутри себя как поле `self.quest`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В квест-комнате нужно создавать копии карточек квестов — например, для архивного каталога. При этом у комнаты есть вложенный объект `Props`, и важно понимать, копируется ли он вместе с комнатой или остаётся общим. Реализуйте в классе `Room` методы `__copy__` и `__deepcopy__`. Создайте комнату с вложенным объектом `Props` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — реквизит общий (`shallow.props is room.props` → `True`), при глубоком — свой (`deep.props is room.props` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В квест-комнате есть разные объекты, у которых есть метод `describe()` — квесты, комнаты, игроки, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Quest`, но и с любым другим классом, у которого есть `describe()` (например, классом `Room`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В квест-комнате есть расписание `Schedule`, и хочется перебирать квесты в нём в цикле `for` — так же удобно, как список. Сейчас `for quest in schedule:` не работает. Реализуйте в классе `Schedule` магические методы `__iter__` и `__next__` так, чтобы по расписанию можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте расписание с несколькими квестами и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(schedule)` должен возвращать список всех квестов.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В расписании квесты должны быть доступны по индексу — как элементы списка: `schedule[0]` возвращает первый квест, `schedule[1] = quest` заменяет второй. Реализуйте в классе `Schedule` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `schedule[0]` — получение квеста по индексу, `schedule[1] = new_quest` — замена квеста, срезы `schedule[0:2]` — получение подсписка, `for quest in schedule` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `quest in schedule`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В квест-комнате есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных бронирований её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `BookingDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with BookingDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В квест-комнате есть разные способы расчёта стоимости игры: фиксированная (5000 ₽), по сложности (лёгкий — 3000, средний — 5000, сложный — 8000) и «для постоянных игроков» (скидка 15% от фиксированной). Реализуйте паттерн «Стратегия»: базовый класс `PricingStrategy` с методом `calculate(quest)`, и три наследника — `FixedPricing`, `DifficultyPricing`, `LoyaltyPricing`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `LoyaltyPricing`, чтобы вызвать реализацию `FixedPricing` и применить скидку. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В квест-комнате появился новый тип квеста — приключенческий (`AdventureQuest`). Если добавлять его через изменение базового класса `QuestItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `QuestItem` с методом `price()`; наследников `EasyQuest`, `MediumQuest`; функцию `process_item(item: QuestItem)`, которая работает с любым квестом через родительский интерфейс. Добавьте `AdventureQuest` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В квест-комнате есть служебный объект `PriceCalculator`, который должен уметь рассчитывать стоимость игры по сложности. Удобно, чтобы его можно было вызывать как функцию: `calculator(difficulty)`, а не `calculator.calculate(difficulty)`. Реализуйте в классе `PriceCalculator` магический метод `__call__(self, difficulty)`, который возвращает стоимость (например, 3000 для лёгкого, 5000 для среднего, 8000 для сложного). Продемонстрируйте, что объект можно вызывать как функцию: `calc = PriceCalculator(); calc("средний")`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В квест-комнате есть класс `Quest`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Quest` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования квестов по названию и сложности): два вызова `Quest("Тайна старого особняка", "средний", 60)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В квест-комнате есть базовый класс `QuestItem` с методом `get_info()`. Нужно добавлять к объекту дополнительную информацию — например, «новинка», «для опытных», «с актёром». При этом нельзя менять сам класс `QuestItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `QuestItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `NewDecorator` и `ActorDecorator`, которые можно комбинировать: `NewDecorator(ActorDecorator(quest))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В квест-комнате появилась внешняя система бронирования, которая возвращает данные о квестах в другом формате — например, в виде словаря `{"name": "Тайна старого особняка", "level": "средний", "minutes": 60}`. А весь остальной код квест-комнаты работает с классом `Quest`, у которого поля `title`, `difficulty`, `duration`. Реализуйте паттерн «Адаптер»: класс `QuestAdapter`, который принимает словарь и предоставляет интерфейс `Quest` — свойства `title`, `difficulty`, `duration` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Quest` — например, передать его в функцию, которая работает с квестом. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В квест-комнате есть базовый класс `QuestItem`, и постепенно в нём накопились методы, которые нужны не всем объектам: `play()` (только для квестов), `prepare()` (только для комнат), `book()` (только для бронируемых). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Playable`, `Preparable`, `Bookable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `BookingService`, который зависит не от конкретных классов (`HorrorQuest`, `DetectiveQuest`), а от абстракции `Bookable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип квеста (например, `AdventureQuest`, реализующий только `Bookable`) без изменения `BookingService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

## 🟢 Уровень 1. Новичок (задачи 1–10)

**Что изучается:** класс, объект, конструктор, инкапсуляция, наследование, полиморфизм, модификаторы доступа, статические и абстрактные методы, композиция, переопределение, имитация перегрузки.

**Задача 1 (создание класса, конструктор, метод).** В планетарии ведётся учёт шоу. Создайте класс `Show` с полями `title` (название), `topic` (тема: космос, звёзды, планеты) и `duration` (длительность в минутах). Реализуйте конструктор `__init__`, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который возвращает строку с информацией о шоу в формате: `Название — тема ..., длительность ... мин`. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: `Путешествие к Марсу — тема планеты, длительность 45 мин`. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр `self` в методах.

**Задача 2 (инкапсуляция).** У каждого билета в планетарий есть уникальный код, который нельзя менять или читать напрямую. Создайте класс `PlanetariumTicket`, в котором код билета хранится в приватном поле `__ticket_code`. Реализуйте метод `set_ticket_code(value)`, предназначенный только для установки кода при продаже, и метод `get_ticket_code()` для чтения значения. Повторная попытка изменить код после продажи должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте билет, установите код, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать `ticket.__ticket_code` — будет ошибка `AttributeError`, но через name mangling (`ticket._PlanetariumTicket__ticket_code`) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

**Задача 3 (наследование).** В планетарии есть разные типы шоу. Создайте базовый класс `SpaceShow` с методом `get_show_type()`, возвращающим строку «Шоу». Унаследуйте от него классы `StarShow` и `PlanetShow`, переопределив в каждом метод `get_show_type()` так, чтобы он возвращал «Шоу о звёздах» и «Шоу о планетах» соответственно. Создайте по одному объекту каждого класса и выведите их тип, чтобы убедиться, что каждый класс возвращает своё значение. В комментариях объясните, что произойдёт, если в наследнике не переопределять метод, и как вызвать родительскую версию через `super()`.

**Задача 4 (полиморфизм).** Планетарий рассчитывает стоимость билета для разных категорий посетителей. Создайте классы `AdultTicket`, `ChildTicket` и `StudentTicket`, у каждого из которых есть метод `price()`, возвращающий стоимость: для взрослого — 700, для детского — 350, для студенческого — 500. Напишите функцию `show_price(ticket)`, которая принимает любой билет и выводит его стоимость. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

**Задача 5 (модификаторы доступа).** Создайте класс `PlanetariumDirector` (директор планетария), в котором есть три поля: `name` — публичное (доступно всем), `_dome` — защищённое (купол, доступно внутри класса и наследникам), `__salary` — приватное (зарплата, доступна только внутри класса). Продемонстрируйте обращение к каждому из полей извне класса: к `name` — свободно, к `_dome` — можно, но не рекомендуется, к `__salary` — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (`_PlanetariumDirector__salary`) — так вы увидите, что «приватность» в Python условна. Объясните в комментариях, почему приватное поле защищает данные сотрудника и как это связано с принципом инкапсуляции.

**Задача 6 (статические методы).** Планетарию нужна утилита для проверки времени сеанса, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс `PlanetariumUtils` со статическим методом `is_valid_time(hour)`, который возвращает `True`, если час находится в диапазоне от 10 до 22, и `False` — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: `PlanetariumUtils.is_valid_time(15)`. Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `PlanetariumItem` с абстрактным методом `describe()`. Этот класс описывает общее понятие «объект планетария», но не может существовать сам по себе — у каждого конкретного объекта должно быть своё описание. Унаследуйте от него классы `Show` и `Telescope`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `PlanetariumItem` создать нельзя (будет ошибка `TypeError`), а объекты наследников — можно. Подсказка: используйте `from abc import ABC, abstractmethod`. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

**Задача 8 (композиция).** У каждого шоу есть проектор — шоу не может проходить без него, и проектор создаётся вместе с шоу. Создайте класс `Projector` с методом `show()`, который выводит строку «Проектор включён». Создайте класс `Show`, который содержит объект класса `Projector` внутри себя как поле (композиция) — объект `Projector` создаётся прямо в конструкторе шоу. Метод `show_show()` у шоу должен выводить название и вызывать метод `show()` у его проектора, демонстрируя, что шоу «владеет» своим проектором. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

**Задача 9 (переопределение методов).** В планетарии есть базовый класс `PlanetariumItem` с методом `get_info()`, который возвращает строку «Объект планетария». Унаследуйте от него класс `StarShow` и переопределите метод `get_info()` так, чтобы он возвращал строку «Шоу о звёздах: <название>». Создайте объект класса `StarShow`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

**Задача 10 (имитация перегрузки).** Кассиру планетария нужен метод для продажи билетов, который мог бы работать в разных ситуациях: иногда известно только шоу, иногда шоу и количество билетов, иногда ещё и категория посетителя. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `CashDesk` с методом `sell(show, count=None, category=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Продан билет на шоу: <title>», «Продано <count> билетов на шоу: <title>» или «Продано <count> билетов категории <category> на шоу <title>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

## 🔵 Уровень 2. Ученик (задачи 11–18)

**Что изучается:** магические методы (`__str__`, `__repr__`, `__eq__`, `__hash__`, `__add__`, `__len__`), свойства `@property`, классовые методы `@classmethod`, классовые и экземплярные атрибуты, собственные исключения, агрегация и ассоциация.

**Задача 11 (магические методы `__str__` и `__repr__`).** Шоу часто выводят на экран — в афише, билетах, отчётах. По умолчанию `print(show)` выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс `Show` магический метод `__str__`, который возвращает строку вида «Шоу "Путешествие к Марсу" (планеты, 45 мин)». Убедитесь, что при вызове `print(show)` выводится именно эта строка. Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Show('Путешествие к Марсу', 'планеты', 45)` — оно используется при выводе списка шоу и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

**Задача 12 (магические методы `__eq__` и `__hash__`).** В планетарии может оказаться несколько записей об одном и том же шоу — например, при повторном добавлении в репертуар. Нужно, чтобы программа понимала, что это «одно и то же шоу», и не дублировала его в афише. Реализуйте в классе `Show` магический метод `__eq__`, чтобы два шоу считались равными, если совпадают их название и тема. Дополнительно реализуйте `__hash__`, чтобы шоу можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых шоу попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.

**Задача 13 (магические методы `__add__` и `__len__`).** В планетарии есть программы, и иногда нужно объединять их (например, при объединении двух тематических блоков) или узнавать, сколько шоу в программе. Создайте класс `Program` (программа), в котором хранится список шоу. Реализуйте магический метод `__add__` для объединения двух программ в одну (возвращается **новая** программа со всеми шоу) и метод `__len__`, возвращающий количество шоу. Продемонстрируйте работу: создайте две программы, объедините их через `p3 = p1 + p2` и выведите `len(p3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

**Задача 14 (свойство `@property`).** В планетарии важно, чтобы длительность шоу была корректной — нельзя допустить отрицательное значение или ноль. Перепишите класс `Show` так, чтобы поле `duration` было реализовано через свойство `@property`. Проверки: длительность должна быть целым числом и строго больше нуля. При нарушении должно выбрасываться исключение `ValueError` с понятным сообщением. Покажите, что обращение к `show.duration` работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: `show.duration = -10` → ошибка, `show.duration = 0` → ошибка, `show.duration = "abc"` → ошибка, `show.duration = 45` → успех. Объясните, чем `@property` удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

**Задача 15 (классовый метод `@classmethod`).** В планетарий часто поступают данные о шоу в виде строк — например, из файла или базы данных: `"Путешествие к Марсу;планеты;45"`. Добавьте в класс `Show` классовый метод `from_string(cls, data)`, который создаёт объект шоу из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание шоу через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.

**Задача 16 (классовые vs экземплярные атрибуты).** Планетарию нужно вести общий счётчик всех шоу — сколько всего шоу зарегистрировано в системе. Добавьте в класс `Show` классовый атрибут `count = 0`, который увеличивается в конструкторе при создании каждого нового шоу. Убедитесь, что `count` общий для всех объектов класса, а не для отдельного экземпляра: создайте несколько шоу и выведите `Show.count`. Дополнительно покажите, что если попытаться изменить `count` через экземпляр (`show.count = 100`), создастся отдельный атрибут экземпляра, а классовый останется прежним. Объясните разницу между классовыми и экземплярными атрибутами и в каком порядке Python ищет атрибут при обращении `obj.attr`.

**Задача 17 (собственные исключения).** В планетарии возможна ситуация, когда посетитель пытается купить билет на шоу, которого нет в афише. Стандартные исключения тут не подходят — нужно своё, говорящее. Создайте собственное исключение `ShowNotFoundError(Exception)` и используйте его в классе `Planetarium` (планетарий). Метод `find_show(title)` должен возвращать шоу, если оно есть, и выбрасывать `ShowNotFoundError` с понятным сообщением (например, с указанием названия), если его нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

**Задача 18 (агрегация и ассоциация).** В планетарии посетитель посещает шоу — но шоу не принадлежат посетителю, они существуют отдельно и принадлежат планетарию. Реализуйте класс `Visitor` (посетитель), который содержит список посещённых шоу. Шоу при этом существуют отдельно от посетителя — это **агрегация**, а не композиция. Дополнительно реализуйте **ассоциацию**: у каждого шоу есть ссылка на `Planetarium`, в котором оно проходит. Продемонстрируйте, что при удалении посетителя шоу остаются в планетарии. Объясните разницу между композицией (жёсткая связь, часть не может существовать без целого), агрегацией (мягкая связь, часть существует отдельно) и ассоциацией (просто ссылка между независимыми объектами).

## 🟣 Уровень 3. Мастер (задачи 19–24)

**Что изучается:** множественное наследование, MRO, `@dataclass`, миксины, `__slots__`, дескрипторы, метаклассы, паттерн Singleton.

**Задача 19 (множественное наследование и MRO).** В планетарии работает человек, который одновременно и лектор, и оператор проектора — например, универсальный специалист. Создайте два класса-роли: `Lecturer` (лектор) и `Operator` (оператор). У каждого родителя есть метод `get_role()`, возвращающий свою роль. Затем создайте класс `LecturerOperator`, который наследуется от обоих. Продемонстрируйте порядок разрешения методов через `LecturerOperator.__mro__` и объясните, какой метод `get_role()` будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

**Задача 20 (`@dataclass` и миксин).** Класс `Show` содержит только данные и простые методы — его удобно переписать через декоратор `@dataclass`, чтобы не писать вручную конструктор, `__repr__` и `__eq__`. Перепишите класс `Show` через `@dataclass` и сравните с версией из задачи 1. Дополнительно создайте миксин `LoggableMixin` с методом `log()`, который выводит имя класса и текущее время (используйте `from datetime import datetime`). Подключите миксин к классу `Show` через множественное наследование, причём миксин должен идти **первым** в списке родителей: `class Show(LoggableMixin): ...`. Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

**Задача 21 (`__slots__`).** В планетарии может храниться огромное количество записей о шоу — десятки тысяч. При таком объёме важна экономия памяти: каждый объект `Show` по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс `Show` атрибут `__slots__ = ("title", "topic", "duration")`. Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, `show.age_limit = 6` вызовет `AttributeError`). Сравните размер объекта с `__slots__` и без него через `sys.getsizeof` — но учтите, что `getsizeof` показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте `pympler.asizeof` или оцените по количеству полей. Объясните, как `__slots__` экономит память, что теряется при его использовании (наследование, `__dict__`, `@property`) и в каких случаях его стоит применять.

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Show` для полей `duration` (например, диапазон 5–180) и `seats` (диапазон 10–500), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.

**Задача 23 (метаклассы).** В планетарии должен быть только один реестр шоу — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `Planetarium` будет только один экземпляр. Проверьте, что `Planetarium()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

**Задача 24 (паттерн Singleton).** Реализуйте класс `ShowRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

## 🟠 Уровень 4. Архитектор (задачи 25–30)

**Что изучается:** паттерны Factory и Observer, принцип SRP из SOLID, выбор между композицией и наследованием, копирование объектов, утиная типизация.

**Задача 25 (паттерн Factory).** Планетарий регулярно пополняется разными типами шоу — о звёздах, о планетах, о космосе. Создавать их напрямую через конструкторы неудобно: код разрастается, легко ошибиться. Создайте класс `ShowFactory` с методом `create(kind, **kwargs)`, который по строке `"star"`, `"planet"` или `"space"` создаёт соответствующий объект и возвращает его. Неизвестный тип должен вызывать `ValueError`. Продемонстрируйте работу фабрики: создайте по одному шоу каждого типа через фабрику. Объясните, почему фабрика удобнее прямого вызова конструкторов и как этот паттерн связан с принципом Open/Closed (см. задачу 35).

**Задача 26 (паттерн Observer).** Посетители хотят получать уведомления, когда появляется новое шоу или открывается предпродажа. Реализуйте паттерн «Наблюдатель»: класс `Planetarium` хранит список подписчиков (`Visitor`) и при появлении нового шоу уведомляет всех через метод `notify()`. Продемонстрируйте, что при появлении шоу все посетители получают уведомление. Дополнительно реализуйте `unsubscribe()` и покажите, что после отписки посетитель больше не получает уведомлений. Объясните, где в реальных проектах используется этот паттерн (GUI, событийные системы, мессенджеры, pub/sub) и как он помогает избежать жёсткой связи между объектами.

**Задача 27 (принцип SOLID — SRP).** Класс `Planetarium` со временем разросся: он и хранит шоу, и продаёт билеты, и отправляет уведомления, и формирует отчёты. Это нарушает принцип единственной ответственности (SRP) — у класса должна быть одна причина для изменения. Разбейте класс `Planetarium` на несколько классов: `ShowStorage` (хранение шоу и поиск по названию), `TicketService` (продажа билетов, ведение учёта), `NotificationService` (уведомления). Продемонстрируйте работу собранных сервисов на сценарии «посетитель купил билет». Объясните, почему это лучше одного большого класса, как понять, что у класса «одна причина для изменения», и приведите примеры «классов-богов» из реальной жизни.

**Задача 28 (композиция vs наследование).** Полнокупольные шоу (`FullDomeShow`) похожи на обычные шоу, но имеют дополнительные поля — разрешение проекции, число проекторов. Реализуйте одну и ту же задачу двумя способами: через **наследование** (`FullDomeShow(Show)` — переопределяет конструктор и добавляет свои поля) и через **композицию** (`FullDomeShow` содержит объект `Show` внутри себя как поле `self.show`). Сравните плюсы и минусы каждого подхода: наследование даёт переиспользование кода, но создаёт жёсткую связь; композиция гибче, но требует больше кода. Объясните, когда что выбирать, и приведите примеры из реальных проектов и стандартной библиотеки Python.

**Задача 29 (копирование объектов).** В планетарии нужно создавать копии карточек шоу — например, для архивного каталога. При этом у шоу есть вложенный объект `Projector`, и важно понимать, копируется ли он вместе с шоу или остаётся общим. Реализуйте в классе `Show` методы `__copy__` и `__deepcopy__`. Создайте шоу с вложенным объектом `Projector` и продемонстрируйте разницу между поверхностным (`copy.copy`) и глубоким (`copy.deepcopy`) копированием: при поверхностном — проектор общий (`shallow.projector is show.projector` → `True`), при глубоком — свой (`deep.projector is show.projector` → `False`). Объясните, когда какой тип копирования использовать и что делать с циклическими ссылками.

**Задача 30 (утиная типизация).** В планетарии есть разные объекты, у которых есть метод `describe()` — шоу, телескопы, лекторы, а может быть, и что-то ещё, что появится в будущем. Напишите функцию `describe_item(item)`, которая вызывает метод `describe()` у любого объекта. Продемонстрируйте, что функция работает не только с `Show`, но и с любым другим классом, у которого есть `describe()` (например, классом `Telescope`), — без наследования от общего родителя. Объясните принцип утиной типизации: «если это выглядит как утка, плавает как утка и крякает как утка — это утка». Расскажите, как этот принцип отличает Python от языков со строгой типизацией (Java, C++) и какие у него минусы (ошибка видна только во время выполнения).

## 🔴 Уровень 5. Эксперт (задачи 31–35)

**Что изучается:** протоколы итерации и индексации (`__iter__`, `__next__`, `__getitem__`, `__setitem__`), контекстные менеджеры (`__enter__`, `__exit__`), паттерн Strategy, `super()`, принципы OCP и LSP из SOLID.

**Задача 31 (итерация по объекту: `__iter__` и `__next__`).** В планетарии есть программа `Program`, и хочется перебирать шоу в ней в цикле `for` — так же удобно, как список. Сейчас `for show in program:` не работает. Реализуйте в классе `Program` магические методы `__iter__` и `__next__` так, чтобы по программе можно было итерироваться **несколько раз подряд** (это важно: `__iter__` должен возвращать отдельный объект-итератор или генератор, а **не** `self`, иначе повторная итерация сломается). Создайте программу с несколькими шоу и переберите их в цикле `for` дважды подряд. Объясните, чем отличается итератор от итерируемого объекта, зачем нужны оба метода и почему `list(program)` должен возвращать список всех шоу.

**Задача 32 (индексация объекта: `__getitem__` и `__setitem__`).** В программе шоу должны быть доступны по индексу — как элементы списка: `program[0]` возвращает первое шоу, `program[1] = show` заменяет второе. Реализуйте в классе `Program` магические методы `__getitem__` и `__setitem__`. Продемонстрируйте, что теперь работают: `program[0]` — получение шоу по индексу, `program[1] = new_show` — замена шоу, срезы `program[0:2]` — получение подсписка, `for show in program` — итерация (за счёт `__getitem__`, даже без `__iter__`). Дополнительно реализуйте `__delitem__` (удаление по индексу) и `__contains__` (проверка `show in program`). Объясните, как `__getitem__` делает объект «похожим» на список или словарь и почему важно, чтобы он выбрасывал именно `IndexError`, а не какое-то своё исключение.

**Задача 33 (контекстный менеджер: `__enter__` и `__exit__`).** В планетарии есть операции, которые нужно выполнять «аккуратно»: например, при работе с базой данных билетов её нужно открыть в начале и обязательно закрыть в конце — даже если произошла ошибка. Реализуйте класс `TicketDatabase` как контекстный менеджер: методы `__enter__` (открывает базу и возвращает её) и `__exit__` (закрывает базу, даже если было исключение). Продемонстрируйте использование через `with TicketDatabase() as db:`. Дополнительно покажите, что происходит, если внутри `with` возникнет исключение, — база всё равно закроется. Объясните, зачем нужны контекстные менеджеры, как они связаны с конструкцией `try/finally`, что делают аргументы `exc_type`, `exc_value`, `traceback` в `__exit__` и что нужно вернуть, чтобы «проглотить» исключение.

**Задача 34 (паттерн Strategy + `super()`).** В планетарии есть разные способы расчёта стоимости билета: фиксированная (700 ₽), по категории посетителя (взрослый — 700, ребёнок — 350, студент — 500) и «для групп» (скидка 20% от фиксированной). Реализуйте паттерн «Стратегия»: базовый класс `TicketPricingStrategy` с методом `calculate(visitor)`, и три наследника — `FixedPricing`, `CategoryPricing`, `GroupPricing`. Каждый наследник переопределяет `calculate`. Дополнительно используйте `super()` в классе `GroupPricing`, чтобы вызвать реализацию `FixedPricing` и применить скидку. Продемонстрируйте работу всех трёх стратегий через единый интерфейс. Объясните, что такое паттерн «Стратегия», зачем он нужен, как он убирает длинные цепочки `if/elif` и как `super()` помогает переиспользовать код родителя.

**Задача 35 (принципы OCP и LSP из SOLID).** В планетарии появился новый тип шоу — лазерное (`LaserShow`). Если добавлять его через изменение базового класса `PlanetariumItem`, это нарушит принцип **OCP (Open/Closed Principle)** — классы должны быть открыты для расширения, но закрыты для изменения. Дополнительно нужно проверить принцип **LSP (Liskov Substitution Principle)** — объект наследника должен без проблем заменять объект родителя. Реализуйте: базовый класс `PlanetariumItem` с методом `price()`; наследников `StarShow`, `PlanetShow`; функцию `process_item(item: PlanetariumItem)`, которая работает с любым шоу через родительский интерфейс. Добавьте `LaserShow` **без изменения** существующего кода — это и есть OCP. Затем проверьте, что `process_item` одинаково хорошо работает со всеми тремя типами, — это LSP. Объясните оба принципа и приведите примеры их нарушения (классический пример нарушения LSP: «квадрат — не прямоугольник», если сеттеры ширины и высоты меняют оба измерения).

## ⚫ Уровень 6. Гуру (задачи 36–40)

**Что изучается:** объект как функция (`__call__`), разница между `__new__` и `__init__`, паттерны Decorator и Adapter, принципы ISP и DIP из SOLID.

**Задача 36 (объект как функция: `__call__`).** В планетарии есть служебный объект `TicketCalculator`, который должен уметь рассчитывать стоимость билета по категории посетителя. Удобно, чтобы его можно было вызывать как функцию: `calculator(category)`, а не `calculator.calculate(category)`. Реализуйте в классе `TicketCalculator` магический метод `__call__(self, category)`, который возвращает стоимость (например, 350 для ребёнка, 500 для студента, 700 для взрослого). Продемонстрируйте, что объект можно вызывать как функцию: `calc = TicketCalculator(); calc("студент")`. Дополнительно покажите, что такой объект можно передать туда, где ожидается функция-колбэк. Объясните, зачем нужен `__call__`, чем он полезен (функторы, декораторы-классы, колбэки с состоянием) и как он связан с понятием «функтор» в других языках.

**Задача 37 (`__new__` vs `__init__`).** В планетарии есть класс `Show`, и нужно понять, в каком порядке Python создаёт объект. Реализуйте в классе `Show` оба метода — `__new__` и `__init__` — и добавьте в каждый отладочный вывод (`print`), чтобы увидеть порядок вызова. Продемонстрируйте, что `__new__` вызывается **до** `__init__` и отвечает за создание объекта, а `__init__` — за его инициализацию. Дополнительно реализуйте `__new__`, который возвращает уже существующий объект (например, для кэширования шоу по названию и теме): два вызова `Show("Путешествие к Марсу", "планеты", 45)` должны вернуть **один и тот же** объект. Объясните разницу между `__new__` и `__init__`, зачем вообще нужен `__new__` и почему «`__new__` + кэш» легко приводит к странному поведению.

**Задача 38 (паттерн Decorator).** В планетарии есть базовый класс `PlanetariumItem` с методом `get_info()`. Нужно добавлять к шоу дополнительную информацию — например, «премьера», «для детей», «с живой лекцией». При этом нельзя менять сам класс `PlanetariumItem`. Реализуйте паттерн «Декоратор»: класс `ItemDecorator`, который содержит объект `PlanetariumItem` внутри себя и имеет тот же метод `get_info()`, но добавляет к результату дополнительный текст. Создайте декораторы `PremiereDecorator` и `KidsDecorator`, которые можно комбинировать: `PremiereDecorator(KidsDecorator(show))`. Продемонстрируйте работу, а также покажите, что смена порядка декораторов меняет порядок меток в выводе. Объясните, чем паттерн «Декоратор» отличается от наследования и почему он удобнее, когда нужно много комбинаций украшений (иначе пришлось бы создавать 2^n подклассов).

**Задача 39 (паттерн Adapter).** В планетарии появилась внешняя система бронирования, которая возвращает данные о шоу в другом формате — например, в виде словаря `{"name": "Путешествие к Марсу", "subject": "планеты", "minutes": 45}`. А весь остальной код планетария работает с классом `Show`, у которого поля `title`, `topic`, `duration`. Реализуйте паттерн «Адаптер»: класс `ShowAdapter`, который принимает словарь и предоставляет интерфейс `Show` — свойства `title`, `topic`, `duration` и метод `info()`. Продемонстрируйте, что адаптер можно использовать там, где ожидается `Show` — например, передать его в функцию, которая работает с шоу. Объясните, что такое паттерн «Адаптер», зачем он нужен, чем отличается от Facade и Decorator, и как он помогает интегрировать несовместимые интерфейсы без изменения существующего кода.

**Задача 40 (принципы ISP и DIP из SOLID).** В планетарии есть базовый класс `PlanetariumItem`, и постепенно в нём накопились методы, которые нужны не всем объектам: `play()` (только для шоу), `observe()` (только для телескопов), `book()` (только для бронируемых). Это нарушает принцип **ISP (Interface Segregation Principle)** — клиент не должен зависеть от методов, которые он не использует. Разделите интерфейс на несколько маленьких: `Playable`, `Observable`, `Bookable`, и пусть каждый класс реализует только нужные. Дополнительно проверьте принцип **DIP (Dependency Inversion Principle)** — модули высокого уровня не должны зависеть от модулей низкого уровня; оба должны зависеть от абстракций. Реализуйте класс `TicketService`, который зависит не от конкретных классов (`StarShow`, `PlanetShow`), а от абстракции `Playable` (или от `Protocol`). Продемонстрируйте, что можно легко добавить новый тип шоу (например, `LaserShow`, реализующий только `Playable`) без изменения `TicketService`. Объясните оба принципа, приведите примеры их нарушения и расскажите, почему DIP помогает тестировать код (легко подменять зависимости на mock).

# Вариант 31. Цирк

**Задача 1 (создание класса, конструктор, метод).** В цирке ведётся учёт артистов. Создайте класс `Artist` с полями `name` (имя), `genre` (жанр: акробат, клоун, гимнаст) и `experience` (стаж в годах). Реализуйте конструктор, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который выводит информацию об артисте в формате: Имя — жанр ..., стаж ... лет. Создайте один экземпляр класса и вызовите метод.

**Задача 2 (инкапсуляция).** У каждого артиста есть номер контракта, который нельзя менять или читать напрямую. Создайте класс `Contract`, в котором номер контракта хранится в приватном поле `__contract_id`. Реализуйте метод `set_contract_id(value)`, предназначенный только для установки номера при подписании контракта, и метод `get_contract_id()` для чтения значения. Повторная попытка изменить номер после подписания должна игнорироваться.

**Задача 3 (наследование).** В цирке выступают разные категории артистов. Создайте базовый класс `Performer` с методом `get_role()`, возвращающим строку «Артист». Унаследуйте от него классы `Acrobat` и `Clown`, переопределив в каждом метод `get_role()` так, чтобы он возвращал «Акробат» и «Клоун» соответственно. Создайте по одному объекту каждого класса и выведите их роль.

**Задача 4 (полиморфизм).** Цирк рассчитывает длительность номера для разных жанров. Создайте классы `Acrobat`, `Clown` и `Magician`, у каждого из которых есть метод `act_duration()`, возвращающий длительность в минутах: для акробата — 15, для клоуна — 10, для фокусника — 20. Напишите функцию `show_duration(artist)`, которая принимает любого артиста и выводит длительность номера. Убедитесь, что функция одинаково работает со всеми тремя типами объектов.

**Задача 5 (модификаторы доступа).** Создайте класс `CircusDirector` (директор цирка), в котором есть три поля: `name` — публичное, `_arena` — защищённое (арена), `__salary` — приватное (зарплата). Продемонстрируйте обращение к каждому из полей извне класса. Объясните в комментариях, почему к полю `__salary` нельзя обратиться напрямую и как это защищает данные сотрудника.

**Задача 6 (статические методы).** Цирку нужна утилита для проверки возраста зрителя. Создайте класс `CircusUtils` со статическим методом `is_child(age)`, который возвращает `True`, если зрителю меньше 12 лет, и `False` — в противном случае. Вызовите метод, не создавая объекта класса. Создайте класс `ClimbingUtils` со статическим методом `is_adult(age)`, который возвращает `True`, если клиенту не меньше 16 лет, и `False` — в противном случае. Вызовите метод, не создавая объекта класса. Поясните, чем статический метод отличается от обычного.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `ClimbingItem` с абстрактным методом `describe()`. Унаследуйте от него классы `ClimbingRoute` и `Instructor`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `ClimbingItem` создать нельзя, а объекты наследников — можно.

**Задача 8 (композиция).** У каждого клиента есть страховочная система. Создайте класс `Harness` с методом `show()`, который выводит строку «Страховочная система застёгнута». Создайте класс `Climber`, который содержит объект класса `Harness` внутри себя как поле. Метод `show_climber()` у клиента должен выводить имя и вызывать метод `show()` у страховочной системы.

**Задача 9 (переопределение методов).** На скалодроме есть базовый класс `ClimbingItem` с методом `get_info()`, который возвращает строку «Объект скалодрома». Унаследуйте от него класс `HardRoute` и переопределите метод `get_info()` так, чтобы он возвращал строку «Сложная трасса: <название>». Создайте объект класса `HardRoute`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского.

**Задача 10 (перегрузка).** Администратору скалодрома нужен метод для оформления восхождений. Создайте класс `ClimbDesk` с методом `book(...)`, который принимает разное количество аргументов: при вызове `book(climber)` оформляется восхождение, при `book(climber, route)` — восхождение по конкретной трассе, при `book(climber, route, instructor)` — восхождение с инструктором. Метод должен выводить соответствующее сообщение в зависимости от количества переданных аргументов.

# Вариант 40. Верёвочный парк

**Задача 1 (создание класса, конструктор, метод).** В верёвочном парке ведётся учёт маршрутов. Создайте класс `RopeRoute` с полями `name` (название), `level` (уровень: детский, средний, экстрим) и `length` (длина в метрах). Реализуйте конструктор, который принимает эти три значения и сохраняет их в объекте. Добавьте метод `info()`, который выводит информацию о маршруте в формате: Название — уровень ..., длина ... м. Создайте один экземпляр класса и вызовите метод.

**Задача 2 (инкапсуляция).** У каждого участника парка есть номер страховочного троса, который нельзя менять или читать напрямую. Создайте класс `SafetyRope`, в котором номер троса хранится в приватном поле `__rope_id`. Реализуйте метод `set_rope_id(value)`, предназначенный только для установки номера при регистрации участника, и метод `get_rope_id()` для чтения значения. Повторная попытка изменить номер после регистрации должна игнорироваться.

**Задача 3 (наследование).** В верёвочном парке есть разные типы препятствий. Создайте базовый класс `Obstacle` с методом `get_obstacle_type()`, возвращающим строку «Препятствие». Унаследуйте от него классы `RopeBridge` и `ZipLine`, переопределив в каждом метод `get_obstacle_type()` так, чтобы он возвращал «Верёвочный мост» и «Зип-лайн» соответственно. Создайте по одному объекту каждого класса и выведите их тип.

**Задача 4 (полиморфизм).** Верёвочный парк рассчитывает время прохождения для разных маршрутов. Создайте классы `KidsRoute`, `MiddleRoute` и `ExtremeRoute`, у каждого из которых есть метод `passing_time()`, возвращающий время в минутах: для детского — 20, для среднего — 40, для экстрима — 60. Напишите функцию `show_time(route)`, которая принимает любой маршрут и выводит время прохождения. Убедитесь, что функция одинаково работает со всеми тремя типами объектов.

**Задача 5 (модификаторы доступа).** Создайте класс `RopeParkManager` (управляющий парка), в котором есть три поля: `name` — публичное, `_zone` — защищённое (зона), `__salary` — приватное (зарплата). Продемонстрируйте обращение к каждому из полей извне класса. Объясните в комментариях, почему к полю `__salary` нельзя обратиться напрямую и как это защищает данные сотрудника.

**Задача 6 (статические методы).** Верёвочному парку нужна утилита для проверки веса участника. Создайте класс `RopeParkUtils` со статическим методом `is_valid_weight(kg)`, который возвращает `True`, если вес не превышает 100 кг, и `False` — в противном случае. Вызовите метод, не создавая объекта класса. Поясните, чем статический метод отличается от обычного.

**Задача 7 (абстрактные методы).** Создайте абстрактный класс `RopeParkItem` с абстрактным методом `describe()`. Унаследуйте от него классы `RopeRoute` и `Instructor`, каждый из которых обязан реализовать метод `describe()`, возвращающий краткое описание. Проверьте, что объект абстрактного класса `RopeParkItem` создать нельзя, а объекты наследников — можно.

**Задача 8 (композиция).** У каждого участника есть карабин. Создайте класс `Carabiner` с методом `show()`, который выводит строку «Карабин застёгнут». Создайте класс `Participant`, который содержит объект класса `Carabiner` внутри себя как поле. Метод `show_participant()` у участника должен выводить имя и вызывать метод `show()` у карабина.

**Задача 9 (переопределение методов).** В верёвочном парке есть базовый класс `RopeParkItem` с методом `get_info()`, который возвращает строку «Объект парка». Унаследуйте от него класс `ExtremeRoute` и переопределите метод `get_info()` так, чтобы он возвращал строку «Экстрим-маршрут: <название>». Создайте объект класса `ExtremeRoute`, вызовите метод `get_info()` и убедитесь, что используется версия из дочернего класса, а не из родительского.

**Задача 10 (перегрузка).** Инструктору парка нужен метод для регистрации участников. Создайте класс `CheckInDesk` с методом `register(...)`, который принимает разное количество аргументов: при вызове `register(participant)` регистрируется участник, при `register(participant, route)` — участник на конкретный маршрут, при `register(participant, route, instructor)` — участник с инструктором. Метод должен выводить соответствующее сообщение в зависимости от количества переданных аргументов.
