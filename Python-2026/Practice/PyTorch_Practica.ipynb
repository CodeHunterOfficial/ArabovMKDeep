## 🟣 Уровень 2. Autograd и градиенты (11–18)

**Что изучается:** `requires_grad`, `backward()`, `.grad`, `zero_grad`, вычислительный граф, `detach`, `torch.no_grad`, автодифференцирование.

**Задача 11 (`requires_grad`).** Создайте `x = torch.tensor(2.0, requires_grad=True)`. Вычислите `y = x**2 + 3*x`. Вызовите `y.backward()`. Выведите `x.grad`. **Задача 12 (`zero_grad`).** Вызовите `y.backward()` дважды. Выведите `x.grad`. Ожидаемый вывод: `14.0` (градиенты **сложились**). Теперь сделайте `x.grad.zero_()` и повторите — `7.0`. В комментариях поясните: почему PyTorch **накапливает** градиенты (для RNN, gradient accumulation); почему в цикле обучения **обязательно** `optimizer.zero_grad()`; что будет, если забыть (модель «взорвётся»). В комментариях поясните: что делает `detach` (отрезает от графа); зачем (когда нужен только forward, например target-сеть в RL); как это связано с `torch.no_grad()`. Подсказка: `z = y.detach()`.

**Задача 14 (`torch.no_grad`).** Выполните инференс под `with torch.no_grad(): y = x**2`. Проверьте `y.requires_grad`. Ожидаемый вывод: `False` — граф **не** строится. В комментариях поясните: зачем `no_grad` (экономия памяти при инференсе); почему **всегда** используйте его в eval/predict; чем отличается от `detach` (первый — контекст, второй — операция). Подсказка: `with torch.no_grad(): ...`.

**Задача 15 (вычислительный граф).** Создайте `x = torch.tensor(1.0, requires_grad=True)`, `y = x * 2`, `z = y * 3`. Выведите `z.grad_fn`. Ожидаемый вывод: `<MulBackward0>`.
