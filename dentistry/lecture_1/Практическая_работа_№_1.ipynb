**Интуитивная аналогия:**

> Модель анализа тональности — это как **эмоциональный термометр** для текста. Вы даёте ей фразу, и она определяет, негативная она, позитивная или нейтральная. В медицине это может помочь, например, автоматически оценивать удовлетворённость пациентов или выявлять тревожные сигналы в описаниях симптомов.

```python
!pip install transformers
```

4. Во второй ячейке напишите код для загрузки модели и анализа текста:

```python
from transformers import pipeline

# Загружаем модель для анализа тональности (русский язык)
classifier = pipeline("sentiment-analysis",
                      model="blanchefort/rubert-base-cased-sentiment")

# Пример медицинского текста (негативное описание состояния)
text = "Состояние пациента тяжёлое, выраженный болевой синдром, отрицательная динамика."

# Анализируем
result = classifier(text)
print(f"Текст: {text}")
print(f"Тональность: {result[0]['label']}")
print(f"Уверенность: {result[0]['score']*100:.1f}%")
```

5. Запустите ячейки (Shift+Enter).

6. **Проанализируйте результат.** Модель определила тональность текста (негативную, позитивную или нейтральную) и указала уверенность в процентах.
