## 1. ВВЕДЕНИЕ

Современная стоматология, как и вся клиническая медицина, уже не может опираться исключительно на субъективный опыт врача и визуальную оценку. Сегодня каждое утверждение о состоянии тканей полости рта, эффективности лечения или прогнозе заболевания требует **измерения и количественной оценки**. Именно данные превращают клинические впечатления в объективные, воспроизводимые факты, которые можно сравнивать, анализировать и использовать для принятия решений.

В повседневной практике врач-стоматолог сталкивается с множеством стандартизированных инструментов. Индекс гигиены полости рта **OHI-S** (упрощённый индекс гигиены Грина–Вермильона) оценивает количество зубного налёта и зубного камня на поверхностях шести индексных зубов. **Индекс кровоточивости десневой борозды (SBI)** по Мюллеману отражает воспалительную реакцию десны. **Пародонтальный индекс CPITN** (Community Periodontal Index of Treatment Needs) позволяет оценить распространённость и тяжесть заболеваний пародонта в популяции, определяя потребность в лечении. Глубина пародонтального кармана (PPD) и уровень клинического прикрепления (CAL) — ключевые количественные параметры при диагностике пародонтита. Для оценки кариеса используют индекс **КПУ** (сумма кариозных, пломбированных и удалённых зубов) или более детальную систему **ICDAS** (International Caries Detection and Assessment System). Боль и тревожность пациента перед вмешательством измеряются с помощью **визуально-аналоговой шкалы (VAS)** и **стоматологической шкалы тревожности (DAS, MDAS)**. Качество жизни, связанное со здоровьем полости рта, оценивается опросником **OHIP-14** (Oral Health Impact Profile). В ортодонтии используются индексы **ICON**, **PAR** для оценки сложности лечения и его результатов. В имплантологии — шкалы оценки стабильности имплантата (например, **ISQ** по данным резонансно-частотного анализа). В челюстно-лицевой хирургии — шкалы оценки боли, отёка, тризма.

Для врачей смежных специальностей также существуют свои шкалы. Неонатолог использует **шкалу Апгар** для оценки состояния новорождённого, **шкалу Сильвермана** для оценки дыхательных расстройств. Рентгенолог работает с классификациями **BI-RADS** (для молочной железы), **PI-RADS** (для предстательной железы), а также со шкалами оценки качества изображений. Терапевт применяет шкалы **SCORE** (оценка сердечно-сосудистого риска), **CHA₂DS₂-VASc** (риск инсульта при фибрилляции предсердий), **NYHA** (функциональный класс сердечной недостаточности). Медицинский микробиолог опирается на критерии **EUCAST** для определения чувствительности к антибиотикам, индексы микробной обсеменённости, шкалы оценки биоплёнкообразования. Все эти инструменты генерируют структурированные данные, которые могут быть использованы для статистического анализа и построения моделей искусственного интеллекта.

Понимание устройства этих данных даёт врачу ключевые профессиональные преимущества. Во-первых, корректная интерпретация клинических индексов и шкал возможна только при знании того, как они устроены и что означают полученные числа. Во-вторых, критическая оценка научных статей требует умения читать таблицы с описательными статистиками и понимать, о чём говорят цифры (средние, стандартные отклонения, доверительные интервалы). В-третьих, правильная организация данных создаёт базу для работы с моделями искусственного интеллекта — о чём подробно говорилось на предыдущей лекции.

**Ключевая мысль:** данные — это язык, на котором врач говорит с цифрами. Без этого языка невозможны ни серьёзная научная работа, ни обоснованная клиническая практика, ни внедрение современных цифровых технологий.

## 2. СТРОЕНИЕ ТАБЛИЦ ДАННЫХ

Любое клиническое исследование использует таблицы как основной формат хранения информации. Электронные медицинские карты (ЭМК) также организованы в табличные структуры: каждая строка — визит пациента, столбцы — симптомы, диагнозы, назначения, результаты обследований. Понимание этого формата — первый шаг в работе с данными.

### 2.1. Базовый «широкий» формат

В типичной таблице **строки** соответствуют отдельным пациентам (наблюдениям, случаям). Каждая строка описывает одного человека. **Столбцы** представляют собой переменные — характеристики, которые мы измеряем у каждого пациента: клинические параметры, результаты лабораторных исследований, демографические показатели, баллы по шкалам.

Рассмотрим конкретный пример. Предположим, мы собрали данные 30 пациентов с хроническим генерализованным пародонтитом, которым измеряли среднюю глубину пародонтального кармана (PPD, в мм) до лечения. Таблица будет содержать столбцы: `ID` (уникальный номер), `Пол`, `Возраст` (лет), `PPD_средний` (мм). Вот как выглядит фрагмент такой таблицы (полные данные приведены далее в разделе 4):

| ID | Пол | Возраст | PPD_средний (мм) |
|----|-----|---------|-------------------|
| 1  | М   | 34      | 3.8               |
| 2  | Ж   | 28      | 5.2               |
| 3  | М   | 45      | 4.1               |
| …  | …   | …       | …                 |
| 30 | Ж   | 53      | 6.5               |

Однако реальные клинические исследования редко ограничиваются одной переменной. В стоматологической практике мы собираем гораздо больше информации: демографию, несколько клинических индексов, групповые метки (например, метод лечения), а иногда и текстовые описания жалоб или рентгенологических находок. Поэтому таблица может быть значительно шире. Вот как выглядит расширенный вариант для тех же 30 пациентов (показаны первые 3 строки, но мы приводим полный набор столбцов):

| ID | Пол | Возраст | PPD_средний (мм) | Индекс гигиены OHI-S | Кровоточивость SBI (%) | Метод лечения | Курение | Текст: «Жалобы пациента» | Уровень боли VAS (0–10) |
|----|-----|---------|-------------------|----------------------|------------------------|----------------|---------|----------------------------|--------------------------|
| 1  | М   | 34      | 3.8               | 1.2                  | 25                     | Скейлинг       | Нет     | «Кровоточивость при чистке»| 3                        |
| 2  | Ж   | 28      | 5.2               | 2.1                  | 45                     | Лоскутная операция | Да  | «Подвижность зубов»      | 6                        |
| 3  | М   | 45      | 4.1               | 0.9                  | 15                     | Скейлинг       | Нет     | «Запах изо рта»           | 2                        |

Здесь мы видим столбцы разных типов:
- **Количественные:** возраст, PPD_средний, индекс гигиены OHI-S, кровоточивость SBI (%), уровень боли VAS.
- **Номинальные:** пол, метод лечения (скейлинг / лоскутная операция / лазерная терапия), курение (да/нет).
- **Порядковые:** степень тяжести пародонтита (лёгкая → средняя → тяжёлая) — не показана в таблице, но может быть добавлена.
- **Текстовые:** жалобы пациента — их потом придётся кодировать в категории (например, «кровоточивость», «подвижность», «неприятный запах»).

Все статистические пакеты (SPSS, R, Python) и библиотеки машинного обучения ожидают именно такой «прямоугольный» формат: строки — объекты (наблюдения), столбцы — признаки (переменные). Любое отклонение затрудняет анализ. Например, если данные одного пациента разнесены по нескольким строкам, программа не сможет корректно рассчитать корреляцию между переменными.

### 2.2. Другие структуры данных в клинической медицине

Помимо базового «широкого» формата (одна строка = один пациент), существуют и другие способы организации данных, которые используются для решения специфических исследовательских задач.

**«Длинный» формат (long format)** — это альтернативный способ организации данных, где каждое измерение записывается в отдельную строку. Если у вас есть несколько замеров для одного пациента (например, уровень боли VAS измерялся на 1-й, 3-й и 7-й день после операции), то в длинном формате у этого пациента будет **3 строки**. В каждой строке будет указан ID пациента, время замера и само значение. Этот формат незаменим для многих видов статистического анализа, например, для моделей с повторными измерениями (mixed-effects models).

Сравнение форматов:

| ID | Время (день) | Боль VAS | ← длинный формат |
|----|--------------|----------|------------------|
| 1  | 1            | 6        |                  |
| 1  | 3            | 4        |                  |
| 1  | 7            | 2        |                  |
| 2  | 1            | 7        |                  |
| 2  | 3            | 5        |                  |

Выбор между широким и длинным форматом зависит от цели анализа. Широкий формат удобен для чтения и интерпретации человеком — вы сразу видите все показатели одного пациента. Длинный формат предпочтителен для статистических моделей, особенно при работе с повторными измерениями, когда нужно учитывать зависимость наблюдений внутри одного человека (например, динамика боли после удаления зуба мудрости).

**Поперечные (cross-sectional) и продольные (longitudinal) данные** различаются по времени сбора. Поперечные данные — это «срез» в один момент времени, когда мы измеряем группу людей один раз. Это самый распространённый тип данных в клинических исследованиях. Пример из нашей лекции с 30 пациентами и их средними PPD — классические поперечные данные. Мы видим картину «здесь и сейчас», но не знаем, как она менялась.

Продольные данные — это данные, собранные у **одних и тех же** людей в **разные** моменты времени. Это позволяет изучать изменения, развитие и причинно-следственные связи. Пример: измерение глубины пародонтальных карманов у пациентов до лечения, через 3 месяца и через 6 месяцев после пародонтологического лечения. Только так мы можем ответить на вопрос: «Насколько эффективно лечение в динамике?» — поперечный срез этого не покажет.

**Структурированные и неструктурированные данные** различаются по форме представления. Структурированные данные — это чёткая таблица с числами или категориями, они легко интерпретируются и анализируются компьютерами. Это результаты клинических измерений, демографические данные. Неструктурированные данные — это информация без заранее заданной организации: описания рентгенограмм, протоколы операций, ответы на открытые вопросы, фотографии полости рта, аудиозаписи жалоб. Их анализ требует специальных методов, в том числе инструментов ИИ. Например, текстовая жалоба «кровоточивость при чистке» сама по себе не является структурой, но мы можем превратить её в структуру, закодировав в категорию «Гингивит/кровоточивость».

**Иерархические (многоуровневые) данные** возникают, когда данные имеют вложенную структуру. Например, пациенты (1-й уровень) внутри стоматологических клиник (2-й уровень) внутри регионов (3-й уровень). В клинической практике иерархические данные возникают, когда пациенты «вложены» в лечащих врачей, а врачи — в отделения или клиники. Это позволяет учитывать, что пациенты одного врача могут быть похожи друг на друга больше, чем пациенты разных врачей (из-за индивидуальной техники лечения, критериев отбора). Для анализа таких данных нужны специальные многоуровневые модели (multilevel modeling).

**Временные ряды (time-series data)** — это частный случай продольных данных, когда измерения проводятся очень часто (ежедневно, ежечасно) у одного объекта. Например, ежедневные записи уровня боли после имплантации, мониторинг температуры тела после операции, данные фотоплетизмографии во время седации. В стоматологии временные ряды используются для мониторинга состояния в динамике и раннего обнаружения осложнений (например, нарастание отёка или боли).

### 2.3. Схематическое представление

Схематически структура таблицы в широком формате выглядит так:

```mermaid
flowchart LR
    A[Таблица данных<br>широкий формат] --> B[Строки = Пациенты]
    A --> C[Столбцы = Переменные]
    B --> B1["Пациент 1 (ID=001)"]
    B --> B2["Пациент 2 (ID=002)"]
    B --> B3["..."]
    C --> C1["Количественные: возраст, PPD, индекс гигиены"]
    C --> C2["Категориальные: пол, метод лечения, курение"]
    C --> C3["Порядковые: тяжесть пародонтита"]
    C --> C4["Текстовые: жалобы, описания"]
```

Схематически переход от широкого к длинному формату выглядит так:

```mermaid
flowchart TD
    A[Широкий формат<br>1 строка = 1 пациент] --> B["ID 1 | день1 | день3 | день7"]
    A --> C["ID 2 | день1 | день3 | день7"]
    
    D[Длинный формат<br>1 строка = 1 измерение] --> E["ID 1 | время1 | значение"]
    D --> F["ID 1 | время2 | значение"]
    D --> G["ID 1 | время3 | значение"]
    D --> H["ID 2 | время1 | значение"]
```

### 2.4. Итог по структурам данных

Для первого знакомства с данными в стоматологии достаточно понимать базовый табличный формат. Другие структуры — это надстройки, которые используются для более сложных задач:

| Структура | Когда использовать | Пример из стоматологии |
|-----------|---------------------|--------------------------|
| Широкий формат | Для простого описания выборки | Таблица с PPD для 30 пациентов |
| Длинный формат | Для анализа повторных измерений | Уровень боли VAS у 50 пациентов на 4 визитах |
| Поперечные данные | Для изучения текущего состояния | Индекс КПУ у студентов в марте 2026 |
| Продольные данные | Для изучения изменений и развития | Динамика глубины карманов за 6 месяцев лечения |
| Структурированные | Для стандартного статистического анализа | Результаты индексов, демография |
| Неструктурированные | Для анализа текстов, изображений, видео | Протоколы операций, рентгенограммы, фото полости рта |
| Иерархические | Для учёта вложенности данных | Пациенты внутри клиник, пациенты внутри врачей |
| Временные ряды | Для частого мониторинга одного объекта | Ежедневные записи боли после имплантации |

Теперь, когда мы понимаем, как устроена таблица и какие бывают альтернативные структуры, перейдём к классификации переменных — это основа для грамотного анализа данных.

## 3. ТИПЫ ПЕРЕМЕННЫХ

Переменные в таблице делятся на три основных типа. От правильного определения типа зависит выбор статистического метода — это одно из самых частых мест ошибок у начинающих исследователей. Если вы перепутаете тип переменной, вы примените неправильный тест и получите неверные выводы. Поэтому разберём каждый тип подробно, с примерами из реальной клинической практики.

### 3.1. Количественные переменные

Это числа, имеющие смысл величины. Их можно складывать, вычитать, вычислять среднее. Они делятся на два подтипа.

**Непрерывные** — могут принимать любые значения в диапазоне, в том числе дробные. Примеры из стоматологии:
- **Глубина пародонтального кармана (PPD)** — измеряется в мм с точностью до десятых (например, 3.5 мм).
- **Уровень клинического прикрепления (CAL)** — расстояние от цементно-эмалевой границы до дна кармана.
- **Плотность костной ткани** по данным денситометрии (в г/см²).
- **Время анестезии** (в минутах) — сколько действует анестетик.
- **Температура тела** (°C) — при контроле послеоперационного периода.
- **Индекс зубного налёта (PI)** по Силнесс–Лоэ — может принимать значения от 0 до 3 с шагом 0.1 (непрерывная шкала).

**Дискретные** — принимают только целые значения. Примеры из стоматологической практики:
- **Количество кариозных зубов** (компонент «К» индекса КПУ).
- **Число удалённых зубов** (компонент «У»).
- **Количество имплантатов** у пациента.
- **Число посещений стоматолога** за год.
- **Количество зубов с подвижностью III степени**.

В нашем примере возраст, PPD_средний, индекс гигиены OHI-S, кровоточивость SBI (%) — количественные переменные (некоторые дискретны, но часто трактуются как непрерывные для статистических процедур).

### 3.2. Порядковые (ранговые) переменные

Значения упорядочены, но интервалы между ними неравны и не имеют числового смысла. Можно сказать «выше/ниже», но не «насколько выше». Нельзя вычислять среднее!

Примеры из стоматологии:
- **Степень тяжести пародонтита**: лёгкая → средняя → тяжёлая (по классификации AAP/EFP).
- **Степень подвижности зубов** по шкале Миллера: 0 → 1 → 2 → 3.
- **Стадия кариеса по ICDAS**: 0 (здоров) → 1 → 2 → ... → 6.
- **Класс по ASA** (анестезиологический риск): I, II, III, IV, V.
- **Уровень гигиены полости рта**: хорошая / удовлетворительная / неудовлетворительная.
- **Степень открывания рта** (тризм): лёгкий / умеренный / выраженный.

**Важное предостережение:** Визуально-аналоговая шкала боли (VAS) часто ошибочно считается чисто количественной. Строго говоря, это порядковая переменная, потому что расстояния между баллами 2 и 3 не обязательно равны расстоянию между 7 и 8 (восприятие боли субъективно и нелинейно). Однако в клинических исследованиях её часто трактуют как количественную, используя параметрические методы, что допустимо при достаточном размере выборки и нормальном распределении, но требует оговорок. Для строгих работ корректнее использовать непараметрические методы или медиану с IQR.

### 3.3. Номинальные (категориальные) переменные

Это метки без внутреннего порядка. Нельзя сказать «больше» или «меньше» — только «равно» или «не равно». Используются для группировки.

Примеры из стоматологии:
- **Пол**: М / Ж.
- **Диагноз по МКБ-10**: K02.1 (кариес дентина) / K05.3 (хронический пародонтит) / K04.0 (пульпит).
- **Метод лечения**: терапевтическое / хирургическое / ортопедическое / ортодонтическое.
- **Тип реставрации**: композит / амальгама / керамика / стеклоиономер.
- **Наличие вредных привычек**: курение (да/нет), бруксизм (да/нет).
- **Тип прикуса**: ортогнатический / дистальный / мезиальный / открытый / глубокий.
- **Группа крови** (для хирургических вмешательств): A, B, AB, 0.

В нашем примере «Пол», «Метод лечения» и «Курение» — номинальные переменные. Для машинного обучения такие переменные нужно кодировать (например, one-hot encoding), но это тема отдельной лекции.

### 3.4. Как текстовые данные становятся переменными

Текстовые ответы на открытые вопросы или записи в карте — это неструктурированные данные. Но их можно превратить в структурированные через процедуру **кодирования**.

Пример: пациенты описывают жалобы при первичном обращении:
- «Кровоточат дёсны при чистке зубов»
- «Зуб реагирует на холодное»
- «Шатается зуб»
- «Боль при накусывании»

Разрабатываем систему категорий:
1 — Гингивальные симптомы (кровоточивость, отёк)
2 — Гиперчувствительность (реакция на температурные раздражители)
3 — Подвижность зубов
4 — Боль при жевании

Тогда «Кровоточат дёсны...» получает код 1, «Зуб реагирует на холодное» — код 2, «Шатается зуб» — код 3, «Боль при накусывании» — код 4. Теперь это номинальная переменная «Тип жалобы». Так текстовые данные превращаются в данные, доступные для статистического анализа и ИИ.

### 3.5. Пример из клинического протокола

В протоколе обследования пациента перед имплантацией сочетаются все три типа переменных:

| Переменная | Тип | Пример значения |
|------------|-----|-----------------|
| Плотность костной ткани (HU) | Количественный (непрерывный) | 550 HU |
| Количество имплантатов | Количественный (дискретный) | 3 |
| Класс кости по Lekholm–Zarb | Порядковый | Тип II |
| Наличие сахарного диабета | Номинальный | Да / Нет |
| Статус курения | Номинальный | Курит / Не курит / Бросил |
| Тип адентии | Номинальный | Частичная / Полная |

Это показывает, что даже в одном документе мы работаем с разными типами данных, и каждый требует своего подхода к анализу.

### 3.6. Схематическое представление

Диаграмма типов переменных с примерами из стоматологии:

```mermaid
flowchart TD
    A[Типы переменных] --> B[Количественные]
    A --> C[Порядковые]
    A --> D[Номинальные]
    B --> B1["Непрерывные: PPD, CAL, плотность кости"]
    B --> B2["Дискретные: число кариозных зубов, имплантатов"]
    C --> C1["Тяжесть пародонтита, подвижность зубов, класс ASA"]
    D --> D1["Пол, диагноз, метод лечения, тип прикуса"]
```

**Резюме для врача:** если вы перепутаете тип переменной, вы примените неправильный статистический тест. Например, если вы посчитаете среднее для порядковой переменной «степень подвижности зубов» (1, 2, 3) и получите 2.3, это не будет иметь клинического смысла. Для таких данных используйте медиану и непараметрические тесты.

## 4. ОПИСАТЕЛЬНЫЕ СТАТИСТИКИ

Описательные статистики — это «визитная карточка» выборки. Без них любая научная статья неполноценна.

### 4.1. Зачем врачу описательные статистики?

Когда вы читаете статью и видите: «Средняя глубина пародонтального кармана в группе составила 5.2 мм (SD = 1.8)» — вы можете:
- Оценить, насколько тяжёлый пародонтит у участников.
- Понять, насколько группа однородна (если SD мало, все пациенты примерно одинаковы; если велико — есть разброс).
- Сравнить эти данные с вашей собственной практикой (например, у ваших пациентов средний PPD 4.1 мм).
- Решить, применимы ли результаты к вашим пациентам.

Статистики делятся на **меры центра** (что типично) и **меры разброса** (насколько данные отличаются).

### 4.2. Меры центральной тенденции

**Среднее арифметическое** — сумма всех значений, делённая на их количество. Чувствительно к выбросам.

*Пример из практики:* вы измеряете глубину пародонтальных карманов у 10 пациентов. Девять имеют значения от 2 до 4 мм, один пациент — 10 мм (тяжёлый агрессивный пародонтит). Среднее = (2+2.5+3+3+3.5+3.5+4+4+4+10)/10 = 39.5/10 = 3.95 мм. Это не отражает типичного пациента — у большинства значения около 3–4 мм, а один выброс завышает среднее. Если врач решит, что все пациенты имеют глубину около 4 мм, он может недооценить тяжесть состояния конкретного больного.

**Медиана** — значение, которое делит упорядоченный ряд пополам: 50% наблюдений ниже, 50% выше. Устойчива к выбросам. В примере выше сортируем: 2, 2.5, 3, 3, 3.5, 3.5, 4, 4, 4, 10. Центральные значения — 3.5 и 3.5, медиана = 3.5 мм. Это более честная оценка «типичного» пациента. Для порядковых данных и распределений с выбросами рекомендуется использовать именно медиану.

**Мода** — наиболее часто встречающееся значение. В примере выше мода = 3, 3.5, 4 (каждое встречается по 2 раза) — мультимодальное распределение. Мода уникальна тем, что это единственная мера центра, применимая к номинальным переменным. Если мы хотим узнать, какой тип реставрации чаще всего используется в клинике, мы не можем вычислить «среднюю реставрацию», но можем определить моду (например, композит).

### 4.3. Меры разброса

Меры центра говорят о «типичном» значении, но они не дают информации о том, насколько пациенты похожи друг на друга. Для этого нужны меры разброса. Если разброс мал — группа однородна, выводы можно применять с уверенностью. Если разброс велик — группа неоднородна, и среднее может быть обманчивым.

**Дисперсия** — средний квадрат отклонений от среднего. Единицы измерения возведены в квадрат, что делает её трудной для интерпретации. Если мы измеряем глубину кармана в мм, дисперсия будет в «квадратных миллиметрах» — абстрактно. Дисперсия важна для технических расчётов (например, в дисперсионном анализе), но врач в статье её почти не видит — вместо неё приводят стандартное отклонение. Запомните: дисперсия — это промежуточный технический шаг, а не инструмент для клинических выводов.

**Стандартное отклонение (σ, SD)** — корень из дисперсии. Возвращает нас к исходным единицам измерения (мм, баллам, годам). Это самая популярная мера разброса в научных статьях.

*Что показывает:* насколько в среднем значения отклоняются от среднего. Правило «68–95–99.7» (для нормального распределения):
- Около 68% наблюдений лежат в диапазоне **среднее ± 1σ**.
- Около 95% — в диапазоне **среднее ± 2σ**.

*Пример из практики:* вы читаете статью: «В группе из 50 пациентов с хроническим пародонтитом средняя глубина кармана составила 4.8 мм, SD = 1.5 мм».  
Что это значит?  
- 68% пациентов имеют глубину от 3.3 до 6.3 мм (4.8 − 1.5 = 3.3; 4.8 + 1.5 = 6.3).  
- 95% — от 1.8 до 7.8 мм (4.8 − 3 = 1.8; 4.8 + 3 = 7.8).  

Теперь вы понимаете, что группа неоднородна: есть пациенты с лёгким пародонтитом (1.8 мм) и с тяжёлым (7.8 мм). Если ваша клиника специализируется на лёгких случаях, вы не можете напрямую применить выводы этой статьи — выборка слишком широкая. Если же SD маленькое (например, 0.8), то все пациенты примерно одинаковы, и результаты применимы к узкой группе.

*Клинический смысл:* стандартное отклонение помогает оценить, насколько «типичен» средний показатель. Маленькое SD → среднее надёжно. Большое SD → среднее не очень показательно, нужно смотреть на медиану и распределение.

**Размах** = max − min. Показывает, какой диапазон значений охватывает выборка. Полезен, чтобы понять, есть ли в выборке экстремальные случаи (например, очень глубокая пародонтальная кармана, требующая особого внимания). Однако крайне чувствителен к выбросам: если у вас 29 пациентов с PPD от 2 до 6 мм и один пациент с 12 мм, размах станет 10 мм, хотя основная масса данных скучена. Поэтому размах не используют как надёжную меру — он лишь даёт общее представление о границах.

**Интерквартильный размах (IQR)** = Q3 − Q1, где Q1 — 25-й перцентиль, Q3 — 75-й перцентиль. Показывает разброс **средних 50%** данных. В отличие от размаха, IQR игнорирует экстремальные значения (выбросы), что делает его устойчивой мерой для асимметричных распределений.

*Пример из клиники:* в выборке измерений глубины карманов Q1 = 3.5 мм, Q3 = 5.8 мм, IQR = 2.3 мм. Это означает, что **половина пациентов** (с 25-го по 75-й перцентиль) имеют глубину от 3.5 до 5.8 мм — то есть умеренный пародонтит. А те, у кого глубина выше 5.8 мм (верхние 25%) — уже тяжёлая степень. IQR позволяет быстро понять, где находится «ядро» группы.

*Когда используют IQR:* в научных статьях его часто приводят вместе с медианой для порядковых или сильно скошенных данных. Например: «Ме (медиана) = 4.6 мм, IQR = 2.3 мм» — это краткая характеристика, которая говорит о центре и разбросе лучше, чем среднее и SD, если распределение несимметрично.

**Итоговое резюме для врача:**

| Мера | Что показывает | Когда использовать | Чего остерегаться |
|------|----------------|---------------------|-------------------|
| **Дисперсия** | Разброс в квадрате | Почти никогда в интерпретации | Трудна для понимания |
| **Стандартное отклонение** | Среднее отклонение от среднего | Для нормальных распределений | Не работает при выбросах |
| **Размах** | Весь диапазон | Для общего представления о границах | Чувствителен к выбросам |
| **IQR** | Разброс средних 50% | Для порядковых и скошенных данных | Не даёт информации о крайних значениях |

### 4.4. Пример №1: 30 пациентов с глубиной пародонтальных карманов (асимметричное распределение)

**Исходные данные — средняя глубина кармана (мм) у 30 пациентов (в порядке ID):**  
3.8, 5.2, 4.1, 4.8, 5.5, 3.5, 4.2, 5.0, 3.9, 4.6, 5.8, 4.4, 6.2, 3.2, 4.3, 4.9, 4.0, 5.6, 3.7, 4.7, 4.5, 3.4, 5.2, 4.4, 5.9, 3.1, 5.3, 6.3, 3.3, 5.7.

**Отсортированный ряд:**  
3.1, 3.2, 3.3, 3.4, 3.5, 3.7, 3.8, 3.9, 4.0, 4.1, 4.2, 4.3, 4.4, 4.4, 4.5, 4.6, 4.7, 4.8, 4.9, 5.0, 5.2, 5.2, 5.3, 5.5, 5.6, 5.7, 5.8, 5.9, 6.2, 6.3.

**Расчёты:**
- **Среднее** = (сумма всех значений) / 30. Сумма = 141.0 (проверьте: складываем все числа). 141.0 / 30 = **4.70**.
- **Медиана** = (значение на 15-й позиции + значение на 16-й позиции) / 2 = (4.5 + 4.6) / 2 = **4.55**.
- **Мода** — значения 4.4 и 5.2 встречаются дважды, остальные по одному разу → мода = 4.4 и 5.2 (бимодальное).
- **Дисперсия** ≈ 1.07 (рассчитаем точно: сумма квадратов отклонений от среднего = 32.1, делим на 30 = 1.07).
- **Стандартное отклонение** = √1.07 ≈ **1.03**.
- **Размах** = 6.3 − 3.1 = **3.2**.
- **IQR** = Q3 − Q1. Q1 (25-й перцентиль) — значение на позиции 8 (0.25*30=7.5 → берём 8-е) = 3.9. Q3 (75-й перцентиль) — значение на позиции 23 (0.75*30=22.5 → берём 23-е) = 5.3. IQR = 5.3 − 3.9 = **1.4**.

**Интерпретация для стоматолога:**

В выборке значительная вариативность: от 3.1 мм (начальный пародонтит) до 6.3 мм (тяжёлый пародонтит). Среднее (4.70) немного выше медианы (4.55) — это указывает на **положительную асимметрию**: несколько пациентов с очень глубокими карманами «тянут» среднее вверх. σ = 1.03 говорит, что большинство пациентов (около 68%) имеют глубину от 3.67 до 5.73 мм (4.70 ± 1.03). IQR = 1.4 показывает, что средние 50% пациентов сконцентрированы в диапазоне от 3.9 до 5.3 мм.

Если вы работаете в клинике, где преобладают пациенты с лёгкой и средней степенью пародонтита, ваша выборка будет отличаться — и это важно учитывать при интерпретации чужих исследований и сравнении своих результатов.

### 4.5. Пример №2: 30 пациентов с индексом гигиены OHI-S (нормальное распределение)

Те же 30 пациентов, но по упрощённому индексу гигиены Грина–Вермильона (OHI-S, баллы от 0 до 3). Этот пример демонстрирует, как выглядят данные с нормальным распределением.

**Баллы OHI-S:**  
1.8, 2.5, 1.9, 2.2, 2.7, 1.5, 2.0, 2.4, 1.7, 2.1, 2.8, 2.0, 2.9, 1.4, 2.0, 2.3, 1.8, 2.6, 1.6, 2.2, 2.1, 1.5, 2.4, 2.0, 2.7, 1.4, 2.5, 2.9, 1.5, 2.6.

**Отсортированный ряд:**  
1.4, 1.4, 1.5, 1.5, 1.5, 1.6, 1.7, 1.8, 1.8, 1.9, 2.0, 2.0, 2.0, 2.0, 2.1, 2.1, 2.2, 2.2, 2.3, 2.4, 2.4, 2.5, 2.5, 2.6, 2.6, 2.7, 2.7, 2.8, 2.9, 2.9.

**Расчёты:**
- **Среднее** = (сумма всех значений) / 30. Сумма = 63.0. 63.0 / 30 = **2.10**.
- **Медиана** = (значение на 15-й позиции + значение на 16-й позиции) / 2 = (2.0 + 2.1) / 2 = **2.05**.
- **Мода** = 2.0 (встречается 4 раза).
- **Дисперсия** ≈ 0.22.
- **Стандартное отклонение** = √0.22 ≈ **0.47**.
- **Размах** = 2.9 − 1.4 = **1.5**.
- **IQR** ≈ 0.7 (Q1 = 1.8, Q3 = 2.5).

**Интерпретация:**

Здесь среднее (2.10) и медиана (2.05) почти совпадают — это говорит о **симметричном распределении**, близком к нормальному. В отличие от PPD, здесь нет выраженной асимметрии. σ = 0.47 говорит, что большинство пациентов (около 68%) имеют индекс гигиены от 1.63 до 2.57. Это важный вывод: одни клинические параметры (например, OHI-S в общей популяции) часто распределены нормально, другие (глубина карманов в специализированной клинике) — скошены. Если бы мы смотрели только на среднее, мы бы не заметили эту разницу.

### 4.6. Ручной расчёт для подмножества (5 значений)

Чтобы понять арифметику, вычислим статистики для первых 5 значений PPD: **3.8, 5.2, 4.1, 4.8, 5.5**.

- **Среднее** = (3.8+5.2+4.1+4.8+5.5)/5 = 23.4/5 = **4.68**.
- **Медиана** (сортируем: 3.8, 4.1, 4.8, 5.2, 5.5) = **4.8**.
- **Размах** = 5.5 − 3.8 = **1.7**.
- **Дисперсия**: отклонения от 4.68: -0.88, 0.52, -0.58, 0.12, 0.82. Квадраты: 0.7744, 0.2704, 0.3364, 0.0144, 0.6724. Сумма = 2.068. Дисперсия = 2.068/5 = **0.4136**.
- **Стандартное отклонение** = √0.4136 ≈ **0.64**.

Этот ручной расчёт показывает, как «рождаются» цифры в научных статьях, и помогает понять, что стоит за каждой формулой.

### 4.7. Код для расчёта всех статистик (Python)

Код написан максимально прозрачно, чтобы студент понял логику, а не запоминал синтаксис. Каждая команда пояснена.

```python
import math  # для квадратного корня

# Данные (отсортированный список PPD)
data = [3.1, 3.2, 3.3, 3.4, 3.5, 3.7, 3.8, 3.9, 4.0, 4.1,
        4.2, 4.3, 4.4, 4.4, 4.5, 4.6, 4.7, 4.8, 4.9, 5.0,
        5.2, 5.2, 5.3, 5.5, 5.6, 5.7, 5.8, 5.9, 6.2, 6.3]
n = len(data)

# 1. Среднее арифметическое
mean = sum(data) / n
print("Среднее =", round(mean, 2))

# 2. Медиана
if n % 2 == 0:
    median = (data[n//2 - 1] + data[n//2]) / 2
else:
    median = data[n//2]
print("Медиана =", median)

# 3. Мода (подсчёт частот)
freq = {}
for x in data:
    freq[x] = freq.get(x, 0) + 1
max_freq = max(freq.values())
modes = [k for k, v in freq.items() if v == max_freq]
if len(modes) == n:
    print("Мода отсутствует")
else:
    print("Мода =", modes)

# 4. Дисперсия и стандартное отклонение
sum_sq = sum((x - mean) ** 2 for x in data)
variance = sum_sq / n
std_dev = math.sqrt(variance)
print("Дисперсия =", round(variance, 2))
print("Стандартное отклонение =", round(std_dev, 2))

# 5. Размах
print("Размах =", max(data) - min(data))

# 6. Интерквартильный размах (IQR)
mid = n // 2
lower = data[:mid]
upper = data[mid:]

def median_of_list(lst):
    m = len(lst)
    if m % 2 == 0:
        return (lst[m//2 - 1] + lst[m//2]) / 2
    else:
        return lst[m//2]

q1 = median_of_list(lower)
q3 = median_of_list(upper)
iqr = q3 - q1
print("Q1 =", q1, " Q3 =", q3, " IQR =", iqr)
```

Этот код можно использовать для любых числовых данных, просто подставив свой список.

### 5.1. Гистограмма — распределение глубины пародонтальных карманов

**Назначение:** показывает распределение одной количественной переменной. Помогает оценить форму распределения: нормальное, асимметричное, бимодальное.

**Клинический пример:** мы хотим увидеть, как распределена средняя глубина карманов (PPD) в нашей группе из 30 пациентов с пародонтитом. Есть ли асимметрия? Есть ли пациенты с очень глубокими карманами?

**Данные:** берём те же 30 значений PPD, которые использовали в разделе 4.4 (отсортированный ряд).

```python
import matplotlib.pyplot as plt
import seaborn as sns

ppd_data = [3.1, 3.2, 3.3, 3.4, 3.5, 3.7, 3.8, 3.9, 4.0, 4.1,
            4.2, 4.3, 4.4, 4.4, 4.5, 4.6, 4.7, 4.8, 4.9, 5.0,
            5.2, 5.2, 5.3, 5.5, 5.6, 5.7, 5.8, 5.9, 6.2, 6.3]

mean_ppd = 4.70
median_ppd = 4.55

plt.figure(figsize=(8,5))
plt.hist(ppd_data, bins=6, edgecolor='black', color='skyblue', alpha=0.7)
plt.axvline(mean_ppd, color='red', linestyle='dashed', linewidth=2,
            label=f'Среднее = {mean_ppd:.2f}')
plt.axvline(median_ppd, color='green', linestyle='dotted', linewidth=2,
            label=f'Медиана = {median_ppd:.2f}')
plt.title('Распределение глубины пародонтальных карманов (гистограмма)')
plt.xlabel('Средняя глубина кармана, мм')
plt.ylabel('Количество пациентов')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.show()
```

**Что мы видим:** столбцы смещены влево и вытянуты вправо — **положительная асимметрия**. Большинство пациентов имеют глубину от 3.5 до 5.5 мм, но есть несколько человек с глубиной выше 6 мм. Среднее (4.70) больше медианы (4.55), что подтверждает асимметрию. Такой график сразу подсказывает, что для описания этой выборки лучше использовать медиану и IQR, а не среднее.

### 5.2. Ящик с усами — сравнение групп лечения по глубине карманов

**Назначение:** компактно показывает распределение через квартили, медиану и выбросы. Особенно полезен для сравнения нескольких групп.

**Клинический пример:** мы разделили наших 30 пациентов на три группы по 10 человек в зависимости от метода лечения (скейлинг, лоскутная операция, лазерная терапия) и хотим сравнить, отличаются ли группы по глубине карманов через 3 месяца после лечения.

**Данные:** те же 30 значений PPD, но распределённые по группам (мы просто разбили исходный список на три части, предполагая, что после лечения значения могли измениться).

```python
ppd_scaling = [3.2, 3.5, 3.8, 4.0, 4.2, 4.4, 4.6, 4.8, 5.0, 5.5]
ppd_flap = [3.4, 3.6, 3.9, 4.1, 4.3, 4.5, 4.7, 4.9, 5.2, 5.8]
ppd_laser = [3.8, 4.0, 4.2, 4.4, 4.6, 4.8, 5.0, 5.3, 5.6, 6.2]

import pandas as pd
df_box = pd.DataFrame({
    'Метод лечения': ['Скейлинг']*10 + ['Лоскутная операция']*10 + ['Лазер']*10,
    'PPD': ppd_scaling + ppd_flap + ppd_laser
})

plt.figure(figsize=(8,6))
sns.boxplot(x='Метод лечения', y='PPD', data=df_box, palette=['lightblue', 'lightgreen', 'lightcoral'])
plt.title('Сравнение глубины карманов после разных методов лечения')
plt.xlabel('Метод лечения')
plt.ylabel('Средняя глубина кармана, мм')
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.show()
```

**Что мы видим:** медиана в группе скейлинга (~4.3) ниже, чем в группе лоскутной операции (~4.5) и лазера (~4.8). В группе лазера есть выброс (6.2) — возможно, один пациент не ответил на лечение. Boxplot наглядно показывает, что скейлинг может быть эффективнее для уменьшения глубины карманов, но для окончательного вывода нужна статистическая проверка (например, ANOVA или критерий Краскела–Уоллиса).

### 5.3. Диаграмма рассеяния — связь между PPD и индексом гигиены

**Назначение:** показывает связь между двумя количественными переменными. Каждая точка — один пациент. Позволяет увидеть корреляцию, кластеры и нелинейные паттерны.

**Клинический пример:** мы хотим проверить, связаны ли глубина пародонтальных карманов (PPD) и индекс гигиены (OHI-S) у наших 30 пациентов. Гипотеза: они положительно коррелируют (чем хуже гигиена, тем глубже карманы).

**Данные:** берём те же 30 значений PPD и соответствующие им значения OHI-S из раздела 4.5 (в порядке ID).

```python
ppd_scatter = [3.8, 5.2, 4.1, 4.8, 5.5, 3.5, 4.2, 5.0, 3.9, 4.6,
               5.8, 4.4, 6.2, 3.2, 4.3, 4.9, 4.0, 5.6, 3.7, 4.7,
               4.5, 3.4, 5.2, 4.4, 5.9, 3.1, 5.3, 6.3, 3.3, 5.7]

ohis_scatter = [1.8, 2.5, 1.9, 2.2, 2.7, 1.5, 2.0, 2.4, 1.7, 2.1,
                2.8, 2.0, 2.9, 1.4, 2.0, 2.3, 1.8, 2.6, 1.6, 2.2,
                2.1, 1.5, 2.4, 2.0, 2.7, 1.4, 2.5, 2.9, 1.5, 2.6]

import numpy as np
plt.figure(figsize=(8,6))
plt.scatter(ppd_scatter, ohis_scatter, alpha=0.7, color='darkblue')

z = np.polyfit(ppd_scatter, ohis_scatter, 1)
p = np.poly1d(z)
plt.plot(ppd_scatter, p(ppd_scatter), color='red', linestyle='dashed',
         label=f'Тренд: у = {z[0]:.2f}x + {z[1]:.2f}')

plt.title('Связь между глубиной карманов (PPD) и индексом гигиены (OHI-S)')
plt.xlabel('Средняя глубина кармана, мм')
plt.ylabel('Индекс гигиены OHI-S, баллы')
plt.legend()
plt.grid(linestyle='--', alpha=0.3)
plt.show()
```

**Что мы видим:** точки образуют облако, вытянутое снизу вверх слева направо — **положительная корреляция**. Чем глубже карманы, тем хуже гигиена. Коэффициент наклона (≈0.5) показывает, что при увеличении глубины кармана на 1 мм индекс гигиены растёт примерно на 0.5 балла. Разброс точек говорит о том, что связь не идеальна — есть и другие факторы, влияющие на гигиену (курение, генетика, системные заболевания).

### 5.4. Столбчатая диаграмма — сравнение средних значений индекса гигиены в группах

**Назначение:** сравнение средних значений по категориям. Идеально для экспериментальных данных, когда нужно показать различия между группами.

**Клинический пример:** мы хотим наглядно показать разницу в средних значениях индекса гигиены OHI-S между тремя группами пациентов с разной мотивацией к уходу (низкая, средняя, высокая).

**Данные:** сгенерируем средние значения для трёх групп (условные, для иллюстрации).

```python
ohis_low = [2.4, 2.6, 2.5, 2.7, 2.8, 2.9, 2.5, 2.6, 2.7, 2.8]
ohis_medium = [1.8, 1.9, 2.0, 2.1, 2.2, 2.0, 1.9, 2.1, 2.2, 2.3]
ohis_high = [1.2, 1.3, 1.4, 1.5, 1.6, 1.4, 1.5, 1.3, 1.6, 1.7]

mean_low = sum(ohis_low) / len(ohis_low)
mean_medium = sum(ohis_medium) / len(ohis_medium)
mean_high = sum(ohis_high) / len(ohis_high)

groups = ['Низкая мотивация', 'Средняя мотивация', 'Высокая мотивация']
means = [mean_low, mean_medium, mean_high]

plt.figure(figsize=(7,5))
plt.bar(groups, means, color=['lightcoral', 'lightyellow', 'lightgreen'], edgecolor='black')
plt.title('Средний индекс гигиены OHI-S в группах с разной мотивацией')
plt.xlabel('Уровень мотивации')
plt.ylabel('Средний балл OHI-S')
plt.grid(axis='y', linestyle='--', alpha=0.3)
for i, v in enumerate(means):
    plt.text(i, v + 0.05, str(round(v, 2)), ha='center', fontweight='bold')
plt.show()
```

**Что мы видим:** столбцы наглядно показывают, что у пациентов с высокой мотивацией индекс гигиены значительно ниже (лучше), чем у пациентов с низкой мотивацией. Это визуальное подтверждение того, что мотивация влияет на гигиену полости рта. Однако для статистических выводов нужен дисперсионный анализ.

### 5.5. Линейный график — динамика боли после удаления зуба мудрости

**Назначение:** показывает изменение значения во времени или по порядку. Соединяет точки линией, подчёркивая динамику и тренды.

**Клинический пример:** мониторинг уровня боли по визуально-аналоговой шкале (VAS, 0–10) у одного пациента после удаления ретинированного третьего моляра в течение 7 дней. Это помогает врачу отслеживать эффективность обезболивания и вовремя заметить осложнения.

**Данные:** типичная динамика снижения боли с небольшими колебаниями.

```python
days = [1, 2, 3, 4, 5, 6, 7]
pain_vas = [7.0, 5.5, 4.0, 3.0, 2.5, 1.5, 0.8]

plt.figure(figsize=(8,5))
plt.plot(days, pain_vas, marker='o', linewidth=2, color='darkblue')

for i, p in enumerate(pain_vas):
    plt.text(days[i] + 0.05, p + 0.2, str(p), fontsize=9)

plt.title('Динамика боли после удаления зуба мудрости (7 дней)')
plt.xlabel('День после операции')
plt.ylabel('Уровень боли по VAS (0–10)')
plt.ylim(0, 10)
plt.grid(linestyle='--', alpha=0.3)
plt.show()
```

**Что мы видим:** боль постепенно снижается, но есть небольшое плато на 3–4 день (возможно, пик воспалительного отёка). Общий тренд положительный: от 7 до 0.8 баллов за неделю. Линейный график позволяет врачу отслеживать прогресс и корректировать назначения при необходимости (например, усилить анальгетики при неожиданном подъёме боли).

## 6. ПЕРВОЕ ЗНАКОМСТВО С РАЗДЕЛЕНИЕМ ДАННЫХ

Любое правило или модель, созданные на одной выборке, нужно проверять на новых данных, чтобы избежать «подгона» (переобучения). Данные делят на три части:

- **Обучающая (train)** — на ней находят закономерности.
- **Валидационная (validation)** — для настройки параметров (выбор порогов и т.п.).
- **Тестовая (test)** — финальная проверка (только один раз, в конце).

Параллель со стоматологией: разрабатывая новую прогностическую шкалу риска развития периимплантита, мы проверяем её на другой группе пациентов, чем та, на которой создавали. Если проверить на той же — получим завышенные оценки. Обычно используют пропорции 60/20/20 или 70/15/15, сохраняя соотношение групп (пол, возраст, статус курения, тип имплантата) в каждой части — это называется стратификацией.

Схема разделения:

```mermaid
flowchart TD
    A[Исходные данные] --> B[Обучающая 60-70%]
    A --> C[Валидационная 15-20%]
    A --> D[Тестовая 15-20%]
    B --> B1["Поиск закономерностей"]
    C --> C1["Настройка параметров"]
    D --> D1["Финальная проверка"]
```
