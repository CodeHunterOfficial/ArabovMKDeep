### Задание 5. Связь между переменными (scatter plot)

**Описание.** В этом задании вы проверите гипотезу о связи между глубиной пародонтальных карманов (PPD) и индексом гигиены (OHI-S).

**Требуется:**

1. Постройте **диаграмму рассеяния** (scatter plot), где:
   - По оси X — средняя глубина кармана (PPD).
   - По оси Y — индекс гигиены (OHI-S).
   - Каждая точка — один пациент.

2. Добавьте линию тренда (линейную регрессию).

3. Рассчитайте коэффициент корреляции Пирсона между PPD и OHI-S (вручную или с помощью инструмента).

4. Напишите интерпретацию:
   - Есть ли видимая связь между переменными?
   - Какова сила и направление корреляции?
   - Что это может означать для клинической практики (например, важность гигиены для профилактики пародонтита)?

**Что сдать:** график с линией тренда, значение коэффициента корреляции, текстовую интерпретацию.

```python
# Если используете Python:
import numpy as np
from scipy import stats

# Scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(ppd, ohis, alpha=0.7, color='darkblue')

# Линия тренда
z = np.polyfit(ppd, ohis, 1)
p = np.poly1d(z)
plt.plot(ppd, p(ppd), color='red', linestyle='dashed')

# Коэффициент корреляции
corr, p_value = stats.pearsonr(ppd, ohis)
print(f"Корреляция Пирсона: r = {corr:.3f}, p = {p_value:.4f}")

plt.xlabel('Средняя глубина кармана, мм')
plt.ylabel('Индекс гигиены OHI-S, баллы')
plt.title('Связь между глубиной карманов и гигиеной полости рта')
plt.show()
```
