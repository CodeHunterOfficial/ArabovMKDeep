## 2. ЗАЧЕМ ВИЗУАЛИЗИРОВАТЬ: ОСНОВНЫЕ ФУНКЦИИ ГРАФИКОВ

Выделим четыре ключевые функции визуализации в клинических исследованиях.

**1. Разведочный анализ (Exploratory Data Analysis)**

График помогает **увидеть** данные до того, как вы начнёте считать статистики. Выбросы, пропуски, аномалии, форма распределения, нелинейные зависимости — всё это становится видимым.

*Пример из стоматологической практики:* вы построили гистограмму глубины пародонтальных карманов и увидели, что данные имеют две вершины (бимодальное распределение). Это сигнал: возможно, в вашей выборке смешаны две разные группы — пациенты с хроническим генерализованным пародонтитом лёгкой степени и пациенты с агрессивным пародонтитом. Если бы вы не построили график, вы могли бы упустить это важное открытие и неправильно интерпретировать среднее (оно оказалось бы между двумя группами и не отражало реальную картину).

**2. Проверка допущений статистических тестов**

Многие статистические тесты (например, t-тест, ANOVA) требуют, чтобы данные были распределены нормально. График (гистограмма или Q-Q-plot) помогает оценить это визуально — до применения тестов. Если распределение сильно скошено, вы можете выбрать непараметрические тесты (например, Манна–Уитни вместо t-теста).

*Пример:* вы видите на гистограмме, что глубина карманов имеет длинный «хвост» справа (несколько пациентов с очень глубокими карманами). Это значит, что среднее будет смещено вверх, и для описания центра лучше использовать медиану и интерквартильный размах (как мы обсуждали в Лекции 2).

**3. Сравнение групп и условий**

Графики позволяют сравнить несколько групп на одном рисунке. Это особенно полезно в клинических исследованиях, например, при сравнении эффективности разных методов лечения.

*Пример:* ящик с усами для трёх групп лечения пародонтита (скейлинг, лоскутная операция, лазерная терапия) показывает не только средние, но и разброс, и выбросы. Вы сразу видите, какая группа показывает лучшие результаты и насколько однородны данные внутри каждой группы.

**4. Коммуникация результатов**

График — самый эффективный способ донести результаты до коллег, научного сообщества, а иногда и до пациентов [1]. Статья с хорошими графиками читается легче и запоминается лучше. В медицинских журналах графики являются обязательным элементом публикаций [1].

*Пример:* вместо того чтобы писать «средняя глубина пародонтального кармана снизилась с 5.2 мм до 3.1 мм», вы показываете линейный график, где видно плавное снижение в течение 6 месяцев. Это убедительнее и нагляднее.

**Визуализация:** Mermaid-диаграмма с четырьмя блоками-функциями.

```mermaid
flowchart TD
    A[Функции визуализации<br>в клинической медицине] --> B[Разведочный анализ<br>увидеть выбросы, паттерны]
    A --> C[Проверка допущений<br>нормальность распределения]
    A --> D[Сравнение групп<br>методы лечения]
    A --> E[Коммуникация<br>донести результаты до аудитории]
    
    B --> B1["Пример: гистограмма<br>обнаружила бимодальность PPD"]
    C --> C1["Пример: Q-Q-plot<br>оценка нормальности OHI-S"]
    D --> D1["Пример: boxplot<br>сравнение трёх методов"]
    E --> E1["Пример: линейный график<br>динамика заживления"]
```

## 3. КАК ВЫБРАТЬ ТИП ГРАФИКА: АЛГОРИТМ ДЛЯ ВРАЧА

Выбор графика определяется **типом переменных** и **исследовательским вопросом**. Предлагаем простой алгоритм для принятия решения.

```mermaid
flowchart TD
    Start["Что я хочу показать?"]
    
    Start --> Q1["Сколько переменных?"]
    
    Q1 -->|"Одна"| Single["Одна переменная"]
    Q1 -->|"Две"| Two["Две переменные"]
    Q1 -->|"Много"| Multi["Три и более"]
    
    Single --> S1["Какой тип?"]
    S1 -->|"Количественная"| H["Гистограмма<br>или ящик с усами"]
    S1 -->|"Категориальная"| Bar["Столбчатая диаграмма<br>частот"]
    
    Two --> T1["Какие типы?"]
    T1 -->|"Колич + колич"| Scatter["Точечный график<br>scatter plot"]
    T1 -->|"Категор + колич"| Box["Ящик с усами<br>или столбцы со стандартными ошибками"]
    T1 -->|"Категор + категор"| Heat["Тепловая карта<br>сопряжённости"]
    
    Multi --> M1["Что именно?"]
    M1 -->|"Корреляции"| Corr["Тепловая карта<br>корреляционной матрицы"]
    M1 -->|"Динамика"| Line["Линейный график<br>с несколькими линиями"]
    M1 -->|"Многомерные данные"| Facet["Фасетная сетка<br>несколько графиков"]
    
    H --> Result["Вы выбрали график!"]
    Bar --> Result
    Scatter --> Result
    Box --> Result
    Heat --> Result
    Corr --> Result
    Line --> Result
    Facet --> Result
    
    style Start fill:#e1f5fe,stroke:#01579b
    style Result fill:#c8e6c9,stroke:#2e7d32
```

**Краткое пояснение к схеме:**

| Сценарий | Тип графика | Что показывает | Пример из стоматологии |
|----------|-------------|----------------|--------------------------|
| Одна количественная | Гистограмма | Распределение | Распределение глубины карманов в выборке |
| Одна категориальная | Столбчатая диаграмма | Частоты | Сколько пациентов с каждым диагнозом (кариес, пульпит, периодонтит) |
| Две количественные | Точечный график | Связь, корреляция | Связь возраста и глубины карманов |
| Категория + количество | Ящик с усами | Сравнение групп | Глубина карманов по типам лечения |
| Две категории | Тепловая карта | Сопряжённость | Диагноз × метод лечения |
| Корреляции (>2 переменных) | Тепловая карта | Матрица корреляций | Связи между 10 клиническими параметрами |

### 4.1. Гистограмма (Histogram)

**Что это:** график распределения одной количественной переменной. Данные разбиваются на интервалы («корзины», bins), и для каждого интервала показывается частота — сколько значений попало в этот интервал.

**Когда использовать:** чтобы понять форму распределения, проверить нормальность, увидеть выбросы и пробелы.

**Клинический пример:**

> **Задача:** врач-стоматолог хочет понять, как распределена средняя глубина пародонтальных карманов (PPD) у 150 пациентов с хроническим пародонтитом. Есть ли много пациентов с очень глубокими карманами? Распределение симметричное или скошенное?

**Как построить гистограмму (блок-схема):**

```mermaid
flowchart LR
    subgraph Данные
        D["150 пациентов, PPD от 2 до 8 мм"]
    end
    subgraph График
        G["Ось X: интервалы PPD (2-3, 3-4, …)"]
        G2["Ось Y: количество пациентов"]
    end
    subgraph Вывод
        V1["Распределение смещено вправо"]
        V2["Есть выбросы >7 мм"]
    end
    Данные --> График --> Вывод
```

#### 📊 Код для гистограммы (с комментариями)

```python
# 1. Импортируем нужные библиотеки
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Создаём пример данных (150 пациентов, глубина карманов от 2 до 8 мм)
#    В реальности вы загрузите свои данные из файла: data = pd.read_csv('file.csv')
np.random.seed(42)  # чтобы результат был воспроизводимым
ppd_scores = np.random.normal(loc=4.5, scale=1.2, size=150)  # нормальное распределение
ppd_scores = np.clip(ppd_scores, 2, 8)  # обрезаем до клинически возможных значений
data = pd.DataFrame({'PPD': ppd_scores})

# 3. Строим гистограмму
plt.figure(figsize=(10, 6))                 # размер рисунка (ширина, высота)
plt.hist(data['PPD'],                       # данные
         bins=12,                           # количество интервалов
         edgecolor='black',                 # обводка столбцов
         color='skyblue',                   # цвет заливки
         alpha=0.7)                         # прозрачность (0-1)

# 4. Добавляем линии среднего и медианы
mean_val = data['PPD'].mean()
median_val = data['PPD'].median()
plt.axvline(mean_val, color='red', linestyle='--', linewidth=2,
            label=f'Среднее = {mean_val:.1f} мм')
plt.axvline(median_val, color='green', linestyle='--', linewidth=2,
            label=f'Медиана = {median_val:.1f} мм')

# 5. Оформляем по правилам научных публикаций
plt.title('Рисунок 1. Распределение глубины пародонтальных карманов (N=150)',
          fontsize=14, fontweight='bold')
plt.xlabel('Средняя глубина кармана, мм', fontsize=12)
plt.ylabel('Количество пациентов', fontsize=12)
plt.legend()                                # показываем легенду
plt.grid(axis='y', alpha=0.3)               # сетка по Y для удобства
plt.tight_layout()                          # автоматически подгоняет элементы
plt.show()                                  # выводим график на экран
```

**Что можно увидеть на гистограмме:**
- **Нормальное распределение** — колоколообразная форма, симметричная.
- **Скошенность** — «хвост» справа (глубокие карманы) или слева (мелкие карманы). Это важно для выбора мер центральной тенденции (как в Лекции 2).
- **Бимодальность** — две вершины. Это может означать, что в выборке смешаны две разные группы (например, лёгкий и тяжёлый пародонтит).
- **Выбросы** — отдельные столбцы далеко от основной массы (например, пациенты с агрессивным пародонтитом).

**Практические рекомендации:**
- Для небольших выборок (до 30) используйте 5–7 интервалов.
- Для больших выборок (100+) — 10–15 интервалов.
- В научных статьях обычно указывают количество интервалов в примечании к рисунку.

### 4.2. Ящик с усами (Boxplot)

**Что это:** график, показывающий распределение через пять чисел: минимум, первый квартиль (Q1), медиана, третий квартиль (Q3), максимум. «Усы» показывают размах без выбросов, а выбросы рисуются отдельными точками.  
*Примечание:* в большинстве библиотек усы ограничены 1.5×IQR, а не абсолютным минимумом/максимумом — это стандарт.

**Когда использовать:** чтобы сравнить распределения нескольких групп, увидеть выбросы, показать разброс данных.

**Структура ящика с усами:**

```mermaid
flowchart TD
    subgraph Box["Ящик с усами (Boxplot)"]
        Max["Максимум (верхний ус)"]
        Q3["Q3 — 75-й перцентиль"]
        Median["Медиана"]
        Q1["Q1 — 25-й перцентиль"]
        Min["Минимум (нижний ус)"]
        Outlier["Выбросы — точки за усами"]
    end
    style Box fill:#e1f5fe,stroke:#01579b
```

**Клинический пример:**

> **Задача:** стоматолог сравнивает глубину пародонтальных карманов после трёх методов лечения: скейлинг, лоскутная операция, лазерная терапия. Нужно показать не только средние, но и разброс.

#### 📊 Код для ящика с усами (сравнение трёх групп)

```python
# 1. Подготавливаем данные: три группы по 50 пациентов
np.random.seed(42)
group_scaling = np.random.normal(3.5, 0.8, 50)    # скейлинг
group_flap = np.random.normal(3.0, 0.7, 50)       # лоскутная операция
group_laser = np.random.normal(3.3, 0.9, 50)      # лазер
# Собираем в DataFrame
df_box = pd.DataFrame({
    'PPD': np.concatenate([group_scaling, group_flap, group_laser]),
    'Метод': ['Скейлинг']*50 + ['Лоскутная']*50 + ['Лазер']*50
})

# 2. Строим boxplot с помощью seaborn
plt.figure(figsize=(10, 6))
sns.boxplot(data=df_box, x='Метод', y='PPD',
            palette='Set2',            # цветовая палитра (дружелюбная к дальтоникам)
            width=0.5)                # ширина ящиков

# 3. Оформление
plt.title('Рисунок 2. Сравнение глубины карманов после разных методов лечения',
          fontsize=14, fontweight='bold')
plt.xlabel('Метод лечения', fontsize=12)
plt.ylabel('Средняя глубина кармана, мм', fontsize=12)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

**Что можно увидеть на boxplot:**
- **Медиану** (линия внутри ящика) — центральная тенденция (устойчива к выбросам).
- **Разброс** (высота ящика = IQR) — насколько данные разбросаны.
- **Симметричность** — если медиана по центру ящика, распределение симметрично; если смещена — есть скошенность.
- **Выбросы** (точки за усами) — аномальные значения, требующие внимания (например, пациент с неожиданно глубоким карманом после лечения).

**Важное предостережение:** boxplot показывает **грубую** картину. Он не даёт информации о форме распределения внутри ящика (там может быть бимодальность). Поэтому его часто комбинируют с гистограммой или скрипичным графиком (violin plot).

### 4.3. Точечный график (Scatter plot)

**Что это:** график, показывающий связь между двумя количественными переменными. Каждая точка — один пациент. X и Y — значения двух переменных.

**Когда использовать:** чтобы увидеть корреляцию, кластеры, нелинейные зависимости, выбросы.

**Клинический пример:**

> **Задача:** исследователь хочет проверить, есть ли связь между возрастом пациента и глубиной пародонтальных карманов (PPD). Точки показывают каждого пациента.

#### 📊 Код для точечного графика с линией регрессии

```python
# 1. Создаём данные: 50 пациентов, возраст (25–70 лет) и глубина карманов (2–8 мм)
np.random.seed(42)
n = 50
age = np.random.uniform(25, 70, n)                          # X
# глубина кармана = 2 + 0.05 * возраст + шум (положительная корреляция)
ppd = 2 + 0.05 * age + np.random.normal(0, 0.8, n)
ppd = np.clip(ppd, 2, 8)                                    # не выходим за пределы
df_scatter = pd.DataFrame({'Возраст': age, 'PPD': ppd})

# 2. Строим scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(df_scatter['Возраст'], df_scatter['PPD'],
            alpha=0.7, color='blue', s=60)   # s – размер точек

# 3. Добавляем линию регрессии (тренд)
from scipy import stats
slope, intercept, r_value, p_value, std_err = stats.linregress(df_scatter['Возраст'],
                                                                df_scatter['PPD'])
x_vals = np.linspace(df_scatter['Возраст'].min(), df_scatter['Возраст'].max(), 100)
y_vals = slope * x_vals + intercept
plt.plot(x_vals, y_vals, color='red', linestyle='-', linewidth=2,
         label=f'Линия регрессии (r = {r_value:.2f})')

# 4. Оформление
plt.title('Рисунок 3. Связь между возрастом и глубиной пародонтальных карманов',
          fontsize=14, fontweight='bold')
plt.xlabel('Возраст, лет', fontsize=12)
plt.ylabel('Средняя глубина кармана, мм', fontsize=12)
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

**Что можно увидеть на scatter plot:**
- **Корреляцию** — точки вытянуты вдоль линии (положительной или отрицательной).
- **Кластеры** — точки образуют группы (возможно, разные формы пародонтита).
- **Нелинейную связь** — точки образуют дугу или другую сложную форму.
- **Выбросы** — точки, далёкие от основной массы, которые могут искажать корреляцию.

**Дополнительные возможности:**
- **Цвет** точек = категориальная переменная (например, пол) — тогда добавьте параметр `c=df['Пол']`.
- **Размер** точек = третья количественная переменная (например, индекс гигиены) — параметр `s=df['OHI_S']*10`.
- **Линия регрессии** — мы добавили её в коде, она показывает общую тенденцию.

**Важно:** scatter plot показывает **наличие** связи, но не её силу. Для измерения силы связи нужна корреляция (коэффициент Пирсона или Спирмена) [2].

### 4.4. Тепловая карта (Heatmap)

**Что это:** график, где значения представлены цветом. Каждая ячейка — это значение на пересечении двух переменных. Цветовая шкала позволяет мгновенно увидеть паттерны.

**Когда использовать:** для визуализации матриц (корреляционной матрицы, таблиц сопряжённости), когда нужно показать паттерн, а не отдельные числа.

**Два основных типа в медицине:**

**А) Тепловая карта корреляционной матрицы:**

> **Задача:** стоматолог измерил 10 клинических параметров (глубина карманов, уровень клинического прикрепления, индекс гигиены, кровоточивость, возраст, количество зубов и др.) и хочет увидеть, какие параметры сильно коррелируют, а какие — нет.

#### 📊 Код для тепловой карты корреляций

```python
# 1. Создаём искусственные данные для 5 параметров (100 пациентов)
np.random.seed(42)
n_subj = 100
# Генерируем коррелированные данные
ppd = np.random.normal(4.5, 1.2, n_subj)
ohis = 0.6 * ppd + np.random.normal(0, 0.5, n_subj)          # корреляция с гигиеной
cal = 0.8 * ppd + np.random.normal(0, 0.7, n_subj)           # уровень прикрепления
sbi = 0.5 * ppd + np.random.normal(0, 5, n_subj)             # кровоточивость
age = 0.3 * ppd + np.random.normal(45, 10, n_subj)           # возраст

df_corr = pd.DataFrame({
    'PPD': ppd,
    'OHI-S': ohis,
    'CAL': cal,
    'SBI': sbi,
    'Возраст': age
})

# 2. Считаем корреляционную матрицу
corr_matrix = df_corr.corr()

# 3. Строим тепловую карту
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix,
            annot=True,               # показывать числа в ячейках
            fmt='.2f',                # формат чисел (2 знака)
            cmap='coolwarm',          # цветовая схема (красный-белый-синий)
            center=0,                 # центр цветовой шкалы (0 – нет корреляции)
            square=True,              # квадратные ячейки
            linewidths=0.5,           # границы между ячейками
            cbar_kws={'shrink': 0.8}) # размер цветовой шкалы
plt.title('Рисунок 4. Корреляционная матрица клинических параметров (N=100)',
          fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

**Что можно увидеть:**
- Кластеры сильно коррелирующих параметров (например, PPD и CAL — оба отражают тяжесть пародонтита). Это может указывать на то, что они измеряют один и тот же конструкт.
- Параметры, которые почти не коррелируют с другими (изолированные) — возможно, они измеряют что-то уникальное.

**Б) Тепловая карта таблицы сопряжённости:**

> **Задача:** стоматолог изучает связь между диагнозом (кариес, пульпит, периодонтит) и методом лечения (терапевтическое, хирургическое, ортопедическое).

#### 📊 Код для тепловой карты таблицы сопряжённости

```python
# 1. Создаём таблицу частот (например, данные 200 пациентов)
np.random.seed(42)
diagnoses = np.random.choice(['Кариес', 'Пульпит', 'Периодонтит'], size=200, p=[0.5, 0.3, 0.2])
treatment = []
for d in diagnoses:
    if d == 'Кариес':
        treatment.append(np.random.choice(['Пломбирование', 'Реставрация', 'Наблюдение'], p=[0.7, 0.2, 0.1]))
    elif d == 'Пульпит':
        treatment.append(np.random.choice(['Эндодонтия', 'Удаление', 'Пломбирование'], p=[0.6, 0.3, 0.1]))
    else:  # Периодонтит
        treatment.append(np.random.choice(['Скейлинг', 'Лоскутная операция', 'Лазер'], p=[0.4, 0.4, 0.2]))

df_contingency = pd.DataFrame({'Диагноз': diagnoses, 'Лечение': treatment})
# Строим таблицу сопряжённости
contingency_table = pd.crosstab(df_contingency['Диагноз'], df_contingency['Лечение'])

# 2. Тепловая карта
plt.figure(figsize=(8, 5))
sns.heatmap(contingency_table,
            annot=True,            # показывать числа
            fmt='d',               # целые числа
            cmap='Blues',          # голубая палитра
            cbar_kws={'label': 'Количество пациентов'})
plt.title('Рисунок 5. Связь между диагнозом и выбранным методом лечения')
plt.xlabel('Метод лечения')
plt.ylabel('Диагноз')
plt.tight_layout()
plt.show()
```

**Что можно увидеть:** паттерны выбора — например, при пульпите чаще всего выбирают эндодонтию, при периодонтите — скейлинг или лоскутную операцию.

### 4.5. Линейный график (Line plot)

**Что это:** график, показывающий изменение переменной вдоль оси X (обычно время или порядковая переменная). Точки соединяются линией, чтобы подчеркнуть динамику.

**Когда использовать:** для отображения динамики, трендов, изменений во времени.

**Клинический пример:**

> **Задача:** стоматолог отслеживает уровень боли по визуально-аналоговой шкале (VAS) у пациента после удаления ретинированного третьего моляра в течение 7 дней. Нужно показать, как снижается боль.

#### 📊 Код для линейного графика (один пациент)

```python
# 1. Данные: 7 дней, баллы VAS
days = np.arange(1, 8)
vas_scores = [7.0, 5.5, 4.0, 3.5, 2.5, 1.5, 0.8]  # на 4-й день небольшой подъём

# 2. Строим линейный график
plt.figure(figsize=(10, 6))
plt.plot(days, vas_scores, marker='o', linestyle='-', color='blue',
         linewidth=2, markersize=8, label='Уровень боли (VAS)')

# 3. Оформление
plt.title('Рисунок 6. Динамика боли после удаления зуба мудрости (7 дней)',
          fontsize=14, fontweight='bold')
plt.xlabel('День после операции', fontsize=12)
plt.ylabel('Балл по VAS (0–10)', fontsize=12)
plt.xticks(days)                  # показываем все дни
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

**Что можно увидеть на line plot:**
- **Тренд** — возрастание, убывание или стабильность.
- **Скорость изменений** — крутизна наклона (быстрое или медленное снижение боли).
- **Аномалии** — резкие скачки (например, подъём боли на 4-й день, который может указывать на осложнение — альвеолит).

**Варианты с несколькими линиями (сравнение групп):**

```python
# Пример: две группы (скейлинг и лоскутная операция) – средние значения PPD за 6 месяцев
months = np.arange(0, 7)
scaling_means = [5.2, 4.5, 4.0, 3.6, 3.4, 3.2, 3.1]
flap_means = [5.5, 4.2, 3.5, 3.0, 2.8, 2.6, 2.5]

plt.figure(figsize=(10, 6))
plt.plot(months, scaling_means, marker='s', label='Скейлинг', color='green', linewidth=2)
plt.plot(months, flap_means, marker='^', label='Лоскутная операция', color='orange', linewidth=2)
plt.title('Рисунок 7. Динамика глубины карманов в двух группах лечения')
plt.xlabel('Месяц после лечения')
plt.ylabel('Средняя глубина кармана, мм')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

**Доверительные интервалы** можно добавить, используя `plt.fill_between()` (это более сложный вариант, но при желании можно изучить дополнительно).

## 6. ПРАКТИЧЕСКОЕ УПРАЖНЕНИЕ НА ПАРЕ (5–7 минут)

> **Задача:** у вас есть данные 40 пациентов с пародонтитом: глубина пародонтального кармана (PPD) и метод лечения (скейлинг, лоскутная операция, лазер). Нужно показать различие между группами.

### Вопросы для обсуждения

1. **Какой тип графика вы выберете?**  
   – Лучший выбор — **ящик с усами (boxplot)**, так как он показывает медиану, разброс (IQR) и выбросы для каждой группы, что идеально для сравнения.

2. **Что должно быть на осях?**  
   – Ось X: метод лечения (категориальная переменная).  
   – Ось Y: глубина кармана (количественная переменная, мм).

3. **Что вы ожидаете увидеть?**  
   – Например, в группе лоскутной операции медиана ниже, чем в группе скейлинга; в группе лазера — промежуточное значение. Возможны выбросы (пациенты с аномально глубокими карманами).

4. **Как оформить график для научной статьи?**  
   – Добавить номер и название, подписи осей, легенду (если нужно), примечание со статистиками (среднее, стандартное отклонение, количество наблюдений).

### Готовый код для упражнения

```python
# 1. Импорт библиотек
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Создаём пример данных (40 пациентов)
np.random.seed(42)
group_scaling = np.random.normal(4.2, 0.8, 14)      # скейлинг
group_flap = np.random.normal(3.1, 0.7, 13)         # лоскутная операция
group_laser = np.random.normal(3.5, 0.9, 13)        # лазер

df_ex = pd.DataFrame({
    'PPD': np.concatenate([group_scaling, group_flap, group_laser]),
    'Метод': ['Скейлинг']*14 + ['Лоскутная']*13 + ['Лазер']*13
})

# 3. Строим ящик с усами
plt.figure(figsize=(10, 6))
sns.boxplot(data=df_ex, x='Метод', y='PPD',
            palette='Set2', width=0.5)

# 4. Оформление
plt.title('Рисунок А. Глубина пародонтальных карманов после разных методов лечения (N=40)',
          fontsize=14, fontweight='bold')
plt.xlabel('Метод лечения', fontsize=12)
plt.ylabel('Средняя глубина кармана, мм', fontsize=12)
plt.grid(axis='y', alpha=0.3)

# Добавляем примечание со статистиками
means = df_ex.groupby('Метод')['PPD'].mean().round(2)
stds = df_ex.groupby('Метод')['PPD'].std().round(2)
note_text = f"Средние (M ± SD): Скейлинг = {means['Скейлинг']} ± {stds['Скейлинг']} мм, " \
            f"Лоскутная = {means['Лоскутная']} ± {stds['Лоскутная']} мм, " \
            f"Лазер = {means['Лазер']} ± {stds['Лазер']} мм"
plt.figtext(0.5, 0.01, note_text, ha='center', fontsize=10,
            bbox=dict(facecolor='white', alpha=0.8))

plt.tight_layout(rect=[0, 0.05, 1, 1])
plt.show()
```

**Обсуждение альтернатив:**  
Столбчатая диаграмма со стандартными ошибками тоже возможна, но она скрывает распределение внутри группы и не показывает выбросы. Поэтому **boxplot** предпочтительнее для разведочного анализа.

## 7. ОБЗОРНЫЙ КОД ДЛЯ ДЕМОНСТРАЦИИ

Этот код охватывает все основные графики, рассмотренные в лекции. Вы можете скопировать его целиком и запустить в своей среде (Jupyter Notebook, Google Colab и др.). Все шаги подробно прокомментированы на русском.

```python
# ============================================================
# 0. ПОДГОТОВКА (импорт библиотек и создание данных)
# ============================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

np.random.seed(42)
ppd = np.random.normal(4.5, 1.2, 150)
ppd = np.clip(ppd, 2, 8)

group_labels = ['Скейлинг']*50 + ['Лоскутная']*50 + ['Лазер']*50
ppd_group = ppd.copy()
ppd_group[0:50] += 0.5
ppd_group[50:100] -= 0.3
ppd_group[100:150] += 0.2
ppd_group = np.clip(ppd_group, 2, 8)

ohis = 0.6 * ppd_group + np.random.normal(0, 0.5, 150)
ohis = np.clip(ohis, 0, 3)

data = pd.DataFrame({
    'PPD': ppd_group,
    'Группа': group_labels,
    'OHI_S': ohis
})

# ============================================================
# ГРАФИК 1. ГИСТОГРАММА (распределение одной переменной)
# ============================================================
plt.figure(figsize=(10, 6))
plt.hist(data['PPD'], bins=12, edgecolor='black', color='skyblue', alpha=0.7)
plt.axvline(data['PPD'].mean(), color='red', linestyle='--', linewidth=2,
            label=f'Среднее = {data["PPD"].mean():.1f} мм')
plt.axvline(data['PPD'].median(), color='green', linestyle='--', linewidth=2,
            label=f'Медиана = {data["PPD"].median():.1f} мм')
plt.title('Рисунок 1. Распределение глубины пародонтальных карманов (N=150)',
          fontsize=14, fontweight='bold')
plt.xlabel('Глубина кармана, мм', fontsize=12)
plt.ylabel('Количество пациентов', fontsize=12)
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# ============================================================
# ГРАФИК 2. ЯЩИК С УСАМИ (сравнение трёх групп)
# ============================================================
plt.figure(figsize=(10, 6))
sns.boxplot(data=data, x='Группа', y='PPD', palette='Set2', width=0.5)
plt.title('Рисунок 2. Сравнение глубины карманов между группами лечения',
          fontsize=14, fontweight='bold')
plt.xlabel('Метод лечения', fontsize=12)
plt.ylabel('Глубина кармана, мм', fontsize=12)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# ============================================================
# ГРАФИК 3. ТОЧЕЧНЫЙ ГРАФИК (связь двух переменных + регрессия)
# ============================================================
plt.figure(figsize=(10, 6))
plt.scatter(data['OHI_S'], data['PPD'], alpha=0.6, color='blue', s=50)

slope, intercept, r_value, p_value, std_err = stats.linregress(data['OHI_S'],
                                                                data['PPD'])
x_vals = np.linspace(data['OHI_S'].min(), data['OHI_S'].max(), 100)
y_vals = slope * x_vals + intercept
plt.plot(x_vals, y_vals, color='red', linewidth=2,
         label=f'Линия регрессии (r = {r_value:.2f})')

plt.title('Рисунок 3. Связь между индексом гигиены (OHI-S) и глубиной карманов',
          fontsize=14, fontweight='bold')
plt.xlabel('Индекс гигиены OHI-S (баллы)', fontsize=12)
plt.ylabel('Глубина кармана, мм', fontsize=12)
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# ============================================================
# ГРАФИК 4. ТЕПЛОВАЯ КАРТА (корреляционная матрица)
# ============================================================
np.random.seed(42)
sbi = 0.5 * data['PPD'] + np.random.normal(0, 5, 150)
cal = 0.8 * data['PPD'] + np.random.normal(0, 0.7, 150)
age = 0.2 * data['PPD'] + np.random.normal(45, 10, 150)
df_corr = pd.DataFrame({
    'PPD': data['PPD'],
    'OHI-S': data['OHI_S'],
    'SBI': sbi,
    'CAL': cal,
    'Возраст': age
})
corr_matrix = df_corr.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm',
            center=0, square=True, linewidths=0.5,
            cbar_kws={'shrink': 0.8})
plt.title('Рисунок 4. Корреляционная матрица клинических параметров (N=150)',
          fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# ============================================================
# ГРАФИК 5. ЛИНЕЙНЫЙ ГРАФИК (динамика одного пациента)
# ============================================================
days = np.arange(1, 8)
pain_vas = [7.0, 5.5, 4.0, 3.5, 2.5, 1.5, 0.8]
plt.figure(figsize=(10, 6))
plt.plot(days, pain_vas, marker='o', linestyle='-', color='blue',
         linewidth=2, markersize=8, label='Уровень боли')
plt.title('Рисунок 5. Динамика боли после удаления зуба мудрости (7 дней)',
          fontsize=14, fontweight='bold')
plt.xlabel('День после операции', fontsize=12)
plt.ylabel('Балл по VAS (0–10)', fontsize=12)
plt.xticks(days)
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

**Что важно понять из кода (пояснение для студентов):**

| Строка кода | Что делает для врача |
|-------------|----------------------|
| `plt.hist(bins=12)` | Разбивает данные на интервалы и показывает частоту – видим форму распределения |
| `sns.boxplot(x='Группа', y='PPD')` | Сравнивает группы по медиане, разбросу и выбросам |
| `plt.scatter()` + линия регрессии | Показывает связь между двумя переменными и её силу (коэффициент корреляции) |
| `sns.heatmap(corr_matrix, annot=True)` | Визуализирует все попарные корреляции в виде цветовой матрицы |
| `plt.plot(marker='o')` | Отображает изменение показателя во времени |

**Главный вывод:** код — это универсальный шаблон. Вы можете заменить искусственные данные на свои (например, загрузить CSV-файл командой `pd.read_csv()`) и получить готовые графики для своей научной работы. Главное — правильно выбрать тип графика под вашу задачу.

**Основные типы графиков и их задачи:**

| График | Задача | Пример из стоматологии |
|--------|--------|--------------------------|
| Гистограмма | Показать распределение | Распределение PPD в выборке |
| Ящик с усами | Сравнить группы | Глубина карманов по методам лечения |
| Точечный график | Показать связь | Корреляция возраста и PPD |
| Тепловая карта | Показать паттерны | Корреляционная матрица клинических параметров |
| Линейный график | Показать динамику | Снижение боли после операции |

**Этический аспект:** график должен **отражать реальность**, а не создавать иллюзию. Изменение масштаба, обрезание данных, выбор «удобного» типа графика — это манипуляция, которая в медицине может привести к неверным клиническим решениям. Ответственность за честность визуализации лежит на авторе [3].

**Связь с предыдущими лекциями:**
- В Лекции 2 мы научились описывать данные статистиками. Теперь мы умеем показывать их наглядно.
- В Лекции 3 мы научились превращать текст в числа. Теперь эти числа можно визуализировать.
- В будущих лекциях мы будем использовать визуализацию для интерпретации результатов моделей машинного обучения.

**Главная мысль напоследок:**
> График — это не просто иллюстрация. Это инструмент мышления. Хороший график помогает увидеть то, что скрыто в цифрах, и задать правильные вопросы. Плохой график может ввести в заблуждение. Поэтому ответственность врача — делать визуализацию честной, прозрачной и понятной.
