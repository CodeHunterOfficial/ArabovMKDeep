## 1. ВВЕДЕНИЕ

**От слов к смыслу: эволюция NLP в медицине**

Представьте себе такую ситуацию. Вы — врач-стоматолог, и пациент говорит вам: *«Доктор, у меня сильно болит зуб, особенно когда я накусываю, десна опухла, и уже второй день температура 37.5»*. Вы мгновенно понимаете: это признаки острого периодонтита или периостита, есть воспаление, вероятно, гнойный процесс. Вы связываете эти слова в единую клиническую картину, видите контекст, улавливаете оттенки симптомов.

А теперь представьте, что эти же слова попадают в компьютер, который использует **мешок слов** (Bag of Words, из Лекции 3). Что он увидит? Просто набор слов: *«болит»*, *«зуб»*, *«накусывание»*, *«десна»*, *«опухла»*, *«температура»*. Он посчитает частоты, но **не увидит связи** между ними. Для него фраза *«зуб болит при накусывании»* и *«накусывание болит при зуб»* будут почти одинаковыми — хотя смысл кардинально разный!

Именно эту проблему решают современные NLP-модели. Они **понимают контекст**, **учитывают порядок слов** и **извлекают смысл** из текста так, как это делает опытный клиницист.

> **Ключевой пример из стоматологии:**
> Фраза *«зуб не болит»* и *«зуб болит»* отличаются всего одним словом, но для врача это два абсолютно разных состояния. Традиционные методы (мешок слов) увидят только разницу в частотах. Современные модели (BERT, GPT) «поймут» разницу в смысле, потому что они учитывают контекст всего предложения, включая отрицания.

Исследования показывают, что трансформерные модели значительно превосходят традиционные подходы благодаря способности улавливать контекстуальные и семантические связи в тексте. Например, BERT достигает точности **93%** при обнаружении эмоционального дистресса (что применимо для оценки дентофобии), а современные открытые LLM (Gemma-3-27B, Llama-3.3-70B) показывают F1-score **0.818–0.865** при дообучении на клинических данных для скрининга депрессии [5]. Аналогичные подходы могут быть использованы для автоматического анализа жалоб, выявления осложнений и поддержки клинических решений в стоматологии.

## 2. КАК МОДЕЛИ РАБОТАЮТ С ТЕКСТОМ: ОТ СЛОВ К ЧИСЛАМ

Прежде чем погружаться в архитектуру трансформера, давайте разберёмся **с чем именно работает модель** и как текст превращается в числа, которые модель может обрабатывать. Это основа, без которой невозможно понять всё остальное.

---

### 2.1. Векторное представление слов (Word Embeddings)

**Что это:** каждое слово превращается в **вектор** — список чисел (обычно 100–768 чисел на одно слово). Это как «цифровой паспорт» слова.

**Интуитивная аналогия из медицины:**

> Представьте, что вы описываете пациента по множеству параметров: возраст, уровень артериального давления, пульс, температура, лабораторные показатели, рентгенологические признаки... Вы создаёте «числовой портрет» пациента. То же самое модель делает со словами — только параметров у неё сотни, и они отражают не клинические данные, а **смысловые отношения** между словами.

**Ключевое свойство эмбеддингов:** слова со схожим смыслом оказываются **рядом** в этом пространстве.

> **Пример:** в пространстве эмбеддингов:
> - *«пульпит»* и *«периодонтит»* — рядом (оба воспалительные заболевания зубов).
> - *«пульпит»* и *«отбеливание»* — далеко (разные области стоматологии).
> - *«пульпит»* и *«стул»* — очень далеко (несвязанные понятия).

**Как это помогает в стоматологии:** модель может «знать», что *«периодонтит»* связан с *«абсцесс»*, *«гной»* и *«отёк»*, даже если эти слова не встречались вместе в конкретном тексте. Она «понимает» смысловые связи, как опытный клиницист.

---

### 2.2. Word2Vec: как слова обретают смысл

**Word2Vec** — это одна из первых и самых известных моделей для создания эмбеддингов слов, разработанная Google в 2013 году. Её ключевая идея проста: **значение слова определяется его окружением**.

**Интуитивная аналогия:**

> Представьте, что вы изучаете медицинскую литературу, читая тысячи историй болезни. Вы замечаете, что слова *«периодонтит»* и *«абсцесс»* часто встречаются в одних и тех же текстах (рядом с *«отёк»*, *«боль»*, *«гной»*). Постепенно вы начинаете понимать, что эти слова связаны по смыслу. Word2Vec делает то же самое — но математически и в огромном масштабе.

**Как это работает (упрощённо):**

Модель «смотрит» на каждое слово и пытается предсказать слова вокруг него. Если модель угадывает — она «понимает» смысл. В процессе она создаёт векторы — числа, которые отражают эти смысловые связи.

---

### 2.3. Живой пример: находим семантических соседей

Давайте посмотрим, как это работает на практике с **предобученной русскоязычной моделью Word2Vec** (обучена на корпусе текстов из интернета и Википедии) [6].

**Код для демонстрации:**

```python
# Установка библиотеки (выполняется один раз)
!pip install gensim

import gensim.downloader as api

# Загружаем предобученную модель для русского языка
# Это как если бы мы взяли с полки готовый словарь ассоциаций
russian_model = api.load("word2vec-ruscorpora-300")

# ------------------------------------------------------------
# 1. ПОИСК СЕМАНТИЧЕСКИХ СОСЕДЕЙ
# ------------------------------------------------------------
# Находим 10 слов, наиболее близких по смыслу к слову «зуб»
similar_to_tooth = russian_model.most_similar('зуб_NOUN')
print("Слова, близкие к 'зуб':")
for word, score in similar_to_tooth:
    print(f"  {word} — {score:.3f}")
```

**Что мы видим:** модель «понимает», что *«десна»*, *«корень»*, *«эмаль»* и даже *«челюсть»* — семантически близки к *«зуб»* (это части полости рта). А *«пломба»* — связано с лечением зуба. **Числа (0.757, 0.726...) — это косинусное сходство**: чем ближе к 1, тем более похожи слова по смыслу.

**Медицинский смысл этого примера:**
> Если модель «знает», что *«зуб»* близок к *«десна»* и *«корень»*, то она также «знает», что *«периодонтит»* близок к *«абсцесс»*, *«гной»* и *«отёк»*. Это позволяет модели улавливать клинически значимые ассоциации в текстах пациентов.

---

### 2.4. Сравнение семантической близости

Модель может вычислять **степень сходства** между любыми двумя словами:

```python
similar_words = [w for w in russian_model.key_to_index.keys() if w.startswith('пульпит')]
print(similar_words[:10])

# Проверяем наличие нужных слов с тегами
word1 = 'пульпит_NOUN'
word2 = 'периодонтит_NOUN'

if word1 in russian_model.key_to_index and word2 in russian_model.key_to_index:
    similarity = russian_model.similarity(word1, word2)
    print(f"Сходство между 'пульпит' и 'периодонтит': {similarity:.3f}")
else:
    print("Одно из слов с таким тегом отсутствует.")
```

**Что это даёт врачу:** можно количественно оценивать, насколько язык пациента «близок» к описанию определённого заболевания, сравнивая средние расстояния между словами в его жалобах и эталонными кластерами (например, кластер «острая боль», «воспаление»).

---

### 2.5. Поиск «лишнего» слова

Модель может находить слово, которое выбивается из смыслового ряда:

```python
# 3. ПОИСК ЛИШНЕГО СЛОВА
odd_word = russian_model.doesnt_match(
    ["кариес_NOUN", "пульпит_NOUN", "периодонтит_NOUN", "отбеливание_NOUN"]
)
print(f"Лишнее слово: {odd_word}")
# Вывод: отбеливание_NOUN (потому что это не заболевание)
```

**Медицинская аналогия:** это как если бы врач заметил, что среди перечисленных симптомов (боль, отёк, гиперемия, кариес) слово *«кариес»* — это диагноз, а не симптом.

### 2.6. От статических эмбеддингов к контекстуальным

**Важное ограничение Word2Vec:** у слова *всегда один вектор*, независимо от контекста. Слово *«корень»* в предложениях *«корень зуба»* и *«корень растения»* получит **одинаковый** вектор. Это проблема!

**Решение:** современные модели (BERT, GPT) — **контекстуальные**. Они создают вектор для слова **с учётом всего предложения**. Поэтому *«корень»* в разных контекстах получает **разные** векторы.

**Интуитивная аналогия:**
> Word2Vec — это как врач, который знает значение каждого медицинского термина из учебника, но не учитывает контекст. BERT — как опытный клиницист, который понимает, что слово *«боль»* в фразе *«боль при накусывании»* и *«боль в грудной клетке»* означает совершенно разные вещи и требует разного подхода.

### 2.7. Современные модели эмбеддингов 2025-2026 (доступные на Hugging Face)

Для работы с русским языком сегодня доступны десятки моделей, которые можно использовать «из коробки». Вот несколько примеров, применимых в медицинском контексте:

**Для сравнения семантической близости предложений (STS — Semantic Textual Similarity):**

| Модель | Размер эмбеддинга | Что умеет | Ссылка |
|--------|-------------------|-----------|--------|
| **rubert-large-uncased-sts** | 1024 | Оценка похожести русских предложений (STS, поиск парафраз) | sergeyzh/rubert-large-uncased-sts |
| **StaRSE** | 512 | Компактная модель для семантической близости, кластеризации, работает на CPU | BorisTM/starse |

**Пример использования rubert-large-uncased-sts:**

```python
from sentence_transformers import SentenceTransformer

# Загружаем модель для сравнения предложений
model = SentenceTransformer('sergeyzh/rubert-large-uncased-sts')

# Три предложения: два похожих и одно другое
sentences = [
    "Пациент жалуется на сильную боль в зубе при накусывании.",
    "Больной отмечает острую боль при жевании в области моляра.",
    "Сегодня на улице хорошая погода."
]

# Получаем векторы (эмбеддинги) для всех предложений
embeddings = model.encode(sentences)

# Сравниваем все предложения попарно
similarities = model.similarity(embeddings, embeddings)
print("Матрица сходства:")
print(similarities)
```

**Что мы видим:** модель «понимает», что первые два предложения говорят об одном и том же (боль в зубе при накусывании), а третье — о погоде.

**Медицинский смысл:** такая модель может автоматически находить в базе электронных карт похожие клинические случаи. Например, если у вас есть новый пациент с жалобами на боль при накусывании и отёк десны, модель найдёт другие записи с похожими симптомами — для сравнения тактики лечения.

### 2.10. От эмбеддингов к трансформеру

Эмбеддинги (статичные или контекстуальные) — это **входные данные** для модели. Но сами по себе они не решают задач диагностики. Дальше в игру вступает **трансформер** — архитектура, которая обрабатывает эти векторы, уточняет их с помощью механизма внимания и выдаёт финальный результат (классификацию, генерацию, ответ на вопрос).

**Ключевая мысль для запоминания:**

> Word2Vec и подобные модели — это **словарь ассоциаций**. Они знают, какие слова близки по смыслу. BERT и GPT — это **читатели**, которые используют этот словарь, чтобы понять смысл **целого предложения или текста** пациента.

## 3. АРХИТЕКТУРА ТРАНСФОРМЕРА: ИНТУИТИВНОЕ ВВЕДЕНИЕ

Прежде чем говорить о конкретных моделях (BERT, GPT), нужно понять, что у них общего. Почти все современные NLP-модели построены на **архитектуре трансформера** (Transformer). Она была предложена в 2017 году в статье *«Attention is All You Need»* и стала революцией в обработке языка [1].

**Почему трансформер важен:**

До трансформера модели обрабатывали текст последовательно (слово за словом), как мы читаем книгу. Это было медленно и не позволяло улавливать долгосрочные зависимости (связи между словами, которые далеко друг от друга). Трансформер предложил **обрабатывать все слова одновременно** (параллельно), что ускорило обучение и улучшило понимание контекста.

---

### 3.1. Ключевая идея: механизм внимания (Attention)

**Интуитивная аналогия из врачебной практики:**

Представьте, что врач слушает пациента. Пациент говорит: *«Я пришёл к вам, потому что... ну... в последнее время меня беспокоит зуб... он болит, когда я пью холодное, а ещё десна опухла, и я не могу спать от боли...»*

На что врач обращает внимание? Не на все слова одинаково. Он **фокусируется** на ключевых сигналах: *«болит»*, *«холодное»*, *«десна опухла»*, *«не могу спать»*. Он **взвешивает** важность каждого слова: фраза *«я пришёл»* важна, но меньше, чем *«не могу спать от боли»* (это маркер острого воспаления). Он **связывает** слова между собой: *«болит»* связано с *«холодное»* (пульпит) и *«десна опухла»* (периодонтит).

**Это и есть механизм внимания (Attention) в трансформере.** Модель:
- Смотрит на **все слова одновременно**.
- Вычисляет **веса важности** для каждого слова относительно других.
- Фокусируется на **ключевых смысловых связях**.

> **Пример:** в предложении *«пациент с острым пульпитом сообщил о боли при накусывании»*, модель «заметит», что *«пульпитом»* сильно связано с *«боли»*, и даст этим словам высокий вес.

---

### 3.2. Само-внимание (Self-Attention) — внимание к контексту

Трансформер использует **само-внимание** (self-attention): каждое слово в предложении «смотрит» на все остальные слова, чтобы понять свой собственный смысл в контексте.

**Интуитивная аналогия:**

> Слово *«корень»* в предложении *«корень зуба воспалён»* и *«корень растения»* — это одно и то же слово, но смысл разный. Само-внимание позволяет модели «увидеть» окружающие слова и понять, что в первом случае речь о части зуба, а во втором — о части растения.

**В медицинском тексте:**
> Слово *«боль»* в предложении *«боль при накусывании»* и *«боль в грудной клетке»* — разные контексты. Модель через само-внимание «понимает» разницу.

**Как это вычисляется (очень упрощённо):**

Для каждого слова модель вычисляет три вектора:
- **Query (запрос):** «что я ищу?»
- **Key (ключ):** «что у меня есть?»
- **Value (значение):** «что я передаю дальше?»

Затем для каждой пары слов вычисляется «оценка внимания» — насколько запрос одного слова соответствует ключу другого. Чем выше оценка, тем больше внимания модель уделяет этой паре.

---

### 3.3. Многоголовое внимание (Multi-Head Attention)

Модель не ограничивается одним «фокусом внимания». Она использует **несколько голов внимания параллельно** (обычно 8–16). Каждая голова ищет **разные типы связей**:
- Одна голова ищет грамматические связи (подлежащее-сказуемое).
- Другая — семантические (смысловые) связи.
- Третья — эмоциональные маркеры (например, слова боли, страха).

**Интуитивная аналогия:**

> На консилиуме несколько врачей слушают один и тот же случай. Один обращает внимание на симптомы, другой — на анамнез, третий — на рентгенологические данные. Каждый видит что-то своё. Потом они собираются вместе и формируют целостную картину. Это и есть многоголовое внимание.

---

### 3.4. Позиционное кодирование (Positional Encoding)

Трансформер обрабатывает все слова одновременно, но **порядок слов важен**. Как модель «знает», какое слово за каким идёт?

**Решение:** модель добавляет к каждому слову **позиционный вектор** — специальную метку, которая кодирует его положение в предложении.

> **Техническое уточнение:** на самом деле это не просто порядковый номер (1, 2, 3), а сложный математический паттерн (синусоидальные функции), который позволяет модели различать не только порядок, но и расстояние между словами. Но для понимания сути достаточно аналогии с нумерацией карточек.

**Интуитивная аналогия:**

> Представьте, что вы читаете предложение, где все слова написаны на отдельных карточках, и вам нужно понять смысл. Если карточки перемешаны, смысл теряется. Но если на каждой карточке написан номер, вы можете восстановить порядок. Позиционное кодирование — это те самые номера.

---

### 3.5. Полный путь текста через трансформер

Теперь соберём всё вместе. Вот как текст проходит через трансформер:

```mermaid
flowchart TD
    A["📝 Исходный текст<br>«Пациент жалуется на боль в зубе»"]
    
    B["🔤 Токенизация<br>['пациент', 'жалуется', 'на', 'боль', 'в', 'зубе']"]
    
    C["🔢 Векторы слов (Embeddings)<br>каждое слово → 300 чисел"]
    
    D["📍 Добавление позиции<br>вектор + позиция в предложении"]
    
    E["🧠 Слой 1: Внимание<br>каждое слово «смотрит» на другие"]
    
    F["🧠 Слой 1: Нейросеть<br>уточнение векторов"]
    
    G["🔁 Повтор 6-12 раз<br>каждый слой уточняет смысл"]
    
    H["🎯 Выходные векторы<br>слова в контексте всего предложения"]
    
    A --> B --> C --> D --> E --> F --> G --> H
    
    style A fill:#fff9c4,stroke:#f57f17
    style H fill:#c8e6c9,stroke:#2e7d32
```

**Что происходит на каждом шаге:**

1. **Токенизация:** текст разбивается на минимальные единицы.
2. **Векторы слов:** каждое слово получает «цифровой паспорт».
3. **Позиция:** добавляется информация о порядке слов.
4. **Внимание:** слова «общаются» друг с другом, вычисляя важность связей.
5. **Нейросеть:** каждый вектор уточняется через нейронную сеть.
6. **Повтор:** все шаги повторяются несколько раз, чтобы «глубже» понять смысл.
7. **Выход:** каждый токен теперь «понимает» контекст всего предложения.

## 4. ПРЕДОБУЧЕННЫЕ МОДЕЛИ: BERT И GPT

Трансформер — это архитектура (строительные леса). Конкретные модели (BERT, GPT) — это уже построенные здания, которые **обучены на огромных объёмах текстов**.

---

### 4.1. Что такое «предобучение» (Pre-training)?

**Интуитивная аналогия из медицины:**

> Представьте, что врач получает базовое образование: 6 лет учится, читает тысячи книг, изучает анатомию, физиологию, патологию, практикуется на учебных случаях. Это **предобучение**. После этого он приходит в клинику и учится работать с конкретными пациентами — это **дообучение** (fine-tuning).

То же самое с NLP-моделями:
1. **Предобучение:** модель читает миллиарды текстов (медицинские статьи, учебники, истории болезни, форумы) и учится предсказывать слова, понимать контекст, улавливать связи. Это занимает недели и требует огромных вычислительных ресурсов.
2. **Дообучение (fine-tuning):** модель адаптируется под конкретную задачу (например, диагностика периодонтита по описанию) на специальном размеченном наборе данных. Это занимает часы и доступно даже на обычном компьютере.

**Почему это гениально:** вместо того чтобы учить модель с нуля для каждой задачи (что дорого и долго), мы берём уже «образованного помощника» и просто «показываем ему специфику» стоматологии.

---

### 4.2. BERT (Bidirectional Encoder Representations from Transformers)

**Что это:** модель, разработанная Google в 2018 году [2]. Она **читает текст целиком** (слева направо и справа налево) и понимает контекст каждого слова с обеих сторон.

**Ключевая особенность:** BERT — **двунаправленный** (bidirectional). Он учитывает слова, которые стоят **до и после** целевого слова.

**Интуитивная аналогия:**

> Врач слушает пациента и понимает смысл его слов не только из того, что он сказал до этого, но и из того, что он скажет дальше. BERT работает так же: чтобы понять слово *«периодонтит»*, он смотрит и на предыдущие слова, и на последующие.

**Пример из стоматологии:**

> **Текст:** *«Пациент сообщил о сильной [MASK] и отёке десны»*.
>
> BERT, прочитав весь текст, «понимает», что на месте [MASK] должно быть слово *«боли»* (или *«пульсации»*), потому что оно связано с отёком десны. Он учитывает контекст с обеих сторон.

**Применение BERT в стоматологии:**

| Задача | Пример |
|--------|--------|
| **Извлечение симптомов** | Из записи в амбулаторной карте извлечь все упоминания симптомов (боль, отёк, кровоточивость, подвижность) |
| **Классификация текстов** | Определить, относится ли жалоба к острому воспалению или плановому осмотру |
| **Анализ тональности** | Оценить уровень боли или тревожности по тону сообщения пациента в чате |

Исследования показывают впечатляющие результаты: BERT достигает точности **93%** при обнаружении эмоционального дистресса (например, дентофобии), что значительно превосходит предыдущие архитектуры.

---

### 4.3. GPT (Generative Pre-trained Transformer)

**Что это:** модель, разработанная OpenAI (первая версия в 2018, затем GPT-2, GPT-3, GPT-4). В отличие от BERT, GPT — **генеративная** модель. Она не просто анализирует текст, а **генерирует** его.

**Ключевая особенность:** GPT — **однонаправленный** (слева направо). Он предсказывает следующее слово на основе предыдущих. Это делает его идеальным для генерации текста.

**Интуитивная аналогия:**

> Представьте, что пациент говорит: *«Доктор, у меня болит зуб...»*. Хороший врач может предсказать, что дальше будет *«при накусывании»* или *«от холодного»*, опираясь на предыдущий опыт. GPT делает то же самое: он предсказывает следующее слово, опираясь на все предыдущие.

**Пример из стоматологии:**

> **Запрос:** *«Пациент с жалобами на боль в зубе при накусывании, отёк десны»*
>
> GPT может сгенерировать продолжение: *«...На рентгенограмме определяется очаг разрежения костной ткани в области верхушки корня. Диагноз: хронический периодонтит в стадии обострения. Рекомендовано: эндодонтическое лечение...»* — текст, который звучит как заключение врача.

**Применение GPT в стоматологии:**

| Задача | Пример |
|--------|--------|
| **Чат-боты** | Диалог с пациентом для первичной консультации, ответы на вопросы |
| **Генерация резюме** | Составить краткую выжимку из длинной истории болезни |
| **Вопросно-ответные системы** | Ответить на вопрос врача: «Какие факторы риска у этого пациента?» |

**Важное уточнение:** границы между BERT и GPT сейчас стираются. Современные GPT (как GPT-4) отлично понимают контекст и могут извлекать симптомы не хуже BERT, если им правильно дать инструкцию (промпт). Однако BERT легче, быстрее и дешевле для задач классификации, а GPT лучше для диалога и генерации.

---

## 5. КАК МОДЕЛЬ ПРИНИМАЕТ РЕШЕНИЕ: ОТ ТЕКСТА К ДИАГНОЗУ

Теперь, когда мы знаем все компоненты, давайте проследим **полный путь** текста пациента через модель до получения результата. Возьмём конкретную задачу: модель классифицирует текст на «острое воспаление» или «плановый осмотр».

---

### 5.1. Пошаговый разбор

**Шаг 0. Исходный текст жалобы**

```
«У меня опухла щека, зуб болит, особенно ночью, температура 37.8»
```

**Шаг 1. Токенизация — разбивка на токены**

Модель разбивает текст на минимальные единицы.

| Исходное слово | Токены (как видит модель) |
|----------------|---------------------------|
| «опухла» | `['опу', '##хла']` |
| «болит» | `['бол', '##ит']` |
| «температура» | `['темп', '##ера', '##тура']` |

**Зачем это нужно:**
- Модель может обрабатывать слова, которых нет в словаре (например, сленг «флюс» → `['ф', '##люс']`).
- Уменьшается размер словаря (не 100 000 слов, а 30 000 подслов).

**Шаг 2. Превращение токенов в числа (Embedding)**

Каждый токен заменяется на **вектор** — список чисел (обычно 300–768 чисел).

```
«опу» → [0.12, -0.34, 0.56, 0.78, -0.12, ...]
«##хла» → [0.45, 0.23, -0.78, 0.11, 0.34, ...]
«бол» → [0.67, -0.12, 0.34, 0.56, -0.23, ...]
```

**Шаг 3. Добавление позиционного кодирования**

К каждому вектору прибавляется позиционный вектор.

```
Токен 1 («опу») + позиция 1
Токен 2 («##хла») + позиция 2
Токен 3 («бол») + позиция 3
```

**Шаг 4. Проход через слои трансформера**

Векторы проходят через 6–12 слоёв. В каждом слое:

**4А. Многоголовое внимание**

Модель вычисляет, насколько каждое слово связано с каждым другим.

> **Пример:** слово *«болит»* получает высокий вес связи со словом *«ночью»* (потому что ночная боль — характерный симптом пульпита) и низкий вес связи со словом *«щека»* (хотя связь есть, она менее специфична).

**4Б. Нормализация и нейросеть**

Векторы уточняются через нейронную сеть.

**Шаг 5. Финальный этап: классификация**

У BERT есть специальный токен `[CLS]` (от слова *classification*), который «вбирает» в себя смысл всего предложения. Его вектор подаётся на финальный классификатор.

```
Вектор [CLS] → нейронная сеть → вероятность классов
```

**Результат:** `[острое_воспаление: 0.92, плановый_осмотр: 0.08]`

### 5.2. Визуализация полного пути

```mermaid
flowchart TD
    A["📝 Текст жалобы<br>«Опухла щека, зуб болит ночью»"]
    B["🔤 Токенизация<br>['опу', '##хла', 'щека', 'зуб', 'бол', '##ит', 'ночью']"]
    C["🔢 Векторы (Embeddings)<br>каждый токен → 300 чисел"]
    D["📍 Позиционное кодирование<br>вектор + позиция в тексте"]
    E["🧠 Слой 1: Внимание<br>связи между словами"]
    F["🧠 Слой 1: Нейросеть<br>уточнение векторов"]
    G["🔁 Повтор 6-12 раз<br>глубокое понимание контекста"]
    H["🎯 Выход: [CLS] токен<br>«впитал» смысл всего текста"]
    I["📊 Классификатор<br>нейросеть → вероятности"]
    J["✅ Результат<br>острое_воспаление: 0.92"]
    
    A --> B --> C --> D --> E --> F --> G --> H --> I --> J
```

### 6.7. Практический пример: как найти семантических соседей в медицинских данных

Давайте посмотрим, как Word2Vec и современные эмбеддинги могут помочь врачу **находить смысловые связи** между словами в клинических текстах.

**Код для демонстрации:**

```python
# Установка библиотек (выполняется один раз)
# pip install gensim sentence-transformers

import gensim.downloader as api
from sentence_transformers import SentenceTransformer
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# ------------------------------------------------------------
# 1. КЛАССИЧЕСКИЙ Word2Vec: находим семантических соседей
# ------------------------------------------------------------
print("=" * 60)
print("1. ПОИСК СЕМАНТИЧЕСКИХ СОСЕДЕЙ (Word2Vec)")
print("=" * 60)

# Загружаем предобученную модель для русского языка
russian_model = api.load("word2vec-ruscorpora-300")

# Находим 10 слов, наиболее близких по смыслу к слову «пульпит»
similar_words = russian_model.most_similar('пульпит_NOUN', topn=10)
print("Слова, близкие по смыслу к 'пульпит':")
for word, score in similar_words:
    print(f"  {word} — {score:.3f}")

# ------------------------------------------------------------
# 2. СОВРЕМЕННЫЙ ПОДХОД: эмбеддинги предложений
# ------------------------------------------------------------
print("\n" + "=" * 60)
print("2. ПОИСК СМЫСЛОВЫХ СОСЕДЕЙ (Sentence Transformers)")
print("=" * 60)

# Загружаем модель для сравнения предложений
model = SentenceTransformer('sergeyzh/rubert-large-uncased-sts')

# Клинические фразы для поиска соседей
query = "Пациент жалуется на сильную боль в зубе при накусывании"
candidates = [
    "Больной отмечает острую боль при жевании в области моляра",
    "Пациент сообщает о пульсирующей боли ночью",
    "Пациент чувствует себя хорошо, зуб не беспокоит",
    "Жалобы на отёк десны и боль при накусывании",
    "Пациент говорит о повышенной чувствительности к холодному"
]

# Получаем векторы для всех фраз
embeddings = model.encode([query] + candidates)
query_emb = embeddings[0]
candidate_embs = embeddings[1:]

# Вычисляем сходство
similarities = cosine_similarity([query_emb], candidate_embs)[0]

print(f"Запрос: {query}")
print("\nПохожие фразы (от наиболее к наименее похожим):")
for idx in sorted(range(len(similarities)), key=lambda i: similarities[i], reverse=True):
    print(f"  Сходство {similarities[idx]:.3f}: {candidates[idx]}")
```

**Что мы видим в результате:**

1. **Word2Vec** показывает, что *«пульпит»* семантически близок к *«периодонтит»*, *«кариес»*, *«воспаление»*. Это помогает понять, какие термины автоматически связываются в медицинских текстах.

2. **Sentence Transformers** показывает, что фраза про боль при накусывании наиболее похожа на другие фразы про боль при жевании, и совсем не похожа на фразу про хорошее самочувствие.

**Как это применить в стоматологии:**

| Возможность | Пример использования |
|-------------|---------------------|
| **Поиск похожих клинических случаев** | Введите описание симптомов пациента → модель найдёт в базе другие случаи с похожей симптоматикой |
| **Выявление скрытых ассоциаций** | Если у пациента слово *«бормашина»* семантически близко к *«страх»* и *«боль»* (а не к *«лечение»* и *«необходимость»*), это может быть маркером дентофобии |
| **Отслеживание динамики** | Сравните эмбеддинги записей пациента до и после лечения — язык стал ближе к кластеру «выздоровление»? |
| **Автоматическая категоризация** | Группируйте похожие жалобы пациентов для анализа распространённости симптомов |

## 7. ПРИМЕР В СТОМАТОЛОГИИ: ИЗВЛЕЧЕНИЕ СИМПТОМОВ

Рассмотрим **конкретный клинический пример** полного цикла работы NLP-модели.

---

### 7.1. Постановка задачи

Стоматологическая клиника получает сотни сообщений от пациентов через онлайн-чат и записи в амбулаторных картах. Каждая запись — это текст разной длины. Задача: **автоматически извлекать из всех текстов структурированный список симптомов** для быстрой сортировки пациентов по срочности и предварительной диагностики.

---

### 7.2. Исходные данные

**Сообщение пациента (анонимизировано):**

> *«Здравствуйте, у меня болит зуб, особенно когда кусаю, десна опухла, и щека припухла. Боль ноющая, ночью усиливается. Температура 37.4. Что мне делать?»*

---

### 7.3. Этапы обработки с использованием конкретных моделей

**Шаг 1. Очистка текста (Лекция 3)**
- Токенизация, удаление стоп-слов (с сохранением отрицаний), лемматизация.

**Шаг 2. Извлечение симптомов**
- Используем модель **RuBert-tiny2-EmotionsDetected** для выявления эмоциональных маркеров (страх, тревога).
- Используем **rubert-tiny2-mtl-conflict** для анализа интенсивности эмоций и триггеров (например, боль при накусывании как триггер).
- Модель выделяет все упоминания симптомов.

**Шаг 3. Классификация симптомов**
- Используем **rubert-emotion-ru-large** для классификации эмоций (страх перед лечением).
- Каждый симптом относится к категории:
    - **Болевые:** боль при накусывании, ноющая боль, ночное усиление.
    - **Воспалительные:** отёк десны, припухлость щеки.
    - **Системные:** температура 37.4.

**Шаг 4. Оценка срочности**
- Используем **MindGuard-8B** для оценки безопасности (здесь нет суицидального риска, но можно адаптировать для оценки неотложности).
- Модель классифицирует состояние как «требует срочного осмотра» на основе совокупности симптомов.

**Шаг 5. Структурированный результат**

| Тип симптома | Конкретный симптом | Выраженность (оценка модели) |
|--------------|-------------------|------------------------------|
| Болевые | Боль при накусывании | Высокая (8/10) |
| Болевые | Ноющая боль | Средняя (6/10) |
| Болевые | Ночное усиление | Высокая (7/10) |
| Воспалительные | Отёк десны | Высокая (7/10) |
| Воспалительные | Припухлость щеки | Средняя (5/10) |
| Системные | Температура 37.4 | Лёгкая (субфебрильная) |

**Оценка срочности:** ⚠️ Требуется осмотр в ближайшие 24 часа (подозрение на острый периодонтит или периостит).

---

### 7.4. Почему это ценно для стоматолога

- **Экономия времени:** не нужно перечитывать каждое сообщение.
- **Стандартизация:** модель выделяет симптомы единообразно, без пропусков.
- **Приоритизация:** пациенты с острыми состояниями могут быть направлены к врачу в первую очередь.
- **Поддержка принятия решений:** структурированный список помогает быстро оценить состояние.
- **Безопасность:** автоматическая оценка неотложности помогает не пропустить серьёзные осложнения.

```python
# Установка библиотек (выполняется один раз)
# pip install transformers torch sentence-transformers scikit-learn

from transformers import pipeline
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# ------------------------------------------------------------
# ПРИМЕР 1. АНАЛИЗ ЭМОЦИЙ В ТЕКСТЕ ПАЦИЕНТА
# ------------------------------------------------------------
print("=" * 60)
print("1. АНАЛИЗ ЭМОЦИЙ В ТЕКСТЕ ПАЦИЕНТА")
print("=" * 60)

# Загружаем модель для классификации эмоций
emotion_classifier = pipeline(
    "text-classification",
    model="ilyali034/rubert-emotion-ru-large",
    tokenizer="ilyali034/rubert-emotion-ru-large",
    top_k=None  # показываем все эмоции с вероятностями
)

# Текст пациента (реалистичный пример страха перед лечением)
text = "Я очень боюсь идти к стоматологу, у меня паника, сердце колотится, не могу спать"
result = emotion_classifier(text)

print(f"Текст пациента: {text}")
print("\nЭмоции (вероятности):")
for emotion in result[0]:
    if emotion['score'] > 0.1:  # показываем только значимые
        print(f"  {emotion['label']}: {emotion['score']:.3f}")

print("\nИнтерпретация для врача:")
print("- Высокая вероятность страха (fear) — маркер дентофобии")
print("- Тревога (anxiety) — дополнительный компонент")
print("- Рекомендовано: щадящий подход, возможно, премедикация")

# ------------------------------------------------------------
# ПРИМЕР 2. ПОИСК ПОХОЖИХ КЛИНИЧЕСКИХ СЛУЧАЕВ
# ------------------------------------------------------------
print("\n" + "=" * 60)
print("2. ПОИСК ПОХОЖИХ КЛИНИЧЕСКИХ СЛУЧАЕВ")
print("=" * 60)

# Загружаем модель для сравнения текстов
sts_model = SentenceTransformer('sergeyzh/rubert-large-uncased-sts')

# База клинических случаев (гипотетическая)
cases = [
    "Пациент жалуется на острую боль в зубе при накусывании, отёк десны.",
    "Больной отмечает ноющую боль, усиливающуюся ночью.",
    "Пациент сообщает о повышенной чувствительности к холодному.",
    "Пациент чувствует себя хорошо, зуб не беспокоит.",
    "Жалобы на подвижность зуба и кровоточивость дёсен."
]

# Новый пациент
new_case = "Пациент жалуется на боль при жевании и припухлость щеки."

# Получаем векторы для всех текстов
embeddings = sts_model.encode([new_case] + cases)

# Сравниваем новый случай с каждым из базы
similarities = cosine_similarity([embeddings[0]], embeddings[1:])[0]

print(f"Новый случай: {new_case}")
print("\nПохожие случаи (от наиболее к наименее похожим):")
for idx in sorted(range(len(similarities)), key=lambda i: similarities[i], reverse=True):
    print(f"  Сходство {similarities[idx]:.3f}: {cases[idx]}")

print("\nИнтерпретация для врача:")
print("- Найден случай с очень похожей симптоматикой (боль при жевании + отёк)")
print("- Рекомендовано изучить тактику лечения в похожем случае")
print("- Обратить внимание на возможный диагноз: острый периодонтит")

# ------------------------------------------------------------
# ПРИМЕР 3. ПОИСК СЕМАНТИЧЕСКИХ СОСЕДЕЙ ДЛЯ КЛИНИЧЕСКИХ ТЕРМИНОВ
# ------------------------------------------------------------
print("\n" + "=" * 60)
print("3. ПОИСК СЕМАНТИЧЕСКИХ СОСЕДЕЙ ДЛЯ ТЕРМИНОВ")
print("=" * 60)

# Функция для поиска ближайших слов по смыслу
def find_semantic_neighbors(word, model, top_n=10):
    terms = [
        "пульпит", "периодонтит", "абсцесс", "отёк",
        "кариес", "эмаль", "дентин", "гингивит",
        "боль", "страх", "тревога", "анестезия",
        "отбеливание", "винир", "коронка", "имплантат"
    ]
    
    word_emb = model.encode([word])
    term_embs = model.encode(terms)
    
    similarities = cosine_similarity(word_emb, term_embs)[0]
    results = sorted(zip(terms, similarities), key=lambda x: x[1], reverse=True)
    return results[:top_n]

word_model = SentenceTransformer('sergeyzh/rubert-large-uncased-sts')

print("Слова, близкие по смыслу к 'периодонтит':")
neighbors = find_semantic_neighbors("периодонтит", word_model, top_n=8)
for term, score in neighbors:
    print(f"  {term}: {score:.3f}")

print("\nИнтерпретация для врача:")
print("- Модель 'понимает', что 'периодонтит' ближе к 'абсцесс' и 'отёк'")
print("- 'Пульпит' связан, но это разные заболевания")
print("- 'Отбеливание' — далеко, как и ожидалось")
print("- Это помогает автоматически группировать жалобы пациентов по семантическим кластерам")
```

**Пояснение для студентов:**

| Что делает код | Что это значит для врача |
|----------------|--------------------------|
| **Анализ эмоций** | Модель «читает» текст пациента и определяет, какие эмоции в нём присутствуют (страх, тревога и т.д.). Врач получает объективную оценку эмоционального состояния, что важно для выбора подхода (например, седация при дентофобии) |
| **Поиск похожих случаев** | Модель находит в базе другие записи с похожей симптоматикой. Врач может изучить, какая тактика была эффективна в похожих случаях |
| **Семантические соседи** | Модель показывает, какие термины близки по смыслу. Врач может увидеть, как связаны клинические понятия в текстах |

## 9. ЭТИЧЕСКИЕ РИСКИ СОВРЕМЕННЫХ NLP-МОДЕЛЕЙ

Этот раздел критически важен. Мощные модели — это мощные инструменты, но они несут риски.

---

### 9.1. Галлюцинации (Hallucinations)

**Что это:** модель генерирует информацию, которой нет в исходных данных. Она «выдумывает» факты, но выдаёт их за правду.

**Пример в стоматологии:**

> Модель читает историю пациента и на вопрос *«Была ли у пациента аллергия на лидокаин?»* отвечает *«Да, у пациента аллергия на лидокаин»*, хотя в исходном тексте этого не было. Модель просто «придумала» ответ.

**Как с этим бороться:**
- Использовать **RAG (Retrieval-Augmented Generation)** — модель сначала ищет ответ в вашей базе знаний (картах, протоколах), а только потом генерирует ответ, цитируя источник.
- Всегда проверять ответы модели (человеческий контроль).
- Использовать модели с функцией уверенности.

---

### 9.2. Предвзятость (Bias)

**Что это:** модель может быть предвзятой, если обучающие данные содержат стереотипы.

**Пример в стоматологии:**

> Если модель обучена на данных, где большинство пациентов — мужчины с определённым типом патологии, она может ошибаться при анализе текстов женщин или пациентов с другой патологией.

**Как с этим бороться:**
- Использовать разнообразные обучающие данные.
- Проверять модель на разных группах пациентов.
- Понимать ограничения модели и не доверять ей слепо.

---

### 9.3. Конфиденциальность

**Что это:** модели обучаются на данных, которые могут содержать личную информацию. Если модель используется в облаке, данные могут передаваться третьим лицам.

**Решение:** использовать **on-premise** (локальное) развёртывание моделей, где данные не покидают вашу организацию. Многие открытые модели (Serenity-Llama, Llama3-MAGneT) оптимизированы для локального использования. В стоматологии, где данные включают рентгеновские снимки и персональные данные, это особенно важно.

---

### 9.4. Проблема «чёрного ящика»

**Что это:** модели работают как «чёрный ящик» — мы не можем объяснить, почему они выдали именно такой результат.

**Пример в стоматологии:**

> Модель ставит предварительный диагноз «периодонтит». Но почему? Какие слова в тексте пациента привели к этому выводу? Мы не знаем.

**Решение:** использовать интерпретируемые инструменты, такие как **TONY** [6], который объединяет интерпретируемые лексические признаки с современными трансформерами. Также можно использовать методы объяснения (например, SHAP, LIME) для анализа вклада отдельных слов.

---

## 10. ЗАКЛЮЧЕНИЕ

- Трансформер — это архитектура, которая **обрабатывает все слова одновременно** и **фокусируется на ключевых связях** через механизм внимания.
- BERT — двунаправленная модель для **понимания текста** (классификация, извлечение, анализ).
- GPT — генеративная модель для **создания текста** (чат-боты, резюме, ответы на вопросы).
- В 2026 году доступны десятки открытых моделей для медицинских задач — от классификации эмоций до генерации текстов. Для стоматологии можно использовать существующие русскоязычные модели с последующим дообучением на специализированных данных.
- Исследования показывают, что открытые LLM (Gemma-3-27B, Llama-3.3-70B) могут достигать F1-score 0.818–0.865 при дообучении на клинических данных [5].
- **Этический аспект:** модели могут ошибаться (галлюцинации), быть предвзятыми и нарушать конфиденциальность. Использование требует **человеческого контроля** и **критического мышления**.

**Связь с предыдущими лекциями:**
- Лекция 3 (очистка текста) → подготовка данных для модели.
- Лекция 5 (сегодня) → анализ данных моделью.
- Лекция 4 (визуализация) → представление результатов.

**Ключевая мысль:** современные NLP-модели — это **мощные помощники** врача, но не замена. Они расширяют наши возможности, но ответственность за решения всегда остаётся за человеком.
