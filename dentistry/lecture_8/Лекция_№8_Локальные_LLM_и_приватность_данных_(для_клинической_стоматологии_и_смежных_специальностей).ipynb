### 1.1. От облачных моделей к локальным: почему это важно для врача-стоматолога

В Лекции 7 мы научились создавать веб-приложения, которые использовали облачные модели через Hugging Face Inference API. Это было удобно и быстро — не нужно было загружать модели на свой компьютер.

**Но в медицинской практике есть критическое ограничение: данные пациентов нельзя отправлять на сторонние серверы.**

Любая система, работающая с естественным языком, почти неизбежно попадает под действие 152‑ФЗ: пользователи всё равно вводят ФИО, телефоны, e‑mail и другие идентифицирующие данные, даже если на экране написано «не вводите персональные данные». Юридически это означает, что вы как оператор обязаны обеспечить защиту этих данных.

**Интуитивная аналогия из стоматологической практики:**

> Представьте, что вы ведёте приём и записываете всё, что говорит пациент, на диктофон. Но вместо того чтобы хранить записи в сейфе в своём кабинете, вы отправляете их в неизвестную компанию «для анализа». Нарушена ли врачебная тайна? Безусловно. Именно это происходит, когда вы отправляете жалобы, описания симптомов или эпикризы в облачный ChatGPT или аналогичный сервис.

**Проблема облачных сервисов:**

- Данные передаются третьим лицам.
- Нет гарантии, что они не используются для обучения моделей.
- Нарушается 152-ФЗ «О персональных данных» — данные не покидают территорию РФ без согласия.
- Нарушается врачебная тайна (ст. 13 ФЗ № 323-ФЗ «Об основах охраны здоровья граждан в РФ»).

**Решение: локальные LLM** — модели, которые работают на вашем компьютере. Данные **никогда не покидают ваше устройство**. Всё происходит офлайн.

**Что мы узнаем сегодня:**

- Какие преимущества даёт локальный запуск LLM для врача.
- Как установить и настроить **Ollama** — платформу для запуска локальных моделей.
- Как подключить локальные модели к **VS Code** для работы с медицинскими текстами.

- Как настроить **системные промпты** для работы в стоматологическом и общемедицинском контексте.

**Ключевая мысль:** локальные LLM позволяют использовать мощь ИИ, **сохраняя полный контроль над данными пациентов**. Это единственный этически и юридически приемлемый способ применения генеративного ИИ в клинической медицине.

### 1.2. Актуальность для стоматологии и смежных областей

Согласно исследованиям [7], количество врачей, использующих ИИ-инструменты в работе, продолжает расти. Однако большинство из них используют облачные сервисы, не осознавая рисков для конфиденциальности пациентов.

**Ситуация в России:**

- 152-ФЗ «О персональных данных» требует, чтобы обработка персональных данных осуществлялась с соблюдением принципов законности и конфиденциальности [2].
- С 1 сентября 2025 года вступили в силу поправки к 152-ФЗ, касающиеся взаимодействия персональных данных с цифровыми сервисами и искусственным интеллектом.
- Врачебная тайна (ст. 13 ФЗ № 323-ФЗ) запрещает разглашение сведений, составляющих врачебную тайну, без согласия пациента.
- Приказ Минздрава России № 206н от 2023 г. регламентирует использование ИИ в здравоохранении, требуя защиты данных.

**Вывод:** если вы отправляете текст жалобы пациента в облачный API — вы **передаёте** персональные данные и сведения, составляющие врачебную тайну, третьему лицу. Это нарушает 152-ФЗ и ФЗ № 323-ФЗ без явного согласия пациента и договора с оператором. Локальный запуск **полностью исключает** эту проблему.

**Для смежных специальностей:**

- **Неонатолог:** данные о новорождённых особенно чувствительны. Любая передача в облако недопустима. Локальные LLM могут помочь в формировании выписок, анализе дневников наблюдений.
- **Рентгенолог:** заключения по снимкам содержат персональные данные и клинические детали. Локальные модели позволяют автоматически структурировать заключения без риска утечки.
- **Терапевт:** амбулаторные карты, эпикризы — всё это защищено законом. Локальный ИИ может помочь в суммаризации длинных записей.
- **Микробиолог:** результаты антибиотикограмм, описания культуральных свойств — важная информация, которая не должна покидать лабораторию.

## 2. ПРЕИМУЩЕСТВА ЛОКАЛЬНОГО ЗАПУСКА

### 2.1. Приватность и безопасность — главный аргумент

| Критерий | Облачные сервисы (ChatGPT, Claude) | Локальные LLM (Ollama) |
|----------|-------------------------------------|------------------------|
| **Данные покидают компьютер** | ✅ Да | ❌ Нет |
| **Контроль над данными** | У провайдера | Полный, у вас |
| **Соответствие 152-ФЗ** | ❌ Нарушается | ✅ Соблюдается |
| **Врачебная тайна** | ❌ Нарушается | ✅ Соблюдается |
| **Необходимость интернета** | ✅ Да | ❌ Нет (после установки) |
| **Плата за использование** | ✅ Да (или лимиты) | ❌ Нет (бесплатно) |
| **Зависимость от сервиса** | ✅ Да | ❌ Нет |

**Интуитивная аналогия:**

> Локальная LLM — это как вести приём в своём закрытом кабинете. Облачная LLM — как вести приём в открытом коворкинге, где каждый может подойти и послушать.

### 2.2. Другие преимущества локального запуска

- **Экономия:** бесплатно, нет подписок и лимитов на запросы.
- **Независимость:** не нужно интернет-соединение (после загрузки модели).
- **Скорость:** мгновенный отклик (зависит от мощности вашего компьютера).
- **Кастомизация:** можно дообучать модели на своих данных (в рамках приватности).
- **Контроль версий:** вы используете ту модель, которую выбрали, а не ту, которую обновил провайдер.
- **Совместимость с медицинскими стандартами:** возможность настройки промптов под конкретные клинические протоколы.

### 2.3. Схема: локальный vs облачный запуск

```mermaid
flowchart TD
    subgraph CLOUD["☁️ ОБЛАЧНЫЙ ЗАПУСК"]
        direction LR
        C1["Ваш компьютер"] --> C2["Интернет"] --> C3["Сервер провайдера"]
        C3 --> C4["Данные хранятся<br>у третьей стороны"]
        C4 --> C5["❌ Нарушение 152-ФЗ и врачебной тайны"]
    end
    
    subgraph LOCAL["💻 ЛОКАЛЬНЫЙ ЗАПУСК"]
        direction LR
        L1["Ваш компьютер"] --> L2["Ollama + модель"]
        L2 --> L3["Данные НЕ покидают<br>устройство"]
        L3 --> L4["✅ Соответствие законам и врачебной тайне"]
    end
    
    CLOUD -->|Проблема| LOCAL
    
    style CLOUD fill:#ffcdd2,stroke:#c62828
    style LOCAL fill:#c8e6c9,stroke:#2e7d32
```

## 3.1. Что такое Ollama

**Ollama** — это бесплатная платформа с открытым исходным кодом для запуска больших языковых моделей локально. Она позволяет:

- Скачивать (pull) модели одной командой.
- Запускать их на своём компьютере.
- Взаимодействовать с моделями через API (как с облачным сервисом, но локально).

**Интуитивная аналогия:**

> Ollama — это как «магазин приложений» для ИИ-моделей. Вы выбираете модель, скачиваете её на свой компьютер и запускаете. Всё просто, как установка программы.

**Что делает Ollama уникальным:**

- **Простота:** одна команда для скачивания и запуска модели.
- **Универсальность:** работает на Windows, macOS и Linux.
- **Гибкость:** поддерживает множество моделей (Qwen, DeepSeek, Llama, Mistral и др.).
- **API:** предоставляет REST API, совместимый с форматом OpenAI.
- **Кеширование:** модели хранятся локально и загружаются быстро при повторных запусках.

**Системные требования:**

| Требование | Минимальные | Рекомендуемые |
|------------|-------------|---------------|
| **Оперативная память (RAM)** | 8 GB | 16+ GB |
| **Свободное место на диске** | 10 GB | 20+ GB |
| **Видеокарта (GPU)** | Не требуется | NVIDIA GPU с 8+ GB VRAM |
| **Операционная система** | Windows 10+, macOS 11+, Linux | Windows 11, macOS 13+, Ubuntu 22+ |

**Важно:** модели 7B–8B работают на 12–16 GB RAM в режиме CPU. Для ускорения рекомендуется наличие NVIDIA GPU.

### 3.2.1. Установка на Windows

**Пошаговая инструкция:**

**Шаг 1. Скачайте установщик**

1. Откройте браузер и перейдите на [ollama.com/download](https://ollama.com/download).
2. Нажмите **"Download for Windows"**.
3. Файл `OllamaSetup.exe` (примерно 50–100 MB) начнёт скачиваться.
4. Дождитесь завершения загрузки.

**Шаг 2. Запустите установку**

1. Найдите скачанный файл (обычно в папке "Загрузки").
2. Дважды щёлкните по файлу `OllamaSetup.exe`.
3. Если появится предупреждение системы безопасности (UAC), нажмите **"Да"** или **"Run anyway"**.

**Шаг 3. Следуйте инструкциям установщика**

1. Нажмите **"Install"** — установка начнётся.
2. Подождите, пока установка завершится (обычно 1–2 минуты).
3. **Важно:** установщик автоматически добавит Ollama в PATH, чтобы команды работали из терминала.
4. После установки нажмите **"Finish"**.

**Шаг 4. Проверьте, что Ollama запустился**

1. В правом нижнем углу экрана (в трее) появится иконка Ollama (собачка).
2. Это означает, что Ollama работает в фоновом режиме.
3. Если иконки нет — запустите Ollama вручную через меню "Пуск" → "Ollama".

**Шаг 5. Проверьте установку в терминале**

1. Нажмите **Win + R**, введите `cmd` и нажмите Enter.
2. Введите команду:

```bash
ollama --version
# Должно вывести: ollama version 0.x.x
```

**Шаг 6. Если команда не найдена**

1. **Перезагрузите компьютер** — это обновит переменные PATH.
2. Если не помогло, откройте терминал от имени администратора:
   - Нажмите **Win**, найдите "Командная строка".
   - Нажмите правой кнопкой → **"Запуск от имени администратора"**.
3. Попробуйте снова `ollama --version`.

## 3.4. Скачивание (pull) моделей

Ollama использует команду `ollama pull` для загрузки моделей.

**Примерное время загрузки моделей (при скорости 50 Мбит/с):**

| Модель | Размер | Время загрузки |
|--------|--------|----------------|
| Qwen2.5:0.5B | ~2 GB | ~5 минут |
| Qwen2.5:7B | ~8 GB | ~20 минут |
| Qwen2.5:14B | ~16 GB | ~40 минут |

**Базовые команды:**

```bash
# Показать список установленных моделей
ollama list

# Скачать модель Qwen2.5:7B (рекомендуемая)
ollama pull qwen2.5:7b

# Скачать модель DeepSeek-R1:7B
ollama pull deepseek-r1:7b

# Скачать модель Llama3.2:8B
ollama pull llama3.2:8b

# Скачать модель Mistral:7B
ollama pull mistral:7b
```

**Запуск модели в интерактивном режиме:**

```bash
# Запуск модели для диалога
ollama run qwen2.5:7b

# После запуска можно писать сообщения
# Введите текст и нажмите Enter
# Для выхода введите /bye
```

**Пример диалога в терминале:**

```
>>> Привет, как дела?
Привет! Я работаю на русском языке и готов помочь вам.
Что вы хотели бы обсудить?

>>> Что ты умеешь?
Я могу отвечать на вопросы, помогать с анализом текстов,
генерировать идеи и многое другое.

>>> /bye
```

**Другие полезные команды:**

```bash
# Показать информацию о модели
ollama show qwen2.5:7b

# Удалить модель (освободить место)
ollama rm qwen2.5:7b

# Остановить фоновый сервис
ollama stop

# Перезапустить фоновый сервис
ollama restart
```

## 3.5. Схема: архитектура Ollama

```mermaid
flowchart LR
    subgraph USER["🧑 Пользователь (врач)"]
        U1["Терминал<br>ollama run"]
        U2["VS Code<br>через расширение"]
        U3["Streamlit<br>через API"]
        U4["Python<br>через requests"]
    end
    
    subgraph OLLAMA["⚙️ Ollama (localhost:11434)"]
        O1["API сервер<br>REST API"]
        O2["Менеджер моделей<br>на диске"]
        O3["Движок инференса<br>CPU / GPU"]
    end
    
    subgraph MODEL["🧠 Модели на диске"]
        M1["Qwen2.5<br>~8 GB"]
        M2["DeepSeek-R1<br>~8 GB"]
        M3["Llama 3.2<br>~8 GB"]
        M4["Mistral<br>~8 GB"]
    end
    
    USER --> OLLAMA --> MODEL
    
    style USER fill:#e8f5e9,stroke:#2e7d32
    style OLLAMA fill:#e1f5fe,stroke:#01579b
    style MODEL fill:#fff9c4,stroke:#f57f17
```

**Пояснение к схеме:**

- **Ollama** работает как локальный сервер на порту 11434.
- Вы можете взаимодействовать с ним через терминал, VS Code, Streamlit или Python.
- Модели скачиваются и хранятся на диске.
- При запуске модель загружается в память и обрабатывает запросы.
- Ollama автоматически использует GPU, если он доступен (NVIDIA CUDA).

| ✅ VS Code расширение | В Chat выбран Ollama | Модель доступна |

## 6.1. Что такое системный промпт и зачем он нужен

**Системный промпт (system prompt)** — это инструкция, которая задаёт модели **роль, стиль и правила поведения** на протяжении всего диалога. Это не вопрос и не запрос — это «установка» модели, которая действует как постоянный контекст для всех ответов.

**Интуитивная аналогия из медицинской практики:**

> Системный промпт — это как должностная инструкция для ассистента врача. Перед началом работы вы объясняете: «Вы будете помогать мне с документацией, но не будете ставить диагнозы. Вы будете напоминать пациенту о необходимости консультации, но не давать медицинских рекомендаций». Модель получает такую же «установку» перед началом работы.

**Разница между системным и пользовательским промптом:**

| Тип промпта | Кто задаёт | Что определяет | Когда действует |
|-------------|------------|----------------|-----------------|
| **Системный** | Разработчик/врач | Роль, правила, стиль | Весь диалог |
| **Пользовательский** | Пользователь/врач | Конкретный вопрос или запрос | Текущее сообщение |

**Почему системный промпт критически важен для врачей:**

| Причина | Объяснение |
|---------|------------|
| **Этическая безопасность** | Устанавливает границы — модель не пытается ставить диагнозы без контроля |
| **Последовательность** | Модель ведёт себя одинаково на протяжении всей работы |
| **Профессиональный тон** | Задаёт корректный, медицинский, но доступный стиль |
| **Безопасность пациента** | При неотложных состояниях модель знает, как направить к врачу |
| **Соответствие законодательству** | Помогает соблюдать требования к ИИ-инструментам в медицине |

**Важное исследование:** чрезмерно детализированный системный промпт может ограничить гибкость чат-бота, поэтому важен баланс между конкретностью инструкций и свободой модели [6]. Слишком жёсткие правила делают ответы шаблонными, слишком свободные — могут привести к нежелательным высказываниям.

## 6.2. Базовый системный промпт для ассистента стоматолога

Этот промпт подходит для общего использования — когда модель помогает анализировать жалобы, структурировать записи или готовиться к приёму.

**Полный текст промпта:**

```
Ты — ассистент врача-стоматолога. Твоя задача — помогать в анализе жалоб пациентов,
структурировании медицинской документации, но НЕ ставить диагнозы и НЕ давать медицинские рекомендации.

Правила:
1. Всегда напоминай, что твои ответы не заменяют консультацию врача.
2. Не интерпретируй симптомы как окончательный диагноз.
3. Будь вежливым, профессиональным и точным.
4. Если пациент описывает симптомы острого состояния (сильная боль, отёк, температура) — немедленно рекомендуй обратиться к стоматологу.
5. Сохраняй конфиденциальность — не проси и не сохраняй личные данные.

Твой стиль: чёткий, профессиональный, без излишней эмоциональности.
```

**Пример использования:**

> **Пациент:** «У меня болит зуб при накусывании, десна опухла, температура 37.5. Что мне делать?»
>
> **Модель (с системным промптом):** «Судя по описанию, у вас могут быть признаки острого воспаления (возможно, периодонтит). Это состояние требует немедленного осмотра стоматологом. Я не могу поставить диагноз, но рекомендую в ближайшее время обратиться к врачу. До визита можно принять обезболивающее, но не прикладывайте тепло к щеке.»

## 6.3. Промпт для суммаризации медицинских записей

Этот промпт помогает сокращать длинные записи (выписные эпикризы, описания приёма) до ключевых пунктов.

**Текст промпта:**

```
Ты — медицинский редактор. Твоя задача — суммаризировать записи из амбулаторной карты или выписного эпикриза.

Правила:
1. Выдели ключевые элементы: жалобы, анамнез, объективный статус, диагноз, план лечения, рекомендации.
2. Используй медицинскую терминологию, но понятную врачу.
3. Сократи до 4-6 предложений.
4. Не добавляй информацию, которой нет в тексте.
5. Сохрани все важные детали (лекарства, дозировки, особенности).

Верни суммаризацию в структурированном виде.
```

**Пример использования:**

> **Входной текст:** «Пациент К., 35 лет, обратился с жалобами на боли в 16 зубе, усиливающиеся при накусывании, отёк десны в области 16, припухлость щеки справа, повышение температуры до 37.5. Со слов пациента, боль появилась 3 дня назад, постепенно усиливалась, приём анальгетиков даёт кратковременный эффект. Объективно: лицо асимметрично за счёт отёка мягких тканей правой щеки, открывание рта ограничено (2 пальца), слизистая в проекции 16 зуба гиперемирована, отёчна, пальпация резко болезненна, перкуссия 16 резко положительна. Зуб 16 с глубокой кариозной полостью, зондирование дна безболезненно (пульпа некротизирована), подвижность I степени. На рентгенограмме: у верхушки дистального корня 16 определяется очаг разрежения костной ткани с нечёткими контурами размером 5×4 мм. Диагноз: хронический верхушечный периодонтит 16 в стадии обострения, периостит нижней челюсти справа. План: раскрытие полости 16, эндодонтическое лечение, назначение антибиотиков (амоксициллин/клавуланат), НПВС, холод местно. Рекомендовано: при неэффективности — хирургическое лечение (удаление 16).»
>
> **Модель (с промптом суммаризации):** «Пациент 35 лет, жалобы на боль в 16 зубе при накусывании, отёк десны и щеки, температура 37.5. Объективно: отёк мягких тканей, ограничение открывания рта, гиперемия и отёк слизистой в проекции 16, перкуссия резко положительна, зуб с глубокой кариозной полостью, некроз пульпы, подвижность I. Рентгенологически: очаг разрежения у верхушки дистального корня 5×4 мм. Диагноз: обострение хронического верхушечного периодонтита 16, периостит нижней челюсти справа. План: эндодонтическое лечение, антибиотики (амоксициллин/клавуланат), НПВС, холод. При неэффективности — удаление.»

Этот промпт помогает автоматически оценивать эмоциональную окраску отзывов пациентов, что полезно для контроля качества.

**Текст промпта:**

```
Ты — анализатор тональности медицинских отзывов.

Определи тональность отзыва пациента:
- positive (положительный)
- neutral (нейтральный)
- negative (отрицательный)

Правила:
1. Учитывай контекст: если пациент жалуется на боль, но отмечает хорошее отношение персонала, тональность может быть смешанной — укажи dominant.
2. Нейтральным считай информационный отзыв без эмоций.
3. Верни только метку и краткое обоснование (1 предложение).
```

**Пример использования:**

> **Отзыв:** «Удалял зуб мудрости. Сама процедура прошла быстро и без боли, но потом неделю болела челюсть. Врач предупредил, что так может быть. Спасибо за честность.»
>
> **Модель:** «Нейтральный. Несмотря на послеоперационную боль, пациент оценивает информацию и работу врача положительно.»

## 6.6. Промпт для генерации плана лечения (предварительного)

Этот промпт помогает сформировать черновик плана лечения на основе описания клинической ситуации. Важно: план должен быть предварительным и не заменять решение врача.

**Текст промпта:**

```
Ты — помощник врача-стоматолога. На основе предоставленного описания клинической ситуации сформируй предварительный план лечения.

Правила:
1. План должен быть логичным и соответствовать стандартам.
2. Учитывай возможные противопоказания и необходимость дополнительных обследований.
3. Не назначай конкретные препараты без указания врача.
4. Укажи, что план требует подтверждения врачом.
5. Верни план в виде нумерованного списка.
```

**Пример использования:**

> **Описание:** «Пациент 45 лет, жалобы на подвижность зубов, кровоточивость дёсен, неприятный запах. Диагноз: хронический генерализованный пародонтит средней степени.»
>
> **Модель (с промптом):** «Предварительный план лечения (требует подтверждения врачом): 1. Профессиональная гигиена полости рта (снятие зубных отложений). 2. Пародонтологическое лечение: закрытый кюретаж или лоскутные операции при необходимости. 3. Обучение индивидуальной гигиене, подбор средств. 4. Контрольные осмотры каждые 3-6 месяцев. 5. При необходимости — шинирование подвижных зубов. Дополнительно: консультация терапевта для исключения общих заболеваний, влияющих на пародонт.»

## 6.7. Как использовать системные промпты в коде (Ollama API)

В этом разделе мы научимся **писать код на Python**, который отправляет запросы к локальной модели через Ollama API. Это позволяет использовать системные промпты в ваших собственных приложениях.

### 6.7.1. Что такое Ollama API и зачем он нужен

**Ollama API** — это REST-интерфейс, который позволяет взаимодействовать с локальной моделью через HTTP-запросы. Вместо того чтобы запускать модель в терминале (`ollama run`), вы можете отправлять запросы из кода на Python.

**Интуитивная аналогия:**

> Ollama API — это как телефонный номер модели. Вы звоните (отправляете запрос), говорите свой текст, и модель отвечает. При этом модель всегда работает на вашем компьютере, данные никуда не уходят.

**Преимущества использования API:**

| Преимущество | Объяснение |
|--------------|------------|
| **Автоматизация** | Можно обрабатывать сотни запросов автоматически |
| **Интеграция** | Легко подключить к Streamlit, веб-сайту или приложению |
| **Гибкость** | Можно менять модель, температуру, системный промпт для каждого запроса |
| **Программируемость** | Можно создавать сложные сценарии работы |
| **Масштабируемость** | Можно обрабатывать запросы от нескольких пользователей |

**Важно:** Ollama API работает локально на порту `11434`. Убедитесь, что Ollama запущен (`ollama serve`) перед выполнением кода.

### 6.7.3. Полный код для работы с Ollama API (медицинские промпты)

```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
ollama_medical_chat.py — работа с локальной LLM через Ollama API
=========================================================

Этот код позволяет отправлять запросы к локальной модели
с системными промптами для медицинских (стоматологических) задач.

Для запуска:
1. Установите библиотеку: pip install requests
2. Запустите Ollama: ollama serve
3. Выполните: python ollama_medical_chat.py

Режимы запуска:
- python ollama_medical_chat.py              # демонстрация примеров
- python ollama_medical_chat.py --chat       # интерактивный режим
- python ollama_medical_chat.py --chat --model mistral:7b  # выбор модели
"""

import requests
import json
import sys
import time

# ============================================================
# КОНСТАНТЫ И НАСТРОЙКИ
# ============================================================

OLLAMA_URL = "http://localhost:11434/api/chat"
DEFAULT_MODEL = "qwen2.5:7b"
DEFAULT_TEMPERATURE = 0.7
DEFAULT_MAX_TOKENS = 500
DEFAULT_TOP_P = 0.9
DEFAULT_REPEAT_PENALTY = 1.1

# ============================================================
# ОСНОВНЫЕ СИСТЕМНЫЕ ПРОМПТЫ (МЕДИЦИНСКИЕ)
# ============================================================

MEDICAL_PROMPTS = {
    "assistant": """
Ты — ассистент врача-стоматолога. Твоя задача — помогать в анализе жалоб пациентов,
структурировании медицинской документации, но НЕ ставить диагнозы и НЕ давать медицинские рекомендации.

Правила:
1. Всегда напоминай, что твои ответы не заменяют консультацию врача.
2. Не интерпретируй симптомы как окончательный диагноз.
3. Будь вежливым, профессиональным и точным.
4. Если пациент описывает симптомы острого состояния (сильная боль, отёк, температура) — немедленно рекомендуй обратиться к стоматологу.
5. Сохраняй конфиденциальность — не проси и не сохраняй личные данные.

Твой стиль: чёткий, профессиональный, без излишней эмоциональности.
""",

    "summarize": """
Ты — медицинский редактор. Твоя задача — суммаризировать записи из амбулаторной карты или выписного эпикриза.

Правила:
1. Выдели ключевые элементы: жалобы, анамнез, объективный статус, диагноз, план лечения, рекомендации.
2. Используй медицинскую терминологию, но понятную врачу.
3. Сократи до 4-6 предложений.
4. Не добавляй информацию, которой нет в тексте.
5. Сохрани все важные детали (лекарства, дозировки, особенности).

Верни суммаризацию в структурированном виде.
""",

    "plain_language": """
Ты — медицинский переводчик с профессионального языка на понятный для пациента.

Правила:
1. Замени сложные медицинские термины на общедоступные.
2. Сохрани точность и не вводи в заблуждение.
3. Объясни значение терминов, но не упрощай до потери смысла.
4. Используй примеры, если это уместно.
5. Не давай рекомендаций по лечению.

Верни текст, понятный пациенту без медицинского образования.
""",

    "sentiment": """
Ты — анализатор тональности медицинских отзывов.

Определи тональность отзыва пациента:
- positive (положительный)
- neutral (нейтральный)
- negative (отрицательный)

Правила:
1. Учитывай контекст: если пациент жалуется на боль, но отмечает хорошее отношение персонала, тональность может быть смешанной — укажи dominant.
2. Нейтральным считай информационный отзыв без эмоций.
3. Верни только метку и краткое обоснование (1 предложение).
""",

    "treatment_plan": """
Ты — помощник врача-стоматолога. На основе предоставленного описания клинической ситуации сформируй предварительный план лечения.

Правила:
1. План должен быть логичным и соответствовать стандартам.
2. Учитывай возможные противопоказания и необходимость дополнительных обследований.
3. Не назначай конкретные препараты без указания врача.
4. Укажи, что план требует подтверждения врачом.
5. Верни план в виде нумерованного списка.
"""
}

# ============================================================
# ОСНОВНАЯ ФУНКЦИЯ ДЛЯ РАБОТЫ С API
# ============================================================

def get_medical_response(
    user_text: str,
    system_prompt: str,
    model: str = DEFAULT_MODEL,
    temperature: float = DEFAULT_TEMPERATURE,
    max_tokens: int = DEFAULT_MAX_TOKENS,
    top_p: float = DEFAULT_TOP_P,
    repeat_penalty: float = DEFAULT_REPEAT_PENALTY,
    silent: bool = False
) -> tuple:
    """
    Отправляет запрос к локальной LLM через Ollama API с системным промптом.
    Возвращает (ответ, время_выполнения).
    """
    
    payload = {
        "model": model,
        "messages": [
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_text}
        ],
        "stream": False,
        "options": {
            "temperature": temperature,
            "max_tokens": max_tokens,
            "top_p": top_p,
            "repeat_penalty": repeat_penalty
        }
    }
    
    start_time = time.time()
    
    try:
        if not silent:
            print("⏳ Ожидание ответа от модели...", end=" ", flush=True)
        
        response = requests.post(OLLAMA_URL, json=payload, timeout=60)
        response.raise_for_status()
        
        elapsed_time = time.time() - start_time
        if not silent:
            print(f"✅ ({elapsed_time:.1f} сек)")
        
        result = response.json()
        return result["message"]["content"], elapsed_time
        
    except requests.exceptions.ConnectionError:
        return "❌ Ошибка: Ollama не запущен. Запустите 'ollama serve' в терминале.", 0
    except requests.exceptions.Timeout:
        return "❌ Ошибка: Превышено время ожидания ответа от модели (60 секунд).", 0
    except KeyError:
        return "❌ Ошибка: Неожиданный формат ответа от модели.", 0
    except Exception as e:
        return f"❌ Ошибка: {str(e)}", 0


def print_with_separator(text: str, char: str = "=", length: int = 60):
    """Выводит текст с разделителем."""
    print(char * length)
    print(text)
    print(char * length)


def list_available_models():
    """Получает список доступных моделей из Ollama."""
    try:
        response = requests.get("http://localhost:11434/api/tags", timeout=5)
        response.raise_for_status()
        data = response.json()
        return [model["name"] for model in data.get("models", [])]
    except:
        return []


# ============================================================
# ИНТЕРАКТИВНЫЙ РЕЖИМ
# ============================================================

def interactive_chat(model: str = DEFAULT_MODEL):
    """Интерактивный режим чата с выбором медицинского промпта."""
    
    print_with_separator("🦷 Интерактивный чат с локальной LLM (медицина)")
    print(f"📌 Модель: {model}")
    print("📌 Данные НЕ покидают ваш компьютер")
    print()
    
    # Выбор промпта
    print("Выберите режим работы:")
    prompt_options = list(MEDICAL_PROMPTS.keys())
    prompt_names = {
        "assistant": "Ассистент стоматолога",
        "summarize": "Суммаризация медицинских записей",
        "plain_language": "Перевод терминов для пациента",
        "sentiment": "Анализ тональности отзывов",
        "treatment_plan": "Предварительный план лечения"
    }
    for i, key in enumerate(prompt_options, 1):
        print(f"   {i}. {prompt_names.get(key, key)}")
    print("   0. Ввести свой промпт")
    
    while True:
        try:
            choice = input("\nВаш выбор (0-5): ").strip()
            if choice == "0":
                system_prompt = input("Введите ваш системный промпт: ").strip()
                if not system_prompt:
                    print("❌ Промпт не может быть пустым. Использую базовый.")
                    system_prompt = MEDICAL_PROMPTS["assistant"]
                break
            else:
                idx = int(choice) - 1
                if 0 <= idx < len(prompt_options):
                    system_prompt = MEDICAL_PROMPTS[prompt_options[idx]]
                    break
                else:
                    print("❌ Неверный выбор. Попробуйте снова.")
        except ValueError:
            print("❌ Введите число от 0 до 5.")
    
    print("\n" + "=" * 60)
    print("💬 Чат начат. Введите 'выход' для завершения.")
    print("   'смена' - сменить промпт")
    print("   'модель' - сменить модель")
    print("=" * 60 + "\n")
    
    while True:
        user_input = input("👤 Вы: ").strip()
        
        if not user_input:
            continue
            
        if user_input.lower() in ["выход", "exit", "quit"]:
            print("\n🤖 До свидания! Берегите себя.")
            break
        
        if user_input.lower() == "смена":
            # Повторный выбор промпта
            print("Выберите новый режим:")
            for i, key in enumerate(prompt_options, 1):
                print(f"   {i}. {prompt_names.get(key, key)}")
            print("   0. Свой промпт")
            try:
                choice = input("Ваш выбор: ").strip()
                if choice == "0":
                    system_prompt = input("Введите ваш системный промпт: ").strip()
                    if not system_prompt:
                        system_prompt = MEDICAL_PROMPTS["assistant"]
                else:
                    idx = int(choice) - 1
                    if 0 <= idx < len(prompt_options):
                        system_prompt = MEDICAL_PROMPTS[prompt_options[idx]]
                print("✅ Режим изменён.")
            except:
                print("❌ Неверный выбор.")
            continue
        
        if user_input.lower() == "модель":
            new_model = input(f"Введите название модели (текущая: {model}): ").strip()
            if new_model:
                model = new_model
                print(f"✅ Модель изменена на: {model}")
            continue
        
        # Отправка запроса
        response, elapsed = get_medical_response(
            user_input,
            system_prompt,
            model=model,
            silent=False
        )
        
        print(f"🤖 Модель: {response}\n")


# ============================================================
# ДЕМОНСТРАЦИЯ ПРИМЕРОВ
# ============================================================

def run_examples():
    """Запускает демонстрацию примеров."""
    
    print_with_separator("🦷 Локальный ИИ-помощник для врача-стоматолога")
    print("📌 Данные НЕ покидают ваш компьютер")
    print("📌 Модель работает локально через Ollama")
    print()
    
    try:
        requests.get("http://localhost:11434/api/tags", timeout=2)
    except:
        print("❌ Ollama не запущен. Запустите 'ollama serve'")
        return
    
    examples = [
        {
            "name": "Ассистент стоматолога",
            "prompt": MEDICAL_PROMPTS["assistant"],
            "text": "У меня болит зуб при накусывании, десна опухла, температура 37.5. Что мне делать?"
        },
        {
            "name": "Суммаризация медицинской записи",
            "prompt": MEDICAL_PROMPTS["summarize"],
            "text": "Пациент К., 35 лет, обратился с жалобами на боли в 16 зубе, усиливающиеся при накусывании, отёк десны в области 16, припухлость щеки справа, повышение температуры до 37.5. Со слов пациента, боль появилась 3 дня назад, постепенно усиливалась, приём анальгетиков даёт кратковременный эффект. Объективно: лицо асимметрично за счёт отёка мягких тканей правой щеки, открывание рта ограничено (2 пальца), слизистая в проекции 16 зуба гиперемирована, отёчна, пальпация резко болезненна, перкуссия 16 резко положительна. Зуб 16 с глубокой кариозной полостью, зондирование дна безболезненно (пульпа некротизирована), подвижность I степени. На рентгенограмме: у верхушки дистального корня 16 определяется очаг разрежения костной ткани с нечёткими контурами размером 5×4 мм. Диагноз: хронический верхушечный периодонтит 16 в стадии обострения, периостит нижней челюсти справа. План: раскрытие полости 16, эндодонтическое лечение, назначение антибиотиков (амоксициллин/клавуланат), НПВС, холод местно. Рекомендовано: при неэффективности — хирургическое лечение (удаление 16)."
        },
        {
            "name": "Перевод терминов для пациента",
            "prompt": MEDICAL_PROMPTS["plain_language"],
            "text": "У вас диагностирован хронический верхушечный периодонтит 16 зуба. Рекомендуется эндодонтическое лечение с последующей реставрацией."
        },
        {
            "name": "Анализ тональности отзыва",
            "prompt": MEDICAL_PROMPTS["sentiment"],
            "text": "Удалял зуб мудрости. Сама процедура прошла быстро и без боли, но потом неделю болела челюсть. Врач предупредил, что так может быть. Спасибо за честность."
        }
    ]
    
    for i, example in enumerate(examples, 1):
        print(f"\n📌 ПРИМЕР {i}: {example['name']}")
        print("-" * 40)
        print(f"👤 Запрос: {example['text']}\n")
        
        response, elapsed = get_medical_response(
            example['text'],
            example['prompt'],
            silent=False
        )
        print(f"🤖 Модель: {response}\n")
        print(f"⏱️ Время ответа: {elapsed:.1f} сек")
        print()
    
    print_with_separator("✅ Все примеры выполнены!")
    print("\n💡 Для интерактивного режима запустите:")
    print("   python ollama_medical_chat.py --chat")


# ============================================================
# ПАРСИНГ АРГУМЕНТОВ
# ============================================================

def parse_args():
    args = {"chat": False, "model": DEFAULT_MODEL, "list_models": False}
    i = 1
    while i < len(sys.argv):
        arg = sys.argv[i]
        if arg == "--chat":
            args["chat"] = True
        elif arg == "--model" and i + 1 < len(sys.argv):
            args["model"] = sys.argv[i + 1]
            i += 1
        elif arg == "--list-models":
            args["list_models"] = True
        elif arg == "--help" or arg == "-h":
            print(__doc__)
            sys.exit(0)
        i += 1
    return args


if __name__ == "__main__":
    args = parse_args()
    
    if args["list_models"]:
        models = list_available_models()
        if models:
            print("📋 Доступные модели:")
            for model in models:
                print(f"   - {model}")
        else:
            print("❌ Не удалось получить список моделей. Запущен ли Ollama?")
        sys.exit(0)
    
    if args["chat"]:
        interactive_chat(args["model"])
    else:
        run_examples()
```

## 6.8. Таблица: сравнение системных промптов для медицины

| Промпт | Стиль | Основная задача | Кому подходит |
|--------|-------|-----------------|---------------|
| **Ассистент стоматолога** | Чёткий, профессиональный | Анализ жалоб, рекомендации обратиться к врачу | Врачи, регистратура |
| **Суммаризация** | Структурированный | Сокращение эпикризов | Врачи, ординаторы |
| **Перевод терминов** | Доступный, объясняющий | Информирование пациента | Врачи на приёме |
| **Анализ тональности** | Аналитический | Оценка отзывов | Администрация, маркетинг |
| **План лечения** | Практический | Предварительный план | Врачи (черновик) |

---

## 6.9. Основные выводы раздела

1. **Системный промпт** — это инструкция, задающая модели роль, правила и стиль. Он критически важен для безопасного использования в медицине.

2. **Медицинские промпты** должны обязательно включать запрет на постановку диагноза, дисклеймер о необходимости консультации врача, указание на конфиденциальность.

3. **Разные задачи требуют разных промптов.** Для суммаризации — один, для перевода терминов — другой, для анализа тональности — третий.

4. **В коде системный промпт передаётся** через поле `messages` с ролью `system` в API Ollama.

5. **Локальная работа** гарантирует, что данные не покидают устройство, что соответствует 152-ФЗ и врачебной тайне.

6. **Файл для кода** называется `ollama_medical_chat.py`, для запуска требуется библиотека `requests`.

7. **Интерактивный режим** позволяет быстро переключаться между промптами и моделями, что удобно для тестирования.

## 7. ЭТИЧЕСКИЕ РЕКОМЕНДАЦИИ ДЛЯ ВРАЧЕЙ

### 7.1. Что можно и нельзя делать

| ✅ МОЖНО | ❌ НЕЛЬЗЯ |
|----------|-----------|
| Использовать для анализа **анонимизированных** медицинских текстов | Загружать ФИО, адреса, телефоны пациентов |
| Использовать для подготовки черновиков документации | Полагаться на модель в постановке диагноза |
| Использовать для структурирования записей | Использовать как замену клиническому мышлению |
| Использовать в исследовательских целях (с согласованием) | Передавать данные модели без согласия пациента |
| Использовать локально | Использовать облачные аналоги с реальными данными |

```
⚠️ Важно: этот чат-бот использует локальную ИИ-модель.
Ваши данные НЕ покидают ваш компьютер и НЕ передаются третьим лицам.

Тем не менее, помните:
• Это НЕ замена профессиональному врачу.
• Результаты НЕ являются диагнозом.
• При серьёзных симптомах (острая боль, отёк, температура) немедленно обратитесь к стоматологу.
```
