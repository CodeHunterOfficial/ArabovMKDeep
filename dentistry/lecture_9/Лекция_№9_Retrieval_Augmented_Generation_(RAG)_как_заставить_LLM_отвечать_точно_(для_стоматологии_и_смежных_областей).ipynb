## 1. ВВЕДЕНИЕ

### 1.1. От облачных LLM к точным ответам: почему врачу нужен RAG

В Лекции 8 мы научились запускать локальные LLM через Ollama. Мы видели, как они генерируют связные, осмысленные ответы. Но есть одна фундаментальная проблема: **LLM не знают, когда они чего-то не знают**.

**Интуитивная аналогия из стоматологической практики:**

> Представьте, что вы – молодой стоматолог, который только что окончил ординатуру. Вы много читали, но у вас нет доступа к библиотеке. Пациент спрашивает вас о редком заболевании слизистой оболочки полости рта. Вы не знаете точного ответа, но вместо того чтобы признаться, начинаете **уверенно говорить** – и выдаёте ложную информацию, потому что ваш мозг «дорисовывает» недостающие детали. Это и есть **галлюцинация LLM** – модель генерирует убедительный, но ложный ответ, потому что она не умеет говорить «я не знаю».

**Что такое RAG и как он помогает:**

**RAG (Retrieval-Augmented Generation)** – это технология, которая даёт LLM доступ к внешней базе знаний. Вместо того чтобы «гадать», модель сначала **ищет** ответ в ваших документах, а затем **генерирует** ответ на основе найденных фактов.

**Интуитивная аналогия из стоматологии:**

> Представьте, что у вас есть **личный ассистент-библиотекарь**, который:
> 1. Слышит вопрос врача или пациента.
> 2. Мгновенно находит нужный справочник или статью на полке.
> 3. Открывает нужную страницу.
> 4. Показывает вам точную цитату.
> 5. Вы читаете её, **ссылаясь на источник**.
>
> Это и есть RAG – модель больше не «выдумывает», она **ищет и цитирует**.

**Исследования подтверждают эффективность RAG:**

Современные исследования показывают, что RAG значительно снижает количество галлюцинаций у LLM [5]. Модели, использующие RAG, реже «выдумывают» ответы и чаще опираются на факты из проверенных источников.

В стоматологии это особенно важно. Исследования 2025-2026 годов (адаптированные из медицинской сферы) показывают, что RAG-системы достигают точности **до 76–94%** в задачах информационного поиска и поддержки принятия решений в специализированных областях [1], [5]. Например, система Sanity-LLM для чат-ботов психического здоровья продемонстрировала контекстную точность **0.94** [5]. Адаптируя этот подход к стоматологии, можно ожидать высокую точность при поиске информации по клиническим протоколам, диагностическим критериям и рекомендациям.

**Что означают эти цифры для стоматолога:**

| Показатель | Что это значит |
|------------|----------------|
| **Снижение галлюцинаций** | Модель реже выдаёт ложную информацию, чаще опирается на проверенные источники |
| **Точность 76–94%** | В большинстве случаев RAG-система даёт правильные ответы на основе базы знаний |

**Важно понимать:** даже 94% – это не 100%. Это означает, что RAG-система **не заменяет** врача, но является **надёжным помощником**, который может:
- Быстро находить релевантную информацию из проверенных источников (учебники, клинические рекомендации, протоколы).
- Снижать риск ошибок, связанных с человеческой усталостью или забывчивостью.
- Экономить время на поиске информации в бумажных справочниках и статьях.

**Что мы узнаем сегодня:**

- Что такое **галлюцинации** LLM и почему они возникают.
- Как **Retrieval-Augmented Generation (RAG)** решает эту проблему.
- Как работает RAG: **поиск → подстановка → генерация**.
- Что такое **чангкинг**, **эмбеддинги** и **векторное хранилище (Chroma)**.
- Как создать **вопросно-ответную систему** по стоматологическим учебникам и клиническим рекомендациям.

**Ключевая мысль:** RAG – это технология, которая **даёт LLM доступ к внешней базе знаний**. Модель больше не «гадает» – она **ищет** ответ в ваших документах и **цитирует** источники. Это превращает LLM из «болтуна» в **исследователя-ассистента**, который помогает врачу принимать обоснованные решения.

## 2. ПРОБЛЕМА ГАЛЛЮЦИНАЦИЙ LLM

### 2.1. Что такое галлюцинации

**Галлюцинация** – это когда LLM генерирует информацию, которая **звучит правдоподобно, но фактически неверна** или отсутствует в обучающих данных.

**Примеры галлюцинаций в стоматологическом контексте:**

| Вопрос пользователя | Галлюцинация модели | Правильный ответ |
|---------------------|---------------------|------------------|
| «Каковы клинические признаки острого пульпита?» | Модель называет симптомы периодонтита | Острый пульпит: самопроизвольная боль, усиливающаяся ночью, реакция на термические раздражители |
| «Какова рекомендуемая доза артикаина для инфильтрационной анестезии?» | Модель называет завышенную дозу | Максимальная доза 7 мг/кг (с вазоконстриктором) |
| «Какой антибиотик назначают при периодонтите?» | Модель рекомендует неверный препарат | Обычно амоксициллин или метронидазол (по показаниям) |

**Почему LLM галлюцинируют:**

| Причина | Объяснение |
|---------|------------|
| **LLM не знают, чего они не знают** | Они обучены предсказывать следующее слово, а не проверять факты |
| **Обучение на устаревших данных** | Модель может «знать» информацию, которая была актуальна на момент обучения, но уже устарела |
| **Отсутствие доступа к источникам** | Модель не может «посмотреть» в книгу или статью – она полагается только на свои внутренние веса |
| **Вероятностная природа** | LLM генерируют наиболее вероятное продолжение текста, а не наиболее точное |

**Интуитивная аналогия:**

> Галлюцинации LLM – это как если бы вы спросили у коллеги, который никогда не читал первоисточники, но очень уверенно отвечает на все вопросы. Он может быть прав в 80% случаев, но в 20% – он просто «дорисовывает» ответ.

### 2.2. Почему галлюцинации особенно опасны в стоматологии и медицине

| Риск | Что может пойти не так | Последствия |
|------|------------------------|-------------|
| **Диагностические ошибки** | Модель рекомендует неверные диагностические критерии | Неправильный диагноз, неверное лечение |
| **Юридические риски** | Ложная информация может привести к неправильным решениям | Иски, потеря лицензии |
| **Потеря доверия** | Если модель постоянно ошибается, пациенты и коллеги перестают ей доверять | Отказ от использования полезных ИИ-инструментов |
| **Этические нарушения** | Использование ложной информации в клинической практике недопустимо | Нарушение профессиональной этики и врачебной тайны |

### 2.3. Схема: проблема галлюцинаций

## 3. ЧТО ТАКОЕ RAG И КАК ОН РАБОТАЕТ

### 3.1. Определение RAG

**Retrieval-Augmented Generation (RAG)** – это архитектура, которая **дополняет** LLM внешней базой знаний. Перед тем как сгенерировать ответ, система **ищет** релевантные фрагменты в документах и **подставляет** их в промпт.

**Интуитивная аналогия:**

> Представьте, что вы – стоматолог, и у вас есть доступ к **библиотеке** с тысячами книг по стоматологии. Когда пациент задаёт вопрос, вы не отвечаете сходу – вы **идёте к полке**, **находите нужную книгу**, **открываете нужную страницу** и **читаете** ответ. Только после этого вы даёте ответ пациенту, **ссылаясь** на источник. RAG делает то же самое, но в тысячи раз быстрее.

**Три шага RAG:**

```mermaid
flowchart LR
    subgraph STEP1["1️⃣ ПОИСК (Retrieval)"]
        S1["Запрос пользователя"]
        S2["Поиск в векторной БД"]
        S3["Найдены релевантные<br>фрагменты"]
        S1 --> S2 --> S3
    end
    
    subgraph STEP2["2️⃣ ПОДСТАНОВКА (Augmentation)"]
        S4["Фрагменты + запрос"]
        S5["Формирование промпта"]
        S4 --> S5
    end
    
    subgraph STEP3["3️⃣ ГЕНЕРАЦИЯ (Generation)"]
        S6["LLM читает промпт"]
        S7["Генерирует ответ<br>на основе фактов"]
        S6 --> S7
    end
    
    STEP1 --> STEP2 --> STEP3
    
    style STEP1 fill:#e1f5fe,stroke:#01579b
    style STEP2 fill:#fff9c4,stroke:#f57f17
    style STEP3 fill:#c8e6c9,stroke:#2e7d32
```

### 3.2. Как RAG решает проблему галлюцинаций

| Без RAG | С RAG |
|---------|-------|
| Модель «гадает» ответ | Модель **ищет** ответ в документах |
| Нет доступа к источникам | Есть **цитаты** и **ссылки** на источники |
| Не может проверить факты | Может **проверить** по первоисточнику |
| Ответ может быть ложным | Ответ **основан на фактах** из документов |
| Неясно, откуда взят ответ | **Прозрачность** – видно, на каком фрагменте основан ответ |

**Ключевое преимущество:** RAG позволяет **проверять** ответы. Вы всегда можете посмотреть, из какого фрагмента был взят ответ, и убедиться в его достоверности.

### 3.3. Пример RAG в стоматологии

**Вопрос:** *«Каковы клинические признаки острого пульпита?»*

**Без RAG:** Модель генерирует ответ на основе своих весов – может ошибиться.

**С RAG:**
1. Система ищет в базе знаний (например, в учебнике по терапевтической стоматологии).
2. Находит фрагмент: *«Острый пульпит характеризуется самопроизвольной болью, усиливающейся ночью, реакцией на термические раздражители, перкуссия может быть слабоположительной...»*
3. Подставляет этот фрагмент в промпт.
4. LLM генерирует ответ **на основе этого фрагмента**.
5. В ответе есть **ссылка**: *«Источник: Учебник терапевтической стоматологии, глава 5, стр. 123»*.

### 3.4. Схема: RAG vs обычная LLM

```mermaid
flowchart LR
    subgraph WITHOUT["❌ БЕЗ RAG"]
        W1["Вопрос"] --> W2["LLM"]
        W2 --> W3["Ответ (может быть<br>галлюцинацией)"]
    end
    
    subgraph WITH["✅ С RAG"]
        R1["Вопрос"] --> R2["Поиск в<br>векторной БД"]
        R2 --> R3["Релевантные<br>фрагменты"]
        R3 --> R4["Промпт:<br>фрагменты + вопрос"]
        R4 --> R5["LLM"]
        R5 --> R6["Ответ на основе<br>фактов + источники"]
    end
    
    WITHOUT -->|Проблема| WITH
    
    style WITHOUT fill:#ffcdd2,stroke:#c62828
    style WITH fill:#c8e6c9,stroke:#2e7d32
```

## 4.1. Чанкинг (Chunking) — разбиение текста на фрагменты

**Что это:** процесс разбиения большого документа на **небольшие смысловые фрагменты** – «чанки» (chunks). LLM не могут обработать целую книгу целиком (из-за ограничения на длину контекста), поэтому мы разбиваем текст на части, которые модель может «переварить».

**Интуитивная аналогия из стоматологии:**

> Представьте, что у вас есть толстый справочник по стоматологии. Вы не можете держать его открытым целиком – вы открываете его на **одной странице** и читаете **один абзац**. Чанкинг – это как разрезание книги на отдельные страницы или абзацы, чтобы потом легко находить нужный фрагмент.

**Почему чанкинг важен:**

| Причина | Объяснение |
|---------|------------|
| **Ограничение контекста** | LLM могут обработать только ограниченное количество токенов за раз (например, 8192 токена) |
| **Точность поиска** | Маленькие фрагменты легче сравнивать с вопросом – выше точность |
| **Скорость** | Поиск по маленьким фрагментам быстрее |
| **Релевантность** | Можно найти именно тот фрагмент, где содержится ответ |

**Стратегии чанкинга:**

| Стратегия | Описание | Когда использовать | Пример из стоматологии |
|-----------|----------|-------------------|---------------------|
| **Фиксированный размер** | Разбивка на чанки по N токенов (например, 256) | Для простых документов | Тексты без чёткой структуры |
| **По предложениям** | Разбивка по границам предложений | Для художественных текстов | Описания клинических случаев |
| **По абзацам** | Разбивка по абзацам | Для учебников и статей | **Учебники по стоматологии** |
| **Семантический чанкинг** | Разбивка по смысловым границам (с помощью ML) | Для сложных документов | Научные статьи с терминами |

**Ключевое понятие: перекрытие (Overlap)**

При чанкинге важно использовать **перекрытие** – когда чанки частично пересекаются. Это нужно, чтобы важная информация не «потерялась» на границе между чанками.

**Интуитивная аналогия:**

> Представьте, что вы разрезаете книгу на страницы. Если вы разрежете ровно по границе предложения, одно предложение может оказаться на двух страницах. Чтобы этого избежать, вы оставляете небольшое **перекрытие** – последние 1-2 строки предыдущей страницы дублируются на следующей.

**Рекомендация для стоматологических текстов:**

| Параметр | Рекомендация | Почему |
|----------|--------------|--------|
| **Размер чанка** | 256–512 токенов | Достаточно для смыслового фрагмента |
| **Перекрытие** | 50–100 токенов | Чтобы не потерять смысл на границах |
| **Разделители** | Абзацы, предложения | Учитывает структуру текста |

**Пример чанкинга в коде:**

```python
!pip install langchain-text-splitters

from langchain_text_splitters import RecursiveCharacterTextSplitter

# Создаём сплиттер для стоматологических текстов
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,      # размер чанка в символах
    chunk_overlap=50,    # перекрытие между чанками
    separators=["\n\n", "\n", ". ", " ", ""]  # приоритет разделителей
)

# Пример: фрагмент из учебника по терапевтической стоматологии
text = """
Острый пульпит характеризуется самопроизвольной болью, которая усиливается
ночью и при приёме горячей или холодной пищи. При осмотре обнаруживается
глубокая кариозная полость, зондирование дна болезненно. Перкуссия может быть
слабоположительной. Электроодонтодиагностика показывает снижение порога
возбудимости до 20-30 мкА.
"""

# Разбиваем на чанки
chunks = text_splitter.split_text(text)
print(f"Создано {len(chunks)} чанков")
for i, chunk in enumerate(chunks):
    print(f"Чанк {i+1}: {chunk[:100]}...")
```

## 4.2. Эмбеддинги (Embeddings) — превращение текста в числа

**Что это:** **Эмбеддинг** – это вектор чисел (обычно 384–1536 чисел), который **кодирует смысл** текстового фрагмента. Близкие по смыслу тексты имеют **похожие эмбеддинги** – они находятся рядом в многомерном пространстве.

**Интуитивная аналогия из стоматологии:**

> Представьте, что у вас есть карта города, где **смысл** – это географическое положение. Слова «пульпит» и «периодонтит» находятся рядом на этой карте (похожий смысл – воспалительные заболевания), а слово «отбеливание» – далеко. Эмбеддинги – это **координаты** каждого текста на этой карте.

**Как работают эмбеддинги (пошагово):**

```mermaid
flowchart LR
    A["Текст:<br>«Симптомы пульпита»"] --> B["Модель эмбеддингов"]
    B --> C["Вектор:<br>[0.12, -0.34, 0.56, ...]"]
    D["Текст:<br>«Симптомы периодонтита»"] --> B
    B --> E["Вектор:<br>[0.11, -0.32, 0.58, ...]"]
    F["Текст:<br>«Методика отбеливания»"] --> B
    B --> G["Вектор:<br>[0.89, 0.45, -0.12, ...]"]
    
    C -.->|"Близко"| E
    E -.->|"Далеко"| G
    
    style A fill:#e1f5fe,stroke:#01579b
    style D fill:#e1f5fe,stroke:#01579b
    style F fill:#ffcdd2,stroke:#c62828
```

**Почему эмбеддинги важны для RAG:**

| Причина | Объяснение |
|---------|------------|
| **Семантический поиск** | Находят тексты по смыслу, а не по ключевым словам |
| **Скорость** | Поиск по числам (векторам) намного быстрее, чем по тексту |
| **Масштабируемость** | Можно обрабатывать миллионы документов |
| **Мультиязычность** | Некоторые модели работают с разными языками |

**Модели эмбеддингов для русского языка:**

| Модель | Размер вектора | Особенности | Для стоматолога |
|--------|---------------|-------------|---------------|
| **multilingual-e5-large** | 1024 | Поддерживает 100+ языков, включая русский | **Рекомендуется для начала** |
| **bge-m3** | 1024 | Флагманская модель от BAAI | Для высокоточного поиска |
| **rubert-tiny2** | 312 | Компактная и быстрая | Для on-premise систем |
| **Qwen3-Embedding** | 1536 | Мультиязычный RAG, SOTA-метрики | Для сложных RAG-систем |

**Рекомендация для стоматологов:** для начала используйте **multilingual-e5-large** – она хорошо работает с русским языком и доступна бесплатно.

**Пример создания эмбеддингов в коде:**

```python
from sentence_transformers import SentenceTransformer

# Загружаем модель эмбеддингов
embedding_model = SentenceTransformer('intfloat/multilingual-e5-large')

# Тексты для сравнения
texts = [
    "Острый пульпит: самопроизвольная боль, усиливающаяся ночью",
    "Клинические признаки острого пульпита включают боль при накусывании",
    "Методика проведения профессиональной гигиены полости рта"
]

# Создаём эмбеддинги
embeddings = embedding_model.encode(texts)

# Сравниваем близость
from sklearn.metrics.pairwise import cosine_similarity

# Насколько похожи первые два текста?
similarity = cosine_similarity([embeddings[0]], [embeddings[1]])[0][0]
print(f"Сходство текста 1 и 2: {similarity:.3f}")  # Близко к 1.0

# Насколько похожи первый и третий тексты?
similarity = cosine_similarity([embeddings[0]], [embeddings[2]])[0][0]
print(f"Сходство текста 1 и 3: {similarity:.3f}")  # Близко к 0.0
```

**Ожидаемый вывод:**

```
Сходство текста 1 и 2: 0.892
Сходство текста 1 и 3: 0.124
```

**Что это значит:** модель «понимает», что первый и второй тексты говорят об одном (пульпит), а третий – о совершенно другом (гигиена).

## 4.3. Векторное хранилище (Vector Database) — Chroma

**Что это:** база данных, которая **хранит эмбеддинги** и позволяет **быстро искать** похожие векторы.

**Chroma** – это открытая векторная база данных, которая:
- Хранит эмбеддинги на диске.
- Поддерживает поиск по семантической близости.
- Интегрируется с Python и LangChain.
- Бесплатна и проста в использовании.

**Интуитивная аналогия:**

> Представьте, что у вас есть **картотека** всех фрагментов из учебников по стоматологии. Каждая карточка – это фрагмент текста и его «координаты» на карте смыслов (эмбеддинг). Когда вы задаёте вопрос, вы вычисляете его «координаты» и находите **ближайшие** карточки в этой картотеке. Chroma – это и есть такая **умная картотека**.

**Как работает векторный поиск:**

```mermaid
flowchart LR
    subgraph INDEX["📇 ИНДЕКСАЦИЯ (подготовка)"]
        direction TB
        A["Документы"] --> B["Чанки"]
        B --> C["Эмбеддинги"]
        C --> D["Chroma<br>(векторное хранилище)"]
    end
    
    subgraph SEARCH["🔍 ПОИСК (во время запроса)"]
        direction TB
        E["Вопрос"] --> F["Эмбеддинг вопроса"]
        F --> G["Поиск ближайших<br>векторов в Chroma"]
        G --> H["Найденные<br>релевантные чанки"]
    end
    
    INDEX --> SEARCH
    
    style INDEX fill:#e1f5fe,stroke:#01579b
    style SEARCH fill:#fff9c4,stroke:#f57f17
```

**Установка Chroma:**

```bash
pip install chromadb
```

**Пример работы с Chroma в коде (стоматологический):**

```python
import chromadb
from sentence_transformers import SentenceTransformer

# 1. Загружаем эмбеддинг-модель
embedding_model = SentenceTransformer('intfloat/multilingual-e5-large')

# 2. Создаём клиент Chroma (с хранением на диске)
client = chromadb.PersistentClient(path="./dental_db")

# 3. Создаём коллекцию (таблицу) для хранения
collection = client.get_or_create_collection(
    name="dental_knowledge",
    metadata={"hnsw:space": "cosine"}
)

# 4. Добавляем документы
documents = [
    "Острый пульпит: самопроизвольная боль, усиливающаяся ночью, реакция на холодное и горячее. Перкуссия слабоположительная.",
    "Хронический периодонтит: боль при накусывании, отёк десны, возможно образование свища. На рентгенограмме очаг разрежения.",
    "Гингивит: кровоточивость дёсен, отёк десневого края, неприятный запах."
]

embeddings = embedding_model.encode(documents)

for i, (doc, emb) in enumerate(zip(documents, embeddings)):
    collection.add(
        ids=[f"doc_{i}"],
        documents=[doc],
        embeddings=[emb.tolist()]
    )

print(f"✅ Добавлено {len(documents)} документов в Chroma")

# 5. Поиск
query = "Какие признаки пульпита?"
query_embedding = embedding_model.encode([query])[0]

results = collection.query(
    query_embeddings=[query_embedding.tolist()],
    n_results=2
)

print("\n🔍 Результаты поиска по запросу:", query)
for i, doc in enumerate(results['documents'][0]):
    print(f"{i+1}. {doc[:100]}...")
```

## 4.4. Полный пайплайн RAG: от документа к ответу

**Схема: архитектура RAG**

```mermaid
flowchart LR
    subgraph INGESTION["📥 1. ИНДЕКСАЦИЯ (подготовка базы знаний)"]
        direction TB
        A1["📚 Документы<br>(учебники, статьи)"] --> A2["✂️ Чанкинг<br>разбивка на фрагменты"]
        A2 --> A3["🔢 Эмбеддинги<br>превращение в векторы"]
        A3 --> A4["💾 Chroma<br>векторное хранилище"]
    end
    
    subgraph QUERY["🔍 2. ЗАПРОС (поиск в базе)"]
        direction TB
        B1["❓ Вопрос пользователя"] --> B2["🔢 Эмбеддинг вопроса"]
        B2 --> B3["🎯 Поиск похожих<br>чанков в Chroma"]
        B3 --> B4["📄 Найденные<br>релевантные фрагменты"]
    end
    
    subgraph GENERATION["📝 3. ГЕНЕРАЦИЯ (ответ)"]
        direction TB
        C1["📝 Промпт:<br>фрагменты + вопрос"] --> C2["🧠 LLM (Ollama)"]
        C2 --> C3["✅ Ответ на основе<br>фактов + источники"]
    end
    
    INGESTION -->|"Подготовка"| QUERY
    QUERY -->|"Поиск"| GENERATION
    
    style INGESTION fill:#e1f5fe,stroke:#01579b
    style QUERY fill:#fff9c4,stroke:#f57f17
    style GENERATION fill:#c8e6c9,stroke:#2e7d32
```

**Пошаговое объяснение:**

| Шаг | Что происходит | Пример |
|-----|----------------|--------|
| **1. Индексация** | Документы разбиваются на чанки, превращаются в эмбеддинги и сохраняются в Chroma | Учебник по терапевтической стоматологии → тысячи чанков с эмбеддингами |
| **2. Запрос** | Вопрос пользователя превращается в эмбеддинг, ищутся похожие чанки | «Какие признаки острого пульпита?» → находим чанк о симптомах пульпита |
| **3. Генерация** | Найденные чанки подставляются в промпт, LLM генерирует ответ | LLM читает чанк и формулирует ответ на его основе |

**Ключевые выводы по разделу 4:**

1. **Чанкинг** – разбивка текста на фрагменты для удобной обработки. Важно использовать перекрытие, чтобы не потерять смысл.

2. **Эмбеддинги** – превращение смысла текста в числа (векторы). Похожие тексты имеют похожие векторы.

3. **Chroma** – векторная база данных для хранения эмбеддингов и быстрого поиска по смыслу.

4. **Три шага RAG** – индексация (подготовка), запрос (поиск), генерация (ответ).

5. **Для стоматологов** это означает возможность создать персональную базу знаний из учебников, статей и протоколов, которая отвечает на вопросы, ссылаясь на источники.

## 5.1. Вопросно-ответные системы по учебникам и клиническим рекомендациям

**Что это:** RAG-система, которая отвечает на вопросы врача, используя его личную библиотеку учебников, статей и клинических протоколов. Это как иметь **персонального ассистента**, который помнит всё, что вы когда-либо читали, и может мгновенно найти нужную информацию.

**Интуитивная аналогия из стоматологии:**

> Представьте, что вы — стоматолог, и у вас есть **личный библиотекарь**, который:
> - Знает каждую книгу в вашей библиотеке.
> - Помнит, где что лежит.
> - Мгновенно находит нужную страницу.
> - Показывает вам точную цитату.
> - Может ответить на любой вопрос по вашим книгам.
>
> Вы задаёте вопрос: «Каковы критерии диагностики хронического периодонтита?» — и библиотекарь через 2 секунды даёт вам точную цитату из учебника с номером страницы. Это и есть RAG в действии.

**Как это работает в практике стоматолога:**

| Этап | Что делает врач | Что делает RAG-система |
|------|---------------------|----------------------|
| **1. Подготовка** | Загружает учебники, статьи, протоколы в базу знаний | Разбивает тексты на чанки, создаёт эмбеддинги, индексирует |
| **2. Запрос** | Задаёт вопрос о конкретном заболевании, симптоме или методе | Находит релевантные фрагменты в базе знаний |
| **3. Ответ** | Получает точный ответ с цитатой и ссылкой на источник | Генерирует ответ на основе найденных фрагментов |

### 6.2.5. Шаг 4: Создание базы знаний из стоматологических текстов

```python
# Ячейка 4: Создание базы знаний

import chromadb
from sentence_transformers import SentenceTransformer
from langchain_text_splitters import RecursiveCharacterTextSplitter

print("⏳ Загрузка эмбеддинг-модели...")
embedding_model = SentenceTransformer('intfloat/multilingual-e5-large')
print(f"✅ Модель загружена (размер вектора: {embedding_model.get_sentence_embedding_dimension()})")

# Стоматологические документы
documents = [
    "Острый пульпит характеризуется самопроизвольной болью, усиливающейся ночью, реакцией на холодное и горячее. Перкуссия слабоположительная.",
    "Хронический периодонтит: боль при накусывании, отёк десны, возможно образование свища. На рентгенограмме очаг разрежения костной ткани.",
    "Гингивит: кровоточивость дёсен при чистке зубов, отёк десневого края, неприятный запах изо рта.",
    "Кариес: начальный кариес проявляется белым пятном, при прогрессировании образуется полость. Лечение: удаление поражённых тканей и пломбирование.",
    "Профессиональная гигиена полости рта включает снятие зубных отложений ультразвуком и полировку зубов. Рекомендуется проводить каждые 6 месяцев."
]

print("📚 Разбивка текстов на чанки...")
text_splitter = RecursiveCharacterTextSplitter(chunk_size=300, chunk_overlap=50)
chunks = []
for doc in documents:
    chunks.extend(text_splitter.split_text(doc))
print(f"✅ Создано {len(chunks)} чанков")

print("🔢 Создание эмбеддингов...")
chunk_embeddings = embedding_model.encode(chunks, show_progress_bar=True)

print("💾 Сохранение в Chroma...")
client = chromadb.PersistentClient(path="./dental_db")
collection = client.get_or_create_collection(
    name="dental_knowledge",
    metadata={"hnsw:space": "cosine"}
)

for i, (chunk, emb) in enumerate(zip(chunks, chunk_embeddings)):
    collection.add(
        ids=[f"chunk_{i}"],
        documents=[chunk],
        embeddings=[emb.tolist()],
        metadatas=[{"source": f"doc_{i}"}]
    )
print(f"✅ Добавлено {len(chunks)} чанков в векторную базу")
```

---

### 6.2.6. Шаг 5: Функция поиска и генерации ответа

```python
import requests

def rag_query(query, top_k=3):
    print("\n" + "=" * 50)
    print(f"🔍 Вопрос: {query}")
    print("=" * 50)
    
    # Эмбеддинг вопроса
    print("⏳ Превращаем вопрос в эмбеддинг...")
    query_embedding = embedding_model.encode([query])[0]
    
    # Поиск в Chroma
    print(f"🔎 Ищем {top_k} самых похожих фрагментов...")
    results = collection.query(
        query_embeddings=[query_embedding.tolist()],
        n_results=top_k
    )
    
    retrieved_chunks = results['documents'][0]
    retrieved_distances = results['distances'][0]
    
    print(f"📄 Найдено {len(retrieved_chunks)} фрагментов")
    for i, (chunk, dist) in enumerate(zip(retrieved_chunks, retrieved_distances)):
        print(f"   {i+1}. Сходство: {1-dist:.3f} | {chunk[:80]}...")
    
    # Формируем промпт
    print("📝 Формируем промпт для LLM...")
    context = "\n\n---\n\n".join(retrieved_chunks)
    
    prompt = f"""Ты – стоматологический ассистент, который отвечает на вопросы,
основываясь ТОЛЬКО на предоставленных фрагментах из учебников и статей.

Если в фрагментах нет ответа на вопрос, скажи: "В предоставленных материалах
нет информации по этому вопросу."

Используй фрагменты ДОСЛОВНО, если это возможно. Всегда указывай, из какого
фрагмента взята информация.

Фрагменты для ответа:
---
{context}
---
Вопрос пользователя: {query}

Твой ответ:"""

    # Генерация через Ollama
    print("🧠 Генерируем ответ через Ollama...")
    url = "http://localhost:11434/api/generate"
    payload = {
        "model": "llama3.2:1b",
        "prompt": prompt,
        "stream": False,
        "options": {"temperature": 0.3}
    }
    
    response = requests.post(url, json=payload)
    result_json = response.json()
    
    if "response" in result_json:
        answer = result_json["response"]
    else:
        answer = f"⚠️ Ошибка Ollama: {result_json.get('error', 'Неизвестная ошибка')}"
    
    print("✅ Ответ готов!")
    
    return {
        "answer": answer,
        "sources": retrieved_chunks
    }
```

```python
print("\n" + "=" * 60)
print("🧠 ТЕСТИРОВАНИЕ RAG-СИСТЕМЫ")
print("=" * 60)

# Тест 1: есть ответ
print("\n📌 ТЕСТ 1: Вопрос о пульпите")
result = rag_query("Какие симптомы острого пульпита?")
print("\n📊 ОТВЕТ:")
print(result["answer"])

# Тест 2: нет ответа
print("\n📌 ТЕСТ 2: Вопрос, на который нет ответа")
result2 = rag_query("Что такое имплантация?")
print("\n📊 ОТВЕТ:")
print(result2["answer"])

# Тест 3: о гингивите
print("\n📌 ТЕСТ 3: Вопрос о гингивите")
result3 = rag_query("Какие признаки гингивита?")
print("\n📊 ОТВЕТ:")
print(result3["answer"])
```

## 7. ЭТИЧЕСКИЕ АСПЕКТЫ RAG

### 7.1. RAG не панацея

Важно понимать: RAG **уменьшает**, но **не исключает** галлюцинации.

**Остаточные риски:**

| Риск | Что может пойти не так |
|------|------------------------|
| **Неверный поиск** | Система может найти нерелевантные фрагменты |
| **Искажение контекста** | LLM может неверно интерпретировать найденные фрагменты |
| **Устаревшие данные** | Если база знаний не обновляется, ответы могут быть устаревшими |
| **Чувствительные знания** | Неверное извлечение информации из базы может привести к ошибке |

**Правило:** **всегда проверяйте** ответы RAG-системы, особенно в клиническом контексте.

### 7.2. Источники должны быть проверенными

- Используйте только **рецензируемые** источники (учебники, научные статьи, официальные протоколы).
- Не добавляйте в базу знаний непроверенные данные из интернета.
- Регулярно **обновляйте** базу знаний.

### 7.3. Прозрачность для пользователя

Всегда сообщайте пользователю:

- Что ответ сгенерирован ИИ.
- Что ответ основан на конкретных источниках (показывайте ссылки).
- Что это не заменяет профессиональную консультацию.

**Пример дисклеймера для RAG-системы:**

```
⚠️ Этот ответ сгенерирован ИИ-ассистентом на основе предоставленных источников.
Ответы НЕ являются диагнозом и НЕ заменяют консультацию врача.
Всегда проверяйте информацию по первоисточникам.
```

```mermaid
flowchart LR
    CHECK["✅ ЭТИЧЕСКИЙ ЧЕК-ЛИСТ RAG"]
    
    CHECK --> Q1["Источники проверены<br>и рецензированы?"]
    Q1 -->|Да| Q2["База знаний<br>регулярно обновляется?"]
    Q1 -->|Нет| FAIL["❌ НЕ ИСПОЛЬЗОВАТЬ"]
    
    Q2 -->|Да| Q3["Пользователь знает,<br>что это ИИ?"]
    Q2 -->|Нет| FAIL
    
    Q3 -->|Да| Q4["Есть ссылки<br>на источники?"]
    Q3 -->|Нет| FAIL
    
    Q4 -->|Да| PASS["✅ МОЖНО ИСПОЛЬЗОВАТЬ"]
    Q4 -->|Нет| FAIL
    
    style CHECK fill:#fff9c4,stroke:#f57f17
    style PASS fill:#c8e6c9,stroke:#2e7d32
    style FAIL fill:#ffcdd2,stroke:#c62828
```

> RAG – это не просто «ещё одна технология». Это **способ сделать LLM честными**. Вместо того чтобы гадать, модель **ищет** ответ в ваших документах и **показывает** источники. В стоматологии, где цена ошибки высока, это превращает LLM из «рискованного эксперимента» в **надёжного ассистента**.
